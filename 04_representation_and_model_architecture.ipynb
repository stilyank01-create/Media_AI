# Media AI — Representation and Model Architecture

## Purpose

This notebook defines and implements the representation and model architecture of Media AI.

It transforms the structurally prepared corpus and supervision framework established in the preceding notebooks into hierarchical Media AI representations while preserving the relationship between model states and the original language from which they are derived.

The notebook operates across four textual levels:

**Token → Sentence → Paragraph → Article**

and three principal representational spaces:

**Factual → Psychological → Social**

These representations provide the basis for the transformative, recurrent and confluent mechanisms of the Media AI architecture.

## Methodological Role

Notebook 04 forms the architectural bridge between supervision and experimentation.

The preceding notebooks establish:

* the research corpus and source structure;
* prepared and traceable textual units;
* the annotation ontology;
* independent LLM supervision;
* reference-supervision and review pathways.

This notebook does not redefine those structures.

Instead, it constructs the representations required for Media AI modelling and defines how information may propagate and aggregate across textual levels.

The resulting architecture will subsequently be evaluated in Notebook 05 through experiments, baselines, ablations, trajectories, attribution, counterfactual analysis and textual inspection.

## Hierarchical Representation

Media AI treats textual hierarchy as part of the methodology rather than merely as a data-storage structure.

Representations may therefore exist at four connected levels:

[
T^{token}
\rightarrow
T^{sentence}
\rightarrow
T^{paragraph}
\rightarrow
T^{article}
]

Higher-level representations must not be constructed through unexamined averaging of lower-level values.

Aggregation rules must be explicit, reproducible and appropriate to the represented construct.

Sentence-level evidence may contribute differently to paragraph-level state, and paragraph-level developments may contribute differently to the overall article representation.

The architecture must therefore preserve both aggregation and contribution information.

## Representational Spaces

Media AI distinguishes three principal representational spaces.

### Factual Space

The factual space represents the informational contribution of the text.

It concerns externally assessable claims, observations, attribution, quantitative information, evidence references, certainty, uncertainty and related factual structure.

It does not constitute external fact checking or a truth score.

### Psychological Space

The psychological space represents emotional and psychological state and change.

It preserves the distinction between:

* reported emotion;
* transformative emotion;
* immediate emotional state;
* emotional weight;
* uncertainty and confidence where applicable.

Transformative emotion is treated as change introduced by a textual unit rather than as a synonym for emotional intensity.

### Social Space

The social space represents entities, groups, relationships and socially relevant structures expressed or constructed through the text.

Its design must permit overlapping and non-exclusive relationships rather than requiring every entity to belong to a single fixed category.

This provides the basis for later confluent representations.

## Architectural Mechanisms

Three mechanisms are central to the Media AI architecture:

**Transformative representation** describes how a textual unit changes an existing representational state.

**Recurrent representation** describes how state develops sequentially as the article is processed.

**Confluent representation** permits multiple representational and social relationships to coexist and interact without forcing them into a single exclusive structure.

These mechanisms are defined and implemented as architectural components in this notebook.

Their empirical value is not assumed.

Their contribution must later be evaluated experimentally against appropriate baselines and ablations.

## Traceability and Attribution

Interpretability is a methodological requirement of the architecture rather than a post-processing visualisation step.

Every higher-level Media AI representation must retain a traceable relationship to the lower-level textual units that contributed to it.

The intended attribution path is:

[
Article
\rightarrow
Paragraph
\rightarrow
Sentence
\rightarrow
Token
\rightarrow
Original\ Text
]

A quantitative model result must therefore remain inspectable in relation to the language that produced it.

For example, an observed increase in article-level psychological fear should be traceable to contributing paragraphs, sentences and, where supported by the architecture, individual textual elements.

This requirement constrains both representation construction and aggregation.

## Interpretability Framework

Interpretability is incorporated explicitly into the Media AI research framework:

[
Research\ Question
\rightarrow
Hypotheses
\rightarrow
Method
\rightarrow
Experiment
\rightarrow
Evidence
\rightarrow
Interpretation
]

Within this framework:

[
Method
\supset
Classification
+
Attribution
+
Colour\ Inspection
]

and:

[
Evidence
\supset
Scores
+
Trajectories
+
Coloured\ Textual\ Evidence
]

The purpose is to allow a reader to move from a quantitative conclusion back to the textual evidence associated with that conclusion.

Notebook 04 establishes the architectural pathways required for this process.

Notebook 05 will evaluate and inspect those pathways experimentally.

## Research Boundary

This notebook constructs the Media AI representation and model architecture.

It does not treat architectural complexity as evidence of model effectiveness.

In particular, the notebook does not by itself establish that:

* hierarchical aggregation improves performance;
* transformative representations outperform static representations;
* recurrent state improves explanatory or predictive performance;
* confluent representations improve social modelling;
* attribution accurately identifies causal textual influence.

These questions belong to the experimental and evidential stage.

Notebook 04 therefore establishes the architecture required to test them while maintaining structural traceability, methodological reproducibility and separation between representation design and empirical validation.

### Block 8 — Social and Confluent Representation

The architecture must permit:

* entities;
* groups;
* relationships;
* overlapping memberships;
* simultaneous social associations;
* contextual changes in relationships.

An entity must not be forced into one mutually exclusive group where the text supports multiple concurrent relationships.

This block establishes the representational mechanism; its empirical usefulness is evaluated later.

### Block 9 — Transformative State Construction

Defines how textual units contribute changes to an existing Media AI state.

Transformative representation is treated as a state transition rather than as a static emotional or semantic label.

Conceptually:

[
\Delta T_t = f(T_{t-1}, X_t)
]

where the current textual unit may increase, decrease or preserve components of the existing state.

The block must preserve the distinction between:

* observed state;
* introduced change;
* resulting state.

This distinction is required for later trajectory analysis and counterfactual experiments.

### Block 10 — Recurrent State Architecture

Defines the sequential accumulation of Media AI state through an article.

Conceptually:

[
T_t = g(T_{t-1}, \Delta T_t, X_t)
]

The recurrent structure must respect textual order and must not use future textual units when constructing an earlier state.

State trajectories should remain available at sentence and paragraph boundaries.

This block establishes the recurrent architecture and its outputs without treating recurrent behaviour as experimentally validated.

### Block 11 — Hierarchical Aggregation Architecture

Defines how representations move between textual levels:

[
T^{token}
\rightarrow
T^{sentence}
\rightarrow
T^{paragraph}
\rightarrow
T^{article}
]

Aggregation must be explicit and construct-sensitive.

The notebook must not assume that simple arithmetic averaging is an adequate aggregation rule for every representational dimension.

The architecture should support comparison between alternative aggregation mechanisms in Notebook 05, including appropriate baseline aggregation.

Higher-level representations must preserve information about the lower-level units that contributed to them.

### Representation Configuration

The notebook operates across four textual levels:

**Token → Sentence → Paragraph → Article**

and three principal representational spaces:

**Factual → Psychological → Social**

These structures are treated as explicit methodological configuration rather than implicit implementation details.

Later blocks must preserve the distinction between textual hierarchy, representational space and architectural mechanism.

### Aggregation Policy

Hierarchical aggregation is controlled explicitly.

No downstream block may assume that arithmetic averaging is the default methodological solution for moving between token, sentence, paragraph and article representations.

Aggregation mechanisms must be identified, reproducible and traceability-preserving.

Baseline aggregation may be supported for later comparison, but the configured architecture must retain the relationship between higher-level states and the lower-level units that contributed to them.

### Temporal and State Policy

Transformative and recurrent representations must respect textual order.

A state constructed at textual position (t) may depend on the current and preceding permitted information but must not incorporate future textual units.

The notebook therefore preserves the distinction between:

* previous state;
* current textual contribution;
* transformative change;
* resulting state.

Future-context leakage is not permitted in recurrent or transformative state construction.

### Traceability and Attribution Policy

Canonical research text remains the authoritative textual source.

Model-specific tokenisation, embeddings, tensors and other derived representations must remain linked to the original article, paragraph and sentence identifiers.

Higher-level representations must retain sufficient contribution information to support backward inspection through:

[
Article
\rightarrow
Paragraph
\rightarrow
Sentence
\rightarrow
Token
\rightarrow
Original\ Text
]

Attribution support is therefore an architectural requirement of Notebook 04 rather than an optional visualisation feature.

### Supervision Boundary

Canonical supervision produced by Notebook 03 may be consumed as a research input where available.

Notebook 04 must not:

* silently convert missing supervision into numerical zero;
* treat unvalidated provider responses as canonical labels;
* overwrite canonical supervision records;
* modify the annotation ontology;
* infer that supervision confidence constitutes model confidence.

Supervision provenance and value-state semantics must remain preserved when supervision enters a Media AI representation.

Notebook 04 may construct model components and representations required by the Media AI architecture.

It does not use their outputs to establish empirical superiority.

Model comparison, benchmark evaluation, ablation studies, counterfactual experiments and substantive interpretation remain the responsibility of Notebook 05.

This distinction keeps architectural construction separate from experimental evidence.

````python
# =============================================================================
# Media AI — Notebook 04 Environment, Reproducibility and Configuration
# =============================================================================

import sys
import platform
import random
from datetime import datetime, timezone

import numpy as np
import pandas as pd


# -----------------------------------------------------------------------------
# Notebook configuration
# -----------------------------------------------------------------------------

PROJECT_NAME = "Media AI"

NOTEBOOK_NAME = "04_representation_and_model_architecture"
NOTEBOOK_VERSION = "1.0"

REPRESENTATION_SCHEMA_VERSION = "1.0"
ARCHITECTURE_VERSION = "1.0"
AGGREGATION_POLICY_VERSION = "1.0"

RANDOM_SEED = 42

random.seed(
    RANDOM_SEED
)

np.random.seed(
    RANDOM_SEED
)


# -----------------------------------------------------------------------------
# Execution Control Panel
# -----------------------------------------------------------------------------

# All operational switches are defined centrally here.
#
# Later blocks must consume these values and must not redefine them.


# Persistent research-artifact loading.

LOAD_PREPARED_CORPUS = True

LOAD_CANONICAL_SUPERVISION = True


# Model-facing text representation.

BUILD_MODEL_READY_TEXT = False


# Media AI representational spaces.

BUILD_FACTUAL_REPRESENTATION = False

BUILD_PSYCHOLOGICAL_REPRESENTATION = False

BUILD_SOCIAL_REPRESENTATION = False

BUILD_CONFLUENT_REPRESENTATION = False


# Media AI state architecture.

BUILD_TRANSFORMATIVE_STATE = False

BUILD_RECURRENT_STATE = False


# Hierarchical representation.

BUILD_HIERARCHICAL_AGGREGATION = False


# Attribution and traceability.

BUILD_ATTRIBUTION_MAPPINGS = False


# Notebook 05 handover.

BUILD_NOTEBOOK_05_HANDOVER = False


# Persistence of Notebook 04 artefacts.

SAVE_REPRESENTATION_OUTPUTS = False


# -----------------------------------------------------------------------------
# Textual hierarchy configuration
# -----------------------------------------------------------------------------

TEXTUAL_HIERARCHY_LEVELS = (
    "token",
    "sentence",
    "paragraph",
    "article",
)

CANONICAL_ANNOTATION_LEVEL = "sentence"

EXTERNAL_SUPERVISION_REQUEST_LEVEL = "article"


# -----------------------------------------------------------------------------
# Media AI representational spaces
# -----------------------------------------------------------------------------

MEDIA_AI_REPRESENTATION_SPACES = (
    "factual",
    "psychological",
    "social",
)


# -----------------------------------------------------------------------------
# Architectural mechanisms
# -----------------------------------------------------------------------------

MEDIA_AI_ARCHITECTURAL_MECHANISMS = (
    "transformative",
    "recurrent",
    "confluent",
)


# -----------------------------------------------------------------------------
# Hierarchical aggregation policy
# -----------------------------------------------------------------------------

REQUIRE_EXPLICIT_AGGREGATION_METHOD = True

ALLOW_UNDECLARED_SIMPLE_AVERAGING = False

REQUIRE_TRACEABILITY_PRESERVING_AGGREGATION = True

REQUIRE_LOWER_LEVEL_CONTRIBUTION_MAPPING = True


# -----------------------------------------------------------------------------
# Temporal and state policy
# -----------------------------------------------------------------------------

PRESERVE_TEXTUAL_ORDER = True

ALLOW_FUTURE_CONTEXT_IN_STATE_CONSTRUCTION = False

REQUIRE_PREVIOUS_STATE_SEPARATION = True

REQUIRE_TRANSFORMATIVE_CHANGE_SEPARATION = True

REQUIRE_RESULTING_STATE_SEPARATION = True


# -----------------------------------------------------------------------------
# Canonical-text and structural-traceability policy
# -----------------------------------------------------------------------------

PRESERVE_CANONICAL_TEXT = True

PRESERVE_STRUCTURAL_TRACEABILITY = True

OVERWRITE_PREPARED_RECORDS = False

OVERWRITE_CANONICAL_SUPERVISION = False

REQUIRE_TOKEN_TO_SENTENCE_MAPPING = True

REQUIRE_SENTENCE_TO_PARAGRAPH_MAPPING = True

REQUIRE_PARAGRAPH_TO_ARTICLE_MAPPING = True


# -----------------------------------------------------------------------------
# Attribution policy
# -----------------------------------------------------------------------------

REQUIRE_BACKWARD_ATTRIBUTION_PATH = True

ATTRIBUTION_TARGET_PATH = (
    "article",
    "paragraph",
    "sentence",
    "token",
    "original_text",
)

REQUIRE_ATTRIBUTION_PROVENANCE = True


# -----------------------------------------------------------------------------
# Supervision-consumption policy
# -----------------------------------------------------------------------------

REQUIRE_CANONICAL_SUPERVISION_ONLY = True

ALLOW_RAW_PROVIDER_SUPERVISION = False

PRESERVE_SUPERVISION_VALUE_STATES = True

PRESERVE_SUPERVISION_PROVENANCE = True

ALLOW_MISSING_SUPERVISION_AS_ZERO = False

TREAT_SUPERVISION_CONFIDENCE_AS_MODEL_CONFIDENCE = False


# -----------------------------------------------------------------------------
# Model-architecture boundary
# -----------------------------------------------------------------------------

MODEL_TRAINING_ALLOWED = False

MODEL_EVALUATION_ALLOWED = False

ABLATION_EXPERIMENTS_ALLOWED = False

COUNTERFACTUAL_EXPERIMENTS_ALLOWED = False

SUBSTANTIVE_INTERPRETATION_ALLOWED = False


# -----------------------------------------------------------------------------
# Model-specific representation policy
# -----------------------------------------------------------------------------

# Model-specific tokenisation and embeddings belong to later blocks.
#
# They must remain derived analytical structures and must not replace
# the canonical research text.

PERMANENT_MODEL_TOKENISATION = False

MODEL_SPECIFIC_TEXT_OVERWRITE_ALLOWED = False

REQUIRE_MODEL_TOKEN_TRACEABILITY = True


# -----------------------------------------------------------------------------
# Persistence policy
# -----------------------------------------------------------------------------

REQUIRE_ARCHITECTURE_PROVENANCE = True

REQUIRE_REPRESENTATION_PROVENANCE = True

REQUIRE_AGGREGATION_PROVENANCE = True

REQUIRE_RUNTIME_METADATA = True


# -----------------------------------------------------------------------------
# Runtime metadata
# -----------------------------------------------------------------------------

RUNTIME_METADATA = {
    "project":
        PROJECT_NAME,

    "notebook":
        NOTEBOOK_NAME,

    "notebook_version":
        NOTEBOOK_VERSION,

    "representation_schema_version":
        REPRESENTATION_SCHEMA_VERSION,

    "architecture_version":
        ARCHITECTURE_VERSION,

    "aggregation_policy_version":
        AGGREGATION_POLICY_VERSION,

    "python_version":
        sys.version.split()[0],

    "platform":
        platform.platform(),

    "numpy_version":
        np.__version__,

    "pandas_version":
        pd.__version__,

    "random_seed":
        RANDOM_SEED,

    "initialised_at_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),
}


# -----------------------------------------------------------------------------
# Configuration safeguards
# -----------------------------------------------------------------------------

if (
    TEXTUAL_HIERARCHY_LEVELS
    != (
        "token",
        "sentence",
        "paragraph",
        "article",
    )
):
    raise ValueError(
        "Notebook 04 textual hierarchy must remain "
        "Token → Sentence → Paragraph → Article."
    )


if set(
    MEDIA_AI_REPRESENTATION_SPACES
) != {
    "factual",
    "psychological",
    "social",
}:
    raise ValueError(
        "Notebook 04 must preserve the factual, "
        "psychological and social representational spaces."
    )


if set(
    MEDIA_AI_ARCHITECTURAL_MECHANISMS
) != {
    "transformative",
    "recurrent",
    "confluent",
}:
    raise ValueError(
        "Notebook 04 must preserve the transformative, "
        "recurrent and confluent architectural mechanisms."
    )


if not PRESERVE_CANONICAL_TEXT:
    raise ValueError(
        "Notebook 04 must preserve canonical research text."
    )


if not PRESERVE_STRUCTURAL_TRACEABILITY:
    raise ValueError(
        "Notebook 04 must preserve structural traceability."
    )


if OVERWRITE_PREPARED_RECORDS:
    raise ValueError(
        "Notebook 04 must not overwrite prepared corpus records."
    )


if OVERWRITE_CANONICAL_SUPERVISION:
    raise ValueError(
        "Notebook 04 must not overwrite canonical supervision records."
    )


if ALLOW_UNDECLARED_SIMPLE_AVERAGING:
    raise ValueError(
        "Undeclared arithmetic averaging is not permitted "
        "as a hierarchical aggregation policy."
    )


if not REQUIRE_EXPLICIT_AGGREGATION_METHOD:
    raise ValueError(
        "Hierarchical aggregation methods must be explicit."
    )


if not REQUIRE_TRACEABILITY_PRESERVING_AGGREGATION:
    raise ValueError(
        "Hierarchical aggregation must preserve traceability."
    )


if ALLOW_FUTURE_CONTEXT_IN_STATE_CONSTRUCTION:
    raise ValueError(
        "Future textual context is not permitted in transformative "
        "or recurrent state construction."
    )


if not PRESERVE_TEXTUAL_ORDER:
    raise ValueError(
        "Notebook 04 must preserve textual order."
    )


if not REQUIRE_TRANSFORMATIVE_CHANGE_SEPARATION:
    raise ValueError(
        "Transformative change must remain distinct from state."
    )


if not REQUIRE_PREVIOUS_STATE_SEPARATION:
    raise ValueError(
        "Previous state must remain explicitly represented."
    )


if not REQUIRE_RESULTING_STATE_SEPARATION:
    raise ValueError(
        "Resulting state must remain distinct from state change."
    )


if ALLOW_RAW_PROVIDER_SUPERVISION:
    raise ValueError(
        "Raw external-provider supervision must not enter "
        "Notebook 04 representations."
    )


if ALLOW_MISSING_SUPERVISION_AS_ZERO:
    raise ValueError(
        "Missing supervision must remain distinct from numeric zero."
    )


if TREAT_SUPERVISION_CONFIDENCE_AS_MODEL_CONFIDENCE:
    raise ValueError(
        "Supervision confidence must not be interpreted "
        "as Media AI model confidence."
    )


if PERMANENT_MODEL_TOKENISATION:
    raise ValueError(
        "Permanent model-specific tokenisation is not permitted."
    )


if MODEL_SPECIFIC_TEXT_OVERWRITE_ALLOWED:
    raise ValueError(
        "Model-specific text representations must not overwrite "
        "canonical research text."
    )


if MODEL_TRAINING_ALLOWED:
    raise ValueError(
        "Model training is outside the current Notebook 04 execution boundary."
    )


if MODEL_EVALUATION_ALLOWED:
    raise ValueError(
        "Comparative model evaluation belongs to Notebook 05."
    )


if ABLATION_EXPERIMENTS_ALLOWED:
    raise ValueError(
        "Ablation experiments belong to Notebook 05."
    )


if COUNTERFACTUAL_EXPERIMENTS_ALLOWED:
    raise ValueError(
        "Counterfactual experiments belong to Notebook 05."
    )


if SUBSTANTIVE_INTERPRETATION_ALLOWED:
    raise ValueError(
        "Substantive interpretation belongs to Notebook 05."
    )


if not REQUIRE_BACKWARD_ATTRIBUTION_PATH:
    raise ValueError(
        "Backward attribution is a required architectural property."
    )


if (
    ATTRIBUTION_TARGET_PATH
    != (
        "article",
        "paragraph",
        "sentence",
        "token",
        "original_text",
    )
):
    raise ValueError(
        "Attribution target path must remain "
        "Article → Paragraph → Sentence → Token → Original Text."
    )


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Notebook 04 Initialisation")
print("=" * 72)

print(
    f"Notebook                     : "
    f"{NOTEBOOK_NAME}"
)

print(
    f"Notebook version             : "
    f"{NOTEBOOK_VERSION}"
)

print(
    f"Representation schema        : "
    f"{REPRESENTATION_SCHEMA_VERSION}"
)

print(
    f"Architecture version         : "
    f"{ARCHITECTURE_VERSION}"
)

print(
    f"Aggregation policy           : "
    f"{AGGREGATION_POLICY_VERSION}"
)

print(
    f"Random seed                  : "
    f"{RANDOM_SEED}"
)

print("-" * 72)

print("Execution Control Panel")

print(
    f"Load prepared corpus         : "
    f"{LOAD_PREPARED_CORPUS}"
)

print(
    f"Load canonical supervision   : "
    f"{LOAD_CANONICAL_SUPERVISION}"
)

print(
    f"Build model-ready text       : "
    f"{BUILD_MODEL_READY_TEXT}"
)

print(
    f"Build factual representation : "
    f"{BUILD_FACTUAL_REPRESENTATION}"
)

print(
    f"Build psychological repr.    : "
    f"{BUILD_PSYCHOLOGICAL_REPRESENTATION}"
)

print(
    f"Build social representation  : "
    f"{BUILD_SOCIAL_REPRESENTATION}"
)

print(
    f"Build confluent repr.         : "
    f"{BUILD_CONFLUENT_REPRESENTATION}"
)

print(
    f"Build transformative state   : "
    f"{BUILD_TRANSFORMATIVE_STATE}"
)

print(
    f"Build recurrent state        : "
    f"{BUILD_RECURRENT_STATE}"
)

print(
    f"Build hierarchical aggregate : "
    f"{BUILD_HIERARCHICAL_AGGREGATION}"
)

print(
    f"Build attribution mappings   : "
    f"{BUILD_ATTRIBUTION_MAPPINGS}"
)

print(
    f"Build Notebook 05 handover   : "
    f"{BUILD_NOTEBOOK_05_HANDOVER}"
)

print(
    f"Save representation outputs  : "
    f"{SAVE_REPRESENTATION_OUTPUTS}"
)

print("-" * 72)

print("Representation architecture")

print(
    f"Textual hierarchy            : "
    f"{' → '.join(TEXTUAL_HIERARCHY_LEVELS)}"
)

print(
    f"Representation spaces        : "
    f"{', '.join(MEDIA_AI_REPRESENTATION_SPACES)}"
)

print(
    f"Architectural mechanisms     : "
    f"{', '.join(MEDIA_AI_ARCHITECTURAL_MECHANISMS)}"
)

print(
    f"Annotation level             : "
    f"{CANONICAL_ANNOTATION_LEVEL}"
)

print(
    f"Supervision request level    : "
    f"{EXTERNAL_SUPERVISION_REQUEST_LEVEL}"
)

print("-" * 72)

print("Methodological safeguards")

print(
    f"Preserve canonical text      : "
    f"{PRESERVE_CANONICAL_TEXT}"
)

print(
    f"Preserve traceability        : "
    f"{PRESERVE_STRUCTURAL_TRACEABILITY}"
)

print(
    f"Explicit aggregation         : "
    f"{REQUIRE_EXPLICIT_AGGREGATION_METHOD}"
)

print(
    f"Traceable aggregation        : "
    f"{REQUIRE_TRACEABILITY_PRESERVING_AGGREGATION}"
)

print(
    f"Future context allowed       : "
    f"{ALLOW_FUTURE_CONTEXT_IN_STATE_CONSTRUCTION}"
)

print(
    f"Canonical supervision only   : "
    f"{REQUIRE_CANONICAL_SUPERVISION_ONLY}"
)

print(
    f"Missing supervision as zero  : "
    f"{ALLOW_MISSING_SUPERVISION_AS_ZERO}"
)

print(
    f"Backward attribution required: "
    f"{REQUIRE_BACKWARD_ATTRIBUTION_PATH}"
)

print("-" * 72)

print("Experimental boundary")

print(
    f"Model training               : "
    f"{MODEL_TRAINING_ALLOWED}"
)

print(
    f"Model evaluation             : "
    f"{MODEL_EVALUATION_ALLOWED}"
)

print(
    f"Ablation experiments         : "
    f"{ABLATION_EXPERIMENTS_ALLOWED}"
)

print(
    f"Counterfactual experiments   : "
    f"{COUNTERFACTUAL_EXPERIMENTS_ALLOWED}"
)

print(
    f"Substantive interpretation   : "
    f"{SUBSTANTIVE_INTERPRETATION_ALLOWED}"
)

print("=" * 72)

print(
    "Notebook 04 representation and model-architecture "
    "environment initialised successfully."
)

print(
    "Execution controls are centralised in Block 1."
)

print(
    "The Media AI hierarchy, representation spaces and "
    "architectural mechanisms are explicitly configured."
)

print(
    "Traceability-preserving aggregation and backward attribution "
    "are required architectural properties."
)

print(
    "No research artefacts have been loaded by this block."
)

print(
    "No model-specific tokenisation or embedding has been performed."
)

print(
    "No Media AI representation or state has been constructed."
)

print(
    "No model training, experimental evaluation or interpretation "
    "has been performed."
)
````

## Notebook 03 Handover and Research-Artefact Loading

This block loads and validates the research artefacts required by Notebook 04.

The handover separates structurally required corpus data from supervision-dependent annotation data.

Notebook 04 must remain capable of constructing and validating representation architecture when canonical supervision is incomplete, while clearly identifying which downstream components depend on supervision that is not yet available.

### Prepared Corpus

The prepared corpus produced by the preceding data-preparation stage is the authoritative textual input to Notebook 04.

The block loads the prepared corpus when:

`LOAD_PREPARED_CORPUS = True`

The loaded structure must preserve:

* stable Media AI record identifiers;
* article identifiers and metadata;
* paragraph identifiers and ordering;
* sentence identifiers and ordering;
* complete prepared text;
* structural parent-child relationships;
* available source provenance.

Prepared records are treated as immutable research inputs.

Notebook 04 must not modify or overwrite the canonical prepared text.

### Canonical Supervision

Canonical supervision produced by Notebook 03 is loaded when:

`LOAD_CANONICAL_SUPERVISION = True`

Only supervision that has passed the Notebook 03 canonical validation boundary may enter Notebook 04.

Raw provider responses, provider-successful but unvalidated outputs and rejected annotations must not be treated as canonical supervision.

The handover therefore distinguishes between:

* canonical validated supervision;
* supervision that is not yet available;
* incomplete article supervision;
* rejected supervision where diagnostic information is available.

### Partial-Supervision Policy

Complete supervision of the research corpus is not required merely to initialise Notebook 04.

This distinction allows representation architecture to be developed independently of temporary external-supervision constraints.

However, later operations that require canonical supervision must operate only on records for which the required validated annotations are available.

Missing supervision must remain explicitly missing.

It must not be:

* converted to numerical zero;
* imputed without an explicit later methodology;
* inferred from neighbouring annotations;
* replaced with raw provider output;
* interpreted as absence of the represented construct.

### Article-Level Completeness

Although canonical annotation remains sentence-level, Notebook 03 uses complete articles as the external LLM supervision request unit.

Notebook 04 must therefore preserve both levels of information:

**Article-level supervision completeness**

and

**Sentence-level canonical annotation availability**

An article may be classified as fully supervised only when all sentence-level supervision units required by the configured annotation protocol have passed canonical validation for the relevant supervisor or reference source.

Partially supervised articles must remain distinguishable from fully supervised articles.

### Identity Alignment

Canonical supervision must be aligned to the prepared corpus through stable research identifiers rather than through textual matching alone.

At minimum, the handover must verify consistency of:

* Media AI record ID;
* paragraph ID;
* sentence ID;
* paragraph index;
* sentence index;
* article membership.

Text may additionally be checked for consistency, but text matching must not replace identifier-based alignment.

This identity structure provides the basis for later hierarchical representation and attribution.

### Provenance Preservation

Notebook 04 must preserve supervision provenance when canonical annotations are consumed.

Relevant provenance may include:

* supervision-schema version;
* annotation-policy version;
* supervision method;
* supervisor identifier;
* provider;
* model;
* annotation timestamp;
* context scope;
* review status;
* annotation confidence.

These fields remain metadata associated with supervision and must not be silently converted into Media AI model features.

### Handover Validation

The block performs structural checks before research artefacts are exposed to later architectural stages.

Validation should establish:

* whether the prepared corpus was loaded successfully;
* whether canonical supervision was requested;
* whether canonical supervision is available;
* how many canonical sentence annotations are available;
* how many articles are fully, partially or not supervised;
* whether supervision identifiers align with the prepared corpus;
* whether duplicate canonical supervision records exist;
* whether required provenance is present;
* whether prepared text and structural identifiers remain intact.

Incomplete supervision is reported as a research-data state rather than automatically treated as a notebook failure.

Structural corruption, identity conflicts or invalid canonical handover records must fail explicitly.

### Output

This block creates the validated Notebook 04 research-input layer.

Its outputs provide:

* the immutable prepared corpus;
* the available canonical supervision records;
* supervision-coverage diagnostics;
* article-level supervision-completeness status;
* validated identity mappings;
* handover provenance and validation results.

These structures become the controlled inputs to the hierarchical and representational blocks that follow.

No model-specific tokenisation or embedding is performed.

No Media AI representation is constructed.

No transformative or recurrent state is calculated.

No aggregation or attribution is performed.

No model training or experimental evaluation is performed.


````python
# =============================================================================
# Media AI — Notebook 03 Handover and Research-Artefact Loading
# =============================================================================

import io
import json
from copy import deepcopy

import google.auth
from google.colab import auth
from googleapiclient.discovery import build


# -----------------------------------------------------------------------------
# Block 1 availability check
# -----------------------------------------------------------------------------

required_block_1_objects = [
    "LOAD_PREPARED_CORPUS",
    "LOAD_CANONICAL_SUPERVISION",
    "PRESERVE_CANONICAL_TEXT",
    "PRESERVE_STRUCTURAL_TRACEABILITY",
    "REQUIRE_CANONICAL_SUPERVISION_ONLY",
    "ALLOW_RAW_PROVIDER_SUPERVISION",
    "ALLOW_MISSING_SUPERVISION_AS_ZERO",
]

missing_block_1_objects = [
    object_name
    for object_name
    in required_block_1_objects
    if object_name not in globals()
]

if missing_block_1_objects:
    raise NameError(
        "Notebook 04 Block 1 is not initialised. "
        "Missing: "
        f"{missing_block_1_objects}"
    )


# -----------------------------------------------------------------------------
# Persistent-storage configuration
# -----------------------------------------------------------------------------

PROJECT_FOLDER_NAME = "Media_AI"

DATA_FOLDER_NAME = "data"
PREPARED_FOLDER_NAME = "prepared"
ANNOTATED_FOLDER_NAME = "annotated"
GOLD_FOLDER_NAME = "gold"
LOGS_FOLDER_NAME = "logs"

NOTEBOOK_02_HANDOVER_FILENAME = (
    "notebook_02_handover_manifest.json"
)

PREPARED_CORPUS_FILENAME = (
    "notebook_02_prepared_corpus.json"
)

# Notebook 03 persistence remains optional at this stage.
#
# These deterministic names are reserved for later Notebook 03 persistence.
# Their absence must not prevent Notebook 04 structural initialisation.

NOTEBOOK_03_HANDOVER_FILENAME = (
    "notebook_03_handover_manifest.json"
)

NOTEBOOK_03_REFERENCE_DATASET_FILENAME = (
    "notebook_03_reference_dataset.json"
)

NOTEBOOK_03_CANONICAL_SUPERVISION_FILENAME = (
    "notebook_03_canonical_supervision.json"
)


# -----------------------------------------------------------------------------
# Google Drive authorisation
# -----------------------------------------------------------------------------

auth.authenticate_user()

credentials, _ = google.auth.default()

drive_service = build(
    "drive",
    "v3",
    credentials=credentials,
)


# -----------------------------------------------------------------------------
# Google Drive helpers
# -----------------------------------------------------------------------------

def escape_drive_query_value(
    value,
):
    """
    Escape a string for use inside a Google Drive query.
    """

    return str(
        value
    ).replace(
        "\\",
        "\\\\"
    ).replace(
        "'",
        "\\'"
    )


def find_drive_folder(
    folder_name,
    parent_id=None,
):
    """
    Find one non-trashed Google Drive folder by name and optional parent.
    """

    escaped_name = escape_drive_query_value(
        folder_name
    )

    query_parts = [
        f"name = '{escaped_name}'",
        (
            "mimeType = "
            "'application/vnd.google-apps.folder'"
        ),
        "trashed = false",
    ]

    if parent_id is not None:

        escaped_parent = escape_drive_query_value(
            parent_id
        )

        query_parts.append(
            f"'{escaped_parent}' in parents"
        )

    response = (
        drive_service.files()
        .list(
            q=" and ".join(
                query_parts
            ),
            spaces="drive",
            fields="files(id,name,parents)",
            pageSize=100,
        )
        .execute()
    )

    files = response.get(
        "files",
        []
    )

    if not files:
        return None

    return files[0][
        "id"
    ]


def find_drive_file(
    file_name,
    parent_id=None,
):
    """
    Find one non-trashed Google Drive file by name and optional parent.
    """

    escaped_name = escape_drive_query_value(
        file_name
    )

    query_parts = [
        f"name = '{escaped_name}'",
        "trashed = false",
    ]

    if parent_id is not None:

        escaped_parent = escape_drive_query_value(
            parent_id
        )

        query_parts.append(
            f"'{escaped_parent}' in parents"
        )

    response = (
        drive_service.files()
        .list(
            q=" and ".join(
                query_parts
            ),
            spaces="drive",
            fields="files(id,name,mimeType,parents)",
            pageSize=100,
        )
        .execute()
    )

    files = response.get(
        "files",
        []
    )

    if not files:
        return None

    return files[0][
        "id"
    ]


def download_json_from_drive(
    file_id,
):
    """
    Download and decode one JSON research artefact from Google Drive.
    """

    if not file_id:
        raise ValueError(
            "A valid Google Drive file ID is required."
        )

    content = (
        drive_service.files()
        .get_media(
            fileId=file_id
        )
        .execute()
    )

    return json.loads(
        content.decode(
            "utf-8"
        )
    )


# -----------------------------------------------------------------------------
# Resolve persistent Media AI storage
# -----------------------------------------------------------------------------

PROJECT_FOLDER_ID = find_drive_folder(
    PROJECT_FOLDER_NAME
)

if not PROJECT_FOLDER_ID:
    raise FileNotFoundError(
        "The persistent Media_AI project folder "
        "could not be located in Google Drive."
    )


DATA_FOLDER_ID = find_drive_folder(
    DATA_FOLDER_NAME,
    PROJECT_FOLDER_ID,
)

LOGS_FOLDER_ID = find_drive_folder(
    LOGS_FOLDER_NAME,
    PROJECT_FOLDER_ID,
)

if not DATA_FOLDER_ID:
    raise FileNotFoundError(
        "The Media_AI/data folder could not be located."
    )

if not LOGS_FOLDER_ID:
    raise FileNotFoundError(
        "The Media_AI/logs folder could not be located."
    )


PREPARED_DATA_FOLDER_ID = find_drive_folder(
    PREPARED_FOLDER_NAME,
    DATA_FOLDER_ID,
)

ANNOTATED_DATA_FOLDER_ID = find_drive_folder(
    ANNOTATED_FOLDER_NAME,
    DATA_FOLDER_ID,
)

GOLD_DATA_FOLDER_ID = find_drive_folder(
    GOLD_FOLDER_NAME,
    DATA_FOLDER_ID,
)


if (
    LOAD_PREPARED_CORPUS
    and not PREPARED_DATA_FOLDER_ID
):
    raise FileNotFoundError(
        "The Media_AI/data/prepared folder "
        "could not be located."
    )


# -----------------------------------------------------------------------------
# Notebook 02 handover loading
# -----------------------------------------------------------------------------

NOTEBOOK_02_HANDOVER = None
NOTEBOOK_02_HANDOVER_FILE_ID = None

if LOAD_PREPARED_CORPUS:

    NOTEBOOK_02_HANDOVER_FILE_ID = find_drive_file(
        NOTEBOOK_02_HANDOVER_FILENAME,
        LOGS_FOLDER_ID,
    )

    if not NOTEBOOK_02_HANDOVER_FILE_ID:
        raise FileNotFoundError(
            "Notebook 02 handover manifest was not found: "
            f"{NOTEBOOK_02_HANDOVER_FILENAME}"
        )

    NOTEBOOK_02_HANDOVER = (
        download_json_from_drive(
            NOTEBOOK_02_HANDOVER_FILE_ID
        )
    )


# -----------------------------------------------------------------------------
# Prepared-corpus loading
# -----------------------------------------------------------------------------

PREPARED_CORPUS = None
PREPARED_CORPUS_FILE_ID = None
PREPARED_RECORDS = []

if LOAD_PREPARED_CORPUS:

    handover_prepared_corpus = (
        NOTEBOOK_02_HANDOVER.get(
            "prepared_corpus",
            {}
        )
        if isinstance(
            NOTEBOOK_02_HANDOVER,
            dict,
        )
        else {}
    )

    PREPARED_CORPUS_FILE_ID = (
        handover_prepared_corpus.get(
            "file_id"
        )
    )

    if not PREPARED_CORPUS_FILE_ID:

        PREPARED_CORPUS_FILE_ID = (
            find_drive_file(
                PREPARED_CORPUS_FILENAME,
                PREPARED_DATA_FOLDER_ID,
            )
        )

    if not PREPARED_CORPUS_FILE_ID:
        raise FileNotFoundError(
            "Canonical Notebook 02 prepared corpus "
            "could not be located."
        )

    PREPARED_CORPUS = (
        download_json_from_drive(
            PREPARED_CORPUS_FILE_ID
        )
    )

    if (
        PREPARED_CORPUS.get(
            "artifact_type"
        )
        != "media_ai_prepared_corpus"
    ):
        raise ValueError(
            "Loaded prepared corpus has an unexpected "
            "artifact type."
        )

    PREPARED_RECORDS = deepcopy(
        PREPARED_CORPUS.get(
            "records",
            []
        )
    )


# -----------------------------------------------------------------------------
# Reconstruct immutable article → paragraph → sentence views
# -----------------------------------------------------------------------------

ARTICLE_RECORDS = []
PARAGRAPH_RECORDS = []
SENTENCE_RECORDS = []

for article in PREPARED_RECORDS:

    article_id = article.get(
        "media_ai_record_id"
    )

    ARTICLE_RECORDS.append(
        deepcopy(
            article
        )
    )

    for paragraph in article.get(
        "paragraphs",
        []
    ):

        paragraph_record = deepcopy(
            paragraph
        )

        paragraph_record.setdefault(
            "media_ai_record_id",
            article_id,
        )

        PARAGRAPH_RECORDS.append(
            paragraph_record
        )

        for sentence in paragraph.get(
            "sentences",
            []
        ):

            sentence_record = deepcopy(
                sentence
            )

            sentence_record.setdefault(
                "media_ai_record_id",
                article_id,
            )

            sentence_record.setdefault(
                "paragraph_id",
                paragraph.get(
                    "paragraph_id"
                ),
            )

            sentence_record.setdefault(
                "paragraph_index",
                paragraph.get(
                    "paragraph_index"
                ),
            )

            SENTENCE_RECORDS.append(
                sentence_record
            )


# -----------------------------------------------------------------------------
# Prepared hierarchy validation
# -----------------------------------------------------------------------------

ARTICLE_IDS = [
    record.get(
        "media_ai_record_id"
    )
    for record
    in ARTICLE_RECORDS
]

PARAGRAPH_IDS = [
    record.get(
        "paragraph_id"
    )
    for record
    in PARAGRAPH_RECORDS
]

SENTENCE_IDS = [
    record.get(
        "sentence_id"
    )
    for record
    in SENTENCE_RECORDS
]


ARTICLE_IDS_UNIQUE = (
    len(
        ARTICLE_IDS
    )
    == len(
        set(
            ARTICLE_IDS
        )
    )
    and all(
        ARTICLE_IDS
    )
)

PARAGRAPH_IDS_UNIQUE = (
    len(
        PARAGRAPH_IDS
    )
    == len(
        set(
            PARAGRAPH_IDS
        )
    )
    and all(
        PARAGRAPH_IDS
    )
)

SENTENCE_IDS_UNIQUE = (
    len(
        SENTENCE_IDS
    )
    == len(
        set(
            SENTENCE_IDS
        )
    )
    and all(
        SENTENCE_IDS
    )
)


ARTICLE_ID_SET = set(
    ARTICLE_IDS
)

PARAGRAPH_ID_SET = set(
    PARAGRAPH_IDS
)


PARAGRAPH_PARENT_LINKS_VALID = all(
    paragraph.get(
        "media_ai_record_id"
    )
    in ARTICLE_ID_SET
    for paragraph
    in PARAGRAPH_RECORDS
)


SENTENCE_PARENT_LINKS_VALID = all(
    (
        sentence.get(
            "media_ai_record_id"
        )
        in ARTICLE_ID_SET
    )
    and (
        sentence.get(
            "paragraph_id"
        )
        in PARAGRAPH_ID_SET
    )
    for sentence
    in SENTENCE_RECORDS
)


PREPARED_HIERARCHY_VALID = all([
    bool(
        ARTICLE_RECORDS
    ),
    ARTICLE_IDS_UNIQUE,
    PARAGRAPH_IDS_UNIQUE,
    SENTENCE_IDS_UNIQUE,
    PARAGRAPH_PARENT_LINKS_VALID,
    SENTENCE_PARENT_LINKS_VALID,
])


if (
    LOAD_PREPARED_CORPUS
    and not PREPARED_HIERARCHY_VALID
):
    raise ValueError(
        "Prepared-corpus hierarchy failed Notebook 04 "
        "handover validation."
    )


# -----------------------------------------------------------------------------
# Expected prepared-corpus counts from Notebook 02 manifest
# -----------------------------------------------------------------------------

NOTEBOOK_02_EXPECTED_COUNTS = (
    NOTEBOOK_02_HANDOVER.get(
        "counts",
        {}
    )
    if isinstance(
        NOTEBOOK_02_HANDOVER,
        dict,
    )
    else {}
)


ARTICLE_COUNT_MATCHES_HANDOVER = (
    not NOTEBOOK_02_EXPECTED_COUNTS
    or len(
        ARTICLE_RECORDS
    )
    == NOTEBOOK_02_EXPECTED_COUNTS.get(
        "articles"
    )
)


PARAGRAPH_COUNT_MATCHES_HANDOVER = (
    not NOTEBOOK_02_EXPECTED_COUNTS
    or len(
        PARAGRAPH_RECORDS
    )
    == NOTEBOOK_02_EXPECTED_COUNTS.get(
        "paragraphs"
    )
)


SENTENCE_COUNT_MATCHES_HANDOVER = (
    not NOTEBOOK_02_EXPECTED_COUNTS
    or len(
        SENTENCE_RECORDS
    )
    == NOTEBOOK_02_EXPECTED_COUNTS.get(
        "sentences"
    )
)


PREPARED_COUNTS_MATCH_HANDOVER = all([
    ARTICLE_COUNT_MATCHES_HANDOVER,
    PARAGRAPH_COUNT_MATCHES_HANDOVER,
    SENTENCE_COUNT_MATCHES_HANDOVER,
])


if (
    LOAD_PREPARED_CORPUS
    and not PREPARED_COUNTS_MATCH_HANDOVER
):
    raise ValueError(
        "Prepared-corpus counts do not match "
        "the Notebook 02 handover manifest."
    )


# -----------------------------------------------------------------------------
# Canonical Notebook 03 supervision loading
# -----------------------------------------------------------------------------

CANONICAL_SUPERVISION_RECORDS = []

NOTEBOOK_03_HANDOVER = None
NOTEBOOK_03_HANDOVER_FILE_ID = None

CANONICAL_SUPERVISION_SOURCE = "unavailable"

CANONICAL_SUPERVISION_AVAILABLE = False


if LOAD_CANONICAL_SUPERVISION:

    # -------------------------------------------------------------------------
    # Prefer validated supervision already present in the active runtime.
    # -------------------------------------------------------------------------

    runtime_supervision = globals().get(
        "ACCEPTED_SUPERVISION_RECORDS"
    )

    if isinstance(
        runtime_supervision,
        list,
    ) and runtime_supervision:

        CANONICAL_SUPERVISION_RECORDS = deepcopy(
            runtime_supervision
        )

        CANONICAL_SUPERVISION_SOURCE = (
            "active_runtime"
        )

        CANONICAL_SUPERVISION_AVAILABLE = True


    # -------------------------------------------------------------------------
    # Otherwise look for a persisted Notebook 03 handover.
    # -------------------------------------------------------------------------

    else:

        NOTEBOOK_03_HANDOVER_FILE_ID = (
            find_drive_file(
                NOTEBOOK_03_HANDOVER_FILENAME,
                LOGS_FOLDER_ID,
            )
        )

        if NOTEBOOK_03_HANDOVER_FILE_ID:

            NOTEBOOK_03_HANDOVER = (
                download_json_from_drive(
                    NOTEBOOK_03_HANDOVER_FILE_ID
                )
            )

            # -------------------------------------------------------------
            # Resolve a persisted canonical supervision/reference artefact.
            # -------------------------------------------------------------

            candidate_file_id = None

            for section_name in [
                "reference_dataset",
                "canonical_supervision",
                "supervision",
            ]:

                section = (
                    NOTEBOOK_03_HANDOVER.get(
                        section_name,
                        {}
                    )
                )

                if isinstance(
                    section,
                    dict,
                ):

                    candidate_file_id = (
                        section.get(
                            "file_id"
                        )
                    )

                    if candidate_file_id:
                        break


            # -------------------------------------------------------------
            # Deterministic filename fallback.
            # -------------------------------------------------------------

            if not candidate_file_id:

                for (
                    folder_id,
                    file_name,
                ) in [
                    (
                        GOLD_DATA_FOLDER_ID,
                        NOTEBOOK_03_REFERENCE_DATASET_FILENAME,
                    ),
                    (
                        ANNOTATED_DATA_FOLDER_ID,
                        NOTEBOOK_03_CANONICAL_SUPERVISION_FILENAME,
                    ),
                ]:

                    if not folder_id:
                        continue

                    candidate_file_id = (
                        find_drive_file(
                            file_name,
                            folder_id,
                        )
                    )

                    if candidate_file_id:
                        break


            if candidate_file_id:

                supervision_artifact = (
                    download_json_from_drive(
                        candidate_file_id
                    )
                )

                if isinstance(
                    supervision_artifact,
                    dict,
                ):

                    records = (
                        supervision_artifact.get(
                            "records"
                        )
                        or supervision_artifact.get(
                            "annotations"
                        )
                        or supervision_artifact.get(
                            "reference_records"
                        )
                        or []
                    )

                elif isinstance(
                    supervision_artifact,
                    list,
                ):

                    records = (
                        supervision_artifact
                    )

                else:

                    records = []


                if isinstance(
                    records,
                    list,
                ) and records:

                    CANONICAL_SUPERVISION_RECORDS = (
                        deepcopy(
                            records
                        )
                    )

                    CANONICAL_SUPERVISION_SOURCE = (
                        "persistent_notebook_03"
                    )

                    CANONICAL_SUPERVISION_AVAILABLE = True


# -----------------------------------------------------------------------------
# Canonical-supervision boundary validation
# -----------------------------------------------------------------------------

if (
    ALLOW_RAW_PROVIDER_SUPERVISION
):
    raise ValueError(
        "Raw provider supervision is prohibited "
        "from entering Notebook 04."
    )


CANONICAL_SUPERVISION_IDS = []

SUPERVISION_PROVENANCE_COMPLETE = True
SUPERVISION_IDENTITIES_VALID = True


for record in (
    CANONICAL_SUPERVISION_RECORDS
):

    identity = record.get(
        "identity",
        {}
    )

    provenance = record.get(
        "provenance",
        {}
    )

    sentence_id = identity.get(
        "sentence_id"
    )

    media_ai_record_id = identity.get(
        "media_ai_record_id"
    )

    paragraph_id = identity.get(
        "paragraph_id"
    )

    CANONICAL_SUPERVISION_IDS.append(
        sentence_id
    )


    if (
        sentence_id
        not in SENTENCE_ID_SET
        if "SENTENCE_ID_SET" in globals()
        else sentence_id
        not in set(
            SENTENCE_IDS
        )
    ):
        SUPERVISION_IDENTITIES_VALID = False


    if (
        media_ai_record_id
        not in ARTICLE_ID_SET
    ):
        SUPERVISION_IDENTITIES_VALID = False


    if (
        paragraph_id
        not in PARAGRAPH_ID_SET
    ):
        SUPERVISION_IDENTITIES_VALID = False


    required_provenance_fields = [
        "supervision_schema_version",
        "annotation_policy_version",
        "supervision_method",
        "supervisor_identifier",
        "context_scope",
    ]

    if not all(
        provenance.get(
            field
        )
        is not None
        for field
        in required_provenance_fields
    ):
        SUPERVISION_PROVENANCE_COMPLETE = False


SUPERVISION_IDS_UNIQUE = (
    len(
        CANONICAL_SUPERVISION_IDS
    )
    == len(
        set(
            CANONICAL_SUPERVISION_IDS
        )
    )
)


if (
    CANONICAL_SUPERVISION_AVAILABLE
    and not SUPERVISION_IDENTITIES_VALID
):
    raise ValueError(
        "Canonical supervision contains identifiers "
        "that do not align with the prepared corpus."
    )


if (
    CANONICAL_SUPERVISION_AVAILABLE
    and not SUPERVISION_IDS_UNIQUE
):
    raise ValueError(
        "Duplicate canonical supervision sentence IDs "
        "were detected."
    )


if (
    CANONICAL_SUPERVISION_AVAILABLE
    and not SUPERVISION_PROVENANCE_COMPLETE
):
    raise ValueError(
        "Canonical supervision provenance is incomplete."
    )


# -----------------------------------------------------------------------------
# Sentence-level supervision lookup
# -----------------------------------------------------------------------------

CANONICAL_SUPERVISION_BY_SENTENCE_ID = {
    record[
        "identity"
    ][
        "sentence_id"
    ]:
        record

    for record
    in CANONICAL_SUPERVISION_RECORDS

    if record.get(
        "identity",
        {}
    ).get(
        "sentence_id"
    )
}


# -----------------------------------------------------------------------------
# Article-level supervision coverage
# -----------------------------------------------------------------------------

ARTICLE_SUPERVISION_COVERAGE = []

for article in ARTICLE_RECORDS:

    article_id = article.get(
        "media_ai_record_id"
    )

    article_sentence_ids = [
        sentence.get(
            "sentence_id"
        )

        for paragraph
        in article.get(
            "paragraphs",
            []
        )

        for sentence
        in paragraph.get(
            "sentences",
            []
        )
    ]


    supervised_sentence_ids = [
        sentence_id
        for sentence_id
        in article_sentence_ids
        if sentence_id
        in CANONICAL_SUPERVISION_BY_SENTENCE_ID
    ]


    expected_count = len(
        article_sentence_ids
    )

    supervised_count = len(
        supervised_sentence_ids
    )


    if supervised_count == 0:

        coverage_status = (
            "not_supervised"
        )

    elif (
        supervised_count
        == expected_count
    ):

        coverage_status = (
            "fully_supervised"
        )

    else:

        coverage_status = (
            "partially_supervised"
        )


    ARTICLE_SUPERVISION_COVERAGE.append({
        "media_ai_record_id":
            article_id,

        "article_title":
            (
                article.get(
                    "title"
                )
                or article.get(
                    "article_title"
                )
            ),

        "expected_sentences":
            expected_count,

        "canonical_supervision_records":
            supervised_count,

        "coverage_fraction":
            (
                supervised_count
                / expected_count
                if expected_count
                else 0.0
            ),

        "coverage_status":
            coverage_status,
    })


# -----------------------------------------------------------------------------
# Supervision coverage summary
# -----------------------------------------------------------------------------

FULLY_SUPERVISED_ARTICLES = sum(
    1
    for record
    in ARTICLE_SUPERVISION_COVERAGE
    if record[
        "coverage_status"
    ]
    == "fully_supervised"
)


PARTIALLY_SUPERVISED_ARTICLES = sum(
    1
    for record
    in ARTICLE_SUPERVISION_COVERAGE
    if record[
        "coverage_status"
    ]
    == "partially_supervised"
)


UNSUPERVISED_ARTICLES = sum(
    1
    for record
    in ARTICLE_SUPERVISION_COVERAGE
    if record[
        "coverage_status"
    ]
    == "not_supervised"
)


CANONICAL_SUPERVISION_COMPLETE = (
    bool(
        ARTICLE_RECORDS
    )
    and FULLY_SUPERVISED_ARTICLES
    == len(
        ARTICLE_RECORDS
    )
)


# -----------------------------------------------------------------------------
# Notebook 04 research-input layer
# -----------------------------------------------------------------------------

NOTEBOOK_04_RESEARCH_INPUT = {
    "prepared_corpus": {
        "loaded":
            bool(
                PREPARED_RECORDS
            ),

        "artifact_type":
            (
                PREPARED_CORPUS.get(
                    "artifact_type"
                )
                if isinstance(
                    PREPARED_CORPUS,
                    dict,
                )
                else None
            ),

        "article_count":
            len(
                ARTICLE_RECORDS
            ),

        "paragraph_count":
            len(
                PARAGRAPH_RECORDS
            ),

        "sentence_count":
            len(
                SENTENCE_RECORDS
            ),

        "hierarchy_valid":
            PREPARED_HIERARCHY_VALID,

        "counts_match_handover":
            PREPARED_COUNTS_MATCH_HANDOVER,
    },

    "canonical_supervision": {
        "requested":
            LOAD_CANONICAL_SUPERVISION,

        "available":
            CANONICAL_SUPERVISION_AVAILABLE,

        "source":
            CANONICAL_SUPERVISION_SOURCE,

        "sentence_records":
            len(
                CANONICAL_SUPERVISION_RECORDS
            ),

        "fully_supervised_articles":
            FULLY_SUPERVISED_ARTICLES,

        "partially_supervised_articles":
            PARTIALLY_SUPERVISED_ARTICLES,

        "unsupervised_articles":
            UNSUPERVISED_ARTICLES,

        "complete":
            CANONICAL_SUPERVISION_COMPLETE,

        "identity_alignment_valid":
            SUPERVISION_IDENTITIES_VALID,

        "provenance_complete":
            SUPERVISION_PROVENANCE_COMPLETE,
    },
}


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Notebook 03 Handover and Research-Artefact Loading")
print("=" * 72)

print("Persistent storage")

print(
    f"Project folder found        : "
    f"{bool(PROJECT_FOLDER_ID)}"
)

print(
    f"Data folder found           : "
    f"{bool(DATA_FOLDER_ID)}"
)

print(
    f"Prepared folder found       : "
    f"{bool(PREPARED_DATA_FOLDER_ID)}"
)

print(
    f"Logs folder found           : "
    f"{bool(LOGS_FOLDER_ID)}"
)

print("-" * 72)

print("Prepared corpus")

print(
    f"Prepared corpus loaded      : "
    f"{bool(PREPARED_RECORDS)}"
)

print(
    f"Articles                    : "
    f"{len(ARTICLE_RECORDS)}"
)

print(
    f"Paragraphs                  : "
    f"{len(PARAGRAPH_RECORDS)}"
)

print(
    f"Sentences                   : "
    f"{len(SENTENCE_RECORDS)}"
)

print(
    f"Article IDs unique          : "
    f"{ARTICLE_IDS_UNIQUE}"
)

print(
    f"Paragraph IDs unique        : "
    f"{PARAGRAPH_IDS_UNIQUE}"
)

print(
    f"Sentence IDs unique         : "
    f"{SENTENCE_IDS_UNIQUE}"
)

print(
    f"Parent links valid          : "
    f"{PARAGRAPH_PARENT_LINKS_VALID and SENTENCE_PARENT_LINKS_VALID}"
)

print(
    f"Counts match handover       : "
    f"{PREPARED_COUNTS_MATCH_HANDOVER}"
)

print("-" * 72)

print("Canonical supervision")

print(
    f"Supervision requested       : "
    f"{LOAD_CANONICAL_SUPERVISION}"
)

print(
    f"Supervision available       : "
    f"{CANONICAL_SUPERVISION_AVAILABLE}"
)

print(
    f"Supervision source          : "
    f"{CANONICAL_SUPERVISION_SOURCE}"
)

print(
    f"Canonical sentence records  : "
    f"{len(CANONICAL_SUPERVISION_RECORDS)}"
)

print(
    f"Identity alignment valid    : "
    f"{SUPERVISION_IDENTITIES_VALID}"
)

print(
    f"Provenance complete         : "
    f"{SUPERVISION_PROVENANCE_COMPLETE}"
)

print("-" * 72)

print("Article supervision coverage")

print(
    f"Fully supervised articles   : "
    f"{FULLY_SUPERVISED_ARTICLES}"
)

print(
    f"Partially supervised        : "
    f"{PARTIALLY_SUPERVISED_ARTICLES}"
)

print(
    f"Not supervised              : "
    f"{UNSUPERVISED_ARTICLES}"
)

print(
    f"Supervision complete        : "
    f"{CANONICAL_SUPERVISION_COMPLETE}"
)

print("=" * 72)

print(
    "Notebook 04 research artefacts loaded and validated successfully."
)

print(
    "Prepared corpus records remain immutable research inputs."
)

if CANONICAL_SUPERVISION_AVAILABLE:

    print(
        "Available canonical supervision is aligned through stable "
        "research identifiers."
    )

else:

    print(
        "Canonical supervision is currently unavailable or not yet persisted."
    )

    print(
        "This is an allowed Notebook 04 state; supervision-dependent "
        "operations must remain conditional."
    )

print(
    "Missing supervision has not been converted to numeric zero."
)

print(
    "No raw provider supervision has entered Notebook 04."
)

print(
    "No model-specific tokenisation or embedding has been performed."
)

print(
    "No Media AI representation, recurrent state, aggregation or "
    "attribution has been constructed."
)
````

## Hierarchical Text and Identity Validation

This block establishes and validates the hierarchical textual structure used by the Media AI representation and model architecture.

The canonical prepared corpus already provides stable article, paragraph and sentence structures.

Notebook 04 extends this hierarchy with a token-level analytical layer while preserving the canonical research text and all existing structural identities.

The resulting hierarchy is:

**Token → Sentence → Paragraph → Article**

### Canonical Hierarchy

The prepared corpus remains authoritative for:

* article identity;
* paragraph identity and ordering;
* sentence identity and ordering;
* article-to-paragraph relationships;
* paragraph-to-sentence relationships;
* canonical textual content.

Notebook 04 must not reconstruct these identities from text when stable identifiers already exist.

Existing identifiers are inherited unchanged from the prepared corpus.

### Token-Level Analytical Layer

Token is introduced as the lowest representational level required by the Media AI architecture.

Token-level structures may later support:

* model input representation;
* contextual embeddings;
* token-level contribution estimates;
* hierarchical aggregation;
* attribution;
* colour inspection;
* counterfactual analysis.

Tokens are derived analytical units rather than replacements for canonical text.

The introduction of token-level representation must therefore preserve a mapping back to the sentence and original textual span from which each token was derived.

### Tokenisation Boundary

This block defines the structural requirements for token-level representation but does not yet select or execute a model-specific tokenizer.

A later model-facing block may generate model tokens according to the requirements of the selected representation model.

Such tokenisation must remain distinguishable from the canonical sentence text.

Model-specific tokens must not become permanent replacements for the research corpus.

### Stable Identity

Every analytical unit must be traceable through stable structural relationships.

Conceptually:

[
Token
\rightarrow
Sentence\ ID
\rightarrow
Paragraph\ ID
\rightarrow
Media\ AI\ Record\ ID
]

The hierarchy must support traversal in both directions.

This allows the architecture to determine:

* which tokens belong to a sentence;
* which sentences belong to a paragraph;
* which paragraphs belong to an article;
* which lower-level units contributed to a higher-level representation.

### Ordering

Textual order is part of the research structure and must be preserved explicitly.

The hierarchy therefore retains:

* token position within sentence;
* sentence position within paragraph;
* paragraph position within article.

Ordering must be deterministic and must not depend on incidental DataFrame ordering, dictionary ordering or model-batch position.

This requirement is particularly important for later transformative and recurrent state construction.

### Text Preservation

Canonical sentence and article text must remain unchanged by hierarchy construction.

Derived processing may normalise, encode or tokenise text for model use, but such transformations must be stored separately from canonical text.

The architecture must remain capable of returning from any derived representation to the original textual evidence.

### Hierarchical Integrity

The block validates the existing article, paragraph and sentence hierarchy before token-level modelling begins.

Validation includes:

* uniqueness of article identifiers;
* uniqueness of paragraph identifiers;
* uniqueness of sentence identifiers;
* valid paragraph-to-article relationships;
* valid sentence-to-paragraph relationships;
* valid sentence-to-article relationships;
* deterministic paragraph ordering;
* deterministic sentence ordering;
* non-empty canonical sentence text;
* agreement between hierarchical counts and the validated Notebook 04 research-input layer.

Structural conflicts must fail explicitly.

### Token Mapping Contract

Before model-specific tokenisation is permitted, Notebook 04 defines the minimum mapping information that a token-level representation must preserve.

Each derived token must be capable of retaining or resolving:

* token identity or position;
* token text or model-token representation;
* sentence ID;
* paragraph ID;
* Media AI record ID;
* token order within the sentence;
* original-text relationship or character-span information where supported;
* tokenizer or representation provenance.

The exact token structure may depend on the later selected model, but these traceability requirements remain invariant.

### Hierarchical Representation Boundary

This block validates textual structure and defines the identity contract required for later representation.

It does not yet determine how information is aggregated between levels.

In particular, it does not assume:

[
Article =
mean(Paragraphs)
]

or:

[
Paragraph =
mean(Sentences)
]

or:

[
Sentence =
mean(Tokens)
]

Aggregation is a separate architectural problem and is defined explicitly in a later block.

The existence of a textual hierarchy therefore does not imply a particular mathematical aggregation rule.

### Attribution Boundary

The hierarchy established here provides the structural foundation for later attribution.

A higher-level Media AI state must ultimately remain traceable through:

[
Article
\rightarrow
Paragraph
\rightarrow
Sentence
\rightarrow
Token
\rightarrow
Original\ Text
]

This block establishes the identities and mappings required for that path.

It does not yet calculate contribution scores or interpret attribution.

### Output

This block creates:

* validated article-level identities;
* validated paragraph-level identities;
* validated sentence-level identities;
* deterministic hierarchy mappings;
* deterministic textual ordering;
* the token-level identity and mapping contract;
* structural lookup tables required by later blocks;
* hierarchy-validation diagnostics.

The canonical prepared text remains unchanged.

No model-specific tokenisation or embedding is performed.

No factual, psychological or social representation is constructed.

No transformative or recurrent state is calculated.

No hierarchical aggregation or attribution score is calculated.

No model training or experimental evaluation is performed.


````python
# =============================================================================
# Media AI — Hierarchical Text and Identity Validation
# =============================================================================

from copy import deepcopy


# -----------------------------------------------------------------------------
# Block dependency checks
# -----------------------------------------------------------------------------

required_block_2_objects = [
    "NOTEBOOK_04_RESEARCH_INPUT",
    "ARTICLE_RECORDS",
    "PARAGRAPH_RECORDS",
    "SENTENCE_RECORDS",
    "ARTICLE_IDS",
    "PARAGRAPH_IDS",
    "SENTENCE_IDS",
    "PREPARED_HIERARCHY_VALID",
]

missing_block_2_objects = [
    object_name
    for object_name
    in required_block_2_objects
    if object_name not in globals()
]

if missing_block_2_objects:
    raise NameError(
        "Notebook 04 Block 2 is not initialised. "
        "Missing: "
        f"{missing_block_2_objects}"
    )


required_block_1_controls = [
    "TEXTUAL_HIERARCHY_LEVELS",
    "PRESERVE_CANONICAL_TEXT",
    "PRESERVE_STRUCTURAL_TRACEABILITY",
    "PRESERVE_TEXTUAL_ORDER",
    "REQUIRE_TOKEN_TO_SENTENCE_MAPPING",
    "REQUIRE_SENTENCE_TO_PARAGRAPH_MAPPING",
    "REQUIRE_PARAGRAPH_TO_ARTICLE_MAPPING",
    "REQUIRE_MODEL_TOKEN_TRACEABILITY",
]

missing_block_1_controls = [
    object_name
    for object_name
    in required_block_1_controls
    if object_name not in globals()
]

if missing_block_1_controls:
    raise NameError(
        "Required Notebook 04 configuration is unavailable. "
        "Rerun Block 1. Missing: "
        f"{missing_block_1_controls}"
    )


# -----------------------------------------------------------------------------
# Hierarchy-level validation
# -----------------------------------------------------------------------------

EXPECTED_TEXTUAL_HIERARCHY = (
    "token",
    "sentence",
    "paragraph",
    "article",
)

if (
    TEXTUAL_HIERARCHY_LEVELS
    != EXPECTED_TEXTUAL_HIERARCHY
):
    raise ValueError(
        "Configured textual hierarchy does not match the "
        "Notebook 04 methodological hierarchy."
    )


if not PREPARED_HIERARCHY_VALID:
    raise ValueError(
        "The prepared article → paragraph → sentence hierarchy "
        "did not pass Block 2 validation."
    )


# -----------------------------------------------------------------------------
# Stable identity sets
# -----------------------------------------------------------------------------

ARTICLE_ID_SET = set(
    ARTICLE_IDS
)

PARAGRAPH_ID_SET = set(
    PARAGRAPH_IDS
)

SENTENCE_ID_SET = set(
    SENTENCE_IDS
)


# -----------------------------------------------------------------------------
# Canonical lookup tables
# -----------------------------------------------------------------------------

ARTICLE_BY_ID = {
    article[
        "media_ai_record_id"
    ]:
        article

    for article
    in ARTICLE_RECORDS
}


PARAGRAPH_BY_ID = {
    paragraph[
        "paragraph_id"
    ]:
        paragraph

    for paragraph
    in PARAGRAPH_RECORDS
}


SENTENCE_BY_ID = {
    sentence[
        "sentence_id"
    ]:
        sentence

    for sentence
    in SENTENCE_RECORDS
}


# -----------------------------------------------------------------------------
# Canonical sentence-text resolution
# -----------------------------------------------------------------------------

def resolve_canonical_sentence_text(
    sentence_record,
):
    """
    Resolve canonical sentence text without modifying the source record.

    The prepared-corpus field is expected to be explicit. Alternative
    historical field names are recognised only for compatibility.
    """

    candidate_fields = [
        "text",
        "sentence_text",
        "current_sentence",
    ]

    for field_name in candidate_fields:

        value = sentence_record.get(
            field_name
        )

        if (
            isinstance(
                value,
                str,
            )
            and value.strip()
        ):
            return (
                value,
                field_name,
            )

    return (
        None,
        None,
    )


# -----------------------------------------------------------------------------
# Article → paragraph mapping
# -----------------------------------------------------------------------------

ARTICLE_TO_PARAGRAPH_IDS = {
    article_id: []
    for article_id
    in ARTICLE_ID_SET
}


for paragraph in PARAGRAPH_RECORDS:

    article_id = paragraph.get(
        "media_ai_record_id"
    )

    paragraph_id = paragraph.get(
        "paragraph_id"
    )

    if article_id not in ARTICLE_ID_SET:
        raise ValueError(
            "Paragraph references an unknown article: "
            f"{paragraph_id}"
        )

    ARTICLE_TO_PARAGRAPH_IDS[
        article_id
    ].append(
        paragraph_id
    )


# -----------------------------------------------------------------------------
# Paragraph → sentence mapping
# -----------------------------------------------------------------------------

PARAGRAPH_TO_SENTENCE_IDS = {
    paragraph_id: []
    for paragraph_id
    in PARAGRAPH_ID_SET
}


for sentence in SENTENCE_RECORDS:

    paragraph_id = sentence.get(
        "paragraph_id"
    )

    sentence_id = sentence.get(
        "sentence_id"
    )

    if paragraph_id not in PARAGRAPH_ID_SET:
        raise ValueError(
            "Sentence references an unknown paragraph: "
            f"{sentence_id}"
        )

    PARAGRAPH_TO_SENTENCE_IDS[
        paragraph_id
    ].append(
        sentence_id
    )


# -----------------------------------------------------------------------------
# Sentence → paragraph and article mappings
# -----------------------------------------------------------------------------

SENTENCE_TO_PARAGRAPH_ID = {}

SENTENCE_TO_ARTICLE_ID = {}


for sentence in SENTENCE_RECORDS:

    sentence_id = sentence.get(
        "sentence_id"
    )

    paragraph_id = sentence.get(
        "paragraph_id"
    )

    article_id = sentence.get(
        "media_ai_record_id"
    )

    SENTENCE_TO_PARAGRAPH_ID[
        sentence_id
    ] = paragraph_id

    SENTENCE_TO_ARTICLE_ID[
        sentence_id
    ] = article_id


# -----------------------------------------------------------------------------
# Paragraph → article mapping
# -----------------------------------------------------------------------------

PARAGRAPH_TO_ARTICLE_ID = {
    paragraph[
        "paragraph_id"
    ]:
        paragraph[
            "media_ai_record_id"
        ]

    for paragraph
    in PARAGRAPH_RECORDS
}


# -----------------------------------------------------------------------------
# Parent-child mapping validation
# -----------------------------------------------------------------------------

SENTENCE_TO_PARAGRAPH_MAPPING_VALID = all(
    (
        sentence_id
        in SENTENCE_TO_PARAGRAPH_ID
    )
    and (
        SENTENCE_TO_PARAGRAPH_ID[
            sentence_id
        ]
        in PARAGRAPH_ID_SET
    )

    for sentence_id
    in SENTENCE_ID_SET
)


PARAGRAPH_TO_ARTICLE_MAPPING_VALID = all(
    (
        paragraph_id
        in PARAGRAPH_TO_ARTICLE_ID
    )
    and (
        PARAGRAPH_TO_ARTICLE_ID[
            paragraph_id
        ]
        in ARTICLE_ID_SET
    )

    for paragraph_id
    in PARAGRAPH_ID_SET
)


SENTENCE_TO_ARTICLE_MAPPING_VALID = all(
    (
        sentence_id
        in SENTENCE_TO_ARTICLE_ID
    )
    and (
        SENTENCE_TO_ARTICLE_ID[
            sentence_id
        ]
        in ARTICLE_ID_SET
    )

    for sentence_id
    in SENTENCE_ID_SET
)


# -----------------------------------------------------------------------------
# Cross-level parent consistency
# -----------------------------------------------------------------------------

CROSS_LEVEL_PARENT_CONSISTENCY_VALID = True

CROSS_LEVEL_PARENT_ERRORS = []


for sentence_id in SENTENCE_ID_SET:

    paragraph_id = (
        SENTENCE_TO_PARAGRAPH_ID[
            sentence_id
        ]
    )

    direct_article_id = (
        SENTENCE_TO_ARTICLE_ID[
            sentence_id
        ]
    )

    inherited_article_id = (
        PARAGRAPH_TO_ARTICLE_ID.get(
            paragraph_id
        )
    )

    if (
        direct_article_id
        != inherited_article_id
    ):

        CROSS_LEVEL_PARENT_CONSISTENCY_VALID = False

        CROSS_LEVEL_PARENT_ERRORS.append({
            "sentence_id":
                sentence_id,

            "paragraph_id":
                paragraph_id,

            "direct_article_id":
                direct_article_id,

            "paragraph_article_id":
                inherited_article_id,
        })


if not CROSS_LEVEL_PARENT_CONSISTENCY_VALID:
    raise ValueError(
        "Sentence and paragraph article-parent relationships "
        "are inconsistent."
    )


# -----------------------------------------------------------------------------
# Deterministic paragraph ordering
# -----------------------------------------------------------------------------

ARTICLE_PARAGRAPH_ORDER = {}

PARAGRAPH_ORDER_VALID = True
PARAGRAPH_ORDER_ERRORS = []


for article_id in ARTICLE_ID_SET:

    article_paragraphs = [
        paragraph
        for paragraph
        in PARAGRAPH_RECORDS
        if paragraph.get(
            "media_ai_record_id"
        )
        == article_id
    ]


    missing_indices = [
        paragraph.get(
            "paragraph_id"
        )
        for paragraph
        in article_paragraphs
        if not isinstance(
            paragraph.get(
                "paragraph_index"
            ),
            int,
        )
    ]


    if missing_indices:

        PARAGRAPH_ORDER_VALID = False

        PARAGRAPH_ORDER_ERRORS.append({
            "media_ai_record_id":
                article_id,

            "error":
                "missing_or_invalid_paragraph_index",

            "paragraph_ids":
                missing_indices,
        })

        continue


    paragraph_indices = [
        paragraph[
            "paragraph_index"
        ]
        for paragraph
        in article_paragraphs
    ]


    if len(
        paragraph_indices
    ) != len(
        set(
            paragraph_indices
        )
    ):

        PARAGRAPH_ORDER_VALID = False

        PARAGRAPH_ORDER_ERRORS.append({
            "media_ai_record_id":
                article_id,

            "error":
                "duplicate_paragraph_index",
        })

        continue


    ordered_paragraphs = sorted(
        article_paragraphs,
        key=lambda paragraph:
            paragraph[
                "paragraph_index"
            ],
    )


    ARTICLE_PARAGRAPH_ORDER[
        article_id
    ] = [
        paragraph[
            "paragraph_id"
        ]
        for paragraph
        in ordered_paragraphs
    ]


if not PARAGRAPH_ORDER_VALID:
    raise ValueError(
        "Paragraph ordering failed deterministic "
        "hierarchy validation."
    )


# -----------------------------------------------------------------------------
# Deterministic sentence ordering
# -----------------------------------------------------------------------------

PARAGRAPH_SENTENCE_ORDER = {}

SENTENCE_ORDER_VALID = True
SENTENCE_ORDER_ERRORS = []


for paragraph_id in PARAGRAPH_ID_SET:

    paragraph_sentences = [
        sentence
        for sentence
        in SENTENCE_RECORDS
        if sentence.get(
            "paragraph_id"
        )
        == paragraph_id
    ]


    missing_indices = [
        sentence.get(
            "sentence_id"
        )
        for sentence
        in paragraph_sentences
        if not isinstance(
            sentence.get(
                "sentence_index"
            ),
            int,
        )
    ]


    if missing_indices:

        SENTENCE_ORDER_VALID = False

        SENTENCE_ORDER_ERRORS.append({
            "paragraph_id":
                paragraph_id,

            "error":
                "missing_or_invalid_sentence_index",

            "sentence_ids":
                missing_indices,
        })

        continue


    sentence_indices = [
        sentence[
            "sentence_index"
        ]
        for sentence
        in paragraph_sentences
    ]


    if len(
        sentence_indices
    ) != len(
        set(
            sentence_indices
        )
    ):

        SENTENCE_ORDER_VALID = False

        SENTENCE_ORDER_ERRORS.append({
            "paragraph_id":
                paragraph_id,

            "error":
                "duplicate_sentence_index",
        })

        continue


    ordered_sentences = sorted(
        paragraph_sentences,
        key=lambda sentence:
            sentence[
                "sentence_index"
            ],
    )


    PARAGRAPH_SENTENCE_ORDER[
        paragraph_id
    ] = [
        sentence[
            "sentence_id"
        ]
        for sentence
        in ordered_sentences
    ]


if not SENTENCE_ORDER_VALID:
    raise ValueError(
        "Sentence ordering failed deterministic "
        "hierarchy validation."
    )


# -----------------------------------------------------------------------------
# Complete article sentence order
# -----------------------------------------------------------------------------

ARTICLE_SENTENCE_ORDER = {}


for article_id in ARTICLE_ID_SET:

    ordered_sentence_ids = []


    for paragraph_id in (
        ARTICLE_PARAGRAPH_ORDER[
            article_id
        ]
    ):

        ordered_sentence_ids.extend(
            PARAGRAPH_SENTENCE_ORDER[
                paragraph_id
            ]
        )


    ARTICLE_SENTENCE_ORDER[
        article_id
    ] = ordered_sentence_ids


# -----------------------------------------------------------------------------
# Canonical sentence-text validation
# -----------------------------------------------------------------------------

CANONICAL_SENTENCE_TEXT = {}

CANONICAL_SENTENCE_TEXT_FIELD = {}

EMPTY_CANONICAL_SENTENCE_IDS = []


for sentence_id in SENTENCE_IDS:

    sentence_record = (
        SENTENCE_BY_ID[
            sentence_id
        ]
    )

    text, text_field = (
        resolve_canonical_sentence_text(
            sentence_record
        )
    )


    if text is None:

        EMPTY_CANONICAL_SENTENCE_IDS.append(
            sentence_id
        )

        continue


    CANONICAL_SENTENCE_TEXT[
        sentence_id
    ] = text

    CANONICAL_SENTENCE_TEXT_FIELD[
        sentence_id
    ] = text_field


CANONICAL_SENTENCE_TEXT_VALID = (
    len(
        EMPTY_CANONICAL_SENTENCE_IDS
    )
    == 0
)


if (
    PRESERVE_CANONICAL_TEXT
    and not CANONICAL_SENTENCE_TEXT_VALID
):
    raise ValueError(
        "One or more canonical sentence records do not "
        "contain usable text."
    )


# -----------------------------------------------------------------------------
# Hierarchy accounting
# -----------------------------------------------------------------------------

MAPPED_PARAGRAPH_COUNT = sum(
    len(
        paragraph_ids
    )
    for paragraph_ids
    in ARTICLE_TO_PARAGRAPH_IDS.values()
)


MAPPED_SENTENCE_COUNT = sum(
    len(
        sentence_ids
    )
    for sentence_ids
    in PARAGRAPH_TO_SENTENCE_IDS.values()
)


PARAGRAPH_ACCOUNTING_VALID = (
    MAPPED_PARAGRAPH_COUNT
    == len(
        PARAGRAPH_RECORDS
    )
)


SENTENCE_ACCOUNTING_VALID = (
    MAPPED_SENTENCE_COUNT
    == len(
        SENTENCE_RECORDS
    )
)


ARTICLE_SENTENCE_ACCOUNTING_VALID = (
    sum(
        len(
            sentence_ids
        )
        for sentence_ids
        in ARTICLE_SENTENCE_ORDER.values()
    )
    == len(
        SENTENCE_RECORDS
    )
)


# -----------------------------------------------------------------------------
# Block 2 research-input consistency
# -----------------------------------------------------------------------------

block_2_prepared_counts = (
    NOTEBOOK_04_RESEARCH_INPUT.get(
        "prepared_corpus",
        {}
    )
)


BLOCK_2_ARTICLE_COUNT_VALID = (
    len(
        ARTICLE_RECORDS
    )
    == block_2_prepared_counts.get(
        "article_count"
    )
)


BLOCK_2_PARAGRAPH_COUNT_VALID = (
    len(
        PARAGRAPH_RECORDS
    )
    == block_2_prepared_counts.get(
        "paragraph_count"
    )
)


BLOCK_2_SENTENCE_COUNT_VALID = (
    len(
        SENTENCE_RECORDS
    )
    == block_2_prepared_counts.get(
        "sentence_count"
    )
)


BLOCK_2_HIERARCHY_COUNTS_VALID = all([
    BLOCK_2_ARTICLE_COUNT_VALID,
    BLOCK_2_PARAGRAPH_COUNT_VALID,
    BLOCK_2_SENTENCE_COUNT_VALID,
])


if not BLOCK_2_HIERARCHY_COUNTS_VALID:
    raise ValueError(
        "Block 3 hierarchy counts do not reconcile "
        "with the validated Block 2 research-input layer."
    )


# -----------------------------------------------------------------------------
# Token-level identity and traceability contract
# -----------------------------------------------------------------------------

TOKEN_MAPPING_CONTRACT_VERSION = "1.0"


TOKEN_MAPPING_REQUIRED_FIELDS = (
    "token_id",
    "token_position",
    "token_representation",
    "sentence_id",
    "paragraph_id",
    "media_ai_record_id",
    "original_text_start",
    "original_text_end",
    "tokeniser_identifier",
    "tokeniser_version",
)


TOKEN_MAPPING_OPTIONAL_FIELDS = (
    "token_text",
    "model_token_id",
    "special_token",
    "attention_mask_value",
)


TOKEN_MAPPING_CONTRACT = {
    "contract_version":
        TOKEN_MAPPING_CONTRACT_VERSION,

    "hierarchy":
        (
            "token",
            "sentence",
            "paragraph",
            "article",
        ),

    "required_fields":
        TOKEN_MAPPING_REQUIRED_FIELDS,

    "optional_fields":
        TOKEN_MAPPING_OPTIONAL_FIELDS,

    "requirements": {
        "token_to_sentence":
            REQUIRE_TOKEN_TO_SENTENCE_MAPPING,

        "sentence_to_paragraph":
            REQUIRE_SENTENCE_TO_PARAGRAPH_MAPPING,

        "paragraph_to_article":
            REQUIRE_PARAGRAPH_TO_ARTICLE_MAPPING,

        "model_token_traceability":
            REQUIRE_MODEL_TOKEN_TRACEABILITY,

        "canonical_text_preserved":
            PRESERVE_CANONICAL_TEXT,

        "textual_order_preserved":
            PRESERVE_TEXTUAL_ORDER,
    },

    "tokenisation_status":
        "not_performed",
}


# -----------------------------------------------------------------------------
# Tokenisation state
# -----------------------------------------------------------------------------

MODEL_TOKEN_RECORDS = []

MODEL_TOKENISATION_PERFORMED = False


# -----------------------------------------------------------------------------
# Consolidated hierarchical identity registry
# -----------------------------------------------------------------------------

HIERARCHICAL_IDENTITY_REGISTRY = {
    "article_by_id":
        ARTICLE_BY_ID,

    "paragraph_by_id":
        PARAGRAPH_BY_ID,

    "sentence_by_id":
        SENTENCE_BY_ID,

    "article_to_paragraph_ids":
        ARTICLE_TO_PARAGRAPH_IDS,

    "paragraph_to_sentence_ids":
        PARAGRAPH_TO_SENTENCE_IDS,

    "sentence_to_paragraph_id":
        SENTENCE_TO_PARAGRAPH_ID,

    "paragraph_to_article_id":
        PARAGRAPH_TO_ARTICLE_ID,

    "sentence_to_article_id":
        SENTENCE_TO_ARTICLE_ID,

    "article_paragraph_order":
        ARTICLE_PARAGRAPH_ORDER,

    "paragraph_sentence_order":
        PARAGRAPH_SENTENCE_ORDER,

    "article_sentence_order":
        ARTICLE_SENTENCE_ORDER,
}


# -----------------------------------------------------------------------------
# Hierarchy-validation summary
# -----------------------------------------------------------------------------

HIERARCHY_VALIDATION_SUMMARY = {
    "article_ids_unique":
        len(
            ARTICLE_IDS
        )
        == len(
            ARTICLE_ID_SET
        ),

    "paragraph_ids_unique":
        len(
            PARAGRAPH_IDS
        )
        == len(
            PARAGRAPH_ID_SET
        ),

    "sentence_ids_unique":
        len(
            SENTENCE_IDS
        )
        == len(
            SENTENCE_ID_SET
        ),

    "sentence_to_paragraph_valid":
        SENTENCE_TO_PARAGRAPH_MAPPING_VALID,

    "paragraph_to_article_valid":
        PARAGRAPH_TO_ARTICLE_MAPPING_VALID,

    "sentence_to_article_valid":
        SENTENCE_TO_ARTICLE_MAPPING_VALID,

    "cross_level_parent_consistency":
        CROSS_LEVEL_PARENT_CONSISTENCY_VALID,

    "paragraph_order_valid":
        PARAGRAPH_ORDER_VALID,

    "sentence_order_valid":
        SENTENCE_ORDER_VALID,

    "canonical_sentence_text_valid":
        CANONICAL_SENTENCE_TEXT_VALID,

    "paragraph_accounting_valid":
        PARAGRAPH_ACCOUNTING_VALID,

    "sentence_accounting_valid":
        SENTENCE_ACCOUNTING_VALID,

    "article_sentence_accounting_valid":
        ARTICLE_SENTENCE_ACCOUNTING_VALID,

    "block_2_counts_valid":
        BLOCK_2_HIERARCHY_COUNTS_VALID,

    "tokenisation_performed":
        MODEL_TOKENISATION_PERFORMED,
}


HIERARCHICAL_TEXT_AND_IDENTITY_VALID = all([
    HIERARCHY_VALIDATION_SUMMARY[
        "article_ids_unique"
    ],

    HIERARCHY_VALIDATION_SUMMARY[
        "paragraph_ids_unique"
    ],

    HIERARCHY_VALIDATION_SUMMARY[
        "sentence_ids_unique"
    ],

    HIERARCHY_VALIDATION_SUMMARY[
        "sentence_to_paragraph_valid"
    ],

    HIERARCHY_VALIDATION_SUMMARY[
        "paragraph_to_article_valid"
    ],

    HIERARCHY_VALIDATION_SUMMARY[
        "sentence_to_article_valid"
    ],

    HIERARCHY_VALIDATION_SUMMARY[
        "cross_level_parent_consistency"
    ],

    HIERARCHY_VALIDATION_SUMMARY[
        "paragraph_order_valid"
    ],

    HIERARCHY_VALIDATION_SUMMARY[
        "sentence_order_valid"
    ],

    HIERARCHY_VALIDATION_SUMMARY[
        "canonical_sentence_text_valid"
    ],

    HIERARCHY_VALIDATION_SUMMARY[
        "paragraph_accounting_valid"
    ],

    HIERARCHY_VALIDATION_SUMMARY[
        "sentence_accounting_valid"
    ],

    HIERARCHY_VALIDATION_SUMMARY[
        "article_sentence_accounting_valid"
    ],

    HIERARCHY_VALIDATION_SUMMARY[
        "block_2_counts_valid"
    ],
])


if not HIERARCHICAL_TEXT_AND_IDENTITY_VALID:
    raise ValueError(
        "Notebook 04 hierarchical text and identity "
        "validation failed."
    )


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Hierarchical Text and Identity Validation")
print("=" * 72)

print("Canonical hierarchy")

print(
    f"Articles                    : "
    f"{len(ARTICLE_RECORDS)}"
)

print(
    f"Paragraphs                  : "
    f"{len(PARAGRAPH_RECORDS)}"
)

print(
    f"Sentences                   : "
    f"{len(SENTENCE_RECORDS)}"
)

print(
    f"Article IDs unique          : "
    f"{HIERARCHY_VALIDATION_SUMMARY['article_ids_unique']}"
)

print(
    f"Paragraph IDs unique        : "
    f"{HIERARCHY_VALIDATION_SUMMARY['paragraph_ids_unique']}"
)

print(
    f"Sentence IDs unique         : "
    f"{HIERARCHY_VALIDATION_SUMMARY['sentence_ids_unique']}"
)

print("-" * 72)

print("Parent-child relationships")

print(
    f"Sentence → paragraph        : "
    f"{SENTENCE_TO_PARAGRAPH_MAPPING_VALID}"
)

print(
    f"Paragraph → article         : "
    f"{PARAGRAPH_TO_ARTICLE_MAPPING_VALID}"
)

print(
    f"Sentence → article          : "
    f"{SENTENCE_TO_ARTICLE_MAPPING_VALID}"
)

print(
    f"Cross-level consistency     : "
    f"{CROSS_LEVEL_PARENT_CONSISTENCY_VALID}"
)

print("-" * 72)

print("Ordering and text integrity")

print(
    f"Paragraph ordering          : "
    f"{PARAGRAPH_ORDER_VALID}"
)

print(
    f"Sentence ordering           : "
    f"{SENTENCE_ORDER_VALID}"
)

print(
    f"Canonical sentence text     : "
    f"{CANONICAL_SENTENCE_TEXT_VALID}"
)

print(
    f"Paragraph accounting        : "
    f"{PARAGRAPH_ACCOUNTING_VALID}"
)

print(
    f"Sentence accounting         : "
    f"{SENTENCE_ACCOUNTING_VALID}"
)

print(
    f"Article sentence accounting : "
    f"{ARTICLE_SENTENCE_ACCOUNTING_VALID}"
)

print(
    f"Block 2 counts reconcile    : "
    f"{BLOCK_2_HIERARCHY_COUNTS_VALID}"
)

print("-" * 72)

print("Token-level contract")

print(
    f"Contract version            : "
    f"{TOKEN_MAPPING_CONTRACT_VERSION}"
)

print(
    f"Required token fields       : "
    f"{len(TOKEN_MAPPING_REQUIRED_FIELDS)}"
)

print(
    f"Token → sentence required   : "
    f"{REQUIRE_TOKEN_TO_SENTENCE_MAPPING}"
)

print(
    f"Model token traceability    : "
    f"{REQUIRE_MODEL_TOKEN_TRACEABILITY}"
)

print(
    f"Tokenisation performed      : "
    f"{MODEL_TOKENISATION_PERFORMED}"
)

print("-" * 72)

print(
    f"Hierarchy validation passed : "
    f"{HIERARCHICAL_TEXT_AND_IDENTITY_VALID}"
)

print("=" * 72)

print(
    "Canonical article, paragraph and sentence identities "
    "validated successfully."
)

print(
    "Deterministic textual ordering and parent-child mappings "
    "have been established."
)

print(
    "The token-level identity and traceability contract has been defined."
)

print(
    "No model-specific tokenisation has been performed."
)

print(
    "Canonical prepared text remains unchanged."
)

print(
    "No hierarchical aggregation or attribution score has been calculated."
)

print(
    "No Media AI representation or state has been constructed."
)
````

## Media AI Representation Specification

This block defines the formal representation contract used by the Media AI architecture.

It specifies how textual units may be represented in factual, psychological and social spaces without yet constructing model representations.

The representation contract is defined before model-specific encoding, recurrent processing, hierarchical aggregation or experimental evaluation.

### Core Media AI Representation

For any traceable textual unit (u), the Media AI representation is defined as:

[
T_u^{\mathrm{MediaAI}} = {F_u, P_u, S_u}
]

where:

* (T_u^{\mathrm{MediaAI}}) = complete Media AI representation of textual unit (u);
* (u) = one traceable textual unit;
* (F_u) = factual representation of (u);
* (P_u) = psychological representation of (u);
* (S_u) = social and relational representation of (u).

The textual unit (u) may belong to any of the four hierarchical levels:

[
u \in {\text{token}, \text{sentence}, \text{paragraph}, \text{article}}
]

Dictionary:

* `token` = model-facing or analytical token linked to canonical text;
* `sentence` = canonical sentence-level textual unit;
* `paragraph` = ordered collection of canonical sentences;
* `article` = ordered collection of paragraphs representing the complete article.

The three representation spaces remain conceptually distinct even when a later model combines them computationally.

### Representation Principles

Every Media AI representation must:

* preserve the canonical textual hierarchy;
* remain traceable to the language from which it was derived;
* distinguish observed, supervised, derived and unavailable information;
* preserve missingness explicitly;
* preserve representation provenance;
* avoid treating numerical zero as missing information;
* avoid treating confidence as part of the represented construct itself;
* avoid silently mixing representations produced by different methods or models;
* support later hierarchical aggregation;
* support backward attribution.

A Media AI representation is therefore not merely a vector of numbers.

Conceptually:

[
R_u = {V_u, M_u, C_u, Q_u}
]

where:

* (R_u) = complete research representation object for textual unit (u);
* (V_u) = represented values;
* (M_u) = validity and availability information;
* (C_u) = confidence metadata;
* (Q_u) = provenance and construction metadata.

This distinction allows two numerically similar representations to remain different research objects when they were produced by different methods.

---

### Factual Representation

The factual space represents externally assessable informational structure in the text.

It does not represent whether a statement is actually true.

The factual representation is defined as:

[
F_u = {w_u^F, C_u^F}
]

where:

* (F_u) = factual representation of textual unit (u);
* (w_u^F) = factological weight;
* (C_u^F) = vector of factological components.

The factological-component vector is:

[
\begin{aligned}
C_u^F = [&c_{\mathrm{claim}}, c_{\mathrm{event}}, c_{\mathrm{observation}}, c_{\mathrm{quant}}, c_{\mathrm{attrib}},\
&c_{\mathrm{evidence}}, c_{\mathrm{certainty}}, c_{\mathrm{uncertainty}}, c_{\mathrm{speculation}}]
\end{aligned}
]

where:

* (c_{\mathrm{claim}}) = explicit claim;
* (c_{\mathrm{event}}) = reported event or state;
* (c_{\mathrm{observation}}) = reported observation, measurement or perception;
* (c_{\mathrm{quant}}) = quantitative information;
* (c_{\mathrm{attrib}}) = attribution;
* (c_{\mathrm{evidence}}) = evidence reference;
* (c_{\mathrm{certainty}}) = certainty statement;
* (c_{\mathrm{uncertainty}}) = uncertainty statement;
* (c_{\mathrm{speculation}}) = speculation or prediction.

The scalar factological weight satisfies:

[
w_u^F \in [0,1]
]

where (0) represents no meaningful factological contribution under the configured definition and (1) represents maximum factological contribution under the configured scale.

This value is not a probability that the text is true.

Likewise, the nine factual components describe the structure and presentation of factual information. They do not independently establish factual truth or factual accuracy.

The initial factual representation therefore contains one factological-weight dimension and nine structural dimensions.

Where encoded numerically:

[
F_u^{\mathrm{numeric}} \in \mathbb{R}^{10}
]

provided all ten dimensions are available.

Missingness must be represented separately rather than replaced with zero.

---

### Psychological Representation

The psychological space represents emotional state, emotional change and emotional relevance.

Let the configured emotional dimension set be:

[
\begin{aligned}
D_P = {&\mathrm{fear}, \mathrm{anger}, \mathrm{sadness}, \mathrm{disgust}, \mathrm{tension}, \mathrm{uncertainty},\
&\mathrm{hope}, \mathrm{relief}, \mathrm{trust}, \mathrm{compassion}, \mathrm{solidarity}}
\end{aligned}
]

There are eleven dimensions:

[
|D_P| = 11
]

The psychological representation is defined as:

[
P_u = {E_u^{\mathrm{reported}}, \Delta E_u, E_u^{\mathrm{immediate}}, w_u^P}
]

where:

* (P_u) = psychological representation of textual unit (u);
* (E_u^{\mathrm{reported}}) = reported-emotion vector;
* (\Delta E_u) = transformative-emotion vector;
* (E_u^{\mathrm{immediate}}) = immediate emotional-state vector;
* (w_u^P) = emotional weight.

#### Reported Emotion

[
E_u^{\mathrm{reported}}
=======================

[e_1^{\mathrm{reported}}, e_2^{\mathrm{reported}}, \ldots, e_{11}^{\mathrm{reported}}]
]

with:

[
e_i^{\mathrm{reported}} \in [0,1]
]

where:

* (e_i^{\mathrm{reported}}) = reported amount of emotional dimension (i);
* (i) indexes the eleven configured emotional dimensions.

Reported emotion describes emotion represented as belonging to people, groups, organisations or other subjects in the text.

#### Transformative Emotion

[
\Delta E_u = [\Delta e_1, \Delta e_2, \ldots, \Delta e_{11}]
]

with:

[
\Delta e_i \in [-1,1]
]

where:

* (\Delta e_i < 0) = the textual unit reduces emotional dimension (i);
* (\Delta e_i = 0) = no meaningful change in dimension (i);
* (\Delta e_i > 0) = the textual unit increases emotional dimension (i).

Transformative emotion is a change variable.

Therefore:

[
\Delta E_u \neq E_u^{\mathrm{immediate}}
]

#### Immediate Emotional State

[
E_u^{\mathrm{immediate}}
========================

[e_1^{\mathrm{immediate}}, e_2^{\mathrm{immediate}}, \ldots, e_{11}^{\mathrm{immediate}}]
]

with:

[
e_i^{\mathrm{immediate}} \in [0,1]
]

where (e_i^{\mathrm{immediate}}) represents the amount of emotional dimension (i) present after interpreting the current textual unit in its permitted context.

#### Emotional Weight

[
w_u^P \in [0,1]
]

where (w_u^P) represents the strength or importance of emotionally consequential information associated with textual unit (u).

The psychological constructs remain distinct:

[
E_u^{\mathrm{reported}} \neq \Delta E_u \neq E_u^{\mathrm{immediate}}
]

even though all three refer to the same eleven emotional dimensions.

If represented numerically without missing values:

[
P_u^{\mathrm{numeric}} \in \mathbb{R}^{34}
]

because:

[
11 + 11 + 11 + 1 = 34
]

This dimensional count describes the explicit supervised psychological space only.

It does not prevent later addition of learned latent psychological features.

---

### Social Representation

The social space represents actors, groups, memberships and relationships constructed or expressed through the text.

Unlike the factual and psychological spaces, it is not initially forced into one fixed flat vector.

The social representation is defined as:

[
S_u = {A_u, G_u, M_u^S, R_u^S}
]

where:

* (S_u) = social representation of textual unit (u);
* (A_u) = actor or entity set;
* (G_u) = group set;
* (M_u^S) = membership structure connecting actors and groups;
* (R_u^S) = relational structure connecting represented social objects.

#### Actor Set

[
A_u = {a_1, a_2, \ldots, a_n}
]

where:

* (a_i) = one actor, organisation, person, institution or other social entity;
* (n) = number of actors associated with textual unit (u).

The size of (A_u) may vary between textual units.

#### Group Structure

[
G_u = {g_1, g_2, \ldots, g_k}
]

where:

* (g_j) = one identified or constructed social group;
* (k) = number of groups represented at textual unit (u).

The size and identity of the represented group set may vary between textual units.

Concrete group identities are not required to be predefined by this representation contract.

#### Membership Structure

An actor may belong to more than one group.

Membership can therefore be represented as:

[
M_u^S \in [0,1]^{n \times k}
]

where (M_{ij}^S) represents the strength or degree of association between actor (a_i) and group (g_j).

This avoids requiring mutually exclusive group membership and permits overlapping social identities.

Membership is treated as an explicit social component rather than being collapsed into actor or group identity.

#### Relational Structure

The relationship structure may be represented as:

[
R_u^S = {r_{ij}}
]

where (r_{ij}) represents a relation between represented social objects (i) and (j).

A later implementation may represent this structure using:

* relational tensors;
* adjacency matrices;
* graph structures;
* learned relational embeddings;
* another explicitly documented representation.

Block 4 defines the contract but does not choose the final social implementation.

---

### Confluent Representation Boundary

Confluence operates across existing representations.

The initial spaces remain:

[
F_u,\quad P_u,\quad S_u
]

A later confluent representation may be expressed as:

[
K_u = \Phi(F_u, P_u, S_u)
]

where:

* (K_u) = confluent representation;
* (\Phi) = explicitly defined confluent transformation;
* (F_u) = factual input representation;
* (P_u) = psychological input representation;
* (S_u) = social input representation.

The transformation (\Phi) must not make it impossible to recover which representational space contributed to the resulting state.

Confluence therefore combines information without erasing provenance.

---

### Value Semantics

Every scalar representation value must carry an explicit state.

For scalar value (x):

[
X = {x, s_x}
]

where:

* (x) = numerical value or null;
* (s_x) = semantic value state.

The configured value states are:

[
\begin{aligned}
s_x \in {&\text{observed}, \text{insufficient evidence}, \text{not applicable},\
&\text{unavailable}, \text{derived}}
\end{aligned}
]

where:

* `observed` = a substantive numerical value is available;
* `insufficient evidence` = the available text does not support a substantive value;
* `not applicable` = the construct does not apply to this textual unit;
* `unavailable` = the representation has not been produced or the required supervision is absent;
* `derived` = the value has been produced from another valid representation through an explicitly defined transformation.

When:

[
s_x = \text{observed}
]

then (x) must contain a valid numerical value.

When the state does not permit a substantive numerical value, the value must remain unavailable rather than being silently converted to zero.

A numerical zero means actual zero:

[
x = 0 \neq \text{missing}
]

---

### Missingness and Validity Masks

Numerical tensors require an explicit availability mask.

For any numerical representation:

[
X_u = (V_u, M_u)
]

where:

* (V_u) = numerical representation values;
* (M_u) = corresponding validity mask.

For dimension (i):

[
M_{u,i} =
\begin{cases}
1, & \text{if } V_{u,i} \text{ is valid and available},\
0, & \text{otherwise}.
\end{cases}
]

Thus:

* (M_{u,i}=1) = the value is available;
* (M_{u,i}=0) = the value must not be interpreted as substantive data.

The numerical implementation may use `NaN` or another technical placeholder where needed, provided the explicit semantic state is retained.

---

### Confidence

Confidence is represented separately:

[
C_u \in [0,1]
]

where (C_u) is the confidence associated with a supervision or model-generated representation.

Confidence metadata is not part of the represented state itself:

[
C_u \notin T_u^{\mathrm{MediaAI}}
]

unless a later experimental method explicitly introduces it as an input.

A representation therefore conceptually remains:

[
R_u = {T_u^{\mathrm{MediaAI}}, M_u, C_u, Q_u}
]

where:

* (T_u^{\mathrm{MediaAI}}) = substantive Media AI representation;
* (M_u) = validity and missingness structure;
* (C_u) = confidence metadata;
* (Q_u) = provenance metadata.

---

### Representation Source

Every representation must explicitly identify its source.

Let (Z_u) denote representation source.

Then:

[
\begin{aligned}
Z_u \in {&\text{canonical supervision}, \text{human reviewed}, \text{model predicted}, \text{deterministic},\
&\text{transformative}, \text{recurrent}, \text{aggregated}, \text{counterfactual}}
\end{aligned}
]

Dictionary:

* `canonical supervision` = produced from validated canonical machine supervision;
* `human reviewed` = reviewed or established through human annotation;
* `model predicted` = generated by a trained Media AI model;
* `deterministic` = produced by a deterministic transformation;
* `transformative` = produced through an explicitly defined transformative mechanism;
* `recurrent` = produced through sequential state processing;
* `aggregated` = produced by hierarchical aggregation;
* `counterfactual` = produced by a later experimental transformation.

Two representations with identical dimensions but different sources remain different research objects:

[
T_u^{\mathrm{supervised}} \neq T_u^{\mathrm{predicted}}
]

as research objects, even when:

[
T_u^{\mathrm{supervised}} = T_u^{\mathrm{predicted}}
]

numerically.

---

### Representation Level

Every representation also carries a textual level:

[
L_u \in {\text{token}, \text{sentence}, \text{paragraph}, \text{article}}
]

where (L_u) is the hierarchy level of representation (u).

Representation identity therefore includes both source and level:

[
I_u = {ID_u, L_u, Z_u}
]

where:

* (ID_u) = stable textual identity;
* (L_u) = textual hierarchy level;
* (Z_u) = representation source.

---

### Hierarchical Representation

Representations may exist at all four levels:

[
T^{\mathrm{token}},\quad T^{\mathrm{sentence}},\quad
T^{\mathrm{paragraph}},\quad T^{\mathrm{article}}
]

The hierarchical structure is:

[
T^{\mathrm{token}} \rightarrow T^{\mathrm{sentence}}
\rightarrow T^{\mathrm{paragraph}} \rightarrow T^{\mathrm{article}}
]

The arrows represent hierarchical construction, not arithmetic averaging.

Therefore, the architecture does not assume:

[
T^{\mathrm{sentence}} = \operatorname{mean}(T^{\mathrm{token}})
]

nor:

[
T^{\mathrm{paragraph}} = \operatorname{mean}(T^{\mathrm{sentence}})
]

nor:

[
T^{\mathrm{article}} = \operatorname{mean}(T^{\mathrm{paragraph}})
]

A later aggregation mechanism is instead written generically as:

[
T^{\mathrm{higher}}
===================

A(T_1^{\mathrm{lower}}, T_2^{\mathrm{lower}}, \ldots, T_n^{\mathrm{lower}})
]

where:

* (A) = explicitly defined aggregation function;
* (T_i^{\mathrm{lower}}) = contributing lower-level representation;
* (T^{\mathrm{higher}}) = resulting higher-level representation;
* (n) = number of contributing lower-level units.

---

### Transformative Representation

For sequential textual position (t), transformative representation describes change produced by the current textual unit.

Let (T_{t-1}) be the previous state and (X_t) the current textual representation.

The transformative mechanism is:

[
\Delta T_t = \Phi_T(T_{t-1}, X_t)
]

where:

* (T_{t-1}) = state before the current textual unit;
* (X_t) = representation of the current textual unit;
* (\Phi_T) = transformative function;
* (\Delta T_t) = change introduced by the current textual unit.

The resulting state is:

[
T_t = \Psi_T(T_{t-1}, \Delta T_t)
]

where:

* (\Psi_T) = state-update function;
* (T_t) = state after processing textual step (t).

This preserves:

[
T_{t-1} \neq \Delta T_t \neq T_t
]

as distinct objects.

---

### Recurrent Representation

The recurrent mechanism preserves state through textual progression.

Let (H_t) denote recurrent state.

Then:

[
H_t = \Phi_R(H_{t-1}, T_t, \Delta T_t)
]

where:

* (H_{t-1}) = recurrent state before textual step (t);
* (T_t) = representation associated with the current step;
* (\Delta T_t) = transformative change at the current step;
* (\Phi_R) = recurrent update function;
* (H_t) = recurrent state after processing step (t).

The permitted information set is:

[
\mathcal{I}_t = {X_1, X_2, \ldots, X_t}
]

and excludes:

[
X_{t+1}, X_{t+2}, \ldots
]

when preceding-context processing is required.

---

### Social Polarisation Distance Boundary

The social space should eventually support relational distance between groups.

For two groups (g_i) and (g_j):

[
\mathrm{SPD}_t(g_i,g_j)
=======================

D(S_{t,g_i}, S_{t,g_j})
]

where:

* (\mathrm{SPD}_t(g_i,g_j)) = Social Polarisation Distance at textual step (t);
* (S_{t,g_i}) = social representation of group (g_i);
* (S_{t,g_j}) = social representation of group (g_j);
* (D) = explicitly defined relational distance function.

Change in distance is:

[
\Delta \mathrm{SPD}
===================

\mathrm{SPD}*{t_2} - \mathrm{SPD}*{t_1}
]

where:

* (\Delta \mathrm{SPD} > 0) = increasing represented distance;
* (\Delta \mathrm{SPD} < 0) = decreasing represented distance;
* (\Delta \mathrm{SPD} = 0) = no meaningful change under the configured measure.

The sign describes direction of change, not moral value.

Block 4 defines this representational requirement but does not calculate Social Polarisation Distance.

---

### Shape Contract

The architecture does not require all three representational spaces to have the same mathematical shape.

The factual representation is initially:

[
F_u^{\mathrm{numeric}} \in \mathbb{R}^{d_F},
\qquad d_F = 10
]

for one factological weight plus nine factological components.

The explicit psychological representation is initially:

[
P_u^{\mathrm{numeric}} \in \mathbb{R}^{d_P},
\qquad d_P = 34
]

for eleven reported-emotion dimensions, eleven transformative-emotion dimensions, eleven immediate-state dimensions and one emotional-weight dimension.

The fixed-dimensional factual and psychological spaces therefore contain:

[
d_F + d_P = 10 + 34 = 44
]

explicit dimensions before the variable social structure is considered.

The social representation instead has variable structure:

[
S_u = {A_u, G_u, M_u^S, R_u^S}
]

and therefore does not require a fixed (d_S) at this stage.

The complete Media AI representation is therefore not prematurely defined as:

[
T_u^{\mathrm{MediaAI}}
\in
\mathbb{R}^{d_F+d_P+d_S}
]

because (d_S) may be dynamic and relational.

---

### Provenance Contract

Every representation must carry provenance:

[
\begin{aligned}
Q_u = {&q_{\mathrm{record}}, q_{\mathrm{paragraph}}, q_{\mathrm{sentence}},
q_{\mathrm{token}}, q_{\mathrm{level}}, q_{\mathrm{space}},\
&q_{\mathrm{source}}, q_{\mathrm{schema}}, q_{\mathrm{method}},
q_{\mathrm{model}}, q_{\mathrm{supervisor}}, q_{\mathrm{time}}}
\end{aligned}
]

Dictionary:

* (q_{\mathrm{record}}) = Media AI record ID;
* (q_{\mathrm{paragraph}}) = paragraph ID where applicable;
* (q_{\mathrm{sentence}}) = sentence ID where applicable;
* (q_{\mathrm{token}}) = token ID where applicable;
* (q_{\mathrm{level}}) = representation level;
* (q_{\mathrm{space}}) = factual, psychological or social space;
* (q_{\mathrm{source}}) = representation source;
* (q_{\mathrm{schema}}) = representation-schema version;
* (q_{\mathrm{method}}) = construction method;
* (q_{\mathrm{model}}) = model or tokenizer identity where applicable;
* (q_{\mathrm{supervisor}}) = supervision source where applicable;
* (q_{\mathrm{time}}) = construction timestamp when persisted.

---

### Backward Attribution Requirement

Higher-level model states must remain traceable to textual evidence.

The attribution path is:

[
\begin{aligned}
T^{\mathrm{article}} \rightarrow T^{\mathrm{paragraph}}
\rightarrow T^{\mathrm{sentence}}
\rightarrow T^{\mathrm{token}}
\rightarrow \mathrm{Text}^{\mathrm{canonical}}
\end{aligned}
]

A generic attribution structure can be written as:

[
A_{h \leftarrow l}
]

where:

* (h) = higher-level representation;
* (l) = contributing lower-level unit;
* (A_{h \leftarrow l}) = contribution or attribution relationship between them.

For example:

[
A_{\mathrm{article} \leftarrow \mathrm{paragraph}}
]

represents paragraph contribution to an article-level state.

Similarly:

[
A_{\mathrm{paragraph} \leftarrow \mathrm{sentence}}
]

and:

[
A_{\mathrm{sentence} \leftarrow \mathrm{token}}
]

allow progressively finer inspection.

The architecture must preserve the information necessary to calculate such relationships later.

---

### Supervision Boundary

Canonical supervision may initialise selected sentence-level dimensions.

For sentence (s):

[
T_s^{\mathrm{supervised}}
]

may contain available factual and psychological supervision.

Where supervision is absent:

[
M_s = 0
]

for the corresponding unavailable dimensions.

This must not be replaced with:

[
T_s = 0
]

because:

[
\text{missing} \neq 0
]

Raw provider responses and rejected supervision records are not permitted to initialise Media AI representations.

---

### Representation Versioning

The representation contract is versioned by:

[
V_{\mathrm{representation}}
]

where (V_{\mathrm{representation}}) is the configured representation-schema version.

A schema-version change is required when research meaning changes materially, including changes to:

* dimension definitions;
* value ranges;
* missingness semantics;
* representation-space boundaries;
* source semantics;
* shape conventions;
* provenance requirements.

Model implementation changes that do not alter the representation contract may use separate model or architecture versions.

---

### Output

This block creates the formal Media AI representation specification.

Its outputs define:

* the complete Media AI representation object;
* factual-space formulae and dimension dictionary;
* psychological-space formulae and dimension dictionary;
* social actor, group, membership and relational structures;
* confluent-representation boundaries;
* numerical value semantics;
* missingness masks;
* confidence semantics;
* representation-source semantics;
* representation-level semantics;
* hierarchical representation rules;
* transformative-state notation;
* recurrent-state notation;
* Social Polarisation Distance requirements;
* shape contracts;
* provenance dictionaries;
* backward-attribution requirements;
* supervision boundaries;
* representation-versioning rules.

No representation tensor is constructed.

No model-specific tokenisation or embedding is performed.

No social graph or relational embedding is constructed.

No transformative or recurrent state is calculated.

No confluent representation is calculated.

No hierarchical aggregation or attribution score is calculated.

No model training or experimental evaluation is performed.


````python
# =============================================================================
# Media AI — Representation Specification
# =============================================================================

from copy import deepcopy


# -----------------------------------------------------------------------------
# Block dependency checks
# -----------------------------------------------------------------------------

required_block_1_objects = [
    "REPRESENTATION_SCHEMA_VERSION",
    "ARCHITECTURE_VERSION",
    "MEDIA_AI_REPRESENTATION_SPACES",
    "TEXTUAL_HIERARCHY_LEVELS",
    "PRESERVE_SUPERVISION_VALUE_STATES",
    "ALLOW_MISSING_SUPERVISION_AS_ZERO",
    "REQUIRE_BACKWARD_ATTRIBUTION_PATH",
    "ATTRIBUTION_TARGET_PATH",
]

missing_block_1_objects = [
    object_name
    for object_name
    in required_block_1_objects
    if object_name not in globals()
]

if missing_block_1_objects:
    raise NameError(
        "Notebook 04 Block 1 is not initialised. "
        "Missing: "
        f"{missing_block_1_objects}"
    )


required_block_3_objects = [
    "HIERARCHICAL_TEXT_AND_IDENTITY_VALID",
    "HIERARCHICAL_IDENTITY_REGISTRY",
    "TOKEN_MAPPING_CONTRACT",
]

missing_block_3_objects = [
    object_name
    for object_name
    in required_block_3_objects
    if object_name not in globals()
]

if missing_block_3_objects:
    raise NameError(
        "Notebook 04 Block 3 is not initialised. "
        "Missing: "
        f"{missing_block_3_objects}"
    )


if not HIERARCHICAL_TEXT_AND_IDENTITY_VALID:
    raise ValueError(
        "Representation specification requires a valid "
        "hierarchical text and identity structure."
    )


# -----------------------------------------------------------------------------
# Representation-contract configuration
# -----------------------------------------------------------------------------

MEDIA_AI_REPRESENTATION_CONTRACT_VERSION = (
    REPRESENTATION_SCHEMA_VERSION
)

MEDIA_AI_REPRESENTATION_OBJECT_TYPE = (
    "media_ai_representation"
)

MEDIA_AI_REPRESENTATION_FORM = (
    "structured_multispace"
)


# -----------------------------------------------------------------------------
# Representation levels
# -----------------------------------------------------------------------------

MEDIA_AI_REPRESENTATION_LEVELS = (
    "token",
    "sentence",
    "paragraph",
    "article",
)


if (
    MEDIA_AI_REPRESENTATION_LEVELS
    != TEXTUAL_HIERARCHY_LEVELS
):
    raise ValueError(
        "Representation levels must match the validated "
        "Notebook 04 textual hierarchy."
    )


# -----------------------------------------------------------------------------
# Representation spaces
# -----------------------------------------------------------------------------

EXPECTED_REPRESENTATION_SPACES = (
    "factual",
    "psychological",
    "social",
)


if set(
    MEDIA_AI_REPRESENTATION_SPACES
) != set(
    EXPECTED_REPRESENTATION_SPACES
):
    raise ValueError(
        "Media AI representation spaces do not match "
        "the configured architecture."
    )


# -----------------------------------------------------------------------------
# Representation-source semantics
# -----------------------------------------------------------------------------

MEDIA_AI_REPRESENTATION_SOURCES = (
    "canonical_supervision",
    "human_reviewed",
    "model_predicted",
    "deterministic",
    "transformative",
    "recurrent",
    "aggregated",
    "counterfactual",
)


# -----------------------------------------------------------------------------
# Representation value states
# -----------------------------------------------------------------------------

MEDIA_AI_VALUE_STATES = (
    "observed",
    "insufficient_evidence",
    "not_applicable",
    "unavailable",
    "derived",
)


if ALLOW_MISSING_SUPERVISION_AS_ZERO:
    raise ValueError(
        "Missing supervision must not be converted to numeric zero."
    )


# -----------------------------------------------------------------------------
# Factual-space ontology
# -----------------------------------------------------------------------------

FACTUAL_WEIGHT_DIMENSION = (
    "factological_weight"
)


FACTUAL_COMPONENT_DIMENSIONS = (
    "explicit_claim",
    "reported_event_state",
    "observation",
    "quantitative_information",
    "attribution",
    "evidence_reference",
    "certainty_statement",
    "uncertainty_statement",
    "speculation_or_prediction",
)


FACTUAL_DIMENSIONS = (
    FACTUAL_WEIGHT_DIMENSION,
    *FACTUAL_COMPONENT_DIMENSIONS,
)


FACTUAL_DIMENSION_COUNT = len(
    FACTUAL_DIMENSIONS
)


FACTUAL_EXPECTED_DIMENSION_COUNT = 10


if (
    FACTUAL_DIMENSION_COUNT
    != FACTUAL_EXPECTED_DIMENSION_COUNT
):
    raise ValueError(
        "Factual representation must contain "
        "one factological-weight dimension and nine "
        "factological components."
    )


FACTUAL_DIMENSION_INDEX = {
    dimension:
        index

    for index, dimension
    in enumerate(
        FACTUAL_DIMENSIONS
    )
}


FACTUAL_DIMENSION_SPECIFICATION = {
    "factological_weight": {
        "kind":
            "continuous",

        "minimum":
            0.0,

        "maximum":
            1.0,

        "semantic_role":
            "overall_factological_contribution",

        "truth_probability":
            False,
    },

    "explicit_claim": {
        "kind":
            "boolean_or_missing",

        "minimum":
            0.0,

        "maximum":
            1.0,

        "semantic_role":
            "assertive_or_potentially_verifiable_claim",

        "truth_probability":
            False,
    },

    "reported_event_state": {
        "kind":
            "boolean_or_missing",

        "minimum":
            0.0,

        "maximum":
            1.0,

        "semantic_role":
            "reported_event_state_or_condition",

        "truth_probability":
            False,
    },

    "observation": {
        "kind":
            "boolean_or_missing",

        "minimum":
            0.0,

        "maximum":
            1.0,

        "semantic_role":
            "reported_observation_measurement_or_perception",

        "truth_probability":
            False,
    },

    "quantitative_information": {
        "kind":
            "boolean_or_missing",

        "minimum":
            0.0,

        "maximum":
            1.0,

        "semantic_role":
            "quantitative_or_measurement_information",

        "truth_probability":
            False,
    },

    "attribution": {
        "kind":
            "boolean_or_missing",

        "minimum":
            0.0,

        "maximum":
            1.0,

        "semantic_role":
            "information_or_claim_attribution",

        "truth_probability":
            False,
    },

    "evidence_reference": {
        "kind":
            "boolean_or_missing",

        "minimum":
            0.0,

        "maximum":
            1.0,

        "semantic_role":
            "reference_to_supporting_evidence",

        "truth_probability":
            False,
    },

    "certainty_statement": {
        "kind":
            "boolean_or_missing",

        "minimum":
            0.0,

        "maximum":
            1.0,

        "semantic_role":
            "explicit_epistemic_certainty",

        "truth_probability":
            False,
    },

    "uncertainty_statement": {
        "kind":
            "boolean_or_missing",

        "minimum":
            0.0,

        "maximum":
            1.0,

        "semantic_role":
            "explicit_epistemic_uncertainty",

        "truth_probability":
            False,
    },

    "speculation_or_prediction": {
        "kind":
            "boolean_or_missing",

        "minimum":
            0.0,

        "maximum":
            1.0,

        "semantic_role":
            "speculation_inference_or_prediction",

        "truth_probability":
            False,
    },
}


# -----------------------------------------------------------------------------
# Psychological-space ontology
# -----------------------------------------------------------------------------

PSYCHOLOGICAL_EMOTIONAL_DIMENSIONS = (
    "fear",
    "anger",
    "sadness",
    "disgust",
    "tension",
    "uncertainty",
    "hope",
    "relief",
    "trust",
    "compassion",
    "solidarity",
)


PSYCHOLOGICAL_EMOTION_DIMENSION_COUNT = len(
    PSYCHOLOGICAL_EMOTIONAL_DIMENSIONS
)


if (
    PSYCHOLOGICAL_EMOTION_DIMENSION_COUNT
    != 11
):
    raise ValueError(
        "Psychological representation must preserve "
        "the configured eleven emotional dimensions."
    )


PSYCHOLOGICAL_CONSTRUCTS = (
    "reported_emotion",
    "transformative_emotion",
    "immediate_emotional_state",
    "emotional_weight",
)


PSYCHOLOGICAL_DIMENSIONS = (
    tuple(
        f"reported_emotion.{dimension}"
        for dimension
        in PSYCHOLOGICAL_EMOTIONAL_DIMENSIONS
    )
    +
    tuple(
        f"transformative_emotion.{dimension}"
        for dimension
        in PSYCHOLOGICAL_EMOTIONAL_DIMENSIONS
    )
    +
    tuple(
        f"immediate_emotional_state.{dimension}"
        for dimension
        in PSYCHOLOGICAL_EMOTIONAL_DIMENSIONS
    )
    +
    (
        "emotional_weight",
    )
)


PSYCHOLOGICAL_DIMENSION_COUNT = len(
    PSYCHOLOGICAL_DIMENSIONS
)


PSYCHOLOGICAL_EXPECTED_DIMENSION_COUNT = (
    11
    + 11
    + 11
    + 1
)


if (
    PSYCHOLOGICAL_DIMENSION_COUNT
    != PSYCHOLOGICAL_EXPECTED_DIMENSION_COUNT
):
    raise ValueError(
        "Explicit psychological representation must "
        "contain 34 dimensions."
    )


PSYCHOLOGICAL_DIMENSION_INDEX = {
    dimension:
        index

    for index, dimension
    in enumerate(
        PSYCHOLOGICAL_DIMENSIONS
    )
}


PSYCHOLOGICAL_DIMENSION_SPECIFICATION = {}


for dimension in (
    PSYCHOLOGICAL_EMOTIONAL_DIMENSIONS
):

    PSYCHOLOGICAL_DIMENSION_SPECIFICATION[
        f"reported_emotion.{dimension}"
    ] = {
        "kind":
            "continuous",

        "minimum":
            0.0,

        "maximum":
            1.0,

        "construct":
            "reported_emotion",

        "dimension":
            dimension,

        "change_variable":
            False,
    }


    PSYCHOLOGICAL_DIMENSION_SPECIFICATION[
        f"transformative_emotion.{dimension}"
    ] = {
        "kind":
            "continuous",

        "minimum":
            -1.0,

        "maximum":
            1.0,

        "construct":
            "transformative_emotion",

        "dimension":
            dimension,

        "change_variable":
            True,
    }


    PSYCHOLOGICAL_DIMENSION_SPECIFICATION[
        f"immediate_emotional_state.{dimension}"
    ] = {
        "kind":
            "continuous",

        "minimum":
            0.0,

        "maximum":
            1.0,

        "construct":
            "immediate_emotional_state",

        "dimension":
            dimension,

        "change_variable":
            False,
    }


PSYCHOLOGICAL_DIMENSION_SPECIFICATION[
    "emotional_weight"
] = {
    "kind":
        "continuous",

    "minimum":
        0.0,

    "maximum":
        1.0,

    "construct":
        "emotional_weight",

    "dimension":
        "emotional_weight",

    "change_variable":
        False,
}


# -----------------------------------------------------------------------------
# Psychological construct slices
# -----------------------------------------------------------------------------

PSYCHOLOGICAL_CONSTRUCT_SLICES = {
    "reported_emotion":
        (
            0,
            11,
        ),

    "transformative_emotion":
        (
            11,
            22,
        ),

    "immediate_emotional_state":
        (
            22,
            33,
        ),

    "emotional_weight":
        (
            33,
            34,
        ),
}


# -----------------------------------------------------------------------------
# Social-space structural ontology
# -----------------------------------------------------------------------------

SOCIAL_REPRESENTATION_COMPONENTS = (
    "actors",
    "groups",
    "memberships",
    "relationships",
)


SOCIAL_REPRESENTATION_FIXED_DIMENSIONAL = False


SOCIAL_ACTOR_SCHEMA = {
    "actor_id": {
        "required":
            True,

        "description":
            "Stable social-object identifier.",
    },

    "actor_label": {
        "required":
            True,

        "description":
            "Human-readable actor or entity label.",
    },

    "actor_type": {
        "required":
            False,

        "description":
            "Person, organisation, institution or other actor type.",
    },

    "textual_evidence_ids": {
        "required":
            True,

        "description":
            "Traceable textual units supporting the actor representation.",
    },
}


SOCIAL_GROUP_SCHEMA = {
    "group_id": {
        "required":
            True,
    },

    "group_label": {
        "required":
            True,
    },

    "group_type": {
        "required":
            False,
    },

    "textual_evidence_ids": {
        "required":
            True,
    },
}


SOCIAL_MEMBERSHIP_SCHEMA = {
    "actor_id": {
        "required":
            True,
    },

    "group_id": {
        "required":
            True,
    },

    "membership_weight": {
        "required":
            False,
    },

    "membership_state": {
        "required":
            False,
    },

    "textual_evidence_ids": {
        "required":
            True,
    },
}


SOCIAL_RELATIONSHIP_SCHEMA = {
    "source_social_id": {
        "required":
            True,
    },

    "target_social_id": {
        "required":
            True,
    },

    "relationship_type": {
        "required":
            True,
    },

    "relationship_weight": {
        "required":
            False,
    },

    "directional": {
        "required":
            True,
    },

    "textual_evidence_ids": {
        "required":
            True,
    },
}


SOCIAL_REPRESENTATION_SPECIFICATION = {
    "actors": {
        "variable_length":
            True,

        "schema":
            SOCIAL_ACTOR_SCHEMA,
    },

    "groups": {
        "variable_length":
            True,

        "schema":
            SOCIAL_GROUP_SCHEMA,
    },

    "memberships": {
        "variable_length":
            True,

        "schema":
            SOCIAL_MEMBERSHIP_SCHEMA,
    },

    "relationships": {
        "variable_length":
            True,

        "schema":
            SOCIAL_RELATIONSHIP_SCHEMA,
    },

    "overlapping_group_membership_allowed":
        True,

    "fixed_flat_vector_required":
        False,
}


# -----------------------------------------------------------------------------
# Complete Media AI representation formula dictionary
# -----------------------------------------------------------------------------

MEDIA_AI_REPRESENTATION_DICTIONARY = {
    "T_media_ai": {
        "meaning":
            "Complete Media AI representation of one textual unit.",

        "components": (
            "F",
            "P",
            "S",
        ),
    },

    "F": {
        "meaning":
            "Factual representation.",

        "fixed_dimension_count":
            FACTUAL_DIMENSION_COUNT,
    },

    "P": {
        "meaning":
            "Psychological representation.",

        "fixed_dimension_count":
            PSYCHOLOGICAL_DIMENSION_COUNT,
    },

    "S": {
        "meaning":
            "Social and relational representation.",

        "fixed_dimension_count":
            None,

        "variable_structure":
            True,
    },

    "V": {
        "meaning":
            "Numerical representation values.",
    },

    "M": {
        "meaning":
            "Validity or availability mask.",
    },

    "C": {
        "meaning":
            "Confidence metadata.",
    },

    "Q": {
        "meaning":
            "Representation provenance.",
    },

    "Delta_T": {
        "meaning":
            "Transformative state change.",
    },

    "H": {
        "meaning":
            "Recurrent state.",
    },

    "K": {
        "meaning":
            "Confluent representation.",
    },
}


# -----------------------------------------------------------------------------
# Numerical value and mask contract
# -----------------------------------------------------------------------------

NUMERICAL_REPRESENTATION_CONTRACT = {
    "value_array":
        "V",

    "validity_mask":
        "M",

    "mask_available_value":
        1,

    "mask_unavailable_value":
        0,

    "zero_is_substantive":
        True,

    "missing_equals_zero":
        False,

    "technical_missing_placeholder":
        "NaN",

    "preserve_semantic_value_state":
        True,
}


# -----------------------------------------------------------------------------
# Confidence contract
# -----------------------------------------------------------------------------

CONFIDENCE_CONTRACT = {
    "minimum":
        0.0,

    "maximum":
        1.0,

    "part_of_substantive_state":
        False,

    "may_be_used_later_for_weighting":
        True,

    "automatic_weighting_allowed":
        False,
}


# -----------------------------------------------------------------------------
# Representation identity contract
# -----------------------------------------------------------------------------

REPRESENTATION_IDENTITY_REQUIRED_FIELDS = (
    "media_ai_record_id",
    "representation_level",
    "representation_space",
    "representation_source",
    "representation_schema_version",
)


REPRESENTATION_IDENTITY_OPTIONAL_FIELDS = (
    "paragraph_id",
    "sentence_id",
    "token_id",
)


REPRESENTATION_IDENTITY_CONTRACT = {
    "required_fields":
        REPRESENTATION_IDENTITY_REQUIRED_FIELDS,

    "optional_fields":
        REPRESENTATION_IDENTITY_OPTIONAL_FIELDS,

    "permitted_levels":
        MEDIA_AI_REPRESENTATION_LEVELS,

    "permitted_spaces":
        MEDIA_AI_REPRESENTATION_SPACES,

    "permitted_sources":
        MEDIA_AI_REPRESENTATION_SOURCES,
}


# -----------------------------------------------------------------------------
# Representation provenance contract
# -----------------------------------------------------------------------------

REPRESENTATION_PROVENANCE_FIELDS = (
    "media_ai_record_id",
    "paragraph_id",
    "sentence_id",
    "token_id",
    "representation_level",
    "representation_space",
    "representation_source",
    "representation_schema_version",
    "architecture_version",
    "construction_method",
    "model_identifier",
    "model_version",
    "tokeniser_identifier",
    "tokeniser_version",
    "supervisor_identifier",
    "supervision_method",
    "constructed_at_utc",
)


REPRESENTATION_PROVENANCE_CONTRACT = {
    "fields":
        REPRESENTATION_PROVENANCE_FIELDS,

    "preserve_through_aggregation":
        True,

    "preserve_through_recurrence":
        True,

    "preserve_through_confluence":
        True,

    "preserve_through_attribution":
        True,
}


# -----------------------------------------------------------------------------
# Transformative-state contract
# -----------------------------------------------------------------------------

TRANSFORMATIVE_STATE_CONTRACT = {
    "previous_state_symbol":
        "T_previous",

    "current_input_symbol":
        "X_current",

    "state_change_symbol":
        "Delta_T",

    "resulting_state_symbol":
        "T_result",

    "previous_state_distinct":
        True,

    "state_change_distinct":
        True,

    "resulting_state_distinct":
        True,

    "future_context_allowed":
        False,
}


# -----------------------------------------------------------------------------
# Recurrent-state contract
# -----------------------------------------------------------------------------

RECURRENT_STATE_CONTRACT = {
    "previous_recurrent_state":
        "H_previous",

    "current_representation":
        "T_current",

    "current_transform":
        "Delta_T",

    "resulting_recurrent_state":
        "H_current",

    "textual_order_required":
        True,

    "future_context_allowed":
        False,
}


# -----------------------------------------------------------------------------
# Confluent-representation contract
# -----------------------------------------------------------------------------

CONFLUENT_REPRESENTATION_CONTRACT = {
    "symbol":
        "K",

    "input_spaces":
        (
            "factual",
            "psychological",
            "social",
        ),

    "space_identity_preserved":
        True,

    "provenance_preserved":
        True,

    "social_structure_may_be_dynamic":
        True,

    "fixed_flat_concatenation_required":
        False,
}


# -----------------------------------------------------------------------------
# Hierarchical representation contract
# -----------------------------------------------------------------------------

HIERARCHICAL_REPRESENTATION_CONTRACT = {
    "levels":
        MEDIA_AI_REPRESENTATION_LEVELS,

    "construction_path":
        (
            "token",
            "sentence",
            "paragraph",
            "article",
        ),

    "simple_mean_assumed":
        False,

    "explicit_aggregation_required":
        True,

    "lower_level_contribution_mapping_required":
        True,

    "textual_order_preserved":
        True,
}


# -----------------------------------------------------------------------------
# Aggregation formula dictionary
# -----------------------------------------------------------------------------

AGGREGATION_FORMULA_DICTIONARY = {
    "T_lower": {
        "meaning":
            "One lower-level Media AI representation.",
    },

    "T_higher": {
        "meaning":
            "Higher-level representation constructed from lower-level units.",
    },

    "A": {
        "meaning":
            "Explicitly declared aggregation function.",
    },

    "n": {
        "meaning":
            "Number of contributing lower-level textual units.",
    },
}


# -----------------------------------------------------------------------------
# Backward attribution contract
# -----------------------------------------------------------------------------

if not REQUIRE_BACKWARD_ATTRIBUTION_PATH:
    raise ValueError(
        "Backward attribution must remain enabled "
        "for the Media AI representation architecture."
    )


BACKWARD_ATTRIBUTION_CONTRACT = {
    "required":
        True,

    "target_path":
        ATTRIBUTION_TARGET_PATH,

    "higher_to_lower_mapping_required":
        True,

    "canonical_text_terminal":
        "original_text",

    "contribution_score_required_later":
        True,

    "contribution_score_calculated_here":
        False,
}


ATTRIBUTION_FORMULA_DICTIONARY = {
    "A_h_from_l": {
        "meaning":
            "Contribution or attribution relationship from "
            "lower-level unit l to higher-level representation h.",
    },

    "h": {
        "meaning":
            "Higher-level representation.",
    },

    "l": {
        "meaning":
            "Contributing lower-level textual unit.",
    },
}


# -----------------------------------------------------------------------------
# Social Polarisation Distance contract
# -----------------------------------------------------------------------------

SOCIAL_POLARISATION_DISTANCE_CONTRACT = {
    "name":
        "Social Polarisation Distance",

    "abbreviation":
        "SPD",

    "calculated_here":
        False,

    "requires_group_representation":
        True,

    "requires_explicit_distance_function":
        True,

    "signed_change_supported":
        True,

    "direction_is_not_moral_judgement":
        True,
}


SPD_FORMULA_DICTIONARY = {
    "SPD_t": {
        "meaning":
            "Social Polarisation Distance between two groups "
            "at textual step t.",
    },

    "g_i": {
        "meaning":
            "First represented social group.",
    },

    "g_j": {
        "meaning":
            "Second represented social group.",
    },

    "S_t_g_i": {
        "meaning":
            "Social representation of group g_i at step t.",
    },

    "S_t_g_j": {
        "meaning":
            "Social representation of group g_j at step t.",
    },

    "D": {
        "meaning":
            "Explicitly declared relational distance function.",
    },

    "Delta_SPD": {
        "meaning":
            "Change in Social Polarisation Distance between "
            "two textual positions.",
    },
}


# -----------------------------------------------------------------------------
# Supervision-consumption contract
# -----------------------------------------------------------------------------

SUPERVISION_REPRESENTATION_CONTRACT = {
    "canonical_only":
        True,

    "raw_provider_allowed":
        False,

    "rejected_supervision_allowed":
        False,

    "missing_supervision_as_zero":
        False,

    "value_states_preserved":
        PRESERVE_SUPERVISION_VALUE_STATES,

    "supervised_and_predicted_distinct":
        True,

    "supervision_confidence_separate":
        True,
}


# -----------------------------------------------------------------------------
# Representation-space specification
# -----------------------------------------------------------------------------

MEDIA_AI_REPRESENTATION_SPECIFICATION = {
    "contract_type":
        MEDIA_AI_REPRESENTATION_OBJECT_TYPE,

    "contract_version":
        MEDIA_AI_REPRESENTATION_CONTRACT_VERSION,

    "architecture_version":
        ARCHITECTURE_VERSION,

    "representation_form":
        MEDIA_AI_REPRESENTATION_FORM,

    "levels":
        MEDIA_AI_REPRESENTATION_LEVELS,

    "spaces": {
        "factual": {
            "symbol":
                "F",

            "fixed_dimensional":
                True,

            "dimension_count":
                FACTUAL_DIMENSION_COUNT,

            "dimensions":
                FACTUAL_DIMENSIONS,

            "dimension_index":
                deepcopy(
                    FACTUAL_DIMENSION_INDEX
                ),

            "specification":
                deepcopy(
                    FACTUAL_DIMENSION_SPECIFICATION
                ),
        },

        "psychological": {
            "symbol":
                "P",

            "fixed_dimensional":
                True,

            "dimension_count":
                PSYCHOLOGICAL_DIMENSION_COUNT,

            "emotional_dimensions":
                PSYCHOLOGICAL_EMOTIONAL_DIMENSIONS,

            "constructs":
                PSYCHOLOGICAL_CONSTRUCTS,

            "dimensions":
                PSYCHOLOGICAL_DIMENSIONS,

            "dimension_index":
                deepcopy(
                    PSYCHOLOGICAL_DIMENSION_INDEX
                ),

            "construct_slices":
                deepcopy(
                    PSYCHOLOGICAL_CONSTRUCT_SLICES
                ),

            "specification":
                deepcopy(
                    PSYCHOLOGICAL_DIMENSION_SPECIFICATION
                ),
        },

        "social": {
            "symbol":
                "S",

            "fixed_dimensional":
                False,

            "components":
                SOCIAL_REPRESENTATION_COMPONENTS,

            "specification":
                deepcopy(
                    SOCIAL_REPRESENTATION_SPECIFICATION
                ),
        },
    },

    "value_states":
        MEDIA_AI_VALUE_STATES,

    "sources":
        MEDIA_AI_REPRESENTATION_SOURCES,

    "numerical_contract":
        deepcopy(
            NUMERICAL_REPRESENTATION_CONTRACT
        ),

    "confidence_contract":
        deepcopy(
            CONFIDENCE_CONTRACT
        ),

    "identity_contract":
        deepcopy(
            REPRESENTATION_IDENTITY_CONTRACT
        ),

    "provenance_contract":
        deepcopy(
            REPRESENTATION_PROVENANCE_CONTRACT
        ),

    "hierarchical_contract":
        deepcopy(
            HIERARCHICAL_REPRESENTATION_CONTRACT
        ),

    "transformative_contract":
        deepcopy(
            TRANSFORMATIVE_STATE_CONTRACT
        ),

    "recurrent_contract":
        deepcopy(
            RECURRENT_STATE_CONTRACT
        ),

    "confluent_contract":
        deepcopy(
            CONFLUENT_REPRESENTATION_CONTRACT
        ),

    "attribution_contract":
        deepcopy(
            BACKWARD_ATTRIBUTION_CONTRACT
        ),

    "social_polarisation_distance":
        deepcopy(
            SOCIAL_POLARISATION_DISTANCE_CONTRACT
        ),

    "supervision_contract":
        deepcopy(
            SUPERVISION_REPRESENTATION_CONTRACT
        ),
}


# -----------------------------------------------------------------------------
# Formal dimensionality summary
# -----------------------------------------------------------------------------

MEDIA_AI_FIXED_DIMENSION_SUMMARY = {
    "factual":
        FACTUAL_DIMENSION_COUNT,

    "psychological":
        PSYCHOLOGICAL_DIMENSION_COUNT,

    "social":
        None,

    "fixed_dimensions_total_without_social":
        (
            FACTUAL_DIMENSION_COUNT
            + PSYCHOLOGICAL_DIMENSION_COUNT
        ),
}


# -----------------------------------------------------------------------------
# Representation-contract validation
# -----------------------------------------------------------------------------

REPRESENTATION_SPECIFICATION_ERRORS = []


if FACTUAL_DIMENSION_COUNT != 10:

    REPRESENTATION_SPECIFICATION_ERRORS.append(
        "invalid_factual_dimension_count"
    )


if len(
    FACTUAL_COMPONENT_DIMENSIONS
) != 9:

    REPRESENTATION_SPECIFICATION_ERRORS.append(
        "invalid_factual_component_count"
    )


if PSYCHOLOGICAL_DIMENSION_COUNT != 34:

    REPRESENTATION_SPECIFICATION_ERRORS.append(
        "invalid_psychological_dimension_count"
    )


if (
    len(
        FACTUAL_DIMENSION_INDEX
    )
    != FACTUAL_DIMENSION_COUNT
):

    REPRESENTATION_SPECIFICATION_ERRORS.append(
        "invalid_factual_dimension_index"
    )


if (
    len(
        PSYCHOLOGICAL_DIMENSION_INDEX
    )
    != PSYCHOLOGICAL_DIMENSION_COUNT
):

    REPRESENTATION_SPECIFICATION_ERRORS.append(
        "invalid_psychological_dimension_index"
    )


if (
    SOCIAL_REPRESENTATION_COMPONENTS
    != (
        "actors",
        "groups",
        "memberships",
        "relationships",
    )
):

    REPRESENTATION_SPECIFICATION_ERRORS.append(
        "invalid_social_component_structure"
    )


if (
    "memberships"
    not in SOCIAL_REPRESENTATION_SPECIFICATION
):

    REPRESENTATION_SPECIFICATION_ERRORS.append(
        "missing_social_membership_specification"
    )


if (
    SOCIAL_REPRESENTATION_FIXED_DIMENSIONAL
    is not False
):

    REPRESENTATION_SPECIFICATION_ERRORS.append(
        "social_space_prematurely_fixed"
    )


if (
    NUMERICAL_REPRESENTATION_CONTRACT[
        "missing_equals_zero"
    ]
    is not False
):

    REPRESENTATION_SPECIFICATION_ERRORS.append(
        "missingness_zero_conflict"
    )


if (
    CONFIDENCE_CONTRACT[
        "part_of_substantive_state"
    ]
    is not False
):

    REPRESENTATION_SPECIFICATION_ERRORS.append(
        "confidence_incorrectly_embedded_in_state"
    )


if (
    TRANSFORMATIVE_STATE_CONTRACT[
        "future_context_allowed"
    ]
    is not False
):

    REPRESENTATION_SPECIFICATION_ERRORS.append(
        "transformative_future_context_allowed"
    )


if (
    RECURRENT_STATE_CONTRACT[
        "future_context_allowed"
    ]
    is not False
):

    REPRESENTATION_SPECIFICATION_ERRORS.append(
        "recurrent_future_context_allowed"
    )


if (
    HIERARCHICAL_REPRESENTATION_CONTRACT[
        "simple_mean_assumed"
    ]
    is not False
):

    REPRESENTATION_SPECIFICATION_ERRORS.append(
        "implicit_simple_mean_aggregation"
    )


if (
    CONFLUENT_REPRESENTATION_CONTRACT[
        "fixed_flat_concatenation_required"
    ]
    is not False
):

    REPRESENTATION_SPECIFICATION_ERRORS.append(
        "premature_confluent_flattening"
    )


if (
    SUPERVISION_REPRESENTATION_CONTRACT[
        "missing_supervision_as_zero"
    ]
    is not False
):

    REPRESENTATION_SPECIFICATION_ERRORS.append(
        "missing_supervision_zero_conflict"
    )


if tuple(
    BACKWARD_ATTRIBUTION_CONTRACT[
        "target_path"
    ]
) != tuple(
    ATTRIBUTION_TARGET_PATH
):

    REPRESENTATION_SPECIFICATION_ERRORS.append(
        "invalid_attribution_target_path"
    )


MEDIA_AI_REPRESENTATION_SPECIFICATION_VALID = (
    len(
        REPRESENTATION_SPECIFICATION_ERRORS
    )
    == 0
)


if not MEDIA_AI_REPRESENTATION_SPECIFICATION_VALID:

    raise ValueError(
        "Media AI representation specification failed validation: "
        f"{REPRESENTATION_SPECIFICATION_ERRORS}"
    )


# -----------------------------------------------------------------------------
# Construction state
# -----------------------------------------------------------------------------

MEDIA_AI_REPRESENTATIONS_CONSTRUCTED = False

FACTUAL_REPRESENTATIONS_CONSTRUCTED = False

PSYCHOLOGICAL_REPRESENTATIONS_CONSTRUCTED = False

SOCIAL_REPRESENTATIONS_CONSTRUCTED = False

TRANSFORMATIVE_STATE_CALCULATED = False

RECURRENT_STATE_CALCULATED = False

CONFLUENT_REPRESENTATION_CALCULATED = False

HIERARCHICAL_AGGREGATION_PERFORMED = False

ATTRIBUTION_SCORES_CALCULATED = False

SOCIAL_POLARISATION_DISTANCE_CALCULATED = False


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Representation Specification")
print("=" * 72)

print(
    f"Representation schema       : "
    f"{MEDIA_AI_REPRESENTATION_CONTRACT_VERSION}"
)

print(
    f"Architecture version        : "
    f"{ARCHITECTURE_VERSION}"
)

print(
    f"Representation form         : "
    f"{MEDIA_AI_REPRESENTATION_FORM}"
)

print(
    f"Representation levels       : "
    f"{' → '.join(MEDIA_AI_REPRESENTATION_LEVELS)}"
)

print("-" * 72)

print("Factual space")

print(
    f"Factological weight         : "
    f"1"
)

print(
    f"Factological components     : "
    f"{len(FACTUAL_COMPONENT_DIMENSIONS)}"
)

print(
    f"Factual dimensions          : "
    f"{FACTUAL_DIMENSION_COUNT}"
)

print(
    f"Fixed dimensional           : "
    f"{True}"
)

print("-" * 72)

print("Psychological space")

print(
    f"Emotional dimensions        : "
    f"{PSYCHOLOGICAL_EMOTION_DIMENSION_COUNT}"
)

print(
    f"Reported-emotion dimensions : "
    f"{len(PSYCHOLOGICAL_EMOTIONAL_DIMENSIONS)}"
)

print(
    f"Transformative dimensions   : "
    f"{len(PSYCHOLOGICAL_EMOTIONAL_DIMENSIONS)}"
)

print(
    f"Immediate-state dimensions  : "
    f"{len(PSYCHOLOGICAL_EMOTIONAL_DIMENSIONS)}"
)

print(
    f"Emotional-weight dimensions : "
    f"1"
)

print(
    f"Psychological dimensions    : "
    f"{PSYCHOLOGICAL_DIMENSION_COUNT}"
)

print(
    f"Fixed dimensional           : "
    f"{True}"
)

print("-" * 72)

print("Social space")

print(
    f"Structural components       : "
    f"{', '.join(SOCIAL_REPRESENTATION_COMPONENTS)}"
)

print(
    f"Structural component count  : "
    f"{len(SOCIAL_REPRESENTATION_COMPONENTS)}"
)

print(
    f"Overlapping membership      : "
    f"{SOCIAL_REPRESENTATION_SPECIFICATION['overlapping_group_membership_allowed']}"
)

print(
    f"Fixed dimensional           : "
    f"{SOCIAL_REPRESENTATION_FIXED_DIMENSIONAL}"
)

print("-" * 72)

print("Value and source semantics")

print(
    f"Value states                : "
    f"{len(MEDIA_AI_VALUE_STATES)}"
)

print(
    f"Representation sources      : "
    f"{len(MEDIA_AI_REPRESENTATION_SOURCES)}"
)

print(
    f"Missing equals zero         : "
    f"{NUMERICAL_REPRESENTATION_CONTRACT['missing_equals_zero']}"
)

print(
    f"Confidence part of state    : "
    f"{CONFIDENCE_CONTRACT['part_of_substantive_state']}"
)

print("-" * 72)

print("Architecture contracts")

print(
    f"Transformative contract     : "
    f"{bool(TRANSFORMATIVE_STATE_CONTRACT)}"
)

print(
    f"Recurrent contract          : "
    f"{bool(RECURRENT_STATE_CONTRACT)}"
)

print(
    f"Confluent contract          : "
    f"{bool(CONFLUENT_REPRESENTATION_CONTRACT)}"
)

print(
    f"Hierarchical contract       : "
    f"{bool(HIERARCHICAL_REPRESENTATION_CONTRACT)}"
)

print(
    f"Backward attribution        : "
    f"{BACKWARD_ATTRIBUTION_CONTRACT['required']}"
)

print(
    f"SPD contract defined        : "
    f"{bool(SOCIAL_POLARISATION_DISTANCE_CONTRACT)}"
)

print("-" * 72)

print("Fixed-dimensional summary")

print(
    f"Factual                     : "
    f"{MEDIA_AI_FIXED_DIMENSION_SUMMARY['factual']}"
)

print(
    f"Psychological               : "
    f"{MEDIA_AI_FIXED_DIMENSION_SUMMARY['psychological']}"
)

print(
    f"Social                      : "
    f"{MEDIA_AI_FIXED_DIMENSION_SUMMARY['social']}"
)

print(
    f"Fixed total without social  : "
    f"{MEDIA_AI_FIXED_DIMENSION_SUMMARY['fixed_dimensions_total_without_social']}"
)

print("-" * 72)

print(
    f"Specification valid         : "
    f"{MEDIA_AI_REPRESENTATION_SPECIFICATION_VALID}"
)

print(
    f"Representations constructed : "
    f"{MEDIA_AI_REPRESENTATIONS_CONSTRUCTED}"
)

print("=" * 72)

print(
    "Media AI representation specification initialised successfully."
)

print(
    "The factual space contains one factological-weight dimension "
    "and nine explicit factological components."
)

print(
    "Reported event/state and observation are represented "
    "as separate factual dimensions."
)

print(
    "Factual dimensions describe textual factological structure "
    "and do not constitute factual-truth verification."
)

print(
    "Factual and psychological spaces have explicit fixed-dimensional "
    "contracts."
)

print(
    "The social space preserves actors, groups, memberships "
    "and relationships as explicit structural components."
)

print(
    "The social space remains structured and variable-dimensional."
)

print(
    "Missing values remain distinct from substantive numerical zero."
)

print(
    "Confidence and provenance remain separate from substantive state."
)

print(
    "Transformative, recurrent, confluent and hierarchical operations "
    "are defined as architectural contracts only."
)

print(
    "Backward attribution and Social Polarisation Distance requirements "
    "have been specified but not calculated."
)

print(
    "No Media AI representation tensor has been constructed."
)

print(
    "No model-specific tokenisation, model training or evaluation "
    "has been performed."
)
````

## Model-Ready Text and Token Traceability

This block prepares the validated textual hierarchy for later model-specific processing while preserving complete traceability between model-facing tokens and the prepared research text.

The purpose of this block is not to construct Media AI representations.

Its purpose is to establish the model-facing textual layer required by later factual, psychological, social, transformative, recurrent, confluent and attribution mechanisms.

The required traceability path is:

[
Token
\rightarrow
Sentence
\rightarrow
Paragraph
\rightarrow
Article
]

and, where token offsets permit direct reconstruction:

[
Token
\rightarrow
Text\ Span
\rightarrow
Prepared\ Text
]

This block therefore establishes the boundary between the prepared research corpus and model-specific textual processing.

### Model-Ready Text

For each textual unit (u), let:

[
X_u^{text}
]

denote the model-ready textual input derived from the prepared text.

Dictionary:

* (u) = traceable textual unit;
* (X_u^{text}) = model-facing text associated with that unit.

The preparation process must preserve semantic content.

Model-ready preparation may perform only explicitly configured technical transformations required for model processing.

It must not silently:

* paraphrase text;
* summarise text;
* translate text;
* remove substantively meaningful language;
* introduce information not present in the prepared corpus.

The original prepared text remains the reference textual object.

### Text Preservation

Let:

[
X_u^{prepared}
]

represent the prepared research text and:

[
X_u^{model}
]

represent the corresponding model-facing text.

The relationship is:

[
X_u^{prepared}
\rightarrow
X_u^{model}
]

where the transformation must be documented and traceable.

Dictionary:

* (X_u^{prepared}) = prepared research text;
* (X_u^{model}) = model-facing version of that text;
* the arrow represents a controlled technical transformation rather than semantic rewriting.

Where no technical transformation is necessary:

[
X_u^{model}
===========

X_u^{prepared}
]

The block must record whether any transformation has occurred.

### Tokenisation

For a textual unit (u), model-specific tokenisation is represented as:

[
\mathcal{T}(X_u^{model})
========================

[
\tau_1,
\tau_2,
\ldots,
\tau_n
]
]

Dictionary:

* (\mathcal{T}) = tokenizer;
* (X_u^{model}) = model-facing text;
* (\tau_i) = token at position (i);
* (n) = number of produced tokens.

Tokenisation is a technical model operation.

Tokens must not be treated as independent research observations.

Their research meaning remains dependent on their relationship to the prepared textual hierarchy.

### Token Identity

Each token must receive a traceable token identity.

Conceptually:

[
I_{\tau_i}
==========

{
ID_{\tau_i},
ID_s,
ID_p,
ID_a,
i
}
]

Dictionary:

* (I_{\tau_i}) = identity record for token (\tau_i);
* (ID_{\tau_i}) = stable token-level identifier within the tokenisation artefact;
* (ID_s) = parent sentence ID;
* (ID_p) = parent paragraph ID;
* (ID_a) = parent article ID;
* (i) = token position within the relevant token sequence.

Token identities must be deterministic for a fixed prepared corpus, tokenizer and tokenisation configuration.

A change of tokenizer or materially different tokenisation configuration may produce a different token identity space and must therefore be recorded explicitly.

### Token-to-Text Mapping

Where supported by the tokenizer, each token should preserve its corresponding character span:

[
O_{\tau_i}
==========

[
b_i,
e_i
)
]

Dictionary:

* (O_{\tau_i}) = character-offset interval associated with token (\tau_i);
* (b_i) = beginning character offset;
* (e_i) = ending character offset;
* ([b_i,e_i)) = half-open interval in the associated model-facing text.

The corresponding text span is:

[
X_u^{model}[b_i:e_i]
]

This mapping allows later token-level attribution to be projected back onto readable text.

Offset mapping is therefore a preferred tokenizer capability for the Media AI architecture.

If a selected tokenizer cannot provide reliable offsets, the limitation must be recorded rather than silently approximated.

### Special Tokens

Model tokenizers may introduce technical tokens that do not correspond directly to lexical spans in the prepared text.

Let:

[
z_i^{special}
\in
{0,1}
]

where:

* (z_i^{special}=1) identifies a tokenizer-introduced special token;
* (z_i^{special}=0) identifies a token associated with textual content.

Examples may include model-specific beginning, ending, separator or padding tokens.

Special tokens must remain distinguishable from text-bearing tokens.

They must not be assigned fabricated textual evidence spans.

### Attention Mask

Where required by the selected model, the token sequence may include an attention mask:

[
m_i^{attention}
\in
{0,1}
]

Dictionary:

* (m_i^{attention}=1) = token participates in the configured model input;
* (m_i^{attention}=0) = token is excluded from attention under the model input configuration.

The attention mask is a model-input property.

It is distinct from the representation-validity masks defined in Block 4.

Therefore:

[
M^{representation}
\neq
M^{attention}
]

These two masks must not be conflated.

### Token Type Information

Where required by a model architecture, token-type or segment information may also be retained.

Let:

[
g_i
]

denote the model-specific segment or token-type identifier associated with token (i).

Dictionary:

* (g_i) = optional tokenizer/model segment identifier;
* its interpretation depends on the selected model architecture.

If the selected model does not use token-type identifiers, the field may remain unavailable or not applicable.

### Sentence as the Primary Model-Traceability Anchor

Sentence identity remains the primary bridge between the prepared corpus, validated supervision and later model outputs.

For token (\tau_i):

[
\tau_i
\rightarrow
s
]

where (s) is exactly one parent sentence.

The sentence then maps deterministically upward:

[
s
\rightarrow
p
\rightarrow
a
]

where:

* (s) = sentence;
* (p) = paragraph;
* (a) = article.

Therefore every text-bearing model token must ultimately be traceable to an article through the validated hierarchy.

### Article Context and Sentence Targets

The supervision methodology distinguishes between the external supervision request unit and the annotation unit:

[
Article
\rightarrow
Sentence\ Annotations
]

Model preparation must preserve this distinction.

Complete article context may be used where required by a later model architecture, while sentence identities remain available as explicit analytical and supervision targets.

The block must therefore support both:

[
X_a^{model}
]

and:

[
X_s^{model}
]

where:

* (X_a^{model}) = model-ready article text;
* (X_s^{model}) = model-ready sentence text.

This does not imply that both must be supplied simultaneously to every later model.

The choice of model context is an explicit architectural decision.

### Temporal Context Boundary

For sequential or recurrent processing at textual position (t), the permitted information set remains:

[
\mathcal{I}_t
=============

{
X_1,
X_2,
\ldots,
X_t
}
]

Dictionary:

* (\mathcal{I}_t) = textual information available at step (t);
* (X_t) = current textual unit;
* (X_1,\ldots,X_{t-1}) = preceding textual units.

Future textual units:

[
X_{t+1},
X_{t+2},
\ldots
]

must not enter representations assigned to earlier textual steps where the configured architecture requires preceding-context processing.

Full article availability at tokenisation time does not itself authorise future-context leakage during later state construction.

### Sequence Length and Truncation

Model-specific tokenizers and encoders may impose maximum sequence lengths.

Let:

[
n_u
]

denote the number of tokens produced for textual unit (u), and:

[
L_{max}
]

the maximum sequence length supported or configured for a selected model.

Potential overflow occurs when:

[
n_u > L_{max}
]

Dictionary:

* (n_u) = token count for textual unit (u);
* (L_{max}) = configured maximum sequence length.

Substantive text must not be silently discarded.

If a sequence exceeds the configured model limit, the later implementation must use an explicit strategy such as:

* hierarchical processing;
* deterministic chunking;
* overlapping windows;
* another documented long-context strategy.

Any such strategy must preserve sentence and article identity.

### Chunking Contract

If textual unit (u) is divided into model-processing chunks:

[
X_u
\rightarrow
{
C_{u,1},
C_{u,2},
\ldots,
C_{u,k}
}
]

where:

* (C_{u,j}) = chunk (j) derived from textual unit (u);
* (k) = number of chunks.

Each chunk must preserve:

* parent textual identity;
* chunk index;
* token range;
* text span where available;
* tokenizer identity;
* overlap information where applicable.

Chunking is a computational operation and must not redefine the research hierarchy.

A chunk therefore does not become a fifth textual aggregation level.

The Media AI hierarchy remains:

[
Token
\rightarrow
Sentence
\rightarrow
Paragraph
\rightarrow
Article
]

### Tokenizer Provenance

Every tokenisation artefact must record sufficient provenance to reproduce the mapping.

Let:

[
Q_{\mathcal{T}}
===============

{
q_{tokenizer},
q_{version},
q_{configuration},
q_{source},
q_{time}
}
]

Dictionary:

* (Q_{\mathcal{T}}) = tokenizer provenance;
* (q_{tokenizer}) = tokenizer identifier;
* (q_{version}) = tokenizer/model version where available;
* (q_{configuration}) = relevant tokenisation parameters;
* (q_{source}) = model or repository source where applicable;
* (q_{time}) = construction timestamp when persisted.

Tokenizer provenance must remain associated with token identities and later model-derived representations.

### Token Traceability Contract

For every text-bearing token, the architecture must support:

[
\tau_i
\rightarrow
ID_s
\rightarrow
ID_p
\rightarrow
ID_a
]

and, where reliable offsets are available:

[
\tau_i
\rightarrow
O_{\tau_i}
\rightarrow
X^{model}
\rightarrow
X^{prepared}
]

This establishes the technical foundation for later backward attribution.

A later attribution score:

[
A_{s \leftarrow \tau_i}
]

can therefore be connected to the exact textual evidence associated with token (\tau_i).

Dictionary:

* (A_{s \leftarrow \tau_i}) = contribution of token (\tau_i) to a sentence-level state or output;
* (s) = parent sentence;
* (\tau_i) = contributing token.

No attribution score is calculated by this block.

### Coloured Text Inspection Boundary

Later interpretability analysis may project token- or span-level contribution values back onto human-readable text.

Conceptually:

[
A_{\tau}
\rightarrow
Textual\ Highlight
]

where:

* (A_{\tau}) = token-level attribution information;
* `Textual Highlight` = visual representation of contribution over the prepared text.

Block 5 does not assign colours or interpret contribution values.

It establishes the traceability required for Notebook 05 to construct such inspection views without losing the connection to the original textual evidence.

### Representation Boundary

Tokenisation does not itself construct:

[
F_u,
\quad
P_u,
\quad
S_u
]

and does not produce:

[
\Delta T_t,
\quad
H_t,
\quad
K_t
]

Dictionary:

* (F_u) = factual representation;
* (P_u) = psychological representation;
* (S_u) = social representation;
* (\Delta T_t) = transformative state change;
* (H_t) = recurrent state;
* (K_t) = confluent representation.

Tokens and tokenisation metadata are model-facing textual artefacts, not Media AI states.

This separation prevents technical tokenisation choices from being mistaken for substantive representation design.

### Supervision Boundary

Validated supervision remains associated with sentence identities.

Tokenisation must not redistribute sentence-level supervision to individual tokens as if token-level labels had been observed.

Therefore:

[
Y_s
\not\Rightarrow
Y_{\tau_i}
]

where:

* (Y_s) = sentence-level supervision;
* (Y_{\tau_i}) = hypothetical token-level supervision.

Token-level contributions may later be learned or estimated through attribution methods, but they must remain distinguishable from directly supervised labels.

Missing supervision must remain missing and must not be introduced as zero during model-ready text construction.

### Determinism and Reproducibility

For a fixed combination of:

* prepared text;
* tokenizer;
* tokenizer version;
* tokenisation configuration;

the tokenisation artefact should be reproducible.

Conceptually:

[
\mathcal{T}(X;\theta)
\rightarrow
Tokens
]

where:

* (X) = fixed model-facing text;
* (\theta) = fixed tokenizer configuration.

Repeated execution with the same inputs and configuration should produce the same token sequence and identity mapping.

Any non-deterministic preprocessing must be explicitly documented.

### Output

This block prepares the contract for model-ready text and token traceability.

Its outputs define:

* model-ready text semantics;
* text-preservation requirements;
* tokenizer requirements;
* token identities;
* token-to-sentence mappings;
* token-to-paragraph mappings;
* token-to-article mappings;
* character-offset requirements;
* special-token handling;
* attention-mask semantics;
* optional token-type information;
* article-context and sentence-target boundaries;
* temporal-context safeguards;
* sequence-length safeguards;
* chunking requirements;
* tokenizer provenance;
* backward-attribution traceability;
* coloured-text inspection prerequisites;
* supervision boundaries;
* determinism requirements.

Where model-ready text construction is enabled by Block 1, this block may create the corresponding model-facing text and token-traceability artefacts.

No embedding is constructed.

No factual, psychological or social representation is constructed.

No sentence-level supervision is converted into token-level supervision.

No transformative or recurrent state is calculated.

No confluent representation is calculated.

No hierarchical aggregation or attribution score is calculated.

No model training or experimental evaluation is performed.


````python
# =============================================================================
# Media AI — Model-Ready Text and Token Traceability
# =============================================================================

from copy import deepcopy
from datetime import datetime, timezone


# -----------------------------------------------------------------------------
# Block dependency checks
# -----------------------------------------------------------------------------

required_block_1_objects = [
    "BUILD_MODEL_READY_TEXT",
    "PRESERVE_CANONICAL_TEXT",
    "PRESERVE_STRUCTURAL_TRACEABILITY",
    "PRESERVE_TEXTUAL_ORDER",
    "PERMANENT_MODEL_TOKENISATION",
    "MODEL_SPECIFIC_TEXT_OVERWRITE_ALLOWED",
    "REQUIRE_MODEL_TOKEN_TRACEABILITY",
    "REQUIRE_TOKEN_TO_SENTENCE_MAPPING",
    "REQUIRE_SENTENCE_TO_PARAGRAPH_MAPPING",
    "REQUIRE_PARAGRAPH_TO_ARTICLE_MAPPING",
]

missing_block_1_objects = [
    object_name
    for object_name
    in required_block_1_objects
    if object_name not in globals()
]

if missing_block_1_objects:
    raise NameError(
        "Notebook 04 Block 1 is not initialised. "
        "Missing: "
        f"{missing_block_1_objects}"
    )


required_block_3_objects = [
    "HIERARCHICAL_TEXT_AND_IDENTITY_VALID",
    "HIERARCHICAL_IDENTITY_REGISTRY",
    "TOKEN_MAPPING_CONTRACT",
    "TOKEN_MAPPING_REQUIRED_FIELDS",
    "CANONICAL_SENTENCE_TEXT",
    "SENTENCE_BY_ID",
    "SENTENCE_IDS",
    "SENTENCE_TO_PARAGRAPH_ID",
    "SENTENCE_TO_ARTICLE_ID",
    "PARAGRAPH_TO_ARTICLE_ID",
    "ARTICLE_SENTENCE_ORDER",
]

missing_block_3_objects = [
    object_name
    for object_name
    in required_block_3_objects
    if object_name not in globals()
]

if missing_block_3_objects:
    raise NameError(
        "Notebook 04 Block 3 is not initialised. "
        "Missing: "
        f"{missing_block_3_objects}"
    )


required_block_4_objects = [
    "MEDIA_AI_REPRESENTATION_SPECIFICATION_VALID",
    "MEDIA_AI_REPRESENTATION_SPECIFICATION",
    "MEDIA_AI_REPRESENTATION_LEVELS",
]

missing_block_4_objects = [
    object_name
    for object_name
    in required_block_4_objects
    if object_name not in globals()
]

if missing_block_4_objects:
    raise NameError(
        "Notebook 04 Block 4 is not initialised. "
        "Missing: "
        f"{missing_block_4_objects}"
    )


if not HIERARCHICAL_TEXT_AND_IDENTITY_VALID:
    raise ValueError(
        "Model-ready text construction requires a valid "
        "hierarchical text and identity structure."
    )


if not MEDIA_AI_REPRESENTATION_SPECIFICATION_VALID:
    raise ValueError(
        "Model-ready text construction requires a valid "
        "Media AI representation specification."
    )


# -----------------------------------------------------------------------------
# Model-ready text contract configuration
# -----------------------------------------------------------------------------

MODEL_READY_TEXT_CONTRACT_VERSION = "1.0"

TOKEN_TRACEABILITY_CONTRACT_VERSION = "1.0"

MODEL_READY_TEXT_UNIT = "sentence"

MODEL_CONTEXT_LEVELS_SUPPORTED = (
    "sentence",
    "article",
)

TOKENISATION_EXECUTION_STATUS = (
    "not_performed"
)


# -----------------------------------------------------------------------------
# Model-facing text preservation policy
# -----------------------------------------------------------------------------

MODEL_TEXT_TRANSFORMATION_POLICY = {
    "semantic_rewriting_allowed":
        False,

    "paraphrasing_allowed":
        False,

    "summarisation_allowed":
        False,

    "translation_allowed":
        False,

    "substantive_deletion_allowed":
        False,

    "substantive_insertion_allowed":
        False,

    "technical_normalisation_allowed":
        True,

    "prepared_text_remains_authoritative":
        True,
}


if not PRESERVE_CANONICAL_TEXT:
    raise ValueError(
        "Prepared research text must remain preserved."
    )


if MODEL_SPECIFIC_TEXT_OVERWRITE_ALLOWED:
    raise ValueError(
        "Model-facing text must not overwrite prepared research text."
    )


if PERMANENT_MODEL_TOKENISATION:
    raise ValueError(
        "Permanent model-specific tokenisation is not permitted."
    )


# -----------------------------------------------------------------------------
# Tokeniser capability contract
# -----------------------------------------------------------------------------

TOKENISER_CAPABILITY_REQUIREMENTS = {
    "deterministic_for_fixed_input":
        True,

    "preserve_input_order":
        True,

    "return_token_ids":
        True,

    "return_attention_mask":
        True,

    "offset_mapping_preferred":
        True,

    "special_token_identification_required":
        True,

    "token_type_ids_optional":
        True,

    "sentence_traceability_required":
        REQUIRE_TOKEN_TO_SENTENCE_MAPPING,

    "paragraph_traceability_required":
        REQUIRE_SENTENCE_TO_PARAGRAPH_MAPPING,

    "article_traceability_required":
        REQUIRE_PARAGRAPH_TO_ARTICLE_MAPPING,

    "model_token_traceability_required":
        REQUIRE_MODEL_TOKEN_TRACEABILITY,
}


# -----------------------------------------------------------------------------
# Model-ready sentence text construction
# -----------------------------------------------------------------------------

MODEL_READY_SENTENCE_RECORDS = []


for sentence_id in SENTENCE_IDS:

    prepared_text = (
        CANONICAL_SENTENCE_TEXT[
            sentence_id
        ]
    )

    paragraph_id = (
        SENTENCE_TO_PARAGRAPH_ID[
            sentence_id
        ]
    )

    article_id = (
        SENTENCE_TO_ARTICLE_ID[
            sentence_id
        ]
    )

    sentence_record = (
        SENTENCE_BY_ID[
            sentence_id
        ]
    )

    model_ready_record = {
        "record_type":
            "media_ai_model_ready_sentence",

        "contract_version":
            MODEL_READY_TEXT_CONTRACT_VERSION,

        "media_ai_record_id":
            article_id,

        "paragraph_id":
            paragraph_id,

        "sentence_id":
            sentence_id,

        "paragraph_index":
            sentence_record.get(
                "paragraph_index"
            ),

        "sentence_index":
            sentence_record.get(
                "sentence_index"
            ),

        "prepared_text":
            prepared_text,

        "model_text":
            prepared_text,

        "text_transformed":
            False,

        "transformation_method":
            "identity",

        "tokenisation_status":
            "not_performed",

        "token_count":
            None,

        "token_ids":
            None,

        "attention_mask":
            None,

        "offset_mapping":
            None,

        "special_tokens_mask":
            None,

        "token_type_ids":
            None,

        "tokeniser_identifier":
            None,

        "tokeniser_version":
            None,

        "model_identifier":
            None,

        "model_version":
            None,
    }

    MODEL_READY_SENTENCE_RECORDS.append(
        model_ready_record
    )


# -----------------------------------------------------------------------------
# Sentence-level lookup
# -----------------------------------------------------------------------------

MODEL_READY_SENTENCE_BY_ID = {
    record[
        "sentence_id"
    ]:
        record

    for record
    in MODEL_READY_SENTENCE_RECORDS
}


# -----------------------------------------------------------------------------
# Article-level model-ready text construction
# -----------------------------------------------------------------------------

MODEL_READY_ARTICLE_RECORDS = []


for (
    article_id,
    ordered_sentence_ids,
) in ARTICLE_SENTENCE_ORDER.items():

    ordered_sentence_texts = [
        MODEL_READY_SENTENCE_BY_ID[
            sentence_id
        ][
            "model_text"
        ]
        for sentence_id
        in ordered_sentence_ids
    ]


    article_text = " ".join(
        ordered_sentence_texts
    )


    article_record = {
        "record_type":
            "media_ai_model_ready_article",

        "contract_version":
            MODEL_READY_TEXT_CONTRACT_VERSION,

        "media_ai_record_id":
            article_id,

        "sentence_ids":
            deepcopy(
                ordered_sentence_ids
            ),

        "sentence_count":
            len(
                ordered_sentence_ids
            ),

        "model_text":
            article_text,

        "text_transformed":
            False,

        "transformation_method":
            "hierarchical_sentence_concatenation",

        "tokenisation_status":
            "not_performed",

        "token_count":
            None,

        "token_ids":
            None,

        "attention_mask":
            None,

        "offset_mapping":
            None,

        "special_tokens_mask":
            None,

        "token_type_ids":
            None,

        "tokeniser_identifier":
            None,

        "tokeniser_version":
            None,

        "model_identifier":
            None,

        "model_version":
            None,
    }

    MODEL_READY_ARTICLE_RECORDS.append(
        article_record
    )


# -----------------------------------------------------------------------------
# Article-level lookup
# -----------------------------------------------------------------------------

MODEL_READY_ARTICLE_BY_ID = {
    record[
        "media_ai_record_id"
    ]:
        record

    for record
    in MODEL_READY_ARTICLE_RECORDS
}


# -----------------------------------------------------------------------------
# Token identity contract
# -----------------------------------------------------------------------------

TOKEN_IDENTITY_CONTRACT = {
    "contract_version":
        TOKEN_TRACEABILITY_CONTRACT_VERSION,

    "identity_formula":
        (
            "article_id + sentence_id + token_position "
            "+ tokeniser configuration"
        ),

    "required_parent_links": (
        "sentence_id",
        "paragraph_id",
        "media_ai_record_id",
    ),

    "deterministic_for_fixed_tokenisation":
        True,

    "global_token_identity_required":
        True,
}


# -----------------------------------------------------------------------------
# Token record schema
# -----------------------------------------------------------------------------

MODEL_TOKEN_RECORD_SCHEMA = {
    "token_id": {
        "required":
            True,
    },

    "token_position": {
        "required":
            True,
    },

    "token_representation": {
        "required":
            True,
    },

    "token_text": {
        "required":
            False,
    },

    "model_token_id": {
        "required":
            False,
    },

    "sentence_id": {
        "required":
            True,
    },

    "paragraph_id": {
        "required":
            True,
    },

    "media_ai_record_id": {
        "required":
            True,
    },

    "original_text_start": {
        "required":
            True,
    },

    "original_text_end": {
        "required":
            True,
    },

    "special_token": {
        "required":
            True,
    },

    "attention_mask_value": {
        "required":
            False,
    },

    "token_type_id": {
        "required":
            False,
    },

    "tokeniser_identifier": {
        "required":
            True,
    },

    "tokeniser_version": {
        "required":
            True,
    },
}


# -----------------------------------------------------------------------------
# Offset mapping contract
# -----------------------------------------------------------------------------

TOKEN_OFFSET_MAPPING_CONTRACT = {
    "preferred":
        True,

    "interval_convention":
        "[start, end)",

    "start_inclusive":
        True,

    "end_exclusive":
        True,

    "special_tokens_have_text_span":
        False,

    "fabricated_offsets_allowed":
        False,

    "prepared_text_alignment_required":
        True,
}


# -----------------------------------------------------------------------------
# Attention-mask contract
# -----------------------------------------------------------------------------

ATTENTION_MASK_CONTRACT = {
    "mask_name":
        "attention_mask",

    "active_value":
        1,

    "inactive_value":
        0,

    "model_input_property":
        True,

    "representation_validity_mask":
        False,
}


# -----------------------------------------------------------------------------
# Token-type contract
# -----------------------------------------------------------------------------

TOKEN_TYPE_CONTRACT = {
    "required":
        False,

    "model_specific":
        True,

    "not_applicable_allowed":
        True,
}


# -----------------------------------------------------------------------------
# Special-token contract
# -----------------------------------------------------------------------------

SPECIAL_TOKEN_CONTRACT = {
    "must_be_identifiable":
        True,

    "must_not_receive_fabricated_text_span":
        True,

    "textual_evidence":
        False,

    "may_participate_in_model_input":
        True,
}


# -----------------------------------------------------------------------------
# Sequence-length and overflow contract
# -----------------------------------------------------------------------------

SEQUENCE_LENGTH_CONTRACT = {
    "silent_truncation_allowed":
        False,

    "maximum_length_model_specific":
        True,

    "overflow_must_be_detected":
        True,

    "hierarchical_processing_allowed":
        True,

    "chunking_allowed":
        True,

    "overlapping_windows_allowed":
        True,

    "strategy_must_be_explicit":
        True,

    "research_hierarchy_changed_by_chunking":
        False,
}


# -----------------------------------------------------------------------------
# Chunking contract
# -----------------------------------------------------------------------------

CHUNKING_CONTRACT = {
    "chunk_is_research_hierarchy_level":
        False,

    "required_fields": (
        "chunk_id",
        "media_ai_record_id",
        "sentence_ids",
        "chunk_index",
        "token_start",
        "token_end",
        "overlap_previous",
        "overlap_next",
        "tokeniser_identifier",
    ),

    "parent_identity_preserved":
        True,

    "ordering_preserved":
        True,

    "overlap_explicit":
        True,
}


# -----------------------------------------------------------------------------
# Temporal-context contract
# -----------------------------------------------------------------------------

MODEL_TEXT_TEMPORAL_CONTEXT_CONTRACT = {
    "preceding_context_processing":
        True,

    "future_context_for_earlier_state":
        False,

    "complete_article_tokenisation_allowed":
        True,

    "complete_article_availability_implies_future_context_permission":
        False,
}


# -----------------------------------------------------------------------------
# Sentence-supervision boundary
# -----------------------------------------------------------------------------

TOKEN_SUPERVISION_CONTRACT = {
    "sentence_is_supervision_anchor":
        True,

    "sentence_supervision_may_be_copied_to_tokens":
        False,

    "token_labels_directly_observed":
        False,

    "token_attribution_may_be_estimated_later":
        True,

    "missing_supervision_as_zero":
        False,
}


# -----------------------------------------------------------------------------
# Tokeniser provenance contract
# -----------------------------------------------------------------------------

TOKENISER_PROVENANCE_REQUIRED_FIELDS = (
    "tokeniser_identifier",
    "tokeniser_version",
    "tokenisation_configuration",
    "model_identifier",
    "model_version",
    "constructed_at_utc",
)


TOKENISER_PROVENANCE_CONTRACT = {
    "required_fields":
        TOKENISER_PROVENANCE_REQUIRED_FIELDS,

    "must_follow_token_records":
        True,

    "must_follow_model_derived_representations":
        True,

    "must_be_persisted_with_outputs":
        True,
}


# -----------------------------------------------------------------------------
# Backward token traceability contract
# -----------------------------------------------------------------------------

TOKEN_BACKWARD_TRACEABILITY_CONTRACT = {
    "token_to_sentence":
        True,

    "sentence_to_paragraph":
        True,

    "paragraph_to_article":
        True,

    "token_to_text_span_preferred":
        True,

    "token_to_prepared_text_required":
        True,

    "supports_later_colour_inspection":
        True,

    "supports_later_attribution":
        True,
}


# -----------------------------------------------------------------------------
# Model-ready text validation
# -----------------------------------------------------------------------------

MODEL_READY_TEXT_VALIDATION_ERRORS = []


if len(
    MODEL_READY_SENTENCE_RECORDS
) != len(
    SENTENCE_IDS
):

    MODEL_READY_TEXT_VALIDATION_ERRORS.append(
        "sentence_record_count_mismatch"
    )


if len(
    MODEL_READY_ARTICLE_RECORDS
) != len(
    ARTICLE_SENTENCE_ORDER
):

    MODEL_READY_TEXT_VALIDATION_ERRORS.append(
        "article_record_count_mismatch"
    )


for record in (
    MODEL_READY_SENTENCE_RECORDS
):

    sentence_id = record.get(
        "sentence_id"
    )

    paragraph_id = record.get(
        "paragraph_id"
    )

    article_id = record.get(
        "media_ai_record_id"
    )


    if (
        sentence_id
        not in SENTENCE_TO_PARAGRAPH_ID
    ):

        MODEL_READY_TEXT_VALIDATION_ERRORS.append(
            "unknown_sentence_id"
        )

        break


    if (
        SENTENCE_TO_PARAGRAPH_ID[
            sentence_id
        ]
        != paragraph_id
    ):

        MODEL_READY_TEXT_VALIDATION_ERRORS.append(
            "sentence_paragraph_mismatch"
        )

        break


    if (
        SENTENCE_TO_ARTICLE_ID[
            sentence_id
        ]
        != article_id
    ):

        MODEL_READY_TEXT_VALIDATION_ERRORS.append(
            "sentence_article_mismatch"
        )

        break


    if (
        record.get(
            "prepared_text"
        )
        != CANONICAL_SENTENCE_TEXT[
            sentence_id
        ]
    ):

        MODEL_READY_TEXT_VALIDATION_ERRORS.append(
            "prepared_text_mismatch"
        )

        break


    if (
        record.get(
            "model_text"
        )
        != record.get(
            "prepared_text"
        )
    ):

        MODEL_READY_TEXT_VALIDATION_ERRORS.append(
            "unexpected_text_transformation"
        )

        break


    if record.get(
        "text_transformed"
    ):

        MODEL_READY_TEXT_VALIDATION_ERRORS.append(
            "unexpected_text_transformation_flag"
        )

        break


# -----------------------------------------------------------------------------
# Article-order validation
# -----------------------------------------------------------------------------

ARTICLE_MODEL_TEXT_ORDER_VALID = True


for article_record in (
    MODEL_READY_ARTICLE_RECORDS
):

    article_id = article_record[
        "media_ai_record_id"
    ]

    if (
        article_record[
            "sentence_ids"
        ]
        != ARTICLE_SENTENCE_ORDER[
            article_id
        ]
    ):

        ARTICLE_MODEL_TEXT_ORDER_VALID = False

        MODEL_READY_TEXT_VALIDATION_ERRORS.append(
            "article_sentence_order_mismatch"
        )

        break


# -----------------------------------------------------------------------------
# Token contract compatibility with Block 3
# -----------------------------------------------------------------------------

BLOCK_3_TOKEN_FIELDS = set(
    TOKEN_MAPPING_REQUIRED_FIELDS
)


BLOCK_5_TOKEN_FIELDS = {
    field_name
    for (
        field_name,
        field_specification,
    ) in MODEL_TOKEN_RECORD_SCHEMA.items()
    if field_specification.get(
        "required"
    )
}


TOKEN_CONTRACT_FIELDS_COMPATIBLE = (
    BLOCK_3_TOKEN_FIELDS
    .issubset(
        BLOCK_5_TOKEN_FIELDS
    )
)


if not TOKEN_CONTRACT_FIELDS_COMPATIBLE:

    missing_required_fields = sorted(
        BLOCK_3_TOKEN_FIELDS
        - BLOCK_5_TOKEN_FIELDS
    )

    MODEL_READY_TEXT_VALIDATION_ERRORS.append(
        "token_contract_missing_required_fields:"
        f"{missing_required_fields}"
    )


# -----------------------------------------------------------------------------
# Attention-mask separation validation
# -----------------------------------------------------------------------------

if (
    ATTENTION_MASK_CONTRACT[
        "representation_validity_mask"
    ]
    is not False
):

    MODEL_READY_TEXT_VALIDATION_ERRORS.append(
        "attention_and_representation_masks_conflated"
    )


# -----------------------------------------------------------------------------
# Truncation safeguard validation
# -----------------------------------------------------------------------------

if (
    SEQUENCE_LENGTH_CONTRACT[
        "silent_truncation_allowed"
    ]
    is not False
):

    MODEL_READY_TEXT_VALIDATION_ERRORS.append(
        "silent_truncation_allowed"
    )


# -----------------------------------------------------------------------------
# Supervision safeguard validation
# -----------------------------------------------------------------------------

if (
    TOKEN_SUPERVISION_CONTRACT[
        "sentence_supervision_may_be_copied_to_tokens"
    ]
    is not False
):

    MODEL_READY_TEXT_VALIDATION_ERRORS.append(
        "sentence_labels_incorrectly_propagated_to_tokens"
    )


if (
    TOKEN_SUPERVISION_CONTRACT[
        "missing_supervision_as_zero"
    ]
    is not False
):

    MODEL_READY_TEXT_VALIDATION_ERRORS.append(
        "missing_supervision_zero_conflict"
    )


# -----------------------------------------------------------------------------
# Traceability safeguard validation
# -----------------------------------------------------------------------------

if not all([
    TOKEN_BACKWARD_TRACEABILITY_CONTRACT[
        "token_to_sentence"
    ],

    TOKEN_BACKWARD_TRACEABILITY_CONTRACT[
        "sentence_to_paragraph"
    ],

    TOKEN_BACKWARD_TRACEABILITY_CONTRACT[
        "paragraph_to_article"
    ],

    TOKEN_BACKWARD_TRACEABILITY_CONTRACT[
        "token_to_prepared_text_required"
    ],
]):

    MODEL_READY_TEXT_VALIDATION_ERRORS.append(
        "incomplete_token_traceability_contract"
    )


# -----------------------------------------------------------------------------
# Model-ready text contract state
# -----------------------------------------------------------------------------

MODEL_READY_TEXT_CONTRACT_VALID = (
    len(
        MODEL_READY_TEXT_VALIDATION_ERRORS
    )
    == 0
)


if not MODEL_READY_TEXT_CONTRACT_VALID:

    raise ValueError(
        "Model-ready text and token traceability "
        "contract failed validation: "
        f"{MODEL_READY_TEXT_VALIDATION_ERRORS}"
    )


# -----------------------------------------------------------------------------
# Optional tokenisation execution boundary
# -----------------------------------------------------------------------------

MODEL_TOKEN_RECORDS = []

MODEL_TOKENISATION_PERFORMED = False

TOKENISER_CONFIGURED = False

TOKENISER_IDENTIFIER = None
TOKENISER_VERSION = None

MODEL_IDENTIFIER = None
MODEL_VERSION = None

TOKENISATION_PROVENANCE = None


if BUILD_MODEL_READY_TEXT:

    # -------------------------------------------------------------------------
    # Model/tokeniser selection is intentionally deferred.
    #
    # Block 5 establishes the complete tokenisation and traceability contract.
    # A tokenizer must be selected explicitly before model-specific execution.
    # -------------------------------------------------------------------------

    print(
        "Model-ready text construction is enabled, but no model-specific "
        "tokenizer has yet been selected."
    )

    print(
        "Prepared sentence and article text records are available. "
        "Model tokenisation remains deferred."
    )


# -----------------------------------------------------------------------------
# Model-ready text architecture object
# -----------------------------------------------------------------------------

MODEL_READY_TEXT_SPECIFICATION = {
    "contract_version":
        MODEL_READY_TEXT_CONTRACT_VERSION,

    "token_traceability_contract_version":
        TOKEN_TRACEABILITY_CONTRACT_VERSION,

    "model_ready_text_unit":
        MODEL_READY_TEXT_UNIT,

    "supported_context_levels":
        MODEL_CONTEXT_LEVELS_SUPPORTED,

    "text_transformation_policy":
        deepcopy(
            MODEL_TEXT_TRANSFORMATION_POLICY
        ),

    "tokeniser_capability_requirements":
        deepcopy(
            TOKENISER_CAPABILITY_REQUIREMENTS
        ),

    "token_identity_contract":
        deepcopy(
            TOKEN_IDENTITY_CONTRACT
        ),

    "token_record_schema":
        deepcopy(
            MODEL_TOKEN_RECORD_SCHEMA
        ),

    "offset_mapping_contract":
        deepcopy(
            TOKEN_OFFSET_MAPPING_CONTRACT
        ),

    "attention_mask_contract":
        deepcopy(
            ATTENTION_MASK_CONTRACT
        ),

    "token_type_contract":
        deepcopy(
            TOKEN_TYPE_CONTRACT
        ),

    "special_token_contract":
        deepcopy(
            SPECIAL_TOKEN_CONTRACT
        ),

    "sequence_length_contract":
        deepcopy(
            SEQUENCE_LENGTH_CONTRACT
        ),

    "chunking_contract":
        deepcopy(
            CHUNKING_CONTRACT
        ),

    "temporal_context_contract":
        deepcopy(
            MODEL_TEXT_TEMPORAL_CONTEXT_CONTRACT
        ),

    "token_supervision_contract":
        deepcopy(
            TOKEN_SUPERVISION_CONTRACT
        ),

    "tokeniser_provenance_contract":
        deepcopy(
            TOKENISER_PROVENANCE_CONTRACT
        ),

    "backward_traceability_contract":
        deepcopy(
            TOKEN_BACKWARD_TRACEABILITY_CONTRACT
        ),
}


# -----------------------------------------------------------------------------
# Model-ready text summary
# -----------------------------------------------------------------------------

MODEL_READY_TEXT_SUMMARY = {
    "sentence_records":
        len(
            MODEL_READY_SENTENCE_RECORDS
        ),

    "article_records":
        len(
            MODEL_READY_ARTICLE_RECORDS
        ),

    "prepared_text_preserved":
        all(
            record[
                "prepared_text"
            ]
            == record[
                "model_text"
            ]

            for record
            in MODEL_READY_SENTENCE_RECORDS
        ),

    "article_order_valid":
        ARTICLE_MODEL_TEXT_ORDER_VALID,

    "token_contract_compatible":
        TOKEN_CONTRACT_FIELDS_COMPATIBLE,

    "model_ready_text_contract_valid":
        MODEL_READY_TEXT_CONTRACT_VALID,

    "tokeniser_configured":
        TOKENISER_CONFIGURED,

    "tokenisation_performed":
        MODEL_TOKENISATION_PERFORMED,

    "token_records":
        len(
            MODEL_TOKEN_RECORDS
        ),
}


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Model-Ready Text and Token Traceability")
print("=" * 72)

print(
    f"Contract version            : "
    f"{MODEL_READY_TEXT_CONTRACT_VERSION}"
)

print(
    f"Token contract version      : "
    f"{TOKEN_TRACEABILITY_CONTRACT_VERSION}"
)

print(
    f"Primary model-ready unit    : "
    f"{MODEL_READY_TEXT_UNIT}"
)

print(
    f"Supported context levels    : "
    f"{', '.join(MODEL_CONTEXT_LEVELS_SUPPORTED)}"
)

print("-" * 72)

print("Prepared model-facing text")

print(
    f"Sentence records            : "
    f"{len(MODEL_READY_SENTENCE_RECORDS)}"
)

print(
    f"Article records             : "
    f"{len(MODEL_READY_ARTICLE_RECORDS)}"
)

print(
    f"Prepared text preserved     : "
    f"{MODEL_READY_TEXT_SUMMARY['prepared_text_preserved']}"
)

print(
    f"Article ordering valid      : "
    f"{ARTICLE_MODEL_TEXT_ORDER_VALID}"
)

print("-" * 72)

print("Token traceability contract")

print(
    f"Token → sentence            : "
    f"{TOKEN_BACKWARD_TRACEABILITY_CONTRACT['token_to_sentence']}"
)

print(
    f"Sentence → paragraph        : "
    f"{TOKEN_BACKWARD_TRACEABILITY_CONTRACT['sentence_to_paragraph']}"
)

print(
    f"Paragraph → article         : "
    f"{TOKEN_BACKWARD_TRACEABILITY_CONTRACT['paragraph_to_article']}"
)

print(
    f"Offset mapping preferred    : "
    f"{TOKEN_OFFSET_MAPPING_CONTRACT['preferred']}"
)

print(
    f"Special tokens explicit     : "
    f"{SPECIAL_TOKEN_CONTRACT['must_be_identifiable']}"
)

print(
    f"Token contract compatible   : "
    f"{TOKEN_CONTRACT_FIELDS_COMPATIBLE}"
)

print("-" * 72)

print("Sequence safeguards")

print(
    f"Silent truncation allowed   : "
    f"{SEQUENCE_LENGTH_CONTRACT['silent_truncation_allowed']}"
)

print(
    f"Chunking allowed            : "
    f"{SEQUENCE_LENGTH_CONTRACT['chunking_allowed']}"
)

print(
    f"Chunk is hierarchy level    : "
    f"{CHUNKING_CONTRACT['chunk_is_research_hierarchy_level']}"
)

print(
    f"Future context permitted    : "
    f"{MODEL_TEXT_TEMPORAL_CONTEXT_CONTRACT['future_context_for_earlier_state']}"
)

print("-" * 72)

print("Supervision boundary")

print(
    f"Sentence supervision anchor : "
    f"{TOKEN_SUPERVISION_CONTRACT['sentence_is_supervision_anchor']}"
)

print(
    f"Copy sentence labels/token  : "
    f"{TOKEN_SUPERVISION_CONTRACT['sentence_supervision_may_be_copied_to_tokens']}"
)

print(
    f"Missing supervision as zero : "
    f"{TOKEN_SUPERVISION_CONTRACT['missing_supervision_as_zero']}"
)

print("-" * 72)

print("Model-specific execution")

print(
    f"Build model-ready text      : "
    f"{BUILD_MODEL_READY_TEXT}"
)

print(
    f"Tokeniser configured        : "
    f"{TOKENISER_CONFIGURED}"
)

print(
    f"Tokenisation performed      : "
    f"{MODEL_TOKENISATION_PERFORMED}"
)

print(
    f"Token records               : "
    f"{len(MODEL_TOKEN_RECORDS)}"
)

print("-" * 72)

print(
    f"Contract valid              : "
    f"{MODEL_READY_TEXT_CONTRACT_VALID}"
)

print("=" * 72)

print(
    "Model-ready text and token-traceability contract "
    "initialised successfully."
)

print(
    "Prepared sentence and article text remain traceable "
    "through stable research identifiers."
)

print(
    "Model tokenisation is explicitly separated from the "
    "prepared textual hierarchy."
)

print(
    "Chunking, if later required, will not create a new "
    "Media AI hierarchy level."
)

print(
    "Sentence-level supervision has not been propagated "
    "to model tokens."
)

print(
    "No model-specific tokenizer or embedding has been executed."
)

print(
    "No factual, psychological, social, transformative, recurrent "
    "or confluent representation has been constructed."
)

print(
    "No aggregation, attribution score, model training or "
    "experimental evaluation has been performed."
)
````

## Factual Representation

This block defines and, where validated supervision is available, constructs the factual representation used by the Media AI architecture.

The factual space represents the **informational structure and epistemic presentation contributed by a textual unit**.

It does **not** determine whether the represented information is true.

For textual unit \(u\), the conceptual factual representation is:

\[
F_u = \{w_u^F, C_u^F\}
\]

where:

* \(F_u\) = factual representation of textual unit \(u\);
* \(w_u^F\) = factological weight;
* \(C_u^F\) = factological-component vector.

### Factological Weight

Factological weight represents the amount of externally assessable informational content contributed by the textual unit.

It satisfies:

\[
w_u^F \in [0,1]
\]

Dictionary:

* \(w_u^F = 0\) = no meaningful factological contribution under the configured definition;
* \(w_u^F = 1\) = maximum factological contribution under the configured scale.

Factological weight is not:

* a truth probability;
* a fact-checking result;
* a confidence score;
* a measure of source reliability.

A sentence may therefore have high factological weight while the truth of its claims remains unknown.

### Factological Components

The factological-component vector is:

\[
\begin{aligned}
C_u^F = [&c_{\mathrm{claim}}, c_{\mathrm{event}}, c_{\mathrm{observation}}, c_{\mathrm{quant}}, c_{\mathrm{attrib}},\\
&c_{\mathrm{evidence}}, c_{\mathrm{certainty}}, c_{\mathrm{uncertainty}}, c_{\mathrm{speculation}}]
\end{aligned}
\]

Dictionary:

* \(c_{\mathrm{claim}}\) = explicit claim;
* \(c_{\mathrm{event}}\) = reported event or state;
* \(c_{\mathrm{observation}}\) = reported observation, measurement or perception;
* \(c_{\mathrm{quant}}\) = quantitative information;
* \(c_{\mathrm{attrib}}\) = attribution;
* \(c_{\mathrm{evidence}}\) = evidence reference;
* \(c_{\mathrm{certainty}}\) = certainty statement;
* \(c_{\mathrm{uncertainty}}\) = uncertainty statement;
* \(c_{\mathrm{speculation}}\) = speculation or prediction.

For an observed component:

\[
c_i \in \{0,1\}
\]

where \(0\) means that the component is observed as absent and \(1\) means that it is observed as present.

These components describe the **factological structure and epistemic presentation** of information. They do not independently establish whether the represented information is true.

Missing or unavailable component information must remain distinct from observed zero.

### Numerical Factual Representation

When the explicit factual dimensions are numerically available, they form the ten-dimensional vector:

\[
\begin{aligned}
V_u^F = [&w_u^F, c_{\mathrm{claim}}, c_{\mathrm{event}}, c_{\mathrm{observation}}, c_{\mathrm{quant}},\\
&c_{\mathrm{attrib}}, c_{\mathrm{evidence}}, c_{\mathrm{certainty}}, c_{\mathrm{uncertainty}}, c_{\mathrm{speculation}}]
\end{aligned}
\]

Therefore:

\[
V_u^F \in \mathbb{R}^{10}
\]

with:

\[
d_F = 1 + 9 = 10
\]

Dictionary:

* dimension 1 = factological weight;
* dimensions 2–10 = the nine factological components;
* \(d_F\) = number of explicit factual dimensions.

This fixed dimensionality applies only to the explicit factual ontology defined in the current representation schema.

Later learned factual features may exist separately but must not silently redefine these ten dimensions.

### Value-State Representation

Each factual dimension preserves both its value and its semantic state.

For factual dimension \(i\):

\[
X_{u,i}^F = (v_{u,i}^F, s_{u,i}^F)
\]

where:

* \(v_{u,i}^F\) = numerical factual value or unavailable value;
* \(s_{u,i}^F\) = semantic value state.

The permitted states are:

\[
\begin{aligned}
s_{u,i}^F \in \{&\text{observed}, \text{insufficient evidence}, \text{not applicable},\\
&\text{unavailable}, \text{derived}\}
\end{aligned}
\]

Dictionary:

* `observed` = a substantive numerical value is available;
* `insufficient_evidence` = the available evidence does not support assignment;
* `not_applicable` = the construct does not apply;
* `unavailable` = the required factual source has not yet been produced or loaded;
* `derived` = the value was produced through an explicitly defined transformation.

A missing factual value must not be encoded as substantive zero.

### Availability Mask

The factual value vector is accompanied by the availability mask:

\[
M_u^F = [m_{u,1}^F, m_{u,2}^F, \ldots, m_{u,10}^F]
\]

with:

\[
m_{u,i}^F \in \{0,1\}
\]

The mask is defined as:

\[
m_{u,i}^F =
\begin{cases}
1, & \text{if factual dimension } i \text{ contains an available substantive value},\\
0, & \text{otherwise}.
\end{cases}
\]

Dictionary:

* \(M_u^F\) = factual availability mask;
* \(m_{u,i}^F = 1\) = dimension \(i\) may be interpreted numerically;
* \(m_{u,i}^F = 0\) = dimension \(i\) is unavailable for numerical interpretation.

Thus:

\[
v_{u,i}^F = 0,\quad m_{u,i}^F = 1
\]

means **observed zero**, whereas:

\[
m_{u,i}^F = 0
\]

means **no available substantive value**.

Therefore:

\[
\text{missing} \neq 0
\]

### Operational Factual Representation

For computational use, the factual representation must contain more than the numerical vector alone.

We therefore define the operational factual representation as:

\[
\mathcal{F}_u = (V_u^F, M_u^F, Z_u^F, Q_u^F)
\]

where:

* \(V_u^F\) = ten-dimensional factual value vector;
* \(M_u^F\) = factual availability mask;
* \(Z_u^F\) = representation source;
* \(Q_u^F\) = provenance metadata.

The conceptual construct \(F_u\) describes **what factual information is represented**, while \(\mathcal{F}_u\) describes the **operational research object** used by the architecture.

### Representation Source

Every factual representation must identify how it was produced.

Let:

\[
\begin{aligned}
Z_u^F \in \{&\text{validated supervision}, \text{human reviewed}, \text{model predicted},\\
&\text{deterministic}, \text{aggregated}\}
\end{aligned}
\]

Dictionary:

* `validated_supervision` = constructed from validated Notebook 03 supervision;
* `human_reviewed` = established or confirmed through human review;
* `model_predicted` = generated by a later Media AI model;
* `deterministic` = produced through an explicitly defined deterministic rule;
* `aggregated` = constructed from lower-level factual representations.

Representations produced by different sources remain distinct research objects even when their numerical values are identical.

### Sentence-Level Factual Representation

The initial supervision-aligned factual unit is the sentence.

For sentence \(s\):

\[
\mathcal{F}_s = (V_s^F, M_s^F, Z_s^F, Q_s^F)
\]

Validated sentence-level supervision may populate \(V_s^F\) and \(M_s^F\) where available.

If validated supervision is unavailable:

\[
M_s^F = \mathbf{0}_{10}
\]

while the numerical values remain unavailable.

Importantly:

\[
M_s^F = \mathbf{0}_{10} \not\Rightarrow V_s^F = \mathbf{0}_{10}
\]

An unavailable factual representation is therefore not equivalent to a factual representation containing ten observed zeros.

### Partial Supervision

Notebook 04 must support partial factual supervision coverage.

For article \(a\), let its ordered sentence sequence be:

\[
S_a = (s_1, s_2, \ldots, s_n)
\]

Let:

\[
r_s^F \in \{0,1\}
\]

indicate whether sentence \(s\) has usable factual supervision.

Article factual coverage is then:

\[
\gamma_a^F = \frac{\sum_{s \in S_a} r_s^F}{|S_a|}
\]

Dictionary:

* \(\gamma_a^F\) = factual supervision coverage of article \(a\);
* \(r_s^F = 1\) = sentence \(s\) has usable factual supervision;
* \(r_s^F = 0\) = sentence \(s\) does not;
* \(|S_a|\) = number of sentences in article \(a\).

Thus:

\[
0 \leq \gamma_a^F \leq 1
\]

The interpretation is:

* \(\gamma_a^F = 0\) = no factual supervision;
* \(0 < \gamma_a^F < 1\) = partial factual supervision;
* \(\gamma_a^F = 1\) = complete factual supervision coverage.

Partial supervision must remain explicit.

A partially supervised article must not be converted into a fully observed factual sequence through implicit imputation.

### Hierarchical Factual Representation

The Media AI hierarchy permits factual representations at four levels:

\[
F^{\mathrm{token}} \rightarrow F^{\mathrm{sentence}} \rightarrow F^{\mathrm{paragraph}} \rightarrow F^{\mathrm{article}}
\]

The existence of these levels does not imply simple averaging.

For lower-level factual representations \(F_1^{\mathrm{lower}}, \ldots, F_n^{\mathrm{lower}}\), a higher-level representation is produced through an explicit aggregation function:

\[
F^{\mathrm{higher}} = A_F(F_1^{\mathrm{lower}}, F_2^{\mathrm{lower}}, \ldots, F_n^{\mathrm{lower}})
\]

Dictionary:

* \(A_F\) = explicitly defined factual aggregation function;
* \(F_i^{\mathrm{lower}}\) = contributing lower-level factual representation;
* \(F^{\mathrm{higher}}\) = resulting higher-level factual representation.

In general:

\[
A_F(F_1,\ldots,F_n) \neq \frac{1}{n}\sum_{i=1}^{n}F_i
\]

unless simple averaging is explicitly selected and justified for a particular factual dimension.

The aggregation methodology is defined later and is not executed by this block.

### Token-Level Boundary

Sentence-level factual supervision must not be copied directly to tokens.

For token \(\tau_i\) belonging to sentence \(s\):

\[
\tau_i \rightarrow s
\]

but:

\[
F_s^{\mathrm{supervised}} \not\Rightarrow F_{\tau_i}^{\mathrm{supervised}}
\]

Dictionary:

* \(F_s^{\mathrm{supervised}}\) = supervised sentence-level factual representation;
* \(F_{\tau_i}^{\mathrm{supervised}}\) = hypothetical supervised token-level factual representation.

Token-level factual contributions may later be estimated through attribution or another explicit modelling mechanism.

Such estimates must remain distinguishable from directly observed sentence-level supervision.

### Confidence

Supervision confidence remains metadata rather than a substantive factual dimension.

For sentence \(s\):

\[
\kappa_s^F \in [0,1]
\]

where \(\kappa_s^F\) denotes confidence associated with the factual supervision record.

Confidence is not included in the ten-dimensional factual vector:

\[
\kappa_s^F \notin V_s^F
\]

It must also not automatically modify factual values:

\[
V_s^F \not\leftarrow \kappa_s^F V_s^F
\]

Any later confidence-weighting procedure must be defined explicitly.

### Provenance

Each factual representation must preserve sufficient provenance to identify its origin and construction.

Conceptually:

\[
Q_u^F = (ID_u, L_u, Z_u^F, v_{\mathrm{schema}}, m_u, q_u)
\]

Dictionary:

* \(ID_u\) = stable textual identity;
* \(L_u\) = representation level;
* \(Z_u^F\) = factual representation source;
* \(v_{\mathrm{schema}}\) = representation-schema version;
* \(m_u\) = construction method;
* \(q_u\) = source-specific provenance, such as supervisor or model identity.

Provenance must remain associated with the factual representation through later transformation, recurrence, aggregation, attribution and evaluation.

### Truth Boundary

The factual representation is methodologically separate from truth verification.

Let \(\mathrm{Truth}(x)\) denote the externally established truth status of statement \(x\).

Then:

\[
F_u \not\equiv \mathrm{Truth}(x)
\]

and factological weight is not a probability of truth:

\[
w_u^F \neq P(\mathrm{Truth}(x)=\mathrm{True})
\]

A high value of \(w_u^F\) therefore means that the textual unit contains substantial externally assessable informational content, not that its claims are probably true.

Likewise, certainty and uncertainty describe how information is presented or qualified in the text; they do not independently establish whether that information is factually correct.

This distinction is fundamental to the Media AI factual space.

### Attribution Requirement

Higher-level factual states must remain traceable to their contributing textual units.

The required backward path is:

\[
F^{\mathrm{article}} \rightarrow F^{\mathrm{paragraph}} \rightarrow F^{\mathrm{sentence}} \rightarrow \mathrm{Text}
\]

Where later token attribution is available, the path may be extended to:

\[
F^{\mathrm{sentence}} \rightarrow \mathrm{Token} \rightarrow \mathrm{Text}
\]

For a higher-level factual output \(F_h\) and contributing lower-level unit \(l\), later attribution may be represented as:

\[
A_{h \leftarrow l}^F
\]

where:

* \(A_{h \leftarrow l}^F\) = contribution of lower-level textual unit \(l\) to higher-level factual representation \(h\);
* \(h\) = higher-level factual representation;
* \(l\) = contributing lower-level textual unit.

Block 6 preserves the identities and provenance required for this pathway.

It does not calculate attribution scores.

### Current Supervision State

The factual representation architecture must remain valid even when validated Notebook 03 supervision has not yet been persisted or loaded.

If factual supervision is unavailable:

\[
M_s^F = \mathbf{0}_{10}
\]

for the affected sentence, while its factual values remain unavailable.

The representation record can nevertheless preserve:

\[
(ID_s, V_s^F, M_s^F, Z_s^F, Q_s^F)
\]

without fabricating substantive values.

Once validated supervision becomes available, the same representation contract can populate the factual values without changing the architecture.

### Output

This block defines and, where data are available, constructs sentence-level factual representation records.

Its outputs include:

* the ten-dimensional factual value contract;
* factological weight;
* nine factological components;
* separate reported event/state and observation dimensions;
* factual availability masks;
* semantic value states;
* factual representation-source metadata;
* sentence-level factual representation records;
* factual supervision-coverage diagnostics;
* factual provenance;
* hierarchical factual-representation boundaries;
* token-level supervision boundaries;
* truth-verification boundaries;
* backward-attribution requirements.

Where validated supervision is unavailable, factual values remain explicitly unavailable rather than being replaced with zero.

No external fact checking is performed.

No token-level factual supervision is fabricated.

No paragraph-level or article-level factual aggregation is performed.

No psychological or social representation is constructed.

No transformative, recurrent or confluent state is calculated.

No attribution score, model training or experimental evaluation is performed.

````python
# =============================================================================
# Media AI — Factual Representation
# =============================================================================

from datetime import datetime, timezone

import numpy as np


# -----------------------------------------------------------------------------
# Block dependency checks
# -----------------------------------------------------------------------------

required_block_1_objects = [
    "BUILD_FACTUAL_REPRESENTATION",
    "PRESERVE_SUPERVISION_VALUE_STATES",
    "ALLOW_MISSING_SUPERVISION_AS_ZERO",
    "PRESERVE_SUPERVISION_PROVENANCE",
]

missing_block_1_objects = [
    object_name
    for object_name in required_block_1_objects
    if object_name not in globals()
]

if missing_block_1_objects:
    raise NameError(
        "Notebook 04 Block 1 is not initialised. "
        "Missing: "
        f"{missing_block_1_objects}"
    )


required_block_2_objects = [
    "SENTENCE_RECORDS",
    "SENTENCE_IDS",
    "CANONICAL_SUPERVISION_RECORDS",
    "CANONICAL_SUPERVISION_BY_SENTENCE_ID",
    "CANONICAL_SUPERVISION_AVAILABLE",
    "ARTICLE_SUPERVISION_COVERAGE",
]

missing_block_2_objects = [
    object_name
    for object_name in required_block_2_objects
    if object_name not in globals()
]

if missing_block_2_objects:
    raise NameError(
        "Notebook 04 Block 2 is not initialised. "
        "Missing: "
        f"{missing_block_2_objects}"
    )


required_block_3_objects = [
    "HIERARCHICAL_TEXT_AND_IDENTITY_VALID",
    "SENTENCE_BY_ID",
    "SENTENCE_TO_PARAGRAPH_ID",
    "SENTENCE_TO_ARTICLE_ID",
    "ARTICLE_SENTENCE_ORDER",
]

missing_block_3_objects = [
    object_name
    for object_name in required_block_3_objects
    if object_name not in globals()
]

if missing_block_3_objects:
    raise NameError(
        "Notebook 04 Block 3 is not initialised. "
        "Missing: "
        f"{missing_block_3_objects}"
    )


required_block_4_objects = [
    "MEDIA_AI_REPRESENTATION_SPECIFICATION_VALID",
    "REPRESENTATION_SCHEMA_VERSION",
    "FACTUAL_DIMENSIONS",
    "FACTUAL_DIMENSION_COUNT",
    "FACTUAL_DIMENSION_INDEX",
    "FACTUAL_DIMENSION_SPECIFICATION",
    "MEDIA_AI_VALUE_STATES",
    "MEDIA_AI_REPRESENTATION_SOURCES",
]

missing_block_4_objects = [
    object_name
    for object_name in required_block_4_objects
    if object_name not in globals()
]

if missing_block_4_objects:
    raise NameError(
        "Notebook 04 Block 4 is not initialised. "
        "Missing: "
        f"{missing_block_4_objects}"
    )


if not HIERARCHICAL_TEXT_AND_IDENTITY_VALID:
    raise ValueError(
        "Factual representation requires a valid "
        "hierarchical text and identity structure."
    )


if not MEDIA_AI_REPRESENTATION_SPECIFICATION_VALID:
    raise ValueError(
        "Factual representation requires a valid "
        "Media AI representation specification."
    )


if ALLOW_MISSING_SUPERVISION_AS_ZERO:
    raise ValueError(
        "Missing supervision must not be converted to numeric zero."
    )


# -----------------------------------------------------------------------------
# Factual representation configuration
# -----------------------------------------------------------------------------

FACTUAL_REPRESENTATION_VERSION = "1.0"

FACTUAL_REPRESENTATION_LEVEL = "sentence"

FACTUAL_REPRESENTATION_SPACE = "factual"

FACTUAL_REPRESENTATION_SOURCE_VALIDATED = (
    "validated_supervision"
)

FACTUAL_REPRESENTATION_SOURCE_UNAVAILABLE = (
    "unavailable"
)


# -----------------------------------------------------------------------------
# Factual dimension-role validation
# -----------------------------------------------------------------------------

EXPECTED_FACTUAL_DIMENSIONS_BLOCK_6 = (
    "factological_weight",
    "explicit_claim",
    "reported_event_state",
    "observation",
    "quantitative_information",
    "attribution",
    "evidence_reference",
    "certainty_statement",
    "uncertainty_statement",
    "speculation_or_prediction",
)


if FACTUAL_DIMENSION_COUNT != 10:
    raise ValueError(
        "Factual representation must contain exactly ten "
        "explicit dimensions."
    )


if tuple(FACTUAL_DIMENSIONS) != EXPECTED_FACTUAL_DIMENSIONS_BLOCK_6:
    raise ValueError(
        "Block 6 factual dimensions do not match the "
        "Block 4 factual representation contract. "
        f"Expected: {EXPECTED_FACTUAL_DIMENSIONS_BLOCK_6}. "
        f"Received: {tuple(FACTUAL_DIMENSIONS)}."
    )


if FACTUAL_DIMENSIONS[0] != "factological_weight":
    raise ValueError(
        "The first factual dimension must be factological_weight."
    )


FACTUAL_COMPONENT_DIMENSIONS_BLOCK_6 = (
    FACTUAL_DIMENSIONS[1:]
)


if len(FACTUAL_COMPONENT_DIMENSIONS_BLOCK_6) != 9:
    raise ValueError(
        "Factual representation must contain exactly nine "
        "factological components."
    )


# -----------------------------------------------------------------------------
# Value-state helpers
# -----------------------------------------------------------------------------

def factual_unavailable_value_state():
    """
    Return the semantic value state used when validated factual
    supervision is unavailable.
    """

    return "unavailable"


# -----------------------------------------------------------------------------
# Factual vector initialisation
# -----------------------------------------------------------------------------

def create_empty_factual_vector():
    """
    Create one unavailable ten-dimensional factual value vector.
    """

    return np.full(
        FACTUAL_DIMENSION_COUNT,
        np.nan,
        dtype=float,
    )


def create_empty_factual_mask():
    """
    Create one unavailable ten-dimensional factual mask.
    """

    return np.zeros(
        FACTUAL_DIMENSION_COUNT,
        dtype=np.int8,
    )


def create_empty_factual_value_states():
    """
    Create semantic value states for an unavailable factual record.
    """

    return {
        dimension:
            factual_unavailable_value_state()

        for dimension
        in FACTUAL_DIMENSIONS
    }


# -----------------------------------------------------------------------------
# Validated supervision extraction
# -----------------------------------------------------------------------------

def extract_factual_supervision(
    supervision_record,
):
    """
    Extract factual values, masks and semantic states from one validated
    sentence-level supervision record.

    The function never converts missing or unavailable values to zero.
    """

    if not isinstance(
        supervision_record,
        dict,
    ):
        raise TypeError(
            "supervision_record must be a dictionary."
        )


    annotations = supervision_record.get(
        "annotations",
        {}
    )


    factual_values = (
        create_empty_factual_vector()
    )

    factual_mask = (
        create_empty_factual_mask()
    )

    factual_states = (
        create_empty_factual_value_states()
    )


    # -------------------------------------------------------------------------
    # Factological weight
    # -------------------------------------------------------------------------

    weight_record = annotations.get(
        "factological_weight",
        {}
    )

    weight_value = weight_record.get(
        "value"
    )

    weight_state = weight_record.get(
        "value_state"
    )


    if weight_state not in MEDIA_AI_VALUE_STATES:
        raise ValueError(
            "Invalid factological-weight value state: "
            f"{weight_state}"
        )


    weight_index = FACTUAL_DIMENSION_INDEX[
        "factological_weight"
    ]


    factual_states[
        "factological_weight"
    ] = weight_state


    if weight_state == "observed":

        if not isinstance(
            weight_value,
            (int, float),
        ):
            raise ValueError(
                "Observed factological weight must be numeric."
            )


        if not (
            0.0
            <= float(weight_value)
            <= 1.0
        ):
            raise ValueError(
                "Observed factological weight must lie in [0,1]."
            )


        factual_values[
            weight_index
        ] = float(
            weight_value
        )

        factual_mask[
            weight_index
        ] = 1


    else:

        if weight_value is not None:
            raise ValueError(
                "Non-observed factological weight must use null value."
            )


    # -------------------------------------------------------------------------
    # Factological components
    # -------------------------------------------------------------------------

    components = annotations.get(
        "factological_components",
        {}
    )


    for component in (
        FACTUAL_COMPONENT_DIMENSIONS_BLOCK_6
    ):

        component_value = components.get(
            component
        )

        component_index = (
            FACTUAL_DIMENSION_INDEX[
                component
            ]
        )


        if component_value is None:

            factual_states[
                component
            ] = "unavailable"

            factual_values[
                component_index
            ] = np.nan

            factual_mask[
                component_index
            ] = 0

            continue


        if not isinstance(
            component_value,
            bool,
        ):
            raise ValueError(
                "Factological components must be boolean or null. "
                f"Invalid component: {component}"
            )


        factual_states[
            component
        ] = "observed"

        factual_values[
            component_index
        ] = float(
            int(
                component_value
            )
        )

        factual_mask[
            component_index
        ] = 1


    return (
        factual_values,
        factual_mask,
        factual_states,
    )


# -----------------------------------------------------------------------------
# Provenance extraction
# -----------------------------------------------------------------------------

def build_factual_provenance(
    sentence_id,
    representation_source,
    supervision_record=None,
):
    """
    Build factual-representation provenance for one sentence.
    """

    provenance = {
        "media_ai_record_id":
            SENTENCE_TO_ARTICLE_ID[
                sentence_id
            ],

        "paragraph_id":
            SENTENCE_TO_PARAGRAPH_ID[
                sentence_id
            ],

        "sentence_id":
            sentence_id,

        "representation_level":
            FACTUAL_REPRESENTATION_LEVEL,

        "representation_space":
            FACTUAL_REPRESENTATION_SPACE,

        "representation_source":
            representation_source,

        "representation_schema_version":
            REPRESENTATION_SCHEMA_VERSION,

        "factual_representation_version":
            FACTUAL_REPRESENTATION_VERSION,

        "construction_method":
            (
                "validated_supervision_mapping"
                if supervision_record is not None
                else "unavailable_supervision_placeholder"
            ),

        "supervisor_identifier":
            None,

        "supervision_method":
            None,

        "provider":
            None,

        "model":
            None,

        "supervision_schema_version":
            None,

        "annotation_policy_version":
            None,

        "constructed_at_utc":
            datetime.now(
                timezone.utc
            ).isoformat(),
    }


    if supervision_record is not None:

        supervision_provenance = (
            supervision_record.get(
                "provenance",
                {}
            )
        )

        provenance[
            "supervisor_identifier"
        ] = supervision_provenance.get(
            "supervisor_identifier"
        )

        provenance[
            "supervision_method"
        ] = supervision_provenance.get(
            "supervision_method"
        )

        provenance[
            "provider"
        ] = supervision_provenance.get(
            "provider"
        )

        provenance[
            "model"
        ] = supervision_provenance.get(
            "model"
        )

        provenance[
            "supervision_schema_version"
        ] = supervision_provenance.get(
            "supervision_schema_version"
        )

        provenance[
            "annotation_policy_version"
        ] = supervision_provenance.get(
            "annotation_policy_version"
        )


    return provenance


# -----------------------------------------------------------------------------
# Sentence-level factual representation construction
# -----------------------------------------------------------------------------

FACTUAL_SENTENCE_REPRESENTATIONS = []


for sentence_id in SENTENCE_IDS:

    supervision_record = (
        CANONICAL_SUPERVISION_BY_SENTENCE_ID.get(
            sentence_id
        )
    )


    if supervision_record is not None:

        (
            factual_values,
            factual_mask,
            factual_states,
        ) = extract_factual_supervision(
            supervision_record
        )

        representation_source = (
            FACTUAL_REPRESENTATION_SOURCE_VALIDATED
        )


        confidence_record = (
            supervision_record.get(
                "annotations",
                {}
            ).get(
                "annotation_confidence",
                {}
            )
        )

        supervision_confidence = (
            confidence_record.get(
                "value"
            )
        )

        supervision_confidence_state = (
            confidence_record.get(
                "value_state"
            )
        )


    else:

        factual_values = (
            create_empty_factual_vector()
        )

        factual_mask = (
            create_empty_factual_mask()
        )

        factual_states = (
            create_empty_factual_value_states()
        )

        representation_source = (
            FACTUAL_REPRESENTATION_SOURCE_UNAVAILABLE
        )

        supervision_confidence = None

        supervision_confidence_state = (
            "unavailable"
        )


    factual_record = {
        "record_type":
            "media_ai_factual_representation",

        "representation_version":
            FACTUAL_REPRESENTATION_VERSION,

        "media_ai_record_id":
            SENTENCE_TO_ARTICLE_ID[
                sentence_id
            ],

        "paragraph_id":
            SENTENCE_TO_PARAGRAPH_ID[
                sentence_id
            ],

        "sentence_id":
            sentence_id,

        "representation_level":
            FACTUAL_REPRESENTATION_LEVEL,

        "representation_space":
            FACTUAL_REPRESENTATION_SPACE,

        "representation_source":
            representation_source,

        "values":
            factual_values,

        "availability_mask":
            factual_mask,

        "value_states":
            factual_states,

        "supervision_confidence": {
            "value":
                supervision_confidence,

            "value_state":
                supervision_confidence_state,
        },

        "provenance":
            build_factual_provenance(
                sentence_id=
                    sentence_id,

                representation_source=
                    representation_source,

                supervision_record=
                    supervision_record,
            ),
    }


    FACTUAL_SENTENCE_REPRESENTATIONS.append(
        factual_record
    )


# -----------------------------------------------------------------------------
# Sentence-level factual lookup
# -----------------------------------------------------------------------------

FACTUAL_SENTENCE_BY_ID = {
    record[
        "sentence_id"
    ]:
        record

    for record
    in FACTUAL_SENTENCE_REPRESENTATIONS
}


# -----------------------------------------------------------------------------
# Factual representation validation
# -----------------------------------------------------------------------------

FACTUAL_REPRESENTATION_ERRORS = []


for record in (
    FACTUAL_SENTENCE_REPRESENTATIONS
):

    values = record.get(
        "values"
    )

    mask = record.get(
        "availability_mask"
    )

    states = record.get(
        "value_states",
        {}
    )


    if not isinstance(
        values,
        np.ndarray,
    ):

        FACTUAL_REPRESENTATION_ERRORS.append(
            "values_not_numpy_array"
        )

        break


    if values.shape != (
        FACTUAL_DIMENSION_COUNT,
    ):

        FACTUAL_REPRESENTATION_ERRORS.append(
            "invalid_factual_value_shape"
        )

        break


    if not isinstance(
        mask,
        np.ndarray,
    ):

        FACTUAL_REPRESENTATION_ERRORS.append(
            "mask_not_numpy_array"
        )

        break


    if mask.shape != (
        FACTUAL_DIMENSION_COUNT,
    ):

        FACTUAL_REPRESENTATION_ERRORS.append(
            "invalid_factual_mask_shape"
        )

        break


    if not np.all(
        np.isin(
            mask,
            [
                0,
                1,
            ],
        )
    ):

        FACTUAL_REPRESENTATION_ERRORS.append(
            "invalid_factual_mask_values"
        )

        break


    for dimension in FACTUAL_DIMENSIONS:

        if dimension not in states:

            FACTUAL_REPRESENTATION_ERRORS.append(
                "missing_factual_value_state"
            )

            break


        state = states[
            dimension
        ]


        if state not in MEDIA_AI_VALUE_STATES:

            FACTUAL_REPRESENTATION_ERRORS.append(
                "invalid_factual_value_state"
            )

            break


        dimension_index = (
            FACTUAL_DIMENSION_INDEX[
                dimension
            ]
        )


        if state == "observed":

            if mask[
                dimension_index
            ] != 1:

                FACTUAL_REPRESENTATION_ERRORS.append(
                    "observed_value_without_mask"
                )

                break


            if np.isnan(
                values[
                    dimension_index
                ]
            ):

                FACTUAL_REPRESENTATION_ERRORS.append(
                    "observed_value_is_nan"
                )

                break


        else:

            if mask[
                dimension_index
            ] != 0:

                FACTUAL_REPRESENTATION_ERRORS.append(
                    "unavailable_value_with_active_mask"
                )

                break


            if not np.isnan(
                values[
                    dimension_index
                ]
            ):

                FACTUAL_REPRESENTATION_ERRORS.append(
                    "unavailable_value_not_nan"
                )

                break


    if FACTUAL_REPRESENTATION_ERRORS:
        break


# -----------------------------------------------------------------------------
# Source-specific validation
# -----------------------------------------------------------------------------

for record in (
    FACTUAL_SENTENCE_REPRESENTATIONS
):

    source = record.get(
        "representation_source"
    )


    if source == (
        FACTUAL_REPRESENTATION_SOURCE_UNAVAILABLE
    ):

        if np.any(
            record[
                "availability_mask"
            ]
        ):

            FACTUAL_REPRESENTATION_ERRORS.append(
                "unavailable_source_has_observed_values"
            )

            break


    elif source == (
        FACTUAL_REPRESENTATION_SOURCE_VALIDATED
    ):

        if record[
            "sentence_id"
        ] not in (
            CANONICAL_SUPERVISION_BY_SENTENCE_ID
        ):

            FACTUAL_REPRESENTATION_ERRORS.append(
                "validated_source_without_supervision"
            )

            break


    else:

        FACTUAL_REPRESENTATION_ERRORS.append(
            "unexpected_factual_representation_source"
        )

        break


# -----------------------------------------------------------------------------
# Supervision coverage diagnostics
# -----------------------------------------------------------------------------

FACTUAL_SUPERVISED_SENTENCE_COUNT = sum(
    1
    for record
    in FACTUAL_SENTENCE_REPRESENTATIONS
    if record[
        "representation_source"
    ]
    == FACTUAL_REPRESENTATION_SOURCE_VALIDATED
)


FACTUAL_UNAVAILABLE_SENTENCE_COUNT = (
    len(
        FACTUAL_SENTENCE_REPRESENTATIONS
    )
    - FACTUAL_SUPERVISED_SENTENCE_COUNT
)


FACTUAL_SENTENCE_COVERAGE = (
    FACTUAL_SUPERVISED_SENTENCE_COUNT
    / len(
        FACTUAL_SENTENCE_REPRESENTATIONS
    )
    if FACTUAL_SENTENCE_REPRESENTATIONS
    else 0.0
)


# -----------------------------------------------------------------------------
# Article-level factual supervision coverage
# -----------------------------------------------------------------------------

FACTUAL_ARTICLE_COVERAGE = []


for (
    article_id,
    sentence_ids,
) in ARTICLE_SENTENCE_ORDER.items():

    supervised_count = sum(
        1
        for sentence_id
        in sentence_ids
        if FACTUAL_SENTENCE_BY_ID[
            sentence_id
        ][
            "representation_source"
        ]
        == FACTUAL_REPRESENTATION_SOURCE_VALIDATED
    )


    sentence_count = len(
        sentence_ids
    )


    coverage_fraction = (
        supervised_count
        / sentence_count
        if sentence_count
        else 0.0
    )


    if supervised_count == 0:

        coverage_status = (
            "not_supervised"
        )

    elif (
        supervised_count
        == sentence_count
    ):

        coverage_status = (
            "fully_supervised"
        )

    else:

        coverage_status = (
            "partially_supervised"
        )


    FACTUAL_ARTICLE_COVERAGE.append({
        "media_ai_record_id":
            article_id,

        "sentence_count":
            sentence_count,

        "supervised_sentence_count":
            supervised_count,

        "coverage_fraction":
            coverage_fraction,

        "coverage_status":
            coverage_status,
    })


# -----------------------------------------------------------------------------
# Factual availability accounting
# -----------------------------------------------------------------------------

FACTUAL_AVAILABLE_VALUE_COUNT = sum(
    int(
        record[
            "availability_mask"
        ].sum()
    )
    for record
    in FACTUAL_SENTENCE_REPRESENTATIONS
)


FACTUAL_TOTAL_VALUE_CAPACITY = (
    len(
        FACTUAL_SENTENCE_REPRESENTATIONS
    )
    * FACTUAL_DIMENSION_COUNT
)


FACTUAL_VALUE_AVAILABILITY_RATE = (
    FACTUAL_AVAILABLE_VALUE_COUNT
    / FACTUAL_TOTAL_VALUE_CAPACITY
    if FACTUAL_TOTAL_VALUE_CAPACITY
    else 0.0
)


# -----------------------------------------------------------------------------
# Representation status
# -----------------------------------------------------------------------------

FACTUAL_REPRESENTATION_VALID = (
    len(
        FACTUAL_REPRESENTATION_ERRORS
    )
    == 0
)


if not FACTUAL_REPRESENTATION_VALID:

    raise ValueError(
        "Factual representation failed validation: "
        f"{FACTUAL_REPRESENTATION_ERRORS}"
    )


FACTUAL_REPRESENTATIONS_CONSTRUCTED = (
    True
)


FACTUAL_REPRESENTATION_HAS_OBSERVED_VALUES = (
    FACTUAL_AVAILABLE_VALUE_COUNT
    > 0
)


# -----------------------------------------------------------------------------
# Factual representation summary
# -----------------------------------------------------------------------------

FACTUAL_REPRESENTATION_SUMMARY = {
    "representation_version":
        FACTUAL_REPRESENTATION_VERSION,

    "representation_level":
        FACTUAL_REPRESENTATION_LEVEL,

    "dimension_count":
        FACTUAL_DIMENSION_COUNT,

    "component_count":
        len(
            FACTUAL_COMPONENT_DIMENSIONS_BLOCK_6
        ),

    "sentence_records":
        len(
            FACTUAL_SENTENCE_REPRESENTATIONS
        ),

    "supervised_sentence_records":
        FACTUAL_SUPERVISED_SENTENCE_COUNT,

    "unavailable_sentence_records":
        FACTUAL_UNAVAILABLE_SENTENCE_COUNT,

    "sentence_supervision_coverage":
        FACTUAL_SENTENCE_COVERAGE,

    "available_values":
        FACTUAL_AVAILABLE_VALUE_COUNT,

    "total_value_capacity":
        FACTUAL_TOTAL_VALUE_CAPACITY,

    "value_availability_rate":
        FACTUAL_VALUE_AVAILABILITY_RATE,

    "representations_constructed":
        FACTUAL_REPRESENTATIONS_CONSTRUCTED,

    "observed_values_present":
        FACTUAL_REPRESENTATION_HAS_OBSERVED_VALUES,

    "representation_valid":
        FACTUAL_REPRESENTATION_VALID,
}


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Factual Representation")
print("=" * 72)

print(
    f"Representation version      : "
    f"{FACTUAL_REPRESENTATION_VERSION}"
)

print(
    f"Representation level        : "
    f"{FACTUAL_REPRESENTATION_LEVEL}"
)

print(
    f"Factual dimensions          : "
    f"{FACTUAL_DIMENSION_COUNT}"
)

print(
    f"Factological components     : "
    f"{len(FACTUAL_COMPONENT_DIMENSIONS_BLOCK_6)}"
)

print(
    f"Sentence records            : "
    f"{len(FACTUAL_SENTENCE_REPRESENTATIONS)}"
)

print("-" * 72)

print("Supervision coverage")

print(
    f"Validated supervision avail.: "
    f"{CANONICAL_SUPERVISION_AVAILABLE}"
)

print(
    f"Supervised sentences        : "
    f"{FACTUAL_SUPERVISED_SENTENCE_COUNT}"
)

print(
    f"Unavailable sentences       : "
    f"{FACTUAL_UNAVAILABLE_SENTENCE_COUNT}"
)

print(
    f"Sentence coverage           : "
    f"{FACTUAL_SENTENCE_COVERAGE:.4f}"
)

print("-" * 72)

print("Value availability")

print(
    f"Available factual values    : "
    f"{FACTUAL_AVAILABLE_VALUE_COUNT}"
)

print(
    f"Total factual capacity      : "
    f"{FACTUAL_TOTAL_VALUE_CAPACITY}"
)

print(
    f"Value availability rate     : "
    f"{FACTUAL_VALUE_AVAILABILITY_RATE:.4f}"
)

print(
    f"Observed values present     : "
    f"{FACTUAL_REPRESENTATION_HAS_OBSERVED_VALUES}"
)

print("-" * 72)

print("Article coverage")

for article_record in (
    FACTUAL_ARTICLE_COVERAGE
):

    print(
        f"{article_record['media_ai_record_id']}"
    )

    print(
        f"  Sentences                 : "
        f"{article_record['sentence_count']}"
    )

    print(
        f"  Supervised                : "
        f"{article_record['supervised_sentence_count']}"
    )

    print(
        f"  Coverage                  : "
        f"{article_record['coverage_fraction']:.4f}"
    )

    print(
        f"  Status                    : "
        f"{article_record['coverage_status']}"
    )

print("-" * 72)

print("Methodological boundaries")

print(
    "Factual structure = truth  : False"
)

print(
    "Factological weight = truth: False"
)

print(
    "Missing equals zero        : False"
)

print(
    "Confidence part of vector  : False"
)

print("-" * 72)

print(
    f"Representation valid        : "
    f"{FACTUAL_REPRESENTATION_VALID}"
)

print(
    f"Representations constructed : "
    f"{FACTUAL_REPRESENTATIONS_CONSTRUCTED}"
)

print("=" * 72)

print(
    "Sentence-level factual representation records constructed successfully."
)

if FACTUAL_REPRESENTATION_HAS_OBSERVED_VALUES:

    print(
        "Available validated supervision has populated factual values."
    )

else:

    print(
        "Validated factual supervision is currently unavailable."
    )

    print(
        "Factual values therefore remain unavailable with inactive masks."
    )

print(
    "The factual space contains one factological-weight dimension "
    "and nine explicit factological components."
)

print(
    "Reported event/state and observation remain separate factual dimensions."
)

print(
    "Factual dimensions describe factological structure and epistemic "
    "presentation, not externally verified truth."
)

print(
    "Missing factual values have not been converted to numerical zero."
)

print(
    "Factological weight remains separate from external truth verification."
)

print(
    "Sentence-level factual supervision has not been propagated to tokens."
)

print(
    "No paragraph-level or article-level factual aggregation has been performed."
)

print(
    "No psychological, social, transformative, recurrent or confluent "
    "representation has been constructed by this block."
)

print(
    "No attribution score, model training or experimental evaluation "
    "has been performed."
)
````

## Psychological Representation

This block defines and, where validated supervision is available, constructs the psychological representation used by the Media AI architecture.

The psychological space represents emotional state, emotional change and emotional relevance associated with a textual unit.

It preserves three distinct emotional constructs:

(E_u^{reported}), (\Delta E_u), and (E_u^{immediate}),

together with emotional weight (w_u^{P}).

For textual unit (u), the conceptual psychological representation is:

(P_u = {E_u^{reported}, \Delta E_u, E_u^{immediate}, w_u^{P}}),

where:

* (P_u) = psychological representation of textual unit (u);
* (E_u^{reported}) = reported-emotion vector;
* (\Delta E_u) = transformative-emotion vector;
* (E_u^{immediate}) = immediate emotional-state vector;
* (w_u^{P}) = emotional weight.

### Emotional Dimensions

The configured emotional dimension set is:

(D_P = {\text{fear}, \text{anger}, \text{sadness}, \text{disgust}, \text{tension}, \text{uncertainty}, \text{hope}, \text{relief}, \text{trust}, \text{compassion}, \text{solidarity}}).

Thus (|D_P| = 11).

These dimensions remain explicit throughout the psychological representation and are not collapsed into a single sentiment score.

### Reported Emotion

Reported emotion represents emotion attributed by the text to people, groups, organisations or other subjects.

The reported-emotion vector is:

(E_u^{reported} = [e_{u,1}^{reported}, e_{u,2}^{reported}, \ldots, e_{u,11}^{reported}]).

For an observed value:

(e_{u,i}^{reported} \in [0,1]).

Dictionary:

* (e_{u,i}^{reported}=0) = the represented emotion is observed as absent under the configured scale;
* (e_{u,i}^{reported}=1) = maximum represented intensity under the configured scale.

Reported emotion is therefore a state-like descriptive construct rather than a state-change construct.

### Transformative Emotion

Transformative emotion represents the change introduced by the current textual unit relative to the permitted preceding context.

The transformative vector is:

(\Delta E_u = [\Delta e_{u,1}, \Delta e_{u,2}, \ldots, \Delta e_{u,11}]).

For an observed value:

(\Delta e_{u,i} \in [-1,1]).

Dictionary:

* (\Delta e_{u,i}<0) = the current textual unit decreases emotional dimension (i);
* (\Delta e_{u,i}=0) = no meaningful change in emotional dimension (i);
* (\Delta e_{u,i}>0) = the current textual unit increases emotional dimension (i).

Transformative emotion is therefore a change variable.

It must not be interpreted as absolute emotional intensity.

Thus:

(\Delta E_u \neq E_u^{reported})

and:

(\Delta E_u \neq E_u^{immediate}).

### Immediate Emotional State

Immediate emotional state represents the amount of each emotional dimension present after interpreting the current textual unit in its permitted preceding context.

The immediate-state vector is:

(E_u^{immediate} = [e_{u,1}^{immediate}, e_{u,2}^{immediate}, \ldots, e_{u,11}^{immediate}]).

For an observed value:

(e_{u,i}^{immediate} \in [0,1]).

Dictionary:

* (e_{u,i}^{immediate}) = emotional level associated with dimension (i) after processing textual unit (u).

Immediate state is conceptually distinct from both reported emotion and transformative emotion.

### Emotional Weight

Emotional weight represents the overall strength or relevance of emotionally consequential information in the textual unit.

It satisfies:

(w_u^{P} \in [0,1]).

Dictionary:

* (w_u^{P}=0) = no meaningful emotional contribution under the configured definition;
* (w_u^{P}=1) = maximum emotional contribution under the configured scale.

Emotional weight does not replace the emotional vectors.

A sentence may have moderate emotional weight while strongly affecting only one specific emotional dimension.

### Numerical Psychological Representation

When all explicit psychological dimensions are numerically available, the value vector is:

(V_u^{P} = [E_u^{reported}, \Delta E_u, E_u^{immediate}, w_u^{P}]).

The dimensional structure is:

(d_P = 11 + 11 + 11 + 1 = 34).

Therefore:

(V_u^{P} \in \mathbb{R}^{34}).

Dictionary:

* dimensions 1–11 = reported emotion;
* dimensions 12–22 = transformative emotion;
* dimensions 23–33 = immediate emotional state;
* dimension 34 = emotional weight.

These 34 dimensions define the explicit psychological ontology in the current representation schema.

Later latent psychological features may exist separately but must not silently redefine this explicit structure.

### Construct Separation

The three emotional constructs share the same eleven dimensions but represent different research quantities.

For emotional dimension (i):

(e_{u,i}^{reported}), (\Delta e_{u,i}), and (e_{u,i}^{immediate})

must remain separate values.

In particular:

(e_{u,i}^{reported} = e_{u,i}^{immediate})

may occur numerically in some sentences, but this does not make the constructs equivalent.

Likewise:

(\Delta e_{u,i}=0)

means no meaningful change, not absence of emotional state.

### Value-State Representation

Each psychological dimension preserves both its value and semantic state.

For psychological dimension (i):

(X_{u,i}^{P} = (v_{u,i}^{P}, s_{u,i}^{P})),

where:

* (v_{u,i}^{P}) = numerical psychological value or unavailable value;
* (s_{u,i}^{P}) = semantic value state.

The permitted states are:

(s_{u,i}^{P} \in {\text{observed}, \text{insufficient_evidence}, \text{not_applicable}, \text{unavailable}, \text{derived}}).

Missing psychological values must remain distinct from substantive zero.

### Availability Mask

The psychological value vector is accompanied by an availability mask:

(M_u^{P} = [m_{u,1}^{P}, m_{u,2}^{P}, \ldots, m_{u,34}^{P}]),

with:

(m_{u,i}^{P} \in {0,1}).

The mask is defined as:

[
m_{u,i}^{P} =
\begin{cases}
1, & \text{if psychological dimension } i \text{ contains an available substantive value} \
0, & \text{otherwise}
\end{cases}
]

Thus (v_{u,i}^{P}=0,;m_{u,i}^{P}=1) means observed zero, while (m_{u,i}^{P}=0) means the value is unavailable.

Therefore:

(\text{missing} \neq 0).

### Operational Psychological Representation

For computational use, the operational psychological representation is:

(\mathcal{P}_u = (V_u^{P}, M_u^{P}, Z_u^{P}, Q_u^{P})),

where:

* (V_u^{P}) = 34-dimensional psychological value vector;
* (M_u^{P}) = psychological availability mask;
* (Z_u^{P}) = representation source;
* (Q_u^{P}) = provenance metadata.

The conceptual construct (P_u) defines what psychological information is represented, while (\mathcal{P}_u) defines the operational research object.

### Representation Source

Every psychological representation must identify its source.

Let:

(Z_u^{P} \in {\text{validated_supervision}, \text{human_reviewed}, \text{model_predicted}, \text{transformative}, \text{recurrent}, \text{aggregated}}).

Dictionary:

* `validated_supervision` = constructed from validated Notebook 03 supervision;
* `human_reviewed` = established or confirmed through human review;
* `model_predicted` = generated by a later Media AI model;
* `transformative` = generated by an explicitly defined state-change mechanism;
* `recurrent` = generated by sequential state processing;
* `aggregated` = constructed from lower-level psychological representations.

Representations with different sources remain distinct research objects even when their numerical values coincide.

### Sentence-Level Psychological Representation

The initial supervision-aligned psychological unit is the sentence.

For sentence (s):

(\mathcal{P}_s = (V_s^{P}, M_s^{P}, Z_s^{P}, Q_s^{P})).

Validated sentence-level supervision may populate these values where available.

If validated supervision is unavailable:

(M_s^{P} = \mathbf{0}_{34}),

while (V_s^{P}) remains unavailable.

Importantly:

(M_s^{P} = \mathbf{0}*{34} \not\Rightarrow V_s^{P} = \mathbf{0}*{34}).

An unavailable psychological representation is not equivalent to thirty-four observed zeros.

### Partial Supervision

Notebook 04 must support partial psychological supervision coverage.

For article (a), let its ordered sentence sequence be:

(S_a = (s_1, s_2, \ldots, s_n)).

Let (r_s^{P} \in {0,1}) indicate whether sentence (s) has usable psychological supervision.

Psychological supervision coverage is:

(\gamma_a^{P} = \frac{\sum_{s \in S_a} r_s^{P}}{|S_a|}).

Dictionary:

* (\gamma_a^{P}) = psychological supervision coverage of article (a);
* (r_s^{P}=1) = sentence (s) has usable psychological supervision;
* (r_s^{P}=0) = sentence (s) does not;
* (|S_a|) = number of sentences in article (a).

Thus (0 \leq \gamma_a^{P} \leq 1).

The interpretation is:

* (\gamma_a^{P}=0) = no psychological supervision;
* (0<\gamma_a^{P}<1) = partial psychological supervision;
* (\gamma_a^{P}=1) = complete psychological supervision coverage.

Partial coverage must remain explicit.

### Transformative Consistency Boundary

Where both immediate state and transformative emotion are available, the architecture may later test consistency between state change and resulting state.

For emotional dimension (i), a later consistency relation may examine:

(e_{t,i}^{immediate} - e_{t-1,i}^{immediate})

relative to:

(\Delta e_{t,i}).

However, the architecture must not assume:

(\Delta e_{t,i} = e_{t,i}^{immediate} - e_{t-1,i}^{immediate})

as an identity.

The supervised transformative construct is an interpretative state-change estimate, while the numerical difference between consecutive immediate states is a separate derived quantity.

Their relationship is an empirical question for later analysis.

### Hierarchical Psychological Representation

The Media AI hierarchy permits psychological representations at four levels:

(P^{token} \rightarrow P^{sentence} \rightarrow P^{paragraph} \rightarrow P^{article}).

The existence of these levels does not imply simple averaging.

For lower-level psychological representations (P_1^{lower}, \ldots, P_n^{lower}), a higher-level representation is produced through an explicit aggregation function:

(P^{higher} = A_P(P_1^{lower}, P_2^{lower}, \ldots, P_n^{lower})).

Dictionary:

* (A_P) = explicitly defined psychological aggregation function;
* (P_i^{lower}) = contributing lower-level psychological representation;
* (P^{higher}) = resulting higher-level psychological representation.

Different psychological constructs may require different aggregation behaviour.

For example, averaging may have a different interpretation for immediate emotional state than for transformative emotion.

The aggregation methodology is therefore defined separately.

### Token-Level Boundary

Sentence-level psychological supervision must not be copied directly to individual tokens.

For token (\tau_i) belonging to sentence (s):

(\tau_i \rightarrow s),

but:

(P_s^{supervised} \not\Rightarrow P_{\tau_i}^{supervised}).

Token-level psychological contribution may later be estimated through attribution or another explicitly defined model mechanism.

Such estimates must remain distinguishable from directly supervised sentence-level values.

### Confidence

Supervision confidence remains metadata rather than a substantive psychological dimension.

For sentence (s):

(\kappa_s^{P} \in [0,1]).

Confidence is not included in the 34-dimensional psychological vector:

(\kappa_s^{P} \notin V_s^{P}).

It must not automatically modify psychological values:

(V_s^{P} \not\leftarrow \kappa_s^{P}V_s^{P}).

Any later confidence-weighting method must be defined explicitly.

### Provenance

Each psychological representation must preserve sufficient provenance to identify its origin and construction.

Conceptually:

(Q_u^{P} = (ID_u, L_u, Z_u^{P}, v_{schema}, m_u, q_u)).

Dictionary:

* (ID_u) = stable textual identity;
* (L_u) = representation level;
* (Z_u^{P}) = psychological representation source;
* (v_{schema}) = representation-schema version;
* (m_u) = construction method;
* (q_u) = source-specific provenance, such as supervisor or model identity.

Provenance must remain associated with the psychological representation through later transformative, recurrent, confluent, aggregation, attribution and evaluation stages.

### Temporal Boundary

Psychological state construction must preserve textual order.

For sequential step (t), the permitted information set is:

(\mathcal{I}_t = {X_1, X_2, \ldots, X_t}).

Future units (X_{t+1}, X_{t+2}, \ldots) must not influence a psychological state assigned to step (t) where the configured architecture requires preceding-context processing.

This is especially important for transformative and recurrent representations.

### Recurrent Boundary

The psychological representation provides inputs to later recurrent state construction but does not itself constitute recurrent state.

A later psychological recurrent state may be written as:

(H_t^{P} = \Phi_R^{P}(H_{t-1}^{P}, P_t, \Delta P_t)).

Dictionary:

* (H_{t-1}^{P}) = previous psychological recurrent state;
* (P_t) = psychological representation at step (t);
* (\Delta P_t) = psychological change associated with step (t);
* (\Phi_R^{P}) = explicitly defined psychological recurrent update function;
* (H_t^{P}) = resulting psychological recurrent state.

Block 7 does not calculate (H_t^{P}).

### Attribution Requirement

Higher-level psychological states must remain traceable to their contributing textual units.

The required backward path is:

(P^{article} \rightarrow P^{paragraph} \rightarrow P^{sentence} \rightarrow Text).

Where token-level attribution later becomes available:

(P^{sentence} \rightarrow Token \rightarrow Text).

For higher-level psychological representation (h) and lower-level textual unit (l), later contribution may be represented as:

(A_{h \leftarrow l}^{P}).

Dictionary:

* (A_{h \leftarrow l}^{P}) = contribution of lower-level unit (l) to higher-level psychological state (h).

This structure is particularly important for explaining emotional trajectories.

For example, an observed rise in article-level fear should later be traceable to the paragraphs, sentences and potentially tokens most associated with that change.

### Current Supervision State

The psychological representation architecture must remain valid even when validated Notebook 03 supervision has not yet been persisted or loaded.

If psychological supervision is unavailable:

(M_s^{P} = \mathbf{0}_{34})

for the affected sentence, while (V_s^{P}) remains unavailable.

The representation record can nevertheless preserve:

((ID_s, V_s^{P}, M_s^{P}, Z_s^{P}, Q_s^{P}))

without fabricating substantive values.

Once validated supervision becomes available, the same representation contract can populate those values without changing the architecture.

### Output

This block defines and, where data are available, constructs sentence-level psychological representation records.

Its outputs include:

* the eleven configured emotional dimensions;
* reported-emotion representation;
* transformative-emotion representation;
* immediate emotional-state representation;
* emotional weight;
* the 34-dimensional explicit psychological value contract;
* psychological availability masks;
* semantic value states;
* psychological representation-source metadata;
* sentence-level psychological representation records;
* psychological supervision-coverage diagnostics;
* psychological provenance;
* hierarchical psychological-representation boundaries;
* transformative-consistency boundaries;
* token-level supervision boundaries;
* temporal and recurrent boundaries;
* backward-attribution requirements.

Where validated supervision is unavailable, psychological values remain explicitly unavailable rather than being replaced with zero.

No token-level psychological supervision is fabricated.

No paragraph-level or article-level psychological aggregation is performed.

No factual or social representation is modified.

No recurrent or confluent state is calculated.

No attribution score, model training or experimental evaluation is performed.


````python
# =============================================================================
# Media AI — Psychological Representation
# =============================================================================

from copy import deepcopy
from datetime import datetime, timezone

import numpy as np


# -----------------------------------------------------------------------------
# Block dependency checks
# -----------------------------------------------------------------------------

required_block_1_objects = [
    "BUILD_PSYCHOLOGICAL_REPRESENTATION",
    "PRESERVE_SUPERVISION_VALUE_STATES",
    "ALLOW_MISSING_SUPERVISION_AS_ZERO",
    "PRESERVE_SUPERVISION_PROVENANCE",
    "ALLOW_FUTURE_CONTEXT_IN_STATE_CONSTRUCTION",
]

missing_block_1_objects = [
    object_name
    for object_name
    in required_block_1_objects
    if object_name not in globals()
]

if missing_block_1_objects:
    raise NameError(
        "Notebook 04 Block 1 is not initialised. "
        "Missing: "
        f"{missing_block_1_objects}"
    )


required_block_2_objects = [
    "SENTENCE_IDS",
    "CANONICAL_SUPERVISION_BY_SENTENCE_ID",
    "CANONICAL_SUPERVISION_AVAILABLE",
]

missing_block_2_objects = [
    object_name
    for object_name
    in required_block_2_objects
    if object_name not in globals()
]

if missing_block_2_objects:
    raise NameError(
        "Notebook 04 Block 2 is not initialised. "
        "Missing: "
        f"{missing_block_2_objects}"
    )


required_block_3_objects = [
    "HIERARCHICAL_TEXT_AND_IDENTITY_VALID",
    "SENTENCE_BY_ID",
    "SENTENCE_TO_PARAGRAPH_ID",
    "SENTENCE_TO_ARTICLE_ID",
    "ARTICLE_SENTENCE_ORDER",
]

missing_block_3_objects = [
    object_name
    for object_name
    in required_block_3_objects
    if object_name not in globals()
]

if missing_block_3_objects:
    raise NameError(
        "Notebook 04 Block 3 is not initialised. "
        "Missing: "
        f"{missing_block_3_objects}"
    )


required_block_4_objects = [
    "MEDIA_AI_REPRESENTATION_SPECIFICATION_VALID",
    "PSYCHOLOGICAL_EMOTIONAL_DIMENSIONS",
    "PSYCHOLOGICAL_DIMENSIONS",
    "PSYCHOLOGICAL_DIMENSION_COUNT",
    "PSYCHOLOGICAL_DIMENSION_INDEX",
    "PSYCHOLOGICAL_DIMENSION_SPECIFICATION",
    "PSYCHOLOGICAL_CONSTRUCT_SLICES",
    "MEDIA_AI_VALUE_STATES",
]

missing_block_4_objects = [
    object_name
    for object_name
    in required_block_4_objects
    if object_name not in globals()
]

if missing_block_4_objects:
    raise NameError(
        "Notebook 04 Block 4 is not initialised. "
        "Missing: "
        f"{missing_block_4_objects}"
    )


if not HIERARCHICAL_TEXT_AND_IDENTITY_VALID:
    raise ValueError(
        "Psychological representation requires a valid "
        "hierarchical text and identity structure."
    )


if not MEDIA_AI_REPRESENTATION_SPECIFICATION_VALID:
    raise ValueError(
        "Psychological representation requires a valid "
        "Media AI representation specification."
    )


if ALLOW_MISSING_SUPERVISION_AS_ZERO:
    raise ValueError(
        "Missing supervision must not be converted to numeric zero."
    )


if ALLOW_FUTURE_CONTEXT_IN_STATE_CONSTRUCTION:
    raise ValueError(
        "Future context must not enter psychological state construction."
    )


# -----------------------------------------------------------------------------
# Psychological representation configuration
# -----------------------------------------------------------------------------

PSYCHOLOGICAL_REPRESENTATION_VERSION = "1.0"

PSYCHOLOGICAL_REPRESENTATION_LEVEL = "sentence"

PSYCHOLOGICAL_REPRESENTATION_SPACE = "psychological"

PSYCHOLOGICAL_REPRESENTATION_SOURCE_VALIDATED = (
    "validated_supervision"
)

PSYCHOLOGICAL_REPRESENTATION_SOURCE_UNAVAILABLE = (
    "unavailable"
)


# -----------------------------------------------------------------------------
# Psychological dimensional validation
# -----------------------------------------------------------------------------

if len(
    PSYCHOLOGICAL_EMOTIONAL_DIMENSIONS
) != 11:
    raise ValueError(
        "Psychological representation must preserve "
        "the configured eleven emotional dimensions."
    )


if PSYCHOLOGICAL_DIMENSION_COUNT != 34:
    raise ValueError(
        "Psychological representation must contain "
        "exactly 34 explicit dimensions."
    )


EXPECTED_CONSTRUCT_SLICES = {
    "reported_emotion":
        (
            0,
            11,
        ),

    "transformative_emotion":
        (
            11,
            22,
        ),

    "immediate_emotional_state":
        (
            22,
            33,
        ),

    "emotional_weight":
        (
            33,
            34,
        ),
}


if (
    PSYCHOLOGICAL_CONSTRUCT_SLICES
    != EXPECTED_CONSTRUCT_SLICES
):
    raise ValueError(
        "Psychological construct slices do not match "
        "the configured 34-dimensional contract."
    )


# -----------------------------------------------------------------------------
# Psychological vector initialisation
# -----------------------------------------------------------------------------

def create_empty_psychological_vector():
    """
    Create one unavailable 34-dimensional psychological value vector.
    """

    return np.full(
        PSYCHOLOGICAL_DIMENSION_COUNT,
        np.nan,
        dtype=float,
    )


def create_empty_psychological_mask():
    """
    Create one unavailable 34-dimensional psychological mask.
    """

    return np.zeros(
        PSYCHOLOGICAL_DIMENSION_COUNT,
        dtype=np.int8,
    )


def create_empty_psychological_value_states():
    """
    Create semantic value states for an unavailable psychological record.
    """

    return {
        dimension:
            "unavailable"

        for dimension
        in PSYCHOLOGICAL_DIMENSIONS
    }


# -----------------------------------------------------------------------------
# Generic supervised value extraction
# -----------------------------------------------------------------------------

def assign_psychological_scalar(
    values,
    mask,
    states,
    dimension_name,
    value_record,
):
    """
    Validate and assign one scalar psychological supervision value.
    """

    if not isinstance(
        value_record,
        dict,
    ):
        raise TypeError(
            f"{dimension_name} must be represented as a dictionary."
        )


    value = value_record.get(
        "value"
    )

    value_state = value_record.get(
        "value_state"
    )


    if value_state not in MEDIA_AI_VALUE_STATES:
        raise ValueError(
            f"Invalid value state for {dimension_name}: "
            f"{value_state}"
        )


    states[
        dimension_name
    ] = value_state


    dimension_specification = (
        PSYCHOLOGICAL_DIMENSION_SPECIFICATION[
            dimension_name
        ]
    )


    minimum = (
        dimension_specification[
            "minimum"
        ]
    )

    maximum = (
        dimension_specification[
            "maximum"
        ]
    )


    dimension_index = (
        PSYCHOLOGICAL_DIMENSION_INDEX[
            dimension_name
        ]
    )


    if value_state == "observed":

        if not isinstance(
            value,
            (int, float),
        ):
            raise ValueError(
                f"Observed value for {dimension_name} must be numeric."
            )


        numeric_value = float(
            value
        )


        if not (
            minimum
            <= numeric_value
            <= maximum
        ):
            raise ValueError(
                f"{dimension_name} must lie in "
                f"[{minimum}, {maximum}]."
            )


        values[
            dimension_index
        ] = numeric_value

        mask[
            dimension_index
        ] = 1


    else:

        if value is not None:
            raise ValueError(
                f"Non-observed value for {dimension_name} "
                "must use null."
            )


# -----------------------------------------------------------------------------
# Validated supervision extraction
# -----------------------------------------------------------------------------

def extract_psychological_supervision(
    supervision_record,
):
    """
    Extract psychological values, masks and semantic states from one
    validated sentence-level supervision record.
    """

    if not isinstance(
        supervision_record,
        dict,
    ):
        raise TypeError(
            "supervision_record must be a dictionary."
        )


    annotations = supervision_record.get(
        "annotations",
        {}
    )


    psychological_values = (
        create_empty_psychological_vector()
    )

    psychological_mask = (
        create_empty_psychological_mask()
    )

    psychological_states = (
        create_empty_psychological_value_states()
    )


    # -------------------------------------------------------------------------
    # Reported emotion
    # -------------------------------------------------------------------------

    reported_emotion = annotations.get(
        "reported_emotion",
        {}
    )


    for dimension in (
        PSYCHOLOGICAL_EMOTIONAL_DIMENSIONS
    ):

        dimension_name = (
            f"reported_emotion.{dimension}"
        )

        assign_psychological_scalar(
            values=
                psychological_values,

            mask=
                psychological_mask,

            states=
                psychological_states,

            dimension_name=
                dimension_name,

            value_record=
                reported_emotion.get(
                    dimension,
                    {}
                ),
        )


    # -------------------------------------------------------------------------
    # Transformative emotion
    # -------------------------------------------------------------------------

    transformative_emotion = annotations.get(
        "transformative_emotion",
        {}
    )


    for dimension in (
        PSYCHOLOGICAL_EMOTIONAL_DIMENSIONS
    ):

        dimension_name = (
            f"transformative_emotion.{dimension}"
        )

        assign_psychological_scalar(
            values=
                psychological_values,

            mask=
                psychological_mask,

            states=
                psychological_states,

            dimension_name=
                dimension_name,

            value_record=
                transformative_emotion.get(
                    dimension,
                    {}
                ),
        )


    # -------------------------------------------------------------------------
    # Immediate emotional state
    # -------------------------------------------------------------------------

    immediate_emotional_state = annotations.get(
        "immediate_emotional_state",
        {}
    )


    for dimension in (
        PSYCHOLOGICAL_EMOTIONAL_DIMENSIONS
    ):

        dimension_name = (
            f"immediate_emotional_state.{dimension}"
        )

        assign_psychological_scalar(
            values=
                psychological_values,

            mask=
                psychological_mask,

            states=
                psychological_states,

            dimension_name=
                dimension_name,

            value_record=
                immediate_emotional_state.get(
                    dimension,
                    {}
                ),
        )


    # -------------------------------------------------------------------------
    # Emotional weight
    # -------------------------------------------------------------------------

    assign_psychological_scalar(
        values=
            psychological_values,

        mask=
            psychological_mask,

        states=
            psychological_states,

        dimension_name=
            "emotional_weight",

        value_record=
            annotations.get(
                "emotional_weight",
                {}
            ),
    )


    return (
        psychological_values,
        psychological_mask,
        psychological_states,
    )


# -----------------------------------------------------------------------------
# Psychological provenance
# -----------------------------------------------------------------------------

def build_psychological_provenance(
    sentence_id,
    representation_source,
    supervision_record=None,
):
    """
    Build psychological-representation provenance for one sentence.
    """

    provenance = {
        "media_ai_record_id":
            SENTENCE_TO_ARTICLE_ID[
                sentence_id
            ],

        "paragraph_id":
            SENTENCE_TO_PARAGRAPH_ID[
                sentence_id
            ],

        "sentence_id":
            sentence_id,

        "representation_level":
            PSYCHOLOGICAL_REPRESENTATION_LEVEL,

        "representation_space":
            PSYCHOLOGICAL_REPRESENTATION_SPACE,

        "representation_source":
            representation_source,

        "representation_schema_version":
            REPRESENTATION_SCHEMA_VERSION,

        "psychological_representation_version":
            PSYCHOLOGICAL_REPRESENTATION_VERSION,

        "construction_method":
            (
                "validated_supervision_mapping"
                if supervision_record
                else "unavailable_supervision_placeholder"
            ),

        "supervisor_identifier":
            None,

        "supervision_method":
            None,

        "provider":
            None,

        "model":
            None,

        "supervision_schema_version":
            None,

        "annotation_policy_version":
            None,

        "context_scope":
            None,

        "constructed_at_utc":
            datetime.now(
                timezone.utc
            ).isoformat(),
    }


    if supervision_record:

        supervision_provenance = (
            supervision_record.get(
                "provenance",
                {}
            )
        )

        provenance[
            "supervisor_identifier"
        ] = supervision_provenance.get(
            "supervisor_identifier"
        )

        provenance[
            "supervision_method"
        ] = supervision_provenance.get(
            "supervision_method"
        )

        provenance[
            "provider"
        ] = supervision_provenance.get(
            "provider"
        )

        provenance[
            "model"
        ] = supervision_provenance.get(
            "model"
        )

        provenance[
            "supervision_schema_version"
        ] = supervision_provenance.get(
            "supervision_schema_version"
        )

        provenance[
            "annotation_policy_version"
        ] = supervision_provenance.get(
            "annotation_policy_version"
        )

        provenance[
            "context_scope"
        ] = supervision_provenance.get(
            "context_scope"
        )


    return provenance


# -----------------------------------------------------------------------------
# Sentence-level psychological representation construction
# -----------------------------------------------------------------------------

PSYCHOLOGICAL_SENTENCE_REPRESENTATIONS = []


for sentence_id in SENTENCE_IDS:

    supervision_record = (
        CANONICAL_SUPERVISION_BY_SENTENCE_ID.get(
            sentence_id
        )
    )


    if supervision_record is not None:

        (
            psychological_values,
            psychological_mask,
            psychological_states,
        ) = extract_psychological_supervision(
            supervision_record
        )


        representation_source = (
            PSYCHOLOGICAL_REPRESENTATION_SOURCE_VALIDATED
        )


        confidence_record = (
            supervision_record.get(
                "annotations",
                {}
            ).get(
                "annotation_confidence",
                {}
            )
        )


        supervision_confidence = (
            confidence_record.get(
                "value"
            )
        )


        supervision_confidence_state = (
            confidence_record.get(
                "value_state"
            )
        )


    else:

        psychological_values = (
            create_empty_psychological_vector()
        )

        psychological_mask = (
            create_empty_psychological_mask()
        )

        psychological_states = (
            create_empty_psychological_value_states()
        )

        representation_source = (
            PSYCHOLOGICAL_REPRESENTATION_SOURCE_UNAVAILABLE
        )

        supervision_confidence = None

        supervision_confidence_state = (
            "unavailable"
        )


    psychological_record = {
        "record_type":
            "media_ai_psychological_representation",

        "representation_version":
            PSYCHOLOGICAL_REPRESENTATION_VERSION,

        "media_ai_record_id":
            SENTENCE_TO_ARTICLE_ID[
                sentence_id
            ],

        "paragraph_id":
            SENTENCE_TO_PARAGRAPH_ID[
                sentence_id
            ],

        "sentence_id":
            sentence_id,

        "representation_level":
            PSYCHOLOGICAL_REPRESENTATION_LEVEL,

        "representation_space":
            PSYCHOLOGICAL_REPRESENTATION_SPACE,

        "representation_source":
            representation_source,

        "values":
            psychological_values,

        "availability_mask":
            psychological_mask,

        "value_states":
            psychological_states,

        "supervision_confidence": {
            "value":
                supervision_confidence,

            "value_state":
                supervision_confidence_state,
        },

        "provenance":
            build_psychological_provenance(
                sentence_id=
                    sentence_id,

                representation_source=
                    representation_source,

                supervision_record=
                    supervision_record,
            ),
    }


    PSYCHOLOGICAL_SENTENCE_REPRESENTATIONS.append(
        psychological_record
    )


# -----------------------------------------------------------------------------
# Sentence-level lookup
# -----------------------------------------------------------------------------

PSYCHOLOGICAL_SENTENCE_BY_ID = {
    record[
        "sentence_id"
    ]:
        record

    for record
    in PSYCHOLOGICAL_SENTENCE_REPRESENTATIONS
}


# -----------------------------------------------------------------------------
# Psychological representation validation
# -----------------------------------------------------------------------------

PSYCHOLOGICAL_REPRESENTATION_ERRORS = []


for record in (
    PSYCHOLOGICAL_SENTENCE_REPRESENTATIONS
):

    values = record.get(
        "values"
    )

    mask = record.get(
        "availability_mask"
    )

    states = record.get(
        "value_states",
        {}
    )


    if not isinstance(
        values,
        np.ndarray,
    ):

        PSYCHOLOGICAL_REPRESENTATION_ERRORS.append(
            "values_not_numpy_array"
        )

        break


    if values.shape != (
        PSYCHOLOGICAL_DIMENSION_COUNT,
    ):

        PSYCHOLOGICAL_REPRESENTATION_ERRORS.append(
            "invalid_psychological_value_shape"
        )

        break


    if not isinstance(
        mask,
        np.ndarray,
    ):

        PSYCHOLOGICAL_REPRESENTATION_ERRORS.append(
            "mask_not_numpy_array"
        )

        break


    if mask.shape != (
        PSYCHOLOGICAL_DIMENSION_COUNT,
    ):

        PSYCHOLOGICAL_REPRESENTATION_ERRORS.append(
            "invalid_psychological_mask_shape"
        )

        break


    if not np.all(
        np.isin(
            mask,
            [
                0,
                1,
            ],
        )
    ):

        PSYCHOLOGICAL_REPRESENTATION_ERRORS.append(
            "invalid_psychological_mask_values"
        )

        break


    for dimension in PSYCHOLOGICAL_DIMENSIONS:

        if dimension not in states:

            PSYCHOLOGICAL_REPRESENTATION_ERRORS.append(
                "missing_psychological_value_state"
            )

            break


        state = states[
            dimension
        ]


        if state not in MEDIA_AI_VALUE_STATES:

            PSYCHOLOGICAL_REPRESENTATION_ERRORS.append(
                "invalid_psychological_value_state"
            )

            break


        dimension_index = (
            PSYCHOLOGICAL_DIMENSION_INDEX[
                dimension
            ]
        )


        dimension_specification = (
            PSYCHOLOGICAL_DIMENSION_SPECIFICATION[
                dimension
            ]
        )


        if state == "observed":

            if mask[
                dimension_index
            ] != 1:

                PSYCHOLOGICAL_REPRESENTATION_ERRORS.append(
                    "observed_value_without_mask"
                )

                break


            numeric_value = values[
                dimension_index
            ]


            if np.isnan(
                numeric_value
            ):

                PSYCHOLOGICAL_REPRESENTATION_ERRORS.append(
                    "observed_value_is_nan"
                )

                break


            if not (
                dimension_specification[
                    "minimum"
                ]
                <= numeric_value
                <= dimension_specification[
                    "maximum"
                ]
            ):

                PSYCHOLOGICAL_REPRESENTATION_ERRORS.append(
                    "psychological_value_out_of_range"
                )

                break


        else:

            if mask[
                dimension_index
            ] != 0:

                PSYCHOLOGICAL_REPRESENTATION_ERRORS.append(
                    "unavailable_value_with_active_mask"
                )

                break


            if not np.isnan(
                values[
                    dimension_index
                ]
            ):

                PSYCHOLOGICAL_REPRESENTATION_ERRORS.append(
                    "unavailable_value_not_nan"
                )

                break


    if PSYCHOLOGICAL_REPRESENTATION_ERRORS:
        break


# -----------------------------------------------------------------------------
# Source-specific validation
# -----------------------------------------------------------------------------

for record in (
    PSYCHOLOGICAL_SENTENCE_REPRESENTATIONS
):

    source = record.get(
        "representation_source"
    )


    if source == (
        PSYCHOLOGICAL_REPRESENTATION_SOURCE_UNAVAILABLE
    ):

        if np.any(
            record[
                "availability_mask"
            ]
        ):

            PSYCHOLOGICAL_REPRESENTATION_ERRORS.append(
                "unavailable_source_has_observed_values"
            )

            break


    elif source == (
        PSYCHOLOGICAL_REPRESENTATION_SOURCE_VALIDATED
    ):

        if (
            record[
                "sentence_id"
            ]
            not in CANONICAL_SUPERVISION_BY_SENTENCE_ID
        ):

            PSYCHOLOGICAL_REPRESENTATION_ERRORS.append(
                "validated_source_without_supervision"
            )

            break


    else:

        PSYCHOLOGICAL_REPRESENTATION_ERRORS.append(
            "unexpected_psychological_representation_source"
        )

        break


# -----------------------------------------------------------------------------
# Construct-level availability counts
# -----------------------------------------------------------------------------

reported_slice = slice(
    *PSYCHOLOGICAL_CONSTRUCT_SLICES[
        "reported_emotion"
    ]
)

transformative_slice = slice(
    *PSYCHOLOGICAL_CONSTRUCT_SLICES[
        "transformative_emotion"
    ]
)

immediate_slice = slice(
    *PSYCHOLOGICAL_CONSTRUCT_SLICES[
        "immediate_emotional_state"
    ]
)

emotional_weight_slice = slice(
    *PSYCHOLOGICAL_CONSTRUCT_SLICES[
        "emotional_weight"
    ]
)


REPORTED_EMOTION_AVAILABLE_VALUE_COUNT = sum(
    int(
        record[
            "availability_mask"
        ][
            reported_slice
        ].sum()
    )
    for record
    in PSYCHOLOGICAL_SENTENCE_REPRESENTATIONS
)


TRANSFORMATIVE_EMOTION_AVAILABLE_VALUE_COUNT = sum(
    int(
        record[
            "availability_mask"
        ][
            transformative_slice
        ].sum()
    )
    for record
    in PSYCHOLOGICAL_SENTENCE_REPRESENTATIONS
)


IMMEDIATE_EMOTION_AVAILABLE_VALUE_COUNT = sum(
    int(
        record[
            "availability_mask"
        ][
            immediate_slice
        ].sum()
    )
    for record
    in PSYCHOLOGICAL_SENTENCE_REPRESENTATIONS
)


EMOTIONAL_WEIGHT_AVAILABLE_VALUE_COUNT = sum(
    int(
        record[
            "availability_mask"
        ][
            emotional_weight_slice
        ].sum()
    )
    for record
    in PSYCHOLOGICAL_SENTENCE_REPRESENTATIONS
)


# -----------------------------------------------------------------------------
# Supervision coverage diagnostics
# -----------------------------------------------------------------------------

PSYCHOLOGICAL_SUPERVISED_SENTENCE_COUNT = sum(
    1
    for record
    in PSYCHOLOGICAL_SENTENCE_REPRESENTATIONS
    if record[
        "representation_source"
    ]
    == PSYCHOLOGICAL_REPRESENTATION_SOURCE_VALIDATED
)


PSYCHOLOGICAL_UNAVAILABLE_SENTENCE_COUNT = (
    len(
        PSYCHOLOGICAL_SENTENCE_REPRESENTATIONS
    )
    - PSYCHOLOGICAL_SUPERVISED_SENTENCE_COUNT
)


PSYCHOLOGICAL_SENTENCE_COVERAGE = (
    PSYCHOLOGICAL_SUPERVISED_SENTENCE_COUNT
    / len(
        PSYCHOLOGICAL_SENTENCE_REPRESENTATIONS
    )
    if PSYCHOLOGICAL_SENTENCE_REPRESENTATIONS
    else 0.0
)


# -----------------------------------------------------------------------------
# Article-level psychological coverage
# -----------------------------------------------------------------------------

PSYCHOLOGICAL_ARTICLE_COVERAGE = []


for (
    article_id,
    sentence_ids,
) in ARTICLE_SENTENCE_ORDER.items():

    supervised_count = sum(
        1
        for sentence_id
        in sentence_ids
        if PSYCHOLOGICAL_SENTENCE_BY_ID[
            sentence_id
        ][
            "representation_source"
        ]
        == PSYCHOLOGICAL_REPRESENTATION_SOURCE_VALIDATED
    )


    sentence_count = len(
        sentence_ids
    )


    coverage_fraction = (
        supervised_count
        / sentence_count
        if sentence_count
        else 0.0
    )


    if supervised_count == 0:

        coverage_status = (
            "not_supervised"
        )

    elif (
        supervised_count
        == sentence_count
    ):

        coverage_status = (
            "fully_supervised"
        )

    else:

        coverage_status = (
            "partially_supervised"
        )


    PSYCHOLOGICAL_ARTICLE_COVERAGE.append({
        "media_ai_record_id":
            article_id,

        "sentence_count":
            sentence_count,

        "supervised_sentence_count":
            supervised_count,

        "coverage_fraction":
            coverage_fraction,

        "coverage_status":
            coverage_status,
    })


# -----------------------------------------------------------------------------
# Psychological availability accounting
# -----------------------------------------------------------------------------

PSYCHOLOGICAL_AVAILABLE_VALUE_COUNT = sum(
    int(
        record[
            "availability_mask"
        ].sum()
    )
    for record
    in PSYCHOLOGICAL_SENTENCE_REPRESENTATIONS
)


PSYCHOLOGICAL_TOTAL_VALUE_CAPACITY = (
    len(
        PSYCHOLOGICAL_SENTENCE_REPRESENTATIONS
    )
    * PSYCHOLOGICAL_DIMENSION_COUNT
)


PSYCHOLOGICAL_VALUE_AVAILABILITY_RATE = (
    PSYCHOLOGICAL_AVAILABLE_VALUE_COUNT
    / PSYCHOLOGICAL_TOTAL_VALUE_CAPACITY
    if PSYCHOLOGICAL_TOTAL_VALUE_CAPACITY
    else 0.0
)


# -----------------------------------------------------------------------------
# Transformative/immediate consistency preparation
# -----------------------------------------------------------------------------

# No consistency score is calculated here.
#
# The block only preserves the slices required for later comparison between:
#
#   supervised transformative emotion
#
# and
#
#   derived consecutive changes in immediate emotional state.

PSYCHOLOGICAL_CONSISTENCY_INTERFACE = {
    "transformative_slice":
        (
            PSYCHOLOGICAL_CONSTRUCT_SLICES[
                "transformative_emotion"
            ]
        ),

    "immediate_state_slice":
        (
            PSYCHOLOGICAL_CONSTRUCT_SLICES[
                "immediate_emotional_state"
            ]
        ),

    "identity_assumed":
        False,

    "comparison_reserved_for_later":
        True,
}


# -----------------------------------------------------------------------------
# Representation status
# -----------------------------------------------------------------------------

PSYCHOLOGICAL_REPRESENTATION_VALID = (
    len(
        PSYCHOLOGICAL_REPRESENTATION_ERRORS
    )
    == 0
)


if not PSYCHOLOGICAL_REPRESENTATION_VALID:

    raise ValueError(
        "Psychological representation failed validation: "
        f"{PSYCHOLOGICAL_REPRESENTATION_ERRORS}"
    )


PSYCHOLOGICAL_REPRESENTATIONS_CONSTRUCTED = (
    True
)


PSYCHOLOGICAL_REPRESENTATION_HAS_OBSERVED_VALUES = (
    PSYCHOLOGICAL_AVAILABLE_VALUE_COUNT
    > 0
)


# -----------------------------------------------------------------------------
# Psychological representation summary
# -----------------------------------------------------------------------------

PSYCHOLOGICAL_REPRESENTATION_SUMMARY = {
    "representation_version":
        PSYCHOLOGICAL_REPRESENTATION_VERSION,

    "representation_level":
        PSYCHOLOGICAL_REPRESENTATION_LEVEL,

    "dimension_count":
        PSYCHOLOGICAL_DIMENSION_COUNT,

    "sentence_records":
        len(
            PSYCHOLOGICAL_SENTENCE_REPRESENTATIONS
        ),

    "supervised_sentence_records":
        PSYCHOLOGICAL_SUPERVISED_SENTENCE_COUNT,

    "unavailable_sentence_records":
        PSYCHOLOGICAL_UNAVAILABLE_SENTENCE_COUNT,

    "sentence_supervision_coverage":
        PSYCHOLOGICAL_SENTENCE_COVERAGE,

    "available_values":
        PSYCHOLOGICAL_AVAILABLE_VALUE_COUNT,

    "total_value_capacity":
        PSYCHOLOGICAL_TOTAL_VALUE_CAPACITY,

    "value_availability_rate":
        PSYCHOLOGICAL_VALUE_AVAILABILITY_RATE,

    "reported_available_values":
        REPORTED_EMOTION_AVAILABLE_VALUE_COUNT,

    "transformative_available_values":
        TRANSFORMATIVE_EMOTION_AVAILABLE_VALUE_COUNT,

    "immediate_available_values":
        IMMEDIATE_EMOTION_AVAILABLE_VALUE_COUNT,

    "emotional_weight_available_values":
        EMOTIONAL_WEIGHT_AVAILABLE_VALUE_COUNT,

    "representations_constructed":
        PSYCHOLOGICAL_REPRESENTATIONS_CONSTRUCTED,

    "observed_values_present":
        PSYCHOLOGICAL_REPRESENTATION_HAS_OBSERVED_VALUES,

    "representation_valid":
        PSYCHOLOGICAL_REPRESENTATION_VALID,
}


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Psychological Representation")
print("=" * 72)

print(
    f"Representation version      : "
    f"{PSYCHOLOGICAL_REPRESENTATION_VERSION}"
)

print(
    f"Representation level        : "
    f"{PSYCHOLOGICAL_REPRESENTATION_LEVEL}"
)

print(
    f"Emotional dimensions        : "
    f"{len(PSYCHOLOGICAL_EMOTIONAL_DIMENSIONS)}"
)

print(
    f"Psychological dimensions    : "
    f"{PSYCHOLOGICAL_DIMENSION_COUNT}"
)

print(
    f"Sentence records            : "
    f"{len(PSYCHOLOGICAL_SENTENCE_REPRESENTATIONS)}"
)

print("-" * 72)

print("Supervision coverage")

print(
    f"Validated supervision avail.: "
    f"{CANONICAL_SUPERVISION_AVAILABLE}"
)

print(
    f"Supervised sentences        : "
    f"{PSYCHOLOGICAL_SUPERVISED_SENTENCE_COUNT}"
)

print(
    f"Unavailable sentences       : "
    f"{PSYCHOLOGICAL_UNAVAILABLE_SENTENCE_COUNT}"
)

print(
    f"Sentence coverage           : "
    f"{PSYCHOLOGICAL_SENTENCE_COVERAGE:.4f}"
)

print("-" * 72)

print("Construct availability")

print(
    f"Reported emotion values     : "
    f"{REPORTED_EMOTION_AVAILABLE_VALUE_COUNT}"
)

print(
    f"Transformative values       : "
    f"{TRANSFORMATIVE_EMOTION_AVAILABLE_VALUE_COUNT}"
)

print(
    f"Immediate-state values      : "
    f"{IMMEDIATE_EMOTION_AVAILABLE_VALUE_COUNT}"
)

print(
    f"Emotional-weight values     : "
    f"{EMOTIONAL_WEIGHT_AVAILABLE_VALUE_COUNT}"
)

print(
    f"All psychological values    : "
    f"{PSYCHOLOGICAL_AVAILABLE_VALUE_COUNT}"
)

print(
    f"Total psychological capacity: "
    f"{PSYCHOLOGICAL_TOTAL_VALUE_CAPACITY}"
)

print(
    f"Value availability rate     : "
    f"{PSYCHOLOGICAL_VALUE_AVAILABILITY_RATE:.4f}"
)

print(
    f"Observed values present     : "
    f"{PSYCHOLOGICAL_REPRESENTATION_HAS_OBSERVED_VALUES}"
)

print("-" * 72)

print("Article coverage")

for article_record in (
    PSYCHOLOGICAL_ARTICLE_COVERAGE
):

    print(
        f"{article_record['media_ai_record_id']}"
    )

    print(
        f"  Sentences                 : "
        f"{article_record['sentence_count']}"
    )

    print(
        f"  Supervised                : "
        f"{article_record['supervised_sentence_count']}"
    )

    print(
        f"  Coverage                  : "
        f"{article_record['coverage_fraction']:.4f}"
    )

    print(
        f"  Status                    : "
        f"{article_record['coverage_status']}"
    )

print("-" * 72)

print("Methodological boundaries")

print(
    f"Transformative identity     : "
    f"{PSYCHOLOGICAL_CONSISTENCY_INTERFACE['identity_assumed']}"
)

print(
    f"Future context allowed      : "
    f"{ALLOW_FUTURE_CONTEXT_IN_STATE_CONSTRUCTION}"
)

print(
    f"Representation valid        : "
    f"{PSYCHOLOGICAL_REPRESENTATION_VALID}"
)

print(
    f"Representations constructed : "
    f"{PSYCHOLOGICAL_REPRESENTATIONS_CONSTRUCTED}"
)

print("=" * 72)

print(
    "Sentence-level psychological representation records "
    "constructed successfully."
)

if PSYCHOLOGICAL_REPRESENTATION_HAS_OBSERVED_VALUES:

    print(
        "Available validated supervision has populated "
        "psychological values."
    )

else:

    print(
        "Validated psychological supervision is currently unavailable."
    )

    print(
        "Psychological values therefore remain unavailable "
        "with inactive masks."
    )

print(
    "Reported emotion, transformative emotion and immediate "
    "emotional state remain separate constructs."
)

print(
    "Transformative emotion has not been equated with consecutive "
    "differences in immediate emotional state."
)

print(
    "Missing psychological values have not been converted "
    "to numerical zero."
)

print(
    "Sentence-level psychological supervision has not been "
    "propagated to tokens."
)

print(
    "No recurrent state or hierarchical psychological aggregation "
    "has been calculated."
)

print(
    "No attribution score, model training or experimental evaluation "
    "has been performed."
)
````

## Social and Confluent Representation

This block defines the social representation and the confluent representation contract used by the Media AI architecture.

The social space represents **actors, groups, memberships and relationships expressed or implied within the analysed text**.

Unlike the factual and psychological spaces, the social space is not assumed to have a fixed number of dimensions.

The confluent mechanism is defined separately. It provides the architectural interface through which factual, psychological and social information may later interact without collapsing the three representation spaces into a single undifferentiated vector.

### Social Representation

For textual unit (u), the conceptual social representation is (S_u = {A_u, G_u, M_u, R_u}), where:

* (S_u) = social representation of textual unit (u);
* (A_u) = set of identified actors;
* (G_u) = set of identified groups;
* (M_u) = actor-group membership structure;
* (R_u) = relationship structure.

The social representation is therefore a structured relational object rather than a fixed-dimensional numerical vector.

### Actors

Let (A_u = {a_1, a_2, \ldots, a_n}) denote the actors represented in textual unit (u).

An actor may represent a person, organisation, institution, collective entity or another socially relevant participant recognised under the configured social ontology.

Each actor (a_i) must preserve a stable identity where identity resolution is possible.

Conceptually, an actor record may be represented as (a_i = (ID_i, L_i, N_i, T_i, Q_i)), where:

* (ID_i) = stable actor identifier;
* (L_i) = textual label or mention;
* (N_i) = normalised actor identity where available;
* (T_i) = actor type;
* (Q_i) = provenance and resolution metadata.

Distinct textual mentions must not automatically be assumed to refer to the same actor.

Likewise, different surface forms may later be resolved to the same actor only through an explicit identity-resolution procedure.

### Groups

Let (G_u = {g_1, g_2, \ldots, g_m}) denote the social groups represented in textual unit (u).

A group may represent a political, institutional, national, professional, ideological, organisational or other socially meaningful collective category defined by the research ontology.

Conceptually, a group record may be represented as (g_j = (ID_j, L_j, T_j, Q_j)), where:

* (ID_j) = stable group identifier;
* (L_j) = textual or normalised group label;
* (T_j) = group type;
* (Q_j) = provenance metadata.

Group identification must remain traceable to textual evidence or an explicitly defined later inference mechanism.

### Overlapping Membership

Media AI permits overlapping social membership.

For actor (a_i) and group (g_j), membership may be represented as (m_{ij}).

The membership structure is (M_u = {m_{ij}}).

At its simplest, observed membership may satisfy (m_{ij} \in {0,1}).

Dictionary:

* (m_{ij}=1) = actor (a_i) is represented as belonging to group (g_j);
* (m_{ij}=0) = actor (a_i) is explicitly represented as not belonging to group (g_j), where such a negative observation is available.

Unavailable membership information must remain distinct from observed non-membership.

An actor may satisfy (m_{ij}=1) for more than one group.

Therefore, in general, (\sum_j m_{ij} \geq 1) is permitted.

Media AI does not impose mutually exclusive group membership unless a later research design explicitly requires it.

### Relationships

Let (R_u = {r_1, r_2, \ldots, r_k}) denote the set of social relationships represented in textual unit (u).

A relationship may be represented as (r_k = (s_k, p_k, o_k, q_k)), where:

* (s_k) = relationship subject;
* (p_k) = relationship type or predicate;
* (o_k) = relationship object;
* (q_k) = provenance and relationship metadata.

Examples of relationship structure may include support, opposition, conflict, cooperation, attribution, authority, affiliation or another relation defined by the social ontology.

The relation set is extensible, but relationship types must be explicit rather than inferred from an undocumented numerical convention.

### Directionality

Relationships may be directed or undirected.

For a directed relation, (r(a_i,a_j) \neq r(a_j,a_i)) may hold.

For a symmetric relation, (r(a_i,a_j) = r(a_j,a_i)) may hold.

Directionality is therefore a property of the relationship type and must not be assumed globally.

### Signed Relationships

Where the ontology permits relational orientation, a relationship may additionally contain a signed value (\rho_k).

For signed relationships, (\rho_k \in [-1,1]).

Dictionary:

* (\rho_k<0) = negatively oriented relationship;
* (\rho_k=0) = neutral or non-oriented relationship where zero is substantively defined;
* (\rho_k>0) = positively oriented relationship.

A signed relationship value is not required for every relationship type.

Missing orientation must not be represented as substantive zero.

### Social Value States

Social attributes must preserve semantic value state in the same way as factual and psychological representations.

For social attribute (j), (X_{u,j}^{S} = (v_{u,j}^{S}, s_{u,j}^{S})), where:

* (v_{u,j}^{S}) = social value or structured object;
* (s_{u,j}^{S}) = semantic value state.

Permitted states remain (s_{u,j}^{S} \in {\text{observed}, \text{insufficient_evidence}, \text{not_applicable}, \text{unavailable}, \text{derived}}).

Therefore, (\text{missing} \neq 0) also applies to the social space.

### Variable Dimensionality

The number of actors, groups, memberships and relationships depends on the textual unit.

Thus (|A_u|), (|G_u|), (|M_u|) and (|R_u|) may vary between sentences, paragraphs and articles.

Consequently, there is no fixed (d_S) analogous to (d_F=9) or (d_P=34).

The social space is therefore:

(S_u \notin \mathbb{R}^{d_S})

for any globally fixed (d_S) under the current explicit ontology.

A later model may construct fixed-dimensional learned social embeddings, but such embeddings must remain distinct from the explicit structured social representation.

### Operational Social Representation

For computational use, the operational social representation is (\mathcal{S}_u = (S_u, Z_u^{S}, Q_u^{S})), where:

* (S_u) = structured social representation;
* (Z_u^{S}) = representation source;
* (Q_u^{S}) = provenance metadata.

The operational representation must preserve the identities of actors, groups, memberships and relationships rather than reducing them prematurely to aggregate counts.

### Representation Source

Every social representation must identify how it was produced.

Let (Z_u^{S} \in {\text{observed}, \text{human_reviewed}, \text{model_predicted}, \text{deterministic}, \text{resolved}, \text{aggregated}}).

Dictionary:

* `observed` = directly supported by the available textual annotation or structured source;
* `human_reviewed` = established or confirmed through human review;
* `model_predicted` = generated by a later Media AI model;
* `deterministic` = produced through an explicitly defined deterministic procedure;
* `resolved` = produced through explicit actor or group identity resolution;
* `aggregated` = constructed from lower-level social representations.

Different representation sources must remain distinguishable.

### Sentence-Level Social Representation

The initial social representation unit is the sentence.

For sentence (s), (\mathcal{S}_s = (S_s, Z_s^{S}, Q_s^{S})).

Where usable social information is unavailable, the sentence-level record must preserve that state explicitly rather than creating fictitious actors, groups or relationships.

An empty observed social structure and an unavailable social structure are different states.

Thus (S_s = \varnothing) with observed status means that no relevant social structure was identified, whereas unavailable status means that the social structure has not yet been established.

### Identity Resolution Boundary

Actor and group identity resolution is separate from mention detection.

If textual mentions (x_1) and (x_2) occur in the corpus, the architecture must not assume (Identity(x_1)=Identity(x_2)) merely because the strings are similar.

Likewise, identical strings do not guarantee identical entities.

A later resolution function may be written as (a^{*} = R_A(x_1,x_2,\ldots,x_n)), where (R_A) is an explicitly defined actor-resolution procedure.

Identity resolution must preserve the original textual mentions so that resolved social objects remain traceable back to the source text.

### Hierarchical Social Representation

The Media AI hierarchy permits social representations at four levels:

(S^{token} \rightarrow S^{sentence} \rightarrow S^{paragraph} \rightarrow S^{article}).

Higher-level social representation is not obtained by numerical averaging.

For lower-level social structures (S_1^{lower}, \ldots, S_n^{lower}), hierarchical social aggregation may later be expressed as (S^{higher} = A_S(S_1^{lower}, S_2^{lower}, \ldots, S_n^{lower})).

The function (A_S) may involve operations such as:

* actor union with identity preservation;
* group union;
* membership reconciliation;
* relationship accumulation;
* repeated-relation accounting;
* provenance preservation.

The exact aggregation policy is defined in the hierarchical aggregation block.

### Social Polarisation Distance

The social representation provides the structural basis for later Social Polarisation Distance analysis.

Let (g_i) and (g_j) denote two social groups.

A later Social Polarisation Distance may be represented as (D_{SPD}(g_i,g_j)).

The distance may depend on explicitly defined relational, representational or state differences between groups.

At this stage, no specific distance function is assumed.

Therefore (D_{SPD}) is a methodological contract rather than a calculated quantity.

Its later definition must specify:

* the compared social objects;
* the representation space used;
* the distance or divergence function;
* treatment of overlapping membership;
* treatment of missing information;
* temporal level;
* aggregation level.

This prevents Social Polarisation Distance from becoming an undocumented generic similarity score.

### Confluent Representation

The factual, psychological and social spaces remain distinct:

(F_u \neq P_u \neq S_u).

However, Media AI requires a mechanism through which information from these spaces can interact.

The confluent mechanism is represented as (\Phi_C).

For textual unit (u), a later confluent representation may be written as (C_u = \Phi_C(F_u, P_u, S_u)).

Dictionary:

* (F_u) = factual representation;
* (P_u) = psychological representation;
* (S_u) = social representation;
* (\Phi_C) = confluent transformation;
* (C_u) = resulting confluent representation.

The confluent operation must not imply that the three source spaces are semantically interchangeable.

### Confluence Is Not Concatenation by Definition

Simple concatenation is one possible computational baseline, but it is not the definition of confluence.

Therefore, in general:

(C_u \neq [F_u;P_u;S_u]).

A later implementation of (\Phi_C) may use concatenation, attention, gating, cross-space interaction, graph conditioning or another explicitly defined mechanism.

The selected mechanism must be documented and experimentally testable.

### Variable Social Input to Confluence

Because (S_u) is variable-dimensional, the social representation cannot be inserted directly into a fixed-dimensional confluent vector without an explicit transformation.

A later social encoder may therefore construct (H_u^{S} = \Phi_S(S_u)), where:

* (S_u) = explicit structured social representation;
* (\Phi_S) = social encoding function;
* (H_u^{S}) = learned or derived fixed-dimensional social representation.

Confluence may then operate conceptually as (C_u = \Phi_C(H_u^{F}, H_u^{P}, H_u^{S})).

The explicit representations and learned representations must remain distinguishable.

### Missing-Space Boundary

Confluence must not silently interpret an unavailable representation space as a zero vector.

If one or more spaces are unavailable, the confluent mechanism must receive explicit availability information.

Let (M_u^{space} = (m_u^{F},m_u^{P},m_u^{S})), where each component indicates whether the corresponding representation space is available for the required operation.

Thus (m_u^{F}=0), for example, means that factual representation is unavailable for that operation; it does not mean that the factual state is substantively zero.

Any confluent implementation must therefore be mask-aware.

### Temporal Boundary

Where confluence operates on sequential representations, it must preserve the configured temporal information boundary.

At textual step (t), the permitted information set remains (\mathcal{I}_t = {X_1,X_2,\ldots,X_t}).

Future textual units must not influence an earlier confluent state when preceding-context processing is required.

### Recurrent-Confluent Boundary

Confluence and recurrence are separate architectural mechanisms.

Confluence describes interaction across representation spaces.

Recurrence describes state propagation across textual sequence.

Conceptually, a later architecture may use (C_t = \Phi_C(F_t,P_t,S_t)) followed by (H_t = \Phi_R(H_{t-1},C_t)).

Alternatively, recurrent processing may occur within individual spaces before confluence.

The ordering of these operations is an architectural hypothesis and must therefore remain explicit rather than being assumed by this block.

### Attribution Requirement

Confluent outputs must remain attributable to both representation spaces and textual evidence.

A later confluent output (C_u) should support backward attribution of the form (C_u \rightarrow {F_u,P_u,S_u} \rightarrow Text).

Where token-level attribution becomes available, this pathway may extend to (C_u \rightarrow Representation\ Space \rightarrow Sentence \rightarrow Token \rightarrow Text).

This enables later interpretation to distinguish, for example, whether a model output is driven primarily by factual content, psychological framing, social structure or an interaction between them.

### Current Social State

The social representation architecture must remain valid even when no structured social annotations or model-derived social objects have yet been constructed.

In that case, sentence-level social representation records may preserve:

* sentence identity;
* representation level;
* unavailable social state;
* actor container;
* group container;
* membership container;
* relationship container;
* representation source;
* provenance.

No actors, groups or relationships should be fabricated merely to satisfy the representation schema.

### Current Confluent State

This block defines the confluent contract but does not calculate confluent representations.

The factual and psychological representation layers may exist structurally while containing unavailable substantive values, and the social representation may likewise remain unavailable.

Consequently, no meaningful confluent state should be generated until the required source representations and availability conditions are satisfied.

### Output

This block defines and, where information is available, constructs the social representation structure and establishes the confluent representation contract.

Its outputs include:

* actor representation;
* group representation;
* overlapping membership structure;
* relationship representation;
* relationship directionality;
* optional signed relational orientation;
* social semantic value states;
* social representation-source metadata;
* sentence-level social representation records;
* identity-resolution boundaries;
* hierarchical social-representation boundaries;
* Social Polarisation Distance contract;
* confluent representation contract;
* social-encoding boundary;
* missing-space handling requirements;
* recurrent-confluent boundary;
* backward-attribution requirements.

Where social information is unavailable, it remains explicitly unavailable rather than being replaced with fabricated empty observations.

No actor or group identity resolution is performed unless explicitly supported by available inputs.

No Social Polarisation Distance is calculated.

No paragraph-level or article-level social aggregation is performed.

No confluent representation is calculated.

No recurrent state is calculated.

No attribution score, model training or experimental evaluation is performed.


````python
# =============================================================================
# Media AI — Social and Confluent Representation
# =============================================================================

from datetime import datetime, timezone

import numpy as np


# -----------------------------------------------------------------------------
# Block dependency checks
# -----------------------------------------------------------------------------

required_block_1_objects = [
    "BUILD_SOCIAL_REPRESENTATION",
    "BUILD_CONFLUENT_REPRESENTATION",
    "PRESERVE_SUPERVISION_VALUE_STATES",
    "ALLOW_MISSING_SUPERVISION_AS_ZERO",
    "PRESERVE_SUPERVISION_PROVENANCE",
    "ALLOW_FUTURE_CONTEXT_IN_STATE_CONSTRUCTION",
]

missing_block_1_objects = [
    object_name
    for object_name in required_block_1_objects
    if object_name not in globals()
]

if missing_block_1_objects:
    raise NameError(
        "Notebook 04 Block 1 is not initialised. "
        f"Missing: {missing_block_1_objects}"
    )


required_block_2_objects = [
    "SENTENCE_IDS",
    "CANONICAL_SUPERVISION_BY_SENTENCE_ID",
    "CANONICAL_SUPERVISION_AVAILABLE",
]

missing_block_2_objects = [
    object_name
    for object_name in required_block_2_objects
    if object_name not in globals()
]

if missing_block_2_objects:
    raise NameError(
        "Notebook 04 Block 2 is not initialised. "
        f"Missing: {missing_block_2_objects}"
    )


required_block_3_objects = [
    "HIERARCHICAL_TEXT_AND_IDENTITY_VALID",
    "SENTENCE_BY_ID",
    "SENTENCE_TO_PARAGRAPH_ID",
    "SENTENCE_TO_ARTICLE_ID",
    "ARTICLE_SENTENCE_ORDER",
]

missing_block_3_objects = [
    object_name
    for object_name in required_block_3_objects
    if object_name not in globals()
]

if missing_block_3_objects:
    raise NameError(
        "Notebook 04 Block 3 is not initialised. "
        f"Missing: {missing_block_3_objects}"
    )


required_block_4_objects = [
    "MEDIA_AI_REPRESENTATION_SPECIFICATION_VALID",
    "MEDIA_AI_VALUE_STATES",
]

missing_block_4_objects = [
    object_name
    for object_name in required_block_4_objects
    if object_name not in globals()
]

if missing_block_4_objects:
    raise NameError(
        "Notebook 04 Block 4 is not initialised. "
        f"Missing: {missing_block_4_objects}"
    )


required_block_6_objects = [
    "FACTUAL_SENTENCE_REPRESENTATIONS",
    "FACTUAL_SENTENCE_BY_ID",
    "FACTUAL_REPRESENTATION_VALID",
]

missing_block_6_objects = [
    object_name
    for object_name in required_block_6_objects
    if object_name not in globals()
]

if missing_block_6_objects:
    raise NameError(
        "Notebook 04 Block 6 is not initialised. "
        f"Missing: {missing_block_6_objects}"
    )


required_block_7_objects = [
    "PSYCHOLOGICAL_SENTENCE_REPRESENTATIONS",
    "PSYCHOLOGICAL_SENTENCE_BY_ID",
    "PSYCHOLOGICAL_REPRESENTATION_VALID",
]

missing_block_7_objects = [
    object_name
    for object_name in required_block_7_objects
    if object_name not in globals()
]

if missing_block_7_objects:
    raise NameError(
        "Notebook 04 Block 7 is not initialised. "
        f"Missing: {missing_block_7_objects}"
    )


if not HIERARCHICAL_TEXT_AND_IDENTITY_VALID:
    raise ValueError(
        "Social representation requires a valid hierarchical "
        "text and identity structure."
    )


if not MEDIA_AI_REPRESENTATION_SPECIFICATION_VALID:
    raise ValueError(
        "Social representation requires a valid Media AI "
        "representation specification."
    )


if not FACTUAL_REPRESENTATION_VALID:
    raise ValueError(
        "Confluent contract requires a valid factual representation layer."
    )


if not PSYCHOLOGICAL_REPRESENTATION_VALID:
    raise ValueError(
        "Confluent contract requires a valid psychological "
        "representation layer."
    )


if ALLOW_MISSING_SUPERVISION_AS_ZERO:
    raise ValueError(
        "Missing social information must not be converted "
        "to substantive zero."
    )


if ALLOW_FUTURE_CONTEXT_IN_STATE_CONSTRUCTION:
    raise ValueError(
        "Future context must not enter sequential social or "
        "confluent state construction."
    )


# -----------------------------------------------------------------------------
# Social representation configuration
# -----------------------------------------------------------------------------

SOCIAL_REPRESENTATION_VERSION = "1.0"
CONFLUENT_CONTRACT_VERSION = "1.0"

SOCIAL_REPRESENTATION_LEVEL = "sentence"
SOCIAL_REPRESENTATION_SPACE = "social"

SOCIAL_STRUCTURE_COMPONENTS = (
    "actors",
    "groups",
    "memberships",
    "relationships",
)

SOCIAL_STRUCTURE_COMPONENT_COUNT = len(
    SOCIAL_STRUCTURE_COMPONENTS
)

SOCIAL_FIXED_DIMENSIONAL = False
SOCIAL_OVERLAPPING_MEMBERSHIP_ALLOWED = True

SOCIAL_RELATION_DIRECTIONALITY_ALLOWED = True
SOCIAL_SIGNED_RELATIONSHIPS_ALLOWED = True
SOCIAL_RELATIONAL_ORIENTATION_RANGE = (
    -1.0,
    1.0,
)

SOCIAL_REPRESENTATION_SOURCE_UNAVAILABLE = "unavailable"
SOCIAL_REPRESENTATION_SOURCE_OBSERVED = "observed"
SOCIAL_REPRESENTATION_SOURCE_HUMAN = "human_reviewed"
SOCIAL_REPRESENTATION_SOURCE_MODEL = "model_predicted"
SOCIAL_REPRESENTATION_SOURCE_DETERMINISTIC = "deterministic"
SOCIAL_REPRESENTATION_SOURCE_RESOLVED = "resolved"
SOCIAL_REPRESENTATION_SOURCE_AGGREGATED = "aggregated"

SOCIAL_REPRESENTATION_SOURCES = (
    SOCIAL_REPRESENTATION_SOURCE_UNAVAILABLE,
    SOCIAL_REPRESENTATION_SOURCE_OBSERVED,
    SOCIAL_REPRESENTATION_SOURCE_HUMAN,
    SOCIAL_REPRESENTATION_SOURCE_MODEL,
    SOCIAL_REPRESENTATION_SOURCE_DETERMINISTIC,
    SOCIAL_REPRESENTATION_SOURCE_RESOLVED,
    SOCIAL_REPRESENTATION_SOURCE_AGGREGATED,
)


# -----------------------------------------------------------------------------
# Social structure constructors
# -----------------------------------------------------------------------------

def create_empty_social_structure():
    """
    Create an empty social structure container.

    The container itself does not imply that the social structure
    has been observed. Availability is stored separately.
    """

    return {
        "actors": [],
        "groups": [],
        "memberships": [],
        "relationships": [],
    }


def create_social_component_states(
    state="unavailable",
):
    """
    Create semantic states for the four social structure components.
    """

    if state not in MEDIA_AI_VALUE_STATES:
        raise ValueError(
            f"Invalid social value state: {state}"
        )

    return {
        component: state
        for component in SOCIAL_STRUCTURE_COMPONENTS
    }


def create_social_availability_mask(
    available=False,
):
    """
    Create component-level availability indicators.

    These indicators describe availability, not substantive social values.
    """

    value = 1 if available else 0

    return {
        component: value
        for component in SOCIAL_STRUCTURE_COMPONENTS
    }


# -----------------------------------------------------------------------------
# Social provenance
# -----------------------------------------------------------------------------

def build_social_provenance(
    sentence_id,
    representation_source,
    construction_method,
):
    """
    Build provenance for one sentence-level social representation.
    """

    return {
        "media_ai_record_id":
            SENTENCE_TO_ARTICLE_ID[
                sentence_id
            ],

        "paragraph_id":
            SENTENCE_TO_PARAGRAPH_ID[
                sentence_id
            ],

        "sentence_id":
            sentence_id,

        "representation_level":
            SOCIAL_REPRESENTATION_LEVEL,

        "representation_space":
            SOCIAL_REPRESENTATION_SPACE,

        "representation_source":
            representation_source,

        "representation_schema_version":
            REPRESENTATION_SCHEMA_VERSION,

        "social_representation_version":
            SOCIAL_REPRESENTATION_VERSION,

        "construction_method":
            construction_method,

        "identity_resolution_performed":
            False,

        "social_inference_performed":
            False,

        "aggregation_performed":
            False,

        "constructed_at_utc":
            datetime.now(
                timezone.utc
            ).isoformat(),
    }


# -----------------------------------------------------------------------------
# Sentence-level social representation construction
# -----------------------------------------------------------------------------

SOCIAL_SENTENCE_REPRESENTATIONS = []


for sentence_id in SENTENCE_IDS:

    # Notebook 03 currently contains no validated structured social
    # annotation contract. Therefore, no actor, group, membership or
    # relationship is fabricated from the text in this block.
    #
    # An empty container with unavailable state is deliberately distinct
    # from an observed empty social structure.

    social_structure = (
        create_empty_social_structure()
    )

    social_value_states = (
        create_social_component_states(
            state="unavailable"
        )
    )

    social_availability_mask = (
        create_social_availability_mask(
            available=False
        )
    )

    representation_source = (
        SOCIAL_REPRESENTATION_SOURCE_UNAVAILABLE
    )

    social_record = {
        "record_type":
            "media_ai_social_representation",

        "representation_version":
            SOCIAL_REPRESENTATION_VERSION,

        "media_ai_record_id":
            SENTENCE_TO_ARTICLE_ID[
                sentence_id
            ],

        "paragraph_id":
            SENTENCE_TO_PARAGRAPH_ID[
                sentence_id
            ],

        "sentence_id":
            sentence_id,

        "representation_level":
            SOCIAL_REPRESENTATION_LEVEL,

        "representation_space":
            SOCIAL_REPRESENTATION_SPACE,

        "representation_source":
            representation_source,

        "structure":
            social_structure,

        "availability_mask":
            social_availability_mask,

        "value_states":
            social_value_states,

        "provenance":
            build_social_provenance(
                sentence_id=
                    sentence_id,

                representation_source=
                    representation_source,

                construction_method=
                    "unavailable_social_structure_placeholder",
            ),
    }

    SOCIAL_SENTENCE_REPRESENTATIONS.append(
        social_record
    )


# -----------------------------------------------------------------------------
# Sentence-level social lookup
# -----------------------------------------------------------------------------

SOCIAL_SENTENCE_BY_ID = {
    record["sentence_id"]: record
    for record in SOCIAL_SENTENCE_REPRESENTATIONS
}


# -----------------------------------------------------------------------------
# Social representation validation
# -----------------------------------------------------------------------------

SOCIAL_REPRESENTATION_ERRORS = []


for record in SOCIAL_SENTENCE_REPRESENTATIONS:

    structure = record.get(
        "structure"
    )

    mask = record.get(
        "availability_mask"
    )

    states = record.get(
        "value_states"
    )

    source = record.get(
        "representation_source"
    )


    if not isinstance(
        structure,
        dict,
    ):
        SOCIAL_REPRESENTATION_ERRORS.append(
            "social_structure_not_dictionary"
        )
        break


    if set(
        structure.keys()
    ) != set(
        SOCIAL_STRUCTURE_COMPONENTS
    ):
        SOCIAL_REPRESENTATION_ERRORS.append(
            "invalid_social_structure_components"
        )
        break


    for component in SOCIAL_STRUCTURE_COMPONENTS:

        if not isinstance(
            structure[
                component
            ],
            list,
        ):
            SOCIAL_REPRESENTATION_ERRORS.append(
                f"{component}_not_list"
            )
            break

    if SOCIAL_REPRESENTATION_ERRORS:
        break


    if not isinstance(
        mask,
        dict,
    ):
        SOCIAL_REPRESENTATION_ERRORS.append(
            "social_mask_not_dictionary"
        )
        break


    if set(
        mask.keys()
    ) != set(
        SOCIAL_STRUCTURE_COMPONENTS
    ):
        SOCIAL_REPRESENTATION_ERRORS.append(
            "invalid_social_mask_components"
        )
        break


    if not all(
        mask[
            component
        ] in (
            0,
            1,
        )
        for component in SOCIAL_STRUCTURE_COMPONENTS
    ):
        SOCIAL_REPRESENTATION_ERRORS.append(
            "invalid_social_mask_values"
        )
        break


    if not isinstance(
        states,
        dict,
    ):
        SOCIAL_REPRESENTATION_ERRORS.append(
            "social_states_not_dictionary"
        )
        break


    if set(
        states.keys()
    ) != set(
        SOCIAL_STRUCTURE_COMPONENTS
    ):
        SOCIAL_REPRESENTATION_ERRORS.append(
            "invalid_social_state_components"
        )
        break


    for component in SOCIAL_STRUCTURE_COMPONENTS:

        state = states[
            component
        ]

        if state not in MEDIA_AI_VALUE_STATES:
            SOCIAL_REPRESENTATION_ERRORS.append(
                "invalid_social_value_state"
            )
            break


        if state == "observed":

            if mask[
                component
            ] != 1:
                SOCIAL_REPRESENTATION_ERRORS.append(
                    "observed_social_component_without_mask"
                )
                break

        else:

            if mask[
                component
            ] != 0:
                SOCIAL_REPRESENTATION_ERRORS.append(
                    "unavailable_social_component_with_active_mask"
                )
                break

    if SOCIAL_REPRESENTATION_ERRORS:
        break


    if source not in SOCIAL_REPRESENTATION_SOURCES:
        SOCIAL_REPRESENTATION_ERRORS.append(
            "invalid_social_representation_source"
        )
        break


    if (
        source
        == SOCIAL_REPRESENTATION_SOURCE_UNAVAILABLE
    ):

        if any(
            mask.values()
        ):
            SOCIAL_REPRESENTATION_ERRORS.append(
                "unavailable_social_source_has_active_mask"
            )
            break


        if any(
            len(
                structure[
                    component
                ]
            ) > 0
            for component in SOCIAL_STRUCTURE_COMPONENTS
        ):
            SOCIAL_REPRESENTATION_ERRORS.append(
                "unavailable_social_source_has_fabricated_structure"
            )
            break


# -----------------------------------------------------------------------------
# Social object accounting
# -----------------------------------------------------------------------------

SOCIAL_ACTOR_COUNT = sum(
    len(
        record[
            "structure"
        ][
            "actors"
        ]
    )
    for record in SOCIAL_SENTENCE_REPRESENTATIONS
)


SOCIAL_GROUP_COUNT = sum(
    len(
        record[
            "structure"
        ][
            "groups"
        ]
    )
    for record in SOCIAL_SENTENCE_REPRESENTATIONS
)


SOCIAL_MEMBERSHIP_COUNT = sum(
    len(
        record[
            "structure"
        ][
            "memberships"
        ]
    )
    for record in SOCIAL_SENTENCE_REPRESENTATIONS
)


SOCIAL_RELATIONSHIP_COUNT = sum(
    len(
        record[
            "structure"
        ][
            "relationships"
        ]
    )
    for record in SOCIAL_SENTENCE_REPRESENTATIONS
)


# -----------------------------------------------------------------------------
# Social availability accounting
# -----------------------------------------------------------------------------

SOCIAL_AVAILABLE_SENTENCE_COUNT = sum(
    1
    for record in SOCIAL_SENTENCE_REPRESENTATIONS
    if any(
        record[
            "availability_mask"
        ].values()
    )
)


SOCIAL_UNAVAILABLE_SENTENCE_COUNT = (
    len(
        SOCIAL_SENTENCE_REPRESENTATIONS
    )
    - SOCIAL_AVAILABLE_SENTENCE_COUNT
)


SOCIAL_SENTENCE_AVAILABILITY_RATE = (
    SOCIAL_AVAILABLE_SENTENCE_COUNT
    / len(
        SOCIAL_SENTENCE_REPRESENTATIONS
    )
    if SOCIAL_SENTENCE_REPRESENTATIONS
    else 0.0
)


# -----------------------------------------------------------------------------
# Article-level social availability
# -----------------------------------------------------------------------------

SOCIAL_ARTICLE_COVERAGE = []


for (
    article_id,
    sentence_ids,
) in ARTICLE_SENTENCE_ORDER.items():

    available_count = sum(
        1
        for sentence_id in sentence_ids
        if any(
            SOCIAL_SENTENCE_BY_ID[
                sentence_id
            ][
                "availability_mask"
            ].values()
        )
    )


    sentence_count = len(
        sentence_ids
    )


    coverage_fraction = (
        available_count
        / sentence_count
        if sentence_count
        else 0.0
    )


    if available_count == 0:
        coverage_status = "unavailable"

    elif available_count == sentence_count:
        coverage_status = "fully_available"

    else:
        coverage_status = "partially_available"


    SOCIAL_ARTICLE_COVERAGE.append({
        "media_ai_record_id":
            article_id,

        "sentence_count":
            sentence_count,

        "available_sentence_count":
            available_count,

        "coverage_fraction":
            coverage_fraction,

        "coverage_status":
            coverage_status,
    })


# -----------------------------------------------------------------------------
# Social identity-resolution contract
# -----------------------------------------------------------------------------

SOCIAL_IDENTITY_RESOLUTION_CONTRACT = {
    "mention_detection_separate":
        True,

    "actor_resolution_separate":
        True,

    "group_resolution_separate":
        True,

    "surface_form_identity_assumed":
        False,

    "original_mentions_preserved":
        True,

    "overlapping_membership_allowed":
        SOCIAL_OVERLAPPING_MEMBERSHIP_ALLOWED,

    "identity_resolution_performed":
        False,
}


# -----------------------------------------------------------------------------
# Social relationship contract
# -----------------------------------------------------------------------------

SOCIAL_RELATIONSHIP_CONTRACT = {
    "directionality_supported":
        SOCIAL_RELATION_DIRECTIONALITY_ALLOWED,

    "signed_orientation_supported":
        SOCIAL_SIGNED_RELATIONSHIPS_ALLOWED,

    "signed_orientation_minimum":
        SOCIAL_RELATIONAL_ORIENTATION_RANGE[
            0
        ],

    "signed_orientation_maximum":
        SOCIAL_RELATIONAL_ORIENTATION_RANGE[
            1
        ],

    "missing_orientation_equals_zero":
        False,

    "relationship_types_explicit":
        True,
}


# -----------------------------------------------------------------------------
# Social Polarisation Distance contract
# -----------------------------------------------------------------------------

SOCIAL_POLARISATION_DISTANCE_CONTRACT = {
    "defined":
        True,

    "calculated":
        False,

    "group_identity_required":
        True,

    "representation_space_required":
        True,

    "distance_function_required":
        True,

    "overlapping_membership_policy_required":
        True,

    "missing_information_policy_required":
        True,

    "temporal_level_required":
        True,

    "aggregation_level_required":
        True,
}


# -----------------------------------------------------------------------------
# Confluent representation contract
# -----------------------------------------------------------------------------

CONFLUENT_SOURCE_SPACES = (
    "factual",
    "psychological",
    "social",
)


CONFLUENT_REPRESENTATION_CONTRACT = {
    "version":
        CONFLUENT_CONTRACT_VERSION,

    "source_spaces":
        CONFLUENT_SOURCE_SPACES,

    "source_spaces_remain_distinct":
        True,

    "simple_concatenation_is_definition":
        False,

    "mask_aware":
        True,

    "missing_space_equals_zero":
        False,

    "variable_social_input":
        True,

    "social_encoder_required_for_fixed_vector":
        True,

    "temporal_boundary_required":
        True,

    "future_context_allowed":
        False,

    "recurrent_mechanism_separate":
        True,

    "backward_attribution_required":
        True,

    "calculated":
        False,
}


# -----------------------------------------------------------------------------
# Sentence-level confluent readiness interface
# -----------------------------------------------------------------------------

CONFLUENT_SENTENCE_READINESS = []


for sentence_id in SENTENCE_IDS:

    factual_record = (
        FACTUAL_SENTENCE_BY_ID[
            sentence_id
        ]
    )

    psychological_record = (
        PSYCHOLOGICAL_SENTENCE_BY_ID[
            sentence_id
        ]
    )

    social_record = (
        SOCIAL_SENTENCE_BY_ID[
            sentence_id
        ]
    )


    factual_available = bool(
        np.any(
            factual_record[
                "availability_mask"
            ]
        )
    )


    psychological_available = bool(
        np.any(
            psychological_record[
                "availability_mask"
            ]
        )
    )


    social_available = bool(
        any(
            social_record[
                "availability_mask"
            ].values()
        )
    )


    space_availability = {
        "factual":
            factual_available,

        "psychological":
            psychological_available,

        "social":
            social_available,
    }


    all_spaces_available = all(
        space_availability.values()
    )


    any_space_available = any(
        space_availability.values()
    )


    CONFLUENT_SENTENCE_READINESS.append({
        "media_ai_record_id":
            SENTENCE_TO_ARTICLE_ID[
                sentence_id
            ],

        "paragraph_id":
            SENTENCE_TO_PARAGRAPH_ID[
                sentence_id
            ],

        "sentence_id":
            sentence_id,

        "space_availability":
            space_availability,

        "all_spaces_available":
            all_spaces_available,

        "any_space_available":
            any_space_available,

        "confluent_representation_constructed":
            False,
    })


# -----------------------------------------------------------------------------
# Confluent readiness accounting
# -----------------------------------------------------------------------------

CONFLUENT_FULLY_READY_SENTENCE_COUNT = sum(
    1
    for record in CONFLUENT_SENTENCE_READINESS
    if record[
        "all_spaces_available"
    ]
)


CONFLUENT_PARTIALLY_READY_SENTENCE_COUNT = sum(
    1
    for record in CONFLUENT_SENTENCE_READINESS
    if (
        record[
            "any_space_available"
        ]
        and not record[
            "all_spaces_available"
        ]
    )
)


CONFLUENT_UNAVAILABLE_SENTENCE_COUNT = sum(
    1
    for record in CONFLUENT_SENTENCE_READINESS
    if not record[
        "any_space_available"
    ]
)


# -----------------------------------------------------------------------------
# Recurrent-confluent boundary
# -----------------------------------------------------------------------------

RECURRENT_CONFLUENT_BOUNDARY = {
    "confluence_defined_as_cross_space_interaction":
        True,

    "recurrence_defined_as_sequence_propagation":
        True,

    "mechanisms_equivalent":
        False,

    "operation_order_fixed":
        False,

    "operation_order_is_architectural_hypothesis":
        True,
}


# -----------------------------------------------------------------------------
# Attribution contract
# -----------------------------------------------------------------------------

CONFLUENT_ATTRIBUTION_CONTRACT = {
    "required":
        True,

    "space_level_attribution":
        True,

    "sentence_level_attribution":
        True,

    "token_level_attribution_when_available":
        True,

    "required_backward_path":
        (
            "confluent_output"
            " -> representation_space"
            " -> sentence"
            " -> text"
        ),

    "scores_calculated":
        False,
}


# -----------------------------------------------------------------------------
# Representation status
# -----------------------------------------------------------------------------

SOCIAL_REPRESENTATION_VALID = (
    len(
        SOCIAL_REPRESENTATION_ERRORS
    )
    == 0
)


if not SOCIAL_REPRESENTATION_VALID:
    raise ValueError(
        "Social representation failed validation: "
        f"{SOCIAL_REPRESENTATION_ERRORS}"
    )


SOCIAL_REPRESENTATIONS_CONSTRUCTED = True


SOCIAL_REPRESENTATION_HAS_OBSERVED_STRUCTURE = (
    SOCIAL_AVAILABLE_SENTENCE_COUNT
    > 0
)


CONFLUENT_CONTRACT_VALID = all([
    CONFLUENT_REPRESENTATION_CONTRACT[
        "source_spaces_remain_distinct"
    ],
    CONFLUENT_REPRESENTATION_CONTRACT[
        "mask_aware"
    ],
    not CONFLUENT_REPRESENTATION_CONTRACT[
        "missing_space_equals_zero"
    ],
    CONFLUENT_REPRESENTATION_CONTRACT[
        "recurrent_mechanism_separate"
    ],
    CONFLUENT_REPRESENTATION_CONTRACT[
        "backward_attribution_required"
    ],
])


CONFLUENT_REPRESENTATIONS_CONSTRUCTED = False


# -----------------------------------------------------------------------------
# Summary
# -----------------------------------------------------------------------------

SOCIAL_AND_CONFLUENT_SUMMARY = {
    "social_representation_version":
        SOCIAL_REPRESENTATION_VERSION,

    "confluent_contract_version":
        CONFLUENT_CONTRACT_VERSION,

    "social_fixed_dimensional":
        SOCIAL_FIXED_DIMENSIONAL,

    "social_sentence_records":
        len(
            SOCIAL_SENTENCE_REPRESENTATIONS
        ),

    "social_available_sentences":
        SOCIAL_AVAILABLE_SENTENCE_COUNT,

    "social_unavailable_sentences":
        SOCIAL_UNAVAILABLE_SENTENCE_COUNT,

    "social_availability_rate":
        SOCIAL_SENTENCE_AVAILABILITY_RATE,

    "actors":
        SOCIAL_ACTOR_COUNT,

    "groups":
        SOCIAL_GROUP_COUNT,

    "memberships":
        SOCIAL_MEMBERSHIP_COUNT,

    "relationships":
        SOCIAL_RELATIONSHIP_COUNT,

    "social_representation_valid":
        SOCIAL_REPRESENTATION_VALID,

    "social_representations_constructed":
        SOCIAL_REPRESENTATIONS_CONSTRUCTED,

    "social_observed_structure_present":
        SOCIAL_REPRESENTATION_HAS_OBSERVED_STRUCTURE,

    "confluent_contract_valid":
        CONFLUENT_CONTRACT_VALID,

    "confluent_fully_ready_sentences":
        CONFLUENT_FULLY_READY_SENTENCE_COUNT,

    "confluent_partially_ready_sentences":
        CONFLUENT_PARTIALLY_READY_SENTENCE_COUNT,

    "confluent_unavailable_sentences":
        CONFLUENT_UNAVAILABLE_SENTENCE_COUNT,

    "confluent_representations_constructed":
        CONFLUENT_REPRESENTATIONS_CONSTRUCTED,

    "social_polarisation_distance_calculated":
        False,
}


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Social and Confluent Representation")
print("=" * 72)

print(
    f"Social representation ver.  : "
    f"{SOCIAL_REPRESENTATION_VERSION}"
)

print(
    f"Confluent contract version  : "
    f"{CONFLUENT_CONTRACT_VERSION}"
)

print(
    f"Representation level        : "
    f"{SOCIAL_REPRESENTATION_LEVEL}"
)

print(
    f"Social structure components : "
    f"{SOCIAL_STRUCTURE_COMPONENT_COUNT}"
)

print(
    f"Social fixed dimensional    : "
    f"{SOCIAL_FIXED_DIMENSIONAL}"
)

print(
    f"Sentence records            : "
    f"{len(SOCIAL_SENTENCE_REPRESENTATIONS)}"
)

print("-" * 72)

print("Social structure")

print(
    f"Available sentences         : "
    f"{SOCIAL_AVAILABLE_SENTENCE_COUNT}"
)

print(
    f"Unavailable sentences       : "
    f"{SOCIAL_UNAVAILABLE_SENTENCE_COUNT}"
)

print(
    f"Availability rate           : "
    f"{SOCIAL_SENTENCE_AVAILABILITY_RATE:.4f}"
)

print(
    f"Actors                      : "
    f"{SOCIAL_ACTOR_COUNT}"
)

print(
    f"Groups                      : "
    f"{SOCIAL_GROUP_COUNT}"
)

print(
    f"Memberships                 : "
    f"{SOCIAL_MEMBERSHIP_COUNT}"
)

print(
    f"Relationships               : "
    f"{SOCIAL_RELATIONSHIP_COUNT}"
)

print(
    f"Observed structure present  : "
    f"{SOCIAL_REPRESENTATION_HAS_OBSERVED_STRUCTURE}"
)

print("-" * 72)

print("Article social availability")

for article_record in SOCIAL_ARTICLE_COVERAGE:

    print(
        f"{article_record['media_ai_record_id']}"
    )

    print(
        f"  Sentences                 : "
        f"{article_record['sentence_count']}"
    )

    print(
        f"  Available                 : "
        f"{article_record['available_sentence_count']}"
    )

    print(
        f"  Coverage                  : "
        f"{article_record['coverage_fraction']:.4f}"
    )

    print(
        f"  Status                    : "
        f"{article_record['coverage_status']}"
    )

print("-" * 72)

print("Social methodological contracts")

print(
    f"Overlapping membership      : "
    f"{SOCIAL_OVERLAPPING_MEMBERSHIP_ALLOWED}"
)

print(
    f"Identity resolution done    : "
    f"{SOCIAL_IDENTITY_RESOLUTION_CONTRACT['identity_resolution_performed']}"
)

print(
    f"Signed relationships        : "
    f"{SOCIAL_SIGNED_RELATIONSHIPS_ALLOWED}"
)

print(
    f"SPD contract defined        : "
    f"{SOCIAL_POLARISATION_DISTANCE_CONTRACT['defined']}"
)

print(
    f"SPD calculated              : "
    f"{SOCIAL_POLARISATION_DISTANCE_CONTRACT['calculated']}"
)

print("-" * 72)

print("Confluent readiness")

print(
    f"Source spaces               : "
    f"{', '.join(CONFLUENT_SOURCE_SPACES)}"
)

print(
    f"Fully ready sentences       : "
    f"{CONFLUENT_FULLY_READY_SENTENCE_COUNT}"
)

print(
    f"Partially ready sentences   : "
    f"{CONFLUENT_PARTIALLY_READY_SENTENCE_COUNT}"
)

print(
    f"Unavailable sentences       : "
    f"{CONFLUENT_UNAVAILABLE_SENTENCE_COUNT}"
)

print(
    f"Mask-aware                  : "
    f"{CONFLUENT_REPRESENTATION_CONTRACT['mask_aware']}"
)

print(
    f"Concatenation is definition : "
    f"{CONFLUENT_REPRESENTATION_CONTRACT['simple_concatenation_is_definition']}"
)

print(
    f"Confluent state calculated  : "
    f"{CONFLUENT_REPRESENTATIONS_CONSTRUCTED}"
)

print("-" * 72)

print("Architectural boundaries")

print(
    f"Confluence = recurrence     : "
    f"{RECURRENT_CONFLUENT_BOUNDARY['mechanisms_equivalent']}"
)

print(
    f"Operation order fixed       : "
    f"{RECURRENT_CONFLUENT_BOUNDARY['operation_order_fixed']}"
)

print(
    f"Future context allowed      : "
    f"{ALLOW_FUTURE_CONTEXT_IN_STATE_CONSTRUCTION}"
)

print(
    f"Backward attribution req.   : "
    f"{CONFLUENT_ATTRIBUTION_CONTRACT['required']}"
)

print(
    f"Social representation valid : "
    f"{SOCIAL_REPRESENTATION_VALID}"
)

print(
    f"Confluent contract valid    : "
    f"{CONFLUENT_CONTRACT_VALID}"
)

print("=" * 72)

print(
    "Sentence-level social representation records constructed successfully."
)

if SOCIAL_REPRESENTATION_HAS_OBSERVED_STRUCTURE:
    print(
        "Available social information has populated structured social records."
    )
else:
    print(
        "Structured social information is currently unavailable."
    )

    print(
        "Social containers therefore remain empty with explicit unavailable states."
    )

print(
    "An unavailable social structure has not been interpreted as "
    "an observed empty social structure."
)

print(
    "No actors, groups, memberships or relationships have been fabricated."
)

print(
    "Actor and group identity resolution has not been performed."
)

print(
    "Overlapping social membership remains permitted by the architecture."
)

print(
    "Social Polarisation Distance has been defined as a contract "
    "but has not been calculated."
)

print(
    "Factual, psychological and social spaces remain distinct."
)

print(
    "Confluence has not been defined as simple concatenation."
)

print(
    "Missing representation spaces are not interpreted as zero vectors."
)

print(
    "No confluent representation or recurrent state has been calculated."
)

print(
    "No paragraph-level or article-level social aggregation has been performed."
)

print(
    "No attribution score, model training or experimental evaluation "
    "has been performed."
)
````

## Transformative State Construction

This block defines the transformative mechanism of the Media AI architecture.

The transformative mechanism represents **the change introduced when a new textual unit is processed in preceding textual context**.

It does not replace the factual, psychological or social representations defined in the preceding blocks. Instead, it establishes an explicit transformation between the representation state before textual step (t) and the representation state associated with the new textual evidence at step (t).

The central methodological principle is (previous\ state \neq transformation \neq resulting\ state).

These three objects must remain distinguishable throughout the architecture.

### Sequential Information Boundary

Let the ordered textual sequence be (X_1, X_2, \ldots, X_T).

At textual step (t), the permitted information set is (\mathcal{I}_t = {X_1, X_2, \ldots, X_t}).

Future textual units are excluded: (X_{t+1}, \ldots, X_T \notin \mathcal{I}_t).

Therefore, any transformative state associated with step (t) must be constructed without access to later article text.

This preserves the same temporal boundary used during independent supervision.

### Representation State

Let the representation state associated with textual step (t) be (Z_t = {F_t, P_t, S_t}).

Dictionary:

* (F_t) = factual representation;
* (P_t) = psychological representation;
* (S_t) = social representation;
* (Z_t) = structured representation state.

The three spaces remain semantically distinct.

The notation (Z_t) therefore describes a structured representation state and does not require a single numerical vector.

Where a confluent representation later exists, it may provide an additional transformed representation of these spaces, but confluence is not required to define the basic transformative contract.

### Previous State

Immediately before processing textual unit (X_t), the architecture has access to the preceding representation state (Z_{t-1}).

This state represents information established through the permitted textual history (X_1, \ldots, X_{t-1}).

For the first textual unit, no empirical preceding textual state exists.

The initial condition must therefore be represented explicitly rather than silently interpreted as a substantive zero state.

### Current Evidence

The current textual unit contributes new representational evidence (E_t = {F_t^{E}, P_t^{E}, S_t^{E}}).

Dictionary:

* (F_t^{E}) = factual evidence associated with (X_t);
* (P_t^{E}) = psychological evidence associated with (X_t);
* (S_t^{E}) = social evidence associated with (X_t);
* (E_t) = structured current-evidence representation.

The superscript (E) distinguishes **current evidence** from an accumulated or recurrent state.

Current evidence may be partially or completely unavailable.

Unavailable evidence must not be replaced with numerical zero.

### Transformative Mechanism

The transformative mechanism is denoted by (\Phi_T).

At textual step (t), the transformation is (\Delta_t = \Phi_T(Z_{t-1}, E_t)).

Dictionary:

* (Z_{t-1}) = previous representation state;
* (E_t) = current textual evidence;
* (\Phi_T) = transformative mechanism;
* (\Delta_t) = transformation introduced at step (t).

The transformation may contain factual, psychological and social components: (\Delta_t = {\Delta_t^{F}, \Delta_t^{P}, \Delta_t^{S}}).

These components need not share the same dimensional form.

### Transformation Is Not Simple Subtraction

The transformative mechanism is not defined as arithmetic subtraction between consecutive representations.

Therefore, in general, (\Delta_t \neq Z_t - Z_{t-1}).

Simple numerical differences may later be calculated for particular fixed-dimensional variables, but such differences constitute one possible derived diagnostic and not the definition of transformation.

This distinction is especially important for the psychological space.

### Supervised Transformative Emotion

Notebook 03 defines **transformative emotion** as an independently supervised sentence-level psychological construct.

For emotional dimension (e), the supervised value is (\tau_{t,e}^{P} \in [-1,1]).

Dictionary:

* (\tau_{t,e}^{P}<0) = the sentence is assessed as decreasing emotional dimension (e);
* (\tau_{t,e}^{P}=0) = no meaningful change is assessed;
* (\tau_{t,e}^{P}>0) = the sentence is assessed as increasing emotional dimension (e).

The vector of supervised transformative-emotion values is (T_t^{P} = {\tau_{t,1}^{P}, \tau_{t,2}^{P}, \ldots, \tau_{t,11}^{P}}).

This is a supervision signal associated with the current textual step.

It is not automatically identical to the architectural psychological transformation: (T_t^{P} \neq \Delta_t^{P}).

Such an identity could only be introduced explicitly as a later model assumption and would need to be experimentally tested.

### Immediate Emotional State

Let (I_t^{P} = {i_{t,1}^{P}, i_{t,2}^{P}, \ldots, i_{t,11}^{P}}) denote the independently supervised immediate emotional state at textual step (t).

The arithmetic difference (I_t^{P} - I_{t-1}^{P}) may later be compared with supervised transformative emotion or a model-derived psychological transformation.

However, (T_t^{P} \neq I_t^{P} - I_{t-1}^{P}) by definition.

Any relationship among these quantities is therefore an empirical question rather than a methodological assumption.

### Reported Emotion Boundary

Let (R_t^{P}) denote the reported-emotion representation associated with textual unit (X_t).

The three supervised psychological constructs remain distinct: (R_t^{P} \neq T_t^{P} \neq I_t^{P}).

Dictionary:

* (R_t^{P}) = emotion reported or represented in the text;
* (T_t^{P}) = supervised emotional change introduced by the text;
* (I_t^{P}) = supervised immediate emotional state associated with the textual step.

The architectural psychological transformation (\Delta_t^{P}) remains a fourth distinct object.

### Resulting State

The transformation contributes to a resulting state (Z_t^{*} = \Psi_T(Z_{t-1}, E_t, \Delta_t)).

Dictionary:

* (Z_t^{*}) = resulting state after the transformative operation;
* (\Psi_T) = state-update function;
* (Z_{t-1}) = preceding state;
* (E_t) = current evidence;
* (\Delta_t) = transformation.

The explicit separation between (\Phi_T) and (\Psi_T) prevents the transformation itself from being confused with the state that results from it.

At this stage, neither function is assumed to have a specific learned implementation.

### Space-Specific Transformation

Transformation may operate differently in each representation space.

Factual transformation is represented as (\Delta_t^{F} = \Phi_T^{F}(Z_{t-1}^{F}, E_t^{F})).

Psychological transformation is represented as (\Delta_t^{P} = \Phi_T^{P}(Z_{t-1}^{P}, E_t^{P})).

Social transformation is represented as (\Delta_t^{S} = \Phi_T^{S}(Z_{t-1}^{S}, E_t^{S})).

A common transformation rule is not imposed merely for architectural convenience.

### Factual Transformation

A factual transformation may later represent changes such as:

* introduction of new factual content;
* qualification of preceding factual content;
* contradiction or correction;
* addition of contextual detail;
* change in informational density.

The factological representation remains an assessment of externally assessable informational content rather than truth.

Therefore, factual transformation does not itself constitute external fact verification.

### Psychological Transformation

A psychological transformation may later represent changes in emotional or psychological state introduced by the current textual step.

It must preserve (R_t^{P} \neq T_t^{P} \neq I_t^{P} \neq \Delta_t^{P}) as distinct methodological objects rather than treating them as interchangeable quantities.

Supervised transformative emotion may later serve as a target, constraint, explanatory variable or comparison signal for a psychological transformation model.

Its precise role must be specified by the experimental design.

### Social Transformation

A social transformation may later represent changes such as:

* introduction of an actor;
* introduction of a group;
* establishment or removal of membership;
* creation or modification of a relationship;
* change in relational orientation;
* change in represented group structure.

Because the social space is variable-dimensional, social transformation cannot generally be represented as simple vector subtraction.

It may instead require structured operations over actors, groups, memberships and relationships.

### Missing-Evidence Boundary

For each representation space, availability must remain explicit.

Let (M_t^{T} = {m_t^{F}, m_t^{P}, m_t^{S}}) represent transformation-input availability.

An unavailable input does not imply no transformation.

For representation space (X), (m_t^{X}=0 \not\Rightarrow \Delta_t^{X}=0).

Instead, the transformation for that space must remain unavailable unless an explicitly defined later inference mechanism can construct it.

### Zero-Change Boundary

Substantive zero and unavailable transformation are different states.

For a transformation component, (\Delta_{t,j}^{X}=0) means that the transformation has been assessed or derived and indicates no substantive change in component (j).

An unavailable transformation means that the value has not been established.

Therefore, (\text{unavailable} \neq 0).

### Initial-State Boundary

For (t=1), no preceding textual state exists.

The architecture therefore requires an explicit initial-state policy.

Possible later implementations may use:

* a learned initial state;
* a deterministic neutral state;
* an explicit unavailable state;
* another experimentally defined initialisation.

This block does not select among those alternatives.

The first-step transformation must record which initial-state policy is used.

### Sentence-Level Transformative Unit

The initial transformative unit is the sentence.

For sentence (s_t), the transformation record is represented as (\mathcal{T}*t = {ID_t, Z*{t-1}, E_t, \Delta_t, Z_t^{*}, M_t^{T}, Q_t^{T}}).

Dictionary:

* (ID_t) = stable sentence identity;
* (Z_{t-1}) = preceding-state reference;
* (E_t) = current-evidence reference;
* (\Delta_t) = transformation;
* (Z_t^{*}) = resulting-state reference;
* (M_t^{T}) = availability information;
* (Q_t^{T}) = provenance and construction metadata.

References may remain unresolved until later architectural components construct the corresponding states.

### Paragraph and Article Boundaries

Sentence-level transformation is constructed within the established article hierarchy.

Paragraph and article boundaries must remain explicit.

The transition between the final sentence of one paragraph and the first sentence of the next remains part of the same article sequence, but the paragraph boundary must be recorded.

No transformation may cross from one article into another unless a later experimental design explicitly introduces cross-document state.

For the current architecture, (Z_{0}^{article\ j}) is initialised independently for each article.

### Transformative and Recurrent Mechanisms

Transformation and recurrence are related but distinct.

Transformation answers: **What change does the current textual step introduce?**

Recurrence answers: **How is state carried through the textual sequence?**

Conceptually, transformation is (\Delta_t = \Phi_T(Z_{t-1}, E_t)), while recurrence may later be represented as (H_t = \Phi_R(H_{t-1}, E_t, \Delta_t)).

Here (H_t) denotes a recurrent state.

The recurrent mechanism is defined separately.

This block must therefore not silently treat (Z_t^{*}) as a fully specified recurrent hidden state.

### Transformative and Confluent Mechanisms

Transformation and confluence are also distinct.

Transformation operates across textual steps: (t-1 \rightarrow t).

Confluence operates across representation spaces: (F \leftrightarrow P \leftrightarrow S).

A later architecture may construct (C_t = \Phi_C(F_t, P_t, S_t)) and subsequently (\Delta_t = \Phi_T(Z_{t-1}, C_t)).

Alternatively, transformation may occur within individual representation spaces before confluence.

The ordering is not fixed by this block.

It remains an architectural hypothesis to be specified and tested.

### Hierarchical Transformation

Transformative information may later be propagated through the textual hierarchy: (\Delta^{token} \rightarrow \Delta^{sentence} \rightarrow \Delta^{paragraph} \rightarrow \Delta^{article}).

Higher-level transformation must not be defined as an unqualified arithmetic average of lower-level transformations.

A later aggregation function may be written as (\Delta^{higher} = A_T(\Delta_1^{lower}, \Delta_2^{lower}, \ldots, \Delta_n^{lower})).

The aggregation function must preserve:

* textual ordering;
* contribution magnitude;
* representation-space identity;
* availability;
* provenance;
* backward attribution.

The exact aggregation policy is defined later.

### Backward Attribution

Every constructed transformation must remain traceable to the textual evidence that produced it.

The minimum required path is (\Delta_t \rightarrow E_t \rightarrow Sentence_t \rightarrow Text).

Where model-token attribution becomes available, the path may extend to (\Delta_t \rightarrow Representation\ Space \rightarrow Sentence_t \rightarrow Token \rightarrow Text).

For state transitions involving preceding context, attribution may additionally identify relevant contributions from (Z_{t-1}).

This supports later inspection of **which passages contributed to a modelled state change**.

### Counterfactual Compatibility

The transformative architecture must support later counterfactual experiments.

For a textual intervention (X_t \rightarrow X_t'), a later experiment may compare the original transformation (\Delta_t = \Phi_T(Z_{t-1}, E_t)) with the counterfactual transformation (\Delta_t' = \Phi_T(Z_{t-1}, E_t')).

The resulting experimental difference may be represented as (\Delta_t' - \Delta_t).

This subtraction compares compatible transformation outputs; it is not the definition of transformation itself.

This enables Notebook 05 to test whether neutralising, removing or modifying a high-contribution passage changes the predicted transformation.

### Current Execution Boundary

At the current Notebook 04 stage, the transformative mechanism is defined structurally.

Validated factual and psychological supervision may be unavailable, and structured social information may likewise be unavailable.

The block must therefore be capable of constructing traceable sentence-level transformative records without inventing substantive transformations.

Where required source representations are unavailable:

* transformation values remain unavailable;
* availability masks remain inactive;
* previous-state references remain explicit;
* sentence ordering remains preserved;
* article boundaries remain preserved;
* provenance remains available.

No unavailable transformation is converted to substantive zero.

### Output

This block establishes:

* the sequential information boundary;
* previous-state references;
* current-evidence references;
* the transformative mechanism contract;
* factual transformation boundaries;
* psychological transformation boundaries;
* social transformation boundaries;
* the distinction between supervised transformative emotion and architectural transformation;
* the distinction between transformation and resulting state;
* missing-evidence and zero-change semantics;
* initial-state requirements;
* sentence-level transformative records;
* paragraph and article transition boundaries;
* transformative-recurrent separation;
* transformative-confluent separation;
* hierarchical transformation requirements;
* backward-attribution requirements;
* counterfactual compatibility.

At this stage, the block constructs the **transformative-state interface and traceability structure**, not a learned transformative model.

No unavailable transformation is interpreted as zero change.

No supervised transformative-emotion value is assumed to equal a derived state difference.

No future textual context is used.

No cross-article state propagation is performed.

No recurrent hidden state is calculated.

No hierarchical transformative aggregation is performed.

No attribution score is calculated.

No counterfactual experiment is performed.

No model training or experimental evaluation is performed.


````python
# =============================================================================
# Media AI — Transformative State Construction
# =============================================================================

from datetime import datetime, timezone

import numpy as np


# -----------------------------------------------------------------------------
# Block dependency checks
# -----------------------------------------------------------------------------

required_block_1_objects = [
    "BUILD_TRANSFORMATIVE_STATE",
    "PRESERVE_TEXTUAL_ORDER",
    "ALLOW_FUTURE_CONTEXT_IN_STATE_CONSTRUCTION",
    "REQUIRE_PREVIOUS_STATE_SEPARATION",
    "REQUIRE_TRANSFORMATIVE_CHANGE_SEPARATION",
    "REQUIRE_RESULTING_STATE_SEPARATION",
]

missing_block_1_objects = [
    object_name
    for object_name in required_block_1_objects
    if object_name not in globals()
]

if missing_block_1_objects:
    raise NameError(
        "Notebook 04 Block 1 is not initialised. "
        f"Missing: {missing_block_1_objects}"
    )


required_block_3_objects = [
    "HIERARCHICAL_TEXT_AND_IDENTITY_VALID",
    "SENTENCE_BY_ID",
    "SENTENCE_TO_PARAGRAPH_ID",
    "SENTENCE_TO_ARTICLE_ID",
    "ARTICLE_SENTENCE_ORDER",
]

missing_block_3_objects = [
    object_name
    for object_name in required_block_3_objects
    if object_name not in globals()
]

if missing_block_3_objects:
    raise NameError(
        "Notebook 04 Block 3 is not initialised. "
        f"Missing: {missing_block_3_objects}"
    )


required_block_4_objects = [
    "MEDIA_AI_REPRESENTATION_SPECIFICATION_VALID",
    "FACTUAL_DIMENSION_COUNT",
    "PSYCHOLOGICAL_DIMENSION_COUNT",
]

missing_block_4_objects = [
    object_name
    for object_name in required_block_4_objects
    if object_name not in globals()
]

if missing_block_4_objects:
    raise NameError(
        "Notebook 04 Block 4 is not initialised. "
        f"Missing: {missing_block_4_objects}"
    )


required_block_6_objects = [
    "FACTUAL_SENTENCE_BY_ID",
    "FACTUAL_REPRESENTATION_VALID",
]

missing_block_6_objects = [
    object_name
    for object_name in required_block_6_objects
    if object_name not in globals()
]

if missing_block_6_objects:
    raise NameError(
        "Notebook 04 Block 6 is not initialised. "
        f"Missing: {missing_block_6_objects}"
    )


required_block_7_objects = [
    "PSYCHOLOGICAL_SENTENCE_BY_ID",
    "PSYCHOLOGICAL_REPRESENTATION_VALID",
    "PSYCHOLOGICAL_CONSTRUCT_SLICES",
]

missing_block_7_objects = [
    object_name
    for object_name in required_block_7_objects
    if object_name not in globals()
]

if missing_block_7_objects:
    raise NameError(
        "Notebook 04 Block 7 is not initialised. "
        f"Missing: {missing_block_7_objects}"
    )


required_block_8_objects = [
    "SOCIAL_SENTENCE_BY_ID",
    "SOCIAL_REPRESENTATION_VALID",
    "CONFLUENT_CONTRACT_VALID",
]

missing_block_8_objects = [
    object_name
    for object_name in required_block_8_objects
    if object_name not in globals()
]

if missing_block_8_objects:
    raise NameError(
        "Notebook 04 Block 8 is not initialised. "
        f"Missing: {missing_block_8_objects}"
    )


if not HIERARCHICAL_TEXT_AND_IDENTITY_VALID:
    raise ValueError(
        "Transformative state construction requires "
        "a valid textual hierarchy."
    )


if not MEDIA_AI_REPRESENTATION_SPECIFICATION_VALID:
    raise ValueError(
        "Transformative state construction requires "
        "a valid representation specification."
    )


if not FACTUAL_REPRESENTATION_VALID:
    raise ValueError(
        "Transformative state construction requires "
        "a valid factual representation layer."
    )


if not PSYCHOLOGICAL_REPRESENTATION_VALID:
    raise ValueError(
        "Transformative state construction requires "
        "a valid psychological representation layer."
    )


if not SOCIAL_REPRESENTATION_VALID:
    raise ValueError(
        "Transformative state construction requires "
        "a valid social representation layer."
    )


if not PRESERVE_TEXTUAL_ORDER:
    raise ValueError(
        "Transformative state construction requires "
        "preserved textual order."
    )


if ALLOW_FUTURE_CONTEXT_IN_STATE_CONSTRUCTION:
    raise ValueError(
        "Future context must not enter transformative "
        "state construction."
    )


# -----------------------------------------------------------------------------
# Transformative-state configuration
# -----------------------------------------------------------------------------

TRANSFORMATIVE_STATE_VERSION = "1.0"

TRANSFORMATIVE_STATE_LEVEL = "sentence"

TRANSFORMATIVE_SOURCE_SPACES = (
    "factual",
    "psychological",
    "social",
)

TRANSFORMATIVE_INITIAL_STATE_POLICY = (
    "explicit_unavailable"
)

TRANSFORMATIVE_CROSS_ARTICLE_STATE = False

TRANSFORMATIVE_SIMPLE_DIFFERENCE_ASSUMED = False

TRANSFORMATIVE_SUPERVISED_EMOTION_IDENTITY_ASSUMED = False

TRANSFORMATIVE_RESULTING_STATE_IS_RECURRENT_STATE = False


# -----------------------------------------------------------------------------
# Methodological safeguard validation
# -----------------------------------------------------------------------------

if not REQUIRE_PREVIOUS_STATE_SEPARATION:
    raise ValueError(
        "Previous state must remain explicitly separate."
    )


if not REQUIRE_TRANSFORMATIVE_CHANGE_SEPARATION:
    raise ValueError(
        "Transformative change must remain explicitly separate."
    )


if not REQUIRE_RESULTING_STATE_SEPARATION:
    raise ValueError(
        "Resulting state must remain explicitly separate."
    )


if TRANSFORMATIVE_SIMPLE_DIFFERENCE_ASSUMED:
    raise ValueError(
        "Transformative state must not be defined "
        "as simple state subtraction."
    )


if TRANSFORMATIVE_SUPERVISED_EMOTION_IDENTITY_ASSUMED:
    raise ValueError(
        "Supervised transformative emotion must not be "
        "assumed identical to architectural transformation."
    )


if TRANSFORMATIVE_RESULTING_STATE_IS_RECURRENT_STATE:
    raise ValueError(
        "Resulting transformative state must remain "
        "distinct from recurrent hidden state."
    )


# -----------------------------------------------------------------------------
# Source-space availability helpers
# -----------------------------------------------------------------------------

def factual_space_available(
    sentence_id,
):
    """
    Determine whether any factual value is substantively available.
    """

    return bool(
        np.any(
            FACTUAL_SENTENCE_BY_ID[
                sentence_id
            ]["availability_mask"]
        )
    )


def psychological_space_available(
    sentence_id,
):
    """
    Determine whether any psychological value is substantively available.
    """

    return bool(
        np.any(
            PSYCHOLOGICAL_SENTENCE_BY_ID[
                sentence_id
            ]["availability_mask"]
        )
    )


def social_space_available(
    sentence_id,
):
    """
    Determine whether any social structure component is available.
    """

    return bool(
        any(
            SOCIAL_SENTENCE_BY_ID[
                sentence_id
            ]["availability_mask"].values()
        )
    )


def get_space_availability(
    sentence_id,
):
    """
    Return availability for all three Media AI representation spaces.
    """

    return {
        "factual":
            factual_space_available(
                sentence_id
            ),

        "psychological":
            psychological_space_available(
                sentence_id
            ),

        "social":
            social_space_available(
                sentence_id
            ),
    }


# -----------------------------------------------------------------------------
# Current-evidence reference
# -----------------------------------------------------------------------------

def build_current_evidence_reference(
    sentence_id,
):
    """
    Build references to the three sentence-level evidence spaces.
    """

    space_availability = (
        get_space_availability(
            sentence_id
        )
    )

    return {
        "sentence_id":
            sentence_id,

        "factual_reference":
            sentence_id,

        "psychological_reference":
            sentence_id,

        "social_reference":
            sentence_id,

        "space_availability":
            space_availability,

        "all_spaces_available":
            all(
                space_availability.values()
            ),

        "any_space_available":
            any(
                space_availability.values()
            ),
    }


# -----------------------------------------------------------------------------
# Previous-state reference
# -----------------------------------------------------------------------------

def build_previous_state_reference(
    article_id,
    previous_sentence_id,
):
    """
    Build the explicit previous-state reference.

    The first sentence of each article uses an unavailable initial state.
    """

    if previous_sentence_id is None:

        return {
            "state_type":
                "initial_state",

            "initial_state_policy":
                TRANSFORMATIVE_INITIAL_STATE_POLICY,

            "article_id":
                article_id,

            "previous_sentence_id":
                None,

            "state_available":
                False,

            "substantive_zero":
                False,
        }


    return {
        "state_type":
            "preceding_sentence_state_reference",

        "initial_state_policy":
            None,

        "article_id":
            article_id,

        "previous_sentence_id":
            previous_sentence_id,

        "state_available":
            False,

        "substantive_zero":
            False,
    }


# -----------------------------------------------------------------------------
# Empty transformative structure
# -----------------------------------------------------------------------------

def create_unavailable_transformative_structure():
    """
    Create a transformation structure with no fabricated values.
    """

    return {
        "factual": {
            "value":
                None,

            "value_state":
                "unavailable",
        },

        "psychological": {
            "value":
                None,

            "value_state":
                "unavailable",
        },

        "social": {
            "value":
                None,

            "value_state":
                "unavailable",
        },
    }


# -----------------------------------------------------------------------------
# Transformation availability mask
# -----------------------------------------------------------------------------

def create_transformative_availability_mask(
    current_evidence,
):
    """
    Create transformation-space availability.

    Current evidence availability does not automatically mean
    a transformation is constructible, because the transformation
    mechanism itself has not yet been implemented.
    """

    return {
        "factual":
            0,

        "psychological":
            0,

        "social":
            0,
    }


# -----------------------------------------------------------------------------
# Supervised transformative-emotion reference
# -----------------------------------------------------------------------------

def build_supervised_transformative_emotion_reference(
    sentence_id,
):
    """
    Preserve the supervised transformative-emotion slice as a distinct
    reference without treating it as architectural transformation.
    """

    psychological_record = (
        PSYCHOLOGICAL_SENTENCE_BY_ID[
            sentence_id
        ]
    )

    start, end = (
        PSYCHOLOGICAL_CONSTRUCT_SLICES[
            "transformative_emotion"
        ]
    )


    values = psychological_record[
        "values"
    ][
        start:end
    ]


    mask = psychological_record[
        "availability_mask"
    ][
        start:end
    ]


    available = bool(
        np.any(
            mask
        )
    )


    return {
        "available":
            available,

        "psychological_sentence_id":
            sentence_id,

        "slice":
            (
                start,
                end,
            ),

        "identity_with_architectural_transform_assumed":
            False,
    }


# -----------------------------------------------------------------------------
# Immediate-state reference
# -----------------------------------------------------------------------------

def build_immediate_state_reference(
    sentence_id,
):
    """
    Preserve the supervised immediate-emotional-state slice as a
    separate reference.
    """

    psychological_record = (
        PSYCHOLOGICAL_SENTENCE_BY_ID[
            sentence_id
        ]
    )

    start, end = (
        PSYCHOLOGICAL_CONSTRUCT_SLICES[
            "immediate_emotional_state"
        ]
    )


    mask = psychological_record[
        "availability_mask"
    ][
        start:end
    ]


    return {
        "available":
            bool(
                np.any(
                    mask
                )
            ),

        "psychological_sentence_id":
            sentence_id,

        "slice":
            (
                start,
                end,
            ),
    }


# -----------------------------------------------------------------------------
# Reported-emotion reference
# -----------------------------------------------------------------------------

def build_reported_emotion_reference(
    sentence_id,
):
    """
    Preserve reported emotion as a distinct psychological construct.
    """

    psychological_record = (
        PSYCHOLOGICAL_SENTENCE_BY_ID[
            sentence_id
        ]
    )

    start, end = (
        PSYCHOLOGICAL_CONSTRUCT_SLICES[
            "reported_emotion"
        ]
    )


    mask = psychological_record[
        "availability_mask"
    ][
        start:end
    ]


    return {
        "available":
            bool(
                np.any(
                    mask
                )
            ),

        "psychological_sentence_id":
            sentence_id,

        "slice":
            (
                start,
                end,
            ),
    }


# -----------------------------------------------------------------------------
# Transformative provenance
# -----------------------------------------------------------------------------

def build_transformative_provenance(
    sentence_id,
    article_position,
    paragraph_boundary,
    initial_state,
):
    """
    Build provenance for one sentence-level transformative record.
    """

    return {
        "media_ai_record_id":
            SENTENCE_TO_ARTICLE_ID[
                sentence_id
            ],

        "paragraph_id":
            SENTENCE_TO_PARAGRAPH_ID[
                sentence_id
            ],

        "sentence_id":
            sentence_id,

        "representation_level":
            TRANSFORMATIVE_STATE_LEVEL,

        "transformative_state_version":
            TRANSFORMATIVE_STATE_VERSION,

        "construction_method":
            "transformative_interface_only",

        "article_sentence_position":
            article_position,

        "paragraph_boundary":
            paragraph_boundary,

        "initial_state":
            initial_state,

        "future_context_used":
            False,

        "cross_article_state_used":
            False,

        "simple_difference_assumed":
            False,

        "supervised_transformative_identity_assumed":
            False,

        "recurrent_state_calculated":
            False,

        "constructed_at_utc":
            datetime.now(
                timezone.utc
            ).isoformat(),
    }


# -----------------------------------------------------------------------------
# Sentence-level transformative record construction
# -----------------------------------------------------------------------------

TRANSFORMATIVE_SENTENCE_RECORDS = []


for (
    article_id,
    ordered_sentence_ids,
) in ARTICLE_SENTENCE_ORDER.items():

    previous_sentence_id = None
    previous_paragraph_id = None


    for (
        article_position,
        sentence_id,
    ) in enumerate(
        ordered_sentence_ids
    ):

        paragraph_id = (
            SENTENCE_TO_PARAGRAPH_ID[
                sentence_id
            ]
        )


        initial_state = (
            previous_sentence_id
            is None
        )


        paragraph_boundary = (
            previous_paragraph_id
            is not None
            and paragraph_id
            != previous_paragraph_id
        )


        previous_state_reference = (
            build_previous_state_reference(
                article_id=
                    article_id,

                previous_sentence_id=
                    previous_sentence_id,
            )
        )


        current_evidence_reference = (
            build_current_evidence_reference(
                sentence_id
            )
        )


        transformation = (
            create_unavailable_transformative_structure()
        )


        transformation_mask = (
            create_transformative_availability_mask(
                current_evidence_reference
            )
        )


        resulting_state_reference = {
            "state_available":
                False,

            "state_constructed":
                False,

            "recurrent_state":
                False,

            "substantive_zero":
                False,
        }


        transformative_record = {
            "record_type":
                "media_ai_transformative_state",

            "transformative_state_version":
                TRANSFORMATIVE_STATE_VERSION,

            "media_ai_record_id":
                article_id,

            "paragraph_id":
                paragraph_id,

            "sentence_id":
                sentence_id,

            "article_sentence_position":
                article_position,

            "previous_sentence_id":
                previous_sentence_id,

            "initial_state":
                initial_state,

            "paragraph_boundary":
                paragraph_boundary,

            "previous_state_reference":
                previous_state_reference,

            "current_evidence_reference":
                current_evidence_reference,

            "transformation":
                transformation,

            "transformation_availability_mask":
                transformation_mask,

            "resulting_state_reference":
                resulting_state_reference,

            "reported_emotion_reference":
                build_reported_emotion_reference(
                    sentence_id
                ),

            "supervised_transformative_emotion_reference":
                build_supervised_transformative_emotion_reference(
                    sentence_id
                ),

            "immediate_emotional_state_reference":
                build_immediate_state_reference(
                    sentence_id
                ),

            "provenance":
                build_transformative_provenance(
                    sentence_id=
                        sentence_id,

                    article_position=
                        article_position,

                    paragraph_boundary=
                        paragraph_boundary,

                    initial_state=
                        initial_state,
                ),
        }


        TRANSFORMATIVE_SENTENCE_RECORDS.append(
            transformative_record
        )


        previous_sentence_id = (
            sentence_id
        )

        previous_paragraph_id = (
            paragraph_id
        )


# -----------------------------------------------------------------------------
# Sentence-level lookup
# -----------------------------------------------------------------------------

TRANSFORMATIVE_SENTENCE_BY_ID = {
    record[
        "sentence_id"
    ]:
        record

    for record
    in TRANSFORMATIVE_SENTENCE_RECORDS
}


# -----------------------------------------------------------------------------
# Transformative sequence validation
# -----------------------------------------------------------------------------

TRANSFORMATIVE_STATE_ERRORS = []


if len(
    TRANSFORMATIVE_SENTENCE_RECORDS
) != sum(
    len(
        sentence_ids
    )
    for sentence_ids in ARTICLE_SENTENCE_ORDER.values()
):

    TRANSFORMATIVE_STATE_ERRORS.append(
        "transformative_record_count_mismatch"
    )


for (
    article_id,
    ordered_sentence_ids,
) in ARTICLE_SENTENCE_ORDER.items():

    article_records = [
        TRANSFORMATIVE_SENTENCE_BY_ID[
            sentence_id
        ]
        for sentence_id in ordered_sentence_ids
    ]


    for index, record in enumerate(
        article_records
    ):

        expected_previous_sentence_id = (
            ordered_sentence_ids[
                index - 1
            ]
            if index > 0
            else None
        )


        if (
            record[
                "previous_sentence_id"
            ]
            != expected_previous_sentence_id
        ):

            TRANSFORMATIVE_STATE_ERRORS.append(
                "invalid_previous_sentence_reference"
            )

            break


        if (
            index == 0
            and not record[
                "initial_state"
            ]
        ):

            TRANSFORMATIVE_STATE_ERRORS.append(
                "article_initial_state_not_marked"
            )

            break


        if (
            index > 0
            and record[
                "initial_state"
            ]
        ):

            TRANSFORMATIVE_STATE_ERRORS.append(
                "non_initial_sentence_marked_initial"
            )

            break


    if TRANSFORMATIVE_STATE_ERRORS:
        break


# -----------------------------------------------------------------------------
# No-cross-article validation
# -----------------------------------------------------------------------------

for record in TRANSFORMATIVE_SENTENCE_RECORDS:

    previous_sentence_id = (
        record[
            "previous_sentence_id"
        ]
    )


    if previous_sentence_id is None:
        continue


    if (
        SENTENCE_TO_ARTICLE_ID[
            previous_sentence_id
        ]
        != record[
            "media_ai_record_id"
        ]
    ):

        TRANSFORMATIVE_STATE_ERRORS.append(
            "cross_article_state_reference_detected"
        )

        break


# -----------------------------------------------------------------------------
# Transformation-value validation
# -----------------------------------------------------------------------------

for record in TRANSFORMATIVE_SENTENCE_RECORDS:

    transformation = (
        record[
            "transformation"
        ]
    )

    mask = (
        record[
            "transformation_availability_mask"
        ]
    )


    if set(
        transformation.keys()
    ) != set(
        TRANSFORMATIVE_SOURCE_SPACES
    ):

        TRANSFORMATIVE_STATE_ERRORS.append(
            "invalid_transformative_space_structure"
        )

        break


    if set(
        mask.keys()
    ) != set(
        TRANSFORMATIVE_SOURCE_SPACES
    ):

        TRANSFORMATIVE_STATE_ERRORS.append(
            "invalid_transformative_mask_structure"
        )

        break


    for space in TRANSFORMATIVE_SOURCE_SPACES:

        space_record = (
            transformation[
                space
            ]
        )

        if mask[
            space
        ] not in (
            0,
            1,
        ):

            TRANSFORMATIVE_STATE_ERRORS.append(
                "invalid_transformative_mask_value"
            )

            break


        if mask[
            space
        ] == 0:

            if (
                space_record[
                    "value"
                ]
                is not None
            ):

                TRANSFORMATIVE_STATE_ERRORS.append(
                    "unavailable_transform_has_value"
                )

                break


            if (
                space_record[
                    "value_state"
                ]
                != "unavailable"
            ):

                TRANSFORMATIVE_STATE_ERRORS.append(
                    "invalid_unavailable_transform_state"
                )

                break


    if TRANSFORMATIVE_STATE_ERRORS:
        break


# -----------------------------------------------------------------------------
# Psychological construct-separation validation
# -----------------------------------------------------------------------------

for record in TRANSFORMATIVE_SENTENCE_RECORDS:

    supervised_transform_reference = (
        record[
            "supervised_transformative_emotion_reference"
        ]
    )


    if (
        supervised_transform_reference[
            "identity_with_architectural_transform_assumed"
        ]
    ):

        TRANSFORMATIVE_STATE_ERRORS.append(
            "supervised_transform_identity_assumed"
        )

        break


# -----------------------------------------------------------------------------
# Paragraph-boundary accounting
# -----------------------------------------------------------------------------

TRANSFORMATIVE_PARAGRAPH_BOUNDARY_COUNT = sum(
    1
    for record in TRANSFORMATIVE_SENTENCE_RECORDS
    if record[
        "paragraph_boundary"
    ]
)


TRANSFORMATIVE_ARTICLE_INITIAL_STATE_COUNT = sum(
    1
    for record in TRANSFORMATIVE_SENTENCE_RECORDS
    if record[
        "initial_state"
    ]
)


# -----------------------------------------------------------------------------
# Evidence availability accounting
# -----------------------------------------------------------------------------

TRANSFORMATIVE_FACTUAL_EVIDENCE_AVAILABLE = sum(
    1
    for record in TRANSFORMATIVE_SENTENCE_RECORDS
    if record[
        "current_evidence_reference"
    ][
        "space_availability"
    ][
        "factual"
    ]
)


TRANSFORMATIVE_PSYCHOLOGICAL_EVIDENCE_AVAILABLE = sum(
    1
    for record in TRANSFORMATIVE_SENTENCE_RECORDS
    if record[
        "current_evidence_reference"
    ][
        "space_availability"
    ][
        "psychological"
    ]
)


TRANSFORMATIVE_SOCIAL_EVIDENCE_AVAILABLE = sum(
    1
    for record in TRANSFORMATIVE_SENTENCE_RECORDS
    if record[
        "current_evidence_reference"
    ][
        "space_availability"
    ][
        "social"
    ]
)


TRANSFORMATIVE_ANY_EVIDENCE_AVAILABLE = sum(
    1
    for record in TRANSFORMATIVE_SENTENCE_RECORDS
    if record[
        "current_evidence_reference"
    ][
        "any_space_available"
    ]
)


TRANSFORMATIVE_ALL_EVIDENCE_AVAILABLE = sum(
    1
    for record in TRANSFORMATIVE_SENTENCE_RECORDS
    if record[
        "current_evidence_reference"
    ][
        "all_spaces_available"
    ]
)


# -----------------------------------------------------------------------------
# Transformation availability accounting
# -----------------------------------------------------------------------------

TRANSFORMATIVE_AVAILABLE_FACTUAL_COUNT = sum(
    record[
        "transformation_availability_mask"
    ][
        "factual"
    ]
    for record in TRANSFORMATIVE_SENTENCE_RECORDS
)


TRANSFORMATIVE_AVAILABLE_PSYCHOLOGICAL_COUNT = sum(
    record[
        "transformation_availability_mask"
    ][
        "psychological"
    ]
    for record in TRANSFORMATIVE_SENTENCE_RECORDS
)


TRANSFORMATIVE_AVAILABLE_SOCIAL_COUNT = sum(
    record[
        "transformation_availability_mask"
    ][
        "social"
    ]
    for record in TRANSFORMATIVE_SENTENCE_RECORDS
)


# -----------------------------------------------------------------------------
# Transformative contract
# -----------------------------------------------------------------------------

TRANSFORMATIVE_STATE_CONSTRUCTION_CONTRACT = {
    "version":
        TRANSFORMATIVE_STATE_VERSION,

    "level":
        TRANSFORMATIVE_STATE_LEVEL,

    "source_spaces":
        TRANSFORMATIVE_SOURCE_SPACES,

    "previous_state_separate":
        REQUIRE_PREVIOUS_STATE_SEPARATION,

    "transformation_separate":
        REQUIRE_TRANSFORMATIVE_CHANGE_SEPARATION,

    "resulting_state_separate":
        REQUIRE_RESULTING_STATE_SEPARATION,

    "simple_difference_definition":
        False,

    "supervised_transformative_emotion_identity":
        False,

    "future_context_allowed":
        False,

    "cross_article_state":
        False,

    "initial_state_policy":
        TRANSFORMATIVE_INITIAL_STATE_POLICY,

    "resulting_state_is_recurrent_state":
        False,

    "hierarchical_aggregation_performed":
        False,

    "counterfactual_compatible":
        True,

    "backward_attribution_required":
        True,
}


# -----------------------------------------------------------------------------
# Transformative-recurrent boundary
# -----------------------------------------------------------------------------

TRANSFORMATIVE_RECURRENT_BOUNDARY = {
    "transformative_mechanism":
        "change_introduced_by_current_textual_step",

    "recurrent_mechanism":
        "state_propagation_through_textual_sequence",

    "mechanisms_equivalent":
        False,

    "recurrent_state_constructed":
        False,
}


# -----------------------------------------------------------------------------
# Transformative-confluent boundary
# -----------------------------------------------------------------------------

TRANSFORMATIVE_CONFLUENT_BOUNDARY = {
    "transformation_operates_across_textual_steps":
        True,

    "confluence_operates_across_representation_spaces":
        True,

    "mechanisms_equivalent":
        False,

    "operation_order_fixed":
        False,

    "operation_order_is_architectural_hypothesis":
        True,
}


# -----------------------------------------------------------------------------
# Representation status
# -----------------------------------------------------------------------------

TRANSFORMATIVE_STATE_VALID = (
    len(
        TRANSFORMATIVE_STATE_ERRORS
    )
    == 0
)


if not TRANSFORMATIVE_STATE_VALID:
    raise ValueError(
        "Transformative state construction failed validation: "
        f"{TRANSFORMATIVE_STATE_ERRORS}"
    )


TRANSFORMATIVE_STATE_RECORDS_CONSTRUCTED = True

TRANSFORMATIVE_STATE_CALCULATED = False

TRANSFORMATIVE_SUBSTANTIVE_VALUES_PRESENT = any([
    TRANSFORMATIVE_AVAILABLE_FACTUAL_COUNT > 0,
    TRANSFORMATIVE_AVAILABLE_PSYCHOLOGICAL_COUNT > 0,
    TRANSFORMATIVE_AVAILABLE_SOCIAL_COUNT > 0,
])


# -----------------------------------------------------------------------------
# Summary
# -----------------------------------------------------------------------------

TRANSFORMATIVE_STATE_SUMMARY = {
    "version":
        TRANSFORMATIVE_STATE_VERSION,

    "level":
        TRANSFORMATIVE_STATE_LEVEL,

    "sentence_records":
        len(
            TRANSFORMATIVE_SENTENCE_RECORDS
        ),

    "article_initial_states":
        TRANSFORMATIVE_ARTICLE_INITIAL_STATE_COUNT,

    "paragraph_boundaries":
        TRANSFORMATIVE_PARAGRAPH_BOUNDARY_COUNT,

    "factual_evidence_available":
        TRANSFORMATIVE_FACTUAL_EVIDENCE_AVAILABLE,

    "psychological_evidence_available":
        TRANSFORMATIVE_PSYCHOLOGICAL_EVIDENCE_AVAILABLE,

    "social_evidence_available":
        TRANSFORMATIVE_SOCIAL_EVIDENCE_AVAILABLE,

    "any_evidence_available":
        TRANSFORMATIVE_ANY_EVIDENCE_AVAILABLE,

    "all_evidence_available":
        TRANSFORMATIVE_ALL_EVIDENCE_AVAILABLE,

    "factual_transformations_available":
        TRANSFORMATIVE_AVAILABLE_FACTUAL_COUNT,

    "psychological_transformations_available":
        TRANSFORMATIVE_AVAILABLE_PSYCHOLOGICAL_COUNT,

    "social_transformations_available":
        TRANSFORMATIVE_AVAILABLE_SOCIAL_COUNT,

    "records_constructed":
        TRANSFORMATIVE_STATE_RECORDS_CONSTRUCTED,

    "transformative_state_calculated":
        TRANSFORMATIVE_STATE_CALCULATED,

    "substantive_values_present":
        TRANSFORMATIVE_SUBSTANTIVE_VALUES_PRESENT,

    "valid":
        TRANSFORMATIVE_STATE_VALID,
}


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Transformative State Construction")
print("=" * 72)

print(
    f"Transformative state version: "
    f"{TRANSFORMATIVE_STATE_VERSION}"
)

print(
    f"Representation level        : "
    f"{TRANSFORMATIVE_STATE_LEVEL}"
)

print(
    f"Source spaces               : "
    f"{', '.join(TRANSFORMATIVE_SOURCE_SPACES)}"
)

print(
    f"Sentence records            : "
    f"{len(TRANSFORMATIVE_SENTENCE_RECORDS)}"
)

print("-" * 72)

print("Sequential structure")

print(
    f"Article initial states      : "
    f"{TRANSFORMATIVE_ARTICLE_INITIAL_STATE_COUNT}"
)

print(
    f"Initial-state policy        : "
    f"{TRANSFORMATIVE_INITIAL_STATE_POLICY}"
)

print(
    f"Paragraph boundaries        : "
    f"{TRANSFORMATIVE_PARAGRAPH_BOUNDARY_COUNT}"
)

print(
    f"Cross-article state         : "
    f"{TRANSFORMATIVE_CROSS_ARTICLE_STATE}"
)

print(
    f"Future context allowed      : "
    f"{ALLOW_FUTURE_CONTEXT_IN_STATE_CONSTRUCTION}"
)

print("-" * 72)

print("Current evidence availability")

print(
    f"Factual evidence            : "
    f"{TRANSFORMATIVE_FACTUAL_EVIDENCE_AVAILABLE}"
)

print(
    f"Psychological evidence      : "
    f"{TRANSFORMATIVE_PSYCHOLOGICAL_EVIDENCE_AVAILABLE}"
)

print(
    f"Social evidence             : "
    f"{TRANSFORMATIVE_SOCIAL_EVIDENCE_AVAILABLE}"
)

print(
    f"Any source space available  : "
    f"{TRANSFORMATIVE_ANY_EVIDENCE_AVAILABLE}"
)

print(
    f"All source spaces available : "
    f"{TRANSFORMATIVE_ALL_EVIDENCE_AVAILABLE}"
)

print("-" * 72)

print("Transformative availability")

print(
    f"Factual transformations     : "
    f"{TRANSFORMATIVE_AVAILABLE_FACTUAL_COUNT}"
)

print(
    f"Psychological transforms    : "
    f"{TRANSFORMATIVE_AVAILABLE_PSYCHOLOGICAL_COUNT}"
)

print(
    f"Social transformations      : "
    f"{TRANSFORMATIVE_AVAILABLE_SOCIAL_COUNT}"
)

print(
    f"Substantive values present  : "
    f"{TRANSFORMATIVE_SUBSTANTIVE_VALUES_PRESENT}"
)

print("-" * 72)

print("Methodological boundaries")

print(
    f"Simple difference assumed   : "
    f"{TRANSFORMATIVE_SIMPLE_DIFFERENCE_ASSUMED}"
)

print(
    f"Supervised transform = Δ    : "
    f"{TRANSFORMATIVE_SUPERVISED_EMOTION_IDENTITY_ASSUMED}"
)

print(
    f"Resulting state = recurrent : "
    f"{TRANSFORMATIVE_RESULTING_STATE_IS_RECURRENT_STATE}"
)

print(
    f"Transformation = recurrence : "
    f"{TRANSFORMATIVE_RECURRENT_BOUNDARY['mechanisms_equivalent']}"
)

print(
    f"Transformation = confluence : "
    f"{TRANSFORMATIVE_CONFLUENT_BOUNDARY['mechanisms_equivalent']}"
)

print(
    f"Operation order fixed       : "
    f"{TRANSFORMATIVE_CONFLUENT_BOUNDARY['operation_order_fixed']}"
)

print("-" * 72)

print(
    f"Transformative records valid: "
    f"{TRANSFORMATIVE_STATE_VALID}"
)

print(
    f"Records constructed         : "
    f"{TRANSFORMATIVE_STATE_RECORDS_CONSTRUCTED}"
)

print(
    f"Transformative state calc.  : "
    f"{TRANSFORMATIVE_STATE_CALCULATED}"
)

print("=" * 72)

print(
    "Sentence-level transformative-state records constructed successfully."
)

print(
    "Previous state, current evidence, transformation and resulting state "
    "remain separate architectural objects."
)

print(
    "Article initial states are explicit and have not been interpreted "
    "as substantive zero states."
)

print(
    "No cross-article state propagation has been performed."
)

print(
    "Supervised transformative emotion remains distinct from architectural "
    "psychological transformation."
)

print(
    "Transformative emotion has not been defined as the arithmetic difference "
    "between consecutive immediate emotional states."
)

print(
    "Unavailable transformation values have not been converted to zero change."
)

print(
    "No recurrent hidden state has been calculated."
)

print(
    "No hierarchical transformative aggregation or attribution score "
    "has been calculated."
)

print(
    "No counterfactual experiment, model training or experimental evaluation "
    "has been performed."
)
````

## Recurrent State Construction

This block defines the recurrent mechanism of the Media AI architecture.

The recurrent mechanism represents **how Media AI state is carried through an ordered textual sequence**.

It is distinct from the transformative mechanism defined in Block 9.

Transformation answers:

**What change does the current textual unit introduce?**

Recurrence answers:

**What state is carried forward after processing the textual sequence up to the current unit?**

The distinction is therefore:

(transformation \neq recurrence).

### Sequential Recurrent State

Let the ordered sentence sequence of an article be (s_1, s_2, \ldots, s_T).

For textual step (t), let (H_t) denote the recurrent state after processing sentence (s_t).

The generic recurrent update is (H_t = \Phi_R(H_{t-1}, E_t, \Delta_t)).

Dictionary:

* (H_{t-1}) = recurrent state before processing sentence (s_t);
* (E_t) = current sentence evidence;
* (\Delta_t) = transformation introduced by sentence (s_t);
* (\Phi_R) = recurrent state-update mechanism;
* (H_t) = recurrent state after processing sentence (s_t).

The function (\Phi_R) is an architectural contract at this stage.

No specific recurrent neural architecture is assumed.

### Recurrent Information Boundary

At textual step (t), the recurrent state may depend only on the permitted information set (\mathcal{I}_t = {X_1, X_2, \ldots, X_t}).

Future textual units (X_{t+1}, \ldots, X_T) must not influence (H_t).

Therefore:

(H_t = f(X_1, X_2, \ldots, X_t)),

but not:

(H_t = f(X_1, \ldots, X_T))

when the configured methodology requires preceding-context processing.

This preserves the same temporal boundary established in the supervision and transformative stages.

### Initial Recurrent State

For the first sentence of each article, the recurrent process requires an initial state (H_0).

The current architecture does not assume that (H_0 = \mathbf{0}).

Instead, the initial-state policy must remain explicit.

Possible later implementations may use:

* an explicit unavailable initial state;
* a deterministic neutral state;
* a learned initial state;
* a model-specific initialisation.

The selected policy must be recorded as part of recurrent-state provenance.

For the current structural architecture, the initial state remains explicitly unresolved.

### Recurrent State Is Not Current Evidence

The recurrent state is distinct from the current sentence representation.

Thus:

(H_t \neq E_t).

Current evidence describes what the current textual unit contributes.

Recurrent state describes what the architecture carries after processing the sequence up to step (t).

A sentence may therefore introduce limited new evidence while still resulting in a substantial recurrent state because of earlier context.

### Recurrent State Is Not Transformation

The recurrent state is also distinct from the transformation:

(H_t \neq \Delta_t).

The transformation describes change introduced at step (t).

The recurrent state describes accumulated or propagated state after the update.

A large transformation may occur within a low accumulated state, while a small transformation may occur within a high accumulated state.

These quantities must therefore remain separately represented and attributable.

### Resulting Transformative State and Recurrent State

Block 9 introduced the resulting transformative state (Z_t^{*}).

The recurrent state (H_t) must not be assumed identical to (Z_t^{*}).

Therefore:

(H_t \neq Z_t^{*})

unless a later architecture explicitly defines and tests such an equivalence.

A later implementation may use (Z_t^{*}) as an input to recurrent updating, for example (H_t = \Phi_R(H_{t-1}, Z_t^{*})).

Alternatively, recurrence may consume current evidence and transformation separately.

The exact update form remains an architectural hypothesis.

### Space-Specific Recurrent State

The recurrent state may preserve separate factual, psychological and social components.

Conceptually, (H_t = {H_t^{F}, H_t^{P}, H_t^{S}}).

Dictionary:

* (H_t^{F}) = recurrent factual state;
* (H_t^{P}) = recurrent psychological state;
* (H_t^{S}) = recurrent social state.

These components may require different update mechanisms.

For example:

(H_t^{F} = \Phi_R^{F}(H_{t-1}^{F}, E_t^{F}, \Delta_t^{F})),

(H_t^{P} = \Phi_R^{P}(H_{t-1}^{P}, E_t^{P}, \Delta_t^{P})),

and:

(H_t^{S} = \Phi_R^{S}(H_{t-1}^{S}, E_t^{S}, \Delta_t^{S})).

A single shared recurrence rule is not assumed by definition.

### Factual Recurrent State

The factual recurrent state may later preserve accumulated informational structure such as:

* claims introduced earlier in the article;
* attributed sources;
* certainty or uncertainty structure;
* factual density trajectory;
* qualification or correction patterns.

The factual recurrent state does not establish whether those claims are true.

It therefore remains separate from external fact verification.

### Psychological Recurrent State

The psychological recurrent state may later represent accumulated psychological framing through the article.

For example, the state may retain evolving levels or latent representations associated with:

* fear;
* anger;
* tension;
* uncertainty;
* trust;
* compassion;
* solidarity.

The recurrent psychological state must remain distinguishable from:

(R_t^{P}), (T_t^{P}), (I_t^{P}), and (\Delta_t^{P}).

Thus the architecture preserves:

(H_t^{P} \neq R_t^{P} \neq T_t^{P} \neq I_t^{P} \neq \Delta_t^{P}).

These objects may be related empirically, but they are not identical by definition.

### Social Recurrent State

The social recurrent state may later preserve evolving actor and relationship structure through the article.

Examples may include:

* actors introduced earlier;
* recurring actor-group associations;
* evolving alliances;
* conflict structure;
* accumulated relationship evidence;
* changing group polarisation structure.

Because the social representation is variable-dimensional, (H_t^{S}) need not be a simple fixed-dimensional vector.

A later implementation may use a graph state, relational memory, learned embedding or another explicitly defined social-state structure.

### Persistence and Decay

Recurrence does not imply that all prior information must persist indefinitely with equal strength.

A recurrent update mechanism may later include:

* persistence;
* decay;
* forgetting;
* gating;
* reinforcement;
* replacement;
* contextual reweighting.

Conceptually, (H_t = \Phi_R(H_{t-1}, E_t, \Delta_t; \theta_R)), where (\theta_R) denotes the recurrent update parameters or policy.

The architecture therefore permits previous information to weaken or strengthen through the sequence.

No specific decay or gating mechanism is selected by this block.

### Missing-Evidence Boundary

Missing current evidence must not be interpreted as zero evidence.

If one or more representation spaces are unavailable at step (t), the recurrent mechanism must preserve that information explicitly.

Let (M_t^{R} = {m_t^{F}, m_t^{P}, m_t^{S}}) represent recurrent-input availability.

For representation space (X), (m_t^{X}=0) does not imply (E_t^{X}=0), (\Delta_t^{X}=0), or (H_t^{X}=0).

Instead, the recurrent state for that space must remain unavailable or follow an explicitly defined later missing-input policy.

### Recurrent State Availability

A recurrent state may be:

* available;
* partially available;
* unavailable.

For a structured recurrent state, availability may differ between representation spaces.

Thus a sentence may have an available psychological recurrent state while factual or social recurrent components remain unavailable.

The architecture must preserve this distinction rather than requiring all spaces to become available simultaneously.

### Article Boundary

Recurrent state is article-specific in the current architecture.

For article (a), the recurrent sequence is (H_0^{(a)}, H_1^{(a)}, \ldots, H_T^{(a)}).

A new article begins with a new initial state.

Therefore:

(H_T^{(a)} \not\rightarrow H_0^{(b)})

for distinct articles (a) and (b).

No cross-article recurrent state propagation is permitted unless a later research design explicitly introduces it.

### Paragraph Boundary

Paragraph boundaries remain explicit within the recurrent sequence.

The transition from the final sentence of paragraph (p) to the first sentence of paragraph (p+1) may preserve recurrent state because both sentences belong to the same article.

However, the paragraph boundary itself must remain recorded.

This allows later experiments to test whether paragraph boundaries should modify recurrent update behaviour.

### Recurrent State and Confluence

Recurrence and confluence remain separate architectural mechanisms.

Confluence operates across representation spaces.

Recurrence operates across textual sequence.

A later architecture may use sentence-level confluence first:

(C_t = \Phi_C(F_t, P_t, S_t)),

followed by recurrence:

(H_t = \Phi_R(H_{t-1}, C_t, \Delta_t)).

Alternatively, space-specific recurrence may occur before confluence.

For example:

(H_t^{F} = \Phi_R^{F}(H_{t-1}^{F}, F_t)),

(H_t^{P} = \Phi_R^{P}(H_{t-1}^{P}, P_t)),

and (H_t^{S} = \Phi_R^{S}(H_{t-1}^{S}, S_t)),

followed by a later confluent operation.

The ordering is not fixed by this block.

It remains an architectural hypothesis.

### Recurrent State and Transformation

Transformation may influence recurrence, but the architecture does not require recurrence to equal cumulative transformation.

Therefore, in general:

(H_t \neq \sum_{j=1}^{t}\Delta_j).

A recurrent state may involve nonlinear persistence, decay, interaction, gating or other state-update behaviour.

Simple cumulative summation may later be used as a baseline, but it is not the definition of recurrence.

### Recurrent Trajectory

For article (a), the recurrent trajectory is (\mathcal{H}^{(a)} = (H_1^{(a)}, H_2^{(a)}, \ldots, H_T^{(a)})).

Dictionary:

* (\mathcal{H}^{(a)}) = complete recurrent state trajectory of article (a);
* (H_t^{(a)}) = recurrent state after processing textual step (t).

The trajectory enables later analysis of how factual, psychological and social states develop through the article.

For example, a psychological fear trajectory may be represented as (h_{1,fear}^{P}, h_{2,fear}^{P}, \ldots, h_{T,fear}^{P}).

### State Change Along the Recurrent Trajectory

A later diagnostic may compare consecutive recurrent states through (D_t^{H} = H_t - H_{t-1}) where the relevant recurrent representation is numerically compatible.

This quantity is a derived state difference.

It must not be confused with architectural transformation (\Delta_t).

Therefore:

(D_t^{H} \neq \Delta_t)

by definition.

Their relationship may later become an experimental question.

### Hierarchical Recurrent Representation

The recurrent mechanism is initially defined over the ordered sentence sequence.

Later hierarchical states may be constructed at paragraph and article levels.

Conceptually:

(H^{sentence} \rightarrow H^{paragraph} \rightarrow H^{article}).

However, paragraph and article states must not be defined as unqualified averages of sentence recurrent states.

A later aggregation function may be written as (H^{higher} = A_R(H_1^{lower}, H_2^{lower}, \ldots, H_n^{lower})).

The aggregation must preserve:

* temporal order;
* state trajectory;
* availability;
* representation-space identity;
* provenance;
* backward attribution.

### Backward Attribution

Every recurrent state must remain traceable to the textual evidence that contributed to it.

The required path is (H_t \rightarrow {E_1,\ldots,E_t} \rightarrow {Sentence_1,\ldots,Sentence_t} \rightarrow Text).

Where attribution methods permit, the architecture should additionally distinguish the relative contributions of preceding state and current evidence.

For example, a later attribution may decompose a recurrent state as contributions from:

* inherited state (H_{t-1});
* current factual evidence;
* current psychological evidence;
* current social evidence;
* current transformation.

This is necessary for interpreting **why a recurrent trajectory changes at a particular point in the article**.

### Counterfactual Compatibility

The recurrent architecture must support later counterfactual experiments.

For a modified textual unit (X_t'), a later experiment may construct a counterfactual recurrent trajectory (\mathcal{H}').

Comparison with the original trajectory (\mathcal{H}) allows measurement of how a textual intervention changes later states.

A local intervention may therefore affect not only (H_t), but also (H_{t+1}, H_{t+2}, \ldots).

This propagation effect is one of the principal reasons recurrence is modelled separately from immediate transformation.

### Current Execution Boundary

At the current Notebook 04 stage, the recurrent mechanism is defined structurally.

The source factual, psychological and social representations currently contain unavailable substantive values, and the transformative interface contains no calculated transformation values.

The block must therefore be able to construct recurrent-state records and sequential references without fabricating substantive recurrent states.

Where recurrent inputs are unavailable:

* recurrent values remain unavailable;
* availability indicators remain inactive;
* previous recurrent-state references remain explicit;
* sentence ordering remains preserved;
* paragraph boundaries remain preserved;
* article boundaries remain preserved;
* provenance remains available.

No unavailable recurrent state is interpreted as substantive zero.

### Output

This block establishes:

* the recurrent-state contract;
* initial recurrent-state requirements;
* recurrent sequence ordering;
* factual recurrent-state boundaries;
* psychological recurrent-state boundaries;
* social recurrent-state boundaries;
* persistence and decay boundaries;
* missing-input semantics;
* article and paragraph boundaries;
* recurrent-transformative separation;
* recurrent-confluent separation;
* recurrent trajectory structure;
* derived recurrent-state-change boundaries;
* hierarchical recurrent-state requirements;
* backward-attribution requirements;
* counterfactual compatibility.

At this stage, the block constructs the **recurrent-state interface and traceability structure**, not a learned recurrent model.

No unavailable recurrent state is interpreted as zero.

No recurrent state is assumed equal to current evidence, transformation, or the resulting transformative state.

No cross-article recurrent propagation is performed.

No hierarchical recurrent aggregation is performed.

No attribution score is calculated.

No counterfactual experiment is performed.

No model training or experimental evaluation is performed.


````python
# =============================================================================
# Media AI — Recurrent State Construction
# =============================================================================

from datetime import datetime, timezone


# -----------------------------------------------------------------------------
# Block dependency checks
# -----------------------------------------------------------------------------

required_block_1_objects = [
    "BUILD_RECURRENT_STATE",
    "PRESERVE_TEXTUAL_ORDER",
    "ALLOW_FUTURE_CONTEXT_IN_STATE_CONSTRUCTION",
    "REQUIRE_PREVIOUS_STATE_SEPARATION",
    "REQUIRE_TRANSFORMATIVE_CHANGE_SEPARATION",
    "REQUIRE_RESULTING_STATE_SEPARATION",
]

missing_block_1_objects = [
    object_name
    for object_name in required_block_1_objects
    if object_name not in globals()
]

if missing_block_1_objects:
    raise NameError(
        "Notebook 04 Block 1 is not initialised. "
        f"Missing: {missing_block_1_objects}"
    )


required_block_3_objects = [
    "HIERARCHICAL_TEXT_AND_IDENTITY_VALID",
    "SENTENCE_TO_PARAGRAPH_ID",
    "SENTENCE_TO_ARTICLE_ID",
    "ARTICLE_SENTENCE_ORDER",
]

missing_block_3_objects = [
    object_name
    for object_name in required_block_3_objects
    if object_name not in globals()
]

if missing_block_3_objects:
    raise NameError(
        "Notebook 04 Block 3 is not initialised. "
        f"Missing: {missing_block_3_objects}"
    )


required_block_6_objects = [
    "FACTUAL_SENTENCE_BY_ID",
    "FACTUAL_REPRESENTATION_VALID",
]

missing_block_6_objects = [
    object_name
    for object_name in required_block_6_objects
    if object_name not in globals()
]

if missing_block_6_objects:
    raise NameError(
        "Notebook 04 Block 6 is not initialised. "
        f"Missing: {missing_block_6_objects}"
    )


required_block_7_objects = [
    "PSYCHOLOGICAL_SENTENCE_BY_ID",
    "PSYCHOLOGICAL_REPRESENTATION_VALID",
]

missing_block_7_objects = [
    object_name
    for object_name in required_block_7_objects
    if object_name not in globals()
]

if missing_block_7_objects:
    raise NameError(
        "Notebook 04 Block 7 is not initialised. "
        f"Missing: {missing_block_7_objects}"
    )


required_block_8_objects = [
    "SOCIAL_SENTENCE_BY_ID",
    "SOCIAL_REPRESENTATION_VALID",
    "CONFLUENT_CONTRACT_VALID",
]

missing_block_8_objects = [
    object_name
    for object_name in required_block_8_objects
    if object_name not in globals()
]

if missing_block_8_objects:
    raise NameError(
        "Notebook 04 Block 8 is not initialised. "
        f"Missing: {missing_block_8_objects}"
    )


required_block_9_objects = [
    "TRANSFORMATIVE_SENTENCE_BY_ID",
    "TRANSFORMATIVE_STATE_VALID",
    "TRANSFORMATIVE_STATE_CALCULATED",
]

missing_block_9_objects = [
    object_name
    for object_name in required_block_9_objects
    if object_name not in globals()
]

if missing_block_9_objects:
    raise NameError(
        "Notebook 04 Block 9 is not initialised. "
        f"Missing: {missing_block_9_objects}"
    )


if not HIERARCHICAL_TEXT_AND_IDENTITY_VALID:
    raise ValueError(
        "Recurrent state construction requires a valid "
        "textual hierarchy."
    )


if not FACTUAL_REPRESENTATION_VALID:
    raise ValueError(
        "Recurrent state construction requires a valid "
        "factual representation layer."
    )


if not PSYCHOLOGICAL_REPRESENTATION_VALID:
    raise ValueError(
        "Recurrent state construction requires a valid "
        "psychological representation layer."
    )


if not SOCIAL_REPRESENTATION_VALID:
    raise ValueError(
        "Recurrent state construction requires a valid "
        "social representation layer."
    )


if not TRANSFORMATIVE_STATE_VALID:
    raise ValueError(
        "Recurrent state construction requires a valid "
        "transformative-state interface."
    )


if not PRESERVE_TEXTUAL_ORDER:
    raise ValueError(
        "Recurrent state construction requires preserved "
        "textual order."
    )


if ALLOW_FUTURE_CONTEXT_IN_STATE_CONSTRUCTION:
    raise ValueError(
        "Future context must not enter recurrent state construction."
    )


# -----------------------------------------------------------------------------
# Recurrent-state configuration
# -----------------------------------------------------------------------------

RECURRENT_STATE_VERSION = "1.0"

RECURRENT_STATE_LEVEL = "sentence"

RECURRENT_SOURCE_SPACES = (
    "factual",
    "psychological",
    "social",
)

RECURRENT_INITIAL_STATE_POLICY = (
    "explicit_unavailable"
)

RECURRENT_CROSS_ARTICLE_STATE = False

RECURRENT_STATE_EQUALS_CURRENT_EVIDENCE = False

RECURRENT_STATE_EQUALS_TRANSFORMATION = False

RECURRENT_STATE_EQUALS_TRANSFORMATIVE_RESULT = False

RECURRENT_CUMULATIVE_SUM_ASSUMED = False

RECURRENT_OPERATION_ORDER_FIXED = False


# -----------------------------------------------------------------------------
# Methodological safeguard validation
# -----------------------------------------------------------------------------

if not REQUIRE_PREVIOUS_STATE_SEPARATION:
    raise ValueError(
        "Previous state must remain explicitly separate."
    )


if not REQUIRE_TRANSFORMATIVE_CHANGE_SEPARATION:
    raise ValueError(
        "Transformative change must remain explicitly separate."
    )


if not REQUIRE_RESULTING_STATE_SEPARATION:
    raise ValueError(
        "Resulting state must remain explicitly separate."
    )


if RECURRENT_STATE_EQUALS_CURRENT_EVIDENCE:
    raise ValueError(
        "Recurrent state must not be equated with current evidence."
    )


if RECURRENT_STATE_EQUALS_TRANSFORMATION:
    raise ValueError(
        "Recurrent state must not be equated with transformation."
    )


if RECURRENT_STATE_EQUALS_TRANSFORMATIVE_RESULT:
    raise ValueError(
        "Recurrent state must not be equated with the resulting "
        "transformative state."
    )


if RECURRENT_CUMULATIVE_SUM_ASSUMED:
    raise ValueError(
        "Recurrent state must not be defined as cumulative "
        "summation by default."
    )


# -----------------------------------------------------------------------------
# Source-space availability helpers
# -----------------------------------------------------------------------------

def recurrent_factual_input_available(
    sentence_id,
):
    """
    Determine whether factual evidence is available for recurrence.
    """

    return bool(
        FACTUAL_SENTENCE_BY_ID[
            sentence_id
        ][
            "availability_mask"
        ].any()
    )


def recurrent_psychological_input_available(
    sentence_id,
):
    """
    Determine whether psychological evidence is available for recurrence.
    """

    return bool(
        PSYCHOLOGICAL_SENTENCE_BY_ID[
            sentence_id
        ][
            "availability_mask"
        ].any()
    )


def recurrent_social_input_available(
    sentence_id,
):
    """
    Determine whether social evidence is available for recurrence.
    """

    return bool(
        any(
            SOCIAL_SENTENCE_BY_ID[
                sentence_id
            ][
                "availability_mask"
            ].values()
        )
    )


def recurrent_transformation_available(
    sentence_id,
):
    """
    Determine whether any architectural transformation is available.
    """

    transformation_mask = (
        TRANSFORMATIVE_SENTENCE_BY_ID[
            sentence_id
        ][
            "transformation_availability_mask"
        ]
    )

    return bool(
        any(
            transformation_mask.values()
        )
    )


# -----------------------------------------------------------------------------
# Current recurrent input reference
# -----------------------------------------------------------------------------

def build_recurrent_input_reference(
    sentence_id,
):
    """
    Build the recurrent input reference for one sentence.
    """

    factual_available = (
        recurrent_factual_input_available(
            sentence_id
        )
    )

    psychological_available = (
        recurrent_psychological_input_available(
            sentence_id
        )
    )

    social_available = (
        recurrent_social_input_available(
            sentence_id
        )
    )

    transformation_available = (
        recurrent_transformation_available(
            sentence_id
        )
    )


    space_availability = {
        "factual":
            factual_available,

        "psychological":
            psychological_available,

        "social":
            social_available,
    }


    return {
        "sentence_id":
            sentence_id,

        "factual_reference":
            sentence_id,

        "psychological_reference":
            sentence_id,

        "social_reference":
            sentence_id,

        "transformative_reference":
            sentence_id,

        "space_availability":
            space_availability,

        "transformation_available":
            transformation_available,

        "any_space_available":
            any(
                space_availability.values()
            ),

        "all_spaces_available":
            all(
                space_availability.values()
            ),
    }


# -----------------------------------------------------------------------------
# Previous recurrent-state reference
# -----------------------------------------------------------------------------

def build_previous_recurrent_reference(
    article_id,
    previous_sentence_id,
):
    """
    Build the previous recurrent-state reference.

    Each article starts from an explicit unavailable initial state.
    """

    if previous_sentence_id is None:

        return {
            "state_type":
                "initial_recurrent_state",

            "article_id":
                article_id,

            "previous_sentence_id":
                None,

            "initial_state_policy":
                RECURRENT_INITIAL_STATE_POLICY,

            "state_available":
                False,

            "substantive_zero":
                False,
        }


    return {
        "state_type":
            "preceding_recurrent_state_reference",

        "article_id":
            article_id,

        "previous_sentence_id":
            previous_sentence_id,

        "initial_state_policy":
            None,

        "state_available":
            False,

        "substantive_zero":
            False,
    }


# -----------------------------------------------------------------------------
# Empty recurrent-state structure
# -----------------------------------------------------------------------------

def create_unavailable_recurrent_state():
    """
    Create a recurrent-state structure without fabricated values.
    """

    return {
        "factual": {
            "value":
                None,

            "value_state":
                "unavailable",
        },

        "psychological": {
            "value":
                None,

            "value_state":
                "unavailable",
        },

        "social": {
            "value":
                None,

            "value_state":
                "unavailable",
        },
    }


def create_recurrent_availability_mask():
    """
    Create recurrent-state availability indicators.
    """

    return {
        "factual":
            0,

        "psychological":
            0,

        "social":
            0,
    }


# -----------------------------------------------------------------------------
# Recurrent-state provenance
# -----------------------------------------------------------------------------

def build_recurrent_provenance(
    sentence_id,
    article_position,
    paragraph_boundary,
    initial_state,
):
    """
    Build recurrent-state provenance for one sentence.
    """

    return {
        "media_ai_record_id":
            SENTENCE_TO_ARTICLE_ID[
                sentence_id
            ],

        "paragraph_id":
            SENTENCE_TO_PARAGRAPH_ID[
                sentence_id
            ],

        "sentence_id":
            sentence_id,

        "representation_level":
            RECURRENT_STATE_LEVEL,

        "recurrent_state_version":
            RECURRENT_STATE_VERSION,

        "construction_method":
            "recurrent_interface_only",

        "article_sentence_position":
            article_position,

        "paragraph_boundary":
            paragraph_boundary,

        "initial_state":
            initial_state,

        "initial_state_policy":
            (
                RECURRENT_INITIAL_STATE_POLICY
                if initial_state
                else None
            ),

        "future_context_used":
            False,

        "cross_article_state_used":
            False,

        "cumulative_sum_assumed":
            False,

        "recurrent_model_executed":
            False,

        "constructed_at_utc":
            datetime.now(
                timezone.utc
            ).isoformat(),
    }


# -----------------------------------------------------------------------------
# Sentence-level recurrent record construction
# -----------------------------------------------------------------------------

RECURRENT_SENTENCE_RECORDS = []


for (
    article_id,
    ordered_sentence_ids,
) in ARTICLE_SENTENCE_ORDER.items():

    previous_sentence_id = None
    previous_paragraph_id = None


    for (
        article_position,
        sentence_id,
    ) in enumerate(
        ordered_sentence_ids
    ):

        paragraph_id = (
            SENTENCE_TO_PARAGRAPH_ID[
                sentence_id
            ]
        )


        initial_state = (
            previous_sentence_id
            is None
        )


        paragraph_boundary = (
            previous_paragraph_id
            is not None
            and paragraph_id
            != previous_paragraph_id
        )


        previous_recurrent_reference = (
            build_previous_recurrent_reference(
                article_id=
                    article_id,

                previous_sentence_id=
                    previous_sentence_id,
            )
        )


        recurrent_input_reference = (
            build_recurrent_input_reference(
                sentence_id
            )
        )


        recurrent_state = (
            create_unavailable_recurrent_state()
        )


        recurrent_availability_mask = (
            create_recurrent_availability_mask()
        )


        recurrent_record = {
            "record_type":
                "media_ai_recurrent_state",

            "recurrent_state_version":
                RECURRENT_STATE_VERSION,

            "media_ai_record_id":
                article_id,

            "paragraph_id":
                paragraph_id,

            "sentence_id":
                sentence_id,

            "article_sentence_position":
                article_position,

            "previous_sentence_id":
                previous_sentence_id,

            "initial_state":
                initial_state,

            "paragraph_boundary":
                paragraph_boundary,

            "previous_recurrent_state_reference":
                previous_recurrent_reference,

            "current_input_reference":
                recurrent_input_reference,

            "recurrent_state":
                recurrent_state,

            "recurrent_availability_mask":
                recurrent_availability_mask,

            "transformative_state_reference":
                sentence_id,

            "confluent_state_reference":
                None,

            "provenance":
                build_recurrent_provenance(
                    sentence_id=
                        sentence_id,

                    article_position=
                        article_position,

                    paragraph_boundary=
                        paragraph_boundary,

                    initial_state=
                        initial_state,
                ),
        }


        RECURRENT_SENTENCE_RECORDS.append(
            recurrent_record
        )


        previous_sentence_id = (
            sentence_id
        )

        previous_paragraph_id = (
            paragraph_id
        )


# -----------------------------------------------------------------------------
# Sentence-level lookup
# -----------------------------------------------------------------------------

RECURRENT_SENTENCE_BY_ID = {
    record[
        "sentence_id"
    ]:
        record

    for record
    in RECURRENT_SENTENCE_RECORDS
}


# -----------------------------------------------------------------------------
# Article recurrent trajectories
# -----------------------------------------------------------------------------

RECURRENT_ARTICLE_TRAJECTORIES = []


for (
    article_id,
    ordered_sentence_ids,
) in ARTICLE_SENTENCE_ORDER.items():

    trajectory = {
        "media_ai_record_id":
            article_id,

        "sentence_ids":
            list(
                ordered_sentence_ids
            ),

        "state_record_ids":
            list(
                ordered_sentence_ids
            ),

        "state_count":
            len(
                ordered_sentence_ids
            ),

        "trajectory_available":
            False,

        "substantive_states_available":
            0,
    }


    RECURRENT_ARTICLE_TRAJECTORIES.append(
        trajectory
    )


# -----------------------------------------------------------------------------
# Recurrent sequence validation
# -----------------------------------------------------------------------------

RECURRENT_STATE_ERRORS = []


expected_recurrent_record_count = sum(
    len(
        sentence_ids
    )
    for sentence_ids in ARTICLE_SENTENCE_ORDER.values()
)


if len(
    RECURRENT_SENTENCE_RECORDS
) != expected_recurrent_record_count:

    RECURRENT_STATE_ERRORS.append(
        "recurrent_record_count_mismatch"
    )


for (
    article_id,
    ordered_sentence_ids,
) in ARTICLE_SENTENCE_ORDER.items():

    for index, sentence_id in enumerate(
        ordered_sentence_ids
    ):

        record = (
            RECURRENT_SENTENCE_BY_ID[
                sentence_id
            ]
        )


        expected_previous_sentence_id = (
            ordered_sentence_ids[
                index - 1
            ]
            if index > 0
            else None
        )


        if (
            record[
                "previous_sentence_id"
            ]
            != expected_previous_sentence_id
        ):

            RECURRENT_STATE_ERRORS.append(
                "invalid_previous_recurrent_reference"
            )

            break


        if (
            index == 0
            and not record[
                "initial_state"
            ]
        ):

            RECURRENT_STATE_ERRORS.append(
                "article_initial_recurrent_state_not_marked"
            )

            break


        if (
            index > 0
            and record[
                "initial_state"
            ]
        ):

            RECURRENT_STATE_ERRORS.append(
                "non_initial_recurrent_record_marked_initial"
            )

            break


    if RECURRENT_STATE_ERRORS:
        break


# -----------------------------------------------------------------------------
# Cross-article validation
# -----------------------------------------------------------------------------

for record in RECURRENT_SENTENCE_RECORDS:

    previous_sentence_id = (
        record[
            "previous_sentence_id"
        ]
    )


    if previous_sentence_id is None:
        continue


    if (
        SENTENCE_TO_ARTICLE_ID[
            previous_sentence_id
        ]
        != record[
            "media_ai_record_id"
        ]
    ):

        RECURRENT_STATE_ERRORS.append(
            "cross_article_recurrent_reference_detected"
        )

        break


# -----------------------------------------------------------------------------
# Recurrent-state value validation
# -----------------------------------------------------------------------------

for record in RECURRENT_SENTENCE_RECORDS:

    recurrent_state = (
        record[
            "recurrent_state"
        ]
    )

    availability_mask = (
        record[
            "recurrent_availability_mask"
        ]
    )


    if set(
        recurrent_state.keys()
    ) != set(
        RECURRENT_SOURCE_SPACES
    ):

        RECURRENT_STATE_ERRORS.append(
            "invalid_recurrent_space_structure"
        )

        break


    if set(
        availability_mask.keys()
    ) != set(
        RECURRENT_SOURCE_SPACES
    ):

        RECURRENT_STATE_ERRORS.append(
            "invalid_recurrent_mask_structure"
        )

        break


    for space in RECURRENT_SOURCE_SPACES:

        if availability_mask[
            space
        ] not in (
            0,
            1,
        ):

            RECURRENT_STATE_ERRORS.append(
                "invalid_recurrent_mask_value"
            )

            break


        space_state = (
            recurrent_state[
                space
            ]
        )


        if availability_mask[
            space
        ] == 0:

            if (
                space_state[
                    "value"
                ]
                is not None
            ):

                RECURRENT_STATE_ERRORS.append(
                    "unavailable_recurrent_state_has_value"
                )

                break


            if (
                space_state[
                    "value_state"
                ]
                != "unavailable"
            ):

                RECURRENT_STATE_ERRORS.append(
                    "invalid_unavailable_recurrent_state"
                )

                break


    if RECURRENT_STATE_ERRORS:
        break


# -----------------------------------------------------------------------------
# Initial-state accounting
# -----------------------------------------------------------------------------

RECURRENT_ARTICLE_INITIAL_STATE_COUNT = sum(
    1
    for record in RECURRENT_SENTENCE_RECORDS
    if record[
        "initial_state"
    ]
)


RECURRENT_PARAGRAPH_BOUNDARY_COUNT = sum(
    1
    for record in RECURRENT_SENTENCE_RECORDS
    if record[
        "paragraph_boundary"
    ]
)


# -----------------------------------------------------------------------------
# Input availability accounting
# -----------------------------------------------------------------------------

RECURRENT_FACTUAL_INPUT_AVAILABLE = sum(
    1
    for record in RECURRENT_SENTENCE_RECORDS
    if record[
        "current_input_reference"
    ][
        "space_availability"
    ][
        "factual"
    ]
)


RECURRENT_PSYCHOLOGICAL_INPUT_AVAILABLE = sum(
    1
    for record in RECURRENT_SENTENCE_RECORDS
    if record[
        "current_input_reference"
    ][
        "space_availability"
    ][
        "psychological"
    ]
)


RECURRENT_SOCIAL_INPUT_AVAILABLE = sum(
    1
    for record in RECURRENT_SENTENCE_RECORDS
    if record[
        "current_input_reference"
    ][
        "space_availability"
    ][
        "social"
    ]
)


RECURRENT_TRANSFORMATION_INPUT_AVAILABLE = sum(
    1
    for record in RECURRENT_SENTENCE_RECORDS
    if record[
        "current_input_reference"
    ][
        "transformation_available"
    ]
)


RECURRENT_ANY_INPUT_AVAILABLE = sum(
    1
    for record in RECURRENT_SENTENCE_RECORDS
    if record[
        "current_input_reference"
    ][
        "any_space_available"
    ]
)


RECURRENT_ALL_INPUT_AVAILABLE = sum(
    1
    for record in RECURRENT_SENTENCE_RECORDS
    if record[
        "current_input_reference"
    ][
        "all_spaces_available"
    ]
)


# -----------------------------------------------------------------------------
# Recurrent-state availability accounting
# -----------------------------------------------------------------------------

RECURRENT_FACTUAL_STATE_AVAILABLE = sum(
    record[
        "recurrent_availability_mask"
    ][
        "factual"
    ]
    for record in RECURRENT_SENTENCE_RECORDS
)


RECURRENT_PSYCHOLOGICAL_STATE_AVAILABLE = sum(
    record[
        "recurrent_availability_mask"
    ][
        "psychological"
    ]
    for record in RECURRENT_SENTENCE_RECORDS
)


RECURRENT_SOCIAL_STATE_AVAILABLE = sum(
    record[
        "recurrent_availability_mask"
    ][
        "social"
    ]
    for record in RECURRENT_SENTENCE_RECORDS
)


# -----------------------------------------------------------------------------
# Recurrent contract
# -----------------------------------------------------------------------------

RECURRENT_STATE_CONSTRUCTION_CONTRACT = {
    "version":
        RECURRENT_STATE_VERSION,

    "level":
        RECURRENT_STATE_LEVEL,

    "source_spaces":
        RECURRENT_SOURCE_SPACES,

    "initial_state_policy":
        RECURRENT_INITIAL_STATE_POLICY,

    "future_context_allowed":
        False,

    "cross_article_state":
        False,

    "state_equals_current_evidence":
        False,

    "state_equals_transformation":
        False,

    "state_equals_transformative_result":
        False,

    "cumulative_sum_definition":
        False,

    "persistence_allowed":
        True,

    "decay_allowed":
        True,

    "forgetting_allowed":
        True,

    "gating_allowed":
        True,

    "operation_order_fixed":
        False,

    "hierarchical_aggregation_performed":
        False,

    "backward_attribution_required":
        True,

    "counterfactual_compatible":
        True,
}


# -----------------------------------------------------------------------------
# Recurrent-transformative boundary
# -----------------------------------------------------------------------------

RECURRENT_TRANSFORMATIVE_BOUNDARY = {
    "transformation_represents_change":
        True,

    "recurrence_represents_state_propagation":
        True,

    "mechanisms_equivalent":
        False,

    "cumulative_transform_equals_recurrent_state":
        False,
}


# -----------------------------------------------------------------------------
# Recurrent-confluent boundary
# -----------------------------------------------------------------------------

RECURRENT_CONFLUENT_BOUNDARY_BLOCK_10 = {
    "recurrence_operates_across_sequence":
        True,

    "confluence_operates_across_spaces":
        True,

    "mechanisms_equivalent":
        False,

    "operation_order_fixed":
        RECURRENT_OPERATION_ORDER_FIXED,

    "operation_order_is_architectural_hypothesis":
        True,
}


# -----------------------------------------------------------------------------
# Recurrent trajectory contract
# -----------------------------------------------------------------------------

RECURRENT_TRAJECTORY_CONTRACT = {
    "sentence_order_preserved":
        True,

    "article_specific":
        True,

    "cross_article_propagation":
        False,

    "paragraph_boundaries_preserved":
        True,

    "derived_state_difference_allowed_later":
        True,

    "derived_state_difference_equals_transformation":
        False,

    "trajectory_attribution_required":
        True,
}


# -----------------------------------------------------------------------------
# Representation status
# -----------------------------------------------------------------------------

RECURRENT_STATE_VALID = (
    len(
        RECURRENT_STATE_ERRORS
    )
    == 0
)


if not RECURRENT_STATE_VALID:
    raise ValueError(
        "Recurrent state construction failed validation: "
        f"{RECURRENT_STATE_ERRORS}"
    )


RECURRENT_STATE_RECORDS_CONSTRUCTED = True

RECURRENT_STATE_CALCULATED = False

RECURRENT_SUBSTANTIVE_VALUES_PRESENT = any([
    RECURRENT_FACTUAL_STATE_AVAILABLE > 0,
    RECURRENT_PSYCHOLOGICAL_STATE_AVAILABLE > 0,
    RECURRENT_SOCIAL_STATE_AVAILABLE > 0,
])


# -----------------------------------------------------------------------------
# Summary
# -----------------------------------------------------------------------------

RECURRENT_STATE_SUMMARY = {
    "version":
        RECURRENT_STATE_VERSION,

    "level":
        RECURRENT_STATE_LEVEL,

    "sentence_records":
        len(
            RECURRENT_SENTENCE_RECORDS
        ),

    "article_trajectories":
        len(
            RECURRENT_ARTICLE_TRAJECTORIES
        ),

    "article_initial_states":
        RECURRENT_ARTICLE_INITIAL_STATE_COUNT,

    "paragraph_boundaries":
        RECURRENT_PARAGRAPH_BOUNDARY_COUNT,

    "factual_inputs_available":
        RECURRENT_FACTUAL_INPUT_AVAILABLE,

    "psychological_inputs_available":
        RECURRENT_PSYCHOLOGICAL_INPUT_AVAILABLE,

    "social_inputs_available":
        RECURRENT_SOCIAL_INPUT_AVAILABLE,

    "transformative_inputs_available":
        RECURRENT_TRANSFORMATION_INPUT_AVAILABLE,

    "any_inputs_available":
        RECURRENT_ANY_INPUT_AVAILABLE,

    "all_inputs_available":
        RECURRENT_ALL_INPUT_AVAILABLE,

    "factual_states_available":
        RECURRENT_FACTUAL_STATE_AVAILABLE,

    "psychological_states_available":
        RECURRENT_PSYCHOLOGICAL_STATE_AVAILABLE,

    "social_states_available":
        RECURRENT_SOCIAL_STATE_AVAILABLE,

    "records_constructed":
        RECURRENT_STATE_RECORDS_CONSTRUCTED,

    "state_calculated":
        RECURRENT_STATE_CALCULATED,

    "substantive_values_present":
        RECURRENT_SUBSTANTIVE_VALUES_PRESENT,

    "valid":
        RECURRENT_STATE_VALID,
}


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Recurrent State Construction")
print("=" * 72)

print(
    f"Recurrent state version     : "
    f"{RECURRENT_STATE_VERSION}"
)

print(
    f"Representation level        : "
    f"{RECURRENT_STATE_LEVEL}"
)

print(
    f"Source spaces               : "
    f"{', '.join(RECURRENT_SOURCE_SPACES)}"
)

print(
    f"Sentence records            : "
    f"{len(RECURRENT_SENTENCE_RECORDS)}"
)

print(
    f"Article trajectories        : "
    f"{len(RECURRENT_ARTICLE_TRAJECTORIES)}"
)

print("-" * 72)

print("Sequential structure")

print(
    f"Article initial states      : "
    f"{RECURRENT_ARTICLE_INITIAL_STATE_COUNT}"
)

print(
    f"Initial-state policy        : "
    f"{RECURRENT_INITIAL_STATE_POLICY}"
)

print(
    f"Paragraph boundaries        : "
    f"{RECURRENT_PARAGRAPH_BOUNDARY_COUNT}"
)

print(
    f"Cross-article state         : "
    f"{RECURRENT_CROSS_ARTICLE_STATE}"
)

print(
    f"Future context allowed      : "
    f"{ALLOW_FUTURE_CONTEXT_IN_STATE_CONSTRUCTION}"
)

print("-" * 72)

print("Current recurrent inputs")

print(
    f"Factual inputs              : "
    f"{RECURRENT_FACTUAL_INPUT_AVAILABLE}"
)

print(
    f"Psychological inputs        : "
    f"{RECURRENT_PSYCHOLOGICAL_INPUT_AVAILABLE}"
)

print(
    f"Social inputs               : "
    f"{RECURRENT_SOCIAL_INPUT_AVAILABLE}"
)

print(
    f"Transformative inputs       : "
    f"{RECURRENT_TRANSFORMATION_INPUT_AVAILABLE}"
)

print(
    f"Any source space available  : "
    f"{RECURRENT_ANY_INPUT_AVAILABLE}"
)

print(
    f"All source spaces available : "
    f"{RECURRENT_ALL_INPUT_AVAILABLE}"
)

print("-" * 72)

print("Recurrent-state availability")

print(
    f"Factual states              : "
    f"{RECURRENT_FACTUAL_STATE_AVAILABLE}"
)

print(
    f"Psychological states        : "
    f"{RECURRENT_PSYCHOLOGICAL_STATE_AVAILABLE}"
)

print(
    f"Social states               : "
    f"{RECURRENT_SOCIAL_STATE_AVAILABLE}"
)

print(
    f"Substantive values present  : "
    f"{RECURRENT_SUBSTANTIVE_VALUES_PRESENT}"
)

print("-" * 72)

print("Methodological boundaries")

print(
    f"State = current evidence    : "
    f"{RECURRENT_STATE_EQUALS_CURRENT_EVIDENCE}"
)

print(
    f"State = transformation      : "
    f"{RECURRENT_STATE_EQUALS_TRANSFORMATION}"
)

print(
    f"State = transform result    : "
    f"{RECURRENT_STATE_EQUALS_TRANSFORMATIVE_RESULT}"
)

print(
    f"Cumulative sum assumed      : "
    f"{RECURRENT_CUMULATIVE_SUM_ASSUMED}"
)

print(
    f"Recurrence = transformation : "
    f"{RECURRENT_TRANSFORMATIVE_BOUNDARY['mechanisms_equivalent']}"
)

print(
    f"Recurrence = confluence     : "
    f"{RECURRENT_CONFLUENT_BOUNDARY_BLOCK_10['mechanisms_equivalent']}"
)

print(
    f"Operation order fixed       : "
    f"{RECURRENT_CONFLUENT_BOUNDARY_BLOCK_10['operation_order_fixed']}"
)

print("-" * 72)

print(
    f"Recurrent records valid     : "
    f"{RECURRENT_STATE_VALID}"
)

print(
    f"Records constructed         : "
    f"{RECURRENT_STATE_RECORDS_CONSTRUCTED}"
)

print(
    f"Recurrent state calculated  : "
    f"{RECURRENT_STATE_CALCULATED}"
)

print("=" * 72)

print(
    "Sentence-level recurrent-state records constructed successfully."
)

print(
    "Previous recurrent state, current evidence and transformation "
    "remain separate architectural inputs."
)

print(
    "Article initial recurrent states are explicit and have not "
    "been interpreted as substantive zero states."
)

print(
    "Recurrent state has not been equated with current evidence, "
    "transformation or the resulting transformative state."
)

print(
    "Recurrent state has not been defined as cumulative transformation."
)

print(
    "Sentence ordering, paragraph boundaries and article boundaries "
    "remain explicit."
)

print(
    "No cross-article recurrent propagation has been performed."
)

print(
    "No unavailable recurrent state has been converted to zero."
)

print(
    "No hierarchical recurrent aggregation or attribution score "
    "has been calculated."
)

print(
    "No counterfactual experiment, model training or experimental "
    "evaluation has been performed."
)
````

## Block 11 — Group-Aware Confluent Conditioning and Representation Routing

This block defines the group-aware confluent conditioning mechanism that links the global text representation with dynamically detected article-specific groups.

Its purpose is to create and preserve two parallel representation paths:

(X_t^{global}) — the global, group-independent representation of textual unit (t);

(X_{t,g}^{group}) — the representation of the same textual unit conditioned on group (g).

Both paths remain available throughout later factual, psychological, social, transformative, recurrent and hierarchical processing.

The central architectural principle is:

(global\ representation \neq group\text{-}conditioned\ representation).

The group-conditioned path supplements the global path. It does not replace it.

### Global Backbone Representation

For textual unit (t), the pretrained backbone produces a global representation:

(X_t^{global} = Backbone(s_t)).

Dictionary:

* (s_t) = textual unit at step (t);
* (Backbone) = pretrained text encoder;
* (X_t^{global}) = global contextual representation of the textual unit.

The global representation is constructed independently of any particular group identity.

It therefore provides the group-independent textual basis for the downstream representation heads.

### Group Candidate Detection

The architecture does not use a closed predefined list of concrete groups.

Instead, it detects article-specific group candidates from the textual context.

Let:

(G_t = {g_1, g_2, \ldots, g_k})

denote the groups active or relevant at textual step (t).

The number of active groups (k) is variable.

Concrete group identities are therefore dynamically detected rather than globally predefined.

### Group Ontology

Although concrete groups are dynamically detected, the group ontology is predefined.

The ontology specifies what kinds of social entities may qualify as groups.

Possible group types include:

* organisation;
* institution;
* political group;
* national group;
* professional group;
* ideological group;
* social group;
* ad hoc collective;
* other explicitly supported collective categories.

The ontology constrains interpretation without restricting the model to a closed list of group identities.

### Actors and Groups

Actors and groups remain distinct constructs.

An actor may represent a person or individual entity.

A group represents a socially meaningful collective or collective entity.

Thus:

(Actor \neq Group).

An actor may nevertheless be associated with one or more groups.

Overlapping memberships remain permitted.

### Group Detection Supervision

Group detection may later be supervised using structured annotations produced by independent supervisors and, where applicable, human review.

The supervisor may provide:

* group mentions;
* group type;
* actor-group relationships;
* membership relations;
* evidence spans;
* aliases;
* group relevance;
* relationship structure.

The supervisor must also be allowed to return **no socially relevant group**.

Therefore:

(G_t = \varnothing)

is a valid observed outcome.

The architecture must not force the existence of a group merely because group-conditioned processing is available.

### Global-Only Path

Where no relevant group is detected, only the global path remains active.

Thus, if (G_t = \varnothing), downstream processing continues using (X_t^{global}).

No artificial group-conditioned branch is created.

### Group-Conditioned Representation

For each active group (g \in G_t), the confluent conditioning mechanism constructs:

(X_{t,g}^{group} = \Phi_C(X_t^{global}, E_g; \theta_C)).

Dictionary:

* (X_t^{global}) = global textual representation;
* (E_g) = representation of detected group (g);
* (\Phi_C) = group-aware confluent conditioning mechanism;
* (\theta_C) = trainable confluent parameters;
* (X_{t,g}^{group}) = representation conditioned on group (g).

The confluent mechanism therefore operates **before** the factual, psychological and social heads.

### Confluence as Conditioning

In this architecture, confluence is not defined as late fusion of completed factual, psychological and social states.

Instead, confluence provides the contextual transformation that allows the same text to be processed:

* globally;
* relative to a specific group.

Thus:

(X_t^{global} \rightarrow X_{t,g}^{group})

is the fundamental confluent operation at this stage.

### Confluence Is Not Concatenation

Group conditioning is not defined as simple concatenation.

Therefore:

(X_{t,g}^{group} \neq [X_t^{global};E_g])

by definition.

Concatenation may later serve as a baseline, but the trainable confluent mechanism may instead use attention, gating, modulation, cross-attention or another explicitly defined interaction mechanism.

### Shared and Conditioned Information

The group-conditioned representation must preserve information from the global representation while permitting group-specific modulation.

Conceptually:

(X_{t,g}^{group} = X_t^{global} + A_C(X_t^{global}, E_g)).

Here (A_C) denotes a trainable group-conditioning adjustment.

This expression is conceptual rather than a mandatory implementation.

The important requirement is that group conditioning modifies rather than replaces the global textual representation.

### Representation Routing

Both global and group-conditioned representations are routed to the downstream heads.

The global route is:

(X_t^{global} \rightarrow {Head_F, Head_P, Head_S}).

For group (g), the conditioned route is:

(X_{t,g}^{group} \rightarrow {Head_F, Head_P, Head_S}).

Thus each head receives two kinds of input:

* global input;
* group-conditioned input.

### Factual Routing

The factual head produces:

(F_t^{global})

and, for each active group (g):

(F_{t,g}^{group}).

The global factual representation describes the factual structure of the textual unit without conditioning on a particular group.

The group-conditioned factual representation describes how that factual structure is positioned relative to group (g).

### Objective and Subjective Factual Interpretation

The factual branch preserves the distinction between:

(F_t^{objective})

and:

(F_{t,g}^{subjective}).

Here `objective` means **group-independent**, not externally verified truth.

Thus:

(F_t^{objective} \equiv F_t^{global})

and:

(F_{t,g}^{subjective} \equiv F_{t,g}^{group}).

The objective/global representation answers:

**What factual structure is present in the text overall?**

The subjective/group-conditioned representation answers:

**How is that factual structure positioned relative to group (g)?**

### Psychological Routing

The psychological head likewise produces:

(P_t^{global})

and:

(P_{t,g}^{group}).

The global representation describes the overall psychological framing of the textual unit.

The group-conditioned representation describes psychological framing associated with or directed toward group (g).

This permits, for example, fear to increase relative to one group while trust increases relative to another.

### Social Routing

The social head produces:

(S_t^{global})

and:

(S_{t,g}^{group}).

The global social representation describes the overall relational structure.

The group-conditioned social representation describes the relational structure centred on group (g).

The group-conditioned social path may therefore emphasise:

* relationships involving (g);
* memberships involving (g);
* alliances involving (g);
* conflicts involving (g);
* relational orientation around (g).

### Shared Head Ontology

Global and group-conditioned branches must use the same underlying representation ontology.

For example, `explicit_claim`, `uncertainty`, `fear`, or `trust` must retain the same semantic meaning in both global and conditioned outputs.

Therefore, global and group-conditioned heads should preferentially share core parameters.

### Shared-Core Parameterisation

For factual representation, a shared-core design may be written as:

(H_t^{F,global} = \phi_F(X_t^{global};\theta_F))

and:

(H_{t,g}^{F,group} = \phi_F(X_{t,g}^{group};\theta_F) + A_F(X_{t,g}^{group},E_g;\theta_F^{cond})).

Dictionary:

* (\theta_F) = shared factual parameters;
* (\theta_F^{cond}) = group-conditioning parameters;
* (A_F) = factual group-adaptation mechanism.

The same principle may later be applied to psychological and social processing.

### Independent Global Path

The global representation must not be reconstructed from the group-conditioned paths.

Therefore:

(X_t^{global} \neq \frac{1}{|G_t|}\sum_{g \in G_t} X_{t,g}^{group}).

Likewise:

(F_t^{global} \neq \frac{1}{|G_t|}\sum_g F_{t,g}^{group}),

(P_t^{global} \neq \frac{1}{|G_t|}\sum_g P_{t,g}^{group}),

and:

(S_t^{global} \neq A({S_{t,g}^{group}}))

unless such a relationship is explicitly introduced and tested later.

The global path remains independently learned and preserved.

### Group Deviation

The difference between global and group-conditioned representation is an explicit Media AI quantity.

For representation space (X \in {F,P,S}), group deviation is:

(D_{t,g}^{X} = X_{t,g}^{group} - X_t^{global}).

Dictionary:

* (D_{t,g}^{X}) = group deviation in representation space (X);
* (X_t^{global}) = global representation;
* (X_{t,g}^{group}) = group-conditioned representation.

For fixed-dimensional spaces, the deviation remains dimension-specific.

### Signed Group Deviation

Group deviation remains signed.

For dimension (j):

(D_{t,g,j}^{X} > 0)

means the group-conditioned value exceeds the global value.

(D_{t,g,j}^{X} < 0)

means the group-conditioned value is below the global value.

(D_{t,g,j}^{X} = 0)

means no observed deviation in that dimension.

This directional information must not be discarded by immediately converting deviations to absolute values.

### Deviation Magnitude

Where useful, a later derived magnitude may be computed as:

(|D_{t,g}^{X}|).

This measures the overall size of group deviation.

Deviation magnitude does not replace the signed dimensional representation.

### Deviation Is Not Bias

Group deviation is descriptive.

Therefore:

(D_{t,g}^{X} \neq Bias_{t,g}).

A large deviation indicates that group-conditioned representation differs substantially from the global representation.

It does not by itself establish:

* bias;
* discrimination;
* manipulation;
* hostility;
* unfairness.

Interpretation depends on the affected dimensions, textual evidence, context and later experimental analysis.

### Group Relevance

Not every detected group must remain active at every textual step.

Let (a_{t,g} \in [0,1]) denote the relevance of group (g) at step (t).

The relevance score may later influence whether the group-conditioned path is active or how strongly it participates in downstream processing.

The relevance score is not itself a factual, psychological or social representation dimension.

### Sparse Group Routing

To control computational growth, the architecture supports sparse group routing.

Only article-relevant groups need to maintain conditioned state.

For article (a), let (G_a^{active}) denote the active group set.

Group-conditioned processing is performed only for (g \in G_a^{active}).

This prevents the model from maintaining meaningless state for groups unrelated to the article.

### Group Persistence

A detected group may remain relevant across later sentences even when it is not explicitly repeated.

Therefore, group activity may depend on:

* current mention;
* previous mention;
* resolved coreference;
* recurrent relevance;
* explicit deactivation policy.

Group persistence is handled later through the recurrent mechanism.

This block establishes the identity and routing contract only.

### Global Transformative Path

The global representation proceeds through the transformative mechanism:

(X_t^{global} \rightarrow R_t^{global} \rightarrow \Delta_t^{global}).

This applies separately to factual, psychological and social representations.

### Group-Conditioned Transformative Path

Each active group-conditioned representation also proceeds through transformation:

(X_{t,g}^{group} \rightarrow R_{t,g}^{group} \rightarrow \Delta_{t,g}^{group}).

Therefore, both global and group-conditioned framing can change through the article.

### Global Recurrent Path

The global representations maintain recurrent states:

(H_t^{F,global}), (H_t^{P,global}), and (H_t^{S,global}).

These capture the overall factual, psychological and social trajectories of the article.

### Group-Conditioned Recurrent Path

For each active group (g), the architecture also maintains:

(H_{t,g}^{F}), (H_{t,g}^{P}), and (H_{t,g}^{S}).

These states track how the factual, psychological and social representations associated with group (g) evolve through the article.

### Global and Group Recurrent Independence

Global and group-conditioned recurrent states are not required to coincide.

Therefore:

(H_t^{X,global} \neq H_{t,g}^{X})

in general.

The group-conditioned state is a contextual view of the same article, not a duplicate of the global state.

### No Confluent Feedback in Version 1.0

The initial Media AI architecture does not feed the previous confluent group-conditioned state back into the next global backbone representation.

Thus:

(C_{t-1} \not\rightarrow X_t^{global})

in version 1.0.

This keeps the first trainable architecture simpler and improves interpretability.

Confluent feedback may later be tested as an architectural extension.

### Group Identity Traceability

Every group-conditioned representation must preserve:

* stable group identifier;
* group type;
* textual mentions;
* supporting evidence spans;
* article identity;
* sentence identity;
* group-resolution provenance;
* relevance metadata.

This ensures that group-conditioned states remain traceable to the actual language that generated them.

### Missing Group Information

Unavailable group information must remain distinct from an observed absence of groups.

Thus:

(group\ unavailable \neq G_t = \varnothing).

The first means that group detection has not produced a usable result.

The second means that the model or supervisor has explicitly determined that no relevant group is present.

### Training Boundary

This block defines the trainable group-aware routing architecture but does not yet execute model training.

The trainable components introduced conceptually here include:

* group detector;
* group embedding or representation;
* confluent conditioning mechanism;
* group-relevance mechanism;
* optional group-conditioned adaptation parameters.

The factual, psychological and social head parameterisations are defined in the following blocks.

### Output

This block establishes:

* global backbone representation;
* dynamic group detection;
* predefined group ontology;
* actor-group distinction;
* group-conditioned representation construction;
* early confluent conditioning;
* global and group-conditioned routing;
* objective/global and subjective/group-conditioned factual interpretation;
* shared-core head principle;
* independent global representation;
* signed group-deviation representation;
* group-deviation magnitude boundary;
* deviation-versus-bias distinction;
* sparse active-group routing;
* global and group-conditioned transformative paths;
* global and group-conditioned recurrent paths;
* no confluent-feedback policy for version 1.0;
* group identity and evidence traceability;
* trainable confluent parameterisation boundary.

No concrete group identities are globally predefined.

No group is fabricated where none is detected.

No global representation is reconstructed by averaging group-conditioned representations.

No group deviation is automatically interpreted as bias.

No transformative or recurrent state is calculated by this block.

No hierarchical aggregation is performed.

No attribution score is calculated.

No model training or experimental evaluation is performed.


````python
# =============================================================================
# Media AI — Group-Aware Confluent Conditioning and Representation Routing
# =============================================================================

from datetime import datetime, timezone


# -----------------------------------------------------------------------------
# Block dependency checks
# -----------------------------------------------------------------------------

required_block_1_objects = [
    "BUILD_CONFLUENT_REPRESENTATION",
    "BUILD_TRANSFORMATIVE_STATE",
    "BUILD_RECURRENT_STATE",
    "PRESERVE_TEXTUAL_ORDER",
    "ALLOW_FUTURE_CONTEXT_IN_STATE_CONSTRUCTION",
]

missing_block_1_objects = [
    object_name
    for object_name in required_block_1_objects
    if object_name not in globals()
]

if missing_block_1_objects:
    raise NameError(
        "Notebook 04 Block 1 is not initialised. "
        f"Missing: {missing_block_1_objects}"
    )


required_block_3_objects = [
    "HIERARCHICAL_TEXT_AND_IDENTITY_VALID",
    "SENTENCE_IDS",
    "SENTENCE_TO_PARAGRAPH_ID",
    "SENTENCE_TO_ARTICLE_ID",
    "ARTICLE_SENTENCE_ORDER",
]

missing_block_3_objects = [
    object_name
    for object_name in required_block_3_objects
    if object_name not in globals()
]

if missing_block_3_objects:
    raise NameError(
        "Notebook 04 Block 3 is not initialised. "
        f"Missing: {missing_block_3_objects}"
    )


required_block_4_objects = [
    "MEDIA_AI_REPRESENTATION_SPECIFICATION_VALID",
    "FACTUAL_DIMENSION_COUNT",
    "PSYCHOLOGICAL_DIMENSION_COUNT",
]

missing_block_4_objects = [
    object_name
    for object_name in required_block_4_objects
    if object_name not in globals()
]

if missing_block_4_objects:
    raise NameError(
        "Notebook 04 Block 4 is not initialised. "
        f"Missing: {missing_block_4_objects}"
    )


required_block_8_objects = [
    "SOCIAL_REPRESENTATION_VALID",
    "SOCIAL_OVERLAPPING_MEMBERSHIP_ALLOWED",
    "SOCIAL_IDENTITY_RESOLUTION_CONTRACT",
    "CONFLUENT_CONTRACT_VALID",
]

missing_block_8_objects = [
    object_name
    for object_name in required_block_8_objects
    if object_name not in globals()
]

if missing_block_8_objects:
    raise NameError(
        "Notebook 04 Block 8 is not initialised. "
        f"Missing: {missing_block_8_objects}"
    )


required_block_9_objects = [
    "TRANSFORMATIVE_STATE_VALID",
    "TRANSFORMATIVE_STATE_CALCULATED",
]

missing_block_9_objects = [
    object_name
    for object_name in required_block_9_objects
    if object_name not in globals()
]

if missing_block_9_objects:
    raise NameError(
        "Notebook 04 Block 9 is not initialised. "
        f"Missing: {missing_block_9_objects}"
    )


required_block_10_objects = [
    "RECURRENT_STATE_VALID",
    "RECURRENT_STATE_CALCULATED",
]

missing_block_10_objects = [
    object_name
    for object_name in required_block_10_objects
    if object_name not in globals()
]

if missing_block_10_objects:
    raise NameError(
        "Notebook 04 Block 10 is not initialised. "
        f"Missing: {missing_block_10_objects}"
    )


if not HIERARCHICAL_TEXT_AND_IDENTITY_VALID:
    raise ValueError(
        "Group-aware routing requires a valid textual hierarchy."
    )


if not MEDIA_AI_REPRESENTATION_SPECIFICATION_VALID:
    raise ValueError(
        "Group-aware routing requires a valid representation specification."
    )


if not SOCIAL_REPRESENTATION_VALID:
    raise ValueError(
        "Group-aware routing requires a valid social representation contract."
    )


if not CONFLUENT_CONTRACT_VALID:
    raise ValueError(
        "Group-aware routing requires a valid confluent contract."
    )


if not TRANSFORMATIVE_STATE_VALID:
    raise ValueError(
        "Group-aware routing requires a valid transformative-state interface."
    )


if not RECURRENT_STATE_VALID:
    raise ValueError(
        "Group-aware routing requires a valid recurrent-state interface."
    )


if not PRESERVE_TEXTUAL_ORDER:
    raise ValueError(
        "Group-aware routing requires preserved textual order."
    )


if ALLOW_FUTURE_CONTEXT_IN_STATE_CONSTRUCTION:
    raise ValueError(
        "Future context must not enter group-aware routing."
    )


# -----------------------------------------------------------------------------
# Group-aware confluent configuration
# -----------------------------------------------------------------------------

GROUP_AWARE_CONFLUENT_VERSION = "1.0"

GROUP_ROUTING_LEVEL = "sentence"

GROUP_CONDITIONING_SCOPE = "global_and_group_conditioned"

CONFLUENT_POSITION_IN_ARCHITECTURE = "pre_head_group_conditioning"

CONFLUENT_FEEDBACK_ENABLED = False

GLOBAL_PATH_ALWAYS_AVAILABLE = True

GROUP_CONDITIONED_PATH_OPTIONAL = True

GLOBAL_RECONSTRUCTED_FROM_GROUPS = False

GROUP_DEVIATION_AUTOMATICALLY_BIAS = False

GROUP_ROUTING_SPARSE = True

GROUP_IDENTITIES_GLOBALLY_PREDEFINED = False


# -----------------------------------------------------------------------------
# Group ontology
# -----------------------------------------------------------------------------

MEDIA_AI_GROUP_TYPES = (
    "organisation",
    "institution",
    "political_group",
    "national_group",
    "professional_group",
    "ideological_group",
    "social_group",
    "ad_hoc_collective",
    "other",
)


MEDIA_AI_ACTOR_TYPES = (
    "person",
    "individual_entity",
)


ACTOR_GROUP_DISTINCTION_REQUIRED = True

OVERLAPPING_GROUP_MEMBERSHIP_ALLOWED = (
    SOCIAL_OVERLAPPING_MEMBERSHIP_ALLOWED
)


# -----------------------------------------------------------------------------
# Group-detection contract
# -----------------------------------------------------------------------------

GROUP_DETECTION_CONTRACT = {
    "version": GROUP_AWARE_CONFLUENT_VERSION,
    "dynamic_group_detection": True,
    "closed_group_identity_vocabulary": False,
    "group_types_predefined": True,
    "group_identities_predefined": False,
    "no_group_is_valid": True,
    "actor_group_distinction_required": ACTOR_GROUP_DISTINCTION_REQUIRED,
    "overlapping_membership_allowed": OVERLAPPING_GROUP_MEMBERSHIP_ALLOWED,
    "supervisor_support_allowed": True,
    "human_review_allowed": True,
    "evidence_spans_required_when_available": True,
    "identity_resolution_separate": True,
}


# -----------------------------------------------------------------------------
# Group record schema
# -----------------------------------------------------------------------------

GROUP_RECORD_SCHEMA = {
    "group_id": {
        "required": True,
    },
    "group_type": {
        "required": True,
    },
    "surface_mentions": {
        "required": True,
    },
    "normalised_label": {
        "required": False,
    },
    "evidence_spans": {
        "required": False,
    },
    "relevance_score": {
        "required": False,
    },
    "resolution_confidence": {
        "required": False,
    },
    "provenance": {
        "required": True,
    },
}


# -----------------------------------------------------------------------------
# Global representation contract
# -----------------------------------------------------------------------------

GLOBAL_REPRESENTATION_CONTRACT = {
    "scope": "global",
    "group_conditioned": False,
    "group_independent": True,
    "preserved_through_heads": True,
    "preserved_through_transformative_mechanism": True,
    "preserved_through_recurrent_mechanism": True,
    "reconstructed_from_group_paths": False,
}


# -----------------------------------------------------------------------------
# Group-conditioned representation contract
# -----------------------------------------------------------------------------

GROUP_CONDITIONED_REPRESENTATION_CONTRACT = {
    "scope": "group_conditioned",
    "group_conditioned": True,
    "requires_active_group": True,
    "derived_from_global_representation": True,
    "global_representation_preserved": True,
    "preserved_through_heads": True,
    "preserved_through_transformative_mechanism": True,
    "preserved_through_recurrent_mechanism": True,
    "independent_state_allowed": True,
}


# -----------------------------------------------------------------------------
# Trainable confluent-conditioning contract
# -----------------------------------------------------------------------------

GROUP_CONFLUENT_CONDITIONING_CONTRACT = {
    "function_name": "Phi_C",
    "input_global_representation": True,
    "input_group_representation": True,
    "trainable_parameters": True,
    "simple_concatenation_is_definition": False,
    "conditioning_methods_allowed": (
        "attention",
        "cross_attention",
        "gating",
        "feature_modulation",
        "residual_adaptation",
        "learned_interaction",
    ),
    "global_path_retained": True,
    "group_conditioned_path_retained": True,
    "feedback_to_next_global_step": CONFLUENT_FEEDBACK_ENABLED,
}


# -----------------------------------------------------------------------------
# Shared-core head principle
# -----------------------------------------------------------------------------

SHARED_CORE_HEAD_CONTRACT = {
    "factual_shared_core": True,
    "psychological_shared_core": True,
    "social_shared_core_where_applicable": True,
    "group_specific_adaptation_allowed": True,
    "fully_independent_global_and_group_heads_default": False,
    "ontology_consistent_across_scopes": True,
}


# -----------------------------------------------------------------------------
# Objective / subjective factual semantics
# -----------------------------------------------------------------------------

FACTUAL_SCOPE_SEMANTICS = {
    "global_name": "objective",
    "group_conditioned_name": "subjective",
    "objective_means_group_independent": True,
    "objective_means_verified_truth": False,
    "shared_factual_ontology": True,
}


# -----------------------------------------------------------------------------
# Group deviation contract
# -----------------------------------------------------------------------------

GROUP_DEVIATION_CONTRACT = {
    "defined": True,
    "representation_spaces": (
        "factual",
        "psychological",
        "social",
    ),
    "signed": True,
    "magnitude_may_be_derived": True,
    "global_representation_preserved": True,
    "group_representation_preserved": True,
    "automatic_bias_interpretation": False,
    "calculated": False,
}


# -----------------------------------------------------------------------------
# Group relevance contract
# -----------------------------------------------------------------------------

GROUP_RELEVANCE_CONTRACT = {
    "range": (
        0.0,
        1.0,
    ),
    "trainable": True,
    "used_for_sparse_routing": True,
    "is_representation_dimension": False,
    "hard_threshold_required": False,
    "article_persistence_allowed": True,
}


# -----------------------------------------------------------------------------
# Routing helpers
# -----------------------------------------------------------------------------

def create_global_route(sentence_id):
    """
    Create the always-active global routing record.
    """

    return {
        "scope": "global",
        "sentence_id": sentence_id,
        "group_id": None,
        "route_active": True,
        "backbone_representation_available": False,
        "factual_head_route": True,
        "psychological_head_route": True,
        "social_head_route": True,
        "transformative_route": True,
        "recurrent_route": True,
        "hierarchical_route": True,
    }


def create_group_route(sentence_id, group_id):
    """
    Create one group-conditioned routing record.
    """

    return {
        "scope": "group_conditioned",
        "sentence_id": sentence_id,
        "group_id": group_id,
        "route_active": True,
        "group_conditioning_available": False,
        "factual_head_route": True,
        "psychological_head_route": True,
        "social_head_route": True,
        "transformative_route": True,
        "recurrent_route": True,
        "hierarchical_route": True,
    }


# -----------------------------------------------------------------------------
# Current group state
# -----------------------------------------------------------------------------

# No structured group-detection output is currently available in Notebook 04.
# Therefore, concrete group identities are not fabricated.
#
# Every sentence receives a global route.
# Group-conditioned route collections remain empty but structurally valid.

GROUP_AWARE_ROUTING_RECORDS = []


for sentence_id in SENTENCE_IDS:

    record = {
        "record_type": "media_ai_group_aware_routing",
        "group_aware_confluent_version": GROUP_AWARE_CONFLUENT_VERSION,

        "media_ai_record_id": SENTENCE_TO_ARTICLE_ID[
            sentence_id
        ],

        "paragraph_id": SENTENCE_TO_PARAGRAPH_ID[
            sentence_id
        ],

        "sentence_id": sentence_id,

        "global_route": create_global_route(
            sentence_id
        ),

        "detected_groups": [],
        "active_group_ids": [],
        "group_conditioned_routes": [],

        "group_detection_status": "unavailable",
        "group_detection_observed_empty": False,

        "confluent_conditioning_executed": False,
        "group_deviation_calculated": False,

        "provenance": {
            "construction_method": "routing_contract_only",
            "future_context_used": False,
            "confluent_feedback_used": False,
            "groups_fabricated": False,
            "constructed_at_utc": datetime.now(
                timezone.utc
            ).isoformat(),
        },
    }

    GROUP_AWARE_ROUTING_RECORDS.append(
        record
    )


# -----------------------------------------------------------------------------
# Sentence lookup
# -----------------------------------------------------------------------------

GROUP_AWARE_ROUTING_BY_SENTENCE_ID = {
    record["sentence_id"]: record
    for record in GROUP_AWARE_ROUTING_RECORDS
}


# -----------------------------------------------------------------------------
# Article-level active-group interface
# -----------------------------------------------------------------------------

GROUP_AWARE_ARTICLE_ROUTING = []


for article_id, sentence_ids in ARTICLE_SENTENCE_ORDER.items():

    article_record = {
        "media_ai_record_id": article_id,
        "sentence_ids": list(
            sentence_ids
        ),
        "detected_group_ids": [],
        "active_group_ids": [],
        "group_detection_available": False,
        "observed_no_groups": False,
        "global_path_active": True,
        "group_conditioned_paths_active": False,
    }

    GROUP_AWARE_ARTICLE_ROUTING.append(
        article_record
    )


# -----------------------------------------------------------------------------
# Routing validation
# -----------------------------------------------------------------------------

GROUP_AWARE_ROUTING_ERRORS = []


if len(
    GROUP_AWARE_ROUTING_RECORDS
) != len(
    SENTENCE_IDS
):

    GROUP_AWARE_ROUTING_ERRORS.append(
        "routing_record_count_mismatch"
    )


for record in GROUP_AWARE_ROUTING_RECORDS:

    global_route = record[
        "global_route"
    ]

    if not global_route[
        "route_active"
    ]:

        GROUP_AWARE_ROUTING_ERRORS.append(
            "global_route_inactive"
        )

        break


    if global_route[
        "group_id"
    ] is not None:

        GROUP_AWARE_ROUTING_ERRORS.append(
            "global_route_has_group_identity"
        )

        break


    if record[
        "group_detection_status"
    ] == "unavailable":

        if record[
            "detected_groups"
        ]:

            GROUP_AWARE_ROUTING_ERRORS.append(
                "unavailable_group_detection_has_groups"
            )

            break


        if record[
            "active_group_ids"
        ]:

            GROUP_AWARE_ROUTING_ERRORS.append(
                "unavailable_group_detection_has_active_groups"
            )

            break


        if record[
            "group_conditioned_routes"
        ]:

            GROUP_AWARE_ROUTING_ERRORS.append(
                "unavailable_group_detection_has_conditioned_routes"
            )

            break


    if (
        record[
            "group_detection_observed_empty"
        ]
        and record[
            "group_detection_status"
        ] != "observed"
    ):

        GROUP_AWARE_ROUTING_ERRORS.append(
            "observed_empty_group_state_without_observed_status"
        )

        break


# -----------------------------------------------------------------------------
# Global-path safeguard validation
# -----------------------------------------------------------------------------

if GLOBAL_RECONSTRUCTED_FROM_GROUPS:
    GROUP_AWARE_ROUTING_ERRORS.append(
        "global_reconstructed_from_group_paths"
    )


if not GLOBAL_REPRESENTATION_CONTRACT[
    "preserved_through_heads"
]:
    GROUP_AWARE_ROUTING_ERRORS.append(
        "global_representation_not_preserved"
    )


if GROUP_DEVIATION_AUTOMATICALLY_BIAS:
    GROUP_AWARE_ROUTING_ERRORS.append(
        "group_deviation_automatically_interpreted_as_bias"
    )


if CONFLUENT_FEEDBACK_ENABLED:
    GROUP_AWARE_ROUTING_ERRORS.append(
        "confluent_feedback_enabled_in_version_1"
    )


# -----------------------------------------------------------------------------
# Group ontology validation
# -----------------------------------------------------------------------------

if len(
    MEDIA_AI_GROUP_TYPES
) == 0:

    GROUP_AWARE_ROUTING_ERRORS.append(
        "empty_group_ontology"
    )


if GROUP_IDENTITIES_GLOBALLY_PREDEFINED:
    GROUP_AWARE_ROUTING_ERRORS.append(
        "group_identities_should_not_be_globally_predefined"
    )


if not GROUP_DETECTION_CONTRACT[
    "no_group_is_valid"
]:
    GROUP_AWARE_ROUTING_ERRORS.append(
        "no_group_state_not_allowed"
    )


# -----------------------------------------------------------------------------
# Confluent contract validation
# -----------------------------------------------------------------------------

if GROUP_CONFLUENT_CONDITIONING_CONTRACT[
    "simple_concatenation_is_definition"
]:
    GROUP_AWARE_ROUTING_ERRORS.append(
        "confluence_defined_as_concatenation"
    )


if not GROUP_CONFLUENT_CONDITIONING_CONTRACT[
    "trainable_parameters"
]:
    GROUP_AWARE_ROUTING_ERRORS.append(
        "confluent_conditioning_not_trainable"
    )


if not GROUP_CONFLUENT_CONDITIONING_CONTRACT[
    "global_path_retained"
]:
    GROUP_AWARE_ROUTING_ERRORS.append(
        "global_path_not_retained"
    )


if not GROUP_CONFLUENT_CONDITIONING_CONTRACT[
    "group_conditioned_path_retained"
]:
    GROUP_AWARE_ROUTING_ERRORS.append(
        "group_conditioned_path_not_retained"
    )


# -----------------------------------------------------------------------------
# Shared-core validation
# -----------------------------------------------------------------------------

if not SHARED_CORE_HEAD_CONTRACT[
    "ontology_consistent_across_scopes"
]:
    GROUP_AWARE_ROUTING_ERRORS.append(
        "head_ontology_not_consistent"
    )


if SHARED_CORE_HEAD_CONTRACT[
    "fully_independent_global_and_group_heads_default"
]:
    GROUP_AWARE_ROUTING_ERRORS.append(
        "fully_independent_heads_enabled_by_default"
    )


# -----------------------------------------------------------------------------
# Group-deviation validation
# -----------------------------------------------------------------------------

if not GROUP_DEVIATION_CONTRACT[
    "signed"
]:
    GROUP_AWARE_ROUTING_ERRORS.append(
        "group_deviation_not_signed"
    )


if GROUP_DEVIATION_CONTRACT[
    "automatic_bias_interpretation"
]:
    GROUP_AWARE_ROUTING_ERRORS.append(
        "group_deviation_defined_as_bias"
    )


# -----------------------------------------------------------------------------
# Routing accounting
# -----------------------------------------------------------------------------

GROUP_AWARE_GLOBAL_ROUTE_COUNT = sum(
    1
    for record in GROUP_AWARE_ROUTING_RECORDS
    if record[
        "global_route"
    ][
        "route_active"
    ]
)


GROUP_AWARE_DETECTED_GROUP_COUNT = sum(
    len(
        record[
            "detected_groups"
        ]
    )
    for record in GROUP_AWARE_ROUTING_RECORDS
)


GROUP_AWARE_ACTIVE_GROUP_COUNT = sum(
    len(
        record[
            "active_group_ids"
        ]
    )
    for record in GROUP_AWARE_ROUTING_RECORDS
)


GROUP_AWARE_CONDITIONED_ROUTE_COUNT = sum(
    len(
        record[
            "group_conditioned_routes"
        ]
    )
    for record in GROUP_AWARE_ROUTING_RECORDS
)


GROUP_AWARE_OBSERVED_EMPTY_SENTENCE_COUNT = sum(
    1
    for record in GROUP_AWARE_ROUTING_RECORDS
    if record[
        "group_detection_observed_empty"
    ]
)


GROUP_AWARE_UNAVAILABLE_GROUP_SENTENCE_COUNT = sum(
    1
    for record in GROUP_AWARE_ROUTING_RECORDS
    if record[
        "group_detection_status"
    ] == "unavailable"
)


# -----------------------------------------------------------------------------
# Downstream routing contract
# -----------------------------------------------------------------------------

GROUP_AWARE_DOWNSTREAM_ROUTING_CONTRACT = {
    "global_to_factual_head": True,
    "global_to_psychological_head": True,
    "global_to_social_head": True,
    "group_to_factual_head": True,
    "group_to_psychological_head": True,
    "group_to_social_head": True,
    "global_to_transformative": True,
    "group_to_transformative": True,
    "global_to_recurrent": True,
    "group_to_recurrent": True,
    "global_to_hierarchical_aggregation": True,
    "group_to_hierarchical_aggregation": True,
}


# -----------------------------------------------------------------------------
# Trainable parameter map
# -----------------------------------------------------------------------------

GROUP_AWARE_TRAINABLE_PARAMETER_MAP = {
    "group_detector": {
        "trainable": True,
        "implemented": False,
    },

    "group_embedding": {
        "trainable": True,
        "implemented": False,
    },

    "confluent_conditioning": {
        "trainable": True,
        "implemented": False,
    },

    "group_relevance": {
        "trainable": True,
        "implemented": False,
    },

    "group_conditioned_adaptation": {
        "trainable": True,
        "implemented": False,
    },
}


# -----------------------------------------------------------------------------
# Version-1 feedback policy
# -----------------------------------------------------------------------------

GROUP_AWARE_FEEDBACK_POLICY = {
    "version": "1.0",
    "previous_confluent_state_to_global_backbone": False,
    "previous_group_state_to_global_backbone": False,
    "group_persistence_through_recurrence_allowed": True,
    "feedback_extension_reserved": True,
}


# -----------------------------------------------------------------------------
# Representation status
# -----------------------------------------------------------------------------

GROUP_AWARE_ROUTING_VALID = (
    len(
        GROUP_AWARE_ROUTING_ERRORS
    )
    == 0
)


if not GROUP_AWARE_ROUTING_VALID:
    raise ValueError(
        "Group-aware confluent routing failed validation: "
        f"{GROUP_AWARE_ROUTING_ERRORS}"
    )


GROUP_AWARE_ROUTING_RECORDS_CONSTRUCTED = True

GROUP_DETECTION_EXECUTED = False

CONFLUENT_CONDITIONING_EXECUTED = False

GROUP_DEVIATION_CALCULATED = False

TRAINABLE_GROUP_ARCHITECTURE_IMPLEMENTED = False


# -----------------------------------------------------------------------------
# Summary
# -----------------------------------------------------------------------------

GROUP_AWARE_CONFLUENT_SUMMARY = {
    "version": GROUP_AWARE_CONFLUENT_VERSION,

    "routing_level": GROUP_ROUTING_LEVEL,

    "sentence_records": len(
        GROUP_AWARE_ROUTING_RECORDS
    ),

    "global_routes": GROUP_AWARE_GLOBAL_ROUTE_COUNT,

    "detected_groups": GROUP_AWARE_DETECTED_GROUP_COUNT,

    "active_groups": GROUP_AWARE_ACTIVE_GROUP_COUNT,

    "group_conditioned_routes": GROUP_AWARE_CONDITIONED_ROUTE_COUNT,

    "observed_empty_group_sentences":
        GROUP_AWARE_OBSERVED_EMPTY_SENTENCE_COUNT,

    "unavailable_group_sentences":
        GROUP_AWARE_UNAVAILABLE_GROUP_SENTENCE_COUNT,

    "global_path_always_available":
        GLOBAL_PATH_ALWAYS_AVAILABLE,

    "group_conditioned_path_optional":
        GROUP_CONDITIONED_PATH_OPTIONAL,

    "global_reconstructed_from_groups":
        GLOBAL_RECONSTRUCTED_FROM_GROUPS,

    "group_deviation_defined":
        GROUP_DEVIATION_CONTRACT[
            "defined"
        ],

    "group_deviation_calculated":
        GROUP_DEVIATION_CALCULATED,

    "confluent_feedback_enabled":
        CONFLUENT_FEEDBACK_ENABLED,

    "routing_records_constructed":
        GROUP_AWARE_ROUTING_RECORDS_CONSTRUCTED,

    "group_detection_executed":
        GROUP_DETECTION_EXECUTED,

    "confluent_conditioning_executed":
        CONFLUENT_CONDITIONING_EXECUTED,

    "trainable_architecture_implemented":
        TRAINABLE_GROUP_ARCHITECTURE_IMPLEMENTED,

    "valid":
        GROUP_AWARE_ROUTING_VALID,
}


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Group-Aware Confluent Conditioning and Routing")
print("=" * 72)

print(
    f"Architecture version        : "
    f"{GROUP_AWARE_CONFLUENT_VERSION}"
)

print(
    f"Routing level               : "
    f"{GROUP_ROUTING_LEVEL}"
)

print(
    f"Confluent position          : "
    f"{CONFLUENT_POSITION_IN_ARCHITECTURE}"
)

print(
    f"Sentence records            : "
    f"{len(GROUP_AWARE_ROUTING_RECORDS)}"
)

print("-" * 72)

print("Global routing")

print(
    f"Global routes               : "
    f"{GROUP_AWARE_GLOBAL_ROUTE_COUNT}"
)

print(
    f"Global path always active   : "
    f"{GLOBAL_PATH_ALWAYS_AVAILABLE}"
)

print(
    f"Global reconstructed/groups : "
    f"{GLOBAL_RECONSTRUCTED_FROM_GROUPS}"
)

print("-" * 72)

print("Dynamic group routing")

print(
    f"Group identities predefined : "
    f"{GROUP_IDENTITIES_GLOBALLY_PREDEFINED}"
)

print(
    f"Configured group types      : "
    f"{len(MEDIA_AI_GROUP_TYPES)}"
)

print(
    f"Detected groups             : "
    f"{GROUP_AWARE_DETECTED_GROUP_COUNT}"
)

print(
    f"Active groups               : "
    f"{GROUP_AWARE_ACTIVE_GROUP_COUNT}"
)

print(
    f"Conditioned routes          : "
    f"{GROUP_AWARE_CONDITIONED_ROUTE_COUNT}"
)

print(
    f"Observed no-group sentences : "
    f"{GROUP_AWARE_OBSERVED_EMPTY_SENTENCE_COUNT}"
)

print(
    f"Unavailable group sentences : "
    f"{GROUP_AWARE_UNAVAILABLE_GROUP_SENTENCE_COUNT}"
)

print("-" * 72)

print("Confluent conditioning")

print(
    f"Trainable conditioning      : "
    f"{GROUP_CONFLUENT_CONDITIONING_CONTRACT['trainable_parameters']}"
)

print(
    f"Concatenation is definition : "
    f"{GROUP_CONFLUENT_CONDITIONING_CONTRACT['simple_concatenation_is_definition']}"
)

print(
    f"Global path retained        : "
    f"{GROUP_CONFLUENT_CONDITIONING_CONTRACT['global_path_retained']}"
)

print(
    f"Group path retained         : "
    f"{GROUP_CONFLUENT_CONDITIONING_CONTRACT['group_conditioned_path_retained']}"
)

print(
    f"Confluent feedback enabled  : "
    f"{CONFLUENT_FEEDBACK_ENABLED}"
)

print("-" * 72)

print("Group deviation")

print(
    f"Deviation defined           : "
    f"{GROUP_DEVIATION_CONTRACT['defined']}"
)

print(
    f"Signed deviation            : "
    f"{GROUP_DEVIATION_CONTRACT['signed']}"
)

print(
    f"Deviation = bias            : "
    f"{GROUP_DEVIATION_CONTRACT['automatic_bias_interpretation']}"
)

print(
    f"Deviation calculated        : "
    f"{GROUP_DEVIATION_CALCULATED}"
)

print("-" * 72)

print("Downstream routing")

global_all_heads = all([
    GROUP_AWARE_DOWNSTREAM_ROUTING_CONTRACT[
        "global_to_factual_head"
    ],
    GROUP_AWARE_DOWNSTREAM_ROUTING_CONTRACT[
        "global_to_psychological_head"
    ],
    GROUP_AWARE_DOWNSTREAM_ROUTING_CONTRACT[
        "global_to_social_head"
    ],
])

group_all_heads = all([
    GROUP_AWARE_DOWNSTREAM_ROUTING_CONTRACT[
        "group_to_factual_head"
    ],
    GROUP_AWARE_DOWNSTREAM_ROUTING_CONTRACT[
        "group_to_psychological_head"
    ],
    GROUP_AWARE_DOWNSTREAM_ROUTING_CONTRACT[
        "group_to_social_head"
    ],
])

print(
    f"Global → all heads          : "
    f"{global_all_heads}"
)

print(
    f"Group → all heads           : "
    f"{group_all_heads}"
)

print(
    f"Global → transformative     : "
    f"{GROUP_AWARE_DOWNSTREAM_ROUTING_CONTRACT['global_to_transformative']}"
)

print(
    f"Group → transformative      : "
    f"{GROUP_AWARE_DOWNSTREAM_ROUTING_CONTRACT['group_to_transformative']}"
)

print(
    f"Global → recurrent          : "
    f"{GROUP_AWARE_DOWNSTREAM_ROUTING_CONTRACT['global_to_recurrent']}"
)

print(
    f"Group → recurrent           : "
    f"{GROUP_AWARE_DOWNSTREAM_ROUTING_CONTRACT['group_to_recurrent']}"
)

print("-" * 72)

print("Execution state")

print(
    f"Group detection executed    : "
    f"{GROUP_DETECTION_EXECUTED}"
)

print(
    f"Conditioning executed       : "
    f"{CONFLUENT_CONDITIONING_EXECUTED}"
)

print(
    f"Trainable model implemented : "
    f"{TRAINABLE_GROUP_ARCHITECTURE_IMPLEMENTED}"
)

print(
    f"Routing contract valid      : "
    f"{GROUP_AWARE_ROUTING_VALID}"
)

print("=" * 72)

print(
    "Global and group-conditioned representation routes "
    "initialised successfully."
)

print(
    "The global representation remains independently preserved."
)

print(
    "Concrete group identities have not been predefined or fabricated."
)

print(
    "Group-conditioned routes remain unavailable until group detection "
    "produces usable article-specific groups."
)

print(
    "Confluent conditioning is defined as a trainable pre-head mechanism, "
    "not as simple concatenation."
)

print(
    "Global and group-conditioned paths are routed independently through "
    "factual, psychological and social heads."
)

print(
    "Both paths are also preserved for later transformative, recurrent "
    "and hierarchical processing."
)

print(
    "Signed group deviation is defined but has not yet been calculated."
)

print(
    "Group deviation is not automatically interpreted as bias."
)

print(
    "No confluent feedback to the next global textual step is enabled "
    "in architecture version 1.0."
)

print(
    "No group detection, model forward pass, transformation, recurrence, "
    "aggregation or training has been executed by this block."
)
````

## Block 12 — Factual Head and Trainable Parameterisation

This block defines the **trainable factual prediction head** for the Media AI architecture.

Its purpose is to map both the global and group-conditioned backbone representations into the explicit ten-dimensional factual ontology defined in Block 4.

The factual head does **not** determine factual truth. It predicts the **factological structure and epistemic presentation** of the text.

### Input Paths

For textual unit \(t\), the factual head receives:

$$\[
X_t^{global}
\]$$

for the global path, and for every active group \(g\):

\[
X_{t,g}^{group}
\]

for the group-conditioned path.

The two paths share the same factual ontology.

### Shared Factual Core

The factual head uses a shared trainable core:

\[
H_t^{F,global} = \phi_F(X_t^{global}; \theta_F)
\]

and:

\[
H_{t,g}^{F,group} = \phi_F(X_{t,g}^{group}; \theta_F)
\]

where:

* \(\phi_F\) = shared factual feature transformation;
* \(\theta_F\) = shared trainable factual parameters;
* \(H_t^{F,global}\) = hidden factual state for the global path;
* \(H_{t,g}^{F,group}\) = hidden factual state for the group-conditioned path.

Using a shared core keeps the factual dimensions semantically consistent across both scopes.

### Group-Conditioned Adaptation

The group-conditioned path may include an additional learned adaptation:

\[
\widetilde{H}_{t,g}^{F} = H_{t,g}^{F,group} + A_F(X_{t,g}^{group}, E_g; \theta_F^{cond})
\]

where:

* \(A_F\) = group-conditioned factual adaptation mechanism;
* \(E_g\) = representation of group \(g\);
* \(\theta_F^{cond}\) = trainable conditioning parameters.

The global path does not require this group-specific adjustment.

### Global Factual Output

The global factual output is:

\[
F_t^{global} = Head_F(H_t^{F,global}; \theta_F^{out})
\]

with:

\[
F_t^{global} \in \mathbb{R}^{10}
\]

The ten dimensions are:

1. factological weight;
2. explicit claim;
3. reported event/state;
4. observation;
5. quantitative information;
6. attribution;
7. evidence reference;
8. certainty statement;
9. uncertainty statement;
10. speculation or prediction.

### Group-Conditioned Factual Output

For active group \(g\):

\[
F_{t,g}^{group} = Head_F(\widetilde{H}_{t,g}^{F}; \theta_F^{out})
\]

with:

\[
F_{t,g}^{group} \in \mathbb{R}^{10}
\]

The output dimensions have exactly the same meanings as the global dimensions.

The distinction is therefore one of **scope**, not ontology.

### Objective and Subjective Scope

The global factual representation is interpreted as group-independent:

\[
F_t^{objective} \equiv F_t^{global}
\]

The group-conditioned factual representation is interpreted as group-relative:

\[
F_{t,g}^{subjective} \equiv F_{t,g}^{group}
\]

Here, **objective** means group-independent.

It does **not** mean externally verified truth.

### Group Deviation

The factual group deviation is defined as:

\[
D_{t,g}^{F} = F_{t,g}^{group} - F_t^{global}
\]

For dimension \(j\):

\[
D_{t,g,j}^{F} = F_{t,g,j}^{group} - F_{t,j}^{global}
\]

The deviation remains signed.

A positive value means the factual component is stronger in the group-conditioned path than in the global path.

A negative value means it is weaker.

A zero value means no difference in that dimension.

Group deviation is descriptive and must not automatically be interpreted as bias.

### Output Activation Structure

Factological weight is continuous:

\[
\hat{w}_t^F \in [0,1]
\]

and should therefore use a bounded continuous output such as a sigmoid activation.

The nine factual components are binary supervised targets.

For component \(j\):

\[
\hat{c}_{t,j} \in [0,1]
\]

where the output represents the model's predicted probability that the component is present.

A threshold may later convert the probability into a binary prediction for evaluation, but the trainable head should preserve the continuous probability during optimisation.

### Output Vector

The predicted factual vector is:

\[
\hat{F}_t = [\hat{w}_t^F, \hat{c}_{\mathrm{claim}}, \hat{c}_{\mathrm{event}}, \hat{c}_{\mathrm{observation}}, \hat{c}_{\mathrm{quant}}, \hat{c}_{\mathrm{attrib}}, \hat{c}_{\mathrm{evidence}}, \hat{c}_{\mathrm{certainty}}, \hat{c}_{\mathrm{uncertainty}}, \hat{c}_{\mathrm{speculation}}]
\]

The same structure is used for global and group-conditioned outputs.

### Supervision Targets

Where validated factual supervision is available, the model target is:

\[
Y_t^F = [y_t^w, y_{t,1}, \ldots, y_{t,9}]
\]

where:

* \(y_t^w\) = supervised factological weight;
* \(y_{t,j}\) = supervised binary target for factual component \(j\).

Targets remain unavailable when supervision is unavailable.

Missing targets must not be replaced with zero.

### Masked Training

The factual supervision mask is:

\[
M_t^F = [m_{t,1}^F, \ldots, m_{t,10}^F]
\]

Only dimensions with active masks contribute to the loss.

Therefore, the factual loss is mask-aware.

### Factological-Weight Loss

For observed factological weight, a regression loss may be used:

\[
\mathcal{L}_{weight} = \frac{\sum_t m_{t,w}^F \, \ell_{reg}(\hat{w}_t^F, y_t^w)}{\sum_t m_{t,w}^F}
\]

where \(\ell_{reg}\) may later be selected experimentally.

Suitable candidates include:

* mean squared error;
* mean absolute error;
* Huber loss.

The exact regression loss is an experimental hyperparameter and is not fixed by this block.

### Factual-Component Loss

For the nine binary components, the natural baseline loss is binary cross-entropy:

\[
\mathcal{L}_{components} = \frac{\sum_t \sum_{j=1}^{9} m_{t,j}^F \, \mathrm{BCE}(\hat{c}_{t,j}, y_{t,j})}{\sum_t \sum_{j=1}^{9} m_{t,j}^F}
\]

Only observed component targets contribute.

### Combined Factual Loss

The total factual loss may be written as:

\[
\mathcal{L}_F = \lambda_w \mathcal{L}_{weight} + \lambda_c \mathcal{L}_{components}
\]

where:

* \(\lambda_w\) = weight assigned to factological-weight regression;
* \(\lambda_c\) = weight assigned to component classification.

The initial architecture may use:

\[
\lambda_w = 1,\qquad \lambda_c = 1
\]

but these coefficients remain configurable.

### Global and Group Losses

Where group-conditioned factual supervision exists, global and group-conditioned losses remain distinguishable:

\[
\mathcal{L}_F^{global}
\]

and:

\[
\mathcal{L}_F^{group}
\]

The combined training objective may later use:

\[
\mathcal{L}_F^{total} = \lambda_g \mathcal{L}_F^{global} + \lambda_{gc} \mathcal{L}_F^{group}
\]

The weights \(\lambda_g\) and \(\lambda_{gc}\) remain configurable.

Group-conditioned loss must only be calculated for valid active groups with usable supervision.

### Shared Parameters

The factual architecture should preferentially share:

\[
\theta_F
\]

and:

\[
\theta_F^{out}
\]

between global and group-conditioned paths.

This preserves semantic alignment.

The group-conditioned path may additionally use:

\[
\theta_F^{cond}
\]

for contextual adaptation.

Thus:

\[
\theta_F^{group} = \{\theta_F, \theta_F^{out}, \theta_F^{cond}\}
\]

while:

\[
\theta_F^{global} = \{\theta_F, \theta_F^{out}\}
\]

### Parameter Independence

Shared parameters do not imply identical outputs.

In general:

\[
F_t^{global} \neq F_{t,g}^{group}
\]

because the input representations differ and the group-conditioned path may apply additional adaptation.

### Missing Group Path

If no active group exists at step \(t\), the factual head produces only:

\[
F_t^{global}
\]

No group-conditioned factual output is fabricated.

### Truth Boundary

The factual head remains separate from external truth supervision.

Therefore:

\[
\hat{F}_t \neq \widehat{\mathrm{Truth}}_t
\]

The head predicts factual structure, not truth status.

A sentence may therefore receive:

* high factological weight;
* high certainty;
* strong evidence-reference structure;

while still being externally assessed by a supervisor as false or unsupported.

That externally grounded truth signal belongs to a separate supervision or evaluation process.

### Transformative Routing

Both factual outputs continue into the transformative mechanism.

Global:

\[
F_t^{global} \rightarrow \Delta_t^{F,global}
\]

Group-conditioned:

\[
F_{t,g}^{group} \rightarrow \Delta_{t,g}^{F}
\]

The factual head itself does not calculate these transformations.

### Recurrent Routing

The transformed factual states later contribute to:

\[
H_t^{F,global}
\]

and:

\[
H_{t,g}^{F}
\]

through the recurrent mechanism.

The factual head therefore provides the trainable sentence-level factual representation on which later temporal processing operates.

### Attribution Boundary

The factual head must preserve mappings from each predicted dimension back to:

* sentence identity;
* group identity where applicable;
* model input;
* token positions where available;
* supervision source;
* prediction scope.

This permits later attribution of a factual prediction back to textual evidence.

The head does not calculate final attribution scores in this block.

### Trainable Parameter Groups

The factual head introduces the following parameter groups:

* shared factual core parameters \(\theta_F\);
* factual output parameters \(\theta_F^{out}\);
* optional group-conditioned adaptation parameters \(\theta_F^{cond}\).

The backbone parameters are governed by a separate fine-tuning policy and are not automatically assumed to be trainable.

### Output

This block defines the trainable factual-head architecture.

Its outputs specify:

* global factual prediction path;
* group-conditioned factual prediction path;
* shared factual core;
* group-conditioned adaptation;
* ten-dimensional output structure;
* global/objective and group-conditioned/subjective scope;
* signed factual group deviation;
* output activation semantics;
* masked factual supervision;
* factological-weight regression loss;
* factual-component classification loss;
* combined factual loss;
* global and group-conditioned loss boundaries;
* factual parameter groups;
* truth-verification boundary;
* transformative and recurrent routing;
* factual attribution requirements.

No factual model parameters are trained by this block.

No factual forward pass is executed.

No group deviation is calculated.

No transformative or recurrent factual state is calculated.

No hierarchical aggregation is performed.

No attribution score is calculated.

No experimental evaluation is performed.

````python
# =============================================================================
# Media AI — Factual Head and Trainable Parameterisation
# =============================================================================

import torch
import torch.nn as nn
import torch.nn.functional as F


# -----------------------------------------------------------------------------
# Block dependency checks
# -----------------------------------------------------------------------------

required_block_4_objects = [
    "MEDIA_AI_REPRESENTATION_SPECIFICATION_VALID",
    "FACTUAL_DIMENSIONS",
    "FACTUAL_DIMENSION_COUNT",
    "FACTUAL_DIMENSION_INDEX",
]

missing_block_4_objects = [
    object_name
    for object_name in required_block_4_objects
    if object_name not in globals()
]

if missing_block_4_objects:
    raise NameError(
        "Notebook 04 Block 4 is not initialised. "
        "Missing: "
        f"{missing_block_4_objects}"
    )


required_block_6_objects = [
    "FACTUAL_REPRESENTATION_VALID",
    "FACTUAL_SENTENCE_REPRESENTATIONS",
    "FACTUAL_SENTENCE_BY_ID",
]

missing_block_6_objects = [
    object_name
    for object_name in required_block_6_objects
    if object_name not in globals()
]

if missing_block_6_objects:
    raise NameError(
        "Notebook 04 Block 6 is not initialised. "
        "Missing: "
        f"{missing_block_6_objects}"
    )


required_block_11_objects = [
    "GROUP_AWARE_ROUTING_VALID",
    "GROUP_AWARE_ROUTING_RECORDS",
]

missing_block_11_objects = [
    object_name
    for object_name in required_block_11_objects
    if object_name not in globals()
]

if missing_block_11_objects:
    raise NameError(
        "Notebook 04 Block 11 group-aware confluent routing "
        "is not initialised. "
        "Missing: "
        f"{missing_block_11_objects}"
    )


if not MEDIA_AI_REPRESENTATION_SPECIFICATION_VALID:
    raise ValueError(
        "Factual head requires a valid "
        "Media AI representation specification."
    )


if not FACTUAL_REPRESENTATION_VALID:
    raise ValueError(
        "Factual head requires a valid factual representation."
    )


if not GROUP_AWARE_ROUTING_VALID:
    raise ValueError(
        "Factual head requires a valid "
        "group-aware confluent routing contract."
    )


# -----------------------------------------------------------------------------
# Factual-head configuration
# -----------------------------------------------------------------------------

FACTUAL_HEAD_VERSION = "1.0"

FACTUAL_HEAD_LEVEL = "sentence"

FACTUAL_HEAD_SPACE = "factual"

FACTUAL_HEAD_POSITION = "post_confluent_conditioning"

FACTUAL_HEAD_OUTPUT_DIMENSION = (
    FACTUAL_DIMENSION_COUNT
)

FACTUAL_WEIGHT_DIMENSION = (
    "factological_weight"
)

FACTUAL_COMPONENT_DIMENSIONS_HEAD = tuple(
    dimension
    for dimension in FACTUAL_DIMENSIONS
    if dimension != FACTUAL_WEIGHT_DIMENSION
)


EXPECTED_FACTUAL_DIMENSIONS_HEAD = (
    "factological_weight",
    "explicit_claim",
    "reported_event_state",
    "observation",
    "quantitative_information",
    "attribution",
    "evidence_reference",
    "certainty_statement",
    "uncertainty_statement",
    "speculation_or_prediction",
)


if (
    tuple(
        FACTUAL_DIMENSIONS
    )
    != EXPECTED_FACTUAL_DIMENSIONS_HEAD
):
    raise ValueError(
        "Factual-head dimensions do not match the current "
        "Media AI factual representation contract. "
        f"Expected: {EXPECTED_FACTUAL_DIMENSIONS_HEAD}. "
        f"Received: {tuple(FACTUAL_DIMENSIONS)}."
    )


if FACTUAL_HEAD_OUTPUT_DIMENSION != 10:
    raise ValueError(
        "The factual head must produce exactly ten explicit outputs."
    )


if len(
    FACTUAL_COMPONENT_DIMENSIONS_HEAD
) != 9:
    raise ValueError(
        "The factual head must contain exactly nine "
        "factological component outputs."
    )


# -----------------------------------------------------------------------------
# Architectural policy
# -----------------------------------------------------------------------------

FACTUAL_HEAD_SHARED_CORE = True

FACTUAL_HEAD_SHARED_OUTPUT = True

FACTUAL_HEAD_GROUP_ADAPTATION = True

FACTUAL_HEAD_GLOBAL_PATH_REQUIRED = True

FACTUAL_HEAD_GROUP_PATH_OPTIONAL = True

FACTUAL_HEAD_GROUP_DEVIATION_SIGNED = True

FACTUAL_HEAD_GROUP_DEVIATION_IS_BIAS = False

FACTUAL_HEAD_PREDICTS_TRUTH = False

FACTUAL_HEAD_MASK_AWARE = True

FACTUAL_HEAD_CONFIDENCE_IN_TARGET = False

FACTUAL_HEAD_PROPAGATES_TO_TRANSFORMATIVE = True

FACTUAL_HEAD_PROPAGATES_TO_RECURRENT = True

FACTUAL_HEAD_HIERARCHICAL_AGGREGATION = False

FACTUAL_HEAD_ATTRIBUTION_CALCULATED = False


# -----------------------------------------------------------------------------
# Loss configuration
# -----------------------------------------------------------------------------

FACTUAL_WEIGHT_LOSS_NAME = "huber"

FACTUAL_COMPONENT_LOSS_NAME = (
    "binary_cross_entropy"
)

FACTUAL_WEIGHT_LOSS_COEFFICIENT = 1.0

FACTUAL_COMPONENT_LOSS_COEFFICIENT = 1.0

FACTUAL_GLOBAL_LOSS_COEFFICIENT = 1.0

FACTUAL_GROUP_LOSS_COEFFICIENT = 1.0


# -----------------------------------------------------------------------------
# Factual output activation
# -----------------------------------------------------------------------------

def activate_factual_outputs(
    logits,
):
    """
    Convert factual-head logits into the explicit factual output space.

    Dimension 0:
        bounded continuous factological weight in [0,1]

    Dimensions 1-9:
        probabilities for the nine factological components.
    """

    if (
        logits.shape[-1]
        != FACTUAL_HEAD_OUTPUT_DIMENSION
    ):
        raise ValueError(
            "Factual-head logits have invalid final dimension."
        )

    return torch.sigmoid(
        logits
    )


# -----------------------------------------------------------------------------
# Shared factual core
# -----------------------------------------------------------------------------

class SharedFactualCore(nn.Module):
    """
    Shared trainable factual transformation used by both the global
    and group-conditioned paths.

    Input dimensionality is supplied at model instantiation time because
    the backbone or confluent hidden size is not fixed by this block.
    """

    def __init__(
        self,
        input_dim,
        hidden_dim,
        dropout=0.1,
    ):
        super().__init__()

        if input_dim <= 0:
            raise ValueError(
                "input_dim must be greater than zero."
            )

        if hidden_dim <= 0:
            raise ValueError(
                "hidden_dim must be greater than zero."
            )

        if not (
            0.0
            <= float(dropout)
            < 1.0
        ):
            raise ValueError(
                "dropout must lie in [0,1)."
            )


        self.input_dim = int(
            input_dim
        )

        self.hidden_dim = int(
            hidden_dim
        )


        self.network = nn.Sequential(
            nn.Linear(
                self.input_dim,
                self.hidden_dim,
            ),

            nn.GELU(),

            nn.Dropout(
                float(dropout)
            ),

            nn.LayerNorm(
                self.hidden_dim
            ),
        )


    def forward(
        self,
        x,
    ):

        if (
            x.shape[-1]
            != self.input_dim
        ):
            raise ValueError(
                "Shared factual core received an input "
                "with incompatible dimensionality."
            )

        return self.network(
            x
        )


# -----------------------------------------------------------------------------
# Group-conditioned factual adaptation
# -----------------------------------------------------------------------------

class GroupFactualAdapter(nn.Module):
    """
    Optional trainable adaptation for the group-conditioned factual path.

    The adapter produces a residual adjustment to the shared factual
    hidden representation.

    Group embeddings are supplied by the group-aware conditioning
    mechanism and are not fabricated here.
    """

    def __init__(
        self,
        input_dim,
        group_embedding_dim,
        hidden_dim,
        dropout=0.1,
    ):
        super().__init__()

        if input_dim <= 0:
            raise ValueError(
                "input_dim must be greater than zero."
            )

        if group_embedding_dim <= 0:
            raise ValueError(
                "group_embedding_dim must be greater than zero."
            )

        if hidden_dim <= 0:
            raise ValueError(
                "hidden_dim must be greater than zero."
            )

        if not (
            0.0
            <= float(dropout)
            < 1.0
        ):
            raise ValueError(
                "dropout must lie in [0,1)."
            )


        self.input_dim = int(
            input_dim
        )

        self.group_embedding_dim = int(
            group_embedding_dim
        )

        self.hidden_dim = int(
            hidden_dim
        )


        self.network = nn.Sequential(
            nn.Linear(
                self.input_dim
                + self.group_embedding_dim,
                self.hidden_dim,
            ),

            nn.GELU(),

            nn.Dropout(
                float(dropout)
            ),

            nn.Linear(
                self.hidden_dim,
                self.hidden_dim,
            ),
        )


    def forward(
        self,
        group_conditioned_input,
        group_embedding,
    ):

        if (
            group_conditioned_input.shape[-1]
            != self.input_dim
        ):
            raise ValueError(
                "Group-conditioned factual input has "
                "incompatible dimensionality."
            )


        if (
            group_embedding.shape[-1]
            != self.group_embedding_dim
        ):
            raise ValueError(
                "Group embedding has incompatible dimensionality."
            )


        if (
            group_conditioned_input.shape[:-1]
            != group_embedding.shape[:-1]
        ):
            raise ValueError(
                "Group-conditioned input and group embedding "
                "must have compatible leading dimensions."
            )


        adapter_input = torch.cat(
            [
                group_conditioned_input,
                group_embedding,
            ],
            dim=-1,
        )


        return self.network(
            adapter_input
        )


# -----------------------------------------------------------------------------
# Shared factual output layer
# -----------------------------------------------------------------------------

class FactualOutputLayer(nn.Module):
    """
    Shared output layer mapping factual hidden states to the ten
    explicit factual dimensions.
    """

    def __init__(
        self,
        hidden_dim,
    ):
        super().__init__()

        if hidden_dim <= 0:
            raise ValueError(
                "hidden_dim must be greater than zero."
            )


        self.hidden_dim = int(
            hidden_dim
        )


        self.output_layer = nn.Linear(
            self.hidden_dim,
            FACTUAL_HEAD_OUTPUT_DIMENSION,
        )


    def forward(
        self,
        hidden_state,
    ):

        if (
            hidden_state.shape[-1]
            != self.hidden_dim
        ):
            raise ValueError(
                "Factual output layer received a hidden state "
                "with incompatible dimensionality."
            )


        logits = self.output_layer(
            hidden_state
        )


        values = activate_factual_outputs(
            logits
        )


        return {
            "logits":
                logits,

            "values":
                values,
        }


# -----------------------------------------------------------------------------
# Complete global and group-conditioned factual head
# -----------------------------------------------------------------------------

class MediaAIFactualHead(nn.Module):
    """
    Trainable Media AI factual head.

    The global and group-conditioned paths share:
        - factual core parameters
        - factual output parameters

    The group-conditioned path additionally uses:
        - group-conditioned adapter parameters

    No group-conditioned output is produced when no valid group
    representation is supplied.
    """

    def __init__(
        self,
        input_dim,
        hidden_dim,
        group_embedding_dim=None,
        dropout=0.1,
    ):
        super().__init__()


        self.input_dim = int(
            input_dim
        )

        self.hidden_dim = int(
            hidden_dim
        )

        self.group_embedding_dim = (
            int(
                group_embedding_dim
            )
            if group_embedding_dim
            is not None
            else None
        )


        self.shared_core = (
            SharedFactualCore(
                input_dim=
                    self.input_dim,

                hidden_dim=
                    self.hidden_dim,

                dropout=
                    dropout,
            )
        )


        self.output_layer = (
            FactualOutputLayer(
                hidden_dim=
                    self.hidden_dim
            )
        )


        if (
            self.group_embedding_dim
            is not None
        ):

            self.group_adapter = (
                GroupFactualAdapter(
                    input_dim=
                        self.input_dim,

                    group_embedding_dim=
                        self.group_embedding_dim,

                    hidden_dim=
                        self.hidden_dim,

                    dropout=
                        dropout,
                )
            )

        else:

            self.group_adapter = None


    def forward_global(
        self,
        global_input,
    ):
        """
        Produce the group-independent global factual prediction.
        """

        global_hidden = (
            self.shared_core(
                global_input
            )
        )


        global_output = (
            self.output_layer(
                global_hidden
            )
        )


        return {
            "scope":
                "global",

            "hidden_state":
                global_hidden,

            "logits":
                global_output[
                    "logits"
                ],

            "values":
                global_output[
                    "values"
                ],
        }


    def forward_group(
        self,
        group_conditioned_input,
        group_embedding,
    ):
        """
        Produce one group-conditioned factual prediction.
        """

        if self.group_adapter is None:
            raise ValueError(
                "Group-conditioned factual prediction requires "
                "a configured group_embedding_dim."
            )


        group_hidden = (
            self.shared_core(
                group_conditioned_input
            )
        )


        group_adjustment = (
            self.group_adapter(
                group_conditioned_input,
                group_embedding,
            )
        )


        adapted_group_hidden = (
            group_hidden
            + group_adjustment
        )


        group_output = (
            self.output_layer(
                adapted_group_hidden
            )
        )


        return {
            "scope":
                "group_conditioned",

            "hidden_state":
                adapted_group_hidden,

            "shared_hidden_state":
                group_hidden,

            "group_adjustment":
                group_adjustment,

            "logits":
                group_output[
                    "logits"
                ],

            "values":
                group_output[
                    "values"
                ],
        }


    def forward(
        self,
        global_input,
        group_conditioned_input=None,
        group_embedding=None,
    ):
        """
        Execute the global path and, where supplied, the
        group-conditioned path.
        """

        global_output = (
            self.forward_global(
                global_input
            )
        )


        result = {
            "global":
                global_output,

            "group_conditioned":
                None,

            "group_deviation":
                None,
        }


        group_input_present = (
            group_conditioned_input
            is not None
        )

        group_embedding_present = (
            group_embedding
            is not None
        )


        if (
            group_input_present
            != group_embedding_present
        ):
            raise ValueError(
                "Group-conditioned input and group embedding "
                "must either both be supplied or both be absent."
            )


        if (
            group_input_present
            and group_embedding_present
        ):

            group_output = (
                self.forward_group(
                    group_conditioned_input,
                    group_embedding,
                )
            )


            if (
                group_output[
                    "values"
                ].shape
                != global_output[
                    "values"
                ].shape
            ):
                raise ValueError(
                    "Global and group-conditioned factual outputs "
                    "must have compatible shapes before deviation "
                    "is calculated."
                )


            group_deviation = (
                group_output[
                    "values"
                ]
                - global_output[
                    "values"
                ]
            )


            result[
                "group_conditioned"
            ] = group_output

            result[
                "group_deviation"
            ] = group_deviation


        return result


# -----------------------------------------------------------------------------
# Masked factual losses
# -----------------------------------------------------------------------------

def masked_factological_weight_loss(
    predicted_values,
    target_values,
    target_mask,
    loss_name=FACTUAL_WEIGHT_LOSS_NAME,
):
    """
    Calculate the factological-weight loss using only observed targets.

    Returns None when no valid factological-weight target exists.
    """

    weight_index = (
        FACTUAL_DIMENSION_INDEX[
            FACTUAL_WEIGHT_DIMENSION
        ]
    )


    prediction = predicted_values[
        ...,
        weight_index
    ]

    target = target_values[
        ...,
        weight_index
    ]

    mask = target_mask[
        ...,
        weight_index
    ].bool()


    if not torch.any(
        mask
    ):
        return None


    prediction = prediction[
        mask
    ]

    target = target[
        mask
    ]


    if loss_name == "huber":

        return F.smooth_l1_loss(
            prediction,
            target,
        )


    if loss_name == "mse":

        return F.mse_loss(
            prediction,
            target,
        )


    if loss_name == "mae":

        return F.l1_loss(
            prediction,
            target,
        )


    raise ValueError(
        "Unsupported factological-weight loss: "
        f"{loss_name}"
    )


def masked_factual_component_loss(
    predicted_logits,
    target_values,
    target_mask,
):
    """
    Calculate binary cross-entropy for the nine factual components
    using only observed component targets.

    Returns None when no valid component target exists.
    """

    component_indices = [
        FACTUAL_DIMENSION_INDEX[
            dimension
        ]
        for dimension
        in FACTUAL_COMPONENT_DIMENSIONS_HEAD
    ]


    logits = predicted_logits[
        ...,
        component_indices
    ]

    targets = target_values[
        ...,
        component_indices
    ]

    mask = target_mask[
        ...,
        component_indices
    ].bool()


    if not torch.any(
        mask
    ):
        return None


    return F.binary_cross_entropy_with_logits(
        logits[
            mask
        ],
        targets[
            mask
        ],
    )


def combine_factual_losses(
    weight_loss,
    component_loss,
    weight_coefficient=FACTUAL_WEIGHT_LOSS_COEFFICIENT,
    component_coefficient=FACTUAL_COMPONENT_LOSS_COEFFICIENT,
):
    """
    Combine available factual loss terms without fabricating a zero
    loss for unavailable supervision.
    """

    available_terms = []


    if weight_loss is not None:

        available_terms.append(
            float(
                weight_coefficient
            )
            * weight_loss
        )


    if component_loss is not None:

        available_terms.append(
            float(
                component_coefficient
            )
            * component_loss
        )


    if not available_terms:
        return None


    return sum(
        available_terms
    )


def calculate_factual_loss(
    prediction,
    target_values,
    target_mask,
):
    """
    Calculate the complete mask-aware factual loss for one prediction path.
    """

    weight_loss = (
        masked_factological_weight_loss(
            predicted_values=
                prediction[
                    "values"
                ],

            target_values=
                target_values,

            target_mask=
                target_mask,
        )
    )


    component_loss = (
        masked_factual_component_loss(
            predicted_logits=
                prediction[
                    "logits"
                ],

            target_values=
                target_values,

            target_mask=
                target_mask,
        )
    )


    total_loss = (
        combine_factual_losses(
            weight_loss=
                weight_loss,

            component_loss=
                component_loss,
        )
    )


    return {
        "factological_weight_loss":
            weight_loss,

        "component_loss":
            component_loss,

        "total_loss":
            total_loss,
    }


# -----------------------------------------------------------------------------
# Global and group-conditioned loss combination
# -----------------------------------------------------------------------------

def combine_global_and_group_factual_losses(
    global_loss,
    group_loss=None,
    global_coefficient=FACTUAL_GLOBAL_LOSS_COEFFICIENT,
    group_coefficient=FACTUAL_GROUP_LOSS_COEFFICIENT,
):
    """
    Combine global and group-conditioned factual losses.

    Group loss remains optional because group-conditioned supervision
    may legitimately be unavailable.
    """

    available_terms = []


    if global_loss is not None:

        available_terms.append(
            float(
                global_coefficient
            )
            * global_loss
        )


    if group_loss is not None:

        available_terms.append(
            float(
                group_coefficient
            )
            * group_loss
        )


    if not available_terms:
        return None


    return sum(
        available_terms
    )


# -----------------------------------------------------------------------------
# Training-target conversion
# -----------------------------------------------------------------------------

def factual_record_to_training_tensors(
    factual_record,
    device=None,
):
    """
    Convert one Block 6 factual representation record into tensors
    suitable for masked factual-head training.

    Missing values are replaced by a neutral technical placeholder only
    inside the temporary target tensor because the availability mask
    determines whether they participate in the loss.

    The original representation record remains unchanged.
    """

    values = factual_record[
        "values"
    ]

    mask = factual_record[
        "availability_mask"
    ]


    if values.shape != (
        FACTUAL_HEAD_OUTPUT_DIMENSION,
    ):
        raise ValueError(
            "Factual training record has invalid value shape."
        )


    if mask.shape != (
        FACTUAL_HEAD_OUTPUT_DIMENSION,
    ):
        raise ValueError(
            "Factual training record has invalid mask shape."
        )


    safe_values = torch.tensor(
        [
            0.0
            if np.isnan(
                value
            )
            else float(
                value
            )

            for value
            in values
        ],
        dtype=torch.float32,
        device=device,
    )


    mask_tensor = torch.tensor(
        mask,
        dtype=torch.bool,
        device=device,
    )


    return (
        safe_values,
        mask_tensor,
    )


# -----------------------------------------------------------------------------
# Parameter-group helper
# -----------------------------------------------------------------------------

def factual_head_parameter_groups(
    factual_head,
):
    """
    Return explicit factual-head parameter groups for later optimiser
    construction.
    """

    if not isinstance(
        factual_head,
        MediaAIFactualHead,
    ):
        raise TypeError(
            "factual_head must be a MediaAIFactualHead instance."
        )


    parameter_groups = {
        "shared_factual_core":
            list(
                factual_head.shared_core.parameters()
            ),

        "factual_output":
            list(
                factual_head.output_layer.parameters()
            ),

        "group_conditioned_adapter":
            (
                list(
                    factual_head.group_adapter.parameters()
                )
                if factual_head.group_adapter
                is not None
                else []
            ),
    }


    return parameter_groups


# -----------------------------------------------------------------------------
# Architecture validation
# -----------------------------------------------------------------------------

FACTUAL_HEAD_ARCHITECTURE_ERRORS = []


if not FACTUAL_HEAD_SHARED_CORE:
    FACTUAL_HEAD_ARCHITECTURE_ERRORS.append(
        "global_and_group_core_not_shared"
    )


if not FACTUAL_HEAD_SHARED_OUTPUT:
    FACTUAL_HEAD_ARCHITECTURE_ERRORS.append(
        "global_and_group_output_not_shared"
    )


if not FACTUAL_HEAD_GLOBAL_PATH_REQUIRED:
    FACTUAL_HEAD_ARCHITECTURE_ERRORS.append(
        "global_path_not_required"
    )


if FACTUAL_HEAD_PREDICTS_TRUTH:
    FACTUAL_HEAD_ARCHITECTURE_ERRORS.append(
        "factual_head_incorrectly_defined_as_truth_predictor"
    )


if FACTUAL_HEAD_GROUP_DEVIATION_IS_BIAS:
    FACTUAL_HEAD_ARCHITECTURE_ERRORS.append(
        "group_deviation_incorrectly_defined_as_bias"
    )


if not FACTUAL_HEAD_MASK_AWARE:
    FACTUAL_HEAD_ARCHITECTURE_ERRORS.append(
        "factual_head_not_mask_aware"
    )


if FACTUAL_HEAD_CONFIDENCE_IN_TARGET:
    FACTUAL_HEAD_ARCHITECTURE_ERRORS.append(
        "confidence_incorrectly_included_in_target"
    )


if not FACTUAL_HEAD_PROPAGATES_TO_TRANSFORMATIVE:
    FACTUAL_HEAD_ARCHITECTURE_ERRORS.append(
        "factual_output_not_routed_to_transformative"
    )


if not FACTUAL_HEAD_PROPAGATES_TO_RECURRENT:
    FACTUAL_HEAD_ARCHITECTURE_ERRORS.append(
        "factual_output_not_routed_to_recurrent"
    )


if FACTUAL_HEAD_HIERARCHICAL_AGGREGATION:
    FACTUAL_HEAD_ARCHITECTURE_ERRORS.append(
        "hierarchical_aggregation_executed_in_factual_head"
    )


if FACTUAL_HEAD_ATTRIBUTION_CALCULATED:
    FACTUAL_HEAD_ARCHITECTURE_ERRORS.append(
        "attribution_calculated_in_factual_head"
    )


FACTUAL_HEAD_ARCHITECTURE_VALID = (
    len(
        FACTUAL_HEAD_ARCHITECTURE_ERRORS
    )
    == 0
)


if not FACTUAL_HEAD_ARCHITECTURE_VALID:
    raise ValueError(
        "Factual-head architecture failed validation: "
        f"{FACTUAL_HEAD_ARCHITECTURE_ERRORS}"
    )


# -----------------------------------------------------------------------------
# Execution state
# -----------------------------------------------------------------------------

FACTUAL_HEAD_CLASS_DEFINED = True

FACTUAL_HEAD_MODEL_INSTANTIATED = False

FACTUAL_HEAD_FORWARD_PASS_EXECUTED = False

FACTUAL_HEAD_GROUP_DEVIATION_CALCULATED = False

FACTUAL_HEAD_TRAINING_EXECUTED = False

FACTUAL_HEAD_EVALUATION_EXECUTED = False


# -----------------------------------------------------------------------------
# Architecture summary
# -----------------------------------------------------------------------------

FACTUAL_HEAD_SUMMARY = {
    "version":
        FACTUAL_HEAD_VERSION,

    "representation_level":
        FACTUAL_HEAD_LEVEL,

    "position":
        FACTUAL_HEAD_POSITION,

    "output_dimensions":
        FACTUAL_HEAD_OUTPUT_DIMENSION,

    "factological_components":
        len(
            FACTUAL_COMPONENT_DIMENSIONS_HEAD
        ),

    "shared_core":
        FACTUAL_HEAD_SHARED_CORE,

    "shared_output":
        FACTUAL_HEAD_SHARED_OUTPUT,

    "group_adaptation":
        FACTUAL_HEAD_GROUP_ADAPTATION,

    "global_path_required":
        FACTUAL_HEAD_GLOBAL_PATH_REQUIRED,

    "group_path_optional":
        FACTUAL_HEAD_GROUP_PATH_OPTIONAL,

    "signed_group_deviation":
        FACTUAL_HEAD_GROUP_DEVIATION_SIGNED,

    "group_deviation_is_bias":
        FACTUAL_HEAD_GROUP_DEVIATION_IS_BIAS,

    "predicts_truth":
        FACTUAL_HEAD_PREDICTS_TRUTH,

    "mask_aware":
        FACTUAL_HEAD_MASK_AWARE,

    "weight_loss":
        FACTUAL_WEIGHT_LOSS_NAME,

    "component_loss":
        FACTUAL_COMPONENT_LOSS_NAME,

    "architecture_valid":
        FACTUAL_HEAD_ARCHITECTURE_VALID,

    "model_instantiated":
        FACTUAL_HEAD_MODEL_INSTANTIATED,

    "training_executed":
        FACTUAL_HEAD_TRAINING_EXECUTED,
}


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Factual Head and Trainable Parameterisation")
print("=" * 72)

print(
    f"Head version                : "
    f"{FACTUAL_HEAD_VERSION}"
)

print(
    f"Representation level        : "
    f"{FACTUAL_HEAD_LEVEL}"
)

print(
    f"Head position               : "
    f"{FACTUAL_HEAD_POSITION}"
)

print(
    f"Output dimensions           : "
    f"{FACTUAL_HEAD_OUTPUT_DIMENSION}"
)

print(
    f"Factological components     : "
    f"{len(FACTUAL_COMPONENT_DIMENSIONS_HEAD)}"
)

print("-" * 72)

print("Global and group paths")

print(
    f"Global path required        : "
    f"{FACTUAL_HEAD_GLOBAL_PATH_REQUIRED}"
)

print(
    f"Group path optional         : "
    f"{FACTUAL_HEAD_GROUP_PATH_OPTIONAL}"
)

print(
    f"Shared factual core         : "
    f"{FACTUAL_HEAD_SHARED_CORE}"
)

print(
    f"Shared factual output       : "
    f"{FACTUAL_HEAD_SHARED_OUTPUT}"
)

print(
    f"Group adaptation defined    : "
    f"{FACTUAL_HEAD_GROUP_ADAPTATION}"
)

print("-" * 72)

print("Group deviation")

print(
    f"Signed deviation            : "
    f"{FACTUAL_HEAD_GROUP_DEVIATION_SIGNED}"
)

print(
    f"Deviation = bias            : "
    f"{FACTUAL_HEAD_GROUP_DEVIATION_IS_BIAS}"
)

print(
    f"Deviation calculated        : "
    f"{FACTUAL_HEAD_GROUP_DEVIATION_CALCULATED}"
)

print("-" * 72)

print("Training contract")

print(
    f"Mask-aware                  : "
    f"{FACTUAL_HEAD_MASK_AWARE}"
)

print(
    f"Weight loss                 : "
    f"{FACTUAL_WEIGHT_LOSS_NAME}"
)

print(
    f"Component loss              : "
    f"{FACTUAL_COMPONENT_LOSS_NAME}"
)

print(
    f"Weight loss coefficient     : "
    f"{FACTUAL_WEIGHT_LOSS_COEFFICIENT:.1f}"
)

print(
    f"Component loss coefficient  : "
    f"{FACTUAL_COMPONENT_LOSS_COEFFICIENT:.1f}"
)

print("-" * 72)

print("Methodological boundaries")

print(
    f"Factual head predicts truth : "
    f"{FACTUAL_HEAD_PREDICTS_TRUTH}"
)

print(
    f"Confidence part of target   : "
    f"{FACTUAL_HEAD_CONFIDENCE_IN_TARGET}"
)

print(
    f"Routes to transformative    : "
    f"{FACTUAL_HEAD_PROPAGATES_TO_TRANSFORMATIVE}"
)

print(
    f"Routes to recurrent         : "
    f"{FACTUAL_HEAD_PROPAGATES_TO_RECURRENT}"
)

print(
    f"Hierarchical aggregation    : "
    f"{FACTUAL_HEAD_HIERARCHICAL_AGGREGATION}"
)

print(
    f"Attribution calculated      : "
    f"{FACTUAL_HEAD_ATTRIBUTION_CALCULATED}"
)

print("-" * 72)

print("Execution state")

print(
    f"Head class defined          : "
    f"{FACTUAL_HEAD_CLASS_DEFINED}"
)

print(
    f"Model instantiated          : "
    f"{FACTUAL_HEAD_MODEL_INSTANTIATED}"
)

print(
    f"Forward pass executed       : "
    f"{FACTUAL_HEAD_FORWARD_PASS_EXECUTED}"
)

print(
    f"Training executed           : "
    f"{FACTUAL_HEAD_TRAINING_EXECUTED}"
)

print(
    f"Evaluation executed         : "
    f"{FACTUAL_HEAD_EVALUATION_EXECUTED}"
)

print(
    f"Architecture valid          : "
    f"{FACTUAL_HEAD_ARCHITECTURE_VALID}"
)

print("=" * 72)

print(
    "Trainable factual-head architecture defined successfully."
)

print(
    "The global and group-conditioned paths share the factual core "
    "and factual output layer."
)

print(
    "The group-conditioned path may additionally learn a residual "
    "group-specific adaptation."
)

print(
    "Both paths predict the same ten-dimensional factual ontology."
)

print(
    "The global path represents group-independent factual perception."
)

print(
    "The group-conditioned path represents group-relative factual perception."
)

print(
    "Signed group deviation is defined as the group-conditioned output "
    "minus the global output."
)

print(
    "Group deviation is not automatically interpreted as bias."
)

print(
    "Factological weight is treated as a bounded continuous target."
)

print(
    "The nine factological components are treated as binary targets."
)

print(
    "Factual losses are mask-aware and unavailable supervision does not "
    "contribute to optimisation."
)

print(
    "The factual head predicts factological structure and epistemic "
    "presentation, not externally verified truth."
)

print(
    "No factual-head model has been instantiated."
)

print(
    "No forward pass, group deviation, transformation, recurrence, "
    "aggregation, training or evaluation has been executed."
)
````

## Psychological Head and Trainable Parameterisation

This block defines the **trainable psychological prediction head** for the Media AI architecture.

Its purpose is to map both the global and group-conditioned representations into the explicit 34-dimensional psychological space defined by the Media AI representation specification.

The psychological head models the emotional and psychological structure expressed, reported, implied or transformed by the text. It does not determine factual truth, social-group structure or final hierarchical interpretation.

### Input Paths

For textual unit \(t\), the psychological head receives the preserved global representation \(X_t^{global}\) and, where an active group \(g\) exists, the corresponding group-conditioned representation \(X_{t,g}^{group}\).

Both paths are preserved and processed in parallel.

The global path remains active independently of group detection. The group-conditioned path is additional and must not replace or reconstruct the global path.

### Shared Psychological Core

The psychological head uses a shared trainable core for both paths:

$$H_t^{P,global} = \phi_P(X_t^{global}; \theta_P).$$

$$H_{t,g}^{P,group} = \phi_P(X_{t,g}^{group}; \theta_P).$$

Here, \(\phi_P\) denotes the shared psychological feature transformation, \(\theta_P\) denotes its trainable parameters, \(H_t^{P,global}\) is the hidden psychological representation of the global path, and \(H_{t,g}^{P,group}\) is the corresponding hidden representation of the group-conditioned path.

Using a shared core preserves a common psychological ontology across global and group-conditioned perception.

### Group-Conditioned Adaptation

The group-conditioned path may additionally apply a trainable psychological adaptation:

$$\widetilde{H}_{t,g}^{P} = H_{t,g}^{P,group} + A_P(X_{t,g}^{group}, E_g; \theta_P^{cond}).$$

Here, \(A_P\) denotes the group-conditioned psychological adaptation mechanism, \(E_g\) represents active group \(g\), and \(\theta_P^{cond}\) denotes the trainable psychological conditioning parameters.

This adaptation permits the same textual evidence to produce a different psychological representation when interpreted relative to a detected group.

The global psychological path does not require this group-specific adjustment.

### Psychological State Vector

The complete psychological output is represented by the **Psychological State Vector (PSV)**.

The PSV contains 34 dimensions and consists of three 11-dimensional emotional subvectors and one scalar emotional-weight dimension:

$$\mathrm{PSV}_t = [\mathrm{REV}_t,\ \mathrm{TEV}_t,\ \mathrm{IEV}_t,\ \mathrm{EW}_t] \in \mathbb{R}^{34}.$$

The dimensional structure is therefore:

$$\dim(\mathrm{PSV}) = \dim(\mathrm{REV}) + \dim(\mathrm{TEV}) + \dim(\mathrm{IEV}) + \dim(\mathrm{EW}) = 11 + 11 + 11 + 1 = 34.$$

The four psychological components are:

* **Reported Emotion Vector (REV)** — 11 dimensions;
* **Transformative Emotion Vector (TEV)** — 11 dimensions;
* **Immediate Emotion Vector (IEV)** — 11 dimensions;
* **Emotional Weight (EW)** — 1 dimension.

This ordering is inherited directly from the representation contract and must remain stable across training, inference, transformation, recurrence and evaluation.

### Emotional Basis

REV, TEV and IEV share the same ordered 11-dimensional emotional basis defined by the representation specification.

The dimensional order must remain identical across all three vectors.

This shared coordinate system permits direct comparison between reported, transformative and immediate emotional representations without changing the semantic meaning of the underlying dimensions.

The emotional dimensions are not mutually exclusive. Multiple emotional dimensions may therefore be active simultaneously.

Positive and negative emotional dimensions must be treated symmetrically by the architecture.

### Reported Emotion Vector

The **Reported Emotion Vector (REV)** represents emotions explicitly reported, described or attributed by the text.

For textual unit \(t\):

$$\mathrm{REV}_t = [r_{t,1}, r_{t,2}, \ldots, r_{t,11}] \in [0,1]^{11}.$$

Each coordinate corresponds to one emotional dimension in the fixed emotional basis.

The eleven dimensions are therefore:

1. reported fear;
2. reported anger;
3. reported sadness;
4. reported disgust;
5. reported tension;
6. reported uncertainty;
7. reported hope;
8. reported relief;
9. reported trust;
10. reported compassion;
11. reported solidarity.

Reported emotion describes what the text presents as the emotional state of an actor, subject or relevant entity.

It must remain distinct from the psychological response potentially induced by the text.

### Transformative Emotion Vector

The **Transformative Emotion Vector (TEV)** represents the supervised emotional transformation associated with the current textual unit.

For textual unit \(t\):

$$\mathrm{TEV}_t = [\tau_{t,1}, \tau_{t,2}, \ldots, \tau_{t,11}] \in [-1,1]^{11}.$$

Its eleven dimensions are:

1. transformative fear;
2. transformative anger;
3. transformative sadness;
4. transformative disgust;
5. transformative tension;
6. transformative uncertainty;
7. transformative hope;
8. transformative relief;
9. transformative trust;
10. transformative compassion;
11. transformative solidarity.

For any emotional dimension \(j\):

* \(\tau_{t,j} < 0\) indicates a decrease;
* \(\tau_{t,j} = 0\) indicates no meaningful change;
* \(\tau_{t,j} > 0\) indicates an increase.

The TEV remains distinct from the architectural transformative mechanism.

It is also not defined as the arithmetic difference between consecutive Immediate Emotion Vectors:

$$\mathrm{TEV}_t \neq \mathrm{IEV}_t - \mathrm{IEV}_{t-1}.$$

Any relationship between these quantities must be learned or experimentally evaluated rather than assumed by the architecture.

### Immediate Emotion Vector

The **Immediate Emotion Vector (IEV)** represents the psychological state associated with the current textual unit in its permitted context.

For textual unit \(t\):

$$\mathrm{IEV}_t = [i_{t,1}, i_{t,2}, \ldots, i_{t,11}] \in [0,1]^{11}.$$

Its eleven dimensions are:

1. immediate fear;
2. immediate anger;
3. immediate sadness;
4. immediate disgust;
5. immediate tension;
6. immediate uncertainty;
7. immediate hope;
8. immediate relief;
9. immediate trust;
10. immediate compassion;
11. immediate solidarity.

The global IEV represents group-independent psychological interpretation.

The group-conditioned IEV represents the corresponding psychological interpretation relative to an active group.

### Emotional Weight

The **Emotional Weight (EW)** is the scalar psychological-strength dimension:

$$\mathrm{EW}_t = w_t^P \in [0,1].$$

Emotional Weight represents the overall psychological strength or salience associated with the textual unit.

It does not replace, average or collapse the individual emotional dimensions contained in REV, TEV or IEV.

### Global Psychological Output

The global psychological head produces a complete Psychological State Vector:

$$\mathrm{PSV}_t^{global} = Head_P(H_t^{P,global}; \theta_P^{out}) \in \mathbb{R}^{34}.$$

Its internal structure is:

$$\mathrm{PSV}_t^{global} = [\mathrm{REV}_t^{global},\ \mathrm{TEV}_t^{global},\ \mathrm{IEV}_t^{global},\ \mathrm{EW}_t^{global}].$$

The global path represents psychological interpretation without group-specific conditioning.

### Group-Conditioned Psychological Output

For every valid active group \(g\), the psychological head additionally produces:

$$\mathrm{PSV}_{t,g}^{group} = Head_P(\widetilde{H}_{t,g}^{P}; \theta_P^{out}) \in \mathbb{R}^{34}.$$

Its internal structure is:

$$\mathrm{PSV}_{t,g}^{group} = [\mathrm{REV}_{t,g}^{group},\ \mathrm{TEV}_{t,g}^{group},\ \mathrm{IEV}_{t,g}^{group},\ \mathrm{EW}_{t,g}^{group}].$$

The global and group-conditioned outputs use exactly the same psychological ontology and dimensional ordering.

Their distinction is one of interpretative scope rather than dimensional meaning.

### Global and Group-Relative Perception

The global psychological representation is group-independent, whereas the group-conditioned representation captures psychological perception relative to a detected group.

For architectural purposes:

$$\mathrm{PSV}_t^{objective} \equiv \mathrm{PSV}_t^{global}.$$

$$\mathrm{PSV}_{t,g}^{subjective} \equiv \mathrm{PSV}_{t,g}^{group}.$$

Here, **objective** means independent of group conditioning.

It does not imply that the global psychological interpretation constitutes an objective psychological fact about every possible reader.

Similarly, **subjective** identifies group-conditioned perception rather than arbitrary or unsupported interpretation.

### Psychological Group Deviation

The signed psychological group deviation measures the difference between the group-conditioned and global psychological representations:

$$D_{t,g}^{P} = \mathrm{PSV}_{t,g}^{group} - \mathrm{PSV}_t^{global}.$$

For psychological dimension \(j\):

$$D_{t,g,j}^{P} = \mathrm{PSV}_{t,g,j}^{group} - \mathrm{PSV}_{t,j}^{global}.$$

Because the dimensional ordering is shared, deviation can be calculated separately for REV, TEV, IEV and EW.

A positive value indicates that the corresponding psychological dimension is stronger under group conditioning.

A negative value indicates that it is weaker.

A zero value indicates no difference in that dimension.

Psychological group deviation is descriptive. It must not automatically be interpreted as bias, manipulation or polarisation.

### Multi-Dimensional Emotional Activation

The eleven emotional dimensions are not treated as mutually exclusive classes.

For an emotional vector \(E_t\), the architecture therefore does not require:

$$\sum_{j=1}^{11} E_{t,j} = 1.$$

Several emotions may be simultaneously present with different strengths.

The architecture must not impose a probability-simplex constraint across the eleven emotional dimensions.

### Output Activation Structure

The output activations follow the value semantics defined by the representation contract.

REV uses sigmoid activation:

$$\widehat{\mathrm{REV}}_t = \sigma(z_t^{REV}) \in [0,1]^{11}.$$

TEV uses hyperbolic tangent activation:

$$\widehat{\mathrm{TEV}}_t = \tanh(z_t^{TEV}) \in [-1,1]^{11}.$$

IEV uses sigmoid activation:

$$\widehat{\mathrm{IEV}}_t = \sigma(z_t^{IEV}) \in [0,1]^{11}.$$

EW uses sigmoid activation:

$$\widehat{\mathrm{EW}}_t = \sigma(z_t^{EW}) \in [0,1].$$

The complete predicted state is:

$$\widehat{\mathrm{PSV}}_t = [\widehat{\mathrm{REV}}_t,\ \widehat{\mathrm{TEV}}_t,\ \widehat{\mathrm{IEV}}_t,\ \widehat{\mathrm{EW}}_t].$$

### Supervision Targets

Where validated psychological supervision is available, the complete target is:

$$Y_t^P = [Y_t^{REV},\ Y_t^{TEV},\ Y_t^{IEV},\ y_t^{EW}].$$

Here, \(Y_t^{REV}\), \(Y_t^{TEV}\) and \(Y_t^{IEV}\) are 11-dimensional supervised targets, while \(y_t^{EW}\) is the scalar Emotional Weight target.

Targets remain unavailable where psychological supervision is unavailable.

Missing targets must not be converted into substantive numerical zero.

### Masked Training

The psychological supervision mask contains one availability state for each of the 34 psychological dimensions:

$$M_t^P = [m_{t,1}^P,\ldots,m_{t,34}^P].$$

Only dimensions with active supervision masks contribute to the training loss.

This permits partial psychological supervision without fabricating values for unavailable dimensions.

### Reported-Emotion Loss

The REV loss is calculated only over supervised Reported Emotion dimensions:

$$\mathcal{L}_{REV} = \frac{\sum_t \sum_{j=1}^{11} m_{t,j}^{REV}\,\ell_{REV}(\widehat{REV}_{t,j},REV_{t,j})}{\sum_t \sum_{j=1}^{11}m_{t,j}^{REV}}.$$

The baseline loss is Huber regression.

### Transformative-Emotion Loss

The TEV loss is:

$$\mathcal{L}_{TEV} = \frac{\sum_t \sum_{j=1}^{11} m_{t,j}^{TEV}\,\ell_{TEV}(\widehat{TEV}_{t,j},TEV_{t,j})}{\sum_t \sum_{j=1}^{11}m_{t,j}^{TEV}}.$$

The baseline loss is Huber regression over the signed \([-1,1]\) target space.

This loss trains the supervised Transformative Emotion Vector.

It does not by itself define or train the architectural transformative mechanism.

### Immediate-Emotion Loss

The IEV loss is:

$$\mathcal{L}_{IEV} = \frac{\sum_t \sum_{j=1}^{11} m_{t,j}^{IEV}\,\ell_{IEV}(\widehat{IEV}_{t,j},IEV_{t,j})}{\sum_t \sum_{j=1}^{11}m_{t,j}^{IEV}}.$$

Only observed IEV targets contribute to optimisation.

The baseline loss is Huber regression.

### Emotional-Weight Loss

The Emotional Weight loss is calculated only for supervised EW values:

$$\mathcal{L}_{EW} = \frac{\sum_t m_t^{EW}\,\ell_{EW}(\widehat{EW}_t,EW_t)}{\sum_t m_t^{EW}}.$$

The baseline loss is Huber regression.

### Combined Psychological Loss

The complete psychological-head loss combines the four psychological components:

$$\mathcal{L}_P = \lambda_{REV}\mathcal{L}_{REV} + \lambda_{TEV}\mathcal{L}_{TEV} + \lambda_{IEV}\mathcal{L}_{IEV} + \lambda_{EW}\mathcal{L}_{EW}.$$

All loss coefficients remain configurable.

The initial neutral configuration uses:

$$\lambda_{REV} = \lambda_{TEV} = \lambda_{IEV} = \lambda_{EW} = 1.$$

These coefficients may later be adjusted through controlled experiments rather than assumed to represent the optimal relative importance of the psychological components.

### Global and Group-Conditioned Losses

Global and group-conditioned psychological losses remain distinguishable as \(\mathcal{L}_P^{global}\) and \(\mathcal{L}_P^{group}\).

Where valid group-conditioned supervision exists, the combined psychological objective may be defined as:

$$\mathcal{L}_P^{total} = \lambda_g^P\mathcal{L}_P^{global} + \lambda_{gc}^P\mathcal{L}_P^{group}.$$

Group-conditioned loss must only be calculated for valid active groups with usable supervision.

Unavailable groups must not contribute fabricated group targets or losses.

### Shared Parameters

The global and group-conditioned psychological paths share the psychological core and output parameters.

The global parameter set is:

$$\theta_P^{global} = \{\theta_P,\theta_P^{out}\}.$$

The group-conditioned parameter set additionally contains the conditioning parameters:

$$\theta_P^{group} = \{\theta_P,\theta_P^{out},\theta_P^{cond}\}.$$

This preserves a common psychological ontology while permitting group-relative adaptation.

### Parameter Independence

Shared parameters do not imply identical psychological outputs.

In general:

$$\mathrm{PSV}_t^{global} \neq \mathrm{PSV}_{t,g}^{group}.$$

The outputs may differ because their input representations differ and the group-conditioned path may additionally contain a learned adaptation.

### Missing Group Path

If no valid active group exists for textual unit \(t\), the psychological head produces only the global Psychological State Vector.

No group-conditioned PSV, psychological group deviation or group-relative interpretation is fabricated.

### Relationship to the Transformative Mechanism

The **Transformative Emotion Vector** and the **architectural transformative mechanism** are separate concepts.

TEV is an explicit supervised component of the psychological representation.

The architectural transformative mechanism models contextual state transformation.

Therefore:

$$\mathrm{TEV}_t^{supervised} \neq \Delta_t^{P,architectural}.$$

The architecture does not assume that one can be derived mechanically from the other.

Their relationship may later be learned, compared or tested experimentally.

### Transformative Routing

Both psychological paths remain available to the downstream transformative mechanism:

$$\mathrm{PSV}_t^{global} \rightarrow \Delta_t^{P,global}.$$

$$\mathrm{PSV}_{t,g}^{group} \rightarrow \Delta_{t,g}^{P,group}.$$

The psychological head itself does not calculate these architectural transformations.

### Recurrent Routing

The psychological representations and their later transformations remain available to the recurrent mechanism:

$$\mathrm{PSV}_t^{global} \rightarrow H_t^{P,global}.$$

$$\mathrm{PSV}_{t,g}^{group} \rightarrow H_{t,g}^{P,group}.$$

The recurrent state remains a separate architectural object.

It is not identical to REV, TEV, IEV, EW or the architectural transformative state.

### Social Separation

The psychological head does not independently determine social groups, memberships or relationships.

Group identity is supplied through the confluent group-conditioning mechanism.

Social structure remains represented by the separate social head.

This preserves the architectural distinction between psychological interpretation and social structure.

### Attribution Boundary

The psychological head must preserve mappings from every predicted psychological dimension back to:

* sentence identity;
* group identity where applicable;
* psychological vector;
* emotional dimension;
* model input;
* token positions where available;
* supervision source;
* prediction scope.

This permits later backward attribution from PSV, REV, TEV, IEV or EW outputs to textual evidence.

Final attribution scores are not calculated in this block.

### Trainable Parameter Groups

The psychological head introduces three trainable parameter groups:

* shared psychological core parameters \(\theta_P\);
* shared psychological output parameters \(\theta_P^{out}\);
* optional group-conditioned adaptation parameters \(\theta_P^{cond}\).

Backbone parameters remain governed by a separate fine-tuning policy and are not automatically assumed to be trainable.

### Output

This block defines the trainable psychological-head architecture.

Its outputs specify:

* global psychological prediction path;
* group-conditioned psychological prediction path;
* shared psychological core;
* group-conditioned psychological adaptation;
* **Psychological State Vector (PSV)** — 34 dimensions;
* **Reported Emotion Vector (REV)** — 11 dimensions in \([0,1]\);
* **Transformative Emotion Vector (TEV)** — 11 signed dimensions in \([-1,1]\);
* **Immediate Emotion Vector (IEV)** — 11 dimensions in \([0,1]\);
* **Emotional Weight (EW)** — 1 dimension in \([0,1]\);
* common 11-dimensional emotional basis;
* canonical vector order REV → TEV → IEV → EW;
* sigmoid activation for REV and IEV;
* tanh activation for TEV;
* sigmoid activation for EW;
* global and group-relative psychological scope;
* signed psychological group deviation;
* multi-dimensional emotional activation;
* mask-aware psychological supervision;
* Huber regression losses for REV, TEV, IEV and EW;
* combined psychological loss;
* global and group-conditioned loss boundaries;
* psychological parameter groups;
* separation of TEV from the architectural transformative mechanism;
* transformative and recurrent routing;
* separation from social-group detection;
* psychological attribution requirements.

No psychological model parameters are trained by this block.

No psychological forward pass is executed.

No psychological group deviation is calculated.

No architectural transformation is calculated.

No recurrent psychological state is calculated.

No hierarchical aggregation is performed.

No attribution score is calculated.

No experimental evaluation is performed.

````python
# =============================================================================
# Media AI — Psychological Head and Trainable Parameterisation
# =============================================================================

import torch
import torch.nn as nn
import torch.nn.functional as F


# -----------------------------------------------------------------------------
# Block dependency checks
# -----------------------------------------------------------------------------

required_block_4_objects = [
    "MEDIA_AI_REPRESENTATION_SPECIFICATION_VALID",
    "PSYCHOLOGICAL_EMOTIONAL_DIMENSIONS",
    "PSYCHOLOGICAL_DIMENSION_COUNT",
    "PSYCHOLOGICAL_DIMENSIONS",
    "PSYCHOLOGICAL_CONSTRUCT_SLICES",
]

missing_block_4_objects = [
    object_name
    for object_name in required_block_4_objects
    if object_name not in globals()
]

if missing_block_4_objects:
    raise NameError(
        "Notebook 04 Block 4 is not initialised. "
        f"Missing: {missing_block_4_objects}"
    )


required_block_11_objects = [
    "GROUP_AWARE_ROUTING_VALID",
    "GROUP_AWARE_ROUTING_RECORDS",
]

missing_block_11_objects = [
    object_name
    for object_name in required_block_11_objects
    if object_name not in globals()
]

if missing_block_11_objects:
    raise NameError(
        "Notebook 04 Block 11 group-aware confluent routing "
        "is not initialised. "
        f"Missing: {missing_block_11_objects}"
    )


required_block_12_objects = [
    "FACTUAL_HEAD_ARCHITECTURE_VALID",
]

missing_block_12_objects = [
    object_name
    for object_name in required_block_12_objects
    if object_name not in globals()
]

if missing_block_12_objects:
    raise NameError(
        "Notebook 04 Block 12 is not initialised. "
        f"Missing: {missing_block_12_objects}"
    )


if not MEDIA_AI_REPRESENTATION_SPECIFICATION_VALID:
    raise ValueError(
        "Psychological head requires a valid "
        "Media AI representation specification."
    )


if not GROUP_AWARE_ROUTING_VALID:
    raise ValueError(
        "Psychological head requires a valid "
        "group-aware confluent routing contract."
    )


if not FACTUAL_HEAD_ARCHITECTURE_VALID:
    raise ValueError(
        "Psychological head requires the preceding "
        "trainable-head architecture to be valid."
    )


# -----------------------------------------------------------------------------
# Psychological-head configuration
# -----------------------------------------------------------------------------

PSYCHOLOGICAL_HEAD_VERSION = "1.0"

PSYCHOLOGICAL_HEAD_LEVEL = "sentence"

PSYCHOLOGICAL_HEAD_SPACE = "psychological"

PSYCHOLOGICAL_HEAD_POSITION = (
    "post_confluent_conditioning"
)


# -----------------------------------------------------------------------------
# Psychological ontology inherited from Block 4
# -----------------------------------------------------------------------------

PSYCHOLOGICAL_HEAD_EMOTIONAL_DIMENSIONS = tuple(
    PSYCHOLOGICAL_EMOTIONAL_DIMENSIONS
)

PSYCHOLOGICAL_HEAD_EMOTION_DIMENSION_COUNT = len(
    PSYCHOLOGICAL_HEAD_EMOTIONAL_DIMENSIONS
)


if PSYCHOLOGICAL_HEAD_EMOTION_DIMENSION_COUNT != 11:
    raise ValueError(
        "The psychological emotional basis must contain "
        "exactly eleven dimensions."
    )


if PSYCHOLOGICAL_DIMENSION_COUNT != 34:
    raise ValueError(
        "The psychological representation must contain "
        "exactly 34 explicit dimensions."
    )


PSYCHOLOGICAL_HEAD_EMOTION_INDEX = {
    dimension:
        index

    for index, dimension
    in enumerate(
        PSYCHOLOGICAL_HEAD_EMOTIONAL_DIMENSIONS
    )
}


# -----------------------------------------------------------------------------
# Named psychological vectors
# -----------------------------------------------------------------------------

PSYCHOLOGICAL_STATE_VECTOR_NAME = "PSV"

REPORTED_EMOTION_VECTOR_NAME = "REV"

TRANSFORMATIVE_EMOTION_VECTOR_NAME = "TEV"

IMMEDIATE_EMOTION_VECTOR_NAME = "IEV"

EMOTIONAL_WEIGHT_NAME = "EW"


REPORTED_EMOTION_DIMENSION_COUNT = 11

TRANSFORMATIVE_EMOTION_DIMENSION_COUNT = 11

IMMEDIATE_EMOTION_DIMENSION_COUNT = 11

EMOTIONAL_WEIGHT_DIMENSION_COUNT = 1

PSYCHOLOGICAL_STATE_DIMENSION_COUNT = (
    REPORTED_EMOTION_DIMENSION_COUNT
    + TRANSFORMATIVE_EMOTION_DIMENSION_COUNT
    + IMMEDIATE_EMOTION_DIMENSION_COUNT
    + EMOTIONAL_WEIGHT_DIMENSION_COUNT
)


if PSYCHOLOGICAL_STATE_DIMENSION_COUNT != 34:
    raise ValueError(
        "The Psychological State Vector must contain "
        "exactly 34 dimensions."
    )


# -----------------------------------------------------------------------------
# Psychological vector ordering inherited from Block 4
# -----------------------------------------------------------------------------

EXPECTED_PSYCHOLOGICAL_CONSTRUCT_ORDER = (
    "reported_emotion",
    "transformative_emotion",
    "immediate_emotional_state",
    "emotional_weight",
)


if tuple(
    PSYCHOLOGICAL_CONSTRUCT_SLICES.keys()
) != EXPECTED_PSYCHOLOGICAL_CONSTRUCT_ORDER:
    raise ValueError(
        "Block 13 psychological construct order does not "
        "match the Block 4 representation contract."
    )


REV_START, REV_END = (
    PSYCHOLOGICAL_CONSTRUCT_SLICES[
        "reported_emotion"
    ]
)

TEV_START, TEV_END = (
    PSYCHOLOGICAL_CONSTRUCT_SLICES[
        "transformative_emotion"
    ]
)

IEV_START, IEV_END = (
    PSYCHOLOGICAL_CONSTRUCT_SLICES[
        "immediate_emotional_state"
    ]
)

EW_START, EW_END = (
    PSYCHOLOGICAL_CONSTRUCT_SLICES[
        "emotional_weight"
    ]
)


REV_SLICE = slice(
    REV_START,
    REV_END,
)

TEV_SLICE = slice(
    TEV_START,
    TEV_END,
)

IEV_SLICE = slice(
    IEV_START,
    IEV_END,
)

EW_SLICE = slice(
    EW_START,
    EW_END,
)


if (
    REV_END
    - REV_START
) != 11:
    raise ValueError(
        "REV must contain exactly eleven dimensions."
    )


if (
    TEV_END
    - TEV_START
) != 11:
    raise ValueError(
        "TEV must contain exactly eleven dimensions."
    )


if (
    IEV_END
    - IEV_START
) != 11:
    raise ValueError(
        "IEV must contain exactly eleven dimensions."
    )


if (
    EW_END
    - EW_START
) != 1:
    raise ValueError(
        "EW must contain exactly one dimension."
    )


# -----------------------------------------------------------------------------
# Psychological value ranges
# -----------------------------------------------------------------------------

PSYCHOLOGICAL_HEAD_VALUE_RANGES = {
    "REV": (
        0.0,
        1.0,
    ),

    "TEV": (
        -1.0,
        1.0,
    ),

    "IEV": (
        0.0,
        1.0,
    ),

    "EW": (
        0.0,
        1.0,
    ),
}


# -----------------------------------------------------------------------------
# Architectural policy
# -----------------------------------------------------------------------------

PSYCHOLOGICAL_HEAD_SHARED_CORE = True

PSYCHOLOGICAL_HEAD_SHARED_OUTPUT = True

PSYCHOLOGICAL_HEAD_GROUP_ADAPTATION = True

PSYCHOLOGICAL_HEAD_GLOBAL_PATH_REQUIRED = True

PSYCHOLOGICAL_HEAD_GROUP_PATH_OPTIONAL = True

PSYCHOLOGICAL_HEAD_GROUP_DEVIATION_SIGNED = True

PSYCHOLOGICAL_HEAD_GROUP_DEVIATION_IS_BIAS = False

PSYCHOLOGICAL_HEAD_GROUP_DEVIATION_IS_POLARISATION = False

PSYCHOLOGICAL_HEAD_GROUP_DEVIATION_IS_MANIPULATION = False

PSYCHOLOGICAL_HEAD_MASK_AWARE = True

PSYCHOLOGICAL_HEAD_CONFIDENCE_IN_TARGET = False

PSYCHOLOGICAL_HEAD_EMOTIONS_MUTUALLY_EXCLUSIVE = False

PSYCHOLOGICAL_HEAD_SIMPLEX_REQUIRED = False

PSYCHOLOGICAL_HEAD_POSITIVE_NEGATIVE_SYMMETRY = True

PSYCHOLOGICAL_HEAD_TEV_EQUALS_STATE_DIFFERENCE = False

PSYCHOLOGICAL_HEAD_TEV_EQUALS_ARCHITECTURAL_TRANSFORMATION = False

PSYCHOLOGICAL_HEAD_PROPAGATES_TO_TRANSFORMATIVE = True

PSYCHOLOGICAL_HEAD_PROPAGATES_TO_RECURRENT = True

PSYCHOLOGICAL_HEAD_DETECTS_GROUPS = False

PSYCHOLOGICAL_HEAD_HIERARCHICAL_AGGREGATION = False

PSYCHOLOGICAL_HEAD_ATTRIBUTION_CALCULATED = False


# -----------------------------------------------------------------------------
# Output activation policy
# -----------------------------------------------------------------------------

PSYCHOLOGICAL_REV_ACTIVATION = "sigmoid"

PSYCHOLOGICAL_TEV_ACTIVATION = "tanh"

PSYCHOLOGICAL_IEV_ACTIVATION = "sigmoid"

PSYCHOLOGICAL_EW_ACTIVATION = "sigmoid"


def activate_psychological_outputs(
    logits,
):
    """
    Convert psychological-head logits into the explicit
    34-dimensional Psychological State Vector.

    REV:
        continuous bounded values in [0,1]

    TEV:
        signed continuous values in [-1,1]

    IEV:
        continuous bounded values in [0,1]

    EW:
        continuous bounded value in [0,1]

    No emotional subvector is normalised to a probability simplex.
    """

    if (
        logits.shape[-1]
        != PSYCHOLOGICAL_STATE_DIMENSION_COUNT
    ):
        raise ValueError(
            "Psychological-head logits have invalid "
            "final dimensionality."
        )


    rev_values = torch.sigmoid(
        logits[
            ...,
            REV_SLICE
        ]
    )


    tev_values = torch.tanh(
        logits[
            ...,
            TEV_SLICE
        ]
    )


    iev_values = torch.sigmoid(
        logits[
            ...,
            IEV_SLICE
        ]
    )


    ew_values = torch.sigmoid(
        logits[
            ...,
            EW_SLICE
        ]
    )


    values = torch.cat(
        [
            rev_values,
            tev_values,
            iev_values,
            ew_values,
        ],
        dim=-1,
    )


    if (
        values.shape[-1]
        != PSYCHOLOGICAL_STATE_DIMENSION_COUNT
    ):
        raise ValueError(
            "Activated Psychological State Vector has "
            "invalid dimensionality."
        )


    return values


# -----------------------------------------------------------------------------
# Psychological vector decomposition
# -----------------------------------------------------------------------------

def decompose_psychological_state_vector(
    values,
):
    """
    Decompose a 34-dimensional PSV into REV, TEV, IEV and EW
    according to the Block 4 psychological representation contract.
    """

    if (
        values.shape[-1]
        != PSYCHOLOGICAL_STATE_DIMENSION_COUNT
    ):
        raise ValueError(
            "Psychological State Vector must contain "
            "exactly 34 dimensions."
        )


    return {
        "PSV":
            values,

        "REV":
            values[
                ...,
                REV_SLICE
            ],

        "TEV":
            values[
                ...,
                TEV_SLICE
            ],

        "IEV":
            values[
                ...,
                IEV_SLICE
            ],

        "EW":
            values[
                ...,
                EW_SLICE
            ],
    }


# -----------------------------------------------------------------------------
# Shared psychological core
# -----------------------------------------------------------------------------

class SharedPsychologicalCore(nn.Module):
    """
    Shared trainable psychological transformation used by both
    global and group-conditioned paths.
    """

    def __init__(
        self,
        input_dim,
        hidden_dim,
        dropout=0.1,
    ):
        super().__init__()


        if input_dim <= 0:
            raise ValueError(
                "input_dim must be greater than zero."
            )


        if hidden_dim <= 0:
            raise ValueError(
                "hidden_dim must be greater than zero."
            )


        if not (
            0.0
            <= float(
                dropout
            )
            < 1.0
        ):
            raise ValueError(
                "dropout must lie in [0,1)."
            )


        self.input_dim = int(
            input_dim
        )

        self.hidden_dim = int(
            hidden_dim
        )


        self.network = nn.Sequential(
            nn.Linear(
                self.input_dim,
                self.hidden_dim,
            ),

            nn.GELU(),

            nn.Dropout(
                float(
                    dropout
                )
            ),

            nn.LayerNorm(
                self.hidden_dim
            ),
        )


    def forward(
        self,
        x,
    ):

        if (
            x.shape[-1]
            != self.input_dim
        ):
            raise ValueError(
                "Shared psychological core received "
                "an input with incompatible dimensionality."
            )


        return self.network(
            x
        )


# -----------------------------------------------------------------------------
# Group-conditioned psychological adaptation
# -----------------------------------------------------------------------------

class GroupPsychologicalAdapter(nn.Module):
    """
    Trainable residual adaptation for the group-conditioned
    psychological path.

    Group embeddings are supplied by the group-aware confluent
    conditioning mechanism and are not fabricated here.
    """

    def __init__(
        self,
        input_dim,
        group_embedding_dim,
        hidden_dim,
        dropout=0.1,
    ):
        super().__init__()


        if input_dim <= 0:
            raise ValueError(
                "input_dim must be greater than zero."
            )


        if group_embedding_dim <= 0:
            raise ValueError(
                "group_embedding_dim must be greater than zero."
            )


        if hidden_dim <= 0:
            raise ValueError(
                "hidden_dim must be greater than zero."
            )


        if not (
            0.0
            <= float(
                dropout
            )
            < 1.0
        ):
            raise ValueError(
                "dropout must lie in [0,1)."
            )


        self.input_dim = int(
            input_dim
        )

        self.group_embedding_dim = int(
            group_embedding_dim
        )

        self.hidden_dim = int(
            hidden_dim
        )


        self.network = nn.Sequential(
            nn.Linear(
                self.input_dim
                + self.group_embedding_dim,
                self.hidden_dim,
            ),

            nn.GELU(),

            nn.Dropout(
                float(
                    dropout
                )
            ),

            nn.Linear(
                self.hidden_dim,
                self.hidden_dim,
            ),
        )


    def forward(
        self,
        group_conditioned_input,
        group_embedding,
    ):

        if (
            group_conditioned_input.shape[-1]
            != self.input_dim
        ):
            raise ValueError(
                "Group-conditioned psychological input "
                "has incompatible dimensionality."
            )


        if (
            group_embedding.shape[-1]
            != self.group_embedding_dim
        ):
            raise ValueError(
                "Psychological group embedding has "
                "incompatible dimensionality."
            )


        if (
            group_conditioned_input.shape[:-1]
            != group_embedding.shape[:-1]
        ):
            raise ValueError(
                "Group-conditioned psychological input "
                "and group embedding must have compatible "
                "leading dimensions."
            )


        adapter_input = torch.cat(
            [
                group_conditioned_input,
                group_embedding,
            ],
            dim=-1,
        )


        return self.network(
            adapter_input
        )


# -----------------------------------------------------------------------------
# Shared psychological output layer
# -----------------------------------------------------------------------------

class PsychologicalOutputLayer(nn.Module):
    """
    Shared output layer mapping psychological hidden states
    into the complete 34-dimensional PSV.
    """

    def __init__(
        self,
        hidden_dim,
    ):
        super().__init__()


        if hidden_dim <= 0:
            raise ValueError(
                "hidden_dim must be greater than zero."
            )


        self.hidden_dim = int(
            hidden_dim
        )


        self.output_layer = nn.Linear(
            self.hidden_dim,
            PSYCHOLOGICAL_STATE_DIMENSION_COUNT,
        )


    def forward(
        self,
        hidden_state,
    ):

        if (
            hidden_state.shape[-1]
            != self.hidden_dim
        ):
            raise ValueError(
                "Psychological output layer received "
                "a hidden state with incompatible dimensionality."
            )


        logits = self.output_layer(
            hidden_state
        )


        values = activate_psychological_outputs(
            logits
        )


        vectors = (
            decompose_psychological_state_vector(
                values
            )
        )


        return {
            "logits":
                logits,

            "values":
                values,

            "PSV":
                vectors[
                    "PSV"
                ],

            "REV":
                vectors[
                    "REV"
                ],

            "TEV":
                vectors[
                    "TEV"
                ],

            "IEV":
                vectors[
                    "IEV"
                ],

            "EW":
                vectors[
                    "EW"
                ],
        }


# -----------------------------------------------------------------------------
# Complete global and group-conditioned psychological head
# -----------------------------------------------------------------------------

class MediaAIPsychologicalHead(nn.Module):
    """
    Trainable Media AI psychological head.

    Global and group-conditioned paths share:
        - psychological core parameters
        - psychological output parameters

    The group-conditioned path additionally uses:
        - psychological group-adaptation parameters
    """

    def __init__(
        self,
        input_dim,
        hidden_dim,
        group_embedding_dim=None,
        dropout=0.1,
    ):
        super().__init__()


        self.input_dim = int(
            input_dim
        )

        self.hidden_dim = int(
            hidden_dim
        )

        self.group_embedding_dim = (
            int(
                group_embedding_dim
            )
            if group_embedding_dim
            is not None
            else None
        )


        self.shared_core = (
            SharedPsychologicalCore(
                input_dim=
                    self.input_dim,

                hidden_dim=
                    self.hidden_dim,

                dropout=
                    dropout,
            )
        )


        self.output_layer = (
            PsychologicalOutputLayer(
                hidden_dim=
                    self.hidden_dim
            )
        )


        if (
            self.group_embedding_dim
            is not None
        ):

            self.group_adapter = (
                GroupPsychologicalAdapter(
                    input_dim=
                        self.input_dim,

                    group_embedding_dim=
                        self.group_embedding_dim,

                    hidden_dim=
                        self.hidden_dim,

                    dropout=
                        dropout,
                )
            )

        else:

            self.group_adapter = None


    def forward_global(
        self,
        global_input,
    ):
        """
        Produce the group-independent global PSV.
        """

        global_hidden = (
            self.shared_core(
                global_input
            )
        )


        global_output = (
            self.output_layer(
                global_hidden
            )
        )


        return {
            "scope":
                "global",

            "hidden_state":
                global_hidden,

            **global_output,
        }


    def forward_group(
        self,
        group_conditioned_input,
        group_embedding,
    ):
        """
        Produce one group-conditioned PSV.
        """

        if self.group_adapter is None:
            raise ValueError(
                "Group-conditioned psychological prediction "
                "requires a configured group_embedding_dim."
            )


        shared_group_hidden = (
            self.shared_core(
                group_conditioned_input
            )
        )


        group_adjustment = (
            self.group_adapter(
                group_conditioned_input,
                group_embedding,
            )
        )


        adapted_group_hidden = (
            shared_group_hidden
            + group_adjustment
        )


        group_output = (
            self.output_layer(
                adapted_group_hidden
            )
        )


        return {
            "scope":
                "group_conditioned",

            "hidden_state":
                adapted_group_hidden,

            "shared_hidden_state":
                shared_group_hidden,

            "group_adjustment":
                group_adjustment,

            **group_output,
        }


    def forward(
        self,
        global_input,
        group_conditioned_input=None,
        group_embedding=None,
    ):
        """
        Execute the global path and, where valid group conditioning
        exists, the additional group-conditioned path.
        """

        global_output = (
            self.forward_global(
                global_input
            )
        )


        result = {
            "global":
                global_output,

            "group_conditioned":
                None,

            "group_deviation":
                None,

            "group_deviation_vectors":
                None,
        }


        group_input_present = (
            group_conditioned_input
            is not None
        )

        group_embedding_present = (
            group_embedding
            is not None
        )


        if (
            group_input_present
            != group_embedding_present
        ):
            raise ValueError(
                "Group-conditioned psychological input and "
                "group embedding must either both be supplied "
                "or both be absent."
            )


        if (
            group_input_present
            and group_embedding_present
        ):

            group_output = (
                self.forward_group(
                    group_conditioned_input,
                    group_embedding,
                )
            )


            if (
                group_output[
                    "PSV"
                ].shape
                != global_output[
                    "PSV"
                ].shape
            ):
                raise ValueError(
                    "Global and group-conditioned PSVs must "
                    "have compatible shapes before psychological "
                    "group deviation is calculated."
                )


            group_deviation = (
                group_output[
                    "PSV"
                ]
                - global_output[
                    "PSV"
                ]
            )


            group_deviation_vectors = (
                decompose_psychological_state_vector(
                    group_deviation
                )
            )


            result[
                "group_conditioned"
            ] = group_output

            result[
                "group_deviation"
            ] = group_deviation

            result[
                "group_deviation_vectors"
            ] = group_deviation_vectors


        return result


# -----------------------------------------------------------------------------
# Psychological loss configuration
# -----------------------------------------------------------------------------

PSYCHOLOGICAL_REV_LOSS_NAME = "huber"

PSYCHOLOGICAL_TEV_LOSS_NAME = "huber"

PSYCHOLOGICAL_IEV_LOSS_NAME = "huber"

PSYCHOLOGICAL_EW_LOSS_NAME = "huber"


PSYCHOLOGICAL_REV_LOSS_COEFFICIENT = 1.0

PSYCHOLOGICAL_TEV_LOSS_COEFFICIENT = 1.0

PSYCHOLOGICAL_IEV_LOSS_COEFFICIENT = 1.0

PSYCHOLOGICAL_EW_LOSS_COEFFICIENT = 1.0

PSYCHOLOGICAL_GLOBAL_LOSS_COEFFICIENT = 1.0

PSYCHOLOGICAL_GROUP_LOSS_COEFFICIENT = 1.0


# -----------------------------------------------------------------------------
# Generic masked continuous psychological loss
# -----------------------------------------------------------------------------

def masked_continuous_psychological_loss(
    predicted_values,
    target_values,
    target_mask,
    vector_slice,
    loss_name,
):
    """
    Calculate a mask-aware continuous loss for REV, TEV or IEV.

    Returns None when no supervised values are available.
    """

    predictions = predicted_values[
        ...,
        vector_slice
    ]

    targets = target_values[
        ...,
        vector_slice
    ]

    mask = target_mask[
        ...,
        vector_slice
    ].bool()


    if not torch.any(
        mask
    ):
        return None


    predictions = predictions[
        mask
    ]

    targets = targets[
        mask
    ]


    if loss_name == "huber":

        return F.smooth_l1_loss(
            predictions,
            targets,
        )


    if loss_name == "mse":

        return F.mse_loss(
            predictions,
            targets,
        )


    if loss_name == "mae":

        return F.l1_loss(
            predictions,
            targets,
        )


    raise ValueError(
        "Unsupported psychological continuous loss: "
        f"{loss_name}"
    )


# -----------------------------------------------------------------------------
# Emotional-weight loss
# -----------------------------------------------------------------------------

def masked_emotional_weight_loss(
    predicted_values,
    target_values,
    target_mask,
    loss_name=PSYCHOLOGICAL_EW_LOSS_NAME,
):
    """
    Calculate EW loss using only supervised Emotional Weight targets.
    """

    predictions = predicted_values[
        ...,
        EW_SLICE
    ]

    targets = target_values[
        ...,
        EW_SLICE
    ]

    mask = target_mask[
        ...,
        EW_SLICE
    ].bool()


    if not torch.any(
        mask
    ):
        return None


    predictions = predictions[
        mask
    ]

    targets = targets[
        mask
    ]


    if loss_name == "huber":

        return F.smooth_l1_loss(
            predictions,
            targets,
        )


    if loss_name == "mse":

        return F.mse_loss(
            predictions,
            targets,
        )


    if loss_name == "mae":

        return F.l1_loss(
            predictions,
            targets,
        )


    raise ValueError(
        "Unsupported Emotional Weight loss: "
        f"{loss_name}"
    )


# -----------------------------------------------------------------------------
# Combined psychological loss
# -----------------------------------------------------------------------------

def combine_psychological_component_losses(
    rev_loss,
    tev_loss,
    iev_loss,
    ew_loss,
    rev_coefficient=PSYCHOLOGICAL_REV_LOSS_COEFFICIENT,
    tev_coefficient=PSYCHOLOGICAL_TEV_LOSS_COEFFICIENT,
    iev_coefficient=PSYCHOLOGICAL_IEV_LOSS_COEFFICIENT,
    ew_coefficient=PSYCHOLOGICAL_EW_LOSS_COEFFICIENT,
):
    """
    Combine only available psychological loss terms.

    Missing supervision is not interpreted as zero loss.
    """

    available_terms = []


    if rev_loss is not None:

        available_terms.append(
            float(
                rev_coefficient
            )
            * rev_loss
        )


    if tev_loss is not None:

        available_terms.append(
            float(
                tev_coefficient
            )
            * tev_loss
        )


    if iev_loss is not None:

        available_terms.append(
            float(
                iev_coefficient
            )
            * iev_loss
        )


    if ew_loss is not None:

        available_terms.append(
            float(
                ew_coefficient
            )
            * ew_loss
        )


    if not available_terms:
        return None


    return sum(
        available_terms
    )


def calculate_psychological_loss(
    prediction,
    target_values,
    target_mask,
):
    """
    Calculate the complete mask-aware psychological loss
    for one prediction path.
    """

    if (
        target_values.shape[-1]
        != PSYCHOLOGICAL_STATE_DIMENSION_COUNT
    ):
        raise ValueError(
            "Psychological target tensor must contain "
            "exactly 34 dimensions."
        )


    if (
        target_mask.shape[-1]
        != PSYCHOLOGICAL_STATE_DIMENSION_COUNT
    ):
        raise ValueError(
            "Psychological target mask must contain "
            "exactly 34 dimensions."
        )


    rev_loss = (
        masked_continuous_psychological_loss(
            predicted_values=
                prediction[
                    "values"
                ],

            target_values=
                target_values,

            target_mask=
                target_mask,

            vector_slice=
                REV_SLICE,

            loss_name=
                PSYCHOLOGICAL_REV_LOSS_NAME,
        )
    )


    tev_loss = (
        masked_continuous_psychological_loss(
            predicted_values=
                prediction[
                    "values"
                ],

            target_values=
                target_values,

            target_mask=
                target_mask,

            vector_slice=
                TEV_SLICE,

            loss_name=
                PSYCHOLOGICAL_TEV_LOSS_NAME,
        )
    )


    iev_loss = (
        masked_continuous_psychological_loss(
            predicted_values=
                prediction[
                    "values"
                ],

            target_values=
                target_values,

            target_mask=
                target_mask,

            vector_slice=
                IEV_SLICE,

            loss_name=
                PSYCHOLOGICAL_IEV_LOSS_NAME,
        )
    )


    ew_loss = (
        masked_emotional_weight_loss(
            predicted_values=
                prediction[
                    "values"
                ],

            target_values=
                target_values,

            target_mask=
                target_mask,
        )
    )


    total_loss = (
        combine_psychological_component_losses(
            rev_loss=
                rev_loss,

            tev_loss=
                tev_loss,

            iev_loss=
                iev_loss,

            ew_loss=
                ew_loss,
        )
    )


    return {
        "REV_loss":
            rev_loss,

        "TEV_loss":
            tev_loss,

        "IEV_loss":
            iev_loss,

        "EW_loss":
            ew_loss,

        "total_loss":
            total_loss,
    }


# -----------------------------------------------------------------------------
# Global and group-conditioned psychological loss combination
# -----------------------------------------------------------------------------

def combine_global_and_group_psychological_losses(
    global_loss,
    group_loss=None,
    global_coefficient=PSYCHOLOGICAL_GLOBAL_LOSS_COEFFICIENT,
    group_coefficient=PSYCHOLOGICAL_GROUP_LOSS_COEFFICIENT,
):
    """
    Combine global and group-conditioned psychological losses.

    Group loss remains optional because valid group-conditioned
    supervision may be unavailable.
    """

    available_terms = []


    if global_loss is not None:

        available_terms.append(
            float(
                global_coefficient
            )
            * global_loss
        )


    if group_loss is not None:

        available_terms.append(
            float(
                group_coefficient
            )
            * group_loss
        )


    if not available_terms:
        return None


    return sum(
        available_terms
    )


# -----------------------------------------------------------------------------
# Parameter-group helper
# -----------------------------------------------------------------------------

def psychological_head_parameter_groups(
    psychological_head,
):
    """
    Return explicit psychological-head parameter groups
    for later optimiser construction.
    """

    if not isinstance(
        psychological_head,
        MediaAIPsychologicalHead,
    ):
        raise TypeError(
            "psychological_head must be a "
            "MediaAIPsychologicalHead instance."
        )


    return {
        "shared_psychological_core":
            list(
                psychological_head.shared_core.parameters()
            ),

        "psychological_output":
            list(
                psychological_head.output_layer.parameters()
            ),

        "group_conditioned_adapter":
            (
                list(
                    psychological_head.group_adapter.parameters()
                )
                if psychological_head.group_adapter
                is not None
                else []
            ),
    }


# -----------------------------------------------------------------------------
# Architecture validation
# -----------------------------------------------------------------------------

PSYCHOLOGICAL_HEAD_ARCHITECTURE_ERRORS = []


if not PSYCHOLOGICAL_HEAD_SHARED_CORE:
    PSYCHOLOGICAL_HEAD_ARCHITECTURE_ERRORS.append(
        "global_and_group_core_not_shared"
    )


if not PSYCHOLOGICAL_HEAD_SHARED_OUTPUT:
    PSYCHOLOGICAL_HEAD_ARCHITECTURE_ERRORS.append(
        "global_and_group_output_not_shared"
    )


if not PSYCHOLOGICAL_HEAD_GLOBAL_PATH_REQUIRED:
    PSYCHOLOGICAL_HEAD_ARCHITECTURE_ERRORS.append(
        "global_path_not_required"
    )


if PSYCHOLOGICAL_HEAD_GROUP_DEVIATION_IS_BIAS:
    PSYCHOLOGICAL_HEAD_ARCHITECTURE_ERRORS.append(
        "group_deviation_incorrectly_defined_as_bias"
    )


if PSYCHOLOGICAL_HEAD_GROUP_DEVIATION_IS_POLARISATION:
    PSYCHOLOGICAL_HEAD_ARCHITECTURE_ERRORS.append(
        "group_deviation_incorrectly_defined_as_polarisation"
    )


if PSYCHOLOGICAL_HEAD_GROUP_DEVIATION_IS_MANIPULATION:
    PSYCHOLOGICAL_HEAD_ARCHITECTURE_ERRORS.append(
        "group_deviation_incorrectly_defined_as_manipulation"
    )


if not PSYCHOLOGICAL_HEAD_MASK_AWARE:
    PSYCHOLOGICAL_HEAD_ARCHITECTURE_ERRORS.append(
        "psychological_head_not_mask_aware"
    )


if PSYCHOLOGICAL_HEAD_CONFIDENCE_IN_TARGET:
    PSYCHOLOGICAL_HEAD_ARCHITECTURE_ERRORS.append(
        "confidence_incorrectly_included_in_target"
    )


if PSYCHOLOGICAL_HEAD_EMOTIONS_MUTUALLY_EXCLUSIVE:
    PSYCHOLOGICAL_HEAD_ARCHITECTURE_ERRORS.append(
        "emotions_incorrectly_defined_as_mutually_exclusive"
    )


if PSYCHOLOGICAL_HEAD_SIMPLEX_REQUIRED:
    PSYCHOLOGICAL_HEAD_ARCHITECTURE_ERRORS.append(
        "psychological_output_incorrectly_requires_simplex"
    )


if not PSYCHOLOGICAL_HEAD_POSITIVE_NEGATIVE_SYMMETRY:
    PSYCHOLOGICAL_HEAD_ARCHITECTURE_ERRORS.append(
        "positive_negative_emotional_symmetry_not_preserved"
    )


if PSYCHOLOGICAL_HEAD_TEV_EQUALS_STATE_DIFFERENCE:
    PSYCHOLOGICAL_HEAD_ARCHITECTURE_ERRORS.append(
        "TEV_incorrectly_defined_as_immediate_state_difference"
    )


if PSYCHOLOGICAL_HEAD_TEV_EQUALS_ARCHITECTURAL_TRANSFORMATION:
    PSYCHOLOGICAL_HEAD_ARCHITECTURE_ERRORS.append(
        "TEV_incorrectly_equated_with_architectural_transformation"
    )


if not PSYCHOLOGICAL_HEAD_PROPAGATES_TO_TRANSFORMATIVE:
    PSYCHOLOGICAL_HEAD_ARCHITECTURE_ERRORS.append(
        "psychological_output_not_routed_to_transformative"
    )


if not PSYCHOLOGICAL_HEAD_PROPAGATES_TO_RECURRENT:
    PSYCHOLOGICAL_HEAD_ARCHITECTURE_ERRORS.append(
        "psychological_output_not_routed_to_recurrent"
    )


if PSYCHOLOGICAL_HEAD_DETECTS_GROUPS:
    PSYCHOLOGICAL_HEAD_ARCHITECTURE_ERRORS.append(
        "psychological_head_incorrectly_detects_social_groups"
    )


if PSYCHOLOGICAL_HEAD_HIERARCHICAL_AGGREGATION:
    PSYCHOLOGICAL_HEAD_ARCHITECTURE_ERRORS.append(
        "hierarchical_aggregation_executed_in_psychological_head"
    )


if PSYCHOLOGICAL_HEAD_ATTRIBUTION_CALCULATED:
    PSYCHOLOGICAL_HEAD_ARCHITECTURE_ERRORS.append(
        "attribution_calculated_in_psychological_head"
    )


if PSYCHOLOGICAL_REV_ACTIVATION != "sigmoid":
    PSYCHOLOGICAL_HEAD_ARCHITECTURE_ERRORS.append(
        "REV_activation_not_sigmoid"
    )


if PSYCHOLOGICAL_IEV_ACTIVATION != "sigmoid":
    PSYCHOLOGICAL_HEAD_ARCHITECTURE_ERRORS.append(
        "IEV_activation_not_sigmoid"
    )


if PSYCHOLOGICAL_TEV_ACTIVATION != "tanh":
    PSYCHOLOGICAL_HEAD_ARCHITECTURE_ERRORS.append(
        "TEV_activation_not_tanh"
    )


if PSYCHOLOGICAL_EW_ACTIVATION != "sigmoid":
    PSYCHOLOGICAL_HEAD_ARCHITECTURE_ERRORS.append(
        "EW_activation_not_sigmoid"
    )


if PSYCHOLOGICAL_TEV_LOSS_NAME == "binary_cross_entropy":
    PSYCHOLOGICAL_HEAD_ARCHITECTURE_ERRORS.append(
        "signed_TEV_incorrectly_uses_binary_cross_entropy"
    )


PSYCHOLOGICAL_HEAD_ARCHITECTURE_VALID = (
    len(
        PSYCHOLOGICAL_HEAD_ARCHITECTURE_ERRORS
    )
    == 0
)


if not PSYCHOLOGICAL_HEAD_ARCHITECTURE_VALID:
    raise ValueError(
        "Psychological-head architecture failed validation: "
        f"{PSYCHOLOGICAL_HEAD_ARCHITECTURE_ERRORS}"
    )


# -----------------------------------------------------------------------------
# Execution state
# -----------------------------------------------------------------------------

PSYCHOLOGICAL_HEAD_CLASS_DEFINED = True

PSYCHOLOGICAL_HEAD_MODEL_INSTANTIATED = False

PSYCHOLOGICAL_HEAD_FORWARD_PASS_EXECUTED = False

PSYCHOLOGICAL_HEAD_GROUP_DEVIATION_CALCULATED = False

PSYCHOLOGICAL_HEAD_TRAINING_EXECUTED = False

PSYCHOLOGICAL_HEAD_EVALUATION_EXECUTED = False


# -----------------------------------------------------------------------------
# Architecture summary
# -----------------------------------------------------------------------------

PSYCHOLOGICAL_HEAD_SUMMARY = {
    "version":
        PSYCHOLOGICAL_HEAD_VERSION,

    "representation_level":
        PSYCHOLOGICAL_HEAD_LEVEL,

    "position":
        PSYCHOLOGICAL_HEAD_POSITION,

    "PSV_dimensions":
        PSYCHOLOGICAL_STATE_DIMENSION_COUNT,

    "REV_dimensions":
        REPORTED_EMOTION_DIMENSION_COUNT,

    "TEV_dimensions":
        TRANSFORMATIVE_EMOTION_DIMENSION_COUNT,

    "IEV_dimensions":
        IMMEDIATE_EMOTION_DIMENSION_COUNT,

    "EW_dimensions":
        EMOTIONAL_WEIGHT_DIMENSION_COUNT,

    "emotional_basis":
        PSYCHOLOGICAL_HEAD_EMOTIONAL_DIMENSIONS,

    "REV_range":
        PSYCHOLOGICAL_HEAD_VALUE_RANGES[
            "REV"
        ],

    "TEV_range":
        PSYCHOLOGICAL_HEAD_VALUE_RANGES[
            "TEV"
        ],

    "IEV_range":
        PSYCHOLOGICAL_HEAD_VALUE_RANGES[
            "IEV"
        ],

    "EW_range":
        PSYCHOLOGICAL_HEAD_VALUE_RANGES[
            "EW"
        ],

    "shared_core":
        PSYCHOLOGICAL_HEAD_SHARED_CORE,

    "shared_output":
        PSYCHOLOGICAL_HEAD_SHARED_OUTPUT,

    "group_adaptation":
        PSYCHOLOGICAL_HEAD_GROUP_ADAPTATION,

    "global_path_required":
        PSYCHOLOGICAL_HEAD_GLOBAL_PATH_REQUIRED,

    "group_path_optional":
        PSYCHOLOGICAL_HEAD_GROUP_PATH_OPTIONAL,

    "simplex_required":
        PSYCHOLOGICAL_HEAD_SIMPLEX_REQUIRED,

    "TEV_equals_state_difference":
        PSYCHOLOGICAL_HEAD_TEV_EQUALS_STATE_DIFFERENCE,

    "TEV_equals_architectural_transformation":
        PSYCHOLOGICAL_HEAD_TEV_EQUALS_ARCHITECTURAL_TRANSFORMATION,

    "architecture_valid":
        PSYCHOLOGICAL_HEAD_ARCHITECTURE_VALID,

    "model_instantiated":
        PSYCHOLOGICAL_HEAD_MODEL_INSTANTIATED,

    "training_executed":
        PSYCHOLOGICAL_HEAD_TRAINING_EXECUTED,
}


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Psychological Head and Trainable Parameterisation")
print("=" * 72)

print(
    f"Head version                : "
    f"{PSYCHOLOGICAL_HEAD_VERSION}"
)

print(
    f"Representation level        : "
    f"{PSYCHOLOGICAL_HEAD_LEVEL}"
)

print(
    f"Head position               : "
    f"{PSYCHOLOGICAL_HEAD_POSITION}"
)

print("-" * 72)

print("Psychological State Vector")

print(
    f"PSV dimensions              : "
    f"{PSYCHOLOGICAL_STATE_DIMENSION_COUNT}"
)

print(
    f"REV dimensions              : "
    f"{REPORTED_EMOTION_DIMENSION_COUNT}"
)

print(
    f"TEV dimensions              : "
    f"{TRANSFORMATIVE_EMOTION_DIMENSION_COUNT}"
)

print(
    f"IEV dimensions              : "
    f"{IMMEDIATE_EMOTION_DIMENSION_COUNT}"
)

print(
    f"EW dimensions               : "
    f"{EMOTIONAL_WEIGHT_DIMENSION_COUNT}"
)

print(
    f"Emotional basis             : "
    f"{PSYCHOLOGICAL_HEAD_EMOTION_DIMENSION_COUNT}"
)

print("-" * 72)

print("Value ranges and activations")

print(
    f"REV range                   : "
    f"{PSYCHOLOGICAL_HEAD_VALUE_RANGES['REV']}"
)

print(
    f"REV activation              : "
    f"{PSYCHOLOGICAL_REV_ACTIVATION}"
)

print(
    f"TEV range                   : "
    f"{PSYCHOLOGICAL_HEAD_VALUE_RANGES['TEV']}"
)

print(
    f"TEV activation              : "
    f"{PSYCHOLOGICAL_TEV_ACTIVATION}"
)

print(
    f"IEV range                   : "
    f"{PSYCHOLOGICAL_HEAD_VALUE_RANGES['IEV']}"
)

print(
    f"IEV activation              : "
    f"{PSYCHOLOGICAL_IEV_ACTIVATION}"
)

print(
    f"EW range                    : "
    f"{PSYCHOLOGICAL_HEAD_VALUE_RANGES['EW']}"
)

print(
    f"EW activation               : "
    f"{PSYCHOLOGICAL_EW_ACTIVATION}"
)

print("-" * 72)

print("Global and group paths")

print(
    f"Global path required        : "
    f"{PSYCHOLOGICAL_HEAD_GLOBAL_PATH_REQUIRED}"
)

print(
    f"Group path optional         : "
    f"{PSYCHOLOGICAL_HEAD_GROUP_PATH_OPTIONAL}"
)

print(
    f"Shared psychological core   : "
    f"{PSYCHOLOGICAL_HEAD_SHARED_CORE}"
)

print(
    f"Shared psychological output : "
    f"{PSYCHOLOGICAL_HEAD_SHARED_OUTPUT}"
)

print(
    f"Group adaptation defined    : "
    f"{PSYCHOLOGICAL_HEAD_GROUP_ADAPTATION}"
)

print("-" * 72)

print("Emotional semantics")

print(
    f"Mutually exclusive emotions : "
    f"{PSYCHOLOGICAL_HEAD_EMOTIONS_MUTUALLY_EXCLUSIVE}"
)

print(
    f"Probability simplex required: "
    f"{PSYCHOLOGICAL_HEAD_SIMPLEX_REQUIRED}"
)

print(
    f"Positive/negative symmetry  : "
    f"{PSYCHOLOGICAL_HEAD_POSITIVE_NEGATIVE_SYMMETRY}"
)

print(
    f"TEV = IEV difference        : "
    f"{PSYCHOLOGICAL_HEAD_TEV_EQUALS_STATE_DIFFERENCE}"
)

print(
    f"TEV = architectural change  : "
    f"{PSYCHOLOGICAL_HEAD_TEV_EQUALS_ARCHITECTURAL_TRANSFORMATION}"
)

print("-" * 72)

print("Group deviation")

print(
    f"Signed deviation            : "
    f"{PSYCHOLOGICAL_HEAD_GROUP_DEVIATION_SIGNED}"
)

print(
    f"Deviation = bias            : "
    f"{PSYCHOLOGICAL_HEAD_GROUP_DEVIATION_IS_BIAS}"
)

print(
    f"Deviation = polarisation    : "
    f"{PSYCHOLOGICAL_HEAD_GROUP_DEVIATION_IS_POLARISATION}"
)

print(
    f"Deviation = manipulation    : "
    f"{PSYCHOLOGICAL_HEAD_GROUP_DEVIATION_IS_MANIPULATION}"
)

print(
    f"Deviation calculated        : "
    f"{PSYCHOLOGICAL_HEAD_GROUP_DEVIATION_CALCULATED}"
)

print("-" * 72)

print("Training contract")

print(
    f"Mask-aware                  : "
    f"{PSYCHOLOGICAL_HEAD_MASK_AWARE}"
)

print(
    f"REV loss                    : "
    f"{PSYCHOLOGICAL_REV_LOSS_NAME}"
)

print(
    f"TEV loss                    : "
    f"{PSYCHOLOGICAL_TEV_LOSS_NAME}"
)

print(
    f"IEV loss                    : "
    f"{PSYCHOLOGICAL_IEV_LOSS_NAME}"
)

print(
    f"EW loss                     : "
    f"{PSYCHOLOGICAL_EW_LOSS_NAME}"
)

print(
    f"REV loss coefficient        : "
    f"{PSYCHOLOGICAL_REV_LOSS_COEFFICIENT:.1f}"
)

print(
    f"TEV loss coefficient        : "
    f"{PSYCHOLOGICAL_TEV_LOSS_COEFFICIENT:.1f}"
)

print(
    f"IEV loss coefficient        : "
    f"{PSYCHOLOGICAL_IEV_LOSS_COEFFICIENT:.1f}"
)

print(
    f"EW loss coefficient         : "
    f"{PSYCHOLOGICAL_EW_LOSS_COEFFICIENT:.1f}"
)

print("-" * 72)

print("Methodological boundaries")

print(
    f"Confidence part of target   : "
    f"{PSYCHOLOGICAL_HEAD_CONFIDENCE_IN_TARGET}"
)

print(
    f"Psychological head groups   : "
    f"{PSYCHOLOGICAL_HEAD_DETECTS_GROUPS}"
)

print(
    f"Routes to transformative    : "
    f"{PSYCHOLOGICAL_HEAD_PROPAGATES_TO_TRANSFORMATIVE}"
)

print(
    f"Routes to recurrent         : "
    f"{PSYCHOLOGICAL_HEAD_PROPAGATES_TO_RECURRENT}"
)

print(
    f"Hierarchical aggregation    : "
    f"{PSYCHOLOGICAL_HEAD_HIERARCHICAL_AGGREGATION}"
)

print(
    f"Attribution calculated      : "
    f"{PSYCHOLOGICAL_HEAD_ATTRIBUTION_CALCULATED}"
)

print("-" * 72)

print("Execution state")

print(
    f"Head class defined          : "
    f"{PSYCHOLOGICAL_HEAD_CLASS_DEFINED}"
)

print(
    f"Model instantiated          : "
    f"{PSYCHOLOGICAL_HEAD_MODEL_INSTANTIATED}"
)

print(
    f"Forward pass executed       : "
    f"{PSYCHOLOGICAL_HEAD_FORWARD_PASS_EXECUTED}"
)

print(
    f"Training executed           : "
    f"{PSYCHOLOGICAL_HEAD_TRAINING_EXECUTED}"
)

print(
    f"Evaluation executed         : "
    f"{PSYCHOLOGICAL_HEAD_EVALUATION_EXECUTED}"
)

print(
    f"Architecture valid          : "
    f"{PSYCHOLOGICAL_HEAD_ARCHITECTURE_VALID}"
)

print("=" * 72)

print(
    "Trainable psychological-head architecture defined successfully."
)

print(
    "The Psychological State Vector contains 34 dimensions: "
    "11 REV, 11 TEV, 11 IEV and one EW dimension."
)

print(
    "The psychological vector ordering is inherited directly "
    "from the Block 4 representation contract."
)

print(
    "REV, TEV and IEV share the same ordered eleven-dimensional "
    "emotional basis."
)

print(
    "REV and IEV are continuous bounded representations in [0,1]."
)

print(
    "TEV is a signed continuous representation in [-1,1]."
)

print(
    "EW is a continuous bounded scalar in [0,1]."
)

print(
    "REV and IEV use sigmoid activation, TEV uses tanh activation, "
    "and EW uses sigmoid activation."
)

print(
    "REV, TEV, IEV and EW use mask-aware regression losses."
)

print(
    "The global and group-conditioned paths share the psychological "
    "core and psychological output layer."
)

print(
    "The group-conditioned path may additionally learn a residual "
    "group-specific psychological adaptation."
)

print(
    "The global psychological path remains independently preserved."
)

print(
    "Group-conditioned psychological output is additional and does "
    "not replace the global output."
)

print(
    "Psychological group deviation is signed and is not automatically "
    "interpreted as bias, manipulation or polarisation."
)

print(
    "The eleven emotional dimensions are not mutually exclusive and "
    "are not constrained to a probability simplex."
)

print(
    "Positive and negative emotional dimensions are treated "
    "symmetrically by the architectural contract."
)

print(
    "The Transformative Emotion Vector is not defined as the "
    "difference between consecutive Immediate Emotion Vectors."
)

print(
    "The Transformative Emotion Vector remains distinct from the "
    "architectural transformative mechanism."
)

print(
    "Psychological losses are mask-aware and unavailable supervision "
    "does not contribute to optimisation."
)

print(
    "The psychological head does not independently detect social groups."
)

print(
    "No psychological-head model has been instantiated."
)

print(
    "No forward pass, psychological group deviation, architectural "
    "transformation, recurrence, aggregation, training or evaluation "
    "has been executed."
)
````

## Block 14 — Social Head and Trainable Parameterisation

This block defines the **trainable social prediction head** for the Media AI architecture.

Its purpose is to map both the global and group-conditioned representations into the structured social space defined by the Media AI representation specification.

Unlike the factual and psychological spaces, the social space is not represented by a single fixed-dimensional output vector. It preserves four explicit structural components:

* actors;
* groups;
* memberships;
* relationships.

The number of social entities and links may vary between textual units and articles. The social head must therefore predict structured social objects without imposing a fixed number of actors, groups, memberships or relationships.

### Input Paths

For textual unit \(t\), the social head receives the preserved global representation \(X_t^{global}\) and, where an active group \(g\) exists, the corresponding group-conditioned representation \(X_{t,g}^{group}\).

Both paths remain independently available.

The global path represents social structure inferred without conditioning interpretation on a particular detected group.

The group-conditioned path represents the same social ontology under conditioning relative to an active group.

Group conditioning must not replace or reconstruct the global social representation.

### Shared Social Core

The social head uses a shared trainable core for both paths:

$$H_t^{S,global} = \phi_S(X_t^{global}; \theta_S).$$

$$H_{t,g}^{S,group} = \phi_S(X_{t,g}^{group}; \theta_S).$$

Here, \(\phi_S\) denotes the shared social feature transformation, \(\theta_S\) denotes its trainable parameters, \(H_t^{S,global}\) is the hidden social representation of the global path, and \(H_{t,g}^{S,group}\) is the corresponding hidden representation of the group-conditioned path.

Sharing the social core preserves a common structural ontology across global and group-conditioned interpretation.

### Group-Conditioned Social Adaptation

The group-conditioned path may additionally apply a trainable social adaptation:

$$\widetilde{H}_{t,g}^{S} = H_{t,g}^{S,group} + A_S(X_{t,g}^{group}, E_g; \theta_S^{cond}).$$

Here, \(A_S\) denotes the group-conditioned social adaptation mechanism, \(E_g\) represents active group \(g\), and \(\theta_S^{cond}\) denotes the trainable social conditioning parameters.

This permits the model to represent differences between globally identifiable social structure and structure perceived relative to a particular group.

The global social path remains independently preserved.

### Structured Social Representation

The complete social representation for textual unit \(t\) is denoted by:

$$\mathcal{S}_t = (\mathcal{A}_t,\ \mathcal{G}_t,\ \mathcal{M}_t,\ \mathcal{R}_t).$$

Here:

* \(\mathcal{A}_t\) is the set of detected actors;
* \(\mathcal{G}_t\) is the set of detected groups;
* \(\mathcal{M}_t\) is the set of actor–group memberships;
* \(\mathcal{R}_t\) is the set of detected social relationships.

The cardinalities of these sets are not fixed:

$$|\mathcal{A}_t|,\ |\mathcal{G}_t|,\ |\mathcal{M}_t|,\ |\mathcal{R}_t| \ \text{may vary with } t.$$

Accordingly, the social representation has no fixed total dimensionality.

### Actor Component

The **Actor Component** identifies socially relevant actors explicitly present or supported by the textual evidence.

An actor may represent an individual, organisation, institution or other identifiable social entity participating in the represented situation.

For textual unit \(t\):

$$\mathcal{A}_t = \{a_{t,1}, a_{t,2}, \ldots, a_{t,n_t}\}.$$

The number \(n_t\) is determined by the textual evidence and is not predefined.

Each actor record must preserve at minimum:

* actor identity within the article;
* actor type where supported;
* textual evidence span;
* sentence identity;
* availability state;
* supervision source where available.

The head must not fabricate an actor merely to satisfy a fixed output capacity.

### Group Component

The **Group Component** represents socially relevant groups supported by the textual unit or its permitted context.

For textual unit \(t\):

$$\mathcal{G}_t = \{g_{t,1}, g_{t,2}, \ldots, g_{t,k_t}\}.$$

The number \(k_t\) is variable.

Concrete group identities are not predefined by the architecture.

The social head may learn to recognise article-specific groups from validated supervision and model representations, while the supervisor may provide or validate group labels during training and evaluation.

This preserves the earlier architectural decision that group identities should be dynamically detected rather than fabricated from a closed list.

### Group-Type Priors

The architecture may preserve broad configured group types as structural priors or descriptive categories.

Such group types do not constitute predefined concrete group identities.

A configured type may help describe the nature of a detected group, while the actual group represented in an article remains evidence-dependent.

The distinction is therefore:

$$\text{group type} \neq \text{concrete group identity}.$$

### Membership Component

The **Membership Component** represents supported associations between actors and groups.

For actor \(a\) and group \(g\), a membership relation may be represented as:

$$m_t(a,g) \in [0,1].$$

This value represents the modelled strength or support for the actor–group association under the defined supervision semantics.

Membership is not restricted to a single group.

For an actor \(a\):

$$\sum_g m_t(a,g) \neq 1 \ \text{in general}.$$

An actor may therefore belong to, be associated with, or be represented in relation to multiple groups simultaneously.

Overlapping group membership is an explicit architectural requirement.

### Relationship Component

The **Relationship Component** represents supported social relations between actors, groups or actor–group combinations.

The relationship structure can be represented as a set of typed edges:

$$\mathcal{R}_t = \{(u_i,\ r_i,\ v_i,\ w_i)\}_{i=1}^{q_t}.$$

Here, \(u_i\) is the source entity, \(v_i\) is the target entity, \(r_i\) is the relationship type, and \(w_i\) is the modelled relationship strength or support where applicable.

The number \(q_t\) is variable.

Relationships may be directional or non-directional according to their semantic type.

The architecture must not assume symmetry where the relation itself is directional.

### Social Graph Interpretation

The structured social representation may therefore be interpreted as an article-local social graph:

$$\mathcal{G}^{social}_t = (\mathcal{V}_t,\ \mathcal{E}_t).$$

The node set \(\mathcal{V}_t\) contains supported actors and groups, while the edge set \(\mathcal{E}_t\) contains memberships and other social relationships.

This graph interpretation is a representation contract rather than a requirement to use a particular graph-neural-network implementation.

The exact computational implementation may be selected experimentally.

### Global Social Output

The global social path produces:

$$\mathcal{S}_t^{global} = Head_S(H_t^{S,global}; \theta_S^{out}).$$

Its structure is:

$$\mathcal{S}_t^{global} = (\mathcal{A}_t^{global},\ \mathcal{G}_t^{global},\ \mathcal{M}_t^{global},\ \mathcal{R}_t^{global}).$$

The global social output represents group-independent reconstruction of the social structure supported by the text.

### Group-Conditioned Social Output

For every valid active group \(g\), the group-conditioned path additionally produces:

$$\mathcal{S}_{t,g}^{group} = Head_S(\widetilde{H}_{t,g}^{S}; \theta_S^{out}).$$

Its structure is:

$$\mathcal{S}_{t,g}^{group} = (\mathcal{A}_{t,g}^{group},\ \mathcal{G}_{t,g}^{group},\ \mathcal{M}_{t,g}^{group},\ \mathcal{R}_{t,g}^{group}).$$

The global and group-conditioned outputs share the same social ontology.

Their distinction concerns interpretative scope rather than the definition of actors, groups, memberships or relationships.

### Global and Group-Relative Social Perception

The global path provides the reference social representation.

The group-conditioned path permits social interpretation relative to an active group.

For architectural purposes:

$$\mathcal{S}_t^{objective} \equiv \mathcal{S}_t^{global}.$$

$$\mathcal{S}_{t,g}^{subjective} \equiv \mathcal{S}_{t,g}^{group}.$$

As in the factual and psychological heads, **objective** means independent of group conditioning.

It does not imply externally verified social truth.

Similarly, **subjective** denotes group-conditioned social perception rather than arbitrary interpretation.

### Social Group Deviation

Because the social space is structured and variable-dimensional, social deviation cannot generally be defined as direct vector subtraction.

Instead, deviation must be evaluated over aligned social objects and relations.

For a matched social element \(s\), its signed group deviation may be represented as:

$$D_{t,g}^{S}(s) = v_{t,g}^{group}(s) - v_t^{global}(s).$$

Here, \(v(s)\) denotes a comparable scalar property, score or relation strength associated with the aligned social element.

For structural differences, deviation may additionally involve:

* actors present in one path but absent in the other;
* groups present in one path but absent in the other;
* changed membership strengths;
* changed relationship strengths;
* changed relationship types where supported;
* changed salience of otherwise matched social entities.

Social deviation therefore requires an explicit alignment step before numerical comparison.

### Alignment Requirement

Global and group-conditioned social structures must not be compared solely by output position.

Equivalent actors, groups and relationships must first be aligned by identity or validated semantic correspondence.

Only aligned elements may receive directly comparable signed deviations.

Unmatched elements remain explicit structural differences rather than being assigned fabricated numerical differences.

### Social Deviation Is Not Bias

A difference between global and group-conditioned social representations does not automatically constitute bias.

Therefore:

$$D_{t,g}^{S} \neq \mathrm{Bias}_{t,g} \ \text{by definition}.$$

The same restriction applies to manipulation and polarisation.

These higher-order interpretations require separate evidence and later analytical criteria.

### Dynamic Group Detection and Confluent Routing

The social head and confluent mechanism have related but distinct responsibilities.

The confluent mechanism establishes and preserves global and group-conditioned routing.

The social head learns the explicit structured social representation of actors, groups, memberships and relationships.

Detected group information may therefore support group-conditioned routing, while the social head provides the richer social representation used downstream.

The architecture must avoid a circular rule in which a group is treated as valid solely because the same unvalidated prediction was used to create its own conditioning route.

During training, group-conditioned routes must be grounded in validated, supervisor-supported or otherwise admissible group information according to the training protocol.

### Supervision Targets

Validated social supervision is represented structurally rather than as a single fixed tensor.

For textual unit \(t\), the target may be written as:

$$Y_t^S = (Y_t^A,\ Y_t^G,\ Y_t^M,\ Y_t^R).$$

Here:

* \(Y_t^A\) contains actor targets;
* \(Y_t^G\) contains group targets;
* \(Y_t^M\) contains membership targets;
* \(Y_t^R\) contains relationship targets.

Each component may have independent availability.

Missing social supervision must remain unavailable rather than being converted into negative labels or numerical zero.

### Mask-Aware Social Supervision

The social head must distinguish between:

* supervised presence;
* supervised absence;
* unavailable supervision.

An unannotated actor, group, membership or relationship is not automatically a negative example.

This distinction is particularly important for partially annotated social graphs.

Supervision masks or equivalent structural availability indicators must therefore be preserved for each social component.

### Actor Loss

Actor detection may use a supervised entity-identification loss:

$$\mathcal{L}_A = \mathcal{L}_{actor}(\widehat{\mathcal{A}}_t,\mathcal{A}_t;M_t^A).$$

The exact actor-identification objective may depend on the final implementation, such as span-based, token-based or structured prediction.

The architecture does not commit to one implementation in this block.

### Group Loss

Group detection may use:

$$\mathcal{L}_G = \mathcal{L}_{group}(\widehat{\mathcal{G}}_t,\mathcal{G}_t;M_t^G).$$

This objective trains evidence-grounded group recognition.

It must not require every possible group identity to belong to a predefined closed vocabulary.

### Membership Loss

For supervised actor–group pairs, membership learning may use:

$$\mathcal{L}_M = \mathcal{L}_{membership}(\widehat{\mathcal{M}}_t,\mathcal{M}_t;M_t^M).$$

Membership outputs are independent rather than mutually exclusive, permitting overlapping memberships.

### Relationship Loss

For supervised social relations:

$$\mathcal{L}_R = \mathcal{L}_{relationship}(\widehat{\mathcal{R}}_t,\mathcal{R}_t;M_t^R).$$

The loss must preserve directional distinctions for directional relation types.

### Combined Social Loss

The complete social-head objective combines the available structural losses:

$$\mathcal{L}_S = \lambda_A\mathcal{L}_A + \lambda_G\mathcal{L}_G + \lambda_M\mathcal{L}_M + \lambda_R\mathcal{L}_R.$$

The initial neutral configuration may use:

$$\lambda_A = \lambda_G = \lambda_M = \lambda_R = 1.$$

These coefficients remain configurable and may later be adjusted through controlled experiments.

Unavailable supervision components must not contribute fabricated zero-valued targets to the objective.

### Global and Group-Conditioned Losses

Global and group-conditioned social losses remain distinguishable as \(\mathcal{L}_S^{global}\) and \(\mathcal{L}_S^{group}\).

Where valid group-conditioned supervision exists:

$$\mathcal{L}_S^{total} = \lambda_g^S\mathcal{L}_S^{global} + \lambda_{gc}^S\mathcal{L}_S^{group}.$$

Group-conditioned social loss must only be calculated for valid active groups with admissible supervision.

### Shared Parameters

The global and group-conditioned paths share the social core and social output architecture.

The global parameter set is:

$$\theta_S^{global} = \{\theta_S,\theta_S^{out}\}.$$

The group-conditioned parameter set additionally contains:

$$\theta_S^{group} = \{\theta_S,\theta_S^{out},\theta_S^{cond}\}.$$

This preserves a common social ontology while permitting group-relative adaptation.

### Missing Group Path

If no valid active group exists for textual unit \(t\), the social head produces only the global social representation.

No group-conditioned social structure or social deviation is fabricated.

### Relationship to the Factual Head

The social head does not determine whether claims about actors or groups are factually true.

A social entity may be represented because it occurs within the textual social structure even when the factual status of the surrounding claim remains uncertain.

Factual structure and social structure therefore remain separate model spaces.

### Relationship to the Psychological Head

The social head does not assign emotional states merely because an actor or group is detected.

Psychological interpretation remains the responsibility of the psychological head.

The social representation instead supplies explicit entities, memberships and relations that may later be associated with psychological states through the integrated architecture.

### Transformative Routing

Both global and group-conditioned social representations remain available to the downstream transformative mechanism:

$$\mathcal{S}_t^{global} \rightarrow \Delta_t^{S,global}.$$

$$\mathcal{S}_{t,g}^{group} \rightarrow \Delta_{t,g}^{S,group}.$$

The social head itself does not calculate these architectural transformations.

### Recurrent Routing

The social representations and their later transformations remain available to the recurrent mechanism:

$$\mathcal{S}_t^{global} \rightarrow H_t^{S,global}.$$

$$\mathcal{S}_{t,g}^{group} \rightarrow H_{t,g}^{S,group}.$$

The recurrent mechanism may therefore track the development of actors, groups, memberships and relationships across textual sequence.

The recurrent state itself remains separate from the social-head output.

### Social Polarisation Distance Boundary

The social representation provides structural information required for later **Social Polarisation Distance (SPD)** analysis.

However, the presence of multiple groups or differences between global and group-conditioned representations does not itself establish polarisation.

SPD therefore remains a downstream analytical construct.

It is not calculated by the social head.

### Attribution Boundary

Every predicted social object or relation must preserve backward mappings to its supporting textual evidence.

Attribution metadata should permit later reconstruction of:

* article identity;
* sentence identity;
* actor or group identity;
* social component type;
* source and target entities for relationships;
* textual evidence span;
* token positions where available;
* prediction scope;
* active conditioning group where applicable;
* supervision source;
* availability state.

This permits later backward attribution from social structures to the textual evidence supporting them.

Final attribution scores are not calculated in this block.

### Trainable Parameter Groups

The social head introduces trainable parameter groups for:

* shared social core parameters \(\theta_S\);
* shared social output parameters \(\theta_S^{out}\);
* optional group-conditioned adaptation parameters \(\theta_S^{cond}\).

Additional structured-prediction parameters may later be introduced when the precise actor, group, membership and relationship decoders are instantiated.

Backbone parameters remain governed by a separate fine-tuning policy.

### Architectural Boundary

This block defines the **social-head training contract**, not a final social-decoding algorithm.

The architecture fixes:

* what social structures must be represented;
* how global and group-conditioned paths relate;
* what supervision boundaries must be preserved;
* what may be trained;
* what information must remain attributable.

The exact structured decoder may therefore be selected in Notebook 05 without changing the social ontology defined here, provided it satisfies these contracts.

### Output

This block defines the trainable social-head architecture.

Its outputs specify:

* global social prediction path;
* group-conditioned social prediction path;
* shared social core;
* group-conditioned social adaptation;
* variable-dimensional structured social representation;
* actor component;
* group component;
* membership component;
* relationship component;
* overlapping group membership;
* article-local social graph interpretation;
* dynamic rather than predefined concrete group identities;
* distinction between group types and concrete group identities;
* global and group-relative social scope;
* alignment-aware social group deviation;
* mask-aware structured supervision;
* actor, group, membership and relationship loss contracts;
* combined social loss;
* global and group-conditioned loss boundaries;
* social parameter groups;
* separation from factual truth;
* separation from psychological interpretation;
* transformative and recurrent routing;
* Social Polarisation Distance boundary;
* backward social attribution requirements.

No social-head model parameters are trained by this block.

No social-head forward pass is executed.

No concrete group identity is fabricated.

No social group deviation is calculated.

No architectural transformation is calculated.

No recurrent social state is calculated.

No Social Polarisation Distance is calculated.

No hierarchical aggregation is performed.

No attribution score is calculated.

No experimental evaluation is performed.

````python
# =============================================================================
# Media AI — Block 14: Social Head and Trainable Parameterisation
# =============================================================================

import torch
import torch.nn as nn
import torch.nn.functional as F


# -----------------------------------------------------------------------------
# Block dependency checks
# -----------------------------------------------------------------------------

required_block_4_objects = [
    "MEDIA_AI_REPRESENTATION_SPECIFICATION_VALID",
    "SOCIAL_REPRESENTATION_COMPONENTS",
    "SOCIAL_REPRESENTATION_SPECIFICATION",
    "SOCIAL_REPRESENTATION_FIXED_DIMENSIONAL",
]

missing_block_4_objects = [
    object_name
    for object_name in required_block_4_objects
    if object_name not in globals()
]

if missing_block_4_objects:
    raise NameError(
        "Notebook 04 Block 4 is not initialised. "
        f"Missing: {missing_block_4_objects}"
    )


required_block_11_objects = [
    "GROUP_AWARE_ROUTING_VALID",
    "GROUP_AWARE_ROUTING_RECORDS",
]

missing_block_11_objects = [
    object_name
    for object_name in required_block_11_objects
    if object_name not in globals()
]

if missing_block_11_objects:
    raise NameError(
        "Notebook 04 Block 11 group-aware confluent routing "
        "is not initialised. "
        f"Missing: {missing_block_11_objects}"
    )


required_block_12_objects = [
    "FACTUAL_HEAD_ARCHITECTURE_VALID",
]

missing_block_12_objects = [
    object_name
    for object_name in required_block_12_objects
    if object_name not in globals()
]

if missing_block_12_objects:
    raise NameError(
        "Notebook 04 Block 12 is not initialised. "
        f"Missing: {missing_block_12_objects}"
    )


required_block_13_objects = [
    "PSYCHOLOGICAL_HEAD_ARCHITECTURE_VALID",
]

missing_block_13_objects = [
    object_name
    for object_name in required_block_13_objects
    if object_name not in globals()
]

if missing_block_13_objects:
    raise NameError(
        "Notebook 04 Block 13 is not initialised. "
        f"Missing: {missing_block_13_objects}"
    )


if not MEDIA_AI_REPRESENTATION_SPECIFICATION_VALID:
    raise ValueError(
        "Social head requires a valid "
        "Media AI representation specification."
    )


if not GROUP_AWARE_ROUTING_VALID:
    raise ValueError(
        "Social head requires a valid "
        "group-aware confluent routing contract."
    )


if not FACTUAL_HEAD_ARCHITECTURE_VALID:
    raise ValueError(
        "Social head requires a valid factual-head architecture."
    )


if not PSYCHOLOGICAL_HEAD_ARCHITECTURE_VALID:
    raise ValueError(
        "Social head requires a valid psychological-head architecture."
    )


# -----------------------------------------------------------------------------
# Social-head configuration
# -----------------------------------------------------------------------------

SOCIAL_HEAD_VERSION = "1.0"

SOCIAL_HEAD_LEVEL = "sentence"

SOCIAL_HEAD_SPACE = "social"

SOCIAL_HEAD_POSITION = "post_confluent_conditioning"


# -----------------------------------------------------------------------------
# Social ontology inherited from Block 4
# -----------------------------------------------------------------------------

EXPECTED_SOCIAL_COMPONENTS = (
    "actors",
    "groups",
    "memberships",
    "relationships",
)


if tuple(
    SOCIAL_REPRESENTATION_COMPONENTS
) != EXPECTED_SOCIAL_COMPONENTS:
    raise ValueError(
        "Block 14 social components do not match the "
        "Block 4 social representation contract. "
        f"Expected: {EXPECTED_SOCIAL_COMPONENTS}. "
        f"Received: {tuple(SOCIAL_REPRESENTATION_COMPONENTS)}."
    )


if SOCIAL_REPRESENTATION_FIXED_DIMENSIONAL:
    raise ValueError(
        "The social representation must remain variable-dimensional."
    )


SOCIAL_ACTOR_COMPONENT = "actors"

SOCIAL_GROUP_COMPONENT = "groups"

SOCIAL_MEMBERSHIP_COMPONENT = "memberships"

SOCIAL_RELATIONSHIP_COMPONENT = "relationships"


# -----------------------------------------------------------------------------
# Architectural policy
# -----------------------------------------------------------------------------

SOCIAL_HEAD_SHARED_CORE = True

SOCIAL_HEAD_SHARED_OUTPUT_ARCHITECTURE = True

SOCIAL_HEAD_GROUP_ADAPTATION = True

SOCIAL_HEAD_GLOBAL_PATH_REQUIRED = True

SOCIAL_HEAD_GROUP_PATH_OPTIONAL = True

SOCIAL_HEAD_FIXED_TOTAL_DIMENSION = False

SOCIAL_HEAD_CONCRETE_GROUPS_PREDEFINED = False

SOCIAL_HEAD_OVERLAPPING_MEMBERSHIP_ALLOWED = True

SOCIAL_HEAD_RELATIONSHIPS_SIGNED = True

SOCIAL_HEAD_RELATIONSHIPS_DIRECTIONAL_SUPPORTED = True

SOCIAL_HEAD_ALIGNMENT_REQUIRED_FOR_DEVIATION = True

SOCIAL_HEAD_GROUP_DEVIATION_SIGNED = True

SOCIAL_HEAD_GROUP_DEVIATION_IS_BIAS = False

SOCIAL_HEAD_GROUP_DEVIATION_IS_POLARISATION = False

SOCIAL_HEAD_GROUP_DEVIATION_IS_MANIPULATION = False

SOCIAL_HEAD_MASK_AWARE = True

SOCIAL_HEAD_UNANNOTATED_EQUALS_NEGATIVE = False

SOCIAL_HEAD_CONFIDENCE_IN_TARGET = False

SOCIAL_HEAD_DETECTS_CONCRETE_GROUPS = True

SOCIAL_HEAD_SELF_CONFIRMS_GROUP_ROUTES = False

SOCIAL_HEAD_PROPAGATES_TO_TRANSFORMATIVE = True

SOCIAL_HEAD_PROPAGATES_TO_RECURRENT = True

SOCIAL_HEAD_CALCULATES_SPD = False

SOCIAL_HEAD_HIERARCHICAL_AGGREGATION = False

SOCIAL_HEAD_ATTRIBUTION_CALCULATED = False


# -----------------------------------------------------------------------------
# Social structural object names
# -----------------------------------------------------------------------------

SOCIAL_STRUCTURE_NAME = "SSR"

SOCIAL_ACTOR_SET_NAME = "ACT"

SOCIAL_GROUP_SET_NAME = "GRP"

SOCIAL_MEMBERSHIP_SET_NAME = "MEM"

SOCIAL_RELATIONSHIP_SET_NAME = "REL"


SOCIAL_STRUCTURE_COMPONENT_NAMES = (
    SOCIAL_ACTOR_SET_NAME,
    SOCIAL_GROUP_SET_NAME,
    SOCIAL_MEMBERSHIP_SET_NAME,
    SOCIAL_RELATIONSHIP_SET_NAME,
)


# -----------------------------------------------------------------------------
# Social structure helper
# -----------------------------------------------------------------------------

def create_empty_social_structure(
    scope,
):
    """
    Create an explicitly empty social-structure container.

    This object represents an available structural container with
    no decoded social objects yet. It does not imply that the text
    contains no actors, groups, memberships or relationships.
    """

    if scope not in {
        "global",
        "group_conditioned",
    }:
        raise ValueError(
            "Social structure scope must be global "
            "or group_conditioned."
        )


    return {
        "scope":
            scope,

        "actors":
            [],

        "groups":
            [],

        "memberships":
            [],

        "relationships":
            [],

        "decoded":
            False,
    }


# -----------------------------------------------------------------------------
# Shared social core
# -----------------------------------------------------------------------------

class SharedSocialCore(nn.Module):
    """
    Shared trainable social transformation used by both the global
    and group-conditioned paths.
    """

    def __init__(
        self,
        input_dim,
        hidden_dim,
        dropout=0.1,
    ):
        super().__init__()


        if input_dim <= 0:
            raise ValueError(
                "input_dim must be greater than zero."
            )


        if hidden_dim <= 0:
            raise ValueError(
                "hidden_dim must be greater than zero."
            )


        if not (
            0.0
            <= float(dropout)
            < 1.0
        ):
            raise ValueError(
                "dropout must lie in [0,1)."
            )


        self.input_dim = int(
            input_dim
        )

        self.hidden_dim = int(
            hidden_dim
        )


        self.network = nn.Sequential(
            nn.Linear(
                self.input_dim,
                self.hidden_dim,
            ),

            nn.GELU(),

            nn.Dropout(
                float(dropout)
            ),

            nn.LayerNorm(
                self.hidden_dim
            ),
        )


    def forward(
        self,
        x,
    ):

        if (
            x.shape[-1]
            != self.input_dim
        ):
            raise ValueError(
                "Shared social core received an input "
                "with incompatible dimensionality."
            )


        return self.network(
            x
        )


# -----------------------------------------------------------------------------
# Group-conditioned social adaptation
# -----------------------------------------------------------------------------

class GroupSocialAdapter(nn.Module):
    """
    Trainable residual adaptation for the group-conditioned social path.

    Group embeddings are supplied by the validated group-aware
    conditioning mechanism and are not fabricated here.
    """

    def __init__(
        self,
        input_dim,
        group_embedding_dim,
        hidden_dim,
        dropout=0.1,
    ):
        super().__init__()


        if input_dim <= 0:
            raise ValueError(
                "input_dim must be greater than zero."
            )


        if group_embedding_dim <= 0:
            raise ValueError(
                "group_embedding_dim must be greater than zero."
            )


        if hidden_dim <= 0:
            raise ValueError(
                "hidden_dim must be greater than zero."
            )


        if not (
            0.0
            <= float(dropout)
            < 1.0
        ):
            raise ValueError(
                "dropout must lie in [0,1)."
            )


        self.input_dim = int(
            input_dim
        )

        self.group_embedding_dim = int(
            group_embedding_dim
        )

        self.hidden_dim = int(
            hidden_dim
        )


        self.network = nn.Sequential(
            nn.Linear(
                self.input_dim
                + self.group_embedding_dim,
                self.hidden_dim,
            ),

            nn.GELU(),

            nn.Dropout(
                float(dropout)
            ),

            nn.Linear(
                self.hidden_dim,
                self.hidden_dim,
            ),
        )


    def forward(
        self,
        group_conditioned_input,
        group_embedding,
    ):

        if (
            group_conditioned_input.shape[-1]
            != self.input_dim
        ):
            raise ValueError(
                "Group-conditioned social input has "
                "incompatible dimensionality."
            )


        if (
            group_embedding.shape[-1]
            != self.group_embedding_dim
        ):
            raise ValueError(
                "Social group embedding has "
                "incompatible dimensionality."
            )


        if (
            group_conditioned_input.shape[:-1]
            != group_embedding.shape[:-1]
        ):
            raise ValueError(
                "Group-conditioned social input and group embedding "
                "must have compatible leading dimensions."
            )


        adapter_input = torch.cat(
            [
                group_conditioned_input,
                group_embedding,
            ],
            dim=-1,
        )


        return self.network(
            adapter_input
        )


# -----------------------------------------------------------------------------
# Actor candidate scorer
# -----------------------------------------------------------------------------

class SocialActorScorer(nn.Module):
    """
    Score candidate actor representations.

    Candidate generation and textual-span extraction remain separate
    from this architectural block.
    """

    def __init__(
        self,
        hidden_dim,
    ):
        super().__init__()


        if hidden_dim <= 0:
            raise ValueError(
                "hidden_dim must be greater than zero."
            )


        self.hidden_dim = int(
            hidden_dim
        )


        self.scorer = nn.Linear(
            self.hidden_dim,
            1,
        )


    def forward(
        self,
        actor_candidates,
    ):

        if (
            actor_candidates.shape[-1]
            != self.hidden_dim
        ):
            raise ValueError(
                "Actor candidate representations have "
                "incompatible dimensionality."
            )


        logits = self.scorer(
            actor_candidates
        ).squeeze(-1)


        probabilities = torch.sigmoid(
            logits
        )


        return {
            "logits":
                logits,

            "probabilities":
                probabilities,
        }


# -----------------------------------------------------------------------------
# Group candidate scorer
# -----------------------------------------------------------------------------

class SocialGroupScorer(nn.Module):
    """
    Score candidate group representations without requiring a closed
    vocabulary of concrete group identities.
    """

    def __init__(
        self,
        hidden_dim,
    ):
        super().__init__()


        if hidden_dim <= 0:
            raise ValueError(
                "hidden_dim must be greater than zero."
            )


        self.hidden_dim = int(
            hidden_dim
        )


        self.scorer = nn.Linear(
            self.hidden_dim,
            1,
        )


    def forward(
        self,
        group_candidates,
    ):

        if (
            group_candidates.shape[-1]
            != self.hidden_dim
        ):
            raise ValueError(
                "Group candidate representations have "
                "incompatible dimensionality."
            )


        logits = self.scorer(
            group_candidates
        ).squeeze(-1)


        probabilities = torch.sigmoid(
            logits
        )


        return {
            "logits":
                logits,

            "probabilities":
                probabilities,
        }


# -----------------------------------------------------------------------------
# Actor–group membership scorer
# -----------------------------------------------------------------------------

class SocialMembershipScorer(nn.Module):
    """
    Score actor–group membership associations.

    Membership scores are independent and do not form a simplex,
    permitting overlapping memberships.
    """

    def __init__(
        self,
        hidden_dim,
    ):
        super().__init__()


        if hidden_dim <= 0:
            raise ValueError(
                "hidden_dim must be greater than zero."
            )


        self.hidden_dim = int(
            hidden_dim
        )


        self.network = nn.Sequential(
            nn.Linear(
                self.hidden_dim * 2,
                self.hidden_dim,
            ),

            nn.GELU(),

            nn.Linear(
                self.hidden_dim,
                1,
            ),
        )


    def forward(
        self,
        actor_representation,
        group_representation,
    ):

        if (
            actor_representation.shape[-1]
            != self.hidden_dim
        ):
            raise ValueError(
                "Actor representation has incompatible dimensionality."
            )


        if (
            group_representation.shape[-1]
            != self.hidden_dim
        ):
            raise ValueError(
                "Group representation has incompatible dimensionality."
            )


        if (
            actor_representation.shape[:-1]
            != group_representation.shape[:-1]
        ):
            raise ValueError(
                "Actor and group representations must have "
                "compatible leading dimensions."
            )


        pair_representation = torch.cat(
            [
                actor_representation,
                group_representation,
            ],
            dim=-1,
        )


        logits = self.network(
            pair_representation
        ).squeeze(-1)


        membership_strength = torch.sigmoid(
            logits
        )


        return {
            "logits":
                logits,

            "membership_strength":
                membership_strength,
        }


# -----------------------------------------------------------------------------
# Social relationship scorer
# -----------------------------------------------------------------------------

class SocialRelationshipScorer(nn.Module):
    """
    Score typed relationships between aligned social entities.

    The number and semantic identity of relation types are supplied
    when the decoder is instantiated.
    """

    def __init__(
        self,
        hidden_dim,
        relationship_type_count,
    ):
        super().__init__()


        if hidden_dim <= 0:
            raise ValueError(
                "hidden_dim must be greater than zero."
            )


        if relationship_type_count <= 0:
            raise ValueError(
                "relationship_type_count must be greater than zero."
            )


        self.hidden_dim = int(
            hidden_dim
        )

        self.relationship_type_count = int(
            relationship_type_count
        )


        self.network = nn.Sequential(
            nn.Linear(
                self.hidden_dim * 2,
                self.hidden_dim,
            ),

            nn.GELU(),

            nn.Linear(
                self.hidden_dim,
                self.relationship_type_count,
            ),
        )


    def forward(
        self,
        source_representation,
        target_representation,
    ):

        if (
            source_representation.shape[-1]
            != self.hidden_dim
        ):
            raise ValueError(
                "Source social representation has "
                "incompatible dimensionality."
            )


        if (
            target_representation.shape[-1]
            != self.hidden_dim
        ):
            raise ValueError(
                "Target social representation has "
                "incompatible dimensionality."
            )


        if (
            source_representation.shape[:-1]
            != target_representation.shape[:-1]
        ):
            raise ValueError(
                "Source and target representations must have "
                "compatible leading dimensions."
            )


        pair_representation = torch.cat(
            [
                source_representation,
                target_representation,
            ],
            dim=-1,
        )


        relationship_logits = self.network(
            pair_representation
        )


        return {
            "relationship_logits":
                relationship_logits,
        }


# -----------------------------------------------------------------------------
# Complete social head
# -----------------------------------------------------------------------------

class MediaAISocialHead(nn.Module):
    """
    Trainable Media AI social head.

    The social head preserves a variable-dimensional structured output.
    It does not force actors, groups, memberships and relationships
    into a single fixed-dimensional vector.

    Candidate generation and final discrete decoding are intentionally
    deferred to the training/implementation stage.
    """

    def __init__(
        self,
        input_dim,
        hidden_dim,
        group_embedding_dim=None,
        relationship_type_count=None,
        dropout=0.1,
    ):
        super().__init__()


        self.input_dim = int(
            input_dim
        )

        self.hidden_dim = int(
            hidden_dim
        )

        self.group_embedding_dim = (
            int(group_embedding_dim)
            if group_embedding_dim
            is not None
            else None
        )

        self.relationship_type_count = (
            int(relationship_type_count)
            if relationship_type_count
            is not None
            else None
        )


        self.shared_core = SharedSocialCore(
            input_dim=
                self.input_dim,

            hidden_dim=
                self.hidden_dim,

            dropout=
                dropout,
        )


        if (
            self.group_embedding_dim
            is not None
        ):

            self.group_adapter = (
                GroupSocialAdapter(
                    input_dim=
                        self.input_dim,

                    group_embedding_dim=
                        self.group_embedding_dim,

                    hidden_dim=
                        self.hidden_dim,

                    dropout=
                        dropout,
                )
            )

        else:

            self.group_adapter = None


        self.actor_scorer = SocialActorScorer(
            hidden_dim=
                self.hidden_dim
        )


        self.group_scorer = SocialGroupScorer(
            hidden_dim=
                self.hidden_dim
        )


        self.membership_scorer = (
            SocialMembershipScorer(
                hidden_dim=
                    self.hidden_dim
            )
        )


        if (
            self.relationship_type_count
            is not None
        ):

            self.relationship_scorer = (
                SocialRelationshipScorer(
                    hidden_dim=
                        self.hidden_dim,

                    relationship_type_count=
                        self.relationship_type_count,
                )
            )

        else:

            self.relationship_scorer = None


    def forward_global(
        self,
        global_input,
    ):
        """
        Produce the hidden global social representation.

        Structured social decoding is not executed automatically,
        because actor/group candidates are not fabricated by this block.
        """

        global_hidden = self.shared_core(
            global_input
        )


        return {
            "scope":
                "global",

            "hidden_state":
                global_hidden,

            "structure":
                create_empty_social_structure(
                    scope=
                        "global"
                ),
        }


    def forward_group(
        self,
        group_conditioned_input,
        group_embedding,
    ):
        """
        Produce the hidden group-conditioned social representation.
        """

        if self.group_adapter is None:
            raise ValueError(
                "Group-conditioned social prediction requires "
                "a configured group_embedding_dim."
            )


        shared_group_hidden = self.shared_core(
            group_conditioned_input
        )


        group_adjustment = self.group_adapter(
            group_conditioned_input,
            group_embedding,
        )


        adapted_group_hidden = (
            shared_group_hidden
            + group_adjustment
        )


        return {
            "scope":
                "group_conditioned",

            "hidden_state":
                adapted_group_hidden,

            "shared_hidden_state":
                shared_group_hidden,

            "group_adjustment":
                group_adjustment,

            "structure":
                create_empty_social_structure(
                    scope=
                        "group_conditioned"
                ),
        }


    def forward(
        self,
        global_input,
        group_conditioned_input=None,
        group_embedding=None,
    ):
        """
        Execute the global path and, where valid conditioning exists,
        the additional group-conditioned path.

        No concrete social objects are fabricated.
        """

        global_output = self.forward_global(
            global_input
        )


        result = {
            "global":
                global_output,

            "group_conditioned":
                None,

            "social_deviation":
                None,
        }


        group_input_present = (
            group_conditioned_input
            is not None
        )

        group_embedding_present = (
            group_embedding
            is not None
        )


        if (
            group_input_present
            != group_embedding_present
        ):
            raise ValueError(
                "Group-conditioned social input and group embedding "
                "must either both be supplied or both be absent."
            )


        if (
            group_input_present
            and group_embedding_present
        ):

            group_output = self.forward_group(
                group_conditioned_input,
                group_embedding,
            )


            result[
                "group_conditioned"
            ] = group_output


        return result


# -----------------------------------------------------------------------------
# Social-structure alignment helper
# -----------------------------------------------------------------------------

def align_social_objects(
    global_objects,
    group_objects,
    identity_key,
):
    """
    Align global and group-conditioned social objects by explicit identity.

    Unmatched objects remain explicit structural differences.
    No numerical deviation is fabricated for unmatched objects.
    """

    if not isinstance(
        global_objects,
        list,
    ):
        raise TypeError(
            "global_objects must be a list."
        )


    if not isinstance(
        group_objects,
        list,
    ):
        raise TypeError(
            "group_objects must be a list."
        )


    global_by_id = {
        record.get(
            identity_key
        ):
            record

        for record
        in global_objects

        if record.get(
            identity_key
        )
        is not None
    }


    group_by_id = {
        record.get(
            identity_key
        ):
            record

        for record
        in group_objects

        if record.get(
            identity_key
        )
        is not None
    }


    common_ids = (
        set(
            global_by_id
        )
        & set(
            group_by_id
        )
    )


    global_only_ids = (
        set(
            global_by_id
        )
        - set(
            group_by_id
        )
    )


    group_only_ids = (
        set(
            group_by_id
        )
        - set(
            global_by_id
        )
    )


    return {
        "aligned":
            [
                (
                    global_by_id[
                        object_id
                    ],
                    group_by_id[
                        object_id
                    ],
                )

                for object_id
                in sorted(
                    common_ids
                )
            ],

        "global_only":
            [
                global_by_id[
                    object_id
                ]

                for object_id
                in sorted(
                    global_only_ids
                )
            ],

        "group_only":
            [
                group_by_id[
                    object_id
                ]

                for object_id
                in sorted(
                    group_only_ids
                )
            ],
    }


# -----------------------------------------------------------------------------
# Signed aligned social deviation
# -----------------------------------------------------------------------------

def calculate_aligned_social_scalar_deviation(
    global_value,
    group_value,
):
    """
    Calculate signed deviation for one already aligned comparable
    scalar social property.
    """

    if global_value is None:
        return None


    if group_value is None:
        return None


    return float(
        group_value
    ) - float(
        global_value
    )


# -----------------------------------------------------------------------------
# Masked binary social loss
# -----------------------------------------------------------------------------

def masked_binary_social_loss(
    logits,
    targets,
    mask,
):
    """
    Calculate binary cross-entropy only where social supervision
    is explicitly available.

    Unannotated does not automatically mean negative.
    """

    mask = mask.bool()


    if not torch.any(
        mask
    ):
        return None


    return F.binary_cross_entropy_with_logits(
        logits[
            mask
        ],
        targets[
            mask
        ],
    )


# -----------------------------------------------------------------------------
# Masked membership loss
# -----------------------------------------------------------------------------

def masked_social_membership_loss(
    membership_logits,
    membership_targets,
    membership_mask,
):
    """
    Calculate mask-aware membership loss.

    Membership outputs are independent, allowing overlapping groups.
    """

    return masked_binary_social_loss(
        logits=
            membership_logits,

        targets=
            membership_targets,

        mask=
            membership_mask,
    )


# -----------------------------------------------------------------------------
# Masked relationship loss
# -----------------------------------------------------------------------------

def masked_social_relationship_loss(
    relationship_logits,
    relationship_targets,
    relationship_mask,
):
    """
    Calculate a mask-aware relationship classification loss.

    relationship_targets are class indices.

    relationship_mask identifies explicitly supervised relations.
    """

    relationship_mask = (
        relationship_mask.bool()
    )


    if not torch.any(
        relationship_mask
    ):
        return None


    selected_logits = relationship_logits[
        relationship_mask
    ]

    selected_targets = relationship_targets[
        relationship_mask
    ].long()


    return F.cross_entropy(
        selected_logits,
        selected_targets,
    )


# -----------------------------------------------------------------------------
# Combined social loss
# -----------------------------------------------------------------------------

SOCIAL_ACTOR_LOSS_COEFFICIENT = 1.0

SOCIAL_GROUP_LOSS_COEFFICIENT = 1.0

SOCIAL_MEMBERSHIP_LOSS_COEFFICIENT = 1.0

SOCIAL_RELATIONSHIP_LOSS_COEFFICIENT = 1.0

SOCIAL_GLOBAL_LOSS_COEFFICIENT = 1.0

SOCIAL_GROUP_CONDITIONED_LOSS_COEFFICIENT = 1.0


def combine_social_component_losses(
    actor_loss=None,
    group_loss=None,
    membership_loss=None,
    relationship_loss=None,
    actor_coefficient=SOCIAL_ACTOR_LOSS_COEFFICIENT,
    group_coefficient=SOCIAL_GROUP_LOSS_COEFFICIENT,
    membership_coefficient=SOCIAL_MEMBERSHIP_LOSS_COEFFICIENT,
    relationship_coefficient=SOCIAL_RELATIONSHIP_LOSS_COEFFICIENT,
):
    """
    Combine only available social supervision losses.
    """

    available_terms = []


    if actor_loss is not None:

        available_terms.append(
            float(
                actor_coefficient
            )
            * actor_loss
        )


    if group_loss is not None:

        available_terms.append(
            float(
                group_coefficient
            )
            * group_loss
        )


    if membership_loss is not None:

        available_terms.append(
            float(
                membership_coefficient
            )
            * membership_loss
        )


    if relationship_loss is not None:

        available_terms.append(
            float(
                relationship_coefficient
            )
            * relationship_loss
        )


    if not available_terms:
        return None


    return sum(
        available_terms
    )


# -----------------------------------------------------------------------------
# Global and group-conditioned social loss combination
# -----------------------------------------------------------------------------

def combine_global_and_group_social_losses(
    global_loss,
    group_loss=None,
    global_coefficient=SOCIAL_GLOBAL_LOSS_COEFFICIENT,
    group_coefficient=SOCIAL_GROUP_CONDITIONED_LOSS_COEFFICIENT,
):
    """
    Combine global and group-conditioned social losses.

    Group-conditioned loss remains optional.
    """

    available_terms = []


    if global_loss is not None:

        available_terms.append(
            float(
                global_coefficient
            )
            * global_loss
        )


    if group_loss is not None:

        available_terms.append(
            float(
                group_coefficient
            )
            * group_loss
        )


    if not available_terms:
        return None


    return sum(
        available_terms
    )


# -----------------------------------------------------------------------------
# Parameter-group helper
# -----------------------------------------------------------------------------

def social_head_parameter_groups(
    social_head,
):
    """
    Return explicit social-head parameter groups for later optimiser
    construction.
    """

    if not isinstance(
        social_head,
        MediaAISocialHead,
    ):
        raise TypeError(
            "social_head must be a MediaAISocialHead instance."
        )


    parameter_groups = {
        "shared_social_core":
            list(
                social_head.shared_core.parameters()
            ),

        "actor_scorer":
            list(
                social_head.actor_scorer.parameters()
            ),

        "group_scorer":
            list(
                social_head.group_scorer.parameters()
            ),

        "membership_scorer":
            list(
                social_head.membership_scorer.parameters()
            ),

        "relationship_scorer":
            (
                list(
                    social_head.relationship_scorer.parameters()
                )
                if social_head.relationship_scorer
                is not None
                else []
            ),

        "group_conditioned_adapter":
            (
                list(
                    social_head.group_adapter.parameters()
                )
                if social_head.group_adapter
                is not None
                else []
            ),
    }


    return parameter_groups


# -----------------------------------------------------------------------------
# Architecture validation
# -----------------------------------------------------------------------------

SOCIAL_HEAD_ARCHITECTURE_ERRORS = []


if not SOCIAL_HEAD_SHARED_CORE:
    SOCIAL_HEAD_ARCHITECTURE_ERRORS.append(
        "global_and_group_core_not_shared"
    )


if not SOCIAL_HEAD_SHARED_OUTPUT_ARCHITECTURE:
    SOCIAL_HEAD_ARCHITECTURE_ERRORS.append(
        "global_and_group_social_ontology_not_shared"
    )


if not SOCIAL_HEAD_GLOBAL_PATH_REQUIRED:
    SOCIAL_HEAD_ARCHITECTURE_ERRORS.append(
        "global_path_not_required"
    )


if SOCIAL_HEAD_FIXED_TOTAL_DIMENSION:
    SOCIAL_HEAD_ARCHITECTURE_ERRORS.append(
        "social_space_incorrectly_fixed_dimensional"
    )


if SOCIAL_HEAD_CONCRETE_GROUPS_PREDEFINED:
    SOCIAL_HEAD_ARCHITECTURE_ERRORS.append(
        "concrete_groups_incorrectly_predefined"
    )


if not SOCIAL_HEAD_OVERLAPPING_MEMBERSHIP_ALLOWED:
    SOCIAL_HEAD_ARCHITECTURE_ERRORS.append(
        "overlapping_group_membership_not_allowed"
    )


if not SOCIAL_HEAD_ALIGNMENT_REQUIRED_FOR_DEVIATION:
    SOCIAL_HEAD_ARCHITECTURE_ERRORS.append(
        "social_deviation_without_alignment"
    )


if SOCIAL_HEAD_GROUP_DEVIATION_IS_BIAS:
    SOCIAL_HEAD_ARCHITECTURE_ERRORS.append(
        "social_deviation_incorrectly_defined_as_bias"
    )


if SOCIAL_HEAD_GROUP_DEVIATION_IS_POLARISATION:
    SOCIAL_HEAD_ARCHITECTURE_ERRORS.append(
        "social_deviation_incorrectly_defined_as_polarisation"
    )


if SOCIAL_HEAD_GROUP_DEVIATION_IS_MANIPULATION:
    SOCIAL_HEAD_ARCHITECTURE_ERRORS.append(
        "social_deviation_incorrectly_defined_as_manipulation"
    )


if not SOCIAL_HEAD_MASK_AWARE:
    SOCIAL_HEAD_ARCHITECTURE_ERRORS.append(
        "social_head_not_mask_aware"
    )


if SOCIAL_HEAD_UNANNOTATED_EQUALS_NEGATIVE:
    SOCIAL_HEAD_ARCHITECTURE_ERRORS.append(
        "unannotated_social_structure_incorrectly_negative"
    )


if SOCIAL_HEAD_CONFIDENCE_IN_TARGET:
    SOCIAL_HEAD_ARCHITECTURE_ERRORS.append(
        "confidence_incorrectly_included_in_target"
    )


if not SOCIAL_HEAD_DETECTS_CONCRETE_GROUPS:
    SOCIAL_HEAD_ARCHITECTURE_ERRORS.append(
        "social_head_does_not_support_dynamic_group_detection"
    )


if SOCIAL_HEAD_SELF_CONFIRMS_GROUP_ROUTES:
    SOCIAL_HEAD_ARCHITECTURE_ERRORS.append(
        "social_head_allows_self_confirming_group_routes"
    )


if not SOCIAL_HEAD_PROPAGATES_TO_TRANSFORMATIVE:
    SOCIAL_HEAD_ARCHITECTURE_ERRORS.append(
        "social_output_not_routed_to_transformative"
    )


if not SOCIAL_HEAD_PROPAGATES_TO_RECURRENT:
    SOCIAL_HEAD_ARCHITECTURE_ERRORS.append(
        "social_output_not_routed_to_recurrent"
    )


if SOCIAL_HEAD_CALCULATES_SPD:
    SOCIAL_HEAD_ARCHITECTURE_ERRORS.append(
        "SPD_calculated_inside_social_head"
    )


if SOCIAL_HEAD_HIERARCHICAL_AGGREGATION:
    SOCIAL_HEAD_ARCHITECTURE_ERRORS.append(
        "hierarchical_aggregation_executed_in_social_head"
    )


if SOCIAL_HEAD_ATTRIBUTION_CALCULATED:
    SOCIAL_HEAD_ARCHITECTURE_ERRORS.append(
        "attribution_calculated_in_social_head"
    )


SOCIAL_HEAD_ARCHITECTURE_VALID = (
    len(
        SOCIAL_HEAD_ARCHITECTURE_ERRORS
    )
    == 0
)


if not SOCIAL_HEAD_ARCHITECTURE_VALID:
    raise ValueError(
        "Social-head architecture failed validation: "
        f"{SOCIAL_HEAD_ARCHITECTURE_ERRORS}"
    )


# -----------------------------------------------------------------------------
# Execution state
# -----------------------------------------------------------------------------

SOCIAL_HEAD_CLASS_DEFINED = True

SOCIAL_HEAD_MODEL_INSTANTIATED = False

SOCIAL_HEAD_FORWARD_PASS_EXECUTED = False

SOCIAL_HEAD_STRUCTURED_DECODING_EXECUTED = False

SOCIAL_HEAD_GROUP_DETECTION_EXECUTED = False

SOCIAL_HEAD_GROUP_DEVIATION_CALCULATED = False

SOCIAL_HEAD_SPD_CALCULATED = False

SOCIAL_HEAD_TRAINING_EXECUTED = False

SOCIAL_HEAD_EVALUATION_EXECUTED = False


# -----------------------------------------------------------------------------
# Architecture summary
# -----------------------------------------------------------------------------

SOCIAL_HEAD_SUMMARY = {
    "version":
        SOCIAL_HEAD_VERSION,

    "representation_level":
        SOCIAL_HEAD_LEVEL,

    "position":
        SOCIAL_HEAD_POSITION,

    "components":
        SOCIAL_REPRESENTATION_COMPONENTS,

    "fixed_dimensional":
        SOCIAL_HEAD_FIXED_TOTAL_DIMENSION,

    "shared_core":
        SOCIAL_HEAD_SHARED_CORE,

    "shared_output_architecture":
        SOCIAL_HEAD_SHARED_OUTPUT_ARCHITECTURE,

    "group_adaptation":
        SOCIAL_HEAD_GROUP_ADAPTATION,

    "global_path_required":
        SOCIAL_HEAD_GLOBAL_PATH_REQUIRED,

    "group_path_optional":
        SOCIAL_HEAD_GROUP_PATH_OPTIONAL,

    "concrete_groups_predefined":
        SOCIAL_HEAD_CONCRETE_GROUPS_PREDEFINED,

    "dynamic_group_detection":
        SOCIAL_HEAD_DETECTS_CONCRETE_GROUPS,

    "overlapping_membership":
        SOCIAL_HEAD_OVERLAPPING_MEMBERSHIP_ALLOWED,

    "alignment_required":
        SOCIAL_HEAD_ALIGNMENT_REQUIRED_FOR_DEVIATION,

    "mask_aware":
        SOCIAL_HEAD_MASK_AWARE,

    "SPD_calculated":
        SOCIAL_HEAD_CALCULATES_SPD,

    "architecture_valid":
        SOCIAL_HEAD_ARCHITECTURE_VALID,

    "model_instantiated":
        SOCIAL_HEAD_MODEL_INSTANTIATED,

    "training_executed":
        SOCIAL_HEAD_TRAINING_EXECUTED,
}


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Block 14: Social Head and Trainable Parameterisation")
print("=" * 72)

print(
    f"Head version                : "
    f"{SOCIAL_HEAD_VERSION}"
)

print(
    f"Representation level        : "
    f"{SOCIAL_HEAD_LEVEL}"
)

print(
    f"Head position               : "
    f"{SOCIAL_HEAD_POSITION}"
)

print("-" * 72)

print("Social representation")

print(
    f"Structural components       : "
    f"{', '.join(SOCIAL_REPRESENTATION_COMPONENTS)}"
)

print(
    f"Component count             : "
    f"{len(SOCIAL_REPRESENTATION_COMPONENTS)}"
)

print(
    f"Fixed dimensional           : "
    f"{SOCIAL_HEAD_FIXED_TOTAL_DIMENSION}"
)

print(
    f"Concrete groups predefined  : "
    f"{SOCIAL_HEAD_CONCRETE_GROUPS_PREDEFINED}"
)

print(
    f"Dynamic group detection     : "
    f"{SOCIAL_HEAD_DETECTS_CONCRETE_GROUPS}"
)

print(
    f"Overlapping membership      : "
    f"{SOCIAL_HEAD_OVERLAPPING_MEMBERSHIP_ALLOWED}"
)

print("-" * 72)

print("Global and group paths")

print(
    f"Global path required        : "
    f"{SOCIAL_HEAD_GLOBAL_PATH_REQUIRED}"
)

print(
    f"Group path optional         : "
    f"{SOCIAL_HEAD_GROUP_PATH_OPTIONAL}"
)

print(
    f"Shared social core          : "
    f"{SOCIAL_HEAD_SHARED_CORE}"
)

print(
    f"Shared output architecture  : "
    f"{SOCIAL_HEAD_SHARED_OUTPUT_ARCHITECTURE}"
)

print(
    f"Group adaptation defined    : "
    f"{SOCIAL_HEAD_GROUP_ADAPTATION}"
)

print("-" * 72)

print("Structural semantics")

print(
    f"Signed relationships        : "
    f"{SOCIAL_HEAD_RELATIONSHIPS_SIGNED}"
)

print(
    f"Directional supported       : "
    f"{SOCIAL_HEAD_RELATIONSHIPS_DIRECTIONAL_SUPPORTED}"
)

print(
    f"Alignment before deviation  : "
    f"{SOCIAL_HEAD_ALIGNMENT_REQUIRED_FOR_DEVIATION}"
)

print(
    f"Unannotated = negative      : "
    f"{SOCIAL_HEAD_UNANNOTATED_EQUALS_NEGATIVE}"
)

print(
    f"Self-confirming group routes: "
    f"{SOCIAL_HEAD_SELF_CONFIRMS_GROUP_ROUTES}"
)

print("-" * 72)

print("Group deviation")

print(
    f"Signed deviation            : "
    f"{SOCIAL_HEAD_GROUP_DEVIATION_SIGNED}"
)

print(
    f"Deviation = bias            : "
    f"{SOCIAL_HEAD_GROUP_DEVIATION_IS_BIAS}"
)

print(
    f"Deviation = polarisation    : "
    f"{SOCIAL_HEAD_GROUP_DEVIATION_IS_POLARISATION}"
)

print(
    f"Deviation = manipulation    : "
    f"{SOCIAL_HEAD_GROUP_DEVIATION_IS_MANIPULATION}"
)

print(
    f"Deviation calculated        : "
    f"{SOCIAL_HEAD_GROUP_DEVIATION_CALCULATED}"
)

print("-" * 72)

print("Training contract")

print(
    f"Mask-aware                  : "
    f"{SOCIAL_HEAD_MASK_AWARE}"
)

print(
    f"Actor loss coefficient      : "
    f"{SOCIAL_ACTOR_LOSS_COEFFICIENT:.1f}"
)

print(
    f"Group loss coefficient      : "
    f"{SOCIAL_GROUP_LOSS_COEFFICIENT:.1f}"
)

print(
    f"Membership loss coefficient : "
    f"{SOCIAL_MEMBERSHIP_LOSS_COEFFICIENT:.1f}"
)

print(
    f"Relationship loss coeff.    : "
    f"{SOCIAL_RELATIONSHIP_LOSS_COEFFICIENT:.1f}"
)

print("-" * 72)

print("Methodological boundaries")

print(
    f"Confidence part of target   : "
    f"{SOCIAL_HEAD_CONFIDENCE_IN_TARGET}"
)

print(
    f"Routes to transformative    : "
    f"{SOCIAL_HEAD_PROPAGATES_TO_TRANSFORMATIVE}"
)

print(
    f"Routes to recurrent         : "
    f"{SOCIAL_HEAD_PROPAGATES_TO_RECURRENT}"
)

print(
    f"SPD calculated              : "
    f"{SOCIAL_HEAD_CALCULATES_SPD}"
)

print(
    f"Hierarchical aggregation    : "
    f"{SOCIAL_HEAD_HIERARCHICAL_AGGREGATION}"
)

print(
    f"Attribution calculated      : "
    f"{SOCIAL_HEAD_ATTRIBUTION_CALCULATED}"
)

print("-" * 72)

print("Execution state")

print(
    f"Head class defined          : "
    f"{SOCIAL_HEAD_CLASS_DEFINED}"
)

print(
    f"Model instantiated          : "
    f"{SOCIAL_HEAD_MODEL_INSTANTIATED}"
)

print(
    f"Forward pass executed       : "
    f"{SOCIAL_HEAD_FORWARD_PASS_EXECUTED}"
)

print(
    f"Structured decoding executed: "
    f"{SOCIAL_HEAD_STRUCTURED_DECODING_EXECUTED}"
)

print(
    f"Group detection executed    : "
    f"{SOCIAL_HEAD_GROUP_DETECTION_EXECUTED}"
)

print(
    f"Training executed           : "
    f"{SOCIAL_HEAD_TRAINING_EXECUTED}"
)

print(
    f"Evaluation executed         : "
    f"{SOCIAL_HEAD_EVALUATION_EXECUTED}"
)

print(
    f"Architecture valid          : "
    f"{SOCIAL_HEAD_ARCHITECTURE_VALID}"
)

print("=" * 72)

print(
    "Trainable social-head architecture defined successfully."
)

print(
    "The social head preserves actors, groups, memberships and "
    "relationships as variable-dimensional structured outputs."
)

print(
    "Concrete group identities are not predefined."
)

print(
    "Dynamic group detection is supported without requiring a closed "
    "group vocabulary."
)

print(
    "Overlapping actor-group memberships remain explicitly permitted."
)

print(
    "Global and group-conditioned paths share the social core and "
    "social output architecture."
)

print(
    "The group-conditioned path may additionally learn a residual "
    "group-specific social adaptation."
)

print(
    "The global social representation remains independently preserved."
)

print(
    "Social group deviation requires explicit object alignment before "
    "comparable signed differences are calculated."
)

print(
    "Social deviation is not automatically interpreted as bias, "
    "manipulation or polarisation."
)

print(
    "Unannotated social structure is not automatically treated as "
    "negative supervision."
)

print(
    "The social head must not create self-confirming group-conditioned "
    "routes from its own unvalidated group predictions."
)

print(
    "Social Polarisation Distance remains a downstream analytical "
    "construct and is not calculated by this head."
)

print(
    "No social-head model has been instantiated."
)

print(
    "No forward pass, structured decoding, group detection, social "
    "deviation, transformation, recurrence, SPD, aggregation, "
    "training or evaluation has been executed."
)
````


## Block 15 — Transformative Mechanism

### Purpose

This block defines the trainable **transformative mechanism** of the Media AI architecture.

The transformative mechanism models how the factual, psychological and social representations associated with the current textual step may be transformed in relation to the preceding causal context.

It operates after the factual, psychological and social heads and before the recurrent mechanism.

The mechanism is defined at sentence level and preserves separate global and group-conditioned paths.

This block defines the architecture and training contract only. It does not instantiate or train the transformative model and does not update recurrent state.

---

### Transformative representation

For each sentence \(t\), the transformative mechanism operates separately in three representation spaces:

- **factual space**;
- **psychological space**;
- **social space**.

For a representation space \(q\), where \(q \in \{F,P,S\}\), let:

**Zᵠₜ** — the current representation in space \(q\);

**Cₜ** — the permitted causal contextual representation;

**Δᵠₜ** — the learned transformation associated with the current textual step.

The transformation is represented conceptually as:

**Δᵠₜ = Tᵠ(Zᵠₜ, Cₜ; θᵠ)**

The transformation is therefore a learned function of the current representation and its permitted context.

It is **not** defined as simple subtraction between consecutive representation vectors.

Future textual context is not available to the current transformative step.

---

### Factual transformation

The factual transformative path operates on the fixed ten-dimensional factual representation defined by the factual ontology.

Its output describes a learned transformation within factual representational space.

The factual transformation does not represent a change in externally verified truth. It describes transformation in the model's representation of factological structure and epistemic presentation.

Missing factual supervision remains distinct from numerical zero and is excluded from optimisation through masking.

---

### Psychological transformation

The psychological transformative path operates on the 34-dimensional Psychological State Vector.

The Psychological State Vector preserves:

- 11 Reported Emotion Vector dimensions;
- 11 Transformative Emotion Vector dimensions;
- 11 Immediate Emotion Vector dimensions;
- one Emotional Weight dimension.

The architectural psychological transformation is distinct from the **Transformative Emotion Vector (TEV)**.

In particular:

**TEVₜ ≠ Δᴾₜ**

and TEV is not defined as the difference between consecutive Immediate Emotion Vectors:

**TEVₜ ≠ IEVₜ − IEVₜ₋₁**

TEV remains a substantive psychological representation, whereas the architectural transformative mechanism describes learned representational change.

---

### Social transformation

The social transformative path preserves the variable-dimensional structure of the social representation.

The social space continues to represent:

- actors;
- groups;
- memberships;
- relationships.

The social transformation is therefore not forced into a fixed-dimensional final social vector.

A trainable hidden transformation representation may be learned, while structured social transformation requires explicit object alignment and subsequent structured decoding.

Concrete group identities are not predefined.

Overlapping memberships remain permitted.

---

### Space-specific Transformative Weights

The architecture defines three separate **Transformative Weights**:

**TWᶠₜ** — Factual Transformative Weight;

**TWᵖₜ** — Psychological Transformative Weight;

**TWˢₜ** — Social Transformative Weight.

Each Transformative Weight is a learned scalar bounded to:

**0 ≤ TWᵠₜ ≤ 1**

for \(q \in \{F,P,S\}\).

The three Transformative Weights are learned independently and are not shared across representation spaces.

They are not collapsed into a single overall Transformative Weight.

Conceptually:

**TWᵠₜ = σ(Wᵠₜʷ hᵠₜʷ + bᵠₜʷ)**

where the hidden Transformative Weight representation is derived from the corresponding current representation and causal context.

The three weights therefore provide separate estimates of transformative salience in factual, psychological and social representational space.

---

### Transformative Weight semantics

Transformative Weight represents **learned transformative salience**.

It is distinct from the direction and dimensional structure of the transformation itself.

The transformation:

**Δᵠₜ**

describes what changes and in which representational directions.

The mathematical transformation magnitude:

**Mᵠₜ = ‖Δᵠₜ‖₂**

describes the numerical magnitude of a fixed-dimensional transformation.

The Transformative Weight:

**TWᵠₜ**

describes learned transformative salience.

Consequently:

**TWᵠₜ ≠ Mᵠₜ**

Transformative Weight is also distinct from the psychological Emotional Weight:

**TWᵖₜ ≠ EWₜ**

and no factual or social Transformative Weight is interpreted as an Emotional Weight.

This separation allows a transformation to have a particular mathematical magnitude without requiring the model to assign the same level of learned transformative importance to it.

---

### Transformation magnitude

For fixed-dimensional factual and psychological transformations, transformation magnitude may optionally be calculated using the Euclidean norm:

**Mᵠₜ = ‖Δᵠₜ‖₂**

Transformation magnitude is a derived mathematical quantity.

It is not a learned substitute for Transformative Weight.

Because the social representation is structured and variable-dimensional, an equivalent social magnitude is not assumed without an explicit structural alignment and distance definition.

---

### Transformation gate

The factual and psychological transformative paths include a trainable transformation gate.

For transformation space \(q\), the gate regulates how strongly the predicted transformation is propagated toward the recurrent mechanism.

Conceptually:

**Gᵠₜ = σ(Wᵠɢ Zᵠₜ + bᵠɢ)**

and:

**Δ̃ᵠₜ = Gᵠₜ ⊙ Δᵠₜ**

where **Δ̃ᵠₜ** is the gated transformation.

The transformation gate is therefore distinct from Transformative Weight.

The gate controls propagation at the dimensional level, whereas Transformative Weight represents learned space-level transformative salience.

---

### Global and group-conditioned transformation

The global transformative path is always preserved.

Where validated group-conditioned representations are available, an additional group-conditioned transformative path may be constructed.

The global path represents group-independent transformation.

The group-conditioned path represents transformation under a specific detected group condition.

The group-conditioned path may learn a residual group-specific adaptation while retaining the shared transformative core.

The group-conditioned path does not replace the global path.

---

### Transformative group deviation

Where aligned global and group-conditioned transformations are available, signed transformative deviation may be defined.

For a fixed-dimensional representation space \(q\):

**Dᵠₜ,ɢ = Δ̃ᵠₜ,ɢ − Δ̃ᵠₜ**

where:

**Δ̃ᵠₜ** — gated global transformation;

**Δ̃ᵠₜ,ɢ** — gated group-conditioned transformation.

The deviation is signed.

Transformative deviation is not automatically interpreted as:

- bias;
- manipulation;
- polarisation.

Such interpretations require downstream analytical evidence.

For the structured social space, deviation requires explicit object alignment before comparable differences can be calculated.

---

### Causal constraint

The transformative mechanism is causal.

At sentence \(t\), the transformation may depend on the current representation and permitted preceding context.

Future sentences are unavailable.

The architecture therefore enforces:

**Cₜ ⊆ {information available at or before t}**

This prevents future textual information from leaking into the current transformative state.

---

### Training contract

The factual and psychological transformations use mask-aware regression objectives.

Huber loss is the default regression loss for:

- factual transformation;
- psychological transformation;
- Factual Transformative Weight \(TW^F\);
- Psychological Transformative Weight \(TW^P\);
- Social Transformative Weight \(TW^S\).

The social transformation itself remains governed by a structured training objective because the social representation is variable-dimensional.

The three Transformative Weight losses remain separate:

**L_TW = λ_F L_TW_F + λ_P L_TW_P + λ_S L_TW_S**

with the initial coefficients:

**λ_F = λ_P = λ_S = 1**

These coefficients are initial architectural defaults and may later be treated as experimental hyperparameters.

Unavailable supervision does not contribute to optimisation.

Global and group-conditioned losses also remain independently controllable.

---

### Relationship to the recurrent mechanism

The transformative mechanism produces information for subsequent recurrent processing.

It does **not** update recurrent state itself.

The recurrent mechanism may subsequently receive the relevant:

- factual transformation;
- psychological transformation;
- social transformation representation;
- transformation gates;
- Factual Transformative Weight;
- Psychological Transformative Weight;
- Social Transformative Weight.

The precise recurrent update rule is defined separately in the next architectural block.

This separation preserves a distinction between:

**transformation** — what the current textual step changes;

and

**recurrence** — what information persists and influences subsequent textual steps.

---

### Methodological boundaries

The transformative mechanism does not:

- verify factual truth;
- equate factual transformation with truth change;
- equate TEV with architectural psychological transformation;
- define TEV as consecutive IEV subtraction;
- collapse factual, psychological and social transformations into one representation;
- collapse TW_F, TW_P and TW_S into one Transformative Weight;
- equate Transformative Weight with transformation magnitude;
- equate Transformative Weight with Emotional Weight;
- automatically interpret group deviation as bias;
- automatically interpret group deviation as manipulation;
- automatically interpret group deviation as polarisation;
- use future textual information;
- update recurrent state;
- perform hierarchical aggregation;
- calculate backward attribution.

These operations remain conceptually and computationally separate.

---

### Execution boundary

This block defines the trainable transformative architecture only.

At this stage:

- the transformative classes are defined;
- factual, psychological and social transformative paths are specified;
- TW_F, TW_P and TW_S are specified;
- the transformation gate is specified;
- group-conditioned transformative adaptation is specified;
- mask-aware training objectives are specified;
- recurrent routing is specified.

However:

- no transformative model has been instantiated;
- no forward pass has been executed;
- no factual transformation has been calculated;
- no psychological transformation has been calculated;
- no social transformation has been calculated;
- no TW_F has been calculated;
- no TW_P has been calculated;
- no TW_S has been calculated;
- no transformation magnitude has been calculated;
- no group deviation has been calculated;
- no recurrent state has been updated;
- no training has been performed;
- no evaluation has been performed.

Block 15 therefore establishes the transformative mechanism as a trainable architectural contract while preserving the methodological separation between representation, transformation, transformative salience, recurrence and downstream interpretation.

````python
# =============================================================================
# Media AI — Block 15: Transformative Mechanism
# =============================================================================

import torch
import torch.nn as nn
import torch.nn.functional as F


# -----------------------------------------------------------------------------
# Block dependency checks
# -----------------------------------------------------------------------------

required_block_12_objects = [
    "FACTUAL_HEAD_ARCHITECTURE_VALID",
]

missing_block_12_objects = [
    object_name
    for object_name in required_block_12_objects
    if object_name not in globals()
]

if missing_block_12_objects:
    raise NameError(
        "Notebook 04 Block 12 is not initialised. "
        f"Missing: {missing_block_12_objects}"
    )


required_block_13_objects = [
    "PSYCHOLOGICAL_HEAD_ARCHITECTURE_VALID",
    "PSYCHOLOGICAL_STATE_DIMENSION_COUNT",
]

missing_block_13_objects = [
    object_name
    for object_name in required_block_13_objects
    if object_name not in globals()
]

if missing_block_13_objects:
    raise NameError(
        "Notebook 04 Block 13 is not initialised. "
        f"Missing: {missing_block_13_objects}"
    )


required_block_14_objects = [
    "SOCIAL_HEAD_ARCHITECTURE_VALID",
    "SOCIAL_REPRESENTATION_COMPONENTS",
]

missing_block_14_objects = [
    object_name
    for object_name in required_block_14_objects
    if object_name not in globals()
]

if missing_block_14_objects:
    raise NameError(
        "Notebook 04 Block 14 is not initialised. "
        f"Missing: {missing_block_14_objects}"
    )


if not FACTUAL_HEAD_ARCHITECTURE_VALID:
    raise ValueError(
        "Transformative mechanism requires a valid "
        "factual-head architecture."
    )


if not PSYCHOLOGICAL_HEAD_ARCHITECTURE_VALID:
    raise ValueError(
        "Transformative mechanism requires a valid "
        "psychological-head architecture."
    )


if not SOCIAL_HEAD_ARCHITECTURE_VALID:
    raise ValueError(
        "Transformative mechanism requires a valid "
        "social-head architecture."
    )


# -----------------------------------------------------------------------------
# Transformative-mechanism configuration
# -----------------------------------------------------------------------------

TRANSFORMATIVE_MECHANISM_VERSION = "1.1"

TRANSFORMATIVE_MECHANISM_LEVEL = "sentence"

TRANSFORMATIVE_MECHANISM_POSITION = (
    "post_heads_pre_recurrent"
)

TRANSFORMATIVE_SOURCE_SPACES = (
    "factual",
    "psychological",
    "social",
)


# -----------------------------------------------------------------------------
# Space-specific Transformative Weights
# -----------------------------------------------------------------------------

TRANSFORMATIVE_WEIGHT_SPACES = (
    "factual",
    "psychological",
    "social",
)

TRANSFORMATIVE_WEIGHT_FACTUAL_NAME = "TW_F"

TRANSFORMATIVE_WEIGHT_PSYCHOLOGICAL_NAME = "TW_P"

TRANSFORMATIVE_WEIGHT_SOCIAL_NAME = "TW_S"

TRANSFORMATIVE_WEIGHT_RANGE = (
    0.0,
    1.0,
)

TRANSFORMATIVE_WEIGHT_SHARED_ACROSS_SPACES = False

TRANSFORMATIVE_WEIGHT_COLLAPSED_TO_SINGLE_SCALAR = False


# -----------------------------------------------------------------------------
# Architectural policy
# -----------------------------------------------------------------------------

TRANSFORMATIVE_GLOBAL_PATH_REQUIRED = True

TRANSFORMATIVE_GROUP_PATH_OPTIONAL = True

TRANSFORMATIVE_SHARED_CORE = True

TRANSFORMATIVE_GROUP_ADAPTATION = True

TRANSFORMATIVE_CAUSAL_CONTEXT = True

TRANSFORMATIVE_FUTURE_CONTEXT_ALLOWED = False

TRANSFORMATIVE_EQUALS_SIMPLE_DIFFERENCE = False

TRANSFORMATIVE_TEV_EQUALS_ARCHITECTURAL_CHANGE = False

TRANSFORMATIVE_TEV_EQUALS_IEV_DIFFERENCE = False

TRANSFORMATIVE_MASK_AWARE = True

TRANSFORMATIVE_SOCIAL_ALIGNMENT_REQUIRED = True

TRANSFORMATIVE_GROUP_DEVIATION_SIGNED = True

TRANSFORMATIVE_GROUP_DEVIATION_IS_BIAS = False

TRANSFORMATIVE_GROUP_DEVIATION_IS_MANIPULATION = False

TRANSFORMATIVE_GROUP_DEVIATION_IS_POLARISATION = False

TRANSFORMATIVE_GATE_ENABLED = True

TRANSFORMATIVE_WEIGHT_OPTIONAL = True

TRANSFORMATIVE_WEIGHT_EQUALS_EMOTIONAL_WEIGHT = False

TRANSFORMATIVE_MAGNITUDE_OPTIONAL = True

TRANSFORMATIVE_UPDATES_RECURRENT_STATE = False

TRANSFORMATIVE_PROPAGATES_TO_RECURRENT = True

TRANSFORMATIVE_HIERARCHICAL_AGGREGATION = False

TRANSFORMATIVE_ATTRIBUTION_CALCULATED = False


# -----------------------------------------------------------------------------
# Transformative loss configuration
# -----------------------------------------------------------------------------

TRANSFORMATIVE_FACTUAL_LOSS_NAME = "huber"

TRANSFORMATIVE_PSYCHOLOGICAL_LOSS_NAME = "huber"

TRANSFORMATIVE_SOCIAL_LOSS_NAME = "structured"

TRANSFORMATIVE_WEIGHT_FACTUAL_LOSS_NAME = "huber"

TRANSFORMATIVE_WEIGHT_PSYCHOLOGICAL_LOSS_NAME = "huber"

TRANSFORMATIVE_WEIGHT_SOCIAL_LOSS_NAME = "huber"


TRANSFORMATIVE_FACTUAL_LOSS_COEFFICIENT = 1.0

TRANSFORMATIVE_PSYCHOLOGICAL_LOSS_COEFFICIENT = 1.0

TRANSFORMATIVE_SOCIAL_LOSS_COEFFICIENT = 1.0

TRANSFORMATIVE_WEIGHT_FACTUAL_LOSS_COEFFICIENT = 1.0

TRANSFORMATIVE_WEIGHT_PSYCHOLOGICAL_LOSS_COEFFICIENT = 1.0

TRANSFORMATIVE_WEIGHT_SOCIAL_LOSS_COEFFICIENT = 1.0

TRANSFORMATIVE_GLOBAL_LOSS_COEFFICIENT = 1.0

TRANSFORMATIVE_GROUP_LOSS_COEFFICIENT = 1.0


# -----------------------------------------------------------------------------
# Transformative Weight activation
# -----------------------------------------------------------------------------

def activate_transformative_weight(
    logits,
):
    """
    Convert Transformative Weight logits into [0,1].

    This activation is shared semantically across TW_F, TW_P and TW_S,
    while their predictors remain space-specific.
    """

    return torch.sigmoid(
        logits
    )


# -----------------------------------------------------------------------------
# Generic trainable transformation core
# -----------------------------------------------------------------------------

class TransformativeCore(nn.Module):
    """
    Shared trainable transformation core for one representation space.

    The same space-specific core is used for global and
    group-conditioned paths.
    """

    def __init__(
        self,
        input_dim,
        context_dim,
        hidden_dim,
        output_dim,
        dropout=0.1,
    ):
        super().__init__()


        if input_dim <= 0:
            raise ValueError(
                "input_dim must be greater than zero."
            )


        if context_dim <= 0:
            raise ValueError(
                "context_dim must be greater than zero."
            )


        if hidden_dim <= 0:
            raise ValueError(
                "hidden_dim must be greater than zero."
            )


        if output_dim <= 0:
            raise ValueError(
                "output_dim must be greater than zero."
            )


        if not (
            0.0
            <= float(dropout)
            < 1.0
        ):
            raise ValueError(
                "dropout must lie in [0,1)."
            )


        self.input_dim = int(
            input_dim
        )

        self.context_dim = int(
            context_dim
        )

        self.hidden_dim = int(
            hidden_dim
        )

        self.output_dim = int(
            output_dim
        )


        self.network = nn.Sequential(
            nn.Linear(
                self.input_dim
                + self.context_dim,
                self.hidden_dim,
            ),

            nn.GELU(),

            nn.Dropout(
                float(dropout)
            ),

            nn.LayerNorm(
                self.hidden_dim
            ),

            nn.Linear(
                self.hidden_dim,
                self.output_dim,
            ),
        )


    def forward(
        self,
        current_representation,
        contextual_representation,
    ):

        if (
            current_representation.shape[-1]
            != self.input_dim
        ):
            raise ValueError(
                "Current representation has incompatible "
                "dimensionality."
            )


        if (
            contextual_representation.shape[-1]
            != self.context_dim
        ):
            raise ValueError(
                "Contextual representation has incompatible "
                "dimensionality."
            )


        if (
            current_representation.shape[:-1]
            != contextual_representation.shape[:-1]
        ):
            raise ValueError(
                "Current and contextual representations must "
                "have compatible leading dimensions."
            )


        combined = torch.cat(
            [
                current_representation,
                contextual_representation,
            ],
            dim=-1,
        )


        return self.network(
            combined
        )


# -----------------------------------------------------------------------------
# Group-conditioned transformative adapter
# -----------------------------------------------------------------------------

class TransformativeGroupAdapter(nn.Module):
    """
    Residual adaptation for the group-conditioned transformative path.
    """

    def __init__(
        self,
        input_dim,
        group_embedding_dim,
        output_dim,
        hidden_dim,
        dropout=0.1,
    ):
        super().__init__()


        if input_dim <= 0:
            raise ValueError(
                "input_dim must be greater than zero."
            )


        if group_embedding_dim <= 0:
            raise ValueError(
                "group_embedding_dim must be greater than zero."
            )


        if output_dim <= 0:
            raise ValueError(
                "output_dim must be greater than zero."
            )


        if hidden_dim <= 0:
            raise ValueError(
                "hidden_dim must be greater than zero."
            )


        self.input_dim = int(
            input_dim
        )

        self.group_embedding_dim = int(
            group_embedding_dim
        )

        self.output_dim = int(
            output_dim
        )


        self.network = nn.Sequential(
            nn.Linear(
                self.input_dim
                + self.group_embedding_dim,
                hidden_dim,
            ),

            nn.GELU(),

            nn.Dropout(
                float(dropout)
            ),

            nn.Linear(
                hidden_dim,
                self.output_dim,
            ),
        )


    def forward(
        self,
        current_representation,
        group_embedding,
    ):

        if (
            current_representation.shape[-1]
            != self.input_dim
        ):
            raise ValueError(
                "Group-conditioned transformative input has "
                "incompatible dimensionality."
            )


        if (
            group_embedding.shape[-1]
            != self.group_embedding_dim
        ):
            raise ValueError(
                "Group embedding has incompatible dimensionality."
            )


        if (
            current_representation.shape[:-1]
            != group_embedding.shape[:-1]
        ):
            raise ValueError(
                "Transformative input and group embedding must "
                "have compatible leading dimensions."
            )


        combined = torch.cat(
            [
                current_representation,
                group_embedding,
            ],
            dim=-1,
        )


        return self.network(
            combined
        )


# -----------------------------------------------------------------------------
# Transformation gate
# -----------------------------------------------------------------------------

class TransformationGate(nn.Module):
    """
    Trainable gate regulating how strongly a transformation is
    propagated to the recurrent mechanism.
    """

    def __init__(
        self,
        input_dim,
        transformation_dim,
    ):
        super().__init__()


        if input_dim <= 0:
            raise ValueError(
                "input_dim must be greater than zero."
            )


        if transformation_dim <= 0:
            raise ValueError(
                "transformation_dim must be greater than zero."
            )


        self.input_dim = int(
            input_dim
        )

        self.transformation_dim = int(
            transformation_dim
        )


        self.gate = nn.Linear(
            self.input_dim,
            self.transformation_dim,
        )


    def forward(
        self,
        source_representation,
        transformation,
    ):

        if (
            source_representation.shape[-1]
            != self.input_dim
        ):
            raise ValueError(
                "Transformation gate source input has "
                "incompatible dimensionality."
            )


        if (
            transformation.shape[-1]
            != self.transformation_dim
        ):
            raise ValueError(
                "Transformation gate received an incompatible "
                "transformation dimensionality."
            )


        gate_values = torch.sigmoid(
            self.gate(
                source_representation
            )
        )


        gated_transformation = (
            gate_values
            * transformation
        )


        return {
            "gate":
                gate_values,

            "gated_transformation":
                gated_transformation,
        }


# -----------------------------------------------------------------------------
# Space-specific Transformative Weight head
# -----------------------------------------------------------------------------

class TransformativeWeightHead(nn.Module):
    """
    Scalar Transformative Weight predictor for one representation space.

    Separate instances are used for factual, psychological and social
    transformation spaces.

    TW describes learned transformative salience and is not the
    mathematical norm of the transformation vector.
    """

    def __init__(
        self,
        hidden_dim,
        space_name,
    ):
        super().__init__()


        if hidden_dim <= 0:
            raise ValueError(
                "hidden_dim must be greater than zero."
            )


        if space_name not in TRANSFORMATIVE_WEIGHT_SPACES:
            raise ValueError(
                "Invalid Transformative Weight space."
            )


        self.hidden_dim = int(
            hidden_dim
        )

        self.space_name = str(
            space_name
        )


        self.output_layer = nn.Linear(
            self.hidden_dim,
            1,
        )


    def forward(
        self,
        hidden_state,
    ):

        if (
            hidden_state.shape[-1]
            != self.hidden_dim
        ):
            raise ValueError(
                "Transformative Weight head received "
                "incompatible dimensionality."
            )


        logits = self.output_layer(
            hidden_state
        )


        values = activate_transformative_weight(
            logits
        )


        return {
            "space":
                self.space_name,

            "logits":
                logits,

            "values":
                values,
        }


# -----------------------------------------------------------------------------
# Fixed-dimensional transformation module
# -----------------------------------------------------------------------------

class FixedDimensionalTransformativePath(nn.Module):
    """
    Trainable transformative path for fixed-dimensional spaces.

    Used for factual and psychological representations.
    """

    def __init__(
        self,
        input_dim,
        context_dim,
        hidden_dim,
        output_dim,
        space_name,
        group_embedding_dim=None,
        dropout=0.1,
        enable_transformative_weight=True,
    ):
        super().__init__()


        if space_name not in {
            "factual",
            "psychological",
        }:
            raise ValueError(
                "Fixed-dimensional transformative path must be "
                "factual or psychological."
            )


        self.space_name = str(
            space_name
        )

        self.input_dim = int(
            input_dim
        )

        self.context_dim = int(
            context_dim
        )

        self.hidden_dim = int(
            hidden_dim
        )

        self.output_dim = int(
            output_dim
        )

        self.group_embedding_dim = (
            int(group_embedding_dim)
            if group_embedding_dim
            is not None
            else None
        )

        self.enable_transformative_weight = bool(
            enable_transformative_weight
        )


        self.shared_core = TransformativeCore(
            input_dim=
                self.input_dim,

            context_dim=
                self.context_dim,

            hidden_dim=
                self.hidden_dim,

            output_dim=
                self.output_dim,

            dropout=
                dropout,
        )


        self.gate = TransformationGate(
            input_dim=
                self.input_dim,

            transformation_dim=
                self.output_dim,
        )


        if (
            self.group_embedding_dim
            is not None
        ):

            self.group_adapter = (
                TransformativeGroupAdapter(
                    input_dim=
                        self.input_dim,

                    group_embedding_dim=
                        self.group_embedding_dim,

                    output_dim=
                        self.output_dim,

                    hidden_dim=
                        self.hidden_dim,

                    dropout=
                        dropout,
                )
            )

        else:

            self.group_adapter = None


        if self.enable_transformative_weight:

            self.tw_projection = nn.Sequential(
                nn.Linear(
                    self.input_dim
                    + self.context_dim,
                    self.hidden_dim,
                ),

                nn.GELU(),

                nn.LayerNorm(
                    self.hidden_dim
                ),
            )


            self.tw_head = TransformativeWeightHead(
                hidden_dim=
                    self.hidden_dim,

                space_name=
                    self.space_name,
            )

        else:

            self.tw_projection = None

            self.tw_head = None


    def _calculate_transformative_weight(
        self,
        current_representation,
        contextual_representation,
    ):

        if not self.enable_transformative_weight:
            return None


        combined = torch.cat(
            [
                current_representation,
                contextual_representation,
            ],
            dim=-1,
        )


        hidden = self.tw_projection(
            combined
        )


        return self.tw_head(
            hidden
        )


    def forward_global(
        self,
        current_representation,
        contextual_representation,
    ):

        transformation = self.shared_core(
            current_representation,
            contextual_representation,
        )


        gated = self.gate(
            current_representation,
            transformation,
        )


        transformative_weight = (
            self._calculate_transformative_weight(
                current_representation,
                contextual_representation,
            )
        )


        return {
            "scope":
                "global",

            "space":
                self.space_name,

            "transformation":
                transformation,

            "gate":
                gated[
                    "gate"
                ],

            "gated_transformation":
                gated[
                    "gated_transformation"
                ],

            "transformative_weight":
                transformative_weight,
        }


    def forward_group(
        self,
        current_representation,
        contextual_representation,
        group_embedding,
    ):

        if self.group_adapter is None:
            raise ValueError(
                "Group-conditioned transformation requires "
                "a configured group_embedding_dim."
            )


        shared_transformation = self.shared_core(
            current_representation,
            contextual_representation,
        )


        group_adjustment = self.group_adapter(
            current_representation,
            group_embedding,
        )


        transformation = (
            shared_transformation
            + group_adjustment
        )


        gated = self.gate(
            current_representation,
            transformation,
        )


        transformative_weight = (
            self._calculate_transformative_weight(
                current_representation,
                contextual_representation,
            )
        )


        return {
            "scope":
                "group_conditioned",

            "space":
                self.space_name,

            "shared_transformation":
                shared_transformation,

            "group_adjustment":
                group_adjustment,

            "transformation":
                transformation,

            "gate":
                gated[
                    "gate"
                ],

            "gated_transformation":
                gated[
                    "gated_transformation"
                ],

            "transformative_weight":
                transformative_weight,
        }


# -----------------------------------------------------------------------------
# Structured social transformation path
# -----------------------------------------------------------------------------

class StructuredSocialTransformativePath(nn.Module):
    """
    Trainable social transformative path.

    Social transformation remains structurally variable-dimensional.

    A hidden transformation representation and a dedicated social
    Transformative Weight TW_S are learned without forcing the social
    structure itself into a fixed-dimensional final representation.
    """

    def __init__(
        self,
        input_dim,
        context_dim,
        hidden_dim,
        group_embedding_dim=None,
        dropout=0.1,
        enable_transformative_weight=True,
    ):
        super().__init__()


        if input_dim <= 0:
            raise ValueError(
                "input_dim must be greater than zero."
            )


        if context_dim <= 0:
            raise ValueError(
                "context_dim must be greater than zero."
            )


        if hidden_dim <= 0:
            raise ValueError(
                "hidden_dim must be greater than zero."
            )


        self.input_dim = int(
            input_dim
        )

        self.context_dim = int(
            context_dim
        )

        self.hidden_dim = int(
            hidden_dim
        )

        self.group_embedding_dim = (
            int(group_embedding_dim)
            if group_embedding_dim
            is not None
            else None
        )

        self.enable_transformative_weight = bool(
            enable_transformative_weight
        )


        self.shared_core = nn.Sequential(
            nn.Linear(
                self.input_dim
                + self.context_dim,
                self.hidden_dim,
            ),

            nn.GELU(),

            nn.Dropout(
                float(dropout)
            ),

            nn.LayerNorm(
                self.hidden_dim
            ),
        )


        if (
            self.group_embedding_dim
            is not None
        ):

            self.group_adapter = nn.Sequential(
                nn.Linear(
                    self.input_dim
                    + self.group_embedding_dim,
                    self.hidden_dim,
                ),

                nn.GELU(),

                nn.Linear(
                    self.hidden_dim,
                    self.hidden_dim,
                ),
            )

        else:

            self.group_adapter = None


        if self.enable_transformative_weight:

            self.tw_projection = nn.Sequential(
                nn.Linear(
                    self.input_dim
                    + self.context_dim,
                    self.hidden_dim,
                ),

                nn.GELU(),

                nn.LayerNorm(
                    self.hidden_dim
                ),
            )


            self.tw_head = TransformativeWeightHead(
                hidden_dim=
                    self.hidden_dim,

                space_name=
                    "social",
            )

        else:

            self.tw_projection = None

            self.tw_head = None


    def _calculate_transformative_weight(
        self,
        current_representation,
        contextual_representation,
    ):

        if not self.enable_transformative_weight:
            return None


        combined = torch.cat(
            [
                current_representation,
                contextual_representation,
            ],
            dim=-1,
        )


        hidden = self.tw_projection(
            combined
        )


        return self.tw_head(
            hidden
        )


    def forward_global(
        self,
        current_representation,
        contextual_representation,
    ):

        combined = torch.cat(
            [
                current_representation,
                contextual_representation,
            ],
            dim=-1,
        )


        hidden_transformation = self.shared_core(
            combined
        )


        transformative_weight = (
            self._calculate_transformative_weight(
                current_representation,
                contextual_representation,
            )
        )


        return {
            "scope":
                "global",

            "space":
                "social",

            "hidden_transformation":
                hidden_transformation,

            "structured_transformation":
                None,

            "transformative_weight":
                transformative_weight,

            "decoded":
                False,
        }


    def forward_group(
        self,
        current_representation,
        contextual_representation,
        group_embedding,
    ):

        if self.group_adapter is None:
            raise ValueError(
                "Group-conditioned social transformation requires "
                "a configured group_embedding_dim."
            )


        combined = torch.cat(
            [
                current_representation,
                contextual_representation,
            ],
            dim=-1,
        )


        shared_hidden = self.shared_core(
            combined
        )


        group_input = torch.cat(
            [
                current_representation,
                group_embedding,
            ],
            dim=-1,
        )


        group_adjustment = self.group_adapter(
            group_input
        )


        hidden_transformation = (
            shared_hidden
            + group_adjustment
        )


        transformative_weight = (
            self._calculate_transformative_weight(
                current_representation,
                contextual_representation,
            )
        )


        return {
            "scope":
                "group_conditioned",

            "space":
                "social",

            "shared_hidden_transformation":
                shared_hidden,

            "group_adjustment":
                group_adjustment,

            "hidden_transformation":
                hidden_transformation,

            "structured_transformation":
                None,

            "transformative_weight":
                transformative_weight,

            "decoded":
                False,
        }


# -----------------------------------------------------------------------------
# Complete Media AI transformative mechanism
# -----------------------------------------------------------------------------

class MediaAITransformativeMechanism(nn.Module):
    """
    Complete trainable transformative mechanism.

    It preserves:
        - factual transformation + TW_F
        - psychological transformation + TW_P
        - social transformation + TW_S

    No universal Transformative Weight is created.

    No recurrent state update is performed here.
    """

    def __init__(
        self,
        factual_dim,
        psychological_dim,
        social_hidden_input_dim,
        context_dim,
        hidden_dim,
        group_embedding_dim=None,
        dropout=0.1,
        enable_transformative_weights=True,
    ):
        super().__init__()


        if factual_dim <= 0:
            raise ValueError(
                "factual_dim must be greater than zero."
            )


        if psychological_dim <= 0:
            raise ValueError(
                "psychological_dim must be greater than zero."
            )


        if social_hidden_input_dim <= 0:
            raise ValueError(
                "social_hidden_input_dim must be greater than zero."
            )


        if context_dim <= 0:
            raise ValueError(
                "context_dim must be greater than zero."
            )


        if hidden_dim <= 0:
            raise ValueError(
                "hidden_dim must be greater than zero."
            )


        self.factual_dim = int(
            factual_dim
        )

        self.psychological_dim = int(
            psychological_dim
        )

        self.social_hidden_input_dim = int(
            social_hidden_input_dim
        )

        self.context_dim = int(
            context_dim
        )

        self.hidden_dim = int(
            hidden_dim
        )


        self.factual_path = FixedDimensionalTransformativePath(
            input_dim=
                self.factual_dim,

            context_dim=
                self.context_dim,

            hidden_dim=
                self.hidden_dim,

            output_dim=
                self.factual_dim,

            space_name=
                "factual",

            group_embedding_dim=
                group_embedding_dim,

            dropout=
                dropout,

            enable_transformative_weight=
                enable_transformative_weights,
        )


        self.psychological_path = FixedDimensionalTransformativePath(
            input_dim=
                self.psychological_dim,

            context_dim=
                self.context_dim,

            hidden_dim=
                self.hidden_dim,

            output_dim=
                self.psychological_dim,

            space_name=
                "psychological",

            group_embedding_dim=
                group_embedding_dim,

            dropout=
                dropout,

            enable_transformative_weight=
                enable_transformative_weights,
        )


        self.social_path = StructuredSocialTransformativePath(
            input_dim=
                self.social_hidden_input_dim,

            context_dim=
                self.context_dim,

            hidden_dim=
                self.hidden_dim,

            group_embedding_dim=
                group_embedding_dim,

            dropout=
                dropout,

            enable_transformative_weight=
                enable_transformative_weights,
        )


    def forward_global(
        self,
        factual_representation,
        psychological_representation,
        social_hidden_representation,
        contextual_representation,
    ):

        factual_output = self.factual_path.forward_global(
            factual_representation,
            contextual_representation,
        )


        psychological_output = (
            self.psychological_path.forward_global(
                psychological_representation,
                contextual_representation,
            )
        )


        social_output = self.social_path.forward_global(
            social_hidden_representation,
            contextual_representation,
        )


        return {
            "scope":
                "global",

            "factual":
                factual_output,

            "psychological":
                psychological_output,

            "social":
                social_output,

            "transformative_weights": {
                "TW_F":
                    factual_output[
                        "transformative_weight"
                    ],

                "TW_P":
                    psychological_output[
                        "transformative_weight"
                    ],

                "TW_S":
                    social_output[
                        "transformative_weight"
                    ],
            },
        }


    def forward_group(
        self,
        factual_representation,
        psychological_representation,
        social_hidden_representation,
        contextual_representation,
        group_embedding,
    ):

        factual_output = self.factual_path.forward_group(
            factual_representation,
            contextual_representation,
            group_embedding,
        )


        psychological_output = (
            self.psychological_path.forward_group(
                psychological_representation,
                contextual_representation,
                group_embedding,
            )
        )


        social_output = self.social_path.forward_group(
            social_hidden_representation,
            contextual_representation,
            group_embedding,
        )


        return {
            "scope":
                "group_conditioned",

            "factual":
                factual_output,

            "psychological":
                psychological_output,

            "social":
                social_output,

            "transformative_weights": {
                "TW_F":
                    factual_output[
                        "transformative_weight"
                    ],

                "TW_P":
                    psychological_output[
                        "transformative_weight"
                    ],

                "TW_S":
                    social_output[
                        "transformative_weight"
                    ],
            },
        }


    def forward(
        self,
        global_factual,
        global_psychological,
        global_social_hidden,
        global_context,
        group_factual=None,
        group_psychological=None,
        group_social_hidden=None,
        group_context=None,
        group_embedding=None,
    ):

        global_output = self.forward_global(
            factual_representation=
                global_factual,

            psychological_representation=
                global_psychological,

            social_hidden_representation=
                global_social_hidden,

            contextual_representation=
                global_context,
        )


        result = {
            "global":
                global_output,

            "group_conditioned":
                None,

            "group_deviation":
                None,
        }


        group_inputs = [
            group_factual,
            group_psychological,
            group_social_hidden,
            group_context,
            group_embedding,
        ]


        group_input_count = sum(
            value is not None
            for value
            in group_inputs
        )


        if (
            group_input_count
            not in {
                0,
                len(group_inputs),
            }
        ):
            raise ValueError(
                "Group-conditioned transformative inputs must "
                "either all be supplied or all be absent."
            )


        if group_input_count == len(
            group_inputs
        ):

            group_output = self.forward_group(
                factual_representation=
                    group_factual,

                psychological_representation=
                    group_psychological,

                social_hidden_representation=
                    group_social_hidden,

                contextual_representation=
                    group_context,

                group_embedding=
                    group_embedding,
            )


            factual_deviation = (
                group_output[
                    "factual"
                ][
                    "gated_transformation"
                ]
                - global_output[
                    "factual"
                ][
                    "gated_transformation"
                ]
            )


            psychological_deviation = (
                group_output[
                    "psychological"
                ][
                    "gated_transformation"
                ]
                - global_output[
                    "psychological"
                ][
                    "gated_transformation"
                ]
            )


            result[
                "group_conditioned"
            ] = group_output


            result[
                "group_deviation"
            ] = {
                "factual":
                    factual_deviation,

                "psychological":
                    psychological_deviation,

                "social":
                    None,
            }


        return result


# -----------------------------------------------------------------------------
# Transformation magnitude
# -----------------------------------------------------------------------------

def calculate_transformation_magnitude(
    transformation,
    mask=None,
):
    """
    Calculate L2 transformation magnitude for a fixed-dimensional
    transformation.

    Magnitude remains distinct from TW_F, TW_P or TW_S.
    """

    if mask is None:

        return torch.linalg.vector_norm(
            transformation,
            ord=2,
            dim=-1,
        )


    mask = mask.bool()


    if (
        transformation.shape
        != mask.shape
    ):
        raise ValueError(
            "Transformation and mask must have identical shapes."
        )


    masked_transformation = torch.where(
        mask,
        transformation,
        torch.zeros_like(
            transformation
        ),
    )


    valid_count = mask.sum(
        dim=-1
    )


    magnitude = torch.linalg.vector_norm(
        masked_transformation,
        ord=2,
        dim=-1,
    )


    magnitude = torch.where(
        valid_count > 0,
        magnitude,
        torch.full_like(
            magnitude,
            float("nan"),
        ),
    )


    return magnitude


# -----------------------------------------------------------------------------
# Generic masked transformation regression loss
# -----------------------------------------------------------------------------

def masked_transformative_regression_loss(
    predictions,
    targets,
    mask,
    loss_name="huber",
):
    """
    Calculate mask-aware regression loss for fixed-dimensional
    transformative supervision.
    """

    mask = mask.bool()


    if (
        predictions.shape
        != targets.shape
    ):
        raise ValueError(
            "Transformative predictions and targets must "
            "have identical shapes."
        )


    if (
        predictions.shape
        != mask.shape
    ):
        raise ValueError(
            "Transformative predictions and mask must "
            "have identical shapes."
        )


    if not torch.any(
        mask
    ):
        return None


    selected_predictions = predictions[
        mask
    ]

    selected_targets = targets[
        mask
    ]


    if loss_name == "huber":

        return F.smooth_l1_loss(
            selected_predictions,
            selected_targets,
        )


    if loss_name == "mse":

        return F.mse_loss(
            selected_predictions,
            selected_targets,
        )


    if loss_name == "mae":

        return F.l1_loss(
            selected_predictions,
            selected_targets,
        )


    raise ValueError(
        "Unsupported transformative regression loss: "
        f"{loss_name}"
    )


# -----------------------------------------------------------------------------
# Space-specific Transformative Weight loss
# -----------------------------------------------------------------------------

def masked_transformative_weight_loss(
    predictions,
    targets,
    mask,
    loss_name="huber",
):
    """
    Calculate mask-aware Transformative Weight loss.

    Used independently for TW_F, TW_P and TW_S.
    """

    return masked_transformative_regression_loss(
        predictions=
            predictions,

        targets=
            targets,

        mask=
            mask,

        loss_name=
            loss_name,
    )


# -----------------------------------------------------------------------------
# Combined transformative loss
# -----------------------------------------------------------------------------

def combine_transformative_losses(
    factual_loss=None,
    psychological_loss=None,
    social_loss=None,
    tw_f_loss=None,
    tw_p_loss=None,
    tw_s_loss=None,
    factual_coefficient=TRANSFORMATIVE_FACTUAL_LOSS_COEFFICIENT,
    psychological_coefficient=TRANSFORMATIVE_PSYCHOLOGICAL_LOSS_COEFFICIENT,
    social_coefficient=TRANSFORMATIVE_SOCIAL_LOSS_COEFFICIENT,
    tw_f_coefficient=TRANSFORMATIVE_WEIGHT_FACTUAL_LOSS_COEFFICIENT,
    tw_p_coefficient=TRANSFORMATIVE_WEIGHT_PSYCHOLOGICAL_LOSS_COEFFICIENT,
    tw_s_coefficient=TRANSFORMATIVE_WEIGHT_SOCIAL_LOSS_COEFFICIENT,
):
    """
    Combine only available transformative loss terms.

    TW_F, TW_P and TW_S remain independent loss terms.
    """

    available_terms = []


    if factual_loss is not None:

        available_terms.append(
            float(
                factual_coefficient
            )
            * factual_loss
        )


    if psychological_loss is not None:

        available_terms.append(
            float(
                psychological_coefficient
            )
            * psychological_loss
        )


    if social_loss is not None:

        available_terms.append(
            float(
                social_coefficient
            )
            * social_loss
        )


    if tw_f_loss is not None:

        available_terms.append(
            float(
                tw_f_coefficient
            )
            * tw_f_loss
        )


    if tw_p_loss is not None:

        available_terms.append(
            float(
                tw_p_coefficient
            )
            * tw_p_loss
        )


    if tw_s_loss is not None:

        available_terms.append(
            float(
                tw_s_coefficient
            )
            * tw_s_loss
        )


    if not available_terms:
        return None


    return sum(
        available_terms
    )


# -----------------------------------------------------------------------------
# Global and group-conditioned transformative loss combination
# -----------------------------------------------------------------------------

def combine_global_and_group_transformative_losses(
    global_loss,
    group_loss=None,
    global_coefficient=TRANSFORMATIVE_GLOBAL_LOSS_COEFFICIENT,
    group_coefficient=TRANSFORMATIVE_GROUP_LOSS_COEFFICIENT,
):
    """
    Combine global and group-conditioned transformative losses.
    """

    available_terms = []


    if global_loss is not None:

        available_terms.append(
            float(
                global_coefficient
            )
            * global_loss
        )


    if group_loss is not None:

        available_terms.append(
            float(
                group_coefficient
            )
            * group_loss
        )


    if not available_terms:
        return None


    return sum(
        available_terms
    )


# -----------------------------------------------------------------------------
# Parameter-group helper
# -----------------------------------------------------------------------------

def transformative_parameter_groups(
    transformative_mechanism,
):
    """
    Return explicit transformative parameter groups for later optimiser
    construction.
    """

    if not isinstance(
        transformative_mechanism,
        MediaAITransformativeMechanism,
    ):
        raise TypeError(
            "transformative_mechanism must be a "
            "MediaAITransformativeMechanism instance."
        )


    groups = {
        "factual_shared_transform":
            list(
                transformative_mechanism
                .factual_path
                .shared_core
                .parameters()
            ),

        "psychological_shared_transform":
            list(
                transformative_mechanism
                .psychological_path
                .shared_core
                .parameters()
            ),

        "social_shared_transform":
            list(
                transformative_mechanism
                .social_path
                .shared_core
                .parameters()
            ),

        "factual_gate":
            list(
                transformative_mechanism
                .factual_path
                .gate
                .parameters()
            ),

        "psychological_gate":
            list(
                transformative_mechanism
                .psychological_path
                .gate
                .parameters()
            ),
    }


    if (
        transformative_mechanism
        .factual_path
        .tw_head
        is not None
    ):

        groups[
            "TW_F"
        ] = list(
            transformative_mechanism
            .factual_path
            .tw_projection
            .parameters()
        ) + list(
            transformative_mechanism
            .factual_path
            .tw_head
            .parameters()
        )

    else:

        groups[
            "TW_F"
        ] = []


    if (
        transformative_mechanism
        .psychological_path
        .tw_head
        is not None
    ):

        groups[
            "TW_P"
        ] = list(
            transformative_mechanism
            .psychological_path
            .tw_projection
            .parameters()
        ) + list(
            transformative_mechanism
            .psychological_path
            .tw_head
            .parameters()
        )

    else:

        groups[
            "TW_P"
        ] = []


    if (
        transformative_mechanism
        .social_path
        .tw_head
        is not None
    ):

        groups[
            "TW_S"
        ] = list(
            transformative_mechanism
            .social_path
            .tw_projection
            .parameters()
        ) + list(
            transformative_mechanism
            .social_path
            .tw_head
            .parameters()
        )

    else:

        groups[
            "TW_S"
        ] = []


    if (
        transformative_mechanism
        .factual_path
        .group_adapter
        is not None
    ):

        groups[
            "factual_group_adapter"
        ] = list(
            transformative_mechanism
            .factual_path
            .group_adapter
            .parameters()
        )

    else:

        groups[
            "factual_group_adapter"
        ] = []


    if (
        transformative_mechanism
        .psychological_path
        .group_adapter
        is not None
    ):

        groups[
            "psychological_group_adapter"
        ] = list(
            transformative_mechanism
            .psychological_path
            .group_adapter
            .parameters()
        )

    else:

        groups[
            "psychological_group_adapter"
        ] = []


    if (
        transformative_mechanism
        .social_path
        .group_adapter
        is not None
    ):

        groups[
            "social_group_adapter"
        ] = list(
            transformative_mechanism
            .social_path
            .group_adapter
            .parameters()
        )

    else:

        groups[
            "social_group_adapter"
        ] = []


    return groups


# -----------------------------------------------------------------------------
# Architecture validation
# -----------------------------------------------------------------------------

TRANSFORMATIVE_ARCHITECTURE_ERRORS = []


if not TRANSFORMATIVE_GLOBAL_PATH_REQUIRED:
    TRANSFORMATIVE_ARCHITECTURE_ERRORS.append(
        "global_transformative_path_not_required"
    )


if not TRANSFORMATIVE_SHARED_CORE:
    TRANSFORMATIVE_ARCHITECTURE_ERRORS.append(
        "transformative_core_not_shared"
    )


if TRANSFORMATIVE_FUTURE_CONTEXT_ALLOWED:
    TRANSFORMATIVE_ARCHITECTURE_ERRORS.append(
        "future_context_incorrectly_allowed"
    )


if not TRANSFORMATIVE_CAUSAL_CONTEXT:
    TRANSFORMATIVE_ARCHITECTURE_ERRORS.append(
        "causal_context_not_preserved"
    )


if TRANSFORMATIVE_EQUALS_SIMPLE_DIFFERENCE:
    TRANSFORMATIVE_ARCHITECTURE_ERRORS.append(
        "transformation_incorrectly_defined_as_simple_difference"
    )


if TRANSFORMATIVE_TEV_EQUALS_ARCHITECTURAL_CHANGE:
    TRANSFORMATIVE_ARCHITECTURE_ERRORS.append(
        "TEV_incorrectly_equated_with_architectural_transformation"
    )


if TRANSFORMATIVE_TEV_EQUALS_IEV_DIFFERENCE:
    TRANSFORMATIVE_ARCHITECTURE_ERRORS.append(
        "TEV_incorrectly_equated_with_IEV_difference"
    )


if not TRANSFORMATIVE_MASK_AWARE:
    TRANSFORMATIVE_ARCHITECTURE_ERRORS.append(
        "transformative_mechanism_not_mask_aware"
    )


if not TRANSFORMATIVE_SOCIAL_ALIGNMENT_REQUIRED:
    TRANSFORMATIVE_ARCHITECTURE_ERRORS.append(
        "social_transformation_alignment_not_required"
    )


if TRANSFORMATIVE_GROUP_DEVIATION_IS_BIAS:
    TRANSFORMATIVE_ARCHITECTURE_ERRORS.append(
        "transformative_deviation_incorrectly_defined_as_bias"
    )


if TRANSFORMATIVE_GROUP_DEVIATION_IS_MANIPULATION:
    TRANSFORMATIVE_ARCHITECTURE_ERRORS.append(
        "transformative_deviation_incorrectly_defined_as_manipulation"
    )


if TRANSFORMATIVE_GROUP_DEVIATION_IS_POLARISATION:
    TRANSFORMATIVE_ARCHITECTURE_ERRORS.append(
        "transformative_deviation_incorrectly_defined_as_polarisation"
    )


if TRANSFORMATIVE_WEIGHT_EQUALS_EMOTIONAL_WEIGHT:
    TRANSFORMATIVE_ARCHITECTURE_ERRORS.append(
        "transformative_weight_incorrectly_equated_with_emotional_weight"
    )


if TRANSFORMATIVE_WEIGHT_SHARED_ACROSS_SPACES:
    TRANSFORMATIVE_ARCHITECTURE_ERRORS.append(
        "transformative_weights_incorrectly_shared_across_spaces"
    )


if TRANSFORMATIVE_WEIGHT_COLLAPSED_TO_SINGLE_SCALAR:
    TRANSFORMATIVE_ARCHITECTURE_ERRORS.append(
        "transformative_weights_incorrectly_collapsed"
    )


if len(
    TRANSFORMATIVE_WEIGHT_SPACES
) != 3:
    TRANSFORMATIVE_ARCHITECTURE_ERRORS.append(
        "three_space_specific_transformative_weights_not_defined"
    )


if TRANSFORMATIVE_UPDATES_RECURRENT_STATE:
    TRANSFORMATIVE_ARCHITECTURE_ERRORS.append(
        "transformative_block_incorrectly_updates_recurrent_state"
    )


if not TRANSFORMATIVE_PROPAGATES_TO_RECURRENT:
    TRANSFORMATIVE_ARCHITECTURE_ERRORS.append(
        "transformative_output_not_routed_to_recurrent"
    )


if TRANSFORMATIVE_HIERARCHICAL_AGGREGATION:
    TRANSFORMATIVE_ARCHITECTURE_ERRORS.append(
        "hierarchical_aggregation_executed_in_transformative_block"
    )


if TRANSFORMATIVE_ATTRIBUTION_CALCULATED:
    TRANSFORMATIVE_ARCHITECTURE_ERRORS.append(
        "attribution_calculated_in_transformative_block"
    )


TRANSFORMATIVE_ARCHITECTURE_VALID = (
    len(
        TRANSFORMATIVE_ARCHITECTURE_ERRORS
    )
    == 0
)


if not TRANSFORMATIVE_ARCHITECTURE_VALID:
    raise ValueError(
        "Transformative architecture failed validation: "
        f"{TRANSFORMATIVE_ARCHITECTURE_ERRORS}"
    )


# -----------------------------------------------------------------------------
# Execution state
# -----------------------------------------------------------------------------

TRANSFORMATIVE_CLASS_DEFINED = True

TRANSFORMATIVE_MODEL_INSTANTIATED = False

TRANSFORMATIVE_FORWARD_PASS_EXECUTED = False

TRANSFORMATIVE_FACTUAL_CALCULATED = False

TRANSFORMATIVE_PSYCHOLOGICAL_CALCULATED = False

TRANSFORMATIVE_SOCIAL_CALCULATED = False

TRANSFORMATIVE_WEIGHT_FACTUAL_CALCULATED = False

TRANSFORMATIVE_WEIGHT_PSYCHOLOGICAL_CALCULATED = False

TRANSFORMATIVE_WEIGHT_SOCIAL_CALCULATED = False

TRANSFORMATIVE_MAGNITUDE_CALCULATED = False

TRANSFORMATIVE_GROUP_DEVIATION_CALCULATED = False

TRANSFORMATIVE_RECURRENT_STATE_UPDATED = False

TRANSFORMATIVE_TRAINING_EXECUTED = False

TRANSFORMATIVE_EVALUATION_EXECUTED = False


# -----------------------------------------------------------------------------
# Architecture summary
# -----------------------------------------------------------------------------

TRANSFORMATIVE_SUMMARY = {
    "version":
        TRANSFORMATIVE_MECHANISM_VERSION,

    "representation_level":
        TRANSFORMATIVE_MECHANISM_LEVEL,

    "position":
        TRANSFORMATIVE_MECHANISM_POSITION,

    "source_spaces":
        TRANSFORMATIVE_SOURCE_SPACES,

    "transformative_weight_spaces":
        TRANSFORMATIVE_WEIGHT_SPACES,

    "TW_F":
        TRANSFORMATIVE_WEIGHT_FACTUAL_NAME,

    "TW_P":
        TRANSFORMATIVE_WEIGHT_PSYCHOLOGICAL_NAME,

    "TW_S":
        TRANSFORMATIVE_WEIGHT_SOCIAL_NAME,

    "transformative_weight_range":
        TRANSFORMATIVE_WEIGHT_RANGE,

    "transformative_weights_shared":
        TRANSFORMATIVE_WEIGHT_SHARED_ACROSS_SPACES,

    "single_transformative_weight":
        TRANSFORMATIVE_WEIGHT_COLLAPSED_TO_SINGLE_SCALAR,

    "global_path_required":
        TRANSFORMATIVE_GLOBAL_PATH_REQUIRED,

    "group_path_optional":
        TRANSFORMATIVE_GROUP_PATH_OPTIONAL,

    "shared_core":
        TRANSFORMATIVE_SHARED_CORE,

    "group_adaptation":
        TRANSFORMATIVE_GROUP_ADAPTATION,

    "causal_context":
        TRANSFORMATIVE_CAUSAL_CONTEXT,

    "future_context_allowed":
        TRANSFORMATIVE_FUTURE_CONTEXT_ALLOWED,

    "simple_difference":
        TRANSFORMATIVE_EQUALS_SIMPLE_DIFFERENCE,

    "mask_aware":
        TRANSFORMATIVE_MASK_AWARE,

    "gate_enabled":
        TRANSFORMATIVE_GATE_ENABLED,

    "magnitude_optional":
        TRANSFORMATIVE_MAGNITUDE_OPTIONAL,

    "updates_recurrent_state":
        TRANSFORMATIVE_UPDATES_RECURRENT_STATE,

    "routes_to_recurrent":
        TRANSFORMATIVE_PROPAGATES_TO_RECURRENT,

    "architecture_valid":
        TRANSFORMATIVE_ARCHITECTURE_VALID,

    "model_instantiated":
        TRANSFORMATIVE_MODEL_INSTANTIATED,

    "training_executed":
        TRANSFORMATIVE_TRAINING_EXECUTED,
}


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Block 15: Transformative Mechanism")
print("=" * 72)

print(
    f"Mechanism version           : "
    f"{TRANSFORMATIVE_MECHANISM_VERSION}"
)

print(
    f"Representation level        : "
    f"{TRANSFORMATIVE_MECHANISM_LEVEL}"
)

print(
    f"Mechanism position          : "
    f"{TRANSFORMATIVE_MECHANISM_POSITION}"
)

print(
    f"Source spaces               : "
    f"{', '.join(TRANSFORMATIVE_SOURCE_SPACES)}"
)

print("-" * 72)

print("Space-specific Transformative Weights")

print(
    f"Transformative Weight count : "
    f"{len(TRANSFORMATIVE_WEIGHT_SPACES)}"
)

print(
    f"Factual Transformative Wt.  : "
    f"{TRANSFORMATIVE_WEIGHT_FACTUAL_NAME}"
)

print(
    f"Psychological Transform. Wt.: "
    f"{TRANSFORMATIVE_WEIGHT_PSYCHOLOGICAL_NAME}"
)

print(
    f"Social Transformative Wt.   : "
    f"{TRANSFORMATIVE_WEIGHT_SOCIAL_NAME}"
)

print(
    f"Transformative Weight range : "
    f"{TRANSFORMATIVE_WEIGHT_RANGE}"
)

print(
    f"Shared across spaces        : "
    f"{TRANSFORMATIVE_WEIGHT_SHARED_ACROSS_SPACES}"
)

print(
    f"Collapsed to one TW         : "
    f"{TRANSFORMATIVE_WEIGHT_COLLAPSED_TO_SINGLE_SCALAR}"
)

print(
    f"TW = Emotional Weight       : "
    f"{TRANSFORMATIVE_WEIGHT_EQUALS_EMOTIONAL_WEIGHT}"
)

print("-" * 72)

print("Global and group paths")

print(
    f"Global path required        : "
    f"{TRANSFORMATIVE_GLOBAL_PATH_REQUIRED}"
)

print(
    f"Group path optional         : "
    f"{TRANSFORMATIVE_GROUP_PATH_OPTIONAL}"
)

print(
    f"Shared transformative core  : "
    f"{TRANSFORMATIVE_SHARED_CORE}"
)

print(
    f"Group adaptation defined    : "
    f"{TRANSFORMATIVE_GROUP_ADAPTATION}"
)

print("-" * 72)

print("Sequential semantics")

print(
    f"Causal context              : "
    f"{TRANSFORMATIVE_CAUSAL_CONTEXT}"
)

print(
    f"Future context allowed      : "
    f"{TRANSFORMATIVE_FUTURE_CONTEXT_ALLOWED}"
)

print(
    f"Transformation = difference : "
    f"{TRANSFORMATIVE_EQUALS_SIMPLE_DIFFERENCE}"
)

print(
    f"TEV = architectural change  : "
    f"{TRANSFORMATIVE_TEV_EQUALS_ARCHITECTURAL_CHANGE}"
)

print(
    f"TEV = IEV difference        : "
    f"{TRANSFORMATIVE_TEV_EQUALS_IEV_DIFFERENCE}"
)

print("-" * 72)

print("Transformation controls")

print(
    f"Mask-aware                  : "
    f"{TRANSFORMATIVE_MASK_AWARE}"
)

print(
    f"Social alignment required   : "
    f"{TRANSFORMATIVE_SOCIAL_ALIGNMENT_REQUIRED}"
)

print(
    f"Transformation gate         : "
    f"{TRANSFORMATIVE_GATE_ENABLED}"
)

print(
    f"Magnitude optional          : "
    f"{TRANSFORMATIVE_MAGNITUDE_OPTIONAL}"
)

print("-" * 72)

print("Group deviation")

print(
    f"Signed deviation            : "
    f"{TRANSFORMATIVE_GROUP_DEVIATION_SIGNED}"
)

print(
    f"Deviation = bias            : "
    f"{TRANSFORMATIVE_GROUP_DEVIATION_IS_BIAS}"
)

print(
    f"Deviation = manipulation    : "
    f"{TRANSFORMATIVE_GROUP_DEVIATION_IS_MANIPULATION}"
)

print(
    f"Deviation = polarisation    : "
    f"{TRANSFORMATIVE_GROUP_DEVIATION_IS_POLARISATION}"
)

print(
    f"Deviation calculated        : "
    f"{TRANSFORMATIVE_GROUP_DEVIATION_CALCULATED}"
)

print("-" * 72)

print("Training contract")

print(
    f"Factual transform loss      : "
    f"{TRANSFORMATIVE_FACTUAL_LOSS_NAME}"
)

print(
    f"Psychological transform loss: "
    f"{TRANSFORMATIVE_PSYCHOLOGICAL_LOSS_NAME}"
)

print(
    f"Social transform loss       : "
    f"{TRANSFORMATIVE_SOCIAL_LOSS_NAME}"
)

print(
    f"TW_F loss                   : "
    f"{TRANSFORMATIVE_WEIGHT_FACTUAL_LOSS_NAME}"
)

print(
    f"TW_P loss                   : "
    f"{TRANSFORMATIVE_WEIGHT_PSYCHOLOGICAL_LOSS_NAME}"
)

print(
    f"TW_S loss                   : "
    f"{TRANSFORMATIVE_WEIGHT_SOCIAL_LOSS_NAME}"
)

print("-" * 72)

print("Recurrent boundary")

print(
    f"Updates recurrent state     : "
    f"{TRANSFORMATIVE_UPDATES_RECURRENT_STATE}"
)

print(
    f"Routes to recurrent         : "
    f"{TRANSFORMATIVE_PROPAGATES_TO_RECURRENT}"
)

print(
    f"Recurrent state updated     : "
    f"{TRANSFORMATIVE_RECURRENT_STATE_UPDATED}"
)

print("-" * 72)

print("Execution state")

print(
    f"Class defined               : "
    f"{TRANSFORMATIVE_CLASS_DEFINED}"
)

print(
    f"Model instantiated          : "
    f"{TRANSFORMATIVE_MODEL_INSTANTIATED}"
)

print(
    f"Forward pass executed       : "
    f"{TRANSFORMATIVE_FORWARD_PASS_EXECUTED}"
)

print(
    f"Factual transformation      : "
    f"{TRANSFORMATIVE_FACTUAL_CALCULATED}"
)

print(
    f"Psychological transformation: "
    f"{TRANSFORMATIVE_PSYCHOLOGICAL_CALCULATED}"
)

print(
    f"Social transformation       : "
    f"{TRANSFORMATIVE_SOCIAL_CALCULATED}"
)

print(
    f"TW_F calculated             : "
    f"{TRANSFORMATIVE_WEIGHT_FACTUAL_CALCULATED}"
)

print(
    f"TW_P calculated             : "
    f"{TRANSFORMATIVE_WEIGHT_PSYCHOLOGICAL_CALCULATED}"
)

print(
    f"TW_S calculated             : "
    f"{TRANSFORMATIVE_WEIGHT_SOCIAL_CALCULATED}"
)

print(
    f"Transformation magnitude    : "
    f"{TRANSFORMATIVE_MAGNITUDE_CALCULATED}"
)

print(
    f"Training executed           : "
    f"{TRANSFORMATIVE_TRAINING_EXECUTED}"
)

print(
    f"Evaluation executed         : "
    f"{TRANSFORMATIVE_EVALUATION_EXECUTED}"
)

print(
    f"Architecture valid          : "
    f"{TRANSFORMATIVE_ARCHITECTURE_VALID}"
)

print("=" * 72)

print(
    "Trainable transformative architecture defined successfully."
)

print(
    "Factual, psychological and social transformations remain "
    "separate architectural objects."
)

print(
    "Three separate Transformative Weights are preserved: "
    "TW_F, TW_P and TW_S."
)

print(
    "The three Transformative Weights are not collapsed into a "
    "single overall transformation score."
)

print(
    "TW_F represents learned factual transformative salience."
)

print(
    "TW_P represents learned psychological transformative salience."
)

print(
    "TW_S represents learned social transformative salience."
)

print(
    "Transformative Weights remain distinct from mathematical "
    "transformation magnitude."
)

print(
    "Transformative Weights also remain distinct from psychological "
    "Emotional Weight."
)

print(
    "Global and group-conditioned transformative paths are preserved "
    "independently."
)

print(
    "The transformative mechanism is not defined as direct subtraction "
    "between consecutive head outputs."
)

print(
    "TEV remains distinct from both architectural psychological "
    "transformation and consecutive IEV differences."
)

print(
    "The transformative mechanism is causal and does not permit future "
    "context in the current sequential contract."
)

print(
    "Variable-dimensional social transformation retains its structured "
    "representation while supporting an independent TW_S."
)

print(
    "The transformative mechanism routes information to the recurrent "
    "mechanism but does not update recurrent state itself."
)

print(
    "No transformative model has been instantiated."
)

print(
    "No forward pass, transformation, TW_F, TW_P, TW_S, magnitude, "
    "group deviation, recurrence, training or evaluation has been executed."
)
````

## Block 16 — Recurrent Mechanism

### Purpose

This block defines the trainable **recurrent mechanism** of the Media AI architecture.

The recurrent mechanism models how representational information persists and evolves across the ordered sentence sequence of an article.

It operates after the factual, psychological and social heads and after the transformative mechanism defined in Block 15.

The recurrent mechanism is causal. At sentence \(t\), it may use the current representation, the current transformation and the recurrent state inherited from preceding sentences, but it may not use information from future sentences.

The architecture preserves separate recurrent states for the factual, psychological and social representation spaces.

Global and group-conditioned recurrent paths also remain separate.

This block defines the architecture and training contract only. It does not instantiate or train the recurrent model.

---

### Recurrent state spaces

The architecture defines three recurrent state spaces:

- **factual recurrent state**;
- **psychological recurrent state**;
- **social recurrent state**.

For representation space \(q\), where \(q \in \{F,P,S\}\), let:

**\(Z^q_t\)** — current representation;

**\(\tilde{\Delta}^q_t\)** — current gated transformation;

**\(TW^q_t\)** — current space-specific Transformative Weight;

**\(H^q_{t-1}\)** — recurrent state inherited from the preceding textual step;

**\(H^q_t\)** — updated recurrent state.

The recurrent mechanism therefore models:

$$
H^q_t = R^q(H^q_{t-1}, Z^q_t, \tilde{\Delta}^q_t, TW^q_t; \theta^q_R)
$$

The three recurrent states are not collapsed into a single universal recurrent vector.

---

### Factual recurrent state

The factual recurrent state tracks the persistence and evolution of factual and epistemic structure across the sentence sequence.

It receives information from:

- the current factual representation;
- the gated factual transformation;
- the Factual Transformative Weight \(TW^F\);
- the preceding factual recurrent state.

The factual recurrent state does not represent externally verified truth.

It represents the sequential persistence of factological structure and epistemic presentation within the analysed text.

A statement remaining influential in the factual recurrent state therefore does not imply that the statement is true.

---

### Psychological recurrent state

The psychological recurrent state tracks the persistence and evolution of psychological and emotional structure across the sentence sequence.

It receives information from:

- the current 34-dimensional Psychological State Vector;
- the gated psychological transformation;
- the Psychological Transformative Weight \(TW^P\);
- the preceding psychological recurrent state.

This permits emotional effects introduced earlier in the article to remain active beyond the sentence in which they first appear.

The recurrent mechanism therefore distinguishes an immediate psychological state from a psychological state that has accumulated through preceding textual context.

The recurrent state does not replace REV, TEV, IEV or EW.

These remain substantive components of the Psychological State Vector.

---

### Social recurrent state

The social recurrent state tracks the persistence and evolution of social structure across the article.

The social space remains structured and variable-dimensional.

The recurrent mechanism therefore preserves information concerning:

- actors;
- groups;
- memberships;
- relationships.

It may also receive:

- the current hidden social transformation representation;
- the Social Transformative Weight \(TW^S\);
- the preceding social recurrent state.

The social recurrent mechanism must preserve identity alignment through the sequence.

An actor, group, membership or relationship observed at sentence \(t\) must not be treated as a new object at sentence \(t+1\) merely because its surface wording changes.

Entity and group alignment therefore precede recurrent social updating where required.

---

### Persistence and update

The recurrent mechanism must allow information from the preceding state to:

- persist;
- weaken;
- be reinforced;
- be modified;
- or be replaced by sufficiently strong new evidence.

The recurrent state is therefore not a cumulative sum of preceding representations.

A trainable update mechanism determines how much of the preceding state is retained and how much current information is incorporated.

For a fixed-dimensional representation space \(q\), define a recurrent retention gate:

$$
K^q_t =
\sigma
\left(
W^q_K
[H^q_{t-1}, Z^q_t, \tilde{\Delta}^q_t, TW^q_t]
+
b^q_K
\right)
$$

and a candidate recurrent state:

$$
\hat{H}^q_t =
\tanh
\left(
W^q_H
[H^q_{t-1}, Z^q_t, \tilde{\Delta}^q_t, TW^q_t]
+
b^q_H
\right)
$$

The updated recurrent state is then conceptually represented as:

$$
H^q_t =
K^q_t \odot H^q_{t-1}
+
(1-K^q_t) \odot \hat{H}^q_t
$$

The exact parameterisation remains trainable.

The retention gate therefore controls the balance between historical persistence and current-state updating.

---

### Transformative Weight in recurrence

The three Transformative Weights enter the recurrent mechanism independently:

$$
TW^F_t \rightarrow \text{factual recurrence}
$$

$$
TW^P_t \rightarrow \text{psychological recurrence}
$$

$$
TW^S_t \rightarrow \text{social recurrence}
$$

They are not averaged or collapsed into a universal recurrent importance score.

Transformative Weight provides information about the learned transformative salience of the current textual step within its corresponding representation space.

It does not directly determine recurrent persistence.

A high Transformative Weight does not automatically erase the preceding state.

Likewise, a low Transformative Weight does not automatically prevent updating.

The recurrent mechanism learns how Transformative Weight interacts with the current representation, transformation and preceding recurrent state.

---

### Transformation and recurrence

Transformation and recurrence remain distinct architectural operations.

The transformative mechanism asks:

**What representational change is associated with the current textual step?**

The recurrent mechanism asks:

**What representational information remains active after the current textual step?**

Consequently:

$$
\tilde{\Delta}^q_t \neq H^q_t
$$

and:

$$
TW^q_t \neq K^q_t
$$

Transformative Weight therefore does not serve as the recurrent retention gate.

This separation allows a sentence to produce a strong transformation without necessarily eliminating earlier context.

It also allows earlier information to remain recurrently influential even when the current sentence has relatively low transformative salience.

---

### Persistence across multiple sentences

Recurrent state may persist across more than one subsequent sentence.

For example, information introduced at sentence \(t\) may remain represented in:

$$
H^q_{t+1},\ H^q_{t+2},\ \ldots,\ H^q_{t+k}
$$

if the learned recurrent mechanism continues to retain it.

This property is essential for modelling discourse effects that develop over several textual steps.

An emotional association, factual premise, actor relationship or group framing may therefore continue to influence interpretation even when it is not explicitly repeated in every subsequent sentence.

---

### Reinforcement

Repeated or compatible textual signals may reinforce an existing recurrent state.

If the current representation and transformation are compatible with information already preserved in \(H^q_{t-1}\), the recurrent mechanism may strengthen the corresponding state.

Reinforcement is learned rather than assumed.

Repeated textual material does not automatically imply increased importance.

The recurrent mechanism must distinguish meaningful reinforcement from repetition that adds little new representational information.

---

### Attenuation

Previously active information may weaken over the sentence sequence.

Attenuation may occur when:

- subsequent text no longer supports the previous state;
- competing information becomes more salient;
- the learned retention gate reduces persistence;
- the discourse moves to a different representational structure.

Attenuation is therefore learned rather than imposed through a fixed decay constant.

The architecture does not assume that all recurrent information decays at the same rate.

---

### Contradiction and state revision

New textual information may conflict with information represented in the preceding recurrent state.

The recurrent mechanism must permit such information to modify the state rather than simply accumulating contradictory representations indefinitely.

A contradiction may therefore produce:

- partial state revision;
- strong state revision;
- coexistence of competing information where appropriate;
- or attenuation of the preceding state.

The recurrent mechanism itself does not determine factual truth.

Contradictory factual claims may remain represented as competing textual states until external factual verification is performed by a separate process.

---

### Initial recurrent state

At the beginning of an article, no preceding article-specific recurrent state exists.

The initial recurrent states are therefore defined as neutral initial states:

$$
H^F_0,\qquad H^P_0,\qquad H^S_0
$$

These initial states must not encode substantive factual, psychological or social conclusions about the article.

They may be represented by learned or deterministic neutral initialisation, provided that the initialisation does not introduce article-specific evidence.

The choice between learned and deterministic neutral initialisation remains an experimental parameter.

---

### Article boundaries

Recurrent state is reset at article boundaries.

The final recurrent state of one article must not become the initial recurrent state of an unrelated article.

Therefore:

$$
H^q_0(\text{article } j+1)
\neq
H^q_T(\text{article } j)
$$

unless a later experiment explicitly defines a cross-document sequential context.

Architecture version 1.0 does not permit cross-article recurrent carry-over.

This prevents representational leakage between independent articles.

---

### Global and group-conditioned recurrence

The global recurrent path is always preserved.

Where validated group-conditioned routes exist, additional group-conditioned recurrent states may be maintained.

For group \(g\):

$$
H^q_{t,g}
=
R^q_g
\left(
H^q_{t-1,g},
Z^q_{t,g},
\tilde{\Delta}^q_{t,g},
TW^q_{t,g}
\right)
$$

The group-conditioned recurrent state represents the sequential state associated with that group-conditioned path.

It does not replace the global recurrent state.

Concrete group identities must originate from validated group detection and alignment rather than being predefined by the recurrent mechanism.

---

### Recurrent group deviation

Where global and group-conditioned recurrent states are aligned and comparable, signed recurrent deviation may later be defined.

For a fixed-dimensional representation space:

$$
D^{q,R}_{t,g}
=
H^q_{t,g}
-
H^q_t
$$

The deviation describes the difference between group-conditioned and global recurrent states.

It is not automatically interpreted as:

- bias;
- manipulation;
- polarisation.

For social recurrence, deviation requires structural object alignment before meaningful comparison.

Recurrent group deviation is therefore a descriptive intermediate quantity rather than a normative conclusion.

---

### Masking and unavailable information

Missing information remains distinct from substantive numerical zero.

Unavailable supervision must not be converted into negative evidence.

Where recurrent updating depends on a representation component that is unavailable, the architecture must preserve the corresponding mask or availability state.

The recurrent mechanism must therefore distinguish:

- observed zero;
- unavailable value;
- structurally absent object;
- explicitly negative observation.

This distinction is particularly important for factual and social representations.

---

### Causal constraint

The recurrent mechanism is strictly causal.

For sentence \(t\), recurrent updating may depend only on information available at or before sentence \(t\).

Conceptually:

$$
H^q_t = R^q(X_{\leq t})
$$

and not:

$$
H^q_t = R^q(X_{>t})
$$

Future sentences must therefore remain inaccessible during the current recurrent update.

This preserves temporal and methodological validity for subsequent attribution and evaluation.

---

### Training contract

The recurrent mechanism is trainable.

Separate recurrent parameterisations are preserved for factual, psychological and social state spaces.

The architecture does not require the three recurrent spaces to share parameters.

Where supervised recurrent targets are available, fixed-dimensional recurrent states may use mask-aware regression objectives.

Structured social recurrence requires an appropriate structured objective and explicit object alignment.

Training must preserve article boundaries and sentence order.

Sentence sequences must therefore not be randomly reordered within an article during recurrent training.

Unavailable supervision does not contribute to optimisation.

Global and group-conditioned recurrent losses remain independently controllable.

---

### Relationship to hierarchical aggregation

The recurrent mechanism operates across the ordered sentence sequence.

It does not itself perform paragraph-level or article-level hierarchical aggregation.

The recurrent state describes sequential persistence.

Hierarchical aggregation describes how representations are combined across textual levels.

These operations remain separate.

A later hierarchical mechanism may use recurrent outputs when constructing paragraph-level and article-level representations.

---

### Relationship to backward attribution

The recurrent mechanism creates a sequential dependency structure that can later support backward attribution.

If a later sentence-level state depends strongly on information introduced earlier in the article, attribution mechanisms may subsequently trace that influence backwards through the recurrent sequence.

Backward attribution is not calculated in this block.

The recurrent mechanism only preserves the computational dependencies required for later attribution analysis.

---

### Methodological boundaries

The recurrent mechanism does not:

- verify factual truth;
- equate recurrent factual persistence with factual correctness;
- collapse factual, psychological and social recurrent states;
- collapse TW_F, TW_P and TW_S;
- equate Transformative Weight with recurrent retention;
- equate transformation with recurrent state;
- assume that repetition automatically means reinforcement;
- impose a universal fixed decay rate;
- automatically interpret group deviation as bias;
- automatically interpret group deviation as manipulation;
- automatically interpret group deviation as polarisation;
- fabricate group identities;
- treat missing information as numerical zero;
- use future textual information;
- carry recurrent state across independent articles;
- perform hierarchical aggregation;
- calculate backward attribution.

These operations remain conceptually and computationally separate.

---

### Execution boundary

This block defines the trainable recurrent architecture only.

At this stage:

- factual recurrence is specified;
- psychological recurrence is specified;
- structured social recurrence is specified;
- recurrent retention and update are specified;
- TW_F, TW_P and TW_S are routed independently;
- global and group-conditioned recurrent paths are specified;
- article-boundary reset is specified;
- causal sequential processing is specified;
- masking requirements are specified;
- the training contract is specified.

However:

- no recurrent model has been instantiated;
- no initial recurrent state has been generated;
- no recurrent forward pass has been executed;
- no factual recurrent state has been calculated;
- no psychological recurrent state has been calculated;
- no social recurrent state has been calculated;
- no recurrent retention gate has been calculated;
- no recurrent group deviation has been calculated;
- no cross-sentence recurrent propagation has been executed;
- no hierarchical aggregation has been performed;
- no backward attribution has been calculated;
- no training has been performed;
- no evaluation has been performed.

Block 16 therefore establishes the recurrent mechanism as a trainable causal architectural contract while preserving the separation between current representation, transformation, transformative salience, sequential persistence, hierarchical aggregation and downstream interpretation.

````python
# =============================================================================
# Media AI — Block 16: Recurrent Mechanism
# =============================================================================

import torch
import torch.nn as nn
import torch.nn.functional as F


# -----------------------------------------------------------------------------
# Block dependency checks
# -----------------------------------------------------------------------------

required_block_12_objects = [
    "FACTUAL_HEAD_ARCHITECTURE_VALID",
]

missing_block_12_objects = [
    object_name
    for object_name in required_block_12_objects
    if object_name not in globals()
]

if missing_block_12_objects:
    raise NameError(
        "Notebook 04 Block 12 is not initialised. "
        f"Missing: {missing_block_12_objects}"
    )


required_block_13_objects = [
    "PSYCHOLOGICAL_HEAD_ARCHITECTURE_VALID",
    "PSYCHOLOGICAL_STATE_DIMENSION_COUNT",
]

missing_block_13_objects = [
    object_name
    for object_name in required_block_13_objects
    if object_name not in globals()
]

if missing_block_13_objects:
    raise NameError(
        "Notebook 04 Block 13 is not initialised. "
        f"Missing: {missing_block_13_objects}"
    )


required_block_14_objects = [
    "SOCIAL_HEAD_ARCHITECTURE_VALID",
    "SOCIAL_REPRESENTATION_COMPONENTS",
]

missing_block_14_objects = [
    object_name
    for object_name in required_block_14_objects
    if object_name not in globals()
]

if missing_block_14_objects:
    raise NameError(
        "Notebook 04 Block 14 is not initialised. "
        f"Missing: {missing_block_14_objects}"
    )


required_block_15_objects = [
    "TRANSFORMATIVE_ARCHITECTURE_VALID",
    "TRANSFORMATIVE_WEIGHT_SPACES",
    "TRANSFORMATIVE_WEIGHT_FACTUAL_NAME",
    "TRANSFORMATIVE_WEIGHT_PSYCHOLOGICAL_NAME",
    "TRANSFORMATIVE_WEIGHT_SOCIAL_NAME",
]

missing_block_15_objects = [
    object_name
    for object_name in required_block_15_objects
    if object_name not in globals()
]

if missing_block_15_objects:
    raise NameError(
        "Notebook 04 Block 15 is not initialised. "
        f"Missing: {missing_block_15_objects}"
    )


if not FACTUAL_HEAD_ARCHITECTURE_VALID:
    raise ValueError(
        "Recurrent mechanism requires a valid factual-head architecture."
    )


if not PSYCHOLOGICAL_HEAD_ARCHITECTURE_VALID:
    raise ValueError(
        "Recurrent mechanism requires a valid psychological-head architecture."
    )


if not SOCIAL_HEAD_ARCHITECTURE_VALID:
    raise ValueError(
        "Recurrent mechanism requires a valid social-head architecture."
    )


if not TRANSFORMATIVE_ARCHITECTURE_VALID:
    raise ValueError(
        "Recurrent mechanism requires a valid transformative architecture."
    )


# -----------------------------------------------------------------------------
# Recurrent-mechanism configuration
# -----------------------------------------------------------------------------

RECURRENT_MECHANISM_VERSION = "1.0"

RECURRENT_MECHANISM_LEVEL = "sentence"

RECURRENT_MECHANISM_POSITION = "post_transformative"

RECURRENT_STATE_SPACES = (
    "factual",
    "psychological",
    "social",
)

RECURRENT_SEQUENCE_UNIT = "sentence"

RECURRENT_SEQUENCE_ORDER_REQUIRED = True

RECURRENT_ARTICLE_BOUNDARY_RESET = True

RECURRENT_CROSS_ARTICLE_CARRY = False


# -----------------------------------------------------------------------------
# Space-specific recurrent-state names
# -----------------------------------------------------------------------------

RECURRENT_FACTUAL_STATE_NAME = "H_F"

RECURRENT_PSYCHOLOGICAL_STATE_NAME = "H_P"

RECURRENT_SOCIAL_STATE_NAME = "H_S"

RECURRENT_STATES_SHARED_ACROSS_SPACES = False

RECURRENT_STATES_COLLAPSED = False


# -----------------------------------------------------------------------------
# Transformative Weight routing
# -----------------------------------------------------------------------------

RECURRENT_TW_FACTUAL = TRANSFORMATIVE_WEIGHT_FACTUAL_NAME

RECURRENT_TW_PSYCHOLOGICAL = (
    TRANSFORMATIVE_WEIGHT_PSYCHOLOGICAL_NAME
)

RECURRENT_TW_SOCIAL = TRANSFORMATIVE_WEIGHT_SOCIAL_NAME

RECURRENT_TW_COLLAPSED = False

RECURRENT_TW_IS_RETENTION_GATE = False


# -----------------------------------------------------------------------------
# Architectural policy
# -----------------------------------------------------------------------------

RECURRENT_GLOBAL_PATH_REQUIRED = True

RECURRENT_GROUP_PATH_OPTIONAL = True

RECURRENT_GROUP_ADAPTATION = True

RECURRENT_CAUSAL = True

RECURRENT_FUTURE_CONTEXT_ALLOWED = False

RECURRENT_FIXED_DECAY = False

RECURRENT_REPETITION_EQUALS_REINFORCEMENT = False

RECURRENT_CONTRADICTION_REVISION_ALLOWED = True

RECURRENT_MASK_AWARE = True

RECURRENT_SOCIAL_ALIGNMENT_REQUIRED = True

RECURRENT_GROUP_DEVIATION_SIGNED = True

RECURRENT_GROUP_DEVIATION_IS_BIAS = False

RECURRENT_GROUP_DEVIATION_IS_MANIPULATION = False

RECURRENT_GROUP_DEVIATION_IS_POLARISATION = False

RECURRENT_HIERARCHICAL_AGGREGATION = False

RECURRENT_BACKWARD_ATTRIBUTION_CALCULATED = False


# -----------------------------------------------------------------------------
# Initial-state policy
# -----------------------------------------------------------------------------

RECURRENT_INITIAL_STATE_POLICY = "neutral_zero_default"

RECURRENT_LEARNED_INITIAL_STATE_ALLOWED = True

RECURRENT_INITIAL_STATE_ARTICLE_SPECIFIC = False


# -----------------------------------------------------------------------------
# Training contract
# -----------------------------------------------------------------------------

RECURRENT_FACTUAL_LOSS_NAME = "huber"

RECURRENT_PSYCHOLOGICAL_LOSS_NAME = "huber"

RECURRENT_SOCIAL_LOSS_NAME = "structured"

RECURRENT_FACTUAL_LOSS_COEFFICIENT = 1.0

RECURRENT_PSYCHOLOGICAL_LOSS_COEFFICIENT = 1.0

RECURRENT_SOCIAL_LOSS_COEFFICIENT = 1.0

RECURRENT_GLOBAL_LOSS_COEFFICIENT = 1.0

RECURRENT_GROUP_LOSS_COEFFICIENT = 1.0


# -----------------------------------------------------------------------------
# Fixed-dimensional recurrent cell
# -----------------------------------------------------------------------------

class FixedDimensionalRecurrentCell(nn.Module):
    """
    Trainable recurrent cell for a fixed-dimensional representation space.

    The cell receives:
        - preceding recurrent state;
        - current representation;
        - current gated transformation;
        - current space-specific Transformative Weight.

    Transformative Weight is an input to the recurrent update.
    It is not itself the retention gate.
    """

    def __init__(
        self,
        representation_dim,
        transformation_dim,
        recurrent_dim,
        space_name,
        dropout=0.1,
    ):
        super().__init__()


        if space_name not in {
            "factual",
            "psychological",
        }:
            raise ValueError(
                "Fixed-dimensional recurrent space must be "
                "factual or psychological."
            )


        if representation_dim <= 0:
            raise ValueError(
                "representation_dim must be greater than zero."
            )


        if transformation_dim <= 0:
            raise ValueError(
                "transformation_dim must be greater than zero."
            )


        if recurrent_dim <= 0:
            raise ValueError(
                "recurrent_dim must be greater than zero."
            )


        if not (
            0.0 <= float(dropout) < 1.0
        ):
            raise ValueError(
                "dropout must lie in [0,1)."
            )


        self.space_name = str(
            space_name
        )

        self.representation_dim = int(
            representation_dim
        )

        self.transformation_dim = int(
            transformation_dim
        )

        self.recurrent_dim = int(
            recurrent_dim
        )


        self.input_dim = (
            self.recurrent_dim
            + self.representation_dim
            + self.transformation_dim
            + 1
        )


        self.retention_layer = nn.Linear(
            self.input_dim,
            self.recurrent_dim,
        )


        self.candidate_network = nn.Sequential(
            nn.Linear(
                self.input_dim,
                self.recurrent_dim,
            ),

            nn.GELU(),

            nn.Dropout(
                float(dropout)
            ),

            nn.LayerNorm(
                self.recurrent_dim
            ),

            nn.Linear(
                self.recurrent_dim,
                self.recurrent_dim,
            ),

            nn.Tanh(),
        )


    def initial_state(
        self,
        batch_shape,
        device=None,
        dtype=None,
    ):
        """
        Generate a neutral deterministic recurrent state.

        No article-specific information is introduced.
        """

        return torch.zeros(
            *batch_shape,
            self.recurrent_dim,
            device=device,
            dtype=dtype,
        )


    def forward(
        self,
        previous_state,
        current_representation,
        gated_transformation,
        transformative_weight,
    ):

        if previous_state.shape[-1] != self.recurrent_dim:
            raise ValueError(
                "Previous recurrent state has incompatible "
                "dimensionality."
            )


        if (
            current_representation.shape[-1]
            != self.representation_dim
        ):
            raise ValueError(
                "Current representation has incompatible "
                "dimensionality."
            )


        if (
            gated_transformation.shape[-1]
            != self.transformation_dim
        ):
            raise ValueError(
                "Gated transformation has incompatible "
                "dimensionality."
            )


        if transformative_weight.shape[-1] != 1:
            raise ValueError(
                "Transformative Weight must be scalar-valued "
                "on the final dimension."
            )


        leading_shapes = {
            previous_state.shape[:-1],
            current_representation.shape[:-1],
            gated_transformation.shape[:-1],
            transformative_weight.shape[:-1],
        }


        if len(leading_shapes) != 1:
            raise ValueError(
                "All recurrent inputs must have compatible "
                "leading dimensions."
            )


        recurrent_input = torch.cat(
            [
                previous_state,
                current_representation,
                gated_transformation,
                transformative_weight,
            ],
            dim=-1,
        )


        retention_gate = torch.sigmoid(
            self.retention_layer(
                recurrent_input
            )
        )


        candidate_state = self.candidate_network(
            recurrent_input
        )


        current_state = (
            retention_gate
            * previous_state
            +
            (
                1.0
                - retention_gate
            )
            * candidate_state
        )


        return {
            "space":
                self.space_name,

            "previous_state":
                previous_state,

            "retention_gate":
                retention_gate,

            "candidate_state":
                candidate_state,

            "current_state":
                current_state,
        }


# -----------------------------------------------------------------------------
# Group-conditioned fixed-dimensional recurrent adapter
# -----------------------------------------------------------------------------

class RecurrentGroupAdapter(nn.Module):
    """
    Residual group-conditioned adaptation for a recurrent state.

    The global recurrent path remains independently preserved.
    """

    def __init__(
        self,
        recurrent_dim,
        group_embedding_dim,
        hidden_dim,
        dropout=0.1,
    ):
        super().__init__()


        if recurrent_dim <= 0:
            raise ValueError(
                "recurrent_dim must be greater than zero."
            )


        if group_embedding_dim <= 0:
            raise ValueError(
                "group_embedding_dim must be greater than zero."
            )


        if hidden_dim <= 0:
            raise ValueError(
                "hidden_dim must be greater than zero."
            )


        self.recurrent_dim = int(
            recurrent_dim
        )

        self.group_embedding_dim = int(
            group_embedding_dim
        )


        self.network = nn.Sequential(
            nn.Linear(
                self.recurrent_dim
                + self.group_embedding_dim,
                hidden_dim,
            ),

            nn.GELU(),

            nn.Dropout(
                float(dropout)
            ),

            nn.Linear(
                hidden_dim,
                self.recurrent_dim,
            ),
        )


    def forward(
        self,
        recurrent_state,
        group_embedding,
    ):

        if recurrent_state.shape[-1] != self.recurrent_dim:
            raise ValueError(
                "Recurrent state has incompatible dimensionality."
            )


        if (
            group_embedding.shape[-1]
            != self.group_embedding_dim
        ):
            raise ValueError(
                "Group embedding has incompatible dimensionality."
            )


        if (
            recurrent_state.shape[:-1]
            != group_embedding.shape[:-1]
        ):
            raise ValueError(
                "Recurrent state and group embedding must have "
                "compatible leading dimensions."
            )


        adapter_input = torch.cat(
            [
                recurrent_state,
                group_embedding,
            ],
            dim=-1,
        )


        return self.network(
            adapter_input
        )


# -----------------------------------------------------------------------------
# Fixed-dimensional recurrent path
# -----------------------------------------------------------------------------

class FixedDimensionalRecurrentPath(nn.Module):
    """
    Complete factual or psychological recurrent path.

    Global and optional group-conditioned recurrence remain separate.
    """

    def __init__(
        self,
        representation_dim,
        transformation_dim,
        recurrent_dim,
        space_name,
        group_embedding_dim=None,
        dropout=0.1,
    ):
        super().__init__()


        self.space_name = str(
            space_name
        )

        self.recurrent_dim = int(
            recurrent_dim
        )

        self.group_embedding_dim = (
            int(group_embedding_dim)
            if group_embedding_dim is not None
            else None
        )


        self.recurrent_cell = (
            FixedDimensionalRecurrentCell(
                representation_dim=
                    representation_dim,

                transformation_dim=
                    transformation_dim,

                recurrent_dim=
                    recurrent_dim,

                space_name=
                    space_name,

                dropout=
                    dropout,
            )
        )


        if self.group_embedding_dim is not None:

            self.group_adapter = RecurrentGroupAdapter(
                recurrent_dim=
                    recurrent_dim,

                group_embedding_dim=
                    self.group_embedding_dim,

                hidden_dim=
                    recurrent_dim,

                dropout=
                    dropout,
            )

        else:

            self.group_adapter = None


    def initial_state(
        self,
        batch_shape,
        device=None,
        dtype=None,
    ):

        return self.recurrent_cell.initial_state(
            batch_shape=
                batch_shape,

            device=
                device,

            dtype=
                dtype,
        )


    def forward_global(
        self,
        previous_state,
        current_representation,
        gated_transformation,
        transformative_weight,
    ):

        output = self.recurrent_cell(
            previous_state=
                previous_state,

            current_representation=
                current_representation,

            gated_transformation=
                gated_transformation,

            transformative_weight=
                transformative_weight,
        )


        output[
            "scope"
        ] = "global"


        return output


    def forward_group(
        self,
        previous_state,
        current_representation,
        gated_transformation,
        transformative_weight,
        group_embedding,
    ):

        if self.group_adapter is None:
            raise ValueError(
                "Group-conditioned recurrence requires a configured "
                "group_embedding_dim."
            )


        recurrent_output = self.recurrent_cell(
            previous_state=
                previous_state,

            current_representation=
                current_representation,

            gated_transformation=
                gated_transformation,

            transformative_weight=
                transformative_weight,
        )


        group_adjustment = self.group_adapter(
            recurrent_state=
                recurrent_output[
                    "current_state"
                ],

            group_embedding=
                group_embedding,
        )


        group_state = (
            recurrent_output[
                "current_state"
            ]
            + group_adjustment
        )


        recurrent_output[
            "scope"
        ] = "group_conditioned"

        recurrent_output[
            "shared_current_state"
        ] = recurrent_output[
            "current_state"
        ]

        recurrent_output[
            "group_adjustment"
        ] = group_adjustment

        recurrent_output[
            "current_state"
        ] = group_state


        return recurrent_output


# -----------------------------------------------------------------------------
# Structured social recurrent path
# -----------------------------------------------------------------------------

class StructuredSocialRecurrentPath(nn.Module):
    """
    Recurrent mechanism for the variable-dimensional social space.

    The social recurrent state is represented through a trainable hidden
    state while structured actors, groups, memberships and relationships
    remain subject to explicit alignment and later structured decoding.

    TW_S enters the social recurrent update independently.
    """

    def __init__(
        self,
        social_hidden_dim,
        transformation_hidden_dim,
        recurrent_dim,
        group_embedding_dim=None,
        dropout=0.1,
    ):
        super().__init__()


        if social_hidden_dim <= 0:
            raise ValueError(
                "social_hidden_dim must be greater than zero."
            )


        if transformation_hidden_dim <= 0:
            raise ValueError(
                "transformation_hidden_dim must be greater than zero."
            )


        if recurrent_dim <= 0:
            raise ValueError(
                "recurrent_dim must be greater than zero."
            )


        self.social_hidden_dim = int(
            social_hidden_dim
        )

        self.transformation_hidden_dim = int(
            transformation_hidden_dim
        )

        self.recurrent_dim = int(
            recurrent_dim
        )

        self.group_embedding_dim = (
            int(group_embedding_dim)
            if group_embedding_dim is not None
            else None
        )


        self.input_dim = (
            self.recurrent_dim
            + self.social_hidden_dim
            + self.transformation_hidden_dim
            + 1
        )


        self.retention_layer = nn.Linear(
            self.input_dim,
            self.recurrent_dim,
        )


        self.candidate_network = nn.Sequential(
            nn.Linear(
                self.input_dim,
                self.recurrent_dim,
            ),

            nn.GELU(),

            nn.Dropout(
                float(dropout)
            ),

            nn.LayerNorm(
                self.recurrent_dim
            ),

            nn.Linear(
                self.recurrent_dim,
                self.recurrent_dim,
            ),

            nn.Tanh(),
        )


        if self.group_embedding_dim is not None:

            self.group_adapter = RecurrentGroupAdapter(
                recurrent_dim=
                    self.recurrent_dim,

                group_embedding_dim=
                    self.group_embedding_dim,

                hidden_dim=
                    self.recurrent_dim,

                dropout=
                    dropout,
            )

        else:

            self.group_adapter = None


    def initial_state(
        self,
        batch_shape,
        device=None,
        dtype=None,
    ):

        return torch.zeros(
            *batch_shape,
            self.recurrent_dim,
            device=device,
            dtype=dtype,
        )


    def _update(
        self,
        previous_state,
        social_hidden_representation,
        social_hidden_transformation,
        transformative_weight,
    ):

        if previous_state.shape[-1] != self.recurrent_dim:
            raise ValueError(
                "Previous social recurrent state has incompatible "
                "dimensionality."
            )


        if (
            social_hidden_representation.shape[-1]
            != self.social_hidden_dim
        ):
            raise ValueError(
                "Social hidden representation has incompatible "
                "dimensionality."
            )


        if (
            social_hidden_transformation.shape[-1]
            != self.transformation_hidden_dim
        ):
            raise ValueError(
                "Social hidden transformation has incompatible "
                "dimensionality."
            )


        if transformative_weight.shape[-1] != 1:
            raise ValueError(
                "TW_S must be scalar-valued on the final dimension."
            )


        leading_shapes = {
            previous_state.shape[:-1],
            social_hidden_representation.shape[:-1],
            social_hidden_transformation.shape[:-1],
            transformative_weight.shape[:-1],
        }


        if len(leading_shapes) != 1:
            raise ValueError(
                "All social recurrent inputs must have compatible "
                "leading dimensions."
            )


        recurrent_input = torch.cat(
            [
                previous_state,
                social_hidden_representation,
                social_hidden_transformation,
                transformative_weight,
            ],
            dim=-1,
        )


        retention_gate = torch.sigmoid(
            self.retention_layer(
                recurrent_input
            )
        )


        candidate_state = self.candidate_network(
            recurrent_input
        )


        current_state = (
            retention_gate
            * previous_state
            +
            (
                1.0
                - retention_gate
            )
            * candidate_state
        )


        return {
            "space":
                "social",

            "previous_state":
                previous_state,

            "retention_gate":
                retention_gate,

            "candidate_state":
                candidate_state,

            "current_state":
                current_state,

            "structured_state":
                None,

            "structured_decoding_executed":
                False,
        }


    def forward_global(
        self,
        previous_state,
        social_hidden_representation,
        social_hidden_transformation,
        transformative_weight,
    ):

        output = self._update(
            previous_state=
                previous_state,

            social_hidden_representation=
                social_hidden_representation,

            social_hidden_transformation=
                social_hidden_transformation,

            transformative_weight=
                transformative_weight,
        )


        output[
            "scope"
        ] = "global"


        return output


    def forward_group(
        self,
        previous_state,
        social_hidden_representation,
        social_hidden_transformation,
        transformative_weight,
        group_embedding,
    ):

        if self.group_adapter is None:
            raise ValueError(
                "Group-conditioned social recurrence requires "
                "a configured group_embedding_dim."
            )


        output = self._update(
            previous_state=
                previous_state,

            social_hidden_representation=
                social_hidden_representation,

            social_hidden_transformation=
                social_hidden_transformation,

            transformative_weight=
                transformative_weight,
        )


        group_adjustment = self.group_adapter(
            recurrent_state=
                output[
                    "current_state"
                ],

            group_embedding=
                group_embedding,
        )


        group_state = (
            output[
                "current_state"
            ]
            + group_adjustment
        )


        output[
            "scope"
        ] = "group_conditioned"

        output[
            "shared_current_state"
        ] = output[
            "current_state"
        ]

        output[
            "group_adjustment"
        ] = group_adjustment

        output[
            "current_state"
        ] = group_state


        return output


# -----------------------------------------------------------------------------
# Complete Media AI recurrent mechanism
# -----------------------------------------------------------------------------

class MediaAIRecurrentMechanism(nn.Module):
    """
    Complete Media AI recurrent mechanism.

    The mechanism preserves independent:
        - factual recurrent state H_F;
        - psychological recurrent state H_P;
        - social recurrent state H_S.

    TW_F, TW_P and TW_S enter their corresponding recurrent paths
    independently.

    No universal recurrent state or Transformative Weight is created.
    """

    def __init__(
        self,
        factual_dim,
        psychological_dim,
        social_hidden_dim,
        factual_recurrent_dim,
        psychological_recurrent_dim,
        social_recurrent_dim,
        social_transformation_hidden_dim,
        group_embedding_dim=None,
        dropout=0.1,
    ):
        super().__init__()


        if factual_dim <= 0:
            raise ValueError(
                "factual_dim must be greater than zero."
            )


        if psychological_dim <= 0:
            raise ValueError(
                "psychological_dim must be greater than zero."
            )


        if social_hidden_dim <= 0:
            raise ValueError(
                "social_hidden_dim must be greater than zero."
            )


        self.factual_path = FixedDimensionalRecurrentPath(
            representation_dim=
                factual_dim,

            transformation_dim=
                factual_dim,

            recurrent_dim=
                factual_recurrent_dim,

            space_name=
                "factual",

            group_embedding_dim=
                group_embedding_dim,

            dropout=
                dropout,
        )


        self.psychological_path = FixedDimensionalRecurrentPath(
            representation_dim=
                psychological_dim,

            transformation_dim=
                psychological_dim,

            recurrent_dim=
                psychological_recurrent_dim,

            space_name=
                "psychological",

            group_embedding_dim=
                group_embedding_dim,

            dropout=
                dropout,
        )


        self.social_path = StructuredSocialRecurrentPath(
            social_hidden_dim=
                social_hidden_dim,

            transformation_hidden_dim=
                social_transformation_hidden_dim,

            recurrent_dim=
                social_recurrent_dim,

            group_embedding_dim=
                group_embedding_dim,

            dropout=
                dropout,
        )


    def initial_states(
        self,
        batch_shape,
        device=None,
        dtype=None,
    ):
        """
        Generate neutral article-initial recurrent states.
        """

        return {
            "H_F":
                self.factual_path.initial_state(
                    batch_shape=
                        batch_shape,

                    device=
                        device,

                    dtype=
                        dtype,
                ),

            "H_P":
                self.psychological_path.initial_state(
                    batch_shape=
                        batch_shape,

                    device=
                        device,

                    dtype=
                        dtype,
                ),

            "H_S":
                self.social_path.initial_state(
                    batch_shape=
                        batch_shape,

                    device=
                        device,

                    dtype=
                        dtype,
                ),
        }


    def forward_global(
        self,
        previous_states,
        factual_representation,
        psychological_representation,
        social_hidden_representation,
        factual_transformation,
        psychological_transformation,
        social_hidden_transformation,
        tw_f,
        tw_p,
        tw_s,
    ):
        """
        Execute one global recurrent sentence update.
        """

        factual_output = self.factual_path.forward_global(
            previous_state=
                previous_states[
                    "H_F"
                ],

            current_representation=
                factual_representation,

            gated_transformation=
                factual_transformation,

            transformative_weight=
                tw_f,
        )


        psychological_output = (
            self.psychological_path.forward_global(
                previous_state=
                    previous_states[
                        "H_P"
                    ],

                current_representation=
                    psychological_representation,

                gated_transformation=
                    psychological_transformation,

                transformative_weight=
                    tw_p,
            )
        )


        social_output = self.social_path.forward_global(
            previous_state=
                previous_states[
                    "H_S"
                ],

            social_hidden_representation=
                social_hidden_representation,

            social_hidden_transformation=
                social_hidden_transformation,

            transformative_weight=
                tw_s,
        )


        return {
            "scope":
                "global",

            "factual":
                factual_output,

            "psychological":
                psychological_output,

            "social":
                social_output,

            "states": {
                "H_F":
                    factual_output[
                        "current_state"
                    ],

                "H_P":
                    psychological_output[
                        "current_state"
                    ],

                "H_S":
                    social_output[
                        "current_state"
                    ],
            },
        }


    def forward_group(
        self,
        previous_states,
        factual_representation,
        psychological_representation,
        social_hidden_representation,
        factual_transformation,
        psychological_transformation,
        social_hidden_transformation,
        tw_f,
        tw_p,
        tw_s,
        group_embedding,
    ):
        """
        Execute one group-conditioned recurrent sentence update.
        """

        factual_output = self.factual_path.forward_group(
            previous_state=
                previous_states[
                    "H_F"
                ],

            current_representation=
                factual_representation,

            gated_transformation=
                factual_transformation,

            transformative_weight=
                tw_f,

            group_embedding=
                group_embedding,
        )


        psychological_output = (
            self.psychological_path.forward_group(
                previous_state=
                    previous_states[
                        "H_P"
                    ],

                current_representation=
                    psychological_representation,

                gated_transformation=
                    psychological_transformation,

                transformative_weight=
                    tw_p,

                group_embedding=
                    group_embedding,
            )
        )


        social_output = self.social_path.forward_group(
            previous_state=
                previous_states[
                    "H_S"
                ],

            social_hidden_representation=
                social_hidden_representation,

            social_hidden_transformation=
                social_hidden_transformation,

            transformative_weight=
                tw_s,

            group_embedding=
                group_embedding,
        )


        return {
            "scope":
                "group_conditioned",

            "factual":
                factual_output,

            "psychological":
                psychological_output,

            "social":
                social_output,

            "states": {
                "H_F":
                    factual_output[
                        "current_state"
                    ],

                "H_P":
                    psychological_output[
                        "current_state"
                    ],

                "H_S":
                    social_output[
                        "current_state"
                    ],
            },
        }


# -----------------------------------------------------------------------------
# Fixed-dimensional recurrent group deviation
# -----------------------------------------------------------------------------

def calculate_recurrent_group_deviation(
    global_state,
    group_state,
):
    """
    Calculate signed group-conditioned recurrent deviation.

    This is descriptive and is not automatically interpreted as bias,
    manipulation or polarisation.
    """

    if global_state.shape != group_state.shape:
        raise ValueError(
            "Global and group recurrent states must have "
            "identical shapes."
        )


    return (
        group_state
        - global_state
    )


# -----------------------------------------------------------------------------
# Generic masked recurrent regression loss
# -----------------------------------------------------------------------------

def masked_recurrent_regression_loss(
    predictions,
    targets,
    mask,
    loss_name="huber",
):
    """
    Calculate mask-aware regression loss for fixed-dimensional
    recurrent supervision.
    """

    mask = mask.bool()


    if predictions.shape != targets.shape:
        raise ValueError(
            "Recurrent predictions and targets must have "
            "identical shapes."
        )


    if predictions.shape != mask.shape:
        raise ValueError(
            "Recurrent predictions and mask must have "
            "identical shapes."
        )


    if not torch.any(
        mask
    ):
        return None


    selected_predictions = predictions[
        mask
    ]

    selected_targets = targets[
        mask
    ]


    if loss_name == "huber":

        return F.smooth_l1_loss(
            selected_predictions,
            selected_targets,
        )


    if loss_name == "mse":

        return F.mse_loss(
            selected_predictions,
            selected_targets,
        )


    if loss_name == "mae":

        return F.l1_loss(
            selected_predictions,
            selected_targets,
        )


    raise ValueError(
        "Unsupported recurrent regression loss: "
        f"{loss_name}"
    )


# -----------------------------------------------------------------------------
# Combined recurrent loss
# -----------------------------------------------------------------------------

def combine_recurrent_losses(
    factual_loss=None,
    psychological_loss=None,
    social_loss=None,
    factual_coefficient=RECURRENT_FACTUAL_LOSS_COEFFICIENT,
    psychological_coefficient=RECURRENT_PSYCHOLOGICAL_LOSS_COEFFICIENT,
    social_coefficient=RECURRENT_SOCIAL_LOSS_COEFFICIENT,
):
    """
    Combine only available recurrent loss terms.
    """

    available_terms = []


    if factual_loss is not None:

        available_terms.append(
            float(
                factual_coefficient
            )
            * factual_loss
        )


    if psychological_loss is not None:

        available_terms.append(
            float(
                psychological_coefficient
            )
            * psychological_loss
        )


    if social_loss is not None:

        available_terms.append(
            float(
                social_coefficient
            )
            * social_loss
        )


    if not available_terms:
        return None


    return sum(
        available_terms
    )


# -----------------------------------------------------------------------------
# Global and group recurrent loss combination
# -----------------------------------------------------------------------------

def combine_global_and_group_recurrent_losses(
    global_loss,
    group_loss=None,
    global_coefficient=RECURRENT_GLOBAL_LOSS_COEFFICIENT,
    group_coefficient=RECURRENT_GROUP_LOSS_COEFFICIENT,
):
    """
    Combine global and optional group-conditioned recurrent losses.
    """

    available_terms = []


    if global_loss is not None:

        available_terms.append(
            float(
                global_coefficient
            )
            * global_loss
        )


    if group_loss is not None:

        available_terms.append(
            float(
                group_coefficient
            )
            * group_loss
        )


    if not available_terms:
        return None


    return sum(
        available_terms
    )


# -----------------------------------------------------------------------------
# Parameter-group helper
# -----------------------------------------------------------------------------

def recurrent_parameter_groups(
    recurrent_mechanism,
):
    """
    Return explicit recurrent parameter groups for later optimiser
    construction.
    """

    if not isinstance(
        recurrent_mechanism,
        MediaAIRecurrentMechanism,
    ):
        raise TypeError(
            "recurrent_mechanism must be a "
            "MediaAIRecurrentMechanism instance."
        )


    groups = {
        "factual_recurrent":
            list(
                recurrent_mechanism
                .factual_path
                .recurrent_cell
                .parameters()
            ),

        "psychological_recurrent":
            list(
                recurrent_mechanism
                .psychological_path
                .recurrent_cell
                .parameters()
            ),

        "social_recurrent":
            list(
                recurrent_mechanism
                .social_path
                .retention_layer
                .parameters()
            )
            +
            list(
                recurrent_mechanism
                .social_path
                .candidate_network
                .parameters()
            ),
    }


    if (
        recurrent_mechanism
        .factual_path
        .group_adapter
        is not None
    ):

        groups[
            "factual_group_recurrent_adapter"
        ] = list(
            recurrent_mechanism
            .factual_path
            .group_adapter
            .parameters()
        )

    else:

        groups[
            "factual_group_recurrent_adapter"
        ] = []


    if (
        recurrent_mechanism
        .psychological_path
        .group_adapter
        is not None
    ):

        groups[
            "psychological_group_recurrent_adapter"
        ] = list(
            recurrent_mechanism
            .psychological_path
            .group_adapter
            .parameters()
        )

    else:

        groups[
            "psychological_group_recurrent_adapter"
        ] = []


    if (
        recurrent_mechanism
        .social_path
        .group_adapter
        is not None
    ):

        groups[
            "social_group_recurrent_adapter"
        ] = list(
            recurrent_mechanism
            .social_path
            .group_adapter
            .parameters()
        )

    else:

        groups[
            "social_group_recurrent_adapter"
        ] = []


    return groups


# -----------------------------------------------------------------------------
# Architecture validation
# -----------------------------------------------------------------------------

RECURRENT_ARCHITECTURE_ERRORS = []


if tuple(
    RECURRENT_STATE_SPACES
) != (
    "factual",
    "psychological",
    "social",
):
    RECURRENT_ARCHITECTURE_ERRORS.append(
        "recurrent_state_spaces_incorrect"
    )


if RECURRENT_STATES_SHARED_ACROSS_SPACES:
    RECURRENT_ARCHITECTURE_ERRORS.append(
        "recurrent_states_incorrectly_shared"
    )


if RECURRENT_STATES_COLLAPSED:
    RECURRENT_ARCHITECTURE_ERRORS.append(
        "recurrent_states_incorrectly_collapsed"
    )


if RECURRENT_TW_COLLAPSED:
    RECURRENT_ARCHITECTURE_ERRORS.append(
        "transformative_weights_incorrectly_collapsed_in_recurrence"
    )


if RECURRENT_TW_IS_RETENTION_GATE:
    RECURRENT_ARCHITECTURE_ERRORS.append(
        "transformative_weight_incorrectly_defined_as_retention_gate"
    )


if not RECURRENT_GLOBAL_PATH_REQUIRED:
    RECURRENT_ARCHITECTURE_ERRORS.append(
        "global_recurrent_path_not_required"
    )


if not RECURRENT_CAUSAL:
    RECURRENT_ARCHITECTURE_ERRORS.append(
        "recurrent_mechanism_not_causal"
    )


if RECURRENT_FUTURE_CONTEXT_ALLOWED:
    RECURRENT_ARCHITECTURE_ERRORS.append(
        "future_context_incorrectly_allowed"
    )


if RECURRENT_FIXED_DECAY:
    RECURRENT_ARCHITECTURE_ERRORS.append(
        "fixed_recurrent_decay_incorrectly_required"
    )


if RECURRENT_REPETITION_EQUALS_REINFORCEMENT:
    RECURRENT_ARCHITECTURE_ERRORS.append(
        "repetition_incorrectly_equated_with_reinforcement"
    )


if not RECURRENT_CONTRADICTION_REVISION_ALLOWED:
    RECURRENT_ARCHITECTURE_ERRORS.append(
        "contradiction_revision_not_allowed"
    )


if not RECURRENT_MASK_AWARE:
    RECURRENT_ARCHITECTURE_ERRORS.append(
        "recurrent_mechanism_not_mask_aware"
    )


if not RECURRENT_SOCIAL_ALIGNMENT_REQUIRED:
    RECURRENT_ARCHITECTURE_ERRORS.append(
        "social_recurrent_alignment_not_required"
    )


if not RECURRENT_ARTICLE_BOUNDARY_RESET:
    RECURRENT_ARCHITECTURE_ERRORS.append(
        "article_boundary_reset_not_required"
    )


if RECURRENT_CROSS_ARTICLE_CARRY:
    RECURRENT_ARCHITECTURE_ERRORS.append(
        "cross_article_recurrent_carry_incorrectly_enabled"
    )


if RECURRENT_INITIAL_STATE_ARTICLE_SPECIFIC:
    RECURRENT_ARCHITECTURE_ERRORS.append(
        "initial_state_incorrectly_contains_article_specific_information"
    )


if RECURRENT_GROUP_DEVIATION_IS_BIAS:
    RECURRENT_ARCHITECTURE_ERRORS.append(
        "recurrent_deviation_incorrectly_defined_as_bias"
    )


if RECURRENT_GROUP_DEVIATION_IS_MANIPULATION:
    RECURRENT_ARCHITECTURE_ERRORS.append(
        "recurrent_deviation_incorrectly_defined_as_manipulation"
    )


if RECURRENT_GROUP_DEVIATION_IS_POLARISATION:
    RECURRENT_ARCHITECTURE_ERRORS.append(
        "recurrent_deviation_incorrectly_defined_as_polarisation"
    )


if RECURRENT_HIERARCHICAL_AGGREGATION:
    RECURRENT_ARCHITECTURE_ERRORS.append(
        "hierarchical_aggregation_executed_in_recurrent_block"
    )


if RECURRENT_BACKWARD_ATTRIBUTION_CALCULATED:
    RECURRENT_ARCHITECTURE_ERRORS.append(
        "backward_attribution_calculated_in_recurrent_block"
    )


RECURRENT_ARCHITECTURE_VALID = (
    len(
        RECURRENT_ARCHITECTURE_ERRORS
    )
    == 0
)


if not RECURRENT_ARCHITECTURE_VALID:
    raise ValueError(
        "Recurrent architecture failed validation: "
        f"{RECURRENT_ARCHITECTURE_ERRORS}"
    )


# -----------------------------------------------------------------------------
# Execution state
# -----------------------------------------------------------------------------

RECURRENT_CLASS_DEFINED = True

RECURRENT_MODEL_INSTANTIATED = False

RECURRENT_INITIAL_STATE_GENERATED = False

RECURRENT_FORWARD_PASS_EXECUTED = False

RECURRENT_FACTUAL_STATE_CALCULATED = False

RECURRENT_PSYCHOLOGICAL_STATE_CALCULATED = False

RECURRENT_SOCIAL_STATE_CALCULATED = False

RECURRENT_RETENTION_GATE_CALCULATED = False

RECURRENT_GROUP_DEVIATION_CALCULATED = False

RECURRENT_CROSS_SENTENCE_PROPAGATION_EXECUTED = False

RECURRENT_HIERARCHICAL_AGGREGATION_EXECUTED = False

RECURRENT_BACKWARD_ATTRIBUTION_EXECUTED = False

RECURRENT_TRAINING_EXECUTED = False

RECURRENT_EVALUATION_EXECUTED = False


# -----------------------------------------------------------------------------
# Architecture summary
# -----------------------------------------------------------------------------

RECURRENT_SUMMARY = {
    "version":
        RECURRENT_MECHANISM_VERSION,

    "representation_level":
        RECURRENT_MECHANISM_LEVEL,

    "position":
        RECURRENT_MECHANISM_POSITION,

    "state_spaces":
        RECURRENT_STATE_SPACES,

    "factual_state":
        RECURRENT_FACTUAL_STATE_NAME,

    "psychological_state":
        RECURRENT_PSYCHOLOGICAL_STATE_NAME,

    "social_state":
        RECURRENT_SOCIAL_STATE_NAME,

    "states_shared":
        RECURRENT_STATES_SHARED_ACROSS_SPACES,

    "states_collapsed":
        RECURRENT_STATES_COLLAPSED,

    "TW_F":
        RECURRENT_TW_FACTUAL,

    "TW_P":
        RECURRENT_TW_PSYCHOLOGICAL,

    "TW_S":
        RECURRENT_TW_SOCIAL,

    "TW_collapsed":
        RECURRENT_TW_COLLAPSED,

    "TW_is_retention_gate":
        RECURRENT_TW_IS_RETENTION_GATE,

    "causal":
        RECURRENT_CAUSAL,

    "future_context_allowed":
        RECURRENT_FUTURE_CONTEXT_ALLOWED,

    "article_boundary_reset":
        RECURRENT_ARTICLE_BOUNDARY_RESET,

    "cross_article_carry":
        RECURRENT_CROSS_ARTICLE_CARRY,

    "fixed_decay":
        RECURRENT_FIXED_DECAY,

    "mask_aware":
        RECURRENT_MASK_AWARE,

    "architecture_valid":
        RECURRENT_ARCHITECTURE_VALID,

    "model_instantiated":
        RECURRENT_MODEL_INSTANTIATED,

    "training_executed":
        RECURRENT_TRAINING_EXECUTED,
}


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Block 16: Recurrent Mechanism")
print("=" * 72)

print(
    f"Mechanism version           : "
    f"{RECURRENT_MECHANISM_VERSION}"
)

print(
    f"Representation level        : "
    f"{RECURRENT_MECHANISM_LEVEL}"
)

print(
    f"Mechanism position          : "
    f"{RECURRENT_MECHANISM_POSITION}"
)

print(
    f"Sequence unit               : "
    f"{RECURRENT_SEQUENCE_UNIT}"
)

print("-" * 72)

print("Recurrent state spaces")

print(
    f"Recurrent state count       : "
    f"{len(RECURRENT_STATE_SPACES)}"
)

print(
    f"Factual recurrent state     : "
    f"{RECURRENT_FACTUAL_STATE_NAME}"
)

print(
    f"Psychological recurrent st. : "
    f"{RECURRENT_PSYCHOLOGICAL_STATE_NAME}"
)

print(
    f"Social recurrent state      : "
    f"{RECURRENT_SOCIAL_STATE_NAME}"
)

print(
    f"Shared across spaces        : "
    f"{RECURRENT_STATES_SHARED_ACROSS_SPACES}"
)

print(
    f"Collapsed to one state      : "
    f"{RECURRENT_STATES_COLLAPSED}"
)

print("-" * 72)

print("Transformative Weight routing")

print(
    f"Factual TW                  : "
    f"{RECURRENT_TW_FACTUAL}"
)

print(
    f"Psychological TW            : "
    f"{RECURRENT_TW_PSYCHOLOGICAL}"
)

print(
    f"Social TW                   : "
    f"{RECURRENT_TW_SOCIAL}"
)

print(
    f"TW collapsed                : "
    f"{RECURRENT_TW_COLLAPSED}"
)

print(
    f"TW = retention gate         : "
    f"{RECURRENT_TW_IS_RETENTION_GATE}"
)

print("-" * 72)

print("Sequential semantics")

print(
    f"Causal                      : "
    f"{RECURRENT_CAUSAL}"
)

print(
    f"Future context allowed      : "
    f"{RECURRENT_FUTURE_CONTEXT_ALLOWED}"
)

print(
    f"Sentence order required     : "
    f"{RECURRENT_SEQUENCE_ORDER_REQUIRED}"
)

print(
    f"Fixed decay                 : "
    f"{RECURRENT_FIXED_DECAY}"
)

print(
    f"Repetition = reinforcement  : "
    f"{RECURRENT_REPETITION_EQUALS_REINFORCEMENT}"
)

print(
    f"Contradiction revision      : "
    f"{RECURRENT_CONTRADICTION_REVISION_ALLOWED}"
)

print("-" * 72)

print("Article boundaries")

print(
    f"Boundary reset              : "
    f"{RECURRENT_ARTICLE_BOUNDARY_RESET}"
)

print(
    f"Cross-article carry         : "
    f"{RECURRENT_CROSS_ARTICLE_CARRY}"
)

print(
    f"Initial-state policy        : "
    f"{RECURRENT_INITIAL_STATE_POLICY}"
)

print(
    f"Learned initial allowed     : "
    f"{RECURRENT_LEARNED_INITIAL_STATE_ALLOWED}"
)

print(
    f"Article-specific initial    : "
    f"{RECURRENT_INITIAL_STATE_ARTICLE_SPECIFIC}"
)

print("-" * 72)

print("Global and group paths")

print(
    f"Global path required        : "
    f"{RECURRENT_GLOBAL_PATH_REQUIRED}"
)

print(
    f"Group path optional         : "
    f"{RECURRENT_GROUP_PATH_OPTIONAL}"
)

print(
    f"Group adaptation defined    : "
    f"{RECURRENT_GROUP_ADAPTATION}"
)

print(
    f"Social alignment required   : "
    f"{RECURRENT_SOCIAL_ALIGNMENT_REQUIRED}"
)

print("-" * 72)

print("Group deviation")

print(
    f"Signed deviation            : "
    f"{RECURRENT_GROUP_DEVIATION_SIGNED}"
)

print(
    f"Deviation = bias            : "
    f"{RECURRENT_GROUP_DEVIATION_IS_BIAS}"
)

print(
    f"Deviation = manipulation    : "
    f"{RECURRENT_GROUP_DEVIATION_IS_MANIPULATION}"
)

print(
    f"Deviation = polarisation    : "
    f"{RECURRENT_GROUP_DEVIATION_IS_POLARISATION}"
)

print(
    f"Deviation calculated        : "
    f"{RECURRENT_GROUP_DEVIATION_CALCULATED}"
)

print("-" * 72)

print("Training contract")

print(
    f"Mask-aware                  : "
    f"{RECURRENT_MASK_AWARE}"
)

print(
    f"Factual recurrent loss      : "
    f"{RECURRENT_FACTUAL_LOSS_NAME}"
)

print(
    f"Psychological recurrent loss: "
    f"{RECURRENT_PSYCHOLOGICAL_LOSS_NAME}"
)

print(
    f"Social recurrent loss       : "
    f"{RECURRENT_SOCIAL_LOSS_NAME}"
)

print("-" * 72)

print("Methodological boundaries")

print(
    f"Hierarchical aggregation    : "
    f"{RECURRENT_HIERARCHICAL_AGGREGATION}"
)

print(
    f"Backward attribution        : "
    f"{RECURRENT_BACKWARD_ATTRIBUTION_CALCULATED}"
)

print(
    f"Cross-article recurrence    : "
    f"{RECURRENT_CROSS_ARTICLE_CARRY}"
)

print("-" * 72)

print("Execution state")

print(
    f"Class defined               : "
    f"{RECURRENT_CLASS_DEFINED}"
)

print(
    f"Model instantiated          : "
    f"{RECURRENT_MODEL_INSTANTIATED}"
)

print(
    f"Initial state generated     : "
    f"{RECURRENT_INITIAL_STATE_GENERATED}"
)

print(
    f"Forward pass executed       : "
    f"{RECURRENT_FORWARD_PASS_EXECUTED}"
)

print(
    f"Factual recurrent state     : "
    f"{RECURRENT_FACTUAL_STATE_CALCULATED}"
)

print(
    f"Psychological recurrent st. : "
    f"{RECURRENT_PSYCHOLOGICAL_STATE_CALCULATED}"
)

print(
    f"Social recurrent state      : "
    f"{RECURRENT_SOCIAL_STATE_CALCULATED}"
)

print(
    f"Retention gate calculated   : "
    f"{RECURRENT_RETENTION_GATE_CALCULATED}"
)

print(
    f"Cross-sentence propagation  : "
    f"{RECURRENT_CROSS_SENTENCE_PROPAGATION_EXECUTED}"
)

print(
    f"Training executed           : "
    f"{RECURRENT_TRAINING_EXECUTED}"
)

print(
    f"Evaluation executed         : "
    f"{RECURRENT_EVALUATION_EXECUTED}"
)

print(
    f"Architecture valid          : "
    f"{RECURRENT_ARCHITECTURE_VALID}"
)

print("=" * 72)

print(
    "Trainable recurrent architecture defined successfully."
)

print(
    "Factual, psychological and social recurrent states remain "
    "separate architectural objects."
)

print(
    "H_F represents factual sequential persistence."
)

print(
    "H_P represents psychological sequential persistence."
)

print(
    "H_S represents social sequential persistence."
)

print(
    "TW_F, TW_P and TW_S enter their corresponding recurrent "
    "paths independently."
)

print(
    "Transformative Weight is not the recurrent retention gate."
)

print(
    "The recurrent mechanism learns the balance between preceding "
    "state persistence and current-state revision."
)

print(
    "Persistence, reinforcement, attenuation and contradiction "
    "revision are trainable rather than hard-coded."
)

print(
    "The recurrent mechanism is causal and does not permit future "
    "sentence information in the current update."
)

print(
    "Recurrent state is reset at independent article boundaries."
)

print(
    "Global and group-conditioned recurrent paths remain "
    "independently preserved."
)

print(
    "Structured social recurrence preserves the requirement for "
    "actor, group, membership and relationship alignment."
)

print(
    "Recurrent group deviation is signed and is not automatically "
    "interpreted as bias, manipulation or polarisation."
)

print(
    "The recurrent mechanism does not perform hierarchical "
    "aggregation or backward attribution."
)

print(
    "No recurrent model has been instantiated."
)

print(
    "No initial state, forward pass, recurrent state, retention gate, "
    "group deviation, cross-sentence propagation, training or "
    "evaluation has been executed."
)
````

## Block 17 — Architecture Integration, Validation and Notebook 05 Handover

### Purpose

This block integrates and validates the architectural components defined in Notebook 04 and establishes the formal handover contract for Notebook 05.

Notebook 04 has progressively defined the representation, conditioning, predictive-head, transformative and recurrent components of the Media AI architecture.

Block 17 does not introduce an additional substantive modelling mechanism.

Instead, it verifies that the architectural components form a coherent computational sequence and identifies the objects, constraints and methodological boundaries that must be preserved when the architecture is instantiated and trained.

The integrated architecture follows the sentence-level processing sequence:

$$
\text{Textual representation}
\rightarrow
\text{confluent conditioning}
\rightarrow
\text{representation heads}
\rightarrow
\text{transformative mechanism}
\rightarrow
\text{recurrent mechanism}
$$

The resulting sentence-level states may subsequently support hierarchical aggregation, attribution, training and evaluation in Notebook 05.

---

### Integrated architectural sequence

For sentence \(t\), the architecture begins from the contextual textual representation produced by the pretrained semantic backbone.

Let:

**\(X_t\)** — contextual textual representation of sentence \(t\).

The sentence representation enters the confluent conditioning mechanism.

The confluent mechanism preserves a global representation and, where validated groups are available, may additionally construct group-conditioned representations.

Conceptually:

$$
X_t
\rightarrow
\left\{
X^{G}_t,\;
X^{g}_t
\right\}
$$

where:

**\(X^{G}_t\)** — global conditioned representation;

**\(X^{g}_t\)** — optional group-conditioned representation for detected group \(g\).

The global path is mandatory.

The group-conditioned path is additional and must never replace the global path.

---

### Representation heads

Each valid confluent route is processed through the three substantive representation heads:

- factual head;
- psychological head;
- social head.

For route \(r\), where \(r\) may represent the global path or a validated group-conditioned path:

$$
Z^F_{t,r}
=
H_F(X_{t,r})
$$

$$
Z^P_{t,r}
=
H_P(X_{t,r})
$$

$$
Z^S_{t,r}
=
H_S(X_{t,r})
$$

where:

**\(Z^F_{t,r}\)** — factual representation;

**\(Z^P_{t,r}\)** — psychological representation;

**\(Z^S_{t,r}\)** — structured social representation.

The three heads remain substantively distinct.

Their outputs are not collapsed into a single undifferentiated analytical vector.

---

### Factual representation contract

The factual head preserves the ten-dimensional factual ontology defined earlier in Notebook 04.

The factual representation contains:

- one Factological Weight dimension;
- nine explicit factological components.

The factual representation describes factological structure and epistemic presentation.

It does not independently verify external truth.

Missing factual supervision remains distinct from numerical zero.

The factual representation therefore preserves both value and availability information where supervision is present.

---

### Psychological representation contract

The psychological head preserves the 34-dimensional Psychological State Vector.

The vector contains:

- 11 Reported Emotion Vector dimensions;
- 11 Transformative Emotion Vector dimensions;
- 11 Immediate Emotion Vector dimensions;
- one Emotional Weight dimension.

Conceptually:

$$
Z^P_t
=
[
REV_t,\;
TEV_t,\;
IEV_t,\;
EW_t
]
$$

with:

$$
\dim(REV_t)=11
$$

$$
\dim(TEV_t)=11
$$

$$
\dim(IEV_t)=11
$$

$$
\dim(EW_t)=1
$$

and therefore:

$$
\dim(Z^P_t)=34
$$

REV and IEV remain bounded continuous representations.

TEV remains a signed continuous representation.

EW remains a bounded continuous scalar.

The eleven emotional dimensions are not mutually exclusive and are not constrained to a probability simplex.

---

### Social representation contract

The social head preserves a structured and variable-dimensional representation.

The social representation contains four structural component classes:

- actors;
- groups;
- memberships;
- relationships.

Concrete social groups are not predefined.

Group detection remains dynamic.

Overlapping memberships remain permitted.

Relationships may be directional and signed.

Because the social representation is variable-dimensional, it must not be forced into an arbitrary fixed-dimensional final social output solely for architectural convenience.

Trainable hidden representations may be used internally while the explicit structured social representation remains preserved.

---

### Global and group-conditioned paths

The integrated architecture preserves two analytically distinct routing levels:

- global;
- group-conditioned.

The global path is always available.

Group-conditioned paths are created only where usable group information has been detected and validated.

For representation space \(q\):

$$
Z^q_{t,G}
$$

denotes the global representation, while:

$$
Z^q_{t,g}
$$

denotes the representation conditioned on group \(g\).

The group-conditioned representation is additional.

It does not replace or redefine the global representation.

Concrete group identities must originate from validated article-specific detection rather than from a closed predefined ideological or demographic vocabulary.

---

### Group deviation

Where global and group-conditioned representations are aligned and comparable, signed group deviation may be defined.

For a fixed-dimensional representation space \(q\):

$$
D^q_{t,g}
=
Z^q_{t,g}
-
Z^q_{t,G}
$$

The deviation represents a descriptive difference between the group-conditioned and global paths.

It is not automatically equivalent to:

- bias;
- manipulation;
- polarisation.

For the social representation, meaningful deviation requires explicit structural alignment before comparison.

Interpretation of deviation remains a downstream analytical task.

---

### Transformative integration

The outputs of the representation heads enter the transformative mechanism defined in Block 15.

For each representation space:

$$
q \in \{F,P,S\}
$$

the transformative mechanism produces a learned transformation:

$$
\Delta^q_t
=
T^q
\left(
Z^q_t,
C_t;
\theta^q_T
\right)
$$

where \(C_t\) contains only permitted causal context.

The transformation is not defined as direct subtraction between consecutive head outputs.

The transformation is subsequently regulated by a trainable transformation gate.

For fixed-dimensional spaces:

$$
\tilde{\Delta}^q_t
=
G^q_t
\odot
\Delta^q_t
$$

where:

**\(G^q_t\)** — transformation gate;

**\(\tilde{\Delta}^q_t\)** — gated transformation.

---

### Space-specific Transformative Weights

The transformative mechanism preserves three independent Transformative Weights:

$$
TW^F_t
$$

$$
TW^P_t
$$

$$
TW^S_t
$$

These correspond respectively to:

- factual transformative salience;
- psychological transformative salience;
- social transformative salience.

Each Transformative Weight is bounded:

$$
0
\leq
TW^q_t
\leq
1
$$

The three weights are not shared across representation spaces.

They are not collapsed into a single overall Transformative Weight.

Transformative Weight remains distinct from transformation magnitude.

It also remains distinct from psychological Emotional Weight.

In particular:

$$
TW^P_t
\neq
EW_t
$$

---

### Recurrent integration

The gated transformations and space-specific Transformative Weights enter the recurrent mechanism defined in Block 16.

The architecture preserves three recurrent states:

$$
H^F_t
$$

$$
H^P_t
$$

$$
H^S_t
$$

representing factual, psychological and social sequential persistence respectively.

For representation space \(q\), the recurrent update is conceptually defined as:

$$
H^q_t
=
R^q
\left(
H^q_{t-1},
Z^q_t,
\tilde{\Delta}^q_t,
TW^q_t;
\theta^q_R
\right)
$$

The three recurrent states remain separate.

They are not collapsed into a universal recurrent state.

---

### Recurrent retention

For fixed-dimensional recurrent spaces, the recurrent mechanism learns how much preceding state should remain active.

The retention gate is conceptually represented as:

$$
K^q_t
=
\sigma
\left(
W^q_K
[
H^q_{t-1},
Z^q_t,
\tilde{\Delta}^q_t,
TW^q_t
]
+
b^q_K
\right)
$$

The candidate recurrent state is:

$$
\hat{H}^q_t
=
\tanh
\left(
W^q_H
[
H^q_{t-1},
Z^q_t,
\tilde{\Delta}^q_t,
TW^q_t
]
+
b^q_H
\right)
$$

and the recurrent update is:

$$
H^q_t
=
K^q_t
\odot
H^q_{t-1}
+
(1-K^q_t)
\odot
\hat{H}^q_t
$$

Transformative Weight is an input to the recurrent update.

It is not itself the recurrent retention gate.

Therefore:

$$
TW^q_t
\neq
K^q_t
$$

---

### Separation of representation, transformation and recurrence

The integrated architecture preserves three conceptually distinct operations.

**Representation** describes the current sentence-level analytical state:

$$
Z^q_t
$$

**Transformation** describes learned representational change associated with the current textual step:

$$
\tilde{\Delta}^q_t
$$

**Recurrence** describes the information that remains sequentially active after the current textual step:

$$
H^q_t
$$

Consequently:

$$
Z^q_t
\neq
\tilde{\Delta}^q_t
\neq
H^q_t
$$

These objects may interact but must not be treated as interchangeable.

---

### Psychological TEV boundary

The Transformative Emotion Vector remains distinct from the architectural transformative mechanism.

Therefore:

$$
TEV_t
\neq
\Delta^P_t
$$

TEV is also not defined as direct subtraction between consecutive Immediate Emotion Vectors:

$$
TEV_t
\neq
IEV_t
-
IEV_{t-1}
$$

TEV remains a substantive component of the Psychological State Vector.

Architectural psychological transformation remains a learned transformation of psychological representation.

---

### Causal processing contract

The integrated sentence-level architecture is causal.

At sentence \(t\), the model may use information available at or before sentence \(t\).

Future sentences remain inaccessible.

Conceptually:

$$
\mathcal{I}_t
\subseteq
\mathcal{I}_{\leq t}
$$

and:

$$
\mathcal{I}_{>t}
\not\rightarrow
\mathcal{I}_t
$$

This causal constraint applies to:

- transformative processing;
- recurrent processing;
- group-conditioned sequential processing.

Future-context leakage must remain prohibited during training and evaluation wherever the architecture is intended to represent sequential reader exposure.

---

### Article-boundary contract

Recurrent states are article-specific.

At the beginning of each independent article, factual, psychological and social recurrent states are reset.

For article \(j+1\):

$$
H^q_0(j+1)
\neq
H^q_T(j)
$$

as a carry-over relationship.

The final recurrent state of one independent article must not initialise the next independent article.

Cross-document recurrent carry-over is outside the current architecture.

---

### Missing information and masking

The integrated architecture preserves the distinction between:

- observed value;
- observed zero;
- unavailable value;
- structurally absent object;
- explicit negative observation.

Missing supervision must not be converted into numerical zero merely to satisfy tensor construction.

Mask-aware objectives must therefore remain active wherever supervision is incomplete.

For supervised value \(y_i\), prediction \(\hat{y}_i\) and availability mask \(m_i\), optimisation conceptually uses:

$$
\mathcal{L}_{masked}
=
\frac{
\sum_i
m_i
\,
\ell(\hat{y}_i,y_i)
}{
\sum_i m_i
}
$$

where only available supervised observations contribute to the objective.

The exact implementation must also handle the case in which no valid supervised observations are available.

---

### Integrated training objective

Notebook 04 defines multiple trainable architectural components rather than a single undifferentiated prediction head.

Notebook 05 may therefore construct an integrated objective from available component-specific losses.

Conceptually:

$$
\mathcal{L}_{total}
=
\mathcal{L}_{F}
+
\mathcal{L}_{P}
+
\mathcal{L}_{S}
+
\mathcal{L}_{T}
+
\mathcal{L}_{R}
+
\mathcal{L}_{aux}
$$

where:

**\(\mathcal{L}_{F}\)** — factual-head loss;

**\(\mathcal{L}_{P}\)** — psychological-head loss;

**\(\mathcal{L}_{S}\)** — social-head loss;

**\(\mathcal{L}_{T}\)** — transformative-mechanism loss;

**\(\mathcal{L}_{R}\)** — recurrent-mechanism loss;

**\(\mathcal{L}_{aux}\)** — optional explicitly justified auxiliary losses.

More generally, trainable coefficients may be introduced:

$$
\mathcal{L}_{total}
=
\lambda_F\mathcal{L}_F
+
\lambda_P\mathcal{L}_P
+
\lambda_S\mathcal{L}_S
+
\lambda_T\mathcal{L}_T
+
\lambda_R\mathcal{L}_R
+
\lambda_{aux}\mathcal{L}_{aux}
$$

The initial architectural defaults do not establish empirical optimality.

Loss coefficients must therefore remain explicit and auditable experimental parameters.

---

### Training-stage separation

The existence of an integrated architecture does not require every component to be trained simultaneously from the first optimisation step.

Notebook 05 may define a staged training strategy.

Possible stages may include:

- component-level validation;
- supervised head training;
- transformative-mechanism training;
- recurrent training;
- group-conditioned adaptation;
- integrated joint optimisation;
- controlled ablation experiments.

The exact training schedule must be documented before execution.

Training-stage decisions must not silently alter the representational definitions established in Notebook 04.

### Parameter control

Notebook 05 must explicitly determine which parameter groups are trainable during each training stage.

The architecture may contain parameters associated with:

- backbone adaptation;
- confluent conditioning;
- factual head;
- psychological head;
- social head;
- group-conditioned adapters;
- transformative mechanism;
- recurrent mechanism.

Frozen and trainable parameter groups must be recorded explicitly.

No parameter group should become trainable merely because it is reachable by the computational graph.

This distinction is particularly important if the pretrained semantic backbone is frozen or adapted selectively.

### Supervision contract

Training requires explicit supervision availability.

The architecture must not interpret missing labels as negative labels.

Notebook 05 must therefore determine, for each training target:

- whether supervision exists;
- the source of supervision;
- the representation level;
- the target dimensionality or structure;
- the applicable availability mask;
- the corresponding loss function;
- whether the target applies globally, group-conditionally or both.

Where validated supervision is unavailable, the corresponding supervised objective must remain inactive.

The architecture definition itself does not create supervision.

---

### Group-conditioned training boundary

Group-conditioned training requires usable group-conditioned routes.

The existence of a group-conditioned architecture does not imply that such routes are available for every sentence or article.

Group-conditioned optimisation may occur only where:

- a group has been detected;
- the group identity has been sufficiently aligned;
- the relevant route is valid;
- the corresponding supervision or justified training signal is available.

Unavailable group-conditioned routes must not be fabricated.

The global path remains independently trainable and evaluable.

---

### Hierarchical aggregation boundary

Notebook 04 operates primarily at sentence level.

Paragraph-level and article-level aggregation remain downstream operations.

Conceptually, a later hierarchical mechanism may construct:

$$

\right)
$$

for paragraph \(j\), and:

$$

A_A
\left(

P_2,
\ldots,
P_m
\right)
$$

for the article representation.

These equations define only the conceptual boundary.

Block 17 does not implement hierarchical aggregation.

The aggregation method must be specified and validated separately before use.

---

### Attribution boundary

The integrated architecture creates a computational path through which later states may depend on earlier textual steps.

This makes later backward attribution possible.

A future attribution mechanism may examine how a later representation or recurrent state depends on earlier sentences, transformations or recurrent states.

Conceptually:

$$
\text{later analytical state}
\rightarrow
\text{earlier contributing textual steps}
$$

However, attribution is not calculated in this block.

Attribution must remain separate from the predictive architecture so that explanatory analysis does not silently redefine the model outputs themselves.

````python
# =============================================================================
# Media AI — Block 17: Architecture Integration, Validation and
# Notebook 05 Handover
# =============================================================================

from datetime import datetime, timezone
from pathlib import Path
import io
import json

from google.auth import default
from googleapiclient.discovery import build
from googleapiclient.http import MediaIoBaseUpload


# -----------------------------------------------------------------------------
# Block dependency checks
# -----------------------------------------------------------------------------

required_block_12_objects = [
    "FACTUAL_HEAD_ARCHITECTURE_VALID",
]

missing_block_12_objects = [
    object_name
    for object_name in required_block_12_objects
    if object_name not in globals()
]

if missing_block_12_objects:
    raise NameError(
        "Notebook 04 Block 12 is not initialised. "
        f"Missing: {missing_block_12_objects}"
    )


required_block_13_objects = [
    "PSYCHOLOGICAL_HEAD_ARCHITECTURE_VALID",
    "PSYCHOLOGICAL_STATE_DIMENSION_COUNT",
]

missing_block_13_objects = [
    object_name
    for object_name in required_block_13_objects
    if object_name not in globals()
]

if missing_block_13_objects:
    raise NameError(
        "Notebook 04 Block 13 is not initialised. "
        f"Missing: {missing_block_13_objects}"
    )


required_block_14_objects = [
    "SOCIAL_HEAD_ARCHITECTURE_VALID",
    "SOCIAL_REPRESENTATION_COMPONENTS",
]

missing_block_14_objects = [
    object_name
    for object_name in required_block_14_objects
    if object_name not in globals()
]

if missing_block_14_objects:
    raise NameError(
        "Notebook 04 Block 14 is not initialised. "
        f"Missing: {missing_block_14_objects}"
    )


required_block_15_objects = [
    "TRANSFORMATIVE_ARCHITECTURE_VALID",
    "TRANSFORMATIVE_WEIGHT_SPACES",
    "TRANSFORMATIVE_WEIGHT_FACTUAL_NAME",
    "TRANSFORMATIVE_WEIGHT_PSYCHOLOGICAL_NAME",
    "TRANSFORMATIVE_WEIGHT_SOCIAL_NAME",
]

missing_block_15_objects = [
    object_name
    for object_name in required_block_15_objects
    if object_name not in globals()
]

if missing_block_15_objects:
    raise NameError(
        "Notebook 04 Block 15 is not initialised. "
        f"Missing: {missing_block_15_objects}"
    )


required_block_16_objects = [
    "RECURRENT_ARCHITECTURE_VALID",
    "RECURRENT_STATE_SPACES",
    "RECURRENT_FACTUAL_STATE_NAME",
    "RECURRENT_PSYCHOLOGICAL_STATE_NAME",
    "RECURRENT_SOCIAL_STATE_NAME",
]

missing_block_16_objects = [
    object_name
    for object_name in required_block_16_objects
    if object_name not in globals()
]

if missing_block_16_objects:
    raise NameError(
        "Notebook 04 Block 16 is not initialised. "
        f"Missing: {missing_block_16_objects}"
    )


# -----------------------------------------------------------------------------
# Upstream architecture validity
# -----------------------------------------------------------------------------

UPSTREAM_ARCHITECTURE_VALIDITY = {
    "factual_head":
        bool(
            FACTUAL_HEAD_ARCHITECTURE_VALID
        ),

    "psychological_head":
        bool(
            PSYCHOLOGICAL_HEAD_ARCHITECTURE_VALID
        ),

    "social_head":
        bool(
            SOCIAL_HEAD_ARCHITECTURE_VALID
        ),

    "transformative_mechanism":
        bool(
            TRANSFORMATIVE_ARCHITECTURE_VALID
        ),

    "recurrent_mechanism":
        bool(
            RECURRENT_ARCHITECTURE_VALID
        ),
}


invalid_upstream_components = [
    component_name
    for component_name, is_valid
    in UPSTREAM_ARCHITECTURE_VALIDITY.items()
    if not is_valid
]


if invalid_upstream_components:
    raise ValueError(
        "Notebook 04 contains invalid upstream architecture "
        "components: "
        f"{invalid_upstream_components}"
    )


# -----------------------------------------------------------------------------
# Integration configuration
# -----------------------------------------------------------------------------

ARCHITECTURE_INTEGRATION_VERSION = "1.1"

ARCHITECTURE_INTEGRATION_LEVEL = "sentence"

ARCHITECTURE_SEQUENCE = (
    "textual_representation",
    "confluent_conditioning",
    "representation_heads",
    "transformative_mechanism",
    "recurrent_mechanism",
)

REPRESENTATION_SPACES = (
    "factual",
    "psychological",
    "social",
)

REPRESENTATION_HEADS = (
    "factual_head",
    "psychological_head",
    "social_head",
)

GLOBAL_PATH_REQUIRED = True

GROUP_CONDITIONED_PATH_OPTIONAL = True

GLOBAL_PATH_REPLACED_BY_GROUP_PATH = False

REPRESENTATION_SPACES_COLLAPSED = False

GLOBAL_AND_GROUP_PATHS_COLLAPSED = False


# -----------------------------------------------------------------------------
# Representation contracts
# -----------------------------------------------------------------------------

FACTUAL_REPRESENTATION_DIMENSIONS = 10

FACTUAL_REPRESENTATION_IS_TRUTH = False

FACTUAL_MISSING_EQUALS_ZERO = False


PSYCHOLOGICAL_REPRESENTATION_DIMENSIONS = int(
    PSYCHOLOGICAL_STATE_DIMENSION_COUNT
)

PSYCHOLOGICAL_EXPECTED_DIMENSIONS = 34

PSYCHOLOGICAL_REV_DIMENSIONS = 11

PSYCHOLOGICAL_TEV_DIMENSIONS = 11

PSYCHOLOGICAL_IEV_DIMENSIONS = 11

PSYCHOLOGICAL_EW_DIMENSIONS = 1

PSYCHOLOGICAL_EMOTIONS_MUTUALLY_EXCLUSIVE = False

PSYCHOLOGICAL_PROBABILITY_SIMPLEX_REQUIRED = False


SOCIAL_REPRESENTATION_COMPONENTS_INTEGRATED = tuple(
    SOCIAL_REPRESENTATION_COMPONENTS
)

SOCIAL_REPRESENTATION_FIXED_DIMENSIONAL = False

SOCIAL_GROUPS_PREDEFINED = False

SOCIAL_DYNAMIC_GROUP_DETECTION = True

SOCIAL_OVERLAPPING_MEMBERSHIP_ALLOWED = True

SOCIAL_ALIGNMENT_REQUIRED = True


# -----------------------------------------------------------------------------
# Group-conditioned representation contract
# -----------------------------------------------------------------------------

GROUP_DEVIATION_SIGNED = True

GROUP_DEVIATION_IS_BIAS = False

GROUP_DEVIATION_IS_MANIPULATION = False

GROUP_DEVIATION_IS_POLARISATION = False

GROUP_DEVIATION_REQUIRES_ALIGNMENT = True

GROUP_IDENTITIES_PREDEFINED = False


# -----------------------------------------------------------------------------
# Transformative integration contract
# -----------------------------------------------------------------------------

INTEGRATED_TRANSFORMATIVE_SPACES = tuple(
    TRANSFORMATIVE_WEIGHT_SPACES
)

INTEGRATED_TW_F = (
    TRANSFORMATIVE_WEIGHT_FACTUAL_NAME
)

INTEGRATED_TW_P = (
    TRANSFORMATIVE_WEIGHT_PSYCHOLOGICAL_NAME
)

INTEGRATED_TW_S = (
    TRANSFORMATIVE_WEIGHT_SOCIAL_NAME
)

TRANSFORMATIVE_WEIGHT_COUNT = 3

TRANSFORMATIVE_WEIGHTS_COLLAPSED = False

TRANSFORMATIVE_WEIGHTS_SHARED = False

TRANSFORMATIVE_WEIGHT_IS_EMOTIONAL_WEIGHT = False

TRANSFORMATIVE_WEIGHT_IS_TRANSFORMATION_MAGNITUDE = False

TRANSFORMATION_IS_CONSECUTIVE_DIFFERENCE = False

TEV_IS_ARCHITECTURAL_TRANSFORMATION = False

TEV_IS_IEV_DIFFERENCE = False


# -----------------------------------------------------------------------------
# Recurrent integration contract
# -----------------------------------------------------------------------------

INTEGRATED_RECURRENT_SPACES = tuple(
    RECURRENT_STATE_SPACES
)

INTEGRATED_H_F = (
    RECURRENT_FACTUAL_STATE_NAME
)

INTEGRATED_H_P = (
    RECURRENT_PSYCHOLOGICAL_STATE_NAME
)

INTEGRATED_H_S = (
    RECURRENT_SOCIAL_STATE_NAME
)

RECURRENT_STATE_COUNT = 3

RECURRENT_STATES_COLLAPSED_IN_INTEGRATION = False

TRANSFORMATIVE_WEIGHT_IS_RETENTION_GATE = False

RECURRENT_CAUSAL_INTEGRATION = True

RECURRENT_FUTURE_CONTEXT_ALLOWED_IN_INTEGRATION = False

RECURRENT_ARTICLE_BOUNDARY_RESET_INTEGRATION = True

RECURRENT_CROSS_ARTICLE_CARRY_INTEGRATION = False


# -----------------------------------------------------------------------------
# Missing-information and supervision contract
# -----------------------------------------------------------------------------

MASK_AWARE_SUPERVISION_REQUIRED = True

MISSING_SUPERVISION_EQUALS_ZERO = False

MISSING_SUPERVISION_EQUALS_NEGATIVE = False

UNAVAILABLE_SUPERVISION_CONTRIBUTES_TO_LOSS = False

OBSERVED_ZERO_DISTINCT_FROM_MISSING = True

STRUCTURAL_ABSENCE_DISTINCT_FROM_NEGATIVE = True


# -----------------------------------------------------------------------------
# Training-objective contract
# -----------------------------------------------------------------------------

INTEGRATED_LOSS_COMPONENTS = (
    "factual",
    "psychological",
    "social",
    "transformative",
    "recurrent",
)

OPTIONAL_AUXILIARY_LOSS_ALLOWED = True

LOSS_COEFFICIENTS_EXPLICIT = True

LOSS_COEFFICIENTS_AUDITABLE = True

LOSS_COEFFICIENTS_EMPIRICALLY_OPTIMAL = False


DEFAULT_INTEGRATED_LOSS_COEFFICIENTS = {
    "factual": 1.0,
    "psychological": 1.0,
    "social": 1.0,
    "transformative": 1.0,
    "recurrent": 1.0,
}


# -----------------------------------------------------------------------------
# Training-stage contract
# -----------------------------------------------------------------------------

STAGED_TRAINING_ALLOWED = True

JOINT_TRAINING_ALLOWED = True

TRAINING_SCHEDULE_DEFINED = False

TRAINING_SCHEDULE_EXECUTED = False

PARAMETER_GROUP_CONTROL_REQUIRED = True

FROZEN_PARAMETER_RECORD_REQUIRED = True

TRAINABLE_PARAMETER_RECORD_REQUIRED = True

BACKBONE_AUTOMATICALLY_TRAINABLE = False


# -----------------------------------------------------------------------------
# Hierarchical and attribution boundaries
# -----------------------------------------------------------------------------

HIERARCHICAL_AGGREGATION_DEFINED = False

HIERARCHICAL_AGGREGATION_EXECUTED = False

PARAGRAPH_AGGREGATION_EXECUTED = False

ARTICLE_AGGREGATION_EXECUTED = False

BACKWARD_ATTRIBUTION_DEFINED = False

BACKWARD_ATTRIBUTION_EXECUTED = False


# -----------------------------------------------------------------------------
# Evaluation and ablation contract
# -----------------------------------------------------------------------------

EVALUATION_REQUIRED = True

ABLATION_REQUIRED = True

CALIBRATION_WHERE_APPLICABLE = True

ROBUSTNESS_EVALUATION_REQUIRED = True

SUPERVISION_COVERAGE_REPORT_REQUIRED = True

STRUCTURE_APPROPRIATE_METRICS_REQUIRED = True


PLANNED_ABLATIONS = (
    "without_group_conditioned_routing",
    "without_transformative_processing",
    "without_transformative_weights",
    "without_recurrent_processing",
    "without_group_residual_adaptation",
    "global_path_only",
    "individual_representation_heads",
)


# -----------------------------------------------------------------------------
# Reproducibility contract
# -----------------------------------------------------------------------------

PROJECT_RANDOM_SEED = 42

REPRODUCIBILITY_REQUIRED = True

TRAINING_CONFIGURATION_RECORD_REQUIRED = True

CHECKPOINT_SELECTION_RULE_REQUIRED = True

EXPERIMENT_CONFIGURATION_RECOVERABLE = True


REQUIRED_EXPERIMENT_CONFIGURATION_FIELDS = (
    "random_seed",
    "model_configuration",
    "trainable_parameter_groups",
    "frozen_parameter_groups",
    "optimizer",
    "learning_rate",
    "loss_coefficients",
    "batch_construction",
    "sequence_handling",
    "masking_policy",
    "training_duration_or_stopping_criterion",
    "checkpoint_selection_rule",
    "evaluation_configuration",
)


# -----------------------------------------------------------------------------
# Notebook 05 handover contract
# -----------------------------------------------------------------------------

NOTEBOOK_05_NAME = "05_training_evaluation_and_validation"

NOTEBOOK_05_HANDOVER_VERSION = "1.1"

NOTEBOOK_05_HANDOVER_FILENAME = (
    "notebook_04_to_05_handover.json"
)

NOTEBOOK_05_HANDOVER_LOCAL_PATH = (
    Path("/content")
    / NOTEBOOK_05_HANDOVER_FILENAME
)

NOTEBOOK_05_HANDOVER_CREATED_AT_UTC = (
    datetime.now(
        timezone.utc
    ).isoformat()
)


NOTEBOOK_05_REQUIRED_ARCHITECTURAL_CONTRACTS = (
    "representation_specification",
    "factual_ontology",
    "psychological_ontology",
    "social_ontology",
    "global_routing",
    "group_conditioned_routing",
    "confluent_conditioning",
    "factual_head",
    "psychological_head",
    "social_head",
    "transformative_mechanism",
    "TW_F",
    "TW_P",
    "TW_S",
    "recurrent_mechanism",
    "causal_constraints",
    "masking_semantics",
    "article_boundary_reset",
    "group_deviation_semantics",
    "methodological_boundaries",
)


NOTEBOOK_05_RESPONSIBILITIES = (
    "validate_handover",
    "resolve_training_supervision",
    "define_trainable_parameter_groups",
    "define_frozen_parameter_groups",
    "instantiate_model_components",
    "construct_training_sequences",
    "define_optimisation",
    "execute_controlled_forward_passes",
    "calculate_available_losses",
    "train_permitted_components",
    "store_checkpoints",
    "evaluate_trained_behaviour",
    "perform_ablation_analysis",
    "record_experimental_evidence",
)


# -----------------------------------------------------------------------------
# Notebook 04 methodological boundaries
# -----------------------------------------------------------------------------

NOTEBOOK_04_FACTOLOGICAL_EQUALS_TRUTH = False

NOTEBOOK_04_PSYCHOLOGICAL_EQUALS_DIAGNOSIS = False

NOTEBOOK_04_PREDEFINES_CONCRETE_GROUPS = False

NOTEBOOK_04_COLLAPSES_REPRESENTATION_SPACES = False

NOTEBOOK_04_COLLAPSES_GLOBAL_GROUP_PATHS = False

NOTEBOOK_04_COLLAPSES_TRANSFORMATIVE_WEIGHTS = False

NOTEBOOK_04_TW_EQUALS_EW = False

NOTEBOOK_04_TW_EQUALS_RETENTION = False

NOTEBOOK_04_TEV_EQUALS_ARCHITECTURAL_TRANSFORMATION = False

NOTEBOOK_04_TEV_EQUALS_IEV_DIFFERENCE = False

NOTEBOOK_04_GROUP_DEVIATION_EQUALS_BIAS = False

NOTEBOOK_04_GROUP_DEVIATION_EQUALS_MANIPULATION = False

NOTEBOOK_04_GROUP_DEVIATION_EQUALS_POLARISATION = False

NOTEBOOK_04_MISSING_EQUALS_ZERO = False

NOTEBOOK_04_FUTURE_CONTEXT_ALLOWED = False

NOTEBOOK_04_CROSS_ARTICLE_RECURRENCE = False

NOTEBOOK_04_HIERARCHICAL_AGGREGATION_EXECUTED = False

NOTEBOOK_04_BACKWARD_ATTRIBUTION_EXECUTED = False

NOTEBOOK_04_EMPIRICAL_VALIDITY_ESTABLISHED = False


# -----------------------------------------------------------------------------
# Architecture integration validation
# -----------------------------------------------------------------------------

ARCHITECTURE_INTEGRATION_ERRORS = []


if tuple(
    ARCHITECTURE_SEQUENCE
) != (
    "textual_representation",
    "confluent_conditioning",
    "representation_heads",
    "transformative_mechanism",
    "recurrent_mechanism",
):
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "architecture_sequence_incorrect"
    )


if tuple(
    REPRESENTATION_SPACES
) != (
    "factual",
    "psychological",
    "social",
):
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "representation_spaces_incorrect"
    )


if not GLOBAL_PATH_REQUIRED:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "global_path_not_required"
    )


if GLOBAL_PATH_REPLACED_BY_GROUP_PATH:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "global_path_incorrectly_replaced"
    )


if REPRESENTATION_SPACES_COLLAPSED:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "representation_spaces_incorrectly_collapsed"
    )


if GLOBAL_AND_GROUP_PATHS_COLLAPSED:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "global_and_group_paths_incorrectly_collapsed"
    )


if FACTUAL_REPRESENTATION_DIMENSIONS != 10:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "factual_dimension_count_incorrect"
    )


if FACTUAL_REPRESENTATION_IS_TRUTH:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "factual_representation_incorrectly_equated_with_truth"
    )


if FACTUAL_MISSING_EQUALS_ZERO:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "factual_missing_incorrectly_equated_with_zero"
    )


if (
    PSYCHOLOGICAL_REPRESENTATION_DIMENSIONS
    != PSYCHOLOGICAL_EXPECTED_DIMENSIONS
):
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "psychological_dimension_count_incorrect"
    )


if (
    PSYCHOLOGICAL_REV_DIMENSIONS
    + PSYCHOLOGICAL_TEV_DIMENSIONS
    + PSYCHOLOGICAL_IEV_DIMENSIONS
    + PSYCHOLOGICAL_EW_DIMENSIONS
    != PSYCHOLOGICAL_EXPECTED_DIMENSIONS
):
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "psychological_component_dimensions_incorrect"
    )


if PSYCHOLOGICAL_EMOTIONS_MUTUALLY_EXCLUSIVE:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "psychological_emotions_incorrectly_mutually_exclusive"
    )


if PSYCHOLOGICAL_PROBABILITY_SIMPLEX_REQUIRED:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "psychological_probability_simplex_incorrectly_required"
    )


if SOCIAL_REPRESENTATION_FIXED_DIMENSIONAL:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "social_representation_incorrectly_fixed_dimensional"
    )


if SOCIAL_GROUPS_PREDEFINED:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "social_groups_incorrectly_predefined"
    )


if not SOCIAL_DYNAMIC_GROUP_DETECTION:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "dynamic_group_detection_not_preserved"
    )


if not SOCIAL_OVERLAPPING_MEMBERSHIP_ALLOWED:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "overlapping_membership_not_preserved"
    )


if not SOCIAL_ALIGNMENT_REQUIRED:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "social_alignment_not_required"
    )


if not GROUP_DEVIATION_SIGNED:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "group_deviation_not_signed"
    )


if GROUP_DEVIATION_IS_BIAS:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "group_deviation_incorrectly_defined_as_bias"
    )


if GROUP_DEVIATION_IS_MANIPULATION:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "group_deviation_incorrectly_defined_as_manipulation"
    )


if GROUP_DEVIATION_IS_POLARISATION:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "group_deviation_incorrectly_defined_as_polarisation"
    )


if TRANSFORMATIVE_WEIGHT_COUNT != 3:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "transformative_weight_count_incorrect"
    )


if TRANSFORMATIVE_WEIGHTS_COLLAPSED:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "transformative_weights_incorrectly_collapsed"
    )


if TRANSFORMATIVE_WEIGHTS_SHARED:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "transformative_weights_incorrectly_shared"
    )


if TRANSFORMATIVE_WEIGHT_IS_EMOTIONAL_WEIGHT:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "transformative_weight_incorrectly_equated_with_emotional_weight"
    )


if TRANSFORMATIVE_WEIGHT_IS_TRANSFORMATION_MAGNITUDE:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "transformative_weight_incorrectly_equated_with_magnitude"
    )


if TRANSFORMATION_IS_CONSECUTIVE_DIFFERENCE:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "transformation_incorrectly_defined_as_consecutive_difference"
    )


if TEV_IS_ARCHITECTURAL_TRANSFORMATION:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "TEV_incorrectly_defined_as_architectural_transformation"
    )


if TEV_IS_IEV_DIFFERENCE:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "TEV_incorrectly_defined_as_IEV_difference"
    )


if RECURRENT_STATE_COUNT != 3:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "recurrent_state_count_incorrect"
    )


if RECURRENT_STATES_COLLAPSED_IN_INTEGRATION:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "recurrent_states_incorrectly_collapsed"
    )


if TRANSFORMATIVE_WEIGHT_IS_RETENTION_GATE:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "transformative_weight_incorrectly_defined_as_retention_gate"
    )


if not RECURRENT_CAUSAL_INTEGRATION:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "recurrent_integration_not_causal"
    )


if RECURRENT_FUTURE_CONTEXT_ALLOWED_IN_INTEGRATION:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "future_context_incorrectly_allowed"
    )


if not RECURRENT_ARTICLE_BOUNDARY_RESET_INTEGRATION:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "article_boundary_reset_not_preserved"
    )


if RECURRENT_CROSS_ARTICLE_CARRY_INTEGRATION:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "cross_article_recurrence_incorrectly_enabled"
    )


if not MASK_AWARE_SUPERVISION_REQUIRED:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "mask_aware_supervision_not_required"
    )


if MISSING_SUPERVISION_EQUALS_ZERO:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "missing_supervision_incorrectly_equated_with_zero"
    )


if MISSING_SUPERVISION_EQUALS_NEGATIVE:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "missing_supervision_incorrectly_equated_with_negative"
    )


if UNAVAILABLE_SUPERVISION_CONTRIBUTES_TO_LOSS:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "unavailable_supervision_incorrectly_contributes_to_loss"
    )


if not LOSS_COEFFICIENTS_EXPLICIT:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "loss_coefficients_not_explicit"
    )


if not PARAMETER_GROUP_CONTROL_REQUIRED:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "parameter_group_control_not_required"
    )


if BACKBONE_AUTOMATICALLY_TRAINABLE:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "backbone_incorrectly_automatically_trainable"
    )


if HIERARCHICAL_AGGREGATION_EXECUTED:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "hierarchical_aggregation_incorrectly_executed"
    )


if BACKWARD_ATTRIBUTION_EXECUTED:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "backward_attribution_incorrectly_executed"
    )


if PROJECT_RANDOM_SEED != 42:
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "project_random_seed_incorrect"
    )


# -----------------------------------------------------------------------------
# Cross-block consistency validation
# -----------------------------------------------------------------------------

if tuple(
    INTEGRATED_TRANSFORMATIVE_SPACES
) != (
    "factual",
    "psychological",
    "social",
):
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "transformative_spaces_not_aligned"
    )


if tuple(
    INTEGRATED_RECURRENT_SPACES
) != (
    "factual",
    "psychological",
    "social",
):
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "recurrent_spaces_not_aligned"
    )


if INTEGRATED_TW_F != "TW_F":
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "TW_F_name_not_preserved"
    )


if INTEGRATED_TW_P != "TW_P":
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "TW_P_name_not_preserved"
    )


if INTEGRATED_TW_S != "TW_S":
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "TW_S_name_not_preserved"
    )


if INTEGRATED_H_F != "H_F":
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "H_F_name_not_preserved"
    )


if INTEGRATED_H_P != "H_P":
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "H_P_name_not_preserved"
    )


if INTEGRATED_H_S != "H_S":
    ARCHITECTURE_INTEGRATION_ERRORS.append(
        "H_S_name_not_preserved"
    )


# -----------------------------------------------------------------------------
# Integrated architecture validity
# -----------------------------------------------------------------------------

ARCHITECTURE_INTEGRATION_VALID = (
    len(
        ARCHITECTURE_INTEGRATION_ERRORS
    )
    == 0
)


if not ARCHITECTURE_INTEGRATION_VALID:
    raise ValueError(
        "Notebook 04 architecture integration failed validation: "
        f"{ARCHITECTURE_INTEGRATION_ERRORS}"
    )


# -----------------------------------------------------------------------------
# Notebook 05 handover object
# -----------------------------------------------------------------------------

NOTEBOOK_05_HANDOVER = {
    "handover_version":
        NOTEBOOK_05_HANDOVER_VERSION,

    "created_at_utc":
        NOTEBOOK_05_HANDOVER_CREATED_AT_UTC,

    "source_notebook":
        "04_model_architecture",

    "target_notebook":
        NOTEBOOK_05_NAME,

    "architecture_integration_version":
        ARCHITECTURE_INTEGRATION_VERSION,

    "architecture_valid":
        ARCHITECTURE_INTEGRATION_VALID,

    "architecture_complete":
        True,

    "architecture_sequence":
        list(
            ARCHITECTURE_SEQUENCE
        ),

    "representation_spaces":
        list(
            REPRESENTATION_SPACES
        ),

    "factual_dimensions":
        FACTUAL_REPRESENTATION_DIMENSIONS,

    "psychological_dimensions":
        PSYCHOLOGICAL_REPRESENTATION_DIMENSIONS,

    "psychological_vector_order": [
        "REV",
        "TEV",
        "IEV",
        "EW",
    ],

    "psychological_vector_dimensions": {
        "REV": 11,
        "TEV": 11,
        "IEV": 11,
        "EW": 1,
    },

    "social_components":
        list(
            SOCIAL_REPRESENTATION_COMPONENTS_INTEGRATED
        ),

    "social_fixed_dimensional":
        SOCIAL_REPRESENTATION_FIXED_DIMENSIONAL,

    "global_path_required":
        GLOBAL_PATH_REQUIRED,

    "group_path_optional":
        GROUP_CONDITIONED_PATH_OPTIONAL,

    "group_identities_predefined":
        GROUP_IDENTITIES_PREDEFINED,

    "group_deviation": {
        "signed":
            GROUP_DEVIATION_SIGNED,

        "equals_bias":
            GROUP_DEVIATION_IS_BIAS,

        "equals_manipulation":
            GROUP_DEVIATION_IS_MANIPULATION,

        "equals_polarisation":
            GROUP_DEVIATION_IS_POLARISATION,

        "alignment_required":
            GROUP_DEVIATION_REQUIRES_ALIGNMENT,
    },

    "transformative_weights": {
        "factual":
            INTEGRATED_TW_F,

        "psychological":
            INTEGRATED_TW_P,

        "social":
            INTEGRATED_TW_S,
    },

    "transformative_weights_collapsed":
        TRANSFORMATIVE_WEIGHTS_COLLAPSED,

    "transformative_weights_shared":
        TRANSFORMATIVE_WEIGHTS_SHARED,

    "tw_equals_emotional_weight":
        TRANSFORMATIVE_WEIGHT_IS_EMOTIONAL_WEIGHT,

    "tw_equals_transformation_magnitude":
        TRANSFORMATIVE_WEIGHT_IS_TRANSFORMATION_MAGNITUDE,

    "tw_equals_retention_gate":
        TRANSFORMATIVE_WEIGHT_IS_RETENTION_GATE,

    "tev_equals_architectural_transformation":
        TEV_IS_ARCHITECTURAL_TRANSFORMATION,

    "tev_equals_iev_difference":
        TEV_IS_IEV_DIFFERENCE,

    "recurrent_states": {
        "factual":
            INTEGRATED_H_F,

        "psychological":
            INTEGRATED_H_P,

        "social":
            INTEGRATED_H_S,
    },

    "recurrent_states_collapsed":
        RECURRENT_STATES_COLLAPSED_IN_INTEGRATION,

    "causal_processing":
        RECURRENT_CAUSAL_INTEGRATION,

    "future_context_allowed":
        RECURRENT_FUTURE_CONTEXT_ALLOWED_IN_INTEGRATION,

    "article_boundary_reset":
        RECURRENT_ARTICLE_BOUNDARY_RESET_INTEGRATION,

    "cross_article_carry":
        RECURRENT_CROSS_ARTICLE_CARRY_INTEGRATION,

    "mask_aware_supervision":
        MASK_AWARE_SUPERVISION_REQUIRED,

    "missing_supervision_equals_zero":
        MISSING_SUPERVISION_EQUALS_ZERO,

    "missing_supervision_equals_negative":
        MISSING_SUPERVISION_EQUALS_NEGATIVE,

    "unavailable_supervision_enters_loss":
        UNAVAILABLE_SUPERVISION_CONTRIBUTES_TO_LOSS,

    "default_loss_coefficients":
        DEFAULT_INTEGRATED_LOSS_COEFFICIENTS,

    "random_seed":
        PROJECT_RANDOM_SEED,

    "required_architectural_contracts":
        list(
            NOTEBOOK_05_REQUIRED_ARCHITECTURAL_CONTRACTS
        ),

    "notebook_05_responsibilities":
        list(
            NOTEBOOK_05_RESPONSIBILITIES
        ),

    "planned_ablations":
        list(
            PLANNED_ABLATIONS
        ),

    "required_experiment_configuration":
        list(
            REQUIRED_EXPERIMENT_CONFIGURATION_FIELDS
        ),
}


# -----------------------------------------------------------------------------
# JSON serialisation validation
# -----------------------------------------------------------------------------

try:
    NOTEBOOK_05_HANDOVER_JSON = json.dumps(
        NOTEBOOK_05_HANDOVER,
        indent=2,
        ensure_ascii=False,
        sort_keys=True,
    )

except TypeError as exc:
    raise TypeError(
        "Notebook 05 handover is not JSON serialisable."
    ) from exc


# -----------------------------------------------------------------------------
# Local handover persistence
# -----------------------------------------------------------------------------

NOTEBOOK_05_HANDOVER_LOCAL_PATH.write_text(
    NOTEBOOK_05_HANDOVER_JSON,
    encoding="utf-8",
)


if not NOTEBOOK_05_HANDOVER_LOCAL_PATH.exists():
    raise IOError(
        "Notebook 05 handover could not be written locally."
    )


# -----------------------------------------------------------------------------
# Google Drive service
# -----------------------------------------------------------------------------

def get_media_ai_drive_service():
    """
    Reuse an existing Drive service where available.

    Otherwise construct an authenticated Google Drive API service
    from the current Colab credentials.
    """

    existing_service_names = (
        "drive_service",
        "DRIVE_SERVICE",
        "google_drive_service",
    )


    for service_name in existing_service_names:

        if service_name in globals():

            service = globals()[
                service_name
            ]

            if service is not None:
                return service


    credentials, _ = default(
        scopes=[
            "https://www.googleapis.com/auth/drive"
        ]
    )


    return build(
        "drive",
        "v3",
        credentials=credentials,
        cache_discovery=False,
    )


DRIVE_SERVICE_FOR_HANDOVER = (
    get_media_ai_drive_service()
)


# -----------------------------------------------------------------------------
# Persistent Drive upload/update
# -----------------------------------------------------------------------------

def persist_json_handover_to_drive(
    drive_service,
    filename,
    json_text,
):
    """
    Persist the Notebook 04 → Notebook 05 handover to Google Drive.

    If a file with the exact filename already exists, the most recently
    modified matching file is updated.

    Otherwise a new file is created.

    Returns Drive metadata required for Notebook 05 retrieval.
    """

    escaped_filename = filename.replace(
        "'",
        "\\'",
    )


    query = (
        f"name = '{escaped_filename}' "
        "and trashed = false"
    )


    search_result = (
        drive_service
        .files()
        .list(
            q=query,
            spaces="drive",
            fields=(
                "files("
                "id,"
                "name,"
                "mimeType,"
                "createdTime,"
                "modifiedTime"
                ")"
            ),
            orderBy="modifiedTime desc",
            pageSize=20,
        )
        .execute()
    )


    matching_files = search_result.get(
        "files",
        [],
    )


    media = MediaIoBaseUpload(
        io.BytesIO(
            json_text.encode(
                "utf-8"
            )
        ),
        mimetype="application/json",
        resumable=False,
    )


    if matching_files:

        target_file = matching_files[
            0
        ]

        file_id = target_file[
            "id"
        ]


        persisted_file = (
            drive_service
            .files()
            .update(
                fileId=file_id,
                media_body=media,
                fields=(
                    "id,"
                    "name,"
                    "mimeType,"
                    "createdTime,"
                    "modifiedTime,"
                    "size"
                ),
            )
            .execute()
        )


        operation = "updated"


    else:

        file_metadata = {
            "name":
                filename,

            "mimeType":
                "application/json",
        }


        persisted_file = (
            drive_service
            .files()
            .create(
                body=file_metadata,
                media_body=media,
                fields=(
                    "id,"
                    "name,"
                    "mimeType,"
                    "createdTime,"
                    "modifiedTime,"
                    "size"
                ),
            )
            .execute()
        )


        operation = "created"


    return {
        "operation":
            operation,

        "file_id":
            persisted_file.get(
                "id"
            ),

        "name":
            persisted_file.get(
                "name"
            ),

        "mime_type":
            persisted_file.get(
                "mimeType"
            ),

        "created_time":
            persisted_file.get(
                "createdTime"
            ),

        "modified_time":
            persisted_file.get(
                "modifiedTime"
            ),

        "size":
            persisted_file.get(
                "size"
            ),
    }


NOTEBOOK_05_HANDOVER_DRIVE_METADATA = (
    persist_json_handover_to_drive(
        drive_service=
            DRIVE_SERVICE_FOR_HANDOVER,

        filename=
            NOTEBOOK_05_HANDOVER_FILENAME,

        json_text=
            NOTEBOOK_05_HANDOVER_JSON,
    )
)


# -----------------------------------------------------------------------------
# Persistent handover validation
# -----------------------------------------------------------------------------

NOTEBOOK_05_HANDOVER_PERSISTED = bool(
    NOTEBOOK_05_HANDOVER_DRIVE_METADATA.get(
        "file_id"
    )
)


if not NOTEBOOK_05_HANDOVER_PERSISTED:
    raise IOError(
        "Notebook 05 handover was not persisted successfully "
        "to Google Drive."
    )


NOTEBOOK_05_HANDOVER_DRIVE_FILE_ID = (
    NOTEBOOK_05_HANDOVER_DRIVE_METADATA[
        "file_id"
    ]
)


# -----------------------------------------------------------------------------
# Execution state
# -----------------------------------------------------------------------------

ARCHITECTURE_COMPONENTS_INTEGRATED = True

ARCHITECTURE_CONTRACT_VALIDATED = True

NOTEBOOK_05_HANDOVER_CONSTRUCTED = True

NOTEBOOK_05_HANDOVER_LOCAL_COPY_CREATED = True

NOTEBOOK_05_HANDOVER_DRIVE_PERSISTED = (
    NOTEBOOK_05_HANDOVER_PERSISTED
)

INTEGRATED_MODEL_INSTANTIATED = False

INTEGRATED_FORWARD_PASS_EXECUTED = False

TRANSFORMATIVE_FORWARD_PASS_EXECUTED = False

RECURRENT_FORWARD_PASS_EXECUTED_IN_INTEGRATION = False

OPTIMIZER_CREATED = False

TRAINING_EXECUTED = False

MODEL_CHECKPOINT_CREATED = False

ABLATION_EXECUTED = False

EXPERIMENTAL_EVALUATION_EXECUTED = False

EMPIRICAL_VALIDITY_ESTABLISHED = False


# -----------------------------------------------------------------------------
# Final Notebook 04 validation
# -----------------------------------------------------------------------------

NOTEBOOK_04_COMPLETION_ERRORS = []


if not ARCHITECTURE_COMPONENTS_INTEGRATED:
    NOTEBOOK_04_COMPLETION_ERRORS.append(
        "architecture_components_not_integrated"
    )


if not ARCHITECTURE_CONTRACT_VALIDATED:
    NOTEBOOK_04_COMPLETION_ERRORS.append(
        "architecture_contract_not_validated"
    )


if not NOTEBOOK_05_HANDOVER_CONSTRUCTED:
    NOTEBOOK_04_COMPLETION_ERRORS.append(
        "notebook_05_handover_not_constructed"
    )


if not NOTEBOOK_05_HANDOVER_LOCAL_COPY_CREATED:
    NOTEBOOK_04_COMPLETION_ERRORS.append(
        "local_handover_copy_not_created"
    )


if not NOTEBOOK_05_HANDOVER_DRIVE_PERSISTED:
    NOTEBOOK_04_COMPLETION_ERRORS.append(
        "handover_not_persisted_to_drive"
    )


if INTEGRATED_MODEL_INSTANTIATED:
    NOTEBOOK_04_COMPLETION_ERRORS.append(
        "integrated_model_incorrectly_instantiated"
    )


if INTEGRATED_FORWARD_PASS_EXECUTED:
    NOTEBOOK_04_COMPLETION_ERRORS.append(
        "integrated_forward_pass_incorrectly_executed"
    )


if OPTIMIZER_CREATED:
    NOTEBOOK_04_COMPLETION_ERRORS.append(
        "optimizer_incorrectly_created"
    )


if TRAINING_EXECUTED:
    NOTEBOOK_04_COMPLETION_ERRORS.append(
        "training_incorrectly_executed"
    )


if EMPIRICAL_VALIDITY_ESTABLISHED:
    NOTEBOOK_04_COMPLETION_ERRORS.append(
        "empirical_validity_incorrectly_claimed"
    )


NOTEBOOK_04_ARCHITECTURE_COMPLETE = (
    len(
        NOTEBOOK_04_COMPLETION_ERRORS
    )
    == 0
)


if not NOTEBOOK_04_ARCHITECTURE_COMPLETE:
    raise ValueError(
        "Notebook 04 completion validation failed: "
        f"{NOTEBOOK_04_COMPLETION_ERRORS}"
    )


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print(
    "Media AI — Block 17: Architecture Integration, Validation "
    "and Handover"
)
print("=" * 72)

print(
    f"Integration version         : "
    f"{ARCHITECTURE_INTEGRATION_VERSION}"
)

print(
    f"Representation level        : "
    f"{ARCHITECTURE_INTEGRATION_LEVEL}"
)

print(
    f"Architecture stages         : "
    f"{len(ARCHITECTURE_SEQUENCE)}"
)

print("-" * 72)

print("Architecture sequence")

for stage_index, stage_name in enumerate(
    ARCHITECTURE_SEQUENCE,
    start=1,
):
    print(
        f"Stage {stage_index:<2}                    : "
        f"{stage_name}"
    )

print("-" * 72)

print("Representation spaces")

print(
    f"Representation spaces       : "
    f"{len(REPRESENTATION_SPACES)}"
)

print(
    f"Factual dimensions          : "
    f"{FACTUAL_REPRESENTATION_DIMENSIONS}"
)

print(
    f"Psychological dimensions    : "
    f"{PSYCHOLOGICAL_REPRESENTATION_DIMENSIONS}"
)

print(
    f"Social fixed-dimensional    : "
    f"{SOCIAL_REPRESENTATION_FIXED_DIMENSIONAL}"
)

print(
    f"Social component count      : "
    f"{len(SOCIAL_REPRESENTATION_COMPONENTS_INTEGRATED)}"
)

print("-" * 72)

print("Global and group paths")

print(
    f"Global path required        : "
    f"{GLOBAL_PATH_REQUIRED}"
)

print(
    f"Group path optional         : "
    f"{GROUP_CONDITIONED_PATH_OPTIONAL}"
)

print(
    f"Global replaced by group    : "
    f"{GLOBAL_PATH_REPLACED_BY_GROUP_PATH}"
)

print(
    f"Concrete groups predefined  : "
    f"{GROUP_IDENTITIES_PREDEFINED}"
)

print("-" * 72)

print("Transformative integration")

print(
    f"Transformative Weight count : "
    f"{TRANSFORMATIVE_WEIGHT_COUNT}"
)

print(
    f"Factual Transformative Wt.  : "
    f"{INTEGRATED_TW_F}"
)

print(
    f"Psychological Transform. Wt.: "
    f"{INTEGRATED_TW_P}"
)

print(
    f"Social Transformative Wt.   : "
    f"{INTEGRATED_TW_S}"
)

print(
    f"Collapsed to one TW         : "
    f"{TRANSFORMATIVE_WEIGHTS_COLLAPSED}"
)

print(
    f"TW = Emotional Weight       : "
    f"{TRANSFORMATIVE_WEIGHT_IS_EMOTIONAL_WEIGHT}"
)

print(
    f"TW = retention gate         : "
    f"{TRANSFORMATIVE_WEIGHT_IS_RETENTION_GATE}"
)

print("-" * 72)

print("Recurrent integration")

print(
    f"Recurrent state count       : "
    f"{RECURRENT_STATE_COUNT}"
)

print(
    f"Factual recurrent state     : "
    f"{INTEGRATED_H_F}"
)

print(
    f"Psychological recurrent st. : "
    f"{INTEGRATED_H_P}"
)

print(
    f"Social recurrent state      : "
    f"{INTEGRATED_H_S}"
)

print(
    f"Causal processing           : "
    f"{RECURRENT_CAUSAL_INTEGRATION}"
)

print(
    f"Future context allowed      : "
    f"{RECURRENT_FUTURE_CONTEXT_ALLOWED_IN_INTEGRATION}"
)

print(
    f"Article-boundary reset      : "
    f"{RECURRENT_ARTICLE_BOUNDARY_RESET_INTEGRATION}"
)

print(
    f"Cross-article carry         : "
    f"{RECURRENT_CROSS_ARTICLE_CARRY_INTEGRATION}"
)

print("-" * 72)

print("Supervision and masking")

print(
    f"Mask-aware supervision      : "
    f"{MASK_AWARE_SUPERVISION_REQUIRED}"
)

print(
    f"Missing = zero              : "
    f"{MISSING_SUPERVISION_EQUALS_ZERO}"
)

print(
    f"Missing = negative          : "
    f"{MISSING_SUPERVISION_EQUALS_NEGATIVE}"
)

print(
    f"Unavailable enters loss     : "
    f"{UNAVAILABLE_SUPERVISION_CONTRIBUTES_TO_LOSS}"
)

print("-" * 72)

print("Persistent Notebook 05 handover")

print(
    f"Handover version            : "
    f"{NOTEBOOK_05_HANDOVER_VERSION}"
)

print(
    f"Handover filename           : "
    f"{NOTEBOOK_05_HANDOVER_FILENAME}"
)

print(
    f"Local copy created          : "
    f"{NOTEBOOK_05_HANDOVER_LOCAL_COPY_CREATED}"
)

print(
    f"Drive persisted             : "
    f"{NOTEBOOK_05_HANDOVER_DRIVE_PERSISTED}"
)

print(
    f"Drive operation             : "
    f"{NOTEBOOK_05_HANDOVER_DRIVE_METADATA['operation']}"
)

print(
    f"Drive file ID               : "
    f"{NOTEBOOK_05_HANDOVER_DRIVE_FILE_ID}"
)

print("-" * 72)

print("Training handover")

print(
    f"Parameter control required  : "
    f"{PARAMETER_GROUP_CONTROL_REQUIRED}"
)

print(
    f"Backbone auto-trainable     : "
    f"{BACKBONE_AUTOMATICALLY_TRAINABLE}"
)

print(
    f"Project random seed         : "
    f"{PROJECT_RANDOM_SEED}"
)

print("-" * 72)

print("Execution state")

print(
    f"Components integrated       : "
    f"{ARCHITECTURE_COMPONENTS_INTEGRATED}"
)

print(
    f"Contract validated          : "
    f"{ARCHITECTURE_CONTRACT_VALIDATED}"
)

print(
    f"Handover constructed        : "
    f"{NOTEBOOK_05_HANDOVER_CONSTRUCTED}"
)

print(
    f"Handover persisted          : "
    f"{NOTEBOOK_05_HANDOVER_DRIVE_PERSISTED}"
)

print(
    f"Integrated model instantiated: "
    f"{INTEGRATED_MODEL_INSTANTIATED}"
)

print(
    f"Forward pass executed       : "
    f"{INTEGRATED_FORWARD_PASS_EXECUTED}"
)

print(
    f"Optimizer created           : "
    f"{OPTIMIZER_CREATED}"
)

print(
    f"Training executed           : "
    f"{TRAINING_EXECUTED}"
)

print(
    f"Checkpoint created          : "
    f"{MODEL_CHECKPOINT_CREATED}"
)

print(
    f"Ablation executed           : "
    f"{ABLATION_EXECUTED}"
)

print(
    f"Evaluation executed         : "
    f"{EXPERIMENTAL_EVALUATION_EXECUTED}"
)

print(
    f"Empirical validity          : "
    f"{EMPIRICAL_VALIDITY_ESTABLISHED}"
)

print(
    f"Architecture valid          : "
    f"{ARCHITECTURE_INTEGRATION_VALID}"
)

print(
    f"Notebook 04 complete        : "
    f"{NOTEBOOK_04_ARCHITECTURE_COMPLETE}"
)

print("=" * 72)

print(
    "Notebook 04 architectural integration validated successfully."
)

print(
    "The factual, psychological and social representation spaces "
    "remain independently preserved."
)

print(
    "TW_F, TW_P and TW_S remain independent space-specific "
    "Transformative Weights."
)

print(
    "H_F, H_P and H_S remain separate recurrent states."
)

print(
    "Sequential transformative and recurrent processing remains causal."
)

print(
    "Missing supervision remains distinct from zero and negative "
    "supervision."
)

print(
    "Notebook 05 handover was serialised successfully as JSON."
)

print(
    "A local handover copy was created for audit purposes."
)

print(
    "The Notebook 04 to Notebook 05 handover was persisted "
    "successfully to Google Drive."
)

print(
    "Notebook 05 can therefore initialise in an independent "
    "Colab runtime without rerunning Notebook 04."
)

print(
    "No integrated model has been instantiated."
)

print(
    "No integrated forward pass, optimisation, training, checkpoint, "
    "ablation or experimental evaluation has been executed."
)

print(
    "Architectural validity does not constitute empirical validity."
)

print(
    "Notebook 05 handover constructed and persisted successfully."
)
````
