# Media AI: A Transformative, Recurrent and Confluent Architecture for Interpretable Media Analysis

## Abstract

Media texts do more than transmit factual information. Through selection, association, emphasis, sequencing, emotional language and the construction of social groups, a text may progressively transform how information is interpreted. Existing approaches to sentiment analysis, emotion classification and media-bias detection frequently reduce these processes to categorical labels or aggregate scores. Such representations can obscure how framing develops through a text, how multiple emotions coexist, how social groups are constructed and related, and how factual and emotional information interact. They may also provide limited transparency regarding the textual evidence from which a model's conclusions are derived.

**Media AI** investigates whether these processes can be represented computationally through an interpretable multidimensional architecture built upon a pretrained language model. The pretrained model provides the underlying semantic representation rather than being retrained as the principal object of the study. Media AI introduces additional **transformative, recurrent and confluent layers** designed to model how textual information changes psychological and social representations as an article develops.

The **transformative component** is intended to represent changes associated with linguistic framing rather than merely classify isolated passages. It examines how wording, association, uncertainty, emphasis, evaluation and other contextual signals alter psychological and social states relative to the semantic and factological information contained in the text. This distinction is particularly important for separating the reporting of an emotion from language that may itself introduce, reinforce or attenuate an emotional state.

The **recurrent component** models the sequential development of these transformations. Instead of treating sentences or paragraphs as independent observations, it preserves and updates previous states as subsequent text is processed. This permits the study of cumulative processes in which association may establish a frame, uncertainty may subsequently modify it, and emotions such as fear, reassurance, anger, compassion, tension, hope or relief may strengthen, weaken or change direction over the course of an article. The recurrent representation therefore treats framing as a trajectory rather than a static classification.

The **confluent component** addresses the social structure of language. Media texts may simultaneously refer to several explicit, implicit or overlapping groups, while different expressions, entities and pronouns may converge upon the same social representation. The confluent architecture is designed to construct and update these group representations and their relationships rather than assume a single fixed in-group/out-group distinction. This provides the basis for analysing identification, distancing, inclusion, exclusion, solidarity, hostility, integration and polarisation, and for estimating changes in **Social Polarisation Distance** between dynamically represented groups.

Interpretability is treated as an architectural requirement rather than a post-hoc visualisation problem. Media AI therefore uses an explicit tensor representation in which aggregate psychological or social conclusions remain decomposable into their underlying dimensions. The psychological representation covers a multidimensional emotional spectrum rather than a single positive-negative sentiment axis and permits several emotional states to coexist. Positive and negative emotional and social transformations are treated symmetrically so that reassurance, hope, compassion, affiliation and integration are represented alongside fear, anger, tension, hostility and polarisation rather than being treated merely as the absence of negative framing.

The architecture additionally distinguishes **factological weight** from **emotional weight**. Factological weight represents the relative contribution of externally assessable informational content, including explicit claims, observations, quantities, attribution, evidence, certainty and speculation; it is not interpreted as a probability that a statement is true. Emotional weight represents the strength of emotional content and model-estimated emotional transformation and remains decomposable across the full emotional representation. Their relationship is preserved across textual units, allowing the model to examine how the balance between informational and emotional content develops throughout an article.

These representations are maintained hierarchically across token, sentence, paragraph and article levels. Consequently, an article-level conclusion should remain traceable to the dimensions, passages and linguistic elements contributing to it. A colour-based inspection mechanism provides a visual projection of these numerical representations, enabling individual factological, emotional and social dimensions to be examined directly within the source text. Colours constitute an interpretability mechanism rather than independent evidence and do not replace the underlying numerical representation.

The empirical study is designed around a randomly sampled corpus of full-text English-language media articles. Technical eligibility requirements are applied before random sampling, while topic, sentiment, publisher orientation and anticipated framing characteristics are deliberately excluded as acquisition criteria. This separation is intended to reduce researcher-induced selection effects and to prevent the corpus from being constructed around the phenomena that the proposed architecture is expected to detect.

Model development and evaluation follow the predefined sequence **Research Question → Hypotheses → Method → Experiment → Evidence → Interpretation**. The transformative, recurrent and confluent components are evaluated individually and jointly against simpler baselines through controlled comparisons and ablation experiments. Directional symmetry, sequential dependence, group representation, hierarchical attribution and counterfactual transformations form additional targets of evaluation. An independent language model provides scalable secondary supervision, while human-reviewed observations are retained as an independent reference for selected cases and disagreements.

Throughout the study, model-generated quantities are treated as estimates of operationally defined constructs rather than established psychological, social or factual ground truth. Aggregate outputs are therefore not intended to classify articles, publishers or viewpoints as inherently biased, manipulative, trustworthy or untrustworthy. Instead, the study asks whether an explicitly structured and inspectable computational architecture can provide evidence about how factological, emotional and social representations coexist, interact and transform through media text.

The central research objective is consequently not simply to classify emotion or detect media bias, but to investigate whether the **transformation of interpretation through language can itself be represented as a transparent, sequential and relational computational process**.

## Restricted Google Drive Storage Architecture

This block establishes the persistent storage structure for the Media AI research project using the Google Drive API rather than mounting the user's complete Drive as a filesystem.

The purpose of this design is to apply the principle of least privilege. The notebook should create and manage only the Media AI project folders and files required for the research workflow, without requesting broad filesystem-style access to unrelated Drive content.

The block authenticates the current Colab session, connects to the Google Drive API, and then creates the Media AI project hierarchy if it does not already exist. Folder creation is idempotent: rerunning the block should reuse previously created folders rather than generate duplicate project structures.

The storage hierarchy separates immutable source material from processed datasets, annotations, model artefacts, visual outputs and logs. Raw API responses are preserved independently so that preprocessing and modelling choices can later be changed without downloading the same news articles again.

No news API is contacted in this block and no article data are downloaded.

**Input:** Restricted Google Drive authorisation.

**Output:** Dedicated Media AI project folders in Google Drive and a dictionary of Drive folder identifiers used by subsequent notebook blocks.

```python
# =============================================================================
# Media AI — Restricted Google Drive Storage Architecture
# =============================================================================

from google.colab import auth
from googleapiclient.discovery import build


# -----------------------------------------------------------------------------
# Authenticate and connect to Google Drive
# -----------------------------------------------------------------------------

auth.authenticate_user()
drive_service = build("drive", "v3")


# -----------------------------------------------------------------------------
# Helper: find or create one folder inside a parent folder
# -----------------------------------------------------------------------------

def get_or_create_folder(folder_name, parent_id=None):
    """
    Find an existing Google Drive folder by name within the specified parent.
    If it does not exist, create it.

    Returns
    -------
    str
        Google Drive folder ID.
    """

    query_parts = [
        f"name = '{folder_name}'",
        "mimeType = 'application/vnd.google-apps.folder'",
        "trashed = false",
    ]

    if parent_id:
        query_parts.append(f"'{parent_id}' in parents")

    query = " and ".join(query_parts)

    response = (
        drive_service.files()
        .list(
            q=query,
            spaces="drive",
            fields="files(id, name)",
            pageSize=10,
        )
        .execute()
    )

    folders = response.get("files", [])

    if folders:
        return folders[0]["id"]

    metadata = {
        "name": folder_name,
        "mimeType": "application/vnd.google-apps.folder",
    }

    if parent_id:
        metadata["parents"] = [parent_id]

    folder = (
        drive_service.files()
        .create(
            body=metadata,
            fields="id, name",
        )
        .execute()
    )

    return folder["id"]


# -----------------------------------------------------------------------------
# Create project hierarchy
# -----------------------------------------------------------------------------

PROJECT_FOLDER_ID = get_or_create_folder("Media_AI")

DATA_FOLDER_ID = get_or_create_folder("data", PROJECT_FOLDER_ID)

RAW_API_FOLDER_ID = get_or_create_folder("raw_api", DATA_FOLDER_ID)
FULL_TEXT_FOLDER_ID = get_or_create_folder("full_text", DATA_FOLDER_ID)
CLEANED_FOLDER_ID = get_or_create_folder("cleaned", DATA_FOLDER_ID)
SEGMENTED_FOLDER_ID = get_or_create_folder("segmented", DATA_FOLDER_ID)
ANNOTATED_FOLDER_ID = get_or_create_folder("annotated", DATA_FOLDER_ID)
GOLD_FOLDER_ID = get_or_create_folder("gold", DATA_FOLDER_ID)
DEMO_FOLDER_ID = get_or_create_folder("demo", DATA_FOLDER_ID)

MODELS_FOLDER_ID = get_or_create_folder("models", PROJECT_FOLDER_ID)
CHECKPOINTS_FOLDER_ID = get_or_create_folder(
    "checkpoints", MODELS_FOLDER_ID
)
ADAPTERS_FOLDER_ID = get_or_create_folder(
    "adapters", MODELS_FOLDER_ID
)

OUTPUTS_FOLDER_ID = get_or_create_folder("outputs", PROJECT_FOLDER_ID)
METRICS_FOLDER_ID = get_or_create_folder(
    "metrics", OUTPUTS_FOLDER_ID
)
COLOURS_FOLDER_ID = get_or_create_folder(
    "colours", OUTPUTS_FOLDER_ID
)
EXPLANATIONS_FOLDER_ID = get_or_create_folder(
    "explanations", OUTPUTS_FOLDER_ID
)

LOGS_FOLDER_ID = get_or_create_folder("logs", PROJECT_FOLDER_ID)


# -----------------------------------------------------------------------------
# Canonical Drive folder registry
# -----------------------------------------------------------------------------

DRIVE_FOLDERS = {
    "project": PROJECT_FOLDER_ID,
    "data": DATA_FOLDER_ID,
    "raw_api": RAW_API_FOLDER_ID,
    "full_text": FULL_TEXT_FOLDER_ID,
    "cleaned": CLEANED_FOLDER_ID,
    "segmented": SEGMENTED_FOLDER_ID,
    "annotated": ANNOTATED_FOLDER_ID,
    "gold": GOLD_FOLDER_ID,
    "demo": DEMO_FOLDER_ID,
    "models": MODELS_FOLDER_ID,
    "checkpoints": CHECKPOINTS_FOLDER_ID,
    "adapters": ADAPTERS_FOLDER_ID,
    "outputs": OUTPUTS_FOLDER_ID,
    "metrics": METRICS_FOLDER_ID,
    "colours": COLOURS_FOLDER_ID,
    "explanations": EXPLANATIONS_FOLDER_ID,
    "logs": LOGS_FOLDER_ID,
}


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Restricted Google Drive Storage")
print("=" * 72)

for name, folder_id in DRIVE_FOLDERS.items():
    print(f"{name:<15}: OK   {folder_id}")

print("=" * 72)
print("Media AI storage architecture initialised successfully.")
print("No Google Drive filesystem mount was used.")
print("No news API requests have been made.")
```

## GNews Configuration, Credential Isolation and Acquisition Safeguards

This block defines the configuration required for controlled access to the
GNews API. It does not contact the API and does not download article data.

The API credential is read from Google Colab Secrets rather than written
directly into the notebook. This prevents the credential from being exposed
when the notebook is shared through GitHub or with other researchers.

Data acquisition is disabled by default through the `DOWNLOAD_NEW_DATA`
master switch. Consequently, opening the notebook or executing all cells does
not initiate news acquisition unless this setting is deliberately changed.

Three independent hard limits are defined for the pilot acquisition process:

- maximum API requests permitted within one day;
- maximum articles requested in a single API call;
- maximum articles acquired within one day.

The limits are intentionally set below the provider allowance. They function as
research-side safeguards rather than estimates of the provider's own quota.

The current pilot configuration uses no more than 20% of the approved request
and per-request article capacity and imposes an additional conservative daily
article ceiling. These limits will be checked again immediately before every
future API request.

The API key itself is never printed, saved to the research corpus, or included
in notebook outputs. The block reports only whether the expected secret is
available.

**Input:** `GNEWS_API_KEY` stored in Google Colab Secrets.

**Output:** Secure API configuration, disabled-by-default acquisition switch,
and hard-coded pilot usage limits.

No API request is performed by this block.

## Persistent API Usage Ledger and Pre-Request Safety Gate

This block establishes the persistent accounting and safety mechanism that will
govern all future GNews API requests made by the Media AI acquisition notebook.

The notebook must not rely on temporary Colab memory for API usage tracking,
because a runtime restart would otherwise reset local counters and could allow
the same daily research limits to be exceeded unintentionally. Usage records are
therefore stored persistently in the dedicated Media AI Google Drive structure
created in Block 2.

The ledger records the UTC date, number of API requests made, number of articles
requested, number of articles successfully returned, and the timestamp of the
most recent update. The UTC date is used because the provider's daily quota is
defined on a UTC basis.

Before any future API request is sent, a safety function checks the proposed
request against three independent research limits defined in Block 3:

- maximum API requests per day;
- maximum articles requested per API call;
- maximum articles acquired per day.

The request is permitted only when all relevant limits remain satisfied. The
safety gate also requires the `DOWNLOAD_NEW_DATA` master switch to be explicitly
enabled and requires an API credential to be present.

This block does not test whether the current API credential is valid. Credential
validity will be checked separately when the first controlled connection test is
performed. A non-empty placeholder therefore does not authorise data acquisition.

No GNews endpoint is contacted and no news articles are downloaded by this
block.

**Input:** Storage folder identifiers from Block 2 and acquisition limits from
Block 3.

**Output:** Persistent daily usage ledger and reusable pre-request safety
functions for all later acquisition operations.

```python
# =============================================================================
# Media AI — Persistent API Usage Ledger and Pre-Request Safety Gate
# =============================================================================

import io
import json
from datetime import datetime, timezone

from googleapiclient.http import MediaIoBaseDownload, MediaInMemoryUpload


# -----------------------------------------------------------------------------
# Ledger configuration
# -----------------------------------------------------------------------------

USAGE_LEDGER_FILENAME = "gnews_usage_ledger.json"


# -----------------------------------------------------------------------------
# Google Drive helper functions
# -----------------------------------------------------------------------------

def find_file_in_folder(file_name, parent_id):
    """
    Return the first matching file ID within a specific Drive folder.
    Returns None when the file does not exist.
    """

    safe_name = file_name.replace("'", "\\'")

    query = (
        f"name = '{safe_name}' and "
        f"'{parent_id}' in parents and "
        "trashed = false"
    )

    response = (
        drive_service.files()
        .list(
            q=query,
            spaces="drive",
            fields="files(id, name)",
            pageSize=10,
        )
        .execute()
    )

    files = response.get("files", [])
    return files[0]["id"] if files else None


def download_json_file(file_id):
    """
    Download a JSON file from Google Drive and return it as a Python object.
    """

    request = drive_service.files().get_media(fileId=file_id)
    buffer = io.BytesIO()

    downloader = MediaIoBaseDownload(buffer, request)

    done = False
    while not done:
        _, done = downloader.next_chunk()

    buffer.seek(0)
    return json.loads(buffer.read().decode("utf-8"))


def save_json_file(file_name, parent_id, payload):
    """
    Create or update a JSON file within a specific Google Drive folder.
    """

    encoded = json.dumps(
        payload,
        indent=2,
        ensure_ascii=False,
    ).encode("utf-8")

    media = MediaInMemoryUpload(
        encoded,
        mimetype="application/json",
        resumable=False,
    )

    existing_file_id = find_file_in_folder(
        file_name,
        parent_id,
    )

    if existing_file_id:
        drive_service.files().update(
            fileId=existing_file_id,
            media_body=media,
        ).execute()

        return existing_file_id

    metadata = {
        "name": file_name,
        "parents": [parent_id],
    }

    created = (
        drive_service.files()
        .create(
            body=metadata,
            media_body=media,
            fields="id",
        )
        .execute()
    )

    return created["id"]


# -----------------------------------------------------------------------------
# Daily usage ledger
# -----------------------------------------------------------------------------

def current_utc_date():
    return datetime.now(timezone.utc).date().isoformat()


def new_daily_record():
    """
    Create a fresh zero-usage record for the current UTC day.
    """

    now = datetime.now(timezone.utc)

    return {
        "date_utc": now.date().isoformat(),
        "requests_used": 0,
        "articles_requested": 0,
        "articles_returned": 0,
        "last_updated_utc": now.isoformat(),
    }


def load_usage_ledger():
    """
    Load the persistent GNews usage ledger.

    If no ledger exists, or if the stored record belongs to a previous UTC day,
    a fresh daily record is created and saved.
    """

    file_id = find_file_in_folder(
        USAGE_LEDGER_FILENAME,
        LOGS_FOLDER_ID,
    )

    if file_id:
        ledger = download_json_file(file_id)
    else:
        ledger = new_daily_record()

    if ledger.get("date_utc") != current_utc_date():
        ledger = new_daily_record()

    save_json_file(
        USAGE_LEDGER_FILENAME,
        LOGS_FOLDER_ID,
        ledger,
    )

    return ledger


def save_usage_ledger(ledger):
    """
    Persist the current usage record.
    """

    ledger["last_updated_utc"] = datetime.now(
        timezone.utc
    ).isoformat()

    save_json_file(
        USAGE_LEDGER_FILENAME,
        LOGS_FOLDER_ID,
        ledger,
    )


# -----------------------------------------------------------------------------
# Pre-request safety gate
# -----------------------------------------------------------------------------

def check_request_allowed(
    requested_articles,
    require_download_enabled=True,
):
    """
    Validate a proposed API request before it is sent.

    Returns
    -------
    tuple
        (allowed: bool, reason: str, ledger: dict)
    """

    ledger = load_usage_ledger()

    if require_download_enabled and not DOWNLOAD_NEW_DATA:
        return (
            False,
            "DOWNLOAD_NEW_DATA is False.",
            ledger,
        )

    if not GNEWS_API_KEY:
        return (
            False,
            "GNEWS_API_KEY is not available.",
            ledger,
        )

    if requested_articles < 1:
        return (
            False,
            "At least one article must be requested.",
            ledger,
        )

    if requested_articles > MAX_ARTICLES_PER_REQUEST:
        return (
            False,
            "Requested articles exceed the per-request research limit.",
            ledger,
        )

    if ledger["requests_used"] + 1 > MAX_REQUESTS_PER_DAY:
        return (
            False,
            "Daily API request limit would be exceeded.",
            ledger,
        )

    if (
        ledger["articles_returned"] + requested_articles
        > MAX_ARTICLES_PER_DAY
    ):
        return (
            False,
            "Daily article acquisition limit would be exceeded.",
            ledger,
        )

    return (
        True,
        "Request satisfies all current research limits.",
        ledger,
    )


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

usage = load_usage_ledger()

print("=" * 72)
print("Media AI — GNews Persistent Usage Control")
print("=" * 72)

print(f"UTC date                 : {usage['date_utc']}")
print(f"Requests used            : {usage['requests_used']}")
print(f"Articles requested       : {usage['articles_requested']}")
print(f"Articles returned        : {usage['articles_returned']}")
print(f"Requests remaining       : {MAX_REQUESTS_PER_DAY - usage['requests_used']}")
print(
    f"Article capacity remaining: "
    f"{MAX_ARTICLES_PER_DAY - usage['articles_returned']}"
)

print("=" * 72)

allowed, reason, _ = check_request_allowed(
    requested_articles=MAX_ARTICLES_PER_REQUEST
)

print(f"Current acquisition gate : {'OPEN' if allowed else 'CLOSED'}")
print(f"Reason                   : {reason}")

print("=" * 72)
print("Persistent usage ledger initialised successfully.")
print("No GNews API requests have been made.")
```

## Research Sampling Framework and Article Eligibility Criteria

This block defines the sampling methodology used to construct the Media AI
research corpus. The criteria are specified before article acquisition so that
selection decisions are not influenced by the framing characteristics,
emotional content, publisher identity, or topics observed in the retrieved
material.

The pilot follows the principle:

**random by content, selective only by technical usability.**

Articles are therefore not selected according to political orientation,
publisher reputation, topic, sentiment, emotional intensity, anticipated
psychological framing, anticipated social framing, or expected agreement with
the hypotheses of the study.

Technical eligibility is restricted to properties required for the proposed
architecture to operate meaningfully.

An article is eligible for the pilot corpus when:

- its language is English;
- full article text is available;
- the available text contains at least 300 words;
- the record is not an exact duplicate of an already retained article;
- the record is not identified as a strong near-duplicate during subsequent
  deduplication;
- sufficient provenance information is available to identify its source and
  acquisition record.

The minimum-length requirement is methodological rather than editorial.
Media AI is designed to examine sequential transformations across textual
context through transformative, recurrent and confluent processing. Very short
texts provide insufficient sequential structure for evaluating the development
of psychological and social states across an article.

No maximum article length is imposed at the acquisition stage. Long articles
are retained so that article-length effects can later be examined rather than
removed by sampling design. Token-length constraints imposed by the selected
language model will be handled during data preparation rather than during
source acquisition.

Random selection is applied only after technical eligibility has been
established. A fixed random seed is used so that sampling decisions can be
reproduced where the underlying candidate set is unchanged.

Duplicate handling is deliberately separated into two stages. Exact duplicates
can be identified immediately through normalised textual hashing. More complex
near-duplicate detection, including syndicated or lightly rewritten versions of
the same article, will be performed in the data-preparation notebook where
appropriate similarity methods can be evaluated explicitly.

The acquisition pipeline records both accepted and rejected observations,
together with the reason for exclusion. This permits the final corpus to be
audited and allows potential sampling effects introduced by technical
eligibility requirements to be examined later.

These rules define corpus eligibility only. They do not constitute judgements
about factual accuracy, journalistic quality, ideological orientation, or
framing.

**Input:** Article records returned by the future GNews acquisition block.

**Output:** Reusable eligibility settings, text-normalisation utilities,
exact-duplicate identifiers, and article-level technical eligibility checks.

No GNews API request is performed by this block.

```python
# =============================================================================
# Media AI — Research Sampling and Article Eligibility Framework
# =============================================================================

import hashlib
import random
import re
from typing import Any, Dict, Optional


# -----------------------------------------------------------------------------
# Sampling configuration
# -----------------------------------------------------------------------------

TARGET_LANGUAGE = "en"
MIN_ARTICLE_WORDS = 300

# No acquisition-stage maximum length is imposed.
MAX_ARTICLE_WORDS = None

# Sampling remains content-neutral.
FILTER_BY_TOPIC = False
FILTER_BY_SENTIMENT = False
FILTER_BY_PUBLISHER = False
FILTER_BY_POLITICAL_ORIENTATION = False
FILTER_BY_EXPECTED_FRAMING = False

# Reuse the project-wide reproducibility seed defined in Block 1.
SAMPLING_RANDOM_SEED = RANDOM_SEED

sampling_rng = random.Random(SAMPLING_RANDOM_SEED)


# -----------------------------------------------------------------------------
# Text normalisation
# -----------------------------------------------------------------------------

def normalise_article_text(text: Optional[str]) -> str:
    """
    Normalise article text for technical eligibility and exact-duplicate checks.

    The function deliberately performs only conservative transformations:
    - convert non-string values to an empty string;
    - normalise whitespace;
    - remove leading/trailing whitespace.

    Semantic content, punctuation and wording are otherwise preserved.
    """

    if not isinstance(text, str):
        return ""

    text = re.sub(r"\s+", " ", text).strip()
    return text


def count_words(text: Optional[str]) -> int:
    """
    Return a simple whitespace-based word count after text normalisation.
    """

    normalised = normalise_article_text(text)

    if not normalised:
        return 0

    return len(normalised.split())


# -----------------------------------------------------------------------------
# Exact-duplicate fingerprint
# -----------------------------------------------------------------------------

def create_text_fingerprint(text: Optional[str]) -> Optional[str]:
    """
    Create a deterministic SHA-256 fingerprint of normalised article text.

    This fingerprint is intended only for exact-duplicate detection.
    Near-duplicate detection is deferred to the data-preparation notebook.
    """

    normalised = normalise_article_text(text)

    if not normalised:
        return None

    return hashlib.sha256(
        normalised.encode("utf-8")
    ).hexdigest()


# -----------------------------------------------------------------------------
# Technical eligibility assessment
# -----------------------------------------------------------------------------

def assess_article_eligibility(
    article: Dict[str, Any],
) -> Dict[str, Any]:
    """
    Assess whether an article satisfies the pilot's technical eligibility rules.

    Expected fields are intentionally generic so the function can later be
    adapted to the final provider response without changing the research rules.

    Returns
    -------
    dict
        Technical eligibility status, measured properties and exclusion reason.
    """

    language = article.get("language")
    full_text = (
        article.get("content")
        or article.get("full_text")
        or article.get("text")
        or ""
    )

    source = article.get("source")
    url = article.get("url")

    normalised_text = normalise_article_text(full_text)
    word_count = count_words(normalised_text)
    fingerprint = create_text_fingerprint(normalised_text)

    exclusion_reasons = []

    # English-language requirement.
    if language and language.lower() != TARGET_LANGUAGE:
        exclusion_reasons.append("non_english")

    # Full text must be present.
    if not normalised_text:
        exclusion_reasons.append("missing_full_text")

    # Minimum sequential context requirement.
    if normalised_text and word_count < MIN_ARTICLE_WORDS:
        exclusion_reasons.append("below_minimum_word_count")

    # Basic provenance requirement.
    if not source and not url:
        exclusion_reasons.append("insufficient_provenance")

    eligible = len(exclusion_reasons) == 0

    return {
        "eligible": eligible,
        "language": language,
        "word_count": word_count,
        "text_fingerprint": fingerprint,
        "source_available": bool(source),
        "url_available": bool(url),
        "exclusion_reasons": exclusion_reasons,
    }


# -----------------------------------------------------------------------------
# Exact-duplicate check
# -----------------------------------------------------------------------------

def is_exact_duplicate(
    fingerprint: Optional[str],
    existing_fingerprints: set,
) -> bool:
    """
    Determine whether an article fingerprint already exists in the corpus.
    """

    if fingerprint is None:
        return False

    return fingerprint in existing_fingerprints


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Research Sampling Framework")
print("=" * 72)

print(f"Target language                  : {TARGET_LANGUAGE}")
print(f"Minimum article length           : {MIN_ARTICLE_WORDS} words")
print(f"Maximum article length           : {MAX_ARTICLE_WORDS}")
print(f"Fixed sampling seed              : {SAMPLING_RANDOM_SEED}")
print(f"Filter by topic                  : {FILTER_BY_TOPIC}")
print(f"Filter by sentiment              : {FILTER_BY_SENTIMENT}")
print(f"Filter by publisher              : {FILTER_BY_PUBLISHER}")
print(
    f"Filter by political orientation  : "
    f"{FILTER_BY_POLITICAL_ORIENTATION}"
)
print(
    f"Filter by expected framing       : "
    f"{FILTER_BY_EXPECTED_FRAMING}"
)

print("=" * 72)
print("Technical eligibility framework initialised successfully.")
print("Near-duplicate detection is deferred to data preparation.")
print("No GNews API requests have been made.")
```

## Original Source Record Schema and Source Traceability

This block defines the canonical record structure used to preserve GNews
articles acquired for the Media AI research corpus.

A central methodological requirement is the separation of **original source
data** from **derived research data**. Information returned by GNews is therefore
preserved as an unaltered provider record before cleaning, segmentation,
annotation, tensor construction, model supervision, or other analytical
transformations are performed.

The observed GNews article schema contains the fields `content`, `description`,
`id`, `image`, `lang`, `publishedAt`, `source`, `title`, and `url`. These
provider-specific fields are retained inside the preserved source payload while
selected fields are additionally mapped into a standard Media AI acquisition
schema for reproducibility and downstream processing.

This separation establishes a traceable provenance chain:

**external source → acquisition record → derived dataset → model input →
model output → interpretation**

Each acquisition record receives an internal Media AI identifier independent of
the provider's identifier. The record preserves the GNews article identifier,
publication timestamp, source metadata, URL, language, acquisition timestamp,
word count, exact-text fingerprint, technical eligibility assessment, and the
version of the acquisition schema used to construct the record.

The textual fingerprint is used only for exact-duplicate detection. It does not
represent semantic similarity and does not identify syndicated, paraphrased, or
otherwise near-duplicate reporting. More complex deduplication is deferred to
the data-preparation stage.

Technical eligibility is assessed using the predefined acquisition criteria:
English-language content, available full text, a minimum length of 300 words,
and sufficient provenance information. These criteria determine technical
usability only and do not constitute judgements about factual accuracy,
journalistic quality, publisher orientation, sentiment, or framing.

Analytical quantities such as factological weight, emotional weight, emotional
classifications, transformative state, recurrent state, confluent group
representations, Social Polarisation Distance, supervisory labels, and model
confidence are deliberately excluded from the source schema. Such quantities
belong to derived datasets and must never overwrite the original acquisition
record.

The complete GNews provider payload is retained as a separate nested object.
This preserves the exact article record received from the provider and allows
future preprocessing or analytical methods to be revised without altering the
source evidence.

**Input:** GNews article records.

**Output:** Versioned Media AI source records containing provider provenance,
technical eligibility information, reproducibility metadata, and the preserved
original GNews payload.

No GNews API request is performed by this block.

```python
# =============================================================================
# Media AI — Original Source Record Schema and Acquisition Provenance
# =============================================================================

import copy
import uuid
from datetime import datetime, timezone
from typing import Any, Dict


# -----------------------------------------------------------------------------
# Schema configuration
# -----------------------------------------------------------------------------

SOURCE_RECORD_SCHEMA_VERSION = "1.1"

SOURCE_PROVIDER = "GNews"
ACQUISITION_METHOD = "api"


# -----------------------------------------------------------------------------
# Required Media AI source-record fields
# -----------------------------------------------------------------------------

REQUIRED_SOURCE_RECORD_FIELDS = {
    "media_ai_record_id",
    "schema_version",
    "provider",
    "provider_article_id",
    "acquisition_method",
    "acquired_at_utc",
    "published_at",
    "source",
    "source_url",
    "language",
    "title",
    "word_count",
    "text_fingerprint",
    "technical_eligibility",
    "acquisition_context",
    "provider_payload",
}


# -----------------------------------------------------------------------------
# Source-record constructor
# -----------------------------------------------------------------------------

def create_source_record(
    provider_payload: Dict[str, Any],
) -> Dict[str, Any]:
    """
    Construct a canonical Media AI source record from an observed GNews
    article payload.

    The provider payload is deep-copied so that later transformations cannot
    modify the preserved source representation through shared Python references.
    """

    if not isinstance(provider_payload, dict):
        raise TypeError("provider_payload must be a dictionary.")

    preserved_payload = copy.deepcopy(provider_payload)

    # -------------------------------------------------------------------------
    # Observed GNews fields
    # -------------------------------------------------------------------------

    article_text = preserved_payload.get("content") or ""

    provider_article_id = preserved_payload.get("id")
    article_language = preserved_payload.get("lang")
    published_at = preserved_payload.get("publishedAt")

    source = preserved_payload.get("source")
    source_url = preserved_payload.get("url")
    title = preserved_payload.get("title")

    # -------------------------------------------------------------------------
    # Technical eligibility assessment
    # -------------------------------------------------------------------------

    eligibility_input = {
        **preserved_payload,
        "language": article_language,
        "content": article_text,
    }

    eligibility = assess_article_eligibility(
        eligibility_input
    )

    now = datetime.now(timezone.utc)

    # -------------------------------------------------------------------------
    # Canonical Media AI source record
    # -------------------------------------------------------------------------

    record = {
        # Internal identity
        "media_ai_record_id": str(uuid.uuid4()),

        # Schema and acquisition provenance
        "schema_version": SOURCE_RECORD_SCHEMA_VERSION,
        "provider": SOURCE_PROVIDER,
        "provider_article_id": provider_article_id,
        "acquisition_method": ACQUISITION_METHOD,
        "acquired_at_utc": now.isoformat(),

        # Provider-derived descriptors
        "published_at": published_at,
        "source": source,
        "source_url": source_url,
        "language": article_language,
        "title": title,

        # Technical textual descriptors
        "word_count": count_words(article_text),
        "text_fingerprint": create_text_fingerprint(
            article_text
        ),

        # Technical eligibility
        "technical_eligibility": eligibility,

        # Reproducibility context
        "acquisition_context": {
            "project_name": PROJECT_NAME,
            "project_version": PROJECT_VERSION,
            "random_seed": RANDOM_SEED,
            "sampling_seed": SAMPLING_RANDOM_SEED,
            "target_language": TARGET_LANGUAGE,
            "minimum_article_words": MIN_ARTICLE_WORDS,
            "provider": SOURCE_PROVIDER,
            "schema_version": SOURCE_RECORD_SCHEMA_VERSION,
        },

        # Preserved GNews record
        "provider_payload": preserved_payload,
    }

    return record


# -----------------------------------------------------------------------------
# Schema validation
# -----------------------------------------------------------------------------

def validate_source_record(
    record: Dict[str, Any],
) -> Dict[str, Any]:
    """
    Validate structural integrity of a Media AI source record.

    This validates schema completeness only. It does not evaluate factual
    accuracy, journalistic quality, emotional content, or framing.
    """

    if not isinstance(record, dict):
        return {
            "valid": False,
            "missing_fields": [],
            "errors": ["record_is_not_dictionary"],
        }

    missing_fields = sorted(
        REQUIRED_SOURCE_RECORD_FIELDS
        - set(record.keys())
    )

    errors = []

    if missing_fields:
        errors.append("missing_required_fields")

    if (
        record.get("schema_version")
        != SOURCE_RECORD_SCHEMA_VERSION
    ):
        errors.append("unexpected_schema_version")

    if record.get("provider") != SOURCE_PROVIDER:
        errors.append("unexpected_provider")

    if not isinstance(
        record.get("provider_payload"),
        dict,
    ):
        errors.append("provider_payload_not_dictionary")

    return {
        "valid": len(errors) == 0,
        "missing_fields": missing_fields,
        "errors": errors,
    }


# -----------------------------------------------------------------------------
# Synthetic validation using the observed GNews schema
# -----------------------------------------------------------------------------

synthetic_provider_payload = {
    "id": "synthetic-gnews-record",
    "title": "Synthetic Media AI Research Article",
    "description": (
        "Synthetic record used only to validate the acquisition schema."
    ),
    "content": " ".join(
        ["research"] * MIN_ARTICLE_WORDS
    ),
    "url": "https://example.invalid/media-ai-test",
    "image": None,
    "publishedAt": "2026-08-09T00:00:00Z",
    "lang": "en",
    "source": {
        "name": "Synthetic Research Source",
        "url": "https://example.invalid"
    },
}


synthetic_record = create_source_record(
    synthetic_provider_payload
)

schema_test = validate_source_record(
    synthetic_record
)


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Original Source Record Schema")
print("=" * 72)

print(
    f"Schema version             : "
    f"{SOURCE_RECORD_SCHEMA_VERSION}"
)
print(
    f"Provider                   : "
    f"{SOURCE_PROVIDER}"
)
print(
    f"Required fields            : "
    f"{len(REQUIRED_SOURCE_RECORD_FIELDS)}"
)
print(
    f"Synthetic schema valid     : "
    f"{schema_test['valid']}"
)
print(
    f"Synthetic article eligible : "
    f"{synthetic_record['technical_eligibility']['eligible']}"
)
print(
    f"Synthetic word count       : "
    f"{synthetic_record['word_count']}"
)

print("=" * 72)
print("Original source record schema initialised successfully.")
print("Observed GNews fields are mapped explicitly.")
print("Synthetic data only were used for validation.")
print("No GNews API requests have been made.")
```

## Controlled GNews Connection and Response Validation

This block defines the first live interaction with the GNews API. Its purpose is
not to acquire the research corpus, but to verify that the API credential,
endpoint configuration, provider response structure, and previously defined
safety controls operate as expected.

The connection test is deliberately isolated from bulk acquisition. It requests
only a minimal number of articles and is executed only when both conditions are
met:

- a non-placeholder API credential is available;
- the dedicated `RUN_CONNECTION_TEST` switch is explicitly enabled.

Before any request is sent, the pre-request safety gate defined in Block 4 is
called. The test therefore remains subject to the same daily request, per-request
article, and daily article limits as all later acquisition operations.

The response is validated structurally rather than interpreted substantively.
The block checks whether the provider returns the expected top-level article
collection and inspects the field structure of the first returned record. This
step is necessary because the exact provider payload must be observed before the
canonical source-record constructor can be mapped confidently to production data.

No article retrieved by this block is treated as part of the research corpus.
The connection test exists solely to validate technical integration. If a request
is successfully executed, the persistent usage ledger is updated immediately so
that even test traffic remains included in the research-side API accounting.

The API key itself is never printed. Error messages are reported without exposing
credentials or request URLs containing the key.

**Input:** GNews credential, safety configuration, and persistent usage ledger
from earlier blocks.

**Output:** Connection status, structural response summary, first-record field
inventory, and updated API usage ledger if a live request is executed.

Bulk article acquisition is not performed by this block.

```python
# =============================================================================
# Media AI — Controlled GNews Connection and Response Validation
# =============================================================================

import requests
from datetime import datetime, timezone


# -----------------------------------------------------------------------------
# Connection-test configuration
# -----------------------------------------------------------------------------

GNEWS_BASE_URL = "https://gnews.io/api/v4"

# Explicitly disabled by default.
RUN_CONNECTION_TEST = True

# Minimal test request.
CONNECTION_TEST_ARTICLES = 1

# A placeholder value can be used during notebook development,
# but it must never authorise a live request.
PLACEHOLDER_API_KEYS = {
    "",
    "placeholder",
    "your-api-key",
    "your_api_key",
    "test",
}


# -----------------------------------------------------------------------------
# Credential readiness check
# -----------------------------------------------------------------------------

def is_real_api_key(api_key):
    """
    Return True only when the credential is non-empty and not a known placeholder.
    """

    if not isinstance(api_key, str):
        return False

    cleaned = api_key.strip()

    if not cleaned:
        return False

    return cleaned.lower() not in PLACEHOLDER_API_KEYS


# -----------------------------------------------------------------------------
# Usage-ledger update after a successful live request
# -----------------------------------------------------------------------------

def register_api_request(
    ledger,
    articles_requested,
    articles_returned,
):
    """
    Update and persist the daily GNews usage ledger after a completed request.
    """

    ledger["requests_used"] += 1
    ledger["articles_requested"] += int(articles_requested)
    ledger["articles_returned"] += int(articles_returned)

    save_usage_ledger(ledger)


# -----------------------------------------------------------------------------
# Controlled connection test
# -----------------------------------------------------------------------------

def run_gnews_connection_test():
    """
    Perform one minimal GNews request after all safety conditions are satisfied.
    """

    if not RUN_CONNECTION_TEST:
        return {
            "executed": False,
            "status": "SKIPPED",
            "reason": "RUN_CONNECTION_TEST is False.",
        }

    if not is_real_api_key(GNEWS_API_KEY):
        return {
            "executed": False,
            "status": "BLOCKED",
            "reason": "A valid non-placeholder API key is not available.",
        }

    allowed, reason, ledger = check_request_allowed(
        requested_articles=CONNECTION_TEST_ARTICLES,
        require_download_enabled=False,
    )

    if not allowed:
        return {
            "executed": False,
            "status": "BLOCKED",
            "reason": reason,
        }

    endpoint = f"{GNEWS_BASE_URL}/top-headlines"

    params = {
        "lang": TARGET_LANGUAGE,
        "max": CONNECTION_TEST_ARTICLES,
        "apikey": GNEWS_API_KEY,
    }

    try:
        response = requests.get(
            endpoint,
            params=params,
            timeout=30,
        )

        response.raise_for_status()
        payload = response.json()

    except requests.RequestException as exc:
        return {
            "executed": True,
            "status": "ERROR",
            "reason": str(exc),
        }

    except ValueError:
        return {
            "executed": True,
            "status": "ERROR",
            "reason": "Provider response was not valid JSON.",
        }

    articles = payload.get("articles", [])

    if not isinstance(articles, list):
        return {
            "executed": True,
            "status": "ERROR",
            "reason": "Expected 'articles' collection was not returned.",
        }

    register_api_request(
        ledger=ledger,
        articles_requested=CONNECTION_TEST_ARTICLES,
        articles_returned=len(articles),
    )

    first_article_fields = (
        sorted(articles[0].keys())
        if articles
        else []
    )

    return {
        "executed": True,
        "status": "SUCCESS",
        "reason": "Controlled GNews connection succeeded.",
        "articles_returned": len(articles),
        "top_level_fields": sorted(payload.keys()),
        "first_article_fields": first_article_fields,
        "tested_at_utc": datetime.now(
            timezone.utc
        ).isoformat(),
    }


# -----------------------------------------------------------------------------
# Execute or safely skip
# -----------------------------------------------------------------------------

connection_result = run_gnews_connection_test()


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Controlled GNews Connection Test")
print("=" * 72)

print(f"Test status              : {connection_result['status']}")
print(f"Executed                 : {connection_result['executed']}")
print(f"Reason                   : {connection_result['reason']}")

if connection_result["status"] == "SUCCESS":
    print(
        f"Articles returned        : "
        f"{connection_result['articles_returned']}"
    )
    print(
        f"Top-level response fields: "
        f"{connection_result['top_level_fields']}"
    )
    print(
        f"First article fields     : "
        f"{connection_result['first_article_fields']}"
    )

print("=" * 72)

if connection_result["executed"]:
    print("Any completed live request was recorded in the usage ledger.")
else:
    print("No GNews API request has been made.")
```

## Controlled Candidate Acquisition and Original Source Storage

This block performs the first controlled acquisition of candidate media articles
for the Media AI research corpus.

Unlike the connection test in Block 7, records retrieved here are intended to
enter the research data pipeline. Acquisition therefore remains subject to all
previously defined safeguards, including the persistent daily usage ledger, the
per-request article limit, the daily acquisition ceiling, and the
`DOWNLOAD_NEW_DATA` master switch.

The block does not select articles according to topic, sentiment, publisher,
political orientation, emotional content, or expected framing. The provider is
queried using a broad English-language news endpoint so that substantive
selection criteria are not introduced before sampling and eligibility assessment.

Each successful provider response is handled at two levels.

First, the complete API response is preserved as an original acquisition batch.
This retains the provider's returned structure and permits later auditing of the
records as they were received.

Second, every returned article is converted into the canonical Media AI source
record defined in Block 6. Provider-specific fields are mapped explicitly:
`content` supplies the article text, `lang` supplies the provider language,
`publishedAt` preserves the publication timestamp, and `source`, `title`, `url`,
`description`, `image`, and provider `id` remain available inside the preserved
provider payload.

No analytical interpretation is introduced during this stage. Emotional weight,
factological weight, emotional classifications, transformative state,
recurrent state, confluent group representations, Social Polarisation Distance,
and supervisory labels remain excluded from the acquisition records.

The usage ledger is updated immediately after a completed API response. This
ensures that successful acquisition traffic remains recorded even if a later
processing or storage step fails.

Retrieved records are stored in the dedicated `raw_api` research folder in
Google Drive using a UTC timestamp and batch identifier. Existing source data
are not overwritten.

The block reports how many articles were requested, returned, technically
eligible, and stored. Technical eligibility at this stage remains preliminary;
near-duplicate analysis and final corpus preparation are deferred to the next
notebook.

**Input:** Valid GNews credential, acquisition safeguards, usage ledger,
sampling framework, source-record schema, and Google Drive storage.

**Output:** Timestamped original API batch, canonical source records, updated
usage ledger, and acquisition summary.

No request is made unless `DOWNLOAD_NEW_DATA` is explicitly set to `True`.

```python
# =============================================================================
# Media AI — Controlled Candidate Acquisition and Original Source Storage
# =============================================================================

import json
import uuid
from datetime import datetime, timezone

import requests


# -----------------------------------------------------------------------------
# Acquisition configuration
# -----------------------------------------------------------------------------

ACQUISITION_ARTICLES_PER_REQUEST = MAX_ARTICLES_PER_REQUEST

# Broad provider endpoint: no topic or sentiment filtering.
GNEWS_ACQUISITION_ENDPOINT = f"{GNEWS_BASE_URL}/top-headlines"


# -----------------------------------------------------------------------------
# GNews → Media AI source-record mapping
# -----------------------------------------------------------------------------

def create_gnews_source_record(provider_payload):
    """
    Convert one observed GNews article record into the canonical Media AI
    source-record structure while preserving the complete provider payload.
    """

    if not isinstance(provider_payload, dict):
        raise TypeError("provider_payload must be a dictionary.")

    article_language = (
        provider_payload.get("lang")
        or TARGET_LANGUAGE
    )

    # Reuse the Block 6 constructor while explicitly mapping GNews `lang`.
    return create_source_record(
        provider_payload=provider_payload,
        language=article_language,
    )


# -----------------------------------------------------------------------------
# Acquisition-batch storage
# -----------------------------------------------------------------------------

def save_acquisition_batch(
    raw_response,
    source_records,
):
    """
    Save one acquisition batch to the raw_api Google Drive folder.

    A unique timestamp and batch identifier prevent overwriting earlier batches.
    """

    now = datetime.now(timezone.utc)
    batch_id = str(uuid.uuid4())

    timestamp = now.strftime("%Y%m%dT%H%M%SZ")

    batch_payload = {
        "batch_id": batch_id,
        "acquired_at_utc": now.isoformat(),
        "provider": SOURCE_PROVIDER,
        "schema_version": SOURCE_RECORD_SCHEMA_VERSION,
        "raw_response": raw_response,
        "source_records": source_records,
    }

    filename = (
        f"gnews_batch_{timestamp}_{batch_id[:8]}.json"
    )

    file_id = save_json_file(
        file_name=filename,
        parent_id=RAW_API_FOLDER_ID,
        payload=batch_payload,
    )

    return {
        "batch_id": batch_id,
        "filename": filename,
        "file_id": file_id,
        "record_count": len(source_records),
    }


# -----------------------------------------------------------------------------
# Controlled acquisition
# -----------------------------------------------------------------------------

def acquire_gnews_candidates(
    requested_articles=ACQUISITION_ARTICLES_PER_REQUEST,
):
    """
    Perform one controlled candidate-acquisition request.

    All pre-request safeguards are checked before any network activity occurs.
    """

    if not DOWNLOAD_NEW_DATA:
        return {
            "executed": False,
            "status": "SKIPPED",
            "reason": "DOWNLOAD_NEW_DATA is False.",
        }

    if not is_real_api_key(GNEWS_API_KEY):
        return {
            "executed": False,
            "status": "BLOCKED",
            "reason": "A valid non-placeholder API key is not available.",
        }

    allowed, reason, ledger = check_request_allowed(
        requested_articles=requested_articles,
        require_download_enabled=True,
    )

    if not allowed:
        return {
            "executed": False,
            "status": "BLOCKED",
            "reason": reason,
        }

    params = {
        "lang": TARGET_LANGUAGE,
        "max": requested_articles,
        "apikey": GNEWS_API_KEY,
    }

    try:
        response = requests.get(
            GNEWS_ACQUISITION_ENDPOINT,
            params=params,
            timeout=30,
        )

        response.raise_for_status()
        raw_payload = response.json()

    except requests.RequestException as exc:
        return {
            "executed": True,
            "status": "ERROR",
            "reason": str(exc),
        }

    except ValueError:
        return {
            "executed": True,
            "status": "ERROR",
            "reason": "Provider response was not valid JSON.",
        }

    articles = raw_payload.get("articles", [])

    if not isinstance(articles, list):
        return {
            "executed": True,
            "status": "ERROR",
            "reason": "Expected 'articles' collection was not returned.",
        }

    # -------------------------------------------------------------------------
    # Record completed provider usage immediately
    # -------------------------------------------------------------------------

    register_api_request(
        ledger=ledger,
        articles_requested=requested_articles,
        articles_returned=len(articles),
    )

    # -------------------------------------------------------------------------
    # Construct canonical source records
    # -------------------------------------------------------------------------

    source_records = [
        create_gnews_source_record(article)
        for article in articles
    ]

    eligible_records = [
        record
        for record in source_records
        if record["technical_eligibility"]["eligible"]
    ]

    # -------------------------------------------------------------------------
    # Persist complete batch
    # -------------------------------------------------------------------------

    storage_result = save_acquisition_batch(
        raw_response=raw_payload,
        source_records=source_records,
    )

    return {
        "executed": True,
        "status": "SUCCESS",
        "reason": "Candidate acquisition completed successfully.",
        "articles_requested": requested_articles,
        "articles_returned": len(articles),
        "technically_eligible": len(eligible_records),
        "technically_ineligible": (
            len(source_records) - len(eligible_records)
        ),
        "stored_records": len(source_records),
        "batch_id": storage_result["batch_id"],
        "stored_filename": storage_result["filename"],
        "stored_file_id": storage_result["file_id"],
    }


# -----------------------------------------------------------------------------
# Execute or safely skip
# -----------------------------------------------------------------------------

acquisition_result = acquire_gnews_candidates()


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Controlled Candidate Acquisition")
print("=" * 72)

print(f"Status                    : {acquisition_result['status']}")
print(f"Executed                  : {acquisition_result['executed']}")
print(f"Reason                    : {acquisition_result['reason']}")

if acquisition_result["status"] == "SUCCESS":
    print(
        f"Articles requested        : "
        f"{acquisition_result['articles_requested']}"
    )
    print(
        f"Articles returned         : "
        f"{acquisition_result['articles_returned']}"
    )
    print(
        f"Technically eligible      : "
        f"{acquisition_result['technically_eligible']}"
    )
    print(
        f"Technically ineligible    : "
        f"{acquisition_result['technically_ineligible']}"
    )
    print(
        f"Stored source records     : "
        f"{acquisition_result['stored_records']}"
    )
    print(
        f"Stored batch file         : "
        f"{acquisition_result['stored_filename']}"
    )

print("=" * 72)

if acquisition_result["executed"]:
    print("Any completed API request was recorded in the usage ledger.")
else:
    print("No GNews API request has been made.")
```

## Eligibility Review, Exact Deduplication and Reproducible Sampling

This block reviews the source records already acquired and stored in the
`raw_api` research folder. It does not contact GNews and does not create any
additional API usage.

The purpose is to convert the technically eligible acquisition records into a
reproducible candidate corpus while preserving a complete record of exclusions.

The procedure is performed in three stages.

First, each stored Media AI source record is evaluated using the technical
eligibility criteria defined in Block 5. These criteria concern only whether the
record is suitable for the pilot architecture: English-language full text,
minimum length, and sufficient source information. They are not assessments of
factual accuracy, publisher quality, sentiment, ideology, or framing.

Second, exact duplicates are removed using the deterministic textual
fingerprints generated from normalised article content. Where several records
share the same fingerprint, one record is retained and the remaining records are
marked as exact duplicates. This procedure does not attempt to identify
syndicated, paraphrased, or semantically similar articles; near-duplicate
detection remains a separate task for the data-preparation stage.

Third, the remaining eligible and unique records are randomly ordered using the
fixed project seed. This provides a reproducible sampling order without
introducing topic, sentiment, publisher, or framing preferences.

All acquired observations remain auditable. Records excluded for technical
ineligibility or exact duplication are retained in the review output together
with explicit exclusion reasons rather than being silently discarded.

The output of this block is therefore not yet a final modelling dataset. It is a
technically reviewed and reproducibly ordered candidate corpus that can be
passed to the subsequent data-preparation notebook.

**Input:** Original source records stored by Block 8.

**Output:** Reviewed record table, exact-duplicate status, exclusion reasons,
and reproducibly ordered eligible candidate records.

No GNews API request is performed by this block.

```python
# =============================================================================
# Media AI — Eligibility Review, Exact Deduplication and Reproducible Sampling
# =============================================================================

import random


# -----------------------------------------------------------------------------
# Extract source records from the most recent successful acquisition
# -----------------------------------------------------------------------------

if acquisition_result.get("status") != "SUCCESS":
    raise RuntimeError(
        "No successful Block 8 acquisition is available in the current runtime."
    )

# Reuse the source records created during Block 8.
# They were also persisted in the raw_api batch file.
source_records = [
    create_gnews_source_record(article)
    for article in acquisition_result.get("_articles", [])
] if "_articles" in acquisition_result else None


# -----------------------------------------------------------------------------
# Fallback: reconstruct from the successful Block 8 response in memory
# -----------------------------------------------------------------------------

if source_records is None:
    # Block 8 did not expose records directly in its result dictionary,
    # so retrieve the stored acquisition batch from Google Drive.
    batch_file_id = acquisition_result["stored_file_id"]

    stored_batch = download_json_file(batch_file_id)
    source_records = stored_batch.get("source_records", [])


# -----------------------------------------------------------------------------
# Eligibility review
# -----------------------------------------------------------------------------

reviewed_records = []

for record in source_records:

    eligibility = record.get(
        "technical_eligibility",
        {}
    )

    reviewed_records.append({
        "record": record,
        "media_ai_record_id": record.get("media_ai_record_id"),
        "provider_article_id": record.get("provider_article_id"),
        "title": record.get("title"),
        "source_url": record.get("source_url"),
        "word_count": record.get("word_count"),
        "text_fingerprint": record.get("text_fingerprint"),
        "eligible": eligibility.get("eligible", False),
        "exclusion_reasons": list(
            eligibility.get("exclusion_reasons", [])
        ),
        "exact_duplicate": False,
    })


# -----------------------------------------------------------------------------
# Exact deduplication
# -----------------------------------------------------------------------------

seen_fingerprints = set()

for item in reviewed_records:

    fingerprint = item["text_fingerprint"]

    if not item["eligible"]:
        continue

    if fingerprint is None:
        item["eligible"] = False
        item["exclusion_reasons"].append(
            "missing_text_fingerprint"
        )
        continue

    if fingerprint in seen_fingerprints:
        item["exact_duplicate"] = True
        item["eligible"] = False
        item["exclusion_reasons"].append(
            "exact_duplicate"
        )
    else:
        seen_fingerprints.add(fingerprint)


# -----------------------------------------------------------------------------
# Reproducible random ordering
# -----------------------------------------------------------------------------

eligible_unique_records = [
    item["record"]
    for item in reviewed_records
    if item["eligible"]
]

sampling_rng = random.Random(
    SAMPLING_RANDOM_SEED
)

sampling_rng.shuffle(
    eligible_unique_records
)


# -----------------------------------------------------------------------------
# Review summary
# -----------------------------------------------------------------------------

total_records = len(reviewed_records)

eligible_before_dedup = sum(
    1
    for item in reviewed_records
    if (
        item["technical_eligibility"]["eligible"]
        if "technical_eligibility" in item
        else False
    )
)

exact_duplicates = sum(
    1
    for item in reviewed_records
    if item["exact_duplicate"]
)

final_eligible = len(
    eligible_unique_records
)

technical_exclusions = sum(
    1
    for item in reviewed_records
    if (
        not item["record"]
        .get("technical_eligibility", {})
        .get("eligible", False)
    )
)


# -----------------------------------------------------------------------------
# Persist reviewed candidate corpus
# -----------------------------------------------------------------------------

review_payload = {
    "created_at_utc": datetime.now(
        timezone.utc
    ).isoformat(),
    "sampling_seed": SAMPLING_RANDOM_SEED,
    "source_batch_id": acquisition_result["batch_id"],
    "total_source_records": total_records,
    "technical_exclusions": technical_exclusions,
    "exact_duplicates": exact_duplicates,
    "eligible_unique_records": final_eligible,
    "review": [
        {
            key: value
            for key, value in item.items()
            if key != "record"
        }
        for item in reviewed_records
    ],
    "sampled_records": eligible_unique_records,
}

review_filename = (
    f"candidate_review_"
    f"{acquisition_result['batch_id'][:8]}.json"
)

review_file_id = save_json_file(
    file_name=review_filename,
    parent_id=FULL_TEXT_FOLDER_ID,
    payload=review_payload,
)


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Eligibility and Sampling Review")
print("=" * 72)

print(f"Source records reviewed     : {total_records}")
print(f"Technical exclusions        : {technical_exclusions}")
print(f"Exact duplicates            : {exact_duplicates}")
print(f"Eligible unique records     : {final_eligible}")
print(f"Sampling seed               : {SAMPLING_RANDOM_SEED}")
print(f"Stored review file          : {review_filename}")

print("=" * 72)
print("Eligibility review and reproducible sampling completed.")
print("No GNews API requests have been made.")
```
