# Media AI — Notebook 02: Data Sanitisation and Preparation

## Abstract and Methodological Scope

Notebook 02 transforms the technically eligible source records produced by
Notebook 01 into a structured research corpus suitable for subsequent
supervision, annotation and modelling.

The central methodological principle is:

> **Sanitise structure, not meaning.**

The purpose of sanitisation is therefore not to make media text more neutral,
more grammatical, less emotional or easier for the model to interpret.
Sanitisation is restricted to technical transformations required to obtain
consistent, machine-processable textual representations while preserving the
linguistic evidence that Media AI is intended to investigate.

This distinction is particularly important because apparently insignificant
features of a text may contribute to later emotional or social transformations.
Words, punctuation, quotations, repetition, unusual constructions and
context-dependent expressions must not be removed merely because they appear
irrelevant to conventional sentiment or topic analysis.

The original source records created in Notebook 01 remain unchanged. Notebook
02 operates only on derived representations. Every derived record retains a
stable reference to its original source record so that subsequent analytical
outputs can be traced back to the acquired evidence.

### Sanitisation Boundary

Permitted technical operations include, where required:

- Unicode normalisation;
- removal of invalid or non-printing control characters;
- consistent line-ending representation;
- conservative whitespace normalisation;
- removal of clearly identifiable markup that is not part of the article text;
- detection of malformed or unexpectedly structured text;
- validation of required textual fields;
- identification of technical artefacts introduced during acquisition or
  storage.

Sanitisation must not deliberately:

- paraphrase or rewrite article language;
- correct grammar or spelling;
- neutralise emotionally charged wording;
- replace politically or socially sensitive terminology;
- remove quotations because they contain emotional language;
- remove repetition solely because it appears rhetorically intentional;
- convert subjective statements into neutral formulations;
- infer or correct the factual truth of claims;
- alter names, organisations or social-group references;
- remove words because an emotion classifier considers them emotionally
  significant.

Where a transformation could plausibly affect semantic, emotional or social
interpretation, the preferred action is to **flag the observation for review
rather than silently modify it**.

### Original and Derived Representations

Notebook 02 maintains an explicit distinction between:

**Original source representation**

The version preserved by Notebook 01, including the provider payload,
acquisition metadata and original textual fingerprint.

**Derived sanitised representation**

A technically normalised copy created by Notebook 02 for subsequent research
processing.

The relationship is therefore:

Original Source Record  
→ Derived Sanitised Record  
→ Segmented Research Record  
→ Subsequent Annotation and Modelling

The original source representation is never overwritten by the derived version.

### Text Integrity

Sanitisation must be measurable.

For each article, Notebook 02 will record information describing transformations
applied to the text, including the number and type of modifications where
practicable.

Unexpectedly large differences between original and sanitised text will be
flagged for inspection.

This creates a distinction between an intentional technical transformation and
an unexplained change in research evidence.

### Near-Duplicate Detection

Notebook 01 removes only exact duplicates through deterministic textual
fingerprints.

Notebook 02 extends this process by identifying near-duplicate material,
including potentially syndicated or lightly modified versions of substantially
the same article.

Near-duplicate detection is treated separately from sanitisation.

A high similarity score does not automatically prove that two articles are
duplicates. Similarity measurements therefore produce reviewable evidence and
duplicate-group candidates rather than silently deleting observations.

### Hierarchical Segmentation

Prepared articles will be represented hierarchically as:

Article  
→ Paragraph  
→ Sentence

Stable identifiers and source-relative positions will be retained wherever
practicable.

This hierarchy is required because later Media AI components are intended to
analyse transformations at several levels while preserving the ability to trace
an article-level result back to the textual units that contributed to it.

### Tokenisation Boundary

Notebook 02 deliberately does not permanently tokenise the corpus for a
particular pretrained language model.

Tokenisation depends on the backbone architecture and therefore belongs to the
modelling stage.

Notebook 02 preserves model-independent textual units. Notebook 04 will apply
the tokenizer associated with the selected backbone when contextual
representations are generated.

This separation prevents data preparation from becoming unnecessarily coupled
to one pretrained model.

### Relationship to the Media AI Architecture

Notebook 02 does not calculate emotional weights, factological weights,
transformative states, recurrent states or confluent representations.

Its purpose is to preserve and structure the evidence from which those
quantities will later be estimated.

This is especially important for the transformative architecture, which will
process all observed tokens and calculate emotional contributions contextually
and on demand. Notebook 02 must therefore avoid pre-filtering the text into
"emotional" and "non-emotional" vocabulary.

### Reproducibility

All preparation rules are defined before inspecting model outputs.

The notebook uses deterministic configuration wherever possible and records
preparation-policy and schema versions in its outputs.

The resulting corpus should therefore be reproducible from the Notebook 01
handover manifest and the preserved source records.

### Notebook 02 Workflow

The notebook follows the sequence:

Notebook 01 Handover  
→ Input Integrity Verification  
→ Technical Sanitisation  
→ Derived Text Construction  
→ Sanitisation Audit  
→ Near-Duplicate Detection  
→ Hierarchical Segmentation  
→ Quality Diagnostics  
→ Prepared Corpus  
→ Notebook 03 Handover

The output of Notebook 02 is not an emotionally or psychologically labelled
dataset. It is a technically prepared, structurally explicit and traceable
representation of the acquired research corpus.

No GNews API requests are performed by this notebook.

```python
# =============================================================================
# Media AI — Notebook 02 Preparation Configuration and Policy
# =============================================================================

from datetime import datetime, timezone
from pathlib import Path
import platform
import random
import sys


# -----------------------------------------------------------------------------
# Notebook identity
# -----------------------------------------------------------------------------

PROJECT_NAME = "Media AI"
NOTEBOOK_NAME = "02_data_sanitisation_and_preparation"

NOTEBOOK_VERSION = "1.0"
PREPARATION_SCHEMA_VERSION = "1.0"
SANITISATION_POLICY_VERSION = "1.0"

RANDOM_SEED = 42


# -----------------------------------------------------------------------------
# Deterministic configuration
# -----------------------------------------------------------------------------

random.seed(RANDOM_SEED)


# -----------------------------------------------------------------------------
# Preparation policy
# -----------------------------------------------------------------------------

SANITISATION_POLICY = {
    "principle": "sanitise_structure_not_meaning",

    "unicode_normalisation": True,
    "remove_invalid_control_characters": True,
    "normalise_line_endings": True,
    "normalise_whitespace_conservatively": True,
    "remove_confirmed_non_text_markup": True,

    "rewrite_language": False,
    "correct_grammar": False,
    "correct_spelling": False,
    "neutralise_emotional_language": False,
    "remove_emotional_vocabulary": False,
    "remove_sensitive_terminology": False,
    "remove_quotations": False,
    "infer_factual_truth": False,
    "alter_named_entities": False,

    "flag_ambiguous_transformations": True,
    "preserve_original_source_records": True,
}


# -----------------------------------------------------------------------------
# Model-independence policy
# -----------------------------------------------------------------------------

MODEL_PREPARATION_POLICY = {
    "permanent_backbone_tokenisation": False,
    "pre_filter_emotional_tokens": False,
    "pre_filter_non_emotional_tokens": False,

    "preserve_complete_text": True,

    "hierarchical_units": [
        "article",
        "paragraph",
        "sentence",
    ],
}


# -----------------------------------------------------------------------------
# Notebook execution metadata
# -----------------------------------------------------------------------------

NOTEBOOK_02_INITIALISED_AT_UTC = datetime.now(
    timezone.utc
).isoformat()

NOTEBOOK_02_RUNTIME = {
    "project": PROJECT_NAME,
    "notebook": NOTEBOOK_NAME,
    "notebook_version": NOTEBOOK_VERSION,
    "preparation_schema_version": PREPARATION_SCHEMA_VERSION,
    "sanitisation_policy_version": SANITISATION_POLICY_VERSION,
    "random_seed": RANDOM_SEED,
    "python_version": sys.version.split()[0],
    "platform": platform.platform(),
    "initialised_at_utc": NOTEBOOK_02_INITIALISED_AT_UTC,
}


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Notebook 02 Initialisation")
print("=" * 72)

print(f"Notebook                   : {NOTEBOOK_NAME}")
print(f"Notebook version           : {NOTEBOOK_VERSION}")
print(f"Preparation schema         : {PREPARATION_SCHEMA_VERSION}")
print(f"Sanitisation policy        : {SANITISATION_POLICY_VERSION}")
print(f"Random seed                : {RANDOM_SEED}")

print("-" * 72)

print(
    "Core principle             : "
    f"{SANITISATION_POLICY['principle']}"
)

print(
    "Preserve source records    : "
    f"{SANITISATION_POLICY['preserve_original_source_records']}"
)

print(
    "Preserve complete text     : "
    f"{MODEL_PREPARATION_POLICY['preserve_complete_text']}"
)

print(
    "Permanent tokenisation     : "
    f"{MODEL_PREPARATION_POLICY['permanent_backbone_tokenisation']}"
)

print(
    "Pre-filter emotional tokens: "
    f"{MODEL_PREPARATION_POLICY['pre_filter_emotional_tokens']}"
)

print("=" * 72)

print("Notebook 02 preparation policy initialised successfully.")
print("No source records have been modified.")
print("No GNews API requests have been made.")
```

## Notebook 01 Handover and Persistent Storage Access

This block reconnects Notebook 02 to the persistent Media AI research storage
established during Notebook 01 and loads the acquisition handover manifest.

Notebook 02 does not rely on temporary objects remaining in the previous Colab
runtime. Instead, the transition between notebooks is reconstructed from the
persisted research artefacts created at the end of Notebook 01.

This design supports reproducibility and source traceability. A researcher
should be able to close Notebook 01, start a new Colab runtime, open Notebook 02,
and recover the correct preparation input without depending on hidden notebook
state.

Google Drive is accessed through the Google Drive API rather than by mounting
the user's complete Drive filesystem. The notebook locates the dedicated
`Media_AI` project structure and reads only the research files required for the
current workflow.

The principal input is:

`notebook_01_handover_manifest.json`

The manifest identifies the reviewed candidate corpus produced by Notebook 01
and records the expected number of eligible records, source schema version,
sampling configuration, acquisition batch, and relevant storage references.

At this stage, the manifest is loaded but the candidate article content is not
yet sanitised or transformed. The purpose of this block is to establish a
persistent and auditable connection between the acquisition and preparation
stages.

The block performs several preliminary checks:

- the Media AI project folder can be located;
- the expected `logs` folder is available;
- the Notebook 01 handover manifest exists;
- the manifest type and version are readable;
- the next-stage input filename and expected record count are available;
- no API credential is required or accessed.

The Google Drive identifiers stored inside the manifest are research storage
references rather than authentication credentials. They do not grant access to
another user's Google account when the notebook is shared publicly.

**Input:** Restricted Google Drive authorisation and the persistent Notebook 01
handover manifest.

**Output:** Loaded handover manifest, resolved Media AI storage references, and
the expected Notebook 02 candidate-corpus metadata.

No article text is modified and no GNews API request is performed by this block.

```python
# =============================================================================
# Media AI — Notebook 01 Handover and Persistent Storage Access
# =============================================================================

import io
import json

from google.colab import auth
from googleapiclient.discovery import build
from googleapiclient.http import MediaIoBaseDownload


# -----------------------------------------------------------------------------
# Restricted Google Drive authentication
# -----------------------------------------------------------------------------

auth.authenticate_user()

drive_service = build(
    "drive",
    "v3",
    cache_discovery=False,
)


# -----------------------------------------------------------------------------
# Helper: locate one folder inside a specific parent
# -----------------------------------------------------------------------------

def find_drive_folder(folder_name, parent_id=None):
    """
    Locate a Google Drive folder by name.

    Parameters
    ----------
    folder_name : str
        Folder name to locate.

    parent_id : str or None
        Optional parent folder identifier.

    Returns
    -------
    str or None
        Matching folder ID, or None when no matching folder is found.
    """

    safe_name = folder_name.replace("'", "\\'")

    query_parts = [
        f"name = '{safe_name}'",
        "mimeType = 'application/vnd.google-apps.folder'",
        "trashed = false",
    ]

    if parent_id:
        query_parts.append(
            f"'{parent_id}' in parents"
        )

    query = " and ".join(query_parts)

    response = (
        drive_service.files()
        .list(
            q=query,
            spaces="drive",
            fields="files(id, name)",
            pageSize=10,
        )
        .execute()
    )

    folders = response.get("files", [])

    if not folders:
        return None

    return folders[0]["id"]


# -----------------------------------------------------------------------------
# Helper: locate one file inside a specific folder
# -----------------------------------------------------------------------------

def find_drive_file(file_name, parent_id):
    """
    Locate a non-trashed Google Drive file within a specific folder.
    """

    safe_name = file_name.replace("'", "\\'")

    query = (
        f"name = '{safe_name}' and "
        f"'{parent_id}' in parents and "
        "trashed = false"
    )

    response = (
        drive_service.files()
        .list(
            q=query,
            spaces="drive",
            fields="files(id, name, mimeType)",
            pageSize=10,
        )
        .execute()
    )

    files = response.get("files", [])

    if not files:
        return None

    return files[0]["id"]


# -----------------------------------------------------------------------------
# Helper: download JSON
# -----------------------------------------------------------------------------

def download_json_from_drive(file_id):
    """
    Download a JSON research artefact from Google Drive.
    """

    request = drive_service.files().get_media(
        fileId=file_id
    )

    buffer = io.BytesIO()

    downloader = MediaIoBaseDownload(
        buffer,
        request
    )

    done = False

    while not done:
        _, done = downloader.next_chunk()

    buffer.seek(0)

    return json.loads(
        buffer.read().decode("utf-8")
    )


# -----------------------------------------------------------------------------
# Resolve Media AI research folders
# -----------------------------------------------------------------------------

PROJECT_FOLDER_ID = find_drive_folder(
    "Media_AI"
)

if PROJECT_FOLDER_ID is None:
    raise FileNotFoundError(
        "Media_AI project folder was not found."
    )


LOGS_FOLDER_ID = find_drive_folder(
    "logs",
    PROJECT_FOLDER_ID
)

if LOGS_FOLDER_ID is None:
    raise FileNotFoundError(
        "Media_AI/logs folder was not found."
    )


DATA_FOLDER_ID = find_drive_folder(
    "data",
    PROJECT_FOLDER_ID
)

if DATA_FOLDER_ID is None:
    raise FileNotFoundError(
        "Media_AI/data folder was not found."
    )


FULL_TEXT_FOLDER_ID = find_drive_folder(
    "full_text",
    DATA_FOLDER_ID
)

if FULL_TEXT_FOLDER_ID is None:
    raise FileNotFoundError(
        "Media_AI/data/full_text folder was not found."
    )


# -----------------------------------------------------------------------------
# Load Notebook 01 handover manifest
# -----------------------------------------------------------------------------

HANDOVER_FILENAME = (
    "notebook_01_handover_manifest.json"
)

HANDOVER_FILE_ID = find_drive_file(
    HANDOVER_FILENAME,
    LOGS_FOLDER_ID,
)

if HANDOVER_FILE_ID is None:
    raise FileNotFoundError(
        f"{HANDOVER_FILENAME} was not found."
    )


notebook_01_handover = download_json_from_drive(
    HANDOVER_FILE_ID
)


# -----------------------------------------------------------------------------
# Preliminary manifest validation
# -----------------------------------------------------------------------------

expected_manifest_type = (
    "media_ai_acquisition_handover"
)

actual_manifest_type = notebook_01_handover.get(
    "manifest_type"
)

if actual_manifest_type != expected_manifest_type:
    raise ValueError(
        "Unexpected Notebook 01 handover manifest type."
    )


next_stage = notebook_01_handover.get(
    "next_stage",
    {}
)

INPUT_FILENAME = next_stage.get(
    "input_filename"
)

INPUT_FILE_ID = next_stage.get(
    "input_file_id"
)

EXPECTED_INPUT_RECORDS = next_stage.get(
    "input_record_count"
)


if not INPUT_FILENAME:
    raise ValueError(
        "The handover manifest does not specify an input filename."
    )

if not INPUT_FILE_ID:
    raise ValueError(
        "The handover manifest does not specify an input file ID."
    )

if EXPECTED_INPUT_RECORDS is None:
    raise ValueError(
        "The handover manifest does not specify an expected record count."
    )


# -----------------------------------------------------------------------------
# Verify that the reviewed candidate file remains accessible
# -----------------------------------------------------------------------------

candidate_file_metadata = (
    drive_service.files()
    .get(
        fileId=INPUT_FILE_ID,
        fields="id,name,mimeType,trashed",
    )
    .execute()
)

if candidate_file_metadata.get("trashed"):
    raise FileNotFoundError(
        "The candidate corpus referenced by Notebook 01 is in Trash."
    )


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Notebook 01 Handover")
print("=" * 72)

print(
    f"Manifest type              : "
    f"{actual_manifest_type}"
)

print(
    f"Manifest version           : "
    f"{notebook_01_handover.get('manifest_version')}"
)

print(
    f"Preparation input file     : "
    f"{INPUT_FILENAME}"
)

print(
    f"Expected candidate records : "
    f"{EXPECTED_INPUT_RECORDS}"
)

print(
    f"Source schema version      : "
    f"{notebook_01_handover.get('acquisition', {}).get('source_record_schema_version')}"
)

print(
    f"Candidate file accessible  : "
    f"{candidate_file_metadata.get('name') == INPUT_FILENAME}"
)

print("=" * 72)

print("Notebook 01 handover loaded successfully.")
print("Candidate article content has not yet been transformed.")
print("No GNews API requests have been made.")
```

## Input Integrity, Schema Compatibility and Source Traceability Verification

This block verifies that the candidate corpus referenced by the Notebook 01
handover manifest is structurally consistent and traceable to the acquisition
stage before sanitisation or derived-text construction begins.

The verification distinguishes **source integrity** from **schema evolution**.

Notebook 01 currently implements source-record schema version 1.1. The pilot
records handed to Notebook 02 were originally created under schema version 1.0.
The handover manifest therefore records both:

- the current source-record schema implemented by Notebook 01;
- the source-record schema versions actually present in the handed-over records.

Historical source records are not rewritten merely to conform to a later schema.
Doing so would alter the representation originally preserved during acquisition.

Where descriptive fields introduced in a later schema are absent from the
top-level record, they may be read from the unchanged `provider_payload` for
inspection and subsequent derived processing. This does not modify the original
source record.

The critical integrity checks concern whether:

- the number of candidate records matches the Notebook 01 handover;
- every record has a stable Media AI identifier;
- its source schema belongs to the explicitly supported schema history;
- the schema reported by the record is compatible with the schema versions
  declared in the handover manifest;
- the record remains technically eligible;
- source article text is present;
- the original textual fingerprint is present;
- the fingerprint can be reproduced from the preserved provider text;
- no exact duplicate has entered the candidate set.

Provider article identifiers are useful source metadata but are not treated as
a critical integrity requirement for historical schema 1.0 records.

The textual fingerprint is used here as an integrity check rather than as a
semantic-similarity measure. A matching fingerprint indicates that the
conservatively normalised source text has remained unchanged between acquisition
and preparation.

Records that fail critical integrity checks are not passed into the sanitisation
stage.

**Input:** Candidate corpus and schema-history metadata identified by the
Notebook 01 handover manifest.

**Output:** Verified candidate records, schema-compatibility information and the
preparation input approved for subsequent sanitisation.

No candidate article content is transformed and no GNews API request is
performed by this block.

```python
# =============================================================================
# Media AI — Input Integrity, Schema Compatibility and Source Traceability
# =============================================================================

import hashlib
import re


# -----------------------------------------------------------------------------
# Supported source-record schema history
# -----------------------------------------------------------------------------

SUPPORTED_SOURCE_SCHEMA_VERSIONS = {
    "1.0",
    "1.1",
}


# -----------------------------------------------------------------------------
# Read schema information from the Notebook 01 handover
# -----------------------------------------------------------------------------

handover_acquisition = notebook_01_handover.get(
    "acquisition",
    {}
)

CURRENT_SOURCE_SCHEMA_VERSION = (
    handover_acquisition.get(
        "source_record_schema_current"
    )
)

SOURCE_RECORD_SCHEMAS_PRESENT = (
    handover_acquisition.get(
        "source_record_schemas_present",
        []
    )
)


if CURRENT_SOURCE_SCHEMA_VERSION is None:
    raise ValueError(
        "The Notebook 01 handover does not specify "
        "the current source-record schema."
    )

if not SOURCE_RECORD_SCHEMAS_PRESENT:
    raise ValueError(
        "The Notebook 01 handover does not specify "
        "the source-record schemas present."
    )


# -----------------------------------------------------------------------------
# Load reviewed candidate corpus
# -----------------------------------------------------------------------------

candidate_review = download_json_from_drive(
    INPUT_FILE_ID
)

candidate_records = candidate_review.get(
    "sampled_records",
    []
)


# -----------------------------------------------------------------------------
# Conservative normalisation used for fingerprint verification
# -----------------------------------------------------------------------------

def normalise_text_for_integrity(text):
    """
    Reproduce the conservative normalisation used for source fingerprints.

    This function performs whitespace normalisation only and does not sanitise,
    rewrite or otherwise alter semantic content.
    """

    if not isinstance(text, str):
        return ""

    return re.sub(
        r"\s+",
        " ",
        text,
    ).strip()


def calculate_integrity_fingerprint(text):
    """
    Calculate the SHA-256 fingerprint used for source-text integrity checks.
    """

    normalised = normalise_text_for_integrity(
        text
    )

    if not normalised:
        return None

    return hashlib.sha256(
        normalised.encode("utf-8")
    ).hexdigest()


# -----------------------------------------------------------------------------
# Record-level integrity verification
# -----------------------------------------------------------------------------

integrity_results = []

seen_fingerprints = set()

for record in candidate_records:

    provider_payload = record.get(
        "provider_payload",
        {}
    )

    source_text = provider_payload.get(
        "content",
        ""
    )

    stored_fingerprint = record.get(
        "text_fingerprint"
    )

    calculated_fingerprint = (
        calculate_integrity_fingerprint(
            source_text
        )
    )

    technical_eligibility = record.get(
        "technical_eligibility",
        {}
    )

    source_schema_version = record.get(
        "schema_version"
    )

    # -------------------------------------------------------------------------
    # Schema-compatible descriptive fields
    # -------------------------------------------------------------------------

    provider_article_id = (
        record.get("provider_article_id")
        or provider_payload.get("id")
    )

    title = (
        record.get("title")
        or provider_payload.get("title")
    )

    # -------------------------------------------------------------------------
    # Exact-duplicate check
    # -------------------------------------------------------------------------

    fingerprint_duplicate = (
        stored_fingerprint in seen_fingerprints
        if stored_fingerprint
        else False
    )

    if stored_fingerprint:
        seen_fingerprints.add(
            stored_fingerprint
        )

    # -------------------------------------------------------------------------
    # Integrity and compatibility checks
    # -------------------------------------------------------------------------

    checks = {
        "media_ai_record_id_present":
            bool(
                record.get(
                    "media_ai_record_id"
                )
            ),

        "source_schema_supported":
            source_schema_version
            in SUPPORTED_SOURCE_SCHEMA_VERSIONS,

        "source_schema_declared_in_handover":
            source_schema_version
            in SOURCE_RECORD_SCHEMAS_PRESENT,

        "technically_eligible":
            technical_eligibility.get(
                "eligible",
                False
            ),

        "text_present":
            bool(source_text),

        "fingerprint_present":
            bool(stored_fingerprint),

        "fingerprint_matches":
            (
                stored_fingerprint
                == calculated_fingerprint
            ),

        "exact_duplicate_in_candidate_set":
            fingerprint_duplicate,
    }

    critical_checks = [
        checks["media_ai_record_id_present"],
        checks["source_schema_supported"],
        checks["source_schema_declared_in_handover"],
        checks["technically_eligible"],
        checks["text_present"],
        checks["fingerprint_present"],
        checks["fingerprint_matches"],
        not checks[
            "exact_duplicate_in_candidate_set"
        ],
    ]

    integrity_results.append({
        "media_ai_record_id":
            record.get(
                "media_ai_record_id"
            ),

        "provider_article_id":
            provider_article_id,

        "title":
            title,

        "source_schema_version":
            source_schema_version,

        "current_schema_version":
            CURRENT_SOURCE_SCHEMA_VERSION,

        "historical_schema":
            source_schema_version
            != CURRENT_SOURCE_SCHEMA_VERSION,

        "word_count":
            record.get(
                "word_count"
            ),

        "checks":
            checks,

        "integrity_passed":
            all(
                critical_checks
            ),
    })


# -----------------------------------------------------------------------------
# Corpus-level validation
# -----------------------------------------------------------------------------

actual_record_count = len(
    candidate_records
)

record_count_matches = (
    actual_record_count
    == EXPECTED_INPUT_RECORDS
)

all_records_passed = all(
    result["integrity_passed"]
    for result in integrity_results
)

duplicate_count = sum(
    1
    for result in integrity_results
    if result["checks"][
        "exact_duplicate_in_candidate_set"
    ]
)

fingerprint_mismatch_count = sum(
    1
    for result in integrity_results
    if not result["checks"][
        "fingerprint_matches"
    ]
)

historical_schema_count = sum(
    1
    for result in integrity_results
    if result["historical_schema"]
)


# -----------------------------------------------------------------------------
# Verified preparation input
# -----------------------------------------------------------------------------

VERIFIED_CANDIDATE_RECORDS = [
    record
    for record, result
    in zip(
        candidate_records,
        integrity_results
    )
    if result["integrity_passed"]
]


# -----------------------------------------------------------------------------
# Fail safely only for genuine integrity problems
# -----------------------------------------------------------------------------

if not record_count_matches:
    raise ValueError(
        "Candidate record count does not match "
        "the Notebook 01 handover."
    )

if not all_records_passed:
    raise ValueError(
        "One or more candidate records failed "
        "source integrity verification."
    )


# -----------------------------------------------------------------------------
# Verification output
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Input Integrity and Schema Compatibility")
print("=" * 72)

print(
    f"Expected candidate records : "
    f"{EXPECTED_INPUT_RECORDS}"
)

print(
    f"Loaded candidate records   : "
    f"{actual_record_count}"
)

print(
    f"Record count matches       : "
    f"{record_count_matches}"
)

print(
    f"Current source schema      : "
    f"{CURRENT_SOURCE_SCHEMA_VERSION}"
)

print(
    f"Source schemas present     : "
    f"{SOURCE_RECORD_SCHEMAS_PRESENT}"
)

print(
    f"Historical schema records  : "
    f"{historical_schema_count}"
)

print(
    f"Fingerprint mismatches     : "
    f"{fingerprint_mismatch_count}"
)

print(
    f"Exact duplicates detected  : "
    f"{duplicate_count}"
)

print(
    f"Records passing integrity  : "
    f"{len(VERIFIED_CANDIDATE_RECORDS)}"
)

print(
    f"All records passed         : "
    f"{all_records_passed}"
)

print("=" * 72)

print(
    "Notebook 02 input integrity verified successfully."
)

print(
    "Historical schema versions were preserved rather than rewritten."
)

print(
    "No candidate article content has been transformed by this block."
)

print(
    "No GNews API requests have been made by this block."
)
```

## Technical Sanitisation and Derived Text Construction

This block creates the first derived textual representation of the verified
candidate corpus.

The governing principle remains:

> **Sanitise structure, not meaning.**

Sanitisation is restricted to technical transformations that improve structural
consistency and machine-processability without deliberately changing the
semantic, emotional, factual or social content of the article.

The original source records verified in Block 3 remain unchanged. For every
article, Notebook 02 creates a separate derived record containing the sanitised
text, an explicit reference to the original Media AI source record, and a
sanitisation audit describing every transformation applied.

The sanitisation process is intentionally conservative.

Permitted operations include:

- Unicode normalisation using a documented normalisation form;
- normalisation of Windows and legacy line endings;
- removal of invalid null characters;
- removal of non-printing Unicode control characters that do not carry ordinary
  textual meaning;
- preservation of paragraph boundaries;
- conservative normalisation of horizontal whitespace;
- removal of clearly identifiable HTML markup where markup is present in the
  article text;
- conversion of common HTML character entities into their textual
  representation.

The block does not:

- rewrite article wording;
- correct spelling;
- correct grammar;
- paraphrase sentences;
- alter quotations;
- replace emotionally charged expressions;
- remove political, religious, scientific or socially sensitive terminology;
- remove named entities;
- infer factual correctness;
- remove repetition because it appears rhetorical;
- classify words as emotional or non-emotional;
- perform model-specific tokenisation.

### Unicode Normalisation

Unicode permits visually similar text to be represented through different
underlying code-point sequences. Notebook 02 applies a consistent Unicode
normalisation form so that technically equivalent textual representations are
handled consistently.

The operation is structural rather than linguistic. The resulting text is
audited against the original representation.

### Control Characters

Null bytes and non-printing control characters may be introduced through data
transport, storage or encoding irregularities.

Such characters may interfere with downstream text processing while contributing
no intended linguistic information.

They are therefore removed where explicitly identified. Ordinary line breaks,
tabs and textual punctuation are treated separately and are not indiscriminately
deleted.

### Whitespace

Whitespace is normalised conservatively.

Repeated spaces and horizontal tab-like spacing may be collapsed where they
represent technical layout variation rather than linguistic content.

Paragraph separation is preserved because paragraph structure is required for
the later hierarchical representation:

Article  
→ Paragraph  
→ Sentence

Line breaks must therefore not be globally collapsed into a single continuous
string.

### Markup

If article content contains clear HTML tags, those tags may be removed while
their textual content is preserved.

For example:

`<p>The situation changed.</p>`

may become:

`The situation changed.`

The sentence itself is not rewritten.

HTML entities such as `&amp;` may be converted to their textual form.

Markup removal is treated cautiously because angle brackets or unusual
characters can occasionally occur as meaningful text. The sanitisation audit
therefore records whether markup processing altered an article.

### Sanitisation Audit

For every article, the derived record records:

- original Media AI record identifier;
- source schema version;
- sanitisation policy version;
- original character count;
- sanitised character count;
- original and sanitised fingerprints;
- Unicode-normalisation status;
- removed null-character count;
- removed control-character count;
- HTML/markup modification status;
- whitespace modification status;
- whether the sanitised text differs from the original;
- magnitude of the character-count difference;
- warnings requiring later review.

The sanitised fingerprint is distinct from the original-source fingerprint.

The original fingerprint continues to identify the preserved acquisition text.
The derived fingerprint identifies the sanitised representation created in this
notebook.

### Conservative Review Threshold

Large transformations are not silently accepted.

If sanitisation changes an unexpectedly large proportion of the article, the
record is flagged for inspection. The article is not automatically interpreted
as invalid; the flag indicates that the technical transformation deserves human
review before modelling.

### Output

The output of this block is a set of derived sanitised records linked explicitly
to their original source records.

These records remain model-independent. No emotional weights, factological
weights, transformative states, recurrent states, confluent representations or
backbone tokens are created.

**Input:** `VERIFIED_CANDIDATE_RECORDS` from Block 3.

**Output:** Derived sanitised records and article-level sanitisation audit
metadata.

No original source record is overwritten and no GNews API request is performed
by this block.

```python
# =============================================================================
# Media AI — Technical Sanitisation and Derived Text Construction
# =============================================================================

import copy
import hashlib
import html
import re
import unicodedata
from datetime import datetime, timezone


# -----------------------------------------------------------------------------
# Sanitisation configuration
# -----------------------------------------------------------------------------

UNICODE_NORMALISATION_FORM = "NFC"

# Large changes are flagged rather than silently treated as routine.
MAX_CHARACTER_CHANGE_RATIO = 0.05


# -----------------------------------------------------------------------------
# Fingerprint helper
# -----------------------------------------------------------------------------

def calculate_derived_text_fingerprint(text):
    """
    Calculate a SHA-256 fingerprint for a derived textual representation.

    Unlike the source fingerprint from Notebook 01, this fingerprint identifies
    the sanitised text produced by Notebook 02.
    """

    if not isinstance(text, str):
        return None

    if not text:
        return None

    return hashlib.sha256(
        text.encode("utf-8")
    ).hexdigest()


# -----------------------------------------------------------------------------
# Control-character identification
# -----------------------------------------------------------------------------

def remove_disallowed_control_characters(text):
    """
    Remove non-printing Unicode control characters while preserving
    ordinary newline and tab structure where relevant.
    """

    cleaned_characters = []
    removed_count = 0

    allowed_controls = {
        "\n",
        "\t",
    }

    for character in text:

        if character in allowed_controls:
            cleaned_characters.append(
                character
            )
            continue

        category = unicodedata.category(
            character
        )

        if category == "Cc":
            removed_count += 1
            continue

        cleaned_characters.append(
            character
        )

    return (
        "".join(cleaned_characters),
        removed_count,
    )


# -----------------------------------------------------------------------------
# Conservative markup handling
# -----------------------------------------------------------------------------

HTML_TAG_PATTERN = re.compile(
    r"</?[A-Za-z][^>]*>"
)


def remove_confirmed_html_markup(text):
    """
    Remove clearly identifiable HTML tags while preserving their textual
    content.

    This deliberately avoids treating every angle-bracket expression as HTML.
    """

    matches = HTML_TAG_PATTERN.findall(
        text
    )

    cleaned = HTML_TAG_PATTERN.sub(
        "",
        text,
    )

    return (
        cleaned,
        len(matches),
    )


# -----------------------------------------------------------------------------
# Conservative whitespace normalisation
# -----------------------------------------------------------------------------

def normalise_text_whitespace(text):
    """
    Normalise horizontal whitespace while preserving paragraph structure.
    """

    original = text

    # Standardise line endings first.
    text = text.replace(
        "\r\n",
        "\n",
    )

    text = text.replace(
        "\r",
        "\n",
    )

    # Collapse repeated spaces/tabs within each line.
    lines = []

    for line in text.split("\n"):

        line = re.sub(
            r"[ \t]+",
            " ",
            line,
        ).strip()

        lines.append(
            line
        )

    # Preserve paragraph separation while avoiding excessive blank lines.
    text = "\n".join(
        lines
    )

    text = re.sub(
        r"\n{3,}",
        "\n\n",
        text,
    )

    text = text.strip()

    return (
        text,
        text != original,
    )


# -----------------------------------------------------------------------------
# Sanitise one article
# -----------------------------------------------------------------------------

def sanitise_article_text(source_text):
    """
    Apply conservative technical sanitisation to one article.

    Returns the sanitised text and a detailed transformation audit.
    """

    if not isinstance(source_text, str):
        source_text = ""

    original_text = source_text
    original_character_count = len(
        original_text
    )

    audit = {
        "unicode_normalisation_applied":
            False,

        "null_characters_removed":
            0,

        "control_characters_removed":
            0,

        "html_tags_removed":
            0,

        "html_entities_decoded":
            False,

        "whitespace_modified":
            False,

        "warnings":
            [],
    }

    # -------------------------------------------------------------------------
    # Null characters
    # -------------------------------------------------------------------------

    null_count = source_text.count(
        "\x00"
    )

    if null_count:
        source_text = source_text.replace(
            "\x00",
            "",
        )

    audit[
        "null_characters_removed"
    ] = null_count

    # -------------------------------------------------------------------------
    # Unicode normalisation
    # -------------------------------------------------------------------------

    unicode_normalised = unicodedata.normalize(
        UNICODE_NORMALISATION_FORM,
        source_text,
    )

    audit[
        "unicode_normalisation_applied"
    ] = (
        unicode_normalised
        != source_text
    )

    source_text = unicode_normalised

    # -------------------------------------------------------------------------
    # Non-printing control characters
    # -------------------------------------------------------------------------

    (
        source_text,
        control_count,
    ) = remove_disallowed_control_characters(
        source_text
    )

    audit[
        "control_characters_removed"
    ] = control_count

    # -------------------------------------------------------------------------
    # HTML entities
    # -------------------------------------------------------------------------

    entity_decoded_text = html.unescape(
        source_text
    )

    audit[
        "html_entities_decoded"
    ] = (
        entity_decoded_text
        != source_text
    )

    source_text = entity_decoded_text

    # -------------------------------------------------------------------------
    # Clear HTML markup
    # -------------------------------------------------------------------------

    (
        source_text,
        html_tag_count,
    ) = remove_confirmed_html_markup(
        source_text
    )

    audit[
        "html_tags_removed"
    ] = html_tag_count

    # -------------------------------------------------------------------------
    # Conservative whitespace normalisation
    # -------------------------------------------------------------------------

    (
        source_text,
        whitespace_modified,
    ) = normalise_text_whitespace(
        source_text
    )

    audit[
        "whitespace_modified"
    ] = whitespace_modified

    # -------------------------------------------------------------------------
    # Difference analysis
    # -------------------------------------------------------------------------

    sanitised_character_count = len(
        source_text
    )

    character_difference = (
        sanitised_character_count
        - original_character_count
    )

    absolute_character_difference = abs(
        character_difference
    )

    if original_character_count > 0:

        character_change_ratio = (
            absolute_character_difference
            / original_character_count
        )

    else:
        character_change_ratio = 0.0

    text_changed = (
        source_text
        != original_text
    )

    if (
        character_change_ratio
        > MAX_CHARACTER_CHANGE_RATIO
    ):
        audit["warnings"].append(
            "large_character_count_change"
        )

    if not source_text:
        audit["warnings"].append(
            "sanitised_text_empty"
        )

    audit.update({
        "original_character_count":
            original_character_count,

        "sanitised_character_count":
            sanitised_character_count,

        "character_difference":
            character_difference,

        "absolute_character_difference":
            absolute_character_difference,

        "character_change_ratio":
            character_change_ratio,

        "text_changed":
            text_changed,
    })

    return (
        source_text,
        audit,
    )


# -----------------------------------------------------------------------------
# Construct derived sanitised records
# -----------------------------------------------------------------------------

SANITISED_RECORDS = []

for source_record in VERIFIED_CANDIDATE_RECORDS:

    provider_payload = source_record.get(
        "provider_payload",
        {}
    )

    original_text = provider_payload.get(
        "content",
        ""
    )

    (
        sanitised_text,
        sanitisation_audit,
    ) = sanitise_article_text(
        original_text
    )

    original_fingerprint = source_record.get(
        "text_fingerprint"
    )

    sanitised_fingerprint = (
        calculate_derived_text_fingerprint(
            sanitised_text
        )
    )

    derived_record = {
        "media_ai_record_id":
            source_record.get(
                "media_ai_record_id"
            ),

        "provider_article_id":
            (
                source_record.get(
                    "provider_article_id"
                )
                or provider_payload.get(
                    "id"
                )
            ),

        "source_schema_version":
            source_record.get(
                "schema_version"
            ),

        "preparation_schema_version":
            PREPARATION_SCHEMA_VERSION,

        "sanitisation_policy_version":
            SANITISATION_POLICY_VERSION,

        "derived_at_utc":
            datetime.now(
                timezone.utc
            ).isoformat(),

        "title":
            (
                source_record.get(
                    "title"
                )
                or provider_payload.get(
                    "title"
                )
            ),

        "language":
            (
                source_record.get(
                    "language"
                )
                or provider_payload.get(
                    "lang"
                )
            ),

        "source_url":
            (
                source_record.get(
                    "source_url"
                )
                or provider_payload.get(
                    "url"
                )
            ),

        # ---------------------------------------------------------------------
        # Traceability
        # ---------------------------------------------------------------------

        "original_text_fingerprint":
            original_fingerprint,

        "sanitised_text_fingerprint":
            sanitised_fingerprint,

        # ---------------------------------------------------------------------
        # Derived text
        # ---------------------------------------------------------------------

        "sanitised_text":
            sanitised_text,

        # ---------------------------------------------------------------------
        # Transformation audit
        # ---------------------------------------------------------------------

        "sanitisation_audit":
            sanitisation_audit,
    }

    SANITISED_RECORDS.append(
        derived_record
    )


# -----------------------------------------------------------------------------
# Corpus-level sanitisation summary
# -----------------------------------------------------------------------------

sanitised_record_count = len(
    SANITISED_RECORDS
)

changed_record_count = sum(
    1
    for record in SANITISED_RECORDS
    if record[
        "sanitisation_audit"
    ][
        "text_changed"
    ]
)

warning_record_count = sum(
    1
    for record in SANITISED_RECORDS
    if record[
        "sanitisation_audit"
    ][
        "warnings"
    ]
)

total_null_characters_removed = sum(
    record[
        "sanitisation_audit"
    ][
        "null_characters_removed"
    ]
    for record in SANITISED_RECORDS
)

total_control_characters_removed = sum(
    record[
        "sanitisation_audit"
    ][
        "control_characters_removed"
    ]
    for record in SANITISED_RECORDS
)

total_html_tags_removed = sum(
    record[
        "sanitisation_audit"
    ][
        "html_tags_removed"
    ]
    for record in SANITISED_RECORDS
)


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Technical Sanitisation")
print("=" * 72)

print(
    f"Verified source records    : "
    f"{len(VERIFIED_CANDIDATE_RECORDS)}"
)

print(
    f"Derived sanitised records  : "
    f"{sanitised_record_count}"
)

print(
    f"Records with text changes  : "
    f"{changed_record_count}"
)

print(
    f"Records requiring review   : "
    f"{warning_record_count}"
)

print(
    f"Null characters removed    : "
    f"{total_null_characters_removed}"
)

print(
    f"Control characters removed : "
    f"{total_control_characters_removed}"
)

print(
    f"HTML tags removed          : "
    f"{total_html_tags_removed}"
)

print("=" * 72)

print(
    "Technical sanitisation completed successfully."
)

print(
    "Derived sanitised records were created separately from source records."
)

print(
    "Original source records remain unchanged."
)

print(
    "No model-specific tokenisation has been performed."
)

print(
    "No GNews API requests have been made by this block."
)
```

## Sanitisation Audit and Difference Analysis

This block audits the derived textual representations created during technical
sanitisation.

Sanitisation is not expected to modify every article. An unchanged derived text
is a valid and desirable outcome when the source text is already technically
clean.

The purpose of this block is therefore not to search for additional
transformations, but to verify that the sanitisation process behaved as
intended.

For every article, the audit compares the original source representation with
the derived sanitised representation and evaluates:

- original and sanitised character counts;
- original and sanitised word counts;
- whether the textual representation changed;
- absolute character-count difference;
- proportional character-count difference;
- whether Unicode normalisation altered the representation;
- removed null and control characters;
- removed HTML tags;
- decoded HTML entities;
- whitespace modifications;
- sanitisation warnings;
- source and derived fingerprints.

### Fingerprint Interpretation

The original source fingerprint and the sanitised-text fingerprint serve
different purposes.

The source fingerprint was created during acquisition using the source
fingerprinting policy defined in Notebook 01.

The sanitised fingerprint identifies the exact derived text produced by
Notebook 02.

Because the two fingerprints may be calculated under different representation
rules, fingerprint equality is not by itself used to determine whether
sanitisation changed an article.

Instead, the audit compares the actual preserved source text with the sanitised
text directly.

If the texts are identical, the article is classified as unchanged.

### Audit Principle

The audit follows a conservative expectation:

> A technically clean article should pass through sanitisation without
> unnecessary modification.

Zero transformations are therefore not treated as a failure of sanitisation.
They indicate that the sanitisation policy inspected the record and found no
technical alteration necessary.

Any article carrying a sanitisation warning remains explicitly visible for
review before later preparation stages.

### Output

The block creates an article-level sanitisation audit and verifies that:

- every verified source record has exactly one derived sanitised record;
- record identifiers remain traceable;
- no derived record has unexpectedly lost its text;
- the recorded transformation status agrees with direct text comparison;
- all warning states remain visible.

No article text is modified by this block.

**Input:** `VERIFIED_CANDIDATE_RECORDS` and `SANITISED_RECORDS`.

**Output:** `SANITISATION_AUDIT`, corpus-level difference statistics and
sanitisation verification status.

No GNews API request is performed by this block.

```python
# =============================================================================
# Media AI — Sanitisation Audit and Difference Analysis
# =============================================================================

import re


# -----------------------------------------------------------------------------
# Audit helpers
# -----------------------------------------------------------------------------

def count_words_for_audit(text):
    """
    Count whitespace-delimited textual units for descriptive audit purposes.

    This is not model tokenisation and does not create a permanent tokenised
    representation.
    """

    if not isinstance(text, str):
        return 0

    return len(
        re.findall(
            r"\S+",
            text,
        )
    )


# -----------------------------------------------------------------------------
# Index verified source records
# -----------------------------------------------------------------------------

source_records_by_id = {
    record.get("media_ai_record_id"):
        record
    for record in VERIFIED_CANDIDATE_RECORDS
}


# -----------------------------------------------------------------------------
# Index sanitised records
# -----------------------------------------------------------------------------

sanitised_records_by_id = {
    record.get("media_ai_record_id"):
        record
    for record in SANITISED_RECORDS
}


# -----------------------------------------------------------------------------
# Traceability verification
# -----------------------------------------------------------------------------

source_ids = set(
    source_records_by_id.keys()
)

sanitised_ids = set(
    sanitised_records_by_id.keys()
)

missing_sanitised_ids = (
    source_ids
    - sanitised_ids
)

unexpected_sanitised_ids = (
    sanitised_ids
    - source_ids
)

record_mapping_complete = (
    not missing_sanitised_ids
    and not unexpected_sanitised_ids
    and len(VERIFIED_CANDIDATE_RECORDS)
    == len(SANITISED_RECORDS)
)

if not record_mapping_complete:
    raise ValueError(
        "Sanitised-record mapping does not match "
        "the verified source corpus."
    )


# -----------------------------------------------------------------------------
# Article-level sanitisation audit
# -----------------------------------------------------------------------------

SANITISATION_AUDIT = []

for media_ai_record_id in sorted(
    source_ids
):

    source_record = source_records_by_id[
        media_ai_record_id
    ]

    sanitised_record = sanitised_records_by_id[
        media_ai_record_id
    ]

    provider_payload = source_record.get(
        "provider_payload",
        {}
    )

    original_text = provider_payload.get(
        "content",
        ""
    )

    sanitised_text = sanitised_record.get(
        "sanitised_text",
        ""
    )

    recorded_audit = sanitised_record.get(
        "sanitisation_audit",
        {}
    )

    original_character_count = len(
        original_text
    )

    sanitised_character_count = len(
        sanitised_text
    )

    original_word_count = count_words_for_audit(
        original_text
    )

    sanitised_word_count = count_words_for_audit(
        sanitised_text
    )

    direct_text_changed = (
        original_text
        != sanitised_text
    )

    recorded_text_changed = recorded_audit.get(
        "text_changed",
        False,
    )

    transformation_status_matches = (
        direct_text_changed
        == recorded_text_changed
    )

    character_difference = (
        sanitised_character_count
        - original_character_count
    )

    absolute_character_difference = abs(
        character_difference
    )

    if original_character_count > 0:
        character_change_ratio = (
            absolute_character_difference
            / original_character_count
        )
    else:
        character_change_ratio = 0.0

    source_fingerprint = source_record.get(
        "text_fingerprint"
    )

    sanitised_fingerprint = sanitised_record.get(
        "sanitised_text_fingerprint"
    )

    sanitised_text_present = bool(
        sanitised_text
    )

    warnings = recorded_audit.get(
        "warnings",
        [],
    )

    audit_passed = all([
        sanitised_text_present,
        transformation_status_matches,
        isinstance(warnings, list),
    ])

    SANITISATION_AUDIT.append({
        "media_ai_record_id":
            media_ai_record_id,

        "provider_article_id":
            sanitised_record.get(
                "provider_article_id"
            ),

        "title":
            sanitised_record.get(
                "title"
            ),

        "source_schema_version":
            sanitised_record.get(
                "source_schema_version"
            ),

        "original_character_count":
            original_character_count,

        "sanitised_character_count":
            sanitised_character_count,

        "character_difference":
            character_difference,

        "absolute_character_difference":
            absolute_character_difference,

        "character_change_ratio":
            character_change_ratio,

        "original_word_count":
            original_word_count,

        "sanitised_word_count":
            sanitised_word_count,

        "direct_text_changed":
            direct_text_changed,

        "recorded_text_changed":
            recorded_text_changed,

        "transformation_status_matches":
            transformation_status_matches,

        "unicode_normalisation_applied":
            recorded_audit.get(
                "unicode_normalisation_applied",
                False,
            ),

        "null_characters_removed":
            recorded_audit.get(
                "null_characters_removed",
                0,
            ),

        "control_characters_removed":
            recorded_audit.get(
                "control_characters_removed",
                0,
            ),

        "html_tags_removed":
            recorded_audit.get(
                "html_tags_removed",
                0,
            ),

        "html_entities_decoded":
            recorded_audit.get(
                "html_entities_decoded",
                False,
            ),

        "whitespace_modified":
            recorded_audit.get(
                "whitespace_modified",
                False,
            ),

        "warnings":
            warnings,

        "source_fingerprint":
            source_fingerprint,

        "sanitised_fingerprint":
            sanitised_fingerprint,

        "audit_passed":
            audit_passed,
    })


# -----------------------------------------------------------------------------
# Corpus-level difference analysis
# -----------------------------------------------------------------------------

audited_record_count = len(
    SANITISATION_AUDIT
)

unchanged_record_count = sum(
    1
    for record in SANITISATION_AUDIT
    if not record[
        "direct_text_changed"
    ]
)

changed_record_count = (
    audited_record_count
    - unchanged_record_count
)

warning_record_count = sum(
    1
    for record in SANITISATION_AUDIT
    if record["warnings"]
)

transformation_mismatch_count = sum(
    1
    for record in SANITISATION_AUDIT
    if not record[
        "transformation_status_matches"
    ]
)

failed_audit_count = sum(
    1
    for record in SANITISATION_AUDIT
    if not record[
        "audit_passed"
    ]
)

total_original_characters = sum(
    record[
        "original_character_count"
    ]
    for record in SANITISATION_AUDIT
)

total_sanitised_characters = sum(
    record[
        "sanitised_character_count"
    ]
    for record in SANITISATION_AUDIT
)

total_original_words = sum(
    record[
        "original_word_count"
    ]
    for record in SANITISATION_AUDIT
)

total_sanitised_words = sum(
    record[
        "sanitised_word_count"
    ]
    for record in SANITISATION_AUDIT
)


# -----------------------------------------------------------------------------
# Final audit validation
# -----------------------------------------------------------------------------

ALL_SANITISATION_AUDITS_PASSED = (
    record_mapping_complete
    and failed_audit_count == 0
    and transformation_mismatch_count == 0
)

if not ALL_SANITISATION_AUDITS_PASSED:
    raise ValueError(
        "One or more sanitisation audit checks failed."
    )


# -----------------------------------------------------------------------------
# Compact article-level review
# -----------------------------------------------------------------------------

for index, record in enumerate(
    SANITISATION_AUDIT,
    start=1,
):

    print("-" * 72)

    print(
        f"Record {index}"
    )

    print(
        f"Media AI ID               : "
        f"{record['media_ai_record_id']}"
    )

    print(
        f"Title                     : "
        f"{record['title']}"
    )

    print(
        f"Text changed              : "
        f"{record['direct_text_changed']}"
    )

    print(
        f"Characters                : "
        f"{record['original_character_count']} "
        f"→ {record['sanitised_character_count']}"
    )

    print(
        f"Words                     : "
        f"{record['original_word_count']} "
        f"→ {record['sanitised_word_count']}"
    )

    print(
        f"Warnings                  : "
        f"{record['warnings']}"
    )


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Sanitisation Audit and Difference Analysis")
print("=" * 72)

print(
    f"Verified source records    : "
    f"{len(VERIFIED_CANDIDATE_RECORDS)}"
)

print(
    f"Sanitised records audited  : "
    f"{audited_record_count}"
)

print(
    f"Record mapping complete    : "
    f"{record_mapping_complete}"
)

print(
    f"Unchanged records          : "
    f"{unchanged_record_count}"
)

print(
    f"Changed records            : "
    f"{changed_record_count}"
)

print(
    f"Records with warnings      : "
    f"{warning_record_count}"
)

print(
    f"Audit-status mismatches     : "
    f"{transformation_mismatch_count}"
)

print(
    f"Original characters        : "
    f"{total_original_characters}"
)

print(
    f"Sanitised characters       : "
    f"{total_sanitised_characters}"
)

print(
    f"Original words             : "
    f"{total_original_words}"
)

print(
    f"Sanitised words            : "
    f"{total_sanitised_words}"
)

print(
    f"All sanitisation audits pass: "
    f"{ALL_SANITISATION_AUDITS_PASSED}"
)

print("=" * 72)

print(
    "Sanitisation audit completed successfully."
)

print(
    "Unchanged records are retained as valid sanitisation outcomes."
)

print(
    "No article text has been modified by this audit block."
)

print(
    "No model-specific tokenisation has been performed."
)

print(
    "No GNews API requests have been made by this block."
)
```

## Near-Duplicate Detection and Content-Overlap Review

Exact duplicates were already identified during source-integrity verification.
News corpora may nevertheless contain articles that are not textually identical
but reproduce substantially the same content.

Examples include:

- syndicated copies of the same report;
- republished wire-service material;
- articles with minor editorial modifications;
- shortened or extended versions of the same source text;
- records containing substantially overlapping passages.

Such records may distort later analysis if repeated coverage is unintentionally
treated as independent textual evidence.

This block therefore performs a **near-duplicate assessment** on the sanitised
article representations.

### Detection Principle

Near-duplicate detection is performed using lexical similarity rather than
semantic interpretation.

Each article is represented as a set of normalised word n-grams. Pairwise
Jaccard similarity is then calculated between article representations.

For two n-gram sets \(A\) and \(B\), Jaccard similarity is defined as the size
of their intersection divided by the size of their union.

A value close to:

- `0` indicates little lexical overlap;
- `1` indicates very strong lexical overlap.

This measure is intended to identify substantial textual reuse. It is not a
measure of emotional, political, factual or semantic similarity.

Two articles discussing the same event may therefore remain distinct when their
actual textual representations differ substantially.

### Word N-Grams

Single words alone can produce misleading similarity because ordinary news
articles naturally share common vocabulary.

The assessment therefore uses consecutive word sequences rather than isolated
words.

The pilot configuration uses **5-word n-grams**, which provides a conservative
signal of actual textual overlap while remaining interpretable.

No permanent model tokenisation is created. The temporary lexical units used
here exist only for duplicate assessment.

### Similarity Thresholds

Pairwise similarity is classified into review bands.

The thresholds are diagnostic rather than universal statistical constants:

- **low overlap** — no duplicate concern;
- **moderate overlap** — potentially related wording;
- **high overlap** — manual review recommended;
- **very high overlap** — strong near-duplicate candidate.

No article is automatically removed on the basis of a similarity score.

This distinction is important because repeated or similar coverage can itself
be meaningful in media research. Deduplication decisions should therefore remain
explicit and auditable.

### Auditability

For every article pair, the block records:

- both Media AI record identifiers;
- article titles;
- n-gram configuration;
- number of lexical n-grams in each article;
- shared n-gram count;
- Jaccard similarity;
- overlap classification;
- whether manual review is recommended.

The full pairwise review is retained for later quality diagnostics.

### Output

The block produces:

- `NEAR_DUPLICATE_REVIEW`, containing every article pair;
- `NEAR_DUPLICATE_CANDIDATES`, containing only pairs requiring review;
- corpus-level overlap diagnostics.

The sanitised records themselves are not modified or removed.

**Input:** `SANITISED_RECORDS`.

**Output:** Pairwise lexical-overlap diagnostics and explicit near-duplicate
review candidates.

No semantic filtering, emotional weighting, model-specific tokenisation or
GNews API request is performed by this block.

```python
# =============================================================================
# Media AI — Near-Duplicate Detection and Content-Overlap Review
# =============================================================================

import itertools
import re


# -----------------------------------------------------------------------------
# Near-duplicate configuration
# -----------------------------------------------------------------------------

NEAR_DUPLICATE_NGRAM_SIZE = 5

MODERATE_OVERLAP_THRESHOLD = 0.20
HIGH_OVERLAP_THRESHOLD = 0.40
VERY_HIGH_OVERLAP_THRESHOLD = 0.70


# -----------------------------------------------------------------------------
# Temporary lexical normalisation
# -----------------------------------------------------------------------------

def normalise_for_overlap(text):
    """
    Create a temporary lexical representation for near-duplicate assessment.

    This representation is used only for overlap detection and does not replace
    or modify the sanitised article text.
    """

    if not isinstance(text, str):
        return []

    text = text.lower()

    tokens = re.findall(
        r"\b[\w'-]+\b",
        text,
        flags=re.UNICODE,
    )

    return tokens


# -----------------------------------------------------------------------------
# Word n-gram construction
# -----------------------------------------------------------------------------

def create_word_ngrams(
    tokens,
    n=NEAR_DUPLICATE_NGRAM_SIZE,
):
    """
    Construct temporary consecutive word n-grams.
    """

    if len(tokens) < n:
        return set()

    return {
        tuple(
            tokens[index:index + n]
        )
        for index in range(
            len(tokens) - n + 1
        )
    }


# -----------------------------------------------------------------------------
# Jaccard similarity
# -----------------------------------------------------------------------------

def calculate_jaccard_similarity(
    first_set,
    second_set,
):
    """
    Calculate Jaccard similarity between two lexical n-gram sets.
    """

    if not first_set and not second_set:
        return 1.0

    if not first_set or not second_set:
        return 0.0

    intersection = (
        first_set
        & second_set
    )

    union = (
        first_set
        | second_set
    )

    return (
        len(intersection)
        / len(union)
    )


# -----------------------------------------------------------------------------
# Overlap classification
# -----------------------------------------------------------------------------

def classify_overlap(similarity):
    """
    Convert lexical similarity into an interpretable review category.
    """

    if similarity >= VERY_HIGH_OVERLAP_THRESHOLD:
        return (
            "very_high",
            True,
        )

    if similarity >= HIGH_OVERLAP_THRESHOLD:
        return (
            "high",
            True,
        )

    if similarity >= MODERATE_OVERLAP_THRESHOLD:
        return (
            "moderate",
            False,
        )

    return (
        "low",
        False,
    )


# -----------------------------------------------------------------------------
# Construct temporary lexical representations
# -----------------------------------------------------------------------------

overlap_representations = {}

for record in SANITISED_RECORDS:

    media_ai_record_id = record.get(
        "media_ai_record_id"
    )

    sanitised_text = record.get(
        "sanitised_text",
        ""
    )

    tokens = normalise_for_overlap(
        sanitised_text
    )

    ngrams = create_word_ngrams(
        tokens
    )

    overlap_representations[
        media_ai_record_id
    ] = {
        "tokens":
            tokens,

        "ngrams":
            ngrams,
    }


# -----------------------------------------------------------------------------
# Pairwise near-duplicate review
# -----------------------------------------------------------------------------

NEAR_DUPLICATE_REVIEW = []

for first_record, second_record in itertools.combinations(
    SANITISED_RECORDS,
    2,
):

    first_id = first_record.get(
        "media_ai_record_id"
    )

    second_id = second_record.get(
        "media_ai_record_id"
    )

    first_ngrams = overlap_representations[
        first_id
    ][
        "ngrams"
    ]

    second_ngrams = overlap_representations[
        second_id
    ][
        "ngrams"
    ]

    shared_ngrams = (
        first_ngrams
        & second_ngrams
    )

    similarity = calculate_jaccard_similarity(
        first_ngrams,
        second_ngrams,
    )

    (
        overlap_classification,
        manual_review_recommended,
    ) = classify_overlap(
        similarity
    )

    pair_result = {
        "record_id_1":
            first_id,

        "record_id_2":
            second_id,

        "title_1":
            first_record.get(
                "title"
            ),

        "title_2":
            second_record.get(
                "title"
            ),

        "ngram_size":
            NEAR_DUPLICATE_NGRAM_SIZE,

        "ngrams_1":
            len(first_ngrams),

        "ngrams_2":
            len(second_ngrams),

        "shared_ngrams":
            len(shared_ngrams),

        "jaccard_similarity":
            similarity,

        "overlap_classification":
            overlap_classification,

        "manual_review_recommended":
            manual_review_recommended,
    }

    NEAR_DUPLICATE_REVIEW.append(
        pair_result
    )


# -----------------------------------------------------------------------------
# Explicit review candidates
# -----------------------------------------------------------------------------

NEAR_DUPLICATE_CANDIDATES = [
    pair
    for pair in NEAR_DUPLICATE_REVIEW
    if pair[
        "manual_review_recommended"
    ]
]


# -----------------------------------------------------------------------------
# Corpus-level diagnostics
# -----------------------------------------------------------------------------

pair_count = len(
    NEAR_DUPLICATE_REVIEW
)

moderate_overlap_count = sum(
    1
    for pair in NEAR_DUPLICATE_REVIEW
    if pair[
        "overlap_classification"
    ] == "moderate"
)

high_overlap_count = sum(
    1
    for pair in NEAR_DUPLICATE_REVIEW
    if pair[
        "overlap_classification"
    ] == "high"
)

very_high_overlap_count = sum(
    1
    for pair in NEAR_DUPLICATE_REVIEW
    if pair[
        "overlap_classification"
    ] == "very_high"
)

if NEAR_DUPLICATE_REVIEW:

    maximum_similarity = max(
        pair[
            "jaccard_similarity"
        ]
        for pair in NEAR_DUPLICATE_REVIEW
    )

else:
    maximum_similarity = 0.0


# -----------------------------------------------------------------------------
# Pairwise review output
# -----------------------------------------------------------------------------

for index, pair in enumerate(
    NEAR_DUPLICATE_REVIEW,
    start=1,
):

    print("-" * 72)

    print(
        f"Pair {index}"
    )

    print(
        f"Article 1                 : "
        f"{pair['title_1']}"
    )

    print(
        f"Article 2                 : "
        f"{pair['title_2']}"
    )

    print(
        f"Shared {NEAR_DUPLICATE_NGRAM_SIZE}-grams            : "
        f"{pair['shared_ngrams']}"
    )

    print(
        f"Jaccard similarity        : "
        f"{pair['jaccard_similarity']:.4f}"
    )

    print(
        f"Overlap classification    : "
        f"{pair['overlap_classification']}"
    )

    print(
        f"Manual review recommended : "
        f"{pair['manual_review_recommended']}"
    )


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Near-Duplicate Detection")
print("=" * 72)

print(
    f"Sanitised records          : "
    f"{len(SANITISED_RECORDS)}"
)

print(
    f"Article pairs assessed     : "
    f"{pair_count}"
)

print(
    f"N-gram size                : "
    f"{NEAR_DUPLICATE_NGRAM_SIZE}"
)

print(
    f"Moderate-overlap pairs     : "
    f"{moderate_overlap_count}"
)

print(
    f"High-overlap pairs         : "
    f"{high_overlap_count}"
)

print(
    f"Very-high-overlap pairs    : "
    f"{very_high_overlap_count}"
)

print(
    f"Pairs requiring review     : "
    f"{len(NEAR_DUPLICATE_CANDIDATES)}"
)

print(
    f"Maximum similarity         : "
    f"{maximum_similarity:.4f}"
)

print("=" * 72)

print(
    "Near-duplicate assessment completed successfully."
)

print(
    "No records have been removed or modified."
)

print(
    "Similarity scores are diagnostic and do not represent semantic similarity."
)

print(
    "No permanent tokenisation has been created."
)

print(
    "No GNews API requests have been made by this block."
)
```

## Hierarchical Article Segmentation

This block converts each sanitised article into an explicit hierarchical
representation while preserving the complete article text.

The structural hierarchy is:

**Article → Paragraph → Sentence**

This hierarchy provides the observational structure required by later Media AI
stages without yet assigning semantic, emotional, factological, recurrent or
confluent representations.

Segmentation creates derived structural units. It does not replace the complete
sanitised article representation.

Every article therefore retains:

- its complete sanitised text;
- its original Media AI record identifier;
- its source and sanitised fingerprints;
- its preparation and sanitisation metadata.

Paragraphs and sentences are added as derived structural representations linked
back to the article.

Paragraph boundaries are derived from preserved blank-line separation in the
sanitised text. Empty fragments are discarded, but non-empty paragraph content
is not rewritten.

Within each paragraph, sentence boundaries are identified using conservative,
deterministic punctuation rules. The pilot segmenter protects decimal points and
selected common abbreviations before sentence splitting. Temporary protection
markers are removed before any derived sentence record is stored.

The protection mechanism preserves the exact matched source text, including
capitalisation. For example, `Dr.` remains `Dr.` and `U.S.` remains `U.S.`
after segmentation.

The purpose is to establish a reproducible structural baseline rather than to
claim linguistically perfect sentence-boundary detection.

Each structural unit receives a stable identifier derived from its parent Media
AI record and structural position:

`media_ai_record_id`

→ `paragraph_id`

→ `sentence_id`

Paragraph and sentence order are retained through explicit zero-based indices.

Segmentation must not silently change or lose article content. After segmentation,
the article is reconstructed from the derived sentence hierarchy and compared
with a conservatively normalised version of the sanitised source text.

A reconstruction mismatch is treated as an integrity failure because it indicates
that structural processing altered, removed or reordered text.

The structural units created here are not backbone tokens. No transformer
tokenizer, embedding model, emotional classifier, factological model, recurrent
state model or confluent representation is applied.

Later modelling stages may tokenize these textual representations on demand
using the tokenizer associated with the selected backbone.

**Input:** `SANITISED_RECORDS`.

**Output:** `SEGMENTED_RECORDS`, `PARAGRAPH_RECORDS`,
`SENTENCE_RECORDS`, and structural integrity diagnostics.

No permanent model tokenisation, emotional filtering or GNews API request is
performed by this block.

```python
# =============================================================================
# Media AI — Hierarchical Article Segmentation
# =============================================================================

import re


# -----------------------------------------------------------------------------
# Segmentation configuration
# -----------------------------------------------------------------------------

SEGMENTATION_POLICY_VERSION = "1.0"

PROTECTED_ABBREVIATIONS = {
    "mr.",
    "mrs.",
    "ms.",
    "dr.",
    "prof.",
    "sr.",
    "jr.",
    "st.",
    "vs.",
    "etc.",
    "e.g.",
    "i.e.",
    "u.s.",
    "u.k.",
}


# -----------------------------------------------------------------------------
# Paragraph segmentation
# -----------------------------------------------------------------------------

def segment_paragraphs(text):
    """
    Split sanitised article text on preserved blank-line boundaries.

    Non-empty paragraph content is retained without linguistic rewriting.
    """

    if not isinstance(text, str):
        return []

    paragraphs = re.split(
        r"\n\s*\n+",
        text.strip(),
    )

    return [
        paragraph.strip()
        for paragraph in paragraphs
        if paragraph.strip()
    ]


# -----------------------------------------------------------------------------
# Sentence-internal punctuation protection
# -----------------------------------------------------------------------------

def protect_sentence_internal_periods(text):
    """
    Protect periods that should not normally create sentence boundaries.

    The exact matched source text is preserved, including original
    capitalisation.
    """

    protected = text

    # -------------------------------------------------------------------------
    # Protect decimal points
    # -------------------------------------------------------------------------

    protected = re.sub(
        r"(?<=\d)\.(?=\d)",
        "<MEDIA_AI_DECIMAL>",
        protected,
    )

    # -------------------------------------------------------------------------
    # Protect common abbreviations
    # -------------------------------------------------------------------------

    for abbreviation in sorted(
        PROTECTED_ABBREVIATIONS,
        key=len,
        reverse=True,
    ):

        pattern = re.compile(
            re.escape(abbreviation),
            flags=re.IGNORECASE,
        )

        protected = pattern.sub(
            lambda match: match.group(0).replace(
                ".",
                "<MEDIA_AI_PERIOD>",
            ),
            protected,
        )

    return protected


def restore_sentence_internal_periods(text):
    """
    Restore temporary punctuation placeholders.
    """

    return (
        text
        .replace(
            "<MEDIA_AI_DECIMAL>",
            ".",
        )
        .replace(
            "<MEDIA_AI_PERIOD>",
            ".",
        )
    )


# -----------------------------------------------------------------------------
# Sentence segmentation
# -----------------------------------------------------------------------------

def segment_sentences(paragraph):
    """
    Segment one paragraph using deterministic punctuation rules.

    This is a reproducible structural baseline rather than a claim of perfect
    linguistic sentence-boundary detection.
    """

    if not isinstance(paragraph, str):
        return []

    paragraph = paragraph.strip()

    if not paragraph:
        return []

    protected = protect_sentence_internal_periods(
        paragraph
    )

    fragments = re.split(
        r"(?<=[.!?])\s+(?=\S)",
        protected,
    )

    sentences = []

    for fragment in fragments:

        restored = restore_sentence_internal_periods(
            fragment
        ).strip()

        if restored:
            sentences.append(
                restored
            )

    return sentences


# -----------------------------------------------------------------------------
# Reconstruction helper
# -----------------------------------------------------------------------------

def normalise_for_reconstruction(text):
    """
    Create a conservative comparison representation for integrity checks.

    Stored text is not modified by this function.
    """

    if not isinstance(text, str):
        return ""

    return re.sub(
        r"\s+",
        " ",
        text,
    ).strip()


# -----------------------------------------------------------------------------
# Construct hierarchical records
# -----------------------------------------------------------------------------

SEGMENTED_RECORDS = []
PARAGRAPH_RECORDS = []
SENTENCE_RECORDS = []

for record in SANITISED_RECORDS:

    media_ai_record_id = record.get(
        "media_ai_record_id"
    )

    sanitised_text = record.get(
        "sanitised_text",
        ""
    )

    paragraphs = segment_paragraphs(
        sanitised_text
    )

    hierarchical_paragraphs = []

    for paragraph_index, paragraph_text in enumerate(
        paragraphs
    ):

        paragraph_id = (
            f"{media_ai_record_id}"
            f"::p{paragraph_index:04d}"
        )

        sentences = segment_sentences(
            paragraph_text
        )

        hierarchical_sentences = []

        for sentence_index, sentence_text in enumerate(
            sentences
        ):

            sentence_id = (
                f"{paragraph_id}"
                f"::s{sentence_index:04d}"
            )

            sentence_record = {
                "media_ai_record_id":
                    media_ai_record_id,

                "paragraph_id":
                    paragraph_id,

                "sentence_id":
                    sentence_id,

                "paragraph_index":
                    paragraph_index,

                "sentence_index":
                    sentence_index,

                "text":
                    sentence_text,

                "character_count":
                    len(sentence_text),

                "temporary_word_count":
                    len(
                        re.findall(
                            r"\S+",
                            sentence_text,
                        )
                    ),
            }

            hierarchical_sentences.append(
                sentence_record
            )

            SENTENCE_RECORDS.append(
                sentence_record.copy()
            )

        paragraph_record = {
            "media_ai_record_id":
                media_ai_record_id,

            "paragraph_id":
                paragraph_id,

            "paragraph_index":
                paragraph_index,

            "text":
                paragraph_text,

            "character_count":
                len(paragraph_text),

            "sentence_count":
                len(
                    hierarchical_sentences
                ),

            "sentences":
                hierarchical_sentences,
        }

        hierarchical_paragraphs.append(
            paragraph_record
        )

        PARAGRAPH_RECORDS.append({
            key: value
            for key, value
            in paragraph_record.items()
            if key != "sentences"
        })

    # -------------------------------------------------------------------------
    # Reconstruction verification
    # -------------------------------------------------------------------------

    reconstructed_paragraphs = []

    for paragraph in hierarchical_paragraphs:

        reconstructed_paragraph = " ".join(
            sentence["text"]
            for sentence in paragraph[
                "sentences"
            ]
        )

        reconstructed_paragraphs.append(
            reconstructed_paragraph
        )

    reconstructed_text = "\n\n".join(
        reconstructed_paragraphs
    )

    reconstruction_matches = (
        normalise_for_reconstruction(
            sanitised_text
        )
        ==
        normalise_for_reconstruction(
            reconstructed_text
        )
    )

    segmentation_warnings = []

    if not paragraphs:
        segmentation_warnings.append(
            "no_paragraphs_detected"
        )

    if sanitised_text and not any(
        paragraph["sentences"]
        for paragraph
        in hierarchical_paragraphs
    ):
        segmentation_warnings.append(
            "no_sentences_detected"
        )

    if not reconstruction_matches:
        segmentation_warnings.append(
            "reconstruction_mismatch"
        )

    segmented_record = {
        "media_ai_record_id":
            media_ai_record_id,

        "provider_article_id":
            record.get(
                "provider_article_id"
            ),

        "title":
            record.get(
                "title"
            ),

        "language":
            record.get(
                "language"
            ),

        "source_url":
            record.get(
                "source_url"
            ),

        "source_schema_version":
            record.get(
                "source_schema_version"
            ),

        "preparation_schema_version":
            record.get(
                "preparation_schema_version"
            ),

        "sanitisation_policy_version":
            record.get(
                "sanitisation_policy_version"
            ),

        "segmentation_policy_version":
            SEGMENTATION_POLICY_VERSION,

        "original_text_fingerprint":
            record.get(
                "original_text_fingerprint"
            ),

        "sanitised_text_fingerprint":
            record.get(
                "sanitised_text_fingerprint"
            ),

        "sanitised_text":
            sanitised_text,

        "paragraphs":
            hierarchical_paragraphs,

        "segmentation_audit": {
            "paragraph_count":
                len(
                    hierarchical_paragraphs
                ),

            "sentence_count":
                sum(
                    len(
                        paragraph[
                            "sentences"
                        ]
                    )
                    for paragraph
                    in hierarchical_paragraphs
                ),

            "reconstruction_matches":
                reconstruction_matches,

            "warnings":
                segmentation_warnings,
        },
    }

    SEGMENTED_RECORDS.append(
        segmented_record
    )


# -----------------------------------------------------------------------------
# Corpus-level segmentation diagnostics
# -----------------------------------------------------------------------------

segmented_article_count = len(
    SEGMENTED_RECORDS
)

paragraph_count = len(
    PARAGRAPH_RECORDS
)

sentence_count = len(
    SENTENCE_RECORDS
)

reconstruction_mismatch_count = sum(
    1
    for record in SEGMENTED_RECORDS
    if not record[
        "segmentation_audit"
    ][
        "reconstruction_matches"
    ]
)

warning_record_count = sum(
    1
    for record in SEGMENTED_RECORDS
    if record[
        "segmentation_audit"
    ][
        "warnings"
    ]
)

articles_without_paragraphs = sum(
    1
    for record in SEGMENTED_RECORDS
    if record[
        "segmentation_audit"
    ][
        "paragraph_count"
    ] == 0
)

articles_without_sentences = sum(
    1
    for record in SEGMENTED_RECORDS
    if record[
        "segmentation_audit"
    ][
        "sentence_count"
    ] == 0
)


# -----------------------------------------------------------------------------
# Fail safely on structural loss
# -----------------------------------------------------------------------------

if segmented_article_count != len(
    SANITISED_RECORDS
):
    raise ValueError(
        "Segmented article count does not match "
        "the sanitised corpus."
    )

if reconstruction_mismatch_count > 0:
    raise ValueError(
        "One or more segmented articles could not be "
        "reconstructed from their derived hierarchy."
    )


# -----------------------------------------------------------------------------
# Article-level review
# -----------------------------------------------------------------------------

for index, record in enumerate(
    SEGMENTED_RECORDS,
    start=1,
):

    audit = record[
        "segmentation_audit"
    ]

    print("-" * 72)

    print(
        f"Article {index}"
    )

    print(
        f"Media AI ID               : "
        f"{record['media_ai_record_id']}"
    )

    print(
        f"Title                     : "
        f"{record['title']}"
    )

    print(
        f"Paragraphs                : "
        f"{audit['paragraph_count']}"
    )

    print(
        f"Sentences                 : "
        f"{audit['sentence_count']}"
    )

    print(
        f"Reconstruction matches    : "
        f"{audit['reconstruction_matches']}"
    )

    print(
        f"Warnings                  : "
        f"{audit['warnings']}"
    )


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Hierarchical Article Segmentation")
print("=" * 72)

print(
    f"Sanitised articles         : "
    f"{len(SANITISED_RECORDS)}"
)

print(
    f"Segmented articles         : "
    f"{segmented_article_count}"
)

print(
    f"Paragraph records          : "
    f"{paragraph_count}"
)

print(
    f"Sentence records           : "
    f"{sentence_count}"
)

print(
    f"Articles without paragraphs: "
    f"{articles_without_paragraphs}"
)

print(
    f"Articles without sentences : "
    f"{articles_without_sentences}"
)

print(
    f"Reconstruction mismatches  : "
    f"{reconstruction_mismatch_count}"
)

print(
    f"Records with warnings      : "
    f"{warning_record_count}"
)

print("=" * 72)

print(
    "Hierarchical article segmentation completed successfully."
)

print(
    "Complete sanitised article text remains preserved."
)

print(
    "Paragraph and sentence representations remain traceable to source records."
)

print(
    "No permanent model-specific tokenisation has been created."
)

print(
    "No emotional or semantic pre-filtering has been performed."
)

print(
    "No GNews API requests have been made by this block."
)
```

## Dataset Quality Diagnostics

This block evaluates the structural quality of the prepared corpus before it is
persisted and handed to Notebook 03.

The objective is descriptive rather than corrective.

No article, paragraph or sentence is removed automatically because it is short,
long, unusual or associated with a warning. The diagnostics identify structural
conditions that may influence later supervision or modelling and make those
conditions explicit before the next research stage begins.

The assessment combines information generated throughout Notebook 02:

- source-integrity verification;
- sanitisation audit;
- near-duplicate assessment;
- hierarchical segmentation.

The diagnostics examine the corpus at three structural levels:

**Article**

Article-level diagnostics include complete text length, paragraph count,
sentence count, source schema version, sanitisation warnings, segmentation
warnings and whether the article participates in a near-duplicate review pair.

**Paragraph**

Paragraph-level diagnostics include character count and sentence count. Empty
paragraphs should not exist because they were excluded during segmentation.

**Sentence**

Sentence-level diagnostics include character count and temporary descriptive
word count.

These temporary word counts are used only for structural quality assessment.
They are not backbone tokens and do not create a model-specific tokenised
dataset.

### Length Diagnostics

Unusually short or long structural units are flagged using transparent
descriptive thresholds.

These thresholds are not linguistic truths and do not imply that the
corresponding text is invalid. They identify observations that may require
attention during later annotation or modelling.

For the pilot:

- very short sentences contain fewer than 3 whitespace-delimited words;
- long sentences contain more than 60 whitespace-delimited words;
- very short paragraphs contain fewer than 20 words;
- very long paragraphs contain more than 500 words.

The thresholds may be revised for larger corpora after observing empirical
length distributions.

### Warning Consolidation

Warnings generated earlier in Notebook 02 remain visible.

The quality diagnostics therefore do not replace sanitisation or segmentation
audits. Instead, they consolidate relevant signals into an article-level quality
record.

Examples include:

- sanitisation warnings;
- segmentation warnings;
- near-duplicate review recommendations;
- missing titles or source metadata;
- unusually short or long structural units.

### Quality Status

Each article receives a descriptive quality status:

- `clean` — no quality flags detected;
- `review` — one or more non-critical conditions deserve inspection.

A `review` status does not mean that the article should be excluded.

No automatic semantic, political, emotional, factual or publisher-based quality
judgement is introduced.

### Corpus-Level Diagnostics

The block also summarises:

- number of articles;
- number of paragraphs;
- number of sentences;
- article word counts;
- paragraph word counts;
- sentence word counts;
- minimum, median, mean and maximum structural lengths;
- counts of unusually short and long units;
- warning frequencies;
- number of records requiring review.

These statistics describe the prepared pilot corpus and provide a baseline for
future larger acquisition batches.

### Output

The block creates:

- `QUALITY_RECORDS` — article-level structural quality summaries;
- `CORPUS_QUALITY_SUMMARY` — corpus-level descriptive diagnostics.

No records are removed or modified.

**Input:** `SEGMENTED_RECORDS`, `PARAGRAPH_RECORDS`, `SENTENCE_RECORDS`,
`SANITISATION_AUDIT`, and `NEAR_DUPLICATE_REVIEW`.

**Output:** Structural corpus-quality diagnostics suitable for persistence and
handover.

No model-specific tokenisation, semantic filtering, emotional analysis or GNews
API request is performed by this block.

```python
# =============================================================================
# Media AI — Dataset Quality Diagnostics
# =============================================================================

from collections import Counter
import statistics


# -----------------------------------------------------------------------------
# Quality-diagnostic configuration
# -----------------------------------------------------------------------------

QUALITY_POLICY_VERSION = "1.0"

VERY_SHORT_SENTENCE_WORDS = 3
LONG_SENTENCE_WORDS = 60

VERY_SHORT_PARAGRAPH_WORDS = 20
LONG_PARAGRAPH_WORDS = 500


# -----------------------------------------------------------------------------
# Descriptive-statistics helper
# -----------------------------------------------------------------------------

def describe_numeric_values(values):
    """
    Return compact descriptive statistics for a numeric collection.
    """

    values = list(values)

    if not values:
        return {
            "count": 0,
            "minimum": None,
            "median": None,
            "mean": None,
            "maximum": None,
        }

    return {
        "count":
            len(values),

        "minimum":
            min(values),

        "median":
            statistics.median(values),

        "mean":
            statistics.fmean(values),

        "maximum":
            max(values),
    }


# -----------------------------------------------------------------------------
# Index earlier Notebook 02 audits
# -----------------------------------------------------------------------------

sanitisation_audit_by_id = {
    record["media_ai_record_id"]:
        record
    for record in SANITISATION_AUDIT
}


# -----------------------------------------------------------------------------
# Near-duplicate participation
# -----------------------------------------------------------------------------

near_duplicate_review_ids = set()

for pair in NEAR_DUPLICATE_CANDIDATES:

    near_duplicate_review_ids.add(
        pair["record_id_1"]
    )

    near_duplicate_review_ids.add(
        pair["record_id_2"]
    )


# -----------------------------------------------------------------------------
# Paragraph-level descriptive word counts
# -----------------------------------------------------------------------------

paragraph_word_counts = []

paragraph_diagnostics_by_article = {}

for paragraph in PARAGRAPH_RECORDS:

    paragraph_text = paragraph.get(
        "text",
        ""
    )

    paragraph_word_count = count_words_for_audit(
        paragraph_text
    )

    paragraph_word_counts.append(
        paragraph_word_count
    )

    media_ai_record_id = paragraph.get(
        "media_ai_record_id"
    )

    paragraph_diagnostics_by_article.setdefault(
        media_ai_record_id,
        []
    ).append({
        "paragraph_id":
            paragraph.get(
                "paragraph_id"
            ),

        "paragraph_index":
            paragraph.get(
                "paragraph_index"
            ),

        "word_count":
            paragraph_word_count,

        "sentence_count":
            paragraph.get(
                "sentence_count",
                0
            ),

        "very_short":
            paragraph_word_count
            < VERY_SHORT_PARAGRAPH_WORDS,

        "very_long":
            paragraph_word_count
            > LONG_PARAGRAPH_WORDS,
    })


# -----------------------------------------------------------------------------
# Sentence-level diagnostics
# -----------------------------------------------------------------------------

sentence_word_counts = []

sentence_diagnostics_by_article = {}

for sentence in SENTENCE_RECORDS:

    sentence_word_count = sentence.get(
        "temporary_word_count",
        0
    )

    sentence_word_counts.append(
        sentence_word_count
    )

    media_ai_record_id = sentence.get(
        "media_ai_record_id"
    )

    sentence_diagnostics_by_article.setdefault(
        media_ai_record_id,
        []
    ).append({
        "sentence_id":
            sentence.get(
                "sentence_id"
            ),

        "paragraph_index":
            sentence.get(
                "paragraph_index"
            ),

        "sentence_index":
            sentence.get(
                "sentence_index"
            ),

        "word_count":
            sentence_word_count,

        "character_count":
            sentence.get(
                "character_count",
                0
            ),

        "very_short":
            sentence_word_count
            < VERY_SHORT_SENTENCE_WORDS,

        "very_long":
            sentence_word_count
            > LONG_SENTENCE_WORDS,
    })


# -----------------------------------------------------------------------------
# Article-level quality diagnostics
# -----------------------------------------------------------------------------

QUALITY_RECORDS = []

article_word_counts = []

for record in SEGMENTED_RECORDS:

    media_ai_record_id = record.get(
        "media_ai_record_id"
    )

    sanitised_text = record.get(
        "sanitised_text",
        ""
    )

    article_word_count = count_words_for_audit(
        sanitised_text
    )

    article_word_counts.append(
        article_word_count
    )

    segmentation_audit = record.get(
        "segmentation_audit",
        {}
    )

    sanitisation_record = sanitisation_audit_by_id.get(
        media_ai_record_id,
        {}
    )

    article_paragraphs = (
        paragraph_diagnostics_by_article.get(
            media_ai_record_id,
            []
        )
    )

    article_sentences = (
        sentence_diagnostics_by_article.get(
            media_ai_record_id,
            []
        )
    )

    short_paragraph_count = sum(
        1
        for paragraph in article_paragraphs
        if paragraph["very_short"]
    )

    long_paragraph_count = sum(
        1
        for paragraph in article_paragraphs
        if paragraph["very_long"]
    )

    short_sentence_count = sum(
        1
        for sentence in article_sentences
        if sentence["very_short"]
    )

    long_sentence_count = sum(
        1
        for sentence in article_sentences
        if sentence["very_long"]
    )

    quality_flags = []

    # -------------------------------------------------------------------------
    # Metadata completeness
    # -------------------------------------------------------------------------

    if not record.get("title"):
        quality_flags.append(
            "missing_title"
        )

    if not record.get("source_url"):
        quality_flags.append(
            "missing_source_url"
        )

    if not record.get("language"):
        quality_flags.append(
            "missing_language"
        )

    # -------------------------------------------------------------------------
    # Earlier audit warnings
    # -------------------------------------------------------------------------

    sanitisation_warnings = (
        sanitisation_record.get(
            "warnings",
            []
        )
    )

    segmentation_warnings = (
        segmentation_audit.get(
            "warnings",
            []
        )
    )

    if sanitisation_warnings:
        quality_flags.append(
            "sanitisation_warning"
        )

    if segmentation_warnings:
        quality_flags.append(
            "segmentation_warning"
        )

    # -------------------------------------------------------------------------
    # Near-duplicate review
    # -------------------------------------------------------------------------

    near_duplicate_review = (
        media_ai_record_id
        in near_duplicate_review_ids
    )

    if near_duplicate_review:
        quality_flags.append(
            "near_duplicate_review"
        )

    # -------------------------------------------------------------------------
    # Structural-length diagnostics
    # -------------------------------------------------------------------------

    if short_paragraph_count > 0:
        quality_flags.append(
            "very_short_paragraph"
        )

    if long_paragraph_count > 0:
        quality_flags.append(
            "very_long_paragraph"
        )

    if short_sentence_count > 0:
        quality_flags.append(
            "very_short_sentence"
        )

    if long_sentence_count > 0:
        quality_flags.append(
            "very_long_sentence"
        )

    quality_status = (
        "clean"
        if not quality_flags
        else "review"
    )

    QUALITY_RECORDS.append({
        "media_ai_record_id":
            media_ai_record_id,

        "title":
            record.get(
                "title"
            ),

        "source_schema_version":
            record.get(
                "source_schema_version"
            ),

        "quality_policy_version":
            QUALITY_POLICY_VERSION,

        "article_word_count":
            article_word_count,

        "paragraph_count":
            segmentation_audit.get(
                "paragraph_count",
                0
            ),

        "sentence_count":
            segmentation_audit.get(
                "sentence_count",
                0
            ),

        "very_short_paragraph_count":
            short_paragraph_count,

        "very_long_paragraph_count":
            long_paragraph_count,

        "very_short_sentence_count":
            short_sentence_count,

        "very_long_sentence_count":
            long_sentence_count,

        "sanitisation_warnings":
            sanitisation_warnings,

        "segmentation_warnings":
            segmentation_warnings,

        "near_duplicate_review":
            near_duplicate_review,

        "quality_flags":
            quality_flags,

        "quality_status":
            quality_status,
    })


# -----------------------------------------------------------------------------
# Corpus-level warning frequencies
# -----------------------------------------------------------------------------

quality_flag_counter = Counter()

for record in QUALITY_RECORDS:

    quality_flag_counter.update(
        record["quality_flags"]
    )


# -----------------------------------------------------------------------------
# Corpus-level diagnostics
# -----------------------------------------------------------------------------

records_requiring_review = sum(
    1
    for record in QUALITY_RECORDS
    if record[
        "quality_status"
    ] == "review"
)

very_short_sentence_count = sum(
    1
    for sentence in SENTENCE_RECORDS
    if sentence.get(
        "temporary_word_count",
        0
    ) < VERY_SHORT_SENTENCE_WORDS
)

long_sentence_count = sum(
    1
    for sentence in SENTENCE_RECORDS
    if sentence.get(
        "temporary_word_count",
        0
    ) > LONG_SENTENCE_WORDS
)

very_short_paragraph_count = sum(
    1
    for value in paragraph_word_counts
    if value < VERY_SHORT_PARAGRAPH_WORDS
)

long_paragraph_count = sum(
    1
    for value in paragraph_word_counts
    if value > LONG_PARAGRAPH_WORDS
)


CORPUS_QUALITY_SUMMARY = {
    "quality_policy_version":
        QUALITY_POLICY_VERSION,

    "article_count":
        len(
            SEGMENTED_RECORDS
        ),

    "paragraph_count":
        len(
            PARAGRAPH_RECORDS
        ),

    "sentence_count":
        len(
            SENTENCE_RECORDS
        ),

    "article_word_counts":
        describe_numeric_values(
            article_word_counts
        ),

    "paragraph_word_counts":
        describe_numeric_values(
            paragraph_word_counts
        ),

    "sentence_word_counts":
        describe_numeric_values(
            sentence_word_counts
        ),

    "very_short_paragraphs":
        very_short_paragraph_count,

    "very_long_paragraphs":
        long_paragraph_count,

    "very_short_sentences":
        very_short_sentence_count,

    "very_long_sentences":
        long_sentence_count,

    "records_requiring_review":
        records_requiring_review,

    "quality_flag_frequencies":
        dict(
            quality_flag_counter
        ),

    "near_duplicate_pairs_requiring_review":
        len(
            NEAR_DUPLICATE_CANDIDATES
        ),

    "sanitisation_records_with_warnings":
        sum(
            1
            for record in SANITISATION_AUDIT
            if record[
                "warnings"
            ]
        ),

    "segmentation_records_with_warnings":
        sum(
            1
            for record in SEGMENTED_RECORDS
            if record[
                "segmentation_audit"
            ][
                "warnings"
            ]
        ),
}


# -----------------------------------------------------------------------------
# Article-level diagnostic output
# -----------------------------------------------------------------------------

for index, record in enumerate(
    QUALITY_RECORDS,
    start=1,
):

    print("-" * 72)

    print(
        f"Article {index}"
    )

    print(
        f"Media AI ID               : "
        f"{record['media_ai_record_id']}"
    )

    print(
        f"Title                     : "
        f"{record['title']}"
    )

    print(
        f"Article words             : "
        f"{record['article_word_count']}"
    )

    print(
        f"Paragraphs                : "
        f"{record['paragraph_count']}"
    )

    print(
        f"Sentences                 : "
        f"{record['sentence_count']}"
    )

    print(
        f"Quality status            : "
        f"{record['quality_status']}"
    )

    print(
        f"Quality flags             : "
        f"{record['quality_flags']}"
    )


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Dataset Quality Diagnostics")
print("=" * 72)

print(
    f"Articles                   : "
    f"{CORPUS_QUALITY_SUMMARY['article_count']}"
)

print(
    f"Paragraphs                 : "
    f"{CORPUS_QUALITY_SUMMARY['paragraph_count']}"
)

print(
    f"Sentences                  : "
    f"{CORPUS_QUALITY_SUMMARY['sentence_count']}"
)

print(
    f"Records requiring review   : "
    f"{CORPUS_QUALITY_SUMMARY['records_requiring_review']}"
)

print(
    f"Very short paragraphs      : "
    f"{CORPUS_QUALITY_SUMMARY['very_short_paragraphs']}"
)

print(
    f"Very long paragraphs       : "
    f"{CORPUS_QUALITY_SUMMARY['very_long_paragraphs']}"
)

print(
    f"Very short sentences       : "
    f"{CORPUS_QUALITY_SUMMARY['very_short_sentences']}"
)

print(
    f"Very long sentences        : "
    f"{CORPUS_QUALITY_SUMMARY['very_long_sentences']}"
)

print(
    f"Near-duplicate review pairs: "
    f"{CORPUS_QUALITY_SUMMARY['near_duplicate_pairs_requiring_review']}"
)

print("-" * 72)

print(
    "Article word-count summary : "
    f"{CORPUS_QUALITY_SUMMARY['article_word_counts']}"
)

print(
    "Paragraph word summary     : "
    f"{CORPUS_QUALITY_SUMMARY['paragraph_word_counts']}"
)

print(
    "Sentence word summary      : "
    f"{CORPUS_QUALITY_SUMMARY['sentence_word_counts']}"
)

print("=" * 72)

print(
    "Dataset quality diagnostics completed successfully."
)

print(
    "Quality flags are diagnostic and have not removed or modified records."
)

print(
    "No semantic, emotional, political, factual, or publisher-based "
    "quality judgement has been applied."
)

print(
    "No model-specific tokenisation has been created."
)

print(
    "No GNews API requests have been made by this block."
)
```
