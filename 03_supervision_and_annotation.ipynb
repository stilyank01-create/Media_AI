# Media AI — Supervision and Annotation

**Notebook:** `03_supervision_and_annotation.ipynb`  
**Stage:** Supervision and annotation  
**Pipeline position:** Notebook 03 of 05

This notebook defines and constructs the supervision framework used by the
Media AI modelling architecture.

It receives the structurally prepared corpus produced by Notebook 02 and
introduces the annotation concepts required for later model training and
evaluation.

The notebook focuses on:

- defining the annotation schema;
- operationalising the emotional spectrum;
- defining emotional and factological supervision signals;
- distinguishing reported emotion from transformative emotion;
- constructing independent LLM supervision;
- supporting human review and adjudication;
- building a traceable reference dataset for model development.

The complete prepared text remains available throughout the supervision process.
Annotation does not replace, truncate, or overwrite the prepared corpus.

Model architecture, recurrent state computation and model training are outside
the scope of this notebook and are introduced in Notebook 04.

All external, computationally significant, or state-changing operations are
controlled centrally through the **Execution Control Panel** defined in Block 1.

## Environment, Reproducibility and Supervision Configuration

This block initialises the Notebook 03 research environment and defines the
central configuration used throughout the supervision and annotation stage.

The notebook uses a fixed random seed to support reproducible sampling,
annotation ordering and later comparison of supervision experiments.

### Execution Control Panel

All data-loading, external supervision, dataset-construction and persistence
operations are controlled centrally from this block.

Later blocks must read these switches and must not redefine them locally.

Safe configuration and validation operations may run normally. Operations that
load persistent research data, invoke external models, construct new reference
datasets or write research artefacts are explicitly controlled.

External or state-changing operations are disabled by default.

### Supervision Principles

Notebook 03 separates supervision from the original and prepared research
records.

Annotations are derived research artefacts and must not overwrite source or
prepared article content.

The supervision framework will preserve:

- Media AI record identifiers;
- article, paragraph and sentence traceability;
- complete prepared text;
- annotation provenance;
- supervision method and version;
- reviewer or model role where applicable.

Reported emotion and transformative emotion remain distinct annotation
constructs.

Emotional and factological supervision signals will be operationalised
explicitly before any reference dataset is constructed.

No model training, recurrent-state computation or confluent representation is
performed in this notebook.

**Input:** Persisted Notebook 02 prepared corpus and handover manifest.

**Output:** A traceable supervision and annotation dataset suitable for
Notebook 04.

No research data are loaded and no external model or API is called by this
initialisation block.

```python
# =============================================================================
# Media AI — Notebook 03 Environment, Reproducibility and Configuration
# =============================================================================

import sys
import platform
import random
from datetime import datetime, timezone

import numpy as np
import pandas as pd


# -----------------------------------------------------------------------------
# Notebook configuration
# -----------------------------------------------------------------------------

PROJECT_NAME = "Media AI"

NOTEBOOK_NAME = "03_supervision_and_annotation"
NOTEBOOK_VERSION = "1.0"

SUPERVISION_SCHEMA_VERSION = "1.0"
ANNOTATION_POLICY_VERSION = "1.0"

RANDOM_SEED = 42

random.seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)


# -----------------------------------------------------------------------------
# Execution Control Panel
# -----------------------------------------------------------------------------

# All operational switches are defined centrally here.
# Later blocks must consume these values and must not redefine them.


# -----------------------------------------------------------------------------
# Persistent research-data loading
# -----------------------------------------------------------------------------

LOAD_PREPARED_CORPUS = True


# -----------------------------------------------------------------------------
# External LLM supervision
# -----------------------------------------------------------------------------

# Master switch for all external LLM supervision.
#
# If False, no external supervisor may make a request even when an individual
# supervisor is enabled.

RUN_LLM_SUPERVISION = True


# -----------------------------------------------------------------------------
# Individual supervisor switches
# -----------------------------------------------------------------------------

# These switches are subordinate to RUN_LLM_SUPERVISION.

ENABLE_SUPERVISOR_01 = True
ENABLE_SUPERVISOR_02 = False
ENABLE_SUPERVISOR_03 = False
ENABLE_SUPERVISOR_04 = False
ENABLE_SUPERVISOR_05 = False


# -----------------------------------------------------------------------------
# Supervisor configuration
# -----------------------------------------------------------------------------

# Supervisor identities are stable research identifiers.
# Provider and model selection remain configurable independently.

# Supervisor 01 is assigned to the Google Gemini API.
# The exact Gemini model has not yet been selected.

SUPERVISOR_01_PROVIDER = "google"
SUPERVISOR_01_MODEL = "gemini-3.6-flash"

# Additional supervisor slots remain provider- and model-independent until
# comparative experiments are introduced.

SUPERVISOR_02_PROVIDER = None
SUPERVISOR_02_MODEL = None

SUPERVISOR_03_PROVIDER = None
SUPERVISOR_03_MODEL = None

SUPERVISOR_04_PROVIDER = None
SUPERVISOR_04_MODEL = None

SUPERVISOR_05_PROVIDER = None
SUPERVISOR_05_MODEL = None


# -----------------------------------------------------------------------------
# Free-tier and cost safeguards
# -----------------------------------------------------------------------------

# Supervisor 01 is intended to use a Gemini API project configured on the
# Free Tier with billing disabled.
#
# These notebook-side safeguards provide an additional research-control layer
# and must not be interpreted as a substitute for provider-side billing
# configuration.

REQUIRE_FREE_SUPERVISION = True

# Paid API execution is prohibited.

ALLOW_PAID_API_USAGE = False

# Automatic fallback from a free model or quota to any paid service is
# prohibited.

ALLOW_AUTOMATIC_PAID_FALLBACK = False


# -----------------------------------------------------------------------------
# Supervision-volume safeguards
# -----------------------------------------------------------------------------

# Hard maximum number of external supervision requests allowed during one
# notebook execution.
#
# Keep at 1 for the first live supervision test.

MAX_SUPERVISION_REQUESTS_PER_RUN = 1

# Independent research-side daily ceiling.
#
# This does not replace provider-side Free Tier limits. Later blocks must
# enforce the lower of the applicable research and provider limits.

MAX_SUPERVISION_REQUESTS_PER_DAY = 5


# -----------------------------------------------------------------------------
# Human-review artefact generation
# -----------------------------------------------------------------------------

RUN_HUMAN_REVIEW_EXPORT = False


# -----------------------------------------------------------------------------
# Construction of the supervised reference dataset
# -----------------------------------------------------------------------------

BUILD_REFERENCE_DATASET = False


# -----------------------------------------------------------------------------
# Persistence of Notebook 03 annotation and handover artefacts
# -----------------------------------------------------------------------------

SAVE_ANNOTATION_OUTPUTS = True


# -----------------------------------------------------------------------------
# Supervision policy
# -----------------------------------------------------------------------------

PRESERVE_PREPARED_TEXT = True
PRESERVE_STRUCTURAL_TRACEABILITY = True

OVERWRITE_PREPARED_RECORDS = False

REPORTED_AND_TRANSFORMATIVE_EMOTION_SEPARATE = True

ALLOW_SEMANTIC_PREFILTERING = False
ALLOW_EMOTIONAL_TOKEN_PREFILTERING = False

PERMANENT_MODEL_TOKENISATION = False

MODEL_TRAINING_ALLOWED = False
RECURRENT_STATE_CALCULATION_ALLOWED = False
CONFLUENT_REPRESENTATION_ALLOWED = False


# -----------------------------------------------------------------------------
# Annotation provenance policy
# -----------------------------------------------------------------------------

REQUIRE_ANNOTATION_PROVENANCE = True
REQUIRE_SUPERVISION_METHOD = True
REQUIRE_SCHEMA_VERSION = True
REQUIRE_RECORD_TRACEABILITY = True


# -----------------------------------------------------------------------------
# Runtime metadata
# -----------------------------------------------------------------------------

RUNTIME_METADATA = {
    "project":
        PROJECT_NAME,

    "notebook":
        NOTEBOOK_NAME,

    "notebook_version":
        NOTEBOOK_VERSION,

    "supervision_schema":
        SUPERVISION_SCHEMA_VERSION,

    "annotation_policy":
        ANNOTATION_POLICY_VERSION,

    "python_version":
        sys.version.split()[0],

    "platform":
        platform.platform(),

    "numpy_version":
        np.__version__,

    "pandas_version":
        pd.__version__,

    "random_seed":
        RANDOM_SEED,

    "initialised_at_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),
}


# -----------------------------------------------------------------------------
# Configuration safeguards
# -----------------------------------------------------------------------------

if OVERWRITE_PREPARED_RECORDS:
    raise ValueError(
        "Notebook 03 must not overwrite prepared Notebook 02 records."
    )


if not PRESERVE_PREPARED_TEXT:
    raise ValueError(
        "Notebook 03 must preserve complete prepared text."
    )


if not PRESERVE_STRUCTURAL_TRACEABILITY:
    raise ValueError(
        "Notebook 03 must preserve structural traceability."
    )


if not REPORTED_AND_TRANSFORMATIVE_EMOTION_SEPARATE:
    raise ValueError(
        "Reported and transformative emotion must remain "
        "separate supervision constructs."
    )


if ALLOW_EMOTIONAL_TOKEN_PREFILTERING:
    raise ValueError(
        "Emotional token pre-filtering is not permitted "
        "during supervision."
    )


if PERMANENT_MODEL_TOKENISATION:
    raise ValueError(
        "Permanent model-specific tokenisation belongs "
        "to the modelling stage."
    )


if MODEL_TRAINING_ALLOWED:
    raise ValueError(
        "Model training is outside the scope of Notebook 03."
    )


if RECURRENT_STATE_CALCULATION_ALLOWED:
    raise ValueError(
        "Recurrent-state calculation is outside the scope "
        "of Notebook 03."
    )


if CONFLUENT_REPRESENTATION_ALLOWED:
    raise ValueError(
        "Confluent representation is outside the scope "
        "of Notebook 03."
    )


# -----------------------------------------------------------------------------
# Cost-control safeguards
# -----------------------------------------------------------------------------

if REQUIRE_FREE_SUPERVISION and ALLOW_PAID_API_USAGE:
    raise ValueError(
        "Paid API usage cannot be enabled while "
        "REQUIRE_FREE_SUPERVISION is True."
    )


if REQUIRE_FREE_SUPERVISION and ALLOW_AUTOMATIC_PAID_FALLBACK:
    raise ValueError(
        "Automatic paid fallback cannot be enabled while "
        "REQUIRE_FREE_SUPERVISION is True."
    )


if ALLOW_AUTOMATIC_PAID_FALLBACK and not ALLOW_PAID_API_USAGE:
    raise ValueError(
        "Automatic paid fallback cannot be enabled when "
        "paid API usage is prohibited."
    )


if (
    not isinstance(
        MAX_SUPERVISION_REQUESTS_PER_RUN,
        int,
    )
    or MAX_SUPERVISION_REQUESTS_PER_RUN < 1
):
    raise ValueError(
        "MAX_SUPERVISION_REQUESTS_PER_RUN must be a positive integer."
    )


if (
    not isinstance(
        MAX_SUPERVISION_REQUESTS_PER_DAY,
        int,
    )
    or MAX_SUPERVISION_REQUESTS_PER_DAY < 1
):
    raise ValueError(
        "MAX_SUPERVISION_REQUESTS_PER_DAY must be a positive integer."
    )


if (
    MAX_SUPERVISION_REQUESTS_PER_RUN
    > MAX_SUPERVISION_REQUESTS_PER_DAY
):
    raise ValueError(
        "The per-run supervision limit cannot exceed "
        "the daily supervision limit."
    )


# -----------------------------------------------------------------------------
# Supervisor-configuration safeguards
# -----------------------------------------------------------------------------

SUPERVISOR_CONTROL_CONFIGURATION = {
    "supervisor_01": {
        "enabled":
            ENABLE_SUPERVISOR_01,

        "provider":
            SUPERVISOR_01_PROVIDER,

        "model":
            SUPERVISOR_01_MODEL,
    },

    "supervisor_02": {
        "enabled":
            ENABLE_SUPERVISOR_02,

        "provider":
            SUPERVISOR_02_PROVIDER,

        "model":
            SUPERVISOR_02_MODEL,
    },

    "supervisor_03": {
        "enabled":
            ENABLE_SUPERVISOR_03,

        "provider":
            SUPERVISOR_03_PROVIDER,

        "model":
            SUPERVISOR_03_MODEL,
    },

    "supervisor_04": {
        "enabled":
            ENABLE_SUPERVISOR_04,

        "provider":
            SUPERVISOR_04_PROVIDER,

        "model":
            SUPERVISOR_04_MODEL,
    },

    "supervisor_05": {
        "enabled":
            ENABLE_SUPERVISOR_05,

        "provider":
            SUPERVISOR_05_PROVIDER,

        "model":
            SUPERVISOR_05_MODEL,
    },
}


# Enabling a supervisor without selecting a model is allowed while the
# master supervision switch remains False.
#
# Once live supervision is enabled, every enabled supervisor must have both
# provider and model configuration.

if RUN_LLM_SUPERVISION:

    for (
        supervisor_id,
        configuration,
    ) in SUPERVISOR_CONTROL_CONFIGURATION.items():

        if not configuration[
            "enabled"
        ]:
            continue

        if not configuration[
            "provider"
        ]:
            raise ValueError(
                f"{supervisor_id} is enabled for live supervision "
                "but has no provider configured."
            )

        if not configuration[
            "model"
        ]:
            raise ValueError(
                f"{supervisor_id} is enabled for live supervision "
                "but has no model configured."
            )


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Notebook 03 Initialisation")
print("=" * 72)

print(
    f"Notebook                   : "
    f"{NOTEBOOK_NAME}"
)

print(
    f"Notebook version           : "
    f"{NOTEBOOK_VERSION}"
)

print(
    f"Supervision schema         : "
    f"{SUPERVISION_SCHEMA_VERSION}"
)

print(
    f"Annotation policy          : "
    f"{ANNOTATION_POLICY_VERSION}"
)

print(
    f"Random seed                : "
    f"{RANDOM_SEED}"
)

print("-" * 72)

print("Execution Control Panel")

print(
    f"Load prepared corpus       : "
    f"{LOAD_PREPARED_CORPUS}"
)

print(
    f"Run LLM supervision        : "
    f"{RUN_LLM_SUPERVISION}"
)

print(
    f"Supervisor 01 enabled      : "
    f"{ENABLE_SUPERVISOR_01}"
)

print(
    f"Supervisor 02 enabled      : "
    f"{ENABLE_SUPERVISOR_02}"
)

print(
    f"Supervisor 03 enabled      : "
    f"{ENABLE_SUPERVISOR_03}"
)

print(
    f"Supervisor 04 enabled      : "
    f"{ENABLE_SUPERVISOR_04}"
)

print(
    f"Supervisor 05 enabled      : "
    f"{ENABLE_SUPERVISOR_05}"
)

print(
    f"Human review export        : "
    f"{RUN_HUMAN_REVIEW_EXPORT}"
)

print(
    f"Build reference dataset    : "
    f"{BUILD_REFERENCE_DATASET}"
)

print(
    f"Save annotation outputs    : "
    f"{SAVE_ANNOTATION_OUTPUTS}"
)

print("-" * 72)

print("Supervisor configuration")

print(
    f"Supervisor 01 provider     : "
    f"{SUPERVISOR_01_PROVIDER}"
)

print(
    f"Supervisor 01 model set    : "
    f"{SUPERVISOR_01_MODEL is not None}"
)

print(
    f"Supervisor 02 configured   : "
    f"{SUPERVISOR_02_PROVIDER is not None and SUPERVISOR_02_MODEL is not None}"
)

print(
    f"Supervisor 03 configured   : "
    f"{SUPERVISOR_03_PROVIDER is not None and SUPERVISOR_03_MODEL is not None}"
)

print(
    f"Supervisor 04 configured   : "
    f"{SUPERVISOR_04_PROVIDER is not None and SUPERVISOR_04_MODEL is not None}"
)

print(
    f"Supervisor 05 configured   : "
    f"{SUPERVISOR_05_PROVIDER is not None and SUPERVISOR_05_MODEL is not None}"
)

print("-" * 72)

print("Cost and usage safeguards")

print(
    f"Require free supervision   : "
    f"{REQUIRE_FREE_SUPERVISION}"
)

print(
    f"Paid API usage allowed     : "
    f"{ALLOW_PAID_API_USAGE}"
)

print(
    f"Paid fallback allowed      : "
    f"{ALLOW_AUTOMATIC_PAID_FALLBACK}"
)

print(
    f"Max requests / run         : "
    f"{MAX_SUPERVISION_REQUESTS_PER_RUN}"
)

print(
    f"Max requests / day         : "
    f"{MAX_SUPERVISION_REQUESTS_PER_DAY}"
)

print("-" * 72)

print("Supervision boundary")

print(
    f"Preserve prepared text     : "
    f"{PRESERVE_PREPARED_TEXT}"
)

print(
    f"Preserve traceability      : "
    f"{PRESERVE_STRUCTURAL_TRACEABILITY}"
)

print(
    f"Semantic pre-filtering     : "
    f"{ALLOW_SEMANTIC_PREFILTERING}"
)

print(
    f"Emotional pre-filtering    : "
    f"{ALLOW_EMOTIONAL_TOKEN_PREFILTERING}"
)

print(
    f"Permanent tokenisation     : "
    f"{PERMANENT_MODEL_TOKENISATION}"
)

print(
    f"Model training             : "
    f"{MODEL_TRAINING_ALLOWED}"
)

print(
    f"Recurrent-state calculation: "
    f"{RECURRENT_STATE_CALCULATION_ALLOWED}"
)

print(
    f"Confluent representation   : "
    f"{CONFLUENT_REPRESENTATION_ALLOWED}"
)

print("=" * 72)

print(
    "Notebook 03 supervision environment initialised successfully."
)

print(
    "Execution controls are centralised in Block 1."
)

print(
    "Supervisor model selection remains explicit and separate from execution."
)

print(
    "Free-tier and notebook-side supervision limits are active."
)

print(
    "No research data have been loaded by this block."
)

print(
    "No external model or API calls have been made by this block."
)

print(
    "No annotation or model-training operation has been performed."
)
```

## Notebook 02 Handover and Prepared Corpus Loading

This block establishes the persistent research handover from Notebook 02 to
Notebook 03.

Notebook 03 does not depend on temporary Python objects created during the
preparation notebook. Instead, it reconstructs its input from the persisted
Notebook 02 handover manifest and canonical prepared-corpus artefact.

The principal handover file is:

`notebook_02_handover_manifest.json`

The manifest identifies the canonical prepared corpus and records the structural
counts, preparation-policy versions, source-schema history, preservation
guarantees and intended next research stage.

### Execution Control

Prepared-data loading is controlled exclusively by the Block 1 switch:

`LOAD_PREPARED_CORPUS`

If this switch is `False`, this block performs no Google Drive authentication
and does not load any research data.

Later blocks must not redefine the switch.

### Restricted Persistent Storage

When loading is enabled, Google Drive is accessed through the Google Drive API
rather than through a mounted filesystem.

The block locates the dedicated `Media_AI` project storage and reads only the
persistent artefacts required for the Notebook 03 workflow.

Google account credentials are used only for authenticated access to the
research storage and are not written into the notebook or handover artefacts.

### Handover Verification

Before accepting the prepared corpus, the block verifies that:

- the Notebook 02 handover manifest exists;
- its manifest type is `media_ai_preparation_handover`;
- the declared next notebook is `03_supervision_and_annotation`;
- the prepared-corpus file referenced by the manifest remains accessible;
- the prepared corpus identifies itself as `media_ai_prepared_corpus`;
- persisted article, paragraph and sentence counts agree with the manifest;
- complete prepared text remains available;
- no permanent model-specific tokenisation has been introduced;
- no emotional or factological weights have already been assigned.

These checks preserve the methodological boundary between preparation and
supervision.

### Output

When loading is enabled, this block creates:

- `NOTEBOOK_02_HANDOVER`;
- `PREPARED_CORPUS`;
- `PREPARED_ARTICLE_RECORDS`;
- reconstructed article, paragraph and sentence counts;
- verified preparation and preservation metadata.

No supervision labels are created and no external language model is called.

**Input:** Persistent Notebook 02 handover and prepared corpus.

**Output:** Verified Notebook 03 research input.

No GNews API request is performed by this block.

```python
# =============================================================================
# Media AI — Notebook 02 Handover and Prepared Corpus Loading
# =============================================================================

import io
import json


# -----------------------------------------------------------------------------
# Default unloaded state
# -----------------------------------------------------------------------------

NOTEBOOK_02_HANDOVER = None
PREPARED_CORPUS = None
PREPARED_ARTICLE_RECORDS = []

PREPARED_ARTICLE_COUNT = 0
PREPARED_PARAGRAPH_COUNT = 0
PREPARED_SENTENCE_COUNT = 0

PREPARED_CORPUS_LOADED = False
PREPARED_CORPUS_VERIFIED = False


# -----------------------------------------------------------------------------
# Load only when enabled in the central Execution Control Panel
# -----------------------------------------------------------------------------

if LOAD_PREPARED_CORPUS:

    from google.colab import auth
    from googleapiclient.discovery import build
    from googleapiclient.http import MediaIoBaseDownload

    # -------------------------------------------------------------------------
    # Restricted Google Drive authentication
    # -------------------------------------------------------------------------

    auth.authenticate_user()

    drive_service = build(
        "drive",
        "v3",
        cache_discovery=False,
    )

    # -------------------------------------------------------------------------
    # Helper: locate folder
    # -------------------------------------------------------------------------

    def find_drive_folder(
        folder_name,
        parent_id=None,
    ):
        """
        Locate one non-trashed Google Drive folder.
        """

        safe_name = folder_name.replace(
            "'",
            "\\'",
        )

        query_parts = [
            f"name = '{safe_name}'",
            "mimeType = 'application/vnd.google-apps.folder'",
            "trashed = false",
        ]

        if parent_id:
            query_parts.append(
                f"'{parent_id}' in parents"
            )

        response = (
            drive_service.files()
            .list(
                q=" and ".join(query_parts),
                spaces="drive",
                fields="files(id,name)",
                pageSize=10,
            )
            .execute()
        )

        folders = response.get(
            "files",
            []
        )

        if not folders:
            return None

        return folders[0]["id"]

    # -------------------------------------------------------------------------
    # Helper: locate file
    # -------------------------------------------------------------------------

    def find_drive_file(
        file_name,
        parent_id,
    ):
        """
        Locate one non-trashed file inside a specific Google Drive folder.
        """

        safe_name = file_name.replace(
            "'",
            "\\'",
        )

        query = (
            f"name = '{safe_name}' and "
            f"'{parent_id}' in parents and "
            "trashed = false"
        )

        response = (
            drive_service.files()
            .list(
                q=query,
                spaces="drive",
                fields="files(id,name,mimeType)",
                pageSize=10,
            )
            .execute()
        )

        files = response.get(
            "files",
            []
        )

        if not files:
            return None

        return files[0]["id"]

    # -------------------------------------------------------------------------
    # Helper: download JSON
    # -------------------------------------------------------------------------

    def download_json_from_drive(
        file_id,
    ):
        """
        Download and decode one JSON research artefact from Google Drive.
        """

        request = (
            drive_service.files()
            .get_media(
                fileId=file_id
            )
        )

        buffer = io.BytesIO()

        downloader = MediaIoBaseDownload(
            buffer,
            request,
        )

        done = False

        while not done:
            _, done = downloader.next_chunk()

        buffer.seek(0)

        return json.loads(
            buffer.read().decode(
                "utf-8"
            )
        )

    # -------------------------------------------------------------------------
    # Resolve Media AI storage
    # -------------------------------------------------------------------------

    PROJECT_FOLDER_ID = find_drive_folder(
        "Media_AI"
    )

    if PROJECT_FOLDER_ID is None:
        raise FileNotFoundError(
            "Media_AI project folder was not found."
        )

    LOGS_FOLDER_ID = find_drive_folder(
        "logs",
        PROJECT_FOLDER_ID,
    )

    if LOGS_FOLDER_ID is None:
        raise FileNotFoundError(
            "Media_AI/logs folder was not found."
        )

    DATA_FOLDER_ID = find_drive_folder(
        "data",
        PROJECT_FOLDER_ID,
    )

    if DATA_FOLDER_ID is None:
        raise FileNotFoundError(
            "Media_AI/data folder was not found."
        )

    PREPARED_DATA_FOLDER_ID = find_drive_folder(
        "prepared",
        DATA_FOLDER_ID,
    )

    if PREPARED_DATA_FOLDER_ID is None:
        raise FileNotFoundError(
            "Media_AI/data/prepared folder was not found."
        )

    # -------------------------------------------------------------------------
    # Load Notebook 02 handover manifest
    # -------------------------------------------------------------------------

    NOTEBOOK_02_HANDOVER_FILENAME = (
        "notebook_02_handover_manifest.json"
    )

    notebook_02_handover_file_id = (
        find_drive_file(
            NOTEBOOK_02_HANDOVER_FILENAME,
            LOGS_FOLDER_ID,
        )
    )

    if notebook_02_handover_file_id is None:
        raise FileNotFoundError(
            "Notebook 02 handover manifest was not found."
        )

    NOTEBOOK_02_HANDOVER = (
        download_json_from_drive(
            notebook_02_handover_file_id
        )
    )

    # -------------------------------------------------------------------------
    # Validate handover identity
    # -------------------------------------------------------------------------

    if (
        NOTEBOOK_02_HANDOVER.get(
            "manifest_type"
        )
        != "media_ai_preparation_handover"
    ):
        raise ValueError(
            "Unexpected Notebook 02 handover manifest type."
        )

    next_stage = NOTEBOOK_02_HANDOVER.get(
        "next_stage",
        {}
    )

    if (
        next_stage.get(
            "notebook"
        )
        != NOTEBOOK_NAME
    ):
        raise ValueError(
            "Notebook 02 handover does not identify "
            "Notebook 03 as the next research stage."
        )

    # -------------------------------------------------------------------------
    # Resolve canonical prepared corpus
    # -------------------------------------------------------------------------

    prepared_manifest = (
        NOTEBOOK_02_HANDOVER.get(
            "prepared_corpus",
            {}
        )
    )

    prepared_corpus_file_id = (
        prepared_manifest.get(
            "file_id"
        )
    )

    prepared_corpus_filename = (
        prepared_manifest.get(
            "filename"
        )
    )

    if not prepared_corpus_file_id:
        raise ValueError(
            "Notebook 02 handover does not contain "
            "a prepared-corpus file ID."
        )

    if not prepared_corpus_filename:
        raise ValueError(
            "Notebook 02 handover does not contain "
            "a prepared-corpus filename."
        )

    # -------------------------------------------------------------------------
    # Verify file accessibility
    # -------------------------------------------------------------------------

    prepared_file_metadata = (
        drive_service.files()
        .get(
            fileId=prepared_corpus_file_id,
            fields="id,name,trashed",
        )
        .execute()
    )

    if prepared_file_metadata.get(
        "trashed"
    ):
        raise FileNotFoundError(
            "Prepared corpus referenced by Notebook 02 is in Trash."
        )

    if (
        prepared_file_metadata.get(
            "name"
        )
        != prepared_corpus_filename
    ):
        raise ValueError(
            "Prepared-corpus filename does not match "
            "the Notebook 02 handover."
        )

    # -------------------------------------------------------------------------
    # Load canonical prepared corpus
    # -------------------------------------------------------------------------

    PREPARED_CORPUS = (
        download_json_from_drive(
            prepared_corpus_file_id
        )
    )

    if (
        PREPARED_CORPUS.get(
            "artifact_type"
        )
        != "media_ai_prepared_corpus"
    ):
        raise ValueError(
            "Unexpected prepared-corpus artefact type."
        )

    PREPARED_ARTICLE_RECORDS = (
        PREPARED_CORPUS.get(
            "records",
            []
        )
    )

    # -------------------------------------------------------------------------
    # Reconstruct structural counts
    # -------------------------------------------------------------------------

    PREPARED_ARTICLE_COUNT = len(
        PREPARED_ARTICLE_RECORDS
    )

    PREPARED_PARAGRAPH_COUNT = sum(
        len(
            article.get(
                "paragraphs",
                []
            )
        )
        for article
        in PREPARED_ARTICLE_RECORDS
    )

    PREPARED_SENTENCE_COUNT = sum(
        len(
            paragraph.get(
                "sentences",
                []
            )
        )
        for article
        in PREPARED_ARTICLE_RECORDS
        for paragraph
        in article.get(
            "paragraphs",
            []
        )
    )

    # -------------------------------------------------------------------------
    # Manifest-count verification
    # -------------------------------------------------------------------------

    manifest_article_count = (
        prepared_manifest.get(
            "article_count"
        )
    )

    manifest_paragraph_count = (
        prepared_manifest.get(
            "paragraph_count"
        )
    )

    manifest_sentence_count = (
        prepared_manifest.get(
            "sentence_count"
        )
    )

    counts_match = all([
        PREPARED_ARTICLE_COUNT
        == manifest_article_count,

        PREPARED_PARAGRAPH_COUNT
        == manifest_paragraph_count,

        PREPARED_SENTENCE_COUNT
        == manifest_sentence_count,
    ])

    # -------------------------------------------------------------------------
    # Methodological-boundary verification
    # -------------------------------------------------------------------------

    preservation = PREPARED_CORPUS.get(
        "preservation",
        {}
    )

    preservation_valid = all([
        preservation.get(
            "complete_sanitised_text_preserved"
        ) is True,

        preservation.get(
            "source_records_overwritten"
        ) is False,

        preservation.get(
            "article_paragraph_sentence_traceability"
        ) is True,

        preservation.get(
            "permanent_model_tokenisation"
        ) is False,

        preservation.get(
            "emotional_token_prefiltering"
        ) is False,

        preservation.get(
            "emotional_weights_assigned"
        ) is False,

        preservation.get(
            "factological_weights_assigned"
        ) is False,

        preservation.get(
            "recurrent_states_calculated"
        ) is False,

        preservation.get(
            "confluent_representation_calculated"
        ) is False,
    ])

    complete_text_available = all(
        isinstance(
            article.get(
                "sanitised_text"
            ),
            str,
        )
        and bool(
            article.get(
                "sanitised_text"
            )
        )
        for article
        in PREPARED_ARTICLE_RECORDS
    )

    PREPARED_CORPUS_VERIFIED = all([
        counts_match,
        preservation_valid,
        complete_text_available,
    ])

    if not PREPARED_CORPUS_VERIFIED:
        raise ValueError(
            "Prepared corpus failed Notebook 03 "
            "handover verification."
        )

    PREPARED_CORPUS_LOADED = True


# -----------------------------------------------------------------------------
# Verification output
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Notebook 02 Handover")
print("=" * 72)

print(
    f"Load prepared corpus       : "
    f"{LOAD_PREPARED_CORPUS}"
)

print(
    f"Prepared corpus loaded     : "
    f"{PREPARED_CORPUS_LOADED}"
)

if PREPARED_CORPUS_LOADED:

    print(
        f"Handover manifest version  : "
        f"{NOTEBOOK_02_HANDOVER.get('manifest_version')}"
    )

    print(
        f"Prepared corpus schema     : "
        f"{PREPARED_CORPUS.get('schema_version')}"
    )

    print(
        f"Prepared articles          : "
        f"{PREPARED_ARTICLE_COUNT}"
    )

    print(
        f"Prepared paragraphs        : "
        f"{PREPARED_PARAGRAPH_COUNT}"
    )

    print(
        f"Prepared sentences         : "
        f"{PREPARED_SENTENCE_COUNT}"
    )

    print(
        f"Prepared corpus verified   : "
        f"{PREPARED_CORPUS_VERIFIED}"
    )

    print(
        f"Next research stage        : "
        f"{NOTEBOOK_02_HANDOVER.get('next_stage', {}).get('notebook')}"
    )

print("=" * 72)

if PREPARED_CORPUS_LOADED:

    print(
        "Notebook 02 prepared corpus loaded and verified successfully."
    )

    print(
        "Complete prepared text and structural traceability are preserved."
    )

else:

    print(
        "Prepared-corpus loading was skipped by the Execution Control Panel."
    )

print(
    "No supervision labels have been created by this block."
)

print(
    "No external language model or GNews API request has been made by this block."
)
```

## Annotation Schema and Operational Construct Definitions

This block defines the supervision targets used throughout Notebook 03.

The objective is to specify the annotation constructs before any external model,
human reviewer or reference-dataset procedure is allowed to assign labels.

Supervision should evaluate explicitly defined quantities rather than allowing
the supervising model to invent its own interpretation of concepts such as
emotion, transformation, framing or informational content.

The constructs defined here are operational research variables. They are not
treated as direct measurements of human psychological states or objective truth.

### Annotation Unit

Media AI preserves the complete article and its hierarchical structure:

**Article → Paragraph → Sentence**

The primary pilot annotation unit is the **sentence in its preceding article
context**.

A sentence must therefore not be interpreted as an isolated fragment when the
target construct depends on preceding information.

For sentence \(t\), supervision may use:

- the complete article identifier;
- paragraph and sentence position;
- the sentence text;
- preceding textual context;
- relevant person, institution, event or social-group references;
- annotations associated with earlier textual units where methodologically
  appropriate.

The original and prepared text remain unchanged.

---

### Emotional Representation

Media AI uses a multidimensional emotional representation rather than a single
positive-versus-negative sentiment score.

Several emotional processes may coexist.

The pilot emotional ontology contains the following dimensions:

- **fear** — perceived danger, threat, vulnerability or anticipated harm;
- **anger** — hostility, outrage, blame-oriented activation or confrontation;
- **sadness** — loss, suffering, grief, disappointment or emotional decline;
- **disgust** — aversion, revulsion or strong rejection;
- **tension** — unresolved pressure, suspense, instability or conflict;
- **uncertainty** — ambiguity, unpredictability or lack of confidence about what
  is known or what will happen;
- **hope** — expectation of a favourable development or improvement;
- **relief** — reduction of previously established danger, pressure or distress;
- **trust** — confidence in a person, institution, process or claim;
- **compassion** — concern, sympathy or care directed toward another person or
  group;
- **solidarity** — affiliation, common identification or collective support.

The ontology is deliberately multi-label. Fear and compassion, for example, may
increase simultaneously.

Changes to the ontology require a new supervision-schema version rather than a
silent redefinition of an existing dimension.

---

### Reported Emotion

**Reported emotion** represents emotional content attributed by the text to a
person, group or other described subject.

Example:

> Residents said they were terrified.

This may support a high reported-fear annotation because fear is explicitly part
of the reported subject matter.

Reported emotion does not establish that the sentence itself increases fear in
the surrounding textual state.

The annotation question is:

> **Which emotions are represented as belonging to the subjects described by
> this textual unit?**

For each emotional dimension:

\[
E^{reported}_{t,e}\in[0,1]
\]

where higher values represent stronger reported presence or intensity under the
operational definition.

---

### Transformative Emotion

**Transformative emotion** represents the signed change in an emotional
dimension produced by the current textual information relative to preceding
context.

The annotation question is:

> **How does this textual unit change the emotional condition established by the
> preceding context?**

For emotional dimension \(e\):

\[
\Delta E_{t,e}\in[-1,1]
\]

Interpretation:

- values approaching `+1` indicate a strong increase;
- values around `0` indicate little or no meaningful change;
- values approaching `-1` indicate a strong decrease.

The signed range is intentional.

A negative value does not represent a negative amount of emotion. It represents
a **reduction in the corresponding emotional dimension**.

For example, a sentence establishing that a previously reported threat was
unfounded may reduce fear and uncertainty while increasing relief.

Using a non-negative activation such as ReLU directly on the transformation
target would collapse two distinct cases:

- no meaningful change;
- meaningful reduction.

The transformation target therefore remains signed.

The exact mathematical mechanism by which a later model combines previous state
and current transformation is not defined in Notebook 03. Activation functions
such as ReLU, sigmoid, clipping or other bounded state-update mechanisms are
model-design choices to be evaluated in Notebook 04.

---

### Immediate Emotional State

The **immediate emotional state** represents the amount of each emotional
dimension present after the current textual unit has been interpreted in
context.

The annotation question is:

> **What emotional state is present after processing the current textual unit?**

For emotional dimension \(e\):

\[
E^{state}_{t,e}\in[0,1]
\]

The state is therefore non-negative, while the transformation that produced the
change may be positive or negative.

Notebook 03 deliberately does not prescribe an architectural equation connecting
previous state, transformation and resulting state.

That relationship is a modelling hypothesis for Notebook 04.

---

### Recurrent Carry-Forward State

The amount of emotional state transmitted into subsequent context is **not**
directly defined as a Notebook 03 annotation target.

That quantity belongs to the recurrent architecture.

Notebook 04 will investigate how much of a current emotional state persists,
attenuates, reinforces or reverses as subsequent text is processed.

This preserves the conceptual separation between:

- state;
- transformation;
- persistence.

---

### Emotional Weight

**Emotional weight** represents the strength of emotionally consequential
information associated with a textual unit.

It is not sentiment polarity and is not calculated from the number of emotional
words.

A sentence may carry substantial emotional weight through contextual information
without containing explicit emotional vocabulary.

The pilot supervision scale is:

\[
W^{emotion}_t\in[0,1]
\]

where:

- `0` indicates negligible emotional consequence under the operational
  definition;
- `1` indicates very strong emotional consequence.

Emotional weight and emotional direction remain separate.

A textual unit may therefore have high emotional weight while reducing fear and
increasing relief.

---

### Factological Weight

**Factological weight** represents the degree to which a textual unit contributes
externally assessable informational content.

It is intentionally distinct from factual truth.

Relevant informational elements may include:

- explicit claims;
- observations;
- quantities;
- dates and measurable values;
- attribution;
- evidence;
- reported actions or events;
- stated certainty or uncertainty;
- identifiable speculation or prediction.

The pilot scale is:

\[
W^{fact}_t\in[0,1]
\]

Higher values indicate a greater contribution of externally assessable
informational content.

A false claim may still receive high factological weight because it remains an
explicit proposition that can in principle be assessed externally.

Truth verification is a separate research problem.

---

### Factological Composition

Factological weight is accompanied by interpretable multi-label components:

- `explicit_claim`;
- `observation_or_event`;
- `quantitative_information`;
- `attribution`;
- `evidence_reference`;
- `certainty_statement`;
- `uncertainty_statement`;
- `speculation_or_prediction`.

These indicators are not mutually exclusive.

---

### Relationship Between Emotional and Factological Weight

Emotional and factological weight are not opposites and are not required to sum
to one.

A sentence may simultaneously contain:

- high factological and high emotional weight;
- high factological and low emotional weight;
- low factological and high emotional weight;
- low levels of both.

Accordingly:

\[
W^{emotion}_t + W^{fact}_t \neq 1
\]

in general.

---

### Annotation Confidence

Every supervised construct may be accompanied by an annotation-confidence value:

\[
C_t\in[0,1]
\]

Confidence represents the supervisor's certainty that the annotation is
supported by the operational definition and available context.

It is not the probability that the annotation is objectively true.

Low-confidence observations may later be directed to human review or
multi-supervisor adjudication.

---

### Insufficient Evidence

The schema distinguishes:

- a genuine zero or near-zero value;
- `insufficient_evidence`;
- `not_applicable`.

These conditions must not be collapsed into one numeric value.

---

### Annotation Provenance

Every annotation must preserve information about how it was produced.

Required metadata include:

- Media AI record identifier;
- paragraph and sentence identifiers where applicable;
- supervision-schema version;
- annotation-policy version;
- supervision method;
- supervisor role or model identifier;
- annotation timestamp;
- context scope;
- confidence;
- review status.

This enables later comparison between machine supervision, human review and
adjudicated reference labels.

---

### Separation from Model Architecture

Notebook 03 defines **what should be estimated**, not **how the model must
estimate it**.

The supervision schema is therefore independent of the architecture used in
Notebook 04.

Notebook 05 can subsequently compare whether a simple backbone projection,
transformative component, recurrent component or confluent architecture better
predicts the same predefined supervision targets.

### Output

This block creates a versioned machine-readable annotation schema containing:

- emotional dimensions;
- reported-emotion definitions;
- transformative-emotion definitions;
- immediate-state definitions;
- emotional-weight definition;
- factological-weight definition;
- factological component indicators;
- allowed scales and missing-information states;
- confidence rules;
- provenance requirements;
- methodological constraints.

No annotations are generated by this block.

No external model or API is called.

```python
# =============================================================================
# Media AI — Annotation Schema and Operational Construct Definitions
# =============================================================================

from copy import deepcopy


# -----------------------------------------------------------------------------
# Annotation scales
# -----------------------------------------------------------------------------

SIGNED_TRANSFORMATION_SCALE = {
    "minimum": -1.0,
    "maximum": 1.0,
    "neutral": 0.0,
}

NON_NEGATIVE_WEIGHT_SCALE = {
    "minimum": 0.0,
    "maximum": 1.0,
}

CONFIDENCE_SCALE = {
    "minimum": 0.0,
    "maximum": 1.0,
}


# -----------------------------------------------------------------------------
# Missing-information states
# -----------------------------------------------------------------------------

ANNOTATION_VALUE_STATES = {
    "observed",
    "insufficient_evidence",
    "not_applicable",
}


# -----------------------------------------------------------------------------
# Emotional ontology
# -----------------------------------------------------------------------------

EMOTIONAL_DIMENSIONS = {
    "fear": {
        "definition":
            "Perceived danger, threat, vulnerability, or anticipated harm.",
    },

    "anger": {
        "definition":
            "Hostility, outrage, blame-oriented activation, or confrontation.",
    },

    "sadness": {
        "definition":
            "Loss, suffering, grief, disappointment, or emotional decline.",
    },

    "disgust": {
        "definition":
            "Aversion, revulsion, or strong rejection.",
    },

    "tension": {
        "definition":
            "Unresolved pressure, suspense, instability, or conflict.",
    },

    "uncertainty": {
        "definition":
            "Ambiguity, unpredictability, or lack of confidence about what "
            "is known or what will happen.",
    },

    "hope": {
        "definition":
            "Expectation of a favourable development or improvement.",
    },

    "relief": {
        "definition":
            "Reduction of previously established danger, pressure, or distress.",
    },

    "trust": {
        "definition":
            "Confidence in a person, institution, process, or claim.",
    },

    "compassion": {
        "definition":
            "Concern, sympathy, or care directed toward another person or group.",
    },

    "solidarity": {
        "definition":
            "Affiliation, common identification, or collective support.",
    },
}


# -----------------------------------------------------------------------------
# Factological composition
# -----------------------------------------------------------------------------

FACTOLOGICAL_COMPONENTS = {
    "explicit_claim": {
        "definition":
            "An explicit proposition that could in principle be externally "
            "assessed.",
    },

    "observation_or_event": {
        "definition":
            "A described event, action, occurrence, or observable condition.",
    },

    "quantitative_information": {
        "definition":
            "Numerical, temporal, measured, or otherwise quantified information.",
    },

    "attribution": {
        "definition":
            "Identification of the source responsible for a statement or claim.",
    },

    "evidence_reference": {
        "definition":
            "Reference to evidence, documentation, data, investigation, or "
            "supporting material.",
    },

    "certainty_statement": {
        "definition":
            "Language explicitly expressing confidence, confirmation, or "
            "established status.",
    },

    "uncertainty_statement": {
        "definition":
            "Language explicitly expressing uncertainty, doubt, qualification, "
            "or incomplete knowledge.",
    },

    "speculation_or_prediction": {
        "definition":
            "A statement describing a possible, expected, hypothetical, or "
            "future outcome.",
    },
}


# -----------------------------------------------------------------------------
# Core supervision constructs
# -----------------------------------------------------------------------------

SUPERVISION_CONSTRUCTS = {
    "reported_emotion": {
        "type":
            "multidimensional_non_negative",

        "dimensions":
            list(
                EMOTIONAL_DIMENSIONS.keys()
            ),

        "scale":
            deepcopy(
                NON_NEGATIVE_WEIGHT_SCALE
            ),

        "definition":
            "Emotion represented as belonging to a person, group, or subject "
            "described by the text.",

        "question":
            "Which emotions are reported as belonging to the subjects "
            "described by this textual unit?",
    },

    "transformative_emotion": {
        "type":
            "multidimensional_signed_change",

        "dimensions":
            list(
                EMOTIONAL_DIMENSIONS.keys()
            ),

        "scale":
            deepcopy(
                SIGNED_TRANSFORMATION_SCALE
            ),

        "definition":
            "Signed change in emotional state produced by the current textual "
            "unit relative to preceding context.",

        "question":
            "How does this textual unit change the emotional state established "
            "by the preceding context?",
    },

    "immediate_emotional_state": {
        "type":
            "multidimensional_non_negative",

        "dimensions":
            list(
                EMOTIONAL_DIMENSIONS.keys()
            ),

        "scale":
            deepcopy(
                NON_NEGATIVE_WEIGHT_SCALE
            ),

        "definition":
            "Emotional state present after the current transformation has been "
            "incorporated.",

        "question":
            "What emotional state is present after processing the current "
            "textual unit?",
    },

    "emotional_weight": {
        "type":
            "continuous_non_negative",

        "scale":
            deepcopy(
                NON_NEGATIVE_WEIGHT_SCALE
            ),

        "definition":
            "Strength of emotionally consequential information associated with "
            "the textual unit.",

        "question":
            "How emotionally consequential is this textual unit under the "
            "operational definition?",
    },

    "factological_weight": {
        "type":
            "continuous_non_negative",

        "scale":
            deepcopy(
                NON_NEGATIVE_WEIGHT_SCALE
            ),

        "definition":
            "Contribution of externally assessable informational content. "
            "This is not an estimate of truth.",

        "question":
            "How much externally assessable informational content does this "
            "textual unit contribute?",
    },

    "factological_components": {
        "type":
            "multi_label_boolean",

        "dimensions":
            list(
                FACTOLOGICAL_COMPONENTS.keys()
            ),

        "definition":
            "Interpretable indicators describing the type of informational "
            "content contributing to factological weight.",
    },

    "annotation_confidence": {
        "type":
            "continuous_non_negative",

        "scale":
            deepcopy(
                CONFIDENCE_SCALE
            ),

        "definition":
            "Supervisor certainty that the assigned annotation is supported "
            "by the operational definition and available context.",
    },
}


# -----------------------------------------------------------------------------
# Annotation context policy
# -----------------------------------------------------------------------------

ANNOTATION_CONTEXT_POLICY = {
    "primary_annotation_unit":
        "sentence",

    "context_mode":
        "preceding_article_context",

    "complete_article_available":
        True,

    "preserve_article_order":
        True,

    "preserve_paragraph_order":
        True,

    "preserve_sentence_order":
        True,

    "isolated_sentence_annotation_allowed":
        False,

    "semantic_prefiltering_allowed":
        False,

    "emotional_token_prefiltering_allowed":
        False,
}


# -----------------------------------------------------------------------------
# Provenance requirements
# -----------------------------------------------------------------------------

ANNOTATION_PROVENANCE_FIELDS = [
    "media_ai_record_id",
    "paragraph_id",
    "sentence_id",
    "supervision_schema_version",
    "annotation_policy_version",
    "supervision_method",
    "supervisor_role",
    "supervisor_identifier",
    "annotated_at_utc",
    "context_scope",
    "annotation_confidence",
    "review_status",
]


# -----------------------------------------------------------------------------
# Review states
# -----------------------------------------------------------------------------

ANNOTATION_REVIEW_STATES = {
    "unreviewed",
    "review_required",
    "human_reviewed",
    "adjudicated",
}


# -----------------------------------------------------------------------------
# Construct complete annotation schema
# -----------------------------------------------------------------------------

ANNOTATION_SCHEMA = {
    "schema_type":
        "media_ai_supervision_schema",

    "schema_version":
        SUPERVISION_SCHEMA_VERSION,

    "annotation_policy_version":
        ANNOTATION_POLICY_VERSION,

    "primary_annotation_unit":
        ANNOTATION_CONTEXT_POLICY[
            "primary_annotation_unit"
        ],

    "context_policy":
        ANNOTATION_CONTEXT_POLICY,

    "emotional_dimensions":
        EMOTIONAL_DIMENSIONS,

    "factological_components":
        FACTOLOGICAL_COMPONENTS,

    "constructs":
        SUPERVISION_CONSTRUCTS,

    "value_states":
        sorted(
            ANNOTATION_VALUE_STATES
        ),

    "review_states":
        sorted(
            ANNOTATION_REVIEW_STATES
        ),

    "provenance_fields":
        ANNOTATION_PROVENANCE_FIELDS,

    "methodological_constraints": {
        "reported_emotion_distinct_from_transformative_emotion":
            True,

        "emotional_weight_distinct_from_direction":
            True,

        "factological_weight_is_not_truth_probability":
            True,

        "emotional_and_factological_weights_need_not_sum_to_one":
            True,

        "insufficient_evidence_distinct_from_zero":
            True,

        "annotations_are_derived_research_artifacts":
            True,

        "prepared_text_must_not_be_overwritten":
            True,

        "schema_independent_of_model_architecture":
            True,
    },
}


# -----------------------------------------------------------------------------
# Schema validation
# -----------------------------------------------------------------------------

required_constructs = {
    "reported_emotion",
    "transformative_emotion",
    "immediate_emotional_state",
    "emotional_weight",
    "factological_weight",
    "factological_components",
    "annotation_confidence",
}

missing_constructs = (
    required_constructs
    - set(
        ANNOTATION_SCHEMA[
            "constructs"
        ].keys()
    )
)

if missing_constructs:
    raise ValueError(
        "Annotation schema is missing required constructs: "
        f"{sorted(missing_constructs)}"
    )


if (
    ANNOTATION_SCHEMA[
        "constructs"
    ][
        "transformative_emotion"
    ][
        "scale"
    ][
        "minimum"
    ]
    >= 0
):
    raise ValueError(
        "Transformative emotion must support negative state changes."
    )


if (
    ANNOTATION_CONTEXT_POLICY[
        "isolated_sentence_annotation_allowed"
    ]
):
    raise ValueError(
        "Transformative supervision must preserve preceding context."
    )


if (
    ANNOTATION_CONTEXT_POLICY[
        "emotional_token_prefiltering_allowed"
    ]
):
    raise ValueError(
        "Emotional-token pre-filtering is not permitted."
    )


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Annotation Schema")
print("=" * 72)

print(
    f"Supervision schema         : "
    f"{SUPERVISION_SCHEMA_VERSION}"
)

print(
    f"Annotation policy          : "
    f"{ANNOTATION_POLICY_VERSION}"
)

print(
    f"Primary annotation unit    : "
    f"{ANNOTATION_CONTEXT_POLICY['primary_annotation_unit']}"
)

print(
    f"Context mode               : "
    f"{ANNOTATION_CONTEXT_POLICY['context_mode']}"
)

print(
    f"Emotional dimensions       : "
    f"{len(EMOTIONAL_DIMENSIONS)}"
)

print(
    f"Factological components    : "
    f"{len(FACTOLOGICAL_COMPONENTS)}"
)

print(
    f"Supervision constructs     : "
    f"{len(SUPERVISION_CONSTRUCTS)}"
)

print(
    f"Transformative scale       : "
    f"{SIGNED_TRANSFORMATION_SCALE['minimum']} "
    f"to {SIGNED_TRANSFORMATION_SCALE['maximum']}"
)

print(
    f"Weight scale               : "
    f"{NON_NEGATIVE_WEIGHT_SCALE['minimum']} "
    f"to {NON_NEGATIVE_WEIGHT_SCALE['maximum']}"
)

print(
    f"Insufficient evidence state: "
    f"{'insufficient_evidence' in ANNOTATION_VALUE_STATES}"
)

print(
    f"Reported / transformative  : "
    f"{ANNOTATION_SCHEMA['methodological_constraints']['reported_emotion_distinct_from_transformative_emotion']}"
)

print(
    f"Factological ≠ truth       : "
    f"{ANNOTATION_SCHEMA['methodological_constraints']['factological_weight_is_not_truth_probability']}"
)

print(
    f"Architecture-independent   : "
    f"{ANNOTATION_SCHEMA['methodological_constraints']['schema_independent_of_model_architecture']}"
)

print("=" * 72)

print(
    "Annotation schema and supervision constructs initialised successfully."
)

print(
    "No supervision labels have been generated by this block."
)

print(
    "No prepared text has been modified."
)

print(
    "No external model or API calls have been made by this block."
)
```

## Supervision Record Schema and Annotation Template Construction

This block converts the conceptual supervision definitions from Block 3 into a
strict machine-readable annotation format.

The purpose is to ensure that all later supervision methods — including external
LLM supervision, human review and adjudication — produce records with the same
structure and meaning.

The annotation template is defined before any labels are generated.

### Structural Identity

Every annotation record remains linked to the prepared corpus through:

- `media_ai_record_id`;
- `paragraph_id`;
- `sentence_id`;
- paragraph index;
- sentence index.

This preserves the article → paragraph → sentence hierarchy established in
Notebook 02.

### Context Bundle

The primary annotation unit remains the sentence in preceding article context.

For each annotation target, the context bundle contains:

- article title;
- current sentence;
- preceding sentences within the article;
- paragraph and sentence position;
- complete article identifier.

The complete prepared article remains available in the source corpus. The
annotation record stores only the context required for supervision and does not
replace the canonical prepared text.

### Emotion Vectors

The annotation template contains three independent emotion vectors over the same
11-dimensional ontology:

- `reported_emotion`;
- `transformative_emotion`;
- `immediate_emotional_state`.

Reported emotion and immediate emotional state use the non-negative `[0,1]`
scale.

Transformative emotion uses the signed `[-1,1]` scale because it represents
increase or decrease relative to preceding context.

Every emotional dimension also carries an explicit value state so that numeric
zero remains distinguishable from `insufficient_evidence` or `not_applicable`.

### Scalar Weights

The template includes:

- `emotional_weight` in `[0,1]`;
- `factological_weight` in `[0,1]`.

These are independent quantities and do not need to sum to one.

### Factological Components

The eight factological components are represented as explicit boolean indicators.

These indicators describe the type of informational content present and do not
assert whether the underlying claim is true.

### Evidence and Rationale

Every annotation may include concise evidence supporting the assigned labels.

The evidence field should identify the textual or contextual basis for the
annotation without replacing the structured numerical values.

This field is important for:

- later human review;
- disagreement analysis;
- supervisor comparison;
- annotation debugging.

The evidence is an annotation artefact and is never substituted for the original
article text.

### Confidence and Review

Each annotation contains:

- overall annotation confidence;
- review status;
- optional review notes.

Confidence remains distinct from the numeric supervision targets.

### Provenance

Every annotation must identify how it was generated.

Required provenance includes:

- supervision schema version;
- annotation policy version;
- supervision method;
- supervisor role;
- supervisor identifier;
- annotation timestamp;
- context scope.

This supports later comparison between independent LLM supervisors and human
review.

### Template Validation

The block validates that:

- all required emotional dimensions are present;
- all factological components are present;
- signed and non-negative ranges are distinct;
- missing-information states are explicit;
- provenance fields are available;
- the template does not contain model-training or recurrent-state outputs.

### Output

This block creates:

- `ANNOTATION_RECORD_TEMPLATE`;
- helper functions for creating blank annotation records;
- validation functions for completed annotation records;
- supervision context records derived from the prepared sentence hierarchy.

No actual emotional or factological labels are assigned by this block.

No external model or API is called.

```python
# =============================================================================
# Media AI — Supervision Record Schema and Annotation Template Construction
# Block 4 — Canonical Record Contract and Validation
# =============================================================================

from copy import deepcopy
from datetime import datetime, timezone


# -----------------------------------------------------------------------------
# Block configuration
# -----------------------------------------------------------------------------

ANNOTATION_RECORD_SCHEMA_VERSION = "1.0"

ANNOTATION_VALIDATION_CONTRACT_VERSION = "1.1"

DEFAULT_VALUE_STATE = "insufficient_evidence"

DEFAULT_REVIEW_STATUS = "unreviewed"

DEFAULT_SUPERVISION_METHOD = "unassigned"

DEFAULT_SUPERVISOR_ROLE = "unassigned"

DEFAULT_SUPERVISOR_IDENTIFIER = "unassigned"


# -----------------------------------------------------------------------------
# Upstream dependency validation
# -----------------------------------------------------------------------------

required_schema_objects = [
    "SUPERVISION_SCHEMA_VERSION",
    "ANNOTATION_POLICY_VERSION",
    "ANNOTATION_CONTEXT_POLICY",
    "ANNOTATION_VALUE_STATES",
    "ANNOTATION_REVIEW_STATES",
    "EMOTIONAL_DIMENSIONS",
    "FACTOLOGICAL_COMPONENTS",
    "PREPARED_CORPUS_LOADED",
    "PREPARED_ARTICLE_RECORDS",
]


missing_schema_objects = [
    object_name
    for object_name
    in required_schema_objects
    if object_name not in globals()
]


if missing_schema_objects:

    raise NameError(
        "Notebook 03 Block 4 requires the annotation schema and prepared "
        "corpus objects initialised by the preceding blocks. Missing: "
        f"{missing_schema_objects}"
    )


if not isinstance(
    EMOTIONAL_DIMENSIONS,
    dict,
):

    raise TypeError(
        "EMOTIONAL_DIMENSIONS must be a dictionary."
    )


if not isinstance(
    FACTOLOGICAL_COMPONENTS,
    dict,
):

    raise TypeError(
        "FACTOLOGICAL_COMPONENTS must be a dictionary."
    )


if not isinstance(
    ANNOTATION_CONTEXT_POLICY,
    dict,
):

    raise TypeError(
        "ANNOTATION_CONTEXT_POLICY must be a dictionary."
    )


# -----------------------------------------------------------------------------
# Canonical annotation value scales
# -----------------------------------------------------------------------------
#
# These scales are part of the Notebook 03 supervision contract.
#
# They are defined locally in Block 4 because validate_annotation_record()
# depends directly on them.
#
# This prevents the completed-record validator from depending on temporary
# runtime constants that may not exist after a Colab restart.
# -----------------------------------------------------------------------------

NON_NEGATIVE_STATE_SCALE = {
    "minimum":
        0.0,

    "maximum":
        1.0,
}


SIGNED_TRANSFORMATION_SCALE = {
    "minimum":
        -1.0,

    "maximum":
        1.0,
}


NON_NEGATIVE_WEIGHT_SCALE = {
    "minimum":
        0.0,

    "maximum":
        1.0,
}


CONFIDENCE_SCALE = {
    "minimum":
        0.0,

    "maximum":
        1.0,
}


# -----------------------------------------------------------------------------
# Scale-contract validation
# -----------------------------------------------------------------------------

expected_scale_contract = {
    "non_negative_state": {
        "minimum":
            0.0,

        "maximum":
            1.0,
    },

    "signed_transformation": {
        "minimum":
            -1.0,

        "maximum":
            1.0,
    },

    "non_negative_weight": {
        "minimum":
            0.0,

        "maximum":
            1.0,
    },

    "confidence": {
        "minimum":
            0.0,

        "maximum":
            1.0,
    },
}


actual_scale_contract = {
    "non_negative_state":
        NON_NEGATIVE_STATE_SCALE,

    "signed_transformation":
        SIGNED_TRANSFORMATION_SCALE,

    "non_negative_weight":
        NON_NEGATIVE_WEIGHT_SCALE,

    "confidence":
        CONFIDENCE_SCALE,
}


scale_contract_valid = (
    actual_scale_contract
    == expected_scale_contract
)


if not scale_contract_valid:

    raise ValueError(
        "Notebook 03 annotation-value scales do not match "
        "the canonical supervision contract."
    )


# -----------------------------------------------------------------------------
# Machine-readable annotation schema
# -----------------------------------------------------------------------------
#
# This object provides one stable schema descriptor for downstream persistence
# and handover logic.
#
# It does not contain annotation values.
# -----------------------------------------------------------------------------

ANNOTATION_SCHEMA = {
    "schema_type":
        "media_ai_annotation_schema",

    "record_schema_version":
        ANNOTATION_RECORD_SCHEMA_VERSION,

    "validation_contract_version":
        ANNOTATION_VALIDATION_CONTRACT_VERSION,

    "supervision_schema_version":
        SUPERVISION_SCHEMA_VERSION,

    "annotation_policy_version":
        ANNOTATION_POLICY_VERSION,

    "primary_annotation_unit":
        ANNOTATION_CONTEXT_POLICY.get(
            "primary_annotation_unit"
        ),

    "context_mode":
        ANNOTATION_CONTEXT_POLICY.get(
            "context_mode"
        ),

    "emotional_dimensions":
        list(
            EMOTIONAL_DIMENSIONS.keys()
        ),

    "factological_components":
        list(
            FACTOLOGICAL_COMPONENTS.keys()
        ),

    "annotation_value_states":
        sorted(
            ANNOTATION_VALUE_STATES
        ),

    "review_states":
        sorted(
            ANNOTATION_REVIEW_STATES
        ),

    "scales": {
        "reported_emotion":
            deepcopy(
                NON_NEGATIVE_STATE_SCALE
            ),

        "transformative_emotion":
            deepcopy(
                SIGNED_TRANSFORMATION_SCALE
            ),

        "immediate_emotional_state":
            deepcopy(
                NON_NEGATIVE_STATE_SCALE
            ),

        "emotional_weight":
            deepcopy(
                NON_NEGATIVE_WEIGHT_SCALE
            ),

        "factological_weight":
            deepcopy(
                NON_NEGATIVE_WEIGHT_SCALE
            ),

        "annotation_confidence":
            deepcopy(
                CONFIDENCE_SCALE
            ),
    },
}


# -----------------------------------------------------------------------------
# Annotation-value helpers
# -----------------------------------------------------------------------------

def create_scalar_annotation_value():
    """
    Create one blank scalar annotation value.

    Numeric values remain separate from missing-information states.
    """

    return {
        "value":
            None,

        "value_state":
            DEFAULT_VALUE_STATE,
    }


def create_emotion_vector(
    signed=False,
):
    """
    Create one blank emotion vector over the configured emotional ontology.

    The signed argument documents whether the vector will later use the
    signed transformation scale. Blank records themselves contain no values.
    """

    vector = {}


    for dimension in EMOTIONAL_DIMENSIONS:

        vector[
            dimension
        ] = {
            "value":
                None,

            "value_state":
                DEFAULT_VALUE_STATE,
        }


    return vector


def create_factological_component_vector():
    """
    Create one blank multi-label factological-component representation.
    """

    return {
        component:
            None

        for component
        in FACTOLOGICAL_COMPONENTS
    }


# -----------------------------------------------------------------------------
# Canonical blank annotation template
# -----------------------------------------------------------------------------

ANNOTATION_RECORD_TEMPLATE = {
    "record_type":
        "media_ai_supervision_record",

    "record_schema_version":
        ANNOTATION_RECORD_SCHEMA_VERSION,

    "identity": {
        "media_ai_record_id":
            None,

        "paragraph_id":
            None,

        "sentence_id":
            None,

        "paragraph_index":
            None,

        "sentence_index":
            None,
    },

    "context": {
        "article_title":
            None,

        "current_sentence":
            None,

        "preceding_context":
            None,

        "context_scope":
            ANNOTATION_CONTEXT_POLICY[
                "context_mode"
            ],
    },

    "annotations": {
        "reported_emotion":
            create_emotion_vector(
                signed=False
            ),

        "transformative_emotion":
            create_emotion_vector(
                signed=True
            ),

        "immediate_emotional_state":
            create_emotion_vector(
                signed=False
            ),

        "emotional_weight":
            create_scalar_annotation_value(),

        "factological_weight":
            create_scalar_annotation_value(),

        "factological_components":
            create_factological_component_vector(),

        "annotation_confidence":
            create_scalar_annotation_value(),
    },

    "evidence": {
        "summary":
            None,

        "supporting_text":
            [],
    },

    "review": {
        "review_status":
            DEFAULT_REVIEW_STATUS,

        "review_notes":
            None,
    },

    "provenance": {
        "supervision_schema_version":
            SUPERVISION_SCHEMA_VERSION,

        "annotation_policy_version":
            ANNOTATION_POLICY_VERSION,

        "supervision_method":
            DEFAULT_SUPERVISION_METHOD,

        "supervisor_role":
            DEFAULT_SUPERVISOR_ROLE,

        "supervisor_identifier":
            DEFAULT_SUPERVISOR_IDENTIFIER,

        "annotated_at_utc":
            None,

        "context_scope":
            ANNOTATION_CONTEXT_POLICY[
                "context_mode"
            ],
    },
}


# -----------------------------------------------------------------------------
# Build sentence-level supervision contexts
# -----------------------------------------------------------------------------

SUPERVISION_CONTEXT_RECORDS = []


if PREPARED_CORPUS_LOADED:

    for article in PREPARED_ARTICLE_RECORDS:

        media_ai_record_id = article.get(
            "media_ai_record_id"
        )


        article_title = article.get(
            "title"
        )


        preceding_sentences = []


        for paragraph in article.get(
            "paragraphs",
            [],
        ):

            paragraph_id = paragraph.get(
                "paragraph_id"
            )


            paragraph_index = paragraph.get(
                "paragraph_index"
            )


            for sentence in paragraph.get(
                "sentences",
                [],
            ):

                sentence_id = sentence.get(
                    "sentence_id"
                )


                sentence_index = sentence.get(
                    "sentence_index"
                )


                current_sentence = sentence.get(
                    "text",
                    "",
                )


                context_record = {
                    "media_ai_record_id":
                        media_ai_record_id,

                    "paragraph_id":
                        paragraph_id,

                    "sentence_id":
                        sentence_id,

                    "paragraph_index":
                        paragraph_index,

                    "sentence_index":
                        sentence_index,

                    "article_title":
                        article_title,

                    "current_sentence":
                        current_sentence,

                    "preceding_context":
                        " ".join(
                            preceding_sentences
                        ),

                    "preceding_sentence_count":
                        len(
                            preceding_sentences
                        ),

                    "context_scope":
                        ANNOTATION_CONTEXT_POLICY[
                            "context_mode"
                        ],
                }


                SUPERVISION_CONTEXT_RECORDS.append(
                    context_record
                )


                preceding_sentences.append(
                    current_sentence
                )


# -----------------------------------------------------------------------------
# Supervision-context validation
# -----------------------------------------------------------------------------

context_sentence_ids = [
    context_record.get(
        "sentence_id"
    )

    for context_record
    in SUPERVISION_CONTEXT_RECORDS
]


context_ids_present = (
    all(
        context_sentence_ids
    )
    if context_sentence_ids
    else False
)


context_ids_unique = (
    len(
        context_sentence_ids
    )
    == len(
        set(
            context_sentence_ids
        )
    )
    if context_sentence_ids
    else False
)


if (
    PREPARED_CORPUS_LOADED
    and not context_ids_present
):

    raise ValueError(
        "One or more supervision-context records have no sentence ID."
    )


if (
    PREPARED_CORPUS_LOADED
    and not context_ids_unique
):

    raise ValueError(
        "Supervision-context sentence IDs are not unique."
    )


# -----------------------------------------------------------------------------
# Annotation-record constructor
# -----------------------------------------------------------------------------

def create_blank_annotation_record(
    context_record,
):
    """
    Create one blank annotation record linked to one prepared sentence.
    """

    if not isinstance(
        context_record,
        dict,
    ):

        raise TypeError(
            "context_record must be a dictionary."
        )


    record = deepcopy(
        ANNOTATION_RECORD_TEMPLATE
    )


    record[
        "identity"
    ].update({
        "media_ai_record_id":
            context_record.get(
                "media_ai_record_id"
            ),

        "paragraph_id":
            context_record.get(
                "paragraph_id"
            ),

        "sentence_id":
            context_record.get(
                "sentence_id"
            ),

        "paragraph_index":
            context_record.get(
                "paragraph_index"
            ),

        "sentence_index":
            context_record.get(
                "sentence_index"
            ),
    })


    record[
        "context"
    ].update({
        "article_title":
            context_record.get(
                "article_title"
            ),

        "current_sentence":
            context_record.get(
                "current_sentence"
            ),

        "preceding_context":
            context_record.get(
                "preceding_context"
            ),

        "context_scope":
            context_record.get(
                "context_scope"
            ),
    })


    record[
        "provenance"
    ][
        "context_scope"
    ] = context_record.get(
        "context_scope"
    )


    return record


# -----------------------------------------------------------------------------
# Range validation helper
# -----------------------------------------------------------------------------

def value_in_range(
    value,
    scale,
):
    """
    Validate a numeric value against one configured annotation scale.
    """

    if value is None:

        return False


    if isinstance(
        value,
        bool,
    ):

        return False


    if not isinstance(
        value,
        (int, float),
    ):

        return False


    if not isinstance(
        scale,
        dict,
    ):

        return False


    if (
        "minimum"
        not in scale
        or "maximum"
        not in scale
    ):

        return False


    return (
        scale[
            "minimum"
        ]
        <= float(
            value
        )
        <= scale[
            "maximum"
        ]
    )


# -----------------------------------------------------------------------------
# Annotation-value validation helper
# -----------------------------------------------------------------------------

def validate_annotation_value(
    annotation_value,
    scale,
):
    """
    Validate one scalar or emotion-dimension annotation.

    Observed:
        numeric value required within the configured range.

    insufficient_evidence / not_applicable:
        numeric value must be None.
    """

    if not isinstance(
        annotation_value,
        dict,
    ):

        return False


    if set(
        annotation_value.keys()
    ) != {
        "value",
        "value_state",
    }:

        return False


    value_state = annotation_value.get(
        "value_state"
    )


    if (
        value_state
        not in ANNOTATION_VALUE_STATES
    ):

        return False


    value = annotation_value.get(
        "value"
    )


    if value_state == "observed":

        return value_in_range(
            value,
            scale,
        )


    return value is None


# -----------------------------------------------------------------------------
# Completed-record validation
# -----------------------------------------------------------------------------

def validate_annotation_record(
    record,
):
    """
    Validate one completed Media AI supervision record.

    Returns:
        (passed, errors)
    """

    errors = []


    if not isinstance(
        record,
        dict,
    ):

        return (
            False,
            [
                "record_not_dictionary"
            ],
        )


    # -------------------------------------------------------------------------
    # Record identity
    # -------------------------------------------------------------------------

    if (
        record.get(
            "record_type"
        )
        != "media_ai_supervision_record"
    ):

        errors.append(
            "invalid_record_type"
        )


    if (
        record.get(
            "record_schema_version"
        )
        != ANNOTATION_RECORD_SCHEMA_VERSION
    ):

        errors.append(
            "invalid_record_schema_version"
        )


    identity = record.get(
        "identity",
        {},
    )


    for field in [
        "media_ai_record_id",
        "paragraph_id",
        "sentence_id",
    ]:

        if not identity.get(
            field
        ):

            errors.append(
                f"missing_identity_{field}"
            )


    # -------------------------------------------------------------------------
    # Context
    # -------------------------------------------------------------------------

    context = record.get(
        "context",
        {},
    )


    if not isinstance(
        context,
        dict,
    ):

        errors.append(
            "invalid_context_structure"
        )


    else:

        if not isinstance(
            context.get(
                "current_sentence"
            ),
            str,
        ):

            errors.append(
                "invalid_current_sentence"
            )


        if (
            context.get(
                "preceding_context"
            )
            is not None
            and not isinstance(
                context.get(
                    "preceding_context"
                ),
                str,
            )
        ):

            errors.append(
                "invalid_preceding_context"
            )


    # -------------------------------------------------------------------------
    # Emotional vectors
    # -------------------------------------------------------------------------

    annotations = record.get(
        "annotations",
        {},
    )


    if not isinstance(
        annotations,
        dict,
    ):

        return (
            False,
            errors
            + [
                "invalid_annotations_structure"
            ],
        )


    emotion_vector_specs = {
        "reported_emotion":
            NON_NEGATIVE_STATE_SCALE,

        "transformative_emotion":
            SIGNED_TRANSFORMATION_SCALE,

        "immediate_emotional_state":
            NON_NEGATIVE_STATE_SCALE,
    }


    for (
        vector_name,
        scale,
    ) in emotion_vector_specs.items():

        vector = annotations.get(
            vector_name,
            {},
        )


        if not isinstance(
            vector,
            dict,
        ):

            errors.append(
                f"invalid_structure_{vector_name}"
            )

            continue


        if set(
            vector.keys()
        ) != set(
            EMOTIONAL_DIMENSIONS.keys()
        ):

            errors.append(
                f"invalid_dimensions_{vector_name}"
            )

            continue


        for dimension in EMOTIONAL_DIMENSIONS:

            if not validate_annotation_value(
                vector.get(
                    dimension
                ),
                scale,
            ):

                errors.append(
                    f"invalid_{vector_name}_{dimension}"
                )


    # -------------------------------------------------------------------------
    # Scalar annotations
    # -------------------------------------------------------------------------

    scalar_specs = {
        "emotional_weight":
            NON_NEGATIVE_WEIGHT_SCALE,

        "factological_weight":
            NON_NEGATIVE_WEIGHT_SCALE,

        "annotation_confidence":
            CONFIDENCE_SCALE,
    }


    for (
        field,
        scale,
    ) in scalar_specs.items():

        if not validate_annotation_value(
            annotations.get(
                field
            ),
            scale,
        ):

            errors.append(
                f"invalid_{field}"
            )


    # -------------------------------------------------------------------------
    # Factological components
    # -------------------------------------------------------------------------

    factological_components = annotations.get(
        "factological_components",
        {},
    )


    if not isinstance(
        factological_components,
        dict,
    ):

        errors.append(
            "invalid_factological_components"
        )


    elif set(
        factological_components.keys()
    ) != set(
        FACTOLOGICAL_COMPONENTS.keys()
    ):

        errors.append(
            "invalid_factological_components"
        )


    else:

        for (
            component,
            value,
        ) in factological_components.items():

            if value not in {
                True,
                False,
                None,
            }:

                errors.append(
                    f"invalid_factological_component_{component}"
                )


    # -------------------------------------------------------------------------
    # Evidence
    # -------------------------------------------------------------------------

    evidence = record.get(
        "evidence",
        {},
    )


    if not isinstance(
        evidence,
        dict,
    ):

        errors.append(
            "invalid_evidence_structure"
        )


    else:

        if set(
            evidence.keys()
        ) != {
            "summary",
            "supporting_text",
        }:

            errors.append(
                "invalid_evidence_fields"
            )


        summary = evidence.get(
            "summary"
        )


        supporting_text = evidence.get(
            "supporting_text"
        )


        if (
            summary is not None
            and not isinstance(
                summary,
                str,
            )
        ):

            errors.append(
                "invalid_evidence_summary"
            )


        if not isinstance(
            supporting_text,
            list,
        ):

            errors.append(
                "invalid_supporting_text"
            )


        elif not all(
            isinstance(
                item,
                str,
            )

            for item
            in supporting_text
        ):

            errors.append(
                "invalid_supporting_text_item"
            )


    # -------------------------------------------------------------------------
    # Review status
    # -------------------------------------------------------------------------

    review = record.get(
        "review",
        {},
    )


    review_status = review.get(
        "review_status"
    )


    if (
        review_status
        not in ANNOTATION_REVIEW_STATES
    ):

        errors.append(
            "invalid_review_status"
        )


    # -------------------------------------------------------------------------
    # Provenance
    # -------------------------------------------------------------------------

    provenance = record.get(
        "provenance",
        {},
    )


    if not isinstance(
        provenance,
        dict,
    ):

        errors.append(
            "invalid_provenance_structure"
        )


    else:

        if (
            provenance.get(
                "supervision_schema_version"
            )
            != SUPERVISION_SCHEMA_VERSION
        ):

            errors.append(
                "invalid_supervision_schema_version"
            )


        if (
            provenance.get(
                "annotation_policy_version"
            )
            != ANNOTATION_POLICY_VERSION
        ):

            errors.append(
                "invalid_annotation_policy_version"
            )


    return (
        len(
            errors
        )
        == 0,
        errors,
    )


# -----------------------------------------------------------------------------
# Construct blank pilot annotation records
# -----------------------------------------------------------------------------

BLANK_ANNOTATION_RECORDS = [
    create_blank_annotation_record(
        context_record
    )

    for context_record
    in SUPERVISION_CONTEXT_RECORDS
]


# -----------------------------------------------------------------------------
# Template-level validation
# -----------------------------------------------------------------------------

template_dimensions_valid = all([
    set(
        ANNOTATION_RECORD_TEMPLATE[
            "annotations"
        ][
            "reported_emotion"
        ].keys()
    )
    == set(
        EMOTIONAL_DIMENSIONS.keys()
    ),

    set(
        ANNOTATION_RECORD_TEMPLATE[
            "annotations"
        ][
            "transformative_emotion"
        ].keys()
    )
    == set(
        EMOTIONAL_DIMENSIONS.keys()
    ),

    set(
        ANNOTATION_RECORD_TEMPLATE[
            "annotations"
        ][
            "immediate_emotional_state"
        ].keys()
    )
    == set(
        EMOTIONAL_DIMENSIONS.keys()
    ),
])


factological_template_valid = (
    set(
        ANNOTATION_RECORD_TEMPLATE[
            "annotations"
        ][
            "factological_components"
        ].keys()
    )
    == set(
        FACTOLOGICAL_COMPONENTS.keys()
    )
)


if not template_dimensions_valid:

    raise ValueError(
        "Annotation template emotional dimensions "
        "do not match the supervision schema."
    )


if not factological_template_valid:

    raise ValueError(
        "Annotation template factological components "
        "do not match the supervision schema."
    )


# -----------------------------------------------------------------------------
# Blank-record validation
# -----------------------------------------------------------------------------
#
# Blank records intentionally contain insufficient_evidence with null values.
# They should therefore satisfy the structural annotation contract.
# -----------------------------------------------------------------------------

blank_record_validation_results = [
    validate_annotation_record(
        record
    )

    for record
    in BLANK_ANNOTATION_RECORDS
]


blank_records_valid = all(
    passed
    for (
        passed,
        _
    ) in blank_record_validation_results
)


if not blank_records_valid:

    first_blank_errors = next(
        errors

        for (
            passed,
            errors,
        ) in blank_record_validation_results

        if not passed
    )


    raise ValueError(
        "One or more blank annotation records failed structural validation. "
        f"First errors: {first_blank_errors}"
    )


# -----------------------------------------------------------------------------
# Block validity
# -----------------------------------------------------------------------------

ANNOTATION_RECORD_SCHEMA_VALID = all([
    scale_contract_valid,
    template_dimensions_valid,
    factological_template_valid,
    context_ids_present,
    context_ids_unique,
    blank_records_valid,
])


if not ANNOTATION_RECORD_SCHEMA_VALID:

    raise ValueError(
        "Notebook 03 Block 4 annotation-record contract is invalid."
    )


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)

print(
    "Media AI — Supervision Record Schema"
)

print("=" * 72)


print(
    f"Record schema version      : "
    f"{ANNOTATION_RECORD_SCHEMA_VERSION}"
)

print(
    f"Validation contract        : "
    f"{ANNOTATION_VALIDATION_CONTRACT_VERSION}"
)

print(
    f"Prepared corpus loaded     : "
    f"{PREPARED_CORPUS_LOADED}"
)


print("-" * 72)

print(
    "Canonical scales"
)

print(
    f"Reported emotion scale     : "
    f"{NON_NEGATIVE_STATE_SCALE}"
)

print(
    f"Transformative scale       : "
    f"{SIGNED_TRANSFORMATION_SCALE}"
)

print(
    f"Immediate-state scale      : "
    f"{NON_NEGATIVE_STATE_SCALE}"
)

print(
    f"Weight scale               : "
    f"{NON_NEGATIVE_WEIGHT_SCALE}"
)

print(
    f"Confidence scale           : "
    f"{CONFIDENCE_SCALE}"
)

print(
    f"Scale contract valid       : "
    f"{scale_contract_valid}"
)


print("-" * 72)

print(
    "Structural representation"
)

print(
    f"Supervision contexts       : "
    f"{len(SUPERVISION_CONTEXT_RECORDS)}"
)

print(
    f"Blank annotation records   : "
    f"{len(BLANK_ANNOTATION_RECORDS)}"
)

print(
    f"Context IDs present        : "
    f"{context_ids_present}"
)

print(
    f"Context IDs unique         : "
    f"{context_ids_unique}"
)

print(
    f"Emotional dimensions       : "
    f"{len(EMOTIONAL_DIMENSIONS)}"
)

print(
    f"Factological components    : "
    f"{len(FACTOLOGICAL_COMPONENTS)}"
)


print("-" * 72)

print(
    "Schema validation"
)

print(
    f"Template dimensions valid  : "
    f"{template_dimensions_valid}"
)

print(
    f"Factological schema valid  : "
    f"{factological_template_valid}"
)

print(
    f"Blank records valid        : "
    f"{blank_records_valid}"
)

print(
    f"Annotation schema defined  : "
    f"{isinstance(ANNOTATION_SCHEMA, dict)}"
)

print(
    f"Record schema valid        : "
    f"{ANNOTATION_RECORD_SCHEMA_VALID}"
)


print("-" * 72)

print(
    f"Primary annotation unit    : "
    f"{ANNOTATION_CONTEXT_POLICY['primary_annotation_unit']}"
)

print(
    f"Context mode               : "
    f"{ANNOTATION_CONTEXT_POLICY['context_mode']}"
)


print("=" * 72)


print(
    "Supervision record schema and annotation templates "
    "initialised successfully."
)

print(
    "Canonical numerical scales are now defined locally in Block 4."
)

print(
    "Completed-record validation therefore no longer depends on "
    "undefined runtime scale objects."
)

print(
    "The machine-readable ANNOTATION_SCHEMA object is available for "
    "downstream persistence and handover validation."
)

print(
    "Blank records contain no supervision labels."
)

print(
    "Prepared text and structural identifiers remain unchanged."
)

print(
    "No LLM supervision or human adjudication has been performed."
)

print(
    "No external model or API calls have been made by this block."
)

print("=" * 72)
```

## Independent LLM Supervision Protocol

This block defines the protocol used by external language models to generate
independent supervision annotations.

The protocol is specified before any external model is called.

Its purpose is to ensure that different supervising models receive the same
construct definitions, context structure, output requirements and methodological
constraints.

### Independent Supervision

Each supervising model must annotate the prepared text independently.

A supervisor must not receive:

* another supervisor's annotations;
* consensus labels;
* human-review decisions;
* model predictions produced by the Media AI architecture;
* external fact-checking results unless a later experimental protocol explicitly introduces them.

This prevents annotations from being conditioned on labels or conclusions that
are intended to remain independent supervision signals.

### Supervision Unit and Request Unit

The primary annotation unit remains the **sentence**.

For external LLM supervision, however, the request unit is the **complete
article**.

Each article-level request contains the ordered sentence sequence together with
stable Media AI article, paragraph and sentence identifiers.

The supervisor must return exactly one annotation for every supplied sentence.

This distinction allows the complete emotional and factological trajectory of
an article to be supervised within one external request while preserving
sentence-level annotation, structural traceability and later model training.

### Temporal Context

For sentence (t), the annotation must depend only on:

* article title;
* sentences preceding sentence (t);
* the current sentence;
* paragraph and sentence position.

The complete article may be supplied within the same external request so that
all sentence-level annotations can be generated as one coherent article
trajectory.

However, later sentences must not influence the annotation assigned to an
earlier sentence.

The supervisor must therefore process the article sequentially and preserve
causal ordering when estimating reported emotion, transformative emotion and
immediate emotional state.

This is particularly important for transformative-emotion supervision because
the annotation represents the change introduced at the current textual step.

### Annotation Tasks

For each sentence, the supervisor is asked to estimate:

1. **Reported emotion**

   * eleven non-negative emotional dimensions in `[0,1]`.

2. **Transformative emotion**

   * eleven signed emotional changes in `[-1,1]`;
   * negative values represent decreases;
   * zero represents no meaningful change;
   * positive values represent increases.

3. **Immediate emotional state**

   * eleven non-negative emotional dimensions in `[0,1]`.

4. **Emotional weight**

   * one value in `[0,1]`.

5. **Factological weight**

   * one value in `[0,1]`.

6. **Factological components**

   * eight boolean indicators.

7. **Annotation confidence**

   * one value in `[0,1]`.

The supervisor may also return `insufficient_evidence` or `not_applicable`
where permitted by the schema.

Numeric zero must not be used as a substitute for missing evidence.

### Factological Boundary

Factological weight is not a truth score.

The supervisor must assess how much externally assessable informational content
the sentence contributes, not whether the information is actually correct.

External fact checking, web search and retrieval are outside the supervision
task unless a later experimental protocol explicitly introduces them.

### Emotional Boundary

Reported emotion, transformative emotion and immediate emotional state must be
evaluated separately.

The supervisor must not infer transformative emotion merely from emotional
keywords.

A sentence may reduce an emotional dimension even when it contains emotionally
charged language, and it may change emotional state without containing explicit
emotion words.

### Evidence

The supervisor should provide concise evidence explaining the principal basis
for each sentence-level annotation.

Evidence must be grounded only in:

* the current sentence;
* supplied preceding article context.

Later sentences in the article must not be used as evidence for an earlier
annotation.

Evidence should not introduce unsupported external information.

The evidence field is retained for review and disagreement analysis but is not
itself a numerical supervision target.

### Structured Output

Supervisors must return machine-readable JSON conforming to the Notebook 03
annotation schema.

Each article-level response must contain exactly one sentence-level annotation
for every supplied sentence identifier.

The returned sentence identifiers must preserve the supplied article order and
must remain unchanged.

Free-form prose outside the structured response is not part of the annotation
contract.

Provider responses are later parsed, checked for complete article coverage,
validated structurally and numerically, and converted into standardised Media AI
supervision records before they are accepted.

### Failure Policy

A provider response is not accepted merely because a request succeeded.

Responses must pass structural and numerical validation.

For article-level supervision, a response is not considered complete unless all
expected sentence identifiers are returned exactly once.

Missing, duplicated or unexpected sentence annotations cause the article-level
response to fail completeness validation.

Invalid responses are retained diagnostically and may later be:

* retried;
* routed to another supervisor;
* marked for human review.

They must not silently enter the reference dataset.

Article-level acceptance is atomic: a partially valid article response must not
silently create a partially supervised article trajectory.

### Execution Control

Actual LLM supervision is controlled exclusively by the Block 1 switch:

`RUN_LLM_SUPERVISION`

This block defines the supervision protocol regardless of the switch value.

When the switch is `False`, no external model is contacted.

Request-volume limits are also defined centrally in Block 1 and are interpreted
as **article-level external supervision requests** once Block 6 is migrated to
the article-request architecture.

### Migration Compatibility

For auditability and migration safety, the existing sentence-level supervision
payload collection remains available as:

`LLM_SUPERVISION_PAYLOADS`

The new article-level external supervision payload collection is created as:

`LLM_ARTICLE_SUPERVISION_PAYLOADS`

The sentence-level collection preserves the existing annotation-unit structure,
while the article-level collection becomes the external request unit.

This allows pilot supervision already generated under the sentence-request
architecture to remain interpretable and traceable.

### Output

This block creates:

* a versioned LLM supervision protocol;
* canonical supervisor instructions;
* machine-readable output specifications;
* sentence-level supervision payloads for traceability and compatibility;
* article-level external supervision payloads containing ordered sentence units;
* article-to-sentence request mappings;
* protocol-validation checks.

No external model or API call is performed by this block.


```python
# =============================================================================
# Media AI — Independent LLM Supervision Protocol
# =============================================================================

import json
from copy import deepcopy


# -----------------------------------------------------------------------------
# Protocol configuration
# -----------------------------------------------------------------------------

LLM_SUPERVISION_PROTOCOL_VERSION = "1.1"

LLM_SUPERVISION_ANNOTATION_UNIT = "sentence"
LLM_SUPERVISION_REQUEST_UNIT = "article"

ALLOW_FUTURE_CONTEXT_FOR_ANNOTATION = False
ALLOW_EXTERNAL_FACT_CHECKING = False
ALLOW_OTHER_SUPERVISOR_LABELS = False
ALLOW_MEDIA_AI_MODEL_PREDICTIONS = False

REQUIRE_STRUCTURED_OUTPUT = True
REQUIRE_EVIDENCE = True
REQUIRE_CONFIDENCE = True


# -----------------------------------------------------------------------------
# Canonical supervision instructions
# -----------------------------------------------------------------------------

LLM_SUPERVISION_INSTRUCTIONS = """
You are acting as an independent research supervisor for the Media AI project.

You will receive one complete article represented as an ordered sequence of
sentence units.

The annotation unit is the sentence.

The external supervision request unit is the complete article.

Process the article sequentially.

For each sentence, use only:

1. the article title;
2. sentences appearing before the current sentence;
3. the current sentence;
4. the supplied paragraph and sentence position.

Later sentences may be present in the article-level request, but they must not
influence the annotation assigned to an earlier sentence.

Do not assume access to annotations produced by another supervisor.

Do not perform external fact checking.

Do not use Media AI model predictions.

For every supplied sentence, return exactly one structured annotation and
preserve the supplied sentence identifier unchanged.

The annotation constructs are:

1. Reported emotion

   Estimate the emotion represented as belonging to people, groups, or subjects
   described by the text.

   Dimensions:
   fear, anger, sadness, disgust, tension, uncertainty, hope, relief, trust,
   compassion, solidarity.

   Each observed value must be between 0 and 1.

2. Transformative emotion

   Estimate how the current sentence changes each emotional dimension relative
   to the preceding article context.

   Each observed value must be between -1 and 1.

   Negative values mean the emotion decreases.
   Zero means no meaningful change.
   Positive values mean the emotion increases.

   This is a change variable, not an emotional-state amount.

3. Immediate emotional state

   Estimate the amount of each emotional dimension present after interpreting
   the current sentence in the supplied preceding context.

   Each observed value must be between 0 and 1.

4. Emotional weight

   Estimate the strength of emotionally consequential information in the
   current sentence.

   The observed value must be between 0 and 1.

5. Factological weight

   Estimate how much externally assessable informational content the current
   sentence contributes.

   The observed value must be between 0 and 1.

   This is NOT a truth probability.

   A claim can have high factological weight even when its truth has not been
   verified.

6. Factological components

   Indicate whether each component is present:

   explicit_claim
   observation_or_event
   quantitative_information
   attribution
   evidence_reference
   certainty_statement
   uncertainty_statement
   speculation_or_prediction

7. Annotation confidence

   Estimate your confidence that the annotation is supported by the supplied
   text and operational definitions.

   The observed value must be between 0 and 1.

For numeric constructs, distinguish an observed value of zero from missing
information.

Use one of these value states:

observed
insufficient_evidence
not_applicable

If value_state is observed, provide a numeric value within the required range.

If value_state is insufficient_evidence or not_applicable, return null for the
numeric value.

Provide concise evidence grounded only in the current sentence and preceding
article context available at that sentence.

Do not use later sentences as evidence for earlier annotations.

Do not introduce unsupported external information.

Return only structured JSON matching the requested output format.

Return annotations for every supplied sentence exactly once and in the supplied
article order.
""".strip()


# -----------------------------------------------------------------------------
# Machine-readable sentence-level output specification
# -----------------------------------------------------------------------------

LLM_OUTPUT_SPECIFICATION = {
    "reported_emotion": {
        dimension: {
            "value": "number_or_null",
            "value_state": "annotation_value_state",
        }
        for dimension
        in EMOTIONAL_DIMENSIONS
    },

    "transformative_emotion": {
        dimension: {
            "value": "number_or_null",
            "value_state": "annotation_value_state",
        }
        for dimension
        in EMOTIONAL_DIMENSIONS
    },

    "immediate_emotional_state": {
        dimension: {
            "value": "number_or_null",
            "value_state": "annotation_value_state",
        }
        for dimension
        in EMOTIONAL_DIMENSIONS
    },

    "emotional_weight": {
        "value": "number_or_null",
        "value_state": "annotation_value_state",
    },

    "factological_weight": {
        "value": "number_or_null",
        "value_state": "annotation_value_state",
    },

    "factological_components": {
        component: "boolean_or_null"
        for component
        in FACTOLOGICAL_COMPONENTS
    },

    "annotation_confidence": {
        "value": "number_or_null",
        "value_state": "annotation_value_state",
    },

    "evidence": {
        "summary": "string_or_null",
        "supporting_text": [
            "string"
        ],
    },
}


# -----------------------------------------------------------------------------
# Machine-readable article-level output specification
# -----------------------------------------------------------------------------

LLM_ARTICLE_OUTPUT_SPECIFICATION = {
    "media_ai_record_id":
        "string",

    "sentence_annotations": [
        {
            "sentence_id":
                "string",

            "annotation":
                deepcopy(
                    LLM_OUTPUT_SPECIFICATION
                ),
        }
    ],
}


# -----------------------------------------------------------------------------
# Construct one sentence-level supervision payload
# -----------------------------------------------------------------------------

def create_llm_supervision_payload(
    context_record,
):
    """
    Construct one provider-independent sentence-level supervision payload.

    Sentence-level payloads are retained for traceability, inspection and
    compatibility with pilot supervision generated before article-level
    request execution is introduced.
    """

    if not isinstance(
        context_record,
        dict,
    ):
        raise TypeError(
            "context_record must be a dictionary."
        )

    payload = {
        "payload_type":
            "media_ai_sentence_supervision_payload",

        "protocol": {
            "protocol_name":
                "media_ai_independent_llm_supervision",

            "protocol_version":
                LLM_SUPERVISION_PROTOCOL_VERSION,

            "supervision_schema_version":
                SUPERVISION_SCHEMA_VERSION,

            "annotation_policy_version":
                ANNOTATION_POLICY_VERSION,

            "annotation_unit":
                LLM_SUPERVISION_ANNOTATION_UNIT,

            "request_unit":
                LLM_SUPERVISION_REQUEST_UNIT,
        },

        "identity": {
            "media_ai_record_id":
                context_record.get(
                    "media_ai_record_id"
                ),

            "paragraph_id":
                context_record.get(
                    "paragraph_id"
                ),

            "sentence_id":
                context_record.get(
                    "sentence_id"
                ),

            "paragraph_index":
                context_record.get(
                    "paragraph_index"
                ),

            "sentence_index":
                context_record.get(
                    "sentence_index"
                ),
        },

        "context": {
            "article_title":
                context_record.get(
                    "article_title"
                ),

            "preceding_context":
                context_record.get(
                    "preceding_context"
                ),

            "current_sentence":
                context_record.get(
                    "current_sentence"
                ),

            "context_scope":
                context_record.get(
                    "context_scope"
                ),
        },

        "instructions":
            LLM_SUPERVISION_INSTRUCTIONS,

        "output_specification":
            deepcopy(
                LLM_OUTPUT_SPECIFICATION
            ),
    }

    return payload


# -----------------------------------------------------------------------------
# Construct all sentence-level supervision payloads
# -----------------------------------------------------------------------------

LLM_SUPERVISION_PAYLOADS = [
    create_llm_supervision_payload(
        context_record
    )
    for context_record
    in SUPERVISION_CONTEXT_RECORDS
]


# -----------------------------------------------------------------------------
# Validate one sentence-level supervision payload
# -----------------------------------------------------------------------------

def validate_llm_supervision_payload(
    payload,
):
    """
    Validate one provider-independent sentence-level supervision payload.
    """

    errors = []

    if not isinstance(
        payload,
        dict,
    ):
        return (
            False,
            ["payload_not_dictionary"],
        )

    protocol = payload.get(
        "protocol",
        {}
    )

    if (
        protocol.get(
            "protocol_version"
        )
        != LLM_SUPERVISION_PROTOCOL_VERSION
    ):
        errors.append(
            "invalid_protocol_version"
        )

    if (
        protocol.get(
            "annotation_unit"
        )
        != LLM_SUPERVISION_ANNOTATION_UNIT
    ):
        errors.append(
            "invalid_annotation_unit"
        )

    if (
        protocol.get(
            "request_unit"
        )
        != LLM_SUPERVISION_REQUEST_UNIT
    ):
        errors.append(
            "invalid_request_unit"
        )

    identity = payload.get(
        "identity",
        {}
    )

    for field in [
        "media_ai_record_id",
        "paragraph_id",
        "sentence_id",
    ]:

        if not identity.get(
            field
        ):
            errors.append(
                f"missing_identity_{field}"
            )

    context = payload.get(
        "context",
        {}
    )

    if not isinstance(
        context.get(
            "current_sentence"
        ),
        str,
    ):
        errors.append(
            "missing_current_sentence"
        )

    if (
        context.get(
            "context_scope"
        )
        != ANNOTATION_CONTEXT_POLICY[
            "context_mode"
        ]
    ):
        errors.append(
            "invalid_context_scope"
        )

    if not payload.get(
        "instructions"
    ):
        errors.append(
            "missing_supervision_instructions"
        )

    output_specification = payload.get(
        "output_specification",
        {}
    )

    required_output_fields = {
        "reported_emotion",
        "transformative_emotion",
        "immediate_emotional_state",
        "emotional_weight",
        "factological_weight",
        "factological_components",
        "annotation_confidence",
        "evidence",
    }

    if set(
        output_specification.keys()
    ) != required_output_fields:
        errors.append(
            "invalid_output_specification"
        )

    return (
        len(errors) == 0,
        errors,
    )


# -----------------------------------------------------------------------------
# Validate all sentence-level supervision payloads
# -----------------------------------------------------------------------------

LLM_PAYLOAD_VALIDATION_RESULTS = []

for payload in LLM_SUPERVISION_PAYLOADS:

    passed, errors = (
        validate_llm_supervision_payload(
            payload
        )
    )

    LLM_PAYLOAD_VALIDATION_RESULTS.append({
        "sentence_id":
            payload[
                "identity"
            ][
                "sentence_id"
            ],

        "passed":
            passed,

        "errors":
            errors,
    })


INVALID_LLM_PAYLOADS = [
    result
    for result
    in LLM_PAYLOAD_VALIDATION_RESULTS
    if not result[
        "passed"
    ]
]


# -----------------------------------------------------------------------------
# Group sentence payloads by article
# -----------------------------------------------------------------------------

ARTICLE_SUPERVISION_GROUPS = {}

for payload in LLM_SUPERVISION_PAYLOADS:

    identity = payload[
        "identity"
    ]

    context = payload[
        "context"
    ]

    media_ai_record_id = (
        identity[
            "media_ai_record_id"
        ]
    )

    if (
        media_ai_record_id
        not in ARTICLE_SUPERVISION_GROUPS
    ):

        ARTICLE_SUPERVISION_GROUPS[
            media_ai_record_id
        ] = {
            "media_ai_record_id":
                media_ai_record_id,

            "article_title":
                context.get(
                    "article_title"
                ),

            "sentence_units":
                [],
        }

    ARTICLE_SUPERVISION_GROUPS[
        media_ai_record_id
    ][
        "sentence_units"
    ].append(
        {
            "media_ai_record_id":
                media_ai_record_id,

            "paragraph_id":
                identity.get(
                    "paragraph_id"
                ),

            "sentence_id":
                identity.get(
                    "sentence_id"
                ),

            "paragraph_index":
                identity.get(
                    "paragraph_index"
                ),

            "sentence_index":
                identity.get(
                    "sentence_index"
                ),

            "preceding_context":
                context.get(
                    "preceding_context"
                ),

            "current_sentence":
                context.get(
                    "current_sentence"
                ),

            "context_scope":
                context.get(
                    "context_scope"
                ),
        }
    )


# -----------------------------------------------------------------------------
# Preserve stable sentence order within each article
# -----------------------------------------------------------------------------

for article_group in (
    ARTICLE_SUPERVISION_GROUPS.values()
):

    article_group[
        "sentence_units"
    ] = sorted(
        article_group[
            "sentence_units"
        ],
        key=lambda unit: (
            unit.get(
                "paragraph_index",
                0,
            ),
            unit.get(
                "sentence_index",
                0,
            ),
        ),
    )


# -----------------------------------------------------------------------------
# Construct one article-level external supervision payload
# -----------------------------------------------------------------------------

def create_article_supervision_payload(
    article_group,
):
    """
    Construct one provider-independent article-level supervision payload.
    """

    if not isinstance(
        article_group,
        dict,
    ):
        raise TypeError(
            "article_group must be a dictionary."
        )

    sentence_units = deepcopy(
        article_group.get(
            "sentence_units",
            []
        )
    )

    expected_sentence_ids = [
        sentence_unit.get(
            "sentence_id"
        )
        for sentence_unit
        in sentence_units
    ]

    return {
        "payload_type":
            "media_ai_article_supervision_payload",

        "protocol": {
            "protocol_name":
                "media_ai_independent_llm_supervision",

            "protocol_version":
                LLM_SUPERVISION_PROTOCOL_VERSION,

            "supervision_schema_version":
                SUPERVISION_SCHEMA_VERSION,

            "annotation_policy_version":
                ANNOTATION_POLICY_VERSION,

            "annotation_unit":
                LLM_SUPERVISION_ANNOTATION_UNIT,

            "request_unit":
                LLM_SUPERVISION_REQUEST_UNIT,
        },

        "identity": {
            "media_ai_record_id":
                article_group.get(
                    "media_ai_record_id"
                ),
        },

        "article": {
            "article_title":
                article_group.get(
                    "article_title"
                ),

            "sentence_count":
                len(
                    sentence_units
                ),

            "expected_sentence_ids":
                expected_sentence_ids,

            "sentence_units":
                sentence_units,
        },

        "instructions":
            LLM_SUPERVISION_INSTRUCTIONS,

        "output_specification":
            deepcopy(
                LLM_ARTICLE_OUTPUT_SPECIFICATION
            ),
    }


# -----------------------------------------------------------------------------
# Construct all article-level external supervision payloads
# -----------------------------------------------------------------------------

LLM_ARTICLE_SUPERVISION_PAYLOADS = [
    create_article_supervision_payload(
        article_group
    )
    for article_group
    in ARTICLE_SUPERVISION_GROUPS.values()
]


# -----------------------------------------------------------------------------
# Validate one article-level supervision payload
# -----------------------------------------------------------------------------

def validate_article_supervision_payload(
    payload,
):
    """
    Validate one provider-independent article-level supervision payload.
    """

    errors = []

    if not isinstance(
        payload,
        dict,
    ):
        return (
            False,
            ["payload_not_dictionary"],
        )

    protocol = payload.get(
        "protocol",
        {}
    )

    if (
        protocol.get(
            "protocol_version"
        )
        != LLM_SUPERVISION_PROTOCOL_VERSION
    ):
        errors.append(
            "invalid_protocol_version"
        )

    if (
        protocol.get(
            "annotation_unit"
        )
        != LLM_SUPERVISION_ANNOTATION_UNIT
    ):
        errors.append(
            "invalid_annotation_unit"
        )

    if (
        protocol.get(
            "request_unit"
        )
        != LLM_SUPERVISION_REQUEST_UNIT
    ):
        errors.append(
            "invalid_request_unit"
        )

    identity = payload.get(
        "identity",
        {}
    )

    media_ai_record_id = (
        identity.get(
            "media_ai_record_id"
        )
    )

    if not media_ai_record_id:
        errors.append(
            "missing_media_ai_record_id"
        )

    article = payload.get(
        "article",
        {}
    )

    sentence_units = article.get(
        "sentence_units",
        []
    )

    expected_sentence_ids = article.get(
        "expected_sentence_ids",
        []
    )

    if not isinstance(
        sentence_units,
        list,
    ) or not sentence_units:
        errors.append(
            "article_has_no_sentence_units"
        )

    actual_sentence_ids = [
        sentence_unit.get(
            "sentence_id"
        )
        for sentence_unit
        in sentence_units
    ]

    if any(
        sentence_id is None
        for sentence_id
        in actual_sentence_ids
    ):
        errors.append(
            "missing_sentence_id"
        )

    if len(
        actual_sentence_ids
    ) != len(
        set(
            actual_sentence_ids
        )
    ):
        errors.append(
            "duplicate_sentence_ids"
        )

    if (
        actual_sentence_ids
        != expected_sentence_ids
    ):
        errors.append(
            "sentence_identity_or_order_mismatch"
        )

    if (
        article.get(
            "sentence_count"
        )
        != len(
            sentence_units
        )
    ):
        errors.append(
            "sentence_count_mismatch"
        )

    for sentence_unit in sentence_units:

        if (
            sentence_unit.get(
                "media_ai_record_id"
            )
            != media_ai_record_id
        ):
            errors.append(
                "cross_article_sentence_identity"
            )

            break

        if not isinstance(
            sentence_unit.get(
                "current_sentence"
            ),
            str,
        ):
            errors.append(
                "missing_sentence_text"
            )

            break

        if (
            sentence_unit.get(
                "context_scope"
            )
            != ANNOTATION_CONTEXT_POLICY[
                "context_mode"
            ]
        ):
            errors.append(
                "invalid_sentence_context_scope"
            )

            break

    if not payload.get(
        "instructions"
    ):
        errors.append(
            "missing_supervision_instructions"
        )

    output_specification = payload.get(
        "output_specification",
        {}
    )

    if set(
        output_specification.keys()
    ) != {
        "media_ai_record_id",
        "sentence_annotations",
    }:
        errors.append(
            "invalid_article_output_specification"
        )

    return (
        len(errors) == 0,
        errors,
    )


# -----------------------------------------------------------------------------
# Validate all article-level supervision payloads
# -----------------------------------------------------------------------------

ARTICLE_PAYLOAD_VALIDATION_RESULTS = []

for payload in (
    LLM_ARTICLE_SUPERVISION_PAYLOADS
):

    passed, errors = (
        validate_article_supervision_payload(
            payload
        )
    )

    ARTICLE_PAYLOAD_VALIDATION_RESULTS.append({
        "media_ai_record_id":
            payload[
                "identity"
            ][
                "media_ai_record_id"
            ],

        "sentence_count":
            payload[
                "article"
            ][
                "sentence_count"
            ],

        "passed":
            passed,

        "errors":
            errors,
    })


INVALID_ARTICLE_SUPERVISION_PAYLOADS = [
    result
    for result
    in ARTICLE_PAYLOAD_VALIDATION_RESULTS
    if not result[
        "passed"
    ]
]


# -----------------------------------------------------------------------------
# Cross-payload accounting
# -----------------------------------------------------------------------------

SENTENCE_SUPERVISION_PAYLOAD_COUNT = (
    len(
        LLM_SUPERVISION_PAYLOADS
    )
)

ARTICLE_SUPERVISION_PAYLOAD_COUNT = (
    len(
        LLM_ARTICLE_SUPERVISION_PAYLOADS
    )
)

ARTICLE_SUPERVISION_SENTENCE_COUNT = sum(
    payload[
        "article"
    ][
        "sentence_count"
    ]
    for payload
    in LLM_ARTICLE_SUPERVISION_PAYLOADS
)

ARTICLE_SENTENCE_ACCOUNTING_VALID = (
    SENTENCE_SUPERVISION_PAYLOAD_COUNT
    == ARTICLE_SUPERVISION_SENTENCE_COUNT
)


# -----------------------------------------------------------------------------
# Protocol safeguards
# -----------------------------------------------------------------------------

if ALLOW_FUTURE_CONTEXT_FOR_ANNOTATION:
    raise ValueError(
        "Future article context is not permitted "
        "when assigning supervision to an earlier sentence."
    )


if ALLOW_EXTERNAL_FACT_CHECKING:
    raise ValueError(
        "External fact checking is outside the "
        "current supervision protocol."
    )


if ALLOW_OTHER_SUPERVISOR_LABELS:
    raise ValueError(
        "Independent supervisors must not receive "
        "another supervisor's labels."
    )


if ALLOW_MEDIA_AI_MODEL_PREDICTIONS:
    raise ValueError(
        "Supervisors must not receive Media AI model predictions."
    )


if INVALID_LLM_PAYLOADS:
    raise ValueError(
        "One or more sentence-level LLM supervision payloads "
        "failed validation."
    )


if INVALID_ARTICLE_SUPERVISION_PAYLOADS:
    raise ValueError(
        "One or more article-level LLM supervision payloads "
        "failed validation."
    )


if not ARTICLE_SENTENCE_ACCOUNTING_VALID:
    raise ValueError(
        "Sentence-level and article-level supervision payload "
        "counts do not reconcile."
    )


# -----------------------------------------------------------------------------
# Execution state
# -----------------------------------------------------------------------------

LLM_SUPERVISION_EXECUTED = False

if RUN_LLM_SUPERVISION:

    # Provider/API execution is intentionally implemented in Block 6.
    #
    # Block 5 defines and validates the supervision protocol only.
    #
    # Keeping provider execution separate ensures that protocol construction
    # itself cannot accidentally trigger external model usage.

    print(
        "LLM supervision is enabled in the Execution Control Panel."
    )

    print(
        "No provider call is implemented in Block 5; "
        "execution is deferred to Block 6."
    )


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Independent LLM Supervision Protocol")
print("=" * 72)

print(
    f"Protocol version             : "
    f"{LLM_SUPERVISION_PROTOCOL_VERSION}"
)

print(
    f"Annotation unit              : "
    f"{LLM_SUPERVISION_ANNOTATION_UNIT}"
)

print(
    f"External request unit        : "
    f"{LLM_SUPERVISION_REQUEST_UNIT}"
)

print(
    f"Supervision enabled          : "
    f"{RUN_LLM_SUPERVISION}"
)

print("-" * 72)

print(
    f"Sentence supervision payloads: "
    f"{SENTENCE_SUPERVISION_PAYLOAD_COUNT}"
)

print(
    f"Valid sentence payloads      : "
    f"{SENTENCE_SUPERVISION_PAYLOAD_COUNT - len(INVALID_LLM_PAYLOADS)}"
)

print(
    f"Invalid sentence payloads    : "
    f"{len(INVALID_LLM_PAYLOADS)}"
)

print(
    f"Article supervision payloads : "
    f"{ARTICLE_SUPERVISION_PAYLOAD_COUNT}"
)

print(
    f"Valid article payloads       : "
    f"{ARTICLE_SUPERVISION_PAYLOAD_COUNT - len(INVALID_ARTICLE_SUPERVISION_PAYLOADS)}"
)

print(
    f"Invalid article payloads     : "
    f"{len(INVALID_ARTICLE_SUPERVISION_PAYLOADS)}"
)

print(
    f"Sentences in article payloads: "
    f"{ARTICLE_SUPERVISION_SENTENCE_COUNT}"
)

print(
    f"Article/sentence accounting  : "
    f"{ARTICLE_SENTENCE_ACCOUNTING_VALID}"
)

print("-" * 72)

print(
    f"Future context for annotation: "
    f"{ALLOW_FUTURE_CONTEXT_FOR_ANNOTATION}"
)

print(
    f"External fact checking       : "
    f"{ALLOW_EXTERNAL_FACT_CHECKING}"
)

print(
    f"Other supervisor labels      : "
    f"{ALLOW_OTHER_SUPERVISOR_LABELS}"
)

print(
    f"Media AI predictions         : "
    f"{ALLOW_MEDIA_AI_MODEL_PREDICTIONS}"
)

print(
    f"Structured output required   : "
    f"{REQUIRE_STRUCTURED_OUTPUT}"
)

print(
    f"Evidence required            : "
    f"{REQUIRE_EVIDENCE}"
)

print(
    f"Confidence required          : "
    f"{REQUIRE_CONFIDENCE}"
)

print("-" * 72)

print(
    f"LLM supervision executed     : "
    f"{LLM_SUPERVISION_EXECUTED}"
)

print("=" * 72)

print(
    "Independent LLM supervision protocol initialised successfully."
)

print(
    "Sentence-level annotation units are preserved for traceability "
    "and compatibility."
)

print(
    "Complete articles are prepared as the external LLM request unit."
)

print(
    "Later sentences must not influence supervision assigned to "
    "earlier sentences."
)

print(
    "No external model or API calls have been made by this block."
)
```

## Controlled LLM Supervisor Registry and Execution

This block implements the controlled execution layer for independent LLM
supervision.

The supervision protocol itself was defined in Block 5. Block 6 does not change
the annotation ontology, temporal context policy or output contract.

Its purpose is to:

* define a stable registry of independent supervisors;
* configure one initial supervisor for the pilot;
* preserve placeholders for additional supervisors;
* enforce the Block 1 execution switches and usage safeguards;
* validate credential and model readiness before external calls;
* execute article-level supervision independently for each enabled supervisor;
* retain raw provider responses separately by supervisor;
* prevent one supervisor from observing another supervisor's output;
* preserve article-level resume protection so successfully supervised articles
  are not submitted again unintentionally.

### Supervisor Registry

The notebook supports five supervisor slots:

* `supervisor_01`;
* `supervisor_02`;
* `supervisor_03`;
* `supervisor_04`;
* `supervisor_05`.

The slot identifiers are stable research identifiers and are deliberately
separate from provider or model names.

This allows a provider or model to be replaced later without changing the
supervisor identity used in the research pipeline.

Only configured and enabled supervisors may execute.

### Execution Hierarchy

External LLM execution requires two levels of permission:

1. `RUN_LLM_SUPERVISION = True`
2. the corresponding `ENABLE_SUPERVISOR_XX = True`

If the master switch is `False`, no supervisor may make an external request even
when its individual switch is enabled.

Execution is additionally constrained by the cost and usage safeguards defined
centrally in Block 1.

### Provider Configuration

Supervisor 01 is configured as the initial pilot supervisor.

The provider and model remain explicit metadata.

API credentials are read from the runtime environment rather than written into
the notebook.

The notebook therefore does not commit API keys or private credentials to the
repository.

Provider, model and supervision metadata remain explicit in the raw result so
that every external supervision request remains traceable to the supervisor and
model that produced it.

### Annotation Unit and Execution Unit

The annotation unit remains the **sentence**.

The external supervision request unit is now the **complete article**.

Block 6 therefore consumes:

`LLM_ARTICLE_SUPERVISION_PAYLOADS`

created and validated in Block 5.

One external provider request corresponds to one complete article.

The provider must return one sentence-level annotation for every sentence in
that article while preserving the supplied sentence identifiers and article
order.

This architecture allows the complete emotional and factological trajectory of
an article to be generated coherently within one request while preserving the
sentence-level targets required by the Media AI research design.

### Temporal Execution Boundary

Although the complete article is supplied within one request, annotations must
respect the temporal context policy defined in Block 5.

For sentence (t), later sentences must not influence the annotation assigned to
that sentence.

The article is therefore treated as an ordered supervision sequence rather than
as an unrestricted full-text classification task.

Block 6 does not redefine or weaken this constraint.

### Independent Response Storage

Responses are stored separately by supervisor.

The structure is conceptually:

`LLM_SUPERVISION_RAW_RESULTS`

→ `supervisor_01`

→ article-level response records

→ sentence-level annotations returned within each article response

This separation is maintained before comparison, consensus or adjudication.

Later blocks may compare supervisors, but Block 6 does not merge their labels.

### Controlled Pilot Execution

The prepared supervision corpus contains multiple article-level external
supervision payloads.

The request-volume controls defined in Block 1 are interpreted as limits on
**external article requests**, not individual sentence annotations.

For example:

`MAX_SUPERVISION_REQUESTS_PER_RUN = 1`

permits at most one complete article to be submitted during that notebook run.

The number of sentence-level annotations returned by that article request does
not increase the notebook-side request count.

The execution limit controls external request volume only. It does not change
the prepared corpus, annotation schema or number of sentence-level supervision
targets.

### Free-Tier and Cost Safeguards

Block 6 must respect the supervision cost policy defined in Block 1.

External execution is permitted only when the configured supervisor satisfies
the applicable readiness and cost-policy checks.

Notebook-side request ceilings are independent research safeguards and do not
replace provider-side rate, quota or Free Tier restrictions.

Where both apply, execution must remain within the more restrictive applicable
limit.

Automatic migration to paid API usage or a paid fallback is not permitted when
disabled by the Block 1 configuration.

### Article-Level Resume Protection

Successful supervision is tracked at the external request unit.

Because one request now corresponds to one complete article, resume protection
uses stable article identifiers rather than successful sentence identifiers.

Before selecting a new article for supervision, Block 6 checks whether that
article has already been successfully supervised by the same supervisor.

Previously successful article-supervisor combinations are excluded from new
requests.

This prevents accidental duplicate external calls while preserving independent
supervision across different supervisors.

A successful article for `supervisor_01` therefore remains independently
eligible for another configured supervisor if that supervisor has not yet
processed it.

### Raw Response Principle

A successful provider request does not automatically create an accepted Media AI
annotation.

Block 6 stores provider output as a raw article-level supervision result.

The returned sentence-level annotations are not accepted into the canonical
research dataset at this stage.

Article completeness checks, schema parsing and canonical annotation validation
occur in Block 7.

This preserves the distinction:

**provider response → article completeness validation → sentence-level schema validation → accepted supervision records**

### Failure Handling

Request failures, malformed provider responses and unavailable credentials are
recorded explicitly.

Failures are not silently converted into annotations.

A failed article remains available for retry or later review.

A provider request that succeeds technically but returns an incomplete,
duplicated or otherwise invalid set of sentence annotations is not considered a
canonical supervision success merely because the API request completed.

Acceptance remains the responsibility of Block 7.

### Request Accounting

Request accounting is maintained at article-request level.

Diagnostics distinguish, where applicable:

* requests selected;
* requests attempted;
* requests completed;
* requests successful;
* requests failed;
* requests blocked;
* previously successful articles;
* eligible unsupervised articles;
* remaining notebook-side daily request capacity.

Sentence counts may additionally be reported for diagnostic purposes, but they
must not be interpreted as external API request counts.

### Output

This block creates:

* `SUPERVISOR_REGISTRY`;
* enabled-supervisor metadata;
* controlled article-level provider execution functions;
* article-level resume-protection state;
* `LLM_SUPERVISION_RAW_RESULTS`;
* provider and model provenance;
* request-level execution diagnostics.

Actual external calls occur only when enabled by Block 1.

Raw provider outputs remain separated by supervisor identity.

No canonical annotation is accepted by this block.

No supervisor comparison, consensus construction or human adjudication is
performed by this block.


```python
# =============================================================================
# Media AI — Controlled LLM Supervisor Registry and Execution
# Block 6 — Restart-Safe Article-Level Supervision
# =============================================================================

import io
import json
from copy import deepcopy
from datetime import datetime, timezone

from googleapiclient.http import MediaIoBaseUpload


# -----------------------------------------------------------------------------
# Controlled-execution configuration
# -----------------------------------------------------------------------------

SUPERVISOR_REGISTRY_VERSION = "1.5"

RAW_SUPERVISION_STORE_VERSION = "1.0"

RAW_SUPERVISION_STORE_FILENAME = (
    "notebook_03_raw_supervision_results.json"
)

LLM_USAGE_LEDGER_VERSION = "2.0"

PROVIDER_SCHEMA_VERSION = "1.1"

PROVIDER_SCHEMA_STRATEGY = (
    "shallow_article_envelope_with_annotation_json"
)

LLM_USAGE_LEDGER_FILENAME = (
    "notebook_03_llm_usage_ledger.json"
)


# -----------------------------------------------------------------------------
# Block 1 control availability check
# -----------------------------------------------------------------------------

required_execution_controls = [
    "RUN_LLM_SUPERVISION",

    "ENABLE_SUPERVISOR_01",
    "ENABLE_SUPERVISOR_02",
    "ENABLE_SUPERVISOR_03",
    "ENABLE_SUPERVISOR_04",
    "ENABLE_SUPERVISOR_05",

    "SUPERVISOR_01_PROVIDER",
    "SUPERVISOR_01_MODEL",

    "SUPERVISOR_02_PROVIDER",
    "SUPERVISOR_02_MODEL",

    "SUPERVISOR_03_PROVIDER",
    "SUPERVISOR_03_MODEL",

    "SUPERVISOR_04_PROVIDER",
    "SUPERVISOR_04_MODEL",

    "SUPERVISOR_05_PROVIDER",
    "SUPERVISOR_05_MODEL",

    "REQUIRE_FREE_SUPERVISION",
    "ALLOW_PAID_API_USAGE",
    "ALLOW_AUTOMATIC_PAID_FALLBACK",

    "MAX_SUPERVISION_REQUESTS_PER_RUN",
    "MAX_SUPERVISION_REQUESTS_PER_DAY",
]


missing_execution_controls = [
    control
    for control in required_execution_controls
    if control not in globals()
]


if missing_execution_controls:
    raise NameError(
        "Block 1 execution controls are not initialised. "
        "Rerun Notebook 03 Block 1. Missing: "
        f"{missing_execution_controls}"
    )


# -----------------------------------------------------------------------------
# Block 2 storage availability check
# -----------------------------------------------------------------------------

required_storage_objects = [
    "drive_service",
    "LOGS_FOLDER_ID",
    "find_drive_file",
    "download_json_from_drive",
]


missing_storage_objects = [
    object_name
    for object_name in required_storage_objects
    if object_name not in globals()
]


if missing_storage_objects:
    raise NameError(
        "Required persistent-storage objects are not initialised. "
        "Rerun Notebook 03 Block 2. Missing: "
        f"{missing_storage_objects}"
    )


# -----------------------------------------------------------------------------
# Block 5 protocol availability check
# -----------------------------------------------------------------------------

required_protocol_objects = [
    "LLM_ARTICLE_SUPERVISION_PAYLOADS",
    "LLM_SUPERVISION_INSTRUCTIONS",
    "LLM_SUPERVISION_PROTOCOL_VERSION",
    "LLM_SUPERVISION_ANNOTATION_UNIT",
    "LLM_SUPERVISION_REQUEST_UNIT",
    "ANNOTATION_VALUE_STATES",
    "EMOTIONAL_DIMENSIONS",
    "FACTOLOGICAL_COMPONENTS",
    "SUPERVISION_SCHEMA_VERSION",
    "ANNOTATION_POLICY_VERSION",
]


missing_protocol_objects = [
    object_name
    for object_name in required_protocol_objects
    if object_name not in globals()
]


if missing_protocol_objects:
    raise NameError(
        "Required supervision protocol objects are not initialised. "
        "Rerun Notebook 03 Blocks 3–5. Missing: "
        f"{missing_protocol_objects}"
    )


if LLM_SUPERVISION_REQUEST_UNIT != "article":
    raise ValueError(
        "Block 6 requires article-level supervision payloads."
    )


if LLM_SUPERVISION_ANNOTATION_UNIT != "sentence":
    raise ValueError(
        "The Media AI annotation unit must remain the sentence."
    )


# -----------------------------------------------------------------------------
# Google GenAI SDK readiness
# -----------------------------------------------------------------------------

try:

    from google import genai

    GOOGLE_GENAI_AVAILABLE = True

except ImportError:

    genai = None

    GOOGLE_GENAI_AVAILABLE = False


# -----------------------------------------------------------------------------
# Gemini credential readiness
# -----------------------------------------------------------------------------

GEMINI_API_KEY = None

GEMINI_CREDENTIAL_READY = False

GEMINI_SECRET_STATUS = "NOT_LOADED"


try:

    from google.colab import userdata

    GEMINI_API_KEY = userdata.get(
        "GEMINI_API_KEY"
    )

    GEMINI_CREDENTIAL_READY = bool(
        isinstance(
            GEMINI_API_KEY,
            str,
        )
        and GEMINI_API_KEY.strip()
    )

    GEMINI_SECRET_STATUS = (
        "AVAILABLE"
        if GEMINI_CREDENTIAL_READY
        else "UNAVAILABLE"
    )


except Exception:

    GEMINI_API_KEY = None

    GEMINI_CREDENTIAL_READY = False

    GEMINI_SECRET_STATUS = "UNAVAILABLE"


# -----------------------------------------------------------------------------
# Research-approved Free Tier model allow-list
# -----------------------------------------------------------------------------

# Models must be manually verified against current provider documentation
# before being added to this list.
#
# The allow-list is an internal research safeguard.
# It does not independently inspect Google billing configuration.

FREE_TIER_APPROVED_SUPERVISOR_MODELS = {
    "gemini-3.6-flash",
}


# -----------------------------------------------------------------------------
# Stable five-slot supervisor registry
# -----------------------------------------------------------------------------

SUPERVISOR_REGISTRY = {
    "supervisor_01": {
        "enabled":
            ENABLE_SUPERVISOR_01,

        "provider":
            SUPERVISOR_01_PROVIDER,

        "model":
            SUPERVISOR_01_MODEL,

        "role":
            "independent_llm_supervisor",

        "implementation_status":
            "implemented",
    },

    "supervisor_02": {
        "enabled":
            ENABLE_SUPERVISOR_02,

        "provider":
            SUPERVISOR_02_PROVIDER,

        "model":
            SUPERVISOR_02_MODEL,

        "role":
            "independent_llm_supervisor",

        "implementation_status":
            "placeholder",
    },

    "supervisor_03": {
        "enabled":
            ENABLE_SUPERVISOR_03,

        "provider":
            SUPERVISOR_03_PROVIDER,

        "model":
            SUPERVISOR_03_MODEL,

        "role":
            "independent_llm_supervisor",

        "implementation_status":
            "placeholder",
    },

    "supervisor_04": {
        "enabled":
            ENABLE_SUPERVISOR_04,

        "provider":
            SUPERVISOR_04_PROVIDER,

        "model":
            SUPERVISOR_04_MODEL,

        "role":
            "independent_llm_supervisor",

        "implementation_status":
            "placeholder",
    },

    "supervisor_05": {
        "enabled":
            ENABLE_SUPERVISOR_05,

        "provider":
            SUPERVISOR_05_PROVIDER,

        "model":
            SUPERVISOR_05_MODEL,

        "role":
            "independent_llm_supervisor",

        "implementation_status":
            "placeholder",
    },
}


# -----------------------------------------------------------------------------
# Registry validation
# -----------------------------------------------------------------------------

EXPECTED_SUPERVISOR_IDS = {
    "supervisor_01",
    "supervisor_02",
    "supervisor_03",
    "supervisor_04",
    "supervisor_05",
}


if (
    set(
        SUPERVISOR_REGISTRY.keys()
    )
    != EXPECTED_SUPERVISOR_IDS
):
    raise ValueError(
        "Supervisor registry does not contain "
        "the expected five stable supervisor slots."
    )


# -----------------------------------------------------------------------------
# Free-tier model validation
# -----------------------------------------------------------------------------

def supervisor_free_tier_approved(
    supervisor_configuration,
):
    """
    Return True only when the configured model has been explicitly approved
    for free-tier supervision by the research configuration.
    """

    model = supervisor_configuration.get(
        "model"
    )

    return (
        model
        in FREE_TIER_APPROVED_SUPERVISOR_MODELS
    )


# -----------------------------------------------------------------------------
# Persistent article-level usage-ledger construction
# -----------------------------------------------------------------------------

def create_empty_llm_usage_ledger():
    """
    Create a new UTC-day article-level supervision usage ledger.

    The external request unit is the article.

    Sentence-level annotations remain the research annotation unit but do not
    independently increment article-request counters.
    """

    now = datetime.now(
        timezone.utc
    )

    return {
        "ledger_type":
            "media_ai_llm_supervision_usage",

        "ledger_version":
            LLM_USAGE_LEDGER_VERSION,

        "date_utc":
            now.date().isoformat(),

        "request_unit":
            "article",

        "article_requests_reserved":
            0,

        "article_requests_completed":
            0,

        "article_requests_successful":
            0,

        "article_requests_failed":
            0,

        "pending_article_ids_by_supervisor": {
            supervisor_id: []
            for supervisor_id
            in sorted(
                EXPECTED_SUPERVISOR_IDS
            )
        },

        "validated_article_ids_by_supervisor": {
            supervisor_id: []
            for supervisor_id
            in sorted(
                EXPECTED_SUPERVISOR_IDS
            )
        },

        "legacy_sentence_request_totals": {
            "requests_reserved":
                0,

            "requests_completed":
                0,

            "requests_successful":
                0,

            "requests_failed":
                0,
        },

        "legacy_successful_sentence_ids_by_supervisor": {
            supervisor_id: []
            for supervisor_id
            in sorted(
                EXPECTED_SUPERVISOR_IDS
            )
        },

        "last_updated_utc":
            now.isoformat(),
    }


# -----------------------------------------------------------------------------
# Usage-ledger schema migration
# -----------------------------------------------------------------------------

def normalise_llm_usage_ledger(
    ledger,
):
    """
    Upgrade an existing sentence-level usage ledger to the article-level
    request architecture without deleting historical request accounting.

    Previous sentence-level requests are retained as legacy audit information.

    They are not converted into completed article supervision because the
    earlier sentence requests did not represent atomic article-level
    supervision.
    """

    if not isinstance(
        ledger,
        dict,
    ):
        return create_empty_llm_usage_ledger()


    # -------------------------------------------------------------------------
    # Already migrated article-level ledger
    # -------------------------------------------------------------------------

    if (
        ledger.get(
            "request_unit"
        )
        == "article"
    ):

        ledger[
            "ledger_version"
        ] = LLM_USAGE_LEDGER_VERSION

        ledger.setdefault(
            "article_requests_reserved",
            0,
        )

        ledger.setdefault(
            "article_requests_completed",
            0,
        )

        ledger.setdefault(
            "article_requests_successful",
            0,
        )

        ledger.setdefault(
            "article_requests_failed",
            0,
        )

        ledger.setdefault(
            "legacy_sentence_request_totals",
            {
                "requests_reserved": 0,
                "requests_completed": 0,
                "requests_successful": 0,
                "requests_failed": 0,
            },
        )

        ledger.setdefault(
            "legacy_successful_sentence_ids_by_supervisor",
            {},
        )

        pending_ids = ledger.setdefault(
            "pending_article_ids_by_supervisor",
            {},
        )

        validated_ids = ledger.setdefault(
            "validated_article_ids_by_supervisor",
            {},
        )

        legacy_sentence_ids = ledger[
            "legacy_successful_sentence_ids_by_supervisor"
        ]

        for supervisor_id in sorted(
            EXPECTED_SUPERVISOR_IDS
        ):

            pending_ids.setdefault(
                supervisor_id,
                [],
            )

            validated_ids.setdefault(
                supervisor_id,
                [],
            )

            legacy_sentence_ids.setdefault(
                supervisor_id,
                [],
            )

            pending_ids[
                supervisor_id
            ] = list(
                dict.fromkeys(
                    pending_ids[
                        supervisor_id
                    ]
                )
            )

            validated_ids[
                supervisor_id
            ] = list(
                dict.fromkeys(
                    validated_ids[
                        supervisor_id
                    ]
                )
            )

            legacy_sentence_ids[
                supervisor_id
            ] = list(
                dict.fromkeys(
                    legacy_sentence_ids[
                        supervisor_id
                    ]
                )
            )

        return ledger


    # -------------------------------------------------------------------------
    # Migration from earlier sentence-request ledger
    # -------------------------------------------------------------------------

    migrated = create_empty_llm_usage_ledger()

    migrated[
        "date_utc"
    ] = ledger.get(
        "date_utc",
        migrated[
            "date_utc"
        ],
    )

    migrated[
        "legacy_sentence_request_totals"
    ] = {
        "requests_reserved":
            ledger.get(
                "requests_reserved",
                0,
            ),

        "requests_completed":
            ledger.get(
                "requests_completed",
                0,
            ),

        "requests_successful":
            ledger.get(
                "requests_successful",
                0,
            ),

        "requests_failed":
            ledger.get(
                "requests_failed",
                0,
            ),
    }

    old_successful_sentence_ids = (
        ledger.get(
            "successful_sentence_ids_by_supervisor",
            {},
        )
    )

    for supervisor_id in sorted(
        EXPECTED_SUPERVISOR_IDS
    ):

        migrated[
            "legacy_successful_sentence_ids_by_supervisor"
        ][
            supervisor_id
        ] = list(
            dict.fromkeys(
                old_successful_sentence_ids.get(
                    supervisor_id,
                    [],
                )
            )
        )

    migrated[
        "last_updated_utc"
    ] = datetime.now(
        timezone.utc
    ).isoformat()

    return migrated


# -----------------------------------------------------------------------------
# Persistent usage-ledger storage
# -----------------------------------------------------------------------------

def save_llm_usage_ledger(
    ledger,
):
    """
    Persist the Notebook 03 article-level LLM supervision usage ledger.
    """

    encoded = json.dumps(
        ledger,
        ensure_ascii=False,
        indent=2,
    ).encode(
        "utf-8"
    )

    media = MediaIoBaseUpload(
        io.BytesIO(
            encoded
        ),
        mimetype="application/json",
        resumable=False,
    )

    existing_file_id = find_drive_file(
        LLM_USAGE_LEDGER_FILENAME,
        LOGS_FOLDER_ID,
    )

    if existing_file_id:

        drive_service.files().update(
            fileId=existing_file_id,
            media_body=media,
            fields="id",
        ).execute()

        return existing_file_id

    metadata = {
        "name":
            LLM_USAGE_LEDGER_FILENAME,

        "parents":
            [
                LOGS_FOLDER_ID
            ],
    }

    created = (
        drive_service.files()
        .create(
            body=metadata,
            media_body=media,
            fields="id",
        )
        .execute()
    )

    return created[
        "id"
    ]


# -----------------------------------------------------------------------------
# Persistent usage-ledger loading
# -----------------------------------------------------------------------------

def load_llm_usage_ledger():
    """
    Load the persistent article-level supervision ledger.

    A new article-level ledger is created automatically for a new UTC day.
    """

    today_utc = datetime.now(
        timezone.utc
    ).date().isoformat()

    existing_file_id = find_drive_file(
        LLM_USAGE_LEDGER_FILENAME,
        LOGS_FOLDER_ID,
    )

    if not existing_file_id:

        return create_empty_llm_usage_ledger()

    ledger = download_json_from_drive(
        existing_file_id
    )

    if (
        ledger.get(
            "date_utc"
        )
        != today_utc
    ):
        return create_empty_llm_usage_ledger()

    migrated_ledger = (
        normalise_llm_usage_ledger(
            ledger
        )
    )

    save_llm_usage_ledger(
        migrated_ledger
    )

    return migrated_ledger


LLM_USAGE_LEDGER = (
    load_llm_usage_ledger()
)


# -----------------------------------------------------------------------------
# Persistent raw-supervision result store
# -----------------------------------------------------------------------------

def create_empty_raw_supervision_store():
    """
    Create the persistent raw-supervision store.

    The store preserves provider responses independently of the Colab runtime.
    """

    now = datetime.now(
        timezone.utc
    )

    return {
        "store_type":
            "media_ai_raw_llm_supervision",

        "store_version":
            RAW_SUPERVISION_STORE_VERSION,

        "created_at_utc":
            now.isoformat(),

        "last_updated_utc":
            now.isoformat(),

        "results_by_supervisor": {
            supervisor_id: []
            for supervisor_id
            in sorted(
                EXPECTED_SUPERVISOR_IDS
            )
        },
    }


def normalise_raw_supervision_store(
    store,
):
    """
    Normalise an existing persistent raw-supervision store.
    """

    if not isinstance(
        store,
        dict,
    ):
        return create_empty_raw_supervision_store()


    store[
        "store_version"
    ] = RAW_SUPERVISION_STORE_VERSION


    results_by_supervisor = store.setdefault(
        "results_by_supervisor",
        {},
    )


    for supervisor_id in sorted(
        EXPECTED_SUPERVISOR_IDS
    ):

        results_by_supervisor.setdefault(
            supervisor_id,
            [],
        )


        if not isinstance(
            results_by_supervisor[
                supervisor_id
            ],
            list,
        ):

            results_by_supervisor[
                supervisor_id
            ] = []


    store.setdefault(
        "created_at_utc",
        datetime.now(
            timezone.utc
        ).isoformat(),
    )


    store[
        "last_updated_utc"
    ] = datetime.now(
        timezone.utc
    ).isoformat()


    return store


def save_raw_supervision_store(
    store,
):
    """
    Persist raw article-level supervision results to Google Drive.
    """

    encoded = json.dumps(
        store,
        ensure_ascii=False,
        indent=2,
        default=str,
    ).encode(
        "utf-8"
    )


    media = MediaIoBaseUpload(
        io.BytesIO(
            encoded
        ),
        mimetype="application/json",
        resumable=False,
    )


    existing_file_id = find_drive_file(
        RAW_SUPERVISION_STORE_FILENAME,
        LOGS_FOLDER_ID,
    )


    if existing_file_id:

        drive_service.files().update(
            fileId=
                existing_file_id,

            media_body=
                media,

            fields=
                "id",
        ).execute()


        return existing_file_id


    metadata = {
        "name":
            RAW_SUPERVISION_STORE_FILENAME,

        "parents": [
            LOGS_FOLDER_ID
        ],
    }


    created = (
        drive_service
        .files()
        .create(
            body=
                metadata,

            media_body=
                media,

            fields=
                "id",
        )
        .execute()
    )


    return created[
        "id"
    ]


def load_raw_supervision_store():
    """
    Load previously persisted raw supervision results.

    Unlike the daily usage ledger, raw supervision results are historical
    research artefacts and are therefore not reset at UTC-day boundaries.
    """

    existing_file_id = find_drive_file(
        RAW_SUPERVISION_STORE_FILENAME,
        LOGS_FOLDER_ID,
    )


    if not existing_file_id:

        store = (
            create_empty_raw_supervision_store()
        )


        file_id = save_raw_supervision_store(
            store
        )


        return (
            store,
            file_id,
        )


    store = download_json_from_drive(
        existing_file_id
    )


    store = normalise_raw_supervision_store(
        store
    )


    save_raw_supervision_store(
        store
    )


    return (
        store,
        existing_file_id,
    )


(
    RAW_SUPERVISION_STORE,
    RAW_SUPERVISION_STORE_FILE_ID,
) = load_raw_supervision_store()


# -----------------------------------------------------------------------------
# Raw-result identity
# -----------------------------------------------------------------------------

def raw_supervision_result_identity(
    raw_result,
):
    """
    Construct a deterministic audit identity for one raw result.

    Provider response ID is preferred where available.
    """

    execution = raw_result.get(
        "execution",
        {}
    )


    provider_response_id = (
        execution.get(
            "provider_response_id"
        )
    )


    if provider_response_id:

        return (
            "provider_response",
            str(
                provider_response_id
            ),
        )


    return (
        "request",
        str(
            raw_result.get(
                "supervisor_id"
            )
        ),
        str(
            raw_result.get(
                "identity",
                {}
            ).get(
                "media_ai_record_id"
            )
        ),
        str(
            raw_result.get(
                "requested_at_utc"
            )
        ),
    )


# -----------------------------------------------------------------------------
# Persistent raw-result append
# -----------------------------------------------------------------------------

def persist_raw_supervision_result(
    raw_result,
):
    """
    Persist one raw supervision result immediately after provider execution.

    Existing identical audit records are not duplicated.
    """

    supervisor_id = raw_result.get(
        "supervisor_id"
    )


    if (
        supervisor_id
        not in EXPECTED_SUPERVISOR_IDS
    ):
        raise ValueError(
            "Raw supervision result contains an invalid supervisor ID."
        )


    supervisor_results = (
        RAW_SUPERVISION_STORE[
            "results_by_supervisor"
        ].setdefault(
            supervisor_id,
            [],
        )
    )


    new_identity = (
        raw_supervision_result_identity(
            raw_result
        )
    )


    existing_identities = {
        raw_supervision_result_identity(
            existing_result
        )

        for existing_result
        in supervisor_results

        if isinstance(
            existing_result,
            dict,
        )
    }


    if (
        new_identity
        not in existing_identities
    ):

        supervisor_results.append(
            deepcopy(
                raw_result
            )
        )


    RAW_SUPERVISION_STORE[
        "last_updated_utc"
    ] = datetime.now(
        timezone.utc
    ).isoformat()


    save_raw_supervision_store(
        RAW_SUPERVISION_STORE
    )


# -----------------------------------------------------------------------------
# Restore raw results into current runtime
# -----------------------------------------------------------------------------

LLM_SUPERVISION_RAW_RESULTS = {
    supervisor_id:
        deepcopy(
            RAW_SUPERVISION_STORE[
                "results_by_supervisor"
            ].get(
                supervisor_id,
                [],
            )
        )

    for supervisor_id
    in sorted(
        EXPECTED_SUPERVISOR_IDS
    )
}


PERSISTED_RAW_RESULT_COUNT = sum(
    len(
        results
    )

    for results
    in LLM_SUPERVISION_RAW_RESULTS.values()
)


PERSISTED_SUCCESSFUL_ARTICLE_PAIRS = {
    (
        result.get(
            "supervisor_id"
        ),
        result.get(
            "identity",
            {}
        ).get(
            "media_ai_record_id"
        ),
    )

    for results
    in LLM_SUPERVISION_RAW_RESULTS.values()

    for result
    in results

    if (
        isinstance(
            result,
            dict,
        )
        and result.get(
            "execution",
            {}
        ).get(
            "status"
        )
        == "SUCCESS"
        and result.get(
            "identity",
            {}
        ).get(
            "media_ai_record_id"
        )
    )
}


# -----------------------------------------------------------------------------
# Reconcile orphaned pending articles
# -----------------------------------------------------------------------------

def reconcile_pending_articles_with_raw_store():
    """
    Release pending article-supervisor pairs for which no recoverable
    provider-successful raw result exists.

    This handles historical runtime-only successful requests that were entered
    into the usage ledger before raw-result persistence was introduced.

    Releasing such an orphaned pending pair permits one controlled retry.
    """

    pending_ids = (
        LLM_USAGE_LEDGER[
            "pending_article_ids_by_supervisor"
        ]
    )


    released_pairs = []


    for supervisor_id in sorted(
        EXPECTED_SUPERVISOR_IDS
    ):

        supervisor_pending = list(
            pending_ids.get(
                supervisor_id,
                [],
            )
        )


        retained_pending = []


        for media_ai_record_id in (
            supervisor_pending
        ):

            pair = (
                supervisor_id,
                media_ai_record_id,
            )


            if (
                pair
                in PERSISTED_SUCCESSFUL_ARTICLE_PAIRS
            ):

                retained_pending.append(
                    media_ai_record_id
                )


            else:

                released_pairs.append(
                    {
                        "supervisor_id":
                            supervisor_id,

                        "media_ai_record_id":
                            media_ai_record_id,

                        "reason":
                            "pending_without_persisted_successful_raw_response",
                    }
                )


        pending_ids[
            supervisor_id
        ] = retained_pending


    if released_pairs:

        LLM_USAGE_LEDGER[
            "last_updated_utc"
        ] = datetime.now(
            timezone.utc
        ).isoformat()


        save_llm_usage_ledger(
            LLM_USAGE_LEDGER
        )


    return released_pairs


ORPHANED_PENDING_ARTICLES_RELEASED = (
    reconcile_pending_articles_with_raw_store()
)

# -----------------------------------------------------------------------------
# Synchronise article validation from Block 7 when available
# -----------------------------------------------------------------------------

def synchronise_runtime_article_validation():
    """
    Synchronise article-level validation decisions produced by Block 7.

    Provider-successful article responses are initially recorded as pending.

    A valid Block 7 article result promotes the article from pending to
    validated.

    A rejected article result removes the article from pending so it remains
    eligible for a later retry.
    """

    if (
        "ARTICLE_RESPONSE_VALIDATION_RESULTS"
        not in globals()
    ):
        return {
            "validated_imported":
                0,

            "rejected_released":
                0,
        }

    pending_ids = (
        LLM_USAGE_LEDGER[
            "pending_article_ids_by_supervisor"
        ]
    )

    validated_ids = (
        LLM_USAGE_LEDGER[
            "validated_article_ids_by_supervisor"
        ]
    )

    validated_imported = 0

    rejected_released = 0


    for result in (
        ARTICLE_RESPONSE_VALIDATION_RESULTS
    ):

        supervisor_id = result.get(
            "supervisor_id"
        )

        media_ai_record_id = result.get(
            "media_ai_record_id"
        )

        valid = result.get(
            "valid"
        )

        if (
            supervisor_id
            not in EXPECTED_SUPERVISOR_IDS
            or not media_ai_record_id
        ):
            continue


        pending_for_supervisor = (
            pending_ids.setdefault(
                supervisor_id,
                [],
            )
        )

        validated_for_supervisor = (
            validated_ids.setdefault(
                supervisor_id,
                [],
            )
        )


        if valid is True:

            if (
                media_ai_record_id
                not in validated_for_supervisor
            ):

                validated_for_supervisor.append(
                    media_ai_record_id
                )

                validated_imported += 1


            if (
                media_ai_record_id
                in pending_for_supervisor
            ):

                pending_for_supervisor.remove(
                    media_ai_record_id
                )


        elif valid is False:

            if (
                media_ai_record_id
                in pending_for_supervisor
            ):

                pending_for_supervisor.remove(
                    media_ai_record_id
                )

                rejected_released += 1


    if (
        validated_imported
        or rejected_released
    ):

        LLM_USAGE_LEDGER[
            "last_updated_utc"
        ] = datetime.now(
            timezone.utc
        ).isoformat()

        save_llm_usage_ledger(
            LLM_USAGE_LEDGER
        )


    return {
        "validated_imported":
            validated_imported,

        "rejected_released":
            rejected_released,
    }


RUNTIME_ARTICLE_VALIDATION_SYNC = (
    synchronise_runtime_article_validation()
)


# -----------------------------------------------------------------------------
# Article-level usage-limit verification
# -----------------------------------------------------------------------------

def remaining_daily_supervision_capacity():
    """
    Return remaining article-request capacity for the current UTC day.
    """

    requests_reserved_today = (
        LLM_USAGE_LEDGER.get(
            "article_requests_reserved",
            0,
        )
    )

    return max(
        0,
        (
            MAX_SUPERVISION_REQUESTS_PER_DAY
            - requests_reserved_today
        ),
    )


def check_supervision_request_allowed():
    """
    Verify article-level daily request availability.
    """

    if (
        remaining_daily_supervision_capacity()
        <= 0
    ):
        return (
            False,
            "Daily article supervision request limit reached.",
        )

    return (
        True,
        "Article request allowed.",
    )


# -----------------------------------------------------------------------------
# Article request reservation
# -----------------------------------------------------------------------------

def reserve_supervision_request():
    """
    Reserve one article-level external request before provider execution.
    """

    allowed, reason = (
        check_supervision_request_allowed()
    )

    if not allowed:
        return (
            False,
            reason,
        )

    LLM_USAGE_LEDGER[
        "article_requests_reserved"
    ] += 1

    LLM_USAGE_LEDGER[
        "last_updated_utc"
    ] = datetime.now(
        timezone.utc
    ).isoformat()

    save_llm_usage_ledger(
        LLM_USAGE_LEDGER
    )

    return (
        True,
        "Article request reserved.",
    )


# -----------------------------------------------------------------------------
# Completed article-request accounting
# -----------------------------------------------------------------------------

def register_supervision_result(
    success,
    supervisor_id,
    media_ai_record_id,
):
    """
    Record completion of one article-level provider request.

    A provider-successful article is marked as pending until Block 7 confirms
    complete and valid sentence-level coverage.
    """

    LLM_USAGE_LEDGER[
        "article_requests_completed"
    ] += 1

    if success:

        LLM_USAGE_LEDGER[
            "article_requests_successful"
        ] += 1

        pending_ids = (
            LLM_USAGE_LEDGER[
                "pending_article_ids_by_supervisor"
            ].setdefault(
                supervisor_id,
                [],
            )
        )

        if (
            media_ai_record_id
            and media_ai_record_id
            not in pending_ids
        ):

            pending_ids.append(
                media_ai_record_id
            )

    else:

        LLM_USAGE_LEDGER[
            "article_requests_failed"
        ] += 1

    LLM_USAGE_LEDGER[
        "last_updated_utc"
    ] = datetime.now(
        timezone.utc
    ).isoformat()

    save_llm_usage_ledger(
        LLM_USAGE_LEDGER
    )


# -----------------------------------------------------------------------------
# Supervisor readiness
# -----------------------------------------------------------------------------

SUPERVISOR_READINESS = {}


for (
    supervisor_id,
    configuration,
) in SUPERVISOR_REGISTRY.items():

    enabled = configuration.get(
        "enabled",
        False,
    )

    provider = configuration.get(
        "provider"
    )

    model = configuration.get(
        "model"
    )

    configured = all([
        bool(
            provider
        ),
        bool(
            model
        ),
    ])

    implemented = (
        configuration.get(
            "implementation_status"
        )
        == "implemented"
    )

    free_tier_approved = (
        supervisor_free_tier_approved(
            configuration
        )
        if configured
        else False
    )


    if provider == "google":

        credential_ready = (
            GEMINI_CREDENTIAL_READY
        )

        sdk_ready = (
            GOOGLE_GENAI_AVAILABLE
        )

    else:

        credential_ready = False

        sdk_ready = False


    cost_policy_ready = all([
        REQUIRE_FREE_SUPERVISION,
        not ALLOW_PAID_API_USAGE,
        not ALLOW_AUTOMATIC_PAID_FALLBACK,
        free_tier_approved,
    ])


    executable = all([
        RUN_LLM_SUPERVISION,
        enabled,
        configured,
        implemented,
        credential_ready,
        sdk_ready,
        cost_policy_ready,
    ])


    SUPERVISOR_READINESS[
        supervisor_id
    ] = {
        "enabled":
            enabled,

        "configured":
            configured,

        "implemented":
            implemented,

        "credential_ready":
            credential_ready,

        "sdk_ready":
            sdk_ready,

        "free_tier_approved":
            free_tier_approved,

        "cost_policy_ready":
            cost_policy_ready,

        "executable":
            executable,
    }


# -----------------------------------------------------------------------------
# Enabled-supervisor resolution
# -----------------------------------------------------------------------------

ENABLED_SUPERVISORS = {
    supervisor_id:
        configuration

    for (
        supervisor_id,
        configuration,
    ) in SUPERVISOR_REGISTRY.items()

    if configuration[
        "enabled"
    ]
}


# -----------------------------------------------------------------------------
# Executable-supervisor resolution
# -----------------------------------------------------------------------------

SUPERVISORS_ALLOWED_TO_EXECUTE = {
    supervisor_id:
        SUPERVISOR_REGISTRY[
            supervisor_id
        ]

    for supervisor_id
    in SUPERVISOR_REGISTRY

    if SUPERVISOR_READINESS[
        supervisor_id
    ][
        "executable"
    ]
}


# -----------------------------------------------------------------------------
# Canonical JSON Schema helpers
# -----------------------------------------------------------------------------
#
# These schemas continue to define the full Media AI annotation contract.
#
# They are retained locally for methodological reference and downstream
# validation.
#
# They are NOT sent in full to Gemini in version 1.4.
# -----------------------------------------------------------------------------

def build_annotation_value_schema(
    minimum,
    maximum,
):
    """
    Build the canonical schema for one bounded annotation value.
    """

    return {
        "type":
            "object",

        "properties": {
            "value": {
                "type": [
                    "number",
                    "null",
                ],

                "minimum":
                    minimum,

                "maximum":
                    maximum,
            },

            "value_state": {
                "type":
                    "string",

                "enum":
                    sorted(
                        ANNOTATION_VALUE_STATES
                    ),
            },
        },

        "required": [
            "value",
            "value_state",
        ],

        "additionalProperties":
            False,
    }


def build_emotion_vector_schema(
    minimum,
    maximum,
):
    """
    Build one complete canonical emotion-vector schema.
    """

    return {
        "type":
            "object",

        "properties": {
            dimension:
                build_annotation_value_schema(
                    minimum,
                    maximum,
                )

            for dimension
            in EMOTIONAL_DIMENSIONS
        },

        "required":
            list(
                EMOTIONAL_DIMENSIONS.keys()
            ),

        "additionalProperties":
            False,
    }


# -----------------------------------------------------------------------------
# Full canonical sentence-level annotation JSON Schema
# -----------------------------------------------------------------------------

LLM_SENTENCE_ANNOTATION_JSON_SCHEMA = {
    "type":
        "object",

    "properties": {
        "reported_emotion":
            build_emotion_vector_schema(
                minimum=0.0,
                maximum=1.0,
            ),

        "transformative_emotion":
            build_emotion_vector_schema(
                minimum=-1.0,
                maximum=1.0,
            ),

        "immediate_emotional_state":
            build_emotion_vector_schema(
                minimum=0.0,
                maximum=1.0,
            ),

        "emotional_weight":
            build_annotation_value_schema(
                minimum=0.0,
                maximum=1.0,
            ),

        "factological_weight":
            build_annotation_value_schema(
                minimum=0.0,
                maximum=1.0,
            ),

        "factological_components": {
            "type":
                "object",

            "properties": {
                component: {
                    "type": [
                        "boolean",
                        "null",
                    ],
                }

                for component
                in FACTOLOGICAL_COMPONENTS
            },

            "required":
                list(
                    FACTOLOGICAL_COMPONENTS.keys()
                ),

            "additionalProperties":
                False,
        },

        "annotation_confidence":
            build_annotation_value_schema(
                minimum=0.0,
                maximum=1.0,
            ),

        "evidence": {
            "type":
                "object",

            "properties": {
                "summary": {
                    "type": [
                        "string",
                        "null",
                    ],
                },

                "supporting_text": {
                    "type":
                        "array",

                    "items": {
                        "type":
                            "string",
                    },
                },
            },

            "required": [
                "summary",
                "supporting_text",
            ],

            "additionalProperties":
                False,
        },
    },

    "required": [
        "reported_emotion",
        "transformative_emotion",
        "immediate_emotional_state",
        "emotional_weight",
        "factological_weight",
        "factological_components",
        "annotation_confidence",
        "evidence",
    ],

    "additionalProperties":
        False,
}


# -----------------------------------------------------------------------------
# Lightweight provider article-output JSON Schema
# -----------------------------------------------------------------------------
#
# Version 1.4 deliberately avoids sending the full deeply nested canonical
# annotation schema to the provider.
#
# Gemini is asked to guarantee only:
#
#   article identity
#   sentence identity
#   one JSON-encoded annotation object for every sentence
#
# The JSON annotation string is parsed locally and remains subject to the
# complete Media AI validation contract downstream.
# -----------------------------------------------------------------------------

def build_article_output_json_schema(
    supervision_payload,
):
    """
    Build the lightweight provider-side structured-output schema.

    The provider schema constrains article identity, sentence identity,
    sentence count and the presence of one annotation JSON string per
    sentence.

    It does not replace the full canonical annotation schema.
    """

    article = supervision_payload.get(
        "article",
        {}
    )

    expected_sentence_ids = article.get(
        "expected_sentence_ids",
        []
    )

    expected_sentence_count = len(
        expected_sentence_ids
    )


    if expected_sentence_count <= 0:
        raise ValueError(
            "Article supervision payload contains no expected sentence IDs."
        )


    return {
        "type":
            "object",

        "properties": {
            "media_ai_record_id": {
                "type":
                    "string",

                "description":
                    "Media AI record identifier supplied in the request.",
            },

            "sentence_annotations": {
                "type":
                    "array",

                "minItems":
                    expected_sentence_count,

                "maxItems":
                    expected_sentence_count,

                "items": {
                    "type":
                        "object",

                    "properties": {
                        "sentence_id": {
                            "type":
                                "string",

                            "enum":
                                expected_sentence_ids,

                            "description":
                                "Sentence identifier supplied in the request.",
                        },

                        "annotation_json": {
                            "type":
                                "string",

                            "description": (
                                "Valid JSON text containing the complete "
                                "Media AI sentence annotation object."
                            ),
                        },
                    },

                    "required": [
                        "sentence_id",
                        "annotation_json",
                    ],

                    "additionalProperties":
                        False,
                },
            },
        },

        "required": [
            "media_ai_record_id",
            "sentence_annotations",
        ],

        "additionalProperties":
            False,
    }


# -----------------------------------------------------------------------------
# Provider-independent article supervisor input
# -----------------------------------------------------------------------------

def build_supervisor_input_text(
    supervision_payload,
):
    """
    Convert one validated article supervision payload into provider-neutral
    text.

    The complete article is supplied once as an ordered sentence sequence.
    """

    identity = supervision_payload.get(
        "identity",
        {}
    )

    article = supervision_payload.get(
        "article",
        {}
    )

    sentence_units = article.get(
        "sentence_units",
        []
    )


    ordered_sentence_units = [
        {
            "paragraph_id":
                unit.get(
                    "paragraph_id"
                ),

            "sentence_id":
                unit.get(
                    "sentence_id"
                ),

            "paragraph_index":
                unit.get(
                    "paragraph_index"
                ),

            "sentence_index":
                unit.get(
                    "sentence_index"
                ),

            "text":
                unit.get(
                    "current_sentence"
                ),
        }

        for unit
        in sentence_units
    ]


    article_sequence_json = json.dumps(
        ordered_sentence_units,
        ensure_ascii=False,
        indent=2,
    )


    required_annotation_fields = [
        "reported_emotion",
        "transformative_emotion",
        "immediate_emotional_state",
        "emotional_weight",
        "factological_weight",
        "factological_components",
        "annotation_confidence",
        "evidence",
    ]


    required_annotation_fields_text = (
        ", ".join(
            required_annotation_fields
        )
    )


    return (
        f"{LLM_SUPERVISION_INSTRUCTIONS}\n\n"

        "MEDIA AI ARTICLE SUPERVISION UNIT\n\n"

        f"Media AI record ID: "
        f"{identity.get('media_ai_record_id')}\n\n"

        f"Article title:\n"
        f"{article.get('article_title')}\n\n"

        f"Expected sentence count: "
        f"{article.get('sentence_count')}\n\n"

        "Ordered article sentence sequence:\n"
        f"{article_sequence_json}\n\n"

        "PROVIDER OUTPUT ENVELOPE\n\n"

        "Return exactly one sentence-level annotation for every supplied "
        "sentence identifier, in the same order.\n\n"

        "Return the Media AI record ID exactly as supplied.\n\n"

        "For each sentence, place the COMPLETE Media AI annotation object "
        "inside the `annotation_json` field as valid JSON text.\n\n"

        "The JSON object encoded inside `annotation_json` must contain "
        "exactly these top-level annotation fields:\n"
        f"{required_annotation_fields_text}\n\n"

        "The internal annotation structure, dimensions, value ranges, "
        "value states, factological components and evidence requirements "
        "must follow the Media AI supervision instructions above.\n\n"

        "Do not omit dimensions because their values are unavailable. "
        "Represent unavailable values using the annotation value-state "
        "contract defined in the supervision instructions.\n\n"

        "Return only the structured article output required by the "
        "provider response schema."
    )


# -----------------------------------------------------------------------------
# Provider-output normalisation
# -----------------------------------------------------------------------------

def normalise_provider_article_output(
    provider_output,
    supervision_payload,
):
    """
    Parse annotation_json values from the lightweight provider envelope.

    Returns the original downstream article structure:

        {
            media_ai_record_id,
            sentence_annotations: [
                {
                    sentence_id,
                    annotation: {...}
                }
            ]
        }

    Full semantic/schema acceptance remains deferred to Block 7.
    """

    if not isinstance(
        provider_output,
        dict,
    ):
        raise TypeError(
            "Gemini article output must decode to a dictionary."
        )


    media_ai_record_id = provider_output.get(
        "media_ai_record_id"
    )


    sentence_annotations = provider_output.get(
        "sentence_annotations"
    )


    if not isinstance(
        sentence_annotations,
        list,
    ):
        raise TypeError(
            "Gemini article output does not contain a valid "
            "sentence_annotations list."
        )


    expected_article = supervision_payload.get(
        "article",
        {}
    )

    expected_sentence_ids = expected_article.get(
        "expected_sentence_ids",
        []
    )


    normalised_annotations = []


    for item in sentence_annotations:

        if not isinstance(
            item,
            dict,
        ):
            raise TypeError(
                "One or more provider sentence annotations "
                "are not dictionaries."
            )


        sentence_id = item.get(
            "sentence_id"
        )


        annotation_json = item.get(
            "annotation_json"
        )


        if not isinstance(
            annotation_json,
            str,
        ):
            raise TypeError(
                "Provider annotation_json must be a string."
            )


        if not annotation_json.strip():
            raise ValueError(
                "Provider annotation_json cannot be empty."
            )


        try:

            annotation = json.loads(
                annotation_json
            )

        except json.JSONDecodeError as exc:

            raise ValueError(
                "One provider annotation_json value could not "
                "be decoded as JSON."
            ) from exc


        if not isinstance(
            annotation,
            dict,
        ):
            raise TypeError(
                "Decoded sentence annotation must be a dictionary."
            )


        normalised_annotations.append(
            {
                "sentence_id":
                    sentence_id,

                "annotation":
                    annotation,
            }
        )


    # -------------------------------------------------------------------------
    # Lightweight structural checks
    # -------------------------------------------------------------------------
    #
    # These checks prevent obvious transport/envelope corruption.
    #
    # Full annotation validation remains Block 7 responsibility.
    # -------------------------------------------------------------------------

    returned_sentence_ids = [
        item.get(
            "sentence_id"
        )
        for item
        in normalised_annotations
    ]


    if len(
        returned_sentence_ids
    ) != len(
        expected_sentence_ids
    ):
        raise ValueError(
            "Provider returned an unexpected number of "
            "sentence annotations."
        )


    if set(
        returned_sentence_ids
    ) != set(
        expected_sentence_ids
    ):
        raise ValueError(
            "Provider sentence identifiers do not match "
            "the requested article sentence identifiers."
        )


    return {
        "media_ai_record_id":
            media_ai_record_id,

        "sentence_annotations":
            normalised_annotations,
    }


# -----------------------------------------------------------------------------
# Gemini article-level supervisor execution
# -----------------------------------------------------------------------------

def execute_gemini_supervision(
    supervisor_id,
    supervisor_configuration,
    supervision_payload,
):
    """
    Execute one complete-article Gemini supervision request.

    Provider-side structured output uses a deliberately shallow envelope.

    The complete annotation object is encoded as annotation_json and parsed
    locally before downstream canonical validation.

    A provider-successful article remains pending until Block 7 validates
    complete sentence-level coverage and all sentence annotations.
    """

    if not GOOGLE_GENAI_AVAILABLE:

        return {
            "executed":
                False,

            "status":
                "BLOCKED",

            "reason":
                "The google-genai Python SDK is unavailable.",
        }


    if not GEMINI_CREDENTIAL_READY:

        return {
            "executed":
                False,

            "status":
                "BLOCKED",

            "reason":
                "GEMINI_API_KEY is unavailable in Colab Secrets.",
        }


    model_name = (
        supervisor_configuration.get(
            "model"
        )
    )


    if not model_name:

        return {
            "executed":
                False,

            "status":
                "BLOCKED",

            "reason":
                "No Gemini supervisor model is configured.",
        }


    if (
        REQUIRE_FREE_SUPERVISION
        and model_name
        not in FREE_TIER_APPROVED_SUPERVISOR_MODELS
    ):

        return {
            "executed":
                False,

            "status":
                "BLOCKED",

            "reason":
                "Configured Gemini model is not approved "
                "by the research Free Tier allow-list.",
        }


    if ALLOW_PAID_API_USAGE:

        return {
            "executed":
                False,

            "status":
                "BLOCKED",

            "reason":
                "Paid API usage is prohibited by the current "
                "research configuration.",
        }


    if ALLOW_AUTOMATIC_PAID_FALLBACK:

        return {
            "executed":
                False,

            "status":
                "BLOCKED",

            "reason":
                "Automatic paid fallback is prohibited by the "
                "current research configuration.",
        }


    media_ai_record_id = (
        supervision_payload.get(
            "identity",
            {}
        ).get(
            "media_ai_record_id"
        )
    )


    if not media_ai_record_id:

        return {
            "executed":
                False,

            "status":
                "BLOCKED",

            "reason":
                "Article supervision payload has no Media AI record ID.",
        }


    allowed, reason = (
        reserve_supervision_request()
    )


    if not allowed:

        return {
            "executed":
                False,

            "status":
                "BLOCKED",

            "reason":
                reason,
        }


    client = genai.Client(
        api_key=
            GEMINI_API_KEY
    )


    user_input = (
        build_supervisor_input_text(
            supervision_payload
        )
    )


    article_output_schema = (
        build_article_output_json_schema(
            supervision_payload
        )
    )


    try:

        # ---------------------------------------------------------------------
        # Current Gemini Interactions structured-output contract
        # ---------------------------------------------------------------------

        interaction = (
            client.interactions.create(
                model=
                    model_name,

                input=
                    user_input,

                response_format={
                    "type":
                        "text",

                    "mime_type":
                        "application/json",

                    "schema":
                        article_output_schema,
                },
            )
        )


        raw_output_text = (
            interaction.output_text
        )


        if not isinstance(
            raw_output_text,
            str,
        ):

            raise TypeError(
                "Gemini interaction did not return textual output."
            )


        if not raw_output_text.strip():

            raise ValueError(
                "Gemini interaction returned an empty structured response."
            )


        try:

            provider_parsed_output = json.loads(
                raw_output_text
            )

        except json.JSONDecodeError as exc:

            raise ValueError(
                "Gemini structured response could not "
                "be decoded as JSON."
            ) from exc


        # ---------------------------------------------------------------------
        # Convert lightweight provider envelope to downstream article structure
        # ---------------------------------------------------------------------

        parsed_output = (
            normalise_provider_article_output(
                provider_output=
                    provider_parsed_output,

                supervision_payload=
                    supervision_payload,
            )
        )


        register_supervision_result(
            success=True,
            supervisor_id=
                supervisor_id,
            media_ai_record_id=
                media_ai_record_id,
        )


        return {
            "executed":
                True,

            "status":
                "SUCCESS",

            "reason":
                "Gemini article supervision request completed successfully.",

            "provider":
                "google",

            "model":
                model_name,

            "provider_response_id":
                getattr(
                    interaction,
                    "id",
                    None,
                ),

            "provider_schema_version":
                PROVIDER_SCHEMA_VERSION,

            "provider_schema_strategy":
                PROVIDER_SCHEMA_STRATEGY,

            "raw_output_text":
                raw_output_text,

            "provider_parsed_output":
                provider_parsed_output,

            "parsed_output":
                parsed_output,

            "article_sentence_count":
                supervision_payload.get(
                    "article",
                    {}
                ).get(
                    "sentence_count"
                ),
        }


    except Exception as exc:

        register_supervision_result(
            success=False,
            supervisor_id=
                supervisor_id,
            media_ai_record_id=
                media_ai_record_id,
        )


        return {
            "executed":
                True,

            "status":
                "ERROR",

            "reason":
                str(
                    exc
                ),

            "error_type":
                type(
                    exc
                ).__name__,

            "provider":
                "google",

            "model":
                model_name,

            "provider_schema_version":
                PROVIDER_SCHEMA_VERSION,

            "provider_schema_strategy":
                PROVIDER_SCHEMA_STRATEGY,

            "article_sentence_count":
                supervision_payload.get(
                    "article",
                    {}
                ).get(
                    "sentence_count"
                ),
        }


# -----------------------------------------------------------------------------
# Provider routing
# -----------------------------------------------------------------------------

def execute_supervisor_request(
    supervisor_id,
    supervisor_configuration,
    supervision_payload,
):
    """
    Route one complete-article supervision request to its configured provider.
    """

    provider = (
        supervisor_configuration.get(
            "provider"
        )
    )


    if provider == "google":

        return execute_gemini_supervision(
            supervisor_id=
                supervisor_id,

            supervisor_configuration=
                supervisor_configuration,

            supervision_payload=
                supervision_payload,
        )


    return {
        "executed":
            False,

        "status":
            "BLOCKED",

        "reason":
            "Configured supervisor provider is not implemented.",
    }


# -----------------------------------------------------------------------------
# Article-level selection diagnostics
# -----------------------------------------------------------------------------

SUPERVISION_SELECTION_SUMMARY = {
    supervisor_id: {
        "validated_articles":
            len(
                LLM_USAGE_LEDGER[
                    "validated_article_ids_by_supervisor"
                ].get(
                    supervisor_id,
                    [],
                )
            ),

        "pending_articles":
            len(
                LLM_USAGE_LEDGER[
                    "pending_article_ids_by_supervisor"
                ].get(
                    supervisor_id,
                    [],
                )
            ),

        "eligible_unsupervised":
            0,

        "selected_this_run":
            0,

        "selected_sentence_count":
            0,
    }

    for supervisor_id
    in SUPERVISOR_REGISTRY
}


# -----------------------------------------------------------------------------
# Controlled article-level supervisor execution with resume protection
# -----------------------------------------------------------------------------

RUN_REQUESTS_ATTEMPTED = 0

RUN_SENTENCES_INCLUDED = 0


if RUN_LLM_SUPERVISION:

    remaining_run_capacity = min(
        MAX_SUPERVISION_REQUESTS_PER_RUN,
        remaining_daily_supervision_capacity(),
    )


    for (
        supervisor_id,
        configuration,
    ) in SUPERVISORS_ALLOWED_TO_EXECUTE.items():

        if remaining_run_capacity <= 0:
            break


        validated_article_ids = set(
            LLM_USAGE_LEDGER[
                "validated_article_ids_by_supervisor"
            ].get(
                supervisor_id,
                [],
            )
        )


        pending_article_ids = set(
            LLM_USAGE_LEDGER[
                "pending_article_ids_by_supervisor"
            ].get(
                supervisor_id,
                [],
            )
        )


        already_processed_article_ids = (
            validated_article_ids
            | pending_article_ids
        )


        eligible_payloads = [
            payload
            for payload
            in LLM_ARTICLE_SUPERVISION_PAYLOADS
            if payload.get(
                "identity",
                {}
            ).get(
                "media_ai_record_id"
            )
            not in already_processed_article_ids
        ]


        SUPERVISION_SELECTION_SUMMARY[
            supervisor_id
        ][
            "eligible_unsupervised"
        ] = len(
            eligible_payloads
        )


        requested_payloads = (
            eligible_payloads[
                :remaining_run_capacity
            ]
        )


        SUPERVISION_SELECTION_SUMMARY[
            supervisor_id
        ][
            "selected_this_run"
        ] = len(
            requested_payloads
        )


        SUPERVISION_SELECTION_SUMMARY[
            supervisor_id
        ][
            "selected_sentence_count"
        ] = sum(
            payload.get(
                "article",
                {}
            ).get(
                "sentence_count",
                0,
            )

            for payload
            in requested_payloads
        )


        for payload in requested_payloads:

            if remaining_run_capacity <= 0:
                break


            request_started_at = (
                datetime.now(
                    timezone.utc
                ).isoformat()
            )


            execution_result = (
                execute_supervisor_request(
                    supervisor_id=
                        supervisor_id,

                    supervisor_configuration=
                        configuration,

                    supervision_payload=
                        payload,
                )
            )


            raw_result = {
                "supervisor_id":
                    supervisor_id,

                "provider":
                    configuration.get(
                        "provider"
                    ),

                "model":
                    configuration.get(
                        "model"
                    ),

                "request_unit":
                    "article",

                "annotation_unit":
                    "sentence",

                "requested_at_utc":
                    request_started_at,

                "identity":
                    deepcopy(
                        payload.get(
                            "identity",
                            {}
                        )
                    ),

                "article_metadata": {
                    "article_title":
                        payload.get(
                            "article",
                            {}
                        ).get(
                            "article_title"
                        ),

                    "sentence_count":
                        payload.get(
                            "article",
                            {}
                        ).get(
                            "sentence_count"
                        ),

                    "expected_sentence_ids":
                        deepcopy(
                            payload.get(
                                "article",
                                {}
                            ).get(
                                "expected_sentence_ids",
                                [],
                            )
                        ),
                },

                "protocol_version":
                    LLM_SUPERVISION_PROTOCOL_VERSION,

                "supervision_schema_version":
                    SUPERVISION_SCHEMA_VERSION,

                "annotation_policy_version":
                    ANNOTATION_POLICY_VERSION,

                "provider_schema_version":
                    PROVIDER_SCHEMA_VERSION,

                "provider_schema_strategy":
                    PROVIDER_SCHEMA_STRATEGY,

                "execution":
                    execution_result,
            }


            LLM_SUPERVISION_RAW_RESULTS[
                supervisor_id
            ].append(
                raw_result
            )


            # -----------------------------------------------------------------
            # Persist raw provider result immediately
            # -----------------------------------------------------------------

            persist_raw_supervision_result(
                raw_result
            )


            if execution_result.get(
                "executed"
            ):

                RUN_REQUESTS_ATTEMPTED += 1

                RUN_SENTENCES_INCLUDED += (
                    payload.get(
                        "article",
                        {}
                    ).get(
                        "sentence_count",
                        0,
                    )
                )

                remaining_run_capacity -= 1


# -----------------------------------------------------------------------------
# Execution diagnostics
# -----------------------------------------------------------------------------

SUPERVISION_EXECUTION_SUMMARY = {}


for supervisor_id in SUPERVISOR_REGISTRY:

    results = (
        LLM_SUPERVISION_RAW_RESULTS[
            supervisor_id
        ]
    )


    successful = sum(
        1
        for result in results
        if result.get(
            "execution",
            {}
        ).get(
            "status"
        )
        == "SUCCESS"
    )


    errors = sum(
        1
        for result in results
        if result.get(
            "execution",
            {}
        ).get(
            "status"
        )
        == "ERROR"
    )


    blocked = sum(
        1
        for result in results
        if result.get(
            "execution",
            {}
        ).get(
            "status"
        )
        == "BLOCKED"
    )


    sentence_annotations_requested = sum(
        result.get(
            "article_metadata",
            {}
        ).get(
            "sentence_count",
            0,
        )

        for result
        in results
    )


    SUPERVISION_EXECUTION_SUMMARY[
        supervisor_id
    ] = {
        "requests_recorded":
            len(
                results
            ),

        "successful":
            successful,

        "errors":
            errors,

        "blocked":
            blocked,

        "sentence_annotations_requested":
            sentence_annotations_requested,
    }


# -----------------------------------------------------------------------------
# Final persistent usage state
# -----------------------------------------------------------------------------

FINAL_LLM_USAGE = deepcopy(
    LLM_USAGE_LEDGER
)


FINAL_DAILY_CAPACITY = (
    remaining_daily_supervision_capacity()
)


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)

print(
    "Media AI — Controlled LLM Supervisor Execution"
)

print("=" * 72)


print(
    f"Registry version           : "
    f"{SUPERVISOR_REGISTRY_VERSION}"
)

print(
    f"Provider schema version    : "
    f"{PROVIDER_SCHEMA_VERSION}"
)

print(
    f"Provider schema strategy   : "
    f"{PROVIDER_SCHEMA_STRATEGY}"
)

print(
    f"Master supervision switch  : "
    f"{RUN_LLM_SUPERVISION}"
)

print(
    f"External request unit      : "
    f"{LLM_SUPERVISION_REQUEST_UNIT}"
)

print(
    f"Annotation unit            : "
    f"{LLM_SUPERVISION_ANNOTATION_UNIT}"
)

print(
    f"Article payloads available : "
    f"{len(LLM_ARTICLE_SUPERVISION_PAYLOADS)}"
)

print(
    f"Configured supervisor slots: "
    f"{len(SUPERVISOR_REGISTRY)}"
)

print(
    f"Enabled supervisor slots   : "
    f"{len(ENABLED_SUPERVISORS)}"
)

print(
    f"Executable supervisors     : "
    f"{len(SUPERVISORS_ALLOWED_TO_EXECUTE)}"
)


print("-" * 72)

print(
    "Structured-output boundary"
)

print(
    f"Full canonical schema local: "
    f"{True}"
)

print(
    f"Full canonical schema sent : "
    f"{False}"
)

print(
    f"Provider envelope shallow  : "
    f"{True}"
)

print(
    f"Annotation JSON local parse : "
    f"{True}"
)

print(
    f"Canonical validation later : "
    f"{True}"
)


print("-" * 72)

print(
    "Cost and article-request controls"
)

print(
    f"Require free supervision   : "
    f"{REQUIRE_FREE_SUPERVISION}"
)

print(
    f"Paid API usage allowed     : "
    f"{ALLOW_PAID_API_USAGE}"
)

print(
    f"Paid fallback allowed      : "
    f"{ALLOW_AUTOMATIC_PAID_FALLBACK}"
)

print(
    f"Max article requests / run : "
    f"{MAX_SUPERVISION_REQUESTS_PER_RUN}"
)

print(
    f"Max article requests / day : "
    f"{MAX_SUPERVISION_REQUESTS_PER_DAY}"
)

print(
    f"Article requests this run  : "
    f"{RUN_REQUESTS_ATTEMPTED}"
)

print(
    f"Sentences included this run: "
    f"{RUN_SENTENCES_INCLUDED}"
)

print(
    f"Article requests reserved  : "
    f"{FINAL_LLM_USAGE['article_requests_reserved']}"
)

print(
    f"Article requests completed : "
    f"{FINAL_LLM_USAGE['article_requests_completed']}"
)

print(
    f"Provider-successful        : "
    f"{FINAL_LLM_USAGE['article_requests_successful']}"
)

print(
    f"Provider-failed            : "
    f"{FINAL_LLM_USAGE['article_requests_failed']}"
)

print(
    f"Remaining article capacity : "
    f"{FINAL_DAILY_CAPACITY}"
)


print("-" * 72)

print(
    "Legacy sentence-request audit"
)


legacy_totals = (
    FINAL_LLM_USAGE[
        "legacy_sentence_request_totals"
    ]
)


print(
    f"Legacy requests reserved   : "
    f"{legacy_totals['requests_reserved']}"
)

print(
    f"Legacy requests completed  : "
    f"{legacy_totals['requests_completed']}"
)

print(
    f"Legacy requests successful : "
    f"{legacy_totals['requests_successful']}"
)

print(
    f"Legacy requests failed     : "
    f"{legacy_totals['requests_failed']}"
)


print("-" * 72)

print(
    "Article-level resume protection"
)

print(
    f"Validated IDs imported     : "
    f"{RUNTIME_ARTICLE_VALIDATION_SYNC['validated_imported']}"
)

print(
    f"Rejected pending released  : "
    f"{RUNTIME_ARTICLE_VALIDATION_SYNC['rejected_released']}"
)


for supervisor_id in sorted(
    SUPERVISOR_REGISTRY
):

    selection = (
        SUPERVISION_SELECTION_SUMMARY[
            supervisor_id
        ]
    )


    print(
        f"{supervisor_id}"
    )

    print(
        f"  Validated articles       : "
        f"{selection['validated_articles']}"
    )

    print(
        f"  Pending articles         : "
        f"{selection['pending_articles']}"
    )

    print(
        f"  Eligible unsupervised    : "
        f"{selection['eligible_unsupervised']}"
    )

    print(
        f"  Selected this run        : "
        f"{selection['selected_this_run']}"
    )

    print(
        f"  Sentences selected       : "
        f"{selection['selected_sentence_count']}"
    )


print("-" * 72)

print(
    "Gemini runtime"
)

print(
    f"Google GenAI SDK available : "
    f"{GOOGLE_GENAI_AVAILABLE}"
)

print(
    f"Gemini credential ready    : "
    f"{GEMINI_CREDENTIAL_READY}"
)

print(
    f"Gemini secret status       : "
    f"{GEMINI_SECRET_STATUS}"
)


print("-" * 72)


for supervisor_id in sorted(
    SUPERVISOR_REGISTRY
):

    configuration = (
        SUPERVISOR_REGISTRY[
            supervisor_id
        ]
    )

    readiness = (
        SUPERVISOR_READINESS[
            supervisor_id
        ]
    )

    execution_summary = (
        SUPERVISION_EXECUTION_SUMMARY[
            supervisor_id
        ]
    )


    print(
        f"{supervisor_id}"
    )

    print(
        f"  Enabled                  : "
        f"{configuration['enabled']}"
    )

    print(
        f"  Provider                 : "
        f"{configuration['provider']}"
    )

    print(
        f"  Model                    : "
        f"{configuration['model']}"
    )

    print(
        f"  Configured               : "
        f"{readiness['configured']}"
    )

    print(
        f"  Implementation           : "
        f"{configuration['implementation_status']}"
    )

    print(
        f"  Credential ready         : "
        f"{readiness['credential_ready']}"
    )

    print(
        f"  SDK ready                : "
        f"{readiness['sdk_ready']}"
    )

    print(
        f"  Free Tier approved       : "
        f"{readiness['free_tier_approved']}"
    )

    print(
        f"  Cost policy ready        : "
        f"{readiness['cost_policy_ready']}"
    )

    print(
        f"  Executable               : "
        f"{readiness['executable']}"
    )

    print(
        f"  Article requests recorded: "
        f"{execution_summary['requests_recorded']}"
    )

    print(
        f"  Successful               : "
        f"{execution_summary['successful']}"
    )

    print(
        f"  Errors                   : "
        f"{execution_summary['errors']}"
    )

    print(
        f"  Blocked                  : "
        f"{execution_summary['blocked']}"
    )

    print(
        f"  Sentence targets         : "
        f"{execution_summary['sentence_annotations_requested']}"
    )

    print("-" * 72)


print(
    "Complete articles are the external LLM request unit."
)

print(
    "Sentence-level annotations remain the Media AI supervision unit."
)

print(
    "Provider-side structured output now enforces only the lightweight "
    "article and sentence envelope."
)

print(
    "Complete Media AI sentence annotations are returned as JSON text "
    "inside that provider envelope and parsed locally."
)

print(
    "The full canonical annotation schema remains a local validation "
    "contract and has not been weakened."
)

print(
    "Provider-successful articles remain pending until Block 7 "
    "validates complete sentence-level coverage."
)

print(
    "Validated article-supervisor pairs are excluded from later requests."
)

print(
    "Earlier sentence-level API usage is retained separately for audit."
)

print(
    "Raw supervisor outputs remain separated by supervisor identity."
)


if RUN_LLM_SUPERVISION:

    print(
        "Controlled Gemini supervision was attempted only for "
        "eligible article-level request units."
    )

else:

    print(
        "LLM supervision was skipped by the Execution Control Panel."
    )


print(
    "No canonical sentence annotation has been accepted by this block."
)

print(
    "No supervisor comparison or consensus has been performed."
)

print("=" * 72)
```

```python
# =============================================================================
# Media AI — Final Audit Contract Diagnostic
# =============================================================================

print("=" * 72)
print("Media AI — Final Audit Contract Diagnostic")
print("=" * 72)

print("ANNOTATION_SCHEMA")
print("-" * 72)

for key in [
    "schema_type",
    "schema_version",
    "annotation_policy_version",
    "primary_annotation_unit",
    "emotional_dimensions",
    "factological_components",
    "constructs",
    "context_policy",
    "methodological_constraints",
]:

    value = ANNOTATION_SCHEMA.get(
        key
    )

    print(
        f"{key:<28}: "
        f"type={type(value).__name__}"
    )

    if key in {
        "schema_type",
        "schema_version",
        "annotation_policy_version",
        "primary_annotation_unit",
    }:
        print(
            f"  value                    : "
            f"{value}"
        )


print("-" * 72)

print("Numerical scales")
print("-" * 72)

scale_objects = {
    "NON_NEGATIVE_STATE_SCALE":
        NON_NEGATIVE_STATE_SCALE,

    "SIGNED_TRANSFORMATION_SCALE":
        SIGNED_TRANSFORMATION_SCALE,

    "NON_NEGATIVE_WEIGHT_SCALE":
        NON_NEGATIVE_WEIGHT_SCALE,

    "CONFIDENCE_SCALE":
        CONFIDENCE_SCALE,
}


for name, value in scale_objects.items():

    print(
        f"{name:<32}: "
        f"type={type(value).__name__}"
    )

    print(
        f"  value                    : "
        f"{value}"
    )

    if isinstance(
        value,
        dict,
    ):
        print(
            f"  keys                     : "
            f"{sorted(value.keys())}"
        )


print("-" * 72)

print("Template validation")
print("-" * 72)

print(
    f"template_dimensions_valid   : "
    f"{template_dimensions_valid}"
)

print(
    f"factological_template_valid : "
    f"{factological_template_valid}"
)

print(
    f"blank_records_valid         : "
    f"{blank_records_valid}"
)

print(
    f"ANNOTATION_RECORD_SCHEMA_VALID: "
    f"{ANNOTATION_RECORD_SCHEMA_VALID}"
)

print("=" * 72)
```

```python
# =============================================================================
# Media AI — Controlled LLM Supervisor Registry and Execution
# Block 6 — Restart-Safe Article-Level Supervision
# =============================================================================

import io
import json
from copy import deepcopy
from datetime import datetime, timezone

from googleapiclient.http import MediaIoBaseUpload


# -----------------------------------------------------------------------------
# Controlled-execution configuration
# -----------------------------------------------------------------------------

SUPERVISOR_REGISTRY_VERSION = "1.5"

LLM_USAGE_LEDGER_VERSION = "2.0"

LLM_USAGE_LEDGER_FILENAME = (
    "notebook_03_llm_usage_ledger.json"
)

RAW_SUPERVISION_STORE_VERSION = "1.0"

RAW_SUPERVISION_STORE_FILENAME = (
    "notebook_03_raw_supervision_results.json"
)

PROVIDER_SCHEMA_VERSION = "1.1"

PROVIDER_SCHEMA_STRATEGY = (
    "shallow_article_envelope_with_annotation_json"
)


# -----------------------------------------------------------------------------
# Block 1 control availability check
# -----------------------------------------------------------------------------

required_execution_controls = [
    "RUN_LLM_SUPERVISION",

    "ENABLE_SUPERVISOR_01",
    "ENABLE_SUPERVISOR_02",
    "ENABLE_SUPERVISOR_03",
    "ENABLE_SUPERVISOR_04",
    "ENABLE_SUPERVISOR_05",

    "SUPERVISOR_01_PROVIDER",
    "SUPERVISOR_01_MODEL",

    "SUPERVISOR_02_PROVIDER",
    "SUPERVISOR_02_MODEL",

    "SUPERVISOR_03_PROVIDER",
    "SUPERVISOR_03_MODEL",

    "SUPERVISOR_04_PROVIDER",
    "SUPERVISOR_04_MODEL",

    "SUPERVISOR_05_PROVIDER",
    "SUPERVISOR_05_MODEL",

    "REQUIRE_FREE_SUPERVISION",
    "ALLOW_PAID_API_USAGE",
    "ALLOW_AUTOMATIC_PAID_FALLBACK",

    "MAX_SUPERVISION_REQUESTS_PER_RUN",
    "MAX_SUPERVISION_REQUESTS_PER_DAY",
]


missing_execution_controls = [
    control
    for control in required_execution_controls
    if control not in globals()
]


if missing_execution_controls:

    raise NameError(
        "Block 1 execution controls are not initialised. "
        "Rerun Notebook 03 Block 1. Missing: "
        f"{missing_execution_controls}"
    )


# -----------------------------------------------------------------------------
# Block 2 persistent-storage availability check
# -----------------------------------------------------------------------------

required_storage_objects = [
    "drive_service",
    "LOGS_FOLDER_ID",
    "find_drive_file",
    "download_json_from_drive",
]


missing_storage_objects = [
    object_name
    for object_name in required_storage_objects
    if object_name not in globals()
]


if missing_storage_objects:

    raise NameError(
        "Required persistent-storage objects are not initialised. "
        "Rerun Notebook 03 Block 2. Missing: "
        f"{missing_storage_objects}"
    )


# -----------------------------------------------------------------------------
# Block 5 supervision-protocol availability check
# -----------------------------------------------------------------------------

required_protocol_objects = [
    "LLM_ARTICLE_SUPERVISION_PAYLOADS",
    "LLM_SUPERVISION_INSTRUCTIONS",
    "LLM_SUPERVISION_PROTOCOL_VERSION",
    "LLM_SUPERVISION_ANNOTATION_UNIT",
    "LLM_SUPERVISION_REQUEST_UNIT",
    "ANNOTATION_VALUE_STATES",
    "EMOTIONAL_DIMENSIONS",
    "FACTOLOGICAL_COMPONENTS",
    "SUPERVISION_SCHEMA_VERSION",
    "ANNOTATION_POLICY_VERSION",
]


missing_protocol_objects = [
    object_name
    for object_name in required_protocol_objects
    if object_name not in globals()
]


if missing_protocol_objects:

    raise NameError(
        "Required supervision protocol objects are not initialised. "
        "Rerun Notebook 03 Blocks 3–5. Missing: "
        f"{missing_protocol_objects}"
    )


if LLM_SUPERVISION_REQUEST_UNIT != "article":

    raise ValueError(
        "Block 6 requires article-level supervision payloads."
    )


if LLM_SUPERVISION_ANNOTATION_UNIT != "sentence":

    raise ValueError(
        "The Media AI annotation unit must remain the sentence."
    )


# -----------------------------------------------------------------------------
# Google GenAI SDK readiness
# -----------------------------------------------------------------------------

try:

    from google import genai

    GOOGLE_GENAI_AVAILABLE = True

except ImportError:

    genai = None

    GOOGLE_GENAI_AVAILABLE = False


# -----------------------------------------------------------------------------
# Gemini credential readiness
# -----------------------------------------------------------------------------

GEMINI_API_KEY = None

GEMINI_CREDENTIAL_READY = False

GEMINI_SECRET_STATUS = "NOT_LOADED"


try:

    from google.colab import userdata

    GEMINI_API_KEY = userdata.get(
        "GEMINI_API_KEY"
    )

    GEMINI_CREDENTIAL_READY = bool(
        isinstance(
            GEMINI_API_KEY,
            str,
        )
        and GEMINI_API_KEY.strip()
    )

    GEMINI_SECRET_STATUS = (
        "AVAILABLE"
        if GEMINI_CREDENTIAL_READY
        else "UNAVAILABLE"
    )


except Exception:

    GEMINI_API_KEY = None

    GEMINI_CREDENTIAL_READY = False

    GEMINI_SECRET_STATUS = "UNAVAILABLE"


# -----------------------------------------------------------------------------
# Research-approved Free Tier model allow-list
# -----------------------------------------------------------------------------

FREE_TIER_APPROVED_SUPERVISOR_MODELS = {
    "gemini-3.6-flash",
}


# -----------------------------------------------------------------------------
# Stable five-slot supervisor registry
# -----------------------------------------------------------------------------

SUPERVISOR_REGISTRY = {
    "supervisor_01": {
        "enabled":
            ENABLE_SUPERVISOR_01,

        "provider":
            SUPERVISOR_01_PROVIDER,

        "model":
            SUPERVISOR_01_MODEL,

        "role":
            "independent_llm_supervisor",

        "implementation_status":
            "implemented",
    },

    "supervisor_02": {
        "enabled":
            ENABLE_SUPERVISOR_02,

        "provider":
            SUPERVISOR_02_PROVIDER,

        "model":
            SUPERVISOR_02_MODEL,

        "role":
            "independent_llm_supervisor",

        "implementation_status":
            "placeholder",
    },

    "supervisor_03": {
        "enabled":
            ENABLE_SUPERVISOR_03,

        "provider":
            SUPERVISOR_03_PROVIDER,

        "model":
            SUPERVISOR_03_MODEL,

        "role":
            "independent_llm_supervisor",

        "implementation_status":
            "placeholder",
    },

    "supervisor_04": {
        "enabled":
            ENABLE_SUPERVISOR_04,

        "provider":
            SUPERVISOR_04_PROVIDER,

        "model":
            SUPERVISOR_04_MODEL,

        "role":
            "independent_llm_supervisor",

        "implementation_status":
            "placeholder",
    },

    "supervisor_05": {
        "enabled":
            ENABLE_SUPERVISOR_05,

        "provider":
            SUPERVISOR_05_PROVIDER,

        "model":
            SUPERVISOR_05_MODEL,

        "role":
            "independent_llm_supervisor",

        "implementation_status":
            "placeholder",
    },
}


EXPECTED_SUPERVISOR_IDS = {
    "supervisor_01",
    "supervisor_02",
    "supervisor_03",
    "supervisor_04",
    "supervisor_05",
}


if (
    set(
        SUPERVISOR_REGISTRY.keys()
    )
    != EXPECTED_SUPERVISOR_IDS
):

    raise ValueError(
        "Supervisor registry does not contain "
        "the expected five stable supervisor slots."
    )


# -----------------------------------------------------------------------------
# Free-tier model validation
# -----------------------------------------------------------------------------

def supervisor_free_tier_approved(
    supervisor_configuration,
):
    """
    Return True only when the configured model has been explicitly approved
    by the internal research Free Tier allow-list.
    """

    model = supervisor_configuration.get(
        "model"
    )

    return (
        model
        in FREE_TIER_APPROVED_SUPERVISOR_MODELS
    )


# =============================================================================
# Persistent article-level usage ledger
# =============================================================================


# -----------------------------------------------------------------------------
# Empty usage ledger
# -----------------------------------------------------------------------------

def create_empty_llm_usage_ledger():
    """
    Create a new UTC-day article-level supervision usage ledger.

    External request unit:
        article

    Research annotation unit:
        sentence
    """

    now = datetime.now(
        timezone.utc
    )

    return {
        "ledger_type":
            "media_ai_llm_supervision_usage",

        "ledger_version":
            LLM_USAGE_LEDGER_VERSION,

        "date_utc":
            now.date().isoformat(),

        "request_unit":
            "article",

        "article_requests_reserved":
            0,

        "article_requests_completed":
            0,

        "article_requests_successful":
            0,

        "article_requests_failed":
            0,

        "pending_article_ids_by_supervisor": {
            supervisor_id: []
            for supervisor_id
            in sorted(
                EXPECTED_SUPERVISOR_IDS
            )
        },

        "validated_article_ids_by_supervisor": {
            supervisor_id: []
            for supervisor_id
            in sorted(
                EXPECTED_SUPERVISOR_IDS
            )
        },

        "legacy_sentence_request_totals": {
            "requests_reserved":
                0,

            "requests_completed":
                0,

            "requests_successful":
                0,

            "requests_failed":
                0,
        },

        "legacy_successful_sentence_ids_by_supervisor": {
            supervisor_id: []
            for supervisor_id
            in sorted(
                EXPECTED_SUPERVISOR_IDS
            )
        },

        "last_updated_utc":
            now.isoformat(),
    }


# -----------------------------------------------------------------------------
# Usage-ledger migration
# -----------------------------------------------------------------------------

def normalise_llm_usage_ledger(
    ledger,
):
    """
    Upgrade an existing ledger to the current article-level structure without
    deleting historical sentence-level accounting.
    """

    if not isinstance(
        ledger,
        dict,
    ):

        return create_empty_llm_usage_ledger()


    # -------------------------------------------------------------------------
    # Already article-level
    # -------------------------------------------------------------------------

    if (
        ledger.get(
            "request_unit"
        )
        == "article"
    ):

        ledger[
            "ledger_version"
        ] = LLM_USAGE_LEDGER_VERSION


        ledger.setdefault(
            "article_requests_reserved",
            0,
        )

        ledger.setdefault(
            "article_requests_completed",
            0,
        )

        ledger.setdefault(
            "article_requests_successful",
            0,
        )

        ledger.setdefault(
            "article_requests_failed",
            0,
        )


        ledger.setdefault(
            "legacy_sentence_request_totals",
            {
                "requests_reserved": 0,
                "requests_completed": 0,
                "requests_successful": 0,
                "requests_failed": 0,
            },
        )


        ledger.setdefault(
            "legacy_successful_sentence_ids_by_supervisor",
            {},
        )


        pending_ids = ledger.setdefault(
            "pending_article_ids_by_supervisor",
            {},
        )


        validated_ids = ledger.setdefault(
            "validated_article_ids_by_supervisor",
            {},
        )


        legacy_sentence_ids = ledger[
            "legacy_successful_sentence_ids_by_supervisor"
        ]


        for supervisor_id in sorted(
            EXPECTED_SUPERVISOR_IDS
        ):

            pending_ids.setdefault(
                supervisor_id,
                [],
            )

            validated_ids.setdefault(
                supervisor_id,
                [],
            )

            legacy_sentence_ids.setdefault(
                supervisor_id,
                [],
            )


            pending_ids[
                supervisor_id
            ] = list(
                dict.fromkeys(
                    pending_ids[
                        supervisor_id
                    ]
                )
            )


            validated_ids[
                supervisor_id
            ] = list(
                dict.fromkeys(
                    validated_ids[
                        supervisor_id
                    ]
                )
            )


            legacy_sentence_ids[
                supervisor_id
            ] = list(
                dict.fromkeys(
                    legacy_sentence_ids[
                        supervisor_id
                    ]
                )
            )


        return ledger


    # -------------------------------------------------------------------------
    # Migration from earlier sentence-level ledger
    # -------------------------------------------------------------------------

    migrated = create_empty_llm_usage_ledger()


    migrated[
        "date_utc"
    ] = ledger.get(
        "date_utc",
        migrated[
            "date_utc"
        ],
    )


    migrated[
        "legacy_sentence_request_totals"
    ] = {
        "requests_reserved":
            ledger.get(
                "requests_reserved",
                0,
            ),

        "requests_completed":
            ledger.get(
                "requests_completed",
                0,
            ),

        "requests_successful":
            ledger.get(
                "requests_successful",
                0,
            ),

        "requests_failed":
            ledger.get(
                "requests_failed",
                0,
            ),
    }


    old_successful_sentence_ids = (
        ledger.get(
            "successful_sentence_ids_by_supervisor",
            {},
        )
    )


    for supervisor_id in sorted(
        EXPECTED_SUPERVISOR_IDS
    ):

        migrated[
            "legacy_successful_sentence_ids_by_supervisor"
        ][
            supervisor_id
        ] = list(
            dict.fromkeys(
                old_successful_sentence_ids.get(
                    supervisor_id,
                    [],
                )
            )
        )


    migrated[
        "last_updated_utc"
    ] = datetime.now(
        timezone.utc
    ).isoformat()


    return migrated


# -----------------------------------------------------------------------------
# Save usage ledger
# -----------------------------------------------------------------------------

def save_llm_usage_ledger(
    ledger,
):
    """
    Persist the article-level supervision usage ledger.
    """

    encoded = json.dumps(
        ledger,
        ensure_ascii=False,
        indent=2,
    ).encode(
        "utf-8"
    )


    media = MediaIoBaseUpload(
        io.BytesIO(
            encoded
        ),
        mimetype="application/json",
        resumable=False,
    )


    existing_file_id = find_drive_file(
        LLM_USAGE_LEDGER_FILENAME,
        LOGS_FOLDER_ID,
    )


    if existing_file_id:

        drive_service.files().update(
            fileId=
                existing_file_id,

            media_body=
                media,

            fields=
                "id",
        ).execute()


        return existing_file_id


    metadata = {
        "name":
            LLM_USAGE_LEDGER_FILENAME,

        "parents": [
            LOGS_FOLDER_ID
        ],
    }


    created = (
        drive_service
        .files()
        .create(
            body=
                metadata,

            media_body=
                media,

            fields=
                "id",
        )
        .execute()
    )


    return created[
        "id"
    ]


# -----------------------------------------------------------------------------
# Load usage ledger
# -----------------------------------------------------------------------------

def load_llm_usage_ledger():
    """
    Load the current UTC-day article-level usage ledger.

    Daily counters reset on a new UTC day.
    """

    today_utc = datetime.now(
        timezone.utc
    ).date().isoformat()


    existing_file_id = find_drive_file(
        LLM_USAGE_LEDGER_FILENAME,
        LOGS_FOLDER_ID,
    )


    if not existing_file_id:

        return create_empty_llm_usage_ledger()


    ledger = download_json_from_drive(
        existing_file_id
    )


    if (
        ledger.get(
            "date_utc"
        )
        != today_utc
    ):

        return create_empty_llm_usage_ledger()


    migrated_ledger = (
        normalise_llm_usage_ledger(
            ledger
        )
    )


    save_llm_usage_ledger(
        migrated_ledger
    )


    return migrated_ledger


LLM_USAGE_LEDGER = (
    load_llm_usage_ledger()
)


# =============================================================================
# Persistent raw-supervision response store
# =============================================================================


# -----------------------------------------------------------------------------
# Empty raw store
# -----------------------------------------------------------------------------

def create_empty_raw_supervision_store():
    """
    Create a persistent raw-supervision store.

    Unlike the usage ledger, this store is historical and is not reset
    on a new UTC day.
    """

    now = datetime.now(
        timezone.utc
    )


    return {
        "store_type":
            "media_ai_raw_llm_supervision",

        "store_version":
            RAW_SUPERVISION_STORE_VERSION,

        "created_at_utc":
            now.isoformat(),

        "last_updated_utc":
            now.isoformat(),

        "results_by_supervisor": {
            supervisor_id: []
            for supervisor_id
            in sorted(
                EXPECTED_SUPERVISOR_IDS
            )
        },
    }


# -----------------------------------------------------------------------------
# Normalise raw store
# -----------------------------------------------------------------------------

def normalise_raw_supervision_store(
    store,
):
    """
    Normalise an existing persistent raw-supervision store.
    """

    if not isinstance(
        store,
        dict,
    ):

        return create_empty_raw_supervision_store()


    store[
        "store_version"
    ] = RAW_SUPERVISION_STORE_VERSION


    results_by_supervisor = store.setdefault(
        "results_by_supervisor",
        {},
    )


    for supervisor_id in sorted(
        EXPECTED_SUPERVISOR_IDS
    ):

        results_by_supervisor.setdefault(
            supervisor_id,
            [],
        )


        if not isinstance(
            results_by_supervisor[
                supervisor_id
            ],
            list,
        ):

            results_by_supervisor[
                supervisor_id
            ] = []


    store.setdefault(
        "created_at_utc",
        datetime.now(
            timezone.utc
        ).isoformat(),
    )


    store[
        "last_updated_utc"
    ] = datetime.now(
        timezone.utc
    ).isoformat()


    return store


# -----------------------------------------------------------------------------
# Save raw store
# -----------------------------------------------------------------------------

def save_raw_supervision_store(
    store,
):
    """
    Persist raw article-level supervision responses to Google Drive.
    """

    encoded = json.dumps(
        store,
        ensure_ascii=False,
        indent=2,
        default=str,
    ).encode(
        "utf-8"
    )


    media = MediaIoBaseUpload(
        io.BytesIO(
            encoded
        ),
        mimetype="application/json",
        resumable=False,
    )


    existing_file_id = find_drive_file(
        RAW_SUPERVISION_STORE_FILENAME,
        LOGS_FOLDER_ID,
    )


    if existing_file_id:

        drive_service.files().update(
            fileId=
                existing_file_id,

            media_body=
                media,

            fields=
                "id",
        ).execute()


        return existing_file_id


    metadata = {
        "name":
            RAW_SUPERVISION_STORE_FILENAME,

        "parents": [
            LOGS_FOLDER_ID
        ],
    }


    created = (
        drive_service
        .files()
        .create(
            body=
                metadata,

            media_body=
                media,

            fields=
                "id",
        )
        .execute()
    )


    return created[
        "id"
    ]


# -----------------------------------------------------------------------------
# Load raw store
# -----------------------------------------------------------------------------

def load_raw_supervision_store():
    """
    Load persistent raw article-level supervision results.
    """

    existing_file_id = find_drive_file(
        RAW_SUPERVISION_STORE_FILENAME,
        LOGS_FOLDER_ID,
    )


    if not existing_file_id:

        store = (
            create_empty_raw_supervision_store()
        )


        file_id = save_raw_supervision_store(
            store
        )


        return (
            store,
            file_id,
        )


    store = download_json_from_drive(
        existing_file_id
    )


    store = normalise_raw_supervision_store(
        store
    )


    save_raw_supervision_store(
        store
    )


    return (
        store,
        existing_file_id,
    )


(
    RAW_SUPERVISION_STORE,
    RAW_SUPERVISION_STORE_FILE_ID,
) = load_raw_supervision_store()


# -----------------------------------------------------------------------------
# Raw-result identity
# -----------------------------------------------------------------------------

def raw_supervision_result_identity(
    raw_result,
):
    """
    Construct a stable audit identity for one raw supervision response.
    """

    execution = raw_result.get(
        "execution",
        {}
    )


    provider_response_id = (
        execution.get(
            "provider_response_id"
        )
    )


    if provider_response_id:

        return (
            "provider_response",
            str(
                provider_response_id
            ),
        )


    return (
        "request",
        str(
            raw_result.get(
                "supervisor_id"
            )
        ),
        str(
            raw_result.get(
                "identity",
                {}
            ).get(
                "media_ai_record_id"
            )
        ),
        str(
            raw_result.get(
                "requested_at_utc"
            )
        ),
    )


# -----------------------------------------------------------------------------
# Persist one raw result
# -----------------------------------------------------------------------------

def persist_raw_supervision_result(
    raw_result,
):
    """
    Persist one raw provider result without duplicating the same audit record.
    """

    supervisor_id = raw_result.get(
        "supervisor_id"
    )


    if (
        supervisor_id
        not in EXPECTED_SUPERVISOR_IDS
    ):

        raise ValueError(
            "Raw supervision result contains an invalid supervisor ID."
        )


    supervisor_results = (
        RAW_SUPERVISION_STORE[
            "results_by_supervisor"
        ].setdefault(
            supervisor_id,
            [],
        )
    )


    new_identity = (
        raw_supervision_result_identity(
            raw_result
        )
    )


    existing_identities = {
        raw_supervision_result_identity(
            existing_result
        )

        for existing_result
        in supervisor_results

        if isinstance(
            existing_result,
            dict,
        )
    }


    if (
        new_identity
        not in existing_identities
    ):

        supervisor_results.append(
            deepcopy(
                raw_result
            )
        )


    RAW_SUPERVISION_STORE[
        "last_updated_utc"
    ] = datetime.now(
        timezone.utc
    ).isoformat()


    save_raw_supervision_store(
        RAW_SUPERVISION_STORE
    )


# -----------------------------------------------------------------------------
# Restore persisted raw results into current runtime
# -----------------------------------------------------------------------------

LLM_SUPERVISION_RAW_RESULTS = {
    supervisor_id:
        deepcopy(
            RAW_SUPERVISION_STORE[
                "results_by_supervisor"
            ].get(
                supervisor_id,
                [],
            )
        )

    for supervisor_id
    in sorted(
        EXPECTED_SUPERVISOR_IDS
    )
}


RECOVERED_RAW_RESULT_COUNT = sum(
    len(
        results
    )
    for results
    in LLM_SUPERVISION_RAW_RESULTS.values()
)


# -----------------------------------------------------------------------------
# Successful persisted article-supervisor pairs
# -----------------------------------------------------------------------------

def get_successful_persisted_article_pairs():
    """
    Return all provider-successful article-supervisor pairs preserved
    in the raw store.
    """

    return {
        (
            result.get(
                "supervisor_id"
            ),
            result.get(
                "identity",
                {}
            ).get(
                "media_ai_record_id"
            ),
        )

        for results
        in RAW_SUPERVISION_STORE[
            "results_by_supervisor"
        ].values()

        for result
        in results

        if (
            isinstance(
                result,
                dict,
            )
            and result.get(
                "execution",
                {}
            ).get(
                "status"
            )
            == "SUCCESS"
            and result.get(
                "identity",
                {}
            ).get(
                "media_ai_record_id"
            )
        )
    }


PERSISTED_SUCCESSFUL_ARTICLE_PAIRS = (
    get_successful_persisted_article_pairs()
)


# -----------------------------------------------------------------------------
# Reconcile orphaned pending articles
# -----------------------------------------------------------------------------

def reconcile_pending_articles_with_raw_store():
    """
    Release historical pending pairs for which no recoverable provider-successful
    raw result exists.

    This repairs runtime-only historical successes created before the raw-store
    persistence mechanism existed.
    """

    pending_ids = (
        LLM_USAGE_LEDGER[
            "pending_article_ids_by_supervisor"
        ]
    )


    released_pairs = []


    for supervisor_id in sorted(
        EXPECTED_SUPERVISOR_IDS
    ):

        supervisor_pending = list(
            pending_ids.get(
                supervisor_id,
                [],
            )
        )


        retained_pending = []


        for media_ai_record_id in (
            supervisor_pending
        ):

            pair = (
                supervisor_id,
                media_ai_record_id,
            )


            if (
                pair
                in PERSISTED_SUCCESSFUL_ARTICLE_PAIRS
            ):

                retained_pending.append(
                    media_ai_record_id
                )


            else:

                released_pairs.append(
                    {
                        "supervisor_id":
                            supervisor_id,

                        "media_ai_record_id":
                            media_ai_record_id,

                        "reason":
                            "pending_without_persisted_successful_raw_response",
                    }
                )


        pending_ids[
            supervisor_id
        ] = retained_pending


    if released_pairs:

        LLM_USAGE_LEDGER[
            "last_updated_utc"
        ] = datetime.now(
            timezone.utc
        ).isoformat()


        save_llm_usage_ledger(
            LLM_USAGE_LEDGER
        )


    return released_pairs


ORPHANED_PENDING_ARTICLES_RELEASED = (
    reconcile_pending_articles_with_raw_store()
)


# =============================================================================
# Synchronise article validation from Block 7 when available
# =============================================================================


def synchronise_runtime_article_validation():
    """
    Synchronise article-level Block 7 validation decisions with the usage ledger.

    Valid:
        pending -> validated

    Invalid:
        pending -> released for later retry
    """

    if (
        "ARTICLE_RESPONSE_VALIDATION_RESULTS"
        not in globals()
    ):

        return {
            "validated_imported":
                0,

            "rejected_released":
                0,
        }


    pending_ids = (
        LLM_USAGE_LEDGER[
            "pending_article_ids_by_supervisor"
        ]
    )


    validated_ids = (
        LLM_USAGE_LEDGER[
            "validated_article_ids_by_supervisor"
        ]
    )


    validated_imported = 0

    rejected_released = 0


    for result in (
        ARTICLE_RESPONSE_VALIDATION_RESULTS
    ):

        supervisor_id = result.get(
            "supervisor_id"
        )


        media_ai_record_id = result.get(
            "media_ai_record_id"
        )


        valid = result.get(
            "valid"
        )


        if (
            supervisor_id
            not in EXPECTED_SUPERVISOR_IDS
            or not media_ai_record_id
        ):

            continue


        pending_for_supervisor = (
            pending_ids.setdefault(
                supervisor_id,
                [],
            )
        )


        validated_for_supervisor = (
            validated_ids.setdefault(
                supervisor_id,
                [],
            )
        )


        if valid is True:

            if (
                media_ai_record_id
                not in validated_for_supervisor
            ):

                validated_for_supervisor.append(
                    media_ai_record_id
                )

                validated_imported += 1


            if (
                media_ai_record_id
                in pending_for_supervisor
            ):

                pending_for_supervisor.remove(
                    media_ai_record_id
                )


        elif valid is False:

            if (
                media_ai_record_id
                in pending_for_supervisor
            ):

                pending_for_supervisor.remove(
                    media_ai_record_id
                )

                rejected_released += 1


    if (
        validated_imported
        or rejected_released
    ):

        LLM_USAGE_LEDGER[
            "last_updated_utc"
        ] = datetime.now(
            timezone.utc
        ).isoformat()


        save_llm_usage_ledger(
            LLM_USAGE_LEDGER
        )


    return {
        "validated_imported":
            validated_imported,

        "rejected_released":
            rejected_released,
    }


RUNTIME_ARTICLE_VALIDATION_SYNC = (
    synchronise_runtime_article_validation()
)


# =============================================================================
# Usage limits and accounting
# =============================================================================


# -----------------------------------------------------------------------------
# Remaining daily capacity
# -----------------------------------------------------------------------------

def remaining_daily_supervision_capacity():
    """
    Return remaining article-request capacity for the current UTC day.
    """

    requests_reserved_today = (
        LLM_USAGE_LEDGER.get(
            "article_requests_reserved",
            0,
        )
    )


    return max(
        0,
        (
            MAX_SUPERVISION_REQUESTS_PER_DAY
            - requests_reserved_today
        ),
    )


# -----------------------------------------------------------------------------
# Request permission
# -----------------------------------------------------------------------------

def check_supervision_request_allowed():
    """
    Verify article-level daily request availability.
    """

    if (
        remaining_daily_supervision_capacity()
        <= 0
    ):

        return (
            False,
            "Daily article supervision request limit reached.",
        )


    return (
        True,
        "Article request allowed.",
    )


# -----------------------------------------------------------------------------
# Reserve article request
# -----------------------------------------------------------------------------

def reserve_supervision_request():
    """
    Reserve one article-level external request before provider execution.
    """

    allowed, reason = (
        check_supervision_request_allowed()
    )


    if not allowed:

        return (
            False,
            reason,
        )


    LLM_USAGE_LEDGER[
        "article_requests_reserved"
    ] += 1


    LLM_USAGE_LEDGER[
        "last_updated_utc"
    ] = datetime.now(
        timezone.utc
    ).isoformat()


    save_llm_usage_ledger(
        LLM_USAGE_LEDGER
    )


    return (
        True,
        "Article request reserved.",
    )


# -----------------------------------------------------------------------------
# Register completed provider request
# -----------------------------------------------------------------------------

def register_supervision_result(
    success,
    supervisor_id,
    media_ai_record_id,
):
    """
    Record completion of one article-level provider request.

    Provider-successful results are placed in pending state until Block 7
    validates the full sentence-level article response.
    """

    LLM_USAGE_LEDGER[
        "article_requests_completed"
    ] += 1


    if success:

        LLM_USAGE_LEDGER[
            "article_requests_successful"
        ] += 1


        pending_ids = (
            LLM_USAGE_LEDGER[
                "pending_article_ids_by_supervisor"
            ].setdefault(
                supervisor_id,
                [],
            )
        )


        if (
            media_ai_record_id
            and media_ai_record_id
            not in pending_ids
        ):

            pending_ids.append(
                media_ai_record_id
            )


    else:

        LLM_USAGE_LEDGER[
            "article_requests_failed"
        ] += 1


    LLM_USAGE_LEDGER[
        "last_updated_utc"
    ] = datetime.now(
        timezone.utc
    ).isoformat()


    save_llm_usage_ledger(
        LLM_USAGE_LEDGER
    )


# =============================================================================
# Supervisor readiness
# =============================================================================


SUPERVISOR_READINESS = {}


for (
    supervisor_id,
    configuration,
) in SUPERVISOR_REGISTRY.items():

    enabled = configuration.get(
        "enabled",
        False,
    )


    provider = configuration.get(
        "provider"
    )


    model = configuration.get(
        "model"
    )


    configured = all([
        bool(
            provider
        ),
        bool(
            model
        ),
    ])


    implemented = (
        configuration.get(
            "implementation_status"
        )
        == "implemented"
    )


    free_tier_approved = (
        supervisor_free_tier_approved(
            configuration
        )
        if configured
        else False
    )


    if provider == "google":

        credential_ready = (
            GEMINI_CREDENTIAL_READY
        )

        sdk_ready = (
            GOOGLE_GENAI_AVAILABLE
        )


    else:

        credential_ready = False

        sdk_ready = False


    cost_policy_ready = all([
        REQUIRE_FREE_SUPERVISION,
        not ALLOW_PAID_API_USAGE,
        not ALLOW_AUTOMATIC_PAID_FALLBACK,
        free_tier_approved,
    ])


    executable = all([
        RUN_LLM_SUPERVISION,
        enabled,
        configured,
        implemented,
        credential_ready,
        sdk_ready,
        cost_policy_ready,
    ])


    SUPERVISOR_READINESS[
        supervisor_id
    ] = {
        "enabled":
            enabled,

        "configured":
            configured,

        "implemented":
            implemented,

        "credential_ready":
            credential_ready,

        "sdk_ready":
            sdk_ready,

        "free_tier_approved":
            free_tier_approved,

        "cost_policy_ready":
            cost_policy_ready,

        "executable":
            executable,
    }


ENABLED_SUPERVISORS = {
    supervisor_id:
        configuration

    for (
        supervisor_id,
        configuration,
    ) in SUPERVISOR_REGISTRY.items()

    if configuration[
        "enabled"
    ]
}


SUPERVISORS_ALLOWED_TO_EXECUTE = {
    supervisor_id:
        SUPERVISOR_REGISTRY[
            supervisor_id
        ]

    for supervisor_id
    in SUPERVISOR_REGISTRY

    if SUPERVISOR_READINESS[
        supervisor_id
    ][
        "executable"
    ]
}


# =============================================================================
# Canonical annotation-schema definitions
# =============================================================================


def build_annotation_value_schema(
    minimum,
    maximum,
):
    """
    Build the canonical schema for one bounded annotation value.
    """

    return {
        "type":
            "object",

        "properties": {
            "value": {
                "type": [
                    "number",
                    "null",
                ],

                "minimum":
                    minimum,

                "maximum":
                    maximum,
            },

            "value_state": {
                "type":
                    "string",

                "enum":
                    sorted(
                        ANNOTATION_VALUE_STATES
                    ),
            },
        },

        "required": [
            "value",
            "value_state",
        ],

        "additionalProperties":
            False,
    }


def build_emotion_vector_schema(
    minimum,
    maximum,
):
    """
    Build one complete canonical emotion-vector schema.
    """

    return {
        "type":
            "object",

        "properties": {
            dimension:
                build_annotation_value_schema(
                    minimum,
                    maximum,
                )

            for dimension
            in EMOTIONAL_DIMENSIONS
        },

        "required":
            list(
                EMOTIONAL_DIMENSIONS.keys()
            ),

        "additionalProperties":
            False,
    }


LLM_SENTENCE_ANNOTATION_JSON_SCHEMA = {
    "type":
        "object",

    "properties": {
        "reported_emotion":
            build_emotion_vector_schema(
                minimum=0.0,
                maximum=1.0,
            ),

        "transformative_emotion":
            build_emotion_vector_schema(
                minimum=-1.0,
                maximum=1.0,
            ),

        "immediate_emotional_state":
            build_emotion_vector_schema(
                minimum=0.0,
                maximum=1.0,
            ),

        "emotional_weight":
            build_annotation_value_schema(
                minimum=0.0,
                maximum=1.0,
            ),

        "factological_weight":
            build_annotation_value_schema(
                minimum=0.0,
                maximum=1.0,
            ),

        "factological_components": {
            "type":
                "object",

            "properties": {
                component: {
                    "type": [
                        "boolean",
                        "null",
                    ],
                }

                for component
                in FACTOLOGICAL_COMPONENTS
            },

            "required":
                list(
                    FACTOLOGICAL_COMPONENTS.keys()
                ),

            "additionalProperties":
                False,
        },

        "annotation_confidence":
            build_annotation_value_schema(
                minimum=0.0,
                maximum=1.0,
            ),

        "evidence": {
            "type":
                "object",

            "properties": {
                "summary": {
                    "type": [
                        "string",
                        "null",
                    ],
                },

                "supporting_text": {
                    "type":
                        "array",

                    "items": {
                        "type":
                            "string",
                    },
                },
            },

            "required": [
                "summary",
                "supporting_text",
            ],

            "additionalProperties":
                False,
        },
    },

    "required": [
        "reported_emotion",
        "transformative_emotion",
        "immediate_emotional_state",
        "emotional_weight",
        "factological_weight",
        "factological_components",
        "annotation_confidence",
        "evidence",
    ],

    "additionalProperties":
        False,
}


# =============================================================================
# Lightweight provider structured-output schema
# =============================================================================


def build_article_output_json_schema(
    supervision_payload,
):
    """
    Build the deliberately shallow provider-side JSON schema.

    The provider guarantees:
        article identity;
        sentence identity;
        exact sentence count;
        one annotation JSON string per sentence.

    Full Media AI canonical validation remains local.
    """

    article = supervision_payload.get(
        "article",
        {}
    )


    expected_sentence_ids = article.get(
        "expected_sentence_ids",
        []
    )


    expected_sentence_count = len(
        expected_sentence_ids
    )


    if expected_sentence_count <= 0:

        raise ValueError(
            "Article supervision payload contains no expected sentence IDs."
        )


    return {
        "type":
            "object",

        "properties": {
            "media_ai_record_id": {
                "type":
                    "string",
            },

            "sentence_annotations": {
                "type":
                    "array",

                "minItems":
                    expected_sentence_count,

                "maxItems":
                    expected_sentence_count,

                "items": {
                    "type":
                        "object",

                    "properties": {
                        "sentence_id": {
                            "type":
                                "string",

                            "enum":
                                expected_sentence_ids,
                        },

                        "annotation_json": {
                            "type":
                                "string",
                        },
                    },

                    "required": [
                        "sentence_id",
                        "annotation_json",
                    ],

                    "additionalProperties":
                        False,
                },
            },
        },

        "required": [
            "media_ai_record_id",
            "sentence_annotations",
        ],

        "additionalProperties":
            False,
    }


# -----------------------------------------------------------------------------
# Provider-independent article input
# -----------------------------------------------------------------------------

def build_supervisor_input_text(
    supervision_payload,
):
    """
    Convert one validated article supervision payload into provider-neutral
    text.
    """

    identity = supervision_payload.get(
        "identity",
        {}
    )


    article = supervision_payload.get(
        "article",
        {}
    )


    sentence_units = article.get(
        "sentence_units",
        []
    )


    ordered_sentence_units = [
        {
            "paragraph_id":
                unit.get(
                    "paragraph_id"
                ),

            "sentence_id":
                unit.get(
                    "sentence_id"
                ),

            "paragraph_index":
                unit.get(
                    "paragraph_index"
                ),

            "sentence_index":
                unit.get(
                    "sentence_index"
                ),

            "text":
                unit.get(
                    "current_sentence"
                ),
        }

        for unit
        in sentence_units
    ]


    article_sequence_json = json.dumps(
        ordered_sentence_units,
        ensure_ascii=False,
        indent=2,
    )


    return (
        f"{LLM_SUPERVISION_INSTRUCTIONS}\n\n"

        "MEDIA AI ARTICLE SUPERVISION UNIT\n\n"

        f"Media AI record ID: "
        f"{identity.get('media_ai_record_id')}\n\n"

        f"Article title:\n"
        f"{article.get('article_title')}\n\n"

        f"Expected sentence count: "
        f"{article.get('sentence_count')}\n\n"

        "Ordered article sentence sequence:\n"
        f"{article_sequence_json}\n\n"

        "Return exactly one sentence-level annotation for every supplied "
        "sentence identifier, in the same order.\n\n"

        "Return the Media AI record ID exactly as supplied.\n\n"

        "For each sentence, return the COMPLETE Media AI sentence annotation "
        "inside `annotation_json` as valid JSON text.\n\n"

        "The JSON object inside `annotation_json` must contain exactly these "
        "top-level fields:\n"

        "reported_emotion, transformative_emotion, "
        "immediate_emotional_state, emotional_weight, "
        "factological_weight, factological_components, "
        "annotation_confidence, evidence.\n\n"

        "Follow the complete Media AI supervision instructions above for "
        "all emotional dimensions, value ranges, value states, factual "
        "components and evidence requirements.\n\n"

        "Return only the structured article response required by the "
        "provider response schema."
    )


# -----------------------------------------------------------------------------
# Normalise provider envelope
# -----------------------------------------------------------------------------

def normalise_provider_article_output(
    provider_output,
    supervision_payload,
):
    """
    Decode the annotation_json strings and convert the provider envelope into
    the downstream article structure expected by Block 7.
    """

    if not isinstance(
        provider_output,
        dict,
    ):

        raise TypeError(
            "Gemini article output must decode to a dictionary."
        )


    media_ai_record_id = provider_output.get(
        "media_ai_record_id"
    )


    sentence_annotations = provider_output.get(
        "sentence_annotations"
    )


    if not isinstance(
        sentence_annotations,
        list,
    ):

        raise TypeError(
            "Gemini article output does not contain a valid "
            "sentence_annotations list."
        )


    expected_sentence_ids = (
        supervision_payload.get(
            "article",
            {}
        ).get(
            "expected_sentence_ids",
            [],
        )
    )


    normalised_annotations = []


    for item in sentence_annotations:

        if not isinstance(
            item,
            dict,
        ):

            raise TypeError(
                "Provider sentence annotation item is not a dictionary."
            )


        sentence_id = item.get(
            "sentence_id"
        )


        annotation_json = item.get(
            "annotation_json"
        )


        if not isinstance(
            annotation_json,
            str,
        ):

            raise TypeError(
                "Provider annotation_json must be a string."
            )


        if not annotation_json.strip():

            raise ValueError(
                "Provider annotation_json cannot be empty."
            )


        try:

            annotation = json.loads(
                annotation_json
            )


        except json.JSONDecodeError as exc:

            raise ValueError(
                "One provider annotation_json value could not "
                "be decoded as JSON."
            ) from exc


        if not isinstance(
            annotation,
            dict,
        ):

            raise TypeError(
                "Decoded sentence annotation must be a dictionary."
            )


        normalised_annotations.append(
            {
                "sentence_id":
                    sentence_id,

                "annotation":
                    annotation,
            }
        )


    returned_sentence_ids = [
        item.get(
            "sentence_id"
        )
        for item
        in normalised_annotations
    ]


    if (
        len(
            returned_sentence_ids
        )
        != len(
            expected_sentence_ids
        )
    ):

        raise ValueError(
            "Provider returned an unexpected number of sentence annotations."
        )


    if (
        set(
            returned_sentence_ids
        )
        != set(
            expected_sentence_ids
        )
    ):

        raise ValueError(
            "Provider sentence identifiers do not match the requested "
            "article sentence identifiers."
        )


    return {
        "media_ai_record_id":
            media_ai_record_id,

        "sentence_annotations":
            normalised_annotations,
    }


# =============================================================================
# Gemini article-level supervisor execution
# =============================================================================


def execute_gemini_supervision(
    supervisor_id,
    supervisor_configuration,
    supervision_payload,
):
    """
    Execute one complete-article Gemini supervision request.

    Provider-side structured output uses a shallow envelope.

    The complete sentence annotation is returned as JSON text and decoded
    locally before Block 7 performs canonical validation.
    """

    if not GOOGLE_GENAI_AVAILABLE:

        return {
            "executed":
                False,

            "status":
                "BLOCKED",

            "reason":
                "The google-genai Python SDK is unavailable.",
        }


    if not GEMINI_CREDENTIAL_READY:

        return {
            "executed":
                False,

            "status":
                "BLOCKED",

            "reason":
                "GEMINI_API_KEY is unavailable in Colab Secrets.",
        }


    model_name = (
        supervisor_configuration.get(
            "model"
        )
    )


    if not model_name:

        return {
            "executed":
                False,

            "status":
                "BLOCKED",

            "reason":
                "No Gemini supervisor model is configured.",
        }


    if (
        REQUIRE_FREE_SUPERVISION
        and model_name
        not in FREE_TIER_APPROVED_SUPERVISOR_MODELS
    ):

        return {
            "executed":
                False,

            "status":
                "BLOCKED",

            "reason":
                "Configured Gemini model is not approved "
                "by the research Free Tier allow-list.",
        }


    if ALLOW_PAID_API_USAGE:

        return {
            "executed":
                False,

            "status":
                "BLOCKED",

            "reason":
                "Paid API usage is prohibited by the current "
                "research configuration.",
        }


    if ALLOW_AUTOMATIC_PAID_FALLBACK:

        return {
            "executed":
                False,

            "status":
                "BLOCKED",

            "reason":
                "Automatic paid fallback is prohibited by the "
                "current research configuration.",
        }


    media_ai_record_id = (
        supervision_payload.get(
            "identity",
            {}
        ).get(
            "media_ai_record_id"
        )
    )


    if not media_ai_record_id:

        return {
            "executed":
                False,

            "status":
                "BLOCKED",

            "reason":
                "Article supervision payload has no Media AI record ID.",
        }


    allowed, reason = (
        reserve_supervision_request()
    )


    if not allowed:

        return {
            "executed":
                False,

            "status":
                "BLOCKED",

            "reason":
                reason,
        }


    client = genai.Client(
        api_key=
            GEMINI_API_KEY
    )


    user_input = (
        build_supervisor_input_text(
            supervision_payload
        )
    )


    article_output_schema = (
        build_article_output_json_schema(
            supervision_payload
        )
    )


    try:

        interaction = (
            client.interactions.create(
                model=
                    model_name,

                input=
                    user_input,

                response_format={
                    "type":
                        "text",

                    "mime_type":
                        "application/json",

                    "schema":
                        article_output_schema,
                },
            )
        )


        raw_output_text = (
            interaction.output_text
        )


        if not isinstance(
            raw_output_text,
            str,
        ):

            raise TypeError(
                "Gemini interaction did not return textual output."
            )


        if not raw_output_text.strip():

            raise ValueError(
                "Gemini interaction returned an empty structured response."
            )


        try:

            provider_parsed_output = json.loads(
                raw_output_text
            )


        except json.JSONDecodeError as exc:

            raise ValueError(
                "Gemini structured response could not be decoded as JSON."
            ) from exc


        parsed_output = (
            normalise_provider_article_output(
                provider_output=
                    provider_parsed_output,

                supervision_payload=
                    supervision_payload,
            )
        )


        register_supervision_result(
            success=
                True,

            supervisor_id=
                supervisor_id,

            media_ai_record_id=
                media_ai_record_id,
        )


        return {
            "executed":
                True,

            "status":
                "SUCCESS",

            "reason":
                "Gemini article supervision request completed successfully.",

            "provider":
                "google",

            "model":
                model_name,

            "provider_response_id":
                getattr(
                    interaction,
                    "id",
                    None,
                ),

            "provider_schema_version":
                PROVIDER_SCHEMA_VERSION,

            "provider_schema_strategy":
                PROVIDER_SCHEMA_STRATEGY,

            "raw_output_text":
                raw_output_text,

            "provider_parsed_output":
                provider_parsed_output,

            "parsed_output":
                parsed_output,

            "article_sentence_count":
                supervision_payload.get(
                    "article",
                    {}
                ).get(
                    "sentence_count"
                ),
        }


    except Exception as exc:

        register_supervision_result(
            success=
                False,

            supervisor_id=
                supervisor_id,

            media_ai_record_id=
                media_ai_record_id,
        )


        return {
            "executed":
                True,

            "status":
                "ERROR",

            "reason":
                str(
                    exc
                ),

            "error_type":
                type(
                    exc
                ).__name__,

            "provider":
                "google",

            "model":
                model_name,

            "provider_schema_version":
                PROVIDER_SCHEMA_VERSION,

            "provider_schema_strategy":
                PROVIDER_SCHEMA_STRATEGY,

            "article_sentence_count":
                supervision_payload.get(
                    "article",
                    {}
                ).get(
                    "sentence_count"
                ),
        }


# -----------------------------------------------------------------------------
# Provider routing
# -----------------------------------------------------------------------------

def execute_supervisor_request(
    supervisor_id,
    supervisor_configuration,
    supervision_payload,
):
    """
    Route one complete-article request to its configured provider.
    """

    provider = (
        supervisor_configuration.get(
            "provider"
        )
    )


    if provider == "google":

        return execute_gemini_supervision(
            supervisor_id=
                supervisor_id,

            supervisor_configuration=
                supervisor_configuration,

            supervision_payload=
                supervision_payload,
        )


    return {
        "executed":
            False,

        "status":
            "BLOCKED",

        "reason":
            "Configured supervisor provider is not implemented.",
    }


# =============================================================================
# Selection diagnostics
# =============================================================================


SUPERVISION_SELECTION_SUMMARY = {
    supervisor_id: {
        "validated_articles":
            len(
                LLM_USAGE_LEDGER[
                    "validated_article_ids_by_supervisor"
                ].get(
                    supervisor_id,
                    [],
                )
            ),

        "pending_articles":
            len(
                LLM_USAGE_LEDGER[
                    "pending_article_ids_by_supervisor"
                ].get(
                    supervisor_id,
                    [],
                )
            ),

        "persisted_successes":
            sum(
                1
                for (
                    stored_supervisor_id,
                    _
                )
                in PERSISTED_SUCCESSFUL_ARTICLE_PAIRS
                if stored_supervisor_id
                == supervisor_id
            ),

        "eligible_unsupervised":
            0,

        "selected_this_run":
            0,

        "selected_sentence_count":
            0,
    }

    for supervisor_id
    in SUPERVISOR_REGISTRY
}


# =============================================================================
# Controlled execution with persistent resume protection
# =============================================================================


RUN_REQUESTS_ATTEMPTED = 0

RUN_SENTENCES_INCLUDED = 0


if RUN_LLM_SUPERVISION:

    remaining_run_capacity = min(
        MAX_SUPERVISION_REQUESTS_PER_RUN,
        remaining_daily_supervision_capacity(),
    )


    for (
        supervisor_id,
        configuration,
    ) in SUPERVISORS_ALLOWED_TO_EXECUTE.items():

        if remaining_run_capacity <= 0:

            break


        validated_article_ids = set(
            LLM_USAGE_LEDGER[
                "validated_article_ids_by_supervisor"
            ].get(
                supervisor_id,
                [],
            )
        )


        pending_article_ids = set(
            LLM_USAGE_LEDGER[
                "pending_article_ids_by_supervisor"
            ].get(
                supervisor_id,
                [],
            )
        )


        persisted_successful_article_ids = {
            media_ai_record_id

            for (
                stored_supervisor_id,
                media_ai_record_id
            )
            in PERSISTED_SUCCESSFUL_ARTICLE_PAIRS

            if (
                stored_supervisor_id
                == supervisor_id
                and media_ai_record_id
            )
        }


        already_processed_article_ids = (
            validated_article_ids
            | pending_article_ids
            | persisted_successful_article_ids
        )


        eligible_payloads = [
            payload

            for payload
            in LLM_ARTICLE_SUPERVISION_PAYLOADS

            if payload.get(
                "identity",
                {}
            ).get(
                "media_ai_record_id"
            )
            not in already_processed_article_ids
        ]


        SUPERVISION_SELECTION_SUMMARY[
            supervisor_id
        ][
            "eligible_unsupervised"
        ] = len(
            eligible_payloads
        )


        requested_payloads = (
            eligible_payloads[
                :remaining_run_capacity
            ]
        )


        SUPERVISION_SELECTION_SUMMARY[
            supervisor_id
        ][
            "selected_this_run"
        ] = len(
            requested_payloads
        )


        SUPERVISION_SELECTION_SUMMARY[
            supervisor_id
        ][
            "selected_sentence_count"
        ] = sum(
            payload.get(
                "article",
                {}
            ).get(
                "sentence_count",
                0,
            )

            for payload
            in requested_payloads
        )


        for payload in requested_payloads:

            if remaining_run_capacity <= 0:

                break


            request_started_at = (
                datetime.now(
                    timezone.utc
                ).isoformat()
            )


            execution_result = (
                execute_supervisor_request(
                    supervisor_id=
                        supervisor_id,

                    supervisor_configuration=
                        configuration,

                    supervision_payload=
                        payload,
                )
            )


            raw_result = {
                "supervisor_id":
                    supervisor_id,

                "provider":
                    configuration.get(
                        "provider"
                    ),

                "model":
                    configuration.get(
                        "model"
                    ),

                "request_unit":
                    "article",

                "annotation_unit":
                    "sentence",

                "requested_at_utc":
                    request_started_at,

                "identity":
                    deepcopy(
                        payload.get(
                            "identity",
                            {}
                        )
                    ),

                "article_metadata": {
                    "article_title":
                        payload.get(
                            "article",
                            {}
                        ).get(
                            "article_title"
                        ),

                    "sentence_count":
                        payload.get(
                            "article",
                            {}
                        ).get(
                            "sentence_count"
                        ),

                    "expected_sentence_ids":
                        deepcopy(
                            payload.get(
                                "article",
                                {}
                            ).get(
                                "expected_sentence_ids",
                                [],
                            )
                        ),
                },

                "protocol_version":
                    LLM_SUPERVISION_PROTOCOL_VERSION,

                "supervision_schema_version":
                    SUPERVISION_SCHEMA_VERSION,

                "annotation_policy_version":
                    ANNOTATION_POLICY_VERSION,

                "provider_schema_version":
                    PROVIDER_SCHEMA_VERSION,

                "provider_schema_strategy":
                    PROVIDER_SCHEMA_STRATEGY,

                "execution":
                    execution_result,
            }


            # -----------------------------------------------------------------
            # Current runtime
            # -----------------------------------------------------------------

            LLM_SUPERVISION_RAW_RESULTS[
                supervisor_id
            ].append(
                raw_result
            )


            # -----------------------------------------------------------------
            # Persistent storage immediately after execution
            # -----------------------------------------------------------------

            persist_raw_supervision_result(
                raw_result
            )


            # -----------------------------------------------------------------
            # Runtime request accounting
            # -----------------------------------------------------------------

            if execution_result.get(
                "executed"
            ):

                RUN_REQUESTS_ATTEMPTED += 1


                RUN_SENTENCES_INCLUDED += (
                    payload.get(
                        "article",
                        {}
                    ).get(
                        "sentence_count",
                        0,
                    )
                )


                remaining_run_capacity -= 1


# =============================================================================
# Post-execution persistent state
# =============================================================================


FINAL_RAW_RESULT_COUNT = sum(
    len(
        results
    )
    for results
    in RAW_SUPERVISION_STORE[
        "results_by_supervisor"
    ].values()
)


FINAL_PERSISTED_SUCCESSFUL_ARTICLE_PAIRS = (
    get_successful_persisted_article_pairs()
)


# =============================================================================
# Execution diagnostics
# =============================================================================


SUPERVISION_EXECUTION_SUMMARY = {}


for supervisor_id in SUPERVISOR_REGISTRY:

    results = (
        LLM_SUPERVISION_RAW_RESULTS[
            supervisor_id
        ]
    )


    successful = sum(
        1

        for result
        in results

        if result.get(
            "execution",
            {}
        ).get(
            "status"
        )
        == "SUCCESS"
    )


    errors = sum(
        1

        for result
        in results

        if result.get(
            "execution",
            {}
        ).get(
            "status"
        )
        == "ERROR"
    )


    blocked = sum(
        1

        for result
        in results

        if result.get(
            "execution",
            {}
        ).get(
            "status"
        )
        == "BLOCKED"
    )


    sentence_annotations_requested = sum(
        result.get(
            "article_metadata",
            {}
        ).get(
            "sentence_count",
            0,
        )

        for result
        in results
    )


    SUPERVISION_EXECUTION_SUMMARY[
        supervisor_id
    ] = {
        "requests_recorded":
            len(
                results
            ),

        "successful":
            successful,

        "errors":
            errors,

        "blocked":
            blocked,

        "sentence_annotations_requested":
            sentence_annotations_requested,
    }


# -----------------------------------------------------------------------------
# Final usage state
# -----------------------------------------------------------------------------

FINAL_LLM_USAGE = deepcopy(
    LLM_USAGE_LEDGER
)


FINAL_DAILY_CAPACITY = (
    remaining_daily_supervision_capacity()
)


# =============================================================================
# Verification
# =============================================================================


print("=" * 72)

print(
    "Media AI — Controlled LLM Supervisor Execution"
)

print("=" * 72)


print(
    f"Registry version           : "
    f"{SUPERVISOR_REGISTRY_VERSION}"
)

print(
    f"Provider schema version    : "
    f"{PROVIDER_SCHEMA_VERSION}"
)

print(
    f"Provider schema strategy   : "
    f"{PROVIDER_SCHEMA_STRATEGY}"
)

print(
    f"Master supervision switch  : "
    f"{RUN_LLM_SUPERVISION}"
)

print(
    f"External request unit      : "
    f"{LLM_SUPERVISION_REQUEST_UNIT}"
)

print(
    f"Annotation unit            : "
    f"{LLM_SUPERVISION_ANNOTATION_UNIT}"
)

print(
    f"Article payloads available : "
    f"{len(LLM_ARTICLE_SUPERVISION_PAYLOADS)}"
)

print(
    f"Configured supervisor slots: "
    f"{len(SUPERVISOR_REGISTRY)}"
)

print(
    f"Enabled supervisor slots   : "
    f"{len(ENABLED_SUPERVISORS)}"
)

print(
    f"Executable supervisors     : "
    f"{len(SUPERVISORS_ALLOWED_TO_EXECUTE)}"
)


print("-" * 72)

print(
    "Structured-output boundary"
)

print(
    f"Full canonical schema local: "
    f"{True}"
)

print(
    f"Full canonical schema sent : "
    f"{False}"
)

print(
    f"Provider envelope shallow  : "
    f"{True}"
)

print(
    f"Annotation JSON local parse : "
    f"{True}"
)

print(
    f"Canonical validation later : "
    f"{True}"
)


print("-" * 72)

print(
    "Persistent raw supervision"
)

print(
    f"Raw store version          : "
    f"{RAW_SUPERVISION_STORE_VERSION}"
)

print(
    f"Raw store filename         : "
    f"{RAW_SUPERVISION_STORE_FILENAME}"
)

print(
    f"Raw store file ID          : "
    f"{RAW_SUPERVISION_STORE_FILE_ID}"
)

print(
    f"Recovered raw results      : "
    f"{RECOVERED_RAW_RESULT_COUNT}"
)

print(
    f"Recovered successes        : "
    f"{len(PERSISTED_SUCCESSFUL_ARTICLE_PAIRS)}"
)

print(
    f"Orphaned pending released  : "
    f"{len(ORPHANED_PENDING_ARTICLES_RELEASED)}"
)

print(
    f"Final raw result count     : "
    f"{FINAL_RAW_RESULT_COUNT}"
)

print(
    f"Final persisted successes  : "
    f"{len(FINAL_PERSISTED_SUCCESSFUL_ARTICLE_PAIRS)}"
)


print("-" * 72)

print(
    "Cost and article-request controls"
)

print(
    f"Require free supervision   : "
    f"{REQUIRE_FREE_SUPERVISION}"
)

print(
    f"Paid API usage allowed     : "
    f"{ALLOW_PAID_API_USAGE}"
)

print(
    f"Paid fallback allowed      : "
    f"{ALLOW_AUTOMATIC_PAID_FALLBACK}"
)

print(
    f"Max article requests / run : "
    f"{MAX_SUPERVISION_REQUESTS_PER_RUN}"
)

print(
    f"Max article requests / day : "
    f"{MAX_SUPERVISION_REQUESTS_PER_DAY}"
)

print(
    f"Article requests this run  : "
    f"{RUN_REQUESTS_ATTEMPTED}"
)

print(
    f"Sentences included this run: "
    f"{RUN_SENTENCES_INCLUDED}"
)

print(
    f"Article requests reserved  : "
    f"{FINAL_LLM_USAGE['article_requests_reserved']}"
)

print(
    f"Article requests completed : "
    f"{FINAL_LLM_USAGE['article_requests_completed']}"
)

print(
    f"Provider-successful        : "
    f"{FINAL_LLM_USAGE['article_requests_successful']}"
)

print(
    f"Provider-failed            : "
    f"{FINAL_LLM_USAGE['article_requests_failed']}"
)

print(
    f"Remaining article capacity : "
    f"{FINAL_DAILY_CAPACITY}"
)


print("-" * 72)

print(
    "Article-level resume protection"
)

print(
    f"Validated IDs imported     : "
    f"{RUNTIME_ARTICLE_VALIDATION_SYNC['validated_imported']}"
)

print(
    f"Rejected pending released  : "
    f"{RUNTIME_ARTICLE_VALIDATION_SYNC['rejected_released']}"
)


for supervisor_id in sorted(
    SUPERVISOR_REGISTRY
):

    selection = (
        SUPERVISION_SELECTION_SUMMARY[
            supervisor_id
        ]
    )


    print(
        f"{supervisor_id}"
    )

    print(
        f"  Validated articles       : "
        f"{selection['validated_articles']}"
    )

    print(
        f"  Pending articles         : "
        f"{selection['pending_articles']}"
    )

    print(
        f"  Persisted successes      : "
        f"{selection['persisted_successes']}"
    )

    print(
        f"  Eligible unsupervised    : "
        f"{selection['eligible_unsupervised']}"
    )

    print(
        f"  Selected this run        : "
        f"{selection['selected_this_run']}"
    )

    print(
        f"  Sentences selected       : "
        f"{selection['selected_sentence_count']}"
    )


print("-" * 72)

print(
    "Gemini runtime"
)

print(
    f"Google GenAI SDK available : "
    f"{GOOGLE_GENAI_AVAILABLE}"
)

print(
    f"Gemini credential ready    : "
    f"{GEMINI_CREDENTIAL_READY}"
)

print(
    f"Gemini secret status       : "
    f"{GEMINI_SECRET_STATUS}"
)


print("-" * 72)


for supervisor_id in sorted(
    SUPERVISOR_REGISTRY
):

    configuration = (
        SUPERVISOR_REGISTRY[
            supervisor_id
        ]
    )


    readiness = (
        SUPERVISOR_READINESS[
            supervisor_id
        ]
    )


    execution_summary = (
        SUPERVISION_EXECUTION_SUMMARY[
            supervisor_id
        ]
    )


    print(
        f"{supervisor_id}"
    )

    print(
        f"  Enabled                  : "
        f"{configuration['enabled']}"
    )

    print(
        f"  Provider                 : "
        f"{configuration['provider']}"
    )

    print(
        f"  Model                    : "
        f"{configuration['model']}"
    )

    print(
        f"  Configured               : "
        f"{readiness['configured']}"
    )

    print(
        f"  Implementation           : "
        f"{configuration['implementation_status']}"
    )

    print(
        f"  Credential ready         : "
        f"{readiness['credential_ready']}"
    )

    print(
        f"  SDK ready                : "
        f"{readiness['sdk_ready']}"
    )

    print(
        f"  Free Tier approved       : "
        f"{readiness['free_tier_approved']}"
    )

    print(
        f"  Cost policy ready        : "
        f"{readiness['cost_policy_ready']}"
    )

    print(
        f"  Executable               : "
        f"{readiness['executable']}"
    )

    print(
        f"  Raw results in runtime   : "
        f"{execution_summary['requests_recorded']}"
    )

    print(
        f"  Successful               : "
        f"{execution_summary['successful']}"
    )

    print(
        f"  Errors                   : "
        f"{execution_summary['errors']}"
    )

    print(
        f"  Blocked                  : "
        f"{execution_summary['blocked']}"
    )

    print(
        f"  Sentence targets         : "
        f"{execution_summary['sentence_annotations_requested']}"
    )

    print("-" * 72)


print(
    "Complete articles remain the external LLM request unit."
)

print(
    "Sentence-level annotations remain the Media AI supervision unit."
)

print(
    "Provider-side structured output uses only a lightweight article "
    "and sentence envelope."
)

print(
    "Complete Media AI sentence annotations are returned as JSON text "
    "and decoded locally."
)

print(
    "The full canonical annotation schema remains a local validation contract."
)

print(
    "Raw supervision responses are persisted independently of the "
    "Colab runtime."
)

print(
    "Provider-successful article responses can therefore be recovered "
    "after a runtime restart."
)

print(
    "Persisted provider-successful article responses are excluded from "
    "later requests even if the daily usage ledger has reset."
)


if ORPHANED_PENDING_ARTICLES_RELEASED:

    print(
        "Historical pending article-supervisor pairs without recoverable "
        "successful raw responses were released for controlled retry."
    )


print(
    "Provider-successful articles remain pending until Block 7 validates "
    "complete sentence-level coverage."
)

print(
    "Validated article-supervisor pairs remain excluded from later requests."
)

print(
    "Earlier sentence-level API usage remains retained separately for audit."
)

print(
    "Raw supervisor outputs remain separated by supervisor identity."
)


if RUN_LLM_SUPERVISION:

    print(
        "Controlled Gemini supervision was permitted only for eligible "
        "article-level request units."
    )

else:

    print(
        "LLM supervision was skipped by the Execution Control Panel."
    )


print(
    "No canonical sentence annotation has been accepted by this block."
)

print(
    "No supervisor comparison or consensus has been performed."
)

print("=" * 72)
```

## Raw Supervision Parsing and Canonical Annotation Validation

This block converts successful raw LLM supervision responses into canonical
Media AI annotation records.

It establishes the formal acceptance boundary between an external supervisor and
the research dataset.

A successful provider request does **not** automatically constitute an accepted
annotation.

The processing sequence is:

**raw provider response**

→ **structured response extraction**

→ **canonical annotation construction**

→ **schema and range validation**

→ **accepted supervision record or rejected diagnostic record**

### Canonicalisation

Provider-specific response structures are not propagated into the reference
dataset.

For each successful supervision response, this block creates a canonical Media
AI supervision record using the annotation template defined in Block 4.

The canonical record preserves:

- Media AI record identifier;
- paragraph and sentence identifiers;
- sentence position;
- supplied supervision context;
- reported-emotion vector;
- transformative-emotion vector;
- immediate-emotional-state vector;
- emotional weight;
- factological weight;
- factological component indicators;
- annotation confidence;
- evidence;
- supervisor identity;
- provider and model information;
- supervision and annotation schema versions;
- annotation timestamp;
- review state.

### Validation

Every candidate annotation is validated before acceptance.

Validation checks include:

- record identity;
- schema versions;
- all eleven emotional dimensions;
- `[0,1]` reported-emotion values;
- `[-1,1]` transformative-emotion values;
- `[0,1]` immediate-state values;
- emotional-weight range;
- factological-weight range;
- annotation-confidence range;
- explicit missing-information states;
- all eight factological components;
- evidence structure;
- supervision provenance;
- supervisor identity.

A value marked `observed` must contain a numeric value within the corresponding
range.

Values marked `insufficient_evidence` or `not_applicable` must contain `null`
rather than a numeric substitute.

### Rejection Policy

Invalid annotations are not repaired silently.

A rejected response remains available in the validation diagnostics together
with the reasons for rejection.

This permits later:

- controlled retry;
- supervisor replacement;
- human review;
- protocol debugging.

Rejected annotations must not enter the reference dataset.

### Independent Supervisor Preservation

Accepted records remain grouped by stable supervisor identifier.

No averaging, consensus, voting or adjudication is performed in this block.

This preserves the independence required for later inter-supervisor comparison.

### Review Status

Machine-generated annotations enter the canonical store with:

`review_status = "unreviewed"`

Low-confidence or otherwise questionable records may later be routed to review,
but Block 7 does not adjudicate them.

### Disabled-Supervision Behaviour

When `RUN_LLM_SUPERVISION = False`, Block 6 produces no raw supervision
responses.

Block 7 must therefore complete successfully with:

- zero raw responses;
- zero accepted annotations;
- zero rejected annotations.

An empty result is a valid execution state and does not indicate a pipeline
failure.

### Output

This block creates:

- `CANONICAL_SUPERVISION_RECORDS_BY_SUPERVISOR`;
- `ACCEPTED_SUPERVISION_RECORDS`;
- `REJECTED_SUPERVISION_RECORDS`;
- `SUPERVISION_VALIDATION_RESULTS`;
- corpus-level validation diagnostics.

No external model or API call is performed by this block.

```python
# =============================================================================
# Media AI — Raw Supervision Parsing and Canonical Annotation Validation
# Block 7 — Article-Level Response Expansion and Sentence-Level Validation
# =============================================================================

from copy import deepcopy
from datetime import datetime, timezone


# -----------------------------------------------------------------------------
# Block configuration
# -----------------------------------------------------------------------------

CANONICAL_SUPERVISION_VALIDATION_VERSION = "1.2"

EXPECTED_REQUEST_UNIT = "article"

EXPECTED_ANNOTATION_UNIT = "sentence"

EVIDENCE_NORMALISATION_POLICY = (
    "provider_string_to_supporting_text"
)


# -----------------------------------------------------------------------------
# Block dependency checks
# -----------------------------------------------------------------------------

required_block_4_objects = [
    "ANNOTATION_RECORD_TEMPLATE",
    "ANNOTATION_RECORD_SCHEMA_VERSION",
    "create_blank_annotation_record",
    "validate_annotation_record",
]


missing_block_4_objects = [
    object_name
    for object_name in required_block_4_objects
    if object_name not in globals()
]


if missing_block_4_objects:
    raise NameError(
        "Block 4 annotation-record objects are not initialised. "
        "Rerun Notebook 03 Block 4. Missing: "
        f"{missing_block_4_objects}"
    )


required_block_6_objects = [
    "SUPERVISOR_REGISTRY",
    "LLM_SUPERVISION_RAW_RESULTS",
    "LLM_SUPERVISION_PROTOCOL_VERSION",
    "SUPERVISION_SCHEMA_VERSION",
    "ANNOTATION_POLICY_VERSION",
]


missing_block_6_objects = [
    object_name
    for object_name in required_block_6_objects
    if object_name not in globals()
]


if missing_block_6_objects:
    raise NameError(
        "Block 6 supervision-execution objects are not initialised. "
        "Rerun Notebook 03 Block 6. Missing: "
        f"{missing_block_6_objects}"
    )


if "SUPERVISION_CONTEXT_RECORDS" not in globals():
    raise NameError(
        "SUPERVISION_CONTEXT_RECORDS is unavailable. "
        "Rerun the Notebook 03 context-construction block."
    )


# -----------------------------------------------------------------------------
# Context lookup
# -----------------------------------------------------------------------------

SUPERVISION_CONTEXT_BY_SENTENCE_ID = {
    context_record["sentence_id"]:
        context_record

    for context_record
    in SUPERVISION_CONTEXT_RECORDS
}


# -----------------------------------------------------------------------------
# Canonical sentence annotation structural validation
# -----------------------------------------------------------------------------

def validate_provider_annotation_structure(
    provider_annotation,
):
    """
    Verify that one parsed sentence annotation contains the expected
    top-level Media AI supervision fields.

    Full canonical validation is performed later.
    """

    errors = []


    if not isinstance(
        provider_annotation,
        dict,
    ):
        return (
            False,
            [
                "provider_annotation_not_dictionary"
            ],
        )


    required_fields = {
        "reported_emotion",
        "transformative_emotion",
        "immediate_emotional_state",
        "emotional_weight",
        "factological_weight",
        "factological_components",
        "annotation_confidence",
        "evidence",
    }


    output_fields = set(
        provider_annotation.keys()
    )


    missing_fields = (
        required_fields
        - output_fields
    )


    unexpected_fields = (
        output_fields
        - required_fields
    )


    if missing_fields:

        errors.append(
            "missing_provider_fields:"
            + ",".join(
                sorted(
                    missing_fields
                )
            )
        )


    if unexpected_fields:

        errors.append(
            "unexpected_provider_fields:"
            + ",".join(
                sorted(
                    unexpected_fields
                )
            )
        )


    return (
        len(errors) == 0,
        errors,
    )


# -----------------------------------------------------------------------------
# Evidence normalisation
# -----------------------------------------------------------------------------

def normalise_provider_evidence(
    provider_evidence,
):
    """
    Convert provider evidence into the canonical Media AI evidence structure.

    Canonical evidence structure:
        {
            "summary": str | None,
            "supporting_text": list[str]
        }

    A plain provider string is preserved as supporting text.
    No summary is invented.

    Returns:
        canonical_evidence, errors
    """

    if isinstance(
        provider_evidence,
        dict,
    ):

        return (
            deepcopy(
                provider_evidence
            ),
            [],
        )


    if isinstance(
        provider_evidence,
        str,
    ):

        return (
            {
                "summary":
                    None,

                "supporting_text": [
                    provider_evidence
                ],
            },
            [],
        )


    if provider_evidence is None:

        return (
            {
                "summary":
                    None,

                "supporting_text":
                    [],
            },
            [],
        )


    return (
        None,
        [
            "unsupported_evidence_structure"
        ],
    )


# -----------------------------------------------------------------------------
# Article-level response validation
# -----------------------------------------------------------------------------

def validate_article_response_envelope(
    raw_result,
):
    """
    Validate the successful Block 6 article-level response envelope.

    This checks transport-level and article-level structure only.

    Sentence annotation semantics are validated independently afterward.
    """

    errors = []


    if not isinstance(
        raw_result,
        dict,
    ):
        return (
            False,
            [
                "raw_result_not_dictionary"
            ],
        )


    if (
        raw_result.get(
            "request_unit"
        )
        != EXPECTED_REQUEST_UNIT
    ):

        errors.append(
            "invalid_request_unit"
        )


    if (
        raw_result.get(
            "annotation_unit"
        )
        != EXPECTED_ANNOTATION_UNIT
    ):

        errors.append(
            "invalid_annotation_unit"
        )


    execution = raw_result.get(
        "execution",
        {}
    )


    if not isinstance(
        execution,
        dict,
    ):

        errors.append(
            "execution_not_dictionary"
        )

        return (
            False,
            errors,
        )


    if (
        execution.get(
            "status"
        )
        != "SUCCESS"
    ):

        errors.append(
            "provider_request_not_successful"
        )

        return (
            False,
            errors,
        )


    parsed_output = execution.get(
        "parsed_output"
    )


    if not isinstance(
        parsed_output,
        dict,
    ):

        errors.append(
            "parsed_output_not_dictionary"
        )

        return (
            False,
            errors,
        )


    media_ai_record_id = (
        parsed_output.get(
            "media_ai_record_id"
        )
    )


    sentence_annotations = (
        parsed_output.get(
            "sentence_annotations"
        )
    )


    if not media_ai_record_id:

        errors.append(
            "missing_media_ai_record_id"
        )


    if not isinstance(
        sentence_annotations,
        list,
    ):

        errors.append(
            "sentence_annotations_not_list"
        )


    return (
        len(errors) == 0,
        errors,
    )


# -----------------------------------------------------------------------------
# Article-level completeness validation
# -----------------------------------------------------------------------------

def validate_article_sentence_coverage(
    raw_result,
):
    """
    Validate complete sentence-level coverage for one article response.

    Requirements:
        - returned article ID matches requested article ID;
        - returned sentence count matches expected sentence count;
        - sentence IDs match exactly;
        - sentence IDs are unique;
        - sentence ordering matches the requested article order.
    """

    errors = []


    identity = raw_result.get(
        "identity",
        {}
    )


    expected_media_ai_record_id = (
        identity.get(
            "media_ai_record_id"
        )
    )


    article_metadata = raw_result.get(
        "article_metadata",
        {}
    )


    expected_sentence_ids = list(
        article_metadata.get(
            "expected_sentence_ids",
            [],
        )
    )


    expected_sentence_count = (
        article_metadata.get(
            "sentence_count"
        )
    )


    parsed_output = (
        raw_result.get(
            "execution",
            {}
        ).get(
            "parsed_output",
            {},
        )
    )


    returned_media_ai_record_id = (
        parsed_output.get(
            "media_ai_record_id"
        )
    )


    sentence_annotations = (
        parsed_output.get(
            "sentence_annotations",
            [],
        )
    )


    if (
        returned_media_ai_record_id
        != expected_media_ai_record_id
    ):

        errors.append(
            "media_ai_record_id_mismatch"
        )


    if not isinstance(
        sentence_annotations,
        list,
    ):

        errors.append(
            "sentence_annotations_not_list"
        )

        return (
            False,
            errors,
        )


    returned_sentence_ids = [
        item.get(
            "sentence_id"
        )
        if isinstance(
            item,
            dict,
        )
        else None

        for item
        in sentence_annotations
    ]


    if (
        expected_sentence_count
        is not None
        and len(
            sentence_annotations
        )
        != expected_sentence_count
    ):

        errors.append(
            "sentence_count_mismatch"
        )


    if len(
        returned_sentence_ids
    ) != len(
        set(
            returned_sentence_ids
        )
    ):

        errors.append(
            "duplicate_sentence_ids"
        )


    if set(
        returned_sentence_ids
    ) != set(
        expected_sentence_ids
    ):

        errors.append(
            "sentence_id_set_mismatch"
        )


    if (
        returned_sentence_ids
        != expected_sentence_ids
    ):

        errors.append(
            "sentence_order_mismatch"
        )


    return (
        len(errors) == 0,
        errors,
    )


# -----------------------------------------------------------------------------
# Canonical sentence-record construction
# -----------------------------------------------------------------------------

def canonicalise_sentence_annotation(
    raw_result,
    sentence_annotation_item,
):
    """
    Convert one successful article-level sentence annotation into
    a canonical Media AI supervision record.

    Provider evidence is deterministically normalised into the canonical
    evidence structure when necessary.

    Returns:
        canonical_record, errors
    """

    errors = []


    if not isinstance(
        sentence_annotation_item,
        dict,
    ):

        return (
            None,
            [
                "sentence_annotation_item_not_dictionary"
            ],
        )


    sentence_id = (
        sentence_annotation_item.get(
            "sentence_id"
        )
    )


    provider_annotation = (
        sentence_annotation_item.get(
            "annotation"
        )
    )


    if not sentence_id:

        return (
            None,
            [
                "missing_sentence_id"
            ],
        )


    structure_valid, structure_errors = (
        validate_provider_annotation_structure(
            provider_annotation
        )
    )


    if not structure_valid:

        return (
            None,
            structure_errors,
        )


    context_record = (
        SUPERVISION_CONTEXT_BY_SENTENCE_ID.get(
            sentence_id
        )
    )


    if context_record is None:

        return (
            None,
            [
                "supervision_context_not_found"
            ],
        )


    # -------------------------------------------------------------------------
    # Start from canonical Block 4 template
    # -------------------------------------------------------------------------

    canonical_record = (
        create_blank_annotation_record(
            context_record
        )
    )


    # -------------------------------------------------------------------------
    # Copy structured sentence supervision outputs
    # -------------------------------------------------------------------------

    annotations = canonical_record[
        "annotations"
    ]


    annotations[
        "reported_emotion"
    ] = deepcopy(
        provider_annotation[
            "reported_emotion"
        ]
    )


    annotations[
        "transformative_emotion"
    ] = deepcopy(
        provider_annotation[
            "transformative_emotion"
        ]
    )


    annotations[
        "immediate_emotional_state"
    ] = deepcopy(
        provider_annotation[
            "immediate_emotional_state"
        ]
    )


    annotations[
        "emotional_weight"
    ] = deepcopy(
        provider_annotation[
            "emotional_weight"
        ]
    )


    annotations[
        "factological_weight"
    ] = deepcopy(
        provider_annotation[
            "factological_weight"
        ]
    )


    annotations[
        "factological_components"
    ] = deepcopy(
        provider_annotation[
            "factological_components"
        ]
    )


    annotations[
        "annotation_confidence"
    ] = deepcopy(
        provider_annotation[
            "annotation_confidence"
        ]
    )


    # -------------------------------------------------------------------------
    # Evidence normalisation
    # -------------------------------------------------------------------------

    provider_evidence = (
        provider_annotation.get(
            "evidence"
        )
    )


    canonical_evidence, evidence_errors = (
        normalise_provider_evidence(
            provider_evidence
        )
    )


    if evidence_errors:

        return (
            None,
            evidence_errors,
        )


    canonical_record[
        "evidence"
    ] = canonical_evidence


    # -------------------------------------------------------------------------
    # Machine-supervision review state
    # -------------------------------------------------------------------------

    canonical_record[
        "review"
    ][
        "review_status"
    ] = "unreviewed"


    canonical_record[
        "review"
    ][
        "review_notes"
    ] = None


    # -------------------------------------------------------------------------
    # Canonical provenance
    # -------------------------------------------------------------------------

    execution = raw_result.get(
        "execution",
        {}
    )


    supervisor_id = raw_result.get(
        "supervisor_id"
    )


    provider = raw_result.get(
        "provider"
    )


    model = raw_result.get(
        "model"
    )


    canonical_record[
        "provenance"
    ].update({
        "supervision_method":
            "independent_llm_supervision",

        "supervisor_role":
            "independent_llm_supervisor",

        "supervisor_identifier":
            supervisor_id,

        "provider":
            provider,

        "model":
            model,

        "provider_response_id":
            execution.get(
                "provider_response_id"
            ),

        "annotated_at_utc":
            datetime.now(
                timezone.utc
            ).isoformat(),

        "context_scope":
            context_record.get(
                "context_scope"
            ),

        "annotation_confidence":
            deepcopy(
                provider_annotation[
                    "annotation_confidence"
                ]
            ),

        "review_status":
            "unreviewed",

        "request_unit":
            "article",

        "annotation_unit":
            "sentence",

        "media_ai_record_id":
            raw_result.get(
                "identity",
                {}
            ).get(
                "media_ai_record_id"
            ),

        "sentence_id":
            sentence_id,

        "provider_schema_version":
            raw_result.get(
                "provider_schema_version"
            ),

        "provider_schema_strategy":
            raw_result.get(
                "provider_schema_strategy"
            ),

        "evidence_normalisation_policy":
            EVIDENCE_NORMALISATION_POLICY,
    })


    # -------------------------------------------------------------------------
    # Preserve protocol metadata
    # -------------------------------------------------------------------------

    canonical_record[
        "protocol"
    ] = {
        "protocol_version":
            raw_result.get(
                "protocol_version"
            ),

        "supervision_schema_version":
            raw_result.get(
                "supervision_schema_version"
            ),

        "annotation_policy_version":
            raw_result.get(
                "annotation_policy_version"
            ),
    }


    return (
        canonical_record,
        errors,
    )


# -----------------------------------------------------------------------------
# Additional canonical validation
# -----------------------------------------------------------------------------

def validate_canonical_supervision_record(
    record,
):
    """
    Apply Block 4 schema validation plus supervision-specific
    provenance and evidence checks.
    """

    errors = []


    block_4_passed, block_4_errors = (
        validate_annotation_record(
            record
        )
    )


    if not block_4_passed:

        errors.extend(
            block_4_errors
        )


    # -------------------------------------------------------------------------
    # Evidence
    # -------------------------------------------------------------------------

    evidence = record.get(
        "evidence",
        {}
    )


    if not isinstance(
        evidence,
        dict,
    ):

        errors.append(
            "invalid_evidence_structure"
        )


    else:

        summary = evidence.get(
            "summary"
        )


        supporting_text = evidence.get(
            "supporting_text"
        )


        if (
            summary is not None
            and not isinstance(
                summary,
                str,
            )
        ):

            errors.append(
                "invalid_evidence_summary"
            )


        if not isinstance(
            supporting_text,
            list,
        ):

            errors.append(
                "invalid_supporting_text"
            )


        elif not all(
            isinstance(
                item,
                str,
            )

            for item
            in supporting_text
        ):

            errors.append(
                "invalid_supporting_text_item"
            )


    # -------------------------------------------------------------------------
    # Provenance
    # -------------------------------------------------------------------------

    provenance = record.get(
        "provenance",
        {}
    )


    supervisor_identifier = (
        provenance.get(
            "supervisor_identifier"
        )
    )


    if (
        supervisor_identifier
        not in SUPERVISOR_REGISTRY
    ):

        errors.append(
            "invalid_supervisor_identifier"
        )


    if (
        provenance.get(
            "supervision_method"
        )
        != "independent_llm_supervision"
    ):

        errors.append(
            "invalid_supervision_method"
        )


    if not provenance.get(
        "annotated_at_utc"
    ):

        errors.append(
            "missing_annotation_timestamp"
        )


    if (
        provenance.get(
            "request_unit"
        )
        != "article"
    ):

        errors.append(
            "invalid_provenance_request_unit"
        )


    if (
        provenance.get(
            "annotation_unit"
        )
        != "sentence"
    ):

        errors.append(
            "invalid_provenance_annotation_unit"
        )


    # -------------------------------------------------------------------------
    # Protocol versions
    # -------------------------------------------------------------------------

    protocol = record.get(
        "protocol",
        {}
    )


    if (
        protocol.get(
            "supervision_schema_version"
        )
        != SUPERVISION_SCHEMA_VERSION
    ):

        errors.append(
            "protocol_supervision_schema_mismatch"
        )


    if (
        protocol.get(
            "annotation_policy_version"
        )
        != ANNOTATION_POLICY_VERSION
    ):

        errors.append(
            "protocol_annotation_policy_mismatch"
        )


    if (
        protocol.get(
            "protocol_version"
        )
        != LLM_SUPERVISION_PROTOCOL_VERSION
    ):

        errors.append(
            "protocol_version_mismatch"
        )


    return (
        len(errors) == 0,
        errors,
    )


# -----------------------------------------------------------------------------
# Canonical supervision stores
# -----------------------------------------------------------------------------

CANONICAL_SUPERVISION_RECORDS_BY_SUPERVISOR = {
    supervisor_id: []
    for supervisor_id
    in SUPERVISOR_REGISTRY
}


ACCEPTED_SUPERVISION_RECORDS = []

REJECTED_SUPERVISION_RECORDS = []

SUPERVISION_VALIDATION_RESULTS = []

ARTICLE_RESPONSE_VALIDATION_RESULTS = []


# -----------------------------------------------------------------------------
# Process raw article-level supervision results
# -----------------------------------------------------------------------------

for supervisor_id in sorted(
    LLM_SUPERVISION_RAW_RESULTS
):

    supervisor_results = (
        LLM_SUPERVISION_RAW_RESULTS[
            supervisor_id
        ]
    )


    for raw_result in supervisor_results:

        execution = raw_result.get(
            "execution",
            {}
        )


        execution_status = (
            execution.get(
                "status"
            )
        )


        media_ai_record_id = (
            raw_result.get(
                "identity",
                {}
            ).get(
                "media_ai_record_id"
            )
        )


        expected_sentence_ids = list(
            raw_result.get(
                "article_metadata",
                {}
            ).get(
                "expected_sentence_ids",
                [],
            )
        )


        # ---------------------------------------------------------------------
        # Non-successful provider requests remain rejected diagnostics
        # ---------------------------------------------------------------------

        if execution_status != "SUCCESS":

            rejection = {
                "supervisor_id":
                    supervisor_id,

                "media_ai_record_id":
                    media_ai_record_id,

                "sentence_id":
                    None,

                "execution_status":
                    execution_status,

                "accepted":
                    False,

                "errors": [
                    "provider_request_not_successful"
                ],

                "raw_result":
                    deepcopy(
                        raw_result
                    ),
            }


            REJECTED_SUPERVISION_RECORDS.append(
                rejection
            )


            SUPERVISION_VALIDATION_RESULTS.append({
                "supervisor_id":
                    supervisor_id,

                "media_ai_record_id":
                    media_ai_record_id,

                "sentence_id":
                    None,

                "accepted":
                    False,

                "errors":
                    rejection[
                        "errors"
                    ],
            })


            ARTICLE_RESPONSE_VALIDATION_RESULTS.append({
                "supervisor_id":
                    supervisor_id,

                "media_ai_record_id":
                    media_ai_record_id,

                "valid":
                    False,

                "errors":
                    rejection[
                        "errors"
                    ],
            })


            continue


        # ---------------------------------------------------------------------
        # Article envelope validation
        # ---------------------------------------------------------------------

        article_envelope_valid, article_envelope_errors = (
            validate_article_response_envelope(
                raw_result
            )
        )


        if not article_envelope_valid:

            rejection = {
                "supervisor_id":
                    supervisor_id,

                "media_ai_record_id":
                    media_ai_record_id,

                "sentence_id":
                    None,

                "execution_status":
                    execution_status,

                "accepted":
                    False,

                "errors":
                    article_envelope_errors,

                "raw_result":
                    deepcopy(
                        raw_result
                    ),
            }


            REJECTED_SUPERVISION_RECORDS.append(
                rejection
            )


            SUPERVISION_VALIDATION_RESULTS.append({
                "supervisor_id":
                    supervisor_id,

                "media_ai_record_id":
                    media_ai_record_id,

                "sentence_id":
                    None,

                "accepted":
                    False,

                "errors":
                    article_envelope_errors,
            })


            ARTICLE_RESPONSE_VALIDATION_RESULTS.append({
                "supervisor_id":
                    supervisor_id,

                "media_ai_record_id":
                    media_ai_record_id,

                "valid":
                    False,

                "errors":
                    article_envelope_errors,
            })


            continue


        # ---------------------------------------------------------------------
        # Article sentence-coverage validation
        # ---------------------------------------------------------------------

        coverage_valid, coverage_errors = (
            validate_article_sentence_coverage(
                raw_result
            )
        )


        if not coverage_valid:

            rejection = {
                "supervisor_id":
                    supervisor_id,

                "media_ai_record_id":
                    media_ai_record_id,

                "sentence_id":
                    None,

                "execution_status":
                    execution_status,

                "accepted":
                    False,

                "errors":
                    coverage_errors,

                "raw_result":
                    deepcopy(
                        raw_result
                    ),
            }


            REJECTED_SUPERVISION_RECORDS.append(
                rejection
            )


            SUPERVISION_VALIDATION_RESULTS.append({
                "supervisor_id":
                    supervisor_id,

                "media_ai_record_id":
                    media_ai_record_id,

                "sentence_id":
                    None,

                "accepted":
                    False,

                "errors":
                    coverage_errors,
            })


            ARTICLE_RESPONSE_VALIDATION_RESULTS.append({
                "supervisor_id":
                    supervisor_id,

                "media_ai_record_id":
                    media_ai_record_id,

                "valid":
                    False,

                "errors":
                    coverage_errors,
            })


            continue


        # ---------------------------------------------------------------------
        # Expand article response into sentence annotations
        # ---------------------------------------------------------------------

        parsed_output = execution[
            "parsed_output"
        ]


        sentence_annotations = parsed_output[
            "sentence_annotations"
        ]


        article_sentence_acceptance = []

        article_sentence_errors = []


        for sentence_annotation_item in sentence_annotations:

            sentence_id = (
                sentence_annotation_item.get(
                    "sentence_id"
                )
            )


            # -----------------------------------------------------------------
            # Canonicalise
            # -----------------------------------------------------------------

            canonical_record, canonicalisation_errors = (
                canonicalise_sentence_annotation(
                    raw_result=
                        raw_result,

                    sentence_annotation_item=
                        sentence_annotation_item,
                )
            )


            if canonicalisation_errors:

                rejection = {
                    "supervisor_id":
                        supervisor_id,

                    "media_ai_record_id":
                        media_ai_record_id,

                    "sentence_id":
                        sentence_id,

                    "execution_status":
                        execution_status,

                    "accepted":
                        False,

                    "errors":
                        canonicalisation_errors,

                    "raw_sentence_annotation":
                        deepcopy(
                            sentence_annotation_item
                        ),

                    "raw_result":
                        deepcopy(
                            raw_result
                        ),
                }


                REJECTED_SUPERVISION_RECORDS.append(
                    rejection
                )


                SUPERVISION_VALIDATION_RESULTS.append({
                    "supervisor_id":
                        supervisor_id,

                    "media_ai_record_id":
                        media_ai_record_id,

                    "sentence_id":
                        sentence_id,

                    "accepted":
                        False,

                    "errors":
                        canonicalisation_errors,
                })


                article_sentence_acceptance.append(
                    False
                )


                article_sentence_errors.extend(
                    [
                        f"{sentence_id}:{error}"
                        for error
                        in canonicalisation_errors
                    ]
                )


                continue


            # -----------------------------------------------------------------
            # Canonical validation
            # -----------------------------------------------------------------

            passed, validation_errors = (
                validate_canonical_supervision_record(
                    canonical_record
                )
            )


            if not passed:

                rejection = {
                    "supervisor_id":
                        supervisor_id,

                    "media_ai_record_id":
                        media_ai_record_id,

                    "sentence_id":
                        sentence_id,

                    "execution_status":
                        execution_status,

                    "accepted":
                        False,

                    "errors":
                        validation_errors,

                    "canonical_candidate":
                        deepcopy(
                            canonical_record
                        ),

                    "raw_sentence_annotation":
                        deepcopy(
                            sentence_annotation_item
                        ),

                    "raw_result":
                        deepcopy(
                            raw_result
                        ),
                }


                REJECTED_SUPERVISION_RECORDS.append(
                    rejection
                )


                SUPERVISION_VALIDATION_RESULTS.append({
                    "supervisor_id":
                        supervisor_id,

                    "media_ai_record_id":
                        media_ai_record_id,

                    "sentence_id":
                        sentence_id,

                    "accepted":
                        False,

                    "errors":
                        validation_errors,
                })


                article_sentence_acceptance.append(
                    False
                )


                article_sentence_errors.extend(
                    [
                        f"{sentence_id}:{error}"
                        for error
                        in validation_errors
                    ]
                )


                continue


            # -----------------------------------------------------------------
            # Accept canonical sentence annotation
            # -----------------------------------------------------------------

            CANONICAL_SUPERVISION_RECORDS_BY_SUPERVISOR[
                supervisor_id
            ].append(
                canonical_record
            )


            ACCEPTED_SUPERVISION_RECORDS.append(
                canonical_record
            )


            SUPERVISION_VALIDATION_RESULTS.append({
                "supervisor_id":
                    supervisor_id,

                "media_ai_record_id":
                    media_ai_record_id,

                "sentence_id":
                    sentence_id,

                "accepted":
                    True,

                "errors":
                    [],
            })


            article_sentence_acceptance.append(
                True
            )


        # ---------------------------------------------------------------------
        # Article-level validation result
        # ---------------------------------------------------------------------

        article_valid = bool(
            len(
                article_sentence_acceptance
            )
            == len(
                expected_sentence_ids
            )
            and all(
                article_sentence_acceptance
            )
        )


        ARTICLE_RESPONSE_VALIDATION_RESULTS.append({
            "supervisor_id":
                supervisor_id,

            "media_ai_record_id":
                media_ai_record_id,

            "valid":
                article_valid,

            "expected_sentence_count":
                len(
                    expected_sentence_ids
                ),

            "accepted_sentence_count":
                sum(
                    article_sentence_acceptance
                ),

            "errors":
                article_sentence_errors,
        })


# -----------------------------------------------------------------------------
# Validation diagnostics
# -----------------------------------------------------------------------------

raw_article_response_count = sum(
    len(
        results
    )
    for results
    in LLM_SUPERVISION_RAW_RESULTS.values()
)


accepted_count = len(
    ACCEPTED_SUPERVISION_RECORDS
)


rejected_count = len(
    REJECTED_SUPERVISION_RECORDS
)


validation_result_count = len(
    SUPERVISION_VALIDATION_RESULTS
)


article_validation_count = len(
    ARTICLE_RESPONSE_VALIDATION_RESULTS
)


validated_article_count = sum(
    1
    for result
    in ARTICLE_RESPONSE_VALIDATION_RESULTS
    if result.get(
        "valid"
    )
)


rejected_article_count = sum(
    1
    for result
    in ARTICLE_RESPONSE_VALIDATION_RESULTS
    if not result.get(
        "valid"
    )
)


accepted_by_supervisor = {
    supervisor_id:
        len(
            CANONICAL_SUPERVISION_RECORDS_BY_SUPERVISOR[
                supervisor_id
            ]
        )

    for supervisor_id
    in SUPERVISOR_REGISTRY
}


rejected_by_supervisor = {
    supervisor_id:
        sum(
            1
            for record
            in REJECTED_SUPERVISION_RECORDS
            if record[
                "supervisor_id"
            ]
            == supervisor_id
        )

    for supervisor_id
    in SUPERVISOR_REGISTRY
}


# -----------------------------------------------------------------------------
# Sentence-accounting verification
# -----------------------------------------------------------------------------

validation_accounting_valid = (
    validation_result_count
    == accepted_count
    + rejected_count
)


if not validation_accounting_valid:

    raise ValueError(
        "Supervision validation accounting is inconsistent."
    )


# -----------------------------------------------------------------------------
# Article-accounting verification
# -----------------------------------------------------------------------------

article_validation_accounting_valid = (
    raw_article_response_count
    == article_validation_count
    == validated_article_count
    + rejected_article_count
)


if not article_validation_accounting_valid:

    raise ValueError(
        "Article supervision validation accounting is inconsistent."
    )


# -----------------------------------------------------------------------------
# Current validation state
# -----------------------------------------------------------------------------

ARTICLE_RESPONSE_VALIDATION_VALID = (
    validation_accounting_valid
    and article_validation_accounting_valid
)


CANONICAL_SUPERVISION_VALID = (
    ARTICLE_RESPONSE_VALIDATION_VALID
)


# -----------------------------------------------------------------------------
# Verification output
# -----------------------------------------------------------------------------

print("=" * 72)

print(
    "Media AI — Canonical Supervision Validation"
)

print("=" * 72)


print(
    f"Validation version         : "
    f"{CANONICAL_SUPERVISION_VALIDATION_VERSION}"
)

print(
    f"Evidence normalisation     : "
    f"{EVIDENCE_NORMALISATION_POLICY}"
)

print(
    f"Raw article responses      : "
    f"{raw_article_response_count}"
)

print(
    f"Article validation results : "
    f"{article_validation_count}"
)

print(
    f"Validated articles         : "
    f"{validated_article_count}"
)

print(
    f"Rejected articles          : "
    f"{rejected_article_count}"
)


print("-" * 72)


print(
    f"Accepted annotations       : "
    f"{accepted_count}"
)

print(
    f"Rejected annotations       : "
    f"{rejected_count}"
)

print(
    f"Sentence validation results: "
    f"{validation_result_count}"
)

print(
    f"Sentence accounting valid  : "
    f"{validation_accounting_valid}"
)

print(
    f"Article accounting valid   : "
    f"{article_validation_accounting_valid}"
)


print("-" * 72)


for supervisor_id in sorted(
    SUPERVISOR_REGISTRY
):

    print(
        f"{supervisor_id}"
    )

    print(
        f"  Accepted sentences       : "
        f"{accepted_by_supervisor[supervisor_id]}"
    )

    print(
        f"  Rejected diagnostics     : "
        f"{rejected_by_supervisor[supervisor_id]}"
    )


print("-" * 72)

print(
    "Article response diagnostics"
)


for index, result in enumerate(
    ARTICLE_RESPONSE_VALIDATION_RESULTS,
    start=1,
):

    print(
        f"  Article result {index}"
    )

    print(
        f"    Supervisor             : "
        f"{result.get('supervisor_id')}"
    )

    print(
        f"    Media AI record ID     : "
        f"{result.get('media_ai_record_id')}"
    )

    print(
        f"    Valid                  : "
        f"{result.get('valid')}"
    )


    if (
        result.get(
            "expected_sentence_count"
        )
        is not None
    ):

        print(
            f"    Expected sentences     : "
            f"{result.get('expected_sentence_count')}"
        )


    if (
        result.get(
            "accepted_sentence_count"
        )
        is not None
    ):

        print(
            f"    Accepted sentences     : "
            f"{result.get('accepted_sentence_count')}"
        )


    if result.get(
        "errors"
    ):

        print(
            f"    Errors                 : "
            f"{result.get('errors')}"
        )


print("-" * 72)


if rejected_count:

    print(
        "Rejected-response diagnostics"
    )


    for index, rejection in enumerate(
        REJECTED_SUPERVISION_RECORDS,
        start=1,
    ):

        print(
            f"  Rejection {index}"
        )

        print(
            f"    Supervisor             : "
            f"{rejection.get('supervisor_id')}"
        )

        print(
            f"    Media AI record ID     : "
            f"{rejection.get('media_ai_record_id')}"
        )

        print(
            f"    Sentence ID            : "
            f"{rejection.get('sentence_id')}"
        )

        print(
            f"    Errors                 : "
            f"{rejection.get('errors')}"
        )


print("=" * 72)


if raw_article_response_count == 0:

    print(
        "No raw supervision article responses were available "
        "for validation."
    )

    print(
        "This is an expected state while LLM supervision remains disabled."
    )


else:

    print(
        "Raw article-level supervision responses were expanded into "
        "sentence-level annotations and processed through canonical validation."
    )


print(
    "Provider evidence strings were deterministically normalised into "
    "canonical supporting-text structures without inventing summaries."
)

print(
    "Complete article identity, sentence coverage and sentence order "
    "were validated before sentence annotations were accepted."
)

print(
    "Only schema-valid sentence annotations were accepted into "
    "canonical supervision."
)

print(
    "Rejected sentence or article responses remain diagnostic and "
    "have not entered the reference dataset."
)

print(
    "Article-level provider requests remain distinct from "
    "sentence-level Media AI supervision records."
)

print(
    "Supervisor outputs remain independent; no comparison or consensus "
    "has been performed."
)

print(
    "No external model or API calls have been made by this validation block."
)

print("=" * 72)
```

## Supervisor Agreement and Disagreement Analysis

This block evaluates agreement and disagreement across independent LLM
supervisors.

The comparison operates only on canonical annotations that have already passed
Block 7 validation.

Raw provider responses and rejected annotations are excluded from agreement
analysis.

### Comparison Unit

Supervisors are compared only when they have independently annotated the same:

- Media AI record;
- paragraph;
- sentence.

The sentence identifier is therefore the primary alignment key.

### Independent-Supervision Principle

Supervisor outputs remain independent.

This block does not:

- expose one supervisor's labels to another;
- average annotations into a consensus target;
- perform majority voting;
- adjudicate disagreements;
- overwrite individual annotations.

It only measures the degree to which independently generated annotations agree.

### Continuous Constructs

For continuous supervision variables, agreement is measured through absolute
difference.

This includes:

- reported emotion;
- transformative emotion;
- immediate emotional state;
- emotional weight;
- factological weight;
- annotation confidence.

For emotional vectors, differences are calculated separately for every emotional
dimension and may also be summarised across dimensions.

For two observed values \(x\) and \(y\):

\[
D(x,y)=|x-y|
\]

Smaller values represent closer agreement.

The block does not treat missing values as zero.

Comparisons are made only where both supervisors provide an `observed` value.

### Transformative Direction Agreement

Transformative emotion contains signed information.

In addition to absolute numeric difference, the block therefore records whether
two supervisors agree on the direction of change:

- decrease;
- approximately neutral;
- increase.

A small neutral band around zero is used to avoid interpreting tiny numerical
differences as substantive directional changes.

The pilot neutral threshold is configurable.

### Factological Components

Boolean factological components are compared through exact agreement:

- both `True`;
- both `False`;
- disagreement;
- unavailable comparison when either value is `null`.

### Evidence and Rationale

Evidence text is preserved with each individual annotation but is not converted
into a numeric agreement score in this block.

Later qualitative analysis may examine whether supervisors reached similar
numeric conclusions for different reasons.

### Pairwise Comparison

With \(n\) supervisors, all available supervisor pairs may be compared for each
shared sentence.

The architecture therefore supports the five-slot registry introduced in
Block 6 without requiring all five supervisors to be active.

### Review Signals

Substantial disagreement is treated as a review signal rather than an error.

A sentence may be flagged when:

- continuous annotations diverge beyond a configured threshold;
- transformative direction differs;
- factological component labels disagree;
- supervisors provide materially different confidence levels.

These signals can later guide human review or adjudication.

### No-Supervision and Single-Supervisor States

The block must also handle early pilot states correctly.

If no accepted annotations exist, agreement analysis returns an empty but valid
result.

If only one supervisor has annotated a sentence, that annotation remains valid
but no inter-supervisor agreement score can be calculated.

### Output

This block creates:

- `SUPERVISOR_PAIRWISE_COMPARISONS`;
- `SUPERVISOR_DISAGREEMENT_FLAGS`;
- `SUPERVISOR_AGREEMENT_SUMMARY`;
- per-construct and per-dimension agreement diagnostics.

No consensus or adjudicated reference label is created.

No external model or API call is performed by this block.

```python
# =============================================================================
# Media AI — Supervisor Agreement and Disagreement Analysis
# =============================================================================

from collections import defaultdict
from itertools import combinations
import statistics


# -----------------------------------------------------------------------------
# Agreement-analysis configuration
# -----------------------------------------------------------------------------

SUPERVISOR_AGREEMENT_POLICY_VERSION = "1.0"

# Values within this range around zero are treated as approximately neutral
# for transformative-direction comparison.
TRANSFORMATIVE_NEUTRAL_THRESHOLD = 0.10

# Diagnostic disagreement thresholds.
CONTINUOUS_DISAGREEMENT_THRESHOLD = 0.35
CONFIDENCE_DISAGREEMENT_THRESHOLD = 0.30


# -----------------------------------------------------------------------------
# Dependency checks
# -----------------------------------------------------------------------------

required_agreement_objects = [
    "CANONICAL_SUPERVISION_RECORDS_BY_SUPERVISOR",
    "SUPERVISOR_REGISTRY",
    "EMOTIONAL_DIMENSIONS",
    "FACTOLOGICAL_COMPONENTS",
]

missing_agreement_objects = [
    object_name
    for object_name in required_agreement_objects
    if object_name not in globals()
]

if missing_agreement_objects:
    raise NameError(
        "Required canonical-supervision objects are not initialised. "
        "Rerun Notebook 03 Blocks 3–7. Missing: "
        f"{missing_agreement_objects}"
    )


# -----------------------------------------------------------------------------
# Helpers
# -----------------------------------------------------------------------------

def get_observed_value(annotation_value):
    """
    Return a numeric annotation only when its value state is observed.
    """

    if not isinstance(
        annotation_value,
        dict,
    ):
        return None

    if (
        annotation_value.get(
            "value_state"
        )
        != "observed"
    ):
        return None

    value = annotation_value.get(
        "value"
    )

    if not isinstance(
        value,
        (int, float),
    ):
        return None

    return float(value)


def classify_transformative_direction(
    value,
):
    """
    Convert a signed transformative value into decrease / neutral / increase.
    """

    if value is None:
        return None

    if value < -TRANSFORMATIVE_NEUTRAL_THRESHOLD:
        return "decrease"

    if value > TRANSFORMATIVE_NEUTRAL_THRESHOLD:
        return "increase"

    return "neutral"


def compare_scalar_annotation_values(
    first_value,
    second_value,
):
    """
    Compare two scalar annotation objects where both provide observed values.
    """

    first_numeric = get_observed_value(
        first_value
    )

    second_numeric = get_observed_value(
        second_value
    )

    if (
        first_numeric is None
        or second_numeric is None
    ):
        return {
            "comparable":
                False,

            "value_1":
                first_numeric,

            "value_2":
                second_numeric,

            "absolute_difference":
                None,
        }

    difference = abs(
        first_numeric
        - second_numeric
    )

    return {
        "comparable":
            True,

        "value_1":
            first_numeric,

        "value_2":
            second_numeric,

        "absolute_difference":
            difference,
    }


# -----------------------------------------------------------------------------
# Index accepted records by sentence and supervisor
# -----------------------------------------------------------------------------

annotations_by_sentence = defaultdict(
    dict
)

for supervisor_id, records in (
    CANONICAL_SUPERVISION_RECORDS_BY_SUPERVISOR.items()
):

    for record in records:

        sentence_id = (
            record.get(
                "identity",
                {}
            ).get(
                "sentence_id"
            )
        )

        if not sentence_id:
            continue

        annotations_by_sentence[
            sentence_id
        ][
            supervisor_id
        ] = record


# -----------------------------------------------------------------------------
# Pairwise comparison containers
# -----------------------------------------------------------------------------

SUPERVISOR_PAIRWISE_COMPARISONS = []
SUPERVISOR_DISAGREEMENT_FLAGS = []


# -----------------------------------------------------------------------------
# Pairwise sentence-level comparison
# -----------------------------------------------------------------------------

for sentence_id, supervisor_records in (
    annotations_by_sentence.items()
):

    available_supervisors = sorted(
        supervisor_records.keys()
    )

    if len(
        available_supervisors
    ) < 2:
        continue

    for (
        supervisor_1,
        supervisor_2,
    ) in combinations(
        available_supervisors,
        2,
    ):

        record_1 = supervisor_records[
            supervisor_1
        ]

        record_2 = supervisor_records[
            supervisor_2
        ]

        annotations_1 = record_1.get(
            "annotations",
            {}
        )

        annotations_2 = record_2.get(
            "annotations",
            {}
        )

        comparison = {
            "sentence_id":
                sentence_id,

            "media_ai_record_id":
                record_1.get(
                    "identity",
                    {}
                ).get(
                    "media_ai_record_id"
                ),

            "supervisor_1":
                supervisor_1,

            "supervisor_2":
                supervisor_2,

            "reported_emotion":
                {},

            "transformative_emotion":
                {},

            "immediate_emotional_state":
                {},

            "scalar_annotations":
                {},

            "factological_components":
                {},

            "disagreement_flags":
                [],
        }

        # ---------------------------------------------------------------------
        # Reported emotion
        # ---------------------------------------------------------------------

        for dimension in EMOTIONAL_DIMENSIONS:

            result = (
                compare_scalar_annotation_values(
                    annotations_1.get(
                        "reported_emotion",
                        {}
                    ).get(
                        dimension
                    ),

                    annotations_2.get(
                        "reported_emotion",
                        {}
                    ).get(
                        dimension
                    ),
                )
            )

            comparison[
                "reported_emotion"
            ][
                dimension
            ] = result

            if (
                result[
                    "comparable"
                ]
                and result[
                    "absolute_difference"
                ]
                > CONTINUOUS_DISAGREEMENT_THRESHOLD
            ):
                comparison[
                    "disagreement_flags"
                ].append(
                    f"reported_emotion:{dimension}"
                )

        # ---------------------------------------------------------------------
        # Transformative emotion
        # ---------------------------------------------------------------------

        for dimension in EMOTIONAL_DIMENSIONS:

            result = (
                compare_scalar_annotation_values(
                    annotations_1.get(
                        "transformative_emotion",
                        {}
                    ).get(
                        dimension
                    ),

                    annotations_2.get(
                        "transformative_emotion",
                        {}
                    ).get(
                        dimension
                    ),
                )
            )

            if result[
                "comparable"
            ]:

                direction_1 = (
                    classify_transformative_direction(
                        result[
                            "value_1"
                        ]
                    )
                )

                direction_2 = (
                    classify_transformative_direction(
                        result[
                            "value_2"
                        ]
                    )
                )

                direction_agreement = (
                    direction_1
                    == direction_2
                )

            else:

                direction_1 = None
                direction_2 = None
                direction_agreement = None

            result[
                "direction_1"
            ] = direction_1

            result[
                "direction_2"
            ] = direction_2

            result[
                "direction_agreement"
            ] = direction_agreement

            comparison[
                "transformative_emotion"
            ][
                dimension
            ] = result

            if (
                result[
                    "comparable"
                ]
                and result[
                    "absolute_difference"
                ]
                > CONTINUOUS_DISAGREEMENT_THRESHOLD
            ):
                comparison[
                    "disagreement_flags"
                ].append(
                    f"transformative_emotion:{dimension}:magnitude"
                )

            if (
                result[
                    "direction_agreement"
                ]
                is False
            ):
                comparison[
                    "disagreement_flags"
                ].append(
                    f"transformative_emotion:{dimension}:direction"
                )

        # ---------------------------------------------------------------------
        # Immediate emotional state
        # ---------------------------------------------------------------------

        for dimension in EMOTIONAL_DIMENSIONS:

            result = (
                compare_scalar_annotation_values(
                    annotations_1.get(
                        "immediate_emotional_state",
                        {}
                    ).get(
                        dimension
                    ),

                    annotations_2.get(
                        "immediate_emotional_state",
                        {}
                    ).get(
                        dimension
                    ),
                )
            )

            comparison[
                "immediate_emotional_state"
            ][
                dimension
            ] = result

            if (
                result[
                    "comparable"
                ]
                and result[
                    "absolute_difference"
                ]
                > CONTINUOUS_DISAGREEMENT_THRESHOLD
            ):
                comparison[
                    "disagreement_flags"
                ].append(
                    f"immediate_emotional_state:{dimension}"
                )

        # ---------------------------------------------------------------------
        # Scalar weights and confidence
        # ---------------------------------------------------------------------

        scalar_fields = [
            "emotional_weight",
            "factological_weight",
            "annotation_confidence",
        ]

        for field in scalar_fields:

            result = (
                compare_scalar_annotation_values(
                    annotations_1.get(
                        field
                    ),

                    annotations_2.get(
                        field
                    ),
                )
            )

            comparison[
                "scalar_annotations"
            ][
                field
            ] = result

            threshold = (
                CONFIDENCE_DISAGREEMENT_THRESHOLD
                if field
                == "annotation_confidence"
                else CONTINUOUS_DISAGREEMENT_THRESHOLD
            )

            if (
                result[
                    "comparable"
                ]
                and result[
                    "absolute_difference"
                ]
                > threshold
            ):
                comparison[
                    "disagreement_flags"
                ].append(
                    field
                )

        # ---------------------------------------------------------------------
        # Factological components
        # ---------------------------------------------------------------------

        components_1 = annotations_1.get(
            "factological_components",
            {}
        )

        components_2 = annotations_2.get(
            "factological_components",
            {}
        )

        for component in FACTOLOGICAL_COMPONENTS:

            value_1 = components_1.get(
                component
            )

            value_2 = components_2.get(
                component
            )

            comparable = (
                value_1
                in {
                    True,
                    False,
                }
                and value_2
                in {
                    True,
                    False,
                }
            )

            agreement = (
                value_1
                == value_2
                if comparable
                else None
            )

            comparison[
                "factological_components"
            ][
                component
            ] = {
                "comparable":
                    comparable,

                "value_1":
                    value_1,

                "value_2":
                    value_2,

                "agreement":
                    agreement,
            }

            if agreement is False:

                comparison[
                    "disagreement_flags"
                ].append(
                    f"factological_component:{component}"
                )

        # ---------------------------------------------------------------------
        # Pair-level review signal
        # ---------------------------------------------------------------------

        comparison[
            "review_required"
        ] = bool(
            comparison[
                "disagreement_flags"
            ]
        )

        SUPERVISOR_PAIRWISE_COMPARISONS.append(
            comparison
        )

        if comparison[
            "review_required"
        ]:

            SUPERVISOR_DISAGREEMENT_FLAGS.append({
                "sentence_id":
                    sentence_id,

                "media_ai_record_id":
                    comparison[
                        "media_ai_record_id"
                    ],

                "supervisor_1":
                    supervisor_1,

                "supervisor_2":
                    supervisor_2,

                "flags":
                    list(
                        comparison[
                            "disagreement_flags"
                        ]
                    ),
            })


# -----------------------------------------------------------------------------
# Aggregate agreement diagnostics
# -----------------------------------------------------------------------------

continuous_differences = []

transformative_direction_comparisons = 0
transformative_direction_agreements = 0

factological_component_comparisons = 0
factological_component_agreements = 0


for comparison in SUPERVISOR_PAIRWISE_COMPARISONS:

    # -------------------------------------------------------------------------
    # Emotional-vector differences
    # -------------------------------------------------------------------------

    for field in [
        "reported_emotion",
        "transformative_emotion",
        "immediate_emotional_state",
    ]:

        for result in comparison[
            field
        ].values():

            if (
                result.get(
                    "comparable"
                )
                and result.get(
                    "absolute_difference"
                )
                is not None
            ):
                continuous_differences.append(
                    result[
                        "absolute_difference"
                    ]
                )

    # -------------------------------------------------------------------------
    # Scalar differences
    # -------------------------------------------------------------------------

    for result in comparison[
        "scalar_annotations"
    ].values():

        if (
            result.get(
                "comparable"
            )
            and result.get(
                "absolute_difference"
            )
            is not None
        ):
            continuous_differences.append(
                result[
                    "absolute_difference"
                ]
            )

    # -------------------------------------------------------------------------
    # Transformative-direction agreement
    # -------------------------------------------------------------------------

    for result in comparison[
        "transformative_emotion"
    ].values():

        if (
            result.get(
                "direction_agreement"
            )
            is not None
        ):

            transformative_direction_comparisons += 1

            if result[
                "direction_agreement"
            ]:

                transformative_direction_agreements += 1

    # -------------------------------------------------------------------------
    # Factological-component agreement
    # -------------------------------------------------------------------------

    for result in comparison[
        "factological_components"
    ].values():

        if result.get(
            "comparable"
        ):

            factological_component_comparisons += 1

            if result[
                "agreement"
            ]:

                factological_component_agreements += 1


# -----------------------------------------------------------------------------
# Aggregate summary
# -----------------------------------------------------------------------------

if continuous_differences:

    mean_absolute_difference = (
        statistics.fmean(
            continuous_differences
        )
    )

    median_absolute_difference = (
        statistics.median(
            continuous_differences
        )
    )

    maximum_absolute_difference = max(
        continuous_differences
    )

else:

    mean_absolute_difference = None
    median_absolute_difference = None
    maximum_absolute_difference = None


if transformative_direction_comparisons:

    transformative_direction_agreement_rate = (
        transformative_direction_agreements
        / transformative_direction_comparisons
    )

else:

    transformative_direction_agreement_rate = None


if factological_component_comparisons:

    factological_component_agreement_rate = (
        factological_component_agreements
        / factological_component_comparisons
    )

else:

    factological_component_agreement_rate = None


sentences_with_multiple_supervisors = sum(
    1
    for supervisor_records
    in annotations_by_sentence.values()
    if len(
        supervisor_records
    ) >= 2
)


SUPERVISOR_AGREEMENT_SUMMARY = {
    "agreement_policy_version":
        SUPERVISOR_AGREEMENT_POLICY_VERSION,

    "accepted_annotations":
        len(
            ACCEPTED_SUPERVISION_RECORDS
        ),

    "sentences_with_annotations":
        len(
            annotations_by_sentence
        ),

    "sentences_with_multiple_supervisors":
        sentences_with_multiple_supervisors,

    "pairwise_comparisons":
        len(
            SUPERVISOR_PAIRWISE_COMPARISONS
        ),

    "pairwise_reviews_required":
        len(
            SUPERVISOR_DISAGREEMENT_FLAGS
        ),

    "continuous_values_compared":
        len(
            continuous_differences
        ),

    "mean_absolute_difference":
        mean_absolute_difference,

    "median_absolute_difference":
        median_absolute_difference,

    "maximum_absolute_difference":
        maximum_absolute_difference,

    "transformative_direction_comparisons":
        transformative_direction_comparisons,

    "transformative_direction_agreement_rate":
        transformative_direction_agreement_rate,

    "factological_component_comparisons":
        factological_component_comparisons,

    "factological_component_agreement_rate":
        factological_component_agreement_rate,

    "consensus_constructed":
        False,

    "adjudication_performed":
        False,
}


# -----------------------------------------------------------------------------
# Verification output
# -----------------------------------------------------------------------------

print("=" * 72)
print("Media AI — Supervisor Agreement and Disagreement Analysis")
print("=" * 72)

print(
    f"Agreement policy version   : "
    f"{SUPERVISOR_AGREEMENT_POLICY_VERSION}"
)

print(
    f"Accepted annotations       : "
    f"{SUPERVISOR_AGREEMENT_SUMMARY['accepted_annotations']}"
)

print(
    f"Sentences with annotations : "
    f"{SUPERVISOR_AGREEMENT_SUMMARY['sentences_with_annotations']}"
)

print(
    f"Multi-supervisor sentences : "
    f"{SUPERVISOR_AGREEMENT_SUMMARY['sentences_with_multiple_supervisors']}"
)

print(
    f"Pairwise comparisons       : "
    f"{SUPERVISOR_AGREEMENT_SUMMARY['pairwise_comparisons']}"
)

print(
    f"Pairs requiring review     : "
    f"{SUPERVISOR_AGREEMENT_SUMMARY['pairwise_reviews_required']}"
)

print("-" * 72)

print(
    f"Continuous values compared : "
    f"{SUPERVISOR_AGREEMENT_SUMMARY['continuous_values_compared']}"
)

print(
    f"Mean absolute difference   : "
    f"{SUPERVISOR_AGREEMENT_SUMMARY['mean_absolute_difference']}"
)

print(
    f"Median absolute difference : "
    f"{SUPERVISOR_AGREEMENT_SUMMARY['median_absolute_difference']}"
)

print(
    f"Maximum absolute difference: "
    f"{SUPERVISOR_AGREEMENT_SUMMARY['maximum_absolute_difference']}"
)

print("-" * 72)

print(
    f"Transform direction tests  : "
    f"{SUPERVISOR_AGREEMENT_SUMMARY['transformative_direction_comparisons']}"
)

print(
    f"Transform direction agree  : "
    f"{SUPERVISOR_AGREEMENT_SUMMARY['transformative_direction_agreement_rate']}"
)

print(
    f"Factological comparisons   : "
    f"{SUPERVISOR_AGREEMENT_SUMMARY['factological_component_comparisons']}"
)

print(
    f"Factological agreement     : "
    f"{SUPERVISOR_AGREEMENT_SUMMARY['factological_component_agreement_rate']}"
)

print("-" * 72)

print(
    f"Consensus constructed      : "
    f"{SUPERVISOR_AGREEMENT_SUMMARY['consensus_constructed']}"
)

print(
    f"Adjudication performed     : "
    f"{SUPERVISOR_AGREEMENT_SUMMARY['adjudication_performed']}"
)

print("=" * 72)


if len(
    SUPERVISOR_PAIRWISE_COMPARISONS
) == 0:

    print(
        "No pairwise supervisor comparison is currently available."
    )

    print(
        "At least two accepted independent annotations for the same sentence "
        "are required for agreement analysis."
    )

else:

    print(
        "Independent supervisor annotations were compared successfully."
    )


print(
    "Disagreement remains a review signal and has not been converted into exclusion."
)

print(
    "No consensus or adjudicated reference labels have been created."
)

print(
    "No external model or API calls have been made by this block."
)
```

## Reference Dataset Construction and Persistent Annotation Artefacts

This block constructs the Notebook 03 reference-dataset representation from
validated supervision records.

The reference dataset is the formal supervision artefact intended for later
model training and evaluation.

Its construction remains controlled by the Block 1 switch:

`BUILD_REFERENCE_DATASET`

Persistence is controlled separately by:

`SAVE_ANNOTATION_OUTPUTS`

These switches are intentionally independent.

A reference dataset may therefore be constructed in memory without being written
to persistent storage.

### Reference-Label Principle

Independent LLM annotations are supervision evidence, not automatically
authoritative reference labels.

The dataset therefore distinguishes between:

- independent machine supervision;
- human-reviewed supervision;
- adjudicated reference supervision.

This distinction is preserved explicitly rather than collapsing all labels into
one field.

### Pilot Reference Policy

For the initial pilot, validated independent LLM annotations may be retained as
machine-supervision records.

They must not be represented as adjudicated human reference labels unless a
later review step actually performs that function.

Where several supervisors annotate the same sentence, their annotations remain
separate.

Block 9 does not create consensus by:

- averaging;
- majority voting;
- confidence weighting;
- model ranking;
- implicit selection of one supervisor.

Consensus and adjudication require an explicit methodological rule.

### Reference Dataset Unit

The reference dataset preserves the sentence-level unit established throughout
Notebook 03.

Each record retains:

- Media AI record identifier;
- paragraph identifier;
- sentence identifier;
- paragraph and sentence indices;
- sentence context;
- annotation provenance;
- supervision source;
- review status;
- reference-label status.

The complete prepared article remains available through the canonical Notebook
02 corpus and is not duplicated unnecessarily.

### Reference Status

Every supervision record receives an explicit reference status.

The initial states are:

- `machine_supervision` — validated independent LLM supervision;
- `human_reviewed` — reviewed by a human but not necessarily adjudicated;
- `adjudicated_reference` — accepted as an explicit reference label;
- `not_reference_ready` — valid supervision that should not yet serve as a
  reference target.

This status is separate from annotation validity.

A record can be schema-valid while still not being reference-ready.

### Human Review Boundary

Human-review export is controlled separately through:

`RUN_HUMAN_REVIEW_EXPORT`

If human review has not been performed, Block 9 must not invent
`human_reviewed` or `adjudicated_reference` status.

### Persistence

When `SAVE_ANNOTATION_OUTPUTS = True`, Notebook 03 may persist:

- the supervision schema;
- the LLM supervision protocol;
- validated canonical annotations;
- rejected-response diagnostics;
- supervisor-agreement diagnostics;
- the constructed reference dataset;
- a Notebook 03 handover manifest for Notebook 04.

Credentials, API keys and private authentication information must never be
written into these artefacts.

### Empty-State Behaviour

If no canonical annotations exist, reference-dataset construction must remain a
valid operation.

When construction is disabled, no reference dataset is produced.

When construction is enabled but no accepted annotations exist, an empty
versioned dataset may be created for structural verification, but it must
explicitly report zero reference records.

### Notebook 04 Boundary

Notebook 04 should receive only persisted, versioned supervision artefacts.

It must not depend on temporary provider responses or runtime-only annotation
objects.

The Notebook 03 handover will therefore identify:

- the prepared corpus;
- supervision schema;
- reference dataset;
- annotation counts;
- supervisor configuration metadata;
- review and adjudication state.

### Output

This block creates, when enabled:

- `REFERENCE_DATASET`;
- `REFERENCE_DATASET_RECORDS`;
- reference-status diagnostics;
- optional persistent Notebook 03 artefacts;
- preliminary Notebook 04 handover metadata.

No model training, recurrence or confluent-state calculation is performed.

No external LLM or GNews API request is performed by this block.

```python
# =============================================================================
# Media AI — Reference Dataset Construction and Persistent Annotation Artefacts
# =============================================================================

import io
import json
from copy import deepcopy
from datetime import datetime, timezone


# -----------------------------------------------------------------------------
# Block configuration
# -----------------------------------------------------------------------------

REFERENCE_DATASET_SCHEMA_VERSION = "1.0"

ANNOTATION_PERSISTENCE_VERSION = "1.1"

NOTEBOOK_03_HANDOVER_VERSION = "1.1"


REFERENCE_STATUS_VALUES = {
    "machine_supervision",
    "human_reviewed",
    "adjudicated_reference",
    "not_reference_ready",
}


REFERENCE_DATASET_FILENAME = (
    "notebook_03_reference_dataset.json"
)


ACCEPTED_ANNOTATIONS_FILENAME = (
    "notebook_03_accepted_annotations.json"
)


REJECTED_ANNOTATIONS_FILENAME = (
    "notebook_03_rejected_annotations.json"
)


SUPERVISION_SCHEMA_FILENAME = (
    "notebook_03_supervision_schema.json"
)


SUPERVISOR_AGREEMENT_FILENAME = (
    "notebook_03_supervisor_agreement.json"
)


NOTEBOOK_03_HANDOVER_FILENAME = (
    "notebook_03_handover_manifest.json"
)


# -----------------------------------------------------------------------------
# Dependency checks
# -----------------------------------------------------------------------------

required_reference_objects = [
    "ACCEPTED_SUPERVISION_RECORDS",
    "REJECTED_SUPERVISION_RECORDS",
    "SUPERVISOR_AGREEMENT_SUMMARY",
    "ANNOTATION_SCHEMA",
    "LLM_SUPERVISION_PROTOCOL_VERSION",
    "SUPERVISION_SCHEMA_VERSION",
    "ANNOTATION_POLICY_VERSION",
    "SUPERVISOR_REGISTRY_VERSION",
    "PROJECT_NAME",
    "NOTEBOOK_NAME",
    "NOTEBOOK_VERSION",
]


missing_reference_objects = [
    object_name
    for object_name
    in required_reference_objects
    if object_name not in globals()
]


if missing_reference_objects:

    raise NameError(
        "Required Notebook 03 supervision objects are not initialised. "
        "Rerun the required Notebook 03 blocks. Missing: "
        f"{missing_reference_objects}"
    )


# -----------------------------------------------------------------------------
# Persistence dependency checks
# -----------------------------------------------------------------------------

if SAVE_ANNOTATION_OUTPUTS:

    required_persistence_objects = [
        "drive_service",
        "find_drive_file",
        "find_drive_folder",
        "download_json_from_drive",
        "DATA_FOLDER_ID",
        "LOAD_PREPARED_CORPUS",
    ]


    missing_persistence_objects = [
        object_name
        for object_name
        in required_persistence_objects
        if object_name not in globals()
    ]


    if missing_persistence_objects:

        raise NameError(
            "Persistent annotation output requires the Notebook 03 "
            "storage layer. Missing: "
            f"{missing_persistence_objects}"
        )


# -----------------------------------------------------------------------------
# Default reference-dataset state
# -----------------------------------------------------------------------------

REFERENCE_DATASET = None

REFERENCE_DATASET_RECORDS = []

REFERENCE_DATASET_CONSTRUCTED = False

REFERENCE_DATASET_PERSISTED = False


# -----------------------------------------------------------------------------
# Default annotation-persistence state
# -----------------------------------------------------------------------------

ANNOTATION_FOLDER_ID = None

ANNOTATION_OUTPUTS_PERSISTED = False

PERSISTED_ANNOTATION_RECORD_COUNT = 0

PERSISTED_REJECTED_RECORD_COUNT = 0

PERSISTED_NOTEBOOK_03_FILES = {}


# -----------------------------------------------------------------------------
# Default handover state
# -----------------------------------------------------------------------------

NOTEBOOK_03_HANDOVER = None

NOTEBOOK_03_HANDOVER_PERSISTED = False

NOTEBOOK_03_HANDOVER_FILE_ID = None

NOTEBOOK_04_HANDOVER_READY = False


# -----------------------------------------------------------------------------
# Determine reference status
# -----------------------------------------------------------------------------

def determine_reference_status(
    canonical_record,
):
    """
    Determine the current reference status of one valid canonical annotation.

    Machine supervision remains distinct from human review and adjudication.
    """

    review = canonical_record.get(
        "review",
        {}
    )


    review_status = review.get(
        "review_status"
    )


    if review_status == "adjudicated":

        return "adjudicated_reference"


    if review_status == "human_reviewed":

        return "human_reviewed"


    if review_status == "unreviewed":

        return "machine_supervision"


    return "not_reference_ready"


# -----------------------------------------------------------------------------
# Construct one reference-dataset record
# -----------------------------------------------------------------------------

def create_reference_dataset_record(
    canonical_record,
):
    """
    Convert one canonical supervision record into the versioned reference
    dataset representation.

    This conversion does not change its methodological status.
    """

    identity = deepcopy(
        canonical_record.get(
            "identity",
            {}
        )
    )


    context = deepcopy(
        canonical_record.get(
            "context",
            {}
        )
    )


    annotations = deepcopy(
        canonical_record.get(
            "annotations",
            {}
        )
    )


    evidence = deepcopy(
        canonical_record.get(
            "evidence",
            {}
        )
    )


    review = deepcopy(
        canonical_record.get(
            "review",
            {}
        )
    )


    provenance = deepcopy(
        canonical_record.get(
            "provenance",
            {}
        )
    )


    reference_status = (
        determine_reference_status(
            canonical_record
        )
    )


    if (
        reference_status
        not in REFERENCE_STATUS_VALUES
    ):

        raise ValueError(
            "Invalid reference status produced."
        )


    return {
        "record_type":
            "media_ai_reference_supervision_record",

        "reference_dataset_schema_version":
            REFERENCE_DATASET_SCHEMA_VERSION,

        "identity":
            identity,

        "context":
            context,

        "annotations":
            annotations,

        "evidence":
            evidence,

        "review":
            review,

        "provenance":
            provenance,

        "reference_status":
            reference_status,

        "reference_ready":
            reference_status
            == "adjudicated_reference",
    }


# -----------------------------------------------------------------------------
# Controlled reference-dataset construction
# -----------------------------------------------------------------------------

if BUILD_REFERENCE_DATASET:

    REFERENCE_DATASET_RECORDS = [
        create_reference_dataset_record(
            record
        )

        for record
        in ACCEPTED_SUPERVISION_RECORDS
    ]


    reference_created_at = (
        datetime.now(
            timezone.utc
        ).isoformat()
    )


    REFERENCE_DATASET = {
        "artifact_type":
            "media_ai_reference_dataset",

        "schema_version":
            REFERENCE_DATASET_SCHEMA_VERSION,

        "created_at_utc":
            reference_created_at,

        "project": {
            "name":
                PROJECT_NAME,

            "notebook":
                NOTEBOOK_NAME,

            "notebook_version":
                NOTEBOOK_VERSION,
        },

        "supervision": {
            "supervision_schema_version":
                SUPERVISION_SCHEMA_VERSION,

            "annotation_policy_version":
                ANNOTATION_POLICY_VERSION,

            "llm_supervision_protocol_version":
                LLM_SUPERVISION_PROTOCOL_VERSION,

            "supervisor_registry_version":
                SUPERVISOR_REGISTRY_VERSION,

            "agreement_policy_version":
                SUPERVISOR_AGREEMENT_SUMMARY.get(
                    "agreement_policy_version"
                ),
        },

        "counts": {
            "accepted_supervision_records":
                len(
                    ACCEPTED_SUPERVISION_RECORDS
                ),

            "reference_dataset_records":
                len(
                    REFERENCE_DATASET_RECORDS
                ),

            "machine_supervision":
                sum(
                    1
                    for record
                    in REFERENCE_DATASET_RECORDS
                    if record[
                        "reference_status"
                    ]
                    == "machine_supervision"
                ),

            "human_reviewed":
                sum(
                    1
                    for record
                    in REFERENCE_DATASET_RECORDS
                    if record[
                        "reference_status"
                    ]
                    == "human_reviewed"
                ),

            "adjudicated_reference":
                sum(
                    1
                    for record
                    in REFERENCE_DATASET_RECORDS
                    if record[
                        "reference_status"
                    ]
                    == "adjudicated_reference"
                ),

            "not_reference_ready":
                sum(
                    1
                    for record
                    in REFERENCE_DATASET_RECORDS
                    if record[
                        "reference_status"
                    ]
                    == "not_reference_ready"
                ),
        },

        "agreement_summary":
            deepcopy(
                SUPERVISOR_AGREEMENT_SUMMARY
            ),

        "records":
            REFERENCE_DATASET_RECORDS,

        "methodological_guarantees": {
            "raw_provider_outputs_are_reference_labels":
                False,

            "independent_supervisor_records_preserved":
                True,

            "consensus_constructed":
                False,

            "human_review_inferred":
                False,

            "adjudication_inferred":
                False,

            "prepared_text_overwritten":
                False,

            "model_training_performed":
                False,
        },
    }


    REFERENCE_DATASET_CONSTRUCTED = True


# -----------------------------------------------------------------------------
# JSON persistence helper
# -----------------------------------------------------------------------------

def save_or_replace_json_file(
    file_name,
    parent_id,
    payload,
):
    """
    Persist one JSON research artefact using a deterministic filename.
    """

    encoded = json.dumps(
        payload,
        ensure_ascii=False,
        indent=2,
        default=str,
    ).encode(
        "utf-8"
    )


    media = MediaIoBaseUpload(
        io.BytesIO(
            encoded
        ),
        mimetype="application/json",
        resumable=False,
    )


    existing_file_id = find_drive_file(
        file_name,
        parent_id,
    )


    if existing_file_id:

        updated = (
            drive_service.files()
            .update(
                fileId=
                    existing_file_id,

                media_body=
                    media,

                fields=
                    "id,name",
            )
            .execute()
        )


        return updated[
            "id"
        ]


    metadata = {
        "name":
            file_name,

        "parents": [
            parent_id
        ],
    }


    created = (
        drive_service.files()
        .create(
            body=
                metadata,

            media_body=
                media,

            fields=
                "id,name",
        )
        .execute()
    )


    return created[
        "id"
    ]


# -----------------------------------------------------------------------------
# Persistent annotation folder
# -----------------------------------------------------------------------------

if SAVE_ANNOTATION_OUTPUTS:

    if not LOAD_PREPARED_CORPUS:

        raise ValueError(
            "Persistent annotation output requires "
            "prepared-corpus storage access."
        )


    ANNOTATION_FOLDER_ID = find_drive_folder(
        "annotations",
        DATA_FOLDER_ID,
    )


    if ANNOTATION_FOLDER_ID is None:

        metadata = {
            "name":
                "annotations",

            "mimeType":
                "application/vnd.google-apps.folder",

            "parents": [
                DATA_FOLDER_ID
            ],
        }


        created_folder = (
            drive_service.files()
            .create(
                body=
                    metadata,

                fields=
                    "id",
            )
            .execute()
        )


        ANNOTATION_FOLDER_ID = (
            created_folder[
                "id"
            ]
        )


# -----------------------------------------------------------------------------
# Persist Notebook 03 annotation artefacts
# -----------------------------------------------------------------------------

if SAVE_ANNOTATION_OUTPUTS:

    # -------------------------------------------------------------------------
    # Supervision schema
    # -------------------------------------------------------------------------

    schema_file_id = (
        save_or_replace_json_file(
            file_name=
                SUPERVISION_SCHEMA_FILENAME,

            parent_id=
                ANNOTATION_FOLDER_ID,

            payload=
                ANNOTATION_SCHEMA,
        )
    )


    PERSISTED_NOTEBOOK_03_FILES[
        "supervision_schema"
    ] = {
        "filename":
            SUPERVISION_SCHEMA_FILENAME,

        "file_id":
            schema_file_id,
    }


    # -------------------------------------------------------------------------
    # Accepted canonical machine-supervision annotations
    # -------------------------------------------------------------------------

    accepted_payload = {
        "artifact_type":
            "media_ai_accepted_supervision",

        "artifact_version":
            ANNOTATION_PERSISTENCE_VERSION,

        "created_at_utc":
            datetime.now(
                timezone.utc
            ).isoformat(),

        "record_count":
            len(
                ACCEPTED_SUPERVISION_RECORDS
            ),

        "reference_dataset":
            False,

        "records":
            ACCEPTED_SUPERVISION_RECORDS,
    }


    accepted_file_id = (
        save_or_replace_json_file(
            file_name=
                ACCEPTED_ANNOTATIONS_FILENAME,

            parent_id=
                ANNOTATION_FOLDER_ID,

            payload=
                accepted_payload,
        )
    )


    PERSISTED_NOTEBOOK_03_FILES[
        "accepted_annotations"
    ] = {
        "filename":
            ACCEPTED_ANNOTATIONS_FILENAME,

        "file_id":
            accepted_file_id,
    }


    # -------------------------------------------------------------------------
    # Rejected supervision diagnostics
    # -------------------------------------------------------------------------

    rejected_payload = {
        "artifact_type":
            "media_ai_rejected_supervision",

        "artifact_version":
            ANNOTATION_PERSISTENCE_VERSION,

        "created_at_utc":
            datetime.now(
                timezone.utc
            ).isoformat(),

        "record_count":
            len(
                REJECTED_SUPERVISION_RECORDS
            ),

        "records":
            REJECTED_SUPERVISION_RECORDS,
    }


    rejected_file_id = (
        save_or_replace_json_file(
            file_name=
                REJECTED_ANNOTATIONS_FILENAME,

            parent_id=
                ANNOTATION_FOLDER_ID,

            payload=
                rejected_payload,
        )
    )


    PERSISTED_NOTEBOOK_03_FILES[
        "rejected_annotations"
    ] = {
        "filename":
            REJECTED_ANNOTATIONS_FILENAME,

        "file_id":
            rejected_file_id,
    }


    # -------------------------------------------------------------------------
    # Agreement diagnostics
    # -------------------------------------------------------------------------

    agreement_payload = {
        "artifact_type":
            "media_ai_supervisor_agreement",

        "artifact_version":
            ANNOTATION_PERSISTENCE_VERSION,

        "created_at_utc":
            datetime.now(
                timezone.utc
            ).isoformat(),

        "summary":
            SUPERVISOR_AGREEMENT_SUMMARY,

        "pairwise_comparisons":
            (
                SUPERVISOR_PAIRWISE_COMPARISONS
                if "SUPERVISOR_PAIRWISE_COMPARISONS"
                in globals()
                else []
            ),

        "disagreement_flags":
            (
                SUPERVISOR_DISAGREEMENT_FLAGS
                if "SUPERVISOR_DISAGREEMENT_FLAGS"
                in globals()
                else []
            ),
    }


    agreement_file_id = (
        save_or_replace_json_file(
            file_name=
                SUPERVISOR_AGREEMENT_FILENAME,

            parent_id=
                ANNOTATION_FOLDER_ID,

            payload=
                agreement_payload,
        )
    )


    PERSISTED_NOTEBOOK_03_FILES[
        "supervisor_agreement"
    ] = {
        "filename":
            SUPERVISOR_AGREEMENT_FILENAME,

        "file_id":
            agreement_file_id,
    }


    # -------------------------------------------------------------------------
    # Optional reference dataset
    # -------------------------------------------------------------------------

    if REFERENCE_DATASET_CONSTRUCTED:

        reference_dataset_file_id = (
            save_or_replace_json_file(
                file_name=
                    REFERENCE_DATASET_FILENAME,

                parent_id=
                    ANNOTATION_FOLDER_ID,

                payload=
                    REFERENCE_DATASET,
            )
        )


        PERSISTED_NOTEBOOK_03_FILES[
            "reference_dataset"
        ] = {
            "filename":
                REFERENCE_DATASET_FILENAME,

            "file_id":
                reference_dataset_file_id,
        }


        REFERENCE_DATASET_PERSISTED = True


# -----------------------------------------------------------------------------
# Verify persisted annotation artefacts
# -----------------------------------------------------------------------------

if SAVE_ANNOTATION_OUTPUTS:

    accepted_persisted = (
        download_json_from_drive(
            PERSISTED_NOTEBOOK_03_FILES[
                "accepted_annotations"
            ][
                "file_id"
            ]
        )
    )


    rejected_persisted = (
        download_json_from_drive(
            PERSISTED_NOTEBOOK_03_FILES[
                "rejected_annotations"
            ][
                "file_id"
            ]
        )
    )


    persisted_accepted_records = (
        accepted_persisted.get(
            "records",
            [],
        )
    )


    persisted_rejected_records = (
        rejected_persisted.get(
            "records",
            [],
        )
    )


    PERSISTED_ANNOTATION_RECORD_COUNT = len(
        persisted_accepted_records
    )


    PERSISTED_REJECTED_RECORD_COUNT = len(
        persisted_rejected_records
    )


    annotation_record_count_matches = (
        PERSISTED_ANNOTATION_RECORD_COUNT
        == len(
            ACCEPTED_SUPERVISION_RECORDS
        )
    )


    rejected_record_count_matches = (
        PERSISTED_REJECTED_RECORD_COUNT
        == len(
            REJECTED_SUPERVISION_RECORDS
        )
    )


    persisted_sentence_ids = [
        record.get(
            "identity",
            {}
        ).get(
            "sentence_id"
        )

        for record
        in persisted_accepted_records
    ]


    runtime_sentence_ids = [
        record.get(
            "identity",
            {}
        ).get(
            "sentence_id"
        )

        for record
        in ACCEPTED_SUPERVISION_RECORDS
    ]


    persisted_identity_matches = (
        persisted_sentence_ids
        == runtime_sentence_ids
    )


    ANNOTATION_OUTPUTS_PERSISTED = all([
        annotation_record_count_matches,
        rejected_record_count_matches,
        persisted_identity_matches,
        PERSISTED_ANNOTATION_RECORD_COUNT > 0,
    ])


    if not ANNOTATION_OUTPUTS_PERSISTED:

        raise ValueError(
            "Persisted Notebook 03 annotation artefacts failed "
            "post-write verification."
        )


# -----------------------------------------------------------------------------
# Reference-status diagnostics
# -----------------------------------------------------------------------------

if REFERENCE_DATASET_CONSTRUCTED:

    reference_status_counts = {
        status:
            sum(
                1
                for record
                in REFERENCE_DATASET_RECORDS
                if record[
                    "reference_status"
                ]
                == status
            )

        for status
        in sorted(
            REFERENCE_STATUS_VALUES
        )
    }


else:

    reference_status_counts = {
        status:
            0

        for status
        in sorted(
            REFERENCE_STATUS_VALUES
        )
    }


# -----------------------------------------------------------------------------
# Machine-supervision handover state
# -----------------------------------------------------------------------------

MACHINE_SUPERVISION_AVAILABLE = (
    len(
        ACCEPTED_SUPERVISION_RECORDS
    )
    > 0
)


REFERENCE_DATASET_AVAILABLE = all([
    REFERENCE_DATASET_CONSTRUCTED,
    len(
        REFERENCE_DATASET_RECORDS
    )
    > 0,
])


# -----------------------------------------------------------------------------
# Notebook 03 → Notebook 04 handover
# -----------------------------------------------------------------------------

if (
    SAVE_ANNOTATION_OUTPUTS
    and ANNOTATION_OUTPUTS_PERSISTED
):

    handover_created_at = (
        datetime.now(
            timezone.utc
        ).isoformat()
    )


    NOTEBOOK_03_HANDOVER = {
        "artifact_type":
            "media_ai_notebook_handover",

        "handover_version":
            NOTEBOOK_03_HANDOVER_VERSION,

        "created_at_utc":
            handover_created_at,

        "source_notebook":
            NOTEBOOK_NAME,

        "source_notebook_version":
            NOTEBOOK_VERSION,

        "target_notebook":
            "04_model_architecture",

        "project":
            PROJECT_NAME,

        "supervision": {
            "supervision_schema_version":
                SUPERVISION_SCHEMA_VERSION,

            "annotation_policy_version":
                ANNOTATION_POLICY_VERSION,

            "llm_supervision_protocol_version":
                LLM_SUPERVISION_PROTOCOL_VERSION,

            "supervisor_registry_version":
                SUPERVISOR_REGISTRY_VERSION,
        },

        "counts": {
            "accepted_machine_supervision_records":
                len(
                    ACCEPTED_SUPERVISION_RECORDS
                ),

            "rejected_supervision_records":
                len(
                    REJECTED_SUPERVISION_RECORDS
                ),

            "validated_articles":
                sum(
                    1
                    for result
                    in ARTICLE_RESPONSE_VALIDATION_RESULTS
                    if result.get(
                        "valid"
                    )
                )
                if "ARTICLE_RESPONSE_VALIDATION_RESULTS"
                in globals()
                else None,
        },

        "availability": {
            "machine_supervision":
                MACHINE_SUPERVISION_AVAILABLE,

            "reference_dataset":
                REFERENCE_DATASET_AVAILABLE,

            "human_review":
                False,

            "adjudicated_reference":
                False,
        },

        "agreement": {
            "multiple_supervisors_available":
                SUPERVISOR_AGREEMENT_SUMMARY.get(
                    "sentences_with_multiple_supervisors",
                    0,
                )
                > 0,

            "consensus_constructed":
                SUPERVISOR_AGREEMENT_SUMMARY.get(
                    "consensus_constructed",
                    False,
                ),

            "adjudication_performed":
                SUPERVISOR_AGREEMENT_SUMMARY.get(
                    "adjudication_performed",
                    False,
                ),
        },

        "files":
            deepcopy(
                PERSISTED_NOTEBOOK_03_FILES
            ),

        "methodological_boundaries": {
            "machine_supervision_is_reference_dataset":
                False,

            "machine_supervision_is_human_review":
                False,

            "machine_supervision_is_adjudication":
                False,

            "reference_dataset_required_for_handover":
                False,

            "model_training_performed":
                False,

            "recurrent_state_calculated":
                False,

            "confluent_representation_calculated":
                False,
        },

        "next_stage":
            "model_architecture",
    }


    NOTEBOOK_03_HANDOVER_FILE_ID = (
        save_or_replace_json_file(
            file_name=
                NOTEBOOK_03_HANDOVER_FILENAME,

            parent_id=
                ANNOTATION_FOLDER_ID,

            payload=
                NOTEBOOK_03_HANDOVER,
        )
    )


    PERSISTED_NOTEBOOK_03_FILES[
        "handover_manifest"
    ] = {
        "filename":
            NOTEBOOK_03_HANDOVER_FILENAME,

        "file_id":
            NOTEBOOK_03_HANDOVER_FILE_ID,
    }


    persisted_handover = (
        download_json_from_drive(
            NOTEBOOK_03_HANDOVER_FILE_ID
        )
    )


    NOTEBOOK_03_HANDOVER_PERSISTED = all([
        isinstance(
            persisted_handover,
            dict,
        ),

        persisted_handover.get(
            "handover_version"
        )
        == NOTEBOOK_03_HANDOVER_VERSION,

        persisted_handover.get(
            "availability",
            {}
        ).get(
            "machine_supervision"
        )
        is True,

        persisted_handover.get(
            "counts",
            {}
        ).get(
            "accepted_machine_supervision_records"
        )
        == len(
            ACCEPTED_SUPERVISION_RECORDS
        ),
    ])


# -----------------------------------------------------------------------------
# Final Notebook 04 handover readiness
# -----------------------------------------------------------------------------

NOTEBOOK_04_HANDOVER_READY = all([
    MACHINE_SUPERVISION_AVAILABLE,
    ANNOTATION_OUTPUTS_PERSISTED,
    NOTEBOOK_03_HANDOVER_PERSISTED,
])


# -----------------------------------------------------------------------------
# Verification
# -----------------------------------------------------------------------------

print("=" * 72)

print(
    "Media AI — Reference Dataset Construction and Persistent Annotation Artefacts"
)

print("=" * 72)


print(
    f"Build reference dataset       : "
    f"{BUILD_REFERENCE_DATASET}"
)

print(
    f"Save annotation outputs       : "
    f"{SAVE_ANNOTATION_OUTPUTS}"
)

print(
    f"Accepted annotations          : "
    f"{len(ACCEPTED_SUPERVISION_RECORDS)}"
)


print("-" * 72)

print(
    "Machine-supervision persistence"
)

print(
    f"Machine supervision available : "
    f"{MACHINE_SUPERVISION_AVAILABLE}"
)

print(
    f"Annotation outputs persisted  : "
    f"{ANNOTATION_OUTPUTS_PERSISTED}"
)

print(
    f"Persisted annotation records  : "
    f"{PERSISTED_ANNOTATION_RECORD_COUNT}"
)

print(
    f"Persisted rejected records    : "
    f"{PERSISTED_REJECTED_RECORD_COUNT}"
)


print("-" * 72)

print(
    "Reference dataset"
)

print(
    f"Reference dataset built       : "
    f"{REFERENCE_DATASET_CONSTRUCTED}"
)

print(
    f"Reference records             : "
    f"{len(REFERENCE_DATASET_RECORDS)}"
)

print(
    f"Reference dataset persisted   : "
    f"{REFERENCE_DATASET_PERSISTED}"
)

print(
    f"Reference dataset available   : "
    f"{REFERENCE_DATASET_AVAILABLE}"
)


print("-" * 72)

print(
    "Reference status"
)


for status in sorted(
    REFERENCE_STATUS_VALUES
):

    print(
        f"  {status:<23}: "
        f"{reference_status_counts[status]}"
    )


print("-" * 72)

print(
    "Notebook 03 → Notebook 04 handover"
)

print(
    f"Handover version              : "
    f"{NOTEBOOK_03_HANDOVER_VERSION}"
)

print(
    f"Handover manifest persisted   : "
    f"{NOTEBOOK_03_HANDOVER_PERSISTED}"
)

print(
    f"Handover manifest file ID     : "
    f"{NOTEBOOK_03_HANDOVER_FILE_ID}"
)

print(
    f"Notebook 04 handover ready    : "
    f"{NOTEBOOK_04_HANDOVER_READY}"
)


print("-" * 72)

print(
    "Persisted artefacts"
)


for (
    artefact_name,
    artefact_info,
) in sorted(
    PERSISTED_NOTEBOOK_03_FILES.items()
):

    print(
        f"  {artefact_name:<24}: "
        f"{artefact_info.get('filename')}"
    )


print("=" * 72)


if not BUILD_REFERENCE_DATASET:

    print(
        "Reference-dataset construction was skipped by the "
        "Execution Control Panel."
    )


else:

    print(
        "Reference-dataset construction was enabled by the "
        "Execution Control Panel."
    )


if ANNOTATION_OUTPUTS_PERSISTED:

    print(
        "Canonical machine-supervision annotations were persisted "
        "independently of reference-dataset construction."
    )


if NOTEBOOK_03_HANDOVER_PERSISTED:

    print(
        "Notebook 03 handover was persisted and verified successfully."
    )


print(
    "Independent machine supervision has not been treated as "
    "adjudicated human reference."
)

print(
    "Machine supervision remains available to downstream architecture "
    "without being promoted to a reference dataset."
)

print(
    "No consensus, human review, or adjudication has been inferred."
)

print(
    "No model training, recurrent-state calculation, or confluent "
    "representation has been performed."
)

print(
    "No external model or API calls have been made by this block."
)

print("=" * 72)
```

## Final Notebook 03 Audit and Notebook 04 Handover Readiness

This block performs the final audit of Notebook 03.

It verifies the supervision and annotation infrastructure created throughout the
notebook and determines whether the project is ready to hand a supervised
reference dataset to Notebook 04.

Two completion states are intentionally distinguished:

1. **Notebook 03 pipeline completeness**
2. **Notebook 04 data-handover readiness**

These states are not equivalent.

The Notebook 03 supervision pipeline may be fully implemented and internally
consistent even when external supervision has not yet been executed.

Notebook 04 should not, however, be declared data-ready until valid supervision
records have been created, accepted, assembled into the reference dataset and
persisted.

### Pipeline Audit

The final audit verifies that:

- the Notebook 02 prepared corpus is loaded and valid;
- the annotation schema is initialised;
- all supervision constructs remain versioned;
- the sentence-level annotation template exists;
- the number of supervision contexts matches the prepared sentence count;
- the independent LLM supervision protocol exists;
- all provider-independent supervision payloads are valid;
- the five-slot supervisor registry is present;
- the master and individual supervisor controls remain visible;
- canonical response-validation infrastructure exists;
- supervisor-comparison infrastructure exists;
- reference-dataset construction infrastructure exists.

### Supervision State

The audit reports the current external-supervision state separately.

This includes:

- master LLM supervision switch;
- enabled supervisors;
- configured supervisors;
- executable supervisors;
- raw supervision responses;
- accepted canonical annotations;
- rejected responses.

A zero-response state is valid when external supervision is disabled.

### Agreement State

Supervisor agreement cannot be calculated until at least two independent
supervisors have produced valid annotations for the same sentence.

The final audit therefore reports whether agreement analysis is:

- structurally available;
- empirically populated.

Absence of pairwise comparisons before multi-supervisor execution is not an
error.

### Reference Dataset State

The audit distinguishes:

- construction enabled or disabled;
- dataset constructed or not constructed;
- number of machine-supervision records;
- human-reviewed records;
- adjudicated reference records;
- persistence status.

Independent LLM annotations are not silently promoted to adjudicated reference
labels.

### Notebook 04 Readiness

Notebook 04 data-handover readiness requires, at minimum:

- at least one accepted canonical supervision record;
- a constructed reference dataset;
- at least one reference-dataset record;
- persisted Notebook 03 supervision artefacts;
- a valid Notebook 04 handover manifest.

The exact minimum dataset size required for meaningful training is a separate
experimental question.

This readiness criterion only determines whether Notebook 04 has a valid
supervision artefact to consume.

### Methodological Boundary

The final audit verifies that Notebook 03 has not:

- overwritten prepared source text;
- introduced permanent model-specific tokenisation;
- performed model training;
- calculated recurrent carry-forward state;
- calculated confluent representations;
- inferred consensus without an explicit rule;
- inferred human review or adjudication.

### Output

This block creates:

- `NOTEBOOK_03_PIPELINE_COMPLETE`;
- `NOTEBOOK_04_DATA_READY`;
- a final supervision-state summary;
- explicit readiness blockers where applicable.

No external model or API request is performed by this block.

```python
# =============================================================================
# Media AI — Final Notebook 03 Audit and Notebook 04 Handover Readiness
# =============================================================================


# -----------------------------------------------------------------------------
# Final-audit configuration
# -----------------------------------------------------------------------------

FINAL_AUDIT_VERSION = "1.4"

EXPECTED_SUPERVISOR_SLOT_COUNT = 5

EXPECTED_PRIMARY_ANNOTATION_UNIT = "sentence"

EXPECTED_EXTERNAL_REQUEST_UNIT = "article"

EXPECTED_LLM_SUPERVISION_PROTOCOL_VERSION = "1.1"

EXPECTED_NOTEBOOK_03_HANDOVER_VERSION = "1.1"


# -----------------------------------------------------------------------------
# Required pipeline objects
# -----------------------------------------------------------------------------

required_pipeline_objects = [
    # Prepared corpus
    "PREPARED_CORPUS",
    "PREPARED_ARTICLE_RECORDS",
    "PREPARED_CORPUS_LOADED",
    "PREPARED_CORPUS_VERIFIED",
    "PREPARED_ARTICLE_COUNT",
    "PREPARED_PARAGRAPH_COUNT",
    "PREPARED_SENTENCE_COUNT",

    # Annotation schema and record contract
    "ANNOTATION_SCHEMA",
    "ANNOTATION_RECORD_TEMPLATE",
    "ANNOTATION_RECORD_SCHEMA_VERSION",
    "ANNOTATION_RECORD_SCHEMA_VALID",
    "SUPERVISION_SCHEMA_VERSION",
    "ANNOTATION_POLICY_VERSION",
    "SUPERVISION_CONTEXT_RECORDS",
    "BLANK_ANNOTATION_RECORDS",
    "EMOTIONAL_DIMENSIONS",
    "FACTOLOGICAL_COMPONENTS",
    "SUPERVISION_CONSTRUCTS",
    "NON_NEGATIVE_STATE_SCALE",
    "SIGNED_TRANSFORMATION_SCALE",
    "NON_NEGATIVE_WEIGHT_SCALE",
    "CONFIDENCE_SCALE",
    "template_dimensions_valid",
    "factological_template_valid",
    "blank_records_valid",
    "context_ids_present",
    "context_ids_unique",

    # Supervision protocol
    "LLM_SUPERVISION_PROTOCOL_VERSION",
    "LLM_SUPERVISION_REQUEST_UNIT",
    "LLM_SUPERVISION_ANNOTATION_UNIT",
    "LLM_SUPERVISION_PAYLOADS",
    "INVALID_LLM_PAYLOADS",
    "LLM_ARTICLE_SUPERVISION_PAYLOADS",

    # Supervisor execution
    "RUN_LLM_SUPERVISION",
    "SUPERVISOR_REGISTRY",
    "SUPERVISOR_READINESS",
    "LLM_SUPERVISION_RAW_RESULTS",

    # Canonical validation
    "CANONICAL_SUPERVISION_RECORDS_BY_SUPERVISOR",
    "ACCEPTED_SUPERVISION_RECORDS",
    "REJECTED_SUPERVISION_RECORDS",
    "ARTICLE_RESPONSE_VALIDATION_RESULTS",
    "SUPERVISION_VALIDATION_RESULTS",
    "validation_accounting_valid",
    "article_validation_accounting_valid",
    "CANONICAL_SUPERVISION_VALID",

    # Agreement
    "SUPERVISOR_AGREEMENT_SUMMARY",
    "SUPERVISOR_PAIRWISE_COMPARISONS",

    # Reference layer
    "REFERENCE_DATASET_SCHEMA_VERSION",
    "REFERENCE_STATUS_VALUES",
    "REFERENCE_DATASET_RECORDS",
    "REFERENCE_DATASET_CONSTRUCTED",
    "REFERENCE_DATASET_PERSISTED",
    "REFERENCE_DATASET_AVAILABLE",
    "BUILD_REFERENCE_DATASET",

    # Persistent machine supervision
    "SAVE_ANNOTATION_OUTPUTS",
    "MACHINE_SUPERVISION_AVAILABLE",
    "ANNOTATION_OUTPUTS_PERSISTED",
    "PERSISTED_ANNOTATION_RECORD_COUNT",
    "PERSISTED_REJECTED_RECORD_COUNT",
    "PERSISTED_NOTEBOOK_03_FILES",

    # Notebook 03 → Notebook 04 handover
    "NOTEBOOK_03_HANDOVER",
    "NOTEBOOK_03_HANDOVER_PERSISTED",
    "NOTEBOOK_03_HANDOVER_FILE_ID",
    "NOTEBOOK_04_HANDOVER_READY",
]


missing_pipeline_objects = [
    object_name
    for object_name
    in required_pipeline_objects
    if object_name not in globals()
]


if missing_pipeline_objects:

    raise NameError(
        "Required Notebook 03 pipeline objects are not initialised. "
        "Rerun the required Notebook 03 blocks. Missing: "
        f"{missing_pipeline_objects}"
    )


# -----------------------------------------------------------------------------
# Prepared-corpus verification
# -----------------------------------------------------------------------------

prepared_input_verified = all([
    PREPARED_CORPUS_LOADED
    is True,

    PREPARED_CORPUS_VERIFIED
    is True,

    isinstance(
        PREPARED_ARTICLE_RECORDS,
        list,
    ),

    PREPARED_ARTICLE_COUNT
    == len(
        PREPARED_ARTICLE_RECORDS
    ),

    PREPARED_SENTENCE_COUNT
    == len(
        SUPERVISION_CONTEXT_RECORDS
    ),
])


# -----------------------------------------------------------------------------
# Annotation-schema verification
# -----------------------------------------------------------------------------
#
# ANNOTATION_SCHEMA is the conceptual supervision schema.
#
# Numerical scales belong to the separate completed-record validation
# contract and are therefore not expected to be embedded in this object.
# -----------------------------------------------------------------------------

annotation_schema_verified = all([
    isinstance(
        ANNOTATION_SCHEMA,
        dict,
    ),

    ANNOTATION_SCHEMA.get(
        "schema_type"
    )
    == "media_ai_supervision_schema",

    ANNOTATION_SCHEMA.get(
        "schema_version"
    )
    == SUPERVISION_SCHEMA_VERSION,

    ANNOTATION_SCHEMA.get(
        "annotation_policy_version"
    )
    == ANNOTATION_POLICY_VERSION,

    ANNOTATION_SCHEMA.get(
        "primary_annotation_unit"
    )
    == EXPECTED_PRIMARY_ANNOTATION_UNIT,

    isinstance(
        ANNOTATION_SCHEMA.get(
            "emotional_dimensions"
        ),
        dict,
    ),

    set(
        ANNOTATION_SCHEMA.get(
            "emotional_dimensions",
            {},
        ).keys()
    )
    == set(
        EMOTIONAL_DIMENSIONS.keys()
    ),

    isinstance(
        ANNOTATION_SCHEMA.get(
            "factological_components"
        ),
        dict,
    ),

    set(
        ANNOTATION_SCHEMA.get(
            "factological_components",
            {},
        ).keys()
    )
    == set(
        FACTOLOGICAL_COMPONENTS.keys()
    ),

    isinstance(
        ANNOTATION_SCHEMA.get(
            "constructs"
        ),
        dict,
    ),

    len(
        ANNOTATION_SCHEMA.get(
            "constructs",
            {},
        )
    )
    == len(
        SUPERVISION_CONSTRUCTS
    ),

    isinstance(
        ANNOTATION_SCHEMA.get(
            "context_policy"
        ),
        dict,
    ),

    isinstance(
        ANNOTATION_SCHEMA.get(
            "methodological_constraints"
        ),
        dict,
    ),
])


# -----------------------------------------------------------------------------
# Numerical validation-contract verification
# -----------------------------------------------------------------------------
#
# Contractual numerical bounds are validated independently of the conceptual
# annotation schema.
#
# Additional descriptive fields are permitted. In particular, the signed
# transformation scale explicitly defines a neutral point at zero.
# -----------------------------------------------------------------------------

numerical_scale_contract_verified = all([
    isinstance(
        NON_NEGATIVE_STATE_SCALE,
        dict,
    ),

    NON_NEGATIVE_STATE_SCALE.get(
        "minimum"
    )
    == 0.0,

    NON_NEGATIVE_STATE_SCALE.get(
        "maximum"
    )
    == 1.0,

    isinstance(
        SIGNED_TRANSFORMATION_SCALE,
        dict,
    ),

    SIGNED_TRANSFORMATION_SCALE.get(
        "minimum"
    )
    == -1.0,

    SIGNED_TRANSFORMATION_SCALE.get(
        "maximum"
    )
    == 1.0,

    SIGNED_TRANSFORMATION_SCALE.get(
        "neutral",
        0.0,
    )
    == 0.0,

    isinstance(
        NON_NEGATIVE_WEIGHT_SCALE,
        dict,
    ),

    NON_NEGATIVE_WEIGHT_SCALE.get(
        "minimum"
    )
    == 0.0,

    NON_NEGATIVE_WEIGHT_SCALE.get(
        "maximum"
    )
    == 1.0,

    isinstance(
        CONFIDENCE_SCALE,
        dict,
    ),

    CONFIDENCE_SCALE.get(
        "minimum"
    )
    == 0.0,

    CONFIDENCE_SCALE.get(
        "maximum"
    )
    == 1.0,
])


# -----------------------------------------------------------------------------
# Annotation-template verification
# -----------------------------------------------------------------------------

annotation_template_verified = all([
    ANNOTATION_RECORD_TEMPLATE.get(
        "record_type"
    )
    == "media_ai_supervision_record",

    ANNOTATION_RECORD_TEMPLATE.get(
        "record_schema_version"
    )
    == ANNOTATION_RECORD_SCHEMA_VERSION,

    template_dimensions_valid
    is True,

    factological_template_valid
    is True,

    blank_records_valid
    is True,

    ANNOTATION_RECORD_SCHEMA_VALID
    is True,
])


# -----------------------------------------------------------------------------
# Supervision-context verification
# -----------------------------------------------------------------------------

supervision_context_count = len(
    SUPERVISION_CONTEXT_RECORDS
)


blank_annotation_count = len(
    BLANK_ANNOTATION_RECORDS
)


context_counts_verified = all([
    supervision_context_count
    == PREPARED_SENTENCE_COUNT,

    blank_annotation_count
    == PREPARED_SENTENCE_COUNT,

    context_ids_present
    is True,

    context_ids_unique
    is True,
])


# -----------------------------------------------------------------------------
# LLM supervision protocol verification
# -----------------------------------------------------------------------------
#
# Concrete protocol objects produced by the supervision protocol block are
# validated directly.
#
# Historical loose ALLOW_* runtime variables are deliberately not required.
# -----------------------------------------------------------------------------

sentence_payload_count = len(
    LLM_SUPERVISION_PAYLOADS
)


invalid_sentence_payload_count = len(
    INVALID_LLM_PAYLOADS
)


article_payload_count = len(
    LLM_ARTICLE_SUPERVISION_PAYLOADS
)


article_payload_sentence_count = sum(
    payload.get(
        "article",
        {},
    ).get(
        "sentence_count",
        0,
    )

    for payload
    in LLM_ARTICLE_SUPERVISION_PAYLOADS
)


article_payload_ids = [
    payload.get(
        "identity",
        {},
    ).get(
        "media_ai_record_id"
    )

    for payload
    in LLM_ARTICLE_SUPERVISION_PAYLOADS
]


article_payload_ids_present = (
    bool(
        article_payload_ids
    )
    and all(
        article_payload_ids
    )
)


article_payload_ids_unique = (
    article_payload_ids_present
    and len(
        article_payload_ids
    )
    == len(
        set(
            article_payload_ids
        )
    )
)


llm_protocol_verified = all([
    LLM_SUPERVISION_PROTOCOL_VERSION
    == EXPECTED_LLM_SUPERVISION_PROTOCOL_VERSION,

    LLM_SUPERVISION_REQUEST_UNIT
    == EXPECTED_EXTERNAL_REQUEST_UNIT,

    LLM_SUPERVISION_ANNOTATION_UNIT
    == EXPECTED_PRIMARY_ANNOTATION_UNIT,

    sentence_payload_count
    == PREPARED_SENTENCE_COUNT,

    invalid_sentence_payload_count
    == 0,

    article_payload_count
    == PREPARED_ARTICLE_COUNT,

    article_payload_sentence_count
    == PREPARED_SENTENCE_COUNT,

    article_payload_ids_present,

    article_payload_ids_unique,
])


# -----------------------------------------------------------------------------
# Supervisor-registry verification
# -----------------------------------------------------------------------------

supervisor_registry_verified = all([
    len(
        SUPERVISOR_REGISTRY
    )
    == EXPECTED_SUPERVISOR_SLOT_COUNT,

    len(
        set(
            SUPERVISOR_REGISTRY.keys()
        )
    )
    == EXPECTED_SUPERVISOR_SLOT_COUNT,

    "supervisor_01"
    in SUPERVISOR_REGISTRY,

    "supervisor_05"
    in SUPERVISOR_REGISTRY,
])


enabled_supervisor_ids = [
    supervisor_id

    for (
        supervisor_id,
        configuration,
    )
    in SUPERVISOR_REGISTRY.items()

    if configuration.get(
        "enabled",
        False,
    )
]


configured_supervisor_ids = [
    supervisor_id

    for (
        supervisor_id,
        configuration,
    )
    in SUPERVISOR_REGISTRY.items()

    if all([
        bool(
            configuration.get(
                "provider"
            )
        ),

        bool(
            configuration.get(
                "model"
            )
        ),
    ])
]


executable_supervisor_ids = [
    supervisor_id

    for (
        supervisor_id,
        readiness,
    )
    in SUPERVISOR_READINESS.items()

    if readiness.get(
        "executable",
        False,
    )
]


# -----------------------------------------------------------------------------
# Raw article-level supervision verification
# -----------------------------------------------------------------------------

final_raw_article_response_count = sum(
    len(
        supervisor_results
    )

    for supervisor_results
    in LLM_SUPERVISION_RAW_RESULTS.values()
)


successful_raw_article_count = sum(
    1

    for supervisor_results
    in LLM_SUPERVISION_RAW_RESULTS.values()

    for result
    in supervisor_results

    if result.get(
        "execution",
        {},
    ).get(
        "status"
    )
    == "SUCCESS"
)


error_raw_article_count = sum(
    1

    for supervisor_results
    in LLM_SUPERVISION_RAW_RESULTS.values()

    for result
    in supervisor_results

    if result.get(
        "execution",
        {},
    ).get(
        "status"
    )
    == "ERROR"
)


blocked_raw_article_count = sum(
    1

    for supervisor_results
    in LLM_SUPERVISION_RAW_RESULTS.values()

    for result
    in supervisor_results

    if result.get(
        "execution",
        {},
    ).get(
        "status"
    )
    == "BLOCKED"
)


other_raw_article_count = (
    final_raw_article_response_count
    - successful_raw_article_count
    - error_raw_article_count
    - blocked_raw_article_count
)


raw_article_accounting_verified = all([
    final_raw_article_response_count
    >= 1,

    other_raw_article_count
    == 0,

    final_raw_article_response_count
    == (
        successful_raw_article_count
        + error_raw_article_count
        + blocked_raw_article_count
    ),
])


# -----------------------------------------------------------------------------
# Canonical sentence-level supervision verification
# -----------------------------------------------------------------------------

final_accepted_count = len(
    ACCEPTED_SUPERVISION_RECORDS
)


final_rejected_count = len(
    REJECTED_SUPERVISION_RECORDS
)


sentence_validation_result_count = len(
    SUPERVISION_VALIDATION_RESULTS
)


accepted_sentence_ids = [
    record.get(
        "identity",
        {},
    ).get(
        "sentence_id"
    )

    for record
    in ACCEPTED_SUPERVISION_RECORDS
]


accepted_sentence_ids_present = (
    bool(
        accepted_sentence_ids
    )
    and all(
        accepted_sentence_ids
    )
)


accepted_sentence_ids_unique = (
    accepted_sentence_ids_present
    and len(
        accepted_sentence_ids
    )
    == len(
        set(
            accepted_sentence_ids
        )
    )
)


# -----------------------------------------------------------------------------
# Article-validation verification
# -----------------------------------------------------------------------------

article_validation_result_count = len(
    ARTICLE_RESPONSE_VALIDATION_RESULTS
)


validated_article_count = sum(
    1

    for result
    in ARTICLE_RESPONSE_VALIDATION_RESULTS

    if result.get(
        "valid"
    )
    is True
)


rejected_article_count = sum(
    1

    for result
    in ARTICLE_RESPONSE_VALIDATION_RESULTS

    if result.get(
        "valid"
    )
    is False
)


article_validation_verified = all([
    article_validation_accounting_valid
    is True,

    article_validation_result_count
    == (
        validated_article_count
        + rejected_article_count
    ),

    validated_article_count
    > 0,
])


# -----------------------------------------------------------------------------
# Canonical-validation verification
# -----------------------------------------------------------------------------

canonical_validation_verified = all([
    CANONICAL_SUPERVISION_VALID
    is True,

    validation_accounting_valid
    is True,

    final_accepted_count
    > 0,

    final_rejected_count
    >= 0,

    sentence_validation_result_count
    == (
        final_accepted_count
        + final_rejected_count
    ),

    accepted_sentence_ids_present,

    accepted_sentence_ids_unique,

    raw_article_accounting_verified,

    article_validation_verified,
])


# -----------------------------------------------------------------------------
# Supervisor-accounting verification
# -----------------------------------------------------------------------------

canonical_records_by_supervisor_count = sum(
    len(
        records
    )

    for records
    in CANONICAL_SUPERVISION_RECORDS_BY_SUPERVISOR.values()
)


supervisor_annotation_accounting_verified = (
    canonical_records_by_supervisor_count
    == final_accepted_count
)


# -----------------------------------------------------------------------------
# Agreement-layer verification
# -----------------------------------------------------------------------------

pairwise_comparison_count = len(
    SUPERVISOR_PAIRWISE_COMPARISONS
)


multi_supervisor_sentence_count = (
    SUPERVISOR_AGREEMENT_SUMMARY.get(
        "sentences_with_multiple_supervisors",
        0,
    )
)


agreement_layer_verified = all([
    isinstance(
        SUPERVISOR_AGREEMENT_SUMMARY,
        dict,
    ),

    SUPERVISOR_AGREEMENT_SUMMARY.get(
        "accepted_annotations"
    )
    == final_accepted_count,

    SUPERVISOR_AGREEMENT_SUMMARY.get(
        "sentences_with_annotations"
    )
    == len(
        set(
            accepted_sentence_ids
        )
    ),

    SUPERVISOR_AGREEMENT_SUMMARY.get(
        "consensus_constructed"
    )
    is False,

    SUPERVISOR_AGREEMENT_SUMMARY.get(
        "adjudication_performed"
    )
    is False,

    supervisor_annotation_accounting_verified,
])


# -----------------------------------------------------------------------------
# Reference-dataset verification
# -----------------------------------------------------------------------------
#
# Reference-dataset construction remains optional.
#
# Notebook 04 may consume validated persistent machine supervision without
# treating it as a human-reviewed or adjudicated reference dataset.
# -----------------------------------------------------------------------------

reference_record_count = len(
    REFERENCE_DATASET_RECORDS
)


machine_supervision_reference_count = sum(
    1

    for record
    in REFERENCE_DATASET_RECORDS

    if record.get(
        "reference_status"
    )
    == "machine_supervision"
)


human_reviewed_count = sum(
    1

    for record
    in REFERENCE_DATASET_RECORDS

    if record.get(
        "reference_status"
    )
    == "human_reviewed"
)


adjudicated_reference_count = sum(
    1

    for record
    in REFERENCE_DATASET_RECORDS

    if record.get(
        "reference_status"
    )
    == "adjudicated_reference"
)


not_reference_ready_count = sum(
    1

    for record
    in REFERENCE_DATASET_RECORDS

    if record.get(
        "reference_status"
    )
    == "not_reference_ready"
)


reference_layer_verified = all([
    REFERENCE_DATASET_SCHEMA_VERSION
    == "1.0",

    isinstance(
        REFERENCE_DATASET_RECORDS,
        list,
    ),

    REFERENCE_STATUS_VALUES
    == {
        "machine_supervision",
        "human_reviewed",
        "adjudicated_reference",
        "not_reference_ready",
    },

    isinstance(
        REFERENCE_DATASET_AVAILABLE,
        bool,
    ),

    REFERENCE_DATASET_AVAILABLE
    == all([
        REFERENCE_DATASET_CONSTRUCTED,
        reference_record_count > 0,
    ]),
])


# -----------------------------------------------------------------------------
# Persistent machine-supervision verification
# -----------------------------------------------------------------------------

required_persisted_artefacts = {
    "accepted_annotations",
    "rejected_annotations",
    "supervision_schema",
    "supervisor_agreement",
    "handover_manifest",
}


persisted_artefact_names = set(
    PERSISTED_NOTEBOOK_03_FILES.keys()
)


persisted_artefact_set_verified = (
    required_persisted_artefacts
    .issubset(
        persisted_artefact_names
    )
)


accepted_persistence_verified = (
    PERSISTED_ANNOTATION_RECORD_COUNT
    == final_accepted_count
)


rejected_persistence_verified = (
    PERSISTED_REJECTED_RECORD_COUNT
    == final_rejected_count
)


machine_supervision_persistence_verified = all([
    SAVE_ANNOTATION_OUTPUTS
    is True,

    MACHINE_SUPERVISION_AVAILABLE
    is True,

    ANNOTATION_OUTPUTS_PERSISTED
    is True,

    final_accepted_count
    > 0,

    accepted_persistence_verified,

    rejected_persistence_verified,

    persisted_artefact_set_verified,
])


# -----------------------------------------------------------------------------
# Notebook 03 → Notebook 04 handover verification
# -----------------------------------------------------------------------------

handover_availability = (
    NOTEBOOK_03_HANDOVER.get(
        "availability",
        {},
    )
)


handover_counts = (
    NOTEBOOK_03_HANDOVER.get(
        "counts",
        {},
    )
)


handover_agreement = (
    NOTEBOOK_03_HANDOVER.get(
        "agreement",
        {},
    )
)


handover_boundaries = (
    NOTEBOOK_03_HANDOVER.get(
        "methodological_boundaries",
        {},
    )
)


handover_manifest_verified = all([
    isinstance(
        NOTEBOOK_03_HANDOVER,
        dict,
    ),

    NOTEBOOK_03_HANDOVER.get(
        "artifact_type"
    )
    == "media_ai_notebook_handover",

    NOTEBOOK_03_HANDOVER.get(
        "handover_version"
    )
    == EXPECTED_NOTEBOOK_03_HANDOVER_VERSION,

    handover_availability.get(
        "machine_supervision"
    )
    is True,

    handover_availability.get(
        "reference_dataset"
    )
    == REFERENCE_DATASET_AVAILABLE,

    handover_availability.get(
        "human_review"
    )
    is False,

    handover_availability.get(
        "adjudicated_reference"
    )
    is False,

    handover_counts.get(
        "accepted_machine_supervision_records"
    )
    == final_accepted_count,

    handover_counts.get(
        "rejected_supervision_records"
    )
    == final_rejected_count,

    handover_agreement.get(
        "consensus_constructed"
    )
    is False,

    handover_agreement.get(
        "adjudication_performed"
    )
    is False,

    NOTEBOOK_03_HANDOVER_PERSISTED
    is True,

    NOTEBOOK_03_HANDOVER_FILE_ID
    is not None,
])


# -----------------------------------------------------------------------------
# Methodological-boundary verification
# -----------------------------------------------------------------------------
#
# The persistent handover manifest is the authoritative downstream
# methodological contract.
# -----------------------------------------------------------------------------

methodological_boundary_verified = all([
    handover_boundaries.get(
        "machine_supervision_is_reference_dataset"
    )
    is False,

    handover_boundaries.get(
        "machine_supervision_is_human_review"
    )
    is False,

    handover_boundaries.get(
        "machine_supervision_is_adjudication"
    )
    is False,

    handover_boundaries.get(
        "reference_dataset_required_for_handover"
    )
    is False,

    handover_boundaries.get(
        "model_training_performed"
    )
    is False,

    handover_boundaries.get(
        "recurrent_state_calculated"
    )
    is False,

    handover_boundaries.get(
        "confluent_representation_calculated"
    )
    is False,
])


# -----------------------------------------------------------------------------
# Notebook 03 pipeline-completion criterion
# -----------------------------------------------------------------------------

audit_checks = {
    "prepared_input_verified":
        prepared_input_verified,

    "annotation_schema_verified":
        annotation_schema_verified,

    "numerical_scale_contract_verified":
        numerical_scale_contract_verified,

    "annotation_template_verified":
        annotation_template_verified,

    "context_counts_verified":
        context_counts_verified,

    "llm_protocol_verified":
        llm_protocol_verified,

    "supervisor_registry_verified":
        supervisor_registry_verified,

    "canonical_validation_verified":
        canonical_validation_verified,

    "agreement_layer_verified":
        agreement_layer_verified,

    "reference_layer_verified":
        reference_layer_verified,

    "machine_supervision_persistence_verified":
        machine_supervision_persistence_verified,

    "handover_manifest_verified":
        handover_manifest_verified,

    "methodological_boundary_verified":
        methodological_boundary_verified,
}


NOTEBOOK_03_PIPELINE_COMPLETE = all(
    audit_checks.values()
)


# -----------------------------------------------------------------------------
# Notebook 04 readiness blockers
# -----------------------------------------------------------------------------

NOTEBOOK_04_READINESS_BLOCKERS = []


if not prepared_input_verified:

    NOTEBOOK_04_READINESS_BLOCKERS.append(
        "prepared_input_not_verified"
    )


if not annotation_schema_verified:

    NOTEBOOK_04_READINESS_BLOCKERS.append(
        "annotation_schema_not_verified"
    )


if not numerical_scale_contract_verified:

    NOTEBOOK_04_READINESS_BLOCKERS.append(
        "numerical_scale_contract_not_verified"
    )


if not annotation_template_verified:

    NOTEBOOK_04_READINESS_BLOCKERS.append(
        "annotation_template_not_verified"
    )


if not llm_protocol_verified:

    NOTEBOOK_04_READINESS_BLOCKERS.append(
        "llm_protocol_not_verified"
    )


if not canonical_validation_verified:

    NOTEBOOK_04_READINESS_BLOCKERS.append(
        "canonical_supervision_not_verified"
    )


if final_accepted_count == 0:

    NOTEBOOK_04_READINESS_BLOCKERS.append(
        "no_accepted_supervision_records"
    )


if not MACHINE_SUPERVISION_AVAILABLE:

    NOTEBOOK_04_READINESS_BLOCKERS.append(
        "machine_supervision_not_available"
    )


if not ANNOTATION_OUTPUTS_PERSISTED:

    NOTEBOOK_04_READINESS_BLOCKERS.append(
        "annotation_outputs_not_persisted"
    )


if not accepted_persistence_verified:

    NOTEBOOK_04_READINESS_BLOCKERS.append(
        "persisted_annotation_count_mismatch"
    )


if not rejected_persistence_verified:

    NOTEBOOK_04_READINESS_BLOCKERS.append(
        "persisted_rejected_count_mismatch"
    )


if not persisted_artefact_set_verified:

    NOTEBOOK_04_READINESS_BLOCKERS.append(
        "required_persisted_artefacts_missing"
    )


if not NOTEBOOK_03_HANDOVER_PERSISTED:

    NOTEBOOK_04_READINESS_BLOCKERS.append(
        "handover_manifest_not_persisted"
    )


if not handover_manifest_verified:

    NOTEBOOK_04_READINESS_BLOCKERS.append(
        "handover_manifest_not_verified"
    )


if not methodological_boundary_verified:

    NOTEBOOK_04_READINESS_BLOCKERS.append(
        "methodological_boundary_not_verified"
    )


if not NOTEBOOK_04_HANDOVER_READY:

    NOTEBOOK_04_READINESS_BLOCKERS.append(
        "notebook_04_handover_not_ready"
    )


NOTEBOOK_04_DATA_READY = (
    len(
        NOTEBOOK_04_READINESS_BLOCKERS
    )
    == 0
)


# -----------------------------------------------------------------------------
# Final Notebook 03 overview
# -----------------------------------------------------------------------------

print("=" * 72)

print(
    "Media AI — Notebook 03 Final Audit"
)

print("=" * 72)


print(
    f"Final audit version          : "
    f"{FINAL_AUDIT_VERSION}"
)


print("-" * 72)

print(
    "Prepared supervision input"
)

print(
    f"  Articles                   : "
    f"{PREPARED_ARTICLE_COUNT}"
)

print(
    f"  Paragraphs                 : "
    f"{PREPARED_PARAGRAPH_COUNT}"
)

print(
    f"  Sentences                  : "
    f"{PREPARED_SENTENCE_COUNT}"
)

print(
    f"  Prepared corpus verified   : "
    f"{prepared_input_verified}"
)


print("-" * 72)

print(
    "Supervision schema"
)

print(
    f"  Schema type                : "
    f"{ANNOTATION_SCHEMA.get('schema_type')}"
)

print(
    f"  Schema version             : "
    f"{ANNOTATION_SCHEMA.get('schema_version')}"
)

print(
    f"  Annotation policy          : "
    f"{ANNOTATION_SCHEMA.get('annotation_policy_version')}"
)

print(
    f"  Primary annotation unit    : "
    f"{ANNOTATION_SCHEMA.get('primary_annotation_unit')}"
)

print(
    f"  Emotional dimensions       : "
    f"{len(EMOTIONAL_DIMENSIONS)}"
)

print(
    f"  Factological components    : "
    f"{len(FACTOLOGICAL_COMPONENTS)}"
)

print(
    f"  Supervision constructs     : "
    f"{len(SUPERVISION_CONSTRUCTS)}"
)

print(
    f"  Conceptual schema verified : "
    f"{annotation_schema_verified}"
)


print("-" * 72)

print(
    "Numerical validation contract"
)

print(
    f"  Reported-emotion range     : "
    f"{NON_NEGATIVE_STATE_SCALE['minimum']} "
    f"to {NON_NEGATIVE_STATE_SCALE['maximum']}"
)

print(
    f"  Transformative range       : "
    f"{SIGNED_TRANSFORMATION_SCALE['minimum']} "
    f"to {SIGNED_TRANSFORMATION_SCALE['maximum']}"
)

print(
    f"  Transformative neutral     : "
    f"{SIGNED_TRANSFORMATION_SCALE.get('neutral')}"
)

print(
    f"  Weight range               : "
    f"{NON_NEGATIVE_WEIGHT_SCALE['minimum']} "
    f"to {NON_NEGATIVE_WEIGHT_SCALE['maximum']}"
)

print(
    f"  Confidence range           : "
    f"{CONFIDENCE_SCALE['minimum']} "
    f"to {CONFIDENCE_SCALE['maximum']}"
)

print(
    f"  Scale contract verified    : "
    f"{numerical_scale_contract_verified}"
)


print("-" * 72)

print(
    "Annotation infrastructure"
)

print(
    f"  Record schema version      : "
    f"{ANNOTATION_RECORD_SCHEMA_VERSION}"
)

print(
    f"  Supervision contexts       : "
    f"{supervision_context_count}"
)

print(
    f"  Blank annotation records   : "
    f"{blank_annotation_count}"
)

print(
    f"  Context IDs present        : "
    f"{context_ids_present}"
)

print(
    f"  Context IDs unique         : "
    f"{context_ids_unique}"
)

print(
    f"  Template dimensions valid  : "
    f"{template_dimensions_valid}"
)

print(
    f"  Factological schema valid  : "
    f"{factological_template_valid}"
)

print(
    f"  Blank records valid        : "
    f"{blank_records_valid}"
)

print(
    f"  Record schema valid        : "
    f"{ANNOTATION_RECORD_SCHEMA_VALID}"
)

print(
    f"  Template verified          : "
    f"{annotation_template_verified}"
)


print("-" * 72)

print(
    "LLM supervision protocol"
)

print(
    f"  Protocol version           : "
    f"{LLM_SUPERVISION_PROTOCOL_VERSION}"
)

print(
    f"  External request unit      : "
    f"{LLM_SUPERVISION_REQUEST_UNIT}"
)

print(
    f"  Annotation unit            : "
    f"{LLM_SUPERVISION_ANNOTATION_UNIT}"
)

print(
    f"  Sentence payloads          : "
    f"{sentence_payload_count}"
)

print(
    f"  Invalid sentence payloads  : "
    f"{invalid_sentence_payload_count}"
)

print(
    f"  Article payloads           : "
    f"{article_payload_count}"
)

print(
    f"  Sentences in article units : "
    f"{article_payload_sentence_count}"
)

print(
    f"  Article IDs present        : "
    f"{article_payload_ids_present}"
)

print(
    f"  Article IDs unique         : "
    f"{article_payload_ids_unique}"
)

print(
    f"  Protocol verified          : "
    f"{llm_protocol_verified}"
)


print("-" * 72)

print(
    "Supervisor execution"
)

print(
    f"  Master switch              : "
    f"{RUN_LLM_SUPERVISION}"
)

print(
    f"  Supervisor slots           : "
    f"{len(SUPERVISOR_REGISTRY)}"
)

print(
    f"  Enabled supervisors        : "
    f"{enabled_supervisor_ids}"
)

print(
    f"  Configured supervisors     : "
    f"{configured_supervisor_ids}"
)

print(
    f"  Executable supervisors     : "
    f"{executable_supervisor_ids}"
)

print(
    f"  Raw article responses      : "
    f"{final_raw_article_response_count}"
)

print(
    f"  Successful article calls   : "
    f"{successful_raw_article_count}"
)

print(
    f"  Error article calls        : "
    f"{error_raw_article_count}"
)

print(
    f"  Blocked article calls      : "
    f"{blocked_raw_article_count}"
)

print(
    f"  Raw accounting valid       : "
    f"{raw_article_accounting_verified}"
)


print("-" * 72)

print(
    "Canonical supervision"
)

print(
    f"  Validated articles         : "
    f"{validated_article_count}"
)

print(
    f"  Rejected articles          : "
    f"{rejected_article_count}"
)

print(
    f"  Accepted annotations       : "
    f"{final_accepted_count}"
)

print(
    f"  Rejected annotations       : "
    f"{final_rejected_count}"
)

print(
    f"  Sentence validations       : "
    f"{sentence_validation_result_count}"
)

print(
    f"  Sentence IDs present       : "
    f"{accepted_sentence_ids_present}"
)

print(
    f"  Sentence IDs unique        : "
    f"{accepted_sentence_ids_unique}"
)

print(
    f"  Supervisor accounting      : "
    f"{supervisor_annotation_accounting_verified}"
)

print(
    f"  Canonical validation valid : "
    f"{canonical_validation_verified}"
)


print("-" * 72)

print(
    "Supervisor comparison"
)

print(
    f"  Multi-supervisor sentences : "
    f"{multi_supervisor_sentence_count}"
)

print(
    f"  Pairwise comparisons       : "
    f"{pairwise_comparison_count}"
)

print(
    f"  Consensus constructed      : "
    f"{SUPERVISOR_AGREEMENT_SUMMARY['consensus_constructed']}"
)

print(
    f"  Adjudication performed     : "
    f"{SUPERVISOR_AGREEMENT_SUMMARY['adjudication_performed']}"
)

print(
    f"  Agreement layer verified   : "
    f"{agreement_layer_verified}"
)


print("-" * 72)

print(
    "Persistent machine supervision"
)

print(
    f"  Machine supervision avail. : "
    f"{MACHINE_SUPERVISION_AVAILABLE}"
)

print(
    f"  Annotation outputs saved   : "
    f"{ANNOTATION_OUTPUTS_PERSISTED}"
)

print(
    f"  Persisted annotations      : "
    f"{PERSISTED_ANNOTATION_RECORD_COUNT}"
)

print(
    f"  Persisted rejected records : "
    f"{PERSISTED_REJECTED_RECORD_COUNT}"
)

print(
    f"  Required artefacts present : "
    f"{persisted_artefact_set_verified}"
)

print(
    f"  Persistence verified       : "
    f"{machine_supervision_persistence_verified}"
)


print("-" * 72)

print(
    "Reference dataset"
)

print(
    f"  Construction enabled       : "
    f"{BUILD_REFERENCE_DATASET}"
)

print(
    f"  Dataset constructed        : "
    f"{REFERENCE_DATASET_CONSTRUCTED}"
)

print(
    f"  Reference records          : "
    f"{reference_record_count}"
)

print(
    f"  Machine-supervision refs   : "
    f"{machine_supervision_reference_count}"
)

print(
    f"  Human reviewed             : "
    f"{human_reviewed_count}"
)

print(
    f"  Adjudicated reference      : "
    f"{adjudicated_reference_count}"
)

print(
    f"  Not reference ready        : "
    f"{not_reference_ready_count}"
)

print(
    f"  Dataset persisted          : "
    f"{REFERENCE_DATASET_PERSISTED}"
)

print(
    f"  Dataset available          : "
    f"{REFERENCE_DATASET_AVAILABLE}"
)

print(
    f"  Reference layer verified   : "
    f"{reference_layer_verified}"
)


print("-" * 72)

print(
    "Notebook 03 → Notebook 04 handover"
)

print(
    f"  Handover version           : "
    f"{NOTEBOOK_03_HANDOVER.get('handover_version')}"
)

print(
    f"  Manifest persisted         : "
    f"{NOTEBOOK_03_HANDOVER_PERSISTED}"
)

print(
    f"  Manifest verified          : "
    f"{handover_manifest_verified}"
)

print(
    f"  Manifest file ID           : "
    f"{NOTEBOOK_03_HANDOVER_FILE_ID}"
)

print(
    f"  Methodological boundary    : "
    f"{methodological_boundary_verified}"
)

print(
    f"  Notebook 04 handover ready : "
    f"{NOTEBOOK_04_HANDOVER_READY}"
)


print("-" * 72)

print(
    "Audit checks"
)


for (
    check_name,
    passed,
) in audit_checks.items():

    print(
        f"  {check_name:<42}: "
        f"{passed}"
    )


print("-" * 72)

print(
    "Completion state"
)

print(
    f"  Notebook 03 pipeline complete: "
    f"{NOTEBOOK_03_PIPELINE_COMPLETE}"
)

print(
    f"  Notebook 04 data ready       : "
    f"{NOTEBOOK_04_DATA_READY}"
)


if NOTEBOOK_04_READINESS_BLOCKERS:

    print(
        "  Notebook 04 readiness blockers:"
    )


    for blocker in (
        NOTEBOOK_04_READINESS_BLOCKERS
    ):

        print(
            f"    - {blocker}"
        )


print("=" * 72)


# -----------------------------------------------------------------------------
# Final audit result
# -----------------------------------------------------------------------------
#
# The complete diagnostic state is printed before any exception is raised.
# This preserves direct visibility of every failed contract.
# -----------------------------------------------------------------------------

if not NOTEBOOK_03_PIPELINE_COMPLETE:

    audit_failures = [
        name

        for (
            name,
            passed,
        )
        in audit_checks.items()

        if not passed
    ]


    print(
        "Notebook 03 pipeline audit did not pass."
    )

    print(
        "Failed checks:"
    )


    for failure in audit_failures:

        print(
            f"  - {failure}"
        )


    print("=" * 72)


    raise ValueError(
        "Notebook 03 pipeline audit failed. "
        "Failed checks: "
        f"{audit_failures}"
    )


print(
    "Notebook 03 supervision and annotation pipeline audit passed."
)


if NOTEBOOK_04_DATA_READY:

    print(
        "Persisted canonical machine-supervision artefacts are "
        "verified and ready for Notebook 04."
    )


else:

    print(
        "Notebook 04 data handover is not ready."
    )


print(
    "Machine supervision and reference-dataset status are "
    "reported separately."
)

print(
    "A constructed reference dataset is not required for the "
    "Notebook 03 to Notebook 04 handover."
)

print(
    "Independent machine supervision has not been treated as "
    "human-reviewed or adjudicated reference."
)

print(
    "No consensus, human review, or adjudication has been inferred."
)

print(
    "The persisted handover manifest confirms that no model training, "
    "recurrent-state calculation, or confluent representation has occurred."
)

print(
    "No external model or API calls have been made by this audit block."
)

print("=" * 72)
```

## Supervision State and Native Handover Readiness

This block performs a runtime integrity check of the validated supervision state before the final Notebook 03 audit. Its purpose is to confirm that canonical machine-supervision records remain structurally aligned with the prepared sentence contexts and that the native Notebook 03 → Notebook 04 handover is available under the active persistence controls.

The check operates exclusively on the supervision objects produced within Notebook 03. It does not reconstruct sentence records through a generic exporter, generate synthetic observations, or infer reference status from machine-generated annotations.

The block verifies:

- alignment between accepted canonical annotations and the original supervision contexts;
- presence and uniqueness of sentence identifiers;
- consistency of annotation counts across supervisor-specific canonical records;
- article-level validation outcomes;
- the distinction between machine supervision and a human-reviewed or adjudicated reference dataset;
- the state of annotation-output persistence; and
- readiness of the native Notebook 04 handover.

Machine supervision is treated as a validated computational research artefact rather than as ground-truth reference data. Consequently, successful persistence is sufficient for downstream use in Notebook 04, while construction of a reference dataset remains an independent and optional process.

This block is diagnostic only. It does not modify prepared text, create supervision labels, perform consensus or adjudication, train a model, calculate recurrent states, construct confluent representations, or make external model or API calls.

```python
# =============================================================================
# Media AI — Notebook 03 Supervision State and Native Handover Readiness
# =============================================================================

from collections import Counter


# -----------------------------------------------------------------------------
# Block configuration
# -----------------------------------------------------------------------------

SUPERVISION_HANDOVER_CHECK_VERSION = "1.1"


# -----------------------------------------------------------------------------
# Required Notebook 03 objects
# -----------------------------------------------------------------------------

required_objects = [
    "SUPERVISION_CONTEXT_RECORDS",
    "ACCEPTED_SUPERVISION_RECORDS",
    "CANONICAL_SUPERVISION_RECORDS_BY_SUPERVISOR",
    "ARTICLE_RESPONSE_VALIDATION_RESULTS",
]


missing_objects = [
    object_name
    for object_name in required_objects
    if object_name not in globals()
]


if missing_objects:
    raise NameError(
        "Notebook 03 supervision state is incomplete. "
        "Rerun the required Notebook 03 blocks. Missing: "
        f"{missing_objects}"
    )


# -----------------------------------------------------------------------------
# Validate runtime object types
# -----------------------------------------------------------------------------

if not isinstance(
    SUPERVISION_CONTEXT_RECORDS,
    list,
):
    raise TypeError(
        "SUPERVISION_CONTEXT_RECORDS must be a list."
    )


if not isinstance(
    ACCEPTED_SUPERVISION_RECORDS,
    list,
):
    raise TypeError(
        "ACCEPTED_SUPERVISION_RECORDS must be a list."
    )


if not isinstance(
    CANONICAL_SUPERVISION_RECORDS_BY_SUPERVISOR,
    dict,
):
    raise TypeError(
        "CANONICAL_SUPERVISION_RECORDS_BY_SUPERVISOR must be a dictionary."
    )


if not isinstance(
    ARTICLE_RESPONSE_VALIDATION_RESULTS,
    list,
):
    raise TypeError(
        "ARTICLE_RESPONSE_VALIDATION_RESULTS must be a list."
    )


# -----------------------------------------------------------------------------
# Supervision-context state
# -----------------------------------------------------------------------------

context_record_count = len(
    SUPERVISION_CONTEXT_RECORDS
)


context_sentence_ids = [
    record.get(
        "sentence_id"
    )
    for record
    in SUPERVISION_CONTEXT_RECORDS
    if isinstance(
        record,
        dict,
    )
]


context_sentence_ids_valid = (
    len(
        context_sentence_ids
    )
    == context_record_count
    and all(
        context_sentence_ids
    )
    and len(
        context_sentence_ids
    )
    == len(
        set(
            context_sentence_ids
        )
    )
)


# -----------------------------------------------------------------------------
# Accepted canonical supervision state
# -----------------------------------------------------------------------------

accepted_annotation_count = len(
    ACCEPTED_SUPERVISION_RECORDS
)


accepted_sentence_ids = []


for record in ACCEPTED_SUPERVISION_RECORDS:

    if not isinstance(
        record,
        dict,
    ):
        continue

    sentence_id = (
        record.get(
            "provenance",
            {}
        ).get(
            "sentence_id"
        )
    )


    if sentence_id is None:

        sentence_id = record.get(
            "sentence_id"
        )


    accepted_sentence_ids.append(
        sentence_id
    )


accepted_sentence_ids_present = bool(
    accepted_sentence_ids
) and all(
    accepted_sentence_ids
)


accepted_sentence_ids_unique = (
    accepted_sentence_ids_present
    and len(
        accepted_sentence_ids
    )
    == len(
        set(
            accepted_sentence_ids
        )
    )
)


accepted_context_alignment = (
    accepted_sentence_ids_present
    and set(
        accepted_sentence_ids
    ).issubset(
        set(
            context_sentence_ids
        )
    )
)


# -----------------------------------------------------------------------------
# Supervisor accounting
# -----------------------------------------------------------------------------

accepted_by_supervisor = {
    supervisor_id:
        len(
            records
        )

    for supervisor_id, records
    in CANONICAL_SUPERVISION_RECORDS_BY_SUPERVISOR.items()
}


supervisor_annotation_total = sum(
    accepted_by_supervisor.values()
)


supervisor_accounting_valid = (
    supervisor_annotation_total
    == accepted_annotation_count
)


# -----------------------------------------------------------------------------
# Article validation state
# -----------------------------------------------------------------------------

validated_article_results = [
    result
    for result
    in ARTICLE_RESPONSE_VALIDATION_RESULTS
    if isinstance(
        result,
        dict,
    )
    and result.get(
        "valid"
    )
]


validated_article_count = len(
    validated_article_results
)


rejected_article_count = sum(
    1
    for result
    in ARTICLE_RESPONSE_VALIDATION_RESULTS
    if isinstance(
        result,
        dict,
    )
    and not result.get(
        "valid"
    )
)


validated_media_ai_record_ids = [
    result.get(
        "media_ai_record_id"
    )
    for result
    in validated_article_results
    if result.get(
        "media_ai_record_id"
    )
]


# -----------------------------------------------------------------------------
# Reference-dataset state
# -----------------------------------------------------------------------------

build_reference_dataset = bool(
    globals().get(
        "BUILD_REFERENCE_DATASET",
        False,
    )
)


save_annotation_outputs = bool(
    globals().get(
        "SAVE_ANNOTATION_OUTPUTS",
        False,
    )
)


reference_dataset_built = bool(
    globals().get(
        "REFERENCE_DATASET_BUILT",
        False,
    )
)


reference_dataset_persisted = bool(
    globals().get(
        "REFERENCE_DATASET_PERSISTED",
        False,
    )
)


# -----------------------------------------------------------------------------
# Native handover state
# -----------------------------------------------------------------------------

machine_supervision_available = (
    accepted_annotation_count > 0
)


canonical_supervision_valid = bool(
    globals().get(
        "CANONICAL_SUPERVISION_VALID",
        False,
    )
)


annotation_persistence_requested = (
    save_annotation_outputs
)


annotation_persistence_complete = bool(
    globals().get(
        "ANNOTATION_OUTPUTS_PERSISTED",
        False,
    )
)


notebook_04_handover_ready = bool(
    globals().get(
        "NOTEBOOK_04_HANDOVER_READY",
        False,
    )
)


# -----------------------------------------------------------------------------
# Structural validity
# -----------------------------------------------------------------------------

runtime_supervision_valid = all([
    context_record_count > 0,
    context_sentence_ids_valid,
    accepted_annotation_count > 0,
    accepted_sentence_ids_present,
    accepted_sentence_ids_unique,
    accepted_context_alignment,
    supervisor_accounting_valid,
    validated_article_count > 0,
    canonical_supervision_valid,
])


if not runtime_supervision_valid:
    raise ValueError(
        "Notebook 03 runtime supervision state failed structural validation."
    )


# -----------------------------------------------------------------------------
# Verification output
# -----------------------------------------------------------------------------

print("=" * 72)

print(
    "Media AI — Notebook 03 Supervision State and Native Handover Readiness"
)

print("=" * 72)


print(
    f"Check version              : "
    f"{SUPERVISION_HANDOVER_CHECK_VERSION}"
)


print("-" * 72)

print(
    "Supervision context"
)

print(
    f"Context sentence records   : "
    f"{context_record_count}"
)

print(
    f"Sentence IDs valid         : "
    f"{context_sentence_ids_valid}"
)


print("-" * 72)

print(
    "Canonical supervision"
)

print(
    f"Accepted annotations       : "
    f"{accepted_annotation_count}"
)

print(
    f"Sentence IDs present       : "
    f"{accepted_sentence_ids_present}"
)

print(
    f"Sentence IDs unique        : "
    f"{accepted_sentence_ids_unique}"
)

print(
    f"Context alignment          : "
    f"{accepted_context_alignment}"
)

print(
    f"Canonical supervision valid: "
    f"{canonical_supervision_valid}"
)


print("-" * 72)

print(
    "Supervisor accounting"
)


for supervisor_id in sorted(
    accepted_by_supervisor
):

    print(
        f"{supervisor_id:<27}: "
        f"{accepted_by_supervisor[supervisor_id]}"
    )


print(
    f"Supervisor accounting valid: "
    f"{supervisor_accounting_valid}"
)


print("-" * 72)

print(
    "Article validation"
)

print(
    f"Validated articles         : "
    f"{validated_article_count}"
)

print(
    f"Rejected articles          : "
    f"{rejected_article_count}"
)


print("-" * 72)

print(
    "Reference and persistence controls"
)

print(
    f"Build reference dataset    : "
    f"{build_reference_dataset}"
)

print(
    f"Save annotation outputs    : "
    f"{save_annotation_outputs}"
)

print(
    f"Reference dataset built    : "
    f"{reference_dataset_built}"
)

print(
    f"Reference dataset persisted: "
    f"{reference_dataset_persisted}"
)


print("-" * 72)

print(
    "Native Notebook 04 handover"
)

print(
    f"Machine supervision avail. : "
    f"{machine_supervision_available}"
)

print(
    f"Persistence requested      : "
    f"{annotation_persistence_requested}"
)

print(
    f"Persistence complete       : "
    f"{annotation_persistence_complete}"
)

print(
    f"Notebook 04 handover ready : "
    f"{notebook_04_handover_ready}"
)


print("-" * 72)

print(
    "Execution state"
)

print(
    f"Runtime supervision valid  : "
    f"{runtime_supervision_valid}"
)

print(
    f"Generic data exporter used : "
    f"{False}"
)

print(
    f"Synthetic records created  : "
    f"{False}"
)


print("=" * 72)


print(
    "Notebook 03 runtime supervision state validated successfully."
)

print(
    "The block uses the native Notebook 03 supervision objects and does "
    "not search for a generic SOURCE_SENTENCE_RECORDS object."
)

print(
    "Accepted canonical machine-supervision records remain distinct from "
    "an adjudicated or human-reviewed reference dataset."
)

print(
    "No reference status has been inferred from machine supervision alone."
)

print(
    "No synthetic sentence records or alternative handover format "
    "have been created."
)


if not save_annotation_outputs:

    print(
        "Annotation-output persistence remains disabled by the "
        "Execution Control Panel."
    )

    print(
        "The validated supervision records therefore remain available "
        "in the current runtime but are not yet persistently handed over."
    )


elif not annotation_persistence_complete:

    print(
        "Annotation-output persistence has been requested but has not "
        "yet been confirmed as complete."
    )


if not notebook_04_handover_ready:

    print(
        "Notebook 04 handover is not yet ready."
    )


print(
    "No external model or API calls have been made by this block."
)

print("=" * 72)
```
