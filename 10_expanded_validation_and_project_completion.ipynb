# Media AI — Expanded Validation and Project Completion

## Purpose

This notebook completes the first validation phase of Media AI.

It restores the complete and validated architecture established through Notebooks 01–09 and evaluates whether its principal structural, transformative and recurrent properties remain valid beyond the controlled single-article pilot used during architectural construction.

The notebook therefore extends validation from the original canonical article to a broader multi-article setting while preserving the architectural contracts already established.

The completed architecture operates across four textual levels:

**Token → Sentence → Paragraph → Article**

and three principal representational spaces:

**Factual → Psychological → Social**

Within these spaces, the validated architecture combines:

- hierarchical textual representation;
- factual, psychological and social representation pathways;
- trained transformative mechanisms;
- learned dimension-wise Transformative Weights;
- pathway-specific recurrent state;
- additive recurrent updating;
- deterministic article-boundary state reset;
- causal sequential execution without future-context access.

The purpose of this notebook is not to redesign these mechanisms.

Instead, it determines whether the completed architecture remains coherent, deterministic and structurally valid when executed across a broader controlled corpus.

## Methodological Role

Notebook 10 forms the completion boundary of the first Media AI validation project.

The preceding notebooks established:

- the research corpus and source structure;
- prepared and traceable textual units;
- the annotation ontology;
- independent LLM supervision;

- persistent architectural checkpointing and reproducible restoration.

Notebook 09 concluded this architectural sequence by producing the canonical persistent handover:

- `notebook_09_final_checkpoint.pt`;
- `notebook_09_final_metadata.json`.

These artefacts contain the complete validated Media AI architecture required by this notebook.

Notebook 10 does not reconstruct the architecture from earlier notebooks and does not redefine its established contracts.

Instead, it restores the canonical Notebook 09 handover and subjects the resulting architecture to expanded validation.

## Validation Scope

The notebook extends validation from the controlled 16-sentence architectural pilot to a broader multi-article corpus.

The principal validation questions are:

1. Can the complete Notebook 09 architecture be restored exactly from its persistent handover?
2. Can additional articles be represented under the established input and pathway contracts?
3. Can recurrent execution proceed independently through each article while preserving canonical sentence order?
4. Are recurrent states reset deterministically at every article boundary?
5. Do factual, psychological and social pathways preserve their established dimensional and causal contracts?
6. Do learned Transformative Weights remain finite, bounded and correctly aligned to active and deferred dimensions?
7. Does expanded execution remain deterministic and numerically stable?
8. Does the architecture preserve strict separation between article-specific runtime state and learned model state?
9. Are the results sufficient to close the first architectural validation phase and support transition to expanded-corpus training?

The objective is therefore **expanded architectural validation**, not final model training.

## Inherited Architecture

Notebook 10 inherits the complete Notebook 09 architecture without structural modification.

The persisted model contains:

- **894,003 representation parameters**;
- **8,187 transformative parameters**;
- **25 learned Transformative-Weight parameters**;
- **902,215 total persisted model parameters**.

The pathway dimensions remain:

- factual: **10**;
- psychological: **34**;
- social: **7**.

The complete Transformative-Weight space contains **51 structural dimensions**:

- **25 active dimensions** governed by learned Transformative-Weight parameters;
- **26 deferred dimensions** retained structurally with effective weight fixed at zero under the current validation contract.

The inherited recurrent update remains additive:

$$
\mathbf{h}_{t}^{(p)}
=
\mathbf{h}_{t-1}^{(p)}
+
\mathbf{w}^{(p)}
\odot
\Delta\mathbf{r}_{t}^{(p)},
$$

where:

- $p \in \{\mathrm{factual},\mathrm{psychological},\mathrm{social}\}$ denotes the representation pathway;
- $\mathbf{h}_{t}^{(p)}$ is the pathway-specific recurrent state at sentence position $t$;
- $\mathbf{w}^{(p)}$ is the learned effective Transformative-Weight vector;
- $\Delta\mathbf{r}_{t}^{(p)}$ is the candidate transformation produced by the inherited transformative mechanism;
- $\odot$ denotes element-wise multiplication.

Every article begins from the deterministic initial condition:

$$
\mathbf{h}_{0}^{(p)}=\mathbf{0}.
$$

Recurrent state may propagate between sentences within an article but must never propagate across article boundaries.

## Expanded Multi-Article Validation

Expanded validation must preserve article identity and canonical sentence identity throughout execution.

For an article containing the ordered sentence sequence

$$
\mathcal{A}_i =
\left(
s_{i,1},
s_{i,2},
\ldots,
s_{i,T_i}
\right),
$$

recurrent execution proceeds strictly in canonical order:

$$
s_{i,1}
\rightarrow
s_{i,2}
\rightarrow
\cdots
\rightarrow
s_{i,T_i}.
$$

At the transition to a new article $\mathcal{A}_{i+1}$, all pathway-specific recurrent states are discarded and reconstructed from the deterministic zero initial condition.

Accordingly,

$$
\mathbf{h}_{i,T_i}^{(p)}
\not\rightarrow
\mathbf{h}_{i+1,0}^{(p)}.
$$

This separation is fundamental to the current architecture.

Article-specific recurrent trajectories are runtime states and are not persistent trainable model parameters.

## Validation Principles

The notebook follows several strict methodological constraints.

### Canonical restoration

The Notebook 09 checkpoint and metadata are the authoritative architectural source.

Earlier notebooks must not be rerun merely to reconstruct model state already contained in the persistent handover.

### No architectural redesign

Representation, transformative, Transformative-Weight and recurrent contracts remain unchanged during expanded validation.

Any future architectural modification belongs to a subsequent experimental phase.

### Article independence

Every article begins from independently constructed zero recurrent states.

Cross-article recurrent propagation is prohibited.

### Causal execution

Only the current sentence representation and preceding recurrent state may contribute to the current recurrent transition.

Future sentence information must not enter the forward path.

### Pathway independence

Factual, psychological and social recurrent pathways remain structurally independent under the current architecture.

### Deferred-dimension preservation

Deferred dimensions remain structurally present but receive zero Transformative-Weight-gated candidate contribution.

### Deterministic validation

Repeated execution under identical inputs and model state must reproduce identical outputs within the established numerical contract.

### No supervision substitution

Supervision targets may be used for evaluation where appropriate but must never be substituted for model representations or recurrent inputs.

## Project Completion Criterion

Notebook 10 closes the first Media AI validation project only if the expanded validation confirms that the architecture remains:

- restorable;
- structurally coherent;
- dimensionally valid;
- causal;
- article-boundary safe;
- pathway-consistent;
- numerically finite;
- deterministic;
- persistable;
- suitable for subsequent expanded-corpus training.

Successful completion does not imply that Media AI is fully trained or externally validated.

It establishes that the architecture has passed the controlled validation phase required before larger-scale optimisation and evaluation.

# Block 1 — Notebook 09 → Notebook 10 Handover Restoration and Initialisation Audit

This block establishes the controlled initialisation boundary for Notebook 10 by restoring the complete validated Media AI architecture persisted at the conclusion of Notebook 09.

Notebook 10 begins from the final Notebook 09 checkpoint rather than reconstructing the architecture from earlier notebook stages. The checkpoint and its accompanying metadata are retrieved directly from the persistent Google Drive handover using the canonical file identifiers established by Notebook 09. Before model restoration, the block validates artefact identity, byte count, SHA-256 integrity, checkpoint–metadata consistency and the Notebook 09 → Notebook 10 bootstrap contract.

### Inherited Model Architecture

The restored model contains three parameterised architectural components:

- the **representation architecture**, containing 894,003 parameters;
- the **transformative mechanisms**, containing 8,187 parameters;
- the learned **Transformative-Weight architecture**, containing 49 trainable parameters at the conclusion of Notebook 09.

The complete inherited parameter count is therefore

$$
N_{\mathrm{total}}
=
N_{\mathrm{representation}}
+
N_{\mathrm{transformative}}
+
N_{\mathrm{TW}}
=
894{,}003
+
8{,}187
+
49
=
902{,}239.
$$

The Transformative-Weight architecture spans 51 structural dimensions across the factual, psychological and social representational spaces. Of these, 49 dimensions are activation-eligible and contain learned latent parameters, while 2 social dimensions remain deferred under the current architectural contract.

The inherited pathway dimensions are

$$
d_{\mathrm{factual}} = 10,
\qquad
d_{\mathrm{psychological}} = 34,
\qquad
d_{\mathrm{social}} = 7.
$$

All factual and psychological dimensions are active. Within the social pathway, dimensions indexed by 1 and 3 remain deferred, while dimensions 0, 2, 4, 5 and 6 are active.

### Transformative-Weight Restoration

For each active dimension, the persisted latent Transformative-Weight parameter $\alpha$ is restored directly from the Notebook 09 checkpoint. Its effective bounded weight is reconstructed using the inherited mapping

$$
w = \sigma(\alpha),
$$

where $\sigma(\cdot)$ denotes the sigmoid function.

Deferred dimensions remain structurally present but retain an effective Transformative Weight of exactly zero.

The block verifies that the reconstructed effective weights are finite, bounded and consistent with the persisted Notebook 09 effective-weight state. This ensures that Notebook 10 begins from the learned Transformative-Weight configuration produced by the controlled pilot optimisation rather than from a newly initialised gating state.

### Recurrent-State Contract

The recurrent architecture itself contains no persistent article-specific state. Instead, Notebook 10 inherits the causal recurrent-update contract established in Notebook 09:

$$
\mathbf{h}_{t}
=
\mathbf{h}_{t-1}
+
\mathbf{w}
\odot
\mathbf{c}_{t},
$$

where $\mathbf{h}_{t-1}$ is the preceding pathway-specific recurrent state, $\mathbf{w}$ is the effective Transformative-Weight vector, $\mathbf{c}_{t}$ is the candidate transformation produced by the corresponding transformative mechanism, and $\odot$ denotes element-wise multiplication.

Every article begins from a deterministic zero state,

$$
\mathbf{h}_{0}=\mathbf{0},
$$

and recurrent state is prohibited from propagating across article boundaries. Cross-pathway recurrence and future-context access likewise remain prohibited.

Article-specific recurrent tensors are therefore runtime states rather than learned model parameters and are not restored from the persistent checkpoint.

### Restoration and Freeze Validation

After reconstruction, the block verifies exact restoration of the representation, transformative and Transformative-Weight parameter states. It additionally validates device placement, parameter counts, pathway contracts, effective-weight reconstruction and checkpoint–metadata consistency.

All inherited parameters are then frozen and all inherited modules are placed in evaluation mode. This establishes an immutable Notebook 09 baseline before Notebook 10 introduces any expanded validation or training procedure.

Importantly, the optimisation performed in Notebook 09 remains classified as **controlled pilot architectural training**. It demonstrated that the Transformative-Weight mechanism is learnable under the validated recurrent objective, but it is not interpreted as final model training or final empirical performance.

### Execution Boundary

This block performs restoration and validation only.

It does **not**:

- load the expanded corpus;
- execute a representation, transformative or Transformative-Weight forward path;
- instantiate an article recurrent state;
- execute a recurrent update;
- calculate a training loss;
- create an optimiser;
- execute backpropagation;
- update any model parameter;
- perform expanded training.

The purpose of Block 1 is therefore to establish a verified and reproducible architectural starting point for Notebook 10.

Once this handover is validated, Notebook 10 may proceed to expanded-corpus inventory, eligibility assessment and construction of the multi-article validation source required for the final experimental stage.

```python
# =============================================================================
# Media AI — Notebook 10
# Block 1: Notebook 09 → 10 Handover Restoration
#          and Initialisation Audit
# =============================================================================

from copy import deepcopy
from datetime import datetime, timezone

import hashlib
import io
import json

import numpy as np
import torch
import torch.nn as nn

from google.colab import auth
import google.auth

from googleapiclient.discovery import build
from googleapiclient.http import MediaIoBaseDownload


# =============================================================================
# Block identity
# =============================================================================

NOTEBOOK_10 = (
    "10_expanded_validation_and_project_completion"
)

NOTEBOOK_10_VERSION = "1.0"

NOTEBOOK_10_BLOCK_1 = 1

NOTEBOOK_10_BLOCK_1_NAME = (
    "Notebook 09 → 10 Handover Restoration "
    "and Initialisation Audit"
)

NOTEBOOK_10_BLOCK_1_VERSION = "1.1"

BLOCK_1_EXECUTED_AT_UTC = (
    datetime.now(
        timezone.utc
    ).isoformat()
)


# =============================================================================
# Project reproducibility
# =============================================================================

NOTEBOOK_10_RANDOM_SEED = 42


np.random.seed(
    NOTEBOOK_10_RANDOM_SEED
)


torch.manual_seed(
    NOTEBOOK_10_RANDOM_SEED
)


if torch.cuda.is_available():

    torch.cuda.manual_seed_all(
        NOTEBOOK_10_RANDOM_SEED
    )


# =============================================================================
# Controlled execution device
# =============================================================================
#
# Notebook 09 completed on CPU and its final persisted states are CPU-safe.
# Block 1 therefore restores on CPU first and preserves CPU execution for the
# initial audit. Later blocks may explicitly change the execution device.
# =============================================================================

DEVICE = torch.device(
    "cpu"
)

DEFAULT_DTYPE = torch.float32


# =============================================================================
# Canonical Notebook 09 → 10 persistent identifiers
# =============================================================================
#
# These IDs were created and validated by Notebook 09 Block 13.
#
# No filename discovery is used.
# =============================================================================

NOTEBOOK_09_TO_10_CHECKPOINT_FILENAME = (
    "notebook_09_final_checkpoint.pt"
)

NOTEBOOK_09_TO_10_METADATA_FILENAME = (
    "notebook_09_final_metadata.json"
)


NOTEBOOK_09_TO_10_CHECKPOINT_DRIVE_FILE_ID = (
    "1h2uRHtNfAFPlKNMZqwiZvGROewn0b2-R"
)

NOTEBOOK_09_TO_10_METADATA_DRIVE_FILE_ID = (
    "1V0eeY-QdeD9htWWYbESU42qbak5hV8Ms"
)


# =============================================================================
# Expected inherited architecture
# =============================================================================
#
# These values reproduce the validated Notebook 09 Block 13 completion
# contract. Checkpoint and metadata must additionally agree exactly.
# =============================================================================

BLOCK_1_EXPECTED_REPRESENTATION_PARAMETERS = 894_003

BLOCK_1_EXPECTED_TRANSFORMATIVE_PARAMETERS = 8_187

BLOCK_1_EXPECTED_TRANSFORMATIVE_WEIGHT_PARAMETERS = 49

BLOCK_1_EXPECTED_TOTAL_PARAMETERS = 902_239


BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS = {
    "factual":
        10,

    "psychological":
        34,

    "social":
        7,
}


BLOCK_1_EXPECTED_ACTIVE_DIMENSIONS = {
    "factual":
        tuple(
            range(
                10
            )
        ),

    "psychological":
        tuple(
            range(
                34
            )
        ),

    "social":
        (
            0,
            2,
            4,
            5,
            6,
        ),
}


BLOCK_1_EXPECTED_DEFERRED_DIMENSIONS = {
    "factual":
        tuple(),

    "psychological":
        tuple(),

    "social":
        (
            1,
            3,
        ),
}


BLOCK_1_EXPECTED_STRUCTURAL_WEIGHT_DIMENSIONS = int(
    sum(
        BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS.values()
    )
)


BLOCK_1_EXPECTED_ACTIVE_WEIGHT_DIMENSIONS = int(
    sum(
        len(
            BLOCK_1_EXPECTED_ACTIVE_DIMENSIONS[
                pathway_name
            ]
        )

        for pathway_name
        in BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS
    )
)


BLOCK_1_EXPECTED_DEFERRED_WEIGHT_DIMENSIONS = int(
    sum(
        len(
            BLOCK_1_EXPECTED_DEFERRED_DIMENSIONS[
                pathway_name
            ]
        )

        for pathway_name
        in BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS
    )
)


BLOCK_1_EXPECTED_WEIGHT_PARAMETER_COUNTS = {
    pathway_name:
        len(
            BLOCK_1_EXPECTED_ACTIVE_DIMENSIONS[
                pathway_name
            ]
        )

    for pathway_name
    in BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS
}


if not all(
    [
        (
            BLOCK_1_EXPECTED_STRUCTURAL_WEIGHT_DIMENSIONS
            ==
            51
        ),

        (
            BLOCK_1_EXPECTED_ACTIVE_WEIGHT_DIMENSIONS
            ==
            BLOCK_1_EXPECTED_TRANSFORMATIVE_WEIGHT_PARAMETERS
        ),

        (
            BLOCK_1_EXPECTED_ACTIVE_WEIGHT_DIMENSIONS
            +
            BLOCK_1_EXPECTED_DEFERRED_WEIGHT_DIMENSIONS
            ==
            BLOCK_1_EXPECTED_STRUCTURAL_WEIGHT_DIMENSIONS
        ),

        (
            BLOCK_1_EXPECTED_TOTAL_PARAMETERS
            ==
            BLOCK_1_EXPECTED_REPRESENTATION_PARAMETERS
            +
            BLOCK_1_EXPECTED_TRANSFORMATIVE_PARAMETERS
            +
            BLOCK_1_EXPECTED_TRANSFORMATIVE_WEIGHT_PARAMETERS
        ),
    ]
):

    raise RuntimeError(
        "Notebook 10 Block 1 expected Notebook 09 architecture "
        "contract is internally inconsistent."
    )


# =============================================================================
# Execution boundary
# =============================================================================

NOTEBOOK_10_BLOCK_1_EXPANDED_CORPUS_LOADED = False

NOTEBOOK_10_BLOCK_1_REPRESENTATION_FORWARD_EXECUTED = False

NOTEBOOK_10_BLOCK_1_TRANSFORMATIVE_FORWARD_EXECUTED = False

NOTEBOOK_10_BLOCK_1_WEIGHT_FORWARD_EXECUTED = False

NOTEBOOK_10_BLOCK_1_RECURRENT_STATE_CREATED = False

NOTEBOOK_10_BLOCK_1_RECURRENT_UPDATE_EXECUTED = False

NOTEBOOK_10_BLOCK_1_LOSS_CALCULATED = False

NOTEBOOK_10_BLOCK_1_OPTIMIZER_CREATED = False

NOTEBOOK_10_BLOCK_1_ZERO_GRAD_EXECUTED = False

NOTEBOOK_10_BLOCK_1_BACKWARD_PASS_EXECUTED = False

NOTEBOOK_10_BLOCK_1_PARAMETER_UPDATE_EXECUTED = False

NOTEBOOK_10_BLOCK_1_TRAINING_EXECUTED = False


# =============================================================================
# Google authentication
# =============================================================================
#
# This is the same narrow Colab authentication pattern already proven in the
# preceding Media AI notebooks:
#
#   auth.authenticate_user()
#   google.auth.default()
#   build(... credentials=...)
#
# No Drive filesystem mount is used.
# No metadata-service fallback is intentionally introduced.
# =============================================================================

auth.authenticate_user()


BLOCK_1_CREDENTIALS, _ = (
    google.auth.default()
)


NOTEBOOK_10_DRIVE_SERVICE = build(
    "drive",
    "v3",
    credentials=
        BLOCK_1_CREDENTIALS,
    cache_discovery=
        False,
)


BLOCK_1_COLAB_AUTHENTICATED = True


BLOCK_1_DRIVE_SERVICE_INITIALISED = (
    NOTEBOOK_10_DRIVE_SERVICE
    is not None
)


if not BLOCK_1_DRIVE_SERVICE_INITIALISED:

    raise RuntimeError(
        "Notebook 10 could not initialise the Google Drive API service."
    )


# =============================================================================
# Drive helpers
# =============================================================================

def block_1_get_drive_metadata(
    drive_service,
    file_id,
):

    return (
        drive_service.files()
        .get(
            fileId=
                file_id,

            fields=(
                "id,"
                "name,"
                "size,"
                "modifiedTime,"
                "mimeType,"
                "parents"
            ),
        )
        .execute()
    )


def block_1_download_drive_bytes(
    drive_service,
    file_id,
):

    request = (
        drive_service.files()
        .get_media(
            fileId=
                file_id
        )
    )


    buffer = io.BytesIO()


    downloader = MediaIoBaseDownload(
        buffer,
        request,
    )


    complete = False


    while not complete:

        _, complete = downloader.next_chunk()


    buffer.seek(
        0
    )


    return buffer.read()


def block_1_sha256(
    payload,
):

    return hashlib.sha256(
        payload
    ).hexdigest()


# =============================================================================
# Locate the two canonical artefacts by direct ID
# =============================================================================

BLOCK_1_CHECKPOINT_DRIVE_METADATA = (
    block_1_get_drive_metadata(
        NOTEBOOK_10_DRIVE_SERVICE,
        NOTEBOOK_09_TO_10_CHECKPOINT_DRIVE_FILE_ID,
    )
)


BLOCK_1_METADATA_DRIVE_METADATA = (
    block_1_get_drive_metadata(
        NOTEBOOK_10_DRIVE_SERVICE,
        NOTEBOOK_09_TO_10_METADATA_DRIVE_FILE_ID,
    )
)


BLOCK_1_CHECKPOINT_LOCATED = (
    BLOCK_1_CHECKPOINT_DRIVE_METADATA.get(
        "id"
    )
    ==
    NOTEBOOK_09_TO_10_CHECKPOINT_DRIVE_FILE_ID
)


BLOCK_1_METADATA_LOCATED = (
    BLOCK_1_METADATA_DRIVE_METADATA.get(
        "id"
    )
    ==
    NOTEBOOK_09_TO_10_METADATA_DRIVE_FILE_ID
)


BLOCK_1_CHECKPOINT_FILENAME_VALID = (
    BLOCK_1_CHECKPOINT_DRIVE_METADATA.get(
        "name"
    )
    ==
    NOTEBOOK_09_TO_10_CHECKPOINT_FILENAME
)


BLOCK_1_METADATA_FILENAME_VALID = (
    BLOCK_1_METADATA_DRIVE_METADATA.get(
        "name"
    )
    ==
    NOTEBOOK_09_TO_10_METADATA_FILENAME
)


if not all(
    [
        BLOCK_1_CHECKPOINT_LOCATED,
        BLOCK_1_METADATA_LOCATED,
        BLOCK_1_CHECKPOINT_FILENAME_VALID,
        BLOCK_1_METADATA_FILENAME_VALID,
    ]
):

    raise RuntimeError(
        "Notebook 09 → 10 direct persistent artefact identity "
        "validation failed."
    )


# =============================================================================
# Download metadata first
# =============================================================================
#
# The metadata contains the authoritative checkpoint SHA-256 digest and byte
# count. It is therefore parsed before the checkpoint is deserialised.
# =============================================================================

BLOCK_1_METADATA_BYTES = (
    block_1_download_drive_bytes(
        NOTEBOOK_10_DRIVE_SERVICE,
        NOTEBOOK_09_TO_10_METADATA_DRIVE_FILE_ID,
    )
)


if not BLOCK_1_METADATA_BYTES:

    raise RuntimeError(
        "Notebook 09 final metadata artefact is empty."
    )


BLOCK_1_METADATA_DOWNLOADED = True


BLOCK_1_METADATA_DRIVE_BYTE_COUNT_VALID = (
    int(
        BLOCK_1_METADATA_DRIVE_METADATA.get(
            "size",
            -1,
        )
    )
    ==
    len(
        BLOCK_1_METADATA_BYTES
    )
)


if not BLOCK_1_METADATA_DRIVE_BYTE_COUNT_VALID:

    raise RuntimeError(
        "Notebook 09 metadata download byte count does not match "
        "Google Drive metadata."
    )


# =============================================================================
# Parse metadata
# =============================================================================

NOTEBOOK_10_INHERITED_METADATA = json.loads(
    BLOCK_1_METADATA_BYTES.decode(
        "utf-8"
    )
)


BLOCK_1_METADATA_JSON_PARSED = isinstance(
    NOTEBOOK_10_INHERITED_METADATA,
    dict,
)


if not BLOCK_1_METADATA_JSON_PARSED:

    raise TypeError(
        "Notebook 09 final metadata must parse to a JSON object."
    )


# =============================================================================
# Metadata identity contract
# =============================================================================

BLOCK_1_METADATA_IDENTITY_VALID = all(
    [
        (
            NOTEBOOK_10_INHERITED_METADATA.get(
                "artifact_type"
            )
            ==
            "media_ai_notebook_09_final_metadata"
        ),

        (
            NOTEBOOK_10_INHERITED_METADATA.get(
                "metadata_schema_version"
            )
            ==
            "1.0"
        ),

        (
            NOTEBOOK_10_INHERITED_METADATA.get(
                "notebook_number"
            )
            ==
            9
        ),

        (
            NOTEBOOK_10_INHERITED_METADATA.get(
                "completion_block"
            )
            ==
            13
        ),
    ]
)


if not BLOCK_1_METADATA_IDENTITY_VALID:

    raise RuntimeError(
        "Notebook 09 final metadata identity contract is invalid."
    )


# =============================================================================
# Metadata bootstrap contract
# =============================================================================

BLOCK_1_METADATA_BOOTSTRAP = (
    NOTEBOOK_10_INHERITED_METADATA.get(
        "notebook_10_bootstrap",
        {},
    )
)


BLOCK_1_METADATA_BOOTSTRAP_VALID = all(
    [
        (
            BLOCK_1_METADATA_BOOTSTRAP.get(
                "source_notebook"
            )
            ==
            9
        ),

        (
            BLOCK_1_METADATA_BOOTSTRAP.get(
                "checkpoint_filename"
            )
            ==
            NOTEBOOK_09_TO_10_CHECKPOINT_FILENAME
        ),

        (
            BLOCK_1_METADATA_BOOTSTRAP.get(
                "checkpoint_drive_file_id"
            )
            ==
            NOTEBOOK_09_TO_10_CHECKPOINT_DRIVE_FILE_ID
        ),

        (
            BLOCK_1_METADATA_BOOTSTRAP.get(
                "metadata_filename"
            )
            ==
            NOTEBOOK_09_TO_10_METADATA_FILENAME
        ),

        (
            BLOCK_1_METADATA_BOOTSTRAP.get(
                "metadata_drive_file_id"
            )
            ==
            NOTEBOOK_09_TO_10_METADATA_DRIVE_FILE_ID
        ),

        (
            BLOCK_1_METADATA_BOOTSTRAP.get(
                "architecture_complete"
            )
            is True
        ),

        (
            BLOCK_1_METADATA_BOOTSTRAP.get(
                "expanded_corpus_training_required"
            )
            is True
        ),

        (
            BLOCK_1_METADATA_BOOTSTRAP.get(
                "reconstruct_upstream_notebooks"
            )
            is False
        ),
    ]
)


if not BLOCK_1_METADATA_BOOTSTRAP_VALID:

    raise RuntimeError(
        "Notebook 09 final metadata does not reproduce the "
        "Notebook 10 bootstrap contract."
    )


# =============================================================================
# Canonical checkpoint identity from metadata
# =============================================================================

BLOCK_1_METADATA_CHECKPOINT_CONTRACT = (
    NOTEBOOK_10_INHERITED_METADATA.get(
        "checkpoint",
        {},
    )
)


BLOCK_1_EXPECTED_CHECKPOINT_SHA256 = (
    BLOCK_1_METADATA_CHECKPOINT_CONTRACT.get(
        "sha256"
    )
)


BLOCK_1_EXPECTED_CHECKPOINT_BYTE_COUNT = (
    BLOCK_1_METADATA_CHECKPOINT_CONTRACT.get(
        "byte_count"
    )
)


BLOCK_1_METADATA_CHECKPOINT_IDENTITY_VALID = all(
    [
        (
            BLOCK_1_METADATA_CHECKPOINT_CONTRACT.get(
                "filename"
            )
            ==
            NOTEBOOK_09_TO_10_CHECKPOINT_FILENAME
        ),

        (
            BLOCK_1_METADATA_CHECKPOINT_CONTRACT.get(
                "drive_file_id"
            )
            ==
            NOTEBOOK_09_TO_10_CHECKPOINT_DRIVE_FILE_ID
        ),

        isinstance(
            BLOCK_1_EXPECTED_CHECKPOINT_SHA256,
            str,
        ),

        (
            len(
                BLOCK_1_EXPECTED_CHECKPOINT_SHA256
            )
            ==
            64
        ),

        isinstance(
            BLOCK_1_EXPECTED_CHECKPOINT_BYTE_COUNT,
            int,
        ),

        (
            BLOCK_1_EXPECTED_CHECKPOINT_BYTE_COUNT
            >
            0
        ),
    ]
)


if not BLOCK_1_METADATA_CHECKPOINT_IDENTITY_VALID:

    raise RuntimeError(
        "Notebook 09 metadata does not contain a valid checkpoint "
        "identity and integrity contract."
    )


# =============================================================================
# Download checkpoint
# =============================================================================

BLOCK_1_CHECKPOINT_BYTES = (
    block_1_download_drive_bytes(
        NOTEBOOK_10_DRIVE_SERVICE,
        NOTEBOOK_09_TO_10_CHECKPOINT_DRIVE_FILE_ID,
    )
)


if not BLOCK_1_CHECKPOINT_BYTES:

    raise RuntimeError(
        "Notebook 09 final checkpoint artefact is empty."
    )


BLOCK_1_CHECKPOINT_DOWNLOADED = True


# =============================================================================
# Validate checkpoint before deserialisation
# =============================================================================

BLOCK_1_CHECKPOINT_SHA256 = (
    block_1_sha256(
        BLOCK_1_CHECKPOINT_BYTES
    )
)


BLOCK_1_CHECKPOINT_SHA256_VALID = (
    BLOCK_1_CHECKPOINT_SHA256
    ==
    BLOCK_1_EXPECTED_CHECKPOINT_SHA256
)


BLOCK_1_CHECKPOINT_METADATA_BYTE_COUNT_VALID = (
    len(
        BLOCK_1_CHECKPOINT_BYTES
    )
    ==
    BLOCK_1_EXPECTED_CHECKPOINT_BYTE_COUNT
)


BLOCK_1_CHECKPOINT_DRIVE_BYTE_COUNT_VALID = (
    int(
        BLOCK_1_CHECKPOINT_DRIVE_METADATA.get(
            "size",
            -1,
        )
    )
    ==
    len(
        BLOCK_1_CHECKPOINT_BYTES
    )
)


if not all(
    [
        BLOCK_1_CHECKPOINT_SHA256_VALID,
        BLOCK_1_CHECKPOINT_METADATA_BYTE_COUNT_VALID,
        BLOCK_1_CHECKPOINT_DRIVE_BYTE_COUNT_VALID,
    ]
):

    raise RuntimeError(
        "Notebook 09 final checkpoint failed pre-deserialisation "
        "integrity validation."
    )


# =============================================================================
# Deserialize checkpoint only after hash validation
# =============================================================================

NOTEBOOK_10_INHERITED_CHECKPOINT = torch.load(
    io.BytesIO(
        BLOCK_1_CHECKPOINT_BYTES
    ),
    map_location=
        "cpu",
    weights_only=
        False,
)


BLOCK_1_CHECKPOINT_PARSED = isinstance(
    NOTEBOOK_10_INHERITED_CHECKPOINT,
    dict,
)


if not BLOCK_1_CHECKPOINT_PARSED:

    raise TypeError(
        "Notebook 09 final checkpoint must deserialize to a dictionary."
    )


# =============================================================================
# Checkpoint identity
# =============================================================================

BLOCK_1_CHECKPOINT_IDENTITY_VALID = all(
    [
        (
            NOTEBOOK_10_INHERITED_CHECKPOINT.get(
                "artifact_type"
            )
            ==
            "media_ai_notebook_09_final_checkpoint"
        ),

        (
            NOTEBOOK_10_INHERITED_CHECKPOINT.get(
                "checkpoint_schema_version"
            )
            ==
            "1.0"
        ),

        (
            NOTEBOOK_10_INHERITED_CHECKPOINT.get(
                "notebook_number"
            )
            ==
            9
        ),

        (
            NOTEBOOK_10_INHERITED_CHECKPOINT.get(
                "notebook_block"
            )
            ==
            13
        ),
    ]
)


if not BLOCK_1_CHECKPOINT_IDENTITY_VALID:

    raise RuntimeError(
        "Notebook 09 final checkpoint identity contract is invalid."
    )


# =============================================================================
# Checkpoint top-level state contract
# =============================================================================

BLOCK_1_REQUIRED_CHECKPOINT_KEYS = {
    "artifact_type",
    "checkpoint_schema_version",
    "architecture",
    "recurrent_contract",
    "representation_state",
    "transformative_state",
    "transformative_weight_state",
    "effective_transformative_weights",
    "pilot_training",
    "completion",
}


BLOCK_1_CHECKPOINT_REQUIRED_KEYS_VALID = (
    BLOCK_1_REQUIRED_CHECKPOINT_KEYS
    .issubset(
        set(
            NOTEBOOK_10_INHERITED_CHECKPOINT.keys()
        )
    )
)


if not BLOCK_1_CHECKPOINT_REQUIRED_KEYS_VALID:

    missing_keys = (
        BLOCK_1_REQUIRED_CHECKPOINT_KEYS
        -
        set(
            NOTEBOOK_10_INHERITED_CHECKPOINT.keys()
        )
    )


    raise RuntimeError(
        "Notebook 09 final checkpoint is missing required state. "
        f"Missing keys: {sorted(missing_keys)}"
    )


# =============================================================================
# Architecture contract from checkpoint
# =============================================================================

BLOCK_1_CHECKPOINT_ARCHITECTURE = (
    NOTEBOOK_10_INHERITED_CHECKPOINT[
        "architecture"
    ]
)


BLOCK_1_CHECKPOINT_PATHWAY_DIMENSIONS = {
    pathway_name:
        int(
            dimension
        )

    for (
        pathway_name,
        dimension,
    ) in BLOCK_1_CHECKPOINT_ARCHITECTURE[
        "pathway_dimensions"
    ].items()
}


BLOCK_1_CHECKPOINT_ACTIVE_DIMENSIONS = {
    pathway_name:
        tuple(
            int(
                index
            )

            for index
            in indices
        )

    for (
        pathway_name,
        indices,
    ) in BLOCK_1_CHECKPOINT_ARCHITECTURE[
        "active_dimension_indices"
    ].items()
}


BLOCK_1_CHECKPOINT_DEFERRED_DIMENSIONS = {
    pathway_name:
        tuple(
            int(
                index
            )

            for index
            in indices
        )

    for (
        pathway_name,
        indices,
    ) in BLOCK_1_CHECKPOINT_ARCHITECTURE[
        "deferred_dimension_indices"
    ].items()
}


BLOCK_1_PATHWAY_DIMENSIONS_VALID = (
    BLOCK_1_CHECKPOINT_PATHWAY_DIMENSIONS
    ==
    BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS
)


BLOCK_1_ACTIVE_DIMENSIONS_VALID = (
    BLOCK_1_CHECKPOINT_ACTIVE_DIMENSIONS
    ==
    BLOCK_1_EXPECTED_ACTIVE_DIMENSIONS
)


BLOCK_1_DEFERRED_DIMENSIONS_VALID = (
    BLOCK_1_CHECKPOINT_DEFERRED_DIMENSIONS
    ==
    BLOCK_1_EXPECTED_DEFERRED_DIMENSIONS
)


# =============================================================================
# Active / deferred partition audit
# =============================================================================

BLOCK_1_DIMENSION_PARTITIONS_VALID = {}


for (
    pathway_name,
    pathway_dim,
) in BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS.items():

    active_set = set(
        BLOCK_1_CHECKPOINT_ACTIVE_DIMENSIONS[
            pathway_name
        ]
    )


    deferred_set = set(
        BLOCK_1_CHECKPOINT_DEFERRED_DIMENSIONS[
            pathway_name
        ]
    )


    BLOCK_1_DIMENSION_PARTITIONS_VALID[
        pathway_name
    ] = all(
        [
            active_set.isdisjoint(
                deferred_set
            ),

            (
                active_set
                |
                deferred_set
            )
            ==
            set(
                range(
                    pathway_dim
                )
            ),
        ]
    )


BLOCK_1_ALL_DIMENSION_PARTITIONS_VALID = all(
    BLOCK_1_DIMENSION_PARTITIONS_VALID.values()
)


if not all(
    [
        BLOCK_1_PATHWAY_DIMENSIONS_VALID,
        BLOCK_1_ACTIVE_DIMENSIONS_VALID,
        BLOCK_1_DEFERRED_DIMENSIONS_VALID,
        BLOCK_1_ALL_DIMENSION_PARTITIONS_VALID,
    ]
):

    raise RuntimeError(
        "Notebook 09 pathway dimensions or active/deferred "
        "Transformative-Weight partitions are invalid."
    )


# =============================================================================
# Parameter-count contract
# =============================================================================

BLOCK_1_CHECKPOINT_PARAMETER_COUNTS_VALID = all(
    [
        (
            BLOCK_1_CHECKPOINT_ARCHITECTURE.get(
                "representation_parameter_count"
            )
            ==
            BLOCK_1_EXPECTED_REPRESENTATION_PARAMETERS
        ),

        (
            BLOCK_1_CHECKPOINT_ARCHITECTURE.get(
                "transformative_parameter_count"
            )
            ==
            BLOCK_1_EXPECTED_TRANSFORMATIVE_PARAMETERS
        ),

        (
            BLOCK_1_CHECKPOINT_ARCHITECTURE.get(
                "transformative_weight_parameter_count"
            )
            ==
            BLOCK_1_EXPECTED_TRANSFORMATIVE_WEIGHT_PARAMETERS
        ),

        (
            BLOCK_1_CHECKPOINT_ARCHITECTURE.get(
                "total_parameter_count"
            )
            ==
            BLOCK_1_EXPECTED_TOTAL_PARAMETERS
        ),

        (
            BLOCK_1_CHECKPOINT_ARCHITECTURE.get(
                "structural_weight_dimensions"
            )
            ==
            BLOCK_1_EXPECTED_STRUCTURAL_WEIGHT_DIMENSIONS
        ),

        (
            BLOCK_1_CHECKPOINT_ARCHITECTURE.get(
                "active_weight_dimensions"
            )
            ==
            BLOCK_1_EXPECTED_ACTIVE_WEIGHT_DIMENSIONS
        ),

        (
            BLOCK_1_CHECKPOINT_ARCHITECTURE.get(
                "deferred_weight_dimensions"
            )
            ==
            BLOCK_1_EXPECTED_DEFERRED_WEIGHT_DIMENSIONS
        ),
    ]
)


if not BLOCK_1_CHECKPOINT_PARAMETER_COUNTS_VALID:

    raise RuntimeError(
        "Notebook 09 checkpoint parameter-count contract is invalid."
    )


# =============================================================================
# Metadata architecture contract
# =============================================================================

BLOCK_1_METADATA_ARCHITECTURE = (
    NOTEBOOK_10_INHERITED_METADATA.get(
        "architecture",
        {},
    )
)


BLOCK_1_METADATA_PATHWAY_DIMENSIONS = {
    pathway_name:
        int(
            dimension
        )

    for (
        pathway_name,
        dimension,
    ) in BLOCK_1_METADATA_ARCHITECTURE.get(
        "pathway_dimensions",
        {},
    ).items()
}


BLOCK_1_METADATA_ACTIVE_DIMENSIONS = {
    pathway_name:
        tuple(
            int(
                index
            )

            for index
            in indices
        )

    for (
        pathway_name,
        indices,
    ) in BLOCK_1_METADATA_ARCHITECTURE.get(
        "active_dimension_indices",
        {},
    ).items()
}


BLOCK_1_METADATA_DEFERRED_DIMENSIONS = {
    pathway_name:
        tuple(
            int(
                index
            )

            for index
            in indices
        )

    for (
        pathway_name,
        indices,
    ) in BLOCK_1_METADATA_ARCHITECTURE.get(
        "deferred_dimension_indices",
        {},
    ).items()
}


BLOCK_1_METADATA_ARCHITECTURE_VALID = all(
    [
        (
            BLOCK_1_METADATA_ARCHITECTURE.get(
                "representation_parameters"
            )
            ==
            BLOCK_1_EXPECTED_REPRESENTATION_PARAMETERS
        ),

        (
            BLOCK_1_METADATA_ARCHITECTURE.get(
                "transformative_parameters"
            )
            ==
            BLOCK_1_EXPECTED_TRANSFORMATIVE_PARAMETERS
        ),

        (
            BLOCK_1_METADATA_ARCHITECTURE.get(
                "transformative_weight_parameters"
            )
            ==
            BLOCK_1_EXPECTED_TRANSFORMATIVE_WEIGHT_PARAMETERS
        ),

        (
            BLOCK_1_METADATA_ARCHITECTURE.get(
                "total_parameters"
            )
            ==
            BLOCK_1_EXPECTED_TOTAL_PARAMETERS
        ),

        (
            BLOCK_1_METADATA_PATHWAY_DIMENSIONS
            ==
            BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS
        ),

        (
            BLOCK_1_METADATA_ACTIVE_DIMENSIONS
            ==
            BLOCK_1_EXPECTED_ACTIVE_DIMENSIONS
        ),

        (
            BLOCK_1_METADATA_DEFERRED_DIMENSIONS
            ==
            BLOCK_1_EXPECTED_DEFERRED_DIMENSIONS
        ),
    ]
)


if not BLOCK_1_METADATA_ARCHITECTURE_VALID:

    raise RuntimeError(
        "Notebook 09 final metadata architecture contract is invalid."
    )


# =============================================================================
# Cross-artefact contract
# =============================================================================

BLOCK_1_CROSS_ARTEFACT_VALID = all(
    [
        (
            BLOCK_1_CHECKPOINT_PATHWAY_DIMENSIONS
            ==
            BLOCK_1_METADATA_PATHWAY_DIMENSIONS
        ),

        (
            BLOCK_1_CHECKPOINT_ACTIVE_DIMENSIONS
            ==
            BLOCK_1_METADATA_ACTIVE_DIMENSIONS
        ),

        (
            BLOCK_1_CHECKPOINT_DEFERRED_DIMENSIONS
            ==
            BLOCK_1_METADATA_DEFERRED_DIMENSIONS
        ),

        (
            BLOCK_1_CHECKPOINT_ARCHITECTURE[
                "representation_parameter_count"
            ]
            ==
            BLOCK_1_METADATA_ARCHITECTURE[
                "representation_parameters"
            ]
        ),

        (
            BLOCK_1_CHECKPOINT_ARCHITECTURE[
                "transformative_parameter_count"
            ]
            ==
            BLOCK_1_METADATA_ARCHITECTURE[
                "transformative_parameters"
            ]
        ),

        (
            BLOCK_1_CHECKPOINT_ARCHITECTURE[
                "transformative_weight_parameter_count"
            ]
            ==
            BLOCK_1_METADATA_ARCHITECTURE[
                "transformative_weight_parameters"
            ]
        ),

        (
            BLOCK_1_CHECKPOINT_ARCHITECTURE[
                "total_parameter_count"
            ]
            ==
            BLOCK_1_METADATA_ARCHITECTURE[
                "total_parameters"
            ]
        ),

        (
            NOTEBOOK_10_INHERITED_CHECKPOINT[
                "recurrent_contract"
            ]
            ==
            NOTEBOOK_10_INHERITED_METADATA[
                "recurrent_contract"
            ]
        ),

        (
            BLOCK_1_CHECKPOINT_ARCHITECTURE.get(
                "structural_weight_dimensions"
            )
            ==
            BLOCK_1_EXPECTED_STRUCTURAL_WEIGHT_DIMENSIONS
        ),

        (
            BLOCK_1_CHECKPOINT_ARCHITECTURE.get(
                "active_weight_dimensions"
            )
            ==
            BLOCK_1_EXPECTED_ACTIVE_WEIGHT_DIMENSIONS
        ),

        (
            BLOCK_1_CHECKPOINT_ARCHITECTURE.get(
                "deferred_weight_dimensions"
            )
            ==
            BLOCK_1_EXPECTED_DEFERRED_WEIGHT_DIMENSIONS
        ),
    ]
)


if not BLOCK_1_CROSS_ARTEFACT_VALID:

    raise RuntimeError(
        "Notebook 09 checkpoint and metadata disagree on "
        "the inherited architecture."
    )


# =============================================================================
# Deterministic representation architecture reconstruction
# =============================================================================

class Notebook10Backbone(
    nn.Module
):

    def __init__(
        self,
        input_dim=768,
        output_dim=768,
    ):

        super().__init__()


        self.projection = nn.Linear(
            input_dim,
            output_dim,
        )


    def forward(
        self,
        x,
    ):

        return self.projection(
            x
        )


class Notebook10GlobalConfluent(
    nn.Module
):

    def __init__(
        self,
        input_dim=768,
        output_dim=256,
        dropout=0.10,
    ):

        super().__init__()


        self.global_projection = nn.Sequential(
            nn.Linear(
                input_dim,
                output_dim,
            ),

            nn.GELU(),

            nn.Dropout(
                dropout
            ),

            nn.LayerNorm(
                output_dim
            ),
        )


    def forward(
        self,
        x,
    ):

        return self.global_projection(
            x
        )


class Notebook10RepresentationCore(
    nn.Module
):

    def __init__(
        self,
        input_dim=256,
        hidden_dim=128,
        dropout=0.10,
    ):

        super().__init__()


        self.network = nn.Sequential(
            nn.Linear(
                input_dim,
                hidden_dim,
            ),

            nn.GELU(),

            nn.Dropout(
                dropout
            ),

            nn.LayerNorm(
                hidden_dim
            ),
        )


    def forward(
        self,
        x,
    ):

        return self.network(
            x
        )


class Notebook10RepresentationHead(
    nn.Module
):

    def __init__(
        self,
        output_dim,
        input_dim=256,
        hidden_dim=128,
        dropout=0.10,
    ):

        super().__init__()


        self.core = Notebook10RepresentationCore(
            input_dim=
                input_dim,

            hidden_dim=
                hidden_dim,

            dropout=
                dropout,
        )


        self.output_layer = nn.Linear(
            hidden_dim,
            output_dim,
        )


    def forward(
        self,
        x,
    ):

        return self.output_layer(
            self.core(
                x
            )
        )


# =============================================================================
# Deterministic transformative architecture reconstruction
# =============================================================================

class Notebook10TransformativeMechanism(
    nn.Module
):

    def __init__(
        self,
        pathway_dim,
        dropout=0.10,
    ):

        super().__init__()


        input_dim = (
            pathway_dim
            *
            2
        )


        hidden_dim = input_dim


        self.pathway_dim = int(
            pathway_dim
        )


        self.transform_network = nn.Sequential(
            nn.Linear(
                input_dim,
                hidden_dim,
            ),

            nn.GELU(),

            nn.Dropout(
                dropout
            ),

            nn.LayerNorm(
                hidden_dim
            ),

            nn.Linear(
                hidden_dim,
                pathway_dim,
            ),
        )


    def forward(
        self,
        current_representation,
        preceding_state,
    ):

        combined = torch.cat(
            (
                current_representation,
                preceding_state,
            ),
            dim=
                -1,
        )


        return self.transform_network(
            combined
        )


# =============================================================================
# Instantiate representation and transformative modules
# =============================================================================

NOTEBOOK_10_RESTORED_BACKBONE = (
    Notebook10Backbone()
)


NOTEBOOK_10_RESTORED_GLOBAL_CONFLUENT = (
    Notebook10GlobalConfluent()
)


NOTEBOOK_10_RESTORED_PSYCHOLOGICAL_HEAD = (
    Notebook10RepresentationHead(
        output_dim=
            BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS[
                "psychological"
            ]
    )
)


NOTEBOOK_10_RESTORED_FACTUAL_HEAD = (
    Notebook10RepresentationHead(
        output_dim=
            BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS[
                "factual"
            ]
    )
)


NOTEBOOK_10_RESTORED_SOCIAL_HEAD = (
    Notebook10RepresentationHead(
        output_dim=
            BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS[
                "social"
            ]
    )
)


NOTEBOOK_10_FACTUAL_TRANSFORMATIVE_MECHANISM = (
    Notebook10TransformativeMechanism(
        pathway_dim=
            BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS[
                "factual"
            ]
    )
)


NOTEBOOK_10_PSYCHOLOGICAL_TRANSFORMATIVE_MECHANISM = (
    Notebook10TransformativeMechanism(
        pathway_dim=
            BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS[
                "psychological"
            ]
    )
)


NOTEBOOK_10_SOCIAL_TRANSFORMATIVE_MECHANISM = (
    Notebook10TransformativeMechanism(
        pathway_dim=
            BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS[
                "social"
            ]
    )
)


NOTEBOOK_10_REPRESENTATION_MODULES = {
    "backbone":
        NOTEBOOK_10_RESTORED_BACKBONE,

    "global_confluent":
        NOTEBOOK_10_RESTORED_GLOBAL_CONFLUENT,

    "psychological_head":
        NOTEBOOK_10_RESTORED_PSYCHOLOGICAL_HEAD,

    "factual_head":
        NOTEBOOK_10_RESTORED_FACTUAL_HEAD,

    "social_head":
        NOTEBOOK_10_RESTORED_SOCIAL_HEAD,
}


NOTEBOOK_10_TRANSFORMATIVE_MODULES = {
    "factual":
        NOTEBOOK_10_FACTUAL_TRANSFORMATIVE_MECHANISM,

    "psychological":
        NOTEBOOK_10_PSYCHOLOGICAL_TRANSFORMATIVE_MECHANISM,

    "social":
        NOTEBOOK_10_SOCIAL_TRANSFORMATIVE_MECHANISM,
}


# =============================================================================
# Restore representation state
# =============================================================================

BLOCK_1_CHECKPOINT_REPRESENTATION_STATE = (
    NOTEBOOK_10_INHERITED_CHECKPOINT[
        "representation_state"
    ]
)


for (
    module_name,
    module,
) in NOTEBOOK_10_REPRESENTATION_MODULES.items():

    if module_name not in BLOCK_1_CHECKPOINT_REPRESENTATION_STATE:

        raise RuntimeError(
            "Notebook 09 checkpoint is missing representation state for "
            f"'{module_name}'."
        )


    module.load_state_dict(
        BLOCK_1_CHECKPOINT_REPRESENTATION_STATE[
            module_name
        ],
        strict=
            True,
    )


# =============================================================================
# Restore transformative state
# =============================================================================

BLOCK_1_CHECKPOINT_TRANSFORMATIVE_STATE = (
    NOTEBOOK_10_INHERITED_CHECKPOINT[
        "transformative_state"
    ]
)


for (
    pathway_name,
    module,
) in NOTEBOOK_10_TRANSFORMATIVE_MODULES.items():

    if pathway_name not in BLOCK_1_CHECKPOINT_TRANSFORMATIVE_STATE:

        raise RuntimeError(
            "Notebook 09 checkpoint is missing transformative state for "
            f"'{pathway_name}'."
        )


    module.load_state_dict(
        BLOCK_1_CHECKPOINT_TRANSFORMATIVE_STATE[
            pathway_name
        ],
        strict=
            True,
    )


# =============================================================================
# Exact state helper
# =============================================================================

def block_1_state_dict_exact(
    expected_state,
    observed_state,
):

    if (
        set(
            expected_state.keys()
        )
        !=
        set(
            observed_state.keys()
        )
    ):

        return False


    for key in expected_state:

        expected = expected_state[
            key
        ]


        observed = observed_state[
            key
        ]


        if not (
            torch.is_tensor(
                expected
            )
            and
            torch.is_tensor(
                observed
            )
        ):

            return False


        if not torch.equal(
            expected.detach().cpu(),
            observed.detach().cpu(),
        ):

            return False


    return True


# =============================================================================
# Exact representation / transformative restoration validation
# =============================================================================

BLOCK_1_REPRESENTATION_STATE_EXACT = all(
    block_1_state_dict_exact(
        BLOCK_1_CHECKPOINT_REPRESENTATION_STATE[
            module_name
        ],
        module.state_dict(),
    )

    for (
        module_name,
        module,
    ) in NOTEBOOK_10_REPRESENTATION_MODULES.items()
)


BLOCK_1_TRANSFORMATIVE_STATE_EXACT = all(
    block_1_state_dict_exact(
        BLOCK_1_CHECKPOINT_TRANSFORMATIVE_STATE[
            pathway_name
        ],
        module.state_dict(),
    )

    for (
        pathway_name,
        module,
    ) in NOTEBOOK_10_TRANSFORMATIVE_MODULES.items()
)


if not all(
    [
        BLOCK_1_REPRESENTATION_STATE_EXACT,
        BLOCK_1_TRANSFORMATIVE_STATE_EXACT,
    ]
):

    raise RuntimeError(
        "Notebook 10 could not reproduce the persisted representation "
        "or transformative state exactly."
    )


# =============================================================================
# Transformative-Weight persisted-state inspection
# =============================================================================
#
# Notebook 09 persists one learned latent parameter vector per pathway.
#
# We deliberately inspect the persisted state rather than assuming that its
# parameter key has a particular name.
# =============================================================================

BLOCK_1_CHECKPOINT_WEIGHT_STATE = (
    NOTEBOOK_10_INHERITED_CHECKPOINT[
        "transformative_weight_state"
    ]
)


def block_1_resolve_latent_tensor(
    pathway_name,
    state_dict,
    expected_active_count,
):

    if not isinstance(
        state_dict,
        dict,
    ):

        raise TypeError(
            f"Persisted Transformative-Weight state for "
            f"{pathway_name} is not dictionary-like."
        )


    tensor_entries = {
        key:
            value

        for (
            key,
            value,
        ) in state_dict.items()

        if torch.is_tensor(
            value
        )
    }


    preferred_candidates = [
        (
            key,
            value,
        )

        for (
            key,
            value,
        ) in tensor_entries.items()

        if (
            value.is_floating_point()
            and
            value.numel()
            ==
            expected_active_count
            and
            (
                "alpha"
                in key.lower()
                or
                "latent"
                in key.lower()
            )
        )
    ]


    if len(
        preferred_candidates
    ) == 1:

        return preferred_candidates[
            0
        ]


    general_candidates = [
        (
            key,
            value,
        )

        for (
            key,
            value,
        ) in tensor_entries.items()

        if (
            value.is_floating_point()
            and
            value.numel()
            ==
            expected_active_count
        )
    ]


    if len(
        general_candidates
    ) == 1:

        return general_candidates[
            0
        ]


    state_description = {
        key:
            {
                "shape":
                    tuple(
                        value.shape
                    ),

                "dtype":
                    str(
                        value.dtype
                    ),

                "floating":
                    bool(
                        value.is_floating_point()
                    ),
            }

        for (
            key,
            value,
        ) in tensor_entries.items()
    }


    raise RuntimeError(
        "Notebook 10 could not identify exactly one learned latent "
        f"Transformative-Weight tensor for pathway '{pathway_name}'. "
        f"Persisted state: {state_description}"
    )


BLOCK_1_LATENT_SOURCE_KEYS = {}

BLOCK_1_LATENT_WEIGHT_TENSORS = {}


for pathway_name in BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS:

    (
        latent_key,
        latent_tensor,
    ) = block_1_resolve_latent_tensor(
        pathway_name=
            pathway_name,

        state_dict=
            BLOCK_1_CHECKPOINT_WEIGHT_STATE[
                pathway_name
            ],

        expected_active_count=
            len(
                BLOCK_1_EXPECTED_ACTIVE_DIMENSIONS[
                    pathway_name
                ]
            ),
    )


    BLOCK_1_LATENT_SOURCE_KEYS[
        pathway_name
    ] = latent_key


    BLOCK_1_LATENT_WEIGHT_TENSORS[
        pathway_name
    ] = (
        latent_tensor.detach()
        .cpu()
        .clone()
        .reshape(
            -1
        )
    )


# =============================================================================
# Deterministic Transformative-Weight reconstruction
# =============================================================================

class Notebook10TransformativeWeight(
    nn.Module
):

    def __init__(
        self,
        structural_dim,
        active_indices,
        latent_values,
    ):

        super().__init__()


        self.structural_dim = int(
            structural_dim
        )


        self.active_indices = tuple(
            int(
                index
            )

            for index
            in active_indices
        )


        if latent_values.numel() != len(
            self.active_indices
        ):

            raise ValueError(
                "Transformative-Weight latent parameter count does not "
                "match the active-dimension count."
            )


        self.alpha = nn.Parameter(
            latent_values.detach()
            .clone()
            .to(
                dtype=
                    DEFAULT_DTYPE
            )
            .reshape(
                -1
            )
        )


    def forward(
        self,
    ):

        effective_weight = torch.zeros(
            self.structural_dim,
            dtype=
                self.alpha.dtype,
            device=
                self.alpha.device,
        )


        active_index_tensor = torch.tensor(
            self.active_indices,
            dtype=
                torch.long,
            device=
                self.alpha.device,
        )


        effective_weight = effective_weight.index_copy(
            0,
            active_index_tensor,
            torch.sigmoid(
                self.alpha
            ),
        )


        return effective_weight


NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES = {
    pathway_name:
        Notebook10TransformativeWeight(
            structural_dim=
                BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS[
                    pathway_name
                ],

            active_indices=
                BLOCK_1_EXPECTED_ACTIVE_DIMENSIONS[
                    pathway_name
                ],

            latent_values=
                BLOCK_1_LATENT_WEIGHT_TENSORS[
                    pathway_name
                ],
        )

    for pathway_name
    in BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS
}


# =============================================================================
# Persisted effective Transformative-Weight contract
# =============================================================================

BLOCK_1_PERSISTED_EFFECTIVE_WEIGHTS = {
    pathway_name:
        NOTEBOOK_10_INHERITED_CHECKPOINT[
            "effective_transformative_weights"
        ][
            pathway_name
        ]
        .detach()
        .cpu()
        .clone()

    for pathway_name
    in BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS
}


# =============================================================================
# Validate reconstructed effective weights without invoking module.forward()
# =============================================================================

BLOCK_1_RECONSTRUCTED_EFFECTIVE_WEIGHTS = {}


for pathway_name in BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS:

    structural_dim = (
        BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS[
            pathway_name
        ]
    )


    active_indices = (
        BLOCK_1_EXPECTED_ACTIVE_DIMENSIONS[
            pathway_name
        ]
    )


    latent_values = (
        BLOCK_1_LATENT_WEIGHT_TENSORS[
            pathway_name
        ]
    )


    reconstructed_weight = torch.zeros(
        structural_dim,
        dtype=
            DEFAULT_DTYPE,
    )


    reconstructed_weight[
        list(
            active_indices
        )
    ] = torch.sigmoid(
        latent_values.to(
            dtype=
                DEFAULT_DTYPE
        )
    )


    BLOCK_1_RECONSTRUCTED_EFFECTIVE_WEIGHTS[
        pathway_name
    ] = reconstructed_weight


BLOCK_1_EFFECTIVE_WEIGHT_IDENTITY_VALID = all(
    torch.equal(
        BLOCK_1_RECONSTRUCTED_EFFECTIVE_WEIGHTS[
            pathway_name
        ],
        BLOCK_1_PERSISTED_EFFECTIVE_WEIGHTS[
            pathway_name
        ].to(
            dtype=
                DEFAULT_DTYPE
        ),
    )

    for pathway_name
    in BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS
)


if not BLOCK_1_EFFECTIVE_WEIGHT_IDENTITY_VALID:

    raise RuntimeError(
        "Notebook 10 reconstructed Transformative Weights do not "
        "reproduce the persisted Notebook 09 effective-weight state."
    )


# =============================================================================
# Transformative-Weight numerical validation
# =============================================================================

BLOCK_1_EFFECTIVE_WEIGHTS_FINITE = {}

BLOCK_1_EFFECTIVE_WEIGHTS_BOUNDED = {}

BLOCK_1_DEFERRED_WEIGHTS_ZERO = {}


for pathway_name in BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS:

    effective_weight = (
        BLOCK_1_RECONSTRUCTED_EFFECTIVE_WEIGHTS[
            pathway_name
        ]
    )


    deferred_indices = (
        BLOCK_1_EXPECTED_DEFERRED_DIMENSIONS[
            pathway_name
        ]
    )


    BLOCK_1_EFFECTIVE_WEIGHTS_FINITE[
        pathway_name
    ] = bool(
        torch.isfinite(
            effective_weight
        ).all().item()
    )


    BLOCK_1_EFFECTIVE_WEIGHTS_BOUNDED[
        pathway_name
    ] = bool(
        (
            (
                effective_weight
                >=
                0.0
            )
            &
            (
                effective_weight
                <=
                1.0
            )
        ).all().item()
    )


    BLOCK_1_DEFERRED_WEIGHTS_ZERO[
        pathway_name
    ] = torch.equal(
        effective_weight[
            list(
                deferred_indices
            )
        ],
        torch.zeros(
            len(
                deferred_indices
            ),
            dtype=
                effective_weight.dtype,
        ),
    )


BLOCK_1_ALL_EFFECTIVE_WEIGHTS_FINITE = all(
    BLOCK_1_EFFECTIVE_WEIGHTS_FINITE.values()
)


BLOCK_1_ALL_EFFECTIVE_WEIGHTS_BOUNDED = all(
    BLOCK_1_EFFECTIVE_WEIGHTS_BOUNDED.values()
)


BLOCK_1_ALL_DEFERRED_WEIGHTS_ZERO = all(
    BLOCK_1_DEFERRED_WEIGHTS_ZERO.values()
)


if not all(
    [
        BLOCK_1_ALL_EFFECTIVE_WEIGHTS_FINITE,
        BLOCK_1_ALL_EFFECTIVE_WEIGHTS_BOUNDED,
        BLOCK_1_ALL_DEFERRED_WEIGHTS_ZERO,
    ]
):

    raise RuntimeError(
        "Notebook 10 inherited Transformative-Weight numerical "
        "contract is invalid."
    )


# =============================================================================
# Move restored architecture to controlled device
# =============================================================================

for module_collection in (
    NOTEBOOK_10_REPRESENTATION_MODULES,
    NOTEBOOK_10_TRANSFORMATIVE_MODULES,
    NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
):

    for module in module_collection.values():

        module.to(
            device=
                DEVICE,
            dtype=
                DEFAULT_DTYPE,
        )


# =============================================================================
# Freeze complete inherited architecture
# =============================================================================

for module_collection in (
    NOTEBOOK_10_REPRESENTATION_MODULES,
    NOTEBOOK_10_TRANSFORMATIVE_MODULES,
    NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
):

    for module in module_collection.values():

        module.eval()


        for parameter in module.parameters():

            parameter.requires_grad_(
                False
            )


# =============================================================================
# Freeze / evaluation-mode audit
# =============================================================================

BLOCK_1_ALL_PARAMETERS_FROZEN = all(
    not parameter.requires_grad

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


BLOCK_1_ALL_MODULES_EVAL = all(
    not module.training

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()
)


if not all(
    [
        BLOCK_1_ALL_PARAMETERS_FROZEN,
        BLOCK_1_ALL_MODULES_EVAL,
    ]
):

    raise RuntimeError(
        "Notebook 10 inherited architecture is not fully frozen "
        "and in evaluation mode."
    )


# =============================================================================
# Actual parameter accounting
# =============================================================================

BLOCK_1_REPRESENTATION_PARAMETER_COUNTS = {
    module_name:
        int(
            sum(
                parameter.numel()

                for parameter
                in module.parameters()
            )
        )

    for (
        module_name,
        module,
    ) in NOTEBOOK_10_REPRESENTATION_MODULES.items()
}


BLOCK_1_TRANSFORMATIVE_PARAMETER_COUNTS = {
    pathway_name:
        int(
            sum(
                parameter.numel()

                for parameter
                in module.parameters()
            )
        )

    for (
        pathway_name,
        module,
    ) in NOTEBOOK_10_TRANSFORMATIVE_MODULES.items()
}


BLOCK_1_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNTS = {
    pathway_name:
        int(
            sum(
                parameter.numel()

                for parameter
                in module.parameters()
            )
        )

    for (
        pathway_name,
        module,
    ) in NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES.items()
}


BLOCK_1_REPRESENTATION_PARAMETER_COUNT = int(
    sum(
        BLOCK_1_REPRESENTATION_PARAMETER_COUNTS.values()
    )
)


BLOCK_1_TRANSFORMATIVE_PARAMETER_COUNT = int(
    sum(
        BLOCK_1_TRANSFORMATIVE_PARAMETER_COUNTS.values()
    )
)


BLOCK_1_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT = int(
    sum(
        BLOCK_1_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNTS.values()
    )
)


BLOCK_1_TOTAL_PARAMETER_COUNT = int(
    BLOCK_1_REPRESENTATION_PARAMETER_COUNT
    +
    BLOCK_1_TRANSFORMATIVE_PARAMETER_COUNT
    +
    BLOCK_1_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT
)


BLOCK_1_PARAMETER_COUNTS_VALID = all(
    [
        (
            BLOCK_1_REPRESENTATION_PARAMETER_COUNT
            ==
            BLOCK_1_EXPECTED_REPRESENTATION_PARAMETERS
        ),

        (
            BLOCK_1_TRANSFORMATIVE_PARAMETER_COUNT
            ==
            BLOCK_1_EXPECTED_TRANSFORMATIVE_PARAMETERS
        ),

        (
            BLOCK_1_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT
            ==
            BLOCK_1_EXPECTED_TRANSFORMATIVE_WEIGHT_PARAMETERS
        ),

        (
            BLOCK_1_TOTAL_PARAMETER_COUNT
            ==
            BLOCK_1_EXPECTED_TOTAL_PARAMETERS
        ),
    ]
)


if not BLOCK_1_PARAMETER_COUNTS_VALID:

    raise RuntimeError(
        "Notebook 10 restored parameter counts do not reproduce "
        "the canonical Notebook 09 architecture."
    )


# =============================================================================
# Expected per-module parameter counts
# =============================================================================

BLOCK_1_EXPECTED_REPRESENTATION_PARAMETER_COUNTS = {
    "backbone":
        590_592,

    "global_confluent":
        197_376,

    "psychological_head":
        37_538,

    "factual_head":
        34_442,

    "social_head":
        34_055,
}


BLOCK_1_EXPECTED_TRANSFORMATIVE_PARAMETER_COUNTS = {
    "factual":
        670,

    "psychological":
        7_174,

    "social":
        343,
}


BLOCK_1_PER_MODULE_PARAMETER_COUNTS_VALID = all(
    [
        (
            BLOCK_1_REPRESENTATION_PARAMETER_COUNTS
            ==
            BLOCK_1_EXPECTED_REPRESENTATION_PARAMETER_COUNTS
        ),

        (
            BLOCK_1_TRANSFORMATIVE_PARAMETER_COUNTS
            ==
            BLOCK_1_EXPECTED_TRANSFORMATIVE_PARAMETER_COUNTS
        ),

        (
            BLOCK_1_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNTS
            ==
            BLOCK_1_EXPECTED_WEIGHT_PARAMETER_COUNTS
        ),
    ]
)


if not BLOCK_1_PER_MODULE_PARAMETER_COUNTS_VALID:

    raise RuntimeError(
        "Notebook 10 per-module parameter accounting does not "
        "reproduce the validated architecture."
    )


# =============================================================================
# Recurrent contract restoration
# =============================================================================

NOTEBOOK_10_RECURRENT_CONTRACT = deepcopy(
    NOTEBOOK_10_INHERITED_CHECKPOINT[
        "recurrent_contract"
    ]
)


BLOCK_1_EXPECTED_RECURRENT_CONTRACT = {
    "update_type":
        "additive",

    "initial_state_policy":
        "deterministic_zero",

    "article_boundary_reset":
        True,

    "cross_article_propagation":
        False,

    "cross_pathway_recurrence":
        False,

    "future_context_allowed":
        False,
}


BLOCK_1_RECURRENT_CONTRACT_VALID = (
    NOTEBOOK_10_RECURRENT_CONTRACT
    ==
    BLOCK_1_EXPECTED_RECURRENT_CONTRACT
)


if not BLOCK_1_RECURRENT_CONTRACT_VALID:

    raise RuntimeError(
        "Notebook 09 → 10 recurrent-state contract is invalid."
    )


# =============================================================================
# Pilot-training methodological boundary
# =============================================================================

BLOCK_1_CHECKPOINT_PILOT_TRAINING = (
    NOTEBOOK_10_INHERITED_CHECKPOINT[
        "pilot_training"
    ]
)


BLOCK_1_METADATA_PILOT_TRAINING = (
    NOTEBOOK_10_INHERITED_METADATA[
        "pilot_training"
    ]
)


BLOCK_1_PILOT_TRAINING_BOUNDARY_VALID = all(
    [
        (
            BLOCK_1_CHECKPOINT_PILOT_TRAINING.get(
                "final_training"
            )
            is False
        ),

        (
            BLOCK_1_CHECKPOINT_PILOT_TRAINING.get(
                "expanded_corpus_training_required"
            )
            is True
        ),

        (
            BLOCK_1_METADATA_PILOT_TRAINING.get(
                "final_training"
            )
            is False
        ),

        (
            BLOCK_1_METADATA_PILOT_TRAINING.get(
                "expanded_corpus_training_required"
            )
            is True
        ),
    ]
)


if not BLOCK_1_PILOT_TRAINING_BOUNDARY_VALID:

    raise RuntimeError(
        "Notebook 09 pilot-training methodological boundary "
        "was not preserved."
    )


# =============================================================================
# Completion state
# =============================================================================

BLOCK_1_CHECKPOINT_COMPLETION = (
    NOTEBOOK_10_INHERITED_CHECKPOINT[
        "completion"
    ]
)


BLOCK_1_METADATA_COMPLETION = (
    NOTEBOOK_10_INHERITED_METADATA[
        "completion"
    ]
)


BLOCK_1_NOTEBOOK_09_COMPLETION_VALID = all(
    [
        (
            BLOCK_1_CHECKPOINT_COMPLETION.get(
                "architecture_complete"
            )
            is True
        ),

        (
            BLOCK_1_CHECKPOINT_COMPLETION.get(
                "controlled_training_valid"
            )
            is True
        ),

        (
            BLOCK_1_CHECKPOINT_COMPLETION.get(
                "posttraining_recurrent_validation_valid"
            )
            is True
        ),

        (
            BLOCK_1_CHECKPOINT_COMPLETION.get(
                "notebook_09_complete"
            )
            is True
        ),

        (
            BLOCK_1_METADATA_COMPLETION.get(
                "notebook_09_complete"
            )
            is True
        ),

        (
            BLOCK_1_METADATA_COMPLETION.get(
                "architecture_complete"
            )
            is True
        ),

        (
            BLOCK_1_METADATA_COMPLETION.get(
                "posttraining_validation_valid"
            )
            is True
        ),

        (
            BLOCK_1_METADATA_COMPLETION.get(
                "notebook_10_ready"
            )
            is True
        ),
    ]
)


if not BLOCK_1_NOTEBOOK_09_COMPLETION_VALID:

    raise RuntimeError(
        "Notebook 09 completion state is not valid for Notebook 10."
    )


# =============================================================================
# Post-device exact state validation
# =============================================================================
#
# DEVICE is intentionally CPU in Block 1, allowing exact comparison against
# the CPU-safe checkpoint tensors.
# =============================================================================

BLOCK_1_POST_DEVICE_REPRESENTATION_EXACT = all(
    block_1_state_dict_exact(
        BLOCK_1_CHECKPOINT_REPRESENTATION_STATE[
            module_name
        ],
        module.state_dict(),
    )

    for (
        module_name,
        module,
    ) in NOTEBOOK_10_REPRESENTATION_MODULES.items()
)


BLOCK_1_POST_DEVICE_TRANSFORMATIVE_EXACT = all(
    block_1_state_dict_exact(
        BLOCK_1_CHECKPOINT_TRANSFORMATIVE_STATE[
            pathway_name
        ],
        module.state_dict(),
    )

    for (
        pathway_name,
        module,
    ) in NOTEBOOK_10_TRANSFORMATIVE_MODULES.items()
)


BLOCK_1_POST_DEVICE_WEIGHT_LATENT_EXACT = all(
    torch.equal(
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES[
            pathway_name
        ].alpha.detach()
        .cpu(),

        BLOCK_1_LATENT_WEIGHT_TENSORS[
            pathway_name
        ].to(
            dtype=
                DEFAULT_DTYPE
        ),
    )

    for pathway_name
    in BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS
)


if not all(
    [
        BLOCK_1_POST_DEVICE_REPRESENTATION_EXACT,
        BLOCK_1_POST_DEVICE_TRANSFORMATIVE_EXACT,
        BLOCK_1_POST_DEVICE_WEIGHT_LATENT_EXACT,
    ]
):

    raise RuntimeError(
        "Notebook 10 post-device restored model state does not "
        "reproduce the Notebook 09 checkpoint exactly."
    )


# =============================================================================
# Gradient-state validation
# =============================================================================

BLOCK_1_ALL_GRADIENTS_ABSENT = all(
    parameter.grad is None

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


if not BLOCK_1_ALL_GRADIENTS_ABSENT:

    raise RuntimeError(
        "Unexpected gradients exist after Notebook 10 restoration."
    )


# =============================================================================
# Expose stable Notebook 10 architecture globals
# =============================================================================

NOTEBOOK_10_PATHWAY_DIMENSIONS = deepcopy(
    BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS
)


NOTEBOOK_10_ACTIVE_DIMENSION_INDICES = deepcopy(
    BLOCK_1_EXPECTED_ACTIVE_DIMENSIONS
)


NOTEBOOK_10_DEFERRED_DIMENSION_INDICES = deepcopy(
    BLOCK_1_EXPECTED_DEFERRED_DIMENSIONS
)


NOTEBOOK_10_REPRESENTATION_PARAMETER_COUNT = (
    BLOCK_1_REPRESENTATION_PARAMETER_COUNT
)


NOTEBOOK_10_TRANSFORMATIVE_PARAMETER_COUNT = (
    BLOCK_1_TRANSFORMATIVE_PARAMETER_COUNT
)


NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT = (
    BLOCK_1_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT
)


NOTEBOOK_10_TOTAL_PARAMETER_COUNT = (
    BLOCK_1_TOTAL_PARAMETER_COUNT
)


NOTEBOOK_10_INHERITED_EFFECTIVE_TRANSFORMATIVE_WEIGHTS = {
    pathway_name:
        BLOCK_1_RECONSTRUCTED_EFFECTIVE_WEIGHTS[
            pathway_name
        ].detach()
        .cpu()
        .clone()

    for pathway_name
    in BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS
}


NOTEBOOK_10_NOTEBOOK_09_PILOT_PRETRAIN_OBJECTIVE = float(
    BLOCK_1_CHECKPOINT_PILOT_TRAINING[
        "pretraining_total_objective"
    ]
)


NOTEBOOK_10_NOTEBOOK_09_PILOT_POSTTRAIN_OBJECTIVE = float(
    BLOCK_1_CHECKPOINT_PILOT_TRAINING[
        "posttraining_total_objective"
    ]
)


# =============================================================================
# Explicit Block 1 prohibited-operation validation
# =============================================================================

BLOCK_1_EXECUTION_BOUNDARY_VALID = all(
    [
        NOTEBOOK_10_BLOCK_1_EXPANDED_CORPUS_LOADED is False,

        NOTEBOOK_10_BLOCK_1_REPRESENTATION_FORWARD_EXECUTED is False,

        NOTEBOOK_10_BLOCK_1_TRANSFORMATIVE_FORWARD_EXECUTED is False,

        NOTEBOOK_10_BLOCK_1_WEIGHT_FORWARD_EXECUTED is False,

        NOTEBOOK_10_BLOCK_1_RECURRENT_STATE_CREATED is False,

        NOTEBOOK_10_BLOCK_1_RECURRENT_UPDATE_EXECUTED is False,

        NOTEBOOK_10_BLOCK_1_LOSS_CALCULATED is False,

        NOTEBOOK_10_BLOCK_1_OPTIMIZER_CREATED is False,

        NOTEBOOK_10_BLOCK_1_ZERO_GRAD_EXECUTED is False,

        NOTEBOOK_10_BLOCK_1_BACKWARD_PASS_EXECUTED is False,

        NOTEBOOK_10_BLOCK_1_PARAMETER_UPDATE_EXECUTED is False,

        NOTEBOOK_10_BLOCK_1_TRAINING_EXECUTED is False,
    ]
)


# =============================================================================
# Final Block 1 validation
# =============================================================================

NOTEBOOK_10_BLOCK_1_ERRORS = []


BLOCK_1_VALIDATION_CHECKS = {
    "drive_service_not_initialised":
        BLOCK_1_DRIVE_SERVICE_INITIALISED,

    "checkpoint_not_located":
        BLOCK_1_CHECKPOINT_LOCATED,

    "metadata_not_located":
        BLOCK_1_METADATA_LOCATED,

    "checkpoint_filename_invalid":
        BLOCK_1_CHECKPOINT_FILENAME_VALID,

    "metadata_filename_invalid":
        BLOCK_1_METADATA_FILENAME_VALID,

    "metadata_download_invalid":
        BLOCK_1_METADATA_DOWNLOADED,

    "metadata_drive_byte_count_invalid":
        BLOCK_1_METADATA_DRIVE_BYTE_COUNT_VALID,

    "metadata_json_invalid":
        BLOCK_1_METADATA_JSON_PARSED,

    "metadata_identity_invalid":
        BLOCK_1_METADATA_IDENTITY_VALID,

    "metadata_bootstrap_invalid":
        BLOCK_1_METADATA_BOOTSTRAP_VALID,

    "metadata_checkpoint_identity_invalid":
        BLOCK_1_METADATA_CHECKPOINT_IDENTITY_VALID,

    "checkpoint_download_invalid":
        BLOCK_1_CHECKPOINT_DOWNLOADED,

    "checkpoint_sha256_invalid":
        BLOCK_1_CHECKPOINT_SHA256_VALID,

    "checkpoint_metadata_byte_count_invalid":
        BLOCK_1_CHECKPOINT_METADATA_BYTE_COUNT_VALID,

    "checkpoint_drive_byte_count_invalid":
        BLOCK_1_CHECKPOINT_DRIVE_BYTE_COUNT_VALID,

    "checkpoint_parse_invalid":
        BLOCK_1_CHECKPOINT_PARSED,

    "checkpoint_identity_invalid":
        BLOCK_1_CHECKPOINT_IDENTITY_VALID,

    "checkpoint_required_keys_invalid":
        BLOCK_1_CHECKPOINT_REQUIRED_KEYS_VALID,

    "checkpoint_parameter_counts_invalid":
        BLOCK_1_CHECKPOINT_PARAMETER_COUNTS_VALID,

    "metadata_architecture_invalid":
        BLOCK_1_METADATA_ARCHITECTURE_VALID,

    "cross_artefact_invalid":
        BLOCK_1_CROSS_ARTEFACT_VALID,

    "pathway_dimensions_invalid":
        BLOCK_1_PATHWAY_DIMENSIONS_VALID,

    "active_dimensions_invalid":
        BLOCK_1_ACTIVE_DIMENSIONS_VALID,

    "deferred_dimensions_invalid":
        BLOCK_1_DEFERRED_DIMENSIONS_VALID,

    "dimension_partitions_invalid":
        BLOCK_1_ALL_DIMENSION_PARTITIONS_VALID,

    "representation_state_not_exact":
        BLOCK_1_REPRESENTATION_STATE_EXACT,

    "transformative_state_not_exact":
        BLOCK_1_TRANSFORMATIVE_STATE_EXACT,

    "effective_weight_identity_invalid":
        BLOCK_1_EFFECTIVE_WEIGHT_IDENTITY_VALID,

    "effective_weights_not_finite":
        BLOCK_1_ALL_EFFECTIVE_WEIGHTS_FINITE,

    "effective_weights_not_bounded":
        BLOCK_1_ALL_EFFECTIVE_WEIGHTS_BOUNDED,

    "deferred_weights_not_zero":
        BLOCK_1_ALL_DEFERRED_WEIGHTS_ZERO,

    "all_parameters_not_frozen":
        BLOCK_1_ALL_PARAMETERS_FROZEN,

    "all_modules_not_eval":
        BLOCK_1_ALL_MODULES_EVAL,

    "parameter_counts_invalid":
        BLOCK_1_PARAMETER_COUNTS_VALID,

    "per_module_parameter_counts_invalid":
        BLOCK_1_PER_MODULE_PARAMETER_COUNTS_VALID,

    "recurrent_contract_invalid":
        BLOCK_1_RECURRENT_CONTRACT_VALID,

    "pilot_training_boundary_invalid":
        BLOCK_1_PILOT_TRAINING_BOUNDARY_VALID,

    "notebook_09_completion_invalid":
        BLOCK_1_NOTEBOOK_09_COMPLETION_VALID,

    "post_device_representation_not_exact":
        BLOCK_1_POST_DEVICE_REPRESENTATION_EXACT,

    "post_device_transformative_not_exact":
        BLOCK_1_POST_DEVICE_TRANSFORMATIVE_EXACT,

    "post_device_weight_latent_not_exact":
        BLOCK_1_POST_DEVICE_WEIGHT_LATENT_EXACT,

    "unexpected_gradients":
        BLOCK_1_ALL_GRADIENTS_ABSENT,

    "execution_boundary_invalid":
        BLOCK_1_EXECUTION_BOUNDARY_VALID,
}


for (
    error_name,
    condition,
) in BLOCK_1_VALIDATION_CHECKS.items():

    if not condition:

        NOTEBOOK_10_BLOCK_1_ERRORS.append(
            error_name
        )


# =============================================================================
# Final Notebook 10 initialisation state
# =============================================================================

NOTEBOOK_09_TO_10_HANDOVER_VALID = (
    len(
        NOTEBOOK_10_BLOCK_1_ERRORS
    )
    ==
    0
)


NOTEBOOK_10_INITIALISED = (
    NOTEBOOK_09_TO_10_HANDOVER_VALID
)


NOTEBOOK_10_EXPANDED_CORPUS_INVENTORY_READY = (
    NOTEBOOK_10_INITIALISED
)


NOTEBOOK_10_BLOCK_1_VALID = (
    NOTEBOOK_10_INITIALISED
)


if not NOTEBOOK_10_BLOCK_1_VALID:

    raise RuntimeError(
        "Notebook 10 Block 1 Notebook 09 → 10 restoration failed. "
        f"Errors: {NOTEBOOK_10_BLOCK_1_ERRORS}"
    )


NOTEBOOK_10_BLOCK_1_COMPLETE = True


# =============================================================================
# Block summary
# =============================================================================

NOTEBOOK_10_BLOCK_1_SUMMARY = {
    "block":
        NOTEBOOK_10_BLOCK_1,

    "block_name":
        NOTEBOOK_10_BLOCK_1_NAME,

    "block_version":
        NOTEBOOK_10_BLOCK_1_VERSION,

    "checkpoint_filename":
        NOTEBOOK_09_TO_10_CHECKPOINT_FILENAME,

    "checkpoint_drive_file_id":
        NOTEBOOK_09_TO_10_CHECKPOINT_DRIVE_FILE_ID,

    "checkpoint_sha256":
        BLOCK_1_CHECKPOINT_SHA256,

    "metadata_filename":
        NOTEBOOK_09_TO_10_METADATA_FILENAME,

    "metadata_drive_file_id":
        NOTEBOOK_09_TO_10_METADATA_DRIVE_FILE_ID,

    "representation_parameters":
        NOTEBOOK_10_REPRESENTATION_PARAMETER_COUNT,

    "transformative_parameters":
        NOTEBOOK_10_TRANSFORMATIVE_PARAMETER_COUNT,

    "transformative_weight_parameters":
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT,

    "total_parameters":
        NOTEBOOK_10_TOTAL_PARAMETER_COUNT,

    "pathway_dimensions":
        deepcopy(
            NOTEBOOK_10_PATHWAY_DIMENSIONS
        ),

    "active_dimension_indices":
        deepcopy(
            NOTEBOOK_10_ACTIVE_DIMENSION_INDICES
        ),

    "deferred_dimension_indices":
        deepcopy(
            NOTEBOOK_10_DEFERRED_DIMENSION_INDICES
        ),

    "recurrent_contract":
        deepcopy(
            NOTEBOOK_10_RECURRENT_CONTRACT
        ),

    "pilot_pretraining_objective":
        NOTEBOOK_10_NOTEBOOK_09_PILOT_PRETRAIN_OBJECTIVE,

    "pilot_posttraining_objective":
        NOTEBOOK_10_NOTEBOOK_09_PILOT_POSTTRAIN_OBJECTIVE,

    "expanded_corpus_training_required":
        True,

    "handover_valid":
        NOTEBOOK_09_TO_10_HANDOVER_VALID,

    "expanded_corpus_inventory_ready":
        NOTEBOOK_10_EXPANDED_CORPUS_INVENTORY_READY,

    "block_valid":
        NOTEBOOK_10_BLOCK_1_VALID,

    "block_complete":
        NOTEBOOK_10_BLOCK_1_COMPLETE,
}


# =============================================================================
# Final report
# =============================================================================

print("=" * 72)

print(
    "Media AI — Notebook 10, Block 1: "
    "Notebook 09 → 10 Handover Restoration "
    "and Initialisation Audit"
)

print("=" * 72)

print(
    f"Notebook                     : "
    f"{NOTEBOOK_10}"
)

print(
    f"Notebook version             : "
    f"{NOTEBOOK_10_VERSION}"
)

print(
    f"Block version                : "
    f"{NOTEBOOK_10_BLOCK_1_VERSION}"
)

print(
    f"Project random seed          : "
    f"{NOTEBOOK_10_RANDOM_SEED}"
)

print(
    f"Execution device             : "
    f"{DEVICE}"
)

print("-" * 72)

print(
    "Google authentication"
)

print(
    f"Colab authenticated          : "
    f"{BLOCK_1_COLAB_AUTHENTICATED}"
)

print(
    f"Drive service initialised    : "
    f"{BLOCK_1_DRIVE_SERVICE_INITIALISED}"
)

print("-" * 72)

print(
    "Persistent Notebook 09 → 10 handover"
)

print(
    f"Checkpoint filename          : "
    f"{NOTEBOOK_09_TO_10_CHECKPOINT_FILENAME}"
)

print(
    f"Checkpoint Drive file ID     : "
    f"{NOTEBOOK_09_TO_10_CHECKPOINT_DRIVE_FILE_ID}"
)

print(
    f"Checkpoint located           : "
    f"{BLOCK_1_CHECKPOINT_LOCATED}"
)

print(
    f"Checkpoint downloaded        : "
    f"{BLOCK_1_CHECKPOINT_DOWNLOADED}"
)

print(
    f"Checkpoint SHA-256 valid     : "
    f"{BLOCK_1_CHECKPOINT_SHA256_VALID}"
)

print(
    f"Checkpoint byte count valid  : "
    f"{BLOCK_1_CHECKPOINT_METADATA_BYTE_COUNT_VALID}"
)

print(
    f"Checkpoint size              : "
    f"{len(BLOCK_1_CHECKPOINT_BYTES) / (1024 ** 2):.3f} MB"
)

print(
    f"Metadata filename            : "
    f"{NOTEBOOK_09_TO_10_METADATA_FILENAME}"
)

print(
    f"Metadata Drive file ID       : "
    f"{NOTEBOOK_09_TO_10_METADATA_DRIVE_FILE_ID}"
)

print(
    f"Metadata located             : "
    f"{BLOCK_1_METADATA_LOCATED}"
)

print(
    f"Metadata downloaded          : "
    f"{BLOCK_1_METADATA_DOWNLOADED}"
)

print(
    f"Metadata JSON parsed         : "
    f"{BLOCK_1_METADATA_JSON_PARSED}"
)

print(
    f"Metadata size                : "
    f"{len(BLOCK_1_METADATA_BYTES) / 1024:.3f} KB"
)

print("-" * 72)

print(
    "Restored architecture"
)

print(
    f"Representation parameters    : "
    f"{NOTEBOOK_10_REPRESENTATION_PARAMETER_COUNT:,}"
)

print(
    f"Transformative parameters    : "
    f"{NOTEBOOK_10_TRANSFORMATIVE_PARAMETER_COUNT:,}"
)

print(
    f"Transformative Weight params : "
    f"{NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT}"
)

print(
    f"Combined restored params     : "
    f"{NOTEBOOK_10_TOTAL_PARAMETER_COUNT:,}"
)

print(
    f"Structural weight dimensions : "
    f"{BLOCK_1_EXPECTED_STRUCTURAL_WEIGHT_DIMENSIONS}"
)

print(
    f"Active weight dimensions     : "
    f"{BLOCK_1_EXPECTED_ACTIVE_WEIGHT_DIMENSIONS}"
)

print(
    f"Deferred weight dimensions   : "
    f"{BLOCK_1_EXPECTED_DEFERRED_WEIGHT_DIMENSIONS}"
)

print("-" * 72)

print(
    "Pathway contracts"
)

for pathway_name in BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS:

    print(
        f"{pathway_name:<14} dimension          : "
        f"{NOTEBOOK_10_PATHWAY_DIMENSIONS[pathway_name]}"
    )

    print(
        f"{pathway_name:<14} active dimensions  : "
        f"{list(NOTEBOOK_10_ACTIVE_DIMENSION_INDICES[pathway_name])}"
    )

    print(
        f"{pathway_name:<14} deferred dimensions: "
        f"{list(NOTEBOOK_10_DEFERRED_DIMENSION_INDICES[pathway_name])}"
    )

    print(
        f"{pathway_name:<14} latent source key  : "
        f"{BLOCK_1_LATENT_SOURCE_KEYS[pathway_name]}"
    )

    print(
        f"{pathway_name:<14} weight finite      : "
        f"{BLOCK_1_EFFECTIVE_WEIGHTS_FINITE[pathway_name]}"
    )

    print(
        f"{pathway_name:<14} weight bounded     : "
        f"{BLOCK_1_EFFECTIVE_WEIGHTS_BOUNDED[pathway_name]}"
    )

    print(
        f"{pathway_name:<14} deferred zero      : "
        f"{BLOCK_1_DEFERRED_WEIGHTS_ZERO[pathway_name]}"
    )

print("-" * 72)

print(
    "Recurrent contract"
)

print(
    f"Update type                  : "
    f"{NOTEBOOK_10_RECURRENT_CONTRACT['update_type']}"
)

print(
    f"Initial-state policy         : "
    f"{NOTEBOOK_10_RECURRENT_CONTRACT['initial_state_policy']}"
)

print(
    f"Article-boundary reset       : "
    f"{NOTEBOOK_10_RECURRENT_CONTRACT['article_boundary_reset']}"
)

print(
    f"Cross-article propagation    : "
    f"{NOTEBOOK_10_RECURRENT_CONTRACT['cross_article_propagation']}"
)

print(
    f"Cross-pathway recurrence     : "
    f"{NOTEBOOK_10_RECURRENT_CONTRACT['cross_pathway_recurrence']}"
)

print(
    f"Future context allowed       : "
    f"{NOTEBOOK_10_RECURRENT_CONTRACT['future_context_allowed']}"
)

print("-" * 72)

print(
    "Restoration validation"
)

print(
    f"Metadata identity valid      : "
    f"{BLOCK_1_METADATA_IDENTITY_VALID}"
)

print(
    f"Checkpoint identity valid    : "
    f"{BLOCK_1_CHECKPOINT_IDENTITY_VALID}"
)

print(
    f"Bootstrap contract valid     : "
    f"{BLOCK_1_METADATA_BOOTSTRAP_VALID}"
)

print(
    f"Cross-artefact valid         : "
    f"{BLOCK_1_CROSS_ARTEFACT_VALID}"
)

print(
    f"Parameter counts valid       : "
    f"{BLOCK_1_PARAMETER_COUNTS_VALID}"
)

print(
    f"Pathway contract valid       : "
    f"{BLOCK_1_ALL_DIMENSION_PARTITIONS_VALID}"
)

print(
    f"Representation state exact   : "
    f"{BLOCK_1_REPRESENTATION_STATE_EXACT}"
)

print(
    f"Transformative state exact   : "
    f"{BLOCK_1_TRANSFORMATIVE_STATE_EXACT}"
)

print(
    f"Weight state exact           : "
    f"{BLOCK_1_EFFECTIVE_WEIGHT_IDENTITY_VALID}"
)

print(
    f"Post-device repr. exact      : "
    f"{BLOCK_1_POST_DEVICE_REPRESENTATION_EXACT}"
)

print(
    f"Post-device transform exact  : "
    f"{BLOCK_1_POST_DEVICE_TRANSFORMATIVE_EXACT}"
)

print(
    f"Post-device weight exact     : "
    f"{BLOCK_1_POST_DEVICE_WEIGHT_LATENT_EXACT}"
)

print(
    f"All parameters frozen        : "
    f"{BLOCK_1_ALL_PARAMETERS_FROZEN}"
)

print(
    f"All modules eval             : "
    f"{BLOCK_1_ALL_MODULES_EVAL}"
)

print(
    f"All gradients absent         : "
    f"{BLOCK_1_ALL_GRADIENTS_ABSENT}"
)

print("-" * 72)

print(
    "Inherited pilot boundary"
)

print(
    f"Pilot pretraining objective  : "
    f"{NOTEBOOK_10_NOTEBOOK_09_PILOT_PRETRAIN_OBJECTIVE:.8f}"
)

print(
    f"Pilot posttraining objective : "
    f"{NOTEBOOK_10_NOTEBOOK_09_PILOT_POSTTRAIN_OBJECTIVE:.8f}"
)

print(
    f"Final training claimed       : False"
)

print(
    f"Expanded training required   : True"
)

print("-" * 72)

print(
    "Notebook 10 execution boundary"
)

print(
    f"Expanded corpus loaded       : "
    f"{NOTEBOOK_10_BLOCK_1_EXPANDED_CORPUS_LOADED}"
)

print(
    f"Representation forward       : "
    f"{NOTEBOOK_10_BLOCK_1_REPRESENTATION_FORWARD_EXECUTED}"
)

print(
    f"Transform forward            : "
    f"{NOTEBOOK_10_BLOCK_1_TRANSFORMATIVE_FORWARD_EXECUTED}"
)

print(
    f"Weight forward               : "
    f"{NOTEBOOK_10_BLOCK_1_WEIGHT_FORWARD_EXECUTED}"
)

print(
    f"Recurrent state created      : "
    f"{NOTEBOOK_10_BLOCK_1_RECURRENT_STATE_CREATED}"
)

print(
    f"Recurrent update executed    : "
    f"{NOTEBOOK_10_BLOCK_1_RECURRENT_UPDATE_EXECUTED}"
)

print(
    f"Loss calculated              : "
    f"{NOTEBOOK_10_BLOCK_1_LOSS_CALCULATED}"
)

print(
    f"Optimizer created            : "
    f"{NOTEBOOK_10_BLOCK_1_OPTIMIZER_CREATED}"
)

print(
    f"Backward pass executed       : "
    f"{NOTEBOOK_10_BLOCK_1_BACKWARD_PASS_EXECUTED}"
)

print(
    f"Parameter update executed    : "
    f"{NOTEBOOK_10_BLOCK_1_PARAMETER_UPDATE_EXECUTED}"
)

print(
    f"Training executed            : "
    f"{NOTEBOOK_10_BLOCK_1_TRAINING_EXECUTED}"
)

print("-" * 72)

print(
    f"Notebook 09 → 10 valid       : "
    f"{NOTEBOOK_09_TO_10_HANDOVER_VALID}"
)

print(
    f"Notebook 10 initialised      : "
    f"{NOTEBOOK_10_INITIALISED}"
)

print(
    f"Expanded inventory ready     : "
    f"{NOTEBOOK_10_EXPANDED_CORPUS_INVENTORY_READY}"
)

print(
    f"Block valid                  : "
    f"{NOTEBOOK_10_BLOCK_1_VALID}"
)

print(
    f"Block complete               : "
    f"{NOTEBOOK_10_BLOCK_1_COMPLETE}"
)

print("=" * 72)

print(
    "Notebook 10 initialised successfully from the persistent "
    "Notebook 09 → Notebook 10 handover."
)

print(
    "The canonical Notebook 09 checkpoint was retrieved directly by "
    "Drive file ID and its SHA-256 digest was validated before "
    "deserialisation."
)

print(
    f"The validated {NOTEBOOK_10_REPRESENTATION_PARAMETER_COUNT:,}-parameter "
    "representation architecture, "
    f"{NOTEBOOK_10_TRANSFORMATIVE_PARAMETER_COUNT:,}-parameter transformative "
    f"architecture and {NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT} "
    "learned Transformative-Weight parameter(s) were restored successfully."
)

print(
    f"The complete inherited {NOTEBOOK_10_TOTAL_PARAMETER_COUNT:,}-parameter "
    "architecture is frozen and in evaluation mode at the Notebook 10 "
    "initialisation boundary."
)

print(
    "Factual, psychological and social pathway dimensions and their "
    "active/deferred Transformative-Weight partitions reproduce the "
    "canonical Notebook 09 contract."
)

print(
    "The learned effective Transformative Weights reproduce the "
    "persisted Notebook 09 state, remain finite and bounded, and preserve "
    "all deferred dimensions at exactly zero."
)

print(
    "The additive recurrent contract, deterministic zero article reset, "
    "cross-article isolation, pathway independence and no-future-context "
    "policy were inherited without modification."
)

print(
    "The Notebook 09 controlled optimisation remains explicitly classified "
    "as pilot architectural training rather than final model training."
)

print(
    "No expanded corpus, model forward path, recurrent state, loss, "
    "optimiser, backward pass, parameter update or training was executed "
    "by Block 1."
)

print(
    "Notebook 10 may now proceed to expanded-corpus inventory, "
    "eligibility and multi-article validation-source construction."
)

print("=" * 72)
```

```python
# =============================================================================
# Media AI — Notebook 10
# Block 1: Notebook 09 → 10 Handover Restoration
#          and Initialisation Audit
# =============================================================================

from copy import deepcopy
from datetime import datetime, timezone

import hashlib
import io
import json

import numpy as np
import torch
import torch.nn as nn

from google.colab import auth
import google.auth

from googleapiclient.discovery import build
from googleapiclient.http import MediaIoBaseDownload


# =============================================================================
# Block identity
# =============================================================================

NOTEBOOK_10 = (
    "10_expanded_validation_and_project_completion"
)

NOTEBOOK_10_VERSION = "1.0"

NOTEBOOK_10_BLOCK_1 = 1

NOTEBOOK_10_BLOCK_1_NAME = (
    "Notebook 09 → 10 Handover Restoration "
    "and Initialisation Audit"
)

NOTEBOOK_10_BLOCK_1_VERSION = "1.2"

BLOCK_1_EXECUTED_AT_UTC = (
    datetime.now(
        timezone.utc
    ).isoformat()
)


# =============================================================================
# Project reproducibility
# =============================================================================

NOTEBOOK_10_RANDOM_SEED = 42


np.random.seed(
    NOTEBOOK_10_RANDOM_SEED
)


torch.manual_seed(
    NOTEBOOK_10_RANDOM_SEED
)


if torch.cuda.is_available():

    torch.cuda.manual_seed_all(
        NOTEBOOK_10_RANDOM_SEED
    )


# =============================================================================
# Controlled execution device
# =============================================================================
#
# Notebook 09 completed on CPU and its final persisted states are CPU-safe.
# Block 1 therefore restores on CPU first and preserves CPU execution for the
# initial audit. Later blocks may explicitly change the execution device.
# =============================================================================

DEVICE = torch.device(
    "cpu"
)

DEFAULT_DTYPE = torch.float32


# =============================================================================
# Canonical Notebook 09 → 10 persistent identifiers
# =============================================================================
#
# These IDs were created and validated by Notebook 09 Block 13.
#
# No filename discovery is used.
# =============================================================================

NOTEBOOK_09_TO_10_CHECKPOINT_FILENAME = (
    "notebook_09_final_checkpoint.pt"
)

NOTEBOOK_09_TO_10_METADATA_FILENAME = (
    "notebook_09_final_metadata.json"
)


NOTEBOOK_09_TO_10_CHECKPOINT_DRIVE_FILE_ID = (
    "1h2uRHtNfAFPlKNMZqwiZvGROewn0b2-R"
)

NOTEBOOK_09_TO_10_METADATA_DRIVE_FILE_ID = (
    "1V0eeY-QdeD9htWWYbESU42qbak5hV8Ms"
)


# =============================================================================
# Expected inherited architecture
# =============================================================================
#
# These values reproduce the validated Notebook 09 Block 13 completion
# contract. Checkpoint and metadata must additionally agree exactly.
# =============================================================================

BLOCK_1_EXPECTED_REPRESENTATION_PARAMETERS = 894_003

BLOCK_1_EXPECTED_TRANSFORMATIVE_PARAMETERS = 8_187

BLOCK_1_EXPECTED_TRANSFORMATIVE_WEIGHT_PARAMETERS = 49

BLOCK_1_EXPECTED_TOTAL_PARAMETERS = 902_239


BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS = {
    "factual":
        10,

    "psychological":
        34,

    "social":
        7,
}


BLOCK_1_EXPECTED_ACTIVE_DIMENSIONS = {
    "factual":
        tuple(
            range(
                10
            )
        ),

    "psychological":
        tuple(
            range(
                34
            )
        ),

    "social":
        (
            0,
            2,
            4,
            5,
            6,
        ),
}


BLOCK_1_EXPECTED_DEFERRED_DIMENSIONS = {
    "factual":
        tuple(),

    "psychological":
        tuple(),

    "social":
        (
            1,
            3,
        ),
}


BLOCK_1_EXPECTED_STRUCTURAL_WEIGHT_DIMENSIONS = int(
    sum(
        BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS.values()
    )
)


BLOCK_1_EXPECTED_ACTIVE_WEIGHT_DIMENSIONS = int(
    sum(
        len(
            BLOCK_1_EXPECTED_ACTIVE_DIMENSIONS[
                pathway_name
            ]
        )

        for pathway_name
        in BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS
    )
)


BLOCK_1_EXPECTED_DEFERRED_WEIGHT_DIMENSIONS = int(
    sum(
        len(
            BLOCK_1_EXPECTED_DEFERRED_DIMENSIONS[
                pathway_name
            ]
        )

        for pathway_name
        in BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS
    )
)


BLOCK_1_EXPECTED_WEIGHT_PARAMETER_COUNTS = {
    pathway_name:
        len(
            BLOCK_1_EXPECTED_ACTIVE_DIMENSIONS[
                pathway_name
            ]
        )

    for pathway_name
    in BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS
}


if not all(
    [
        (
            BLOCK_1_EXPECTED_STRUCTURAL_WEIGHT_DIMENSIONS
            ==
            51
        ),

        (
            BLOCK_1_EXPECTED_ACTIVE_WEIGHT_DIMENSIONS
            ==
            BLOCK_1_EXPECTED_TRANSFORMATIVE_WEIGHT_PARAMETERS
        ),

        (
            BLOCK_1_EXPECTED_ACTIVE_WEIGHT_DIMENSIONS
            +
            BLOCK_1_EXPECTED_DEFERRED_WEIGHT_DIMENSIONS
            ==
            BLOCK_1_EXPECTED_STRUCTURAL_WEIGHT_DIMENSIONS
        ),

        (
            BLOCK_1_EXPECTED_TOTAL_PARAMETERS
            ==
            BLOCK_1_EXPECTED_REPRESENTATION_PARAMETERS
            +
            BLOCK_1_EXPECTED_TRANSFORMATIVE_PARAMETERS
            +
            BLOCK_1_EXPECTED_TRANSFORMATIVE_WEIGHT_PARAMETERS
        ),
    ]
):

    raise RuntimeError(
        "Notebook 10 Block 1 expected Notebook 09 architecture "
        "contract is internally inconsistent."
    )


# =============================================================================
# Execution boundary
# =============================================================================

NOTEBOOK_10_BLOCK_1_EXPANDED_CORPUS_LOADED = False

NOTEBOOK_10_BLOCK_1_REPRESENTATION_FORWARD_EXECUTED = False

NOTEBOOK_10_BLOCK_1_TRANSFORMATIVE_FORWARD_EXECUTED = False

NOTEBOOK_10_BLOCK_1_WEIGHT_FORWARD_EXECUTED = False

NOTEBOOK_10_BLOCK_1_RECURRENT_STATE_CREATED = False

NOTEBOOK_10_BLOCK_1_RECURRENT_UPDATE_EXECUTED = False

NOTEBOOK_10_BLOCK_1_LOSS_CALCULATED = False

NOTEBOOK_10_BLOCK_1_OPTIMIZER_CREATED = False

NOTEBOOK_10_BLOCK_1_ZERO_GRAD_EXECUTED = False

NOTEBOOK_10_BLOCK_1_BACKWARD_PASS_EXECUTED = False

NOTEBOOK_10_BLOCK_1_PARAMETER_UPDATE_EXECUTED = False

NOTEBOOK_10_BLOCK_1_TRAINING_EXECUTED = False


# =============================================================================
# Google authentication
# =============================================================================
#
# This is the same narrow Colab authentication pattern already proven in the
# preceding Media AI notebooks:
#
#   auth.authenticate_user()
#   google.auth.default()
#   build(... credentials=...)
#
# No Drive filesystem mount is used.
# No metadata-service fallback is intentionally introduced.
# =============================================================================

auth.authenticate_user()


BLOCK_1_CREDENTIALS, _ = (
    google.auth.default()
)


NOTEBOOK_10_DRIVE_SERVICE = build(
    "drive",
    "v3",
    credentials=
        BLOCK_1_CREDENTIALS,
    cache_discovery=
        False,
)


BLOCK_1_COLAB_AUTHENTICATED = True


BLOCK_1_DRIVE_SERVICE_INITIALISED = (
    NOTEBOOK_10_DRIVE_SERVICE
    is not None
)


if not BLOCK_1_DRIVE_SERVICE_INITIALISED:

    raise RuntimeError(
        "Notebook 10 could not initialise the Google Drive API service."
    )


# =============================================================================
# Drive helpers
# =============================================================================

def block_1_get_drive_metadata(
    drive_service,
    file_id,
):

    return (
        drive_service.files()
        .get(
            fileId=
                file_id,

            fields=(
                "id,"
                "name,"
                "size,"
                "modifiedTime,"
                "mimeType,"
                "parents"
            ),
        )
        .execute()
    )


def block_1_download_drive_bytes(
    drive_service,
    file_id,
):

    request = (
        drive_service.files()
        .get_media(
            fileId=
                file_id
        )
    )


    buffer = io.BytesIO()


    downloader = MediaIoBaseDownload(
        buffer,
        request,
    )


    complete = False


    while not complete:

        _, complete = downloader.next_chunk()


    buffer.seek(
        0
    )


    return buffer.read()


def block_1_sha256(
    payload,
):

    return hashlib.sha256(
        payload
    ).hexdigest()


# =============================================================================
# Locate the two canonical artefacts by direct ID
# =============================================================================

BLOCK_1_CHECKPOINT_DRIVE_METADATA = (
    block_1_get_drive_metadata(
        NOTEBOOK_10_DRIVE_SERVICE,
        NOTEBOOK_09_TO_10_CHECKPOINT_DRIVE_FILE_ID,
    )
)


BLOCK_1_METADATA_DRIVE_METADATA = (
    block_1_get_drive_metadata(
        NOTEBOOK_10_DRIVE_SERVICE,
        NOTEBOOK_09_TO_10_METADATA_DRIVE_FILE_ID,
    )
)


BLOCK_1_CHECKPOINT_LOCATED = (
    BLOCK_1_CHECKPOINT_DRIVE_METADATA.get(
        "id"
    )
    ==
    NOTEBOOK_09_TO_10_CHECKPOINT_DRIVE_FILE_ID
)


BLOCK_1_METADATA_LOCATED = (
    BLOCK_1_METADATA_DRIVE_METADATA.get(
        "id"
    )
    ==
    NOTEBOOK_09_TO_10_METADATA_DRIVE_FILE_ID
)


BLOCK_1_CHECKPOINT_FILENAME_VALID = (
    BLOCK_1_CHECKPOINT_DRIVE_METADATA.get(
        "name"
    )
    ==
    NOTEBOOK_09_TO_10_CHECKPOINT_FILENAME
)


BLOCK_1_METADATA_FILENAME_VALID = (
    BLOCK_1_METADATA_DRIVE_METADATA.get(
        "name"
    )
    ==
    NOTEBOOK_09_TO_10_METADATA_FILENAME
)


if not all(
    [
        BLOCK_1_CHECKPOINT_LOCATED,
        BLOCK_1_METADATA_LOCATED,
        BLOCK_1_CHECKPOINT_FILENAME_VALID,
        BLOCK_1_METADATA_FILENAME_VALID,
    ]
):

    raise RuntimeError(
        "Notebook 09 → 10 direct persistent artefact identity "
        "validation failed."
    )


# =============================================================================
# Download metadata first
# =============================================================================
#
# The metadata contains the authoritative checkpoint SHA-256 digest and byte
# count. It is therefore parsed before the checkpoint is deserialised.
# =============================================================================

BLOCK_1_METADATA_BYTES = (
    block_1_download_drive_bytes(
        NOTEBOOK_10_DRIVE_SERVICE,
        NOTEBOOK_09_TO_10_METADATA_DRIVE_FILE_ID,
    )
)


if not BLOCK_1_METADATA_BYTES:

    raise RuntimeError(
        "Notebook 09 final metadata artefact is empty."
    )


BLOCK_1_METADATA_DOWNLOADED = True


BLOCK_1_METADATA_DRIVE_BYTE_COUNT_VALID = (
    int(
        BLOCK_1_METADATA_DRIVE_METADATA.get(
            "size",
            -1,
        )
    )
    ==
    len(
        BLOCK_1_METADATA_BYTES
    )
)


if not BLOCK_1_METADATA_DRIVE_BYTE_COUNT_VALID:

    raise RuntimeError(
        "Notebook 09 metadata download byte count does not match "
        "Google Drive metadata."
    )


# =============================================================================
# Parse metadata
# =============================================================================

NOTEBOOK_10_INHERITED_METADATA = json.loads(
    BLOCK_1_METADATA_BYTES.decode(
        "utf-8"
    )
)


BLOCK_1_METADATA_JSON_PARSED = isinstance(
    NOTEBOOK_10_INHERITED_METADATA,
    dict,
)


if not BLOCK_1_METADATA_JSON_PARSED:

    raise TypeError(
        "Notebook 09 final metadata must parse to a JSON object."
    )


# =============================================================================
# Metadata identity contract
# =============================================================================

BLOCK_1_METADATA_IDENTITY_VALID = all(
    [
        (
            NOTEBOOK_10_INHERITED_METADATA.get(
                "artifact_type"
            )
            ==
            "media_ai_notebook_09_final_metadata"
        ),

        (
            NOTEBOOK_10_INHERITED_METADATA.get(
                "metadata_schema_version"
            )
            ==
            "1.0"
        ),

        (
            NOTEBOOK_10_INHERITED_METADATA.get(
                "notebook_number"
            )
            ==
            9
        ),

        (
            NOTEBOOK_10_INHERITED_METADATA.get(
                "completion_block"
            )
            ==
            13
        ),
    ]
)


if not BLOCK_1_METADATA_IDENTITY_VALID:

    raise RuntimeError(
        "Notebook 09 final metadata identity contract is invalid."
    )


# =============================================================================
# Metadata bootstrap contract
# =============================================================================

BLOCK_1_METADATA_BOOTSTRAP = (
    NOTEBOOK_10_INHERITED_METADATA.get(
        "notebook_10_bootstrap",
        {},
    )
)


BLOCK_1_METADATA_BOOTSTRAP_VALID = all(
    [
        (
            BLOCK_1_METADATA_BOOTSTRAP.get(
                "source_notebook"
            )
            ==
            9
        ),

        (
            BLOCK_1_METADATA_BOOTSTRAP.get(
                "checkpoint_filename"
            )
            ==
            NOTEBOOK_09_TO_10_CHECKPOINT_FILENAME
        ),

        (
            BLOCK_1_METADATA_BOOTSTRAP.get(
                "checkpoint_drive_file_id"
            )
            ==
            NOTEBOOK_09_TO_10_CHECKPOINT_DRIVE_FILE_ID
        ),

        (
            BLOCK_1_METADATA_BOOTSTRAP.get(
                "metadata_filename"
            )
            ==
            NOTEBOOK_09_TO_10_METADATA_FILENAME
        ),

        (
            BLOCK_1_METADATA_BOOTSTRAP.get(
                "metadata_drive_file_id"
            )
            ==
            NOTEBOOK_09_TO_10_METADATA_DRIVE_FILE_ID
        ),

        (
            BLOCK_1_METADATA_BOOTSTRAP.get(
                "architecture_complete"
            )
            is True
        ),

        (
            BLOCK_1_METADATA_BOOTSTRAP.get(
                "expanded_corpus_training_required"
            )
            is True
        ),

        (
            BLOCK_1_METADATA_BOOTSTRAP.get(
                "reconstruct_upstream_notebooks"
            )
            is False
        ),
    ]
)


if not BLOCK_1_METADATA_BOOTSTRAP_VALID:

    raise RuntimeError(
        "Notebook 09 final metadata does not reproduce the "
        "Notebook 10 bootstrap contract."
    )


# =============================================================================
# Canonical checkpoint identity from metadata
# =============================================================================

BLOCK_1_METADATA_CHECKPOINT_CONTRACT = (
    NOTEBOOK_10_INHERITED_METADATA.get(
        "checkpoint",
        {},
    )
)


BLOCK_1_EXPECTED_CHECKPOINT_SHA256 = (
    BLOCK_1_METADATA_CHECKPOINT_CONTRACT.get(
        "sha256"
    )
)


BLOCK_1_EXPECTED_CHECKPOINT_BYTE_COUNT = (
    BLOCK_1_METADATA_CHECKPOINT_CONTRACT.get(
        "byte_count"
    )
)


BLOCK_1_METADATA_CHECKPOINT_IDENTITY_VALID = all(
    [
        (
            BLOCK_1_METADATA_CHECKPOINT_CONTRACT.get(
                "filename"
            )
            ==
            NOTEBOOK_09_TO_10_CHECKPOINT_FILENAME
        ),

        (
            BLOCK_1_METADATA_CHECKPOINT_CONTRACT.get(
                "drive_file_id"
            )
            ==
            NOTEBOOK_09_TO_10_CHECKPOINT_DRIVE_FILE_ID
        ),

        isinstance(
            BLOCK_1_EXPECTED_CHECKPOINT_SHA256,
            str,
        ),

        (
            len(
                BLOCK_1_EXPECTED_CHECKPOINT_SHA256
            )
            ==
            64
        ),

        isinstance(
            BLOCK_1_EXPECTED_CHECKPOINT_BYTE_COUNT,
            int,
        ),

        (
            BLOCK_1_EXPECTED_CHECKPOINT_BYTE_COUNT
            >
            0
        ),
    ]
)


if not BLOCK_1_METADATA_CHECKPOINT_IDENTITY_VALID:

    raise RuntimeError(
        "Notebook 09 metadata does not contain a valid checkpoint "
        "identity and integrity contract."
    )


# =============================================================================
# Download checkpoint
# =============================================================================

BLOCK_1_CHECKPOINT_BYTES = (
    block_1_download_drive_bytes(
        NOTEBOOK_10_DRIVE_SERVICE,
        NOTEBOOK_09_TO_10_CHECKPOINT_DRIVE_FILE_ID,
    )
)


if not BLOCK_1_CHECKPOINT_BYTES:

    raise RuntimeError(
        "Notebook 09 final checkpoint artefact is empty."
    )


BLOCK_1_CHECKPOINT_DOWNLOADED = True


# =============================================================================
# Validate checkpoint before deserialisation
# =============================================================================

BLOCK_1_CHECKPOINT_SHA256 = (
    block_1_sha256(
        BLOCK_1_CHECKPOINT_BYTES
    )
)


BLOCK_1_CHECKPOINT_SHA256_VALID = (
    BLOCK_1_CHECKPOINT_SHA256
    ==
    BLOCK_1_EXPECTED_CHECKPOINT_SHA256
)


BLOCK_1_CHECKPOINT_METADATA_BYTE_COUNT_VALID = (
    len(
        BLOCK_1_CHECKPOINT_BYTES
    )
    ==
    BLOCK_1_EXPECTED_CHECKPOINT_BYTE_COUNT
)


BLOCK_1_CHECKPOINT_DRIVE_BYTE_COUNT_VALID = (
    int(
        BLOCK_1_CHECKPOINT_DRIVE_METADATA.get(
            "size",
            -1,
        )
    )
    ==
    len(
        BLOCK_1_CHECKPOINT_BYTES
    )
)


if not all(
    [
        BLOCK_1_CHECKPOINT_SHA256_VALID,
        BLOCK_1_CHECKPOINT_METADATA_BYTE_COUNT_VALID,
        BLOCK_1_CHECKPOINT_DRIVE_BYTE_COUNT_VALID,
    ]
):

    raise RuntimeError(
        "Notebook 09 final checkpoint failed pre-deserialisation "
        "integrity validation."
    )


# =============================================================================
# Deserialize checkpoint only after hash validation
# =============================================================================

NOTEBOOK_10_INHERITED_CHECKPOINT = torch.load(
    io.BytesIO(
        BLOCK_1_CHECKPOINT_BYTES
    ),
    map_location=
        "cpu",
    weights_only=
        False,
)


BLOCK_1_CHECKPOINT_PARSED = isinstance(
    NOTEBOOK_10_INHERITED_CHECKPOINT,
    dict,
)


if not BLOCK_1_CHECKPOINT_PARSED:

    raise TypeError(
        "Notebook 09 final checkpoint must deserialize to a dictionary."
    )


# =============================================================================
# Checkpoint identity
# =============================================================================

BLOCK_1_CHECKPOINT_IDENTITY_VALID = all(
    [
        (
            NOTEBOOK_10_INHERITED_CHECKPOINT.get(
                "artifact_type"
            )
            ==
            "media_ai_notebook_09_final_checkpoint"
        ),

        (
            NOTEBOOK_10_INHERITED_CHECKPOINT.get(
                "checkpoint_schema_version"
            )
            ==
            "1.0"
        ),

        (
            NOTEBOOK_10_INHERITED_CHECKPOINT.get(
                "notebook_number"
            )
            ==
            9
        ),

        (
            NOTEBOOK_10_INHERITED_CHECKPOINT.get(
                "notebook_block"
            )
            ==
            13
        ),
    ]
)


if not BLOCK_1_CHECKPOINT_IDENTITY_VALID:

    raise RuntimeError(
        "Notebook 09 final checkpoint identity contract is invalid."
    )


# =============================================================================
# Checkpoint top-level state contract
# =============================================================================

BLOCK_1_REQUIRED_CHECKPOINT_KEYS = {
    "artifact_type",
    "checkpoint_schema_version",
    "architecture",
    "recurrent_contract",
    "representation_state",
    "transformative_state",
    "transformative_weight_state",
    "effective_transformative_weights",
    "pilot_training",
    "completion",
}


BLOCK_1_CHECKPOINT_REQUIRED_KEYS_VALID = (
    BLOCK_1_REQUIRED_CHECKPOINT_KEYS
    .issubset(
        set(
            NOTEBOOK_10_INHERITED_CHECKPOINT.keys()
        )
    )
)


if not BLOCK_1_CHECKPOINT_REQUIRED_KEYS_VALID:

    missing_keys = (
        BLOCK_1_REQUIRED_CHECKPOINT_KEYS
        -
        set(
            NOTEBOOK_10_INHERITED_CHECKPOINT.keys()
        )
    )


    raise RuntimeError(
        "Notebook 09 final checkpoint is missing required state. "
        f"Missing keys: {sorted(missing_keys)}"
    )


# =============================================================================
# Architecture contract from checkpoint
# =============================================================================

BLOCK_1_CHECKPOINT_ARCHITECTURE = (
    NOTEBOOK_10_INHERITED_CHECKPOINT[
        "architecture"
    ]
)


BLOCK_1_CHECKPOINT_PATHWAY_DIMENSIONS = {
    pathway_name:
        int(
            dimension
        )

    for (
        pathway_name,
        dimension,
    ) in BLOCK_1_CHECKPOINT_ARCHITECTURE[
        "pathway_dimensions"
    ].items()
}


BLOCK_1_CHECKPOINT_ACTIVE_DIMENSIONS = {
    pathway_name:
        tuple(
            int(
                index
            )

            for index
            in indices
        )

    for (
        pathway_name,
        indices,
    ) in BLOCK_1_CHECKPOINT_ARCHITECTURE[
        "active_dimension_indices"
    ].items()
}


BLOCK_1_CHECKPOINT_DEFERRED_DIMENSIONS = {
    pathway_name:
        tuple(
            int(
                index
            )

            for index
            in indices
        )

    for (
        pathway_name,
        indices,
    ) in BLOCK_1_CHECKPOINT_ARCHITECTURE[
        "deferred_dimension_indices"
    ].items()
}


BLOCK_1_PATHWAY_DIMENSIONS_VALID = (
    BLOCK_1_CHECKPOINT_PATHWAY_DIMENSIONS
    ==
    BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS
)


BLOCK_1_ACTIVE_DIMENSIONS_VALID = (
    BLOCK_1_CHECKPOINT_ACTIVE_DIMENSIONS
    ==
    BLOCK_1_EXPECTED_ACTIVE_DIMENSIONS
)


BLOCK_1_DEFERRED_DIMENSIONS_VALID = (
    BLOCK_1_CHECKPOINT_DEFERRED_DIMENSIONS
    ==
    BLOCK_1_EXPECTED_DEFERRED_DIMENSIONS
)


# =============================================================================
# Active / deferred partition audit
# =============================================================================

BLOCK_1_DIMENSION_PARTITIONS_VALID = {}


for (
    pathway_name,
    pathway_dim,
) in BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS.items():

    active_set = set(
        BLOCK_1_CHECKPOINT_ACTIVE_DIMENSIONS[
            pathway_name
        ]
    )


    deferred_set = set(
        BLOCK_1_CHECKPOINT_DEFERRED_DIMENSIONS[
            pathway_name
        ]
    )


    BLOCK_1_DIMENSION_PARTITIONS_VALID[
        pathway_name
    ] = all(
        [
            active_set.isdisjoint(
                deferred_set
            ),

            (
                active_set
                |
                deferred_set
            )
            ==
            set(
                range(
                    pathway_dim
                )
            ),
        ]
    )


BLOCK_1_ALL_DIMENSION_PARTITIONS_VALID = all(
    BLOCK_1_DIMENSION_PARTITIONS_VALID.values()
)


if not all(
    [
        BLOCK_1_PATHWAY_DIMENSIONS_VALID,
        BLOCK_1_ACTIVE_DIMENSIONS_VALID,
        BLOCK_1_DEFERRED_DIMENSIONS_VALID,
        BLOCK_1_ALL_DIMENSION_PARTITIONS_VALID,
    ]
):

    raise RuntimeError(
        "Notebook 09 pathway dimensions or active/deferred "
        "Transformative-Weight partitions are invalid."
    )


# =============================================================================
# Parameter-count contract
# =============================================================================

BLOCK_1_CHECKPOINT_PARAMETER_COUNTS_VALID = all(
    [
        (
            BLOCK_1_CHECKPOINT_ARCHITECTURE.get(
                "representation_parameter_count"
            )
            ==
            BLOCK_1_EXPECTED_REPRESENTATION_PARAMETERS
        ),

        (
            BLOCK_1_CHECKPOINT_ARCHITECTURE.get(
                "transformative_parameter_count"
            )
            ==
            BLOCK_1_EXPECTED_TRANSFORMATIVE_PARAMETERS
        ),

        (
            BLOCK_1_CHECKPOINT_ARCHITECTURE.get(
                "transformative_weight_parameter_count"
            )
            ==
            BLOCK_1_EXPECTED_TRANSFORMATIVE_WEIGHT_PARAMETERS
        ),

        (
            BLOCK_1_CHECKPOINT_ARCHITECTURE.get(
                "total_parameter_count"
            )
            ==
            BLOCK_1_EXPECTED_TOTAL_PARAMETERS
        ),

        (
            BLOCK_1_CHECKPOINT_ARCHITECTURE.get(
                "structural_weight_dimensions"
            )
            ==
            BLOCK_1_EXPECTED_STRUCTURAL_WEIGHT_DIMENSIONS
        ),

        (
            BLOCK_1_CHECKPOINT_ARCHITECTURE.get(
                "active_weight_dimensions"
            )
            ==
            BLOCK_1_EXPECTED_ACTIVE_WEIGHT_DIMENSIONS
        ),

        (
            BLOCK_1_CHECKPOINT_ARCHITECTURE.get(
                "deferred_weight_dimensions"
            )
            ==
            BLOCK_1_EXPECTED_DEFERRED_WEIGHT_DIMENSIONS
        ),
    ]
)


if not BLOCK_1_CHECKPOINT_PARAMETER_COUNTS_VALID:

    raise RuntimeError(
        "Notebook 09 checkpoint parameter-count contract is invalid."
    )


# =============================================================================
# Standalone persisted model-state schema validation
# =============================================================================
#
# Notebook 10 Block 1 must be independently executable in a fresh runtime.
# It may depend only on the persisted Notebook 09 checkpoint/metadata and on
# libraries/authentication initialised inside this block.
#
# Representation and transformative state collections have stable top-level
# module identities and are validated exactly at that level.
#
# Transformative-Weight state is validated semantically: each pathway must
# contain exactly one learnable floating latent vector of the expected active
# dimensionality. The persisted implementation key may be `alpha_active`,
# whereas Notebook 10 reconstructs the runtime parameter as `alpha`.
# =============================================================================

BLOCK_1_PERSISTED_REPRESENTATION_STATE = (
    NOTEBOOK_10_INHERITED_CHECKPOINT.get(
        "representation_state"
    )
)


BLOCK_1_PERSISTED_TRANSFORMATIVE_STATE = (
    NOTEBOOK_10_INHERITED_CHECKPOINT.get(
        "transformative_state"
    )
)


BLOCK_1_PERSISTED_WEIGHT_STATE = (
    NOTEBOOK_10_INHERITED_CHECKPOINT.get(
        "transformative_weight_state"
    )
)


BLOCK_1_PERSISTED_EFFECTIVE_WEIGHTS = (
    NOTEBOOK_10_INHERITED_CHECKPOINT.get(
        "effective_transformative_weights"
    )
)


BLOCK_1_EXPECTED_REPRESENTATION_STATE_KEYS = {
    "backbone",
    "global_confluent",
    "psychological_head",
    "factual_head",
    "social_head",
}


BLOCK_1_EXPECTED_TRANSFORMATIVE_STATE_KEYS = set(
    BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS.keys()
)


BLOCK_1_REPRESENTATION_STATE_SCHEMA_VALID = all(
    [
        isinstance(
            BLOCK_1_PERSISTED_REPRESENTATION_STATE,
            dict,
        ),

        (
            set(
                BLOCK_1_PERSISTED_REPRESENTATION_STATE.keys()
            )
            ==
            BLOCK_1_EXPECTED_REPRESENTATION_STATE_KEYS
        ),

        all(
            isinstance(
                state_dict,
                dict,
            )
            and
            len(
                state_dict
            )
            >
            0

            for state_dict
            in BLOCK_1_PERSISTED_REPRESENTATION_STATE.values()
        ),
    ]
)


BLOCK_1_TRANSFORMATIVE_STATE_SCHEMA_VALID = all(
    [
        isinstance(
            BLOCK_1_PERSISTED_TRANSFORMATIVE_STATE,
            dict,
        ),

        (
            set(
                BLOCK_1_PERSISTED_TRANSFORMATIVE_STATE.keys()
            )
            ==
            BLOCK_1_EXPECTED_TRANSFORMATIVE_STATE_KEYS
        ),

        all(
            isinstance(
                state_dict,
                dict,
            )
            and
            len(
                state_dict
            )
            >
            0

            for state_dict
            in BLOCK_1_PERSISTED_TRANSFORMATIVE_STATE.values()
        ),
    ]
)


def block_1_persisted_latent_candidates(
    state_dict,
    expected_active_count,
):

    if not isinstance(
        state_dict,
        dict,
    ):

        return []


    return [
        (
            tensor_name,
            tensor,
        )

        for (
            tensor_name,
            tensor,
        ) in state_dict.items()

        if (
            torch.is_tensor(
                tensor
            )
            and
            tensor.is_floating_point()
            and
            tensor.numel()
            ==
            expected_active_count
        )
    ]


BLOCK_1_WEIGHT_STATE_SCHEMA_AUDIT = {}

BLOCK_1_WEIGHT_STATE_SCHEMA_VALID = all(
    [
        isinstance(
            BLOCK_1_PERSISTED_WEIGHT_STATE,
            dict,
        ),

        (
            set(
                BLOCK_1_PERSISTED_WEIGHT_STATE.keys()
            )
            ==
            set(
                BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS.keys()
            )
        ),
    ]
)


if BLOCK_1_WEIGHT_STATE_SCHEMA_VALID:

    for pathway_name in BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS:

        expected_active_count = len(
            BLOCK_1_EXPECTED_ACTIVE_DIMENSIONS[
                pathway_name
            ]
        )


        state_dict = (
            BLOCK_1_PERSISTED_WEIGHT_STATE[
                pathway_name
            ]
        )


        latent_candidates = (
            block_1_persisted_latent_candidates(
                state_dict,
                expected_active_count,
            )
        )


        preferred_candidates = [
            (
                tensor_name,
                tensor,
            )

            for (
                tensor_name,
                tensor,
            ) in latent_candidates

            if (
                "alpha"
                in
                tensor_name.lower()
                or
                "latent"
                in
                tensor_name.lower()
            )
        ]


        resolved_candidates = (
            preferred_candidates
            if preferred_candidates
            else latent_candidates
        )


        pathway_schema_valid = (
            len(
                resolved_candidates
            )
            ==
            1
        )


        BLOCK_1_WEIGHT_STATE_SCHEMA_AUDIT[
            pathway_name
        ] = {
            "expected_active_count":
                expected_active_count,

            "persisted_keys":
                tuple(
                    state_dict.keys()
                )
                if isinstance(
                    state_dict,
                    dict,
                )
                else tuple(),

            "candidate_count":
                len(
                    latent_candidates
                ),

            "preferred_candidate_count":
                len(
                    preferred_candidates
                ),

            "resolved_candidate_count":
                len(
                    resolved_candidates
                ),

            "schema_valid":
                pathway_schema_valid,
        }


        if not pathway_schema_valid:

            BLOCK_1_WEIGHT_STATE_SCHEMA_VALID = False


BLOCK_1_EFFECTIVE_WEIGHT_SCHEMA_VALID = all(
    [
        isinstance(
            BLOCK_1_PERSISTED_EFFECTIVE_WEIGHTS,
            dict,
        ),

        (
            set(
                BLOCK_1_PERSISTED_EFFECTIVE_WEIGHTS.keys()
            )
            ==
            set(
                BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS.keys()
            )
        ),

        all(
            torch.is_tensor(
                BLOCK_1_PERSISTED_EFFECTIVE_WEIGHTS[
                    pathway_name
                ]
            )
            and
            (
                BLOCK_1_PERSISTED_EFFECTIVE_WEIGHTS[
                    pathway_name
                ].numel()
                ==
                BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS[
                    pathway_name
                ]
            )

            for pathway_name
            in BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS
        ),
    ]
)


BLOCK_1_PERSISTED_STATE_SCHEMA_VALID = all(
    [
        BLOCK_1_REPRESENTATION_STATE_SCHEMA_VALID,
        BLOCK_1_TRANSFORMATIVE_STATE_SCHEMA_VALID,
        BLOCK_1_WEIGHT_STATE_SCHEMA_VALID,
        BLOCK_1_EFFECTIVE_WEIGHT_SCHEMA_VALID,
    ]
)


if not BLOCK_1_PERSISTED_STATE_SCHEMA_VALID:

    raise RuntimeError(
        "Notebook 09 persisted model-state schema is incompatible with "
        "the validated Notebook 09 → 10 architecture contract. "
        f"Representation schema valid: "
        f"{BLOCK_1_REPRESENTATION_STATE_SCHEMA_VALID}; "
        f"Transformative schema valid: "
        f"{BLOCK_1_TRANSFORMATIVE_STATE_SCHEMA_VALID}; "
        f"Transformative-Weight schema valid: "
        f"{BLOCK_1_WEIGHT_STATE_SCHEMA_VALID}; "
        f"Effective-weight schema valid: "
        f"{BLOCK_1_EFFECTIVE_WEIGHT_SCHEMA_VALID}; "
        f"Weight audit: {BLOCK_1_WEIGHT_STATE_SCHEMA_AUDIT}"
    )


# =============================================================================
# Metadata architecture contract
# =============================================================================

BLOCK_1_METADATA_ARCHITECTURE = (
    NOTEBOOK_10_INHERITED_METADATA.get(
        "architecture",
        {},
    )
)


BLOCK_1_METADATA_PATHWAY_DIMENSIONS = {
    pathway_name:
        int(
            dimension
        )

    for (
        pathway_name,
        dimension,
    ) in BLOCK_1_METADATA_ARCHITECTURE.get(
        "pathway_dimensions",
        {},
    ).items()
}


BLOCK_1_METADATA_ACTIVE_DIMENSIONS = {
    pathway_name:
        tuple(
            int(
                index
            )

            for index
            in indices
        )

    for (
        pathway_name,
        indices,
    ) in BLOCK_1_METADATA_ARCHITECTURE.get(
        "active_dimension_indices",
        {},
    ).items()
}


BLOCK_1_METADATA_DEFERRED_DIMENSIONS = {
    pathway_name:
        tuple(
            int(
                index
            )

            for index
            in indices
        )

    for (
        pathway_name,
        indices,
    ) in BLOCK_1_METADATA_ARCHITECTURE.get(
        "deferred_dimension_indices",
        {},
    ).items()
}


BLOCK_1_METADATA_ARCHITECTURE_VALID = all(
    [
        (
            BLOCK_1_METADATA_ARCHITECTURE.get(
                "representation_parameters"
            )
            ==
            BLOCK_1_EXPECTED_REPRESENTATION_PARAMETERS
        ),

        (
            BLOCK_1_METADATA_ARCHITECTURE.get(
                "transformative_parameters"
            )
            ==
            BLOCK_1_EXPECTED_TRANSFORMATIVE_PARAMETERS
        ),

        (
            BLOCK_1_METADATA_ARCHITECTURE.get(
                "transformative_weight_parameters"
            )
            ==
            BLOCK_1_EXPECTED_TRANSFORMATIVE_WEIGHT_PARAMETERS
        ),

        (
            BLOCK_1_METADATA_ARCHITECTURE.get(
                "total_parameters"
            )
            ==
            BLOCK_1_EXPECTED_TOTAL_PARAMETERS
        ),

        (
            BLOCK_1_METADATA_PATHWAY_DIMENSIONS
            ==
            BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS
        ),

        (
            BLOCK_1_METADATA_ACTIVE_DIMENSIONS
            ==
            BLOCK_1_EXPECTED_ACTIVE_DIMENSIONS
        ),

        (
            BLOCK_1_METADATA_DEFERRED_DIMENSIONS
            ==
            BLOCK_1_EXPECTED_DEFERRED_DIMENSIONS
        ),
    ]
)


if not BLOCK_1_METADATA_ARCHITECTURE_VALID:

    raise RuntimeError(
        "Notebook 09 final metadata architecture contract is invalid."
    )


# =============================================================================
# Cross-artefact contract
# =============================================================================

BLOCK_1_CROSS_ARTEFACT_VALID = all(
    [
        (
            BLOCK_1_CHECKPOINT_PATHWAY_DIMENSIONS
            ==
            BLOCK_1_METADATA_PATHWAY_DIMENSIONS
        ),

        (
            BLOCK_1_CHECKPOINT_ACTIVE_DIMENSIONS
            ==
            BLOCK_1_METADATA_ACTIVE_DIMENSIONS
        ),

        (
            BLOCK_1_CHECKPOINT_DEFERRED_DIMENSIONS
            ==
            BLOCK_1_METADATA_DEFERRED_DIMENSIONS
        ),

        (
            BLOCK_1_CHECKPOINT_ARCHITECTURE[
                "representation_parameter_count"
            ]
            ==
            BLOCK_1_METADATA_ARCHITECTURE[
                "representation_parameters"
            ]
        ),

        (
            BLOCK_1_CHECKPOINT_ARCHITECTURE[
                "transformative_parameter_count"
            ]
            ==
            BLOCK_1_METADATA_ARCHITECTURE[
                "transformative_parameters"
            ]
        ),

        (
            BLOCK_1_CHECKPOINT_ARCHITECTURE[
                "transformative_weight_parameter_count"
            ]
            ==
            BLOCK_1_METADATA_ARCHITECTURE[
                "transformative_weight_parameters"
            ]
        ),

        (
            BLOCK_1_CHECKPOINT_ARCHITECTURE[
                "total_parameter_count"
            ]
            ==
            BLOCK_1_METADATA_ARCHITECTURE[
                "total_parameters"
            ]
        ),

        (
            NOTEBOOK_10_INHERITED_CHECKPOINT[
                "recurrent_contract"
            ]
            ==
            NOTEBOOK_10_INHERITED_METADATA[
                "recurrent_contract"
            ]
        ),

        (
            BLOCK_1_CHECKPOINT_ARCHITECTURE.get(
                "structural_weight_dimensions"
            )
            ==
            BLOCK_1_EXPECTED_STRUCTURAL_WEIGHT_DIMENSIONS
        ),

        (
            BLOCK_1_CHECKPOINT_ARCHITECTURE.get(
                "active_weight_dimensions"
            )
            ==
            BLOCK_1_EXPECTED_ACTIVE_WEIGHT_DIMENSIONS
        ),

        (
            BLOCK_1_CHECKPOINT_ARCHITECTURE.get(
                "deferred_weight_dimensions"
            )
            ==
            BLOCK_1_EXPECTED_DEFERRED_WEIGHT_DIMENSIONS
        ),
    ]
)


if not BLOCK_1_CROSS_ARTEFACT_VALID:

    raise RuntimeError(
        "Notebook 09 checkpoint and metadata disagree on "
        "the inherited architecture."
    )


# =============================================================================
# Deterministic representation architecture reconstruction
# =============================================================================

class Notebook10Backbone(
    nn.Module
):

    def __init__(
        self,
        input_dim=768,
        output_dim=768,
    ):

        super().__init__()


        self.projection = nn.Linear(
            input_dim,
            output_dim,
        )


    def forward(
        self,
        x,
    ):

        return self.projection(
            x
        )


class Notebook10GlobalConfluent(
    nn.Module
):

    def __init__(
        self,
        input_dim=768,
        output_dim=256,
        dropout=0.10,
    ):

        super().__init__()


        self.global_projection = nn.Sequential(
            nn.Linear(
                input_dim,
                output_dim,
            ),

            nn.GELU(),

            nn.Dropout(
                dropout
            ),

            nn.LayerNorm(
                output_dim
            ),
        )


    def forward(
        self,
        x,
    ):

        return self.global_projection(
            x
        )


class Notebook10RepresentationCore(
    nn.Module
):

    def __init__(
        self,
        input_dim=256,
        hidden_dim=128,
        dropout=0.10,
    ):

        super().__init__()


        self.network = nn.Sequential(
            nn.Linear(
                input_dim,
                hidden_dim,
            ),

            nn.GELU(),

            nn.Dropout(
                dropout
            ),

            nn.LayerNorm(
                hidden_dim
            ),
        )


    def forward(
        self,
        x,
    ):

        return self.network(
            x
        )


class Notebook10RepresentationHead(
    nn.Module
):

    def __init__(
        self,
        output_dim,
        input_dim=256,
        hidden_dim=128,
        dropout=0.10,
    ):

        super().__init__()


        self.core = Notebook10RepresentationCore(
            input_dim=
                input_dim,

            hidden_dim=
                hidden_dim,

            dropout=
                dropout,
        )


        self.output_layer = nn.Linear(
            hidden_dim,
            output_dim,
        )


    def forward(
        self,
        x,
    ):

        return self.output_layer(
            self.core(
                x
            )
        )


# =============================================================================
# Deterministic transformative architecture reconstruction
# =============================================================================

class Notebook10TransformativeMechanism(
    nn.Module
):

    def __init__(
        self,
        pathway_dim,
        dropout=0.10,
    ):

        super().__init__()


        input_dim = (
            pathway_dim
            *
            2
        )


        hidden_dim = input_dim


        self.pathway_dim = int(
            pathway_dim
        )


        self.transform_network = nn.Sequential(
            nn.Linear(
                input_dim,
                hidden_dim,
            ),

            nn.GELU(),

            nn.Dropout(
                dropout
            ),

            nn.LayerNorm(
                hidden_dim
            ),

            nn.Linear(
                hidden_dim,
                pathway_dim,
            ),
        )


    def forward(
        self,
        current_representation,
        preceding_state,
    ):

        combined = torch.cat(
            (
                current_representation,
                preceding_state,
            ),
            dim=
                -1,
        )


        return self.transform_network(
            combined
        )


# =============================================================================
# Instantiate representation and transformative modules
# =============================================================================

NOTEBOOK_10_RESTORED_BACKBONE = (
    Notebook10Backbone()
)


NOTEBOOK_10_RESTORED_GLOBAL_CONFLUENT = (
    Notebook10GlobalConfluent()
)


NOTEBOOK_10_RESTORED_PSYCHOLOGICAL_HEAD = (
    Notebook10RepresentationHead(
        output_dim=
            BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS[
                "psychological"
            ]
    )
)


NOTEBOOK_10_RESTORED_FACTUAL_HEAD = (
    Notebook10RepresentationHead(
        output_dim=
            BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS[
                "factual"
            ]
    )
)


NOTEBOOK_10_RESTORED_SOCIAL_HEAD = (
    Notebook10RepresentationHead(
        output_dim=
            BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS[
                "social"
            ]
    )
)


NOTEBOOK_10_FACTUAL_TRANSFORMATIVE_MECHANISM = (
    Notebook10TransformativeMechanism(
        pathway_dim=
            BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS[
                "factual"
            ]
    )
)


NOTEBOOK_10_PSYCHOLOGICAL_TRANSFORMATIVE_MECHANISM = (
    Notebook10TransformativeMechanism(
        pathway_dim=
            BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS[
                "psychological"
            ]
    )
)


NOTEBOOK_10_SOCIAL_TRANSFORMATIVE_MECHANISM = (
    Notebook10TransformativeMechanism(
        pathway_dim=
            BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS[
                "social"
            ]
    )
)


NOTEBOOK_10_REPRESENTATION_MODULES = {
    "backbone":
        NOTEBOOK_10_RESTORED_BACKBONE,

    "global_confluent":
        NOTEBOOK_10_RESTORED_GLOBAL_CONFLUENT,

    "psychological_head":
        NOTEBOOK_10_RESTORED_PSYCHOLOGICAL_HEAD,

    "factual_head":
        NOTEBOOK_10_RESTORED_FACTUAL_HEAD,

    "social_head":
        NOTEBOOK_10_RESTORED_SOCIAL_HEAD,
}


NOTEBOOK_10_TRANSFORMATIVE_MODULES = {
    "factual":
        NOTEBOOK_10_FACTUAL_TRANSFORMATIVE_MECHANISM,

    "psychological":
        NOTEBOOK_10_PSYCHOLOGICAL_TRANSFORMATIVE_MECHANISM,

    "social":
        NOTEBOOK_10_SOCIAL_TRANSFORMATIVE_MECHANISM,
}


# =============================================================================
# Restore representation state
# =============================================================================

BLOCK_1_CHECKPOINT_REPRESENTATION_STATE = (
    BLOCK_1_PERSISTED_REPRESENTATION_STATE
)


for (
    module_name,
    module,
) in NOTEBOOK_10_REPRESENTATION_MODULES.items():

    if module_name not in BLOCK_1_CHECKPOINT_REPRESENTATION_STATE:

        raise RuntimeError(
            "Notebook 09 checkpoint is missing representation state for "
            f"'{module_name}'."
        )


    module.load_state_dict(
        BLOCK_1_CHECKPOINT_REPRESENTATION_STATE[
            module_name
        ],
        strict=
            True,
    )


# =============================================================================
# Restore transformative state
# =============================================================================

BLOCK_1_CHECKPOINT_TRANSFORMATIVE_STATE = (
    BLOCK_1_PERSISTED_TRANSFORMATIVE_STATE
)


for (
    pathway_name,
    module,
) in NOTEBOOK_10_TRANSFORMATIVE_MODULES.items():

    if pathway_name not in BLOCK_1_CHECKPOINT_TRANSFORMATIVE_STATE:

        raise RuntimeError(
            "Notebook 09 checkpoint is missing transformative state for "
            f"'{pathway_name}'."
        )


    module.load_state_dict(
        BLOCK_1_CHECKPOINT_TRANSFORMATIVE_STATE[
            pathway_name
        ],
        strict=
            True,
    )


# =============================================================================
# Exact state helper
# =============================================================================

def block_1_state_dict_exact(
    expected_state,
    observed_state,
):

    if (
        set(
            expected_state.keys()
        )
        !=
        set(
            observed_state.keys()
        )
    ):

        return False


    for key in expected_state:

        expected = expected_state[
            key
        ]


        observed = observed_state[
            key
        ]


        if not (
            torch.is_tensor(
                expected
            )
            and
            torch.is_tensor(
                observed
            )
        ):

            return False


        if not torch.equal(
            expected.detach().cpu(),
            observed.detach().cpu(),
        ):

            return False


    return True


# =============================================================================
# Exact representation / transformative restoration validation
# =============================================================================

BLOCK_1_REPRESENTATION_STATE_EXACT = all(
    block_1_state_dict_exact(
        BLOCK_1_CHECKPOINT_REPRESENTATION_STATE[
            module_name
        ],
        module.state_dict(),
    )

    for (
        module_name,
        module,
    ) in NOTEBOOK_10_REPRESENTATION_MODULES.items()
)


BLOCK_1_TRANSFORMATIVE_STATE_EXACT = all(
    block_1_state_dict_exact(
        BLOCK_1_CHECKPOINT_TRANSFORMATIVE_STATE[
            pathway_name
        ],
        module.state_dict(),
    )

    for (
        pathway_name,
        module,
    ) in NOTEBOOK_10_TRANSFORMATIVE_MODULES.items()
)


if not all(
    [
        BLOCK_1_REPRESENTATION_STATE_EXACT,
        BLOCK_1_TRANSFORMATIVE_STATE_EXACT,
    ]
):

    raise RuntimeError(
        "Notebook 10 could not reproduce the persisted representation "
        "or transformative state exactly."
    )


# =============================================================================
# Transformative-Weight persisted-state inspection
# =============================================================================
#
# Notebook 09 persists one learned latent parameter vector per pathway.
#
# We deliberately inspect the persisted state rather than assuming that its
# parameter key has a particular name.
# =============================================================================

BLOCK_1_CHECKPOINT_WEIGHT_STATE = (
    BLOCK_1_PERSISTED_WEIGHT_STATE
)


def block_1_resolve_latent_tensor(
    pathway_name,
    state_dict,
    expected_active_count,
):

    if not isinstance(
        state_dict,
        dict,
    ):

        raise TypeError(
            f"Persisted Transformative-Weight state for "
            f"{pathway_name} is not dictionary-like."
        )


    tensor_entries = {
        key:
            value

        for (
            key,
            value,
        ) in state_dict.items()

        if torch.is_tensor(
            value
        )
    }


    preferred_candidates = [
        (
            key,
            value,
        )

        for (
            key,
            value,
        ) in tensor_entries.items()

        if (
            value.is_floating_point()
            and
            value.numel()
            ==
            expected_active_count
            and
            (
                "alpha"
                in key.lower()
                or
                "latent"
                in key.lower()
            )
        )
    ]


    if len(
        preferred_candidates
    ) == 1:

        return preferred_candidates[
            0
        ]


    general_candidates = [
        (
            key,
            value,
        )

        for (
            key,
            value,
        ) in tensor_entries.items()

        if (
            value.is_floating_point()
            and
            value.numel()
            ==
            expected_active_count
        )
    ]


    if len(
        general_candidates
    ) == 1:

        return general_candidates[
            0
        ]


    state_description = {
        key:
            {
                "shape":
                    tuple(
                        value.shape
                    ),

                "dtype":
                    str(
                        value.dtype
                    ),

                "floating":
                    bool(
                        value.is_floating_point()
                    ),
            }

        for (
            key,
            value,
        ) in tensor_entries.items()
    }


    raise RuntimeError(
        "Notebook 10 could not identify exactly one learned latent "
        f"Transformative-Weight tensor for pathway '{pathway_name}'. "
        f"Persisted state: {state_description}"
    )


BLOCK_1_LATENT_SOURCE_KEYS = {}

BLOCK_1_LATENT_WEIGHT_TENSORS = {}


for pathway_name in BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS:

    (
        latent_key,
        latent_tensor,
    ) = block_1_resolve_latent_tensor(
        pathway_name=
            pathway_name,

        state_dict=
            BLOCK_1_CHECKPOINT_WEIGHT_STATE[
                pathway_name
            ],

        expected_active_count=
            len(
                BLOCK_1_EXPECTED_ACTIVE_DIMENSIONS[
                    pathway_name
                ]
            ),
    )


    BLOCK_1_LATENT_SOURCE_KEYS[
        pathway_name
    ] = latent_key


    BLOCK_1_LATENT_WEIGHT_TENSORS[
        pathway_name
    ] = (
        latent_tensor.detach()
        .cpu()
        .clone()
        .reshape(
            -1
        )
    )


# =============================================================================
# Deterministic Transformative-Weight reconstruction
# =============================================================================

class Notebook10TransformativeWeight(
    nn.Module
):

    def __init__(
        self,
        structural_dim,
        active_indices,
        latent_values,
    ):

        super().__init__()


        self.structural_dim = int(
            structural_dim
        )


        self.active_indices = tuple(
            int(
                index
            )

            for index
            in active_indices
        )


        if latent_values.numel() != len(
            self.active_indices
        ):

            raise ValueError(
                "Transformative-Weight latent parameter count does not "
                "match the active-dimension count."
            )


        self.alpha = nn.Parameter(
            latent_values.detach()
            .clone()
            .to(
                dtype=
                    DEFAULT_DTYPE
            )
            .reshape(
                -1
            )
        )


    def forward(
        self,
    ):

        effective_weight = torch.zeros(
            self.structural_dim,
            dtype=
                self.alpha.dtype,
            device=
                self.alpha.device,
        )


        active_index_tensor = torch.tensor(
            self.active_indices,
            dtype=
                torch.long,
            device=
                self.alpha.device,
        )


        effective_weight = effective_weight.index_copy(
            0,
            active_index_tensor,
            torch.sigmoid(
                self.alpha
            ),
        )


        return effective_weight


NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES = {
    pathway_name:
        Notebook10TransformativeWeight(
            structural_dim=
                BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS[
                    pathway_name
                ],

            active_indices=
                BLOCK_1_EXPECTED_ACTIVE_DIMENSIONS[
                    pathway_name
                ],

            latent_values=
                BLOCK_1_LATENT_WEIGHT_TENSORS[
                    pathway_name
                ],
        )

    for pathway_name
    in BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS
}


# =============================================================================
# Persisted effective Transformative-Weight contract
# =============================================================================

BLOCK_1_PERSISTED_EFFECTIVE_WEIGHTS = {
    pathway_name:
        NOTEBOOK_10_INHERITED_CHECKPOINT[
            "effective_transformative_weights"
        ][
            pathway_name
        ]
        .detach()
        .cpu()
        .clone()

    for pathway_name
    in BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS
}


# =============================================================================
# Validate reconstructed effective weights without invoking module.forward()
# =============================================================================

BLOCK_1_RECONSTRUCTED_EFFECTIVE_WEIGHTS = {}


for pathway_name in BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS:

    structural_dim = (
        BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS[
            pathway_name
        ]
    )


    active_indices = (
        BLOCK_1_EXPECTED_ACTIVE_DIMENSIONS[
            pathway_name
        ]
    )


    latent_values = (
        BLOCK_1_LATENT_WEIGHT_TENSORS[
            pathway_name
        ]
    )


    reconstructed_weight = torch.zeros(
        structural_dim,
        dtype=
            DEFAULT_DTYPE,
    )


    reconstructed_weight[
        list(
            active_indices
        )
    ] = torch.sigmoid(
        latent_values.to(
            dtype=
                DEFAULT_DTYPE
        )
    )


    BLOCK_1_RECONSTRUCTED_EFFECTIVE_WEIGHTS[
        pathway_name
    ] = reconstructed_weight


BLOCK_1_EFFECTIVE_WEIGHT_IDENTITY_VALID = all(
    torch.equal(
        BLOCK_1_RECONSTRUCTED_EFFECTIVE_WEIGHTS[
            pathway_name
        ],
        BLOCK_1_PERSISTED_EFFECTIVE_WEIGHTS[
            pathway_name
        ].to(
            dtype=
                DEFAULT_DTYPE
        ),
    )

    for pathway_name
    in BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS
)


if not BLOCK_1_EFFECTIVE_WEIGHT_IDENTITY_VALID:

    raise RuntimeError(
        "Notebook 10 reconstructed Transformative Weights do not "
        "reproduce the persisted Notebook 09 effective-weight state."
    )


# =============================================================================
# Transformative-Weight numerical validation
# =============================================================================

BLOCK_1_EFFECTIVE_WEIGHTS_FINITE = {}

BLOCK_1_EFFECTIVE_WEIGHTS_BOUNDED = {}

BLOCK_1_DEFERRED_WEIGHTS_ZERO = {}


for pathway_name in BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS:

    effective_weight = (
        BLOCK_1_RECONSTRUCTED_EFFECTIVE_WEIGHTS[
            pathway_name
        ]
    )


    deferred_indices = (
        BLOCK_1_EXPECTED_DEFERRED_DIMENSIONS[
            pathway_name
        ]
    )


    BLOCK_1_EFFECTIVE_WEIGHTS_FINITE[
        pathway_name
    ] = bool(
        torch.isfinite(
            effective_weight
        ).all().item()
    )


    BLOCK_1_EFFECTIVE_WEIGHTS_BOUNDED[
        pathway_name
    ] = bool(
        (
            (
                effective_weight
                >=
                0.0
            )
            &
            (
                effective_weight
                <=
                1.0
            )
        ).all().item()
    )


    BLOCK_1_DEFERRED_WEIGHTS_ZERO[
        pathway_name
    ] = torch.equal(
        effective_weight[
            list(
                deferred_indices
            )
        ],
        torch.zeros(
            len(
                deferred_indices
            ),
            dtype=
                effective_weight.dtype,
        ),
    )


BLOCK_1_ALL_EFFECTIVE_WEIGHTS_FINITE = all(
    BLOCK_1_EFFECTIVE_WEIGHTS_FINITE.values()
)


BLOCK_1_ALL_EFFECTIVE_WEIGHTS_BOUNDED = all(
    BLOCK_1_EFFECTIVE_WEIGHTS_BOUNDED.values()
)


BLOCK_1_ALL_DEFERRED_WEIGHTS_ZERO = all(
    BLOCK_1_DEFERRED_WEIGHTS_ZERO.values()
)


if not all(
    [
        BLOCK_1_ALL_EFFECTIVE_WEIGHTS_FINITE,
        BLOCK_1_ALL_EFFECTIVE_WEIGHTS_BOUNDED,
        BLOCK_1_ALL_DEFERRED_WEIGHTS_ZERO,
    ]
):

    raise RuntimeError(
        "Notebook 10 inherited Transformative-Weight numerical "
        "contract is invalid."
    )


# =============================================================================
# Move restored architecture to controlled device
# =============================================================================

for module_collection in (
    NOTEBOOK_10_REPRESENTATION_MODULES,
    NOTEBOOK_10_TRANSFORMATIVE_MODULES,
    NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
):

    for module in module_collection.values():

        module.to(
            device=
                DEVICE,
            dtype=
                DEFAULT_DTYPE,
        )


# =============================================================================
# Freeze complete inherited architecture
# =============================================================================

for module_collection in (
    NOTEBOOK_10_REPRESENTATION_MODULES,
    NOTEBOOK_10_TRANSFORMATIVE_MODULES,
    NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
):

    for module in module_collection.values():

        module.eval()


        for parameter in module.parameters():

            parameter.requires_grad_(
                False
            )


# =============================================================================
# Freeze / evaluation-mode audit
# =============================================================================

BLOCK_1_ALL_PARAMETERS_FROZEN = all(
    not parameter.requires_grad

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


BLOCK_1_ALL_MODULES_EVAL = all(
    not module.training

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()
)


if not all(
    [
        BLOCK_1_ALL_PARAMETERS_FROZEN,
        BLOCK_1_ALL_MODULES_EVAL,
    ]
):

    raise RuntimeError(
        "Notebook 10 inherited architecture is not fully frozen "
        "and in evaluation mode."
    )


# =============================================================================
# Actual parameter accounting
# =============================================================================

BLOCK_1_REPRESENTATION_PARAMETER_COUNTS = {
    module_name:
        int(
            sum(
                parameter.numel()

                for parameter
                in module.parameters()
            )
        )

    for (
        module_name,
        module,
    ) in NOTEBOOK_10_REPRESENTATION_MODULES.items()
}


BLOCK_1_TRANSFORMATIVE_PARAMETER_COUNTS = {
    pathway_name:
        int(
            sum(
                parameter.numel()

                for parameter
                in module.parameters()
            )
        )

    for (
        pathway_name,
        module,
    ) in NOTEBOOK_10_TRANSFORMATIVE_MODULES.items()
}


BLOCK_1_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNTS = {
    pathway_name:
        int(
            sum(
                parameter.numel()

                for parameter
                in module.parameters()
            )
        )

    for (
        pathway_name,
        module,
    ) in NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES.items()
}


BLOCK_1_REPRESENTATION_PARAMETER_COUNT = int(
    sum(
        BLOCK_1_REPRESENTATION_PARAMETER_COUNTS.values()
    )
)


BLOCK_1_TRANSFORMATIVE_PARAMETER_COUNT = int(
    sum(
        BLOCK_1_TRANSFORMATIVE_PARAMETER_COUNTS.values()
    )
)


BLOCK_1_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT = int(
    sum(
        BLOCK_1_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNTS.values()
    )
)


BLOCK_1_TOTAL_PARAMETER_COUNT = int(
    BLOCK_1_REPRESENTATION_PARAMETER_COUNT
    +
    BLOCK_1_TRANSFORMATIVE_PARAMETER_COUNT
    +
    BLOCK_1_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT
)


BLOCK_1_PARAMETER_COUNTS_VALID = all(
    [
        (
            BLOCK_1_REPRESENTATION_PARAMETER_COUNT
            ==
            BLOCK_1_EXPECTED_REPRESENTATION_PARAMETERS
        ),

        (
            BLOCK_1_TRANSFORMATIVE_PARAMETER_COUNT
            ==
            BLOCK_1_EXPECTED_TRANSFORMATIVE_PARAMETERS
        ),

        (
            BLOCK_1_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT
            ==
            BLOCK_1_EXPECTED_TRANSFORMATIVE_WEIGHT_PARAMETERS
        ),

        (
            BLOCK_1_TOTAL_PARAMETER_COUNT
            ==
            BLOCK_1_EXPECTED_TOTAL_PARAMETERS
        ),
    ]
)


if not BLOCK_1_PARAMETER_COUNTS_VALID:

    raise RuntimeError(
        "Notebook 10 restored parameter counts do not reproduce "
        "the canonical Notebook 09 architecture."
    )


# =============================================================================
# Expected per-module parameter counts
# =============================================================================

BLOCK_1_EXPECTED_REPRESENTATION_PARAMETER_COUNTS = {
    "backbone":
        590_592,

    "global_confluent":
        197_376,

    "psychological_head":
        37_538,

    "factual_head":
        34_442,

    "social_head":
        34_055,
}


BLOCK_1_EXPECTED_TRANSFORMATIVE_PARAMETER_COUNTS = {
    "factual":
        670,

    "psychological":
        7_174,

    "social":
        343,
}


BLOCK_1_PER_MODULE_PARAMETER_COUNTS_VALID = all(
    [
        (
            BLOCK_1_REPRESENTATION_PARAMETER_COUNTS
            ==
            BLOCK_1_EXPECTED_REPRESENTATION_PARAMETER_COUNTS
        ),

        (
            BLOCK_1_TRANSFORMATIVE_PARAMETER_COUNTS
            ==
            BLOCK_1_EXPECTED_TRANSFORMATIVE_PARAMETER_COUNTS
        ),

        (
            BLOCK_1_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNTS
            ==
            BLOCK_1_EXPECTED_WEIGHT_PARAMETER_COUNTS
        ),
    ]
)


if not BLOCK_1_PER_MODULE_PARAMETER_COUNTS_VALID:

    raise RuntimeError(
        "Notebook 10 per-module parameter accounting does not "
        "reproduce the validated architecture."
    )


# =============================================================================
# Recurrent contract restoration
# =============================================================================

NOTEBOOK_10_RECURRENT_CONTRACT = deepcopy(
    NOTEBOOK_10_INHERITED_CHECKPOINT[
        "recurrent_contract"
    ]
)


BLOCK_1_EXPECTED_RECURRENT_CONTRACT = {
    "update_type":
        "additive",

    "initial_state_policy":
        "deterministic_zero",

    "article_boundary_reset":
        True,

    "cross_article_propagation":
        False,

    "cross_pathway_recurrence":
        False,

    "future_context_allowed":
        False,
}


BLOCK_1_RECURRENT_CONTRACT_VALID = (
    NOTEBOOK_10_RECURRENT_CONTRACT
    ==
    BLOCK_1_EXPECTED_RECURRENT_CONTRACT
)


if not BLOCK_1_RECURRENT_CONTRACT_VALID:

    raise RuntimeError(
        "Notebook 09 → 10 recurrent-state contract is invalid."
    )


# =============================================================================
# Pilot-training methodological boundary
# =============================================================================

BLOCK_1_CHECKPOINT_PILOT_TRAINING = (
    NOTEBOOK_10_INHERITED_CHECKPOINT[
        "pilot_training"
    ]
)


BLOCK_1_METADATA_PILOT_TRAINING = (
    NOTEBOOK_10_INHERITED_METADATA[
        "pilot_training"
    ]
)


BLOCK_1_PILOT_TRAINING_BOUNDARY_VALID = all(
    [
        (
            BLOCK_1_CHECKPOINT_PILOT_TRAINING.get(
                "final_training"
            )
            is False
        ),

        (
            BLOCK_1_CHECKPOINT_PILOT_TRAINING.get(
                "expanded_corpus_training_required"
            )
            is True
        ),

        (
            BLOCK_1_METADATA_PILOT_TRAINING.get(
                "final_training"
            )
            is False
        ),

        (
            BLOCK_1_METADATA_PILOT_TRAINING.get(
                "expanded_corpus_training_required"
            )
            is True
        ),
    ]
)


if not BLOCK_1_PILOT_TRAINING_BOUNDARY_VALID:

    raise RuntimeError(
        "Notebook 09 pilot-training methodological boundary "
        "was not preserved."
    )


# =============================================================================
# Completion state
# =============================================================================

BLOCK_1_CHECKPOINT_COMPLETION = (
    NOTEBOOK_10_INHERITED_CHECKPOINT[
        "completion"
    ]
)


BLOCK_1_METADATA_COMPLETION = (
    NOTEBOOK_10_INHERITED_METADATA[
        "completion"
    ]
)


BLOCK_1_NOTEBOOK_09_COMPLETION_VALID = all(
    [
        (
            BLOCK_1_CHECKPOINT_COMPLETION.get(
                "architecture_complete"
            )
            is True
        ),

        (
            BLOCK_1_CHECKPOINT_COMPLETION.get(
                "controlled_training_valid"
            )
            is True
        ),

        (
            BLOCK_1_CHECKPOINT_COMPLETION.get(
                "posttraining_recurrent_validation_valid"
            )
            is True
        ),

        (
            BLOCK_1_CHECKPOINT_COMPLETION.get(
                "notebook_09_complete"
            )
            is True
        ),

        (
            BLOCK_1_METADATA_COMPLETION.get(
                "notebook_09_complete"
            )
            is True
        ),

        (
            BLOCK_1_METADATA_COMPLETION.get(
                "architecture_complete"
            )
            is True
        ),

        (
            BLOCK_1_METADATA_COMPLETION.get(
                "posttraining_validation_valid"
            )
            is True
        ),

        (
            BLOCK_1_METADATA_COMPLETION.get(
                "notebook_10_ready"
            )
            is True
        ),
    ]
)


if not BLOCK_1_NOTEBOOK_09_COMPLETION_VALID:

    raise RuntimeError(
        "Notebook 09 completion state is not valid for Notebook 10."
    )


# =============================================================================
# Post-device exact state validation
# =============================================================================
#
# DEVICE is intentionally CPU in Block 1, allowing exact comparison against
# the CPU-safe checkpoint tensors.
# =============================================================================

BLOCK_1_POST_DEVICE_REPRESENTATION_EXACT = all(
    block_1_state_dict_exact(
        BLOCK_1_CHECKPOINT_REPRESENTATION_STATE[
            module_name
        ],
        module.state_dict(),
    )

    for (
        module_name,
        module,
    ) in NOTEBOOK_10_REPRESENTATION_MODULES.items()
)


BLOCK_1_POST_DEVICE_TRANSFORMATIVE_EXACT = all(
    block_1_state_dict_exact(
        BLOCK_1_CHECKPOINT_TRANSFORMATIVE_STATE[
            pathway_name
        ],
        module.state_dict(),
    )

    for (
        pathway_name,
        module,
    ) in NOTEBOOK_10_TRANSFORMATIVE_MODULES.items()
)


BLOCK_1_POST_DEVICE_WEIGHT_LATENT_EXACT = all(
    torch.equal(
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES[
            pathway_name
        ].alpha.detach()
        .cpu(),

        BLOCK_1_LATENT_WEIGHT_TENSORS[
            pathway_name
        ].to(
            dtype=
                DEFAULT_DTYPE
        ),
    )

    for pathway_name
    in BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS
)


if not all(
    [
        BLOCK_1_POST_DEVICE_REPRESENTATION_EXACT,
        BLOCK_1_POST_DEVICE_TRANSFORMATIVE_EXACT,
        BLOCK_1_POST_DEVICE_WEIGHT_LATENT_EXACT,
    ]
):

    raise RuntimeError(
        "Notebook 10 post-device restored model state does not "
        "reproduce the Notebook 09 checkpoint exactly."
    )


# =============================================================================
# Gradient-state validation
# =============================================================================

BLOCK_1_ALL_GRADIENTS_ABSENT = all(
    parameter.grad is None

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


if not BLOCK_1_ALL_GRADIENTS_ABSENT:

    raise RuntimeError(
        "Unexpected gradients exist after Notebook 10 restoration."
    )


# =============================================================================
# Expose stable Notebook 10 architecture globals
# =============================================================================

NOTEBOOK_10_PATHWAY_DIMENSIONS = deepcopy(
    BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS
)


NOTEBOOK_10_ACTIVE_DIMENSION_INDICES = deepcopy(
    BLOCK_1_EXPECTED_ACTIVE_DIMENSIONS
)


NOTEBOOK_10_DEFERRED_DIMENSION_INDICES = deepcopy(
    BLOCK_1_EXPECTED_DEFERRED_DIMENSIONS
)


NOTEBOOK_10_REPRESENTATION_PARAMETER_COUNT = (
    BLOCK_1_REPRESENTATION_PARAMETER_COUNT
)


NOTEBOOK_10_TRANSFORMATIVE_PARAMETER_COUNT = (
    BLOCK_1_TRANSFORMATIVE_PARAMETER_COUNT
)


NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT = (
    BLOCK_1_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT
)


NOTEBOOK_10_TOTAL_PARAMETER_COUNT = (
    BLOCK_1_TOTAL_PARAMETER_COUNT
)


NOTEBOOK_10_INHERITED_EFFECTIVE_TRANSFORMATIVE_WEIGHTS = {
    pathway_name:
        BLOCK_1_RECONSTRUCTED_EFFECTIVE_WEIGHTS[
            pathway_name
        ].detach()
        .cpu()
        .clone()

    for pathway_name
    in BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS
}


NOTEBOOK_10_NOTEBOOK_09_PILOT_PRETRAIN_OBJECTIVE = float(
    BLOCK_1_CHECKPOINT_PILOT_TRAINING[
        "pretraining_total_objective"
    ]
)


NOTEBOOK_10_NOTEBOOK_09_PILOT_POSTTRAIN_OBJECTIVE = float(
    BLOCK_1_CHECKPOINT_PILOT_TRAINING[
        "posttraining_total_objective"
    ]
)


# =============================================================================
# Explicit Block 1 prohibited-operation validation
# =============================================================================

BLOCK_1_EXECUTION_BOUNDARY_VALID = all(
    [
        NOTEBOOK_10_BLOCK_1_EXPANDED_CORPUS_LOADED is False,

        NOTEBOOK_10_BLOCK_1_REPRESENTATION_FORWARD_EXECUTED is False,

        NOTEBOOK_10_BLOCK_1_TRANSFORMATIVE_FORWARD_EXECUTED is False,

        NOTEBOOK_10_BLOCK_1_WEIGHT_FORWARD_EXECUTED is False,

        NOTEBOOK_10_BLOCK_1_RECURRENT_STATE_CREATED is False,

        NOTEBOOK_10_BLOCK_1_RECURRENT_UPDATE_EXECUTED is False,

        NOTEBOOK_10_BLOCK_1_LOSS_CALCULATED is False,

        NOTEBOOK_10_BLOCK_1_OPTIMIZER_CREATED is False,

        NOTEBOOK_10_BLOCK_1_ZERO_GRAD_EXECUTED is False,

        NOTEBOOK_10_BLOCK_1_BACKWARD_PASS_EXECUTED is False,

        NOTEBOOK_10_BLOCK_1_PARAMETER_UPDATE_EXECUTED is False,

        NOTEBOOK_10_BLOCK_1_TRAINING_EXECUTED is False,
    ]
)


# =============================================================================
# Final Block 1 validation
# =============================================================================

NOTEBOOK_10_BLOCK_1_ERRORS = []


BLOCK_1_VALIDATION_CHECKS = {
    "drive_service_not_initialised":
        BLOCK_1_DRIVE_SERVICE_INITIALISED,

    "checkpoint_not_located":
        BLOCK_1_CHECKPOINT_LOCATED,

    "metadata_not_located":
        BLOCK_1_METADATA_LOCATED,

    "checkpoint_filename_invalid":
        BLOCK_1_CHECKPOINT_FILENAME_VALID,

    "metadata_filename_invalid":
        BLOCK_1_METADATA_FILENAME_VALID,

    "metadata_download_invalid":
        BLOCK_1_METADATA_DOWNLOADED,

    "metadata_drive_byte_count_invalid":
        BLOCK_1_METADATA_DRIVE_BYTE_COUNT_VALID,

    "metadata_json_invalid":
        BLOCK_1_METADATA_JSON_PARSED,

    "metadata_identity_invalid":
        BLOCK_1_METADATA_IDENTITY_VALID,

    "metadata_bootstrap_invalid":
        BLOCK_1_METADATA_BOOTSTRAP_VALID,

    "metadata_checkpoint_identity_invalid":
        BLOCK_1_METADATA_CHECKPOINT_IDENTITY_VALID,

    "checkpoint_download_invalid":
        BLOCK_1_CHECKPOINT_DOWNLOADED,

    "checkpoint_sha256_invalid":
        BLOCK_1_CHECKPOINT_SHA256_VALID,

    "checkpoint_metadata_byte_count_invalid":
        BLOCK_1_CHECKPOINT_METADATA_BYTE_COUNT_VALID,

    "checkpoint_drive_byte_count_invalid":
        BLOCK_1_CHECKPOINT_DRIVE_BYTE_COUNT_VALID,

    "checkpoint_parse_invalid":
        BLOCK_1_CHECKPOINT_PARSED,

    "checkpoint_identity_invalid":
        BLOCK_1_CHECKPOINT_IDENTITY_VALID,

    "checkpoint_required_keys_invalid":
        BLOCK_1_CHECKPOINT_REQUIRED_KEYS_VALID,

    "checkpoint_parameter_counts_invalid":
        BLOCK_1_CHECKPOINT_PARAMETER_COUNTS_VALID,

    "metadata_architecture_invalid":
        BLOCK_1_METADATA_ARCHITECTURE_VALID,

    "cross_artefact_invalid":
        BLOCK_1_CROSS_ARTEFACT_VALID,

    "pathway_dimensions_invalid":
        BLOCK_1_PATHWAY_DIMENSIONS_VALID,

    "active_dimensions_invalid":
        BLOCK_1_ACTIVE_DIMENSIONS_VALID,

    "deferred_dimensions_invalid":
        BLOCK_1_DEFERRED_DIMENSIONS_VALID,

    "dimension_partitions_invalid":
        BLOCK_1_ALL_DIMENSION_PARTITIONS_VALID,

    "representation_state_not_exact":
        BLOCK_1_REPRESENTATION_STATE_EXACT,

    "transformative_state_not_exact":
        BLOCK_1_TRANSFORMATIVE_STATE_EXACT,

    "effective_weight_identity_invalid":
        BLOCK_1_EFFECTIVE_WEIGHT_IDENTITY_VALID,

    "effective_weights_not_finite":
        BLOCK_1_ALL_EFFECTIVE_WEIGHTS_FINITE,

    "effective_weights_not_bounded":
        BLOCK_1_ALL_EFFECTIVE_WEIGHTS_BOUNDED,

    "deferred_weights_not_zero":
        BLOCK_1_ALL_DEFERRED_WEIGHTS_ZERO,

    "all_parameters_not_frozen":
        BLOCK_1_ALL_PARAMETERS_FROZEN,

    "all_modules_not_eval":
        BLOCK_1_ALL_MODULES_EVAL,

    "parameter_counts_invalid":
        BLOCK_1_PARAMETER_COUNTS_VALID,

    "per_module_parameter_counts_invalid":
        BLOCK_1_PER_MODULE_PARAMETER_COUNTS_VALID,

    "recurrent_contract_invalid":
        BLOCK_1_RECURRENT_CONTRACT_VALID,

    "pilot_training_boundary_invalid":
        BLOCK_1_PILOT_TRAINING_BOUNDARY_VALID,

    "notebook_09_completion_invalid":
        BLOCK_1_NOTEBOOK_09_COMPLETION_VALID,

    "post_device_representation_not_exact":
        BLOCK_1_POST_DEVICE_REPRESENTATION_EXACT,

    "post_device_transformative_not_exact":
        BLOCK_1_POST_DEVICE_TRANSFORMATIVE_EXACT,

    "post_device_weight_latent_not_exact":
        BLOCK_1_POST_DEVICE_WEIGHT_LATENT_EXACT,

    "unexpected_gradients":
        BLOCK_1_ALL_GRADIENTS_ABSENT,

    "execution_boundary_invalid":
        BLOCK_1_EXECUTION_BOUNDARY_VALID,
}


for (
    error_name,
    condition,
) in BLOCK_1_VALIDATION_CHECKS.items():

    if not condition:

        NOTEBOOK_10_BLOCK_1_ERRORS.append(
            error_name
        )


# =============================================================================
# Final Notebook 10 initialisation state
# =============================================================================

NOTEBOOK_09_TO_10_HANDOVER_VALID = (
    len(
        NOTEBOOK_10_BLOCK_1_ERRORS
    )
    ==
    0
)


NOTEBOOK_10_INITIALISED = (
    NOTEBOOK_09_TO_10_HANDOVER_VALID
)


NOTEBOOK_10_EXPANDED_CORPUS_INVENTORY_READY = (
    NOTEBOOK_10_INITIALISED
)


NOTEBOOK_10_BLOCK_1_VALID = (
    NOTEBOOK_10_INITIALISED
)


if not NOTEBOOK_10_BLOCK_1_VALID:

    raise RuntimeError(
        "Notebook 10 Block 1 Notebook 09 → 10 restoration failed. "
        f"Errors: {NOTEBOOK_10_BLOCK_1_ERRORS}"
    )


NOTEBOOK_10_BLOCK_1_COMPLETE = True


# =============================================================================
# Block summary
# =============================================================================

NOTEBOOK_10_BLOCK_1_SUMMARY = {
    "block":
        NOTEBOOK_10_BLOCK_1,

    "block_name":
        NOTEBOOK_10_BLOCK_1_NAME,

    "block_version":
        NOTEBOOK_10_BLOCK_1_VERSION,

    "checkpoint_filename":
        NOTEBOOK_09_TO_10_CHECKPOINT_FILENAME,

    "checkpoint_drive_file_id":
        NOTEBOOK_09_TO_10_CHECKPOINT_DRIVE_FILE_ID,

    "checkpoint_sha256":
        BLOCK_1_CHECKPOINT_SHA256,

    "metadata_filename":
        NOTEBOOK_09_TO_10_METADATA_FILENAME,

    "metadata_drive_file_id":
        NOTEBOOK_09_TO_10_METADATA_DRIVE_FILE_ID,

    "representation_parameters":
        NOTEBOOK_10_REPRESENTATION_PARAMETER_COUNT,

    "transformative_parameters":
        NOTEBOOK_10_TRANSFORMATIVE_PARAMETER_COUNT,

    "transformative_weight_parameters":
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT,

    "total_parameters":
        NOTEBOOK_10_TOTAL_PARAMETER_COUNT,

    "pathway_dimensions":
        deepcopy(
            NOTEBOOK_10_PATHWAY_DIMENSIONS
        ),

    "active_dimension_indices":
        deepcopy(
            NOTEBOOK_10_ACTIVE_DIMENSION_INDICES
        ),

    "deferred_dimension_indices":
        deepcopy(
            NOTEBOOK_10_DEFERRED_DIMENSION_INDICES
        ),

    "recurrent_contract":
        deepcopy(
            NOTEBOOK_10_RECURRENT_CONTRACT
        ),

    "pilot_pretraining_objective":
        NOTEBOOK_10_NOTEBOOK_09_PILOT_PRETRAIN_OBJECTIVE,

    "pilot_posttraining_objective":
        NOTEBOOK_10_NOTEBOOK_09_PILOT_POSTTRAIN_OBJECTIVE,

    "expanded_corpus_training_required":
        True,

    "handover_valid":
        NOTEBOOK_09_TO_10_HANDOVER_VALID,

    "expanded_corpus_inventory_ready":
        NOTEBOOK_10_EXPANDED_CORPUS_INVENTORY_READY,

    "block_valid":
        NOTEBOOK_10_BLOCK_1_VALID,

    "block_complete":
        NOTEBOOK_10_BLOCK_1_COMPLETE,
}


NOTEBOOK_10_BLOCK_1_STANDALONE_INITIALISATION_VALID = (
    BLOCK_1_PERSISTED_STATE_SCHEMA_VALID
)


# =============================================================================
# Final report
# =============================================================================

print("=" * 72)

print(
    "Media AI — Notebook 10, Block 1: "
    "Notebook 09 → 10 Handover Restoration "
    "and Initialisation Audit"
)

print("=" * 72)

print(
    f"Notebook                     : "
    f"{NOTEBOOK_10}"
)

print(
    f"Notebook version             : "
    f"{NOTEBOOK_10_VERSION}"
)

print(
    f"Block version                : "
    f"{NOTEBOOK_10_BLOCK_1_VERSION}"
)

print(
    f"Project random seed          : "
    f"{NOTEBOOK_10_RANDOM_SEED}"
)

print(
    f"Execution device             : "
    f"{DEVICE}"
)

print("-" * 72)

print(
    "Google authentication"
)

print(
    f"Colab authenticated          : "
    f"{BLOCK_1_COLAB_AUTHENTICATED}"
)

print(
    f"Drive service initialised    : "
    f"{BLOCK_1_DRIVE_SERVICE_INITIALISED}"
)

print("-" * 72)

print(
    "Persistent Notebook 09 → 10 handover"
)

print(
    f"Checkpoint filename          : "
    f"{NOTEBOOK_09_TO_10_CHECKPOINT_FILENAME}"
)

print(
    f"Checkpoint Drive file ID     : "
    f"{NOTEBOOK_09_TO_10_CHECKPOINT_DRIVE_FILE_ID}"
)

print(
    f"Checkpoint located           : "
    f"{BLOCK_1_CHECKPOINT_LOCATED}"
)

print(
    f"Checkpoint downloaded        : "
    f"{BLOCK_1_CHECKPOINT_DOWNLOADED}"
)

print(
    f"Checkpoint SHA-256 valid     : "
    f"{BLOCK_1_CHECKPOINT_SHA256_VALID}"
)

print(
    f"Checkpoint byte count valid  : "
    f"{BLOCK_1_CHECKPOINT_METADATA_BYTE_COUNT_VALID}"
)

print(
    f"Checkpoint size              : "
    f"{len(BLOCK_1_CHECKPOINT_BYTES) / (1024 ** 2):.3f} MB"
)

print(
    f"Metadata filename            : "
    f"{NOTEBOOK_09_TO_10_METADATA_FILENAME}"
)

print(
    f"Metadata Drive file ID       : "
    f"{NOTEBOOK_09_TO_10_METADATA_DRIVE_FILE_ID}"
)

print(
    f"Metadata located             : "
    f"{BLOCK_1_METADATA_LOCATED}"
)

print(
    f"Metadata downloaded          : "
    f"{BLOCK_1_METADATA_DOWNLOADED}"
)

print(
    f"Metadata JSON parsed         : "
    f"{BLOCK_1_METADATA_JSON_PARSED}"
)

print(
    f"Metadata size                : "
    f"{len(BLOCK_1_METADATA_BYTES) / 1024:.3f} KB"
)

print("-" * 72)

print(
    "Restored architecture"
)

print(
    f"Representation parameters    : "
    f"{NOTEBOOK_10_REPRESENTATION_PARAMETER_COUNT:,}"
)

print(
    f"Transformative parameters    : "
    f"{NOTEBOOK_10_TRANSFORMATIVE_PARAMETER_COUNT:,}"
)

print(
    f"Transformative Weight params : "
    f"{NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT}"
)

print(
    f"Combined restored params     : "
    f"{NOTEBOOK_10_TOTAL_PARAMETER_COUNT:,}"
)

print(
    f"Structural weight dimensions : "
    f"{BLOCK_1_EXPECTED_STRUCTURAL_WEIGHT_DIMENSIONS}"
)

print(
    f"Active weight dimensions     : "
    f"{BLOCK_1_EXPECTED_ACTIVE_WEIGHT_DIMENSIONS}"
)

print(
    f"Deferred weight dimensions   : "
    f"{BLOCK_1_EXPECTED_DEFERRED_WEIGHT_DIMENSIONS}"
)

print("-" * 72)

print(
    "Pathway contracts"
)

for pathway_name in BLOCK_1_EXPECTED_PATHWAY_DIMENSIONS:

    print(
        f"{pathway_name:<14} dimension          : "
        f"{NOTEBOOK_10_PATHWAY_DIMENSIONS[pathway_name]}"
    )

    print(
        f"{pathway_name:<14} active dimensions  : "
        f"{list(NOTEBOOK_10_ACTIVE_DIMENSION_INDICES[pathway_name])}"
    )

    print(
        f"{pathway_name:<14} deferred dimensions: "
        f"{list(NOTEBOOK_10_DEFERRED_DIMENSION_INDICES[pathway_name])}"
    )

    print(
        f"{pathway_name:<14} latent source key  : "
        f"{BLOCK_1_LATENT_SOURCE_KEYS[pathway_name]}"
    )

    print(
        f"{pathway_name:<14} weight finite      : "
        f"{BLOCK_1_EFFECTIVE_WEIGHTS_FINITE[pathway_name]}"
    )

    print(
        f"{pathway_name:<14} weight bounded     : "
        f"{BLOCK_1_EFFECTIVE_WEIGHTS_BOUNDED[pathway_name]}"
    )

    print(
        f"{pathway_name:<14} deferred zero      : "
        f"{BLOCK_1_DEFERRED_WEIGHTS_ZERO[pathway_name]}"
    )

print("-" * 72)

print(
    "Recurrent contract"
)

print(
    f"Update type                  : "
    f"{NOTEBOOK_10_RECURRENT_CONTRACT['update_type']}"
)

print(
    f"Initial-state policy         : "
    f"{NOTEBOOK_10_RECURRENT_CONTRACT['initial_state_policy']}"
)

print(
    f"Article-boundary reset       : "
    f"{NOTEBOOK_10_RECURRENT_CONTRACT['article_boundary_reset']}"
)

print(
    f"Cross-article propagation    : "
    f"{NOTEBOOK_10_RECURRENT_CONTRACT['cross_article_propagation']}"
)

print(
    f"Cross-pathway recurrence     : "
    f"{NOTEBOOK_10_RECURRENT_CONTRACT['cross_pathway_recurrence']}"
)

print(
    f"Future context allowed       : "
    f"{NOTEBOOK_10_RECURRENT_CONTRACT['future_context_allowed']}"
)

print("-" * 72)

print(
    "Restoration validation"
)

print(
    f"Metadata identity valid      : "
    f"{BLOCK_1_METADATA_IDENTITY_VALID}"
)

print(
    f"Checkpoint identity valid    : "
    f"{BLOCK_1_CHECKPOINT_IDENTITY_VALID}"
)

print(
    f"Bootstrap contract valid     : "
    f"{BLOCK_1_METADATA_BOOTSTRAP_VALID}"
)

print(
    f"Cross-artefact valid         : "
    f"{BLOCK_1_CROSS_ARTEFACT_VALID}"
)

print(
    f"Parameter counts valid       : "
    f"{BLOCK_1_PARAMETER_COUNTS_VALID}"
)

print(
    f"Pathway contract valid       : "
    f"{BLOCK_1_ALL_DIMENSION_PARTITIONS_VALID}"
)

print(
    f"Representation state exact   : "
    f"{BLOCK_1_REPRESENTATION_STATE_EXACT}"
)

print(
    f"Transformative state exact   : "
    f"{BLOCK_1_TRANSFORMATIVE_STATE_EXACT}"
)

print(
    f"Weight state exact           : "
    f"{BLOCK_1_EFFECTIVE_WEIGHT_IDENTITY_VALID}"
)

print(
    f"Post-device repr. exact      : "
    f"{BLOCK_1_POST_DEVICE_REPRESENTATION_EXACT}"
)

print(
    f"Post-device transform exact  : "
    f"{BLOCK_1_POST_DEVICE_TRANSFORMATIVE_EXACT}"
)

print(
    f"Post-device weight exact     : "
    f"{BLOCK_1_POST_DEVICE_WEIGHT_LATENT_EXACT}"
)

print(
    f"All parameters frozen        : "
    f"{BLOCK_1_ALL_PARAMETERS_FROZEN}"
)

print(
    f"All modules eval             : "
    f"{BLOCK_1_ALL_MODULES_EVAL}"
)

print(
    f"All gradients absent         : "
    f"{BLOCK_1_ALL_GRADIENTS_ABSENT}"
)

print("-" * 72)

print(
    "Inherited pilot boundary"
)

print(
    f"Pilot pretraining objective  : "
    f"{NOTEBOOK_10_NOTEBOOK_09_PILOT_PRETRAIN_OBJECTIVE:.8f}"
)

print(
    f"Pilot posttraining objective : "
    f"{NOTEBOOK_10_NOTEBOOK_09_PILOT_POSTTRAIN_OBJECTIVE:.8f}"
)

print(
    f"Final training claimed       : False"
)

print(
    f"Expanded training required   : True"
)

print("-" * 72)

print(
    "Notebook 10 execution boundary"
)

print(
    f"Expanded corpus loaded       : "
    f"{NOTEBOOK_10_BLOCK_1_EXPANDED_CORPUS_LOADED}"
)

print(
    f"Representation forward       : "
    f"{NOTEBOOK_10_BLOCK_1_REPRESENTATION_FORWARD_EXECUTED}"
)

print(
    f"Transform forward            : "
    f"{NOTEBOOK_10_BLOCK_1_TRANSFORMATIVE_FORWARD_EXECUTED}"
)

print(
    f"Weight forward               : "
    f"{NOTEBOOK_10_BLOCK_1_WEIGHT_FORWARD_EXECUTED}"
)

print(
    f"Recurrent state created      : "
    f"{NOTEBOOK_10_BLOCK_1_RECURRENT_STATE_CREATED}"
)

print(
    f"Recurrent update executed    : "
    f"{NOTEBOOK_10_BLOCK_1_RECURRENT_UPDATE_EXECUTED}"
)

print(
    f"Loss calculated              : "
    f"{NOTEBOOK_10_BLOCK_1_LOSS_CALCULATED}"
)

print(
    f"Optimizer created            : "
    f"{NOTEBOOK_10_BLOCK_1_OPTIMIZER_CREATED}"
)

print(
    f"Backward pass executed       : "
    f"{NOTEBOOK_10_BLOCK_1_BACKWARD_PASS_EXECUTED}"
)

print(
    f"Parameter update executed    : "
    f"{NOTEBOOK_10_BLOCK_1_PARAMETER_UPDATE_EXECUTED}"
)

print(
    f"Training executed            : "
    f"{NOTEBOOK_10_BLOCK_1_TRAINING_EXECUTED}"
)

print("-" * 72)

print(
    f"Notebook 09 → 10 valid       : "
    f"{NOTEBOOK_09_TO_10_HANDOVER_VALID}"
)

print(
    f"Notebook 10 initialised      : "
    f"{NOTEBOOK_10_INITIALISED}"
)

print(
    f"Expanded inventory ready     : "
    f"{NOTEBOOK_10_EXPANDED_CORPUS_INVENTORY_READY}"
)

print(
    f"Block valid                  : "
    f"{NOTEBOOK_10_BLOCK_1_VALID}"
)

print(
    f"Block complete               : "
    f"{NOTEBOOK_10_BLOCK_1_COMPLETE}"
)

print("=" * 72)

print(
    "Notebook 10 initialised successfully from the persistent "
    "Notebook 09 → Notebook 10 handover."
)

print(
    "The canonical Notebook 09 checkpoint was retrieved directly by "
    "Drive file ID and its SHA-256 digest was validated before "
    "deserialisation."
)

print(
    f"The validated {NOTEBOOK_10_REPRESENTATION_PARAMETER_COUNT:,}-parameter "
    "representation architecture, "
    f"{NOTEBOOK_10_TRANSFORMATIVE_PARAMETER_COUNT:,}-parameter transformative "
    f"architecture and {NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT} "
    "learned Transformative-Weight parameter(s) were restored successfully."
)

print(
    f"The complete inherited {NOTEBOOK_10_TOTAL_PARAMETER_COUNT:,}-parameter "
    "architecture is frozen and in evaluation mode at the Notebook 10 "
    "initialisation boundary."
)

print(
    "Factual, psychological and social pathway dimensions and their "
    "active/deferred Transformative-Weight partitions reproduce the "
    "canonical Notebook 09 contract."
)

print(
    "The learned effective Transformative Weights reproduce the "
    "persisted Notebook 09 state, remain finite and bounded, and preserve "
    "all deferred dimensions at exactly zero."
)

print(
    "The additive recurrent contract, deterministic zero article reset, "
    "cross-article isolation, pathway independence and no-future-context "
    "policy were inherited without modification."
)

print(
    "The Notebook 09 controlled optimisation remains explicitly classified "
    "as pilot architectural training rather than final model training."
)

print(
    "No expanded corpus, model forward path, recurrent state, loss, "
    "optimiser, backward pass, parameter update or training was executed "
    "by Block 1."
)

print(
    "Notebook 10 may now proceed to expanded-corpus inventory, "
    "eligibility and multi-article validation-source construction."
)

print("=" * 72)
```

# Block 2 — Expanded-Corpus Inventory, Eligibility and Multi-Article Validation-Source Construction

## Purpose

This block establishes the expanded-corpus input contract required for multi-article validation and subsequent Transformative-Weight training.

Notebook 09 completed the Media AI architecture and validated its recurrent behaviour on a controlled single-article pilot. Block 1 of the present notebook restored that architecture exactly from the persistent Notebook 09 handover and confirmed that expanded-corpus training remains methodologically required.

The present block therefore moves from **architectural restoration** to **expanded-corpus readiness**.

Its purpose is to identify, recover and validate the available article-level textual material and determine which records can participate safely in the next phase of Media AI validation and training.

No model forward execution, recurrent-state propagation, objective calculation or parameter optimisation is performed in this block.

## Expanded-Corpus Role

The controlled Notebook 09 experiment used one canonical article containing 16 sentences.

At the beginning of Notebook 10, the validated runtime may still contain only the single canonical pilot article inherited from the preceding notebooks. This does not invalidate the present inventory block. Instead, the block must distinguish between successful corpus inventory and sufficient corpus expansion. If fewer than two eligible articles are available, expanded-corpus and multi-article readiness must remain false and the requirement for additional corpus construction must be reported explicitly.

That article was sufficient to validate:

- the representation architecture;
- the transformative mechanisms;
- the Transformative-Weight architecture;
- additive recurrent-state propagation;
- article-boundary reset behaviour;
- causal sentence traversal;
- supervision alignment;
- differentiable objective construction;
- controlled Transformative-Weight optimisation.

It is not sufficient for substantive model training or multi-article validation.

Notebook 10 therefore requires a broader collection of independently identifiable articles while preserving the structural and causal contracts already established.

The expanded corpus must retain the hierarchy

**Article → Paragraph → Sentence**

and each sentence must remain traceable to its source article and canonical within-article position.

## Canonical Identity Contract

Every eligible sentence must expose sufficient structural identity to support deterministic reconstruction of its article sequence.

At minimum, the block must be able to establish:

- a unique article identity;
- a unique sentence identity;
- paragraph membership where available;
- sentence position within the canonical article sequence;
- deterministic ordering within each article.

Sentence identities must be unique across the expanded corpus.

Article identities must permit sentences to be grouped without ambiguity, and canonical ordering must be reconstructable without using future information.

The block must not infer article membership from semantic similarity or model output.

## Source Recovery Policy

Expanded-corpus sources must be recovered from validated project artefacts or other explicitly identified corpus sources.

The block should prefer persisted structured artefacts that preserve article and sentence identity over reconstructed or heuristic alternatives.

Source recovery must distinguish between:

1. **textual corpus records**;
2. **sentence and article identity metadata**;
3. **contextual representation artefacts**;
4. **supervision artefacts**.

These components must not be treated as interchangeable.

In particular, supervision targets must never be substituted for textual or representational model inputs.

No synthetic article, sentence, representation or supervision record may be introduced to satisfy an incomplete source contract.

## Article Eligibility Contract

An article is eligible for expanded multi-article processing only when its sentence sequence can be reconstructed deterministically.

For each candidate article, the block should validate:

- article identity;
- non-empty sentence membership;
- valid sentence identities;
- unique sentence identities;
- valid sequence positions;
- unique sequence positions within the article;
- complete deterministic sentence ordering;
- absence of cross-article identity collisions.

Articles failing the structural identity contract must remain explicitly excluded rather than silently repaired.

Eligibility decisions must therefore remain auditable at article level.

## Contextual Representation Availability

Where persisted contextual sentence representations are available, they must be aligned to the expanded canonical sentence inventory through sentence identity rather than row position alone.

The representation contract inherited from the validated project architecture is:

$$
\mathbf{x}_{i,t} \in \mathbb{R}^{768}
$$

where $\mathbf{x}_{i,t}$ denotes the contextual representation of sentence $t$ in article $i$.

For each available contextual representation, the block should validate:

- dimensionality of 768;
- numerical finiteness;
- sentence-identity alignment;
- article-identity consistency;
- canonical sequence compatibility;
- absence of duplicated representation records.

Representation availability and article structural eligibility must be reported separately.

An article may therefore be structurally valid while remaining unavailable for later model execution because one or more required sentence representations are missing.

## Supervision Availability Inventory

The expanded-corpus inventory should also determine whether validated supervision exists for the three Media AI pathways:

**Factual → Psychological → Social**

For article $i$ and sentence $t$, pathway supervision may be represented as

$$
\mathbf{y}^{(p)}_{i,t} \in \mathbb{R}^{d_p}
$$

with pathway dimensions

$$
d_F = 10, \qquad d_P = 34, \qquad d_S = 7.
$$

Supervision availability may remain partial.

Missing supervision must remain explicitly masked and must not be converted into zero-valued targets.

The block should therefore distinguish between:

- structurally eligible sentences;
- representation-ready sentences;
- factual-supervised dimensions;
- psychological-supervised dimensions;
- social-supervised dimensions.

No requirement is imposed at this stage that every sentence contain complete supervision across every pathway.

## Multi-Article Boundary Contract

Each article constitutes an independent recurrent sequence.

For article $i$, the recurrent state must eventually begin from the deterministic zero condition

$$
\mathbf{h}^{(p)}_{i,0} = \mathbf{0},
$$

for each pathway $p$.

No recurrent state may propagate from article $i$ into article $i+1$.

The expanded-corpus inventory must therefore preserve explicit article boundaries so that later blocks can enforce

$$
\mathbf{h}^{(p)}_{i+1,0} = \mathbf{0}
$$

independently of the final recurrent state of the preceding article.

This block validates that such boundaries can be reconstructed but does **not** create or propagate recurrent states.

## Canonical Multi-Article Ordering

Sentence order must be preserved within each article.

Ordering between independent articles does not define recurrent context and must not create cross-article dependence.

Consequently:

- sentence shuffling within an article is prohibited;
- cross-article recurrent propagation is prohibited;
- future sentences must not influence earlier sentence states;
- article-level batching may only be introduced later if these constraints remain preserved.

The present block should construct a deterministic multi-article inventory suitable for subsequent controlled execution.

## Eligibility Tiers

To avoid conflating different forms of readiness, the block should distinguish at least three levels:

**Structurally eligible**

The article has a complete and deterministic canonical sentence sequence.

**Representation ready**

The structurally eligible article has a valid aligned 768-dimensional contextual representation for every required sentence.

**Supervision eligible**

The representation-ready article contains sufficient validated supervision for at least one currently active Transformative-Weight objective.

These categories should remain explicit so that missing supervision does not incorrectly invalidate otherwise usable corpus material.

## Validation Requirements

The block should confirm that:

- expanded corpus sources can be identified;
- article identities are valid;
- sentence identities are valid and globally unique;
- canonical within-article ordering is deterministic;
- article boundaries are explicit;
- eligible articles contain non-empty sentence sequences;
- available contextual representations align by sentence identity;
- representation dimensionality remains 768;
- available representation values are finite;
- supervision availability can be inventoried without substitution;
- missing supervision remains distinguishable from genuine zero values;
- no synthetic fallback is used;
- no future context is introduced;
- no cross-article recurrent relationship is constructed.

The resulting inventory must be sufficient to identify the subset of articles that can proceed safely to controlled multi-article representation and recurrent validation.

## Execution Boundary

This block performs only corpus recovery, identity validation, alignment inspection, eligibility classification and multi-article source construction.

It does **not**:

- execute the frozen representation architecture;
- execute a transformative mechanism;
- execute a Transformative-Weight forward pass;
- construct recurrent states;
- propagate recurrent states;
- calculate a training objective;
- create an optimiser;
- execute a backward pass;
- update model parameters;
- begin expanded-corpus training.

The complete 902,215-parameter architecture restored by Block 1 therefore remains frozen and unchanged.

## Expected Outcome

At completion, Notebook 10 should possess a validated expanded-corpus inventory in which every candidate article has an explicit structural and modelling-readiness status.

The block should report:

- number of candidate articles;
- number of candidate sentences;
- structurally eligible articles and sentences;
- representation-ready articles and sentences;
- supervision-eligible articles and sentences;
- excluded articles and exclusion reasons;
- contextual representation coverage;
- factual supervision coverage;
- psychological supervision coverage;
- social supervision coverage;
- article-boundary validity;
- canonical sequence validity;
- multi-article validation readiness.

Successful completion establishes the source contract required for the next stage:

**controlled expanded-corpus representation and multi-article recurrent validation.**

```python
# =============================================================================
# Media AI — Notebook 10
# Block 2: Expanded-Corpus Inventory, Eligibility
#          and Multi-Article Validation-Source Construction
# =============================================================================

from copy import deepcopy
from datetime import datetime, timezone

import io
import json
import math

import numpy as np
import torch

from googleapiclient.http import MediaIoBaseDownload


# =============================================================================
# Block identity
# =============================================================================

NOTEBOOK_10_BLOCK_2 = 2

NOTEBOOK_10_BLOCK_2_NAME = (
    "Expanded-Corpus Inventory, Eligibility "
    "and Multi-Article Validation-Source Construction"
)

NOTEBOOK_10_BLOCK_2_VERSION = "1.1"

BLOCK_2_EXECUTED_AT_UTC = (
    datetime.now(
        timezone.utc
    ).isoformat()
)


# =============================================================================
# Block 1 dependency
# =============================================================================

BLOCK_2_BLOCK_1_DEPENDENCY_VALID = all(
    [
        globals().get(
            "NOTEBOOK_10_BLOCK_1_COMPLETE",
            False,
        )
        is True,

        globals().get(
            "NOTEBOOK_10_BLOCK_1_VALID",
            False,
        )
        is True,

        globals().get(
            "NOTEBOOK_10_INITIALISED",
            False,
        )
        is True,

        globals().get(
            "NOTEBOOK_09_TO_10_HANDOVER_VALID",
            False,
        )
        is True,
    ]
)


if not BLOCK_2_BLOCK_1_DEPENDENCY_VALID:

    raise RuntimeError(
        "Notebook 10 Block 2 requires a valid completed Block 1 "
        "Notebook 09 → 10 handover."
    )


# =============================================================================
# Immutable Block 1 model-state baseline
# =============================================================================
#
# Block 1 reconstructs the Notebook 09 Transformative-Weight modules using a
# canonical Notebook 10 parameter name (`alpha`) even though the persisted
# Notebook 09 state may use a different implementation key such as
# `alpha_active`.
#
# Block 2 must therefore validate model immutability against the restored
# Block 1 runtime state, not by comparing Notebook 10 module state-dict keys
# directly with the raw Notebook 09 checkpoint implementation keys.
# =============================================================================

def block_2_snapshot_module_collection(
    module_collection,
):

    return {
        module_name:
            {
                tensor_name:
                    tensor.detach()
                    .cpu()
                    .clone()

                for (
                    tensor_name,
                    tensor,
                ) in module.state_dict().items()
            }

        for (
            module_name,
            module,
        ) in module_collection.items()
    }


def block_2_nested_state_exact(
    state_before,
    state_after,
):

    if state_before.keys() != state_after.keys():

        return False


    for module_name in state_before:

        if (
            state_before[
                module_name
            ].keys()
            !=
            state_after[
                module_name
            ].keys()
        ):

            return False


        for tensor_name in state_before[
            module_name
        ]:

            if not torch.equal(
                state_before[
                    module_name
                ][
                    tensor_name
                ],
                state_after[
                    module_name
                ][
                    tensor_name
                ],
            ):

                return False


    return True


BLOCK_2_REPRESENTATION_STATE_BASELINE = (
    block_2_snapshot_module_collection(
        NOTEBOOK_10_REPRESENTATION_MODULES
    )
)


BLOCK_2_TRANSFORMATIVE_STATE_BASELINE = (
    block_2_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_MODULES
    )
)


BLOCK_2_WEIGHT_STATE_BASELINE = (
    block_2_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES
    )
)


# =============================================================================
# Canonical inherited architecture contract
# =============================================================================

BLOCK_2_PATHWAY_DIMENSIONS = deepcopy(
    NOTEBOOK_10_PATHWAY_DIMENSIONS
)


BLOCK_2_ACTIVE_DIMENSION_INDICES = deepcopy(
    NOTEBOOK_10_ACTIVE_DIMENSION_INDICES
)


BLOCK_2_DEFERRED_DIMENSION_INDICES = deepcopy(
    NOTEBOOK_10_DEFERRED_DIMENSION_INDICES
)


BLOCK_2_EXPECTED_EMBEDDING_DIMENSION = 768

BLOCK_2_MINIMUM_MULTI_ARTICLE_COUNT = 2


# =============================================================================
# Execution boundary
# =============================================================================

NOTEBOOK_10_BLOCK_2_REPRESENTATION_FORWARD_EXECUTED = False

NOTEBOOK_10_BLOCK_2_TRANSFORMATIVE_FORWARD_EXECUTED = False

NOTEBOOK_10_BLOCK_2_WEIGHT_FORWARD_EXECUTED = False

NOTEBOOK_10_BLOCK_2_RECURRENT_STATE_CREATED = False

NOTEBOOK_10_BLOCK_2_RECURRENT_UPDATE_EXECUTED = False

NOTEBOOK_10_BLOCK_2_LOSS_CALCULATED = False

NOTEBOOK_10_BLOCK_2_OPTIMIZER_CREATED = False

NOTEBOOK_10_BLOCK_2_ZERO_GRAD_EXECUTED = False

NOTEBOOK_10_BLOCK_2_BACKWARD_PASS_EXECUTED = False

NOTEBOOK_10_BLOCK_2_PARAMETER_UPDATE_EXECUTED = False

NOTEBOOK_10_BLOCK_2_TRAINING_EXECUTED = False


# =============================================================================
# General helpers
# =============================================================================

def block_2_first_present(
    mapping,
    keys,
):

    if not isinstance(
        mapping,
        dict,
    ):

        return None


    for key in keys:

        if (
            key in mapping
            and
            mapping[key] is not None
        ):

            return mapping[key]


    return None


def block_2_normalise_identifier(
    value,
):

    if value is None:

        return None


    if isinstance(
        value,
        np.generic,
    ):

        value = value.item()


    value = str(
        value
    ).strip()


    if not value:

        return None


    return value


def block_2_normalise_position(
    value,
):

    if value is None:

        return None


    if isinstance(
        value,
        np.generic,
    ):

        value = value.item()


    if isinstance(
        value,
        bool,
    ):

        return None


    try:

        numeric_value = float(
            value
        )

    except (
        TypeError,
        ValueError,
    ):

        return None


    if not math.isfinite(
        numeric_value
    ):

        return None


    if not numeric_value.is_integer():

        return None


    return int(
        numeric_value
    )


def block_2_to_numpy_vector(
    value,
):

    if value is None:

        return None


    if torch.is_tensor(
        value
    ):

        array = (
            value.detach()
            .cpu()
            .numpy()
        )

    else:

        try:

            array = np.asarray(
                value,
                dtype=
                    np.float32,
            )

        except (
            TypeError,
            ValueError,
        ):

            return None


    array = np.asarray(
        array
    )


    if (
        array.ndim
        ==
        2
        and
        1
        in array.shape
    ):

        array = array.reshape(
            -1
        )


    if array.ndim != 1:

        return None


    return array.astype(
        np.float32,
        copy=
            False,
    )


def block_2_to_numpy_target(
    value,
    dimension,
):

    array = block_2_to_numpy_vector(
        value
    )


    if array is None:

        return None


    if array.shape != (
        dimension,
    ):

        return None


    return array


def block_2_extract_records(
    payload,
):

    if isinstance(
        payload,
        list,
    ):

        if all(
            isinstance(
                record,
                dict,
            )

            for record
            in payload
        ):

            return payload


        return None


    if not isinstance(
        payload,
        dict,
    ):

        return None


    candidate_keys = (
        "records",
        "sentence_records",
        "sentences",
        "items",
        "data",
        "contextual_sentence_representations",
        "representations",
    )


    for key in candidate_keys:

        candidate = payload.get(
            key
        )


        if (
            isinstance(
                candidate,
                list,
            )
            and
            all(
                isinstance(
                    record,
                    dict,
                )

                for record
                in candidate
            )
        ):

            return candidate


    return None


def block_2_download_drive_json(
    drive_service,
    file_id,
):

    request = (
        drive_service.files()
        .get_media(
            fileId=
                file_id
        )
    )


    buffer = io.BytesIO()


    downloader = MediaIoBaseDownload(
        buffer,
        request,
    )


    complete = False


    while not complete:

        _, complete = downloader.next_chunk()


    buffer.seek(
        0
    )


    raw_bytes = buffer.read()


    if not raw_bytes:

        raise RuntimeError(
            f"Drive JSON artefact {file_id} is empty."
        )


    return json.loads(
        raw_bytes.decode(
            "utf-8"
        )
    )


# =============================================================================
# Runtime source discovery
# =============================================================================
#
# Block 2 inventories existing structured corpus material only.
#
# It does not use:
#   - supervision as textual input;
#   - supervision as representation input;
#   - synthetic sentence records;
#   - semantic reconstruction of article identity.
#
# Candidate runtime variables are inspected conservatively.
# =============================================================================

BLOCK_2_RUNTIME_SOURCE_NAMES = (
    # Notebook 10 / Notebook 09 likely canonical sources
    "NOTEBOOK_10_CANONICAL_SENTENCE_RECORDS",
    "NOTEBOOK_09_CANONICAL_SENTENCE_RECORDS",
    "NOTEBOOK_09_ARTICLE_SENTENCE_RECORDS",
    "NOTEBOOK_09_CONTEXTUAL_SENTENCE_RECORDS",

    # Notebook 05 contextual artefact runtime variants
    "NOTEBOOK_05_CONTEXTUAL_SENTENCE_REPRESENTATIONS",
    "NOTEBOOK_05_CONTEXTUAL_SENTENCE_RECORDS",
    "NOTEBOOK_05_CONTEXTUAL_REPRESENTATION_RECORDS",

    # Generic validated project names
    "CONTEXTUAL_SENTENCE_REPRESENTATIONS",
    "CONTEXTUAL_SENTENCE_RECORDS",
    "CANONICAL_SENTENCE_RECORDS",
)


BLOCK_2_RUNTIME_SOURCES_INSPECTED = []

BLOCK_2_RUNTIME_RECORD_COLLECTIONS = []


for source_name in BLOCK_2_RUNTIME_SOURCE_NAMES:

    if source_name not in globals():

        continue


    source_payload = globals()[
        source_name
    ]


    source_records = block_2_extract_records(
        source_payload
    )


    BLOCK_2_RUNTIME_SOURCES_INSPECTED.append(
        {
            "source_name":
                source_name,

            "payload_type":
                type(
                    source_payload
                ).__name__,

            "records_found":
                (
                    len(
                        source_records
                    )
                    if source_records is not None
                    else 0
                ),
        }
    )


    if source_records:

        BLOCK_2_RUNTIME_RECORD_COLLECTIONS.append(
            {
                "source_name":
                    source_name,

                "records":
                    source_records,
            }
        )


# =============================================================================
# Known validated contextual representation artefact
# =============================================================================
#
# Notebook 09 Block 8 recovered the canonical Notebook 05 contextual
# sentence representation artefact:
#
#   notebook_05_contextual_sentence_representations.json
#
# Drive file ID:
#   11_j_CebqBpMSDLlccGwGROr5hklTBYRm
#
# This is a validated project artefact, not a synthetic fallback.
#
# It is used here only if a suitable runtime collection has not already
# been recovered.
# =============================================================================

BLOCK_2_NOTEBOOK_05_CONTEXTUAL_FILENAME = (
    "notebook_05_contextual_sentence_representations.json"
)

BLOCK_2_NOTEBOOK_05_CONTEXTUAL_DRIVE_FILE_ID = (
    "11_j_CebqBpMSDLlccGwGROr5hklTBYRm"
)


BLOCK_2_PERSISTED_CONTEXTUAL_SOURCE_USED = False

BLOCK_2_PERSISTED_CONTEXTUAL_PAYLOAD = None

BLOCK_2_PERSISTED_CONTEXTUAL_RECORDS = None


if not BLOCK_2_RUNTIME_RECORD_COLLECTIONS:

    if (
        "NOTEBOOK_10_DRIVE_SERVICE"
        not in globals()
    ):

        raise RuntimeError(
            "Notebook 10 Drive service from Block 1 is unavailable."
        )


    BLOCK_2_PERSISTED_CONTEXTUAL_PAYLOAD = (
        block_2_download_drive_json(
            NOTEBOOK_10_DRIVE_SERVICE,
            BLOCK_2_NOTEBOOK_05_CONTEXTUAL_DRIVE_FILE_ID,
        )
    )


    BLOCK_2_PERSISTED_CONTEXTUAL_RECORDS = (
        block_2_extract_records(
            BLOCK_2_PERSISTED_CONTEXTUAL_PAYLOAD
        )
    )


    if not BLOCK_2_PERSISTED_CONTEXTUAL_RECORDS:

        raise RuntimeError(
            "Notebook 10 Block 2 could not recover any structured "
            "sentence records from the validated Notebook 05 "
            "contextual representation artefact."
        )


    BLOCK_2_RUNTIME_RECORD_COLLECTIONS.append(
        {
            "source_name":
                BLOCK_2_NOTEBOOK_05_CONTEXTUAL_FILENAME,

            "records":
                BLOCK_2_PERSISTED_CONTEXTUAL_RECORDS,
        }
    )


    BLOCK_2_PERSISTED_CONTEXTUAL_SOURCE_USED = True


# =============================================================================
# Select the strongest available structured source
# =============================================================================
#
# Selection is based only on explicit structural information:
# article identity, sentence identity, sequence position and representation.
#
# No semantic scoring is used.
# =============================================================================

BLOCK_2_ARTICLE_ID_KEYS = (
    "article_id",
    "article_uuid",
    "document_id",
    "source_article_id",
)


BLOCK_2_SENTENCE_ID_KEYS = (
    "sentence_id",
    "sentence_uuid",
    "id",
)


BLOCK_2_PARAGRAPH_ID_KEYS = (
    "paragraph_id",
    "paragraph_uuid",
)


BLOCK_2_SEQUENCE_POSITION_KEYS = (
    "sequence_position",
    "sentence_position",
    "sentence_index",
    "position",
    "sequence_index",
)


BLOCK_2_REPRESENTATION_KEYS = (
    "contextual_representation",
    "contextual_embedding",
    "representation",
    "embedding",
    "sentence_embedding",
    "vector",
)


def block_2_collection_score(
    records,
):

    if not records:

        return (
            0,
            0,
            0,
            0,
            0,
        )


    article_count = 0

    sentence_count = 0

    position_count = 0

    representation_count = 0


    for record in records:

        article_id = block_2_normalise_identifier(
            block_2_first_present(
                record,
                BLOCK_2_ARTICLE_ID_KEYS,
            )
        )


        sentence_id = block_2_normalise_identifier(
            block_2_first_present(
                record,
                BLOCK_2_SENTENCE_ID_KEYS,
            )
        )


        sequence_position = block_2_normalise_position(
            block_2_first_present(
                record,
                BLOCK_2_SEQUENCE_POSITION_KEYS,
            )
        )


        representation = block_2_to_numpy_vector(
            block_2_first_present(
                record,
                BLOCK_2_REPRESENTATION_KEYS,
            )
        )


        article_count += int(
            article_id
            is not None
        )


        sentence_count += int(
            sentence_id
            is not None
        )


        position_count += int(
            sequence_position
            is not None
        )


        representation_count += int(
            representation is not None
            and
            representation.shape
            ==
            (
                BLOCK_2_EXPECTED_EMBEDDING_DIMENSION,
            )
        )


    return (
        article_count,
        sentence_count,
        position_count,
        representation_count,
        len(
            records
        ),
    )


BLOCK_2_SOURCE_CANDIDATES = []


for collection in BLOCK_2_RUNTIME_RECORD_COLLECTIONS:

    score = block_2_collection_score(
        collection[
            "records"
        ]
    )


    BLOCK_2_SOURCE_CANDIDATES.append(
        {
            "source_name":
                collection[
                    "source_name"
                ],

            "records":
                collection[
                    "records"
                ],

            "score":
                score,
        }
    )


if not BLOCK_2_SOURCE_CANDIDATES:

    raise RuntimeError(
        "Notebook 10 Block 2 found no candidate structured corpus source."
    )


BLOCK_2_SOURCE_CANDIDATES = sorted(
    BLOCK_2_SOURCE_CANDIDATES,
    key=
        lambda item:
            item[
                "score"
            ],
    reverse=
        True,
)


BLOCK_2_SELECTED_SOURCE = (
    BLOCK_2_SOURCE_CANDIDATES[
        0
    ]
)


BLOCK_2_SELECTED_SOURCE_NAME = (
    BLOCK_2_SELECTED_SOURCE[
        "source_name"
    ]
)


BLOCK_2_SELECTED_RAW_RECORDS = (
    BLOCK_2_SELECTED_SOURCE[
        "records"
    ]
)


# =============================================================================
# Canonical sentence inventory construction
# =============================================================================

BLOCK_2_CANONICAL_RECORDS = []

BLOCK_2_RECORD_EXCLUSIONS = []


for raw_index, raw_record in enumerate(
    BLOCK_2_SELECTED_RAW_RECORDS
):

    article_id = block_2_normalise_identifier(
        block_2_first_present(
            raw_record,
            BLOCK_2_ARTICLE_ID_KEYS,
        )
    )


    sentence_id = block_2_normalise_identifier(
        block_2_first_present(
            raw_record,
            BLOCK_2_SENTENCE_ID_KEYS,
        )
    )


    paragraph_id = block_2_normalise_identifier(
        block_2_first_present(
            raw_record,
            BLOCK_2_PARAGRAPH_ID_KEYS,
        )
    )


    sequence_position = block_2_normalise_position(
        block_2_first_present(
            raw_record,
            BLOCK_2_SEQUENCE_POSITION_KEYS,
        )
    )


    representation = block_2_to_numpy_vector(
        block_2_first_present(
            raw_record,
            BLOCK_2_REPRESENTATION_KEYS,
        )
    )


    exclusion_reasons = []


    if article_id is None:

        exclusion_reasons.append(
            "missing_article_identity"
        )


    if sentence_id is None:

        exclusion_reasons.append(
            "missing_sentence_identity"
        )


    if sequence_position is None:

        exclusion_reasons.append(
            "missing_or_invalid_sequence_position"
        )


    if exclusion_reasons:

        BLOCK_2_RECORD_EXCLUSIONS.append(
            {
                "raw_index":
                    raw_index,

                "sentence_id":
                    sentence_id,

                "article_id":
                    article_id,

                "reasons":
                    exclusion_reasons,
            }
        )


        continue


    representation_shape_valid = (
        representation is not None
        and
        representation.shape
        ==
        (
            BLOCK_2_EXPECTED_EMBEDDING_DIMENSION,
        )
    )


    representation_finite = (
        representation_shape_valid
        and
        bool(
            np.isfinite(
                representation
            ).all()
        )
    )


    BLOCK_2_CANONICAL_RECORDS.append(
        {
            "article_id":
                article_id,

            "paragraph_id":
                paragraph_id,

            "sentence_id":
                sentence_id,

            "sequence_position":
                sequence_position,

            "contextual_representation":
                (
                    representation.copy()
                    if representation is not None
                    else None
                ),

            "representation_shape_valid":
                bool(
                    representation_shape_valid
                ),

            "representation_finite":
                bool(
                    representation_finite
                ),

            "source_name":
                BLOCK_2_SELECTED_SOURCE_NAME,

            "source_row":
                raw_index,
        }
    )


if not BLOCK_2_CANONICAL_RECORDS:

    raise RuntimeError(
        "Notebook 10 Block 2 recovered a structured source, "
        "but no records satisfied the minimum article/sentence/"
        "sequence identity contract."
    )


# =============================================================================
# Global sentence-identity audit
# =============================================================================

BLOCK_2_SENTENCE_IDS = [
    record[
        "sentence_id"
    ]

    for record
    in BLOCK_2_CANONICAL_RECORDS
]


BLOCK_2_SENTENCE_IDS_VALID = all(
    sentence_id is not None

    for sentence_id
    in BLOCK_2_SENTENCE_IDS
)


BLOCK_2_SENTENCE_IDS_GLOBALLY_UNIQUE = (
    len(
        BLOCK_2_SENTENCE_IDS
    )
    ==
    len(
        set(
            BLOCK_2_SENTENCE_IDS
        )
    )
)


# =============================================================================
# Group records by article
# =============================================================================

BLOCK_2_ARTICLE_RECORDS = {}


for record in BLOCK_2_CANONICAL_RECORDS:

    BLOCK_2_ARTICLE_RECORDS.setdefault(
        record[
            "article_id"
        ],
        [],
    ).append(
        record
    )


BLOCK_2_CANDIDATE_ARTICLE_COUNT = len(
    BLOCK_2_ARTICLE_RECORDS
)


BLOCK_2_CANDIDATE_SENTENCE_COUNT = len(
    BLOCK_2_CANONICAL_RECORDS
)


# =============================================================================
# Article-level structural eligibility
# =============================================================================

BLOCK_2_ARTICLE_AUDIT = {}

BLOCK_2_STRUCTURALLY_ELIGIBLE_ARTICLES = {}

BLOCK_2_STRUCTURALLY_EXCLUDED_ARTICLES = {}


for (
    article_id,
    article_records,
) in BLOCK_2_ARTICLE_RECORDS.items():

    sentence_ids = [
        record[
            "sentence_id"
        ]

        for record
        in article_records
    ]


    sequence_positions = [
        record[
            "sequence_position"
        ]

        for record
        in article_records
    ]


    non_empty = (
        len(
            article_records
        )
        >
        0
    )


    sentence_ids_valid = all(
        sentence_id is not None

        for sentence_id
        in sentence_ids
    )


    sentence_ids_unique = (
        len(
            sentence_ids
        )
        ==
        len(
            set(
                sentence_ids
            )
        )
    )


    sequence_positions_valid = all(
        isinstance(
            position,
            int,
        )
        and
        position
        >=
        0

        for position
        in sequence_positions
    )


    sequence_positions_unique = (
        len(
            sequence_positions
        )
        ==
        len(
            set(
                sequence_positions
            )
        )
    )


    sorted_records = sorted(
        article_records,
        key=
            lambda record:
                (
                    record[
                        "sequence_position"
                    ],
                    record[
                        "sentence_id"
                    ],
                ),
    )


    sorted_positions = [
        record[
            "sequence_position"
        ]

        for record
        in sorted_records
    ]


    if sorted_positions:

        zero_based_contiguous = (
            sorted_positions
            ==
            list(
                range(
                    len(
                        sorted_positions
                    )
                )
            )
        )


        one_based_contiguous = (
            sorted_positions
            ==
            list(
                range(
                    1,
                    len(
                        sorted_positions
                    )
                    +
                    1
                )
            )
        )

    else:

        zero_based_contiguous = False

        one_based_contiguous = False


    canonical_order_valid = (
        zero_based_contiguous
        or
        one_based_contiguous
    )


    if zero_based_contiguous:

        sequence_position_base = (
            "zero_based"
        )

    elif one_based_contiguous:

        sequence_position_base = (
            "one_based"
        )

    else:

        sequence_position_base = (
            None
        )


    exclusion_reasons = []


    if not non_empty:

        exclusion_reasons.append(
            "empty_article"
        )


    if not sentence_ids_valid:

        exclusion_reasons.append(
            "invalid_sentence_identity"
        )


    if not sentence_ids_unique:

        exclusion_reasons.append(
            "duplicate_sentence_identity_within_article"
        )


    if not sequence_positions_valid:

        exclusion_reasons.append(
            "invalid_sequence_position"
        )


    if not sequence_positions_unique:

        exclusion_reasons.append(
            "duplicate_sequence_position"
        )


    if not canonical_order_valid:

        exclusion_reasons.append(
            "non_contiguous_canonical_sequence"
        )


    structurally_eligible = (
        len(
            exclusion_reasons
        )
        ==
        0
    )


    BLOCK_2_ARTICLE_AUDIT[
        article_id
    ] = {
        "article_id":
            article_id,

        "sentence_count":
            len(
                article_records
            ),

        "sentence_ids_valid":
            sentence_ids_valid,

        "sentence_ids_unique":
            sentence_ids_unique,

        "sequence_positions_valid":
            sequence_positions_valid,

        "sequence_positions_unique":
            sequence_positions_unique,

        "canonical_order_valid":
            canonical_order_valid,

        "sequence_position_base":
            sequence_position_base,

        "structurally_eligible":
            structurally_eligible,

        "exclusion_reasons":
            exclusion_reasons,
    }


    if structurally_eligible:

        BLOCK_2_STRUCTURALLY_ELIGIBLE_ARTICLES[
            article_id
        ] = sorted_records

    else:

        BLOCK_2_STRUCTURALLY_EXCLUDED_ARTICLES[
            article_id
        ] = exclusion_reasons


# =============================================================================
# Cross-article identity collision audit
# =============================================================================

BLOCK_2_SENTENCE_TO_ARTICLE = {}

BLOCK_2_CROSS_ARTICLE_SENTENCE_COLLISIONS = {}


for (
    article_id,
    article_records,
) in BLOCK_2_ARTICLE_RECORDS.items():

    for record in article_records:

        sentence_id = record[
            "sentence_id"
        ]


        if sentence_id in BLOCK_2_SENTENCE_TO_ARTICLE:

            previous_article_id = (
                BLOCK_2_SENTENCE_TO_ARTICLE[
                    sentence_id
                ]
            )


            if (
                previous_article_id
                !=
                article_id
            ):

                BLOCK_2_CROSS_ARTICLE_SENTENCE_COLLISIONS[
                    sentence_id
                ] = {
                    previous_article_id,
                    article_id,
                }

        else:

            BLOCK_2_SENTENCE_TO_ARTICLE[
                sentence_id
            ] = article_id


BLOCK_2_NO_CROSS_ARTICLE_IDENTITY_COLLISIONS = (
    len(
        BLOCK_2_CROSS_ARTICLE_SENTENCE_COLLISIONS
    )
    ==
    0
)


# =============================================================================
# Representation readiness
# =============================================================================

BLOCK_2_REPRESENTATION_READY_ARTICLES = {}

BLOCK_2_REPRESENTATION_NOT_READY_ARTICLES = {}


for (
    article_id,
    article_records,
) in BLOCK_2_STRUCTURALLY_ELIGIBLE_ARTICLES.items():

    representation_complete = all(
        record[
            "contextual_representation"
        ]
        is not None

        for record
        in article_records
    )


    representation_shapes_valid = all(
        record[
            "representation_shape_valid"
        ]

        for record
        in article_records
    )


    representation_values_finite = all(
        record[
            "representation_finite"
        ]

        for record
        in article_records
    )


    representation_ready = all(
        [
            representation_complete,
            representation_shapes_valid,
            representation_values_finite,
        ]
    )


    BLOCK_2_ARTICLE_AUDIT[
        article_id
    ][
        "representation_complete"
    ] = representation_complete


    BLOCK_2_ARTICLE_AUDIT[
        article_id
    ][
        "representation_shapes_valid"
    ] = representation_shapes_valid


    BLOCK_2_ARTICLE_AUDIT[
        article_id
    ][
        "representation_values_finite"
    ] = representation_values_finite


    BLOCK_2_ARTICLE_AUDIT[
        article_id
    ][
        "representation_ready"
    ] = representation_ready


    if representation_ready:

        BLOCK_2_REPRESENTATION_READY_ARTICLES[
            article_id
        ] = article_records

    else:

        reasons = []


        if not representation_complete:

            reasons.append(
                "missing_contextual_representation"
            )


        if not representation_shapes_valid:

            reasons.append(
                "invalid_contextual_representation_shape"
            )


        if not representation_values_finite:

            reasons.append(
                "non_finite_contextual_representation"
            )


        BLOCK_2_REPRESENTATION_NOT_READY_ARTICLES[
            article_id
        ] = reasons


# =============================================================================
# Representation matrices
# =============================================================================

BLOCK_2_ARTICLE_REPRESENTATION_MATRICES = {}


for (
    article_id,
    article_records,
) in BLOCK_2_REPRESENTATION_READY_ARTICLES.items():

    matrix = np.stack(
        [
            record[
                "contextual_representation"
            ]

            for record
            in article_records
        ],
        axis=
            0,
    ).astype(
        np.float32,
        copy=
            False,
    )


    if (
        matrix.ndim
        !=
        2
        or
        matrix.shape[
            1
        ]
        !=
        BLOCK_2_EXPECTED_EMBEDDING_DIMENSION
    ):

        raise RuntimeError(
            f"Article {article_id} produced an invalid "
            "contextual representation matrix."
        )


    if not np.isfinite(
        matrix
    ).all():

        raise RuntimeError(
            f"Article {article_id} produced non-finite "
            "contextual representation values."
        )


    BLOCK_2_ARTICLE_REPRESENTATION_MATRICES[
        article_id
    ] = matrix


# =============================================================================
# Optional supervision inventory
# =============================================================================
#
# Block 2 does not require complete supervision.
#
# It inventories supervision only when it is already exposed in the selected
# sentence records. It does not use supervision to construct article identity
# or contextual representations.
#
# More specialised persisted supervision artefacts may be aligned in a later
# dedicated block if the expanded corpus requires them.
# =============================================================================

BLOCK_2_SUPERVISION_KEYS = {
    "factual":
        (
            "factual_supervision",
            "factual_target",
            "factual_targets",
            "factual",
        ),

    "psychological":
        (
            "psychological_supervision",
            "psychological_target",
            "psychological_targets",
            "psychological",
        ),

    "social":
        (
            "social_supervision",
            "social_target",
            "social_targets",
            "social",
        ),
}


BLOCK_2_SUPERVISION_MASK_KEYS = {
    "factual":
        (
            "factual_mask",
            "factual_availability",
            "factual_available",
        ),

    "psychological":
        (
            "psychological_mask",
            "psychological_availability",
            "psychological_available",
        ),

    "social":
        (
            "social_mask",
            "social_availability",
            "social_available",
        ),
}


BLOCK_2_SUPERVISION_AVAILABLE_DIMENSIONS = {
    pathway_name:
        0

    for pathway_name
    in BLOCK_2_PATHWAY_DIMENSIONS
}


BLOCK_2_SUPERVISION_TOTAL_DIMENSIONS = {
    pathway_name:
        (
            BLOCK_2_CANDIDATE_SENTENCE_COUNT
            *
            BLOCK_2_PATHWAY_DIMENSIONS[
                pathway_name
            ]
        )

    for pathway_name
    in BLOCK_2_PATHWAY_DIMENSIONS
}


BLOCK_2_SENTENCE_SUPERVISION = {}


for record in BLOCK_2_CANONICAL_RECORDS:

    raw_record = (
        BLOCK_2_SELECTED_RAW_RECORDS[
            record[
                "source_row"
            ]
        ]
    )


    sentence_id = record[
        "sentence_id"
    ]


    BLOCK_2_SENTENCE_SUPERVISION[
        sentence_id
    ] = {}


    for pathway_name in (
        "factual",
        "psychological",
        "social",
    ):

        dimension = (
            BLOCK_2_PATHWAY_DIMENSIONS[
                pathway_name
            ]
        )


        raw_target = block_2_first_present(
            raw_record,
            BLOCK_2_SUPERVISION_KEYS[
                pathway_name
            ],
        )


        target = block_2_to_numpy_target(
            raw_target,
            dimension,
        )


        raw_mask = block_2_first_present(
            raw_record,
            BLOCK_2_SUPERVISION_MASK_KEYS[
                pathway_name
            ],
        )


        if target is None:

            target_array = np.full(
                dimension,
                np.nan,
                dtype=
                    np.float32,
            )


            availability_mask = np.zeros(
                dimension,
                dtype=
                    bool,
            )

        else:

            target_array = target.astype(
                np.float32,
                copy=
                    True,
            )


            if raw_mask is None:

                availability_mask = np.isfinite(
                    target_array
                )

            else:

                mask_array = np.asarray(
                    raw_mask
                ).reshape(
                    -1
                )


                if mask_array.shape != (
                    dimension,
                ):

                    availability_mask = np.isfinite(
                        target_array
                    )

                else:

                    availability_mask = (
                        mask_array.astype(
                            bool
                        )
                        &
                        np.isfinite(
                            target_array
                        )
                    )


        BLOCK_2_SUPERVISION_AVAILABLE_DIMENSIONS[
            pathway_name
        ] += int(
            availability_mask.sum()
        )


        BLOCK_2_SENTENCE_SUPERVISION[
            sentence_id
        ][
            pathway_name
        ] = {
            "target":
                target_array,

            "availability_mask":
                availability_mask,
        }


# =============================================================================
# Supervision eligibility
# =============================================================================
#
# An article is supervision eligible here only if:
#   1. it is representation ready; and
#   2. at least one validated supervised element exists in a currently
#      active Transformative-Weight dimension.
#
# This is an inventory classification, not an objective calculation.
# =============================================================================

BLOCK_2_SUPERVISION_ELIGIBLE_ARTICLES = {}

BLOCK_2_SUPERVISION_NOT_ELIGIBLE_ARTICLES = {}


for (
    article_id,
    article_records,
) in BLOCK_2_REPRESENTATION_READY_ARTICLES.items():

    pathway_active_counts = {}


    for pathway_name in (
        "factual",
        "psychological",
        "social",
    ):

        active_indices = np.asarray(
            BLOCK_2_ACTIVE_DIMENSION_INDICES[
                pathway_name
            ],
            dtype=
                np.int64,
        )


        active_supervised_count = 0


        for record in article_records:

            sentence_id = record[
                "sentence_id"
            ]


            availability_mask = (
                BLOCK_2_SENTENCE_SUPERVISION[
                    sentence_id
                ][
                    pathway_name
                ][
                    "availability_mask"
                ]
            )


            active_supervised_count += int(
                availability_mask[
                    active_indices
                ].sum()
            )


        pathway_active_counts[
            pathway_name
        ] = active_supervised_count


    supervision_eligible = any(
        count
        >
        0

        for count
        in pathway_active_counts.values()
    )


    BLOCK_2_ARTICLE_AUDIT[
        article_id
    ][
        "active_supervision_counts"
    ] = pathway_active_counts


    BLOCK_2_ARTICLE_AUDIT[
        article_id
    ][
        "supervision_eligible"
    ] = supervision_eligible


    if supervision_eligible:

        BLOCK_2_SUPERVISION_ELIGIBLE_ARTICLES[
            article_id
        ] = article_records

    else:

        BLOCK_2_SUPERVISION_NOT_ELIGIBLE_ARTICLES[
            article_id
        ] = (
            "no_active_dimension_supervision_in_selected_source"
        )


# =============================================================================
# Aggregate corpus counts
# =============================================================================

BLOCK_2_STRUCTURALLY_ELIGIBLE_ARTICLE_COUNT = len(
    BLOCK_2_STRUCTURALLY_ELIGIBLE_ARTICLES
)


BLOCK_2_STRUCTURALLY_ELIGIBLE_SENTENCE_COUNT = sum(
    len(
        records
    )

    for records
    in BLOCK_2_STRUCTURALLY_ELIGIBLE_ARTICLES.values()
)


BLOCK_2_REPRESENTATION_READY_ARTICLE_COUNT = len(
    BLOCK_2_REPRESENTATION_READY_ARTICLES
)


BLOCK_2_REPRESENTATION_READY_SENTENCE_COUNT = sum(
    len(
        records
    )

    for records
    in BLOCK_2_REPRESENTATION_READY_ARTICLES.values()
)


BLOCK_2_SUPERVISION_ELIGIBLE_ARTICLE_COUNT = len(
    BLOCK_2_SUPERVISION_ELIGIBLE_ARTICLES
)


BLOCK_2_SUPERVISION_ELIGIBLE_SENTENCE_COUNT = sum(
    len(
        records
    )

    for records
    in BLOCK_2_SUPERVISION_ELIGIBLE_ARTICLES.values()
)


BLOCK_2_STRUCTURALLY_EXCLUDED_ARTICLE_COUNT = len(
    BLOCK_2_STRUCTURALLY_EXCLUDED_ARTICLES
)


# =============================================================================
# Representation coverage
# =============================================================================

BLOCK_2_REPRESENTATION_AVAILABLE_SENTENCE_COUNT = sum(
    int(
        record[
            "representation_shape_valid"
        ]
        and
        record[
            "representation_finite"
        ]
    )

    for record
    in BLOCK_2_CANONICAL_RECORDS
)


BLOCK_2_REPRESENTATION_COVERAGE = (
    BLOCK_2_REPRESENTATION_AVAILABLE_SENTENCE_COUNT
    /
    BLOCK_2_CANDIDATE_SENTENCE_COUNT
)


# =============================================================================
# Supervision coverage
# =============================================================================

BLOCK_2_SUPERVISION_COVERAGE = {}


for pathway_name in BLOCK_2_PATHWAY_DIMENSIONS:

    total_dimensions = (
        BLOCK_2_SUPERVISION_TOTAL_DIMENSIONS[
            pathway_name
        ]
    )


    available_dimensions = (
        BLOCK_2_SUPERVISION_AVAILABLE_DIMENSIONS[
            pathway_name
        ]
    )


    BLOCK_2_SUPERVISION_COVERAGE[
        pathway_name
    ] = (
        available_dimensions
        /
        total_dimensions
        if total_dimensions
        >
        0
        else 0.0
    )


# =============================================================================
# Multi-article readiness
# =============================================================================

BLOCK_2_EXPANDED_CORPUS_AVAILABLE = (
    BLOCK_2_STRUCTURALLY_ELIGIBLE_ARTICLE_COUNT
    >=
    BLOCK_2_MINIMUM_MULTI_ARTICLE_COUNT
)


BLOCK_2_MULTI_ARTICLE_CORPUS_AVAILABLE = (
    BLOCK_2_REPRESENTATION_READY_ARTICLE_COUNT
    >=
    BLOCK_2_MINIMUM_MULTI_ARTICLE_COUNT
)


BLOCK_2_ADDITIONAL_CORPUS_REQUIRED = (
    not
    BLOCK_2_MULTI_ARTICLE_CORPUS_AVAILABLE
)


BLOCK_2_MULTI_ARTICLE_VALIDATION_READY = all(
    [
        BLOCK_2_MULTI_ARTICLE_CORPUS_AVAILABLE,
        BLOCK_2_SENTENCE_IDS_GLOBALLY_UNIQUE,
        BLOCK_2_NO_CROSS_ARTICLE_IDENTITY_COLLISIONS,
    ]
)


# =============================================================================
# Article-boundary and canonical-sequence contract
# =============================================================================

BLOCK_2_ARTICLE_BOUNDARIES_EXPLICIT = all(
    record[
        "article_id"
    ]
    is not None

    for record
    in BLOCK_2_CANONICAL_RECORDS
)


BLOCK_2_CANONICAL_SEQUENCE_VALID = all(
    audit[
        "canonical_order_valid"
    ]

    for (
        article_id,
        audit,
    ) in BLOCK_2_ARTICLE_AUDIT.items()

    if audit[
        "structurally_eligible"
    ]
)


BLOCK_2_ARTICLE_BOUNDARY_CONTRACT_VALID = all(
    [
        BLOCK_2_ARTICLE_BOUNDARIES_EXPLICIT,
        BLOCK_2_NO_CROSS_ARTICLE_IDENTITY_COLLISIONS,
        BLOCK_2_SENTENCE_IDS_GLOBALLY_UNIQUE,
    ]
)


# =============================================================================
# Construct deterministic multi-article inventory
# =============================================================================
#
# Article ordering is deterministic for storage and audit only.
# It does not define recurrent context between articles.
# =============================================================================

NOTEBOOK_10_EXPANDED_CORPUS_INVENTORY = []


for article_id in sorted(
    BLOCK_2_ARTICLE_RECORDS.keys()
):

    audit = deepcopy(
        BLOCK_2_ARTICLE_AUDIT[
            article_id
        ]
    )


    NOTEBOOK_10_EXPANDED_CORPUS_INVENTORY.append(
        audit
    )


NOTEBOOK_10_STRUCTURALLY_ELIGIBLE_ARTICLE_RECORDS = {
    article_id:
        deepcopy(
            records
        )

    for (
        article_id,
        records,
    ) in BLOCK_2_STRUCTURALLY_ELIGIBLE_ARTICLES.items()
}


NOTEBOOK_10_REPRESENTATION_READY_ARTICLE_RECORDS = {
    article_id:
        deepcopy(
            records
        )

    for (
        article_id,
        records,
    ) in BLOCK_2_REPRESENTATION_READY_ARTICLES.items()
}


NOTEBOOK_10_ARTICLE_REPRESENTATION_MATRICES = {
    article_id:
        matrix.copy()

    for (
        article_id,
        matrix,
    ) in BLOCK_2_ARTICLE_REPRESENTATION_MATRICES.items()
}


NOTEBOOK_10_SENTENCE_SUPERVISION_INVENTORY = deepcopy(
    BLOCK_2_SENTENCE_SUPERVISION
)


# =============================================================================
# Model-state immutability validation
# =============================================================================
#
# Compare the end-of-Block-2 runtime state against the exact restored Block 1
# baseline captured before corpus inventory work began.
# =============================================================================

BLOCK_2_REPRESENTATION_STATE_AFTER = (
    block_2_snapshot_module_collection(
        NOTEBOOK_10_REPRESENTATION_MODULES
    )
)


BLOCK_2_TRANSFORMATIVE_STATE_AFTER = (
    block_2_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_MODULES
    )
)


BLOCK_2_WEIGHT_STATE_AFTER = (
    block_2_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES
    )
)


BLOCK_2_REPRESENTATION_STATE_UNCHANGED = (
    block_2_nested_state_exact(
        BLOCK_2_REPRESENTATION_STATE_BASELINE,
        BLOCK_2_REPRESENTATION_STATE_AFTER,
    )
)


BLOCK_2_TRANSFORMATIVE_STATE_UNCHANGED = (
    block_2_nested_state_exact(
        BLOCK_2_TRANSFORMATIVE_STATE_BASELINE,
        BLOCK_2_TRANSFORMATIVE_STATE_AFTER,
    )
)


BLOCK_2_WEIGHT_STATE_UNCHANGED = (
    block_2_nested_state_exact(
        BLOCK_2_WEIGHT_STATE_BASELINE,
        BLOCK_2_WEIGHT_STATE_AFTER,
    )
)


# Additional semantic identity check for reconstructed effective weights.
# This guards against a hypothetical unchanged latent tensor being interpreted
# under a different active-index mapping.
with torch.no_grad():

    BLOCK_2_EFFECTIVE_WEIGHT_IDENTITY_UNCHANGED = all(
        torch.equal(
            NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES[
                pathway_name
            ]()
            .detach()
            .cpu(),

            NOTEBOOK_10_INHERITED_EFFECTIVE_TRANSFORMATIVE_WEIGHTS[
                pathway_name
            ]
            .detach()
            .cpu(),
        )

        for pathway_name
        in BLOCK_2_PATHWAY_DIMENSIONS
    )


BLOCK_2_ALL_PARAMETERS_FROZEN = all(
    not parameter.requires_grad

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


BLOCK_2_ALL_MODULES_EVAL = all(
    not module.training

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()
)


BLOCK_2_ALL_GRADIENTS_ABSENT = all(
    parameter.grad is None

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


# =============================================================================
# Execution-boundary validation
# =============================================================================

BLOCK_2_EXECUTION_BOUNDARY_VALID = all(
    [
        NOTEBOOK_10_BLOCK_2_REPRESENTATION_FORWARD_EXECUTED is False,

        NOTEBOOK_10_BLOCK_2_TRANSFORMATIVE_FORWARD_EXECUTED is False,

        NOTEBOOK_10_BLOCK_2_WEIGHT_FORWARD_EXECUTED is False,

        NOTEBOOK_10_BLOCK_2_RECURRENT_STATE_CREATED is False,

        NOTEBOOK_10_BLOCK_2_RECURRENT_UPDATE_EXECUTED is False,

        NOTEBOOK_10_BLOCK_2_LOSS_CALCULATED is False,

        NOTEBOOK_10_BLOCK_2_OPTIMIZER_CREATED is False,

        NOTEBOOK_10_BLOCK_2_ZERO_GRAD_EXECUTED is False,

        NOTEBOOK_10_BLOCK_2_BACKWARD_PASS_EXECUTED is False,

        NOTEBOOK_10_BLOCK_2_PARAMETER_UPDATE_EXECUTED is False,

        NOTEBOOK_10_BLOCK_2_TRAINING_EXECUTED is False,
    ]
)


# =============================================================================
# Final Block 2 validity
# =============================================================================
#
# IMPORTANT:
#
# Block validity does not require >= 2 articles.
#
# A one-article inventory is a valid scientific result. It simply means
# that corpus expansion is still required before multi-article validation.
# =============================================================================

NOTEBOOK_10_BLOCK_2_ERRORS = []


BLOCK_2_VALIDATION_CHECKS = {
    "block_1_dependency_invalid":
        BLOCK_2_BLOCK_1_DEPENDENCY_VALID,

    "no_candidate_articles":
        (
            BLOCK_2_CANDIDATE_ARTICLE_COUNT
            >
            0
        ),

    "no_candidate_sentences":
        (
            BLOCK_2_CANDIDATE_SENTENCE_COUNT
            >
            0
        ),

    "sentence_ids_invalid":
        BLOCK_2_SENTENCE_IDS_VALID,

    "sentence_ids_not_globally_unique":
        BLOCK_2_SENTENCE_IDS_GLOBALLY_UNIQUE,

    "cross_article_identity_collision":
        BLOCK_2_NO_CROSS_ARTICLE_IDENTITY_COLLISIONS,

    "no_structurally_eligible_article":
        (
            BLOCK_2_STRUCTURALLY_ELIGIBLE_ARTICLE_COUNT
            >
            0
        ),

    "article_boundary_contract_invalid":
        BLOCK_2_ARTICLE_BOUNDARY_CONTRACT_VALID,

    "canonical_sequence_invalid":
        BLOCK_2_CANONICAL_SEQUENCE_VALID,

    "representation_state_changed":
        BLOCK_2_REPRESENTATION_STATE_UNCHANGED,

    "transformative_state_changed":
        BLOCK_2_TRANSFORMATIVE_STATE_UNCHANGED,

    "weight_state_changed":
        BLOCK_2_WEIGHT_STATE_UNCHANGED,

    "effective_weight_identity_changed":
        BLOCK_2_EFFECTIVE_WEIGHT_IDENTITY_UNCHANGED,

    "parameters_not_frozen":
        BLOCK_2_ALL_PARAMETERS_FROZEN,

    "modules_not_eval":
        BLOCK_2_ALL_MODULES_EVAL,

    "unexpected_gradients":
        BLOCK_2_ALL_GRADIENTS_ABSENT,

    "execution_boundary_invalid":
        BLOCK_2_EXECUTION_BOUNDARY_VALID,
}


for (
    error_name,
    condition,
) in BLOCK_2_VALIDATION_CHECKS.items():

    if not condition:

        NOTEBOOK_10_BLOCK_2_ERRORS.append(
            error_name
        )


NOTEBOOK_10_BLOCK_2_VALID = (
    len(
        NOTEBOOK_10_BLOCK_2_ERRORS
    )
    ==
    0
)


if not NOTEBOOK_10_BLOCK_2_VALID:

    raise RuntimeError(
        "Notebook 10 Block 2 expanded-corpus inventory failed. "
        f"Errors: {NOTEBOOK_10_BLOCK_2_ERRORS}"
    )


NOTEBOOK_10_BLOCK_2_COMPLETE = True


# =============================================================================
# Stable readiness globals for subsequent blocks
# =============================================================================

NOTEBOOK_10_EXPANDED_CORPUS_AVAILABLE = (
    BLOCK_2_EXPANDED_CORPUS_AVAILABLE
)


NOTEBOOK_10_MULTI_ARTICLE_CORPUS_AVAILABLE = (
    BLOCK_2_MULTI_ARTICLE_CORPUS_AVAILABLE
)


NOTEBOOK_10_ADDITIONAL_CORPUS_REQUIRED = (
    BLOCK_2_ADDITIONAL_CORPUS_REQUIRED
)


NOTEBOOK_10_MULTI_ARTICLE_VALIDATION_READY = (
    BLOCK_2_MULTI_ARTICLE_VALIDATION_READY
)


# =============================================================================
# Block summary
# =============================================================================

NOTEBOOK_10_BLOCK_2_SUMMARY = {
    "block":
        NOTEBOOK_10_BLOCK_2,

    "block_name":
        NOTEBOOK_10_BLOCK_2_NAME,

    "block_version":
        NOTEBOOK_10_BLOCK_2_VERSION,

    "selected_source":
        BLOCK_2_SELECTED_SOURCE_NAME,

    "candidate_articles":
        BLOCK_2_CANDIDATE_ARTICLE_COUNT,

    "candidate_sentences":
        BLOCK_2_CANDIDATE_SENTENCE_COUNT,

    "structurally_eligible_articles":
        BLOCK_2_STRUCTURALLY_ELIGIBLE_ARTICLE_COUNT,

    "structurally_eligible_sentences":
        BLOCK_2_STRUCTURALLY_ELIGIBLE_SENTENCE_COUNT,

    "representation_ready_articles":
        BLOCK_2_REPRESENTATION_READY_ARTICLE_COUNT,

    "representation_ready_sentences":
        BLOCK_2_REPRESENTATION_READY_SENTENCE_COUNT,

    "supervision_eligible_articles":
        BLOCK_2_SUPERVISION_ELIGIBLE_ARTICLE_COUNT,

    "supervision_eligible_sentences":
        BLOCK_2_SUPERVISION_ELIGIBLE_SENTENCE_COUNT,

    "representation_coverage":
        BLOCK_2_REPRESENTATION_COVERAGE,

    "supervision_coverage":
        deepcopy(
            BLOCK_2_SUPERVISION_COVERAGE
        ),

    "expanded_corpus_available":
        BLOCK_2_EXPANDED_CORPUS_AVAILABLE,

    "multi_article_corpus_available":
        BLOCK_2_MULTI_ARTICLE_CORPUS_AVAILABLE,

    "additional_corpus_required":
        BLOCK_2_ADDITIONAL_CORPUS_REQUIRED,

    "multi_article_validation_ready":
        BLOCK_2_MULTI_ARTICLE_VALIDATION_READY,

    "representation_state_unchanged":
        BLOCK_2_REPRESENTATION_STATE_UNCHANGED,

    "transformative_state_unchanged":
        BLOCK_2_TRANSFORMATIVE_STATE_UNCHANGED,

    "weight_state_unchanged":
        BLOCK_2_WEIGHT_STATE_UNCHANGED,

    "effective_weight_identity_unchanged":
        BLOCK_2_EFFECTIVE_WEIGHT_IDENTITY_UNCHANGED,

    "block_valid":
        NOTEBOOK_10_BLOCK_2_VALID,

    "block_complete":
        NOTEBOOK_10_BLOCK_2_COMPLETE,
}


# =============================================================================
# Final report
# =============================================================================

print("=" * 72)

print(
    "Media AI — Notebook 10, Block 2: "
    "Expanded-Corpus Inventory, Eligibility "
    "and Multi-Article Validation-Source Construction"
)

print("=" * 72)

print(
    f"Block version                : "
    f"{NOTEBOOK_10_BLOCK_2_VERSION}"
)

print("-" * 72)

print(
    "Corpus source"
)

print(
    f"Selected source              : "
    f"{BLOCK_2_SELECTED_SOURCE_NAME}"
)

print(
    f"Persisted contextual fallback: "
    f"{BLOCK_2_PERSISTED_CONTEXTUAL_SOURCE_USED}"
)

print(
    f"Candidate sources inspected  : "
    f"{len(BLOCK_2_SOURCE_CANDIDATES)}"
)

print("-" * 72)

print(
    "Candidate corpus inventory"
)

print(
    f"Candidate articles           : "
    f"{BLOCK_2_CANDIDATE_ARTICLE_COUNT}"
)

print(
    f"Candidate sentences          : "
    f"{BLOCK_2_CANDIDATE_SENTENCE_COUNT}"
)

print(
    f"Record exclusions            : "
    f"{len(BLOCK_2_RECORD_EXCLUSIONS)}"
)

print(
    f"Sentence IDs valid           : "
    f"{BLOCK_2_SENTENCE_IDS_VALID}"
)

print(
    f"Sentence IDs globally unique : "
    f"{BLOCK_2_SENTENCE_IDS_GLOBALLY_UNIQUE}"
)

print(
    f"Cross-article collisions     : "
    f"{len(BLOCK_2_CROSS_ARTICLE_SENTENCE_COLLISIONS)}"
)

print("-" * 72)

print(
    "Structural eligibility"
)

print(
    f"Eligible articles            : "
    f"{BLOCK_2_STRUCTURALLY_ELIGIBLE_ARTICLE_COUNT}"
)

print(
    f"Eligible sentences           : "
    f"{BLOCK_2_STRUCTURALLY_ELIGIBLE_SENTENCE_COUNT}"
)

print(
    f"Excluded articles            : "
    f"{BLOCK_2_STRUCTURALLY_EXCLUDED_ARTICLE_COUNT}"
)

print(
    f"Article boundaries explicit  : "
    f"{BLOCK_2_ARTICLE_BOUNDARIES_EXPLICIT}"
)

print(
    f"Canonical sequence valid     : "
    f"{BLOCK_2_CANONICAL_SEQUENCE_VALID}"
)

print("-" * 72)

print(
    "Contextual representation availability"
)

print(
    f"Expected dimension           : "
    f"{BLOCK_2_EXPECTED_EMBEDDING_DIMENSION}"
)

print(
    f"Representation-ready articles: "
    f"{BLOCK_2_REPRESENTATION_READY_ARTICLE_COUNT}"
)

print(
    f"Representation-ready sentences: "
    f"{BLOCK_2_REPRESENTATION_READY_SENTENCE_COUNT}"
)

print(
    f"Representation coverage      : "
    f"{BLOCK_2_REPRESENTATION_COVERAGE:.2%}"
)

print("-" * 72)

print(
    "Supervision inventory in selected source"
)

for pathway_name in BLOCK_2_PATHWAY_DIMENSIONS:

    print(
        f"{pathway_name:<14} available dimensions: "
        f"{BLOCK_2_SUPERVISION_AVAILABLE_DIMENSIONS[pathway_name]} "
        f"/ "
        f"{BLOCK_2_SUPERVISION_TOTAL_DIMENSIONS[pathway_name]}"
    )

    print(
        f"{pathway_name:<14} coverage            : "
        f"{BLOCK_2_SUPERVISION_COVERAGE[pathway_name]:.2%}"
    )

print(
    f"Supervision-eligible articles: "
    f"{BLOCK_2_SUPERVISION_ELIGIBLE_ARTICLE_COUNT}"
)

print(
    f"Supervision-eligible sentences: "
    f"{BLOCK_2_SUPERVISION_ELIGIBLE_SENTENCE_COUNT}"
)

print("-" * 72)

print(
    "Expanded-corpus readiness"
)

print(
    f"Minimum article count        : "
    f"{BLOCK_2_MINIMUM_MULTI_ARTICLE_COUNT}"
)

print(
    f"Expanded corpus available    : "
    f"{BLOCK_2_EXPANDED_CORPUS_AVAILABLE}"
)

print(
    f"Multi-article corpus available: "
    f"{BLOCK_2_MULTI_ARTICLE_CORPUS_AVAILABLE}"
)

print(
    f"Additional corpus required   : "
    f"{BLOCK_2_ADDITIONAL_CORPUS_REQUIRED}"
)

print(
    f"Multi-article validation ready: "
    f"{BLOCK_2_MULTI_ARTICLE_VALIDATION_READY}"
)

print("-" * 72)

print(
    "Inherited model state"
)

print(
    f"Representation unchanged     : "
    f"{BLOCK_2_REPRESENTATION_STATE_UNCHANGED}"
)

print(
    f"Transformative unchanged     : "
    f"{BLOCK_2_TRANSFORMATIVE_STATE_UNCHANGED}"
)

print(
    f"Transform Weight unchanged   : "
    f"{BLOCK_2_WEIGHT_STATE_UNCHANGED}"
)

print(
    f"Effective Weight identity    : "
    f"{BLOCK_2_EFFECTIVE_WEIGHT_IDENTITY_UNCHANGED}"
)

print(
    f"All parameters frozen        : "
    f"{BLOCK_2_ALL_PARAMETERS_FROZEN}"
)

print(
    f"All modules eval             : "
    f"{BLOCK_2_ALL_MODULES_EVAL}"
)

print(
    f"All gradients absent         : "
    f"{BLOCK_2_ALL_GRADIENTS_ABSENT}"
)

print("-" * 72)

print(
    "Execution boundary"
)

print(
    f"Representation forward       : "
    f"{NOTEBOOK_10_BLOCK_2_REPRESENTATION_FORWARD_EXECUTED}"
)

print(
    f"Transform forward            : "
    f"{NOTEBOOK_10_BLOCK_2_TRANSFORMATIVE_FORWARD_EXECUTED}"
)

print(
    f"Weight forward               : "
    f"{NOTEBOOK_10_BLOCK_2_WEIGHT_FORWARD_EXECUTED}"
)

print(
    f"Recurrent state created      : "
    f"{NOTEBOOK_10_BLOCK_2_RECURRENT_STATE_CREATED}"
)

print(
    f"Recurrent update executed    : "
    f"{NOTEBOOK_10_BLOCK_2_RECURRENT_UPDATE_EXECUTED}"
)

print(
    f"Loss calculated              : "
    f"{NOTEBOOK_10_BLOCK_2_LOSS_CALCULATED}"
)

print(
    f"Optimizer created            : "
    f"{NOTEBOOK_10_BLOCK_2_OPTIMIZER_CREATED}"
)

print(
    f"Backward pass executed       : "
    f"{NOTEBOOK_10_BLOCK_2_BACKWARD_PASS_EXECUTED}"
)

print(
    f"Parameter update executed    : "
    f"{NOTEBOOK_10_BLOCK_2_PARAMETER_UPDATE_EXECUTED}"
)

print(
    f"Training executed            : "
    f"{NOTEBOOK_10_BLOCK_2_TRAINING_EXECUTED}"
)

print("-" * 72)

print(
    f"Article-boundary valid       : "
    f"{BLOCK_2_ARTICLE_BOUNDARY_CONTRACT_VALID}"
)

print(
    f"Inventory valid              : "
    f"{NOTEBOOK_10_BLOCK_2_VALID}"
)

print(
    f"Block complete               : "
    f"{NOTEBOOK_10_BLOCK_2_COMPLETE}"
)

print("=" * 72)


if BLOCK_2_MULTI_ARTICLE_VALIDATION_READY:

    print(
        "A valid multi-article expanded corpus is available."
    )

    print(
        "Notebook 10 may proceed to controlled expanded-corpus "
        "representation and multi-article recurrent validation."
    )

else:

    print(
        "The current corpus inventory was validated successfully, "
        "but it is not yet sufficient for multi-article validation."
    )

    print(
        f"The validated runtime currently provides "
        f"{BLOCK_2_REPRESENTATION_READY_ARTICLE_COUNT} "
        f"representation-ready article(s)."
    )

    print(
        "Additional independently identifiable article material is "
        "therefore required before expanded multi-article execution."
    )

    print(
        "This corpus-sufficiency result does not invalidate Block 2."
    )

    print(
        "Notebook 10 should next proceed to controlled additional-corpus "
        "construction and preparation."
    )


print("=" * 72)
```

# Block 3 — Additional-Corpus Construction and Multi-Article Source Preparation

## Purpose

This block constructs the additional article material required to move Notebook 10 beyond the single-article pilot corpus identified in Block 2.

Block 2 established that the currently available validated source contains:

- **1 article**;
- **16 sentences**;
- complete structural identity;
- valid canonical sequence ordering;
- complete 768-dimensional contextual representations;
- no embedded factual, psychological or social supervision;
- insufficient article count for multi-article validation.

The present block therefore addresses the specific corpus-sufficiency limitation identified by Block 2.

Its purpose is to recover or construct additional independently identifiable article material, preserve deterministic textual structure, and prepare a validated multi-article source suitable for later representation and recurrent execution.

No model training is performed in this block.

## Methodological Role

Block 3 is a corpus-construction block.

It does not modify the Media AI architecture restored in Block 1 and does not alter the eligibility definitions established in Block 2.

Instead, it extends the available textual source from

$$
N_{\mathrm{articles}} = 1
$$

toward a genuinely multi-article corpus satisfying

$$
N_{\mathrm{articles}} \ge 2.
$$

The objective is not merely to increase sentence count.

The additional material must introduce **independent article boundaries** so that later blocks can validate:

- repeated zero-state initialisation;
- independent recurrent trajectories;
- absence of cross-article state propagation;
- deterministic article-specific execution;
- model stability across more than one textual sequence.

## Source Policy

Additional articles must originate from genuine project corpus material or from another explicitly identified source accepted for the validation project.

The block must not create synthetic articles solely to satisfy the minimum article-count requirement.

Permitted sources include:

- previously prepared project corpus artefacts;
- validated raw or cleaned article collections from earlier notebooks;
- persisted article-level project data containing traceable source identity;
- additional manually supplied article material where source identity and structural boundaries remain explicit.

The block must prefer existing project artefacts over manual reconstruction when those artefacts are available.

## No Synthetic Fallback

The expanded corpus must represent actual independent text.

The block must not:

- split the original 16-sentence pilot article into artificial sub-articles;
- duplicate the pilot article under different identifiers;
- generate synthetic sentences;
- perturb existing sentences and treat them as new articles;
- use model-generated text as independent validation data;
- convert supervision targets into textual records;
- infer additional article boundaries from semantic similarity.

Any such procedure would create artificial independence and would invalidate later article-boundary validation.

## Article Identity Contract

Every additional article must expose a stable and unique article identifier.

For article $i$,

$$
\mathcal{A}_i
=
\left(
s_{i,1},
s_{i,2},
\ldots,
s_{i,T_i}
\right),
$$

where each sentence $s_{i,t}$ belongs to exactly one article.

The article identifier must satisfy:

- non-null identity;
- uniqueness at article level;
- deterministic association with all member sentences;
- no collision with the existing pilot article;
- no dependence on row position alone.

Article identity must be preserved throughout all subsequent processing.

## Sentence Identity Contract

Every sentence in the expanded corpus must expose a globally unique sentence identifier.

For any two different sentence records,

$$
s_{i,t} \neq s_{j,u},
$$

the identifiers must satisfy

$$
\mathrm{id}(s_{i,t})
\neq
\mathrm{id}(s_{j,u}).
$$

This requirement applies both within and across articles.

Sentence identity must remain stable across:

- corpus preparation;
- contextual representation generation;
- supervision alignment;
- recurrent execution;
- evaluation.

## Paragraph Identity

Paragraph membership should be retained where available.

For each sentence, the preferred structural hierarchy is:

$$
\mathrm{Article}
\rightarrow
\mathrm{Paragraph}
\rightarrow
\mathrm{Sentence}.
$$

Paragraph identity is not required for recurrent-state propagation itself, but preserving it maintains the original corpus hierarchy and supports later structural analysis.

Missing paragraph identity must remain explicitly missing rather than being inferred without evidence.

## Canonical Sequence Construction

Within every article, sentence order must be deterministic.

For article $i$ containing $T_i$ sentences,

$$
s_{i,1}
\rightarrow
s_{i,2}
\rightarrow
\cdots
\rightarrow
s_{i,T_i}.
$$

Canonical sequence positions must be:

- numeric;
- unique within the article;
- contiguous;
- consistently zero-based or one-based within that article;
- sufficient to reproduce the original textual order exactly.

The block may normalise sequence positions to a common internal convention after validating their original structure.

If normalisation is performed, the original sequence position should remain recoverable for audit purposes.

## Textual Integrity

Additional article records must preserve the textual content required for contextual representation generation.

Each eligible sentence should contain either:

- the canonical sentence text itself; or
- a validated persistent representation already aligned to the sentence identity.

Where raw sentence text is available, the block should preserve it unchanged.

Text must not be rewritten, paraphrased or normalised semantically merely to simplify model processing.

Only structural cleaning already established by the preceding corpus-preparation methodology may be reused.

## Contextual Representation Contract

The Media AI representation architecture expects contextual sentence inputs in

$$
\mathbb{R}^{768}.
$$

Where additional articles already possess validated contextual sentence representations, those representations may be reused if their identity and encoder contract can be verified.

Where representations do not yet exist, they must later be generated under the established contextual-encoder contract.

The expected encoder remains:

**sentence-transformers/all-mpnet-base-v2**

with

$$
d_{\mathrm{embedding}} = 768.
$$

This block should distinguish between:

- additional articles with existing validated contextual representations;
- additional articles requiring contextual representation generation.

It should not silently mix incompatible embedding sources.

## Encoder Compatibility

Any reused contextual representation artefact must be checked for compatibility with the inherited representation contract.

At minimum, the following properties must be validated where metadata are available:

- encoder provider;
- encoder model;
- embedding dimension;
- normalisation policy;
- sentence-identity alignment;
- article-identity alignment.

If encoder provenance cannot be verified, the representation should be treated as unavailable rather than assumed compatible.

## Representation Generation Boundary

If new contextual representations are required, Block 3 may prepare the required textual input records and the generation contract.

The actual representation generation may occur in this block only if it can be performed under the exact established encoder contract without changing the architecture.

Whether generation is executed here or in the following block, the output must remain a sentence-level matrix

$$
\mathbf{X}_i
\in
\mathbb{R}^{T_i \times 768}
$$

for each article $i$.

No recurrent execution is permitted during contextual representation preparation.

## Corpus Merge Contract

The original pilot article and the newly recovered articles must be merged without altering the identity or ordering of the pilot records.

Let the original article set be

$$
\mathcal{C}_{\mathrm{pilot}}
=
\{\mathcal{A}_1\},
$$

and the newly prepared article set be

$$
\mathcal{C}_{\mathrm{new}}
=
\{
\mathcal{A}_2,
\ldots,
\mathcal{A}_N
\}.
$$

The expanded validation corpus becomes

$$
\mathcal{C}_{\mathrm{expanded}}
=
\mathcal{C}_{\mathrm{pilot}}
\cup
\mathcal{C}_{\mathrm{new}}.
$$

The merge must preserve:

- article identity;
- sentence identity;
- sentence order;
- paragraph identity where available;
- contextual representation alignment;
- provenance.

No article may appear twice.

## Corpus Sufficiency

The minimum condition for multi-article validation is

$$
N_{\mathrm{eligible\ articles}} \ge 2.
$$

This minimum is sufficient only to demonstrate cross-article architectural behaviour.

It is not sufficient for substantive training or generalisation claims.

The block should therefore distinguish between:

**Multi-article validation sufficiency**

Enough independently identifiable articles exist to validate repeated recurrent resets and independent trajectories.

**Expanded-training sufficiency**

Enough diverse material exists for meaningful larger-corpus optimisation.

The latter threshold is intentionally not satisfied merely by reaching two articles.

## Diversity Inventory

Where source metadata are available, the block should retain descriptive information such as:

- source;
- publication;
- article date;
- topic;
- section;
- language;
- author or provider identity where already present in the corpus.

These properties are not used as model inputs unless explicitly defined elsewhere.

Their purpose is to support later assessment of corpus diversity and prevent the expanded corpus from being treated as representative merely because it contains more records.

## Supervision Status

Block 3 does not require factual, psychological or social supervision to construct the additional textual corpus.

The primary objective is to establish valid multi-article model input.

For each new article, supervision status should therefore be classified separately as:

- supervision available;
- supervision partially available;
- supervision unavailable;
- supervision alignment not yet assessed.

Missing supervision must not block structural corpus construction.

However, supervision availability must remain traceable because later expanded training requires a dedicated alignment stage.

## Existing Pilot Supervision

The absence of supervision in the Notebook 05 contextual representation artefact does not imply that the pilot article lacks supervision.

The pilot supervision used in Notebook 09 was recovered from separate validated factual, psychological and social supervision artefacts.

Block 3 must preserve this distinction:

**contextual representation source ≠ supervision source.**

The same separation should be maintained for newly added articles.

## Article-Level Eligibility

An additional article is structurally eligible only when:

- article identity is valid;
- at least one sentence is present;
- sentence identities are valid;
- sentence identities are unique;
- sequence positions are valid;
- sequence positions are unique;
- canonical sentence ordering is deterministic;
- no global sentence-identity collision occurs.

Representation readiness is evaluated separately.

## Representation Readiness

A structurally eligible article is representation ready when every required sentence has:

- one aligned contextual representation;
- dimensionality exactly 768;
- finite numeric values;
- verified sentence identity;
- verified article identity.

If any required sentence representation is missing, the article remains structurally eligible but is not yet representation ready.

## Multi-Article Source Readiness

The constructed source is ready for expanded recurrent validation only if:

- at least two articles are structurally eligible;
- at least two articles are representation ready;
- sentence identities are globally unique;
- article identities are unique;
- canonical sequence ordering is valid for every ready article;
- article boundaries are explicit;
- representation dimensionality is consistent;
- representation values are finite;
- no synthetic fallback has been used.

```python
# =============================================================================
# Media AI — Notebook 10
# Block 3: Additional-Corpus Construction
#          and Multi-Article Source Preparation
# =============================================================================

from copy import deepcopy
from datetime import datetime, timezone

import io
import json
import math

import numpy as np
import torch

from googleapiclient.http import MediaIoBaseDownload


# =============================================================================
# Block identity
# =============================================================================

NOTEBOOK_10_BLOCK_3 = 3

NOTEBOOK_10_BLOCK_3_NAME = (
    "Additional-Corpus Construction "
    "and Multi-Article Source Preparation"
)

NOTEBOOK_10_BLOCK_3_VERSION = "1.2"

BLOCK_3_EXECUTED_AT_UTC = (
    datetime.now(
        timezone.utc
    ).isoformat()
)


# =============================================================================
# Block dependencies
# =============================================================================

BLOCK_3_PREREQUISITES_VALID = all(
    [
        globals().get(
            "NOTEBOOK_10_BLOCK_1_COMPLETE",
            False,
        )
        is True,

        globals().get(
            "NOTEBOOK_10_BLOCK_1_VALID",
            False,
        )
        is True,

        globals().get(
            "NOTEBOOK_10_BLOCK_2_COMPLETE",
            False,
        )
        is True,

        globals().get(
            "NOTEBOOK_10_BLOCK_2_VALID",
            False,
        )
        is True,

        globals().get(
            "NOTEBOOK_10_INITIALISED",
            False,
        )
        is True,

        "NOTEBOOK_10_REPRESENTATION_READY_ARTICLE_RECORDS"
        in globals(),

        "NOTEBOOK_10_DRIVE_SERVICE"
        in globals(),

        "NOTEBOOK_10_REPRESENTATION_MODULES"
        in globals(),

        "NOTEBOOK_10_TRANSFORMATIVE_MODULES"
        in globals(),

        "NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES"
        in globals(),
    ]
)


if not BLOCK_3_PREREQUISITES_VALID:

    raise RuntimeError(
        "Notebook 10 Block 3 requires valid completed Blocks 1 and 2."
    )


# =============================================================================
# Immutable inherited model-state baseline
# =============================================================================
#
# Notebook 10 Block 1 reconstructs the Transformative-Weight modules under the
# canonical Notebook 10 parameter key (`alpha`), while the persisted Notebook 09
# checkpoint may use a different implementation key such as `alpha_active`.
#
# Block 3 therefore validates immutability against the exact Notebook 10 runtime
# state inherited at the start of this block, not by comparing reconstructed
# Notebook 10 state-dict keys directly with raw Notebook 09 checkpoint keys.
# =============================================================================

def block_3_snapshot_module_collection(
    module_collection,
):

    return {
        module_name:
            {
                tensor_name:
                    tensor.detach()
                    .cpu()
                    .clone()

                for (
                    tensor_name,
                    tensor,
                ) in module.state_dict().items()
            }

        for (
            module_name,
            module,
        ) in module_collection.items()
    }


def block_3_nested_state_exact(
    state_before,
    state_after,
):

    if state_before.keys() != state_after.keys():

        return False


    for module_name in state_before:

        if (
            state_before[
                module_name
            ].keys()
            !=
            state_after[
                module_name
            ].keys()
        ):

            return False


        for tensor_name in state_before[
            module_name
        ]:

            if not torch.equal(
                state_before[
                    module_name
                ][
                    tensor_name
                ],
                state_after[
                    module_name
                ][
                    tensor_name
                ],
            ):

                return False


    return True


BLOCK_3_REPRESENTATION_STATE_BASELINE = (
    block_3_snapshot_module_collection(
        NOTEBOOK_10_REPRESENTATION_MODULES
    )
)


BLOCK_3_TRANSFORMATIVE_STATE_BASELINE = (
    block_3_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_MODULES
    )
)


BLOCK_3_WEIGHT_STATE_BASELINE = (
    block_3_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES
    )
)


# =============================================================================
# Canonical contextual-representation contract
# =============================================================================

BLOCK_3_EXPECTED_EMBEDDING_PROVIDER = (
    "sentence-transformers"
)

BLOCK_3_EXPECTED_EMBEDDING_MODEL = (
    "sentence-transformers/all-mpnet-base-v2"
)

BLOCK_3_EXPECTED_EMBEDDING_DIMENSION = 768

BLOCK_3_EXPECTED_NORMALISED = True

BLOCK_3_MINIMUM_MULTI_ARTICLE_COUNT = 2


# =============================================================================
# Execution boundary
# =============================================================================

NOTEBOOK_10_BLOCK_3_MEDIA_AI_REPRESENTATION_FORWARD_EXECUTED = False

NOTEBOOK_10_BLOCK_3_TRANSFORMATIVE_FORWARD_EXECUTED = False

NOTEBOOK_10_BLOCK_3_WEIGHT_FORWARD_EXECUTED = False

NOTEBOOK_10_BLOCK_3_RECURRENT_STATE_CREATED = False

NOTEBOOK_10_BLOCK_3_RECURRENT_UPDATE_EXECUTED = False

NOTEBOOK_10_BLOCK_3_LOSS_CALCULATED = False

NOTEBOOK_10_BLOCK_3_OPTIMIZER_CREATED = False

NOTEBOOK_10_BLOCK_3_BACKWARD_PASS_EXECUTED = False

NOTEBOOK_10_BLOCK_3_PARAMETER_UPDATE_EXECUTED = False

NOTEBOOK_10_BLOCK_3_TRAINING_EXECUTED = False

NOTEBOOK_10_BLOCK_3_SYNTHETIC_ARTICLE_CREATED = False

NOTEBOOK_10_BLOCK_3_PILOT_ARTICLE_SPLIT = False

NOTEBOOK_10_BLOCK_3_PILOT_ARTICLE_DUPLICATED = False

NOTEBOOK_10_BLOCK_3_SUPERVISION_USED_AS_TEXT = False

NOTEBOOK_10_BLOCK_3_SUPERVISION_USED_AS_REPRESENTATION = False


# =============================================================================
# General scalar helpers
# =============================================================================

def block_3_normalise_identifier(
    value,
):

    if value is None:

        return None


    if isinstance(
        value,
        np.generic,
    ):

        value = value.item()


    value = str(
        value
    ).strip()


    if not value:

        return None


    return value


def block_3_normalise_position(
    value,
):

    if value is None:

        return None


    if isinstance(
        value,
        np.generic,
    ):

        value = value.item()


    if isinstance(
        value,
        bool,
    ):

        return None


    try:

        numeric_value = float(
            value
        )

    except (
        TypeError,
        ValueError,
    ):

        return None


    if not math.isfinite(
        numeric_value
    ):

        return None


    if not numeric_value.is_integer():

        return None


    return int(
        numeric_value
    )


def block_3_normalise_text(
    value,
):

    if value is None:

        return None


    if not isinstance(
        value,
        str,
    ):

        return None


    value = value.strip()


    if not value:

        return None


    return value


def block_3_first_present(
    mapping,
    keys,
):

    if not isinstance(
        mapping,
        dict,
    ):

        return None


    for key in keys:

        if (
            key in mapping
            and
            mapping[
                key
            ]
            is not None
        ):

            return mapping[
                key
            ]


    return None


# =============================================================================
# Numeric-vector helper
# =============================================================================

def block_3_to_vector(
    value,
):

    if value is None:

        return None


    if torch.is_tensor(
        value
    ):

        array = (
            value.detach()
            .cpu()
            .numpy()
        )

    else:

        try:

            array = np.asarray(
                value,
                dtype=
                    np.float32,
            )

        except (
            TypeError,
            ValueError,
        ):

            return None


    if (
        array.ndim
        ==
        2
        and
        1
        in array.shape
    ):

        array = array.reshape(
            -1
        )


    if array.ndim != 1:

        return None


    return array.astype(
        np.float32,
        copy=
            False,
    )


# =============================================================================
# Exact nested-object comparison helper
# =============================================================================
#
# Required because the pilot records contain NumPy arrays.
#
# Plain dictionary equality is invalid for dictionaries containing
# multi-element NumPy arrays because array equality returns an array rather
# than a scalar Boolean.
# =============================================================================

def block_3_objects_exact(
    object_a,
    object_b,
):

    # -------------------------------------------------------------------------
    # NumPy arrays
    # -------------------------------------------------------------------------
    if (
        isinstance(
            object_a,
            np.ndarray,
        )
        or
        isinstance(
            object_b,
            np.ndarray,
        )
    ):

        if not (
            isinstance(
                object_a,
                np.ndarray,
            )
            and
            isinstance(
                object_b,
                np.ndarray,
            )
        ):

            return False


        if (
            object_a.dtype
            !=
            object_b.dtype
        ):

            return False


        if (
            object_a.shape
            !=
            object_b.shape
        ):

            return False


        return bool(
            np.array_equal(
                object_a,
                object_b,
                equal_nan=
                    True,
            )
        )


    # -------------------------------------------------------------------------
    # PyTorch tensors
    # -------------------------------------------------------------------------
    if (
        torch.is_tensor(
            object_a
        )
        or
        torch.is_tensor(
            object_b
        )
    ):

        if not (
            torch.is_tensor(
                object_a
            )
            and
            torch.is_tensor(
                object_b
            )
        ):

            return False


        if (
            object_a.dtype
            !=
            object_b.dtype
        ):

            return False


        if (
            tuple(
                object_a.shape
            )
            !=
            tuple(
                object_b.shape
            )
        ):

            return False


        return bool(
            torch.equal(
                object_a.detach().cpu(),
                object_b.detach().cpu(),
            )
        )


    # -------------------------------------------------------------------------
    # Dictionaries
    # -------------------------------------------------------------------------
    if (
        isinstance(
            object_a,
            dict,
        )
        or
        isinstance(
            object_b,
            dict,
        )
    ):

        if not (
            isinstance(
                object_a,
                dict,
            )
            and
            isinstance(
                object_b,
                dict,
            )
        ):

            return False


        if (
            set(
                object_a.keys()
            )
            !=
            set(
                object_b.keys()
            )
        ):

            return False


        return all(
            block_3_objects_exact(
                object_a[
                    key
                ],
                object_b[
                    key
                ],
            )

            for key
            in object_a
        )


    # -------------------------------------------------------------------------
    # Lists and tuples
    # -------------------------------------------------------------------------
    if (
        isinstance(
            object_a,
            (
                list,
                tuple,
            ),
        )
        or
        isinstance(
            object_b,
            (
                list,
                tuple,
            ),
        )
    ):

        if not (
            isinstance(
                object_a,
                (
                    list,
                    tuple,
                ),
            )
            and
            isinstance(
                object_b,
                (
                    list,
                    tuple,
                ),
            )
        ):

            return False


        if (
            len(
                object_a
            )
            !=
            len(
                object_b
            )
        ):

            return False


        return all(
            block_3_objects_exact(
                value_a,
                value_b,
            )

            for (
                value_a,
                value_b,
            ) in zip(
                object_a,
                object_b,
            )
        )


    # -------------------------------------------------------------------------
    # Sets
    # -------------------------------------------------------------------------
    if (
        isinstance(
            object_a,
            set,
        )
        or
        isinstance(
            object_b,
            set,
        )
    ):

        if not (
            isinstance(
                object_a,
                set,
            )
            and
            isinstance(
                object_b,
                set,
            )
        ):

            return False


        return (
            object_a
            ==
            object_b
        )


    # -------------------------------------------------------------------------
    # NumPy scalar values
    # -------------------------------------------------------------------------
    if isinstance(
        object_a,
        np.generic,
    ):

        object_a = object_a.item()


    if isinstance(
        object_b,
        np.generic,
    ):

        object_b = object_b.item()


    # -------------------------------------------------------------------------
    # Scalar fallback
    # -------------------------------------------------------------------------
    try:

        comparison = (
            object_a
            ==
            object_b
        )


        if isinstance(
            comparison,
            (
                bool,
                np.bool_,
            ),
        ):

            return bool(
                comparison
            )


        return False


    except Exception:

        return False


# =============================================================================
# Google Drive byte download helper
# =============================================================================

def block_3_download_drive_bytes(
    drive_service,
    file_id,
):

    request = (
        drive_service.files()
        .get_media(
            fileId=
                file_id
        )
    )


    buffer = io.BytesIO()


    downloader = MediaIoBaseDownload(
        buffer,
        request,
    )


    complete = False


    while not complete:

        _, complete = downloader.next_chunk()


    buffer.seek(
        0
    )


    return buffer.read()


# =============================================================================
# Canonical structural-key vocabulary
# =============================================================================

BLOCK_3_ARTICLE_ID_KEYS = (
    "article_id",
    "article_uuid",
    "document_id",
    "document_uuid",
    "source_article_id",
)


BLOCK_3_SENTENCE_ID_KEYS = (
    "sentence_id",
    "sentence_uuid",
)


BLOCK_3_PARAGRAPH_ID_KEYS = (
    "paragraph_id",
    "paragraph_uuid",
)


BLOCK_3_SEQUENCE_POSITION_KEYS = (
    "sequence_position",
    "sentence_position",
    "sentence_index",
    "sequence_index",
    "position",
)


BLOCK_3_TEXT_KEYS = (
    "sentence_text",
    "text",
    "sentence",
    "cleaned_text",
    "canonical_text",
    "content",
)


BLOCK_3_REPRESENTATION_KEYS = (
    "contextual_representation",
    "contextual_embedding",
    "sentence_representation",
    "sentence_embedding",
    "representation",
    "embedding",
    "vector",
)


# =============================================================================
# Preserve the existing validated pilot corpus
# =============================================================================

BLOCK_3_PILOT_ARTICLE_RECORDS = {
    article_id:
        deepcopy(
            records
        )

    for (
        article_id,
        records,
    ) in NOTEBOOK_10_REPRESENTATION_READY_ARTICLE_RECORDS.items()
}


BLOCK_3_PILOT_ARTICLE_RECORDS_BEFORE = deepcopy(
    BLOCK_3_PILOT_ARTICLE_RECORDS
)


BLOCK_3_PILOT_ARTICLE_COUNT = len(
    BLOCK_3_PILOT_ARTICLE_RECORDS
)


BLOCK_3_PILOT_SENTENCE_COUNT = sum(
    len(
        records
    )

    for records
    in BLOCK_3_PILOT_ARTICLE_RECORDS.values()
)


if BLOCK_3_PILOT_ARTICLE_COUNT < 1:

    raise RuntimeError(
        "Notebook 10 Block 3 requires at least one validated "
        "pilot article from Block 2."
    )


BLOCK_3_PILOT_ARTICLE_IDS = set(
    BLOCK_3_PILOT_ARTICLE_RECORDS.keys()
)


BLOCK_3_PILOT_SENTENCE_IDS = {
    record[
        "sentence_id"
    ]

    for records
    in BLOCK_3_PILOT_ARTICLE_RECORDS.values()

    for record
    in records
}


# =============================================================================
# Validate pilot article identity before source discovery
# =============================================================================

BLOCK_3_PILOT_ARTICLE_IDENTITIES_VALID = all(
    (
        article_id
        is not None
        and
        len(
            records
        )
        >
        0
    )

    for (
        article_id,
        records,
    ) in BLOCK_3_PILOT_ARTICLE_RECORDS.items()
)


BLOCK_3_PILOT_SENTENCE_IDENTITIES_VALID = all(
    (
        record.get(
            "sentence_id"
        )
        is not None
    )

    for records
    in BLOCK_3_PILOT_ARTICLE_RECORDS.values()

    for record
    in records
)


if not all(
    [
        BLOCK_3_PILOT_ARTICLE_IDENTITIES_VALID,
        BLOCK_3_PILOT_SENTENCE_IDENTITIES_VALID,
    ]
):

    raise RuntimeError(
        "Notebook 10 Block 3 received an invalid pilot corpus "
        "from Block 2."
    )


# =============================================================================
# Runtime corpus-source discovery
# =============================================================================
#
# Only plausible corpus-like structured variables are inspected.
#
# Notebook 10 state itself is excluded to prevent recursive reuse of the
# inventory being constructed by this notebook.
# =============================================================================

BLOCK_3_RUNTIME_SOURCE_KEYWORDS = (
    "corpus",
    "article",
    "sentence",
    "prepared",
    "cleaned",
    "text",
)


BLOCK_3_RUNTIME_EXCLUDED_PREFIXES = (
    "BLOCK_1_",
    "BLOCK_2_",
    "BLOCK_3_",
    "NOTEBOOK_10_",
)


BLOCK_3_RUNTIME_CANDIDATES = []


for (
    variable_name,
    variable_value,
) in list(
    globals().items()
):

    if variable_name.startswith(
        BLOCK_3_RUNTIME_EXCLUDED_PREFIXES
    ):

        continue


    variable_name_lower = (
        variable_name.lower()
    )


    if not any(
        keyword
        in
        variable_name_lower

        for keyword
        in BLOCK_3_RUNTIME_SOURCE_KEYWORDS
    ):

        continue


    if not isinstance(
        variable_value,
        (
            dict,
            list,
        ),
    ):

        continue


    BLOCK_3_RUNTIME_CANDIDATES.append(
        {
            "source_name":
                variable_name,

            "payload":
                variable_value,

            "source_type":
                "runtime",
        }
    )


# =============================================================================
# Drive candidate discovery
# =============================================================================
#
# This searches only for structured project artefacts that might contain
# genuine textual records.
#
# Training checkpoints, metadata and supervision/target artefacts are excluded.
# =============================================================================

BLOCK_3_DRIVE_QUERY_TERMS = (
    "notebook_",
    "corpus",
    "article",
    "prepared",
)


BLOCK_3_DRIVE_FILE_RESULTS = {}


for query_term in BLOCK_3_DRIVE_QUERY_TERMS:

    result = (
        NOTEBOOK_10_DRIVE_SERVICE.files()
        .list(
            q=(
                "trashed = false "
                f"and name contains '{query_term}'"
            ),

            spaces=
                "drive",

            fields=(
                "files("
                "id,"
                "name,"
                "size,"
                "mimeType,"
                "modifiedTime"
                ")"
            ),

            orderBy=
                "modifiedTime desc",

            pageSize=
                100,
        )
        .execute()
    )


    for file_info in result.get(
        "files",
        [],
    ):

        BLOCK_3_DRIVE_FILE_RESULTS[
            file_info[
                "id"
            ]
        ] = file_info


# =============================================================================
# Drive candidate filtering
# =============================================================================

BLOCK_3_EXCLUDED_FILENAME_TERMS = (
    "checkpoint",
    "metadata",
    "supervision",
    "target_contract",
    "target",
    "annotation",
    "optimizer",
    "optimiser",
    "training_history",
)


BLOCK_3_ALLOWED_STRUCTURED_SUFFIXES = (
    ".json",
    ".jsonl",
)


BLOCK_3_MAX_SOURCE_BYTES = (
    25
    *
    1024
    *
    1024
)


BLOCK_3_DRIVE_CANDIDATE_FILES = []


for file_info in BLOCK_3_DRIVE_FILE_RESULTS.values():

    filename = str(
        file_info.get(
            "name",
            "",
        )
    )


    filename_lower = filename.lower()


    if filename in (
        BLOCK_2_NOTEBOOK_05_CONTEXTUAL_FILENAME,
        NOTEBOOK_09_TO_10_METADATA_FILENAME,
    ):

        continue


    if any(
        excluded_term
        in
        filename_lower

        for excluded_term
        in BLOCK_3_EXCLUDED_FILENAME_TERMS
    ):

        continue


    if not filename_lower.endswith(
        BLOCK_3_ALLOWED_STRUCTURED_SUFFIXES
    ):

        continue


    try:

        file_size = int(
            file_info.get(
                "size",
                0,
            )
        )

    except (
        TypeError,
        ValueError,
    ):

        file_size = 0


    if (
        file_size
        <=
        0
        or
        file_size
        >
        BLOCK_3_MAX_SOURCE_BYTES
    ):

        continue


    BLOCK_3_DRIVE_CANDIDATE_FILES.append(
        file_info
    )


# =============================================================================
# Structured-payload parsing
# =============================================================================

def block_3_parse_structured_bytes(
    raw_bytes,
    filename,
):

    if not raw_bytes:

        raise RuntimeError(
            f"Structured source '{filename}' is empty."
        )


    filename_lower = filename.lower()


    if filename_lower.endswith(
        ".json"
    ):

        return json.loads(
            raw_bytes.decode(
                "utf-8"
            )
        )


    if filename_lower.endswith(
        ".jsonl"
    ):

        records = []


        decoded_text = raw_bytes.decode(
            "utf-8"
        )


        for (
            line_number,
            line,
        ) in enumerate(
            decoded_text.splitlines()
        ):

            line = line.strip()


            if not line:

                continue


            try:

                record = json.loads(
                    line
                )

            except json.JSONDecodeError as exc:

                raise RuntimeError(
                    f"Invalid JSONL record in '{filename}' "
                    f"at line {line_number + 1}."
                ) from exc


            records.append(
                record
            )


        return records


    raise RuntimeError(
        f"Unsupported structured source '{filename}'."
    )


# =============================================================================
# Metadata scalar search
# =============================================================================

def block_3_find_scalar_values(
    payload,
    wanted_keys,
    max_depth=5,
):

    found_values = []


    def recurse(
        value,
        depth,
    ):

        if depth > max_depth:

            return


        if isinstance(
            value,
            dict,
        ):

            for (
                key,
                child,
            ) in value.items():

                if (
                    key.lower()
                    in
                    wanted_keys
                    and
                    isinstance(
                        child,
                        (
                            str,
                            int,
                            float,
                            bool,
                        ),
                    )
                ):

                    found_values.append(
                        child
                    )


                if isinstance(
                    child,
                    (
                        dict,
                        list,
                    ),
                ):

                    recurse(
                        child,
                        depth + 1,
                    )


        elif isinstance(
            value,
            list,
        ):

            for child in value[:250]:

                if isinstance(
                    child,
                    (
                        dict,
                        list,
                    ),
                ):

                    recurse(
                        child,
                        depth + 1,
                    )


    recurse(
        payload,
        0,
    )


    return found_values


# =============================================================================
# Encoder-provenance audit
# =============================================================================

def block_3_encoder_contract(
    payload,
):

    model_values = block_3_find_scalar_values(
        payload,
        {
            "model",
            "model_name",
            "encoder_model",
            "embedding_model",
        },
    )


    provider_values = block_3_find_scalar_values(
        payload,
        {
            "provider",
            "encoder_provider",
            "embedding_provider",
        },
    )


    dimension_values = block_3_find_scalar_values(
        payload,
        {
            "embedding_dimension",
            "embedding_dim",
            "representation_dimension",
        },
    )


    normalised_values = block_3_find_scalar_values(
        payload,
        {
            "normalised",
            "normalized",
            "normalize_embeddings",
        },
    )


    model_match = any(
        str(
            value
        ).strip()
        ==
        BLOCK_3_EXPECTED_EMBEDDING_MODEL

        for value
        in model_values
    )


    provider_match = any(
        str(
            value
        ).strip()
        ==
        BLOCK_3_EXPECTED_EMBEDDING_PROVIDER

        for value
        in provider_values
    )


    dimension_match = any(
        block_3_normalise_position(
            value
        )
        ==
        BLOCK_3_EXPECTED_EMBEDDING_DIMENSION

        for value
        in dimension_values
    )


    normalised_match = any(
        value
        is True

        for value
        in normalised_values
    )


    # -------------------------------------------------------------------------
    # For representation reuse, the model and dimension must be explicit.
    #
    # Provider and normalisation are audited separately because older project
    # artefacts may omit one of those metadata fields.
    # -------------------------------------------------------------------------

    verified_for_reuse = all(
        [
            model_match,
            dimension_match,
        ]
    )


    return {
        "model_match":
            model_match,

        "provider_match":
            provider_match,

        "dimension_match":
            dimension_match,

        "normalised_match":
            normalised_match,

        "verified_for_reuse":
            verified_for_reuse,
    }


# =============================================================================
# Recursive sentence-record extraction
# =============================================================================
#
# Article and paragraph identity may be inherited from enclosing dictionary
# objects.
#
# Generic "id" is intentionally not accepted as sentence identity because it
# can refer to article, paragraph or source records and would make sentence
# identity ambiguous.
# =============================================================================

def block_3_extract_sentence_records(
    payload,
    source_name,
    representation_contract_verified,
):

    extracted_records = []


    def recurse(
        value,
        inherited_article_id=None,
        inherited_paragraph_id=None,
        depth=0,
    ):

        if depth > 8:

            return


        if isinstance(
            value,
            dict,
        ):

            local_article_id = block_3_normalise_identifier(
                block_3_first_present(
                    value,
                    BLOCK_3_ARTICLE_ID_KEYS,
                )
            )


            if local_article_id is None:

                local_article_id = (
                    inherited_article_id
                )


            local_paragraph_id = block_3_normalise_identifier(
                block_3_first_present(
                    value,
                    BLOCK_3_PARAGRAPH_ID_KEYS,
                )
            )


            if local_paragraph_id is None:

                local_paragraph_id = (
                    inherited_paragraph_id
                )


            sentence_id = block_3_normalise_identifier(
                block_3_first_present(
                    value,
                    BLOCK_3_SENTENCE_ID_KEYS,
                )
            )


            sequence_position = block_3_normalise_position(
                block_3_first_present(
                    value,
                    BLOCK_3_SEQUENCE_POSITION_KEYS,
                )
            )


            sentence_text = block_3_normalise_text(
                block_3_first_present(
                    value,
                    BLOCK_3_TEXT_KEYS,
                )
            )


            representation = block_3_to_vector(
                block_3_first_present(
                    value,
                    BLOCK_3_REPRESENTATION_KEYS,
                )
            )


            sentence_like = all(
                [
                    local_article_id
                    is not None,

                    sentence_id
                    is not None,

                    sequence_position
                    is not None,

                    (
                        sentence_text
                        is not None
                        or
                        representation
                        is not None
                    ),
                ]
            )


            if sentence_like:

                representation_shape_valid = (
                    representation
                    is not None
                    and
                    representation.shape
                    ==
                    (
                        BLOCK_3_EXPECTED_EMBEDDING_DIMENSION,
                    )
                )


                representation_finite = (
                    representation_shape_valid
                    and
                    bool(
                        np.isfinite(
                            representation
                        ).all()
                    )
                )


                representation_reusable = all(
                    [
                        representation_shape_valid,
                        representation_finite,
                        representation_contract_verified,
                    ]
                )


                extracted_records.append(
                    {
                        "article_id":
                            local_article_id,

                        "paragraph_id":
                            local_paragraph_id,

                        "sentence_id":
                            sentence_id,

                        "sequence_position":
                            sequence_position,

                        "sentence_text":
                            sentence_text,

                        "contextual_representation":
                            (
                                representation.copy()
                                if representation_reusable
                                else None
                            ),

                        "representation_present":
                            (
                                representation
                                is not None
                            ),

                        "representation_shape_valid":
                            bool(
                                representation_shape_valid
                            ),

                        "representation_finite":
                            bool(
                                representation_finite
                            ),

                        "representation_contract_verified":
                            bool(
                                representation_contract_verified
                            ),

                        "representation_reusable":
                            bool(
                                representation_reusable
                            ),

                        "source_name":
                            source_name,
                    }
                )


            for child in value.values():

                if isinstance(
                    child,
                    (
                        dict,
                        list,
                    ),
                ):

                    recurse(
                        child,
                        inherited_article_id=
                            local_article_id,

                        inherited_paragraph_id=
                            local_paragraph_id,

                        depth=
                            depth + 1,
                    )


        elif isinstance(
            value,
            list,
        ):

            for child in value:

                if isinstance(
                    child,
                    (
                        dict,
                        list,
                    ),
                ):

                    recurse(
                        child,
                        inherited_article_id=
                            inherited_article_id,

                        inherited_paragraph_id=
                            inherited_paragraph_id,

                        depth=
                            depth + 1,
                    )


    recurse(
        payload
    )


    return extracted_records


# =============================================================================
# Source audit containers
# =============================================================================

BLOCK_3_SOURCE_AUDIT = []

BLOCK_3_EXTRACTED_RECORDS = []

BLOCK_3_SOURCE_READ_ERRORS = []


# =============================================================================
# Inspect runtime candidates
# =============================================================================

for runtime_source in BLOCK_3_RUNTIME_CANDIDATES:

    try:

        payload = runtime_source[
            "payload"
        ]


        encoder_contract = (
            block_3_encoder_contract(
                payload
            )
        )


        extracted_records = (
            block_3_extract_sentence_records(
                payload=
                    payload,

                source_name=
                    runtime_source[
                        "source_name"
                    ],

                representation_contract_verified=
                    encoder_contract[
                        "verified_for_reuse"
                    ],
            )
        )


        BLOCK_3_SOURCE_AUDIT.append(
            {
                "source_name":
                    runtime_source[
                        "source_name"
                    ],

                "source_type":
                    "runtime",

                "encoder_contract":
                    encoder_contract,

                "records_extracted":
                    len(
                        extracted_records
                    ),

                "read_valid":
                    True,
            }
        )


        BLOCK_3_EXTRACTED_RECORDS.extend(
            extracted_records
        )


    except Exception as exc:

        BLOCK_3_SOURCE_READ_ERRORS.append(
            {
                "source_name":
                    runtime_source[
                        "source_name"
                    ],

                "source_type":
                    "runtime",

                "error_type":
                    type(
                        exc
                    ).__name__,

                "error":
                    str(
                        exc
                    ),
            }
        )


# =============================================================================
# Inspect Drive candidate sources
# =============================================================================

for file_info in BLOCK_3_DRIVE_CANDIDATE_FILES:

    file_id = file_info[
        "id"
    ]


    filename = file_info[
        "name"
    ]


    try:

        raw_bytes = block_3_download_drive_bytes(
            NOTEBOOK_10_DRIVE_SERVICE,
            file_id,
        )


        payload = block_3_parse_structured_bytes(
            raw_bytes,
            filename,
        )


        encoder_contract = (
            block_3_encoder_contract(
                payload
            )
        )


        extracted_records = (
            block_3_extract_sentence_records(
                payload=
                    payload,

                source_name=
                    filename,

                representation_contract_verified=
                    encoder_contract[
                        "verified_for_reuse"
                    ],
            )
        )


        BLOCK_3_SOURCE_AUDIT.append(
            {
                "source_name":
                    filename,

                "source_type":
                    "drive",

                "drive_file_id":
                    file_id,

                "encoder_contract":
                    encoder_contract,

                "records_extracted":
                    len(
                        extracted_records
                    ),

                "read_valid":
                    True,
            }
        )


        BLOCK_3_EXTRACTED_RECORDS.extend(
            extracted_records
        )


    except Exception as exc:

        BLOCK_3_SOURCE_READ_ERRORS.append(
            {
                "source_name":
                    filename,

                "source_type":
                    "drive",

                "drive_file_id":
                    file_id,

                "error_type":
                    type(
                        exc
                    ).__name__,

                "error":
                    str(
                        exc
                    ),
            }
        )


# =============================================================================
# Remove records belonging to the existing pilot article
# =============================================================================
#
# A recovered copy of the pilot article is not an additional independent
# article.
# =============================================================================

BLOCK_3_NON_PILOT_EXTRACTED_RECORDS = []

BLOCK_3_PILOT_DUPLICATE_RECORD_COUNT = 0


for record in BLOCK_3_EXTRACTED_RECORDS:

    article_id = record[
        "article_id"
    ]


    sentence_id = record[
        "sentence_id"
    ]


    if (
        article_id
        in
        BLOCK_3_PILOT_ARTICLE_IDS
    ):

        BLOCK_3_PILOT_DUPLICATE_RECORD_COUNT += 1

        continue


    if (
        sentence_id
        in
        BLOCK_3_PILOT_SENTENCE_IDS
    ):

        BLOCK_3_PILOT_DUPLICATE_RECORD_COUNT += 1

        continue


    BLOCK_3_NON_PILOT_EXTRACTED_RECORDS.append(
        record
    )


# =============================================================================
# Merge repeated evidence for the same sentence
# =============================================================================
#
# The same sentence may appear in more than one project artefact.
#
# It may be merged only when article identity and sequence position agree.
# =============================================================================

BLOCK_3_MERGED_SENTENCE_RECORDS = {}

BLOCK_3_CONFLICTING_SENTENCE_RECORDS = {}


for record in BLOCK_3_NON_PILOT_EXTRACTED_RECORDS:

    sentence_id = record[
        "sentence_id"
    ]


    if (
        sentence_id
        not in
        BLOCK_3_MERGED_SENTENCE_RECORDS
    ):

        merged_record = deepcopy(
            record
        )


        merged_record[
            "source_names"
        ] = [
            record[
                "source_name"
            ]
        ]


        BLOCK_3_MERGED_SENTENCE_RECORDS[
            sentence_id
        ] = merged_record


        continue


    existing = (
        BLOCK_3_MERGED_SENTENCE_RECORDS[
            sentence_id
        ]
    )


    # -------------------------------------------------------------------------
    # Structural identity must agree
    # -------------------------------------------------------------------------

    structural_match = all(
        [
            (
                existing[
                    "article_id"
                ]
                ==
                record[
                    "article_id"
                ]
            ),

            (
                existing[
                    "sequence_position"
                ]
                ==
                record[
                    "sequence_position"
                ]
            ),
        ]
    )


    if not structural_match:

        BLOCK_3_CONFLICTING_SENTENCE_RECORDS[
            sentence_id
        ] = {
            "reason":
                "structural_identity_conflict",

            "existing_article_id":
                existing[
                    "article_id"
                ],

            "new_article_id":
                record[
                    "article_id"
                ],

            "existing_sequence_position":
                existing[
                    "sequence_position"
                ],

            "new_sequence_position":
                record[
                    "sequence_position"
                ],
        }


        continue


    # -------------------------------------------------------------------------
    # Preserve paragraph identity when one source provides it
    # -------------------------------------------------------------------------

    if (
        existing.get(
            "paragraph_id"
        )
        is None
        and
        record.get(
            "paragraph_id"
        )
        is not None
    ):

        existing[
            "paragraph_id"
        ] = record[
            "paragraph_id"
        ]


    # -------------------------------------------------------------------------
    # Merge textual evidence
    # -------------------------------------------------------------------------

    existing_text = existing.get(
        "sentence_text"
    )


    new_text = record.get(
        "sentence_text"
    )


    if (
        existing_text is None
        and
        new_text is not None
    ):

        existing[
            "sentence_text"
        ] = new_text


    elif (
        existing_text is not None
        and
        new_text is not None
        and
        existing_text
        !=
        new_text
    ):

        BLOCK_3_CONFLICTING_SENTENCE_RECORDS[
            sentence_id
        ] = {
            "reason":
                "sentence_text_conflict",

            "article_id":
                existing[
                    "article_id"
                ],
        }


        continue


    # -------------------------------------------------------------------------
    # Merge only representations already validated for reuse
    # -------------------------------------------------------------------------

    existing_representation = (
        existing.get(
            "contextual_representation"
        )
    )


    new_representation = (
        record.get(
            "contextual_representation"
        )
    )


    if (
        existing_representation is None
        and
        new_representation is not None
    ):

        existing[
            "contextual_representation"
        ] = new_representation.copy()


        existing[
            "representation_reusable"
        ] = True


        existing[
            "representation_contract_verified"
        ] = True


    elif (
        existing_representation is not None
        and
        new_representation is not None
    ):

        if not np.array_equal(
            existing_representation,
            new_representation,
            equal_nan=
                True,
        ):

            BLOCK_3_CONFLICTING_SENTENCE_RECORDS[
                sentence_id
            ] = {
                "reason":
                    "validated_representation_conflict",

                "article_id":
                    existing[
                        "article_id"
                    ],
            }


            continue


    if (
        record[
            "source_name"
        ]
        not in
        existing[
            "source_names"
        ]
    ):

        existing[
            "source_names"
        ].append(
            record[
                "source_name"
            ]
        )


# =============================================================================
# Remove all conflicted sentence records
# =============================================================================

for sentence_id in BLOCK_3_CONFLICTING_SENTENCE_RECORDS:

    BLOCK_3_MERGED_SENTENCE_RECORDS.pop(
        sentence_id,
        None,
    )


# =============================================================================
# Group additional candidate records by article
# =============================================================================

BLOCK_3_ADDITIONAL_ARTICLE_CANDIDATES = {}


for record in BLOCK_3_MERGED_SENTENCE_RECORDS.values():

    BLOCK_3_ADDITIONAL_ARTICLE_CANDIDATES.setdefault(
        record[
            "article_id"
        ],
        [],
    ).append(
        record
    )


# =============================================================================
# Structural validation of additional articles
# =============================================================================

BLOCK_3_ADDITIONAL_ARTICLE_AUDIT = {}

BLOCK_3_STRUCTURALLY_ELIGIBLE_ADDITIONAL_ARTICLES = {}

BLOCK_3_EXCLUDED_ADDITIONAL_ARTICLES = {}


for (
    article_id,
    article_records,
) in BLOCK_3_ADDITIONAL_ARTICLE_CANDIDATES.items():

    sentence_ids = [
        record[
            "sentence_id"
        ]

        for record
        in article_records
    ]


    sequence_positions = [
        record[
            "sequence_position"
        ]

        for record
        in article_records
    ]


    article_identity_valid = (
        article_id
        is not None
    )


    non_empty = (
        len(
            article_records
        )
        >
        0
    )


    sentence_ids_valid = all(
        sentence_id
        is not None

        for sentence_id
        in sentence_ids
    )


    sentence_ids_unique = (
        len(
            sentence_ids
        )
        ==
        len(
            set(
                sentence_ids
            )
        )
    )


    sequence_positions_valid = all(
        (
            isinstance(
                position,
                int,
            )
            and
            position
            >=
            0
        )

        for position
        in sequence_positions
    )


    sequence_positions_unique = (
        len(
            sequence_positions
        )
        ==
        len(
            set(
                sequence_positions
            )
        )
    )


    sorted_records = sorted(
        article_records,
        key=
            lambda record:
                (
                    record[
                        "sequence_position"
                    ],
                    record[
                        "sentence_id"
                    ],
                ),
    )


    sorted_positions = [
        record[
            "sequence_position"
        ]

        for record
        in sorted_records
    ]


    zero_based_contiguous = (
        sorted_positions
        ==
        list(
            range(
                len(
                    sorted_positions
                )
            )
        )
    )


    one_based_contiguous = (
        sorted_positions
        ==
        list(
            range(
                1,
                len(
                    sorted_positions
                )
                +
                1
            )
        )
    )


    canonical_sequence_valid = (
        zero_based_contiguous
        or
        one_based_contiguous
    )


    if zero_based_contiguous:

        sequence_position_base = (
            "zero_based"
        )

    elif one_based_contiguous:

        sequence_position_base = (
            "one_based"
        )

    else:

        sequence_position_base = None


    sentence_text_available = sum(
        int(
            record.get(
                "sentence_text"
            )
            is not None
        )

        for record
        in sorted_records
    )


    validated_representation_available = sum(
        int(
            record.get(
                "contextual_representation"
            )
            is not None
        )

        for record
        in sorted_records
    )


    exclusion_reasons = []


    if not article_identity_valid:

        exclusion_reasons.append(
            "invalid_article_identity"
        )


    if not non_empty:

        exclusion_reasons.append(
            "empty_article"
        )


    if not sentence_ids_valid:

        exclusion_reasons.append(
            "invalid_sentence_identity"
        )


    if not sentence_ids_unique:

        exclusion_reasons.append(
            "duplicate_sentence_identity"
        )


    if not sequence_positions_valid:

        exclusion_reasons.append(
            "invalid_sequence_position"
        )


    if not sequence_positions_unique:

        exclusion_reasons.append(
            "duplicate_sequence_position"
        )


    if not canonical_sequence_valid:

        exclusion_reasons.append(
            "non_contiguous_sequence"
        )


    structurally_eligible = (
        len(
            exclusion_reasons
        )
        ==
        0
    )


    representation_ready = (
        structurally_eligible
        and
        all(
            (
                record.get(
                    "contextual_representation"
                )
                is not None
                and
                tuple(
                    record[
                        "contextual_representation"
                    ].shape
                )
                ==
                (
                    BLOCK_3_EXPECTED_EMBEDDING_DIMENSION,
                )
                and
                bool(
                    np.isfinite(
                        record[
                            "contextual_representation"
                        ]
                    ).all()
                )
            )

            for record
            in sorted_records
        )
    )


    representation_generation_required = (
        structurally_eligible
        and
        not representation_ready
    )


    representation_generation_possible = (
        representation_generation_required
        and
        all(
            record.get(
                "sentence_text"
            )
            is not None

            for record
            in sorted_records
        )
    )


    BLOCK_3_ADDITIONAL_ARTICLE_AUDIT[
        article_id
    ] = {
        "article_id":
            article_id,

        "sentence_count":
            len(
                sorted_records
            ),

        "article_identity_valid":
            article_identity_valid,

        "sentence_ids_valid":
            sentence_ids_valid,

        "sentence_ids_unique":
            sentence_ids_unique,

        "sequence_positions_valid":
            sequence_positions_valid,

        "sequence_positions_unique":
            sequence_positions_unique,

        "canonical_sequence_valid":
            canonical_sequence_valid,

        "sequence_position_base":
            sequence_position_base,

        "sentence_text_available":
            sentence_text_available,

        "validated_representation_available":
            validated_representation_available,

        "structurally_eligible":
            structurally_eligible,

        "representation_ready":
            representation_ready,

        "representation_generation_required":
            representation_generation_required,

        "representation_generation_possible":
            representation_generation_possible,

        "supervision_status":
            "alignment_not_yet_assessed",

        "exclusion_reasons":
            exclusion_reasons,
    }


    if structurally_eligible:

        BLOCK_3_STRUCTURALLY_ELIGIBLE_ADDITIONAL_ARTICLES[
            article_id
        ] = sorted_records

    else:

        BLOCK_3_EXCLUDED_ADDITIONAL_ARTICLES[
            article_id
        ] = exclusion_reasons


# =============================================================================
# Global additional-corpus identity audit
# =============================================================================

BLOCK_3_ADDITIONAL_SENTENCE_IDS = [
    record[
        "sentence_id"
    ]

    for records
    in BLOCK_3_STRUCTURALLY_ELIGIBLE_ADDITIONAL_ARTICLES.values()

    for record
    in records
]


BLOCK_3_ADDITIONAL_SENTENCE_IDS_UNIQUE = (
    len(
        BLOCK_3_ADDITIONAL_SENTENCE_IDS
    )
    ==
    len(
        set(
            BLOCK_3_ADDITIONAL_SENTENCE_IDS
        )
    )
)


BLOCK_3_NO_PILOT_SENTENCE_COLLISION = (
    set(
        BLOCK_3_ADDITIONAL_SENTENCE_IDS
    )
    .isdisjoint(
        BLOCK_3_PILOT_SENTENCE_IDS
    )
)


BLOCK_3_NO_PILOT_ARTICLE_COLLISION = (
    set(
        BLOCK_3_STRUCTURALLY_ELIGIBLE_ADDITIONAL_ARTICLES.keys()
    )
    .isdisjoint(
        BLOCK_3_PILOT_ARTICLE_IDS
    )
)


# =============================================================================
# Representation-readiness partition
# =============================================================================

BLOCK_3_REPRESENTATION_READY_ADDITIONAL_ARTICLES = {}

BLOCK_3_REPRESENTATION_GENERATION_REQUIRED_ARTICLES = {}

BLOCK_3_REPRESENTATION_GENERATION_BLOCKED_ARTICLES = {}


for (
    article_id,
    records,
) in BLOCK_3_STRUCTURALLY_ELIGIBLE_ADDITIONAL_ARTICLES.items():

    audit = (
        BLOCK_3_ADDITIONAL_ARTICLE_AUDIT[
            article_id
        ]
    )


    if audit[
        "representation_ready"
    ]:

        BLOCK_3_REPRESENTATION_READY_ADDITIONAL_ARTICLES[
            article_id
        ] = records


    elif audit[
        "representation_generation_possible"
    ]:

        BLOCK_3_REPRESENTATION_GENERATION_REQUIRED_ARTICLES[
            article_id
        ] = records


    else:

        BLOCK_3_REPRESENTATION_GENERATION_BLOCKED_ARTICLES[
            article_id
        ] = records


# =============================================================================
# Merge pilot and structurally eligible additional articles
# =============================================================================

NOTEBOOK_10_EXPANDED_ARTICLE_RECORDS = {
    article_id:
        deepcopy(
            records
        )

    for (
        article_id,
        records,
    ) in BLOCK_3_PILOT_ARTICLE_RECORDS.items()
}


for (
    article_id,
    records,
) in BLOCK_3_STRUCTURALLY_ELIGIBLE_ADDITIONAL_ARTICLES.items():

    if (
        article_id
        in
        NOTEBOOK_10_EXPANDED_ARTICLE_RECORDS
    ):

        raise RuntimeError(
            "An additional article identity collides with "
            "the validated pilot article."
        )


    NOTEBOOK_10_EXPANDED_ARTICLE_RECORDS[
        article_id
    ] = deepcopy(
        records
    )


# =============================================================================
# Construct zero-based canonical sequence records
# =============================================================================
#
# Source sequence position is retained separately.
#
# The normalised sequence position is used only as an internal deterministic
# ordering contract for subsequent Notebook 10 blocks.
# =============================================================================

NOTEBOOK_10_EXPANDED_CANONICAL_SENTENCE_RECORDS = []


for article_id in sorted(
    NOTEBOOK_10_EXPANDED_ARTICLE_RECORDS.keys()
):

    sorted_records = sorted(
        NOTEBOOK_10_EXPANDED_ARTICLE_RECORDS[
            article_id
        ],
        key=
            lambda record:
                (
                    record[
                        "sequence_position"
                    ],
                    record[
                        "sentence_id"
                    ],
                ),
    )


    for (
        canonical_position,
        record,
    ) in enumerate(
        sorted_records
    ):

        canonical_record = deepcopy(
            record
        )


        canonical_record[
            "source_sequence_position"
        ] = canonical_record[
            "sequence_position"
        ]


        canonical_record[
            "canonical_sequence_position"
        ] = int(
            canonical_position
        )


        canonical_record[
            "sequence_position"
        ] = int(
            canonical_position
        )


        NOTEBOOK_10_EXPANDED_CANONICAL_SENTENCE_RECORDS.append(
            canonical_record
        )


# =============================================================================
# Complete merged identity audit
# =============================================================================

BLOCK_3_EXPANDED_ARTICLE_IDS = list(
    NOTEBOOK_10_EXPANDED_ARTICLE_RECORDS.keys()
)


BLOCK_3_EXPANDED_SENTENCE_IDS = [
    record[
        "sentence_id"
    ]

    for record
    in NOTEBOOK_10_EXPANDED_CANONICAL_SENTENCE_RECORDS
]


BLOCK_3_EXPANDED_ARTICLE_IDS_UNIQUE = (
    len(
        BLOCK_3_EXPANDED_ARTICLE_IDS
    )
    ==
    len(
        set(
            BLOCK_3_EXPANDED_ARTICLE_IDS
        )
    )
)


BLOCK_3_EXPANDED_SENTENCE_IDS_UNIQUE = (
    len(
        BLOCK_3_EXPANDED_SENTENCE_IDS
    )
    ==
    len(
        set(
            BLOCK_3_EXPANDED_SENTENCE_IDS
        )
    )
)


# =============================================================================
# Validate merged canonical ordering
# =============================================================================

BLOCK_3_EXPANDED_CANONICAL_SEQUENCE_VALID = True


for article_id in BLOCK_3_EXPANDED_ARTICLE_IDS:

    article_records = [
        record

        for record
        in NOTEBOOK_10_EXPANDED_CANONICAL_SENTENCE_RECORDS

        if (
            record[
                "article_id"
            ]
            ==
            article_id
        )
    ]


    observed_positions = [
        record[
            "canonical_sequence_position"
        ]

        for record
        in article_records
    ]


    expected_positions = list(
        range(
            len(
                article_records
            )
        )
    )


    if (
        observed_positions
        !=
        expected_positions
    ):

        BLOCK_3_EXPANDED_CANONICAL_SEQUENCE_VALID = False

        break


# =============================================================================
# Determine representation-ready merged articles
# =============================================================================

BLOCK_3_EXPANDED_REPRESENTATION_READY_ARTICLES = {}


for (
    article_id,
    records,
) in NOTEBOOK_10_EXPANDED_ARTICLE_RECORDS.items():

    representation_ready = all(
        (
            record.get(
                "contextual_representation"
            )
            is not None
            and
            tuple(
                record[
                    "contextual_representation"
                ].shape
            )
            ==
            (
                BLOCK_3_EXPECTED_EMBEDDING_DIMENSION,
            )
            and
            bool(
                np.isfinite(
                    record[
                        "contextual_representation"
                    ]
                ).all()
            )
        )

        for record
        in records
    )


    if representation_ready:

        BLOCK_3_EXPANDED_REPRESENTATION_READY_ARTICLES[
            article_id
        ] = records


# =============================================================================
# Corpus counts
# =============================================================================

BLOCK_3_ADDITIONAL_ARTICLE_COUNT = len(
    BLOCK_3_STRUCTURALLY_ELIGIBLE_ADDITIONAL_ARTICLES
)


BLOCK_3_ADDITIONAL_SENTENCE_COUNT = sum(
    len(
        records
    )

    for records
    in BLOCK_3_STRUCTURALLY_ELIGIBLE_ADDITIONAL_ARTICLES.values()
)


BLOCK_3_TOTAL_ARTICLE_COUNT = len(
    NOTEBOOK_10_EXPANDED_ARTICLE_RECORDS
)


BLOCK_3_TOTAL_SENTENCE_COUNT = len(
    NOTEBOOK_10_EXPANDED_CANONICAL_SENTENCE_RECORDS
)


BLOCK_3_REPRESENTATION_READY_ARTICLE_COUNT = len(
    BLOCK_3_EXPANDED_REPRESENTATION_READY_ARTICLES
)


BLOCK_3_REPRESENTATION_GENERATION_REQUIRED_COUNT = len(
    BLOCK_3_REPRESENTATION_GENERATION_REQUIRED_ARTICLES
)


BLOCK_3_REPRESENTATION_GENERATION_BLOCKED_COUNT = len(
    BLOCK_3_REPRESENTATION_GENERATION_BLOCKED_ARTICLES
)


# =============================================================================
# Source / provenance audit
# =============================================================================

BLOCK_3_SOURCES_WITH_EXTRACTED_RECORDS = [
    source_audit

    for source_audit
    in BLOCK_3_SOURCE_AUDIT

    if (
        source_audit[
            "records_extracted"
        ]
        >
        0
    )
]


BLOCK_3_ADDITIONAL_CORPUS_PROVENANCE_AVAILABLE = (
    len(
        BLOCK_3_SOURCES_WITH_EXTRACTED_RECORDS
    )
    >
    0
)


BLOCK_3_CORPUS_PROVENANCE_AVAILABLE = all(
    [
        BLOCK_3_PILOT_ARTICLE_COUNT
        >
        0,

        (
            BLOCK_3_ADDITIONAL_ARTICLE_COUNT
            ==
            0
            or
            BLOCK_3_ADDITIONAL_CORPUS_PROVENANCE_AVAILABLE
        ),
    ]
)


# =============================================================================
# Synthetic-fallback audit
# =============================================================================

BLOCK_3_SYNTHETIC_FALLBACK_USED = any(
    [
        NOTEBOOK_10_BLOCK_3_SYNTHETIC_ARTICLE_CREATED,
        NOTEBOOK_10_BLOCK_3_PILOT_ARTICLE_SPLIT,
        NOTEBOOK_10_BLOCK_3_PILOT_ARTICLE_DUPLICATED,
    ]
)


# =============================================================================
# Supervision-status inventory
# =============================================================================

NOTEBOOK_10_EXPANDED_ARTICLE_SUPERVISION_STATUS = {}


for article_id in BLOCK_3_PILOT_ARTICLE_IDS:

    NOTEBOOK_10_EXPANDED_ARTICLE_SUPERVISION_STATUS[
        article_id
    ] = (
        "available_in_separate_validated_project_artefacts"
    )


for article_id in BLOCK_3_STRUCTURALLY_ELIGIBLE_ADDITIONAL_ARTICLES:

    NOTEBOOK_10_EXPANDED_ARTICLE_SUPERVISION_STATUS[
        article_id
    ] = (
        "alignment_not_yet_assessed"
    )


# =============================================================================
# Multi-article validation sufficiency
# =============================================================================

BLOCK_3_STRUCTURAL_MULTI_ARTICLE_SUFFICIENCY = (
    BLOCK_3_TOTAL_ARTICLE_COUNT
    >=
    BLOCK_3_MINIMUM_MULTI_ARTICLE_COUNT
)


BLOCK_3_REPRESENTATION_MULTI_ARTICLE_SUFFICIENCY = (
    BLOCK_3_REPRESENTATION_READY_ARTICLE_COUNT
    >=
    BLOCK_3_MINIMUM_MULTI_ARTICLE_COUNT
)


BLOCK_3_MULTI_ARTICLE_VALIDATION_SOURCE_READY = all(
    [
        BLOCK_3_STRUCTURAL_MULTI_ARTICLE_SUFFICIENCY,
        BLOCK_3_REPRESENTATION_MULTI_ARTICLE_SUFFICIENCY,
        BLOCK_3_EXPANDED_ARTICLE_IDS_UNIQUE,
        BLOCK_3_EXPANDED_SENTENCE_IDS_UNIQUE,
        BLOCK_3_EXPANDED_CANONICAL_SEQUENCE_VALID,
        BLOCK_3_NO_PILOT_SENTENCE_COLLISION,
        BLOCK_3_NO_PILOT_ARTICLE_COLLISION,
        BLOCK_3_CORPUS_PROVENANCE_AVAILABLE,
        not
        BLOCK_3_SYNTHETIC_FALLBACK_USED,
    ]
)


# =============================================================================
# Expanded-training sufficiency
# =============================================================================
#
# Notebook 10 validates architecture.
#
# Even if two or more articles are available, that does not establish that
# the corpus is statistically sufficient for substantive model training.
# =============================================================================

BLOCK_3_EXPANDED_TRAINING_SUFFICIENCY = False

BLOCK_3_EXPANDED_TRAINING_SUFFICIENCY_REASON = (
    "not_established_by_validation_project"
)


# =============================================================================
# Remaining corpus requirements
# =============================================================================

BLOCK_3_ADDITIONAL_CORPUS_REQUIRED = (
    not
    BLOCK_3_STRUCTURAL_MULTI_ARTICLE_SUFFICIENCY
)


BLOCK_3_ADDITIONAL_REPRESENTATION_GENERATION_REQUIRED = (
    BLOCK_3_STRUCTURAL_MULTI_ARTICLE_SUFFICIENCY
    and
    not
    BLOCK_3_REPRESENTATION_MULTI_ARTICLE_SUFFICIENCY
    and
    BLOCK_3_REPRESENTATION_GENERATION_REQUIRED_COUNT
    >
    0
)


# =============================================================================
# Exact pilot preservation
# =============================================================================

BLOCK_3_PILOT_ARTICLE_PRESERVED = (
    block_3_objects_exact(
        BLOCK_3_PILOT_ARTICLE_RECORDS,
        BLOCK_3_PILOT_ARTICLE_RECORDS_BEFORE,
    )
)


# =============================================================================
# Inherited model-state validation
# =============================================================================
#
# Compare the post-construction runtime state against the exact Notebook 10
# baseline captured before any corpus-source discovery or preparation.
# =============================================================================

BLOCK_3_REPRESENTATION_STATE_AFTER = (
    block_3_snapshot_module_collection(
        NOTEBOOK_10_REPRESENTATION_MODULES
    )
)


BLOCK_3_TRANSFORMATIVE_STATE_AFTER = (
    block_3_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_MODULES
    )
)


BLOCK_3_WEIGHT_STATE_AFTER = (
    block_3_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES
    )
)


BLOCK_3_REPRESENTATION_STATE_UNCHANGED = (
    block_3_nested_state_exact(
        BLOCK_3_REPRESENTATION_STATE_BASELINE,
        BLOCK_3_REPRESENTATION_STATE_AFTER,
    )
)


BLOCK_3_TRANSFORMATIVE_STATE_UNCHANGED = (
    block_3_nested_state_exact(
        BLOCK_3_TRANSFORMATIVE_STATE_BASELINE,
        BLOCK_3_TRANSFORMATIVE_STATE_AFTER,
    )
)


BLOCK_3_WEIGHT_STATE_UNCHANGED = (
    block_3_nested_state_exact(
        BLOCK_3_WEIGHT_STATE_BASELINE,
        BLOCK_3_WEIGHT_STATE_AFTER,
    )
)


# Semantic identity check for the effective Transformative-Weight vectors.
with torch.no_grad():

    BLOCK_3_EFFECTIVE_WEIGHT_IDENTITY_UNCHANGED = all(
        torch.equal(
            NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES[
                pathway_name
            ]()
            .detach()
            .cpu(),

            NOTEBOOK_10_INHERITED_EFFECTIVE_TRANSFORMATIVE_WEIGHTS[
                pathway_name
            ]
            .detach()
            .cpu(),
        )

        for pathway_name
        in NOTEBOOK_10_PATHWAY_DIMENSIONS
    )


BLOCK_3_ALL_PARAMETERS_FROZEN = all(
    not parameter.requires_grad

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


BLOCK_3_ALL_MODULES_EVAL = all(
    not module.training

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()
)


BLOCK_3_ALL_GRADIENTS_ABSENT = all(
    parameter.grad is None

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


# =============================================================================
# Execution-boundary validation
# =============================================================================

BLOCK_3_EXECUTION_BOUNDARY_VALID = all(
    [
        NOTEBOOK_10_BLOCK_3_MEDIA_AI_REPRESENTATION_FORWARD_EXECUTED
        is False,

        NOTEBOOK_10_BLOCK_3_TRANSFORMATIVE_FORWARD_EXECUTED
        is False,

        NOTEBOOK_10_BLOCK_3_WEIGHT_FORWARD_EXECUTED
        is False,

        NOTEBOOK_10_BLOCK_3_RECURRENT_STATE_CREATED
        is False,

        NOTEBOOK_10_BLOCK_3_RECURRENT_UPDATE_EXECUTED
        is False,

        NOTEBOOK_10_BLOCK_3_LOSS_CALCULATED
        is False,

        NOTEBOOK_10_BLOCK_3_OPTIMIZER_CREATED
        is False,

        NOTEBOOK_10_BLOCK_3_BACKWARD_PASS_EXECUTED
        is False,

        NOTEBOOK_10_BLOCK_3_PARAMETER_UPDATE_EXECUTED
        is False,

        NOTEBOOK_10_BLOCK_3_TRAINING_EXECUTED
        is False,

        NOTEBOOK_10_BLOCK_3_SUPERVISION_USED_AS_TEXT
        is False,

        NOTEBOOK_10_BLOCK_3_SUPERVISION_USED_AS_REPRESENTATION
        is False,
    ]
)


# =============================================================================
# Final Block 3 validity
# =============================================================================
#
# IMPORTANT:
#
# The block does not fail merely because a second article cannot be recovered.
#
# Failure to obtain an expanded corpus is a valid corpus-sufficiency result,
# not an architectural error.
# =============================================================================

NOTEBOOK_10_BLOCK_3_ERRORS = []


BLOCK_3_VALIDATION_CHECKS = {
    "prerequisites_invalid":
        BLOCK_3_PREREQUISITES_VALID,

    "pilot_article_missing":
        (
            BLOCK_3_PILOT_ARTICLE_COUNT
            >=
            1
        ),

    "pilot_article_identity_invalid":
        BLOCK_3_PILOT_ARTICLE_IDENTITIES_VALID,

    "pilot_sentence_identity_invalid":
        BLOCK_3_PILOT_SENTENCE_IDENTITIES_VALID,

    "pilot_article_not_preserved":
        BLOCK_3_PILOT_ARTICLE_PRESERVED,

    "additional_sentence_ids_not_unique":
        BLOCK_3_ADDITIONAL_SENTENCE_IDS_UNIQUE,

    "pilot_sentence_collision":
        BLOCK_3_NO_PILOT_SENTENCE_COLLISION,

    "pilot_article_collision":
        BLOCK_3_NO_PILOT_ARTICLE_COLLISION,

    "expanded_article_ids_not_unique":
        BLOCK_3_EXPANDED_ARTICLE_IDS_UNIQUE,

    "expanded_sentence_ids_not_unique":
        BLOCK_3_EXPANDED_SENTENCE_IDS_UNIQUE,

    "expanded_canonical_sequence_invalid":
        BLOCK_3_EXPANDED_CANONICAL_SEQUENCE_VALID,

    "corpus_provenance_invalid":
        BLOCK_3_CORPUS_PROVENANCE_AVAILABLE,

    "synthetic_fallback_used":
        (
            not
            BLOCK_3_SYNTHETIC_FALLBACK_USED
        ),

    "representation_state_changed":
        BLOCK_3_REPRESENTATION_STATE_UNCHANGED,

    "transformative_state_changed":
        BLOCK_3_TRANSFORMATIVE_STATE_UNCHANGED,

    "weight_state_changed":
        BLOCK_3_WEIGHT_STATE_UNCHANGED,

    "effective_weight_identity_changed":
        BLOCK_3_EFFECTIVE_WEIGHT_IDENTITY_UNCHANGED,

    "parameters_not_frozen":
        BLOCK_3_ALL_PARAMETERS_FROZEN,

    "modules_not_eval":
        BLOCK_3_ALL_MODULES_EVAL,

    "unexpected_gradients":
        BLOCK_3_ALL_GRADIENTS_ABSENT,

    "execution_boundary_invalid":
        BLOCK_3_EXECUTION_BOUNDARY_VALID,
}


for (
    error_name,
    condition,
) in BLOCK_3_VALIDATION_CHECKS.items():

    if not condition:

        NOTEBOOK_10_BLOCK_3_ERRORS.append(
            error_name
        )


NOTEBOOK_10_BLOCK_3_VALID = (
    len(
        NOTEBOOK_10_BLOCK_3_ERRORS
    )
    ==
    0
)


if not NOTEBOOK_10_BLOCK_3_VALID:

    raise RuntimeError(
        "Notebook 10 Block 3 additional-corpus construction failed. "
        f"Errors: {NOTEBOOK_10_BLOCK_3_ERRORS}"
    )


NOTEBOOK_10_BLOCK_3_COMPLETE = True


# =============================================================================
# Stable downstream globals
# =============================================================================

NOTEBOOK_10_ADDITIONAL_ARTICLE_RECORDS = {
    article_id:
        deepcopy(
            records
        )

    for (
        article_id,
        records,
    ) in BLOCK_3_STRUCTURALLY_ELIGIBLE_ADDITIONAL_ARTICLES.items()
}


NOTEBOOK_10_REPRESENTATION_GENERATION_REQUIRED_ARTICLES = {
    article_id:
        deepcopy(
            records
        )

    for (
        article_id,
        records,
    ) in BLOCK_3_REPRESENTATION_GENERATION_REQUIRED_ARTICLES.items()
}


NOTEBOOK_10_EXPANDED_CORPUS_STRUCTURALLY_READY = (
    BLOCK_3_STRUCTURAL_MULTI_ARTICLE_SUFFICIENCY
)


NOTEBOOK_10_EXPANDED_CORPUS_REPRESENTATION_READY = (
    BLOCK_3_REPRESENTATION_MULTI_ARTICLE_SUFFICIENCY
)


NOTEBOOK_10_MULTI_ARTICLE_VALIDATION_SOURCE_READY = (
    BLOCK_3_MULTI_ARTICLE_VALIDATION_SOURCE_READY
)


NOTEBOOK_10_ADDITIONAL_CORPUS_REQUIRED = (
    BLOCK_3_ADDITIONAL_CORPUS_REQUIRED
)


NOTEBOOK_10_ADDITIONAL_REPRESENTATION_GENERATION_REQUIRED = (
    BLOCK_3_ADDITIONAL_REPRESENTATION_GENERATION_REQUIRED
)


NOTEBOOK_10_EXPANDED_TRAINING_SUFFICIENCY = (
    BLOCK_3_EXPANDED_TRAINING_SUFFICIENCY
)


# =============================================================================
# Block summary
# =============================================================================

NOTEBOOK_10_BLOCK_3_SUMMARY = {
    "block":
        NOTEBOOK_10_BLOCK_3,

    "block_name":
        NOTEBOOK_10_BLOCK_3_NAME,

    "block_version":
        NOTEBOOK_10_BLOCK_3_VERSION,

    "pilot_article_count":
        BLOCK_3_PILOT_ARTICLE_COUNT,

    "pilot_sentence_count":
        BLOCK_3_PILOT_SENTENCE_COUNT,

    "runtime_sources_inspected":
        len(
            BLOCK_3_RUNTIME_CANDIDATES
        ),

    "drive_candidate_files":
        len(
            BLOCK_3_DRIVE_CANDIDATE_FILES
        ),

    "sources_with_records":
        len(
            BLOCK_3_SOURCES_WITH_EXTRACTED_RECORDS
        ),

    "source_read_errors":
        len(
            BLOCK_3_SOURCE_READ_ERRORS
        ),

    "additional_article_count":
        BLOCK_3_ADDITIONAL_ARTICLE_COUNT,

    "additional_sentence_count":
        BLOCK_3_ADDITIONAL_SENTENCE_COUNT,

    "total_article_count":
        BLOCK_3_TOTAL_ARTICLE_COUNT,

    "total_sentence_count":
        BLOCK_3_TOTAL_SENTENCE_COUNT,

    "representation_ready_article_count":
        BLOCK_3_REPRESENTATION_READY_ARTICLE_COUNT,

    "representation_generation_required_articles":
        BLOCK_3_REPRESENTATION_GENERATION_REQUIRED_COUNT,

    "representation_generation_blocked_articles":
        BLOCK_3_REPRESENTATION_GENERATION_BLOCKED_COUNT,

    "corpus_provenance_available":
        BLOCK_3_CORPUS_PROVENANCE_AVAILABLE,

    "synthetic_fallback_used":
        BLOCK_3_SYNTHETIC_FALLBACK_USED,

    "structural_multi_article_sufficiency":
        BLOCK_3_STRUCTURAL_MULTI_ARTICLE_SUFFICIENCY,

    "representation_multi_article_sufficiency":
        BLOCK_3_REPRESENTATION_MULTI_ARTICLE_SUFFICIENCY,

    "multi_article_validation_source_ready":
        BLOCK_3_MULTI_ARTICLE_VALIDATION_SOURCE_READY,

    "expanded_training_sufficiency":
        BLOCK_3_EXPANDED_TRAINING_SUFFICIENCY,

    "representation_state_unchanged":
        BLOCK_3_REPRESENTATION_STATE_UNCHANGED,

    "transformative_state_unchanged":
        BLOCK_3_TRANSFORMATIVE_STATE_UNCHANGED,

    "weight_state_unchanged":
        BLOCK_3_WEIGHT_STATE_UNCHANGED,

    "effective_weight_identity_unchanged":
        BLOCK_3_EFFECTIVE_WEIGHT_IDENTITY_UNCHANGED,

    "additional_corpus_required":
        BLOCK_3_ADDITIONAL_CORPUS_REQUIRED,

    "additional_representation_generation_required":
        BLOCK_3_ADDITIONAL_REPRESENTATION_GENERATION_REQUIRED,

    "block_valid":
        NOTEBOOK_10_BLOCK_3_VALID,

    "block_complete":
        NOTEBOOK_10_BLOCK_3_COMPLETE,
}


# =============================================================================
# Final report
# =============================================================================

print("=" * 72)

print(
    "Media AI — Notebook 10, Block 3: "
    "Additional-Corpus Construction "
    "and Multi-Article Source Preparation"
)

print("=" * 72)

print(
    f"Block version                : "
    f"{NOTEBOOK_10_BLOCK_3_VERSION}"
)

print("-" * 72)

print(
    "Existing pilot corpus"
)

print(
    f"Pilot articles               : "
    f"{BLOCK_3_PILOT_ARTICLE_COUNT}"
)

print(
    f"Pilot sentences              : "
    f"{BLOCK_3_PILOT_SENTENCE_COUNT}"
)

print(
    f"Pilot identity preserved     : "
    f"{BLOCK_3_PILOT_ARTICLE_PRESERVED}"
)

print(
    f"Pilot duplicate records seen : "
    f"{BLOCK_3_PILOT_DUPLICATE_RECORD_COUNT}"
)

print("-" * 72)

print(
    "Additional-source discovery"
)

print(
    f"Runtime candidates inspected : "
    f"{len(BLOCK_3_RUNTIME_CANDIDATES)}"
)

print(
    f"Drive candidate files        : "
    f"{len(BLOCK_3_DRIVE_CANDIDATE_FILES)}"
)

print(
    f"Sources with sentence records: "
    f"{len(BLOCK_3_SOURCES_WITH_EXTRACTED_RECORDS)}"
)

print(
    f"Source read errors           : "
    f"{len(BLOCK_3_SOURCE_READ_ERRORS)}"
)

print(
    f"Extracted non-pilot records  : "
    f"{len(BLOCK_3_NON_PILOT_EXTRACTED_RECORDS)}"
)

print(
    f"Conflicting sentence records : "
    f"{len(BLOCK_3_CONFLICTING_SENTENCE_RECORDS)}"
)

print("-" * 72)

print(
    "Additional corpus"
)

print(
    f"Candidate additional articles: "
    f"{len(BLOCK_3_ADDITIONAL_ARTICLE_CANDIDATES)}"
)

print(
    f"Eligible additional articles : "
    f"{BLOCK_3_ADDITIONAL_ARTICLE_COUNT}"
)

print(
    f"Eligible additional sentences: "
    f"{BLOCK_3_ADDITIONAL_SENTENCE_COUNT}"
)

print(
    f"Excluded additional articles : "
    f"{len(BLOCK_3_EXCLUDED_ADDITIONAL_ARTICLES)}"
)

print(
    f"Additional sentence IDs unique: "
    f"{BLOCK_3_ADDITIONAL_SENTENCE_IDS_UNIQUE}"
)

print(
    f"Pilot article collision      : "
    f"{not BLOCK_3_NO_PILOT_ARTICLE_COLLISION}"
)

print(
    f"Pilot sentence collision     : "
    f"{not BLOCK_3_NO_PILOT_SENTENCE_COLLISION}"
)

print("-" * 72)

print(
    "Representation preparation"
)

print(
    f"Expected encoder             : "
    f"{BLOCK_3_EXPECTED_EMBEDDING_MODEL}"
)

print(
    f"Expected dimension           : "
    f"{BLOCK_3_EXPECTED_EMBEDDING_DIMENSION}"
)

print(
    f"Ready additional articles    : "
    f"{len(BLOCK_3_REPRESENTATION_READY_ADDITIONAL_ARTICLES)}"
)

print(
    f"Generation-required articles : "
    f"{BLOCK_3_REPRESENTATION_GENERATION_REQUIRED_COUNT}"
)

print(
    f"Generation-blocked articles  : "
    f"{BLOCK_3_REPRESENTATION_GENERATION_BLOCKED_COUNT}"
)

print("-" * 72)

print(
    "Merged expanded corpus"
)

print(
    f"Total articles               : "
    f"{BLOCK_3_TOTAL_ARTICLE_COUNT}"
)

print(
    f"Total sentences              : "
    f"{BLOCK_3_TOTAL_SENTENCE_COUNT}"
)

print(
    f"Representation-ready articles: "
    f"{BLOCK_3_REPRESENTATION_READY_ARTICLE_COUNT}"
)

print(
    f"Article IDs unique           : "
    f"{BLOCK_3_EXPANDED_ARTICLE_IDS_UNIQUE}"
)

print(
    f"Sentence IDs unique          : "
    f"{BLOCK_3_EXPANDED_SENTENCE_IDS_UNIQUE}"
)

print(
    f"Canonical sequences valid    : "
    f"{BLOCK_3_EXPANDED_CANONICAL_SEQUENCE_VALID}"
)

print(
    f"Corpus provenance available  : "
    f"{BLOCK_3_CORPUS_PROVENANCE_AVAILABLE}"
)

print(
    f"Synthetic fallback used      : "
    f"{BLOCK_3_SYNTHETIC_FALLBACK_USED}"
)

print("-" * 72)

print(
    "Corpus sufficiency"
)

print(
    f"Minimum multi-article count  : "
    f"{BLOCK_3_MINIMUM_MULTI_ARTICLE_COUNT}"
)

print(
    f"Structural multi-article     : "
    f"{BLOCK_3_STRUCTURAL_MULTI_ARTICLE_SUFFICIENCY}"
)

print(
    f"Representation multi-article : "
    f"{BLOCK_3_REPRESENTATION_MULTI_ARTICLE_SUFFICIENCY}"
)

print(
    f"Multi-article source ready   : "
    f"{BLOCK_3_MULTI_ARTICLE_VALIDATION_SOURCE_READY}"
)

print(
    f"Expanded-training sufficient : "
    f"{BLOCK_3_EXPANDED_TRAINING_SUFFICIENCY}"
)

print(
    f"Additional corpus required   : "
    f"{BLOCK_3_ADDITIONAL_CORPUS_REQUIRED}"
)

print(
    f"Representation gen. required : "
    f"{BLOCK_3_ADDITIONAL_REPRESENTATION_GENERATION_REQUIRED}"
)

print("-" * 72)

print(
    "Inherited model state"
)

print(
    f"Representation unchanged     : "
    f"{BLOCK_3_REPRESENTATION_STATE_UNCHANGED}"
)

print(
    f"Transformative unchanged     : "
    f"{BLOCK_3_TRANSFORMATIVE_STATE_UNCHANGED}"
)

print(
    f"Transform Weight unchanged   : "
    f"{BLOCK_3_WEIGHT_STATE_UNCHANGED}"
)

print(
    f"Effective Weight identity    : "
    f"{BLOCK_3_EFFECTIVE_WEIGHT_IDENTITY_UNCHANGED}"
)

print(
    f"All parameters frozen        : "
    f"{BLOCK_3_ALL_PARAMETERS_FROZEN}"
)

print(
    f"All modules eval             : "
    f"{BLOCK_3_ALL_MODULES_EVAL}"
)

print(
    f"All gradients absent         : "
    f"{BLOCK_3_ALL_GRADIENTS_ABSENT}"
)

print("-" * 72)

print(
    "Execution boundary"
)

print(
    f"Media AI repr. forward       : "
    f"{NOTEBOOK_10_BLOCK_3_MEDIA_AI_REPRESENTATION_FORWARD_EXECUTED}"
)

print(
    f"Transform forward            : "
    f"{NOTEBOOK_10_BLOCK_3_TRANSFORMATIVE_FORWARD_EXECUTED}"
)

print(
    f"Weight forward               : "
    f"{NOTEBOOK_10_BLOCK_3_WEIGHT_FORWARD_EXECUTED}"
)

print(
    f"Recurrent state created      : "
    f"{NOTEBOOK_10_BLOCK_3_RECURRENT_STATE_CREATED}"
)

print(
    f"Recurrent update executed    : "
    f"{NOTEBOOK_10_BLOCK_3_RECURRENT_UPDATE_EXECUTED}"
)

print(
    f"Loss calculated              : "
    f"{NOTEBOOK_10_BLOCK_3_LOSS_CALCULATED}"
)

print(
    f"Optimizer created            : "
    f"{NOTEBOOK_10_BLOCK_3_OPTIMIZER_CREATED}"
)

print(
    f"Backward pass executed       : "
    f"{NOTEBOOK_10_BLOCK_3_BACKWARD_PASS_EXECUTED}"
)

print(
    f"Parameter update executed    : "
    f"{NOTEBOOK_10_BLOCK_3_PARAMETER_UPDATE_EXECUTED}"
)

print(
    f"Training executed            : "
    f"{NOTEBOOK_10_BLOCK_3_TRAINING_EXECUTED}"
)

print("-" * 72)

print(
    f"Block valid                  : "
    f"{NOTEBOOK_10_BLOCK_3_VALID}"
)

print(
    f"Block complete               : "
    f"{NOTEBOOK_10_BLOCK_3_COMPLETE}"
)

print("=" * 72)


# =============================================================================
# Completion interpretation
# =============================================================================

if BLOCK_3_MULTI_ARTICLE_VALIDATION_SOURCE_READY:

    print(
        "A genuine multi-article validation source was constructed "
        "successfully without synthetic fallback."
    )

    print(
        f"The expanded source contains "
        f"{BLOCK_3_TOTAL_ARTICLE_COUNT} articles and "
        f"{BLOCK_3_TOTAL_SENTENCE_COUNT} canonical sentences."
    )

    print(
        "At least two independent articles possess valid contextual "
        "representations under the established 768-dimensional contract."
    )

    print(
        "Notebook 10 may proceed to formal multi-article corpus "
        "validation before controlled recurrent execution."
    )


elif BLOCK_3_STRUCTURAL_MULTI_ARTICLE_SUFFICIENCY:

    print(
        "A genuine structurally valid multi-article corpus was recovered."
    )

    print(
        f"The expanded source contains "
        f"{BLOCK_3_TOTAL_ARTICLE_COUNT} articles and "
        f"{BLOCK_3_TOTAL_SENTENCE_COUNT} canonical sentences."
    )

    print(
        "However, fewer than two articles currently possess reusable "
        "validated contextual representations."
    )


    if (
        BLOCK_3_REPRESENTATION_GENERATION_REQUIRED_COUNT
        >
        0
    ):

        print(
            f"{BLOCK_3_REPRESENTATION_GENERATION_REQUIRED_COUNT} "
            "additional article(s) contain complete sentence text and "
            "require contextual representation generation under the "
            "established all-mpnet-base-v2 encoder contract."
        )


    if (
        BLOCK_3_REPRESENTATION_GENERATION_BLOCKED_COUNT
        >
        0
    ):

        print(
            f"{BLOCK_3_REPRESENTATION_GENERATION_BLOCKED_COUNT} "
            "additional article(s) cannot yet be made representation-ready "
            "because complete canonical sentence text or a validated "
            "compatible representation is unavailable."
        )


    print(
        "Notebook 10 should next resolve contextual representation "
        "availability before recurrent validation."
    )


else:

    print(
        "No sufficient genuine additional article corpus was recovered "
        "from the currently accessible validated project sources."
    )

    print(
        f"The validated pilot corpus remains preserved and valid with "
        f"{BLOCK_3_PILOT_ARTICLE_COUNT} article(s) and "
        f"{BLOCK_3_PILOT_SENTENCE_COUNT} sentence(s)."
    )

    print(
        "No pilot splitting, duplication, synthetic text generation "
        "or supervision substitution was used."
    )

    print(
        "Additional independently identifiable article material must "
        "therefore be supplied or recovered before multi-article "
        "validation can proceed."
    )


print("=" * 72)
```

# Block 4 — External Additional-Article Ingestion and Canonical Corpus Extension

## Purpose

This block extends the validated Notebook 10 corpus with one or more genuinely independent articles supplied from outside the currently recovered project artefacts.

Block 3 established that the accessible project sources contain only the original 16-sentence pilot article and no additional independent article material suitable for multi-article validation.

The present block therefore introduces the controlled ingestion pathway required to construct a genuine expanded corpus.

Its purpose is to:

- accept additional article text from an explicitly identified external source;
- preserve article-level provenance;
- construct stable article, paragraph and sentence identities;
- establish deterministic canonical sentence order;
- prepare the new article records for contextual representation generation;
- merge the resulting records with the validated pilot corpus without altering the existing article;
- determine whether the expanded corpus is structurally ready for multi-article validation.

No Media AI recurrent execution or model training is performed in this block.

## Methodological Role

Block 4 represents a deliberate transition from **corpus discovery** to **controlled corpus extension**.

Blocks 2 and 3 established that the current validated project runtime is structurally sound but insufficient for multi-article validation because it contains only one independent article.

The current corpus therefore satisfies

$$
N_{\mathrm{articles}} = 1.
$$

The present block seeks to establish

$$
N_{\mathrm{articles}} \ge 2
$$

using genuinely independent textual material.

The additional article must not be created by splitting, duplicating or perturbing the pilot article.

## External-Source Requirement

Every newly ingested article must originate from a real and independently identifiable source.

The block may accept:

- manually supplied article text;
- an uploaded article file;
- a validated external project corpus record;
- another explicitly identified source whose provenance can be preserved.

The source must remain auditable.

At minimum, the ingestion record should preserve where available:

- article source;
- source identifier;
- title;
- publication;
- publication date;
- author;
- section or topic;
- language;
- original URL or reference;
- ingestion timestamp.

Not every provenance field is mandatory, but the article must remain distinguishable from internally generated or synthetic content.

## No Synthetic Expansion

The block must not satisfy the multi-article requirement through artificial corpus construction.

The following are prohibited:

- splitting the original pilot article into multiple pseudo-articles;
- duplicating the pilot article under a new identifier;
- paraphrasing the pilot article and treating the result as independent text;
- generating a new article with an LLM;
- combining unrelated pilot sentences into artificial article structures;
- using supervision labels as text;
- creating placeholder sentences.

The expanded corpus must represent genuine additional language.

## Canonical Article Identity

Each new article must receive or preserve a stable article identifier.

If the source already exposes an article identifier, that identifier should be retained when suitable.

If no stable identifier exists, the block may construct a deterministic project identifier from stable source information.

The resulting article identifier must:

- be non-null;
- be unique within the expanded corpus;
- not collide with the pilot article;
- remain stable across subsequent notebook execution;
- remain associated with all sentences derived from that article.

## Paragraph Structure

Where the article source preserves paragraph boundaries, those boundaries should be retained.

For article $i$ containing paragraphs

$$
P_{i,1},
P_{i,2},
\ldots,
P_{i,M_i},
$$

each paragraph should receive a stable paragraph identity.

Paragraph identity should support the hierarchy

$$
\mathrm{Article}
\rightarrow
\mathrm{Paragraph}
\rightarrow
\mathrm{Sentence}.
$$

If the supplied source does not preserve paragraph structure, the block may represent the article as a single paragraph, but this must be recorded explicitly rather than inferred silently.

## Sentence Segmentation

The additional article must be segmented into deterministic sentence units.

Sentence segmentation should preserve:

- original sentence wording;
- punctuation;
- textual order;
- paragraph membership;
- source traceability.

The segmentation procedure must not paraphrase or semantically rewrite the article.

Sentence text should be retained exactly apart from previously established structural cleaning rules.

## Sentence Identity

Every ingested sentence must receive a globally unique sentence identifier.

For article $i$ and sentence position $t$,

$$
s_{i,t}
$$

must have an identity satisfying

$$
\mathrm{id}(s_{i,t})
\neq
\mathrm{id}(s_{j,u})
$$

for every distinct sentence record.

Sentence identifiers must not collide with the existing pilot sentence identifiers.

## Canonical Sequence Position

Every sentence must receive a deterministic within-article sequence position.

The canonical internal convention for Notebook 10 should be zero-based:

$$
t \in
\{0,1,\ldots,T_i-1\}.
$$

Accordingly, an article containing $T_i$ sentences is represented as

$$
\mathcal{A}_i
=
\left(
s_{i,0},
s_{i,1},
\ldots,
s_{i,T_i-1}
\right).
$$

The original source order must be preserved exactly.

## Textual Integrity Validation

Before an article can become structurally eligible, the block should confirm:

- non-empty article text;
- at least one non-empty sentence;
- no empty sentence records;
- deterministic sentence ordering;
- unique sentence identities;
- explicit article identity;
- no identity collisions with the existing corpus.

The block should also retain the original article text or a stable source reference where practical so the segmentation remains auditable.

## Contextual Representation Preparation

The inherited Media AI representation architecture expects sentence-level contextual embeddings with dimension

$$
768.
$$

The established encoder contract remains:

**Provider:** `sentence-transformers`

**Model:** `sentence-transformers/all-mpnet-base-v2`

**Embedding dimension:** $768$

**Normalisation:** enabled

For each sentence $s_{i,t}$, the required contextual input is

$$
\mathbf{x}_{i,t}
\in
\mathbb{R}^{768}.
$$

The block may prepare these embeddings directly under the exact established encoder contract or may prepare the canonical sentence records for representation generation in the following block.

Any generated representations must remain aligned by sentence identity.

## Representation Generation Policy

If contextual representations are generated in this block, the process must satisfy:

- exact use of `sentence-transformers/all-mpnet-base-v2`;
- output dimension exactly 768;
- finite numeric values;
- normalisation consistent with the established project contract;
- deterministic sentence-order alignment;
- no Media AI model execution.

The sentence-transformer encoder is an external contextualisation stage and is not the inherited Media AI representation architecture.

Generating sentence embeddings therefore does not constitute a Media AI representation forward pass.

## Article-Level Representation Matrix

For each new article $i$, the contextual representation matrix is

$$
\mathbf{X}_i
=
\begin{bmatrix}
\mathbf{x}_{i,0}\\
\mathbf{x}_{i,1}\\
\vdots\\
\mathbf{x}_{i,T_i-1}
\end{bmatrix}
\in
\mathbb{R}^{T_i \times 768}.
$$

The row order must correspond exactly to canonical sentence order.

## Corpus Merge

The validated pilot article must remain unchanged.

Let the pilot corpus be

$$
\mathcal{C}_{\mathrm{pilot}}
=
\{\mathcal{A}_1\},
$$

and the newly ingested article set be

$$
\mathcal{C}_{\mathrm{new}}
=
\{
\mathcal{A}_2,
\ldots,
\mathcal{A}_N
\}.
$$

The expanded corpus becomes

$$
\mathcal{C}_{\mathrm{expanded}}
=
\mathcal{C}_{\mathrm{pilot}}
\cup
\mathcal{C}_{\mathrm{new}}.
$$

The merge must preserve:

- all original pilot records exactly;
- all new article identities;
- all new sentence identities;
- canonical article boundaries;
- canonical sentence ordering;
- contextual representation alignment;
- source provenance.

## Article Independence

Each ingested article must represent an independent future recurrent sequence.

The existence of a deterministic storage order between articles must not imply recurrent dependence.

Later execution must satisfy

$$
\mathbf{h}_{i+1,0}^{(p)}
=
\mathbf{0}
$$

regardless of

$$
\mathbf{h}_{i,T_i}^{(p)}.
$$

This block preserves article boundaries but does not create recurrent states.

## Supervision Status

Supervision is not required for external article ingestion.

Each newly added article should receive one of the following supervision states:

- `available`;
- `partially_available`;
- `unavailable`;
- `alignment_not_yet_assessed`.

For newly supplied raw article text, the default should be

**`alignment_not_yet_assessed`**

unless validated supervision already exists.

The absence of supervision must not prevent structural or representation readiness.

## Structural Eligibility

A newly ingested article is structurally eligible only if:

- article identity is valid;
- article text is non-empty;
- at least one sentence exists;
- sentence identities are valid;
- sentence identities are globally unique;
- sequence positions are valid;
- sequence positions are contiguous;
- canonical ordering is deterministic;
- article identity does not collide with the pilot corpus;
- sentence identities do not collide with the pilot corpus;
- provenance is retained;
- no synthetic fallback has been used.

## Representation Readiness

A structurally eligible new article becomes representation ready only when every sentence has:

- exactly one contextual representation;
- embedding dimension 768;
- finite values;
- verified sentence alignment;
- verified article alignment.

The block should report representation readiness separately from structural eligibility.

## Multi-Article Validation Readiness

After corpus extension, Notebook 10 is ready for multi-article validation only if:

$$
N_{\mathrm{representation-ready\ articles}}
\ge 2.
$$

In addition:

- article IDs must be unique;
- sentence IDs must be globally unique;
- canonical ordering must be valid for every article;
- article boundaries must be explicit;
- contextual representations must satisfy the established encoder contract;
- the pilot corpus must remain exactly unchanged;
- no synthetic expansion may have occurred.

## Expanded-Training Boundary

Reaching two representation-ready articles is sufficient for architectural multi-article validation.

It is **not** sufficient to establish expanded-training adequacy.

Therefore:

$$
\mathrm{multi\text{-}article\ validation\ ready}
\not\Rightarrow
\mathrm{expanded\ training\ sufficient}.
$$

The latter remains a separate requirement for the subsequent training project.

## Model-State Protection

The complete inherited Media AI architecture must remain unchanged throughout external article ingestion.

The block must preserve exactly:

- **894,003 representation parameters**;
- **8,187 transformative parameters**;
- **25 Transformative-Weight parameters**;
- **902,215 total inherited parameters**.

All inherited modules remain frozen and in evaluation mode.

## Execution Boundary

This block may:

- accept genuine additional article text;
- preserve source provenance;
- construct deterministic article identity;
- preserve or construct paragraph identities;
- perform deterministic sentence segmentation;
- construct globally unique sentence identities;
- establish canonical sentence positions;
- prepare contextual encoder inputs;
- generate 768-dimensional `all-mpnet-base-v2` contextual embeddings;
- validate embedding shape and finiteness;
- merge new article records with the validated pilot corpus;
- classify structural and representation readiness.

This block must not:

- execute the Media AI representation architecture;
- execute transformative mechanisms;
- execute Transformative-Weight forward paths;
- create recurrent states;
- propagate recurrent state;
- calculate recurrent or training objectives;
- create an optimiser;
- execute a backward pass;
- update any Media AI parameter;
- perform model training.

## Expected Outcome

At completion, the block should report:

- supplied additional articles;
- accepted additional articles;
- rejected additional articles;
- rejection reasons;
- pilot article count;
- pilot sentence count;
- additional article count;
- additional sentence count;
- total article count;
- total sentence count;
- paragraph structure status;
- sentence identity validity;
- sentence identity collision status;
- canonical sequence validity;
- contextual representation generation status;
- representation-ready additional articles;
- representation-ready total articles;
- corpus provenance status;
- supervision status;
- synthetic fallback status;
- pilot preservation status;
- multi-article validation readiness;
- expanded-training sufficiency status.

Successful completion establishes a genuine expanded corpus suitable for the next stage:

**formal multi-article corpus validation and controlled recurrent execution.**

```python
# =============================================================================
# Media AI — Notebook 10
# Block 4: External Additional-Article Ingestion
#          and Canonical Corpus Extension
# =============================================================================

from copy import deepcopy
from datetime import datetime, timezone

import hashlib
import re
import uuid

import numpy as np
import torch


# =============================================================================
# Block identity
# =============================================================================

NOTEBOOK_10_BLOCK_4 = 4

NOTEBOOK_10_BLOCK_4_NAME = (
    "External Additional-Article Ingestion "
    "and Canonical Corpus Extension"
)

NOTEBOOK_10_BLOCK_4_VERSION = "1.1"

BLOCK_4_EXECUTED_AT_UTC = (
    datetime.now(
        timezone.utc
    ).isoformat()
)


# =============================================================================
# Prerequisite validation
# =============================================================================

BLOCK_4_PREREQUISITES_VALID = all(
    [
        globals().get(
            "NOTEBOOK_10_BLOCK_1_COMPLETE",
            False,
        )
        is True,

        globals().get(
            "NOTEBOOK_10_BLOCK_1_VALID",
            False,
        )
        is True,

        globals().get(
            "NOTEBOOK_10_BLOCK_2_COMPLETE",
            False,
        )
        is True,

        globals().get(
            "NOTEBOOK_10_BLOCK_2_VALID",
            False,
        )
        is True,

        globals().get(
            "NOTEBOOK_10_BLOCK_3_COMPLETE",
            False,
        )
        is True,

        globals().get(
            "NOTEBOOK_10_BLOCK_3_VALID",
            False,
        )
        is True,

        "NOTEBOOK_10_EXPANDED_ARTICLE_RECORDS"
        in globals(),

        "NOTEBOOK_10_REPRESENTATION_MODULES"
        in globals(),

        "NOTEBOOK_10_TRANSFORMATIVE_MODULES"
        in globals(),

        "NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES"
        in globals(),
    ]
)


if not BLOCK_4_PREREQUISITES_VALID:

    raise RuntimeError(
        "Notebook 10 Block 4 requires valid completed Blocks 1–3."
    )


# =============================================================================
# Immutable inherited Media AI state baseline
# =============================================================================
#
# Notebook 10 Block 1 reconstructs the Transformative-Weight modules under the
# canonical Notebook 10 parameter key (`alpha`), while the persisted Notebook 09
# checkpoint may use a different implementation key such as `alpha_active`.
#
# Block 4 must therefore validate immutability against the exact Notebook 10
# runtime state inherited at the start of this block, not by directly comparing
# reconstructed Notebook 10 state-dict keys with raw Notebook 09 checkpoint
# implementation keys.
# =============================================================================

def block_4_snapshot_module_collection(
    module_collection,
):

    return {
        module_name:
            {
                tensor_name:
                    tensor.detach()
                    .cpu()
                    .clone()

                for (
                    tensor_name,
                    tensor,
                ) in module.state_dict().items()
            }

        for (
            module_name,
            module,
        ) in module_collection.items()
    }


def block_4_nested_state_exact(
    state_before,
    state_after,
):

    if state_before.keys() != state_after.keys():

        return False


    for module_name in state_before:

        if (
            state_before[
                module_name
            ].keys()
            !=
            state_after[
                module_name
            ].keys()
        ):

            return False


        for tensor_name in state_before[
            module_name
        ]:

            if not torch.equal(
                state_before[
                    module_name
                ][
                    tensor_name
                ],
                state_after[
                    module_name
                ][
                    tensor_name
                ],
            ):

                return False


    return True


BLOCK_4_REPRESENTATION_STATE_BASELINE = (
    block_4_snapshot_module_collection(
        NOTEBOOK_10_REPRESENTATION_MODULES
    )
)


BLOCK_4_TRANSFORMATIVE_STATE_BASELINE = (
    block_4_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_MODULES
    )
)


BLOCK_4_WEIGHT_STATE_BASELINE = (
    block_4_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES
    )
)


# =============================================================================
# Canonical contextual-encoder contract
# =============================================================================

BLOCK_4_EXPECTED_ENCODER_PROVIDER = (
    "sentence-transformers"
)

BLOCK_4_EXPECTED_ENCODER_MODEL = (
    "sentence-transformers/all-mpnet-base-v2"
)

BLOCK_4_EXPECTED_EMBEDDING_DIMENSION = 768

BLOCK_4_EXPECTED_NORMALISED = True

BLOCK_4_MINIMUM_MULTI_ARTICLE_COUNT = 2


# =============================================================================
# External article input contract
# =============================================================================
#
# Supply one or more genuine additional articles through:
#
# NOTEBOOK_10_EXTERNAL_ARTICLE_INPUTS = [
#     {
#         "text": "...full article text...",
#         "title": "...",
#         "source": "...",
#         "source_id": "...",          # optional
#         "publication": "...",        # optional
#         "publication_date": "...",   # optional
#         "author": "...",             # optional
#         "section": "...",            # optional
#         "language": "en",            # optional
#         "url": "...",                # optional
#     }
# ]
#
# The article text must be genuine external material.
#
# Block 4 never generates substitute article text.
# =============================================================================

if (
    "NOTEBOOK_10_EXTERNAL_ARTICLE_INPUTS"
    not in globals()
):

    NOTEBOOK_10_EXTERNAL_ARTICLE_INPUTS = []


if not isinstance(
    NOTEBOOK_10_EXTERNAL_ARTICLE_INPUTS,
    list,
):

    raise TypeError(
        "NOTEBOOK_10_EXTERNAL_ARTICLE_INPUTS must be a list."
    )


# =============================================================================
# Contextual-representation generation policy
# =============================================================================
#
# If True, accepted new articles are encoded immediately with the exact
# established contextual encoder.
#
# If the encoder cannot be loaded, structural ingestion remains valid and
# the affected articles remain representation-generation required.
# =============================================================================

BLOCK_4_GENERATE_CONTEXTUAL_REPRESENTATIONS = True


# =============================================================================
# Execution boundary
# =============================================================================

NOTEBOOK_10_BLOCK_4_EXTERNAL_TEXT_INGESTED = False

NOTEBOOK_10_BLOCK_4_CONTEXTUAL_ENCODER_EXECUTED = False

NOTEBOOK_10_BLOCK_4_MEDIA_AI_REPRESENTATION_FORWARD_EXECUTED = False

NOTEBOOK_10_BLOCK_4_TRANSFORMATIVE_FORWARD_EXECUTED = False

NOTEBOOK_10_BLOCK_4_WEIGHT_FORWARD_EXECUTED = False

NOTEBOOK_10_BLOCK_4_RECURRENT_STATE_CREATED = False

NOTEBOOK_10_BLOCK_4_RECURRENT_UPDATE_EXECUTED = False

NOTEBOOK_10_BLOCK_4_LOSS_CALCULATED = False

NOTEBOOK_10_BLOCK_4_OPTIMIZER_CREATED = False

NOTEBOOK_10_BLOCK_4_BACKWARD_PASS_EXECUTED = False

NOTEBOOK_10_BLOCK_4_PARAMETER_UPDATE_EXECUTED = False

NOTEBOOK_10_BLOCK_4_TRAINING_EXECUTED = False

NOTEBOOK_10_BLOCK_4_SYNTHETIC_TEXT_CREATED = False

NOTEBOOK_10_BLOCK_4_PILOT_ARTICLE_SPLIT = False

NOTEBOOK_10_BLOCK_4_PILOT_ARTICLE_DUPLICATED = False

NOTEBOOK_10_BLOCK_4_SUPERVISION_USED_AS_TEXT = False

NOTEBOOK_10_BLOCK_4_SUPERVISION_USED_AS_REPRESENTATION = False


# =============================================================================
# Exact nested-object comparison
# =============================================================================

def block_4_objects_exact(
    object_a,
    object_b,
):

    if (
        isinstance(
            object_a,
            np.ndarray,
        )
        or
        isinstance(
            object_b,
            np.ndarray,
        )
    ):

        if not (
            isinstance(
                object_a,
                np.ndarray,
            )
            and
            isinstance(
                object_b,
                np.ndarray,
            )
        ):

            return False


        return all(
            [
                object_a.dtype
                ==
                object_b.dtype,

                object_a.shape
                ==
                object_b.shape,

                bool(
                    np.array_equal(
                        object_a,
                        object_b,
                        equal_nan=
                            True,
                    )
                ),
            ]
        )


    if (
        torch.is_tensor(
            object_a
        )
        or
        torch.is_tensor(
            object_b
        )
    ):

        if not (
            torch.is_tensor(
                object_a
            )
            and
            torch.is_tensor(
                object_b
            )
        ):

            return False


        return all(
            [
                object_a.dtype
                ==
                object_b.dtype,

                tuple(
                    object_a.shape
                )
                ==
                tuple(
                    object_b.shape
                ),

                bool(
                    torch.equal(
                        object_a.detach().cpu(),
                        object_b.detach().cpu(),
                    )
                ),
            ]
        )


    if (
        isinstance(
            object_a,
            dict,
        )
        or
        isinstance(
            object_b,
            dict,
        )
    ):

        if not (
            isinstance(
                object_a,
                dict,
            )
            and
            isinstance(
                object_b,
                dict,
            )
        ):

            return False


        if (
            set(
                object_a.keys()
            )
            !=
            set(
                object_b.keys()
            )
        ):

            return False


        return all(
            block_4_objects_exact(
                object_a[
                    key
                ],
                object_b[
                    key
                ],
            )

            for key
            in object_a
        )


    if (
        isinstance(
            object_a,
            (
                list,
                tuple,
            ),
        )
        or
        isinstance(
            object_b,
            (
                list,
                tuple,
            ),
        )
    ):

        if not (
            isinstance(
                object_a,
                (
                    list,
                    tuple,
                ),
            )
            and
            isinstance(
                object_b,
                (
                    list,
                    tuple,
                ),
            )
        ):

            return False


        if (
            len(
                object_a
            )
            !=
            len(
                object_b
            )
        ):

            return False


        return all(
            block_4_objects_exact(
                value_a,
                value_b,
            )

            for (
                value_a,
                value_b,
            ) in zip(
                object_a,
                object_b,
            )
        )


    if isinstance(
        object_a,
        np.generic,
    ):

        object_a = object_a.item()


    if isinstance(
        object_b,
        np.generic,
    ):

        object_b = object_b.item()


    try:

        comparison = (
            object_a
            ==
            object_b
        )


        if isinstance(
            comparison,
            (
                bool,
                np.bool_,
            ),
        ):

            return bool(
                comparison
            )


        return False


    except Exception:

        return False


# =============================================================================
# Text and identity helpers
# =============================================================================

def block_4_clean_text(
    value,
):

    if not isinstance(
        value,
        str,
    ):

        return None


    value = (
        value.replace(
            "\r\n",
            "\n",
        )
        .replace(
            "\r",
            "\n",
        )
        .strip()
    )


    if not value:

        return None


    return value


def block_4_optional_text(
    value,
):

    if value is None:

        return None


    if not isinstance(
        value,
        str,
    ):

        value = str(
            value
        )


    value = value.strip()


    return (
        value
        if value
        else None
    )


def block_4_sha256_text(
    value,
):

    return hashlib.sha256(
        value.encode(
            "utf-8"
        )
    ).hexdigest()


def block_4_stable_article_id(
    article,
    article_text,
):

    explicit_source_id = (
        block_4_optional_text(
            article.get(
                "source_id"
            )
        )
    )


    source = (
        block_4_optional_text(
            article.get(
                "source"
            )
        )
        or
        "external_source"
    )


    title = (
        block_4_optional_text(
            article.get(
                "title"
            )
        )
        or
        "untitled"
    )


    url = (
        block_4_optional_text(
            article.get(
                "url"
            )
        )
        or
        ""
    )


    if explicit_source_id is not None:

        stable_key = (
            f"{source}|"
            f"{explicit_source_id}"
        )

    else:

        text_digest = block_4_sha256_text(
            article_text
        )


        stable_key = (
            f"{source}|"
            f"{title}|"
            f"{url}|"
            f"{text_digest}"
        )


    return str(
        uuid.uuid5(
            uuid.NAMESPACE_URL,
            stable_key,
        )
    )


# =============================================================================
# Deterministic paragraph segmentation
# =============================================================================

def block_4_segment_paragraphs(
    article_text,
):

    raw_paragraphs = re.split(
        r"\n\s*\n+",
        article_text,
    )


    paragraphs = []


    for raw_paragraph in raw_paragraphs:

        paragraph = re.sub(
            r"[ \t]+",
            " ",
            raw_paragraph,
        ).strip()


        if paragraph:

            paragraphs.append(
                paragraph
            )


    if not paragraphs:

        paragraphs = [
            article_text.strip()
        ]


    return paragraphs


# =============================================================================
# Deterministic sentence segmentation
# =============================================================================
#
# This is deliberately conservative and deterministic.
#
# It splits primarily at punctuation followed by whitespace and an apparent
# new sentence boundary.
#
# It does not rewrite sentence text.
# =============================================================================

BLOCK_4_COMMON_ABBREVIATIONS = {
    "mr.",
    "mrs.",
    "ms.",
    "dr.",
    "prof.",
    "sr.",
    "jr.",
    "st.",
    "vs.",
    "etc.",
    "e.g.",
    "i.e.",
    "u.s.",
    "u.k.",
}


def block_4_segment_sentences(
    paragraph_text,
):

    candidate_parts = re.split(
        r"(?<=[.!?])\s+",
        paragraph_text.strip(),
    )


    sentences = []

    buffer = ""


    for candidate in candidate_parts:

        candidate = candidate.strip()


        if not candidate:

            continue


        if buffer:

            candidate = (
                buffer
                +
                " "
                +
                candidate
            )

            buffer = ""


        candidate_lower = candidate.lower()


        last_token = (
            candidate_lower.split()[
                -1
            ]
            if candidate_lower.split()
            else ""
        )


        if (
            last_token
            in
            BLOCK_4_COMMON_ABBREVIATIONS
        ):

            buffer = candidate

            continue


        sentences.append(
            candidate
        )


    if buffer:

        sentences.append(
            buffer
        )


    return [
        sentence

        for sentence
        in sentences

        if sentence.strip()
    ]


# =============================================================================
# Preserve validated corpus from Block 3
# =============================================================================

BLOCK_4_EXISTING_ARTICLE_RECORDS = deepcopy(
    NOTEBOOK_10_EXPANDED_ARTICLE_RECORDS
)


BLOCK_4_EXISTING_ARTICLE_RECORDS_BEFORE = deepcopy(
    BLOCK_4_EXISTING_ARTICLE_RECORDS
)


BLOCK_4_EXISTING_ARTICLE_COUNT = len(
    BLOCK_4_EXISTING_ARTICLE_RECORDS
)


BLOCK_4_EXISTING_SENTENCE_COUNT = sum(
    len(
        records
    )

    for records
    in BLOCK_4_EXISTING_ARTICLE_RECORDS.values()
)


BLOCK_4_EXISTING_ARTICLE_IDS = set(
    BLOCK_4_EXISTING_ARTICLE_RECORDS.keys()
)


BLOCK_4_EXISTING_SENTENCE_IDS = {
    record[
        "sentence_id"
    ]

    for records
    in BLOCK_4_EXISTING_ARTICLE_RECORDS.values()

    for record
    in records
}


# =============================================================================
# External-input audit
# =============================================================================

BLOCK_4_SUPPLIED_ARTICLE_COUNT = len(
    NOTEBOOK_10_EXTERNAL_ARTICLE_INPUTS
)


BLOCK_4_EXTERNAL_INPUT_AVAILABLE = (
    BLOCK_4_SUPPLIED_ARTICLE_COUNT
    >
    0
)


BLOCK_4_ACCEPTED_RAW_ARTICLES = []

BLOCK_4_REJECTED_RAW_ARTICLES = []


for (
    input_index,
    article,
) in enumerate(
    NOTEBOOK_10_EXTERNAL_ARTICLE_INPUTS
):

    rejection_reasons = []


    if not isinstance(
        article,
        dict,
    ):

        BLOCK_4_REJECTED_RAW_ARTICLES.append(
            {
                "input_index":
                    input_index,

                "reasons":
                    [
                        "article_input_not_dictionary"
                    ],
            }
        )

        continue


    article_text = block_4_clean_text(
        article.get(
            "text"
        )
    )


    if article_text is None:

        rejection_reasons.append(
            "missing_or_empty_article_text"
        )


    source = block_4_optional_text(
        article.get(
            "source"
        )
    )


    title = block_4_optional_text(
        article.get(
            "title"
        )
    )


    if source is None:

        rejection_reasons.append(
            "missing_source_provenance"
        )


    if rejection_reasons:

        BLOCK_4_REJECTED_RAW_ARTICLES.append(
            {
                "input_index":
                    input_index,

                "title":
                    title,

                "source":
                    source,

                "reasons":
                    rejection_reasons,
            }
        )

        continue


    article_id = block_4_stable_article_id(
        article,
        article_text,
    )


    if (
        article_id
        in
        BLOCK_4_EXISTING_ARTICLE_IDS
    ):

        BLOCK_4_REJECTED_RAW_ARTICLES.append(
            {
                "input_index":
                    input_index,

                "title":
                    title,

                "source":
                    source,

                "article_id":
                    article_id,

                "reasons":
                    [
                        "article_identity_collision"
                    ],
            }
        )

        continue


    BLOCK_4_ACCEPTED_RAW_ARTICLES.append(
        {
            "input_index":
                input_index,

            "article_id":
                article_id,

            "text":
                article_text,

            "title":
                title,

            "source":
                source,

            "source_id":
                block_4_optional_text(
                    article.get(
                        "source_id"
                    )
                ),

            "publication":
                block_4_optional_text(
                    article.get(
                        "publication"
                    )
                ),

            "publication_date":
                block_4_optional_text(
                    article.get(
                        "publication_date"
                    )
                ),

            "author":
                block_4_optional_text(
                    article.get(
                        "author"
                    )
                ),

            "section":
                block_4_optional_text(
                    article.get(
                        "section"
                    )
                ),

            "language":
                block_4_optional_text(
                    article.get(
                        "language"
                    )
                ),

            "url":
                block_4_optional_text(
                    article.get(
                        "url"
                    )
                ),

            "ingested_at_utc":
                BLOCK_4_EXECUTED_AT_UTC,
        }
    )


NOTEBOOK_10_BLOCK_4_EXTERNAL_TEXT_INGESTED = (
    len(
        BLOCK_4_ACCEPTED_RAW_ARTICLES
    )
    >
    0
)


# =============================================================================
# Detect duplicate newly supplied article identities
# =============================================================================

BLOCK_4_NEW_ARTICLE_IDS = [
    article[
        "article_id"
    ]

    for article
    in BLOCK_4_ACCEPTED_RAW_ARTICLES
]


BLOCK_4_DUPLICATE_NEW_ARTICLE_IDS = {
    article_id

    for article_id
    in BLOCK_4_NEW_ARTICLE_IDS

    if (
        BLOCK_4_NEW_ARTICLE_IDS.count(
            article_id
        )
        >
        1
    )
}


if BLOCK_4_DUPLICATE_NEW_ARTICLE_IDS:

    duplicate_ids = set(
        BLOCK_4_DUPLICATE_NEW_ARTICLE_IDS
    )


    retained = []


    for article in BLOCK_4_ACCEPTED_RAW_ARTICLES:

        if (
            article[
                "article_id"
            ]
            in
            duplicate_ids
        ):

            BLOCK_4_REJECTED_RAW_ARTICLES.append(
                {
                    "input_index":
                        article[
                            "input_index"
                        ],

                    "title":
                        article[
                            "title"
                        ],

                    "source":
                        article[
                            "source"
                        ],

                    "article_id":
                        article[
                            "article_id"
                        ],

                    "reasons":
                        [
                            "duplicate_external_article"
                        ],
                }
            )


        else:

            retained.append(
                article
            )


    BLOCK_4_ACCEPTED_RAW_ARTICLES = retained


# =============================================================================
# Canonical structural construction
# =============================================================================

BLOCK_4_STRUCTURED_NEW_ARTICLES = {}

BLOCK_4_STRUCTURAL_REJECTIONS = []


for article in BLOCK_4_ACCEPTED_RAW_ARTICLES:

    article_id = article[
        "article_id"
    ]


    paragraphs = block_4_segment_paragraphs(
        article[
            "text"
        ]
    )


    sentence_records = []

    global_sentence_position = 0


    for (
        paragraph_index,
        paragraph_text,
    ) in enumerate(
        paragraphs
    ):

        paragraph_id = str(
            uuid.uuid5(
                uuid.UUID(
                    article_id
                ),
                f"paragraph:{paragraph_index}",
            )
        )


        sentences = block_4_segment_sentences(
            paragraph_text
        )


        for (
            paragraph_sentence_index,
            sentence_text,
        ) in enumerate(
            sentences
        ):

            sentence_text = (
                sentence_text.strip()
            )


            if not sentence_text:

                continue


            sentence_id = str(
                uuid.uuid5(
                    uuid.UUID(
                        article_id
                    ),
                    (
                        f"paragraph:{paragraph_index}|"
                        f"sentence:{paragraph_sentence_index}|"
                        f"text:{sentence_text}"
                    ),
                )
            )


            sentence_records.append(
                {
                    "article_id":
                        article_id,

                    "paragraph_id":
                        paragraph_id,

                    "sentence_id":
                        sentence_id,

                    "sequence_position":
                        int(
                            global_sentence_position
                        ),

                    "canonical_sequence_position":
                        int(
                            global_sentence_position
                        ),

                    "paragraph_position":
                        int(
                            paragraph_index
                        ),

                    "paragraph_sentence_position":
                        int(
                            paragraph_sentence_index
                        ),

                    "sentence_text":
                        sentence_text,

                    "contextual_representation":
                        None,

                    "representation_provider":
                        None,

                    "representation_model":
                        None,

                    "representation_dimension":
                        None,

                    "representation_normalised":
                        None,

                    "representation_ready":
                        False,

                    "provenance":
                        {
                            "source":
                                article[
                                    "source"
                                ],

                            "source_id":
                                article[
                                    "source_id"
                                ],

                            "title":
                                article[
                                    "title"
                                ],

                            "publication":
                                article[
                                    "publication"
                                ],

                            "publication_date":
                                article[
                                    "publication_date"
                                ],

                            "author":
                                article[
                                    "author"
                                ],

                            "section":
                                article[
                                    "section"
                                ],

                            "language":
                                article[
                                    "language"
                                ],

                            "url":
                                article[
                                    "url"
                                ],

                            "ingested_at_utc":
                                article[
                                    "ingested_at_utc"
                                ],
                        },

                    "supervision_status":
                        "alignment_not_yet_assessed",

                    "source_type":
                        "external_article",
                }
            )


            global_sentence_position += 1


    rejection_reasons = []


    if not paragraphs:

        rejection_reasons.append(
            "no_paragraphs"
        )


    if not sentence_records:

        rejection_reasons.append(
            "no_sentences"
        )


    sentence_ids = [
        record[
            "sentence_id"
        ]

        for record
        in sentence_records
    ]


    if (
        len(
            sentence_ids
        )
        !=
        len(
            set(
                sentence_ids
            )
        )
    ):

        rejection_reasons.append(
            "duplicate_sentence_identity"
        )


    sequence_positions = [
        record[
            "sequence_position"
        ]

        for record
        in sentence_records
    ]


    if (
        sequence_positions
        !=
        list(
            range(
                len(
                    sentence_records
                )
            )
        )
    ):

        rejection_reasons.append(
            "invalid_canonical_sequence"
        )


    if not set(
        sentence_ids
    ).isdisjoint(
        BLOCK_4_EXISTING_SENTENCE_IDS
    ):

        rejection_reasons.append(
            "pilot_sentence_identity_collision"
        )


    if rejection_reasons:

        BLOCK_4_STRUCTURAL_REJECTIONS.append(
            {
                "article_id":
                    article_id,

                "title":
                    article[
                        "title"
                    ],

                "source":
                    article[
                        "source"
                    ],

                "reasons":
                    rejection_reasons,
            }
        )

        continue


    BLOCK_4_STRUCTURED_NEW_ARTICLES[
        article_id
    ] = {
        "article_metadata":
            deepcopy(
                article
            ),

        "paragraph_count":
            len(
                paragraphs
            ),

        "sentence_count":
            len(
                sentence_records
            ),

        "records":
            sentence_records,
    }


# =============================================================================
# Contextual encoder preparation
# =============================================================================

BLOCK_4_CONTEXTUAL_ENCODER_AVAILABLE = False

BLOCK_4_CONTEXTUAL_ENCODER_ERROR = None

BLOCK_4_CONTEXTUAL_ENCODER = None


if (
    BLOCK_4_GENERATE_CONTEXTUAL_REPRESENTATIONS
    and
    BLOCK_4_STRUCTURED_NEW_ARTICLES
):

    try:

        from sentence_transformers import SentenceTransformer


        BLOCK_4_CONTEXTUAL_ENCODER = (
            SentenceTransformer(
                BLOCK_4_EXPECTED_ENCODER_MODEL,
                device=
                    "cpu",
            )
        )


        BLOCK_4_CONTEXTUAL_ENCODER_AVAILABLE = True


    except Exception as exc:

        BLOCK_4_CONTEXTUAL_ENCODER_AVAILABLE = False


        BLOCK_4_CONTEXTUAL_ENCODER_ERROR = {
            "error_type":
                type(
                    exc
                ).__name__,

            "error":
                str(
                    exc
                ),
        }


# =============================================================================
# Contextual representation generation
# =============================================================================
#
# This invokes the external sentence-transformer encoder only.
#
# It does not execute any Media AI representation, transformative,
# Transformative-Weight or recurrent module.
# =============================================================================

BLOCK_4_REPRESENTATION_GENERATION_AUDIT = {}


if BLOCK_4_CONTEXTUAL_ENCODER_AVAILABLE:

    for (
        article_id,
        article_payload,
    ) in BLOCK_4_STRUCTURED_NEW_ARTICLES.items():

        records = article_payload[
            "records"
        ]


        sentence_texts = [
            record[
                "sentence_text"
            ]

            for record
            in records
        ]


        with torch.no_grad():

            embeddings = (
                BLOCK_4_CONTEXTUAL_ENCODER.encode(
                    sentence_texts,
                    convert_to_numpy=
                        True,
                    normalize_embeddings=
                        True,
                    show_progress_bar=
                        False,
                )
            )


        NOTEBOOK_10_BLOCK_4_CONTEXTUAL_ENCODER_EXECUTED = True


        embeddings = np.asarray(
            embeddings,
            dtype=
                np.float32,
        )


        expected_shape = (
            len(
                records
            ),
            BLOCK_4_EXPECTED_EMBEDDING_DIMENSION,
        )


        shape_valid = (
            embeddings.shape
            ==
            expected_shape
        )


        finite = bool(
            np.isfinite(
                embeddings
            ).all()
        )


        norms = (
            np.linalg.norm(
                embeddings,
                axis=
                    1,
            )
            if shape_valid
            else
            np.asarray(
                [],
                dtype=
                    np.float32,
            )
        )


        normalisation_valid = (
            shape_valid
            and
            bool(
                np.allclose(
                    norms,
                    np.ones_like(
                        norms
                    ),
                    atol=
                        1e-4,
                    rtol=
                        1e-4,
                )
            )
        )


        representation_valid = all(
            [
                shape_valid,
                finite,
                normalisation_valid,
            ]
        )


        if representation_valid:

            for (
                record,
                embedding,
            ) in zip(
                records,
                embeddings,
            ):

                record[
                    "contextual_representation"
                ] = embedding.copy()


                record[
                    "representation_provider"
                ] = (
                    BLOCK_4_EXPECTED_ENCODER_PROVIDER
                )


                record[
                    "representation_model"
                ] = (
                    BLOCK_4_EXPECTED_ENCODER_MODEL
                )


                record[
                    "representation_dimension"
                ] = (
                    BLOCK_4_EXPECTED_EMBEDDING_DIMENSION
                )


                record[
                    "representation_normalised"
                ] = True


                record[
                    "representation_ready"
                ] = True


        BLOCK_4_REPRESENTATION_GENERATION_AUDIT[
            article_id
        ] = {
            "sentence_count":
                len(
                    records
                ),

            "matrix_shape":
                tuple(
                    embeddings.shape
                ),

            "shape_valid":
                shape_valid,

            "finite":
                finite,

            "normalisation_valid":
                normalisation_valid,

            "representation_valid":
                representation_valid,
        }


# =============================================================================
# Article structural / representation readiness
# =============================================================================

BLOCK_4_NEW_ARTICLE_AUDIT = {}

BLOCK_4_ACCEPTED_ADDITIONAL_ARTICLES = {}

BLOCK_4_REPRESENTATION_READY_ADDITIONAL_ARTICLES = {}

BLOCK_4_REPRESENTATION_PENDING_ADDITIONAL_ARTICLES = {}


for (
    article_id,
    article_payload,
) in BLOCK_4_STRUCTURED_NEW_ARTICLES.items():

    records = article_payload[
        "records"
    ]


    sentence_ids = [
        record[
            "sentence_id"
        ]

        for record
        in records
    ]


    sequence_positions = [
        record[
            "sequence_position"
        ]

        for record
        in records
    ]


    sentence_identity_valid = all(
        sentence_id is not None

        for sentence_id
        in sentence_ids
    )


    sentence_identity_unique = (
        len(
            sentence_ids
        )
        ==
        len(
            set(
                sentence_ids
            )
        )
    )


    canonical_sequence_valid = (
        sequence_positions
        ==
        list(
            range(
                len(
                    sequence_positions
                )
            )
        )
    )


    article_collision = (
        article_id
        in
        BLOCK_4_EXISTING_ARTICLE_IDS
    )


    sentence_collision = (
        not
        set(
            sentence_ids
        ).isdisjoint(
            BLOCK_4_EXISTING_SENTENCE_IDS
        )
    )


    provenance_valid = all(
        [
            article_payload[
                "article_metadata"
            ][
                "source"
            ]
            is not None,

            article_payload[
                "article_metadata"
            ][
                "text"
            ]
            is not None,
        ]
    )


    structurally_eligible = all(
        [
            len(
                records
            )
            >
            0,

            sentence_identity_valid,
            sentence_identity_unique,
            canonical_sequence_valid,
            not
            article_collision,
            not
            sentence_collision,
            provenance_valid,
        ]
    )


    representation_ready = (
        structurally_eligible
        and
        all(
            [
                (
                    record[
                        "contextual_representation"
                    ]
                    is not None
                )
                and
                (
                    tuple(
                        record[
                            "contextual_representation"
                        ].shape
                    )
                    ==
                    (
                        BLOCK_4_EXPECTED_EMBEDDING_DIMENSION,
                    )
                )
                and
                bool(
                    np.isfinite(
                        record[
                            "contextual_representation"
                        ]
                    ).all()
                )
                and
                (
                    record[
                        "representation_model"
                    ]
                    ==
                    BLOCK_4_EXPECTED_ENCODER_MODEL
                )
                and
                (
                    record[
                        "representation_normalised"
                    ]
                    is True
                )

                for record
                in records
            ]
        )
    )


    BLOCK_4_NEW_ARTICLE_AUDIT[
        article_id
    ] = {
        "article_id":
            article_id,

        "title":
            article_payload[
                "article_metadata"
            ][
                "title"
            ],

        "source":
            article_payload[
                "article_metadata"
            ][
                "source"
            ],

        "paragraph_count":
            article_payload[
                "paragraph_count"
            ],

        "sentence_count":
            len(
                records
            ),

        "sentence_identity_valid":
            sentence_identity_valid,

        "sentence_identity_unique":
            sentence_identity_unique,

        "canonical_sequence_valid":
            canonical_sequence_valid,

        "article_collision":
            article_collision,

        "sentence_collision":
            sentence_collision,

        "provenance_valid":
            provenance_valid,

        "structurally_eligible":
            structurally_eligible,

        "representation_ready":
            representation_ready,

        "supervision_status":
            "alignment_not_yet_assessed",
    }


    if structurally_eligible:

        BLOCK_4_ACCEPTED_ADDITIONAL_ARTICLES[
            article_id
        ] = deepcopy(
            records
        )


        if representation_ready:

            BLOCK_4_REPRESENTATION_READY_ADDITIONAL_ARTICLES[
                article_id
            ] = deepcopy(
                records
            )

        else:

            BLOCK_4_REPRESENTATION_PENDING_ADDITIONAL_ARTICLES[
                article_id
            ] = deepcopy(
                records
            )


# =============================================================================
# Corpus merge
# =============================================================================

NOTEBOOK_10_EXTENDED_ARTICLE_RECORDS = {
    article_id:
        deepcopy(
            records
        )

    for (
        article_id,
        records,
    ) in BLOCK_4_EXISTING_ARTICLE_RECORDS.items()
}


for (
    article_id,
    records,
) in BLOCK_4_ACCEPTED_ADDITIONAL_ARTICLES.items():

    if article_id in NOTEBOOK_10_EXTENDED_ARTICLE_RECORDS:

        raise RuntimeError(
            "External article identity collision detected "
            "during corpus merge."
        )


    NOTEBOOK_10_EXTENDED_ARTICLE_RECORDS[
        article_id
    ] = deepcopy(
        records
    )


# =============================================================================
# Construct merged canonical sentence collection
# =============================================================================

NOTEBOOK_10_EXTENDED_CANONICAL_SENTENCE_RECORDS = []


for article_id in sorted(
    NOTEBOOK_10_EXTENDED_ARTICLE_RECORDS.keys()
):

    article_records = sorted(
        NOTEBOOK_10_EXTENDED_ARTICLE_RECORDS[
            article_id
        ],
        key=
            lambda record:
                (
                    record[
                        "sequence_position"
                    ],
                    record[
                        "sentence_id"
                    ],
                ),
    )


    for (
        canonical_position,
        record,
    ) in enumerate(
        article_records
    ):

        canonical_record = deepcopy(
            record
        )


        if (
            "source_sequence_position"
            not in
            canonical_record
        ):

            canonical_record[
                "source_sequence_position"
            ] = canonical_record[
                "sequence_position"
            ]


        canonical_record[
            "canonical_sequence_position"
        ] = int(
            canonical_position
        )


        canonical_record[
            "sequence_position"
        ] = int(
            canonical_position
        )


        NOTEBOOK_10_EXTENDED_CANONICAL_SENTENCE_RECORDS.append(
            canonical_record
        )


# =============================================================================
# Global merged identity validation
# =============================================================================

BLOCK_4_EXTENDED_ARTICLE_IDS = list(
    NOTEBOOK_10_EXTENDED_ARTICLE_RECORDS.keys()
)


BLOCK_4_EXTENDED_SENTENCE_IDS = [
    record[
        "sentence_id"
    ]

    for record
    in NOTEBOOK_10_EXTENDED_CANONICAL_SENTENCE_RECORDS
]


BLOCK_4_EXTENDED_ARTICLE_IDS_UNIQUE = (
    len(
        BLOCK_4_EXTENDED_ARTICLE_IDS
    )
    ==
    len(
        set(
            BLOCK_4_EXTENDED_ARTICLE_IDS
        )
    )
)


BLOCK_4_EXTENDED_SENTENCE_IDS_UNIQUE = (
    len(
        BLOCK_4_EXTENDED_SENTENCE_IDS
    )
    ==
    len(
        set(
            BLOCK_4_EXTENDED_SENTENCE_IDS
        )
    )
)


# =============================================================================
# Global canonical-sequence validation
# =============================================================================

BLOCK_4_EXTENDED_CANONICAL_SEQUENCE_VALID = True


for article_id in BLOCK_4_EXTENDED_ARTICLE_IDS:

    article_records = [
        record

        for record
        in NOTEBOOK_10_EXTENDED_CANONICAL_SENTENCE_RECORDS

        if (
            record[
                "article_id"
            ]
            ==
            article_id
        )
    ]


    observed_positions = [
        record[
            "canonical_sequence_position"
        ]

        for record
        in article_records
    ]


    expected_positions = list(
        range(
            len(
                article_records
            )
        )
    )


    if (
        observed_positions
        !=
        expected_positions
    ):

        BLOCK_4_EXTENDED_CANONICAL_SEQUENCE_VALID = False

        break


# =============================================================================
# Determine representation-ready articles across the complete corpus
# =============================================================================

BLOCK_4_EXTENDED_REPRESENTATION_READY_ARTICLES = {}


for (
    article_id,
    records,
) in NOTEBOOK_10_EXTENDED_ARTICLE_RECORDS.items():

    representation_ready = all(
        (
            record.get(
                "contextual_representation"
            )
            is not None
            and
            tuple(
                record[
                    "contextual_representation"
                ].shape
            )
            ==
            (
                BLOCK_4_EXPECTED_EMBEDDING_DIMENSION,
            )
            and
            bool(
                np.isfinite(
                    record[
                        "contextual_representation"
                    ]
                ).all()
            )
        )

        for record
        in records
    )


    if representation_ready:

        BLOCK_4_EXTENDED_REPRESENTATION_READY_ARTICLES[
            article_id
        ] = deepcopy(
            records
        )


# =============================================================================
# Corpus counts
# =============================================================================

BLOCK_4_ACCEPTED_ADDITIONAL_ARTICLE_COUNT = len(
    BLOCK_4_ACCEPTED_ADDITIONAL_ARTICLES
)


BLOCK_4_ACCEPTED_ADDITIONAL_SENTENCE_COUNT = sum(
    len(
        records
    )

    for records
    in BLOCK_4_ACCEPTED_ADDITIONAL_ARTICLES.values()
)


BLOCK_4_REPRESENTATION_READY_ADDITIONAL_ARTICLE_COUNT = len(
    BLOCK_4_REPRESENTATION_READY_ADDITIONAL_ARTICLES
)


BLOCK_4_REPRESENTATION_PENDING_ADDITIONAL_ARTICLE_COUNT = len(
    BLOCK_4_REPRESENTATION_PENDING_ADDITIONAL_ARTICLES
)


BLOCK_4_TOTAL_ARTICLE_COUNT = len(
    NOTEBOOK_10_EXTENDED_ARTICLE_RECORDS
)


BLOCK_4_TOTAL_SENTENCE_COUNT = len(
    NOTEBOOK_10_EXTENDED_CANONICAL_SENTENCE_RECORDS
)


BLOCK_4_TOTAL_REPRESENTATION_READY_ARTICLE_COUNT = len(
    BLOCK_4_EXTENDED_REPRESENTATION_READY_ARTICLES
)


# =============================================================================
# Paragraph-structure audit
# =============================================================================

BLOCK_4_PARAGRAPH_STRUCTURE_PRESERVED = all(
    (
        record.get(
            "paragraph_id"
        )
        is not None
    )

    for records
    in BLOCK_4_ACCEPTED_ADDITIONAL_ARTICLES.values()

    for record
    in records
)


# =============================================================================
# Pilot-corpus preservation
# =============================================================================

BLOCK_4_EXISTING_CORPUS_PRESERVED = (
    block_4_objects_exact(
        BLOCK_4_EXISTING_ARTICLE_RECORDS,
        BLOCK_4_EXISTING_ARTICLE_RECORDS_BEFORE,
    )
)


# =============================================================================
# Synthetic-fallback audit
# =============================================================================

BLOCK_4_SYNTHETIC_FALLBACK_USED = any(
    [
        NOTEBOOK_10_BLOCK_4_SYNTHETIC_TEXT_CREATED,
        NOTEBOOK_10_BLOCK_4_PILOT_ARTICLE_SPLIT,
        NOTEBOOK_10_BLOCK_4_PILOT_ARTICLE_DUPLICATED,
    ]
)


# =============================================================================
# Provenance audit
# =============================================================================

BLOCK_4_NEW_CORPUS_PROVENANCE_VALID = all(
    (
        audit[
            "provenance_valid"
        ]
        is True
    )

    for audit
    in BLOCK_4_NEW_ARTICLE_AUDIT.values()

    if audit[
        "structurally_eligible"
    ]
)


# =============================================================================
# Multi-article readiness
# =============================================================================

BLOCK_4_STRUCTURAL_MULTI_ARTICLE_READY = (
    BLOCK_4_TOTAL_ARTICLE_COUNT
    >=
    BLOCK_4_MINIMUM_MULTI_ARTICLE_COUNT
)


BLOCK_4_REPRESENTATION_MULTI_ARTICLE_READY = (
    BLOCK_4_TOTAL_REPRESENTATION_READY_ARTICLE_COUNT
    >=
    BLOCK_4_MINIMUM_MULTI_ARTICLE_COUNT
)


BLOCK_4_MULTI_ARTICLE_VALIDATION_READY = all(
    [
        BLOCK_4_STRUCTURAL_MULTI_ARTICLE_READY,
        BLOCK_4_REPRESENTATION_MULTI_ARTICLE_READY,
        BLOCK_4_EXTENDED_ARTICLE_IDS_UNIQUE,
        BLOCK_4_EXTENDED_SENTENCE_IDS_UNIQUE,
        BLOCK_4_EXTENDED_CANONICAL_SEQUENCE_VALID,
        BLOCK_4_EXISTING_CORPUS_PRESERVED,
        BLOCK_4_NEW_CORPUS_PROVENANCE_VALID,
        not
        BLOCK_4_SYNTHETIC_FALLBACK_USED,
    ]
)


# =============================================================================
# Expanded-training boundary
# =============================================================================

BLOCK_4_EXPANDED_TRAINING_SUFFICIENCY = False

BLOCK_4_EXPANDED_TRAINING_SUFFICIENCY_REASON = (
    "not_established_by_validation_project"
)


# =============================================================================
# Remaining source / representation requirements
# =============================================================================

BLOCK_4_EXTERNAL_ARTICLE_REQUIRED = (
    not
    BLOCK_4_STRUCTURAL_MULTI_ARTICLE_READY
)


BLOCK_4_CONTEXTUAL_REPRESENTATION_REQUIRED = (
    BLOCK_4_STRUCTURAL_MULTI_ARTICLE_READY
    and
    not
    BLOCK_4_REPRESENTATION_MULTI_ARTICLE_READY
)


# =============================================================================
# Supervision-status inventory
# =============================================================================

NOTEBOOK_10_EXTENDED_ARTICLE_SUPERVISION_STATUS = {}


for article_id in BLOCK_4_EXISTING_ARTICLE_IDS:

    NOTEBOOK_10_EXTENDED_ARTICLE_SUPERVISION_STATUS[
        article_id
    ] = (
        NOTEBOOK_10_EXPANDED_ARTICLE_SUPERVISION_STATUS.get(
            article_id,
            "not_yet_assessed",
        )
    )


for article_id in BLOCK_4_ACCEPTED_ADDITIONAL_ARTICLES:

    NOTEBOOK_10_EXTENDED_ARTICLE_SUPERVISION_STATUS[
        article_id
    ] = (
        "alignment_not_yet_assessed"
    )


# =============================================================================
# Inherited Media AI state validation
# =============================================================================
#
# Compare the end-of-Block-4 runtime state against the exact Notebook 10
# baseline captured before any external ingestion or contextual encoding.
# =============================================================================

BLOCK_4_REPRESENTATION_STATE_AFTER = (
    block_4_snapshot_module_collection(
        NOTEBOOK_10_REPRESENTATION_MODULES
    )
)


BLOCK_4_TRANSFORMATIVE_STATE_AFTER = (
    block_4_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_MODULES
    )
)


BLOCK_4_WEIGHT_STATE_AFTER = (
    block_4_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES
    )
)


BLOCK_4_REPRESENTATION_STATE_UNCHANGED = (
    block_4_nested_state_exact(
        BLOCK_4_REPRESENTATION_STATE_BASELINE,
        BLOCK_4_REPRESENTATION_STATE_AFTER,
    )
)


BLOCK_4_TRANSFORMATIVE_STATE_UNCHANGED = (
    block_4_nested_state_exact(
        BLOCK_4_TRANSFORMATIVE_STATE_BASELINE,
        BLOCK_4_TRANSFORMATIVE_STATE_AFTER,
    )
)


BLOCK_4_WEIGHT_STATE_UNCHANGED = (
    block_4_nested_state_exact(
        BLOCK_4_WEIGHT_STATE_BASELINE,
        BLOCK_4_WEIGHT_STATE_AFTER,
    )
)


# Semantic identity check for reconstructed effective Transformative Weights.
with torch.no_grad():

    BLOCK_4_EFFECTIVE_WEIGHT_IDENTITY_UNCHANGED = all(
        torch.equal(
            NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES[
                pathway_name
            ]()
            .detach()
            .cpu(),

            NOTEBOOK_10_INHERITED_EFFECTIVE_TRANSFORMATIVE_WEIGHTS[
                pathway_name
            ]
            .detach()
            .cpu(),
        )

        for pathway_name
        in NOTEBOOK_10_PATHWAY_DIMENSIONS
    )


BLOCK_4_ALL_PARAMETERS_FROZEN = all(
    not parameter.requires_grad

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


BLOCK_4_ALL_MODULES_EVAL = all(
    not module.training

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()
)


BLOCK_4_ALL_GRADIENTS_ABSENT = all(
    parameter.grad is None

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


# =============================================================================
# Execution-boundary validation
# =============================================================================

BLOCK_4_EXECUTION_BOUNDARY_VALID = all(
    [
        NOTEBOOK_10_BLOCK_4_MEDIA_AI_REPRESENTATION_FORWARD_EXECUTED
        is False,

        NOTEBOOK_10_BLOCK_4_TRANSFORMATIVE_FORWARD_EXECUTED
        is False,

        NOTEBOOK_10_BLOCK_4_WEIGHT_FORWARD_EXECUTED
        is False,

        NOTEBOOK_10_BLOCK_4_RECURRENT_STATE_CREATED
        is False,

        NOTEBOOK_10_BLOCK_4_RECURRENT_UPDATE_EXECUTED
        is False,

        NOTEBOOK_10_BLOCK_4_LOSS_CALCULATED
        is False,

        NOTEBOOK_10_BLOCK_4_OPTIMIZER_CREATED
        is False,

        NOTEBOOK_10_BLOCK_4_BACKWARD_PASS_EXECUTED
        is False,

        NOTEBOOK_10_BLOCK_4_PARAMETER_UPDATE_EXECUTED
        is False,

        NOTEBOOK_10_BLOCK_4_TRAINING_EXECUTED
        is False,

        NOTEBOOK_10_BLOCK_4_SYNTHETIC_TEXT_CREATED
        is False,

        NOTEBOOK_10_BLOCK_4_PILOT_ARTICLE_SPLIT
        is False,

        NOTEBOOK_10_BLOCK_4_PILOT_ARTICLE_DUPLICATED
        is False,

        NOTEBOOK_10_BLOCK_4_SUPERVISION_USED_AS_TEXT
        is False,

        NOTEBOOK_10_BLOCK_4_SUPERVISION_USED_AS_REPRESENTATION
        is False,
    ]
)


# =============================================================================
# Block validity
# =============================================================================
#
# IMPORTANT:
#
# Block 4 remains valid when no external article has yet been supplied.
#
# In that state it establishes the ingestion contract and reports that
# additional external material remains required.
# =============================================================================

NOTEBOOK_10_BLOCK_4_ERRORS = []


BLOCK_4_VALIDATION_CHECKS = {
    "prerequisites_invalid":
        BLOCK_4_PREREQUISITES_VALID,

    "existing_corpus_missing":
        (
            BLOCK_4_EXISTING_ARTICLE_COUNT
            >=
            1
        ),

    "existing_corpus_not_preserved":
        BLOCK_4_EXISTING_CORPUS_PRESERVED,

    "extended_article_ids_not_unique":
        BLOCK_4_EXTENDED_ARTICLE_IDS_UNIQUE,

    "extended_sentence_ids_not_unique":
        BLOCK_4_EXTENDED_SENTENCE_IDS_UNIQUE,

    "extended_canonical_sequence_invalid":
        BLOCK_4_EXTENDED_CANONICAL_SEQUENCE_VALID,

    "new_corpus_provenance_invalid":
        BLOCK_4_NEW_CORPUS_PROVENANCE_VALID,

    "synthetic_fallback_used":
        (
            not
            BLOCK_4_SYNTHETIC_FALLBACK_USED
        ),

    "representation_state_changed":
        BLOCK_4_REPRESENTATION_STATE_UNCHANGED,

    "transformative_state_changed":
        BLOCK_4_TRANSFORMATIVE_STATE_UNCHANGED,

    "weight_state_changed":
        BLOCK_4_WEIGHT_STATE_UNCHANGED,

    "effective_weight_identity_changed":
        BLOCK_4_EFFECTIVE_WEIGHT_IDENTITY_UNCHANGED,

    "parameters_not_frozen":
        BLOCK_4_ALL_PARAMETERS_FROZEN,

    "modules_not_eval":
        BLOCK_4_ALL_MODULES_EVAL,

    "unexpected_gradients":
        BLOCK_4_ALL_GRADIENTS_ABSENT,

    "execution_boundary_invalid":
        BLOCK_4_EXECUTION_BOUNDARY_VALID,
}


for (
    error_name,
    condition,
) in BLOCK_4_VALIDATION_CHECKS.items():

    if not condition:

        NOTEBOOK_10_BLOCK_4_ERRORS.append(
            error_name
        )


NOTEBOOK_10_BLOCK_4_VALID = (
    len(
        NOTEBOOK_10_BLOCK_4_ERRORS
    )
    ==
    0
)


if not NOTEBOOK_10_BLOCK_4_VALID:

    raise RuntimeError(
        "Notebook 10 Block 4 external article ingestion failed. "
        f"Errors: {NOTEBOOK_10_BLOCK_4_ERRORS}"
    )


NOTEBOOK_10_BLOCK_4_COMPLETE = True


# =============================================================================
# Stable downstream globals
# =============================================================================

NOTEBOOK_10_MULTI_ARTICLE_ARTICLE_RECORDS = {
    article_id:
        deepcopy(
            records
        )

    for (
        article_id,
        records,
    ) in NOTEBOOK_10_EXTENDED_ARTICLE_RECORDS.items()
}


NOTEBOOK_10_MULTI_ARTICLE_CANONICAL_SENTENCE_RECORDS = deepcopy(
    NOTEBOOK_10_EXTENDED_CANONICAL_SENTENCE_RECORDS
)


NOTEBOOK_10_MULTI_ARTICLE_REPRESENTATION_READY_ARTICLES = {
    article_id:
        deepcopy(
            records
        )

    for (
        article_id,
        records,
    ) in BLOCK_4_EXTENDED_REPRESENTATION_READY_ARTICLES.items()
}


NOTEBOOK_10_MULTI_ARTICLE_VALIDATION_READY = (
    BLOCK_4_MULTI_ARTICLE_VALIDATION_READY
)


NOTEBOOK_10_EXTERNAL_ARTICLE_REQUIRED = (
    BLOCK_4_EXTERNAL_ARTICLE_REQUIRED
)


NOTEBOOK_10_CONTEXTUAL_REPRESENTATION_REQUIRED = (
    BLOCK_4_CONTEXTUAL_REPRESENTATION_REQUIRED
)


NOTEBOOK_10_EXPANDED_TRAINING_SUFFICIENCY = (
    BLOCK_4_EXPANDED_TRAINING_SUFFICIENCY
)


# =============================================================================
# Block summary
# =============================================================================

NOTEBOOK_10_BLOCK_4_SUMMARY = {
    "block":
        NOTEBOOK_10_BLOCK_4,

    "block_name":
        NOTEBOOK_10_BLOCK_4_NAME,

    "block_version":
        NOTEBOOK_10_BLOCK_4_VERSION,

    "supplied_external_articles":
        BLOCK_4_SUPPLIED_ARTICLE_COUNT,

    "accepted_raw_articles":
        len(
            BLOCK_4_ACCEPTED_RAW_ARTICLES
        ),

    "rejected_raw_articles":
        len(
            BLOCK_4_REJECTED_RAW_ARTICLES
        ),

    "structurally_accepted_articles":
        BLOCK_4_ACCEPTED_ADDITIONAL_ARTICLE_COUNT,

    "structurally_accepted_sentences":
        BLOCK_4_ACCEPTED_ADDITIONAL_SENTENCE_COUNT,

    "representation_ready_additional_articles":
        BLOCK_4_REPRESENTATION_READY_ADDITIONAL_ARTICLE_COUNT,

    "representation_pending_additional_articles":
        BLOCK_4_REPRESENTATION_PENDING_ADDITIONAL_ARTICLE_COUNT,

    "total_articles":
        BLOCK_4_TOTAL_ARTICLE_COUNT,

    "total_sentences":
        BLOCK_4_TOTAL_SENTENCE_COUNT,

    "total_representation_ready_articles":
        BLOCK_4_TOTAL_REPRESENTATION_READY_ARTICLE_COUNT,

    "contextual_encoder_available":
        BLOCK_4_CONTEXTUAL_ENCODER_AVAILABLE,

    "contextual_encoder_executed":
        NOTEBOOK_10_BLOCK_4_CONTEXTUAL_ENCODER_EXECUTED,

    "paragraph_structure_preserved":
        BLOCK_4_PARAGRAPH_STRUCTURE_PRESERVED,

    "existing_corpus_preserved":
        BLOCK_4_EXISTING_CORPUS_PRESERVED,

    "representation_state_unchanged":
        BLOCK_4_REPRESENTATION_STATE_UNCHANGED,

    "transformative_state_unchanged":
        BLOCK_4_TRANSFORMATIVE_STATE_UNCHANGED,

    "weight_state_unchanged":
        BLOCK_4_WEIGHT_STATE_UNCHANGED,

    "effective_weight_identity_unchanged":
        BLOCK_4_EFFECTIVE_WEIGHT_IDENTITY_UNCHANGED,

    "article_ids_unique":
        BLOCK_4_EXTENDED_ARTICLE_IDS_UNIQUE,

    "sentence_ids_unique":
        BLOCK_4_EXTENDED_SENTENCE_IDS_UNIQUE,

    "canonical_sequence_valid":
        BLOCK_4_EXTENDED_CANONICAL_SEQUENCE_VALID,

    "synthetic_fallback_used":
        BLOCK_4_SYNTHETIC_FALLBACK_USED,

    "structural_multi_article_ready":
        BLOCK_4_STRUCTURAL_MULTI_ARTICLE_READY,

    "representation_multi_article_ready":
        BLOCK_4_REPRESENTATION_MULTI_ARTICLE_READY,

    "multi_article_validation_ready":
        BLOCK_4_MULTI_ARTICLE_VALIDATION_READY,

    "expanded_training_sufficiency":
        BLOCK_4_EXPANDED_TRAINING_SUFFICIENCY,

    "external_article_required":
        BLOCK_4_EXTERNAL_ARTICLE_REQUIRED,

    "contextual_representation_required":
        BLOCK_4_CONTEXTUAL_REPRESENTATION_REQUIRED,

    "block_valid":
        NOTEBOOK_10_BLOCK_4_VALID,

    "block_complete":
        NOTEBOOK_10_BLOCK_4_COMPLETE,
}


# =============================================================================
# Final report
# =============================================================================

print("=" * 72)

print(
    "Media AI — Notebook 10, Block 4: "
    "External Additional-Article Ingestion "
    "and Canonical Corpus Extension"
)

print("=" * 72)

print(
    f"Block version                : "
    f"{NOTEBOOK_10_BLOCK_4_VERSION}"
)

print("-" * 72)

print(
    "External article input"
)

print(
    f"Articles supplied            : "
    f"{BLOCK_4_SUPPLIED_ARTICLE_COUNT}"
)

print(
    f"Raw articles accepted        : "
    f"{len(BLOCK_4_ACCEPTED_RAW_ARTICLES)}"
)

print(
    f"Raw articles rejected        : "
    f"{len(BLOCK_4_REJECTED_RAW_ARTICLES)}"
)

print(
    f"External text ingested       : "
    f"{NOTEBOOK_10_BLOCK_4_EXTERNAL_TEXT_INGESTED}"
)

print("-" * 72)

print(
    "Canonical structural preparation"
)

print(
    f"Structurally accepted articles: "
    f"{BLOCK_4_ACCEPTED_ADDITIONAL_ARTICLE_COUNT}"
)

print(
    f"Accepted additional sentences: "
    f"{BLOCK_4_ACCEPTED_ADDITIONAL_SENTENCE_COUNT}"
)

print(
    f"Structural rejections        : "
    f"{len(BLOCK_4_STRUCTURAL_REJECTIONS)}"
)

print(
    f"Paragraph structure preserved: "
    f"{BLOCK_4_PARAGRAPH_STRUCTURE_PRESERVED}"
)

print("-" * 72)

print(
    "Contextual representation preparation"
)

print(
    f"Encoder model                : "
    f"{BLOCK_4_EXPECTED_ENCODER_MODEL}"
)

print(
    f"Embedding dimension          : "
    f"{BLOCK_4_EXPECTED_EMBEDDING_DIMENSION}"
)

print(
    f"Normalisation required       : "
    f"{BLOCK_4_EXPECTED_NORMALISED}"
)

print(
    f"Encoder available            : "
    f"{BLOCK_4_CONTEXTUAL_ENCODER_AVAILABLE}"
)

print(
    f"Encoder executed             : "
    f"{NOTEBOOK_10_BLOCK_4_CONTEXTUAL_ENCODER_EXECUTED}"
)

print(
    f"Ready additional articles    : "
    f"{BLOCK_4_REPRESENTATION_READY_ADDITIONAL_ARTICLE_COUNT}"
)

print(
    f"Pending additional articles  : "
    f"{BLOCK_4_REPRESENTATION_PENDING_ADDITIONAL_ARTICLE_COUNT}"
)

print("-" * 72)

print(
    "Extended corpus"
)

print(
    f"Existing articles            : "
    f"{BLOCK_4_EXISTING_ARTICLE_COUNT}"
)

print(
    f"Existing sentences           : "
    f"{BLOCK_4_EXISTING_SENTENCE_COUNT}"
)

print(
    f"Total articles               : "
    f"{BLOCK_4_TOTAL_ARTICLE_COUNT}"
)

print(
    f"Total sentences              : "
    f"{BLOCK_4_TOTAL_SENTENCE_COUNT}"
)

print(
    f"Representation-ready articles: "
    f"{BLOCK_4_TOTAL_REPRESENTATION_READY_ARTICLE_COUNT}"
)

print(
    f"Article IDs unique           : "
    f"{BLOCK_4_EXTENDED_ARTICLE_IDS_UNIQUE}"
)

print(
    f"Sentence IDs unique          : "
    f"{BLOCK_4_EXTENDED_SENTENCE_IDS_UNIQUE}"
)

print(
    f"Canonical sequences valid    : "
    f"{BLOCK_4_EXTENDED_CANONICAL_SEQUENCE_VALID}"
)

print(
    f"Existing corpus preserved    : "
    f"{BLOCK_4_EXISTING_CORPUS_PRESERVED}"
)

print(
    f"New provenance valid         : "
    f"{BLOCK_4_NEW_CORPUS_PROVENANCE_VALID}"
)

print(
    f"Synthetic fallback used      : "
    f"{BLOCK_4_SYNTHETIC_FALLBACK_USED}"
)

print("-" * 72)

print(
    "Corpus readiness"
)

print(
    f"Structural multi-article     : "
    f"{BLOCK_4_STRUCTURAL_MULTI_ARTICLE_READY}"
)

print(
    f"Representation multi-article : "
    f"{BLOCK_4_REPRESENTATION_MULTI_ARTICLE_READY}"
)

print(
    f"Multi-article validation ready: "
    f"{BLOCK_4_MULTI_ARTICLE_VALIDATION_READY}"
)

print(
    f"External article required    : "
    f"{BLOCK_4_EXTERNAL_ARTICLE_REQUIRED}"
)

print(
    f"Contextual repr. required    : "
    f"{BLOCK_4_CONTEXTUAL_REPRESENTATION_REQUIRED}"
)

print(
    f"Expanded-training sufficient : "
    f"{BLOCK_4_EXPANDED_TRAINING_SUFFICIENCY}"
)

print("-" * 72)

print(
    "Inherited Media AI state"
)

print(
    f"Representation unchanged     : "
    f"{BLOCK_4_REPRESENTATION_STATE_UNCHANGED}"
)

print(
    f"Transformative unchanged     : "
    f"{BLOCK_4_TRANSFORMATIVE_STATE_UNCHANGED}"
)

print(
    f"Transform Weight unchanged   : "
    f"{BLOCK_4_WEIGHT_STATE_UNCHANGED}"
)

print(
    f"Effective Weight identity    : "
    f"{BLOCK_4_EFFECTIVE_WEIGHT_IDENTITY_UNCHANGED}"
)

print(
    f"All parameters frozen        : "
    f"{BLOCK_4_ALL_PARAMETERS_FROZEN}"
)

print(
    f"All modules eval             : "
    f"{BLOCK_4_ALL_MODULES_EVAL}"
)

print(
    f"All gradients absent         : "
    f"{BLOCK_4_ALL_GRADIENTS_ABSENT}"
)

print("-" * 72)

print(
    "Execution boundary"
)

print(
    f"Contextual encoder executed  : "
    f"{NOTEBOOK_10_BLOCK_4_CONTEXTUAL_ENCODER_EXECUTED}"
)

print(
    f"Media AI repr. forward       : "
    f"{NOTEBOOK_10_BLOCK_4_MEDIA_AI_REPRESENTATION_FORWARD_EXECUTED}"
)

print(
    f"Transform forward            : "
    f"{NOTEBOOK_10_BLOCK_4_TRANSFORMATIVE_FORWARD_EXECUTED}"
)

print(
    f"Weight forward               : "
    f"{NOTEBOOK_10_BLOCK_4_WEIGHT_FORWARD_EXECUTED}"
)

print(
    f"Recurrent state created      : "
    f"{NOTEBOOK_10_BLOCK_4_RECURRENT_STATE_CREATED}"
)

print(
    f"Recurrent update executed    : "
    f"{NOTEBOOK_10_BLOCK_4_RECURRENT_UPDATE_EXECUTED}"
)

print(
    f"Loss calculated              : "
    f"{NOTEBOOK_10_BLOCK_4_LOSS_CALCULATED}"
)

print(
    f"Optimizer created            : "
    f"{NOTEBOOK_10_BLOCK_4_OPTIMIZER_CREATED}"
)

print(
    f"Backward pass executed       : "
    f"{NOTEBOOK_10_BLOCK_4_BACKWARD_PASS_EXECUTED}"
)

print(
    f"Parameter update executed    : "
    f"{NOTEBOOK_10_BLOCK_4_PARAMETER_UPDATE_EXECUTED}"
)

print(
    f"Training executed            : "
    f"{NOTEBOOK_10_BLOCK_4_TRAINING_EXECUTED}"
)

print("-" * 72)

print(
    f"Block valid                  : "
    f"{NOTEBOOK_10_BLOCK_4_VALID}"
)

print(
    f"Block complete               : "
    f"{NOTEBOOK_10_BLOCK_4_COMPLETE}"
)

print("=" * 72)


# =============================================================================
# Completion interpretation
# =============================================================================

if BLOCK_4_MULTI_ARTICLE_VALIDATION_READY:

    print(
        "The external article ingestion contract and canonical corpus "
        "extension stage completed successfully."
    )

    print(
        f"The expanded corpus now contains "
        f"{BLOCK_4_TOTAL_ARTICLE_COUNT} independent articles and "
        f"{BLOCK_4_TOTAL_SENTENCE_COUNT} canonical sentences."
    )

    print(
        f"{BLOCK_4_TOTAL_REPRESENTATION_READY_ARTICLE_COUNT} articles "
        "possess valid 768-dimensional contextual representations."
    )

    print(
        "The original validated corpus remains exactly unchanged."
    )

    print(
        "No synthetic article construction, pilot splitting, duplication "
        "or supervision substitution occurred."
    )

    print(
        "Notebook 10 may now proceed to formal multi-article corpus "
        "validation and controlled recurrent execution."
    )


elif BLOCK_4_STRUCTURAL_MULTI_ARTICLE_READY:

    print(
        "A genuine structurally valid multi-article corpus is now available."
    )

    print(
        f"The expanded corpus contains "
        f"{BLOCK_4_TOTAL_ARTICLE_COUNT} articles and "
        f"{BLOCK_4_TOTAL_SENTENCE_COUNT} canonical sentences."
    )

    print(
        "However, fewer than two articles currently possess validated "
        "contextual representations."
    )

    if BLOCK_4_CONTEXTUAL_ENCODER_ERROR is not None:

        print(
            "The contextual encoder could not be initialised during "
            "this execution."
        )

        print(
            f"Encoder error               : "
            f"{BLOCK_4_CONTEXTUAL_ENCODER_ERROR['error_type']}: "
            f"{BLOCK_4_CONTEXTUAL_ENCODER_ERROR['error']}"
        )

    print(
        "Contextual representation preparation must be completed before "
        "multi-article recurrent validation."
    )


else:

    print(
        "No new external article has yet produced a structurally valid "
        "multi-article corpus."
    )

    if not BLOCK_4_EXTERNAL_INPUT_AVAILABLE:

        print(
            "No external article was supplied through "
            "NOTEBOOK_10_EXTERNAL_ARTICLE_INPUTS."
        )

    print(
        "The existing validated pilot corpus remains unchanged."
    )

    print(
        "A genuine independent article must be supplied before "
        "multi-article validation can proceed."
    )


print("=" * 72)
```

# Block 5 — Formal Multi-Article Corpus Validation and Execution-Readiness Audit

## Purpose

This block performs the formal structural and representational validation of the expanded Notebook 10 corpus before any multi-article recurrent execution is allowed.

Block 4 established the controlled ingestion pathway for genuinely independent external article material and defined the conditions under which new articles can be merged with the validated pilot corpus.

The present block determines whether the resulting corpus satisfies the complete multi-article execution contract.

Its purpose is to validate:

- article identity;
- sentence identity;
- paragraph structure where available;
- canonical within-article ordering;
- article-boundary independence;
- contextual representation dimensionality;
- contextual representation finiteness;
- contextual representation alignment;
- encoder compatibility;
- pilot-corpus preservation;
- absence of synthetic corpus expansion;
- readiness for controlled recurrent traversal.

No recurrent state is constructed and no Media AI forward execution is performed in this block.

## Methodological Role

Block 5 forms the final corpus-validation boundary before the completed Media AI architecture is executed across more than one article.

The preceding blocks established:

- exact restoration of the Notebook 09 architecture;
- inventory of the currently available corpus;
- confirmation that the original project material contained only one pilot article;
- a controlled pathway for adding genuinely independent external article text;
- deterministic article, paragraph and sentence identity construction;
- preparation or generation of compatible contextual representations.

Block 5 now determines whether those elements jointly form a valid multi-article execution source.

The distinction is important:

**corpus construction ≠ corpus validation ≠ recurrent execution**

The present block performs the second of these stages.

## Multi-Article Validation Requirement

The minimum architectural requirement for multi-article validation is

$$
N_{\mathrm{representation-ready\ articles}} \ge 2.
$$

This threshold does not establish substantive corpus diversity or training adequacy.

It establishes only that the architecture can be tested across at least two independent article sequences.

If fewer than two representation-ready articles exist, Block 5 may complete its audit, but multi-article execution readiness must remain false.

## Canonical Article Set

Let the validated corpus contain

$$
\mathcal{C}
=
\{
\mathcal{A}_1,
\mathcal{A}_2,
\ldots,
\mathcal{A}_N
\}.
$$

Each article $\mathcal{A}_i$ must represent an independent sequence of canonical sentence records:

$$
\mathcal{A}_i
=
\left(
s_{i,0},
s_{i,1},
\ldots,
s_{i,T_i-1}
\right).
$$

The corpus must preserve explicit article boundaries.

No sentence may belong to more than one article.

## Article Identity Validation

Every article identifier must be:

- non-null;
- stable;
- unique within the corpus;
- distinct from every other article identifier;
- associated consistently with all member sentences.

The pilot article identity must remain unchanged from Blocks 2–4.

Any article identity collision invalidates multi-article execution readiness.

## Sentence Identity Validation

Every sentence identifier must be globally unique.

For any two distinct records,

$$
s_{i,t} \neq s_{j,u},
$$

the corresponding identities must satisfy

$$
\mathrm{id}(s_{i,t})
\neq
\mathrm{id}(s_{j,u}).
$$

Sentence identity must remain stable across:

- corpus construction;
- contextual representation preparation;
- recurrent execution;
- supervision alignment;
- later evaluation.

A sentence-identity collision invalidates the execution source.

## Paragraph-Structure Audit

Paragraph identity should be retained where available.

The preferred hierarchy remains

$$
\mathrm{Article}
\rightarrow
\mathrm{Paragraph}
\rightarrow
\mathrm{Sentence}.
$$

Paragraph structure is not required for recurrent state propagation, but its preservation supports:

- structural traceability;
- later paragraph-level aggregation;
- corpus integrity;
- reproducibility.

If an article is intentionally represented as a single paragraph because no paragraph structure was available at ingestion, this condition should remain explicit.

## Canonical Sequence Validation

Within each article, canonical sentence positions must form a complete zero-based sequence:

$$
0,1,\ldots,T_i-1.
$$

Accordingly,

$$
\mathrm{position}(s_{i,t}) = t.
$$

The block must confirm:

- no missing sequence positions;
- no duplicated sequence positions;
- no negative positions;
- deterministic ordering;
- consistency between stored sequence position and reconstructed canonical order.

Sentence ordering within an article must never depend on model output.

## Article-Boundary Independence

The corpus must expose each article as an independent future recurrent sequence.

The final state of article $i$ must never become the initial state of article $i+1$.

Later recurrent execution must enforce

$$
\mathbf{h}_{i+1,0}^{(p)}
=
\mathbf{0}
$$

for each pathway

$$
p
\in
\{
\mathrm{factual},
\mathrm{psychological},
\mathrm{social}
\}.
$$

Block 5 validates that the corpus structure permits this reset policy.

It does not create the recurrent states themselves.

## Contextual Representation Contract

Every representation-ready sentence must contain one contextual representation

$$
\mathbf{x}_{i,t}
\in
\mathbb{R}^{768}.
$$

For each article,

$$
\mathbf{X}_i
\in
\mathbb{R}^{T_i \times 768}.
$$

The block must validate:

- exactly one representation per sentence;
- dimensionality exactly 768;
- numerical finiteness;
- deterministic sentence alignment;
- deterministic article alignment.

## Encoder Compatibility

Where encoder provenance is available, contextual representations must conform to the established contract:

**Provider:** `sentence-transformers`

**Model:** `sentence-transformers/all-mpnet-base-v2`

**Embedding dimension:** $768$

**Normalisation:** enabled

For newly generated representations, all four properties should be explicitly verifiable.

For the inherited pilot representation artefact, the previously validated Notebook 05 encoder contract remains authoritative.

The block must not silently accept representations from a different model merely because they have dimension 768.

## Normalisation Audit

For representations generated under the current encoder contract, unit normalisation should satisfy approximately

$$
\|\mathbf{x}_{i,t}\|_2 = 1.
$$

Because floating-point arithmetic is finite-precision, numerical equality is evaluated within an explicit tolerance rather than by exact identity.

Normalisation failure should invalidate representation readiness for the affected sentence or article.

## Representation Matrix Construction

For each representation-ready article, Block 5 should construct or validate the matrix

$$
\mathbf{X}_i
=
\begin{bmatrix}
\mathbf{x}_{i,0}\\
\mathbf{x}_{i,1}\\
\vdots\\
\mathbf{x}_{i,T_i-1}
\end{bmatrix}.
$$

The row count must satisfy

$$
\mathrm{rows}(\mathbf{X}_i)=T_i,
$$

and the column count must satisfy

$$
\mathrm{cols}(\mathbf{X}_i)=768.
$$

The matrix must contain only finite values.

## Pilot-Corpus Preservation

The original pilot article must remain exactly unchanged.

Block 5 should validate that:

- pilot article identity is unchanged;
- pilot sentence identities are unchanged;
- pilot canonical order is unchanged;
- pilot contextual representations are unchanged;
- pilot article membership is unchanged.

The expanded corpus must therefore extend the pilot corpus rather than modify it.

## Corpus Provenance

Every article must retain auditable provenance.

For the pilot article, provenance is inherited from the validated project artefacts.

For externally ingested articles, the block should preserve whatever source metadata were supplied, including where available:

- source;
- title;
- publication;
- publication date;
- author;
- language;
- section;
- source identifier;
- URL;
- ingestion timestamp.

Provenance is descriptive metadata and is not automatically used as a model input.

## Synthetic-Fallback Audit

The corpus must remain free of synthetic expansion.

Block 5 must confirm that:

- no pilot article was split into pseudo-articles;
- no pilot article was duplicated;
- no sentence was synthetically generated;
- no supervision value was used as text;
- no model-generated article was introduced as independent validation data.

The validation source must represent genuine independently supplied language.

## Supervision Status

Supervision is not required for Block 5 execution readiness.

Each article may remain classified as:

- `available`;
- `partially_available`;
- `unavailable`;
- `alignment_not_yet_assessed`;
- `available_in_separate_validated_project_artefacts`.

The absence of supervision does not invalidate representation-level recurrent validation.

However, supervision availability must remain separate and traceable because later training and quantitative evaluation require a dedicated alignment contract.

## Structural Readiness

The expanded corpus is structurally ready only if:

- at least two independent articles are present;
- all article identities are valid and unique;
- all sentence identities are valid and globally unique;
- every article contains at least one sentence;
- canonical sequence positions are complete and deterministic;
- article boundaries are explicit;
- pilot identity is preserved;
- provenance is auditable;
- no synthetic fallback has been used.

## Representation Readiness

The expanded corpus is representation ready only if at least two structurally valid articles contain complete compatible contextual representations.

Formally,

$$
N_{\mathrm{representation-ready\ articles}}
\ge 2.
$$

Every ready article must satisfy:

$$
\mathbf{X}_i
\in
\mathbb{R}^{T_i \times 768},
$$

with finite values and valid sentence-level alignment.

## Multi-Article Execution Readiness

The corpus may proceed to recurrent execution only if both conditions hold:

$$
\mathrm{structural\ readiness}
=
\mathrm{True}
$$

and

$$
\mathrm{representation\ readiness}
=
\mathrm{True}.
$$

Therefore,

$$
\mathrm{multi\text{-}article\ execution\ ready}
=
\mathrm{structural\ readiness}
\land
\mathrm{representation\ readiness}.
$$

If either condition is false, recurrent execution must remain blocked.

## Expanded-Training Boundary

Successful multi-article corpus validation does not imply that the corpus is sufficient for expanded training.

Even if

$$
N_{\mathrm{representation-ready\ articles}} \ge 2,
$$

the validation project must continue to report

$$
\mathrm{expanded\ training\ sufficient}
=
\mathrm{False}
$$

unless a later dedicated training-phase corpus contract establishes otherwise.

## Model-State Protection

```python
# =============================================================================
# Media AI — Notebook 10
# Block 5: Formal Multi-Article Corpus Validation
#          and Execution-Readiness Audit
# =============================================================================

from copy import deepcopy
from datetime import datetime, timezone
from collections import Counter, defaultdict

import numpy as np
import torch


# =============================================================================
# Block identity
# =============================================================================

NOTEBOOK_10_BLOCK_5 = 5

NOTEBOOK_10_BLOCK_5_NAME = (
    "Formal Multi-Article Corpus Validation "
    "and Execution-Readiness Audit"
)

NOTEBOOK_10_BLOCK_5_VERSION = "1.0"

BLOCK_5_EXECUTED_AT_UTC = (
    datetime.now(
        timezone.utc
    ).isoformat()
)


# =============================================================================
# Methodological constants
# =============================================================================

BLOCK_5_EXPECTED_ENCODER_PROVIDER = (
    "sentence-transformers"
)

BLOCK_5_EXPECTED_ENCODER_MODEL = (
    "sentence-transformers/all-mpnet-base-v2"
)

BLOCK_5_EXPECTED_EMBEDDING_DIMENSION = 768

BLOCK_5_EXPECTED_NORMALISED = True

BLOCK_5_NORMALISATION_ATOL = 1e-4

BLOCK_5_NORMALISATION_RTOL = 1e-4

BLOCK_5_MINIMUM_MULTI_ARTICLE_COUNT = 2

BLOCK_5_ALLOWED_SUPERVISION_STATUSES = {
    "available",
    "partially_available",
    "unavailable",
    "alignment_not_yet_assessed",
    "available_in_separate_validated_project_artefacts",
}


# =============================================================================
# Prerequisite validation
# =============================================================================

BLOCK_5_REQUIRED_GLOBALS = [
    "NOTEBOOK_10_BLOCK_1_COMPLETE",
    "NOTEBOOK_10_BLOCK_1_VALID",
    "NOTEBOOK_10_BLOCK_2_COMPLETE",
    "NOTEBOOK_10_BLOCK_2_VALID",
    "NOTEBOOK_10_BLOCK_3_COMPLETE",
    "NOTEBOOK_10_BLOCK_3_VALID",
    "NOTEBOOK_10_BLOCK_4_COMPLETE",
    "NOTEBOOK_10_BLOCK_4_VALID",
    "NOTEBOOK_10_EXTENDED_ARTICLE_RECORDS",
    "NOTEBOOK_10_EXTENDED_CANONICAL_SENTENCE_RECORDS",
    "NOTEBOOK_10_REPRESENTATION_MODULES",
    "NOTEBOOK_10_TRANSFORMATIVE_MODULES",
    "NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES",
    "NOTEBOOK_10_INHERITED_EFFECTIVE_TRANSFORMATIVE_WEIGHTS",
]


BLOCK_5_MISSING_PREREQUISITES = [
    object_name

    for object_name
    in BLOCK_5_REQUIRED_GLOBALS

    if object_name not in globals()
]


if BLOCK_5_MISSING_PREREQUISITES:

    raise NameError(
        "Notebook 10 Block 5 prerequisites are not initialised. "
        f"Missing: {BLOCK_5_MISSING_PREREQUISITES}"
    )


BLOCK_5_PREREQUISITES_VALID = all(
    [
        NOTEBOOK_10_BLOCK_1_COMPLETE,
        NOTEBOOK_10_BLOCK_1_VALID,
        NOTEBOOK_10_BLOCK_2_COMPLETE,
        NOTEBOOK_10_BLOCK_2_VALID,
        NOTEBOOK_10_BLOCK_3_COMPLETE,
        NOTEBOOK_10_BLOCK_3_VALID,
        NOTEBOOK_10_BLOCK_4_COMPLETE,
        NOTEBOOK_10_BLOCK_4_VALID,
        isinstance(
            NOTEBOOK_10_EXTENDED_ARTICLE_RECORDS,
            dict,
        ),
        isinstance(
            NOTEBOOK_10_EXTENDED_CANONICAL_SENTENCE_RECORDS,
            list,
        ),
    ]
)


if not BLOCK_5_PREREQUISITES_VALID:

    raise RuntimeError(
        "Notebook 10 Blocks 1–4 must be valid and complete "
        "before formal multi-article corpus validation."
    )


# =============================================================================
# Execution boundary
# =============================================================================

NOTEBOOK_10_BLOCK_5_MEDIA_AI_REPRESENTATION_FORWARD_EXECUTED = False
NOTEBOOK_10_BLOCK_5_TRANSFORMATIVE_FORWARD_EXECUTED = False
NOTEBOOK_10_BLOCK_5_WEIGHT_FORWARD_EXECUTED = False
NOTEBOOK_10_BLOCK_5_RECURRENT_STATE_CREATED = False
NOTEBOOK_10_BLOCK_5_RECURRENT_UPDATE_EXECUTED = False
NOTEBOOK_10_BLOCK_5_LOSS_CALCULATED = False
NOTEBOOK_10_BLOCK_5_OPTIMIZER_CREATED = False
NOTEBOOK_10_BLOCK_5_BACKWARD_PASS_EXECUTED = False
NOTEBOOK_10_BLOCK_5_PARAMETER_UPDATE_EXECUTED = False
NOTEBOOK_10_BLOCK_5_TRAINING_EXECUTED = False


# =============================================================================
# Exact nested-object / model-state helpers
# =============================================================================

def block_5_objects_exact(
    object_a,
    object_b,
):

    if (
        isinstance(object_a, np.ndarray)
        or
        isinstance(object_b, np.ndarray)
    ):

        if not (
            isinstance(object_a, np.ndarray)
            and
            isinstance(object_b, np.ndarray)
        ):

            return False

        return all(
            [
                object_a.dtype == object_b.dtype,
                object_a.shape == object_b.shape,
                bool(
                    np.array_equal(
                        object_a,
                        object_b,
                        equal_nan=True,
                    )
                ),
            ]
        )


    if (
        torch.is_tensor(object_a)
        or
        torch.is_tensor(object_b)
    ):

        if not (
            torch.is_tensor(object_a)
            and
            torch.is_tensor(object_b)
        ):

            return False

        return all(
            [
                object_a.dtype == object_b.dtype,
                tuple(object_a.shape)
                ==
                tuple(object_b.shape),
                bool(
                    torch.equal(
                        object_a.detach().cpu(),
                        object_b.detach().cpu(),
                    )
                ),
            ]
        )


    if (
        isinstance(object_a, dict)
        or
        isinstance(object_b, dict)
    ):

        if not (
            isinstance(object_a, dict)
            and
            isinstance(object_b, dict)
        ):

            return False

        if set(object_a.keys()) != set(object_b.keys()):

            return False

        return all(
            block_5_objects_exact(
                object_a[key],
                object_b[key],
            )

            for key
            in object_a
        )


    if (
        isinstance(object_a, (list, tuple))
        or
        isinstance(object_b, (list, tuple))
    ):

        if not (
            isinstance(object_a, (list, tuple))
            and
            isinstance(object_b, (list, tuple))
        ):

            return False

        if len(object_a) != len(object_b):

            return False

        return all(
            block_5_objects_exact(
                value_a,
                value_b,
            )

            for value_a, value_b
            in zip(object_a, object_b)
        )


    if isinstance(object_a, np.generic):

        object_a = object_a.item()


    if isinstance(object_b, np.generic):

        object_b = object_b.item()


    try:

        comparison = object_a == object_b

        if isinstance(
            comparison,
            (bool, np.bool_),
        ):

            return bool(comparison)

        return False

    except Exception:

        return False


def block_5_snapshot_module_collection(
    module_collection,
):

    return {
        module_name: {
            tensor_name:
                tensor.detach()
                .cpu()
                .clone()

            for tensor_name, tensor
            in module.state_dict().items()
        }

        for module_name, module
        in module_collection.items()
    }


def block_5_nested_state_exact(
    state_before,
    state_after,
):

    if state_before.keys() != state_after.keys():

        return False


    for module_name in state_before:

        if (
            state_before[module_name].keys()
            !=
            state_after[module_name].keys()
        ):

            return False


        for tensor_name in state_before[module_name]:

            if not torch.equal(
                state_before[module_name][tensor_name],
                state_after[module_name][tensor_name],
            ):

                return False


    return True


# =============================================================================
# Immutable inherited model-state baseline
# =============================================================================
#
# As in corrected Blocks 2–4, Block 5 validates Transformative-Weight
# immutability against the exact Notebook 10 runtime state rather than against
# raw Notebook 09 state-dict implementation keys (`alpha_active` vs `alpha`).
# =============================================================================

BLOCK_5_REPRESENTATION_STATE_BASELINE = (
    block_5_snapshot_module_collection(
        NOTEBOOK_10_REPRESENTATION_MODULES
    )
)


BLOCK_5_TRANSFORMATIVE_STATE_BASELINE = (
    block_5_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_MODULES
    )
)


BLOCK_5_WEIGHT_STATE_BASELINE = (
    block_5_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES
    )
)


# =============================================================================
# Preserve the complete Block 4 corpus exactly
# =============================================================================

BLOCK_5_SOURCE_ARTICLE_RECORDS = deepcopy(
    NOTEBOOK_10_EXTENDED_ARTICLE_RECORDS
)


BLOCK_5_SOURCE_CANONICAL_SENTENCE_RECORDS = deepcopy(
    NOTEBOOK_10_EXTENDED_CANONICAL_SENTENCE_RECORDS
)


BLOCK_5_SOURCE_ARTICLE_COUNT = len(
    BLOCK_5_SOURCE_ARTICLE_RECORDS
)


BLOCK_5_SOURCE_SENTENCE_COUNT = len(
    BLOCK_5_SOURCE_CANONICAL_SENTENCE_RECORDS
)


# =============================================================================
# Canonical record grouping
# =============================================================================

BLOCK_5_RECORDS_BY_ARTICLE = defaultdict(list)


for record in BLOCK_5_SOURCE_CANONICAL_SENTENCE_RECORDS:

    if isinstance(record, dict):

        BLOCK_5_RECORDS_BY_ARTICLE[
            record.get("article_id")
        ].append(
            record
        )


# =============================================================================
# Article identity validation
# =============================================================================

BLOCK_5_ARTICLE_IDS = list(
    BLOCK_5_SOURCE_ARTICLE_RECORDS.keys()
)


BLOCK_5_ARTICLE_IDS_NON_NULL = all(
    article_id is not None
    and
    str(article_id).strip() != ""

    for article_id
    in BLOCK_5_ARTICLE_IDS
)


BLOCK_5_ARTICLE_IDS_UNIQUE = (
    len(BLOCK_5_ARTICLE_IDS)
    ==
    len(set(BLOCK_5_ARTICLE_IDS))
)


BLOCK_5_ARTICLE_MEMBERSHIP_VALID = True


for article_id, records in BLOCK_5_SOURCE_ARTICLE_RECORDS.items():

    if not isinstance(records, list) or not records:

        BLOCK_5_ARTICLE_MEMBERSHIP_VALID = False

        continue


    if not all(
        isinstance(record, dict)
        and
        record.get("article_id")
        ==
        article_id

        for record
        in records
    ):

        BLOCK_5_ARTICLE_MEMBERSHIP_VALID = False


BLOCK_5_CANONICAL_ARTICLE_SET_VALID = (
    set(
        BLOCK_5_RECORDS_BY_ARTICLE.keys()
    )
    ==
    set(
        BLOCK_5_ARTICLE_IDS
    )
)


BLOCK_5_ARTICLE_IDENTITY_VALID = all(
    [
        BLOCK_5_ARTICLE_IDS_NON_NULL,
        BLOCK_5_ARTICLE_IDS_UNIQUE,
        BLOCK_5_ARTICLE_MEMBERSHIP_VALID,
        BLOCK_5_CANONICAL_ARTICLE_SET_VALID,
    ]
)


# =============================================================================
# Sentence identity validation
# =============================================================================

BLOCK_5_SENTENCE_IDS = [
    record.get("sentence_id")

    for record
    in BLOCK_5_SOURCE_CANONICAL_SENTENCE_RECORDS
]


BLOCK_5_SENTENCE_IDS_NON_NULL = all(
    sentence_id is not None
    and
    str(sentence_id).strip() != ""

    for sentence_id
    in BLOCK_5_SENTENCE_IDS
)


BLOCK_5_SENTENCE_IDS_UNIQUE = (
    len(BLOCK_5_SENTENCE_IDS)
    ==
    len(set(BLOCK_5_SENTENCE_IDS))
)


BLOCK_5_SENTENCE_MEMBERSHIP_VALID = all(
    record.get("article_id")
    in
    BLOCK_5_SOURCE_ARTICLE_RECORDS

    for record
    in BLOCK_5_SOURCE_CANONICAL_SENTENCE_RECORDS
)


BLOCK_5_SENTENCE_IDENTITY_VALID = all(
    [
        BLOCK_5_SENTENCE_IDS_NON_NULL,
        BLOCK_5_SENTENCE_IDS_UNIQUE,
        BLOCK_5_SENTENCE_MEMBERSHIP_VALID,
    ]
)


# =============================================================================
# Paragraph-structure audit
# =============================================================================

BLOCK_5_PARAGRAPH_STRUCTURE_AVAILABLE_COUNT = 0
BLOCK_5_SINGLE_PARAGRAPH_EXPLICIT_COUNT = 0
BLOCK_5_PARAGRAPH_STRUCTURE_VALID = True
BLOCK_5_ARTICLE_PARAGRAPH_AUDIT = {}


for article_id in BLOCK_5_ARTICLE_IDS:

    records = sorted(
        BLOCK_5_RECORDS_BY_ARTICLE[article_id],
        key=lambda record:
            (
                int(
                    record.get(
                        "canonical_sequence_position",
                        record.get(
                            "sequence_position",
                            -1,
                        ),
                    )
                ),
                str(
                    record.get(
                        "sentence_id",
                        "",
                    )
                ),
            ),
    )


    paragraph_ids = [
        record.get("paragraph_id")

        for record
        in records
    ]


    paragraph_positions = [
        record.get("paragraph_position")

        for record
        in records
    ]


    has_paragraph_ids = all(
        paragraph_id is not None
        and
        str(paragraph_id).strip() != ""

        for paragraph_id
        in paragraph_ids
    )


    available_positions = [
        position

        for position
        in paragraph_positions

        if isinstance(
            position,
            (int, np.integer),
        )
    ]


    positions_non_negative = all(
        int(position) >= 0

        for position
        in available_positions
    )


    if has_paragraph_ids:

        BLOCK_5_PARAGRAPH_STRUCTURE_AVAILABLE_COUNT += 1


    unique_paragraph_ids = {
        paragraph_id

        for paragraph_id
        in paragraph_ids

        if paragraph_id is not None
    }


    single_paragraph = (
        len(unique_paragraph_ids)
        <=
        1
    )


    if single_paragraph:

        BLOCK_5_SINGLE_PARAGRAPH_EXPLICIT_COUNT += 1


    article_paragraph_valid = all(
        [
            len(records) > 0,
            positions_non_negative,
            (
                has_paragraph_ids
                or
                single_paragraph
            ),
        ]
    )


    BLOCK_5_ARTICLE_PARAGRAPH_AUDIT[
        article_id
    ] = {
        "sentence_count":
            len(records),

        "paragraph_ids_available":
            has_paragraph_ids,

        "paragraph_count":
            len(unique_paragraph_ids),

        "single_paragraph_representation":
            single_paragraph,

        "paragraph_positions_non_negative":
            positions_non_negative,

        "valid":
            article_paragraph_valid,
    }


    if not article_paragraph_valid:

        BLOCK_5_PARAGRAPH_STRUCTURE_VALID = False


# =============================================================================
# Canonical sequence validation
# =============================================================================

BLOCK_5_CANONICAL_SEQUENCE_VALID = True
BLOCK_5_ARTICLE_SEQUENCE_AUDIT = {}


for article_id in BLOCK_5_ARTICLE_IDS:

    records = list(
        BLOCK_5_RECORDS_BY_ARTICLE[
            article_id
        ]
    )


    canonical_positions = [
        record.get(
            "canonical_sequence_position",
            record.get(
                "sequence_position"
            ),
        )

        for record
        in records
    ]


    positions_integer = all(
        isinstance(
            position,
            (int, np.integer),
        )

        for position
        in canonical_positions
    )


    if positions_integer:

        integer_positions = [
            int(position)

            for position
            in canonical_positions
        ]

    else:

        integer_positions = []


    positions_non_negative = (
        positions_integer
        and
        all(
            position >= 0

            for position
            in integer_positions
        )
    )


    positions_unique = (
        positions_integer
        and
        len(integer_positions)
        ==
        len(set(integer_positions))
    )


    expected_positions = list(
        range(
            len(records)
        )
    )


    complete_zero_based = (
        positions_integer
        and
        sorted(integer_positions)
        ==
        expected_positions
    )


    deterministic_order = False


    if complete_zero_based:

        ordered_records = sorted(
            records,
            key=lambda record:
                (
                    int(
                        record.get(
                            "canonical_sequence_position",
                            record.get(
                                "sequence_position",
                            ),
                        )
                    ),
                    str(
                        record.get(
                            "sentence_id",
                            "",
                        )
                    ),
                ),
        )


        deterministic_order = (
            [
                int(
                    record.get(
                        "canonical_sequence_position",
                        record.get(
                            "sequence_position",
                        ),
                    )
                )

                for record
                in ordered_records
            ]
            ==
            expected_positions
        )


    article_sequence_valid = all(
        [
            len(records) > 0,
            positions_integer,
            positions_non_negative,
            positions_unique,
            complete_zero_based,
            deterministic_order,
        ]
    )


    BLOCK_5_ARTICLE_SEQUENCE_AUDIT[
        article_id
    ] = {
        "sentence_count":
            len(records),

        "positions_integer":
            positions_integer,

        "positions_non_negative":
            positions_non_negative,

        "positions_unique":
            positions_unique,

        "complete_zero_based":
            complete_zero_based,

        "deterministic_order":
            deterministic_order,

        "valid":
            article_sequence_valid,
    }


    if not article_sequence_valid:

        BLOCK_5_CANONICAL_SEQUENCE_VALID = False


# =============================================================================
# Article-boundary independence contract
# =============================================================================

BLOCK_5_ARTICLE_BOUNDARIES_EXPLICIT = all(
    len(
        BLOCK_5_RECORDS_BY_ARTICLE[
            article_id
        ]
    )
    >
    0

    for article_id
    in BLOCK_5_ARTICLE_IDS
)


BLOCK_5_ARTICLE_BOUNDARY_RESET_PERMITTED = all(
    [
        BLOCK_5_ARTICLE_BOUNDARIES_EXPLICIT,
        BLOCK_5_ARTICLE_IDENTITY_VALID,
        BLOCK_5_CANONICAL_SEQUENCE_VALID,
    ]
)


BLOCK_5_CROSS_ARTICLE_PROPAGATION_REQUIRED = False


BLOCK_5_ARTICLE_BOUNDARY_VALID = all(
    [
        BLOCK_5_ARTICLE_BOUNDARY_RESET_PERMITTED,
        not BLOCK_5_CROSS_ARTICLE_PROPAGATION_REQUIRED,
    ]
)


# =============================================================================
# Contextual representation and encoder audit
# =============================================================================

BLOCK_5_ARTICLE_REPRESENTATION_AUDIT = {}
BLOCK_5_REPRESENTATION_READY_ARTICLES = {}
BLOCK_5_REPRESENTATION_READY_SENTENCE_COUNT = 0

BLOCK_5_ALL_READY_REPRESENTATIONS_DIMENSION_VALID = True
BLOCK_5_ALL_READY_REPRESENTATIONS_FINITE = True
BLOCK_5_ALL_READY_ENCODERS_COMPATIBLE = True
BLOCK_5_ALL_READY_NORMALISATION_VALID = True
BLOCK_5_ALL_READY_ALIGNMENT_VALID = True


for article_id in BLOCK_5_ARTICLE_IDS:

    records = sorted(
        BLOCK_5_RECORDS_BY_ARTICLE[
            article_id
        ],
        key=lambda record:
            int(
                record.get(
                    "canonical_sequence_position",
                    record.get(
                        "sequence_position",
                        -1,
                    ),
                )
            ),
    )


    embeddings = []
    dimensions_valid = True
    finite = True
    encoder_compatible = True
    normalisation_valid = True
    alignment_valid = True
    missing_representation = False


    for expected_position, record in enumerate(records):

        observed_position = record.get(
            "canonical_sequence_position",
            record.get(
                "sequence_position"
            ),
        )


        if (
            not isinstance(
                observed_position,
                (int, np.integer),
            )
            or
            int(observed_position)
            !=
            expected_position
        ):

            alignment_valid = False


        embedding = record.get(
            "contextual_representation"
        )


        if embedding is None:

            missing_representation = True
            dimensions_valid = False
            finite = False
            encoder_compatible = False
            normalisation_valid = False

            continue


        if torch.is_tensor(embedding):

            embedding_array = (
                embedding.detach()
                .cpu()
                .numpy()
            )

        else:

            try:

                embedding_array = np.asarray(
                    embedding,
                    dtype=np.float32,
                )

            except Exception:

                dimensions_valid = False
                finite = False
                encoder_compatible = False
                normalisation_valid = False

                continue


        if (
            embedding_array.shape
            !=
            (
                BLOCK_5_EXPECTED_EMBEDDING_DIMENSION,
            )
        ):

            dimensions_valid = False


        if not bool(
            np.isfinite(
                embedding_array
            ).all()
        ):

            finite = False


        provider = record.get(
            "representation_provider"
        )


        model = record.get(
            "representation_model"
        )


        dimension_metadata = record.get(
            "representation_dimension"
        )


        normalised_metadata = record.get(
            "representation_normalised"
        )


        # Existing Notebook 05 records may carry encoder provenance through
        # their inherited validated corpus contract rather than repeating every
        # field on every sentence. Prefer explicit record metadata when present;
        # otherwise use the established Notebook 10/Notebook 05 contract.
        if provider is not None:

            if provider != BLOCK_5_EXPECTED_ENCODER_PROVIDER:

                encoder_compatible = False


        if model is not None:

            if model != BLOCK_5_EXPECTED_ENCODER_MODEL:

                encoder_compatible = False


        if dimension_metadata is not None:

            try:

                if (
                    int(dimension_metadata)
                    !=
                    BLOCK_5_EXPECTED_EMBEDDING_DIMENSION
                ):

                    encoder_compatible = False

            except Exception:

                encoder_compatible = False


        if normalised_metadata is not None:

            if normalised_metadata is not True:

                encoder_compatible = False


        if (
            embedding_array.shape
            ==
            (
                BLOCK_5_EXPECTED_EMBEDDING_DIMENSION,
            )
            and
            bool(
                np.isfinite(
                    embedding_array
                ).all()
            )
        ):

            norm = float(
                np.linalg.norm(
                    embedding_array
                )
            )


            if not bool(
                np.isclose(
                    norm,
                    1.0,
                    atol=
                        BLOCK_5_NORMALISATION_ATOL,
                    rtol=
                        BLOCK_5_NORMALISATION_RTOL,
                )
            ):

                normalisation_valid = False


        embeddings.append(
            embedding_array.copy()
        )


    complete_representation = all(
        [
            len(records) > 0,
            not missing_representation,
            len(embeddings) == len(records),
            dimensions_valid,
            finite,
            encoder_compatible,
            normalisation_valid,
            alignment_valid,
        ]
    )


    if complete_representation:

        matrix = np.stack(
            embeddings,
            axis=0,
        ).astype(
            np.float32,
            copy=False,
        )


        matrix_shape_valid = (
            matrix.shape
            ==
            (
                len(records),
                BLOCK_5_EXPECTED_EMBEDDING_DIMENSION,
            )
        )


        matrix_finite = bool(
            np.isfinite(
                matrix
            ).all()
        )

    else:

        matrix = None
        matrix_shape_valid = False
        matrix_finite = False


    representation_ready = all(
        [
            complete_representation,
            matrix_shape_valid,
            matrix_finite,
        ]
    )


    BLOCK_5_ARTICLE_REPRESENTATION_AUDIT[
        article_id
    ] = {
        "sentence_count":
            len(records),

        "complete_representation":
            complete_representation,

        "dimension_valid":
            dimensions_valid,

        "finite":
            finite,

        "encoder_compatible":
            encoder_compatible,

        "normalisation_valid":
            normalisation_valid,

        "alignment_valid":
            alignment_valid,

        "matrix_shape":
            (
                tuple(matrix.shape)
                if matrix is not None
                else None
            ),

        "matrix_shape_valid":
            matrix_shape_valid,

        "matrix_finite":
            matrix_finite,

        "representation_ready":
            representation_ready,
    }


    if representation_ready:

        BLOCK_5_REPRESENTATION_READY_ARTICLES[
            article_id
        ] = {
            "records":
                deepcopy(records),

            "matrix":
                matrix.copy(),
        }


        BLOCK_5_REPRESENTATION_READY_SENTENCE_COUNT += (
            len(records)
        )


    if complete_representation:

        BLOCK_5_ALL_READY_REPRESENTATIONS_DIMENSION_VALID &= (
            dimensions_valid
        )

        BLOCK_5_ALL_READY_REPRESENTATIONS_FINITE &= (
            finite
        )

        BLOCK_5_ALL_READY_ENCODERS_COMPATIBLE &= (
            encoder_compatible
        )

        BLOCK_5_ALL_READY_NORMALISATION_VALID &= (
            normalisation_valid
        )

        BLOCK_5_ALL_READY_ALIGNMENT_VALID &= (
            alignment_valid
        )


BLOCK_5_REPRESENTATION_READY_ARTICLE_COUNT = len(
    BLOCK_5_REPRESENTATION_READY_ARTICLES
)


# =============================================================================
# Pilot / pre-extension corpus preservation
# =============================================================================
#
# Block 4 explicitly preserved the incoming Block 3 corpus in
# BLOCK_4_EXISTING_ARTICLE_RECORDS_BEFORE and then extended it.  This is the
# authoritative pre-extension baseline available to Block 5.
# =============================================================================

BLOCK_5_PILOT_BASELINE_AVAILABLE = (
    "BLOCK_4_EXISTING_ARTICLE_RECORDS_BEFORE"
    in globals()
)


if BLOCK_5_PILOT_BASELINE_AVAILABLE:

    BLOCK_5_PILOT_BASELINE = deepcopy(
        BLOCK_4_EXISTING_ARTICLE_RECORDS_BEFORE
    )


    BLOCK_5_PILOT_ARTICLE_IDS = list(
        BLOCK_5_PILOT_BASELINE.keys()
    )


    BLOCK_5_PILOT_ARTICLE_MEMBERSHIP_PRESERVED = all(
        article_id
        in
        BLOCK_5_SOURCE_ARTICLE_RECORDS

        for article_id
        in BLOCK_5_PILOT_ARTICLE_IDS
    )


    BLOCK_5_PILOT_RECORDS_EXACT = all(
        block_5_objects_exact(
            BLOCK_5_PILOT_BASELINE[
                article_id
            ],
            BLOCK_5_SOURCE_ARTICLE_RECORDS[
                article_id
            ],
        )

        for article_id
        in BLOCK_5_PILOT_ARTICLE_IDS

        if article_id
        in
        BLOCK_5_SOURCE_ARTICLE_RECORDS
    )


    BLOCK_5_PILOT_PRESERVED = all(
        [
            BLOCK_5_PILOT_ARTICLE_MEMBERSHIP_PRESERVED,
            BLOCK_5_PILOT_RECORDS_EXACT,
        ]
    )

else:

    # Blocks 2–4 already validated corpus preservation.  Do not fabricate a
    # comparison baseline if an earlier runtime has been cleared.
    BLOCK_5_PILOT_ARTICLE_IDS = []
    BLOCK_5_PILOT_ARTICLE_MEMBERSHIP_PRESERVED = bool(
        globals().get(
            "BLOCK_4_EXISTING_CORPUS_PRESERVED",
            False,
        )
    )
    BLOCK_5_PILOT_RECORDS_EXACT = (
        BLOCK_5_PILOT_ARTICLE_MEMBERSHIP_PRESERVED
    )
    BLOCK_5_PILOT_PRESERVED = (
        BLOCK_5_PILOT_ARTICLE_MEMBERSHIP_PRESERVED
    )


# =============================================================================
# Provenance audit
# =============================================================================

BLOCK_5_ARTICLE_PROVENANCE_AUDIT = {}
BLOCK_5_PROVENANCE_VALID = True


for article_id in BLOCK_5_ARTICLE_IDS:

    records = BLOCK_5_RECORDS_BY_ARTICLE[
        article_id
    ]


    provenance_payloads = [
        record.get(
            "provenance"
        )

        for record
        in records

        if isinstance(
            record.get(
                "provenance"
            ),
            dict,
        )
    ]


    source_types = {
        record.get(
            "source_type"
        )

        for record
        in records

        if record.get(
            "source_type"
        )
        is not None
    }


    # Existing validated project records can inherit provenance from upstream
    # artefacts without duplicating a full provenance dictionary per sentence.
    # New external records must carry explicit provenance.
    external_article = (
        "external_article"
        in
        source_types
    )


    if external_article:

        article_provenance_valid = (
            len(provenance_payloads)
            ==
            len(records)
            and
            all(
                provenance.get(
                    "source"
                )
                is not None

                for provenance
                in provenance_payloads
            )
        )

    else:

        article_provenance_valid = (
            len(records)
            >
            0
        )


    BLOCK_5_ARTICLE_PROVENANCE_AUDIT[
        article_id
    ] = {
        "source_types":
            sorted(
                str(source_type)

                for source_type
                in source_types
            ),

        "explicit_provenance_records":
            len(
                provenance_payloads
            ),

        "external_article":
            external_article,

        "valid":
            article_provenance_valid,
    }


    if not article_provenance_valid:

        BLOCK_5_PROVENANCE_VALID = False


# =============================================================================
# Synthetic-fallback audit
# =============================================================================

BLOCK_5_SYNTHETIC_ARTICLE_CONSTRUCTION = bool(
    globals().get(
        "NOTEBOOK_10_BLOCK_4_SYNTHETIC_TEXT_CREATED",
        False,
    )
)


BLOCK_5_PILOT_ARTICLE_SPLIT = bool(
    globals().get(
        "NOTEBOOK_10_BLOCK_4_PILOT_ARTICLE_SPLIT",
        False,
    )
)


BLOCK_5_PILOT_ARTICLE_DUPLICATED = bool(
    globals().get(
        "NOTEBOOK_10_BLOCK_4_PILOT_ARTICLE_DUPLICATED",
        False,
    )
)


BLOCK_5_SUPERVISION_USED_AS_TEXT = bool(
    globals().get(
        "NOTEBOOK_10_BLOCK_4_SUPERVISION_USED_AS_TEXT",
        False,
    )
)


BLOCK_5_SUPERVISION_USED_AS_REPRESENTATION = bool(
    globals().get(
        "NOTEBOOK_10_BLOCK_4_SUPERVISION_USED_AS_REPRESENTATION",
        False,
    )
)


BLOCK_5_SYNTHETIC_FALLBACK_USED = any(
    [
        BLOCK_5_SYNTHETIC_ARTICLE_CONSTRUCTION,
        BLOCK_5_PILOT_ARTICLE_SPLIT,
        BLOCK_5_PILOT_ARTICLE_DUPLICATED,
        BLOCK_5_SUPERVISION_USED_AS_TEXT,
        BLOCK_5_SUPERVISION_USED_AS_REPRESENTATION,
    ]
)


BLOCK_5_SYNTHETIC_FALLBACK_VALID = (
    not
    BLOCK_5_SYNTHETIC_FALLBACK_USED
)


# =============================================================================
# Supervision-status inventory
# =============================================================================

BLOCK_5_SUPERVISION_STATUS_COUNTER = Counter()


for record in BLOCK_5_SOURCE_CANONICAL_SENTENCE_RECORDS:

    supervision_status = record.get(
        "supervision_status"
    )


    if supervision_status is None:

        supervision_status = (
            "available_in_separate_validated_project_artefacts"
        )


    if (
        supervision_status
        not in
        BLOCK_5_ALLOWED_SUPERVISION_STATUSES
    ):

        supervision_status = (
            "alignment_not_yet_assessed"
        )


    BLOCK_5_SUPERVISION_STATUS_COUNTER[
        supervision_status
    ] += 1


BLOCK_5_SUPERVISION_STATUS_INVENTORY = {
    status:
        int(
            BLOCK_5_SUPERVISION_STATUS_COUNTER.get(
                status,
                0,
            )
        )

    for status
    in sorted(
        BLOCK_5_ALLOWED_SUPERVISION_STATUSES
    )
}


# =============================================================================
# Structural validity per article
# =============================================================================

BLOCK_5_STRUCTURALLY_VALID_ARTICLES = {}


for article_id in BLOCK_5_ARTICLE_IDS:

    structurally_valid = all(
        [
            BLOCK_5_ARTICLE_IDENTITY_VALID,
            BLOCK_5_SENTENCE_IDENTITY_VALID,
            BLOCK_5_ARTICLE_PARAGRAPH_AUDIT[
                article_id
            ][
                "valid"
            ],
            BLOCK_5_ARTICLE_SEQUENCE_AUDIT[
                article_id
            ][
                "valid"
            ],
            len(
                BLOCK_5_RECORDS_BY_ARTICLE[
                    article_id
                ]
            )
            >
            0,
            BLOCK_5_ARTICLE_PROVENANCE_AUDIT[
                article_id
            ][
                "valid"
            ],
        ]
    )


    if structurally_valid:

        BLOCK_5_STRUCTURALLY_VALID_ARTICLES[
            article_id
        ] = deepcopy(
            BLOCK_5_RECORDS_BY_ARTICLE[
                article_id
            ]
        )


BLOCK_5_STRUCTURALLY_VALID_ARTICLE_COUNT = len(
    BLOCK_5_STRUCTURALLY_VALID_ARTICLES
)


# =============================================================================
# Formal readiness classification
# =============================================================================

BLOCK_5_STRUCTURAL_MULTI_ARTICLE_READY = all(
    [
        BLOCK_5_SOURCE_ARTICLE_COUNT
        >=
        BLOCK_5_MINIMUM_MULTI_ARTICLE_COUNT,

        BLOCK_5_STRUCTURALLY_VALID_ARTICLE_COUNT
        >=
        BLOCK_5_MINIMUM_MULTI_ARTICLE_COUNT,

        BLOCK_5_ARTICLE_IDENTITY_VALID,
        BLOCK_5_SENTENCE_IDENTITY_VALID,
        BLOCK_5_PARAGRAPH_STRUCTURE_VALID,
        BLOCK_5_CANONICAL_SEQUENCE_VALID,
        BLOCK_5_ARTICLE_BOUNDARY_VALID,
        BLOCK_5_PILOT_PRESERVED,
        BLOCK_5_PROVENANCE_VALID,
        BLOCK_5_SYNTHETIC_FALLBACK_VALID,
    ]
)


BLOCK_5_REPRESENTATION_MULTI_ARTICLE_READY = (
    BLOCK_5_REPRESENTATION_READY_ARTICLE_COUNT
    >=
    BLOCK_5_MINIMUM_MULTI_ARTICLE_COUNT
)


BLOCK_5_MULTI_ARTICLE_EXECUTION_READY = all(
    [
        BLOCK_5_STRUCTURAL_MULTI_ARTICLE_READY,
        BLOCK_5_REPRESENTATION_MULTI_ARTICLE_READY,
    ]
)


# This remains false by methodological contract.  Multi-article architectural
# validation is not evidence of expanded-training sufficiency.
BLOCK_5_EXPANDED_TRAINING_SUFFICIENT = False


# =============================================================================
# Model-state protection validation
# =============================================================================

BLOCK_5_REPRESENTATION_STATE_AFTER = (
    block_5_snapshot_module_collection(
        NOTEBOOK_10_REPRESENTATION_MODULES
    )
)


BLOCK_5_TRANSFORMATIVE_STATE_AFTER = (
    block_5_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_MODULES
    )
)


BLOCK_5_WEIGHT_STATE_AFTER = (
    block_5_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES
    )
)


BLOCK_5_REPRESENTATION_STATE_UNCHANGED = (
    block_5_nested_state_exact(
        BLOCK_5_REPRESENTATION_STATE_BASELINE,
        BLOCK_5_REPRESENTATION_STATE_AFTER,
    )
)


BLOCK_5_TRANSFORMATIVE_STATE_UNCHANGED = (
    block_5_nested_state_exact(
        BLOCK_5_TRANSFORMATIVE_STATE_BASELINE,
        BLOCK_5_TRANSFORMATIVE_STATE_AFTER,
    )
)


BLOCK_5_WEIGHT_STATE_UNCHANGED = (
    block_5_nested_state_exact(
        BLOCK_5_WEIGHT_STATE_BASELINE,
        BLOCK_5_WEIGHT_STATE_AFTER,
    )
)


with torch.no_grad():

    BLOCK_5_EFFECTIVE_WEIGHT_IDENTITY = all(
        torch.equal(
            NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES[
                pathway_name
            ]()
            .detach()
            .cpu(),

            NOTEBOOK_10_INHERITED_EFFECTIVE_TRANSFORMATIVE_WEIGHTS[
                pathway_name
            ]
            .detach()
            .cpu(),
        )

        for pathway_name
        in NOTEBOOK_10_INHERITED_EFFECTIVE_TRANSFORMATIVE_WEIGHTS
    )


BLOCK_5_ALL_PARAMETERS_FROZEN = all(
    not parameter.requires_grad

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


BLOCK_5_ALL_MODULES_EVAL = all(
    not module.training

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()
)


BLOCK_5_ALL_GRADIENTS_ABSENT = all(
    parameter.grad is None

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


# =============================================================================
# Parameter accounting
# =============================================================================

BLOCK_5_REPRESENTATION_PARAMETER_COUNT = sum(
    parameter.numel()

    for module
    in NOTEBOOK_10_REPRESENTATION_MODULES.values()

    for parameter
    in module.parameters()
)


BLOCK_5_TRANSFORMATIVE_PARAMETER_COUNT = sum(
    parameter.numel()

    for module
    in NOTEBOOK_10_TRANSFORMATIVE_MODULES.values()

    for parameter
    in module.parameters()
)


BLOCK_5_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT = sum(
    parameter.numel()

    for module
    in NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES.values()

    for parameter
    in module.parameters()
)


BLOCK_5_TOTAL_PARAMETER_COUNT = (
    BLOCK_5_REPRESENTATION_PARAMETER_COUNT
    +
    BLOCK_5_TRANSFORMATIVE_PARAMETER_COUNT
    +
    BLOCK_5_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT
)


BLOCK_5_PARAMETER_ACCOUNTING_VALID = all(
    [
        BLOCK_5_REPRESENTATION_PARAMETER_COUNT
        ==
        894003,

        BLOCK_5_TRANSFORMATIVE_PARAMETER_COUNT
        ==
        8187,

        BLOCK_5_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT
        ==
        49,

        BLOCK_5_TOTAL_PARAMETER_COUNT
        ==
        902239,
    ]
)


# =============================================================================
# Execution-boundary validation
# =============================================================================

BLOCK_5_EXECUTION_BOUNDARY_VALID = all(
    [
        not NOTEBOOK_10_BLOCK_5_MEDIA_AI_REPRESENTATION_FORWARD_EXECUTED,
        not NOTEBOOK_10_BLOCK_5_TRANSFORMATIVE_FORWARD_EXECUTED,
        not NOTEBOOK_10_BLOCK_5_WEIGHT_FORWARD_EXECUTED,
        not NOTEBOOK_10_BLOCK_5_RECURRENT_STATE_CREATED,
        not NOTEBOOK_10_BLOCK_5_RECURRENT_UPDATE_EXECUTED,
        not NOTEBOOK_10_BLOCK_5_LOSS_CALCULATED,
        not NOTEBOOK_10_BLOCK_5_OPTIMIZER_CREATED,
        not NOTEBOOK_10_BLOCK_5_BACKWARD_PASS_EXECUTED,
        not NOTEBOOK_10_BLOCK_5_PARAMETER_UPDATE_EXECUTED,
        not NOTEBOOK_10_BLOCK_5_TRAINING_EXECUTED,
    ]
)


# =============================================================================
# Final validation
# =============================================================================

NOTEBOOK_10_BLOCK_5_ERRORS = []


BLOCK_5_VALIDATION_CHECKS = {
    "article_identity_invalid":
        BLOCK_5_ARTICLE_IDENTITY_VALID,

    "sentence_identity_invalid":
        BLOCK_5_SENTENCE_IDENTITY_VALID,

    "paragraph_structure_invalid":
        BLOCK_5_PARAGRAPH_STRUCTURE_VALID,

    "canonical_sequence_invalid":
        BLOCK_5_CANONICAL_SEQUENCE_VALID,

    "article_boundary_invalid":
        BLOCK_5_ARTICLE_BOUNDARY_VALID,

    "pilot_corpus_not_preserved":
        BLOCK_5_PILOT_PRESERVED,

    "provenance_invalid":
        BLOCK_5_PROVENANCE_VALID,

    "synthetic_fallback_detected":
        BLOCK_5_SYNTHETIC_FALLBACK_VALID,

    "representation_state_changed":
        BLOCK_5_REPRESENTATION_STATE_UNCHANGED,

    "transformative_state_changed":
        BLOCK_5_TRANSFORMATIVE_STATE_UNCHANGED,

    "weight_state_changed":
        BLOCK_5_WEIGHT_STATE_UNCHANGED,

    "effective_weight_identity_changed":
        BLOCK_5_EFFECTIVE_WEIGHT_IDENTITY,

    "parameters_not_frozen":
        BLOCK_5_ALL_PARAMETERS_FROZEN,

    "modules_not_eval":
        BLOCK_5_ALL_MODULES_EVAL,

    "gradients_present":
        BLOCK_5_ALL_GRADIENTS_ABSENT,

    "parameter_accounting_invalid":
        BLOCK_5_PARAMETER_ACCOUNTING_VALID,

    "execution_boundary_invalid":
        BLOCK_5_EXECUTION_BOUNDARY_VALID,
}


for error_name, condition in BLOCK_5_VALIDATION_CHECKS.items():

    if not condition:

        NOTEBOOK_10_BLOCK_5_ERRORS.append(
            error_name
        )


# Block 5 is an audit.  A structurally valid corpus with fewer than two ready
# articles may complete successfully while execution readiness remains False.
NOTEBOOK_10_BLOCK_5_VALID = (
    len(
        NOTEBOOK_10_BLOCK_5_ERRORS
    )
    ==
    0
)


if not NOTEBOOK_10_BLOCK_5_VALID:

    raise RuntimeError(
        "Notebook 10 Block 5 formal multi-article corpus validation "
        "failed. "
        f"Errors: {NOTEBOOK_10_BLOCK_5_ERRORS}"
    )


NOTEBOOK_10_BLOCK_5_COMPLETE = True

NOTEBOOK_10_FORMAL_MULTI_ARTICLE_CORPUS_VALIDATED = True

NOTEBOOK_10_MULTI_ARTICLE_EXECUTION_READY = (
    BLOCK_5_MULTI_ARTICLE_EXECUTION_READY
)

NOTEBOOK_10_EXPANDED_TRAINING_SUFFICIENT = (
    BLOCK_5_EXPANDED_TRAINING_SUFFICIENT
)


# =============================================================================
# Canonical Block 5 handover objects for Block 6
# =============================================================================

NOTEBOOK_10_VALIDATED_MULTI_ARTICLE_RECORDS = {
    article_id:
        deepcopy(
            BLOCK_5_RECORDS_BY_ARTICLE[
                article_id
            ]
        )

    for article_id
    in BLOCK_5_ARTICLE_IDS
}


NOTEBOOK_10_VALIDATED_REPRESENTATION_MATRICES = {
    article_id:
        payload[
            "matrix"
        ].copy()

    for article_id, payload
    in BLOCK_5_REPRESENTATION_READY_ARTICLES.items()
}


NOTEBOOK_10_VALIDATED_ARTICLE_IDS = tuple(
    sorted(
        BLOCK_5_ARTICLE_IDS
    )
)


NOTEBOOK_10_REPRESENTATION_READY_ARTICLE_IDS = tuple(
    sorted(
        BLOCK_5_REPRESENTATION_READY_ARTICLES.keys()
    )
)


NOTEBOOK_10_BLOCK_5_SUMMARY = {
    "block":
        NOTEBOOK_10_BLOCK_5,

    "block_name":
        NOTEBOOK_10_BLOCK_5_NAME,

    "version":
        NOTEBOOK_10_BLOCK_5_VERSION,

    "article_count":
        BLOCK_5_SOURCE_ARTICLE_COUNT,

    "sentence_count":
        BLOCK_5_SOURCE_SENTENCE_COUNT,

    "structurally_valid_article_count":
        BLOCK_5_STRUCTURALLY_VALID_ARTICLE_COUNT,

    "representation_ready_article_count":
        BLOCK_5_REPRESENTATION_READY_ARTICLE_COUNT,

    "representation_ready_sentence_count":
        BLOCK_5_REPRESENTATION_READY_SENTENCE_COUNT,

    "article_identity_valid":
        BLOCK_5_ARTICLE_IDENTITY_VALID,

    "sentence_identity_valid":
        BLOCK_5_SENTENCE_IDENTITY_VALID,

    "paragraph_structure_valid":
        BLOCK_5_PARAGRAPH_STRUCTURE_VALID,

    "canonical_sequence_valid":
        BLOCK_5_CANONICAL_SEQUENCE_VALID,

    "article_boundary_valid":
        BLOCK_5_ARTICLE_BOUNDARY_VALID,

    "pilot_preserved":
        BLOCK_5_PILOT_PRESERVED,

    "provenance_valid":
        BLOCK_5_PROVENANCE_VALID,

    "synthetic_fallback_used":
        BLOCK_5_SYNTHETIC_FALLBACK_USED,

    "supervision_status_inventory":
        deepcopy(
            BLOCK_5_SUPERVISION_STATUS_INVENTORY
        ),

    "structural_multi_article_ready":
        BLOCK_5_STRUCTURAL_MULTI_ARTICLE_READY,

    "representation_multi_article_ready":
        BLOCK_5_REPRESENTATION_MULTI_ARTICLE_READY,

    "multi_article_execution_ready":
        BLOCK_5_MULTI_ARTICLE_EXECUTION_READY,

    "expanded_training_sufficient":
        BLOCK_5_EXPANDED_TRAINING_SUFFICIENT,

    "block_valid":
        NOTEBOOK_10_BLOCK_5_VALID,

    "block_complete":
        NOTEBOOK_10_BLOCK_5_COMPLETE,
}


# =============================================================================
# Report
# =============================================================================

print("=" * 72)

print(
    "Media AI — Notebook 10, Block 5: "
    "Formal Multi-Article Corpus Validation and Execution-Readiness Audit"
)

print("=" * 72)

print(
    f"Block version                : "
    f"{NOTEBOOK_10_BLOCK_5_VERSION}"
)

print("-" * 72)

print("Corpus inventory")

print(
    f"Total articles               : "
    f"{BLOCK_5_SOURCE_ARTICLE_COUNT}"
)

print(
    f"Total sentences              : "
    f"{BLOCK_5_SOURCE_SENTENCE_COUNT}"
)

print(
    f"Structurally valid articles  : "
    f"{BLOCK_5_STRUCTURALLY_VALID_ARTICLE_COUNT}"
)

print(
    f"Representation-ready articles: "
    f"{BLOCK_5_REPRESENTATION_READY_ARTICLE_COUNT}"
)

print(
    f"Representation-ready sentences: "
    f"{BLOCK_5_REPRESENTATION_READY_SENTENCE_COUNT}"
)

print("-" * 72)

print("Identity and structure")

print(
    f"Article identity valid       : "
    f"{BLOCK_5_ARTICLE_IDENTITY_VALID}"
)

print(
    f"Sentence identity valid      : "
    f"{BLOCK_5_SENTENCE_IDENTITY_VALID}"
)

print(
    f"Paragraph structure valid    : "
    f"{BLOCK_5_PARAGRAPH_STRUCTURE_VALID}"
)

print(
    f"Canonical sequence valid     : "
    f"{BLOCK_5_CANONICAL_SEQUENCE_VALID}"
)

print(
    f"Article boundaries explicit  : "
    f"{BLOCK_5_ARTICLE_BOUNDARIES_EXPLICIT}"
)

print(
    f"Article-boundary valid       : "
    f"{BLOCK_5_ARTICLE_BOUNDARY_VALID}"
)

print("-" * 72)

print("Contextual representation contract")

print(
    f"Encoder provider             : "
    f"{BLOCK_5_EXPECTED_ENCODER_PROVIDER}"
)

print(
    f"Encoder model                : "
    f"{BLOCK_5_EXPECTED_ENCODER_MODEL}"
)

print(
    f"Embedding dimension          : "
    f"{BLOCK_5_EXPECTED_EMBEDDING_DIMENSION}"
)

print(
    f"Representation dimension valid: "
    f"{BLOCK_5_ALL_READY_REPRESENTATIONS_DIMENSION_VALID}"
)

print(
    f"Representation values finite : "
    f"{BLOCK_5_ALL_READY_REPRESENTATIONS_FINITE}"
)

print(
    f"Encoder compatibility valid  : "
    f"{BLOCK_5_ALL_READY_ENCODERS_COMPATIBLE}"
)

print(
    f"Normalisation valid          : "
    f"{BLOCK_5_ALL_READY_NORMALISATION_VALID}"
)

print(
    f"Sentence alignment valid     : "
    f"{BLOCK_5_ALL_READY_ALIGNMENT_VALID}"
)

print("-" * 72)

print("Corpus integrity")

print(
    f"Pilot corpus preserved       : "
    f"{BLOCK_5_PILOT_PRESERVED}"
)

print(
    f"Provenance valid             : "
    f"{BLOCK_5_PROVENANCE_VALID}"
)

print(
    f"Synthetic fallback used      : "
    f"{BLOCK_5_SYNTHETIC_FALLBACK_USED}"
)

print("Supervision status inventory :")

for (
    supervision_status,
    supervision_count,
) in BLOCK_5_SUPERVISION_STATUS_INVENTORY.items():

    print(
        f"  {supervision_status:<42}: "
        f"{supervision_count}"
    )

print("-" * 72)

print("Readiness")

print(
    f"Minimum article count        : "
    f"{BLOCK_5_MINIMUM_MULTI_ARTICLE_COUNT}"
)

print(
    f"Structural multi-article     : "
    f"{BLOCK_5_STRUCTURAL_MULTI_ARTICLE_READY}"
)

print(
    f"Representation multi-article : "
    f"{BLOCK_5_REPRESENTATION_MULTI_ARTICLE_READY}"
)

print(
    f"Multi-article execution ready: "
    f"{BLOCK_5_MULTI_ARTICLE_EXECUTION_READY}"
)

print(
    f"Expanded-training sufficient : "
    f"{BLOCK_5_EXPANDED_TRAINING_SUFFICIENT}"
)

print("-" * 72)

print("Inherited Media AI state")

print(
    f"Representation parameters    : "
    f"{BLOCK_5_REPRESENTATION_PARAMETER_COUNT:,}"
)

print(
    f"Transformative parameters    : "
    f"{BLOCK_5_TRANSFORMATIVE_PARAMETER_COUNT:,}"
)

print(
    f"Transformative Weight params : "
    f"{BLOCK_5_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT:,}"
)

print(
    f"Total model parameters       : "
    f"{BLOCK_5_TOTAL_PARAMETER_COUNT:,}"
)

print(
    f"Representation unchanged     : "
    f"{BLOCK_5_REPRESENTATION_STATE_UNCHANGED}"
)

print(
    f"Transformative unchanged     : "
    f"{BLOCK_5_TRANSFORMATIVE_STATE_UNCHANGED}"
)

print(
    f"Transform Weight unchanged   : "
    f"{BLOCK_5_WEIGHT_STATE_UNCHANGED}"
)

print(
    f"Effective Weight identity    : "
    f"{BLOCK_5_EFFECTIVE_WEIGHT_IDENTITY}"
)

print(
    f"All parameters frozen        : "
    f"{BLOCK_5_ALL_PARAMETERS_FROZEN}"
)

print(
    f"All modules eval             : "
    f"{BLOCK_5_ALL_MODULES_EVAL}"
)

print(
    f"All gradients absent         : "
    f"{BLOCK_5_ALL_GRADIENTS_ABSENT}"
)

print("-" * 72)

print("Execution boundary")

print(
    f"Media AI repr. forward       : "
    f"{NOTEBOOK_10_BLOCK_5_MEDIA_AI_REPRESENTATION_FORWARD_EXECUTED}"
)

print(
    f"Transform forward            : "
    f"{NOTEBOOK_10_BLOCK_5_TRANSFORMATIVE_FORWARD_EXECUTED}"
)

print(
    f"Weight forward               : "
    f"{NOTEBOOK_10_BLOCK_5_WEIGHT_FORWARD_EXECUTED}"
)

print(
    f"Recurrent state created      : "
    f"{NOTEBOOK_10_BLOCK_5_RECURRENT_STATE_CREATED}"
)

print(
    f"Recurrent update executed    : "
    f"{NOTEBOOK_10_BLOCK_5_RECURRENT_UPDATE_EXECUTED}"
)

print(
    f"Loss calculated              : "
    f"{NOTEBOOK_10_BLOCK_5_LOSS_CALCULATED}"
)

print(
    f"Optimizer created            : "
    f"{NOTEBOOK_10_BLOCK_5_OPTIMIZER_CREATED}"
)

print(
    f"Backward pass executed       : "
    f"{NOTEBOOK_10_BLOCK_5_BACKWARD_PASS_EXECUTED}"
)

print(
    f"Parameter update executed    : "
    f"{NOTEBOOK_10_BLOCK_5_PARAMETER_UPDATE_EXECUTED}"
)

print(
    f"Training executed            : "
    f"{NOTEBOOK_10_BLOCK_5_TRAINING_EXECUTED}"
)

print("-" * 72)

print(
    f"Formal corpus audit valid    : "
    f"{NOTEBOOK_10_FORMAL_MULTI_ARTICLE_CORPUS_VALIDATED}"
)

print(
    f"Block valid                  : "
    f"{NOTEBOOK_10_BLOCK_5_VALID}"
)

print(
    f"Block complete               : "
    f"{NOTEBOOK_10_BLOCK_5_COMPLETE}"
)

print("=" * 72)


if BLOCK_5_MULTI_ARTICLE_EXECUTION_READY:

    print(
        "The validated Notebook 10 corpus satisfies the formal "
        "multi-article execution contract."
    )

    print(
        f"{BLOCK_5_REPRESENTATION_READY_ARTICLE_COUNT} independent "
        f"article(s) and "
        f"{BLOCK_5_REPRESENTATION_READY_SENTENCE_COUNT} canonical "
        "sentence(s) possess complete compatible 768-dimensional "
        "contextual representations."
    )

    print(
        "Article identity, sentence identity, paragraph structure, "
        "canonical ordering, article boundaries, representation "
        "alignment, encoder compatibility, normalisation and provenance "
        "were validated without modifying the inherited Media AI state."
    )

    print(
        "The pre-extension validated corpus remains preserved, and no "
        "synthetic article construction, pilot splitting, duplication "
        "or supervision substitution was introduced."
    )

    print(
        "Expanded-training sufficiency remains False because this block "
        "establishes architectural multi-article readiness only."
    )

    print(
        "Notebook 10 may now proceed to controlled multi-article "
        "recurrent execution and article-boundary reset validation."
    )

else:

    BLOCK_5_READINESS_BLOCKERS = []


    if not BLOCK_5_STRUCTURAL_MULTI_ARTICLE_READY:

        BLOCK_5_READINESS_BLOCKERS.append(
            "structural_multi_article_readiness"
        )


    if not BLOCK_5_REPRESENTATION_MULTI_ARTICLE_READY:

        BLOCK_5_READINESS_BLOCKERS.append(
            "representation_multi_article_readiness"
        )


    print(
        "The formal corpus audit completed without executing the "
        "Media AI architecture, but multi-article recurrent execution "
        "is not yet permitted."
    )

    print(
        "Readiness blockers: "
        f"{BLOCK_5_READINESS_BLOCKERS}"
    )

print("=" * 72)
```

# Block 6 — Controlled Multi-Article Recurrent Execution and Article-Boundary Reset Validation

## Purpose

This block performs the first controlled recurrent execution of the completed Media AI architecture across the formally validated multi-article corpus established in Block 5.

The objective is to verify that the learned transformative architecture operates correctly over complete article sequences while preserving the causal and article-isolation constraints defined throughout the project.

Block 6 therefore moves beyond corpus-readiness validation and executes the factual, psychological and social recurrent pathways across all validated sentences in canonical within-article order.

The block remains strictly a **forward-path validation stage**. It does not construct supervision targets, calculate an optimisation objective, create an optimiser, execute backpropagation or update model parameters.

## Methodological Role

Notebook 09 established the recurrent mechanism and demonstrated its operation within the original pilot architecture.

Notebook 10 now validates that the persisted architecture can execute consistently across the complete multi-article corpus restored and audited in Blocks 1–5.

For each pathway

$$
p \in
\{
\text{factual},
\text{psychological},
\text{social}
\},
$$

each article is treated as an independent ordered sequence.

For article \(a\), the recurrent state is initialised deterministically as

$$
\mathbf{h}^{(p)}_{a,0}
=
\mathbf{0}
\in
\mathbb{R}^{d_p},
$$

where the pathway dimensions inherited from Notebook 09 are

$$
d_{\mathrm{factual}} = 10,
\qquad
d_{\mathrm{psychological}} = 34,
\qquad
d_{\mathrm{social}} = 7.
$$

At sentence position \(t\), the frozen transformative mechanism receives the current pathway representation and the immediately preceding recurrent state:

$$
\mathbf{c}^{(p)}_{a,t}
=
T^{(p)}
\left(
\mathbf{r}^{(p)}_{a,t},
\mathbf{h}^{(p)}_{a,t-1}
\right),
$$

where:

- \(\mathbf{r}^{(p)}_{a,t}\) is the current Media AI pathway representation;
- \(\mathbf{h}^{(p)}_{a,t-1}\) is the preceding recurrent state;
- \(T^{(p)}\) is the frozen pathway-specific transformative mechanism;
- \(\mathbf{c}^{(p)}_{a,t}\) is the resulting candidate transformation.

The learned Transformative Weight is then applied dimension-wise:

$$
\Delta\mathbf{h}^{(p)}_{a,t}
=
\mathbf{w}^{(p)}
\odot
\mathbf{c}^{(p)}_{a,t},
$$

where

$$
\mathbf{w}^{(p)}
\in
[0,1]^{d_p}
$$

is the persisted effective Transformative-Weight vector and \(\odot\) denotes element-wise multiplication.

The recurrent state is updated additively:

$$
\mathbf{h}^{(p)}_{a,t}
=
\mathbf{h}^{(p)}_{a,t-1}
+
\Delta\mathbf{h}^{(p)}_{a,t}.
$$

This recurrence must use only information available at the current or preceding sentence positions. Future sentence representations are never supplied to the current recurrent transition.

## Article-Boundary Isolation

Article boundaries are treated as strict recurrent boundaries.

For every article \(a\),

$$
\mathbf{h}^{(p)}_{a,0}
=
\mathbf{0},
$$

irrespective of the final recurrent state of the preceding article.

Consequently,

$$
\mathbf{h}^{(p)}_{a,0}
\neq
\mathbf{h}^{(p)}_{a-1,T_{a-1}}
$$

except for an accidental numerical equality that carries no computational dependency.

The block must validate the stronger computational condition: the initial state of every article is independently constructed as a fresh deterministic zero tensor and is never derived from, copied from or linked to the final state of another article.

No recurrent information may therefore propagate across article boundaries.

## Canonical Sequence Execution

The corpus validated in Block 5 contains:

- **3 independent articles**;
- **54 canonical sentences**;
- complete compatible **768-dimensional contextual representations**;
- explicit article boundaries;
- validated sentence identities;
- validated within-article sequence positions.

Execution must preserve the canonical sentence order established by the persisted corpus.

Sentence shuffling is prohibited.

Each article is traversed independently from its first canonical sentence to its final canonical sentence before recurrent execution begins for the next article.

## Pathway Independence

The factual, psychological and social recurrent pathways remain structurally independent.

A recurrent state from one representational space must never be supplied to another pathway.

For

$$
p \neq q,
$$

the transition

$$
\mathbf{h}^{(p)}_{a,t-1}
\rightarrow
T^{(q)}
$$

is prohibited.

Each pathway therefore maintains its own dimensionality, Transformative Weight, candidate transformation and recurrent trajectory.

## Deferred Transformative Dimensions

The active and deferred Transformative-Weight partitions inherited from Notebook 09 remain unchanged.

For every deferred dimension \(j\),

$$
w^{(p)}_j = 0.
$$

Therefore its candidate contribution must satisfy

$$
\Delta h^{(p)}_{a,t,j}
=
w^{(p)}_j c^{(p)}_{a,t,j}
=
0.
$$

The block must confirm that deferred dimensions receive no Transformative-Weight-gated recurrent increment during multi-article execution.

The dimensions remain structurally present in the recurrent state; they are not removed from the pathway representation.

## Deterministic Execution Validation

The complete multi-article recurrent execution is repeated under the same validated model state and corpus ordering.

For every article, pathway and sentence position, repeated execution must reproduce the same:

- candidate transformation;
- effective Transformative Weight;
- weighted candidate contribution;
- recurrent state.

For two deterministic executions \(A\) and \(B\),

$$
\mathbf{h}^{(p,A)}_{a,t}
=
\mathbf{h}^{(p,B)}_{a,t}
$$

must hold within the numerical comparison policy used by the notebook.

This provides a corpus-level validation that the persisted architecture behaves deterministically before supervision alignment and expanded training are introduced.

## Model-State Preservation

Block 6 is an execution-only validation stage.

The complete inherited architecture contains:

- **894,003 representation parameters**;
- **8,187 transformative parameters**;
- **49 learned Transformative-Weight parameters**;

for a total of

$$
894{,}003 + 8{,}187 + 49
=
902{,}239
$$

model parameters.

All parameters must remain frozen and exactly unchanged throughout the block.

The block must verify that:

- representation parameters remain unchanged;
- transformative parameters remain unchanged;
- latent Transformative-Weight parameters remain unchanged;
- effective Transformative Weights remain identical to the persisted Notebook 09 state;
- all model modules remain in evaluation mode;
- no parameter gradients are created.

## Execution Boundary

Block 6 may execute:

- validated Media AI representation inputs;
- frozen transformative forward passes;
- Transformative-Weight forward passes;
- weighted candidate transformations;
- recurrent-state construction;
- recurrent-state updates;
- complete article trajectories;
- deterministic repeated trajectories.

Block 6 must not execute:

- supervision alignment;
- transformation-target construction;
- loss calculation;
- optimiser construction;
- gradient calculation;
- backward propagation;
- gradient clipping;
- optimiser steps;
- parameter updates;
- training.

Supervision is not supplied as model input.

## Validation Requirements

The block must explicitly verify:

1. all 3 validated articles are traversed;
2. all 54 canonical sentences are traversed exactly once per execution;
3. canonical within-article ordering is preserved;
4. every article begins from a fresh deterministic zero recurrent state;
5. no recurrent state propagates across article boundaries;
6. factual, psychological and social pathways remain independent;
7. candidate transformations have the correct pathway-specific dimensions;
8. all candidate transformations are finite;
9. learned Transformative Weights retain their validated values;
10. deferred dimensions receive zero weighted candidate contribution;
11. additive recurrent updates are satisfied at every sentence position;
12. all recurrent states remain finite;
13. repeated complete corpus execution is deterministic;
14. no future context is used;
15. all 902,239 model parameters remain unchanged;
16. no gradients, loss, optimiser, backward pass or parameter update are introduced.

## Expected Outcome

Successful completion of Block 6 establishes that the completed Media AI architecture can execute causally and deterministically across multiple independent articles while preserving strict article-boundary isolation.

The result constitutes an **architectural multi-article recurrent validation**, not a model-performance claim.

No inference about generalisation, predictive quality or production readiness is made from the three-article corpus.

If all recurrent, causal, dimensional, deterministic and model-state-preservation checks pass, Notebook 10 may proceed to **Block 7 — Expanded Supervision Alignment and Transformation-Target Construction**, where validated supervision can be aligned to the multi-article recurrent structure in preparation for the final controlled pilot-training stage.

```python
# =============================================================================
# Media AI — Notebook 10
# Block 6: Controlled Multi-Article Recurrent Execution
#          and Article-Boundary Reset Validation
# =============================================================================

from copy import deepcopy
from datetime import datetime, timezone

import numpy as np
import torch


# =============================================================================
# Block identity
# =============================================================================

NOTEBOOK_10_BLOCK_6 = 6

NOTEBOOK_10_BLOCK_6_NAME = (
    "Controlled Multi-Article Recurrent Execution "
    "and Article-Boundary Reset Validation"
)

NOTEBOOK_10_BLOCK_6_VERSION = "1.1"

BLOCK_6_EXECUTED_AT_UTC = (
    datetime.now(
        timezone.utc
    ).isoformat()
)


# =============================================================================
# Prerequisite validation
# =============================================================================
#
# Block 6 is intentionally a downstream execution block. It consumes the
# validated runtime state produced by Blocks 1–5 and must not silently rebuild
# those stages inside this block.
#
# A fresh Colab/Jupyter kernel therefore requires Blocks 1–5 to be rerun first.
# =============================================================================

BLOCK_6_CORE_REQUIRED_GLOBALS = [
    "NOTEBOOK_10_BLOCK_5_COMPLETE",
    "NOTEBOOK_10_BLOCK_5_VALID",
    "NOTEBOOK_10_FORMAL_MULTI_ARTICLE_CORPUS_VALIDATED",
    "NOTEBOOK_10_MULTI_ARTICLE_EXECUTION_READY",
    "NOTEBOOK_10_VALIDATED_MULTI_ARTICLE_RECORDS",
    "NOTEBOOK_10_VALIDATED_REPRESENTATION_MATRICES",
    "NOTEBOOK_10_REPRESENTATION_READY_ARTICLE_IDS",
    "NOTEBOOK_10_REPRESENTATION_MODULES",
    "NOTEBOOK_10_TRANSFORMATIVE_MODULES",
    "NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES",
    "NOTEBOOK_10_PATHWAY_DIMENSIONS",
    "NOTEBOOK_10_ACTIVE_DIMENSION_INDICES",
    "NOTEBOOK_10_DEFERRED_DIMENSION_INDICES",
    "NOTEBOOK_10_INHERITED_EFFECTIVE_TRANSFORMATIVE_WEIGHTS",
]


BLOCK_6_MISSING_PREREQUISITES = [
    object_name

    for object_name
    in BLOCK_6_CORE_REQUIRED_GLOBALS

    if object_name not in globals()
]


BLOCK_6_RUNTIME_STATE_PRESENT = (
    len(
        BLOCK_6_MISSING_PREREQUISITES
    )
    ==
    0
)


if not BLOCK_6_RUNTIME_STATE_PRESENT:

    BLOCK_6_BLOCK5_STATE_MISSING = all(
        object_name not in globals()

        for object_name
        in (
            "NOTEBOOK_10_BLOCK_5_COMPLETE",
            "NOTEBOOK_10_VALIDATED_MULTI_ARTICLE_RECORDS",
            "NOTEBOOK_10_VALIDATED_REPRESENTATION_MATRICES",
        )
    )


    if BLOCK_6_BLOCK5_STATE_MISSING:

        raise RuntimeError(
            "Notebook 10 Block 6 cannot execute because the current "
            "Jupyter/Colab runtime does not contain the validated state from "
            "Blocks 1–5. This normally indicates that the kernel/runtime was "
            "restarted or changed. Rerun Notebook 10 Blocks 1–5 in order in "
            "this same runtime, then execute Block 6 again. "
            f"Missing objects: {BLOCK_6_MISSING_PREREQUISITES}"
        )


    raise RuntimeError(
        "Notebook 10 Block 6 prerequisites are incomplete in the current "
        "runtime. Rerun the missing upstream block(s) before Block 6. "
        f"Missing objects: {BLOCK_6_MISSING_PREREQUISITES}"
    )


BLOCK_6_PREREQUISITES_VALID = all(
    [
        NOTEBOOK_10_BLOCK_5_COMPLETE is True,
        NOTEBOOK_10_BLOCK_5_VALID is True,
        NOTEBOOK_10_FORMAL_MULTI_ARTICLE_CORPUS_VALIDATED is True,
        NOTEBOOK_10_MULTI_ARTICLE_EXECUTION_READY is True,

        isinstance(
            NOTEBOOK_10_VALIDATED_MULTI_ARTICLE_RECORDS,
            dict,
        ),

        isinstance(
            NOTEBOOK_10_VALIDATED_REPRESENTATION_MATRICES,
            dict,
        ),

        len(
            NOTEBOOK_10_REPRESENTATION_READY_ARTICLE_IDS
        )
        >=
        2,

        isinstance(
            NOTEBOOK_10_REPRESENTATION_MODULES,
            dict,
        ),

        isinstance(
            NOTEBOOK_10_TRANSFORMATIVE_MODULES,
            dict,
        ),

        isinstance(
            NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
            dict,
        ),
    ]
)


if not BLOCK_6_PREREQUISITES_VALID:

    raise RuntimeError(
        "Notebook 10 Block 6 requires a valid, representation-ready "
        "multi-article corpus and restored frozen architecture from Blocks 1–5."
    )


# =============================================================================
# Resolve execution device and dtype from restored architecture
# =============================================================================
#
# Device and dtype are runtime properties of the restored model. Block 6 derives
# them directly from the inherited representation modules instead of requiring
# separate global aliases.
# =============================================================================

BLOCK_6_MODEL_PARAMETERS = [
    parameter

    for module
    in NOTEBOOK_10_REPRESENTATION_MODULES.values()

    for parameter
    in module.parameters()
]


if not BLOCK_6_MODEL_PARAMETERS:

    raise RuntimeError(
        "Notebook 10 Block 6 could not resolve model device/dtype because "
        "the restored representation modules contain no parameters."
    )


BLOCK_6_EXECUTION_DEVICE = (
    BLOCK_6_MODEL_PARAMETERS[
        0
    ].device
)


BLOCK_6_EXECUTION_DTYPE = (
    BLOCK_6_MODEL_PARAMETERS[
        0
    ].dtype
)


BLOCK_6_DEVICE_DTYPE_CONSISTENT = all(
    parameter.device
    ==
    BLOCK_6_EXECUTION_DEVICE
    and
    parameter.dtype
    ==
    BLOCK_6_EXECUTION_DTYPE

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


if not BLOCK_6_DEVICE_DTYPE_CONSISTENT:

    raise RuntimeError(
        "Notebook 10 Block 6 inherited modules do not share a consistent "
        "execution device and dtype."
    )


# =============================================================================
# Canonical architectural contract
# =============================================================================

BLOCK_6_EXPECTED_CONTEXTUAL_DIMENSION = 768

BLOCK_6_EXPECTED_PATHWAYS = tuple(
    NOTEBOOK_10_PATHWAY_DIMENSIONS.keys()
)


BLOCK_6_EXPECTED_REPRESENTATION_PARAMETER_COUNT = 894_003

BLOCK_6_EXPECTED_TRANSFORMATIVE_PARAMETER_COUNT = 8_187

BLOCK_6_EXPECTED_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT = 49

BLOCK_6_EXPECTED_TOTAL_PARAMETER_COUNT = 902_239


BLOCK_6_EXPECTED_ARTICLE_COUNT = len(
    NOTEBOOK_10_REPRESENTATION_READY_ARTICLE_IDS
)


BLOCK_6_EXPECTED_SENTENCE_COUNT = sum(
    len(
        NOTEBOOK_10_VALIDATED_MULTI_ARTICLE_RECORDS[
            article_id
        ]
    )

    for article_id
    in NOTEBOOK_10_REPRESENTATION_READY_ARTICLE_IDS
)


# =============================================================================
# Execution boundary
# =============================================================================

NOTEBOOK_10_BLOCK_6_MEDIA_AI_REPRESENTATION_FORWARD_EXECUTED = False

NOTEBOOK_10_BLOCK_6_TRANSFORMATIVE_FORWARD_EXECUTED = False

NOTEBOOK_10_BLOCK_6_WEIGHT_FORWARD_EXECUTED = False

NOTEBOOK_10_BLOCK_6_RECURRENT_STATE_CREATED = False

NOTEBOOK_10_BLOCK_6_RECURRENT_UPDATE_EXECUTED = False

NOTEBOOK_10_BLOCK_6_ARTICLE_TRAJECTORIES_EXECUTED = False

NOTEBOOK_10_BLOCK_6_REPEAT_EXECUTION_PERFORMED = False

NOTEBOOK_10_BLOCK_6_FUTURE_CONTEXT_USED = False

NOTEBOOK_10_BLOCK_6_SUPERVISION_USED_AS_MODEL_INPUT = False

NOTEBOOK_10_BLOCK_6_LOSS_CALCULATED = False

NOTEBOOK_10_BLOCK_6_OPTIMIZER_CREATED = False

NOTEBOOK_10_BLOCK_6_ZERO_GRAD_EXECUTED = False

NOTEBOOK_10_BLOCK_6_BACKWARD_PASS_EXECUTED = False

NOTEBOOK_10_BLOCK_6_GRADIENT_CLIPPING_EXECUTED = False

NOTEBOOK_10_BLOCK_6_OPTIMIZER_STEP_EXECUTED = False

NOTEBOOK_10_BLOCK_6_PARAMETER_UPDATE_EXECUTED = False

NOTEBOOK_10_BLOCK_6_TRAINING_EXECUTED = False


# =============================================================================
# Model-state snapshot helpers
# =============================================================================

def block_6_snapshot_module_collection(
    module_collection,
):

    return {
        module_name:
            {
                tensor_name:
                    tensor.detach()
                    .cpu()
                    .clone()

                for (
                    tensor_name,
                    tensor,
                ) in module.state_dict().items()
            }

        for (
            module_name,
            module,
        ) in module_collection.items()
    }


def block_6_nested_state_exact(
    state_before,
    state_after,
):

    if state_before.keys() != state_after.keys():

        return False


    for module_name in state_before:

        if (
            state_before[
                module_name
            ].keys()
            !=
            state_after[
                module_name
            ].keys()
        ):

            return False


        for tensor_name in state_before[
            module_name
        ]:

            if not torch.equal(
                state_before[
                    module_name
                ][
                    tensor_name
                ],
                state_after[
                    module_name
                ][
                    tensor_name
                ],
            ):

                return False


    return True


# =============================================================================
# Immutable inherited model-state baseline
# =============================================================================

BLOCK_6_REPRESENTATION_STATE_BASELINE = (
    block_6_snapshot_module_collection(
        NOTEBOOK_10_REPRESENTATION_MODULES
    )
)


BLOCK_6_TRANSFORMATIVE_STATE_BASELINE = (
    block_6_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_MODULES
    )
)


BLOCK_6_WEIGHT_STATE_BASELINE = (
    block_6_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES
    )
)


# =============================================================================
# Freeze/evaluation validation before execution
# =============================================================================

BLOCK_6_PARAMETERS_FROZEN_BEFORE = all(
    not parameter.requires_grad

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


BLOCK_6_MODULES_EVAL_BEFORE = all(
    not module.training

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()
)


BLOCK_6_GRADIENTS_ABSENT_BEFORE = all(
    parameter.grad is None

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


if not all(
    [
        BLOCK_6_PARAMETERS_FROZEN_BEFORE,
        BLOCK_6_MODULES_EVAL_BEFORE,
        BLOCK_6_GRADIENTS_ABSENT_BEFORE,
    ]
):

    raise RuntimeError(
        "Notebook 10 Block 6 inherited model state is not in the "
        "required frozen/evaluation/no-gradient condition."
    )


# =============================================================================
# Parameter accounting
# =============================================================================

BLOCK_6_REPRESENTATION_PARAMETER_COUNT = sum(
    parameter.numel()

    for module
    in NOTEBOOK_10_REPRESENTATION_MODULES.values()

    for parameter
    in module.parameters()
)


BLOCK_6_TRANSFORMATIVE_PARAMETER_COUNT = sum(
    parameter.numel()

    for module
    in NOTEBOOK_10_TRANSFORMATIVE_MODULES.values()

    for parameter
    in module.parameters()
)


BLOCK_6_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT = sum(
    parameter.numel()

    for module
    in NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES.values()

    for parameter
    in module.parameters()
)


BLOCK_6_TOTAL_PARAMETER_COUNT = (
    BLOCK_6_REPRESENTATION_PARAMETER_COUNT
    +
    BLOCK_6_TRANSFORMATIVE_PARAMETER_COUNT
    +
    BLOCK_6_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT
)


BLOCK_6_PARAMETER_ACCOUNTING_VALID = all(
    [
        (
            BLOCK_6_REPRESENTATION_PARAMETER_COUNT
            ==
            BLOCK_6_EXPECTED_REPRESENTATION_PARAMETER_COUNT
        ),
        (
            BLOCK_6_TRANSFORMATIVE_PARAMETER_COUNT
            ==
            BLOCK_6_EXPECTED_TRANSFORMATIVE_PARAMETER_COUNT
        ),
        (
            BLOCK_6_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT
            ==
            BLOCK_6_EXPECTED_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT
        ),
        (
            BLOCK_6_TOTAL_PARAMETER_COUNT
            ==
            BLOCK_6_EXPECTED_TOTAL_PARAMETER_COUNT
        ),
    ]
)


if not BLOCK_6_PARAMETER_ACCOUNTING_VALID:

    raise RuntimeError(
        "Notebook 10 Block 6 parameter accounting does not reproduce "
        "the validated Notebook 09 → 10 architecture."
    )


# =============================================================================
# Canonical corpus validation
# =============================================================================

BLOCK_6_ARTICLE_IDS = tuple(
    NOTEBOOK_10_REPRESENTATION_READY_ARTICLE_IDS
)


BLOCK_6_ARTICLE_IDS_VALID = all(
    [
        len(
            BLOCK_6_ARTICLE_IDS
        )
        ==
        BLOCK_6_EXPECTED_ARTICLE_COUNT,

        len(
            BLOCK_6_ARTICLE_IDS
        )
        ==
        len(
            set(
                BLOCK_6_ARTICLE_IDS
            )
        ),

        all(
            article_id
            in
            NOTEBOOK_10_VALIDATED_MULTI_ARTICLE_RECORDS

            for article_id
            in BLOCK_6_ARTICLE_IDS
        ),

        all(
            article_id
            in
            NOTEBOOK_10_VALIDATED_REPRESENTATION_MATRICES

            for article_id
            in BLOCK_6_ARTICLE_IDS
        ),
    ]
)


BLOCK_6_CANONICAL_CORPUS_AUDIT = {}

BLOCK_6_CANONICAL_SEQUENCE_VALID = True

BLOCK_6_CONTEXTUAL_MATRICES_VALID = True

BLOCK_6_TOTAL_SENTENCE_COUNT = 0


for article_id in BLOCK_6_ARTICLE_IDS:

    records = sorted(
        NOTEBOOK_10_VALIDATED_MULTI_ARTICLE_RECORDS[
            article_id
        ],
        key=lambda record:
            int(
                record.get(
                    "canonical_sequence_position",
                    record.get(
                        "sequence_position",
                        -1,
                    ),
                )
            ),
    )


    matrix = np.asarray(
        NOTEBOOK_10_VALIDATED_REPRESENTATION_MATRICES[
            article_id
        ],
        dtype=
            np.float32,
    )


    positions = [
        int(
            record.get(
                "canonical_sequence_position",
                record.get(
                    "sequence_position",
                    -1,
                ),
            )
        )

        for record
        in records
    ]


    sequence_valid = (
        positions
        ==
        list(
            range(
                len(
                    records
                )
            )
        )
    )


    sentence_ids = [
        record.get(
            "sentence_id"
        )

        for record
        in records
    ]


    sentence_identity_valid = all(
        sentence_id is not None

        for sentence_id
        in sentence_ids
    ) and (
        len(
            sentence_ids
        )
        ==
        len(
            set(
                sentence_ids
            )
        )
    )


    matrix_shape_valid = (
        matrix.shape
        ==
        (
            len(
                records
            ),
            BLOCK_6_EXPECTED_CONTEXTUAL_DIMENSION,
        )
    )


    matrix_finite = bool(
        np.isfinite(
            matrix
        ).all()
    )


    article_valid = all(
        [
            len(
                records
            )
            >
            0,
            sequence_valid,
            sentence_identity_valid,
            matrix_shape_valid,
            matrix_finite,
        ]
    )


    BLOCK_6_CANONICAL_CORPUS_AUDIT[
        article_id
    ] = {
        "sentence_count":
            len(
                records
            ),

        "sequence_valid":
            sequence_valid,

        "sentence_identity_valid":
            sentence_identity_valid,

        "matrix_shape":
            tuple(
                matrix.shape
            ),

        "matrix_shape_valid":
            matrix_shape_valid,

        "matrix_finite":
            matrix_finite,

        "article_valid":
            article_valid,
    }


    if not sequence_valid:

        BLOCK_6_CANONICAL_SEQUENCE_VALID = False


    if not all(
        [
            matrix_shape_valid,
            matrix_finite,
        ]
    ):

        BLOCK_6_CONTEXTUAL_MATRICES_VALID = False


    BLOCK_6_TOTAL_SENTENCE_COUNT += len(
        records
    )


BLOCK_6_ALL_SENTENCES_ACCOUNTED_FOR = (
    BLOCK_6_TOTAL_SENTENCE_COUNT
    ==
    BLOCK_6_EXPECTED_SENTENCE_COUNT
)


if not all(
    [
        BLOCK_6_ARTICLE_IDS_VALID,
        BLOCK_6_CANONICAL_SEQUENCE_VALID,
        BLOCK_6_CONTEXTUAL_MATRICES_VALID,
        BLOCK_6_ALL_SENTENCES_ACCOUNTED_FOR,
    ]
):

    raise RuntimeError(
        "Notebook 10 Block 6 canonical corpus contract is invalid."
    )


# =============================================================================
# Resolve representation path
# =============================================================================

BLOCK_6_REQUIRED_REPRESENTATION_MODULE_NAMES = {
    "backbone",
    "global_confluent",
    "factual_head",
    "psychological_head",
    "social_head",
}


BLOCK_6_REPRESENTATION_MODULE_NAMES_VALID = (
    set(
        NOTEBOOK_10_REPRESENTATION_MODULES.keys()
    )
    ==
    BLOCK_6_REQUIRED_REPRESENTATION_MODULE_NAMES
)


if not BLOCK_6_REPRESENTATION_MODULE_NAMES_VALID:

    raise RuntimeError(
        "Notebook 10 Block 6 representation module topology does not "
        "match the validated Notebook 10 restoration contract."
    )


BLOCK_6_BACKBONE = (
    NOTEBOOK_10_REPRESENTATION_MODULES[
        "backbone"
    ]
)


BLOCK_6_GLOBAL_CONFLUENT = (
    NOTEBOOK_10_REPRESENTATION_MODULES[
        "global_confluent"
    ]
)


BLOCK_6_REPRESENTATION_HEADS = {
    "factual":
        NOTEBOOK_10_REPRESENTATION_MODULES[
            "factual_head"
        ],

    "psychological":
        NOTEBOOK_10_REPRESENTATION_MODULES[
            "psychological_head"
        ],

    "social":
        NOTEBOOK_10_REPRESENTATION_MODULES[
            "social_head"
        ],
}


# =============================================================================
# Effective Transformative-Weight validation
# =============================================================================

with torch.inference_mode():

    BLOCK_6_EFFECTIVE_WEIGHTS = {
        pathway_name:
            NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES[
                pathway_name
            ]()
            .detach()
            .clone()

        for pathway_name
        in BLOCK_6_EXPECTED_PATHWAYS
    }


NOTEBOOK_10_BLOCK_6_WEIGHT_FORWARD_EXECUTED = True


BLOCK_6_EFFECTIVE_WEIGHTS_VALID = True

BLOCK_6_EFFECTIVE_WEIGHT_AUDIT = {}


for pathway_name in BLOCK_6_EXPECTED_PATHWAYS:

    pathway_dim = int(
        NOTEBOOK_10_PATHWAY_DIMENSIONS[
            pathway_name
        ]
    )


    weight = (
        BLOCK_6_EFFECTIVE_WEIGHTS[
            pathway_name
        ]
    )


    inherited_weight = (
        NOTEBOOK_10_INHERITED_EFFECTIVE_TRANSFORMATIVE_WEIGHTS[
            pathway_name
        ]
        .to(
            device=
                weight.device,
            dtype=
                weight.dtype,
        )
        .reshape(
            -1
        )
    )


    active_indices = tuple(
        int(
            index
        )

        for index
        in NOTEBOOK_10_ACTIVE_DIMENSION_INDICES[
            pathway_name
        ]
    )


    deferred_indices = tuple(
        int(
            index
        )

        for index
        in NOTEBOOK_10_DEFERRED_DIMENSION_INDICES[
            pathway_name
        ]
    )


    shape_valid = (
        tuple(
            weight.shape
        )
        ==
        (
            pathway_dim,
        )
    )


    finite = bool(
        torch.isfinite(
            weight
        ).all()
        .item()
    )


    bounded = bool(
        (
            (
                weight
                >=
                0.0
            )
            &
            (
                weight
                <=
                1.0
            )
        ).all()
        .item()
    )


    inherited_exact = bool(
        torch.equal(
            weight.detach()
            .cpu(),

            inherited_weight.detach()
            .cpu(),
        )
    )


    active_layout_valid = all(
        0
        <=
        index
        <
        pathway_dim

        for index
        in active_indices
    )


    deferred_layout_valid = all(
        0
        <=
        index
        <
        pathway_dim

        for index
        in deferred_indices
    )


    partition_valid = all(
        [
            set(
                active_indices
            ).isdisjoint(
                deferred_indices
            ),
            (
                set(
                    active_indices
                )
                |
                set(
                    deferred_indices
                )
                ==
                set(
                    range(
                        pathway_dim
                    )
                )
            ),
        ]
    )


    deferred_zero = True


    if deferred_indices:

        deferred_index_tensor = torch.tensor(
            deferred_indices,
            dtype=
                torch.long,
            device=
                weight.device,
        )


        deferred_zero = bool(
            torch.equal(
                weight.index_select(
                    0,
                    deferred_index_tensor,
                ),

                torch.zeros(
                    len(
                        deferred_indices
                    ),
                    dtype=
                        weight.dtype,
                    device=
                        weight.device,
                ),
            )
        )


    pathway_valid = all(
        [
            shape_valid,
            finite,
            bounded,
            inherited_exact,
            active_layout_valid,
            deferred_layout_valid,
            partition_valid,
            deferred_zero,
        ]
    )


    BLOCK_6_EFFECTIVE_WEIGHT_AUDIT[
        pathway_name
    ] = {
        "dimension":
            pathway_dim,

        "active_indices":
            active_indices,

        "deferred_indices":
            deferred_indices,

        "shape_valid":
            shape_valid,

        "finite":
            finite,

        "bounded":
            bounded,

        "inherited_exact":
            inherited_exact,

        "partition_valid":
            partition_valid,

        "deferred_zero":
            deferred_zero,

        "valid":
            pathway_valid,
    }


    if not pathway_valid:

        BLOCK_6_EFFECTIVE_WEIGHTS_VALID = False


if not BLOCK_6_EFFECTIVE_WEIGHTS_VALID:

    raise RuntimeError(
        "Notebook 10 Block 6 inherited Transformative-Weight contract "
        "is invalid."
    )


# =============================================================================
# Fresh-zero recurrent-state helper
# =============================================================================

def block_6_zero_state(
    pathway_name,
):

    pathway_dim = int(
        NOTEBOOK_10_PATHWAY_DIMENSIONS[
            pathway_name
        ]
    )


    return torch.zeros(
        (
            1,
            pathway_dim,
        ),
        dtype=
            BLOCK_6_EXECUTION_DTYPE,
        device=
            BLOCK_6_EXECUTION_DEVICE,
    )


# =============================================================================
# Representation forward helper
# =============================================================================

def block_6_pathway_representations(
    contextual_matrix,
):

    contextual_tensor = torch.as_tensor(
        contextual_matrix,
        dtype=
            BLOCK_6_EXECUTION_DTYPE,
        device=
            BLOCK_6_EXECUTION_DEVICE,
    )


    backbone_output = BLOCK_6_BACKBONE(
        contextual_tensor
    )


    confluent_output = BLOCK_6_GLOBAL_CONFLUENT(
        backbone_output
    )


    pathway_outputs = {
        pathway_name:
            BLOCK_6_REPRESENTATION_HEADS[
                pathway_name
            ](
                confluent_output
            )

        for pathway_name
        in BLOCK_6_EXPECTED_PATHWAYS
    }


    return {
        "contextual":
            contextual_tensor,

        "backbone":
            backbone_output,

        "confluent":
            confluent_output,

        "pathways":
            pathway_outputs,
    }


# =============================================================================
# Complete recurrent corpus execution helper
# =============================================================================

def block_6_execute_corpus(
    execution_label,
):

    execution_result = {
        "execution_label":
            execution_label,

        "articles":
            {},

        "article_order":
            tuple(
                BLOCK_6_ARTICLE_IDS
            ),

        "sentence_count":
            0,
    }


    initial_state_storage_pointers = {
        pathway_name:
            []

        for pathway_name
        in BLOCK_6_EXPECTED_PATHWAYS
    }


    for article_id in BLOCK_6_ARTICLE_IDS:

        records = sorted(
            NOTEBOOK_10_VALIDATED_MULTI_ARTICLE_RECORDS[
                article_id
            ],
            key=lambda record:
                int(
                    record.get(
                        "canonical_sequence_position",
                        record.get(
                            "sequence_position",
                            -1,
                        ),
                    )
                ),
        )


        matrix = np.asarray(
            NOTEBOOK_10_VALIDATED_REPRESENTATION_MATRICES[
                article_id
            ],
            dtype=
                np.float32,
        )


        representation_result = (
            block_6_pathway_representations(
                matrix
            )
        )


        article_result = {
            "article_id":
                article_id,

            "sentence_ids":
                tuple(
                    record[
                        "sentence_id"
                    ]

                    for record
                    in records
                ),

            "sequence_positions":
                tuple(
                    int(
                        record.get(
                            "canonical_sequence_position",
                            record.get(
                                "sequence_position",
                                -1,
                            ),
                        )
                    )

                    for record
                    in records
                ),

            "contextual_shape":
                tuple(
                    representation_result[
                        "contextual"
                    ].shape
                ),

            "backbone_shape":
                tuple(
                    representation_result[
                        "backbone"
                    ].shape
                ),

            "confluent_shape":
                tuple(
                    representation_result[
                        "confluent"
                    ].shape
                ),

            "pathways":
                {},
        }


        for pathway_name in BLOCK_6_EXPECTED_PATHWAYS:

            pathway_dim = int(
                NOTEBOOK_10_PATHWAY_DIMENSIONS[
                    pathway_name
                ]
            )


            current_representations = (
                representation_result[
                    "pathways"
                ][
                    pathway_name
                ]
            )


            if tuple(
                current_representations.shape
            ) != (
                len(
                    records
                ),
                pathway_dim,
            ):

                raise RuntimeError(
                    "Notebook 10 Block 6 pathway representation shape "
                    f"is invalid for '{pathway_name}' in article "
                    f"'{article_id}'."
                )


            preceding_state = (
                block_6_zero_state(
                    pathway_name
                )
            )


            initial_state_storage_pointers[
                pathway_name
            ].append(
                int(
                    preceding_state.data_ptr()
                )
            )


            initial_state = (
                preceding_state.detach()
                .clone()
            )


            weight = (
                BLOCK_6_EFFECTIVE_WEIGHTS[
                    pathway_name
                ]
                .to(
                    device=
                        BLOCK_6_EXECUTION_DEVICE,
                    dtype=
                        BLOCK_6_EXECUTION_DTYPE,
                )
                .reshape(
                    1,
                    pathway_dim,
                )
            )


            deferred_indices = tuple(
                int(
                    index
                )

                for index
                in NOTEBOOK_10_DEFERRED_DIMENSION_INDICES[
                    pathway_name
                ]
            )


            steps = []

            cumulative_weighted_candidate = (
                torch.zeros_like(
                    preceding_state
                )
            )


            for sentence_position in range(
                len(
                    records
                )
            ):

                current_representation = (
                    current_representations[
                        sentence_position:
                        sentence_position
                        +
                        1
                    ]
                )


                previous_state_snapshot = (
                    preceding_state.detach()
                    .clone()
                )


                candidate = (
                    NOTEBOOK_10_TRANSFORMATIVE_MODULES[
                        pathway_name
                    ](
                        current_representation,
                        preceding_state,
                    )
                )


                weighted_candidate = (
                    candidate
                    *
                    weight
                )


                next_state = (
                    preceding_state
                    +
                    weighted_candidate
                )


                cumulative_weighted_candidate = (
                    cumulative_weighted_candidate
                    +
                    weighted_candidate
                )


                deferred_contribution_zero = True

                deferred_state_zero = True


                if deferred_indices:

                    deferred_index_tensor = torch.tensor(
                        deferred_indices,
                        dtype=
                            torch.long,
                        device=
                            BLOCK_6_EXECUTION_DEVICE,
                    )


                    deferred_contribution_zero = bool(
                        torch.equal(
                            weighted_candidate.index_select(
                                1,
                                deferred_index_tensor,
                            ),

                            torch.zeros(
                                (
                                    1,
                                    len(
                                        deferred_indices
                                    ),
                                ),
                                dtype=
                                    BLOCK_6_EXECUTION_DTYPE,
                                device=
                                    BLOCK_6_EXECUTION_DEVICE,
                            ),
                        )
                    )


                    deferred_state_zero = bool(
                        torch.equal(
                            next_state.index_select(
                                1,
                                deferred_index_tensor,
                            ),

                            torch.zeros(
                                (
                                    1,
                                    len(
                                        deferred_indices
                                    ),
                                ),
                                dtype=
                                    BLOCK_6_EXECUTION_DTYPE,
                                device=
                                    BLOCK_6_EXECUTION_DEVICE,
                            ),
                        )
                    )


                additive_identity = bool(
                    torch.equal(
                        next_state,
                        previous_state_snapshot
                        +
                        weighted_candidate,
                    )
                )


                cumulative_identity = bool(
                    torch.equal(
                        next_state,
                        initial_state
                        +
                        cumulative_weighted_candidate,
                    )
                )


                step_record = {
                    "sentence_id":
                        records[
                            sentence_position
                        ][
                            "sentence_id"
                        ],

                    "sequence_position":
                        int(
                            sentence_position
                        ),

                    "current_representation":
                        current_representation.detach()
                        .cpu()
                        .clone(),

                    "preceding_state":
                        previous_state_snapshot.detach()
                        .cpu()
                        .clone(),

                    "candidate_transformation":
                        candidate.detach()
                        .cpu()
                        .clone(),

                    "effective_weight":
                        weight.detach()
                        .cpu()
                        .clone(),

                    "weighted_candidate":
                        weighted_candidate.detach()
                        .cpu()
                        .clone(),

                    "recurrent_state":
                        next_state.detach()
                        .cpu()
                        .clone(),

                    "candidate_finite":
                        bool(
                            torch.isfinite(
                                candidate
                            ).all()
                            .item()
                        ),

                    "weighted_candidate_finite":
                        bool(
                            torch.isfinite(
                                weighted_candidate
                            ).all()
                            .item()
                        ),

                    "state_finite":
                        bool(
                            torch.isfinite(
                                next_state
                            ).all()
                            .item()
                        ),

                    "additive_identity":
                        additive_identity,

                    "cumulative_identity":
                        cumulative_identity,

                    "deferred_contribution_zero":
                        deferred_contribution_zero,

                    "deferred_state_zero":
                        deferred_state_zero,
                }


                steps.append(
                    step_record
                )


                preceding_state = (
                    next_state
                )


            article_result[
                "pathways"
            ][
                pathway_name
            ] = {
                "dimension":
                    pathway_dim,

                "initial_state":
                    initial_state.detach()
                    .cpu()
                    .clone(),

                "initial_state_data_ptr":
                    initial_state_storage_pointers[
                        pathway_name
                    ][
                        -1
                    ],

                "representation_matrix":
                    current_representations.detach()
                    .cpu()
                    .clone(),

                "steps":
                    steps,

                "final_state":
                    preceding_state.detach()
                    .cpu()
                    .clone(),

                "cumulative_weighted_candidate":
                    cumulative_weighted_candidate.detach()
                    .cpu()
                    .clone(),
            }


        execution_result[
            "articles"
        ][
            article_id
        ] = article_result


        execution_result[
            "sentence_count"
        ] += len(
            records
        )


    execution_result[
        "initial_state_storage_pointers"
    ] = {
        pathway_name:
            tuple(
                storage_pointers
            )

        for (
            pathway_name,
            storage_pointers,
        ) in initial_state_storage_pointers.items()
    }


    return execution_result


# =============================================================================
# First complete multi-article execution
# =============================================================================

with torch.inference_mode():

    BLOCK_6_EXECUTION_A = (
        block_6_execute_corpus(
            "execution_a"
        )
    )


NOTEBOOK_10_BLOCK_6_MEDIA_AI_REPRESENTATION_FORWARD_EXECUTED = True

NOTEBOOK_10_BLOCK_6_TRANSFORMATIVE_FORWARD_EXECUTED = True

NOTEBOOK_10_BLOCK_6_RECURRENT_STATE_CREATED = True

NOTEBOOK_10_BLOCK_6_RECURRENT_UPDATE_EXECUTED = True

NOTEBOOK_10_BLOCK_6_ARTICLE_TRAJECTORIES_EXECUTED = True


# =============================================================================
# Second complete deterministic execution
# =============================================================================

with torch.inference_mode():

    BLOCK_6_EXECUTION_B = (
        block_6_execute_corpus(
            "execution_b"
        )
    )


NOTEBOOK_10_BLOCK_6_REPEAT_EXECUTION_PERFORMED = True


# =============================================================================
# Corpus traversal validation
# =============================================================================

BLOCK_6_ARTICLES_TRAVERSED = len(
    BLOCK_6_EXECUTION_A[
        "articles"
    ]
)


BLOCK_6_SENTENCES_TRAVERSED = (
    BLOCK_6_EXECUTION_A[
        "sentence_count"
    ]
)


BLOCK_6_ARTICLE_COUNT_VALID = (
    BLOCK_6_ARTICLES_TRAVERSED
    ==
    BLOCK_6_EXPECTED_ARTICLE_COUNT
)


BLOCK_6_SENTENCE_COUNT_VALID = (
    BLOCK_6_SENTENCES_TRAVERSED
    ==
    BLOCK_6_EXPECTED_SENTENCE_COUNT
)


BLOCK_6_ARTICLE_ORDER_VALID = (
    BLOCK_6_EXECUTION_A[
        "article_order"
    ]
    ==
    tuple(
        BLOCK_6_ARTICLE_IDS
    )
)


# =============================================================================
# Fresh zero-state and cross-article reset validation
# =============================================================================

BLOCK_6_FRESH_ZERO_INITIAL_STATES = True

BLOCK_6_INDEPENDENT_INITIAL_STATE_STORAGE = True

BLOCK_6_CROSS_ARTICLE_RESET_VALID = True


for pathway_name in BLOCK_6_EXPECTED_PATHWAYS:

    initial_states = [
        BLOCK_6_EXECUTION_A[
            "articles"
        ][
            article_id
        ][
            "pathways"
        ][
            pathway_name
        ][
            "initial_state"
        ]

        for article_id
        in BLOCK_6_ARTICLE_IDS
    ]


    for initial_state in initial_states:

        if not torch.equal(
            initial_state,
            torch.zeros_like(
                initial_state
            ),
        ):

            BLOCK_6_FRESH_ZERO_INITIAL_STATES = False


    storage_pointers = (
        BLOCK_6_EXECUTION_A[
            "initial_state_storage_pointers"
        ][
            pathway_name
        ]
    )


    if (
        len(
            storage_pointers
        )
        !=
        len(
            set(
                storage_pointers
            )
        )
    ):

        BLOCK_6_INDEPENDENT_INITIAL_STATE_STORAGE = False


    for article_index in range(
        1,
        len(
            BLOCK_6_ARTICLE_IDS
        ),
    ):

        preceding_article_id = (
            BLOCK_6_ARTICLE_IDS[
                article_index
                -
                1
            ]
        )


        current_article_id = (
            BLOCK_6_ARTICLE_IDS[
                article_index
            ]
        )


        preceding_final_state = (
            BLOCK_6_EXECUTION_A[
                "articles"
            ][
                preceding_article_id
            ][
                "pathways"
            ][
                pathway_name
            ][
                "final_state"
            ]
        )


        current_initial_state = (
            BLOCK_6_EXECUTION_A[
                "articles"
            ][
                current_article_id
            ][
                "pathways"
            ][
                pathway_name
            ][
                "initial_state"
            ]
        )


        if not torch.equal(
            current_initial_state,
            torch.zeros_like(
                current_initial_state
            ),
        ):

            BLOCK_6_CROSS_ARTICLE_RESET_VALID = False


        # This verifies that the reset state is not a storage alias of the
        # preceding article's final state, even if values were ever equal.
        if (
            current_initial_state.data_ptr()
            ==
            preceding_final_state.data_ptr()
        ):

            BLOCK_6_CROSS_ARTICLE_RESET_VALID = False


# =============================================================================
# Pathway and recurrent-step validation
# =============================================================================

BLOCK_6_PATHWAY_AUDIT = {
    pathway_name:
        {
            "candidate_shapes_valid":
                True,

            "candidate_finite":
                True,

            "weighted_candidate_finite":
                True,

            "state_shapes_valid":
                True,

            "state_finite":
                True,

            "effective_weight_stable":
                True,

            "deferred_contribution_zero":
                True,

            "deferred_state_zero":
                True,

            "additive_update":
                True,

            "cumulative_identity":
                True,

            "state_chain_valid":
                True,

            "representation_shapes_valid":
                True,
        }

    for pathway_name
    in BLOCK_6_EXPECTED_PATHWAYS
}


for article_id in BLOCK_6_ARTICLE_IDS:

    for pathway_name in BLOCK_6_EXPECTED_PATHWAYS:

        pathway_dim = int(
            NOTEBOOK_10_PATHWAY_DIMENSIONS[
                pathway_name
            ]
        )


        pathway_result = (
            BLOCK_6_EXECUTION_A[
                "articles"
            ][
                article_id
            ][
                "pathways"
            ][
                pathway_name
            ]
        )


        representation_matrix = (
            pathway_result[
                "representation_matrix"
            ]
        )


        if tuple(
            representation_matrix.shape
        ) != (
            len(
                pathway_result[
                    "steps"
                ]
            ),
            pathway_dim,
        ):

            BLOCK_6_PATHWAY_AUDIT[
                pathway_name
            ][
                "representation_shapes_valid"
            ] = False


        expected_preceding_state = (
            pathway_result[
                "initial_state"
            ]
        )


        for step in pathway_result[
            "steps"
        ]:

            candidate = (
                step[
                    "candidate_transformation"
                ]
            )


            weighted_candidate = (
                step[
                    "weighted_candidate"
                ]
            )


            recurrent_state = (
                step[
                    "recurrent_state"
                ]
            )


            preceding_state = (
                step[
                    "preceding_state"
                ]
            )


            effective_weight = (
                step[
                    "effective_weight"
                ]
            )


            if tuple(
                candidate.shape
            ) != (
                1,
                pathway_dim,
            ):

                BLOCK_6_PATHWAY_AUDIT[
                    pathway_name
                ][
                    "candidate_shapes_valid"
                ] = False


            if tuple(
                recurrent_state.shape
            ) != (
                1,
                pathway_dim,
            ):

                BLOCK_6_PATHWAY_AUDIT[
                    pathway_name
                ][
                    "state_shapes_valid"
                ] = False


            if not step[
                "candidate_finite"
            ]:

                BLOCK_6_PATHWAY_AUDIT[
                    pathway_name
                ][
                    "candidate_finite"
                ] = False


            if not step[
                "weighted_candidate_finite"
            ]:

                BLOCK_6_PATHWAY_AUDIT[
                    pathway_name
                ][
                    "weighted_candidate_finite"
                ] = False


            if not step[
                "state_finite"
            ]:

                BLOCK_6_PATHWAY_AUDIT[
                    pathway_name
                ][
                    "state_finite"
                ] = False


            expected_weight = (
                BLOCK_6_EFFECTIVE_WEIGHTS[
                    pathway_name
                ]
                .detach()
                .cpu()
                .reshape(
                    1,
                    pathway_dim,
                )
            )


            if not torch.equal(
                effective_weight,
                expected_weight,
            ):

                BLOCK_6_PATHWAY_AUDIT[
                    pathway_name
                ][
                    "effective_weight_stable"
                ] = False


            if not step[
                "deferred_contribution_zero"
            ]:

                BLOCK_6_PATHWAY_AUDIT[
                    pathway_name
                ][
                    "deferred_contribution_zero"
                ] = False


            if not step[
                "deferred_state_zero"
            ]:

                BLOCK_6_PATHWAY_AUDIT[
                    pathway_name
                ][
                    "deferred_state_zero"
                ] = False


            if not step[
                "additive_identity"
            ]:

                BLOCK_6_PATHWAY_AUDIT[
                    pathway_name
                ][
                    "additive_update"
                ] = False


            if not step[
                "cumulative_identity"
            ]:

                BLOCK_6_PATHWAY_AUDIT[
                    pathway_name
                ][
                    "cumulative_identity"
                ] = False


            if not torch.equal(
                preceding_state,
                expected_preceding_state,
            ):

                BLOCK_6_PATHWAY_AUDIT[
                    pathway_name
                ][
                    "state_chain_valid"
                ] = False


            expected_preceding_state = (
                recurrent_state
            )


# =============================================================================
# Pathway independence validation
# =============================================================================

BLOCK_6_PATHWAYS_INDEPENDENT = all(
    [
        (
            set(
                NOTEBOOK_10_TRANSFORMATIVE_MODULES.keys()
            )
            ==
            set(
                BLOCK_6_EXPECTED_PATHWAYS
            )
        ),

        (
            set(
                NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES.keys()
            )
            ==
            set(
                BLOCK_6_EXPECTED_PATHWAYS
            )
        ),

        (
            len(
                {
                    id(
                        NOTEBOOK_10_TRANSFORMATIVE_MODULES[
                            pathway_name
                        ]
                    )

                    for pathway_name
                    in BLOCK_6_EXPECTED_PATHWAYS
                }
            )
            ==
            len(
                BLOCK_6_EXPECTED_PATHWAYS
            )
        ),

        (
            len(
                {
                    id(
                        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES[
                            pathway_name
                        ]
                    )

                    for pathway_name
                    in BLOCK_6_EXPECTED_PATHWAYS
                }
            )
            ==
            len(
                BLOCK_6_EXPECTED_PATHWAYS
            )
        ),
    ]
)


# =============================================================================
# Complete deterministic repeat validation
# =============================================================================

def block_6_execution_exact(
    execution_a,
    execution_b,
):

    if (
        execution_a[
            "article_order"
        ]
        !=
        execution_b[
            "article_order"
        ]
    ):

        return False


    if (
        execution_a[
            "sentence_count"
        ]
        !=
        execution_b[
            "sentence_count"
        ]
    ):

        return False


    for article_id in execution_a[
        "article_order"
    ]:

        article_a = (
            execution_a[
                "articles"
            ][
                article_id
            ]
        )


        article_b = (
            execution_b[
                "articles"
            ][
                article_id
            ]
        )


        if (
            article_a[
                "sentence_ids"
            ]
            !=
            article_b[
                "sentence_ids"
            ]
        ):

            return False


        if (
            article_a[
                "sequence_positions"
            ]
            !=
            article_b[
                "sequence_positions"
            ]
        ):

            return False


        for pathway_name in BLOCK_6_EXPECTED_PATHWAYS:

            pathway_a = (
                article_a[
                    "pathways"
                ][
                    pathway_name
                ]
            )


            pathway_b = (
                article_b[
                    "pathways"
                ][
                    pathway_name
                ]
            )


            for tensor_field in (
                "initial_state",
                "representation_matrix",
                "final_state",
                "cumulative_weighted_candidate",
            ):

                if not torch.equal(
                    pathway_a[
                        tensor_field
                    ],
                    pathway_b[
                        tensor_field
                    ],
                ):

                    return False


            if (
                len(
                    pathway_a[
                        "steps"
                    ]
                )
                !=
                len(
                    pathway_b[
                        "steps"
                    ]
                )
            ):

                return False


            for (
                step_a,
                step_b,
            ) in zip(
                pathway_a[
                    "steps"
                ],
                pathway_b[
                    "steps"
                ],
            ):

                if (
                    step_a[
                        "sentence_id"
                    ]
                    !=
                    step_b[
                        "sentence_id"
                    ]
                ):

                    return False


                if (
                    step_a[
                        "sequence_position"
                    ]
                    !=
                    step_b[
                        "sequence_position"
                    ]
                ):

                    return False


                for tensor_field in (
                    "current_representation",
                    "preceding_state",
                    "candidate_transformation",
                    "effective_weight",
                    "weighted_candidate",
                    "recurrent_state",
                ):

                    if not torch.equal(
                        step_a[
                            tensor_field
                        ],
                        step_b[
                            tensor_field
                        ],
                    ):

                        return False


    return True


BLOCK_6_COMPLETE_EXECUTION_DETERMINISTIC = (
    block_6_execution_exact(
        BLOCK_6_EXECUTION_A,
        BLOCK_6_EXECUTION_B,
    )
)


# =============================================================================
# Causal execution validation
# =============================================================================
#
# The execution helper iterates sentence_position = 0..T-1 and passes only
# the current representation and immediately preceding recurrent state to the
# transformative mechanism. No future representation is supplied.
# =============================================================================

BLOCK_6_FUTURE_CONTEXT_USED = (
    NOTEBOOK_10_BLOCK_6_FUTURE_CONTEXT_USED
)


BLOCK_6_SUPERVISION_USED_AS_MODEL_INPUT = (
    NOTEBOOK_10_BLOCK_6_SUPERVISION_USED_AS_MODEL_INPUT
)


BLOCK_6_CAUSAL_CONTRACT_VALID = all(
    [
        BLOCK_6_CANONICAL_SEQUENCE_VALID,
        BLOCK_6_FRESH_ZERO_INITIAL_STATES,
        BLOCK_6_CROSS_ARTICLE_RESET_VALID,
        not
        BLOCK_6_FUTURE_CONTEXT_USED,
        not
        BLOCK_6_SUPERVISION_USED_AS_MODEL_INPUT,
    ]
)


# =============================================================================
# Model-state preservation after execution
# =============================================================================

BLOCK_6_REPRESENTATION_STATE_AFTER = (
    block_6_snapshot_module_collection(
        NOTEBOOK_10_REPRESENTATION_MODULES
    )
)


BLOCK_6_TRANSFORMATIVE_STATE_AFTER = (
    block_6_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_MODULES
    )
)


BLOCK_6_WEIGHT_STATE_AFTER = (
    block_6_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES
    )
)


BLOCK_6_REPRESENTATION_STATE_UNCHANGED = (
    block_6_nested_state_exact(
        BLOCK_6_REPRESENTATION_STATE_BASELINE,
        BLOCK_6_REPRESENTATION_STATE_AFTER,
    )
)


BLOCK_6_TRANSFORMATIVE_STATE_UNCHANGED = (
    block_6_nested_state_exact(
        BLOCK_6_TRANSFORMATIVE_STATE_BASELINE,
        BLOCK_6_TRANSFORMATIVE_STATE_AFTER,
    )
)


BLOCK_6_WEIGHT_STATE_UNCHANGED = (
    block_6_nested_state_exact(
        BLOCK_6_WEIGHT_STATE_BASELINE,
        BLOCK_6_WEIGHT_STATE_AFTER,
    )
)


with torch.inference_mode():

    BLOCK_6_EFFECTIVE_WEIGHT_IDENTITY_UNCHANGED = all(
        torch.equal(
            NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES[
                pathway_name
            ]()
            .detach()
            .cpu(),

            NOTEBOOK_10_INHERITED_EFFECTIVE_TRANSFORMATIVE_WEIGHTS[
                pathway_name
            ]
            .detach()
            .cpu(),
        )

        for pathway_name
        in BLOCK_6_EXPECTED_PATHWAYS
    )


BLOCK_6_ALL_PARAMETERS_FROZEN_AFTER = all(
    not parameter.requires_grad

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


BLOCK_6_ALL_MODULES_EVAL_AFTER = all(
    not module.training

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()
)


BLOCK_6_ALL_GRADIENTS_ABSENT_AFTER = all(
    parameter.grad is None

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


# =============================================================================
# Execution-boundary validation
# =============================================================================

BLOCK_6_EXECUTION_BOUNDARY_VALID = all(
    [
        NOTEBOOK_10_BLOCK_6_MEDIA_AI_REPRESENTATION_FORWARD_EXECUTED
        is True,

        NOTEBOOK_10_BLOCK_6_TRANSFORMATIVE_FORWARD_EXECUTED
        is True,

        NOTEBOOK_10_BLOCK_6_WEIGHT_FORWARD_EXECUTED
        is True,

        NOTEBOOK_10_BLOCK_6_RECURRENT_STATE_CREATED
        is True,

        NOTEBOOK_10_BLOCK_6_RECURRENT_UPDATE_EXECUTED
        is True,

        NOTEBOOK_10_BLOCK_6_ARTICLE_TRAJECTORIES_EXECUTED
        is True,

        NOTEBOOK_10_BLOCK_6_REPEAT_EXECUTION_PERFORMED
        is True,

        NOTEBOOK_10_BLOCK_6_FUTURE_CONTEXT_USED
        is False,

        NOTEBOOK_10_BLOCK_6_SUPERVISION_USED_AS_MODEL_INPUT
        is False,

        NOTEBOOK_10_BLOCK_6_LOSS_CALCULATED
        is False,

        NOTEBOOK_10_BLOCK_6_OPTIMIZER_CREATED
        is False,

        NOTEBOOK_10_BLOCK_6_ZERO_GRAD_EXECUTED
        is False,

        NOTEBOOK_10_BLOCK_6_BACKWARD_PASS_EXECUTED
        is False,

        NOTEBOOK_10_BLOCK_6_GRADIENT_CLIPPING_EXECUTED
        is False,

        NOTEBOOK_10_BLOCK_6_OPTIMIZER_STEP_EXECUTED
        is False,

        NOTEBOOK_10_BLOCK_6_PARAMETER_UPDATE_EXECUTED
        is False,

        NOTEBOOK_10_BLOCK_6_TRAINING_EXECUTED
        is False,
    ]
)


# =============================================================================
# Final pathway validity
# =============================================================================

BLOCK_6_PATHWAY_VALID = {
    pathway_name:
        all(
            BLOCK_6_PATHWAY_AUDIT[
                pathway_name
            ].values()
        )

    for pathway_name
    in BLOCK_6_EXPECTED_PATHWAYS
}


BLOCK_6_ALL_PATHWAYS_VALID = all(
    BLOCK_6_PATHWAY_VALID.values()
)


# =============================================================================
# Final block validation
# =============================================================================

NOTEBOOK_10_BLOCK_6_ERRORS = []


BLOCK_6_VALIDATION_CHECKS = {
    "prerequisites_invalid":
        BLOCK_6_PREREQUISITES_VALID,

    "device_dtype_inconsistent":
        BLOCK_6_DEVICE_DTYPE_CONSISTENT,

    "parameter_accounting_invalid":
        BLOCK_6_PARAMETER_ACCOUNTING_VALID,

    "article_ids_invalid":
        BLOCK_6_ARTICLE_IDS_VALID,

    "canonical_sequence_invalid":
        BLOCK_6_CANONICAL_SEQUENCE_VALID,

    "contextual_matrices_invalid":
        BLOCK_6_CONTEXTUAL_MATRICES_VALID,

    "sentence_count_invalid":
        BLOCK_6_SENTENCE_COUNT_VALID,

    "article_count_invalid":
        BLOCK_6_ARTICLE_COUNT_VALID,

    "article_order_invalid":
        BLOCK_6_ARTICLE_ORDER_VALID,

    "fresh_zero_initial_states_invalid":
        BLOCK_6_FRESH_ZERO_INITIAL_STATES,

    "initial_state_storage_not_independent":
        BLOCK_6_INDEPENDENT_INITIAL_STATE_STORAGE,

    "cross_article_reset_invalid":
        BLOCK_6_CROSS_ARTICLE_RESET_VALID,

    "pathways_not_independent":
        BLOCK_6_PATHWAYS_INDEPENDENT,

    "effective_weights_invalid":
        BLOCK_6_EFFECTIVE_WEIGHTS_VALID,

    "pathway_validation_failed":
        BLOCK_6_ALL_PATHWAYS_VALID,

    "complete_execution_not_deterministic":
        BLOCK_6_COMPLETE_EXECUTION_DETERMINISTIC,

    "causal_contract_invalid":
        BLOCK_6_CAUSAL_CONTRACT_VALID,

    "representation_state_changed":
        BLOCK_6_REPRESENTATION_STATE_UNCHANGED,

    "transformative_state_changed":
        BLOCK_6_TRANSFORMATIVE_STATE_UNCHANGED,

    "weight_state_changed":
        BLOCK_6_WEIGHT_STATE_UNCHANGED,

    "effective_weight_identity_changed":
        BLOCK_6_EFFECTIVE_WEIGHT_IDENTITY_UNCHANGED,

    "parameters_not_frozen":
        BLOCK_6_ALL_PARAMETERS_FROZEN_AFTER,

    "modules_not_eval":
        BLOCK_6_ALL_MODULES_EVAL_AFTER,

    "gradients_present":
        BLOCK_6_ALL_GRADIENTS_ABSENT_AFTER,

    "execution_boundary_invalid":
        BLOCK_6_EXECUTION_BOUNDARY_VALID,
}


for (
    error_name,
    condition,
) in BLOCK_6_VALIDATION_CHECKS.items():

    if not condition:

        NOTEBOOK_10_BLOCK_6_ERRORS.append(
            error_name
        )


NOTEBOOK_10_BLOCK_6_VALID = (
    len(
        NOTEBOOK_10_BLOCK_6_ERRORS
    )
    ==
    0
)


if not NOTEBOOK_10_BLOCK_6_VALID:

    raise RuntimeError(
        "Notebook 10 Block 6 controlled multi-article recurrent "
        "execution failed. "
        f"Errors: {NOTEBOOK_10_BLOCK_6_ERRORS}"
    )


NOTEBOOK_10_BLOCK_6_COMPLETE = True


# =============================================================================
# Stable downstream globals for Block 7
# =============================================================================

NOTEBOOK_10_MULTI_ARTICLE_RECURRENT_TRAJECTORIES = deepcopy(
    BLOCK_6_EXECUTION_A[
        "articles"
    ]
)


NOTEBOOK_10_MULTI_ARTICLE_PATHWAY_REPRESENTATIONS = {
    article_id:
        {
            pathway_name:
                BLOCK_6_EXECUTION_A[
                    "articles"
                ][
                    article_id
                ][
                    "pathways"
                ][
                    pathway_name
                ][
                    "representation_matrix"
                ]
                .clone()

            for pathway_name
            in BLOCK_6_EXPECTED_PATHWAYS
        }

    for article_id
    in BLOCK_6_ARTICLE_IDS
}


NOTEBOOK_10_MULTI_ARTICLE_FINAL_RECURRENT_STATES = {
    article_id:
        {
            pathway_name:
                BLOCK_6_EXECUTION_A[
                    "articles"
                ][
                    article_id
                ][
                    "pathways"
                ][
                    pathway_name
                ][
                    "final_state"
                ]
                .clone()

            for pathway_name
            in BLOCK_6_EXPECTED_PATHWAYS
        }

    for article_id
    in BLOCK_6_ARTICLE_IDS
}


NOTEBOOK_10_MULTI_ARTICLE_RECURRENT_VALIDATION_VALID = (
    NOTEBOOK_10_BLOCK_6_VALID
)


NOTEBOOK_10_EXPANDED_SUPERVISION_ALIGNMENT_READY = (
    NOTEBOOK_10_BLOCK_6_VALID
)


# =============================================================================
# Block summary
# =============================================================================

NOTEBOOK_10_BLOCK_6_SUMMARY = {
    "block":
        NOTEBOOK_10_BLOCK_6,

    "block_name":
        NOTEBOOK_10_BLOCK_6_NAME,

    "block_version":
        NOTEBOOK_10_BLOCK_6_VERSION,

    "article_count":
        BLOCK_6_ARTICLES_TRAVERSED,

    "sentence_count":
        BLOCK_6_SENTENCES_TRAVERSED,

    "pathways":
        tuple(
            BLOCK_6_EXPECTED_PATHWAYS
        ),

    "fresh_zero_initial_states":
        BLOCK_6_FRESH_ZERO_INITIAL_STATES,

    "independent_initial_state_storage":
        BLOCK_6_INDEPENDENT_INITIAL_STATE_STORAGE,

    "cross_article_reset_valid":
        BLOCK_6_CROSS_ARTICLE_RESET_VALID,

    "pathways_independent":
        BLOCK_6_PATHWAYS_INDEPENDENT,

    "complete_execution_deterministic":
        BLOCK_6_COMPLETE_EXECUTION_DETERMINISTIC,

    "causal_contract_valid":
        BLOCK_6_CAUSAL_CONTRACT_VALID,

    "representation_state_unchanged":
        BLOCK_6_REPRESENTATION_STATE_UNCHANGED,

    "transformative_state_unchanged":
        BLOCK_6_TRANSFORMATIVE_STATE_UNCHANGED,

    "weight_state_unchanged":
        BLOCK_6_WEIGHT_STATE_UNCHANGED,

    "effective_weight_identity_unchanged":
        BLOCK_6_EFFECTIVE_WEIGHT_IDENTITY_UNCHANGED,

    "all_parameters_frozen":
        BLOCK_6_ALL_PARAMETERS_FROZEN_AFTER,

    "all_gradients_absent":
        BLOCK_6_ALL_GRADIENTS_ABSENT_AFTER,

    "block_valid":
        NOTEBOOK_10_BLOCK_6_VALID,

    "block_complete":
        NOTEBOOK_10_BLOCK_6_COMPLETE,
}


# =============================================================================
# Final report
# =============================================================================

print("=" * 72)

print(
    "Media AI — Notebook 10, Block 6: "
    "Controlled Multi-Article Recurrent Execution "
    "and Article-Boundary Reset Validation"
)

print("=" * 72)

print(
    f"Block version                : "
    f"{NOTEBOOK_10_BLOCK_6_VERSION}"
)

print(
    f"Execution device             : "
    f"{BLOCK_6_EXECUTION_DEVICE}"
)

print(
    f"Execution dtype              : "
    f"{BLOCK_6_EXECUTION_DTYPE}"
)

print("-" * 72)

print(
    "Canonical corpus traversal"
)

print(
    f"Articles traversed           : "
    f"{BLOCK_6_ARTICLES_TRAVERSED}"
)

print(
    f"Sentences traversed          : "
    f"{BLOCK_6_SENTENCES_TRAVERSED}"
)

print(
    f"Article IDs valid            : "
    f"{BLOCK_6_ARTICLE_IDS_VALID}"
)

print(
    f"Canonical order preserved    : "
    f"{BLOCK_6_CANONICAL_SEQUENCE_VALID}"
)

print(
    f"Contextual matrices valid    : "
    f"{BLOCK_6_CONTEXTUAL_MATRICES_VALID}"
)

print("-" * 72)

print(
    "Article-boundary validation"
)

print(
    f"Fresh zero initial states    : "
    f"{BLOCK_6_FRESH_ZERO_INITIAL_STATES}"
)

print(
    f"Independent reset storage    : "
    f"{BLOCK_6_INDEPENDENT_INITIAL_STATE_STORAGE}"
)

print(
    f"Cross-article reset valid    : "
    f"{BLOCK_6_CROSS_ARTICLE_RESET_VALID}"
)

print(
    f"Future context used          : "
    f"{BLOCK_6_FUTURE_CONTEXT_USED}"
)

print(
    f"Supervision used as input    : "
    f"{BLOCK_6_SUPERVISION_USED_AS_MODEL_INPUT}"
)

print("-" * 72)

print(
    "Pathway recurrent validation"
)

for pathway_name in BLOCK_6_EXPECTED_PATHWAYS:

    audit = (
        BLOCK_6_PATHWAY_AUDIT[
            pathway_name
        ]
    )


    print(
        f"{pathway_name:<14} dimension             : "
        f"{NOTEBOOK_10_PATHWAY_DIMENSIONS[pathway_name]}"
    )

    print(
        f"{pathway_name:<14} active dimensions     : "
        f"{list(NOTEBOOK_10_ACTIVE_DIMENSION_INDICES[pathway_name])}"
    )

    print(
        f"{pathway_name:<14} deferred dimensions   : "
        f"{list(NOTEBOOK_10_DEFERRED_DIMENSION_INDICES[pathway_name])}"
    )

    print(
        f"{pathway_name:<14} representation shapes : "
        f"{audit['representation_shapes_valid']}"
    )

    print(
        f"{pathway_name:<14} candidate shapes      : "
        f"{audit['candidate_shapes_valid']}"
    )

    print(
        f"{pathway_name:<14} candidate finite      : "
        f"{audit['candidate_finite']}"
    )

    print(
        f"{pathway_name:<14} weighted finite       : "
        f"{audit['weighted_candidate_finite']}"
    )

    print(
        f"{pathway_name:<14} state shapes          : "
        f"{audit['state_shapes_valid']}"
    )

    print(
        f"{pathway_name:<14} states finite         : "
        f"{audit['state_finite']}"
    )

    print(
        f"{pathway_name:<14} weight stable         : "
        f"{audit['effective_weight_stable']}"
    )

    print(
        f"{pathway_name:<14} deferred contrib. zero: "
        f"{audit['deferred_contribution_zero']}"
    )

    print(
        f"{pathway_name:<14} deferred state zero   : "
        f"{audit['deferred_state_zero']}"
    )

    print(
        f"{pathway_name:<14} state chain valid     : "
        f"{audit['state_chain_valid']}"
    )

    print(
        f"{pathway_name:<14} additive update       : "
        f"{audit['additive_update']}"
    )

    print(
        f"{pathway_name:<14} cumulative identity   : "
        f"{audit['cumulative_identity']}"
    )


print("-" * 72)

print(
    "Deterministic corpus validation"
)

print(
    f"Pathways independent         : "
    f"{BLOCK_6_PATHWAYS_INDEPENDENT}"
)

print(
    f"Repeated execution performed : "
    f"{NOTEBOOK_10_BLOCK_6_REPEAT_EXECUTION_PERFORMED}"
)

print(
    f"Complete execution deterministic: "
    f"{BLOCK_6_COMPLETE_EXECUTION_DETERMINISTIC}"
)

print(
    f"Causal contract valid        : "
    f"{BLOCK_6_CAUSAL_CONTRACT_VALID}"
)

print("-" * 72)

print(
    "Inherited Media AI state"
)

print(
    f"Representation parameters    : "
    f"{BLOCK_6_REPRESENTATION_PARAMETER_COUNT:,}"
)

print(
    f"Transformative parameters    : "
    f"{BLOCK_6_TRANSFORMATIVE_PARAMETER_COUNT:,}"
)

print(
    f"Transformative Weight params : "
    f"{BLOCK_6_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT:,}"
)

print(
    f"Total model parameters       : "
    f"{BLOCK_6_TOTAL_PARAMETER_COUNT:,}"
)

print(
    f"Representation unchanged     : "
    f"{BLOCK_6_REPRESENTATION_STATE_UNCHANGED}"
)

print(
    f"Transformative unchanged     : "
    f"{BLOCK_6_TRANSFORMATIVE_STATE_UNCHANGED}"
)

print(
    f"Transform Weight unchanged   : "
    f"{BLOCK_6_WEIGHT_STATE_UNCHANGED}"
)

print(
    f"Effective Weight identity    : "
    f"{BLOCK_6_EFFECTIVE_WEIGHT_IDENTITY_UNCHANGED}"
)

print(
    f"All parameters frozen        : "
    f"{BLOCK_6_ALL_PARAMETERS_FROZEN_AFTER}"
)

print(
    f"All modules eval             : "
    f"{BLOCK_6_ALL_MODULES_EVAL_AFTER}"
)

print(
    f"All gradients absent         : "
    f"{BLOCK_6_ALL_GRADIENTS_ABSENT_AFTER}"
)

print("-" * 72)

print(
    "Execution boundary"
)

print(
    f"Representation forward       : "
    f"{NOTEBOOK_10_BLOCK_6_MEDIA_AI_REPRESENTATION_FORWARD_EXECUTED}"
)

print(
    f"Transform forward            : "
    f"{NOTEBOOK_10_BLOCK_6_TRANSFORMATIVE_FORWARD_EXECUTED}"
)

print(
    f"Weight forward               : "
    f"{NOTEBOOK_10_BLOCK_6_WEIGHT_FORWARD_EXECUTED}"
)

print(
    f"Recurrent state created      : "
    f"{NOTEBOOK_10_BLOCK_6_RECURRENT_STATE_CREATED}"
)

print(
    f"Recurrent update executed    : "
    f"{NOTEBOOK_10_BLOCK_6_RECURRENT_UPDATE_EXECUTED}"
)

print(
    f"Article trajectories executed: "
    f"{NOTEBOOK_10_BLOCK_6_ARTICLE_TRAJECTORIES_EXECUTED}"
)

print(
    f"Loss calculated              : "
    f"{NOTEBOOK_10_BLOCK_6_LOSS_CALCULATED}"
)

print(
    f"Optimizer created            : "
    f"{NOTEBOOK_10_BLOCK_6_OPTIMIZER_CREATED}"
)

print(
    f"Backward pass executed       : "
    f"{NOTEBOOK_10_BLOCK_6_BACKWARD_PASS_EXECUTED}"
)

print(
    f"Parameter update executed    : "
    f"{NOTEBOOK_10_BLOCK_6_PARAMETER_UPDATE_EXECUTED}"
)

print(
    f"Training executed            : "
    f"{NOTEBOOK_10_BLOCK_6_TRAINING_EXECUTED}"
)

print("-" * 72)

print(
    f"Multi-article recurrent valid: "
    f"{NOTEBOOK_10_MULTI_ARTICLE_RECURRENT_VALIDATION_VALID}"
)

print(
    f"Block valid                  : "
    f"{NOTEBOOK_10_BLOCK_6_VALID}"
)

print(
    f"Block complete               : "
    f"{NOTEBOOK_10_BLOCK_6_COMPLETE}"
)

print("=" * 72)


# =============================================================================
# Completion interpretation
# =============================================================================

print(
    f"The complete validated corpus of {BLOCK_6_ARTICLES_TRAVERSED} "
    f"article(s) and {BLOCK_6_SENTENCES_TRAVERSED} canonical sentence(s) "
    "was traversed successfully through the frozen Media AI architecture."
)

print(
    "Every article began from independently constructed deterministic "
    "zero recurrent states in the factual, psychological and social pathways."
)

print(
    "No recurrent state was propagated across article boundaries, and each "
    "sentence transition consumed only the current pathway representation "
    "and the immediately preceding pathway-specific recurrent state."
)

print(
    "Candidate transformations, Transformative-Weight-gated contributions "
    "and recurrent states preserved the required pathway dimensions and "
    "remained finite throughout execution."
)

print(
    "Deferred Transformative-Weight dimensions produced zero recurrent "
    "increment and remained zero under the current additive recurrent contract."
)

print(
    "Repeated complete corpus execution reproduced the same candidate "
    "transformations, weighted contributions and recurrent states exactly."
)

print(
    f"The inherited {BLOCK_6_TOTAL_PARAMETER_COUNT:,}-parameter Media AI "
    "architecture remained frozen and exactly unchanged."
)

print(
    "No supervision target, loss, optimiser, backward pass, gradient clipping, "
    "parameter update or training operation was introduced."
)

print(
    "Notebook 10 may now proceed to Block 7 — Expanded Supervision Alignment "
    "and Transformation-Target Construction."
)

print("=" * 72)
```

#Block 7: Expanded Supervision Alignment and Transformation-Target Construction

## Purpose

This block aligns the validated Media AI supervision artefacts with the canonical multi-article corpus established in Blocks 2–5 and structurally traversed through the recurrent architecture in Block 6.

The objective is to construct the supervision tensors and transformation targets required for subsequent expanded recurrent evaluation and controlled optimisation without modifying the inherited Media AI architecture.

Block 7 therefore introduces **supervision into the Notebook 10 execution path**, but only as an external evaluation and optimisation target.

Supervision is never supplied as an input to the representation model, transformative mechanisms, Transformative-Weight modules or recurrent state.

---

## Methodological Role

Notebook 09 demonstrated that the Transformative-Weight mechanism is architecturally learnable under a controlled pilot setting.

Notebook 10 now requires a formally aligned supervision layer before that result can be evaluated beyond the original architectural training procedure.

The validated corpus currently contains:

- **3 independent articles**;
- **54 canonical sentences**;
- persisted **768-dimensional contextual sentence representations**;
- factual, psychological and social Media AI representation pathways;
- frozen transformative mechanisms;
- learned and frozen Transformative Weights;
- validated article-specific recurrent trajectories.

Block 7 connects these canonical sentence identities to the previously validated supervision artefacts while preserving all article boundaries and sentence ordering established upstream.

No supervision value may be inferred merely because a sentence exists in the corpus.

Missing supervision remains explicitly unavailable and is not interpreted as zero.

---

## Supervision Spaces

Supervision is reconstructed independently for the three Media AI representational spaces:

**Factual**

$$
\mathbf{y}^{(F)}_{a,t} \in \mathbb{R}^{10}
$$

**Psychological**

$$
\mathbf{y}^{(P)}_{a,t} \in \mathbb{R}^{34}
$$

**Social**

$$
\mathbf{y}^{(S)}_{a,t} \in \mathbb{R}^{7}
$$

where:

- $a$ identifies the article;
- $t$ identifies the canonical sentence position within that article;
- each vector dimension retains the semantic ordering defined by its validated upstream supervision contract.

The dimensional contracts must reproduce the architecture inherited from Notebook 09 exactly.

---

## Canonical Supervision Alignment

Every supervision record must be aligned to the canonical Notebook 10 corpus through persistent article and sentence identity.

For every pathway $p$ and sentence $(a,t)$, Block 7 constructs:

$$
\mathbf{y}^{(p)}_{a,t}
$$

together with an availability mask:

$$
\mathbf{m}^{(p)}_{a,t} \in \{0,1\}^{d_p}
$$

where $d_p$ is the dimensionality of pathway $p$.

For dimension $j$:

$$
m^{(p)}_{a,t,j}=1
$$

only when a validated supervision value is genuinely available for that sentence and dimension.

Otherwise:

$$
m^{(p)}_{a,t,j}=0
$$

A missing value must therefore remain distinguishable from a genuine supervised value of zero.

---

## Transformation-Target Construction

The transformative mechanism models change relative to the immediately preceding supervised state within the same article.

For each pathway $p$, the signed transformation target is therefore defined as:

$$
\Delta\mathbf{y}^{(p)}_{a,t}
=
\mathbf{y}^{(p)}_{a,t}
-
\mathbf{y}^{(p)}_{a,t-1}
$$

For the first canonical sentence of every article, the preceding supervised state follows the deterministic zero-boundary contract:

$$
\mathbf{y}^{(p)}_{a,-1}
=
\mathbf{0}
$$

and therefore:

$$
\Delta\mathbf{y}^{(p)}_{a,0}
=
\mathbf{y}^{(p)}_{a,0}
$$

This boundary condition mirrors the recurrent architecture validated in Block 6, where every article begins from an independently constructed zero recurrent state.

No transformation target may depend on a sentence belonging to another article.

---

## Mask-Aware Transformation Availability

A transformation target is valid only when the supervision required to define that transformation is available.

For non-initial sentence positions:

$$
\mathbf{m}^{(p,\Delta)}_{a,t}
=
\mathbf{m}^{(p)}_{a,t}
\odot
\mathbf{m}^{(p)}_{a,t-1}
$$

where $\odot$ denotes element-wise multiplication.

For the first sentence of an article, the deterministic zero boundary requires only the current supervision value:

$$
\mathbf{m}^{(p,\Delta)}_{a,0}
=
\mathbf{m}^{(p)}_{a,0}
$$

Consequently, unavailable supervision never enters a transformation objective implicitly.

The distinction between **missing**, **zero**, **positive transformation** and **negative transformation** is preserved.

---

## Transformative-Weight Eligibility

Notebook 09 established that not every structural dimension is currently activation-eligible.

The transformation-objective mask must therefore combine supervision availability with the inherited active Transformative-Weight contract.

For pathway $p$:

$$
\mathbf{m}^{(p,\mathrm{obj})}_{a,t}
=
\mathbf{m}^{(p,\Delta)}_{a,t}
\odot
\mathbf{g}^{(p)}
$$

where:

$$
g^{(p)}_j =
\begin{cases}
1, & j \text{ is activation-eligible} \\
0, & j \text{ is deferred}
\end{cases}
$$

Under the inherited pilot architecture:

- all **10 factual dimensions** are active;
- all **34 psychological dimensions** are active;
- **5 of 7 social dimensions** are active;
- social dimensions **1 and 3** remain deferred.

Deferred dimensions remain structurally present but must not contribute to the transformation objective.

---

## Causal Integrity

Supervision alignment must preserve the causal architecture already validated in Block 6.

For every article:

$$
\Delta\mathbf{y}_{a,t}
=
f\left(
\mathbf{y}_{a,t},
\mathbf{y}_{a,t-1}
\right)
$$

and must never use:

$$
\mathbf{y}_{a,t+1}
$$

or any other future sentence.

The following constraints therefore remain mandatory:

- canonical within-article sentence order is preserved;
- article boundaries remain explicit;
- no cross-article transformation target is constructed;
- no future supervision is used;
- supervision is never supplied as model input;
- no batch shuffling is introduced.

---

## Alignment Audit

Before Block 7 may be considered complete, it must report independently for each pathway:

- supervision tensor shape;
- total structural supervision elements;
- available supervision elements;
- missing supervision elements;
- transformation-target availability;
- objective-active transformation elements;
- active dimensions;
- deferred dimensions;
- article coverage;
- sentence coverage;
- alignment failures, if any.

This audit is required because structural corpus readiness and supervision readiness are distinct properties.

A sentence possessing a valid contextual representation does not imply that every factual, psychological or social supervision dimension is available.

---

## Architectural Preservation

Block 7 is a target-construction block.

The complete inherited Media AI architecture must therefore remain unchanged:

$$
894{,}003
+
8{,}187
+
49
=
902{,}239
$$

parameters.

The following conditions must remain true:

- representation parameters remain frozen;
- transformative parameters remain frozen;
- Transformative-Weight parameters remain frozen;
- effective Transformative Weights remain identical to the Notebook 09 handover;
- all model modules remain in evaluation mode;
- no parameter gradients are created;
- no recurrent model state is persisted as learned state.

---

## Execution Boundary

Block 7 **may**:

- recover validated supervision artefacts;
- align supervision to canonical sentence identities;
- construct pathway-specific supervision tensors;
- construct availability masks;
- derive signed transformation targets;
- construct transformation-target masks;
- apply inherited active/deferred dimensional eligibility;
- audit supervision and transformation coverage.

Block 7 **must not**:

- retrain the representation model;
- retrain the transformative mechanisms;
- update Transformative Weights;
- use supervision as model input;
- execute an optimisation step;
- create a training optimiser;
- execute backward propagation;
- clip gradients;
- update any model parameter;
- claim expanded-model performance.

---

## Expected Outcome

At successful completion, Notebook 10 should possess a canonical, article-aware and mask-aware supervision structure aligned exactly to the validated multi-article corpus.

The resulting transformation targets should provide the formal reference against which the recurrent Media AI trajectories can subsequently be evaluated.

Successful completion of this block establishes:

> **canonical corpus → validated supervision → causal transformation targets → objective-eligible masks**

while preserving the frozen architecture and the causal recurrent contract.

Block 8 may then proceed to **expanded recurrent objective evaluation and pre-training baseline construction**, before any decision is made about further controlled training.

```python
# =============================================================================
# Media AI — Notebook 10
# Block 7: Expanded Supervision Alignment
#          and Transformation-Target Construction
# =============================================================================

from copy import deepcopy
from datetime import datetime, timezone

import io
import json

import numpy as np
import torch

from googleapiclient.http import MediaIoBaseDownload


# =============================================================================
# Block identity
# =============================================================================

NOTEBOOK_10_BLOCK_7 = 7

NOTEBOOK_10_BLOCK_7_NAME = (
    "Expanded Supervision Alignment "
    "and Transformation-Target Construction"
)

NOTEBOOK_10_BLOCK_7_VERSION = "1.1"

BLOCK_7_EXECUTED_AT_UTC = (
    datetime.now(
        timezone.utc
    ).isoformat()
)


# =============================================================================
# Canonical persisted supervision sources
# =============================================================================
#
# These are the same validated supervision artefacts used by the completed
# Notebook 08 / Notebook 09 transformative-learning path.
#
# Direct Drive file IDs are used deliberately. Block 7 therefore does not
# perform ambiguous filename discovery and does not silently substitute a
# different supervision artefact.
# =============================================================================

BLOCK_7_FACTUAL_FILENAME = (
    "notebook_06_factual_target_contract.json"
)

BLOCK_7_FACTUAL_DRIVE_FILE_ID = (
    "1FdTrV_2RoHDRn06JLSx367OD4HffDaAy"
)

BLOCK_7_PSYCHOLOGICAL_SOURCE_NAME = (
    "Notebook 03 accepted supervision"
)

BLOCK_7_PSYCHOLOGICAL_DRIVE_FILE_ID = (
    "1gYiCR-i_avGCwE5O9p921es8Mzbzhvyl"
)

BLOCK_7_SOCIAL_FILENAME = (
    "notebook_07_accepted_social_supervision.json"
)

BLOCK_7_SOCIAL_DRIVE_FILE_ID = (
    "1RMofcYIqIFKH9FqSrAz5hjbicfSNu5ex"
)


# =============================================================================
# Required runtime contract
# =============================================================================
#
# Block 7 consumes the validated multi-article corpus from Block 5 and the
# successful frozen recurrent execution from Block 6.
#
# It intentionally does not rebuild Blocks 1–6.
# =============================================================================

BLOCK_7_REQUIRED_GLOBALS = [
    "NOTEBOOK_10_BLOCK_5_COMPLETE",
    "NOTEBOOK_10_BLOCK_5_VALID",
    "NOTEBOOK_10_BLOCK_6_COMPLETE",
    "NOTEBOOK_10_BLOCK_6_VALID",
    "NOTEBOOK_10_MULTI_ARTICLE_RECURRENT_VALIDATION_VALID",
    "NOTEBOOK_10_EXPANDED_SUPERVISION_ALIGNMENT_READY",
    "NOTEBOOK_10_VALIDATED_MULTI_ARTICLE_RECORDS",
    "NOTEBOOK_10_REPRESENTATION_READY_ARTICLE_IDS",
    "NOTEBOOK_10_REPRESENTATION_MODULES",
    "NOTEBOOK_10_TRANSFORMATIVE_MODULES",
    "NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES",
    "NOTEBOOK_10_PATHWAY_DIMENSIONS",
    "NOTEBOOK_10_ACTIVE_DIMENSION_INDICES",
    "NOTEBOOK_10_DEFERRED_DIMENSION_INDICES",
    "NOTEBOOK_10_INHERITED_EFFECTIVE_TRANSFORMATIVE_WEIGHTS",
    "NOTEBOOK_10_DRIVE_SERVICE",
]


BLOCK_7_MISSING_PREREQUISITES = [
    object_name
    for object_name in BLOCK_7_REQUIRED_GLOBALS
    if object_name not in globals()
]


if BLOCK_7_MISSING_PREREQUISITES:

    raise RuntimeError(
        "Notebook 10 Block 7 cannot execute because the current runtime "
        "does not contain the exact validated Block 6 → Block 7 handover "
        "contract. If the runtime was restarted, rerun Notebook 10 Blocks "
        "1–6 in order. Otherwise inspect the missing handover names below. "
        f"Missing objects: {BLOCK_7_MISSING_PREREQUISITES}"
    )


BLOCK_7_BLOCK_6_HANDOVER_VALID = all(
    [
        NOTEBOOK_10_BLOCK_6_COMPLETE is True,
        NOTEBOOK_10_BLOCK_6_VALID is True,
        NOTEBOOK_10_MULTI_ARTICLE_RECURRENT_VALIDATION_VALID is True,
        NOTEBOOK_10_EXPANDED_SUPERVISION_ALIGNMENT_READY is True,
    ]
)


BLOCK_7_PREREQUISITES_VALID = all(
    [
        NOTEBOOK_10_BLOCK_5_COMPLETE is True,
        NOTEBOOK_10_BLOCK_5_VALID is True,
        NOTEBOOK_10_BLOCK_6_COMPLETE is True,
        NOTEBOOK_10_BLOCK_6_VALID is True,
        NOTEBOOK_10_MULTI_ARTICLE_RECURRENT_VALIDATION_VALID is True,
        NOTEBOOK_10_EXPANDED_SUPERVISION_ALIGNMENT_READY is True,
        isinstance(
            NOTEBOOK_10_VALIDATED_MULTI_ARTICLE_RECORDS,
            dict,
        ),
        isinstance(
            NOTEBOOK_10_REPRESENTATION_MODULES,
            dict,
        ),
        isinstance(
            NOTEBOOK_10_TRANSFORMATIVE_MODULES,
            dict,
        ),
        isinstance(
            NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
            dict,
        ),
    ]
)


if not BLOCK_7_PREREQUISITES_VALID:

    raise RuntimeError(
        "Notebook 10 Block 7 requires valid completed Blocks 5 and 6 "
        "and the frozen Notebook 09 → 10 architecture."
    )


# =============================================================================
# Canonical pathway contract
# =============================================================================

BLOCK_7_PATHWAY_DIMENSIONS = {
    pathway_name:
        int(
            pathway_dim
        )
    for pathway_name, pathway_dim
    in NOTEBOOK_10_PATHWAY_DIMENSIONS.items()
}


BLOCK_7_EXPECTED_PATHWAY_DIMENSIONS = {
    "factual":
        10,
    "psychological":
        34,
    "social":
        7,
}


BLOCK_7_PATHWAY_DIMENSIONS_VALID = (
    BLOCK_7_PATHWAY_DIMENSIONS
    ==
    BLOCK_7_EXPECTED_PATHWAY_DIMENSIONS
)


if not BLOCK_7_PATHWAY_DIMENSIONS_VALID:

    raise RuntimeError(
        "Notebook 10 Block 7 pathway dimensions do not reproduce "
        "the validated Notebook 09 → 10 architecture."
    )


BLOCK_7_ACTIVE_DIMENSION_INDICES = {
    pathway_name:
        tuple(
            int(index)
            for index in indices
        )
    for pathway_name, indices
    in NOTEBOOK_10_ACTIVE_DIMENSION_INDICES.items()
}


BLOCK_7_DEFERRED_DIMENSION_INDICES = {
    pathway_name:
        tuple(
            int(index)
            for index in indices
        )
    for pathway_name, indices
    in NOTEBOOK_10_DEFERRED_DIMENSION_INDICES.items()
}


BLOCK_7_DIMENSION_PARTITIONS_VALID = {}


for pathway_name, pathway_dim in BLOCK_7_PATHWAY_DIMENSIONS.items():

    active_set = set(
        BLOCK_7_ACTIVE_DIMENSION_INDICES[
            pathway_name
        ]
    )

    deferred_set = set(
        BLOCK_7_DEFERRED_DIMENSION_INDICES[
            pathway_name
        ]
    )

    BLOCK_7_DIMENSION_PARTITIONS_VALID[
        pathway_name
    ] = all(
        [
            active_set.isdisjoint(
                deferred_set
            ),
            (
                active_set
                |
                deferred_set
            )
            ==
            set(
                range(
                    pathway_dim
                )
            ),
        ]
    )


BLOCK_7_ALL_DIMENSION_PARTITIONS_VALID = all(
    BLOCK_7_DIMENSION_PARTITIONS_VALID.values()
)


if not BLOCK_7_ALL_DIMENSION_PARTITIONS_VALID:

    raise RuntimeError(
        "Notebook 10 Block 7 active/deferred dimension partitions "
        "are invalid."
    )


# =============================================================================
# Execution boundary
# =============================================================================

NOTEBOOK_10_BLOCK_7_SUPERVISION_RECOVERED = False
NOTEBOOK_10_BLOCK_7_SUPERVISION_ALIGNED = False
NOTEBOOK_10_BLOCK_7_TRANSFORMATION_TARGETS_CONSTRUCTED = False
NOTEBOOK_10_BLOCK_7_OBJECTIVE_MASKS_CONSTRUCTED = False

NOTEBOOK_10_BLOCK_7_REPRESENTATION_FORWARD_EXECUTED = False
NOTEBOOK_10_BLOCK_7_TRANSFORMATIVE_FORWARD_EXECUTED = False
NOTEBOOK_10_BLOCK_7_WEIGHT_FORWARD_EXECUTED = False
NOTEBOOK_10_BLOCK_7_RECURRENT_STATE_CREATED = False
NOTEBOOK_10_BLOCK_7_RECURRENT_UPDATE_EXECUTED = False

NOTEBOOK_10_BLOCK_7_LOSS_CALCULATED = False
NOTEBOOK_10_BLOCK_7_OPTIMIZER_CREATED = False
NOTEBOOK_10_BLOCK_7_ZERO_GRAD_EXECUTED = False
NOTEBOOK_10_BLOCK_7_BACKWARD_PASS_EXECUTED = False
NOTEBOOK_10_BLOCK_7_GRADIENT_CLIPPING_EXECUTED = False
NOTEBOOK_10_BLOCK_7_OPTIMIZER_STEP_EXECUTED = False
NOTEBOOK_10_BLOCK_7_PARAMETER_UPDATE_EXECUTED = False
NOTEBOOK_10_BLOCK_7_TRAINING_EXECUTED = False

NOTEBOOK_10_BLOCK_7_FUTURE_CONTEXT_USED = False
NOTEBOOK_10_BLOCK_7_SUPERVISION_USED_AS_MODEL_INPUT = False


# =============================================================================
# Model-state snapshot helpers
# =============================================================================

def block_7_snapshot_module_collection(
    module_collection,
):

    return {
        module_name:
            {
                tensor_name:
                    tensor.detach()
                    .cpu()
                    .clone()

                for tensor_name, tensor
                in module.state_dict().items()
            }

        for module_name, module
        in module_collection.items()
    }


def block_7_nested_state_exact(
    state_before,
    state_after,
):

    if state_before.keys() != state_after.keys():

        return False

    for module_name in state_before:

        if (
            state_before[
                module_name
            ].keys()
            !=
            state_after[
                module_name
            ].keys()
        ):

            return False

        for tensor_name in state_before[
            module_name
        ]:

            if not torch.equal(
                state_before[
                    module_name
                ][
                    tensor_name
                ],
                state_after[
                    module_name
                ][
                    tensor_name
                ],
            ):

                return False

    return True


BLOCK_7_REPRESENTATION_STATE_BEFORE = (
    block_7_snapshot_module_collection(
        NOTEBOOK_10_REPRESENTATION_MODULES
    )
)

BLOCK_7_TRANSFORMATIVE_STATE_BEFORE = (
    block_7_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_MODULES
    )
)

BLOCK_7_WEIGHT_STATE_BEFORE = (
    block_7_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES
    )
)


# =============================================================================
# Frozen/evaluation-state validation
# =============================================================================

BLOCK_7_ALL_PARAMETERS_FROZEN_BEFORE = all(
    not parameter.requires_grad

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


BLOCK_7_ALL_MODULES_EVAL_BEFORE = all(
    not module.training

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()
)


BLOCK_7_ALL_GRADIENTS_ABSENT_BEFORE = all(
    parameter.grad is None

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


if not all(
    [
        BLOCK_7_ALL_PARAMETERS_FROZEN_BEFORE,
        BLOCK_7_ALL_MODULES_EVAL_BEFORE,
        BLOCK_7_ALL_GRADIENTS_ABSENT_BEFORE,
    ]
):

    raise RuntimeError(
        "Notebook 10 Block 7 inherited architecture is not in the "
        "required frozen/evaluation/no-gradient state."
    )


# =============================================================================
# Canonical multi-article sentence identity
# =============================================================================
#
# The corpus is reconstructed from the validated Block 5 handover. Ordering is
# article-specific and follows the persisted canonical sequence position.
# =============================================================================

BLOCK_7_ARTICLE_IDS = tuple(
    NOTEBOOK_10_REPRESENTATION_READY_ARTICLE_IDS
)


def block_7_sequence_position(
    record,
):

    for key in (
        "canonical_sequence_position",
        "sequence_position",
        "sentence_index",
    ):

        value = record.get(
            key
        )

        if value is not None:

            return int(
                value
            )

    raise RuntimeError(
        "Canonical sentence record has no usable sequence-position field."
    )


BLOCK_7_CANONICAL_RECORDS = []

BLOCK_7_ARTICLE_RECORD_INDICES = {}

BLOCK_7_CANONICAL_SEQUENCE_VALID = True


for article_id in BLOCK_7_ARTICLE_IDS:

    article_records = deepcopy(
        NOTEBOOK_10_VALIDATED_MULTI_ARTICLE_RECORDS[
            article_id
        ]
    )

    article_records = sorted(
        article_records,
        key=
            block_7_sequence_position,
    )

    article_positions = [
        block_7_sequence_position(
            record
        )
        for record in article_records
    ]

    expected_positions = list(
        range(
            len(
                article_records
            )
        )
    )

    if article_positions != expected_positions:

        BLOCK_7_CANONICAL_SEQUENCE_VALID = False

    start_index = len(
        BLOCK_7_CANONICAL_RECORDS
    )

    BLOCK_7_CANONICAL_RECORDS.extend(
        article_records
    )

    end_index = len(
        BLOCK_7_CANONICAL_RECORDS
    )

    BLOCK_7_ARTICLE_RECORD_INDICES[
        article_id
    ] = tuple(
        range(
            start_index,
            end_index,
        )
    )


BLOCK_7_SENTENCE_IDS = tuple(
    record.get(
        "sentence_id"
    )
    for record
    in BLOCK_7_CANONICAL_RECORDS
)


BLOCK_7_ARTICLE_IDS_BY_SENTENCE = tuple(
    record.get(
        "article_id"
    )
    for record
    in BLOCK_7_CANONICAL_RECORDS
)


BLOCK_7_SEQUENCE_POSITIONS = tuple(
    block_7_sequence_position(
        record
    )
    for record
    in BLOCK_7_CANONICAL_RECORDS
)


BLOCK_7_SENTENCE_COUNT = len(
    BLOCK_7_SENTENCE_IDS
)

BLOCK_7_ARTICLE_COUNT = len(
    BLOCK_7_ARTICLE_IDS
)


BLOCK_7_CANONICAL_IDENTITY_VALID = all(
    [
        BLOCK_7_ARTICLE_COUNT > 0,
        BLOCK_7_SENTENCE_COUNT > 0,
        len(
            set(
                BLOCK_7_SENTENCE_IDS
            )
        )
        ==
        BLOCK_7_SENTENCE_COUNT,
        all(
            isinstance(
                sentence_id,
                str,
            )
            and
            bool(
                sentence_id.strip()
            )
            for sentence_id
            in BLOCK_7_SENTENCE_IDS
        ),
        all(
            article_id
            in
            BLOCK_7_ARTICLE_IDS
            for article_id
            in BLOCK_7_ARTICLE_IDS_BY_SENTENCE
        ),
        BLOCK_7_CANONICAL_SEQUENCE_VALID,
        sum(
            len(
                indices
            )
            for indices
            in BLOCK_7_ARTICLE_RECORD_INDICES.values()
        )
        ==
        BLOCK_7_SENTENCE_COUNT,
    ]
)


if not BLOCK_7_CANONICAL_IDENTITY_VALID:

    raise RuntimeError(
        "Notebook 10 Block 7 canonical multi-article sentence identity "
        "is invalid."
    )


# =============================================================================
# Google Drive helpers
# =============================================================================

def block_7_download_drive_bytes(
    drive_service,
    file_id,
):

    request = (
        drive_service.files()
        .get_media(
            fileId=
                file_id
        )
    )

    buffer = io.BytesIO()

    downloader = MediaIoBaseDownload(
        buffer,
        request,
    )

    complete = False

    while not complete:

        _, complete = (
            downloader.next_chunk()
        )

    return buffer.getvalue()


def block_7_load_json_by_id(
    drive_service,
    file_id,
):

    raw_bytes = block_7_download_drive_bytes(
        drive_service=
            drive_service,
        file_id=
            file_id,
    )

    if not raw_bytes:

        raise RuntimeError(
            f"Required persisted JSON artefact {file_id} is empty."
        )

    try:

        payload = json.loads(
            raw_bytes.decode(
                "utf-8"
            )
        )

    except (
        UnicodeDecodeError,
        json.JSONDecodeError,
    ) as exc:

        raise RuntimeError(
            f"Persisted artefact {file_id} is not valid UTF-8 JSON."
        ) from exc

    if not isinstance(
        payload,
        dict,
    ):

        raise RuntimeError(
            f"Persisted artefact {file_id} is not dictionary-like JSON."
        )

    return payload


# =============================================================================
# Restore validated supervision artefacts
# =============================================================================

BLOCK_7_FACTUAL_PAYLOAD = (
    block_7_load_json_by_id(
        NOTEBOOK_10_DRIVE_SERVICE,
        BLOCK_7_FACTUAL_DRIVE_FILE_ID,
    )
)


BLOCK_7_PSYCHOLOGICAL_PAYLOAD = (
    block_7_load_json_by_id(
        NOTEBOOK_10_DRIVE_SERVICE,
        BLOCK_7_PSYCHOLOGICAL_DRIVE_FILE_ID,
    )
)


BLOCK_7_SOCIAL_PAYLOAD = (
    block_7_load_json_by_id(
        NOTEBOOK_10_DRIVE_SERVICE,
        BLOCK_7_SOCIAL_DRIVE_FILE_ID,
    )
)


NOTEBOOK_10_BLOCK_7_SUPERVISION_RECOVERED = True


# =============================================================================
# Canonical supervision record alignment helper
# =============================================================================

def block_7_sentence_id_from_record(
    record,
):

    if not isinstance(
        record,
        dict,
    ):

        return None

    identity = record.get(
        "identity",
        {}
    )

    if isinstance(
        identity,
        dict,
    ):

        sentence_id = identity.get(
            "sentence_id"
        )

        if sentence_id:

            return sentence_id

    return record.get(
        "sentence_id"
    )


def block_7_align_records(
    records,
    source_name,
):

    if not isinstance(
        records,
        list,
    ):

        raise RuntimeError(
            f"{source_name} records are not list-like."
        )

    lookup = {}

    for record_index, record in enumerate(
        records
    ):

        sentence_id = (
            block_7_sentence_id_from_record(
                record
            )
        )

        if not sentence_id:

            raise RuntimeError(
                f"{source_name} record {record_index} has no sentence_id."
            )

        if sentence_id in lookup:

            raise RuntimeError(
                f"{source_name} contains duplicate sentence_id "
                f"{sentence_id!r}."
            )

        lookup[
            sentence_id
        ] = record

    canonical_set = set(
        BLOCK_7_SENTENCE_IDS
    )

    source_set = set(
        lookup.keys()
    )

    missing_ids = sorted(
        canonical_set
        -
        source_set
    )

    if missing_ids:

        raise RuntimeError(
            f"{source_name} does not cover the complete Notebook 10 "
            f"canonical corpus. Missing sentence IDs: {missing_ids[:10]}"
        )

    aligned_records = [
        lookup[
            sentence_id
        ]
        for sentence_id
        in BLOCK_7_SENTENCE_IDS
    ]

    aligned_ids = tuple(
        block_7_sentence_id_from_record(
            record
        )
        for record
        in aligned_records
    )

    if aligned_ids != BLOCK_7_SENTENCE_IDS:

        raise RuntimeError(
            f"{source_name} could not be aligned exactly to the "
            "Notebook 10 canonical sentence order."
        )

    return (
        aligned_records,
        len(
            source_set
            -
            canonical_set
        ),
    )


# =============================================================================
# Factual supervision
# =============================================================================

BLOCK_7_FACTUAL_ARCHITECTURE = (
    BLOCK_7_FACTUAL_PAYLOAD.get(
        "architecture",
        {},
    )
)

BLOCK_7_FACTUAL_DIMENSION_ORDER = tuple(
    BLOCK_7_FACTUAL_ARCHITECTURE.get(
        "dimension_order",
        [],
    )
)

(
    BLOCK_7_FACTUAL_RECORDS,
    BLOCK_7_FACTUAL_UNUSED_RECORD_COUNT,
) = block_7_align_records(
    deepcopy(
        BLOCK_7_FACTUAL_PAYLOAD.get(
            "records",
            [],
        )
    ),
    "Factual supervision",
)


BLOCK_7_FACTUAL_CONTRACT_VALID = all(
    [
        (
            BLOCK_7_FACTUAL_PAYLOAD.get(
                "artefact_type"
            )
            ==
            "media_ai_factual_target_contract"
        ),
        (
            int(
                BLOCK_7_FACTUAL_ARCHITECTURE.get(
                    "factual_dimension",
                    -1,
                )
            )
            ==
            BLOCK_7_PATHWAY_DIMENSIONS[
                "factual"
            ]
        ),
        (
            len(
                BLOCK_7_FACTUAL_DIMENSION_ORDER
            )
            ==
            BLOCK_7_PATHWAY_DIMENSIONS[
                "factual"
            ]
        ),
        (
            len(
                BLOCK_7_FACTUAL_RECORDS
            )
            ==
            BLOCK_7_SENTENCE_COUNT
        ),
    ]
)


if not BLOCK_7_FACTUAL_CONTRACT_VALID:

    raise RuntimeError(
        "Notebook 10 Block 7 factual supervision contract is invalid."
    )


BLOCK_7_FACTUAL_CURRENT_SUPERVISION = np.zeros(
    (
        BLOCK_7_SENTENCE_COUNT,
        BLOCK_7_PATHWAY_DIMENSIONS[
            "factual"
        ],
    ),
    dtype=np.float32,
)

BLOCK_7_FACTUAL_CURRENT_MASK = np.zeros(
    BLOCK_7_FACTUAL_CURRENT_SUPERVISION.shape,
    dtype=bool,
)


for record_index, record in enumerate(
    BLOCK_7_FACTUAL_RECORDS
):

    target = np.asarray(
        record.get(
            "factual_target"
        ),
        dtype=np.float32,
    )

    mask = np.asarray(
        record.get(
            "factual_target_mask"
        ),
        dtype=bool,
    )

    expected_shape = (
        BLOCK_7_PATHWAY_DIMENSIONS[
            "factual"
        ],
    )

    if (
        target.shape
        !=
        expected_shape
        or
        mask.shape
        !=
        expected_shape
    ):

        raise RuntimeError(
            f"Factual supervision record {record_index} has invalid "
            "target/mask dimensions."
        )

    if not np.isfinite(
        target[
            mask
        ]
    ).all():

        raise RuntimeError(
            f"Factual supervision record {record_index} contains "
            "non-finite available values."
        )

    BLOCK_7_FACTUAL_CURRENT_SUPERVISION[
        record_index
    ] = np.where(
        mask,
        target,
        0.0,
    )

    BLOCK_7_FACTUAL_CURRENT_MASK[
        record_index
    ] = mask


# =============================================================================
# Psychological supervision
# =============================================================================

BLOCK_7_PSYCHOLOGICAL_RECORDS_RAW = deepcopy(
    BLOCK_7_PSYCHOLOGICAL_PAYLOAD.get(
        "records",
        [],
    )
)

(
    BLOCK_7_PSYCHOLOGICAL_RECORDS,
    BLOCK_7_PSYCHOLOGICAL_UNUSED_RECORD_COUNT,
) = block_7_align_records(
    BLOCK_7_PSYCHOLOGICAL_RECORDS_RAW,
    "Psychological supervision",
)


BLOCK_7_EMOTION_DIMENSION_ORDER = (
    "fear",
    "anger",
    "sadness",
    "disgust",
    "tension",
    "uncertainty",
    "hope",
    "relief",
    "trust",
    "compassion",
    "solidarity",
)


BLOCK_7_PSYCHOLOGICAL_FAMILY_ORDER = (
    "REV",
    "TEV",
    "IEV",
    "EW",
)


BLOCK_7_PSYCHOLOGICAL_FAMILY_KEYS = {
    "REV":
        "reported_emotion",
    "TEV":
        "transformative_emotion",
    "IEV":
        "immediate_emotional_state",
    "EW":
        "emotional_weight",
}


def block_7_extract_emotion_family(
    records,
    annotation_key,
):

    target = np.zeros(
        (
            BLOCK_7_SENTENCE_COUNT,
            len(
                BLOCK_7_EMOTION_DIMENSION_ORDER
            ),
        ),
        dtype=np.float32,
    )

    mask = np.zeros(
        target.shape,
        dtype=bool,
    )

    for record_index, record in enumerate(
        records
    ):

        annotations = record.get(
            "annotations",
            {},
        )

        family = annotations.get(
            annotation_key,
            {},
        )

        if not isinstance(
            family,
            dict,
        ):

            raise RuntimeError(
                f"Psychological record {record_index} does not contain "
                f"{annotation_key}."
            )

        for dimension_index, emotion_name in enumerate(
            BLOCK_7_EMOTION_DIMENSION_ORDER
        ):

            annotation = family.get(
                emotion_name
            )

            if not isinstance(
                annotation,
                dict,
            ):

                raise RuntimeError(
                    f"Psychological record {record_index} is missing "
                    f"{annotation_key}.{emotion_name}."
                )

            available = (
                annotation.get(
                    "value_state"
                )
                ==
                "observed"
            )

            if available:

                value = annotation.get(
                    "value"
                )

                if value is None:

                    raise RuntimeError(
                        f"Observed psychological value "
                        f"{annotation_key}.{emotion_name} is missing."
                    )

                numeric_value = float(
                    value
                )

                if not np.isfinite(
                    numeric_value
                ):

                    raise RuntimeError(
                        f"Observed psychological value "
                        f"{annotation_key}.{emotion_name} is non-finite."
                    )

                target[
                    record_index,
                    dimension_index,
                ] = numeric_value

                mask[
                    record_index,
                    dimension_index,
                ] = True

    return (
        target,
        mask,
    )


(
    BLOCK_7_REV_TARGETS,
    BLOCK_7_REV_MASK,
) = block_7_extract_emotion_family(
    BLOCK_7_PSYCHOLOGICAL_RECORDS,
    "reported_emotion",
)


(
    BLOCK_7_TEV_TARGETS,
    BLOCK_7_TEV_MASK,
) = block_7_extract_emotion_family(
    BLOCK_7_PSYCHOLOGICAL_RECORDS,
    "transformative_emotion",
)


(
    BLOCK_7_IEV_TARGETS,
    BLOCK_7_IEV_MASK,
) = block_7_extract_emotion_family(
    BLOCK_7_PSYCHOLOGICAL_RECORDS,
    "immediate_emotional_state",
)


BLOCK_7_EW_TARGETS = np.zeros(
    (
        BLOCK_7_SENTENCE_COUNT,
        1,
    ),
    dtype=np.float32,
)

BLOCK_7_EW_MASK = np.zeros(
    BLOCK_7_EW_TARGETS.shape,
    dtype=bool,
)


for record_index, record in enumerate(
    BLOCK_7_PSYCHOLOGICAL_RECORDS
):

    annotations = record.get(
        "annotations",
        {},
    )

    emotional_weight = annotations.get(
        "emotional_weight",
        {},
    )

    if not isinstance(
        emotional_weight,
        dict,
    ):

        raise RuntimeError(
            f"Psychological record {record_index} does not contain "
            "emotional_weight."
        )

    available = (
        emotional_weight.get(
            "value_state"
        )
        ==
        "observed"
    )

    if available:

        value = emotional_weight.get(
            "value"
        )

        if value is None:

            raise RuntimeError(
                f"Observed emotional_weight in record {record_index} "
                "has no value."
            )

        numeric_value = float(
            value
        )

        if not np.isfinite(
            numeric_value
        ):

            raise RuntimeError(
                f"Observed emotional_weight in record {record_index} "
                "is non-finite."
            )

        BLOCK_7_EW_TARGETS[
            record_index,
            0,
        ] = numeric_value

        BLOCK_7_EW_MASK[
            record_index,
            0,
        ] = True


BLOCK_7_PSYCHOLOGICAL_CURRENT_SUPERVISION = np.concatenate(
    [
        BLOCK_7_REV_TARGETS,
        BLOCK_7_TEV_TARGETS,
        BLOCK_7_IEV_TARGETS,
        BLOCK_7_EW_TARGETS,
    ],
    axis=
        1,
).astype(
    np.float32
)


BLOCK_7_PSYCHOLOGICAL_CURRENT_MASK = np.concatenate(
    [
        BLOCK_7_REV_MASK,
        BLOCK_7_TEV_MASK,
        BLOCK_7_IEV_MASK,
        BLOCK_7_EW_MASK,
    ],
    axis=
        1,
).astype(
    bool
)


if (
    BLOCK_7_PSYCHOLOGICAL_CURRENT_SUPERVISION.shape
    !=
    (
        BLOCK_7_SENTENCE_COUNT,
        BLOCK_7_PATHWAY_DIMENSIONS[
            "psychological"
        ],
    )
):

    raise RuntimeError(
        "Notebook 10 Block 7 reconstructed psychological supervision "
        "does not reproduce the 34-dimensional contract."
    )


# =============================================================================
# Social supervision
# =============================================================================

BLOCK_7_SOCIAL_ARCHITECTURE = (
    BLOCK_7_SOCIAL_PAYLOAD.get(
        "architecture",
        {},
    )
)

BLOCK_7_SOCIAL_DIMENSION_ORDER = tuple(
    BLOCK_7_SOCIAL_ARCHITECTURE.get(
        "dimension_order",
        [],
    )
)


BLOCK_7_EXPECTED_SOCIAL_DIMENSION_ORDER = (
    "interpersonal_cohesion",
    "economic_cohesion",
    "political_cohesion",
    "religious_cohesion",
    "group_collective_cohesion",
    "institutional_role_relation",
    "public_audience_relation",
)


(
    BLOCK_7_SOCIAL_RECORDS,
    BLOCK_7_SOCIAL_UNUSED_RECORD_COUNT,
) = block_7_align_records(
    deepcopy(
        BLOCK_7_SOCIAL_PAYLOAD.get(
            "records",
            [],
        )
    ),
    "Social supervision",
)


BLOCK_7_SOCIAL_CONTRACT_VALID = all(
    [
        (
            BLOCK_7_SOCIAL_PAYLOAD.get(
                "artefact_type"
            )
            ==
            "media_ai_accepted_social_supervision"
        ),
        (
            int(
                BLOCK_7_SOCIAL_ARCHITECTURE.get(
                    "representation_dimension",
                    -1,
                )
            )
            ==
            BLOCK_7_PATHWAY_DIMENSIONS[
                "social"
            ]
        ),
        (
            BLOCK_7_SOCIAL_DIMENSION_ORDER
            ==
            BLOCK_7_EXPECTED_SOCIAL_DIMENSION_ORDER
        ),
        (
            len(
                BLOCK_7_SOCIAL_RECORDS
            )
            ==
            BLOCK_7_SENTENCE_COUNT
        ),
    ]
)


if not BLOCK_7_SOCIAL_CONTRACT_VALID:

    raise RuntimeError(
        "Notebook 10 Block 7 accepted social supervision contract "
        "is invalid."
    )


BLOCK_7_SOCIAL_CURRENT_SUPERVISION = np.zeros(
    (
        BLOCK_7_SENTENCE_COUNT,
        BLOCK_7_PATHWAY_DIMENSIONS[
            "social"
        ],
    ),
    dtype=np.float32,
)

BLOCK_7_SOCIAL_CURRENT_MASK = np.zeros(
    BLOCK_7_SOCIAL_CURRENT_SUPERVISION.shape,
    dtype=bool,
)


for record_index, record in enumerate(
    BLOCK_7_SOCIAL_RECORDS
):

    social_annotations = record.get(
        "social_annotations",
        {},
    )

    if not isinstance(
        social_annotations,
        dict,
    ):

        raise RuntimeError(
            f"Social record {record_index} has no "
            "social_annotations dictionary."
        )

    for dimension_index, dimension_name in enumerate(
        BLOCK_7_SOCIAL_DIMENSION_ORDER
    ):

        annotation = social_annotations.get(
            dimension_name
        )

        if not isinstance(
            annotation,
            dict,
        ):

            raise RuntimeError(
                f"Social record {record_index} is missing "
                f"{dimension_name}."
            )

        available = bool(
            annotation.get(
                "available",
                False,
            )
        )

        if available:

            value = annotation.get(
                "value"
            )

            if value is None:

                raise RuntimeError(
                    f"Available social target {dimension_name} "
                    f"in record {record_index} has no value."
                )

            numeric_value = float(
                value
            )

            if not np.isfinite(
                numeric_value
            ):

                raise RuntimeError(
                    f"Available social target {dimension_name} "
                    f"in record {record_index} is non-finite."
                )

            BLOCK_7_SOCIAL_CURRENT_SUPERVISION[
                record_index,
                dimension_index,
            ] = numeric_value

            BLOCK_7_SOCIAL_CURRENT_MASK[
                record_index,
                dimension_index,
            ] = True


# =============================================================================
# Canonical current-state supervision
# =============================================================================

BLOCK_7_CURRENT_SUPERVISION = {
    "factual":
        BLOCK_7_FACTUAL_CURRENT_SUPERVISION,
    "psychological":
        BLOCK_7_PSYCHOLOGICAL_CURRENT_SUPERVISION,
    "social":
        BLOCK_7_SOCIAL_CURRENT_SUPERVISION,
}


BLOCK_7_CURRENT_SUPERVISION_MASKS = {
    "factual":
        BLOCK_7_FACTUAL_CURRENT_MASK,
    "psychological":
        BLOCK_7_PSYCHOLOGICAL_CURRENT_MASK,
    "social":
        BLOCK_7_SOCIAL_CURRENT_MASK,
}


BLOCK_7_CURRENT_SUPERVISION_SHAPES_VALID = all(
    (
        BLOCK_7_CURRENT_SUPERVISION[
            pathway_name
        ].shape
        ==
        (
            BLOCK_7_SENTENCE_COUNT,
            pathway_dim,
        )
        and
        BLOCK_7_CURRENT_SUPERVISION_MASKS[
            pathway_name
        ].shape
        ==
        (
            BLOCK_7_SENTENCE_COUNT,
            pathway_dim,
        )
    )

    for pathway_name, pathway_dim
    in BLOCK_7_PATHWAY_DIMENSIONS.items()
)


BLOCK_7_AVAILABLE_VALUES_FINITE = all(
    bool(
        np.isfinite(
            BLOCK_7_CURRENT_SUPERVISION[
                pathway_name
            ][
                BLOCK_7_CURRENT_SUPERVISION_MASKS[
                    pathway_name
                ]
            ]
        ).all()
    )

    for pathway_name
    in BLOCK_7_PATHWAY_DIMENSIONS
)


if not all(
    [
        BLOCK_7_CURRENT_SUPERVISION_SHAPES_VALID,
        BLOCK_7_AVAILABLE_VALUES_FINITE,
    ]
):

    raise RuntimeError(
        "Notebook 10 Block 7 current-state supervision failed "
        "shape/finiteness validation."
    )


NOTEBOOK_10_BLOCK_7_SUPERVISION_ALIGNED = True


# =============================================================================
# Construct causal transformation targets
# =============================================================================
#
# For the first sentence of each article:
#
#     Δy_(a,0) = y_(a,0) - 0
#
# For every later sentence:
#
#     Δy_(a,t) = y_(a,t) - y_(a,t-1)
#
# A non-initial transformation element is available only when both current and
# immediately preceding supervision are available for that dimension.
# =============================================================================

BLOCK_7_TRANSFORMATION_TARGETS = {
    pathway_name:
        np.zeros(
            (
                BLOCK_7_SENTENCE_COUNT,
                pathway_dim,
            ),
            dtype=np.float32,
        )
    for pathway_name, pathway_dim
    in BLOCK_7_PATHWAY_DIMENSIONS.items()
}


BLOCK_7_TRANSFORMATION_TARGET_MASKS = {
    pathway_name:
        np.zeros(
            (
                BLOCK_7_SENTENCE_COUNT,
                pathway_dim,
            ),
            dtype=bool,
        )
    for pathway_name, pathway_dim
    in BLOCK_7_PATHWAY_DIMENSIONS.items()
}


BLOCK_7_PRECEDING_SUPERVISION = {
    pathway_name:
        np.zeros(
            (
                BLOCK_7_SENTENCE_COUNT,
                pathway_dim,
            ),
            dtype=np.float32,
        )
    for pathway_name, pathway_dim
    in BLOCK_7_PATHWAY_DIMENSIONS.items()
}


BLOCK_7_PRECEDING_SUPERVISION_MASKS = {
    pathway_name:
        np.zeros(
            (
                BLOCK_7_SENTENCE_COUNT,
                pathway_dim,
            ),
            dtype=bool,
        )
    for pathway_name, pathway_dim
    in BLOCK_7_PATHWAY_DIMENSIONS.items()
}


BLOCK_7_ARTICLE_BOUNDARY_TARGETS_VALID = True
BLOCK_7_NO_CROSS_ARTICLE_TARGETS = True


for article_id in BLOCK_7_ARTICLE_IDS:

    article_indices = (
        BLOCK_7_ARTICLE_RECORD_INDICES[
            article_id
        ]
    )

    if not article_indices:

        raise RuntimeError(
            f"Article {article_id!r} has no canonical sentence records."
        )

    for local_position, global_index in enumerate(
        article_indices
    ):

        for pathway_name in BLOCK_7_PATHWAY_DIMENSIONS:

            current = (
                BLOCK_7_CURRENT_SUPERVISION[
                    pathway_name
                ][
                    global_index
                ]
            )

            current_mask = (
                BLOCK_7_CURRENT_SUPERVISION_MASKS[
                    pathway_name
                ][
                    global_index
                ]
            )

            if local_position == 0:

                preceding = np.zeros_like(
                    current
                )

                preceding_mask = np.ones_like(
                    current_mask,
                    dtype=bool,
                )

                target_mask = current_mask.copy()

            else:

                preceding_index = article_indices[
                    local_position
                    -
                    1
                ]

                if (
                    BLOCK_7_ARTICLE_IDS_BY_SENTENCE[
                        preceding_index
                    ]
                    !=
                    article_id
                ):

                    BLOCK_7_NO_CROSS_ARTICLE_TARGETS = False

                preceding = (
                    BLOCK_7_CURRENT_SUPERVISION[
                        pathway_name
                    ][
                        preceding_index
                    ]
                )

                preceding_mask = (
                    BLOCK_7_CURRENT_SUPERVISION_MASKS[
                        pathway_name
                    ][
                        preceding_index
                    ]
                )

                target_mask = (
                    current_mask
                    &
                    preceding_mask
                )

            BLOCK_7_PRECEDING_SUPERVISION[
                pathway_name
            ][
                global_index
            ] = preceding

            BLOCK_7_PRECEDING_SUPERVISION_MASKS[
                pathway_name
            ][
                global_index
            ] = preceding_mask

            signed_target = (
                current
                -
                preceding
            )

            BLOCK_7_TRANSFORMATION_TARGETS[
                pathway_name
            ][
                global_index
            ] = np.where(
                target_mask,
                signed_target,
                0.0,
            )

            BLOCK_7_TRANSFORMATION_TARGET_MASKS[
                pathway_name
            ][
                global_index
            ] = target_mask


# =============================================================================
# Objective-eligible masks
# =============================================================================

BLOCK_7_OBJECTIVE_ELIGIBILITY_VECTORS = {}


for pathway_name, pathway_dim in BLOCK_7_PATHWAY_DIMENSIONS.items():

    eligibility = np.zeros(
        (
            pathway_dim,
        ),
        dtype=bool,
    )

    eligibility[
        list(
            BLOCK_7_ACTIVE_DIMENSION_INDICES[
                pathway_name
            ]
        )
    ] = True

    BLOCK_7_OBJECTIVE_ELIGIBILITY_VECTORS[
        pathway_name
    ] = eligibility


BLOCK_7_OBJECTIVE_MASKS = {
    pathway_name:
        (
            BLOCK_7_TRANSFORMATION_TARGET_MASKS[
                pathway_name
            ]
            &
            BLOCK_7_OBJECTIVE_ELIGIBILITY_VECTORS[
                pathway_name
            ][
                None,
                :
            ]
        )
    for pathway_name
    in BLOCK_7_PATHWAY_DIMENSIONS
}


NOTEBOOK_10_BLOCK_7_TRANSFORMATION_TARGETS_CONSTRUCTED = True
NOTEBOOK_10_BLOCK_7_OBJECTIVE_MASKS_CONSTRUCTED = True


# =============================================================================
# Target and mask validation
# =============================================================================

BLOCK_7_TARGET_SHAPES_VALID = all(
    (
        BLOCK_7_TRANSFORMATION_TARGETS[
            pathway_name
        ].shape
        ==
        (
            BLOCK_7_SENTENCE_COUNT,
            pathway_dim,
        )
        and
        BLOCK_7_TRANSFORMATION_TARGET_MASKS[
            pathway_name
        ].shape
        ==
        (
            BLOCK_7_SENTENCE_COUNT,
            pathway_dim,
        )
        and
        BLOCK_7_OBJECTIVE_MASKS[
            pathway_name
        ].shape
        ==
        (
            BLOCK_7_SENTENCE_COUNT,
            pathway_dim,
        )
    )

    for pathway_name, pathway_dim
    in BLOCK_7_PATHWAY_DIMENSIONS.items()
)


BLOCK_7_AVAILABLE_TARGETS_FINITE = all(
    bool(
        np.isfinite(
            BLOCK_7_TRANSFORMATION_TARGETS[
                pathway_name
            ][
                BLOCK_7_TRANSFORMATION_TARGET_MASKS[
                    pathway_name
                ]
            ]
        ).all()
    )

    for pathway_name
    in BLOCK_7_PATHWAY_DIMENSIONS
)


BLOCK_7_DEFERRED_OBJECTIVE_ZERO = all(
    not bool(
        BLOCK_7_OBJECTIVE_MASKS[
            pathway_name
        ][
            :,
            list(
                BLOCK_7_DEFERRED_DIMENSION_INDICES[
                    pathway_name
                ]
            )
        ].any()
    )
    if BLOCK_7_DEFERRED_DIMENSION_INDICES[
        pathway_name
    ]
    else True

    for pathway_name
    in BLOCK_7_PATHWAY_DIMENSIONS
)


BLOCK_7_FIRST_SENTENCE_BOUNDARY_MASK_VALID = True


for article_id in BLOCK_7_ARTICLE_IDS:

    first_index = (
        BLOCK_7_ARTICLE_RECORD_INDICES[
            article_id
        ][
            0
        ]
    )

    for pathway_name in BLOCK_7_PATHWAY_DIMENSIONS:

        if not np.array_equal(
            BLOCK_7_TRANSFORMATION_TARGET_MASKS[
                pathway_name
            ][
                first_index
            ],
            BLOCK_7_CURRENT_SUPERVISION_MASKS[
                pathway_name
            ][
                first_index
            ],
        ):

            BLOCK_7_FIRST_SENTENCE_BOUNDARY_MASK_VALID = False


BLOCK_7_CAUSAL_TARGET_CONTRACT_VALID = all(
    [
        BLOCK_7_NO_CROSS_ARTICLE_TARGETS,
        BLOCK_7_FIRST_SENTENCE_BOUNDARY_MASK_VALID,
        NOTEBOOK_10_BLOCK_7_FUTURE_CONTEXT_USED is False,
        NOTEBOOK_10_BLOCK_7_SUPERVISION_USED_AS_MODEL_INPUT is False,
    ]
)


if not all(
    [
        BLOCK_7_TARGET_SHAPES_VALID,
        BLOCK_7_AVAILABLE_TARGETS_FINITE,
        BLOCK_7_DEFERRED_OBJECTIVE_ZERO,
        BLOCK_7_CAUSAL_TARGET_CONTRACT_VALID,
    ]
):

    raise RuntimeError(
        "Notebook 10 Block 7 transformation-target or objective-mask "
        "contract failed validation."
    )


# =============================================================================
# Coverage audit
# =============================================================================

BLOCK_7_CURRENT_AVAILABLE_ELEMENTS = {
    pathway_name:
        int(
            BLOCK_7_CURRENT_SUPERVISION_MASKS[
                pathway_name
            ].sum()
        )
    for pathway_name
    in BLOCK_7_PATHWAY_DIMENSIONS
}


BLOCK_7_TRANSFORMATION_AVAILABLE_ELEMENTS = {
    pathway_name:
        int(
            BLOCK_7_TRANSFORMATION_TARGET_MASKS[
                pathway_name
            ].sum()
        )
    for pathway_name
    in BLOCK_7_PATHWAY_DIMENSIONS
}


BLOCK_7_OBJECTIVE_ACTIVE_ELEMENTS = {
    pathway_name:
        int(
            BLOCK_7_OBJECTIVE_MASKS[
                pathway_name
            ].sum()
        )
    for pathway_name
    in BLOCK_7_PATHWAY_DIMENSIONS
}


BLOCK_7_SUPERVISED_SENTENCE_COUNTS = {
    pathway_name:
        int(
            np.any(
                BLOCK_7_CURRENT_SUPERVISION_MASKS[
                    pathway_name
                ],
                axis=
                    1,
            ).sum()
        )
    for pathway_name
    in BLOCK_7_PATHWAY_DIMENSIONS
}


BLOCK_7_SUPERVISED_ARTICLE_COUNTS = {}


for pathway_name in BLOCK_7_PATHWAY_DIMENSIONS:

    supervised_article_count = 0

    for article_id in BLOCK_7_ARTICLE_IDS:

        indices = (
            BLOCK_7_ARTICLE_RECORD_INDICES[
                article_id
            ]
        )

        if bool(
            BLOCK_7_CURRENT_SUPERVISION_MASKS[
                pathway_name
            ][
                list(
                    indices
                )
            ].any()
        ):

            supervised_article_count += 1

    BLOCK_7_SUPERVISED_ARTICLE_COUNTS[
        pathway_name
    ] = supervised_article_count


BLOCK_7_ALL_PATHWAYS_HAVE_SUPERVISION = all(
    BLOCK_7_CURRENT_AVAILABLE_ELEMENTS[
        pathway_name
    ]
    >
    0
    for pathway_name
    in BLOCK_7_PATHWAY_DIMENSIONS
)


BLOCK_7_ALL_PATHWAYS_HAVE_OBJECTIVE_ELEMENTS = all(
    BLOCK_7_OBJECTIVE_ACTIVE_ELEMENTS[
        pathway_name
    ]
    >
    0
    for pathway_name
    in BLOCK_7_PATHWAY_DIMENSIONS
)


if not all(
    [
        BLOCK_7_ALL_PATHWAYS_HAVE_SUPERVISION,
        BLOCK_7_ALL_PATHWAYS_HAVE_OBJECTIVE_ELEMENTS,
    ]
):

    raise RuntimeError(
        "Notebook 10 Block 7 produced no usable supervision/objective "
        "elements for one or more pathways."
    )


# =============================================================================
# Post-construction model-state validation
# =============================================================================

BLOCK_7_REPRESENTATION_STATE_AFTER = (
    block_7_snapshot_module_collection(
        NOTEBOOK_10_REPRESENTATION_MODULES
    )
)

BLOCK_7_TRANSFORMATIVE_STATE_AFTER = (
    block_7_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_MODULES
    )
)

BLOCK_7_WEIGHT_STATE_AFTER = (
    block_7_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES
    )
)


BLOCK_7_REPRESENTATION_UNCHANGED = (
    block_7_nested_state_exact(
        BLOCK_7_REPRESENTATION_STATE_BEFORE,
        BLOCK_7_REPRESENTATION_STATE_AFTER,
    )
)

BLOCK_7_TRANSFORMATIVE_UNCHANGED = (
    block_7_nested_state_exact(
        BLOCK_7_TRANSFORMATIVE_STATE_BEFORE,
        BLOCK_7_TRANSFORMATIVE_STATE_AFTER,
    )
)

BLOCK_7_WEIGHT_UNCHANGED = (
    block_7_nested_state_exact(
        BLOCK_7_WEIGHT_STATE_BEFORE,
        BLOCK_7_WEIGHT_STATE_AFTER,
    )
)


BLOCK_7_ALL_PARAMETERS_FROZEN_AFTER = all(
    not parameter.requires_grad

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


BLOCK_7_ALL_MODULES_EVAL_AFTER = all(
    not module.training

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()
)


BLOCK_7_ALL_GRADIENTS_ABSENT_AFTER = all(
    parameter.grad is None

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


BLOCK_7_MODEL_STATE_VALID = all(
    [
        BLOCK_7_REPRESENTATION_UNCHANGED,
        BLOCK_7_TRANSFORMATIVE_UNCHANGED,
        BLOCK_7_WEIGHT_UNCHANGED,
        BLOCK_7_ALL_PARAMETERS_FROZEN_AFTER,
        BLOCK_7_ALL_MODULES_EVAL_AFTER,
        BLOCK_7_ALL_GRADIENTS_ABSENT_AFTER,
    ]
)


if not BLOCK_7_MODEL_STATE_VALID:

    raise RuntimeError(
        "Notebook 10 Block 7 changed the inherited Media AI model state."
    )


# =============================================================================
# Parameter accounting
# =============================================================================

BLOCK_7_REPRESENTATION_PARAMETER_COUNT = sum(
    parameter.numel()
    for module
    in NOTEBOOK_10_REPRESENTATION_MODULES.values()
    for parameter
    in module.parameters()
)

BLOCK_7_TRANSFORMATIVE_PARAMETER_COUNT = sum(
    parameter.numel()
    for module
    in NOTEBOOK_10_TRANSFORMATIVE_MODULES.values()
    for parameter
    in module.parameters()
)

BLOCK_7_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT = sum(
    parameter.numel()
    for module
    in NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES.values()
    for parameter
    in module.parameters()
)

BLOCK_7_TOTAL_PARAMETER_COUNT = (
    BLOCK_7_REPRESENTATION_PARAMETER_COUNT
    +
    BLOCK_7_TRANSFORMATIVE_PARAMETER_COUNT
    +
    BLOCK_7_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT
)


BLOCK_7_PARAMETER_ACCOUNTING_VALID = all(
    [
        BLOCK_7_REPRESENTATION_PARAMETER_COUNT == 894_003,
        BLOCK_7_TRANSFORMATIVE_PARAMETER_COUNT == 8_187,
        BLOCK_7_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT == 49,
        BLOCK_7_TOTAL_PARAMETER_COUNT == 902_239,
    ]
)


if not BLOCK_7_PARAMETER_ACCOUNTING_VALID:

    raise RuntimeError(
        "Notebook 10 Block 7 parameter accounting changed unexpectedly."
    )


# =============================================================================
# Expose canonical Block 7 handover objects for Block 8
# =============================================================================

NOTEBOOK_10_SUPERVISION_CURRENT_STATE = {
    pathway_name:
        torch.tensor(
            values,
            dtype=
                torch.float32,
        )
    for pathway_name, values
    in BLOCK_7_CURRENT_SUPERVISION.items()
}


NOTEBOOK_10_SUPERVISION_CURRENT_MASKS = {
    pathway_name:
        torch.tensor(
            mask,
            dtype=
                torch.bool,
        )
    for pathway_name, mask
    in BLOCK_7_CURRENT_SUPERVISION_MASKS.items()
}


NOTEBOOK_10_TRANSFORMATION_TARGETS = {
    pathway_name:
        torch.tensor(
            values,
            dtype=
                torch.float32,
        )
    for pathway_name, values
    in BLOCK_7_TRANSFORMATION_TARGETS.items()
}


NOTEBOOK_10_TRANSFORMATION_TARGET_MASKS = {
    pathway_name:
        torch.tensor(
            mask,
            dtype=
                torch.bool,
        )
    for pathway_name, mask
    in BLOCK_7_TRANSFORMATION_TARGET_MASKS.items()
}


NOTEBOOK_10_TRANSFORMATION_OBJECTIVE_MASKS = {
    pathway_name:
        torch.tensor(
            mask,
            dtype=
                torch.bool,
        )
    for pathway_name, mask
    in BLOCK_7_OBJECTIVE_MASKS.items()
}


NOTEBOOK_10_SUPERVISION_CANONICAL_ARTICLE_IDS = tuple(
    BLOCK_7_ARTICLE_IDS
)

NOTEBOOK_10_SUPERVISION_CANONICAL_SENTENCE_IDS = tuple(
    BLOCK_7_SENTENCE_IDS
)

NOTEBOOK_10_SUPERVISION_CANONICAL_SEQUENCE_POSITIONS = tuple(
    BLOCK_7_SEQUENCE_POSITIONS
)

NOTEBOOK_10_SUPERVISION_ARTICLE_RECORD_INDICES = deepcopy(
    BLOCK_7_ARTICLE_RECORD_INDICES
)

NOTEBOOK_10_SUPERVISION_ACTIVE_DIMENSION_INDICES = deepcopy(
    BLOCK_7_ACTIVE_DIMENSION_INDICES
)

NOTEBOOK_10_SUPERVISION_DEFERRED_DIMENSION_INDICES = deepcopy(
    BLOCK_7_DEFERRED_DIMENSION_INDICES
)

NOTEBOOK_10_SUPERVISION_SOCIAL_DIMENSION_NAMES = tuple(
    BLOCK_7_SOCIAL_DIMENSION_ORDER
)


# =============================================================================
# Final validation
# =============================================================================

NOTEBOOK_10_BLOCK_7_ERRORS = []


BLOCK_7_VALIDATION_CHECKS = {
    "prerequisites_invalid":
        BLOCK_7_PREREQUISITES_VALID,
    "block_6_handover_invalid":
        BLOCK_7_BLOCK_6_HANDOVER_VALID,
    "pathway_dimensions_invalid":
        BLOCK_7_PATHWAY_DIMENSIONS_VALID,
    "dimension_partitions_invalid":
        BLOCK_7_ALL_DIMENSION_PARTITIONS_VALID,
    "canonical_identity_invalid":
        BLOCK_7_CANONICAL_IDENTITY_VALID,
    "factual_contract_invalid":
        BLOCK_7_FACTUAL_CONTRACT_VALID,
    "social_contract_invalid":
        BLOCK_7_SOCIAL_CONTRACT_VALID,
    "supervision_shape_invalid":
        BLOCK_7_CURRENT_SUPERVISION_SHAPES_VALID,
    "available_supervision_nonfinite":
        BLOCK_7_AVAILABLE_VALUES_FINITE,
    "target_shape_invalid":
        BLOCK_7_TARGET_SHAPES_VALID,
    "available_target_nonfinite":
        BLOCK_7_AVAILABLE_TARGETS_FINITE,
    "deferred_objective_not_zero":
        BLOCK_7_DEFERRED_OBJECTIVE_ZERO,
    "causal_target_contract_invalid":
        BLOCK_7_CAUSAL_TARGET_CONTRACT_VALID,
    "pathway_supervision_missing":
        BLOCK_7_ALL_PATHWAYS_HAVE_SUPERVISION,
    "pathway_objective_missing":
        BLOCK_7_ALL_PATHWAYS_HAVE_OBJECTIVE_ELEMENTS,
    "model_state_changed":
        BLOCK_7_MODEL_STATE_VALID,
    "parameter_accounting_invalid":
        BLOCK_7_PARAMETER_ACCOUNTING_VALID,
}


for error_name, condition in BLOCK_7_VALIDATION_CHECKS.items():

    if not condition:

        NOTEBOOK_10_BLOCK_7_ERRORS.append(
            error_name
        )


BLOCK_7_PROHIBITED_OPERATIONS = {
    "representation_forward_executed":
        NOTEBOOK_10_BLOCK_7_REPRESENTATION_FORWARD_EXECUTED,
    "transformative_forward_executed":
        NOTEBOOK_10_BLOCK_7_TRANSFORMATIVE_FORWARD_EXECUTED,
    "weight_forward_executed":
        NOTEBOOK_10_BLOCK_7_WEIGHT_FORWARD_EXECUTED,
    "recurrent_state_created":
        NOTEBOOK_10_BLOCK_7_RECURRENT_STATE_CREATED,
    "recurrent_update_executed":
        NOTEBOOK_10_BLOCK_7_RECURRENT_UPDATE_EXECUTED,
    "loss_calculated":
        NOTEBOOK_10_BLOCK_7_LOSS_CALCULATED,
    "optimizer_created":
        NOTEBOOK_10_BLOCK_7_OPTIMIZER_CREATED,
    "zero_grad_executed":
        NOTEBOOK_10_BLOCK_7_ZERO_GRAD_EXECUTED,
    "backward_pass_executed":
        NOTEBOOK_10_BLOCK_7_BACKWARD_PASS_EXECUTED,
    "gradient_clipping_executed":
        NOTEBOOK_10_BLOCK_7_GRADIENT_CLIPPING_EXECUTED,
    "optimizer_step_executed":
        NOTEBOOK_10_BLOCK_7_OPTIMIZER_STEP_EXECUTED,
    "parameter_update_executed":
        NOTEBOOK_10_BLOCK_7_PARAMETER_UPDATE_EXECUTED,
    "training_executed":
        NOTEBOOK_10_BLOCK_7_TRAINING_EXECUTED,
    "future_context_used":
        NOTEBOOK_10_BLOCK_7_FUTURE_CONTEXT_USED,
    "supervision_used_as_model_input":
        NOTEBOOK_10_BLOCK_7_SUPERVISION_USED_AS_MODEL_INPUT,
}


for error_name, operation_executed in BLOCK_7_PROHIBITED_OPERATIONS.items():

    if operation_executed:

        NOTEBOOK_10_BLOCK_7_ERRORS.append(
            error_name
        )


NOTEBOOK_10_EXPANDED_SUPERVISION_ALIGNMENT_VALID = (
    len(
        NOTEBOOK_10_BLOCK_7_ERRORS
    )
    ==
    0
)


NOTEBOOK_10_TRANSFORMATION_TARGET_CONTRACT_VALID = (
    NOTEBOOK_10_EXPANDED_SUPERVISION_ALIGNMENT_VALID
)


NOTEBOOK_10_EXPANDED_OBJECTIVE_EVALUATION_READY = (
    NOTEBOOK_10_TRANSFORMATION_TARGET_CONTRACT_VALID
)


NOTEBOOK_10_BLOCK_7_VALID = (
    NOTEBOOK_10_EXPANDED_SUPERVISION_ALIGNMENT_VALID
)


if not NOTEBOOK_10_BLOCK_7_VALID:

    raise RuntimeError(
        "Notebook 10 Block 7 expanded supervision alignment and "
        "transformation-target construction failed. "
        f"Errors: {NOTEBOOK_10_BLOCK_7_ERRORS}"
    )


NOTEBOOK_10_BLOCK_7_COMPLETE = True


# =============================================================================
# Block summary
# =============================================================================

NOTEBOOK_10_BLOCK_7_SUMMARY = {
    "block":
        NOTEBOOK_10_BLOCK_7,
    "block_name":
        NOTEBOOK_10_BLOCK_7_NAME,
    "version":
        NOTEBOOK_10_BLOCK_7_VERSION,
    "executed_at_utc":
        BLOCK_7_EXECUTED_AT_UTC,
    "article_count":
        BLOCK_7_ARTICLE_COUNT,
    "sentence_count":
        BLOCK_7_SENTENCE_COUNT,
    "block_6_handover_valid":
        BLOCK_7_BLOCK_6_HANDOVER_VALID,
    "pathway_dimensions":
        deepcopy(
            BLOCK_7_PATHWAY_DIMENSIONS
        ),
    "active_dimension_indices":
        deepcopy(
            BLOCK_7_ACTIVE_DIMENSION_INDICES
        ),
    "deferred_dimension_indices":
        deepcopy(
            BLOCK_7_DEFERRED_DIMENSION_INDICES
        ),
    "current_available_elements":
        deepcopy(
            BLOCK_7_CURRENT_AVAILABLE_ELEMENTS
        ),
    "transformation_available_elements":
        deepcopy(
            BLOCK_7_TRANSFORMATION_AVAILABLE_ELEMENTS
        ),
    "objective_active_elements":
        deepcopy(
            BLOCK_7_OBJECTIVE_ACTIVE_ELEMENTS
        ),
    "supervised_sentence_counts":
        deepcopy(
            BLOCK_7_SUPERVISED_SENTENCE_COUNTS
        ),
    "supervised_article_counts":
        deepcopy(
            BLOCK_7_SUPERVISED_ARTICLE_COUNTS
        ),
    "causal_target_contract_valid":
        BLOCK_7_CAUSAL_TARGET_CONTRACT_VALID,
    "model_state_valid":
        BLOCK_7_MODEL_STATE_VALID,
    "parameter_accounting_valid":
        BLOCK_7_PARAMETER_ACCOUNTING_VALID,
    "expanded_objective_evaluation_ready":
        NOTEBOOK_10_EXPANDED_OBJECTIVE_EVALUATION_READY,
    "block_valid":
        NOTEBOOK_10_BLOCK_7_VALID,
    "block_complete":
        NOTEBOOK_10_BLOCK_7_COMPLETE,
}


# =============================================================================
# Final report
# =============================================================================

print("=" * 72)

print(
    "Media AI — Notebook 10, Block 7: "
    "Expanded Supervision Alignment and "
    "Transformation-Target Construction"
)

print("=" * 72)

print(
    f"Block version                : "
    f"{NOTEBOOK_10_BLOCK_7_VERSION}"
)

print("-" * 72)

print(
    "Recovered supervision"
)

print(
    f"Factual source               : "
    f"{BLOCK_7_FACTUAL_FILENAME}"
)

print(
    f"Factual Drive file ID        : "
    f"{BLOCK_7_FACTUAL_DRIVE_FILE_ID}"
)

print(
    f"Psychological source         : "
    f"{BLOCK_7_PSYCHOLOGICAL_SOURCE_NAME}"
)

print(
    f"Psychological Drive file ID  : "
    f"{BLOCK_7_PSYCHOLOGICAL_DRIVE_FILE_ID}"
)

print(
    f"Social source                : "
    f"{BLOCK_7_SOCIAL_FILENAME}"
)

print(
    f"Social Drive file ID         : "
    f"{BLOCK_7_SOCIAL_DRIVE_FILE_ID}"
)

print(
    f"Social dimensions            : "
    f"{list(BLOCK_7_SOCIAL_DIMENSION_ORDER)}"
)

print("-" * 72)

print(
    "Canonical corpus alignment"
)

print(
    f"Articles                     : "
    f"{BLOCK_7_ARTICLE_COUNT}"
)

print(
    f"Sentences                    : "
    f"{BLOCK_7_SENTENCE_COUNT}"
)

print(
    f"Sentence IDs unique          : "
    f"{len(set(BLOCK_7_SENTENCE_IDS)) == BLOCK_7_SENTENCE_COUNT}"
)

print(
    f"Canonical order valid        : "
    f"{BLOCK_7_CANONICAL_SEQUENCE_VALID}"
)

print(
    f"Article boundaries explicit  : "
    f"{bool(BLOCK_7_ARTICLE_RECORD_INDICES)}"
)

print(
    f"Block 6 handover valid       : "
    f"{BLOCK_7_BLOCK_6_HANDOVER_VALID}"
)

print("-" * 72)

print(
    "Current-state supervision"
)

for pathway_name, pathway_dim in BLOCK_7_PATHWAY_DIMENSIONS.items():

    possible_elements = (
        BLOCK_7_SENTENCE_COUNT
        *
        pathway_dim
    )

    print(
        f"{pathway_name:<14} shape              : "
        f"{BLOCK_7_CURRENT_SUPERVISION[pathway_name].shape}"
    )

    print(
        f"{pathway_name:<14} available          : "
        f"{BLOCK_7_CURRENT_AVAILABLE_ELEMENTS[pathway_name]} "
        f"/ {possible_elements}"
    )

    print(
        f"{pathway_name:<14} supervised sentences: "
        f"{BLOCK_7_SUPERVISED_SENTENCE_COUNTS[pathway_name]} "
        f"/ {BLOCK_7_SENTENCE_COUNT}"
    )

    print(
        f"{pathway_name:<14} supervised articles : "
        f"{BLOCK_7_SUPERVISED_ARTICLE_COUNTS[pathway_name]} "
        f"/ {BLOCK_7_ARTICLE_COUNT}"
    )

print("-" * 72)

print(
    "Derived transformation supervision"
)

for pathway_name, pathway_dim in BLOCK_7_PATHWAY_DIMENSIONS.items():

    possible_elements = (
        BLOCK_7_SENTENCE_COUNT
        *
        pathway_dim
    )

    print(
        f"{pathway_name:<14} available          : "
        f"{BLOCK_7_TRANSFORMATION_AVAILABLE_ELEMENTS[pathway_name]} "
        f"/ {possible_elements}"
    )

    print(
        f"{pathway_name:<14} objective active   : "
        f"{BLOCK_7_OBJECTIVE_ACTIVE_ELEMENTS[pathway_name]}"
    )

    print(
        f"{pathway_name:<14} active dimensions  : "
        f"{list(BLOCK_7_ACTIVE_DIMENSION_INDICES[pathway_name])}"
    )

    print(
        f"{pathway_name:<14} deferred dimensions: "
        f"{list(BLOCK_7_DEFERRED_DIMENSION_INDICES[pathway_name])}"
    )

print("-" * 72)

print(
    "Causal target validation"
)

print(
    f"Zero article initial condition: "
    f"{BLOCK_7_FIRST_SENTENCE_BOUNDARY_MASK_VALID}"
)

print(
    f"Cross-article targets        : "
    f"{not BLOCK_7_NO_CROSS_ARTICLE_TARGETS}"
)

print(
    f"Future context used          : "
    f"{NOTEBOOK_10_BLOCK_7_FUTURE_CONTEXT_USED}"
)

print(
    f"Supervision used as input    : "
    f"{NOTEBOOK_10_BLOCK_7_SUPERVISION_USED_AS_MODEL_INPUT}"
)

print(
    f"Causal target contract valid : "
    f"{BLOCK_7_CAUSAL_TARGET_CONTRACT_VALID}"
)

print("-" * 72)

print(
    "Inherited Media AI state"
)

print(
    f"Representation parameters    : "
    f"{BLOCK_7_REPRESENTATION_PARAMETER_COUNT:,}"
)

print(
    f"Transformative parameters    : "
    f"{BLOCK_7_TRANSFORMATIVE_PARAMETER_COUNT:,}"
)

print(
    f"Transformative Weight params : "
    f"{BLOCK_7_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT:,}"
)

print(
    f"Total model parameters       : "
    f"{BLOCK_7_TOTAL_PARAMETER_COUNT:,}"
)

print(
    f"Representation unchanged     : "
    f"{BLOCK_7_REPRESENTATION_UNCHANGED}"
)

print(
    f"Transformative unchanged     : "
    f"{BLOCK_7_TRANSFORMATIVE_UNCHANGED}"
)

print(
    f"Transform Weight unchanged   : "
    f"{BLOCK_7_WEIGHT_UNCHANGED}"
)

print(
    f"All parameters frozen        : "
    f"{BLOCK_7_ALL_PARAMETERS_FROZEN_AFTER}"
)

print(
    f"All modules eval             : "
    f"{BLOCK_7_ALL_MODULES_EVAL_AFTER}"
)

print(
    f"All gradients absent         : "
    f"{BLOCK_7_ALL_GRADIENTS_ABSENT_AFTER}"
)

print("-" * 72)

print(
    "Execution boundary"
)

print(
    f"Supervision recovered        : "
    f"{NOTEBOOK_10_BLOCK_7_SUPERVISION_RECOVERED}"
)

print(
    f"Supervision aligned          : "
    f"{NOTEBOOK_10_BLOCK_7_SUPERVISION_ALIGNED}"
)

print(
    f"Transformation targets built : "
    f"{NOTEBOOK_10_BLOCK_7_TRANSFORMATION_TARGETS_CONSTRUCTED}"
)

print(
    f"Objective masks built        : "
    f"{NOTEBOOK_10_BLOCK_7_OBJECTIVE_MASKS_CONSTRUCTED}"
)

print(
    f"Representation forward       : "
    f"{NOTEBOOK_10_BLOCK_7_REPRESENTATION_FORWARD_EXECUTED}"
)

print(
    f"Transform forward            : "
    f"{NOTEBOOK_10_BLOCK_7_TRANSFORMATIVE_FORWARD_EXECUTED}"
)

print(
    f"Weight forward               : "
    f"{NOTEBOOK_10_BLOCK_7_WEIGHT_FORWARD_EXECUTED}"
)

print(
    f"Recurrent update executed    : "
    f"{NOTEBOOK_10_BLOCK_7_RECURRENT_UPDATE_EXECUTED}"
)

print(
    f"Loss calculated              : "
    f"{NOTEBOOK_10_BLOCK_7_LOSS_CALCULATED}"
)

print(
    f"Optimizer created            : "
    f"{NOTEBOOK_10_BLOCK_7_OPTIMIZER_CREATED}"
)

print(
    f"Backward pass executed       : "
    f"{NOTEBOOK_10_BLOCK_7_BACKWARD_PASS_EXECUTED}"
)

print(
    f"Parameter update executed    : "
    f"{NOTEBOOK_10_BLOCK_7_PARAMETER_UPDATE_EXECUTED}"
)

print(
    f"Training executed            : "
    f"{NOTEBOOK_10_BLOCK_7_TRAINING_EXECUTED}"
)

print("-" * 72)

print(
    f"Expanded supervision valid   : "
    f"{NOTEBOOK_10_EXPANDED_SUPERVISION_ALIGNMENT_VALID}"
)

print(
    f"Target contract valid        : "
    f"{NOTEBOOK_10_TRANSFORMATION_TARGET_CONTRACT_VALID}"
)

print(
    f"Objective evaluation ready   : "
    f"{NOTEBOOK_10_EXPANDED_OBJECTIVE_EVALUATION_READY}"
)

print(
    f"Block valid                  : "
    f"{NOTEBOOK_10_BLOCK_7_VALID}"
)

print(
    f"Block complete               : "
    f"{NOTEBOOK_10_BLOCK_7_COMPLETE}"
)

print("=" * 72)

print(
    f"Validated factual, psychological and social supervision was aligned "
    f"to the canonical Notebook 10 corpus of {BLOCK_7_SENTENCE_COUNT} "
    f"sentence(s) across {BLOCK_7_ARTICLE_COUNT} article(s)."
)

print(
    "Missing supervision remains explicitly masked and distinct from "
    "genuine zero-valued supervision."
)

print(
    "Signed transformation targets were constructed as the current "
    "supervised state minus the immediately preceding supervised state "
    "within each article."
)

print(
    "Every article uses an independent deterministic zero preceding "
    "condition for its first sentence; no transformation target crosses "
    "an article boundary."
)

print(
    "Objective masks combine transformation-target availability with the "
    "inherited active/deferred Transformative-Weight dimensional contract."
)

print(
    "Deferred dimensions remain structurally present but do not contribute "
    "to the future transformation objective."
)

print(
    f"The inherited {BLOCK_7_TOTAL_PARAMETER_COUNT:,}-parameter Media AI "
    "architecture remained frozen, in evaluation mode and exactly unchanged."
)

print(
    "No model forward pass, recurrent update, loss calculation, optimiser, "
    "backward pass, gradient clipping, parameter update or training operation "
    "was executed by Block 7."
)

print(
    "Notebook 10 may now proceed to Block 8 — Expanded Recurrent Objective "
    "Evaluation and Pre-Training Baseline Construction."
)

print("=" * 72)
```

# Block 8 — Expanded Recurrent Objective Evaluation and Pre-Training Baseline Construction

## Purpose

This block evaluates the complete supervised recurrent objective of the validated Notebook 10 corpus before any additional parameter optimisation is permitted.

Blocks 5–7 established that the current corpus is structurally valid, representation-ready, recurrently executable and aligned with validated factual, psychological and social supervision. Block 8 now combines those contracts into the first complete supervised objective evaluation of Notebook 10.

The purpose is not to train the architecture.

Instead, the block establishes an **immutable pre-training reference point** against which all subsequent parameter updates can be evaluated.

The resulting baseline separates the state of the inherited model before Notebook 10 optimisation from the state produced by the controlled training stage that follows.

---

## Methodological Role

For each article, recurrent execution begins from an independently constructed deterministic zero state.

For sentence position \(t\) and pathway \(p\), the frozen transformative mechanism receives the current pathway representation and the immediately preceding recurrent state:

$$
\mathbf{c}^{(p)}_t
=
T^{(p)}
\left(
\mathbf{r}^{(p)}_t,
\mathbf{h}^{(p)}_{t-1}
\right)
$$

where:

- \(\mathbf{r}^{(p)}_t\) is the current sentence representation in pathway \(p\);
- \(\mathbf{h}^{(p)}_{t-1}\) is the preceding recurrent state;
- \(T^{(p)}\) is the inherited transformative mechanism;
- \(\mathbf{c}^{(p)}_t\) is the candidate transformation.

The learned Transformative-Weight vector is then applied dimension-wise:

$$
\widetilde{\mathbf{c}}^{(p)}_t
=
\mathbf{w}^{(p)}
\odot
\mathbf{c}^{(p)}_t
$$

where \(\mathbf{w}^{(p)}\) is the effective Transformative-Weight vector and \(\odot\) denotes element-wise multiplication.

The recurrent state follows the validated additive update:

$$
\mathbf{h}^{(p)}_t
=
\mathbf{h}^{(p)}_{t-1}
+
\widetilde{\mathbf{c}}^{(p)}_t
$$

No recurrent state may propagate across article boundaries.

---

## Transformation-Supervision Objective

Block 7 constructed signed supervision targets describing the observed change between consecutive supervised states:

$$
\Delta\mathbf{y}^{(p)}_t
=
\mathbf{y}^{(p)}_t
-
\mathbf{y}^{(p)}_{t-1}
$$

with an independent deterministic zero preceding condition for the first sentence of every article.

The modelled transformation is therefore compared directly with the validated transformation target:

$$
\widetilde{\mathbf{c}}^{(p)}_t
\approx
\Delta\mathbf{y}^{(p)}_t
$$

This comparison evaluates whether the candidate transformation produced by the frozen transformative mechanism, after application of the inherited Transformative Weights, corresponds to the supervised change associated with the current sentence.

Supervision is used **only as an objective target**. It is never supplied to the representation, transformative or recurrent forward path as model input.

---

## Mask-Aware Pathway Loss

Supervision availability is incomplete and must remain distinct from genuine zero-valued supervision.

For pathway \(p\), sentence \(t\), and semantic dimension \(j\), let

$$
m^{(p)}_{t,j} \in \{0,1\}
$$

denote the objective mask constructed in Block 7.

The mask is active only when:

1. the transformation target is valid and available; and
2. the semantic dimension is currently eligible under the Transformative-Weight activation contract.

The pathway objective is the masked mean-squared error

$$
\mathcal{L}^{(p)}
=
\frac{
\sum_{t,j}
m^{(p)}_{t,j}
\left(
\widetilde{c}^{(p)}_{t,j}
-
\Delta y^{(p)}_{t,j}
\right)^2
}{
\sum_{t,j}m^{(p)}_{t,j}
}
$$

for the factual, psychological and social pathways independently.

Missing supervision contributes nothing to either the numerator or denominator.

Deferred Transformative-Weight dimensions remain structurally represented but are excluded from the objective while their activation status remains deferred.

---

## Combined Pre-Training Objective

The three pathway losses are combined using equal pathway weighting:

$$
\mathcal{L}_{\mathrm{pre}}
=
\frac{1}{3}
\left(
\mathcal{L}^{(\mathrm{factual})}
+
\mathcal{L}^{(\mathrm{psychological})}
+
\mathcal{L}^{(\mathrm{social})}
\right)
$$

Equal pathway weighting prevents differences in pathway dimensionality or supervision density from causing the 34-dimensional psychological pathway, for example, to dominate the factual or social objectives merely because it contains more supervised elements.

The resulting value is the canonical **Notebook 10 pre-training objective**.

It must be persisted in runtime together with the individual pathway losses and objective-active element counts so that subsequent training can be evaluated against exactly the same reference state.

---

## Baseline Immutability

Block 8 establishes the final architectural state immediately before Notebook 10 optimisation.

Accordingly, the block must verify that the complete inherited parameterisation remains unchanged:

- **894,003 representation parameters**;
- **8,187 transformative parameters**;
- **49 active Transformative-Weight parameters**;
- **902,239 total model parameters**.

The representation and transformative modules remain frozen.

The Transformative-Weight parameters must also remain unchanged throughout this block even though they will become the optimisation target in the following training stage.

The effective Transformative-Weight vectors evaluated in Block 8 must therefore reproduce exactly the validated post-Notebook-09 weights inherited by Notebook 10.

---

## Causal and Article-Boundary Contract

Objective evaluation must preserve the causal architecture established previously.

Each article is evaluated independently:

$$
\mathbf{h}^{(p)}_{0}=\mathbf{0}
$$

for every pathway \(p\).

Within an article, recurrent state may propagate only from sentence \(t-1\) to sentence \(t\).

The block must therefore verify that:

- canonical sentence order is preserved;
- every article begins from a fresh zero recurrent state;
- no recurrent state crosses an article boundary;
- no future sentence representation is accessed;
- no supervision target is used as model input;
- pathway recurrence remains independent;
- factual, psychological and social objective masks remain aligned with the canonical sentence sequence.

---

## Deterministic Baseline Validation

The complete objective evaluation should be reproducible under the established deterministic execution contract.

Repeated evaluation from identical article-boundary initial states must reproduce:

- candidate transformations;
- effective Transformative-Weight vectors;
- weighted candidate transformations;
- recurrent trajectories;
- pathway losses;
- the combined pre-training objective.

This provides a stable reference against which Block 9 parameter updates can subsequently be measured.

---

This block **may execute**:

- frozen representation forward passes where required by the validated corpus contract;

- Transformative-Weight forward passes;

- deterministic repeated objective evaluation.

This block **must not execute**:

- backward propagation;
- gradient clipping;
- optimiser steps;
- parameter updates;
- training epochs.

No parameter may change during Block 8.

The objective calculated here is a **measurement**, not a training operation.

---

```python
# =============================================================================
# Media AI — Notebook 10
# Block 8: Expanded Recurrent Objective Evaluation
#          and Pre-Training Baseline Construction
# =============================================================================

from copy import deepcopy
from datetime import datetime, timezone

import torch


# =============================================================================
# Block identity
# =============================================================================

NOTEBOOK_10_BLOCK_8 = 8

NOTEBOOK_10_BLOCK_8_NAME = (
    "Expanded Recurrent Objective Evaluation "
    "and Pre-Training Baseline Construction"
)

NOTEBOOK_10_BLOCK_8_VERSION = "1.0"

BLOCK_8_EXECUTED_AT_UTC = (
    datetime.now(
        timezone.utc
    ).isoformat()
)


# =============================================================================
# Required Block 6 → 7 handover contract
# =============================================================================

BLOCK_8_REQUIRED_GLOBALS = [
    "NOTEBOOK_10_BLOCK_6_COMPLETE",
    "NOTEBOOK_10_BLOCK_6_VALID",
    "NOTEBOOK_10_MULTI_ARTICLE_RECURRENT_VALIDATION_VALID",
    "NOTEBOOK_10_BLOCK_7_COMPLETE",
    "NOTEBOOK_10_BLOCK_7_VALID",
    "NOTEBOOK_10_EXPANDED_SUPERVISION_ALIGNMENT_VALID",
    "NOTEBOOK_10_TRANSFORMATION_TARGET_CONTRACT_VALID",
    "NOTEBOOK_10_EXPANDED_OBJECTIVE_EVALUATION_READY",
    "NOTEBOOK_10_MULTI_ARTICLE_PATHWAY_REPRESENTATIONS",
    "NOTEBOOK_10_TRANSFORMATIVE_MODULES",
    "NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES",
    "NOTEBOOK_10_PATHWAY_DIMENSIONS",
    "NOTEBOOK_10_ACTIVE_DIMENSION_INDICES",
    "NOTEBOOK_10_DEFERRED_DIMENSION_INDICES",
    "NOTEBOOK_10_INHERITED_EFFECTIVE_TRANSFORMATIVE_WEIGHTS",
    "NOTEBOOK_10_TRANSFORMATION_TARGETS",
    "NOTEBOOK_10_TRANSFORMATION_OBJECTIVE_MASKS",
    "NOTEBOOK_10_SUPERVISION_CANONICAL_ARTICLE_IDS",
    "NOTEBOOK_10_SUPERVISION_CANONICAL_SENTENCE_IDS",
    "NOTEBOOK_10_SUPERVISION_ARTICLE_RECORD_INDICES",
    "NOTEBOOK_10_REPRESENTATION_MODULES",
]


BLOCK_8_MISSING_PREREQUISITES = [
    object_name

    for object_name
    in BLOCK_8_REQUIRED_GLOBALS

    if object_name not in globals()
]


if BLOCK_8_MISSING_PREREQUISITES:

    raise RuntimeError(
        "Notebook 10 Block 8 cannot execute because the current runtime "
        "does not contain the exact validated Block 6 → Block 7 handover. "
        "If the runtime was restarted, rerun Notebook 10 Blocks 1–7 in "
        "order in this same runtime. "
        f"Missing objects: {BLOCK_8_MISSING_PREREQUISITES}"
    )


BLOCK_8_PREREQUISITES_VALID = all(
    [
        NOTEBOOK_10_BLOCK_6_COMPLETE is True,
        NOTEBOOK_10_BLOCK_6_VALID is True,
        NOTEBOOK_10_MULTI_ARTICLE_RECURRENT_VALIDATION_VALID is True,
        NOTEBOOK_10_BLOCK_7_COMPLETE is True,
        NOTEBOOK_10_BLOCK_7_VALID is True,
        NOTEBOOK_10_EXPANDED_SUPERVISION_ALIGNMENT_VALID is True,
        NOTEBOOK_10_TRANSFORMATION_TARGET_CONTRACT_VALID is True,
        NOTEBOOK_10_EXPANDED_OBJECTIVE_EVALUATION_READY is True,
        isinstance(
            NOTEBOOK_10_MULTI_ARTICLE_PATHWAY_REPRESENTATIONS,
            dict,
        ),
        isinstance(
            NOTEBOOK_10_TRANSFORMATION_TARGETS,
            dict,
        ),
        isinstance(
            NOTEBOOK_10_TRANSFORMATION_OBJECTIVE_MASKS,
            dict,
        ),
    ]
)


if not BLOCK_8_PREREQUISITES_VALID:

    raise RuntimeError(
        "Notebook 10 Block 8 requires valid completed Blocks 6 and 7."
    )


# =============================================================================
# Canonical pathway contract
# =============================================================================

BLOCK_8_PATHWAY_DIMENSIONS = {
    pathway_name:
        int(
            pathway_dim
        )

    for pathway_name, pathway_dim
    in NOTEBOOK_10_PATHWAY_DIMENSIONS.items()
}


BLOCK_8_EXPECTED_PATHWAY_DIMENSIONS = {
    "factual":
        10,
    "psychological":
        34,
    "social":
        7,
}


BLOCK_8_PATHWAY_DIMENSIONS_VALID = (
    BLOCK_8_PATHWAY_DIMENSIONS
    ==
    BLOCK_8_EXPECTED_PATHWAY_DIMENSIONS
)


BLOCK_8_ACTIVE_DIMENSION_INDICES = {
    pathway_name:
        tuple(
            int(
                index
            )
            for index
            in NOTEBOOK_10_ACTIVE_DIMENSION_INDICES[
                pathway_name
            ]
        )

    for pathway_name
    in BLOCK_8_PATHWAY_DIMENSIONS
}


BLOCK_8_DEFERRED_DIMENSION_INDICES = {
    pathway_name:
        tuple(
            int(
                index
            )
            for index
            in NOTEBOOK_10_DEFERRED_DIMENSION_INDICES[
                pathway_name
            ]
        )

    for pathway_name
    in BLOCK_8_PATHWAY_DIMENSIONS
}


if not BLOCK_8_PATHWAY_DIMENSIONS_VALID:

    raise RuntimeError(
        "Notebook 10 Block 8 pathway dimensions do not reproduce the "
        "validated Notebook 09 → 10 architecture."
    )


# =============================================================================
# Canonical article/sentence contract
# =============================================================================

BLOCK_8_ARTICLE_IDS = tuple(
    NOTEBOOK_10_SUPERVISION_CANONICAL_ARTICLE_IDS
)


BLOCK_8_SENTENCE_IDS = tuple(
    NOTEBOOK_10_SUPERVISION_CANONICAL_SENTENCE_IDS
)


BLOCK_8_ARTICLE_RECORD_INDICES = deepcopy(
    NOTEBOOK_10_SUPERVISION_ARTICLE_RECORD_INDICES
)


BLOCK_8_ARTICLE_COUNT = len(
    BLOCK_8_ARTICLE_IDS
)


BLOCK_8_SENTENCE_COUNT = len(
    BLOCK_8_SENTENCE_IDS
)


BLOCK_8_ARTICLE_INDEX_CONTRACT_VALID = all(
    [
        BLOCK_8_ARTICLE_COUNT > 0,
        BLOCK_8_SENTENCE_COUNT > 0,
        len(
            set(
                BLOCK_8_SENTENCE_IDS
            )
        )
        ==
        BLOCK_8_SENTENCE_COUNT,
        set(
            BLOCK_8_ARTICLE_RECORD_INDICES.keys()
        )
        ==
        set(
            BLOCK_8_ARTICLE_IDS
        ),
        sum(
            len(
                indices
            )
            for indices
            in BLOCK_8_ARTICLE_RECORD_INDICES.values()
        )
        ==
        BLOCK_8_SENTENCE_COUNT,
    ]
)


if not BLOCK_8_ARTICLE_INDEX_CONTRACT_VALID:

    raise RuntimeError(
        "Notebook 10 Block 8 canonical article/sentence index contract "
        "is invalid."
    )


# =============================================================================
# Resolve execution device and dtype from restored modules
# =============================================================================

BLOCK_8_MODEL_PARAMETERS = [
    parameter

    for module
    in NOTEBOOK_10_REPRESENTATION_MODULES.values()

    for parameter
    in module.parameters()
]


if not BLOCK_8_MODEL_PARAMETERS:

    raise RuntimeError(
        "Notebook 10 Block 8 could not resolve model device/dtype."
    )


BLOCK_8_EXECUTION_DEVICE = (
    BLOCK_8_MODEL_PARAMETERS[
        0
    ].device
)


BLOCK_8_EXECUTION_DTYPE = (
    BLOCK_8_MODEL_PARAMETERS[
        0
    ].dtype
)


BLOCK_8_DEVICE_DTYPE_VALID = all(
    parameter.device
    ==
    BLOCK_8_EXECUTION_DEVICE
    and
    parameter.dtype
    ==
    BLOCK_8_EXECUTION_DTYPE

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


if not BLOCK_8_DEVICE_DTYPE_VALID:

    raise RuntimeError(
        "Notebook 10 Block 8 inherited modules do not share a "
        "consistent execution device and dtype."
    )


# =============================================================================
# Execution boundary
# =============================================================================

NOTEBOOK_10_BLOCK_8_REPRESENTATION_FORWARD_EXECUTED = False

NOTEBOOK_10_BLOCK_8_TRANSFORMATIVE_FORWARD_EXECUTED = False

NOTEBOOK_10_BLOCK_8_WEIGHT_FORWARD_EXECUTED = False

NOTEBOOK_10_BLOCK_8_RECURRENT_STATE_CREATED = False

NOTEBOOK_10_BLOCK_8_RECURRENT_UPDATE_EXECUTED = False

NOTEBOOK_10_BLOCK_8_LOSS_CALCULATED = False

NOTEBOOK_10_BLOCK_8_REPEAT_OBJECTIVE_EVALUATED = False

NOTEBOOK_10_BLOCK_8_OPTIMIZER_CREATED = False

NOTEBOOK_10_BLOCK_8_ZERO_GRAD_EXECUTED = False

NOTEBOOK_10_BLOCK_8_BACKWARD_PASS_EXECUTED = False

NOTEBOOK_10_BLOCK_8_GRADIENT_CLIPPING_EXECUTED = False

NOTEBOOK_10_BLOCK_8_OPTIMIZER_STEP_EXECUTED = False

NOTEBOOK_10_BLOCK_8_PARAMETER_UPDATE_EXECUTED = False

NOTEBOOK_10_BLOCK_8_TRAINING_EXECUTED = False

NOTEBOOK_10_BLOCK_8_FUTURE_CONTEXT_USED = False

NOTEBOOK_10_BLOCK_8_SUPERVISION_USED_AS_MODEL_INPUT = False


# =============================================================================
# Model-state snapshot helpers
# =============================================================================

def block_8_snapshot_module_collection(
    module_collection,
):

    return {
        module_name:
            {
                tensor_name:
                    tensor.detach()
                    .cpu()
                    .clone()

                for tensor_name, tensor
                in module.state_dict().items()
            }

        for module_name, module
        in module_collection.items()
    }


def block_8_nested_state_exact(
    state_before,
    state_after,
):

    if state_before.keys() != state_after.keys():

        return False

    for module_name in state_before:

        if (
            state_before[
                module_name
            ].keys()
            !=
            state_after[
                module_name
            ].keys()
        ):

            return False

        for tensor_name in state_before[
            module_name
        ]:

            if not torch.equal(
                state_before[
                    module_name
                ][
                    tensor_name
                ],
                state_after[
                    module_name
                ][
                    tensor_name
                ],
            ):

                return False

    return True


BLOCK_8_REPRESENTATION_STATE_BEFORE = (
    block_8_snapshot_module_collection(
        NOTEBOOK_10_REPRESENTATION_MODULES
    )
)


BLOCK_8_TRANSFORMATIVE_STATE_BEFORE = (
    block_8_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_MODULES
    )
)


BLOCK_8_WEIGHT_STATE_BEFORE = (
    block_8_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES
    )
)


# =============================================================================
# Frozen/evaluation-state audit before objective evaluation
# =============================================================================

BLOCK_8_ALL_PARAMETERS_FROZEN_BEFORE = all(
    not parameter.requires_grad

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


BLOCK_8_ALL_MODULES_EVAL_BEFORE = all(
    not module.training

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()
)


BLOCK_8_ALL_GRADIENTS_ABSENT_BEFORE = all(
    parameter.grad is None

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


if not all(
    [
        BLOCK_8_ALL_PARAMETERS_FROZEN_BEFORE,
        BLOCK_8_ALL_MODULES_EVAL_BEFORE,
        BLOCK_8_ALL_GRADIENTS_ABSENT_BEFORE,
    ]
):

    raise RuntimeError(
        "Notebook 10 Block 8 inherited architecture is not in the "
        "required frozen/evaluation/no-gradient state."
    )


# =============================================================================
# Parameter accounting
# =============================================================================

BLOCK_8_REPRESENTATION_PARAMETER_COUNT = sum(
    parameter.numel()

    for module
    in NOTEBOOK_10_REPRESENTATION_MODULES.values()

    for parameter
    in module.parameters()
)


BLOCK_8_TRANSFORMATIVE_PARAMETER_COUNT = sum(
    parameter.numel()

    for module
    in NOTEBOOK_10_TRANSFORMATIVE_MODULES.values()

    for parameter
    in module.parameters()
)


BLOCK_8_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT = sum(
    parameter.numel()

    for module
    in NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES.values()

    for parameter
    in module.parameters()
)


BLOCK_8_TOTAL_PARAMETER_COUNT = (
    BLOCK_8_REPRESENTATION_PARAMETER_COUNT
    +
    BLOCK_8_TRANSFORMATIVE_PARAMETER_COUNT
    +
    BLOCK_8_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT
)


BLOCK_8_PARAMETER_ACCOUNTING_VALID = all(
    [
        BLOCK_8_REPRESENTATION_PARAMETER_COUNT
        ==
        894_003,

        BLOCK_8_TRANSFORMATIVE_PARAMETER_COUNT
        ==
        8_187,

        BLOCK_8_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT
        ==
        49,

        BLOCK_8_TOTAL_PARAMETER_COUNT
        ==
        902_239,
    ]
)


if not BLOCK_8_PARAMETER_ACCOUNTING_VALID:

    raise RuntimeError(
        "Notebook 10 Block 8 parameter accounting changed unexpectedly."
    )


# =============================================================================
# Validate Block 6 pathway-representation matrices
# =============================================================================

BLOCK_8_PATHWAY_REPRESENTATIONS = deepcopy(
    NOTEBOOK_10_MULTI_ARTICLE_PATHWAY_REPRESENTATIONS
)


BLOCK_8_PATHWAY_REPRESENTATIONS_VALID = True


for article_id in BLOCK_8_ARTICLE_IDS:

    if article_id not in BLOCK_8_PATHWAY_REPRESENTATIONS:

        BLOCK_8_PATHWAY_REPRESENTATIONS_VALID = False

        continue

    article_indices = (
        BLOCK_8_ARTICLE_RECORD_INDICES[
            article_id
        ]
    )

    sentence_count = len(
        article_indices
    )

    for pathway_name, pathway_dim in BLOCK_8_PATHWAY_DIMENSIONS.items():

        if (
            pathway_name
            not in
            BLOCK_8_PATHWAY_REPRESENTATIONS[
                article_id
            ]
        ):

            BLOCK_8_PATHWAY_REPRESENTATIONS_VALID = False

            continue

        matrix = (
            BLOCK_8_PATHWAY_REPRESENTATIONS[
                article_id
            ][
                pathway_name
            ]
        )

        if not torch.is_tensor(
            matrix
        ):

            BLOCK_8_PATHWAY_REPRESENTATIONS_VALID = False

            continue

        if (
            tuple(
                matrix.shape
            )
            !=
            (
                sentence_count,
                pathway_dim,
            )
        ):

            BLOCK_8_PATHWAY_REPRESENTATIONS_VALID = False

        if not bool(
            torch.isfinite(
                matrix
            ).all()
            .item()
        ):

            BLOCK_8_PATHWAY_REPRESENTATIONS_VALID = False


if not BLOCK_8_PATHWAY_REPRESENTATIONS_VALID:

    raise RuntimeError(
        "Notebook 10 Block 8 inherited Block 6 pathway representations "
        "are invalid."
    )


# =============================================================================
# Validate Block 7 target and objective-mask tensors
# =============================================================================

BLOCK_8_TRANSFORMATION_TARGETS = {
    pathway_name:
        NOTEBOOK_10_TRANSFORMATION_TARGETS[
            pathway_name
        ]
        .detach()
        .cpu()
        .clone()

    for pathway_name
    in BLOCK_8_PATHWAY_DIMENSIONS
}


BLOCK_8_OBJECTIVE_MASKS = {
    pathway_name:
        NOTEBOOK_10_TRANSFORMATION_OBJECTIVE_MASKS[
            pathway_name
        ]
        .detach()
        .cpu()
        .clone()
        .to(
            dtype=
                torch.bool
        )

    for pathway_name
    in BLOCK_8_PATHWAY_DIMENSIONS
}


BLOCK_8_TARGET_MASK_CONTRACT_VALID = all(
    [
        all(
            (
                tuple(
                    BLOCK_8_TRANSFORMATION_TARGETS[
                        pathway_name
                    ].shape
                )
                ==
                (
                    BLOCK_8_SENTENCE_COUNT,
                    pathway_dim,
                )
            )
            and
            (
                tuple(
                    BLOCK_8_OBJECTIVE_MASKS[
                        pathway_name
                    ].shape
                )
                ==
                (
                    BLOCK_8_SENTENCE_COUNT,
                    pathway_dim,
                )
            )
            and
            bool(
                torch.isfinite(
                    BLOCK_8_TRANSFORMATION_TARGETS[
                        pathway_name
                    ][
                        BLOCK_8_OBJECTIVE_MASKS[
                            pathway_name
                        ]
                    ]
                ).all()
                .item()
            )

            for pathway_name, pathway_dim
            in BLOCK_8_PATHWAY_DIMENSIONS.items()
        ),

        all(
            not bool(
                BLOCK_8_OBJECTIVE_MASKS[
                    pathway_name
                ][
                    :,
                    list(
                        BLOCK_8_DEFERRED_DIMENSION_INDICES[
                            pathway_name
                        ]
                    )
                ].any()
                .item()
            )
            if BLOCK_8_DEFERRED_DIMENSION_INDICES[
                pathway_name
            ]
            else True

            for pathway_name
            in BLOCK_8_PATHWAY_DIMENSIONS
        ),
    ]
)


if not BLOCK_8_TARGET_MASK_CONTRACT_VALID:

    raise RuntimeError(
        "Notebook 10 Block 8 Block 7 target/mask contract is invalid."
    )


# =============================================================================
# Objective-active element counts
# =============================================================================

BLOCK_8_OBJECTIVE_ACTIVE_ELEMENTS = {
    pathway_name:
        int(
            BLOCK_8_OBJECTIVE_MASKS[
                pathway_name
            ].sum()
            .item()
        )

    for pathway_name
    in BLOCK_8_PATHWAY_DIMENSIONS
}


BLOCK_8_ALL_PATHWAYS_OBJECTIVE_ACTIVE = all(
    BLOCK_8_OBJECTIVE_ACTIVE_ELEMENTS[
        pathway_name
    ]
    >
    0

    for pathway_name
    in BLOCK_8_PATHWAY_DIMENSIONS
)


if not BLOCK_8_ALL_PATHWAYS_OBJECTIVE_ACTIVE:

    raise RuntimeError(
        "Notebook 10 Block 8 has no objective-active supervision for "
        "one or more pathways."
    )


# =============================================================================
# Masked mean-squared-error helper
# =============================================================================

def block_8_masked_mean_squared_error(
    prediction,
    target,
    mask,
):

    if prediction.shape != target.shape:

        raise RuntimeError(
            "Masked-MSE prediction and target shapes do not match."
        )

    if mask.shape != target.shape:

        raise RuntimeError(
            "Masked-MSE mask and target shapes do not match."
        )

    if not bool(
        mask.any()
        .item()
    ):

        raise RuntimeError(
            "Masked-MSE received an empty objective mask."
        )

    error = (
        prediction
        -
        target
    )

    squared_error = (
        error
        *
        error
    )

    return squared_error[
        mask
    ].mean()


# =============================================================================
# Complete causal recurrent-objective evaluation
# =============================================================================

def block_8_evaluate_objective(
    evaluation_label,
):

    pathway_weighted_predictions = {
        pathway_name:
            torch.zeros(
                (
                    BLOCK_8_SENTENCE_COUNT,
                    pathway_dim,
                ),
                dtype=
                    BLOCK_8_EXECUTION_DTYPE,
                device=
                    BLOCK_8_EXECUTION_DEVICE,
            )

        for pathway_name, pathway_dim
        in BLOCK_8_PATHWAY_DIMENSIONS.items()
    }

    recurrent_trajectories = {
        article_id:
            {}

        for article_id
        in BLOCK_8_ARTICLE_IDS
    }

    effective_weights = {}

    with torch.inference_mode():

        for pathway_name, pathway_dim in BLOCK_8_PATHWAY_DIMENSIONS.items():

            weight = (
                NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES[
                    pathway_name
                ]()
                .detach()
            )

            if (
                tuple(
                    weight.shape
                )
                !=
                (
                    pathway_dim,
                )
            ):

                raise RuntimeError(
                    f"Transformative-Weight output shape invalid for "
                    f"{pathway_name}."
                )

            if not bool(
                torch.isfinite(
                    weight
                ).all()
                .item()
            ):

                raise RuntimeError(
                    f"Transformative-Weight output is non-finite for "
                    f"{pathway_name}."
                )

            effective_weights[
                pathway_name
            ] = (
                weight.detach()
                .cpu()
                .clone()
            )

        for article_id in BLOCK_8_ARTICLE_IDS:

            article_indices = (
                BLOCK_8_ARTICLE_RECORD_INDICES[
                    article_id
                ]
            )

            if not article_indices:

                raise RuntimeError(
                    f"Notebook 10 Block 8 article {article_id!r} has "
                    "no canonical sentence indices."
                )

            for pathway_name, pathway_dim in BLOCK_8_PATHWAY_DIMENSIONS.items():

                representation_matrix = (
                    BLOCK_8_PATHWAY_REPRESENTATIONS[
                        article_id
                    ][
                        pathway_name
                    ]
                    .to(
                        device=
                            BLOCK_8_EXECUTION_DEVICE,
                        dtype=
                            BLOCK_8_EXECUTION_DTYPE,
                    )
                )

                if (
                    representation_matrix.shape[
                        0
                    ]
                    !=
                    len(
                        article_indices
                    )
                ):

                    raise RuntimeError(
                        f"Article/pathway representation length mismatch "
                        f"for {article_id!r} / {pathway_name}."
                    )

                weight = (
                    effective_weights[
                        pathway_name
                    ]
                    .to(
                        device=
                            BLOCK_8_EXECUTION_DEVICE,
                        dtype=
                            BLOCK_8_EXECUTION_DTYPE,
                    )
                    .reshape(
                        1,
                        pathway_dim,
                    )
                )

                recurrent_state = torch.zeros(
                    (
                        1,
                        pathway_dim,
                    ),
                    dtype=
                        BLOCK_8_EXECUTION_DTYPE,
                    device=
                        BLOCK_8_EXECUTION_DEVICE,
                )

                initial_state = (
                    recurrent_state.detach()
                    .cpu()
                    .clone()
                )

                steps = []

                for (
                    local_position,
                    global_index,
                ) in enumerate(
                    article_indices
                ):

                    current_representation = (
                        representation_matrix[
                            local_position:
                            local_position
                            +
                            1
                        ]
                    )

                    preceding_state = (
                        recurrent_state
                    )

                    candidate = (
                        NOTEBOOK_10_TRANSFORMATIVE_MODULES[
                            pathway_name
                        ](
                            current_representation,
                            preceding_state,
                        )
                    )

                    weighted_candidate = (
                        candidate
                        *
                        weight
                    )

                    next_state = (
                        preceding_state
                        +
                        weighted_candidate
                    )

                    pathway_weighted_predictions[
                        pathway_name
                    ][
                        global_index:
                        global_index
                        +
                        1
                    ] = weighted_candidate

                    steps.append(
                        {
                            "global_index":
                                int(
                                    global_index
                                ),
                            "candidate":
                                candidate.detach()
                                .cpu()
                                .clone(),
                            "weighted_candidate":
                                weighted_candidate.detach()
                                .cpu()
                                .clone(),
                            "preceding_state":
                                preceding_state.detach()
                                .cpu()
                                .clone(),
                            "recurrent_state":
                                next_state.detach()
                                .cpu()
                                .clone(),
                        }
                    )

                    recurrent_state = next_state

                recurrent_trajectories[
                    article_id
                ][
                    pathway_name
                ] = {
                    "initial_state":
                        initial_state,
                    "steps":
                        steps,
                    "final_state":
                        recurrent_state.detach()
                        .cpu()
                        .clone(),
                }

    pathway_losses = {}

    for pathway_name in BLOCK_8_PATHWAY_DIMENSIONS:

        prediction = (
            pathway_weighted_predictions[
                pathway_name
            ]
            .detach()
            .cpu()
        )

        target = (
            BLOCK_8_TRANSFORMATION_TARGETS[
                pathway_name
            ]
            .to(
                dtype=
                    prediction.dtype,
            )
        )

        mask = (
            BLOCK_8_OBJECTIVE_MASKS[
                pathway_name
            ]
        )

        pathway_losses[
            pathway_name
        ] = float(
            block_8_masked_mean_squared_error(
                prediction=
                    prediction,
                target=
                    target,
                mask=
                    mask,
            )
            .item()
        )

    total_objective = float(
        sum(
            pathway_losses.values()
        )
        /
        len(
            pathway_losses
        )
    )

    return {
        "evaluation_label":
            evaluation_label,
        "weighted_predictions":
            {
                pathway_name:
                    prediction.detach()
                    .cpu()
                    .clone()

                for pathway_name, prediction
                in pathway_weighted_predictions.items()
            },
        "effective_weights":
            effective_weights,
        "recurrent_trajectories":
            recurrent_trajectories,
        "pathway_losses":
            pathway_losses,
        "total_objective":
            total_objective,
    }


# =============================================================================
# First baseline objective evaluation
# =============================================================================

BLOCK_8_EVALUATION_A = (
    block_8_evaluate_objective(
        "baseline_a"
    )
)


NOTEBOOK_10_BLOCK_8_TRANSFORMATIVE_FORWARD_EXECUTED = True
NOTEBOOK_10_BLOCK_8_WEIGHT_FORWARD_EXECUTED = True
NOTEBOOK_10_BLOCK_8_RECURRENT_STATE_CREATED = True
NOTEBOOK_10_BLOCK_8_RECURRENT_UPDATE_EXECUTED = True
NOTEBOOK_10_BLOCK_8_LOSS_CALCULATED = True


# =============================================================================
# Deterministic repeated baseline evaluation
# =============================================================================

BLOCK_8_EVALUATION_B = (
    block_8_evaluate_objective(
        "baseline_b"
    )
)


NOTEBOOK_10_BLOCK_8_REPEAT_OBJECTIVE_EVALUATED = True


# =============================================================================
# Deterministic evaluation comparison
# =============================================================================

def block_8_evaluations_exact(
    evaluation_a,
    evaluation_b,
):

    if (
        evaluation_a[
            "pathway_losses"
        ]
        !=
        evaluation_b[
            "pathway_losses"
        ]
    ):

        return False

    if (
        evaluation_a[
            "total_objective"
        ]
        !=
        evaluation_b[
            "total_objective"
        ]
    ):

        return False

    for pathway_name in BLOCK_8_PATHWAY_DIMENSIONS:

        if not torch.equal(
            evaluation_a[
                "weighted_predictions"
            ][
                pathway_name
            ],
            evaluation_b[
                "weighted_predictions"
            ][
                pathway_name
            ],
        ):

            return False

        if not torch.equal(
            evaluation_a[
                "effective_weights"
            ][
                pathway_name
            ],
            evaluation_b[
                "effective_weights"
            ][
                pathway_name
            ],
        ):

            return False

    for article_id in BLOCK_8_ARTICLE_IDS:

        for pathway_name in BLOCK_8_PATHWAY_DIMENSIONS:

            trajectory_a = (
                evaluation_a[
                    "recurrent_trajectories"
                ][
                    article_id
                ][
                    pathway_name
                ]
            )

            trajectory_b = (
                evaluation_b[
                    "recurrent_trajectories"
                ][
                    article_id
                ][
                    pathway_name
                ]
            )

            if not torch.equal(
                trajectory_a[
                    "initial_state"
                ],
                trajectory_b[
                    "initial_state"
                ],
            ):

                return False

            if not torch.equal(
                trajectory_a[
                    "final_state"
                ],
                trajectory_b[
                    "final_state"
                ],
            ):

                return False

            if (
                len(
                    trajectory_a[
                        "steps"
                    ]
                )
                !=
                len(
                    trajectory_b[
                        "steps"
                    ]
                )
            ):

                return False

            for step_a, step_b in zip(
                trajectory_a[
                    "steps"
                ],
                trajectory_b[
                    "steps"
                ],
            ):

                for tensor_name in (
                    "candidate",
                    "weighted_candidate",
                    "preceding_state",
                    "recurrent_state",
                ):

                    if not torch.equal(
                        step_a[
                            tensor_name
                        ],
                        step_b[
                            tensor_name
                        ],
                    ):

                        return False

    return True


BLOCK_8_BASELINE_DETERMINISTIC = (
    block_8_evaluations_exact(
        BLOCK_8_EVALUATION_A,
        BLOCK_8_EVALUATION_B,
    )
)


# =============================================================================
# Objective values
# =============================================================================

BLOCK_8_PRETRAIN_PATHWAY_LOSSES = deepcopy(
    BLOCK_8_EVALUATION_A[
        "pathway_losses"
    ]
)


BLOCK_8_PRETRAIN_TOTAL_OBJECTIVE = float(
    BLOCK_8_EVALUATION_A[
        "total_objective"
    ]
)


BLOCK_8_OBJECTIVE_FINITE = all(
    [
        all(
            torch.isfinite(
                torch.tensor(
                    pathway_loss,
                    dtype=
                        torch.float64,
                )
            ).item()

            for pathway_loss
            in BLOCK_8_PRETRAIN_PATHWAY_LOSSES.values()
        ),

        torch.isfinite(
            torch.tensor(
                BLOCK_8_PRETRAIN_TOTAL_OBJECTIVE,
                dtype=
                    torch.float64,
            )
        ).item(),
    ]
)


BLOCK_8_EQUAL_PATHWAY_WEIGHTING_VALID = (
    BLOCK_8_PRETRAIN_TOTAL_OBJECTIVE
    ==
    (
        sum(
            BLOCK_8_PRETRAIN_PATHWAY_LOSSES.values()
        )
        /
        len(
            BLOCK_8_PRETRAIN_PATHWAY_LOSSES
        )
    )
)


# =============================================================================
# Article-boundary and causal validation
# =============================================================================

BLOCK_8_FRESH_ZERO_INITIAL_STATES = True
BLOCK_8_STATE_CHAIN_VALID = True
BLOCK_8_ADDITIVE_UPDATE_VALID = True


for article_id in BLOCK_8_ARTICLE_IDS:

    for pathway_name, pathway_dim in BLOCK_8_PATHWAY_DIMENSIONS.items():

        trajectory = (
            BLOCK_8_EVALUATION_A[
                "recurrent_trajectories"
            ][
                article_id
            ][
                pathway_name
            ]
        )

        initial_state = (
            trajectory[
                "initial_state"
            ]
        )

        if not torch.equal(
            initial_state,
            torch.zeros_like(
                initial_state
            ),
        ):

            BLOCK_8_FRESH_ZERO_INITIAL_STATES = False

        expected_preceding = initial_state

        for step in trajectory[
            "steps"
        ]:

            if not torch.equal(
                step[
                    "preceding_state"
                ],
                expected_preceding,
            ):

                BLOCK_8_STATE_CHAIN_VALID = False

            if not torch.equal(
                step[
                    "recurrent_state"
                ],
                step[
                    "preceding_state"
                ]
                +
                step[
                    "weighted_candidate"
                ],
            ):

                BLOCK_8_ADDITIVE_UPDATE_VALID = False

            expected_preceding = (
                step[
                    "recurrent_state"
                ]
            )


BLOCK_8_CAUSAL_CONTRACT_VALID = all(
    [
        BLOCK_8_FRESH_ZERO_INITIAL_STATES,
        BLOCK_8_STATE_CHAIN_VALID,
        BLOCK_8_ADDITIVE_UPDATE_VALID,
        NOTEBOOK_10_BLOCK_8_FUTURE_CONTEXT_USED is False,
        NOTEBOOK_10_BLOCK_8_SUPERVISION_USED_AS_MODEL_INPUT is False,
    ]
)


# =============================================================================
# Effective Transformative-Weight identity
# =============================================================================

BLOCK_8_EFFECTIVE_WEIGHT_IDENTITY_VALID = all(
    torch.equal(
        BLOCK_8_EVALUATION_A[
            "effective_weights"
        ][
            pathway_name
        ],
        NOTEBOOK_10_INHERITED_EFFECTIVE_TRANSFORMATIVE_WEIGHTS[
            pathway_name
        ]
        .detach()
        .cpu(),
    )

    for pathway_name
    in BLOCK_8_PATHWAY_DIMENSIONS
)


# =============================================================================
# Model-state preservation after objective evaluation
# =============================================================================

BLOCK_8_REPRESENTATION_STATE_AFTER = (
    block_8_snapshot_module_collection(
        NOTEBOOK_10_REPRESENTATION_MODULES
    )
)


BLOCK_8_TRANSFORMATIVE_STATE_AFTER = (
    block_8_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_MODULES
    )
)


BLOCK_8_WEIGHT_STATE_AFTER = (
    block_8_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES
    )
)


BLOCK_8_REPRESENTATION_UNCHANGED = (
    block_8_nested_state_exact(
        BLOCK_8_REPRESENTATION_STATE_BEFORE,
        BLOCK_8_REPRESENTATION_STATE_AFTER,
    )
)


BLOCK_8_TRANSFORMATIVE_UNCHANGED = (
    block_8_nested_state_exact(
        BLOCK_8_TRANSFORMATIVE_STATE_BEFORE,
        BLOCK_8_TRANSFORMATIVE_STATE_AFTER,
    )
)


BLOCK_8_WEIGHT_UNCHANGED = (
    block_8_nested_state_exact(
        BLOCK_8_WEIGHT_STATE_BEFORE,
        BLOCK_8_WEIGHT_STATE_AFTER,
    )
)


BLOCK_8_ALL_PARAMETERS_FROZEN_AFTER = all(
    not parameter.requires_grad

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


BLOCK_8_ALL_MODULES_EVAL_AFTER = all(
    not module.training

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()
)


BLOCK_8_ALL_GRADIENTS_ABSENT_AFTER = all(
    parameter.grad is None

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


BLOCK_8_MODEL_STATE_VALID = all(
    [
        BLOCK_8_REPRESENTATION_UNCHANGED,
        BLOCK_8_TRANSFORMATIVE_UNCHANGED,
        BLOCK_8_WEIGHT_UNCHANGED,
        BLOCK_8_EFFECTIVE_WEIGHT_IDENTITY_VALID,
        BLOCK_8_ALL_PARAMETERS_FROZEN_AFTER,
        BLOCK_8_ALL_MODULES_EVAL_AFTER,
        BLOCK_8_ALL_GRADIENTS_ABSENT_AFTER,
    ]
)


# =============================================================================
# Prohibited-operation validation
# =============================================================================

BLOCK_8_PROHIBITED_OPERATIONS = {
    "representation_forward_executed":
        NOTEBOOK_10_BLOCK_8_REPRESENTATION_FORWARD_EXECUTED,

    "optimizer_created":
        NOTEBOOK_10_BLOCK_8_OPTIMIZER_CREATED,

    "zero_grad_executed":
        NOTEBOOK_10_BLOCK_8_ZERO_GRAD_EXECUTED,

    "backward_pass_executed":
        NOTEBOOK_10_BLOCK_8_BACKWARD_PASS_EXECUTED,

    "gradient_clipping_executed":
        NOTEBOOK_10_BLOCK_8_GRADIENT_CLIPPING_EXECUTED,

    "optimizer_step_executed":
        NOTEBOOK_10_BLOCK_8_OPTIMIZER_STEP_EXECUTED,

    "parameter_update_executed":
        NOTEBOOK_10_BLOCK_8_PARAMETER_UPDATE_EXECUTED,

    "training_executed":
        NOTEBOOK_10_BLOCK_8_TRAINING_EXECUTED,

    "future_context_used":
        NOTEBOOK_10_BLOCK_8_FUTURE_CONTEXT_USED,

    "supervision_used_as_model_input":
        NOTEBOOK_10_BLOCK_8_SUPERVISION_USED_AS_MODEL_INPUT,
}


BLOCK_8_PROHIBITED_OPERATIONS_VALID = all(
    not operation_executed

    for operation_executed
    in BLOCK_8_PROHIBITED_OPERATIONS.values()
)


# =============================================================================
# Final validation
# =============================================================================

NOTEBOOK_10_BLOCK_8_ERRORS = []


BLOCK_8_VALIDATION_CHECKS = {
    "prerequisites_invalid":
        BLOCK_8_PREREQUISITES_VALID,

    "pathway_dimensions_invalid":
        BLOCK_8_PATHWAY_DIMENSIONS_VALID,

    "article_index_contract_invalid":
        BLOCK_8_ARTICLE_INDEX_CONTRACT_VALID,

    "device_dtype_invalid":
        BLOCK_8_DEVICE_DTYPE_VALID,

    "parameter_accounting_invalid":
        BLOCK_8_PARAMETER_ACCOUNTING_VALID,

    "pathway_representations_invalid":
        BLOCK_8_PATHWAY_REPRESENTATIONS_VALID,

    "target_mask_contract_invalid":
        BLOCK_8_TARGET_MASK_CONTRACT_VALID,

    "objective_elements_missing":
        BLOCK_8_ALL_PATHWAYS_OBJECTIVE_ACTIVE,

    "objective_nonfinite":
        BLOCK_8_OBJECTIVE_FINITE,

    "equal_pathway_weighting_invalid":
        BLOCK_8_EQUAL_PATHWAY_WEIGHTING_VALID,

    "baseline_not_deterministic":
        BLOCK_8_BASELINE_DETERMINISTIC,

    "causal_contract_invalid":
        BLOCK_8_CAUSAL_CONTRACT_VALID,

    "model_state_invalid":
        BLOCK_8_MODEL_STATE_VALID,

    "prohibited_operation_executed":
        BLOCK_8_PROHIBITED_OPERATIONS_VALID,
}


for error_name, condition in BLOCK_8_VALIDATION_CHECKS.items():

    if not condition:

        NOTEBOOK_10_BLOCK_8_ERRORS.append(
            error_name
        )


NOTEBOOK_10_BLOCK_8_VALID = (
    len(
        NOTEBOOK_10_BLOCK_8_ERRORS
    )
    ==
    0
)


if not NOTEBOOK_10_BLOCK_8_VALID:

    raise RuntimeError(
        "Notebook 10 Block 8 expanded recurrent objective evaluation "
        "failed. "
        f"Errors: {NOTEBOOK_10_BLOCK_8_ERRORS}"
    )


NOTEBOOK_10_BLOCK_8_COMPLETE = True


# =============================================================================
# Stable Block 8 → Block 9 handover
# =============================================================================

NOTEBOOK_10_PRETRAIN_PATHWAY_LOSSES = deepcopy(
    BLOCK_8_PRETRAIN_PATHWAY_LOSSES
)


NOTEBOOK_10_PRETRAIN_TOTAL_OBJECTIVE = float(
    BLOCK_8_PRETRAIN_TOTAL_OBJECTIVE
)


NOTEBOOK_10_PRETRAIN_OBJECTIVE_ACTIVE_ELEMENTS = deepcopy(
    BLOCK_8_OBJECTIVE_ACTIVE_ELEMENTS
)


NOTEBOOK_10_PRETRAIN_WEIGHTED_TRANSFORMATION_PREDICTIONS = {
    pathway_name:
        BLOCK_8_EVALUATION_A[
            "weighted_predictions"
        ][
            pathway_name
        ]
        .clone()

    for pathway_name
    in BLOCK_8_PATHWAY_DIMENSIONS
}


NOTEBOOK_10_PRETRAIN_RECURRENT_TRAJECTORIES = deepcopy(
    BLOCK_8_EVALUATION_A[
        "recurrent_trajectories"
    ]
)


NOTEBOOK_10_PRETRAIN_EFFECTIVE_TRANSFORMATIVE_WEIGHTS = {
    pathway_name:
        BLOCK_8_EVALUATION_A[
            "effective_weights"
        ][
            pathway_name
        ]
        .clone()

    for pathway_name
    in BLOCK_8_PATHWAY_DIMENSIONS
}


NOTEBOOK_10_PRETRAIN_BASELINE_VALID = (
    NOTEBOOK_10_BLOCK_8_VALID
)


NOTEBOOK_10_CONTROLLED_TRAINING_READY = (
    NOTEBOOK_10_PRETRAIN_BASELINE_VALID
)


# =============================================================================
# Block summary
# =============================================================================

NOTEBOOK_10_BLOCK_8_SUMMARY = {
    "block":
        NOTEBOOK_10_BLOCK_8,

    "block_name":
        NOTEBOOK_10_BLOCK_8_NAME,

    "version":
        NOTEBOOK_10_BLOCK_8_VERSION,

    "executed_at_utc":
        BLOCK_8_EXECUTED_AT_UTC,

    "article_count":
        BLOCK_8_ARTICLE_COUNT,

    "sentence_count":
        BLOCK_8_SENTENCE_COUNT,

    "objective_active_elements":
        deepcopy(
            BLOCK_8_OBJECTIVE_ACTIVE_ELEMENTS
        ),

    "pathway_losses":
        deepcopy(
            BLOCK_8_PRETRAIN_PATHWAY_LOSSES
        ),

    "total_objective":
        BLOCK_8_PRETRAIN_TOTAL_OBJECTIVE,

    "baseline_deterministic":
        BLOCK_8_BASELINE_DETERMINISTIC,

    "causal_contract_valid":
        BLOCK_8_CAUSAL_CONTRACT_VALID,

    "model_state_valid":
        BLOCK_8_MODEL_STATE_VALID,

    "training_ready":
        NOTEBOOK_10_CONTROLLED_TRAINING_READY,

    "block_valid":
        NOTEBOOK_10_BLOCK_8_VALID,

    "block_complete":
        NOTEBOOK_10_BLOCK_8_COMPLETE,
}


# =============================================================================
# Final report
# =============================================================================

print("=" * 72)

print(
    "Media AI — Notebook 10, Block 8: "
    "Expanded Recurrent Objective Evaluation "
    "and Pre-Training Baseline Construction"
)

print("=" * 72)

print(
    f"Block version                : "
    f"{NOTEBOOK_10_BLOCK_8_VERSION}"
)

print(
    f"Execution device             : "
    f"{BLOCK_8_EXECUTION_DEVICE}"
)

print(
    f"Execution dtype              : "
    f"{BLOCK_8_EXECUTION_DTYPE}"
)

print("-" * 72)

print(
    "Validated supervision objective"
)

for pathway_name in BLOCK_8_PATHWAY_DIMENSIONS:

    print(
        f"{pathway_name:<14} objective elements : "
        f"{BLOCK_8_OBJECTIVE_ACTIVE_ELEMENTS[pathway_name]}"
    )

print("-" * 72)

print(
    "Pre-training recurrent objective"
)

for pathway_name in BLOCK_8_PATHWAY_DIMENSIONS:

    print(
        f"{pathway_name:<14} masked MSE         : "
        f"{BLOCK_8_PRETRAIN_PATHWAY_LOSSES[pathway_name]:.8f}"
    )

print(
    f"Equal-pathway total objective: "
    f"{BLOCK_8_PRETRAIN_TOTAL_OBJECTIVE:.8f}"
)

print(
    f"Objective finite             : "
    f"{BLOCK_8_OBJECTIVE_FINITE}"
)

print(
    f"Equal weighting valid        : "
    f"{BLOCK_8_EQUAL_PATHWAY_WEIGHTING_VALID}"
)

print("-" * 72)

print(
    "Causal recurrent validation"
)

print(
    f"Articles evaluated           : "
    f"{BLOCK_8_ARTICLE_COUNT}"
)

print(
    f"Sentences evaluated          : "
    f"{BLOCK_8_SENTENCE_COUNT}"
)

print(
    f"Fresh zero initial states    : "
    f"{BLOCK_8_FRESH_ZERO_INITIAL_STATES}"
)

print(
    f"State chain valid            : "
    f"{BLOCK_8_STATE_CHAIN_VALID}"
)

print(
    f"Additive update valid        : "
    f"{BLOCK_8_ADDITIVE_UPDATE_VALID}"
)

print(
    f"Future context used          : "
    f"{NOTEBOOK_10_BLOCK_8_FUTURE_CONTEXT_USED}"
)

print(
    f"Supervision used as input    : "
    f"{NOTEBOOK_10_BLOCK_8_SUPERVISION_USED_AS_MODEL_INPUT}"
)

print(
    f"Causal contract valid        : "
    f"{BLOCK_8_CAUSAL_CONTRACT_VALID}"
)

print("-" * 72)

print(
    "Deterministic baseline validation"
)

print(
    f"Repeated objective evaluated : "
    f"{NOTEBOOK_10_BLOCK_8_REPEAT_OBJECTIVE_EVALUATED}"
)

print(
    f"Baseline deterministic       : "
    f"{BLOCK_8_BASELINE_DETERMINISTIC}"
)

print(
    f"Effective Weight identity    : "
    f"{BLOCK_8_EFFECTIVE_WEIGHT_IDENTITY_VALID}"
)

print("-" * 72)

print(
    "Inherited Media AI state"
)

print(
    f"Representation parameters    : "
    f"{BLOCK_8_REPRESENTATION_PARAMETER_COUNT:,}"
)

print(
    f"Transformative parameters    : "
    f"{BLOCK_8_TRANSFORMATIVE_PARAMETER_COUNT:,}"
)

print(
    f"Transformative Weight params : "
    f"{BLOCK_8_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT:,}"
)

print(
    f"Total model parameters       : "
    f"{BLOCK_8_TOTAL_PARAMETER_COUNT:,}"
)

print(
    f"Representation unchanged     : "
    f"{BLOCK_8_REPRESENTATION_UNCHANGED}"
)

print(
    f"Transformative unchanged     : "
    f"{BLOCK_8_TRANSFORMATIVE_UNCHANGED}"
)

print(
    f"Transform Weight unchanged   : "
    f"{BLOCK_8_WEIGHT_UNCHANGED}"
)

print(
    f"All parameters frozen        : "
    f"{BLOCK_8_ALL_PARAMETERS_FROZEN_AFTER}"
)

print(
    f"All modules eval             : "
    f"{BLOCK_8_ALL_MODULES_EVAL_AFTER}"
)

print(
    f"All gradients absent         : "
    f"{BLOCK_8_ALL_GRADIENTS_ABSENT_AFTER}"
)

print("-" * 72)

print(
    "Execution boundary"
)

print(
    f"Representation forward       : "
    f"{NOTEBOOK_10_BLOCK_8_REPRESENTATION_FORWARD_EXECUTED}"
)

print(
    f"Transform forward            : "
    f"{NOTEBOOK_10_BLOCK_8_TRANSFORMATIVE_FORWARD_EXECUTED}"
)

print(
    f"Weight forward               : "
    f"{NOTEBOOK_10_BLOCK_8_WEIGHT_FORWARD_EXECUTED}"
)

print(
    f"Recurrent state created      : "
    f"{NOTEBOOK_10_BLOCK_8_RECURRENT_STATE_CREATED}"
)

print(
    f"Recurrent update executed    : "
    f"{NOTEBOOK_10_BLOCK_8_RECURRENT_UPDATE_EXECUTED}"
)

print(
    f"Loss calculated              : "
    f"{NOTEBOOK_10_BLOCK_8_LOSS_CALCULATED}"
)

print(
    f"Optimizer created            : "
    f"{NOTEBOOK_10_BLOCK_8_OPTIMIZER_CREATED}"
)

print(
    f"Backward pass executed       : "
    f"{NOTEBOOK_10_BLOCK_8_BACKWARD_PASS_EXECUTED}"
)

print(
    f"Parameter update executed    : "
    f"{NOTEBOOK_10_BLOCK_8_PARAMETER_UPDATE_EXECUTED}"
)

print(
    f"Training executed            : "
    f"{NOTEBOOK_10_BLOCK_8_TRAINING_EXECUTED}"
)

print("-" * 72)

print(
    f"Pre-training baseline valid  : "
    f"{NOTEBOOK_10_PRETRAIN_BASELINE_VALID}"
)

print(
    f"Controlled training ready    : "
    f"{NOTEBOOK_10_CONTROLLED_TRAINING_READY}"
)

print(
    f"Block valid                  : "
    f"{NOTEBOOK_10_BLOCK_8_VALID}"
)

print(
    f"Block complete               : "
    f"{NOTEBOOK_10_BLOCK_8_COMPLETE}"
)

print("=" * 72)

print(
    f"The complete recurrent objective was evaluated across "
    f"{BLOCK_8_ARTICLE_COUNT} article(s) and "
    f"{BLOCK_8_SENTENCE_COUNT} canonical sentence(s)."
)

print(
    "The factual, psychological and social objectives were calculated "
    "using only Block 7 mask-valid, activation-eligible signed "
    "transformation targets."
)

print(
    "Each article began from an independent deterministic zero recurrent "
    "state and no recurrent state propagated across article boundaries."
)

print(
    "Supervision was used only as an objective target and was never supplied "
    "to the Media AI recurrent forward path as model input."
)

print(
    "The complete objective was evaluated twice and reproduced exactly, "
    "establishing a deterministic immutable pre-training baseline."
)

print(
    f"The inherited {BLOCK_8_TOTAL_PARAMETER_COUNT:,}-parameter Media AI "
    "architecture remained frozen and exactly unchanged."
)

print(
    "No optimiser, gradient clearing, backward pass, gradient clipping, "
    "optimiser step, parameter update or training operation was executed."
)

print(
    "Notebook 10 may now proceed to Block 9 — Controlled Expanded Pilot "
    "Training and Parameter-Update Validation."
)

print("=" * 72)
```

# Block 9: Controlled Continued Pilot Training and Parameter-Update Validation

## Purpose

This block performs the first controlled optimisation stage of Notebook 10.

It continues training of the **Transformative-Weight parameters only**, using the validated recurrent objective established in Block 8, while preserving the representation and transformative mechanisms inherited from Notebook 09 as frozen architectural components.

The purpose of this block is not to claim final model training or generalisable performance.

The currently validated Notebook 10 corpus contains the same **3 articles and 54 canonical sentences** represented in the preceding pilot architecture. Accordingly, this stage is treated explicitly as **continued pilot optimisation and parameter-update validation** rather than genuine expanded-corpus training.

Its methodological objective is to determine whether the inherited Transformative-Weight mechanism remains differentiably trainable under the fully reconstructed Notebook 10 recurrent execution contract and whether additional controlled optimisation can reduce the immutable Block 8 baseline objective without modifying any frozen inherited parameter.

---

## Methodological Role

Block 8 established the deterministic pre-training reference objective

$$
\mathcal{L}_{\mathrm{baseline}}
=
\frac{1}{3}
\left(
\mathcal{L}_{F}
+
\mathcal{L}_{P}
+
\mathcal{L}_{S}
\right),
$$

where:

- $\mathcal{L}_{F}$ is the mask-aware factual transformation loss;
- $\mathcal{L}_{P}$ is the mask-aware psychological transformation loss;
- $\mathcal{L}_{S}$ is the mask-aware social transformation loss.

For the current validated corpus, Block 8 established

$$
\mathcal{L}_{\mathrm{baseline}} = 0.20858338.
$$

This value is treated as an **immutable pre-training baseline** for Block 9.

The optimisation performed here must therefore demonstrate parameter learning relative to that fixed reference rather than reconstructing or redefining the baseline after training.

---

## Trainable Parameter Scope

The complete inherited Media AI architecture contains:

- **894,003 representation parameters**;
- **8,187 transformative parameters**;
- **49 Transformative-Weight parameters**.

The representation and transformative parameters remain frozen.

Only the 49 activation-eligible latent Transformative-Weight parameters are supplied to the optimiser.

Let the trainable latent Transformative-Weight parameters for pathway $p$ be

$$
\boldsymbol{\alpha}^{(p)}
\in
\mathbb{R}^{d_p^{\mathrm{active}}}.
$$

Their effective bounded weights remain defined by

$$
\mathbf{w}^{(p)}
=
\sigma\left(\boldsymbol{\alpha}^{(p)}\right),
$$

where $\sigma(\cdot)$ denotes the sigmoid function.

The structural pathway dimensions remain:

- factual: $d_F = 10$;
- psychological: $d_P = 34$;
- social: $d_S = 7$.

The social pathway retains two deferred structural dimensions, which remain fixed at zero and are excluded from optimisation.

No new trainable architecture is introduced in this block.

---

## Recurrent Training Contract

For sentence position $t$ in pathway $p$, the frozen transformative mechanism receives the current representation and immediately preceding recurrent state:

$$
\mathbf{c}^{(p)}_t
=
T^{(p)}
\left(
\mathbf{r}^{(p)}_t,
\mathbf{h}^{(p)}_{t-1}
\right).
$$

The learned Transformative Weight is applied dimension-wise:

$$
\mathbf{g}^{(p)}_t
=
\mathbf{w}^{(p)}
\odot
\mathbf{c}^{(p)}_t.
$$

The recurrent state follows the established additive update:

$$
\mathbf{h}^{(p)}_t
=
\mathbf{h}^{(p)}_{t-1}
+
\mathbf{g}^{(p)}_t.
$$

Every article begins with

$$
\mathbf{h}^{(p)}_0
=
\mathbf{0}.
$$

Recurrent state is propagated only between consecutive sentences belonging to the same article.

It must never cross an article boundary.

The recurrent state must remain differentiably connected across sentence positions during each optimisation forward pass so that gradients reaching the Transformative-Weight parameters correctly reflect their contribution to the recurrent trajectory.

---

## Supervision and Objective

Block 7 established signed transformation targets from validated current-state supervision.

For pathway $p$ and sentence position $t$,

$$
\Delta\mathbf{y}^{(p)}_t
=
\mathbf{y}^{(p)}_t
-
\mathbf{y}^{(p)}_{t-1},
$$

subject to the validated supervision-availability masks and independent zero initial condition at every article boundary.

The predicted transformation is the gated candidate transformation

$$
\widehat{\Delta\mathbf{y}}^{(p)}_t
=
\mathbf{g}^{(p)}_t.
$$

For each pathway, optimisation uses the established mask-aware mean-squared-error objective:

$$
\mathcal{L}_p
=
\frac{
\sum_{t,j}
m^{(p)}_{t,j}
\left(
\widehat{\Delta y}^{(p)}_{t,j}
-
\Delta y^{(p)}_{t,j}
\right)^2
}{
\sum_{t,j}m^{(p)}_{t,j}
}.
$$

Only supervision-valid and activation-eligible dimensions contribute to the objective.

Missing supervision remains distinct from a genuine zero-valued transformation.

The total objective remains equally weighted across the three representational pathways:

$$
\mathcal{L}_{\mathrm{total}}
=
\frac{
\mathcal{L}_{F}
+
\mathcal{L}_{P}
+
\mathcal{L}_{S}
}{3}.
$$

No pathway-specific weighting is introduced.

---

## Controlled Optimisation Policy

Training must use a deterministic and explicitly bounded optimisation policy.

The optimiser is restricted to the 49 Transformative-Weight parameters.

The inherited optimisation family remains:

- **Optimizer:** AdamW
- **Learning rate:** $10^{-3}$
- **Weight decay:** $10^{-4}$
- **Maximum gradient norm:** $1.0$

The controlled epoch count must be fixed explicitly by the implementation and reported in the execution output.

Canonical article and sentence ordering must be preserved throughout optimisation.

Batch shuffling is not permitted.

At every epoch:

1. gradients are cleared;
2. the complete causal recurrent objective is reconstructed;
3. the total mask-aware objective is calculated;
4. backpropagation is executed;
5. gradient finiteness is verified;
6. gradient clipping is applied under the established maximum-norm policy;
7. the optimiser updates only the Transformative-Weight parameters.

No representation or transformative parameter may receive an optimiser update.

---

## Parameter-Update Validation

The block must validate both **learning** and **architectural isolation**.

For each pathway, it must compare the final latent Transformative-Weight state with its pre-training state and establish whether eligible parameters changed.

The effective post-training weights must remain:

- finite;
- bounded by the sigmoid parameterisation;
- aligned with their inherited semantic dimensions;
- exactly zero for every deferred dimension.

The block must additionally verify that:

- representation parameters are bitwise unchanged;
- transformative parameters are bitwise unchanged;
- only Transformative-Weight parameters changed;
- all optimiser parameters belong exclusively to the Transformative-Weight modules.

A reduction in the objective is evidence of successful controlled optimisation, but it is **not** interpreted as evidence of generalisation.

---

## Post-Training Objective Comparison

After the final optimiser step, the complete recurrent objective must be evaluated again under the learned Transformative-Weight state.

The principal comparison is

$$
\Delta\mathcal{L}
=
\mathcal{L}_{\mathrm{post}}
-
\mathcal{L}_{\mathrm{baseline}}.
$$

An improvement corresponds to

$$
\Delta\mathcal{L} < 0.
$$

The final factual, psychological and social pathway objectives must also be reported separately so that improvement or deterioration cannot be concealed by the aggregated objective.

The Block 8 baseline must remain unchanged throughout this comparison.

---

## Causal and Leakage Constraints

The optimisation stage must preserve all previously validated causal restrictions.

Specifically:

- sentence order is preserved;
- article order is deterministic;
- every article begins from a fresh zero recurrent state;
- recurrent state never propagates between articles;
- future sentences are never used when constructing the current recurrent transition;
- supervision targets are never supplied as model inputs;
- missing supervision is masked rather than replaced by zero;
- deferred dimensions remain excluded from the learning objective.

These constraints preserve the distinction between **model state**, **supervision**, and **future information**.

---

## Execution Boundary

This block **does execute training**.

It may therefore:

- execute the frozen transformative forward path;
- execute the Transformative-Weight forward path;
- construct recurrent trajectories;
- calculate the training objective;
- create the optimiser;
- clear gradients;
- execute backward propagation;
- clip gradients;
- update Transformative-Weight parameters.

It must **not**:

- retrain the representation architecture;
- retrain the transformative architecture;
- modify pathway dimensionality;
- alter the recurrent update rule;
- modify active/deferred dimension partitions;
- introduce cross-article recurrence;
- introduce future context;
- substitute supervision for model input;
- claim final or expanded-corpus model performance.

---

## Scientific Interpretation

Successful completion of this block demonstrates that the complete reconstructed Media AI recurrent architecture remains trainable through its intended **Transformative-Weight interface** while preserving the frozen representation and transformative mechanisms.

Because the current validation corpus remains limited to the original three pilot articles, any reduction in objective must be interpreted as **continued optimisation on the pilot corpus**.

It does not establish:

- out-of-sample performance;
- generalisation to unseen articles;
- robustness across publishers or topics;
- final Transformative-Weight estimates;
- production readiness.

These distinctions are essential because architectural learnability and empirical generalisation are separate research claims.

---

## Expected Handover

On successful completion, Block 9 should provide:

- the immutable Block 8 pre-training objective;
- final pathway-specific objectives;
- final total objective;
- objective change relative to Block 8;
- gradient-finiteness and gradient-norm diagnostics;

If all contracts are satisfied, Notebook 10 may proceed to **Block 10 — Post-Training Recurrent Validation, Objective Comparison and Generalisation-Boundary Audit**.

```python
# =============================================================================
# Media AI — Notebook 10
# Block 9: Controlled Continued Pilot Training
#          and Parameter-Update Validation
# =============================================================================

from copy import deepcopy
from datetime import datetime, timezone

import math

import numpy as np
import torch


# =============================================================================
# Block identity
# =============================================================================

NOTEBOOK_10_BLOCK_9 = 9

NOTEBOOK_10_BLOCK_9_NAME = (
    "Controlled Continued Pilot Training "
    "and Parameter-Update Validation"
)

NOTEBOOK_10_BLOCK_9_VERSION = "1.0"

BLOCK_9_EXECUTED_AT_UTC = (
    datetime.now(
        timezone.utc
    ).isoformat()
)


# =============================================================================
# Controlled optimisation policy
# =============================================================================
#
# The policy reproduces the controlled Notebook 09 Transformative-Weight
# optimisation family. This remains continued pilot optimisation on the same
# 3-article / 54-sentence corpus; it is not expanded-corpus training.
# =============================================================================

BLOCK_9_OPTIMIZER_NAME = (
    "AdamW"
)

BLOCK_9_LEARNING_RATE = 1e-3

BLOCK_9_WEIGHT_DECAY = 1e-4

BLOCK_9_MAXIMUM_GRADIENT_NORM = 1.0

BLOCK_9_EPOCHS = 25

BLOCK_9_BASELINE_ATOL = 1e-7

BLOCK_9_BASELINE_RTOL = 0.0


# =============================================================================
# Required Block 8 → 9 handover contract
# =============================================================================

BLOCK_9_REQUIRED_GLOBALS = [
    "NOTEBOOK_10_BLOCK_8_COMPLETE",
    "NOTEBOOK_10_BLOCK_8_VALID",
    "NOTEBOOK_10_PRETRAIN_BASELINE_VALID",
    "NOTEBOOK_10_CONTROLLED_TRAINING_READY",
    "NOTEBOOK_10_PRETRAIN_PATHWAY_LOSSES",
    "NOTEBOOK_10_PRETRAIN_TOTAL_OBJECTIVE",
    "NOTEBOOK_10_PRETRAIN_OBJECTIVE_ACTIVE_ELEMENTS",
    "NOTEBOOK_10_PRETRAIN_EFFECTIVE_TRANSFORMATIVE_WEIGHTS",
    "NOTEBOOK_10_MULTI_ARTICLE_PATHWAY_REPRESENTATIONS",
    "NOTEBOOK_10_TRANSFORMATION_TARGETS",
    "NOTEBOOK_10_TRANSFORMATION_OBJECTIVE_MASKS",
    "NOTEBOOK_10_SUPERVISION_CANONICAL_ARTICLE_IDS",
    "NOTEBOOK_10_SUPERVISION_CANONICAL_SENTENCE_IDS",
    "NOTEBOOK_10_SUPERVISION_ARTICLE_RECORD_INDICES",
    "NOTEBOOK_10_REPRESENTATION_MODULES",
    "NOTEBOOK_10_TRANSFORMATIVE_MODULES",
    "NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES",
    "NOTEBOOK_10_PATHWAY_DIMENSIONS",
    "NOTEBOOK_10_ACTIVE_DIMENSION_INDICES",
    "NOTEBOOK_10_DEFERRED_DIMENSION_INDICES",
]


BLOCK_9_MISSING_PREREQUISITES = [
    object_name

    for object_name
    in BLOCK_9_REQUIRED_GLOBALS

    if object_name not in globals()
]


if BLOCK_9_MISSING_PREREQUISITES:

    raise RuntimeError(
        "Notebook 10 Block 9 cannot execute because the current runtime "
        "does not contain the exact validated Block 8 → Block 9 handover. "
        "If the runtime was restarted, rerun Notebook 10 Blocks 1–8 in "
        "order in this same runtime. "
        f"Missing objects: {BLOCK_9_MISSING_PREREQUISITES}"
    )


BLOCK_9_PREREQUISITES_VALID = all(
    [
        NOTEBOOK_10_BLOCK_8_COMPLETE is True,
        NOTEBOOK_10_BLOCK_8_VALID is True,
        NOTEBOOK_10_PRETRAIN_BASELINE_VALID is True,
        NOTEBOOK_10_CONTROLLED_TRAINING_READY is True,
        isinstance(
            NOTEBOOK_10_PRETRAIN_PATHWAY_LOSSES,
            dict,
        ),
        isinstance(
            NOTEBOOK_10_PRETRAIN_OBJECTIVE_ACTIVE_ELEMENTS,
            dict,
        ),
        isinstance(
            NOTEBOOK_10_MULTI_ARTICLE_PATHWAY_REPRESENTATIONS,
            dict,
        ),
        isinstance(
            NOTEBOOK_10_TRANSFORMATION_TARGETS,
            dict,
        ),
        isinstance(
            NOTEBOOK_10_TRANSFORMATION_OBJECTIVE_MASKS,
            dict,
        ),
    ]
)


if not BLOCK_9_PREREQUISITES_VALID:

    raise RuntimeError(
        "Notebook 10 Block 9 requires a valid immutable Block 8 "
        "pre-training baseline."
    )


# =============================================================================
# Canonical pathway contract
# =============================================================================

BLOCK_9_PATHWAY_DIMENSIONS = {
    pathway_name:
        int(
            pathway_dim
        )

    for pathway_name, pathway_dim
    in NOTEBOOK_10_PATHWAY_DIMENSIONS.items()
}


BLOCK_9_EXPECTED_PATHWAY_DIMENSIONS = {
    "factual":
        10,

    "psychological":
        34,

    "social":
        7,
}


BLOCK_9_PATHWAY_DIMENSIONS_VALID = (
    BLOCK_9_PATHWAY_DIMENSIONS
    ==
    BLOCK_9_EXPECTED_PATHWAY_DIMENSIONS
)


if not BLOCK_9_PATHWAY_DIMENSIONS_VALID:

    raise RuntimeError(
        "Notebook 10 Block 9 pathway dimensions do not reproduce the "
        "validated Notebook 09 → 10 architecture."
    )


BLOCK_9_ACTIVE_DIMENSION_INDICES = {
    pathway_name:
        tuple(
            int(
                index
            )

            for index
            in NOTEBOOK_10_ACTIVE_DIMENSION_INDICES[
                pathway_name
            ]
        )

    for pathway_name
    in BLOCK_9_PATHWAY_DIMENSIONS
}


BLOCK_9_DEFERRED_DIMENSION_INDICES = {
    pathway_name:
        tuple(
            int(
                index
            )

            for index
            in NOTEBOOK_10_DEFERRED_DIMENSION_INDICES[
                pathway_name
            ]
        )

    for pathway_name
    in BLOCK_9_PATHWAY_DIMENSIONS
}


BLOCK_9_DIMENSION_PARTITIONS_VALID = all(
    (
        set(
            BLOCK_9_ACTIVE_DIMENSION_INDICES[
                pathway_name
            ]
        ).isdisjoint(
            BLOCK_9_DEFERRED_DIMENSION_INDICES[
                pathway_name
            ]
        )
        and
        (
            set(
                BLOCK_9_ACTIVE_DIMENSION_INDICES[
                    pathway_name
                ]
            )
            |
            set(
                BLOCK_9_DEFERRED_DIMENSION_INDICES[
                    pathway_name
                ]
            )
            ==
            set(
                range(
                    pathway_dim
                )
            )
        )
    )

    for pathway_name, pathway_dim
    in BLOCK_9_PATHWAY_DIMENSIONS.items()
)


if not BLOCK_9_DIMENSION_PARTITIONS_VALID:

    raise RuntimeError(
        "Notebook 10 Block 9 active/deferred dimensional contract "
        "is invalid."
    )


# =============================================================================
# Canonical article/sentence contract
# =============================================================================

BLOCK_9_ARTICLE_IDS = tuple(
    NOTEBOOK_10_SUPERVISION_CANONICAL_ARTICLE_IDS
)


BLOCK_9_SENTENCE_IDS = tuple(
    NOTEBOOK_10_SUPERVISION_CANONICAL_SENTENCE_IDS
)


BLOCK_9_ARTICLE_RECORD_INDICES = deepcopy(
    NOTEBOOK_10_SUPERVISION_ARTICLE_RECORD_INDICES
)


BLOCK_9_ARTICLE_COUNT = len(
    BLOCK_9_ARTICLE_IDS
)


BLOCK_9_SENTENCE_COUNT = len(
    BLOCK_9_SENTENCE_IDS
)


BLOCK_9_CANONICAL_GLOBAL_INDICES = tuple(
    global_index

    for article_id
    in BLOCK_9_ARTICLE_IDS

    for global_index
    in BLOCK_9_ARTICLE_RECORD_INDICES[
        article_id
    ]
)


BLOCK_9_CANONICAL_INDEX_CONTRACT_VALID = all(
    [
        BLOCK_9_ARTICLE_COUNT > 0,
        BLOCK_9_SENTENCE_COUNT > 0,
        len(
            set(
                BLOCK_9_SENTENCE_IDS
            )
        )
        ==
        BLOCK_9_SENTENCE_COUNT,
        set(
            BLOCK_9_ARTICLE_RECORD_INDICES.keys()
        )
        ==
        set(
            BLOCK_9_ARTICLE_IDS
        ),
        BLOCK_9_CANONICAL_GLOBAL_INDICES
        ==
        tuple(
            range(
                BLOCK_9_SENTENCE_COUNT
            )
        ),
    ]
)


if not BLOCK_9_CANONICAL_INDEX_CONTRACT_VALID:

    raise RuntimeError(
        "Notebook 10 Block 9 canonical article/sentence order is invalid."
    )


# =============================================================================
# Resolve execution device and dtype
# =============================================================================

BLOCK_9_REFERENCE_PARAMETERS = [
    parameter

    for module
    in NOTEBOOK_10_REPRESENTATION_MODULES.values()

    for parameter
    in module.parameters()
]


if not BLOCK_9_REFERENCE_PARAMETERS:

    raise RuntimeError(
        "Notebook 10 Block 9 could not resolve model device/dtype."
    )


BLOCK_9_EXECUTION_DEVICE = (
    BLOCK_9_REFERENCE_PARAMETERS[
        0
    ].device
)


BLOCK_9_EXECUTION_DTYPE = (
    BLOCK_9_REFERENCE_PARAMETERS[
        0
    ].dtype
)


BLOCK_9_DEVICE_DTYPE_VALID = all(
    parameter.device
    ==
    BLOCK_9_EXECUTION_DEVICE
    and
    parameter.dtype
    ==
    BLOCK_9_EXECUTION_DTYPE

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


if not BLOCK_9_DEVICE_DTYPE_VALID:

    raise RuntimeError(
        "Notebook 10 Block 9 inherited modules do not share one "
        "execution device/dtype."
    )


# =============================================================================
# Execution-state flags
# =============================================================================

NOTEBOOK_10_BLOCK_9_REPRESENTATION_FORWARD_EXECUTED = False

NOTEBOOK_10_BLOCK_9_TRANSFORMATIVE_FORWARD_EXECUTED = False

NOTEBOOK_10_BLOCK_9_WEIGHT_FORWARD_EXECUTED = False

NOTEBOOK_10_BLOCK_9_RECURRENT_STATE_CREATED = False

NOTEBOOK_10_BLOCK_9_RECURRENT_UPDATE_EXECUTED = False

NOTEBOOK_10_BLOCK_9_LOSS_CALCULATED = False

NOTEBOOK_10_BLOCK_9_OPTIMIZER_CREATED = False

NOTEBOOK_10_BLOCK_9_ZERO_GRAD_EXECUTED = False

NOTEBOOK_10_BLOCK_9_BACKWARD_PASS_EXECUTED = False

NOTEBOOK_10_BLOCK_9_GRADIENT_CLIPPING_EXECUTED = False

NOTEBOOK_10_BLOCK_9_OPTIMIZER_STEP_EXECUTED = False

NOTEBOOK_10_BLOCK_9_PARAMETER_UPDATE_EXECUTED = False

NOTEBOOK_10_BLOCK_9_TRAINING_EXECUTED = False

NOTEBOOK_10_BLOCK_9_FUTURE_CONTEXT_USED = False

NOTEBOOK_10_BLOCK_9_SUPERVISION_USED_AS_MODEL_INPUT = False

NOTEBOOK_10_BLOCK_9_BATCH_SHUFFLING_USED = False


# =============================================================================
# Exact state helpers
# =============================================================================

def block_9_snapshot_module_collection(
    module_collection,
):

    return {
        module_name:
            {
                tensor_name:
                    tensor.detach()
                    .cpu()
                    .clone()

                for tensor_name, tensor
                in module.state_dict().items()
            }

        for module_name, module
        in module_collection.items()
    }


def block_9_nested_state_exact(
    state_before,
    state_after,
):

    if state_before.keys() != state_after.keys():

        return False

    for module_name in state_before:

        if (
            state_before[
                module_name
            ].keys()
            !=
            state_after[
                module_name
            ].keys()
        ):

            return False

        for tensor_name in state_before[
            module_name
        ]:

            if not torch.equal(
                state_before[
                    module_name
                ][
                    tensor_name
                ],
                state_after[
                    module_name
                ][
                    tensor_name
                ],
            ):

                return False

    return True


# =============================================================================
# Immutable pre-training model state
# =============================================================================

BLOCK_9_REPRESENTATION_STATE_BEFORE = (
    block_9_snapshot_module_collection(
        NOTEBOOK_10_REPRESENTATION_MODULES
    )
)


BLOCK_9_TRANSFORMATIVE_STATE_BEFORE = (
    block_9_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_MODULES
    )
)


BLOCK_9_WEIGHT_STATE_BEFORE = (
    block_9_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES
    )
)


# =============================================================================
# Parameter accounting
# =============================================================================

BLOCK_9_REPRESENTATION_PARAMETER_COUNT = sum(
    parameter.numel()

    for module
    in NOTEBOOK_10_REPRESENTATION_MODULES.values()

    for parameter
    in module.parameters()
)


BLOCK_9_TRANSFORMATIVE_PARAMETER_COUNT = sum(
    parameter.numel()

    for module
    in NOTEBOOK_10_TRANSFORMATIVE_MODULES.values()

    for parameter
    in module.parameters()
)


BLOCK_9_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT = sum(
    parameter.numel()

    for module
    in NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES.values()

    for parameter
    in module.parameters()
)


BLOCK_9_TOTAL_PARAMETER_COUNT = (
    BLOCK_9_REPRESENTATION_PARAMETER_COUNT
    +
    BLOCK_9_TRANSFORMATIVE_PARAMETER_COUNT
    +
    BLOCK_9_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT
)


BLOCK_9_PARAMETER_ACCOUNTING_VALID = all(
    [
        BLOCK_9_REPRESENTATION_PARAMETER_COUNT
        ==
        894_003,

        BLOCK_9_TRANSFORMATIVE_PARAMETER_COUNT
        ==
        8_187,

        BLOCK_9_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT
        ==
        49,

        BLOCK_9_TOTAL_PARAMETER_COUNT
        ==
        902_239,
    ]
)


if not BLOCK_9_PARAMETER_ACCOUNTING_VALID:

    raise RuntimeError(
        "Notebook 10 Block 9 parameter accounting changed unexpectedly."
    )


# =============================================================================
# Validate and prepare fixed representations, targets and masks
# =============================================================================

BLOCK_9_PATHWAY_REPRESENTATIONS = deepcopy(
    NOTEBOOK_10_MULTI_ARTICLE_PATHWAY_REPRESENTATIONS
)


BLOCK_9_TARGETS = {
    pathway_name:
        NOTEBOOK_10_TRANSFORMATION_TARGETS[
            pathway_name
        ]
        .detach()
        .to(
            device=
                BLOCK_9_EXECUTION_DEVICE,
            dtype=
                BLOCK_9_EXECUTION_DTYPE,
        )
        .clone()

    for pathway_name
    in BLOCK_9_PATHWAY_DIMENSIONS
}


BLOCK_9_OBJECTIVE_MASKS = {
    pathway_name:
        NOTEBOOK_10_TRANSFORMATION_OBJECTIVE_MASKS[
            pathway_name
        ]
        .detach()
        .to(
            device=
                BLOCK_9_EXECUTION_DEVICE,
            dtype=
                torch.bool,
        )
        .clone()

    for pathway_name
    in BLOCK_9_PATHWAY_DIMENSIONS
}


BLOCK_9_FIXED_INPUT_CONTRACT_VALID = True


for article_id in BLOCK_9_ARTICLE_IDS:

    if article_id not in BLOCK_9_PATHWAY_REPRESENTATIONS:

        BLOCK_9_FIXED_INPUT_CONTRACT_VALID = False

        continue

    article_length = len(
        BLOCK_9_ARTICLE_RECORD_INDICES[
            article_id
        ]
    )

    for pathway_name, pathway_dim in BLOCK_9_PATHWAY_DIMENSIONS.items():

        matrix = (
            BLOCK_9_PATHWAY_REPRESENTATIONS[
                article_id
            ].get(
                pathway_name
            )
        )

        if not torch.is_tensor(
            matrix
        ):

            BLOCK_9_FIXED_INPUT_CONTRACT_VALID = False

            continue

        if (
            tuple(
                matrix.shape
            )
            !=
            (
                article_length,
                pathway_dim,
            )
        ):

            BLOCK_9_FIXED_INPUT_CONTRACT_VALID = False

        if not bool(
            torch.isfinite(
                matrix
            ).all()
            .item()
        ):

            BLOCK_9_FIXED_INPUT_CONTRACT_VALID = False


for pathway_name, pathway_dim in BLOCK_9_PATHWAY_DIMENSIONS.items():

    if (
        tuple(
            BLOCK_9_TARGETS[
                pathway_name
            ].shape
        )
        !=
        (
            BLOCK_9_SENTENCE_COUNT,
            pathway_dim,
        )
    ):

        BLOCK_9_FIXED_INPUT_CONTRACT_VALID = False

    if (
        tuple(
            BLOCK_9_OBJECTIVE_MASKS[
                pathway_name
            ].shape
        )
        !=
        (
            BLOCK_9_SENTENCE_COUNT,
            pathway_dim,
        )
    ):

        BLOCK_9_FIXED_INPUT_CONTRACT_VALID = False

    if not bool(
        BLOCK_9_OBJECTIVE_MASKS[
            pathway_name
        ].any()
        .item()
    ):

        BLOCK_9_FIXED_INPUT_CONTRACT_VALID = False

    if not bool(
        torch.isfinite(
            BLOCK_9_TARGETS[
                pathway_name
            ][
                BLOCK_9_OBJECTIVE_MASKS[
                    pathway_name
                ]
            ]
        ).all()
        .item()
    ):

        BLOCK_9_FIXED_INPUT_CONTRACT_VALID = False


if not BLOCK_9_FIXED_INPUT_CONTRACT_VALID:

    raise RuntimeError(
        "Notebook 10 Block 9 fixed representation/target/mask contract "
        "is invalid."
    )


# =============================================================================
# Immutable Block 8 baseline
# =============================================================================

BLOCK_9_IMMUTABLE_BASELINE_PATHWAY_LOSSES = {
    pathway_name:
        float(
            NOTEBOOK_10_PRETRAIN_PATHWAY_LOSSES[
                pathway_name
            ]
        )

    for pathway_name
    in BLOCK_9_PATHWAY_DIMENSIONS
}


BLOCK_9_IMMUTABLE_BASELINE_TOTAL_OBJECTIVE = float(
    NOTEBOOK_10_PRETRAIN_TOTAL_OBJECTIVE
)


BLOCK_9_OBJECTIVE_ACTIVE_ELEMENTS = {
    pathway_name:
        int(
            BLOCK_9_OBJECTIVE_MASKS[
                pathway_name
            ].sum()
            .item()
        )

    for pathway_name
    in BLOCK_9_PATHWAY_DIMENSIONS
}


BLOCK_9_OBJECTIVE_COUNTS_MATCH_BLOCK_8 = all(
    BLOCK_9_OBJECTIVE_ACTIVE_ELEMENTS[
        pathway_name
    ]
    ==
    int(
        NOTEBOOK_10_PRETRAIN_OBJECTIVE_ACTIVE_ELEMENTS[
            pathway_name
        ]
    )

    for pathway_name
    in BLOCK_9_PATHWAY_DIMENSIONS
)


if not BLOCK_9_OBJECTIVE_COUNTS_MATCH_BLOCK_8:

    raise RuntimeError(
        "Notebook 10 Block 9 objective-active counts do not reproduce "
        "the immutable Block 8 baseline contract."
    )


# =============================================================================
# Trainable parameter scope
# =============================================================================
#
# Representation and transformative parameters remain frozen.
# Only the 49 Transformative-Weight latent parameters become trainable.
# =============================================================================

for module in NOTEBOOK_10_REPRESENTATION_MODULES.values():

    module.eval()

    for parameter in module.parameters():

        parameter.requires_grad_(
            False
        )


for module in NOTEBOOK_10_TRANSFORMATIVE_MODULES.values():

    module.eval()

    for parameter in module.parameters():

        parameter.requires_grad_(
            False
        )


for module in NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES.values():

    module.eval()

    for parameter in module.parameters():

        parameter.requires_grad_(
            True
        )


BLOCK_9_OPTIMIZER_PARAMETERS = [
    parameter

    for module
    in NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES.values()

    for parameter
    in module.parameters()

    if parameter.requires_grad
]


BLOCK_9_OPTIMIZER_PARAMETER_COUNT = sum(
    parameter.numel()

    for parameter
    in BLOCK_9_OPTIMIZER_PARAMETERS
)


BLOCK_9_REPRESENTATION_FROZEN_BEFORE = all(
    not parameter.requires_grad

    for module
    in NOTEBOOK_10_REPRESENTATION_MODULES.values()

    for parameter
    in module.parameters()
)


BLOCK_9_TRANSFORMATIVE_FROZEN_BEFORE = all(
    not parameter.requires_grad

    for module
    in NOTEBOOK_10_TRANSFORMATIVE_MODULES.values()

    for parameter
    in module.parameters()
)


BLOCK_9_WEIGHT_PARAMETERS_TRAINABLE_BEFORE = all(
    parameter.requires_grad

    for parameter
    in BLOCK_9_OPTIMIZER_PARAMETERS
)


BLOCK_9_PARAMETER_SCOPE_VALID = all(
    [
        BLOCK_9_REPRESENTATION_FROZEN_BEFORE,
        BLOCK_9_TRANSFORMATIVE_FROZEN_BEFORE,
        BLOCK_9_WEIGHT_PARAMETERS_TRAINABLE_BEFORE,
        BLOCK_9_OPTIMIZER_PARAMETER_COUNT
        ==
        49,
    ]
)


if not BLOCK_9_PARAMETER_SCOPE_VALID:

    raise RuntimeError(
        "Notebook 10 Block 9 trainable parameter scope is invalid."
    )


# =============================================================================
# Differentiable recurrent objective
# =============================================================================
#
# The recurrent state is not detached inside an article. This preserves the
# differentiable effect of Transformative Weights through the causal recurrent
# trajectory. A fresh zero state is constructed at every article boundary.
# =============================================================================

def block_9_forward_objective():

    pathway_outputs = {}

    for pathway_name, pathway_dim in BLOCK_9_PATHWAY_DIMENSIONS.items():

        global_indices = []

        candidate_rows = []

        effective_weight_rows = []

        gated_prediction_rows = []

        recurrent_state_rows = []

        for article_id in BLOCK_9_ARTICLE_IDS:

            article_indices = (
                BLOCK_9_ARTICLE_RECORD_INDICES[
                    article_id
                ]
            )

            representation_matrix = (
                BLOCK_9_PATHWAY_REPRESENTATIONS[
                    article_id
                ][
                    pathway_name
                ]
                .to(
                    device=
                        BLOCK_9_EXECUTION_DEVICE,
                    dtype=
                        BLOCK_9_EXECUTION_DTYPE,
                )
            )

            previous_state = torch.zeros(
                (
                    1,
                    pathway_dim,
                ),
                dtype=
                    BLOCK_9_EXECUTION_DTYPE,
                device=
                    BLOCK_9_EXECUTION_DEVICE,
            )

            for local_position, global_index in enumerate(
                article_indices
            ):

                current_representation = (
                    representation_matrix[
                        local_position:
                        local_position
                        +
                        1
                    ]
                )

                candidate = (
                    NOTEBOOK_10_TRANSFORMATIVE_MODULES[
                        pathway_name
                    ](
                        current_representation,
                        previous_state,
                    )
                )

                effective_weight = (
                    NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES[
                        pathway_name
                    ]()
                )

                gated_prediction = (
                    effective_weight.unsqueeze(
                        0
                    )
                    *
                    candidate
                )

                updated_state = (
                    previous_state
                    +
                    gated_prediction
                )

                global_indices.append(
                    int(
                        global_index
                    )
                )

                candidate_rows.append(
                    candidate
                )

                effective_weight_rows.append(
                    effective_weight
                )

                gated_prediction_rows.append(
                    gated_prediction
                )

                recurrent_state_rows.append(
                    updated_state
                )

                previous_state = updated_state

        if tuple(
            global_indices
        ) != tuple(
            range(
                BLOCK_9_SENTENCE_COUNT
            )
        ):

            raise RuntimeError(
                f"Notebook 10 Block 9 recurrent traversal order is "
                f"invalid for pathway {pathway_name!r}."
            )

        candidate_matrix = torch.cat(
            candidate_rows,
            dim=
                0,
        )

        effective_weight_matrix = torch.stack(
            effective_weight_rows,
            dim=
                0,
        )

        gated_prediction_matrix = torch.cat(
            gated_prediction_rows,
            dim=
                0,
        )

        recurrent_state_matrix = torch.cat(
            recurrent_state_rows,
            dim=
                0,
        )

        target_matrix = (
            BLOCK_9_TARGETS[
                pathway_name
            ]
        )

        objective_mask = (
            BLOCK_9_OBJECTIVE_MASKS[
                pathway_name
            ]
        )

        squared_error = (
            gated_prediction_matrix
            -
            target_matrix
        ).pow(
            2
        )

        masked_error = squared_error[
            objective_mask
        ]

        if masked_error.numel() == 0:

            raise RuntimeError(
                f"No objective-active supervision exists for "
                f"{pathway_name}."
            )

        pathway_loss = (
            masked_error.mean()
        )

        pathway_outputs[
            pathway_name
        ] = {
            "candidate_matrix":
                candidate_matrix,

            "effective_weight_matrix":
                effective_weight_matrix,

            "gated_prediction_matrix":
                gated_prediction_matrix,

            "recurrent_state_matrix":
                recurrent_state_matrix,

            "pathway_loss":
                pathway_loss,
        }

    total_objective = torch.stack(
        [
            pathway_outputs[
                pathway_name
            ][
                "pathway_loss"
            ]

            for pathway_name
            in BLOCK_9_PATHWAY_DIMENSIONS
        ],
        dim=
            0,
    ).mean()

    return (
        pathway_outputs,
        total_objective,
    )


# =============================================================================
# Re-evaluate immutable Block 8 baseline before optimiser creation
# =============================================================================

with torch.no_grad():

    (
        BLOCK_9_BASELINE_FORWARD,
        BLOCK_9_BASELINE_TOTAL_OBJECTIVE_TENSOR,
    ) = block_9_forward_objective()


BLOCK_9_REEVALUATED_BASELINE_PATHWAY_LOSSES = {
    pathway_name:
        float(
            BLOCK_9_BASELINE_FORWARD[
                pathway_name
            ][
                "pathway_loss"
            ]
            .detach()
            .cpu()
            .item()
        )

    for pathway_name
    in BLOCK_9_PATHWAY_DIMENSIONS
}


BLOCK_9_REEVALUATED_BASELINE_TOTAL_OBJECTIVE = float(
    BLOCK_9_BASELINE_TOTAL_OBJECTIVE_TENSOR
    .detach()
    .cpu()
    .item()
)


BLOCK_9_BLOCK_8_BASELINE_MATCH = all(
    [
        np.isclose(
            BLOCK_9_REEVALUATED_BASELINE_TOTAL_OBJECTIVE,
            BLOCK_9_IMMUTABLE_BASELINE_TOTAL_OBJECTIVE,
            rtol=
                BLOCK_9_BASELINE_RTOL,
            atol=
                BLOCK_9_BASELINE_ATOL,
        ),

        all(
            np.isclose(
                BLOCK_9_REEVALUATED_BASELINE_PATHWAY_LOSSES[
                    pathway_name
                ],
                BLOCK_9_IMMUTABLE_BASELINE_PATHWAY_LOSSES[
                    pathway_name
                ],
                rtol=
                    BLOCK_9_BASELINE_RTOL,
                atol=
                    BLOCK_9_BASELINE_ATOL,
            )

            for pathway_name
            in BLOCK_9_PATHWAY_DIMENSIONS
        ),
    ]
)


if not BLOCK_9_BLOCK_8_BASELINE_MATCH:

    raise RuntimeError(
        "Notebook 10 Block 9 pre-training objective does not reproduce "
        "the immutable Block 8 baseline."
    )


# =============================================================================
# Initial latent and effective Transformative-Weight state
# =============================================================================

BLOCK_9_INITIAL_LATENT_WEIGHT_STATE = {
    pathway_name:
        {
            tensor_name:
                tensor.detach()
                .cpu()
                .clone()

            for tensor_name, tensor
            in module.state_dict().items()
        }

    for pathway_name, module
    in NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES.items()
}


with torch.no_grad():

    BLOCK_9_INITIAL_EFFECTIVE_WEIGHTS = {
        pathway_name:
            NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES[
                pathway_name
            ]()
            .detach()
            .cpu()
            .clone()

        for pathway_name
        in BLOCK_9_PATHWAY_DIMENSIONS
    }


BLOCK_9_INITIAL_EFFECTIVE_WEIGHTS_MATCH_BLOCK_8 = all(
    torch.equal(
        BLOCK_9_INITIAL_EFFECTIVE_WEIGHTS[
            pathway_name
        ],
        NOTEBOOK_10_PRETRAIN_EFFECTIVE_TRANSFORMATIVE_WEIGHTS[
            pathway_name
        ]
        .detach()
        .cpu(),
    )

    for pathway_name
    in BLOCK_9_PATHWAY_DIMENSIONS
)


if not BLOCK_9_INITIAL_EFFECTIVE_WEIGHTS_MATCH_BLOCK_8:

    raise RuntimeError(
        "Notebook 10 Block 9 initial effective weights do not reproduce "
        "the immutable Block 8 baseline state."
    )


# =============================================================================
# Optimiser construction and exact scope validation
# =============================================================================

BLOCK_9_OPTIMIZER = torch.optim.AdamW(
    BLOCK_9_OPTIMIZER_PARAMETERS,
    lr=
        BLOCK_9_LEARNING_RATE,
    weight_decay=
        BLOCK_9_WEIGHT_DECAY,
)


NOTEBOOK_10_BLOCK_9_OPTIMIZER_CREATED = True


BLOCK_9_OPTIMIZER_PARAMETER_IDS = {
    id(
        parameter
    )

    for parameter_group
    in BLOCK_9_OPTIMIZER.param_groups

    for parameter
    in parameter_group[
        "params"
    ]
}


BLOCK_9_EXPECTED_WEIGHT_PARAMETER_IDS = {
    id(
        parameter
    )

    for parameter
    in BLOCK_9_OPTIMIZER_PARAMETERS
}


BLOCK_9_REPRESENTATION_PARAMETER_IDS = {
    id(
        parameter
    )

    for module
    in NOTEBOOK_10_REPRESENTATION_MODULES.values()

    for parameter
    in module.parameters()
}


BLOCK_9_TRANSFORMATIVE_PARAMETER_IDS = {
    id(
        parameter
    )

    for module
    in NOTEBOOK_10_TRANSFORMATIVE_MODULES.values()

    for parameter
    in module.parameters()
}


BLOCK_9_OPTIMIZER_SCOPE_EXACT = (
    BLOCK_9_OPTIMIZER_PARAMETER_IDS
    ==
    BLOCK_9_EXPECTED_WEIGHT_PARAMETER_IDS
)


BLOCK_9_OPTIMIZER_EXCLUDES_REPRESENTATION = (
    BLOCK_9_OPTIMIZER_PARAMETER_IDS.isdisjoint(
        BLOCK_9_REPRESENTATION_PARAMETER_IDS
    )
)


BLOCK_9_OPTIMIZER_EXCLUDES_TRANSFORMATIVE = (
    BLOCK_9_OPTIMIZER_PARAMETER_IDS.isdisjoint(
        BLOCK_9_TRANSFORMATIVE_PARAMETER_IDS
    )
)


BLOCK_9_OPTIMIZER_SCOPE_VALID = all(
    [
        BLOCK_9_OPTIMIZER_SCOPE_EXACT,
        BLOCK_9_OPTIMIZER_EXCLUDES_REPRESENTATION,
        BLOCK_9_OPTIMIZER_EXCLUDES_TRANSFORMATIVE,
        sum(
            parameter.numel()

            for parameter_group
            in BLOCK_9_OPTIMIZER.param_groups

            for parameter
            in parameter_group[
                "params"
            ]
        )
        ==
        49,
    ]
)


if not BLOCK_9_OPTIMIZER_SCOPE_VALID:

    raise RuntimeError(
        "Notebook 10 Block 9 optimiser parameter scope is invalid."
    )


# =============================================================================
# Controlled training
# =============================================================================

BLOCK_9_TRAINING_HISTORY = []

BLOCK_9_ALL_GRADIENTS_FINITE = True

BLOCK_9_MAXIMUM_OBSERVED_GRADIENT_NORM = 0.0

BLOCK_9_ALL_EPOCH_OBJECTIVES_FINITE = True


for epoch_index in range(
    BLOCK_9_EPOCHS
):

    BLOCK_9_OPTIMIZER.zero_grad(
        set_to_none=
            True
    )

    NOTEBOOK_10_BLOCK_9_ZERO_GRAD_EXECUTED = True

    (
        epoch_forward,
        epoch_total_objective,
    ) = block_9_forward_objective()

    NOTEBOOK_10_BLOCK_9_TRANSFORMATIVE_FORWARD_EXECUTED = True
    NOTEBOOK_10_BLOCK_9_WEIGHT_FORWARD_EXECUTED = True
    NOTEBOOK_10_BLOCK_9_RECURRENT_STATE_CREATED = True
    NOTEBOOK_10_BLOCK_9_RECURRENT_UPDATE_EXECUTED = True
    NOTEBOOK_10_BLOCK_9_LOSS_CALCULATED = True

    if not bool(
        torch.isfinite(
            epoch_total_objective
        ).item()
    ):

        BLOCK_9_ALL_EPOCH_OBJECTIVES_FINITE = False

        raise RuntimeError(
            f"Notebook 10 Block 9 epoch {epoch_index + 1} "
            "produced a non-finite objective."
        )

    epoch_total_objective.backward()

    NOTEBOOK_10_BLOCK_9_BACKWARD_PASS_EXECUTED = True

    epoch_gradients_finite = all(
        (
            parameter.grad is not None
            and
            bool(
                torch.isfinite(
                    parameter.grad
                ).all()
                .item()
            )
        )

        for parameter
        in BLOCK_9_OPTIMIZER_PARAMETERS
    )

    if not epoch_gradients_finite:

        BLOCK_9_ALL_GRADIENTS_FINITE = False

        raise RuntimeError(
            f"Notebook 10 Block 9 epoch {epoch_index + 1} "
            "produced missing or non-finite Transformative-Weight gradients."
        )

    preclip_gradient_norm = (
        torch.nn.utils.clip_grad_norm_(
            BLOCK_9_OPTIMIZER_PARAMETERS,
            max_norm=
                BLOCK_9_MAXIMUM_GRADIENT_NORM,
        )
    )

    NOTEBOOK_10_BLOCK_9_GRADIENT_CLIPPING_EXECUTED = True

    preclip_gradient_norm_value = float(
        preclip_gradient_norm.detach()
        .cpu()
        .item()
    )


    if not math.isfinite(
        preclip_gradient_norm_value
    ):

        BLOCK_9_ALL_GRADIENTS_FINITE = False

        raise RuntimeError(
            f"Notebook 10 Block 9 epoch {epoch_index + 1} "
            "produced a non-finite gradient norm."
        )

    BLOCK_9_MAXIMUM_OBSERVED_GRADIENT_NORM = max(
        BLOCK_9_MAXIMUM_OBSERVED_GRADIENT_NORM,
        preclip_gradient_norm_value,
    )

    BLOCK_9_OPTIMIZER.step()

    NOTEBOOK_10_BLOCK_9_OPTIMIZER_STEP_EXECUTED = True
    NOTEBOOK_10_BLOCK_9_PARAMETER_UPDATE_EXECUTED = True

    epoch_pathway_losses = {
        pathway_name:
            float(
                epoch_forward[
                    pathway_name
                ][
                    "pathway_loss"
                ]
                .detach()
                .cpu()
                .item()
            )

        for pathway_name
        in BLOCK_9_PATHWAY_DIMENSIONS
    }

    BLOCK_9_TRAINING_HISTORY.append(
        {
            "epoch":
                int(
                    epoch_index
                    +
                    1
                ),

            "total_objective":
                float(
                    epoch_total_objective.detach()
                    .cpu()
                    .item()
                ),

            "pathway_losses":
                epoch_pathway_losses,

            "preclip_gradient_norm":
                preclip_gradient_norm_value,
        }
    )


NOTEBOOK_10_BLOCK_9_TRAINING_EXECUTED = True


# =============================================================================
# Clear residual gradients after training
# =============================================================================

BLOCK_9_OPTIMIZER.zero_grad(
    set_to_none=
        True
)


NOTEBOOK_10_BLOCK_9_ZERO_GRAD_EXECUTED = True


BLOCK_9_RESIDUAL_GRADIENTS_CLEARED = all(
    parameter.grad is None

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


# =============================================================================
# Final post-training objective evaluation
# =============================================================================

with torch.no_grad():

    (
        BLOCK_9_FINAL_FORWARD,
        BLOCK_9_FINAL_TOTAL_OBJECTIVE_TENSOR,
    ) = block_9_forward_objective()


BLOCK_9_FINAL_PATHWAY_LOSSES = {
    pathway_name:
        float(
            BLOCK_9_FINAL_FORWARD[
                pathway_name
            ][
                "pathway_loss"
            ]
            .detach()
            .cpu()
            .item()
        )

    for pathway_name
    in BLOCK_9_PATHWAY_DIMENSIONS
}


BLOCK_9_FINAL_TOTAL_OBJECTIVE = float(
    BLOCK_9_FINAL_TOTAL_OBJECTIVE_TENSOR
    .detach()
    .cpu()
    .item()
)


BLOCK_9_FINAL_OBJECTIVE_FINITE = all(
    [
        math.isfinite(
            BLOCK_9_FINAL_TOTAL_OBJECTIVE
        ),

        all(
            math.isfinite(
                pathway_loss
            )

            for pathway_loss
            in BLOCK_9_FINAL_PATHWAY_LOSSES.values()
        ),
    ]
)


if not BLOCK_9_FINAL_OBJECTIVE_FINITE:

    raise RuntimeError(
        "Notebook 10 Block 9 final objective is non-finite."
    )


# =============================================================================
# Objective comparison against immutable Block 8 baseline
# =============================================================================

BLOCK_9_OBJECTIVE_DELTA = (
    BLOCK_9_FINAL_TOTAL_OBJECTIVE
    -
    BLOCK_9_IMMUTABLE_BASELINE_TOTAL_OBJECTIVE
)


BLOCK_9_OBJECTIVE_IMPROVED = (
    BLOCK_9_OBJECTIVE_DELTA
    <
    0.0
)


BLOCK_9_PATHWAY_LOSS_DELTAS = {
    pathway_name:
        (
            BLOCK_9_FINAL_PATHWAY_LOSSES[
                pathway_name
            ]
            -
            BLOCK_9_IMMUTABLE_BASELINE_PATHWAY_LOSSES[
                pathway_name
            ]
        )

    for pathway_name
    in BLOCK_9_PATHWAY_DIMENSIONS
}


# =============================================================================
# Final latent/effective Transformative-Weight state
# =============================================================================

BLOCK_9_FINAL_WEIGHT_STATE = (
    block_9_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES
    )
)


with torch.no_grad():

    BLOCK_9_FINAL_EFFECTIVE_WEIGHTS = {
        pathway_name:
            NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES[
                pathway_name
            ]()
            .detach()
            .cpu()
            .clone()

        for pathway_name
        in BLOCK_9_PATHWAY_DIMENSIONS
    }


BLOCK_9_WEIGHT_PARAMETER_DELTAS = {}

BLOCK_9_WEIGHT_PARAMETERS_UPDATED = {}

BLOCK_9_EFFECTIVE_WEIGHTS_FINITE = {}

BLOCK_9_EFFECTIVE_WEIGHTS_BOUNDED = {}

BLOCK_9_DEFERRED_WEIGHTS_ZERO = {}


for pathway_name, pathway_dim in BLOCK_9_PATHWAY_DIMENSIONS.items():

    before_state = (
        BLOCK_9_WEIGHT_STATE_BEFORE[
            pathway_name
        ]
    )

    after_state = (
        BLOCK_9_FINAL_WEIGHT_STATE[
            pathway_name
        ]
    )

    if (
        before_state.keys()
        !=
        after_state.keys()
    ):

        raise RuntimeError(
            f"Notebook 10 Block 9 Transformative-Weight state schema "
            f"changed for {pathway_name}."
        )

    tensor_deltas = []

    for tensor_name in before_state:

        before_tensor = (
            before_state[
                tensor_name
            ]
        )

        after_tensor = (
            after_state[
                tensor_name
            ]
        )

        if before_tensor.shape != after_tensor.shape:

            raise RuntimeError(
                f"Notebook 10 Block 9 Transformative-Weight tensor "
                f"shape changed for {pathway_name}.{tensor_name}."
            )

        if before_tensor.numel() > 0:

            tensor_deltas.append(
                float(
                    (
                        after_tensor
                        -
                        before_tensor
                    )
                    .abs()
                    .max()
                    .item()
                )
            )

    maximum_delta = (
        max(
            tensor_deltas
        )
        if tensor_deltas
        else 0.0
    )

    BLOCK_9_WEIGHT_PARAMETER_DELTAS[
        pathway_name
    ] = maximum_delta

    BLOCK_9_WEIGHT_PARAMETERS_UPDATED[
        pathway_name
    ] = (
        maximum_delta
        >
        0.0
    )

    effective_weight = (
        BLOCK_9_FINAL_EFFECTIVE_WEIGHTS[
            pathway_name
        ]
    )

    BLOCK_9_EFFECTIVE_WEIGHTS_FINITE[
        pathway_name
    ] = bool(
        torch.isfinite(
            effective_weight
        ).all()
        .item()
    )

    BLOCK_9_EFFECTIVE_WEIGHTS_BOUNDED[
        pathway_name
    ] = bool(
        (
            (
                effective_weight
                >=
                0.0
            )
            &
            (
                effective_weight
                <=
                1.0
            )
        ).all()
        .item()
    )

    deferred_indices = (
        BLOCK_9_DEFERRED_DIMENSION_INDICES[
            pathway_name
        ]
    )

    if deferred_indices:

        BLOCK_9_DEFERRED_WEIGHTS_ZERO[
            pathway_name
        ] = bool(
            torch.equal(
                effective_weight[
                    list(
                        deferred_indices
                    )
                ],
                torch.zeros(
                    len(
                        deferred_indices
                    ),
                    dtype=
                        effective_weight.dtype,
                ),
            )
        )

    else:

        BLOCK_9_DEFERRED_WEIGHTS_ZERO[
            pathway_name
        ] = True


BLOCK_9_ALL_WEIGHT_PATHWAYS_UPDATED = all(
    BLOCK_9_WEIGHT_PARAMETERS_UPDATED.values()
)


BLOCK_9_FINAL_WEIGHT_NUMERICAL_CONTRACT_VALID = all(
    [
        all(
            BLOCK_9_EFFECTIVE_WEIGHTS_FINITE.values()
        ),

        all(
            BLOCK_9_EFFECTIVE_WEIGHTS_BOUNDED.values()
        ),

        all(
            BLOCK_9_DEFERRED_WEIGHTS_ZERO.values()
        ),
    ]
)


# =============================================================================
# Exact inherited architecture validation after training
# =============================================================================

BLOCK_9_REPRESENTATION_STATE_AFTER = (
    block_9_snapshot_module_collection(
        NOTEBOOK_10_REPRESENTATION_MODULES
    )
)


BLOCK_9_TRANSFORMATIVE_STATE_AFTER = (
    block_9_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_MODULES
    )
)


BLOCK_9_REPRESENTATION_UNCHANGED = (
    block_9_nested_state_exact(
        BLOCK_9_REPRESENTATION_STATE_BEFORE,
        BLOCK_9_REPRESENTATION_STATE_AFTER,
    )
)


BLOCK_9_TRANSFORMATIVE_UNCHANGED = (
    block_9_nested_state_exact(
        BLOCK_9_TRANSFORMATIVE_STATE_BEFORE,
        BLOCK_9_TRANSFORMATIVE_STATE_AFTER,
    )
)


BLOCK_9_WEIGHT_STATE_CHANGED = (
    not
    block_9_nested_state_exact(
        BLOCK_9_WEIGHT_STATE_BEFORE,
        BLOCK_9_FINAL_WEIGHT_STATE,
    )
)


BLOCK_9_REPRESENTATION_FROZEN_AFTER = all(
    not parameter.requires_grad

    for module
    in NOTEBOOK_10_REPRESENTATION_MODULES.values()

    for parameter
    in module.parameters()
)


BLOCK_9_TRANSFORMATIVE_FROZEN_AFTER = all(
    not parameter.requires_grad

    for module
    in NOTEBOOK_10_TRANSFORMATIVE_MODULES.values()

    for parameter
    in module.parameters()
)


BLOCK_9_WEIGHT_PARAMETERS_TRAINABLE_AFTER = all(
    parameter.requires_grad

    for module
    in NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES.values()

    for parameter
    in module.parameters()
)


BLOCK_9_ALL_MODULES_EVAL_AFTER = all(
    not module.training

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()
)


# =============================================================================
# Causal/leakage validation
# =============================================================================

BLOCK_9_CAUSAL_TRAINING_CONTRACT_VALID = all(
    [
        BLOCK_9_CANONICAL_GLOBAL_INDICES
        ==
        tuple(
            range(
                BLOCK_9_SENTENCE_COUNT
            )
        ),

        NOTEBOOK_10_BLOCK_9_FUTURE_CONTEXT_USED
        is False,

        NOTEBOOK_10_BLOCK_9_SUPERVISION_USED_AS_MODEL_INPUT
        is False,

        NOTEBOOK_10_BLOCK_9_BATCH_SHUFFLING_USED
        is False,
    ]
)


# =============================================================================
# Execution-state validation
# =============================================================================

BLOCK_9_EXECUTION_STATE_VALID = all(
    [
        NOTEBOOK_10_BLOCK_9_REPRESENTATION_FORWARD_EXECUTED
        is False,

        NOTEBOOK_10_BLOCK_9_TRANSFORMATIVE_FORWARD_EXECUTED
        is True,

        NOTEBOOK_10_BLOCK_9_WEIGHT_FORWARD_EXECUTED
        is True,

        NOTEBOOK_10_BLOCK_9_RECURRENT_STATE_CREATED
        is True,

        NOTEBOOK_10_BLOCK_9_RECURRENT_UPDATE_EXECUTED
        is True,

        NOTEBOOK_10_BLOCK_9_LOSS_CALCULATED
        is True,

        NOTEBOOK_10_BLOCK_9_OPTIMIZER_CREATED
        is True,

        NOTEBOOK_10_BLOCK_9_ZERO_GRAD_EXECUTED
        is True,

        NOTEBOOK_10_BLOCK_9_BACKWARD_PASS_EXECUTED
        is True,

        NOTEBOOK_10_BLOCK_9_GRADIENT_CLIPPING_EXECUTED
        is True,

        NOTEBOOK_10_BLOCK_9_OPTIMIZER_STEP_EXECUTED
        is True,

        NOTEBOOK_10_BLOCK_9_PARAMETER_UPDATE_EXECUTED
        is True,

        NOTEBOOK_10_BLOCK_9_TRAINING_EXECUTED
        is True,
    ]
)


# =============================================================================
# Final validation
# =============================================================================

NOTEBOOK_10_BLOCK_9_ERRORS = []


BLOCK_9_VALIDATION_CHECKS = {
    "prerequisites_invalid":
        BLOCK_9_PREREQUISITES_VALID,

    "pathway_dimensions_invalid":
        BLOCK_9_PATHWAY_DIMENSIONS_VALID,

    "dimension_partitions_invalid":
        BLOCK_9_DIMENSION_PARTITIONS_VALID,

    "canonical_index_contract_invalid":
        BLOCK_9_CANONICAL_INDEX_CONTRACT_VALID,

    "device_dtype_invalid":
        BLOCK_9_DEVICE_DTYPE_VALID,

    "parameter_accounting_invalid":
        BLOCK_9_PARAMETER_ACCOUNTING_VALID,

    "fixed_input_contract_invalid":
        BLOCK_9_FIXED_INPUT_CONTRACT_VALID,

    "objective_counts_changed":
        BLOCK_9_OBJECTIVE_COUNTS_MATCH_BLOCK_8,

    "parameter_scope_invalid":
        BLOCK_9_PARAMETER_SCOPE_VALID,

    "block_8_baseline_mismatch":
        BLOCK_9_BLOCK_8_BASELINE_MATCH,

    "initial_weights_mismatch":
        BLOCK_9_INITIAL_EFFECTIVE_WEIGHTS_MATCH_BLOCK_8,

    "optimizer_scope_invalid":
        BLOCK_9_OPTIMIZER_SCOPE_VALID,

    "epoch_objective_nonfinite":
        BLOCK_9_ALL_EPOCH_OBJECTIVES_FINITE,

    "gradient_nonfinite":
        BLOCK_9_ALL_GRADIENTS_FINITE,

    "residual_gradients_present":
        BLOCK_9_RESIDUAL_GRADIENTS_CLEARED,

    "final_objective_nonfinite":
        BLOCK_9_FINAL_OBJECTIVE_FINITE,

    "objective_not_improved":
        BLOCK_9_OBJECTIVE_IMPROVED,

    "weight_pathway_not_updated":
        BLOCK_9_ALL_WEIGHT_PATHWAYS_UPDATED,

    "weight_state_not_changed":
        BLOCK_9_WEIGHT_STATE_CHANGED,

    "final_weight_contract_invalid":
        BLOCK_9_FINAL_WEIGHT_NUMERICAL_CONTRACT_VALID,

    "representation_state_changed":
        BLOCK_9_REPRESENTATION_UNCHANGED,

    "transformative_state_changed":
        BLOCK_9_TRANSFORMATIVE_UNCHANGED,

    "representation_not_frozen":
        BLOCK_9_REPRESENTATION_FROZEN_AFTER,

    "transformative_not_frozen":
        BLOCK_9_TRANSFORMATIVE_FROZEN_AFTER,

    "weight_parameters_not_trainable":
        BLOCK_9_WEIGHT_PARAMETERS_TRAINABLE_AFTER,

    "modules_not_eval":
        BLOCK_9_ALL_MODULES_EVAL_AFTER,

    "causal_training_contract_invalid":
        BLOCK_9_CAUSAL_TRAINING_CONTRACT_VALID,

    "execution_state_invalid":
        BLOCK_9_EXECUTION_STATE_VALID,
}


for error_name, condition in BLOCK_9_VALIDATION_CHECKS.items():

    if not condition:

        NOTEBOOK_10_BLOCK_9_ERRORS.append(
            error_name
        )


NOTEBOOK_10_CONTROLLED_CONTINUED_PILOT_TRAINING_VALID = (
    len(
        NOTEBOOK_10_BLOCK_9_ERRORS
    )
    ==
    0
)


NOTEBOOK_10_BLOCK_9_VALID = (
    NOTEBOOK_10_CONTROLLED_CONTINUED_PILOT_TRAINING_VALID
)


if not NOTEBOOK_10_BLOCK_9_VALID:

    raise RuntimeError(
        "Notebook 10 Block 9 controlled continued pilot training failed. "
        f"Errors: {NOTEBOOK_10_BLOCK_9_ERRORS}"
    )


NOTEBOOK_10_BLOCK_9_COMPLETE = True


# =============================================================================
# Stable Block 9 → 10 handover
# =============================================================================

NOTEBOOK_10_TRAINING_HISTORY = deepcopy(
    BLOCK_9_TRAINING_HISTORY
)


NOTEBOOK_10_POSTTRAIN_PATHWAY_LOSSES = deepcopy(
    BLOCK_9_FINAL_PATHWAY_LOSSES
)


NOTEBOOK_10_POSTTRAIN_TOTAL_OBJECTIVE = float(
    BLOCK_9_FINAL_TOTAL_OBJECTIVE
)


NOTEBOOK_10_POSTTRAIN_OBJECTIVE_DELTA = float(
    BLOCK_9_OBJECTIVE_DELTA
)


NOTEBOOK_10_POSTTRAIN_PATHWAY_LOSS_DELTAS = deepcopy(
    BLOCK_9_PATHWAY_LOSS_DELTAS
)


NOTEBOOK_10_TRAINED_TRANSFORMATIVE_WEIGHT_STATE = deepcopy(
    BLOCK_9_FINAL_WEIGHT_STATE
)


NOTEBOOK_10_TRAINED_EFFECTIVE_TRANSFORMATIVE_WEIGHTS = {
    pathway_name:
        effective_weight.clone()

    for pathway_name, effective_weight
    in BLOCK_9_FINAL_EFFECTIVE_WEIGHTS.items()
}


NOTEBOOK_10_POSTTRAIN_VALIDATION_READY = (
    NOTEBOOK_10_BLOCK_9_VALID
)


# =============================================================================
# Block summary
# =============================================================================

NOTEBOOK_10_BLOCK_9_SUMMARY = {
    "block":
        NOTEBOOK_10_BLOCK_9,

    "block_name":
        NOTEBOOK_10_BLOCK_9_NAME,

    "version":
        NOTEBOOK_10_BLOCK_9_VERSION,

    "executed_at_utc":
        BLOCK_9_EXECUTED_AT_UTC,

    "training_classification":
        "continued_pilot_optimisation",

    "article_count":
        BLOCK_9_ARTICLE_COUNT,

    "sentence_count":
        BLOCK_9_SENTENCE_COUNT,

    "optimizer":
        BLOCK_9_OPTIMIZER_NAME,

    "learning_rate":
        BLOCK_9_LEARNING_RATE,

    "weight_decay":
        BLOCK_9_WEIGHT_DECAY,

    "maximum_gradient_norm":
        BLOCK_9_MAXIMUM_GRADIENT_NORM,

    "epochs":
        BLOCK_9_EPOCHS,

    "optimizer_parameter_count":
        BLOCK_9_OPTIMIZER_PARAMETER_COUNT,

    "objective_active_elements":
        deepcopy(
            BLOCK_9_OBJECTIVE_ACTIVE_ELEMENTS
        ),

    "pretraining_pathway_losses":
        deepcopy(
            BLOCK_9_IMMUTABLE_BASELINE_PATHWAY_LOSSES
        ),

    "pretraining_total_objective":
        BLOCK_9_IMMUTABLE_BASELINE_TOTAL_OBJECTIVE,

    "posttraining_pathway_losses":
        deepcopy(
            BLOCK_9_FINAL_PATHWAY_LOSSES
        ),

    "posttraining_total_objective":
        BLOCK_9_FINAL_TOTAL_OBJECTIVE,

    "objective_delta":
        BLOCK_9_OBJECTIVE_DELTA,

    "objective_improved":
        BLOCK_9_OBJECTIVE_IMPROVED,

    "maximum_observed_gradient_norm":
        BLOCK_9_MAXIMUM_OBSERVED_GRADIENT_NORM,

    "weight_parameter_deltas":
        deepcopy(
            BLOCK_9_WEIGHT_PARAMETER_DELTAS
        ),

    "representation_unchanged":
        BLOCK_9_REPRESENTATION_UNCHANGED,

    "transformative_unchanged":
        BLOCK_9_TRANSFORMATIVE_UNCHANGED,

    "weight_state_changed":
        BLOCK_9_WEIGHT_STATE_CHANGED,

    "controlled_training_valid":
        NOTEBOOK_10_CONTROLLED_CONTINUED_PILOT_TRAINING_VALID,

    "posttraining_validation_ready":
        NOTEBOOK_10_POSTTRAIN_VALIDATION_READY,

    "block_valid":
        NOTEBOOK_10_BLOCK_9_VALID,

    "block_complete":
        NOTEBOOK_10_BLOCK_9_COMPLETE,
}


# =============================================================================
# Final report
# =============================================================================

print("=" * 72)

print(
    "Media AI — Notebook 10, Block 9: "
    "Controlled Continued Pilot Training "
    "and Parameter-Update Validation"
)

print("=" * 72)

print(
    f"Block version                : "
    f"{NOTEBOOK_10_BLOCK_9_VERSION}"
)

print(
    f"Training classification      : "
    f"continued_pilot_optimisation"
)

print(
    f"Execution device             : "
    f"{BLOCK_9_EXECUTION_DEVICE}"
)

print(
    f"Execution dtype              : "
    f"{BLOCK_9_EXECUTION_DTYPE}"
)

print("-" * 72)

print(
    "Controlled optimisation policy"
)

print(
    f"Optimizer                    : "
    f"{BLOCK_9_OPTIMIZER_NAME}"
)

print(
    f"Learning rate                : "
    f"{BLOCK_9_LEARNING_RATE}"
)

print(
    f"Weight decay                 : "
    f"{BLOCK_9_WEIGHT_DECAY}"
)

print(
    f"Maximum gradient norm        : "
    f"{BLOCK_9_MAXIMUM_GRADIENT_NORM}"
)

print(
    f"Epochs                       : "
    f"{BLOCK_9_EPOCHS}"
)

print(
    f"Sequence order preserved     : "
    f"{BLOCK_9_CANONICAL_INDEX_CONTRACT_VALID}"
)

print(
    f"Batch shuffling used         : "
    f"{NOTEBOOK_10_BLOCK_9_BATCH_SHUFFLING_USED}"
)

print(
    f"Future context used          : "
    f"{NOTEBOOK_10_BLOCK_9_FUTURE_CONTEXT_USED}"
)

print("-" * 72)

print(
    "Parameter scope"
)

print(
    f"Representation parameters    : "
    f"{BLOCK_9_REPRESENTATION_PARAMETER_COUNT:,}"
)

print(
    f"Transformative parameters    : "
    f"{BLOCK_9_TRANSFORMATIVE_PARAMETER_COUNT:,}"
)

print(
    f"Transformative Weight params : "
    f"{BLOCK_9_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT:,}"
)

print(
    f"Total model parameters       : "
    f"{BLOCK_9_TOTAL_PARAMETER_COUNT:,}"
)

print(
    f"Optimizer parameter count    : "
    f"{BLOCK_9_OPTIMIZER_PARAMETER_COUNT}"
)

print(
    f"Optimizer scope exact        : "
    f"{BLOCK_9_OPTIMIZER_SCOPE_VALID}"
)

print("-" * 72)

print(
    "Objective-active supervision"
)

for pathway_name in BLOCK_9_PATHWAY_DIMENSIONS:

    print(
        f"{pathway_name:<14} target elements    : "
        f"{BLOCK_9_OBJECTIVE_ACTIVE_ELEMENTS[pathway_name]}"
    )

print("-" * 72)

print(
    "Controlled training objective"
)

print(
    f"Immutable Block 8 baseline   : "
    f"{BLOCK_9_IMMUTABLE_BASELINE_TOTAL_OBJECTIVE:.8f}"
)

print(
    f"Re-evaluated baseline        : "
    f"{BLOCK_9_REEVALUATED_BASELINE_TOTAL_OBJECTIVE:.8f}"
)

print(
    f"Baseline match               : "
    f"{BLOCK_9_BLOCK_8_BASELINE_MATCH}"
)

print(
    f"Final total objective        : "
    f"{BLOCK_9_FINAL_TOTAL_OBJECTIVE:.8f}"
)

print(
    f"Objective delta              : "
    f"{BLOCK_9_OBJECTIVE_DELTA:.8f}"
)

print(
    f"Objective improved           : "
    f"{BLOCK_9_OBJECTIVE_IMPROVED}"
)

print("-" * 72)

print(
    "Pathway losses"
)

for pathway_name in BLOCK_9_PATHWAY_DIMENSIONS:

    print(
        f"{pathway_name:<14} initial masked MSE : "
        f"{BLOCK_9_IMMUTABLE_BASELINE_PATHWAY_LOSSES[pathway_name]:.8f}"
    )

    print(
        f"{pathway_name:<14} final masked MSE   : "
        f"{BLOCK_9_FINAL_PATHWAY_LOSSES[pathway_name]:.8f}"
    )

    print(
        f"{pathway_name:<14} loss delta         : "
        f"{BLOCK_9_PATHWAY_LOSS_DELTAS[pathway_name]:.8f}"
    )

print("-" * 72)

print(
    "Gradient and parameter validation"
)

print(
    f"All gradients finite         : "
    f"{BLOCK_9_ALL_GRADIENTS_FINITE}"
)

print(
    f"Maximum observed grad norm   : "
    f"{BLOCK_9_MAXIMUM_OBSERVED_GRADIENT_NORM:.8f}"
)

for pathway_name in BLOCK_9_PATHWAY_DIMENSIONS:

    print(
        f"{pathway_name:<14} max latent delta   : "
        f"{BLOCK_9_WEIGHT_PARAMETER_DELTAS[pathway_name]:.10f}"
    )

    print(
        f"{pathway_name:<14} parameters updated : "
        f"{BLOCK_9_WEIGHT_PARAMETERS_UPDATED[pathway_name]}"
    )

print("-" * 72)

print(
    "Final Transformative-Weight validation"
)

for pathway_name in BLOCK_9_PATHWAY_DIMENSIONS:

    active_indices = (
        BLOCK_9_ACTIVE_DIMENSION_INDICES[
            pathway_name
        ]
    )

    final_effective = (
        BLOCK_9_FINAL_EFFECTIVE_WEIGHTS[
            pathway_name
        ]
    )

    initial_effective = (
        BLOCK_9_INITIAL_EFFECTIVE_WEIGHTS[
            pathway_name
        ]
    )

    print(
        f"{pathway_name:<14} active indices    : "
        f"{list(active_indices)}"
    )

    print(
        f"{pathway_name:<14} initial weights   : "
        f"{[round(float(initial_effective[index]), 6) for index in active_indices]}"
    )

    print(
        f"{pathway_name:<14} final weights     : "
        f"{[round(float(final_effective[index]), 6) for index in active_indices]}"
    )

    print(
        f"{pathway_name:<14} finite            : "
        f"{BLOCK_9_EFFECTIVE_WEIGHTS_FINITE[pathway_name]}"
    )

    print(
        f"{pathway_name:<14} bounded           : "
        f"{BLOCK_9_EFFECTIVE_WEIGHTS_BOUNDED[pathway_name]}"
    )

    print(
        f"{pathway_name:<14} deferred zero     : "
        f"{BLOCK_9_DEFERRED_WEIGHTS_ZERO[pathway_name]}"
    )

print("-" * 72)

print(
    "Inherited architecture validation"
)

print(
    f"Representation unchanged     : "
    f"{BLOCK_9_REPRESENTATION_UNCHANGED}"
)

print(
    f"Transformative unchanged     : "
    f"{BLOCK_9_TRANSFORMATIVE_UNCHANGED}"
)

print(
    f"Transformative Weight changed: "
    f"{BLOCK_9_WEIGHT_STATE_CHANGED}"
)

print(
    f"Representation frozen        : "
    f"{BLOCK_9_REPRESENTATION_FROZEN_AFTER}"
)

print(
    f"Transformative frozen        : "
    f"{BLOCK_9_TRANSFORMATIVE_FROZEN_AFTER}"
)

print(
    f"Weight parameters trainable  : "
    f"{BLOCK_9_WEIGHT_PARAMETERS_TRAINABLE_AFTER}"
)

print(
    f"Residual gradients cleared   : "
    f"{BLOCK_9_RESIDUAL_GRADIENTS_CLEARED}"
)

print("-" * 72)

print(
    "Execution state"
)

print(
    f"Representation forward       : "
    f"{NOTEBOOK_10_BLOCK_9_REPRESENTATION_FORWARD_EXECUTED}"
)

print(
    f"Transform forward            : "
    f"{NOTEBOOK_10_BLOCK_9_TRANSFORMATIVE_FORWARD_EXECUTED}"
)

print(
    f"Weight forward               : "
    f"{NOTEBOOK_10_BLOCK_9_WEIGHT_FORWARD_EXECUTED}"
)

print(
    f"Recurrent state created      : "
    f"{NOTEBOOK_10_BLOCK_9_RECURRENT_STATE_CREATED}"
)

print(
    f"Recurrent update executed    : "
    f"{NOTEBOOK_10_BLOCK_9_RECURRENT_UPDATE_EXECUTED}"
)

print(
    f"Loss calculated              : "
    f"{NOTEBOOK_10_BLOCK_9_LOSS_CALCULATED}"
)

print(
    f"Optimizer created            : "
    f"{NOTEBOOK_10_BLOCK_9_OPTIMIZER_CREATED}"
)

print(
    f"Backward pass executed       : "
    f"{NOTEBOOK_10_BLOCK_9_BACKWARD_PASS_EXECUTED}"
)

print(
    f"Gradient clipping executed   : "
    f"{NOTEBOOK_10_BLOCK_9_GRADIENT_CLIPPING_EXECUTED}"
)

print(
    f"Optimizer step executed      : "
    f"{NOTEBOOK_10_BLOCK_9_OPTIMIZER_STEP_EXECUTED}"
)

print(
    f"Parameter update executed    : "
    f"{NOTEBOOK_10_BLOCK_9_PARAMETER_UPDATE_EXECUTED}"
)

print(
    f"Training executed            : "
    f"{NOTEBOOK_10_BLOCK_9_TRAINING_EXECUTED}"
)

print("-" * 72)

print(
    f"Controlled training valid    : "
    f"{NOTEBOOK_10_CONTROLLED_CONTINUED_PILOT_TRAINING_VALID}"
)

print(
    f"Post-training validation ready: "
    f"{NOTEBOOK_10_POSTTRAIN_VALIDATION_READY}"
)

print(
    f"Block valid                  : "
    f"{NOTEBOOK_10_BLOCK_9_VALID}"
)

print(
    f"Block complete               : "
    f"{NOTEBOOK_10_BLOCK_9_COMPLETE}"
)

print("=" * 72)

print(
    f"The {BLOCK_9_OPTIMIZER_PARAMETER_COUNT} activation-eligible "
    "Transformative-Weight parameter(s) were optimised through the complete "
    f"causal corpus of {BLOCK_9_SENTENCE_COUNT} sentence(s) across "
    f"{BLOCK_9_ARTICLE_COUNT} article(s)."
)

print(
    "The recurrent state remained differentiably connected across sentence "
    "positions within each article, while every article began from a fresh "
    "deterministic zero state."
)

print(
    "Only mask-valid and activation-eligible signed transformation targets "
    "contributed to the equal-pathway optimisation objective."
)

print(
    "The immutable Block 8 baseline was reproduced before optimiser creation "
    "and remained the fixed reference for the post-training comparison."
)

print(
    f"The controlled objective changed by "
    f"{BLOCK_9_OBJECTIVE_DELTA:.8f} under the "
    f"{BLOCK_9_EPOCHS}-epoch AdamW policy."
)

print(
    "The 894,003 representation parameters and 8,187 transformative "
    "parameters remained frozen and exactly unchanged."
)

print(
    "Only the 49 latent Transformative-Weight parameter(s) were supplied "
    "to the optimiser and received parameter updates."
)

print(
    "This result demonstrates continued pilot optimisation and parameter-scope "
    "isolation on the existing three-article corpus; it is not interpreted as "
    "expanded-corpus training or evidence of generalisation."
)

print(
    "Notebook 10 may now proceed to Block 10 — Post-Training Recurrent "
    "Validation, Objective Comparison and Generalisation-Boundary Audit."
)

print("=" * 72)
```

# Block 10 — Post-Training Recurrent Validation, Objective Comparison and Generalisation-Boundary Audit

## Purpose

This block performs the final post-training validation of the controlled Transformative-Weight optimisation executed in Block 9.

Block 9 demonstrated that the activation-eligible Transformative-Weight parameters can be updated under the established causal recurrent objective while preserving the frozen representation and transformative architecture. The present block does not continue optimisation. Instead, it evaluates the resulting trained state as a fixed model state and determines whether the post-training recurrent behaviour remains structurally, numerically and causally valid.

The block therefore separates **parameter learning** from **post-training model validation**.

Its principal objectives are to:

1. freeze the Transformative-Weight parameters learned in Block 9;
2. execute the complete multi-article recurrent pathway using the trained weights;
3. verify deterministic article-level recurrent behaviour after training;
4. recompute the factual, psychological and social masked objectives;
5. compare the resulting objective values against the immutable Block 8 pre-training baseline;
6. confirm that the Block 9 improvement is reproducible without additional parameter updates;
7. verify preservation of article-boundary resets, pathway independence and causal ordering;
8. confirm that deferred Transformative-Weight dimensions remain structurally inactive;
9. verify that the inherited representation and transformative parameters remain exactly unchanged; and
10. establish the explicit scientific boundary between demonstrated pilot learnability and untested generalisation.

---

## Post-Training Model State

The architecture entering this block consists of the frozen representation and transformative components inherited from Notebook 09 together with the Transformative-Weight state learned in Block 9.

The complete parameterisation remains:

- **Representation architecture:** 894,003 parameters
- **Transformative architecture:** 8,187 parameters
- **Transformative-Weight architecture:** 49 learned latent parameters
- **Complete Media AI architecture:** 902,239 parameters

No parameter is eligible for optimisation in this block.

The learned Transformative-Weight parameters are treated as fixed post-training model state and all model components are placed in evaluation mode before recurrent validation begins.

This establishes a strict execution boundary:

**Block 9 learns; Block 10 validates.**

---

## Post-Training Recurrent Execution

For each article $a$, each pathway begins from an independent deterministic zero state:

$$
\mathbf{s}^{(p,a)}_0 = \mathbf{0},
$$

where $p$ denotes the factual, psychological or social pathway.

For sentence position $t$, the frozen transformative mechanism receives the current pathway representation and the immediately preceding recurrent state:

$$
\mathbf{c}^{(p,a)}_t
=
T^{(p)}
\left(
\mathbf{r}^{(p,a)}_t,
\mathbf{s}^{(p,a)}_{t-1}
\right).
$$

The learned effective Transformative-Weight vector is then applied dimension-wise:

$$
\tilde{\mathbf{c}}^{(p,a)}_t
=
\mathbf{w}^{(p)}
\odot
\mathbf{c}^{(p,a)}_t.
$$

The recurrent state follows the established additive update:

$$
\mathbf{s}^{(p,a)}_t
=
\mathbf{s}^{(p,a)}_{t-1}
+
\tilde{\mathbf{c}}^{(p,a)}_t.
$$

The state is reset to zero at every article boundary. No recurrent information may propagate between independent articles.

The block must therefore verify that the learned Transformative Weights alter only the magnitude of activation-eligible candidate transformations and do not modify the causal recurrent architecture itself.

---

## Post-Training Objective Evaluation

The post-training recurrent outputs are evaluated against the validated signed transformation targets constructed in Block 7.

For pathway $p$, the masked mean-squared transformation objective remains:

$$
\mathcal{L}_p
=
\frac{
\sum_{a,t,d}
m^{(p,a)}_{t,d}
\left(
\tilde{c}^{(p,a)}_{t,d}
-
\Delta y^{(p,a)}_{t,d}
\right)^2
}{
\sum_{a,t,d}
m^{(p,a)}_{t,d}
},
$$

where:

- $\tilde{c}^{(p,a)}_{t,d}$ is the Transformative-Weight-gated candidate transformation;
- $\Delta y^{(p,a)}_{t,d}$ is the validated signed transformation target;
- $m^{(p,a)}_{t,d}$ is the combined supervision-availability and activation-eligibility mask.

The complete objective remains equally weighted across the three representational pathways:

$$
\mathcal{L}_{\mathrm{total}}
=
\frac{1}{3}
\left(
\mathcal{L}_{\mathrm{factual}}
+
\mathcal{L}_{\mathrm{psychological}}
+
\mathcal{L}_{\mathrm{social}}
\right).
$$

No supervision value is supplied as recurrent model input. Supervision is used exclusively for post-training objective evaluation.

---

## Immutable Baseline Comparison

The principal reference remains the deterministic pre-training baseline established in Block 8:

$$
\mathcal{L}_{\mathrm{baseline}}
=
0.20858338.
$$

Block 9 produced a post-training objective of approximately:

$$
\mathcal{L}_{\mathrm{trained}}
=
0.20655662.
$$

Block 10 must independently reproduce the trained objective from the frozen post-training state.

The objective change is evaluated as:

$$
\Delta\mathcal{L}
=
\mathcal{L}_{\mathrm{post}}
-
\mathcal{L}_{\mathrm{baseline}}.
$$

A negative value indicates improvement relative to the immutable Block 8 baseline.

This comparison is descriptive of controlled optimisation on the existing pilot corpus. It is not interpreted as an estimate of out-of-sample performance.

---

## Deterministic Post-Training Validation

The complete post-training recurrent evaluation is executed repeatedly without parameter modification.

For identical article representations, initial states and learned Transformative Weights, the resulting:

- candidate transformations;
- effective Transformative Weights;
- weighted candidate transformations;
- recurrent state trajectories;
- pathway objectives; and
- total objective

must reproduce deterministically within the numerical precision established by the notebook.

This validation distinguishes a genuine persistent learned model state from an optimisation result dependent on transient runtime state.

---

## Architectural Invariance

The block verifies that training in Block 9 did not alter architectural components outside the intended parameter scope.

The following must remain exactly unchanged relative to the inherited reference state:

- representation backbone;
- global confluent module;
- factual representation head;
- psychological representation head;
- social representation head;
- factual transformative module;
- psychological transformative module; and
- social transformative module.

Only the 49 activation-eligible Transformative-Weight latent parameters may differ from their pre-Block-9 values.

Deferred dimensions must remain fixed at an effective weight of zero.

All model parameters are frozen during Block 10.

---

## Causal and Article-Boundary Audit

The post-training trajectory must continue to satisfy the causal recurrent contract established earlier in the architecture.

The block explicitly validates that:

- canonical sentence order is preserved;
- every article begins from a fresh zero recurrent state;
- no state propagates across article boundaries;
- factual, psychological and social recurrent states remain independent;
- no future sentence representation is accessed;
- supervision is never used as recurrent model input;
- recurrent updates remain additive; and
- deferred dimensions contribute no recurrent increment.

Learning the Transformative Weights must therefore leave the underlying causal architecture unchanged.

---

## Generalisation-Boundary Audit

The current corpus contains three independent articles and 54 canonical sentences.

Although this is sufficient to demonstrate multi-article execution, causal recurrent behaviour and controlled parameter learnability, it is **not sufficient to establish generalisation**.

The Block 9 optimisation and the Block 10 post-training evaluation operate on the same validated pilot corpus.

Consequently, the observed reduction in objective establishes:

> **controlled in-sample architectural learnability under the Media AI recurrent contract**

but does not establish:

- out-of-sample predictive performance;
- robustness across unseen articles;
- cross-topic generalisation;
- cross-source generalisation;
- temporal generalisation;
- population-level statistical validity; or
- production readiness.

These claims require a substantially larger corpus together with explicit train, validation and held-out test partitions.

The distinction is retained as part of the final pilot research record rather than inferred from objective improvement alone.

---

## Execution Boundary

This block performs:

- post-training Transformative-Weight forward execution;
- transformative forward execution;
- article-level recurrent trajectory reconstruction;
- masked pathway objective calculation;
- total objective calculation;
- deterministic repeated evaluation;
- pre-training versus post-training objective comparison;
- causal-contract validation;
- parameter-state validation; and
- generalisation-boundary auditing.

This block does **not** perform:

- optimiser construction;
- gradient clearing;
- backward propagation;
- gradient clipping;
- optimiser stepping;
- parameter updating;
- additional training;
- representation-model training;
- transformative-module training; or
- claims of out-of-sample generalisation.

---

## Expected Outcome

A successful Block 10 should establish that:

1. the Block 9 trained Transformative-Weight state can be evaluated independently without further optimisation;
2. the post-training objective remains below the immutable Block 8 baseline;
3. the factual, psychological and social post-training objectives remain finite and reproducible;
4. the complete recurrent trajectories remain deterministic;
5. article-boundary resets and causal state continuity remain valid;
6. all deferred dimensions remain exactly inactive;
7. the representation and transformative architectures remain exactly unchanged;
8. all 902,239 model parameters are frozen during post-training validation; and
9. the observed improvement is explicitly classified as pilot in-sample optimisation rather than evidence of generalisation.

Successful completion of this block establishes a validated post-training state for the controlled Media AI pilot architecture.

Notebook 10 may then proceed to the final pilot evaluation, architectural completion audit and persistent project handover.

```python
# =============================================================================
# Media AI — Notebook 10
# Block 10: Post-Training Recurrent Validation, Objective Comparison
#           and Generalisation-Boundary Audit
# =============================================================================

from copy import deepcopy
from datetime import datetime, timezone

import math

import torch


# =============================================================================
# Block identity
# =============================================================================

NOTEBOOK_10_BLOCK_10 = 10

NOTEBOOK_10_BLOCK_10_NAME = (
    "Post-Training Recurrent Validation, Objective Comparison "
    "and Generalisation-Boundary Audit"
)

NOTEBOOK_10_BLOCK_10_VERSION = "1.1"

BLOCK_10_EXECUTED_AT_UTC = (
    datetime.now(
        timezone.utc
    ).isoformat()
)


# =============================================================================
# Exact Block 8 / Block 9 handover contract
# =============================================================================
#
# IMPORTANT:
# Block 10 consumes the exact stable globals exported by the validated
# Block 8 and Block 9 implementations. It does not search for speculative
# aliases and does not reconstruct earlier notebook stages.
# =============================================================================

BLOCK_10_REQUIRED_GLOBALS = [
    "NOTEBOOK_10_BLOCK_8_COMPLETE",
    "NOTEBOOK_10_BLOCK_8_VALID",
    "NOTEBOOK_10_PRETRAIN_BASELINE_VALID",
    "NOTEBOOK_10_PRETRAIN_PATHWAY_LOSSES",
    "NOTEBOOK_10_PRETRAIN_TOTAL_OBJECTIVE",
    "NOTEBOOK_10_PRETRAIN_OBJECTIVE_ACTIVE_ELEMENTS",
    "NOTEBOOK_10_BLOCK_9_COMPLETE",
    "NOTEBOOK_10_BLOCK_9_VALID",
    "NOTEBOOK_10_CONTROLLED_CONTINUED_PILOT_TRAINING_VALID",
    "NOTEBOOK_10_POSTTRAIN_VALIDATION_READY",
    "NOTEBOOK_10_POSTTRAIN_PATHWAY_LOSSES",
    "NOTEBOOK_10_POSTTRAIN_TOTAL_OBJECTIVE",
    "NOTEBOOK_10_POSTTRAIN_OBJECTIVE_DELTA",
    "NOTEBOOK_10_TRAINED_TRANSFORMATIVE_WEIGHT_STATE",
    "NOTEBOOK_10_TRAINED_EFFECTIVE_TRANSFORMATIVE_WEIGHTS",
    "NOTEBOOK_10_MULTI_ARTICLE_PATHWAY_REPRESENTATIONS",
    "NOTEBOOK_10_TRANSFORMATION_TARGETS",
    "NOTEBOOK_10_TRANSFORMATION_OBJECTIVE_MASKS",
    "NOTEBOOK_10_SUPERVISION_CANONICAL_ARTICLE_IDS",
    "NOTEBOOK_10_SUPERVISION_CANONICAL_SENTENCE_IDS",
    "NOTEBOOK_10_SUPERVISION_ARTICLE_RECORD_INDICES",
    "NOTEBOOK_10_REPRESENTATION_MODULES",
    "NOTEBOOK_10_TRANSFORMATIVE_MODULES",
    "NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES",
    "NOTEBOOK_10_PATHWAY_DIMENSIONS",
    "NOTEBOOK_10_ACTIVE_DIMENSION_INDICES",
    "NOTEBOOK_10_DEFERRED_DIMENSION_INDICES",
]


BLOCK_10_MISSING_PREREQUISITES = [
    object_name

    for object_name
    in BLOCK_10_REQUIRED_GLOBALS

    if object_name not in globals()
]


if BLOCK_10_MISSING_PREREQUISITES:

    raise RuntimeError(
        "Notebook 10 Block 10 cannot execute because the exact validated "
        "Block 8 → Block 9 handover is incomplete in the current runtime. "
        "If the runtime was restarted, rerun Notebook 10 Blocks 1–9 in "
        "order. Otherwise inspect the missing names below. "
        f"Missing objects: {BLOCK_10_MISSING_PREREQUISITES}"
    )


BLOCK_10_PREREQUISITES_VALID = all(
    [
        NOTEBOOK_10_BLOCK_8_COMPLETE is True,
        NOTEBOOK_10_BLOCK_8_VALID is True,
        NOTEBOOK_10_PRETRAIN_BASELINE_VALID is True,
        NOTEBOOK_10_BLOCK_9_COMPLETE is True,
        NOTEBOOK_10_BLOCK_9_VALID is True,
        NOTEBOOK_10_CONTROLLED_CONTINUED_PILOT_TRAINING_VALID is True,
        NOTEBOOK_10_POSTTRAIN_VALIDATION_READY is True,
        isinstance(
            NOTEBOOK_10_PRETRAIN_PATHWAY_LOSSES,
            dict,
        ),
        isinstance(
            NOTEBOOK_10_POSTTRAIN_PATHWAY_LOSSES,
            dict,
        ),
        isinstance(
            NOTEBOOK_10_TRAINED_TRANSFORMATIVE_WEIGHT_STATE,
            dict,
        ),
        isinstance(
            NOTEBOOK_10_TRAINED_EFFECTIVE_TRANSFORMATIVE_WEIGHTS,
            dict,
        ),
        isinstance(
            NOTEBOOK_10_MULTI_ARTICLE_PATHWAY_REPRESENTATIONS,
            dict,
        ),
        isinstance(
            NOTEBOOK_10_TRANSFORMATION_TARGETS,
            dict,
        ),
        isinstance(
            NOTEBOOK_10_TRANSFORMATION_OBJECTIVE_MASKS,
            dict,
        ),
    ]
)


if not BLOCK_10_PREREQUISITES_VALID:

    raise RuntimeError(
        "Notebook 10 Block 10 requires valid completed Blocks 8 and 9."
    )


# =============================================================================
# Canonical architecture
# =============================================================================

BLOCK_10_PATHWAY_DIMENSIONS = {
    pathway_name:
        int(
            pathway_dim
        )

    for pathway_name, pathway_dim
    in NOTEBOOK_10_PATHWAY_DIMENSIONS.items()
}


BLOCK_10_EXPECTED_PATHWAY_DIMENSIONS = {
    "factual":
        10,

    "psychological":
        34,

    "social":
        7,
}


BLOCK_10_PATHWAY_DIMENSIONS_VALID = (
    BLOCK_10_PATHWAY_DIMENSIONS
    ==
    BLOCK_10_EXPECTED_PATHWAY_DIMENSIONS
)


if not BLOCK_10_PATHWAY_DIMENSIONS_VALID:

    raise RuntimeError(
        "Notebook 10 Block 10 pathway dimensions do not reproduce the "
        "validated Notebook 09 → 10 architecture."
    )


BLOCK_10_PATHWAYS = tuple(
    BLOCK_10_PATHWAY_DIMENSIONS.keys()
)


BLOCK_10_ACTIVE_DIMENSION_INDICES = {
    pathway_name:
        tuple(
            int(
                index
            )

            for index
            in NOTEBOOK_10_ACTIVE_DIMENSION_INDICES[
                pathway_name
            ]
        )

    for pathway_name
    in BLOCK_10_PATHWAYS
}


BLOCK_10_DEFERRED_DIMENSION_INDICES = {
    pathway_name:
        tuple(
            int(
                index
            )

            for index
            in NOTEBOOK_10_DEFERRED_DIMENSION_INDICES[
                pathway_name
            ]
        )

    for pathway_name
    in BLOCK_10_PATHWAYS
}


BLOCK_10_DIMENSION_PARTITIONS_VALID = all(
    (
        set(
            BLOCK_10_ACTIVE_DIMENSION_INDICES[
                pathway_name
            ]
        ).isdisjoint(
            BLOCK_10_DEFERRED_DIMENSION_INDICES[
                pathway_name
            ]
        )
        and
        (
            set(
                BLOCK_10_ACTIVE_DIMENSION_INDICES[
                    pathway_name
                ]
            )
            |
            set(
                BLOCK_10_DEFERRED_DIMENSION_INDICES[
                    pathway_name
                ]
            )
            ==
            set(
                range(
                    pathway_dim
                )
            )
        )
    )

    for pathway_name, pathway_dim
    in BLOCK_10_PATHWAY_DIMENSIONS.items()
)


if not BLOCK_10_DIMENSION_PARTITIONS_VALID:

    raise RuntimeError(
        "Notebook 10 Block 10 active/deferred dimensional contract "
        "is invalid."
    )


# =============================================================================
# Canonical corpus identity
# =============================================================================

BLOCK_10_ARTICLE_IDS = tuple(
    NOTEBOOK_10_SUPERVISION_CANONICAL_ARTICLE_IDS
)


BLOCK_10_SENTENCE_IDS = tuple(
    NOTEBOOK_10_SUPERVISION_CANONICAL_SENTENCE_IDS
)


BLOCK_10_ARTICLE_RECORD_INDICES = deepcopy(
    NOTEBOOK_10_SUPERVISION_ARTICLE_RECORD_INDICES
)


BLOCK_10_ARTICLE_COUNT = len(
    BLOCK_10_ARTICLE_IDS
)


BLOCK_10_SENTENCE_COUNT = len(
    BLOCK_10_SENTENCE_IDS
)


BLOCK_10_CANONICAL_GLOBAL_INDICES = tuple(
    global_index

    for article_id
    in BLOCK_10_ARTICLE_IDS

    for global_index
    in BLOCK_10_ARTICLE_RECORD_INDICES[
        article_id
    ]
)


BLOCK_10_CANONICAL_CORPUS_VALID = all(
    [
        BLOCK_10_ARTICLE_COUNT > 0,
        BLOCK_10_SENTENCE_COUNT > 0,
        len(
            set(
                BLOCK_10_SENTENCE_IDS
            )
        )
        ==
        BLOCK_10_SENTENCE_COUNT,
        set(
            BLOCK_10_ARTICLE_RECORD_INDICES.keys()
        )
        ==
        set(
            BLOCK_10_ARTICLE_IDS
        ),
        BLOCK_10_CANONICAL_GLOBAL_INDICES
        ==
        tuple(
            range(
                BLOCK_10_SENTENCE_COUNT
            )
        ),
    ]
)


if not BLOCK_10_CANONICAL_CORPUS_VALID:

    raise RuntimeError(
        "Notebook 10 Block 10 canonical article/sentence contract is invalid."
    )


# =============================================================================
# Resolve model device and dtype
# =============================================================================

BLOCK_10_REFERENCE_PARAMETERS = [
    parameter

    for module
    in NOTEBOOK_10_REPRESENTATION_MODULES.values()

    for parameter
    in module.parameters()
]


if not BLOCK_10_REFERENCE_PARAMETERS:

    raise RuntimeError(
        "Notebook 10 Block 10 could not resolve execution device/dtype."
    )


BLOCK_10_EXECUTION_DEVICE = (
    BLOCK_10_REFERENCE_PARAMETERS[
        0
    ].device
)


BLOCK_10_EXECUTION_DTYPE = (
    BLOCK_10_REFERENCE_PARAMETERS[
        0
    ].dtype
)


BLOCK_10_DEVICE_DTYPE_VALID = all(
    parameter.device
    ==
    BLOCK_10_EXECUTION_DEVICE
    and
    parameter.dtype
    ==
    BLOCK_10_EXECUTION_DTYPE

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


if not BLOCK_10_DEVICE_DTYPE_VALID:

    raise RuntimeError(
        "Notebook 10 Block 10 inherited modules do not share one "
        "execution device/dtype."
    )


# =============================================================================
# Execution boundary
# =============================================================================

NOTEBOOK_10_BLOCK_10_REPRESENTATION_FORWARD_EXECUTED = False

NOTEBOOK_10_BLOCK_10_TRANSFORMATIVE_FORWARD_EXECUTED = False

NOTEBOOK_10_BLOCK_10_WEIGHT_FORWARD_EXECUTED = False

NOTEBOOK_10_BLOCK_10_RECURRENT_STATE_CREATED = False

NOTEBOOK_10_BLOCK_10_RECURRENT_UPDATE_EXECUTED = False

NOTEBOOK_10_BLOCK_10_LOSS_CALCULATED = False

NOTEBOOK_10_BLOCK_10_REPEAT_EVALUATION_EXECUTED = False

NOTEBOOK_10_BLOCK_10_OPTIMIZER_CREATED = False

NOTEBOOK_10_BLOCK_10_ZERO_GRAD_EXECUTED = False

NOTEBOOK_10_BLOCK_10_BACKWARD_PASS_EXECUTED = False

NOTEBOOK_10_BLOCK_10_GRADIENT_CLIPPING_EXECUTED = False

NOTEBOOK_10_BLOCK_10_OPTIMIZER_STEP_EXECUTED = False

NOTEBOOK_10_BLOCK_10_PARAMETER_UPDATE_EXECUTED = False

NOTEBOOK_10_BLOCK_10_TRAINING_EXECUTED = False

NOTEBOOK_10_BLOCK_10_FUTURE_CONTEXT_USED = False

NOTEBOOK_10_BLOCK_10_SUPERVISION_USED_AS_MODEL_INPUT = False

NOTEBOOK_10_BLOCK_10_CROSS_ARTICLE_PROPAGATION = False

NOTEBOOK_10_BLOCK_10_CROSS_PATHWAY_RECURRENCE = False


# =============================================================================
# Exact-state helpers
# =============================================================================

def block_10_snapshot_module_collection(
    module_collection,
):

    return {
        module_name:
            {
                tensor_name:
                    tensor.detach()
                    .cpu()
                    .clone()

                for tensor_name, tensor
                in module.state_dict().items()
            }

        for module_name, module
        in module_collection.items()
    }


def block_10_nested_state_exact(
    state_before,
    state_after,
):

    if state_before.keys() != state_after.keys():

        return False

    for module_name in state_before:

        if (
            state_before[
                module_name
            ].keys()
            !=
            state_after[
                module_name
            ].keys()
        ):

            return False

        for tensor_name in state_before[
            module_name
        ]:

            if not torch.equal(
                state_before[
                    module_name
                ][
                    tensor_name
                ],
                state_after[
                    module_name
                ][
                    tensor_name
                ],
            ):

                return False

    return True


# =============================================================================
# Validate current trained weight state against Block 9 handover
# =============================================================================

BLOCK_10_WEIGHT_STATE_AT_ENTRY = (
    block_10_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES
    )
)


BLOCK_10_TRAINED_WEIGHT_STATE_EXACT = (
    block_10_nested_state_exact(
        NOTEBOOK_10_TRAINED_TRANSFORMATIVE_WEIGHT_STATE,
        BLOCK_10_WEIGHT_STATE_AT_ENTRY,
    )
)


if not BLOCK_10_TRAINED_WEIGHT_STATE_EXACT:

    raise RuntimeError(
        "Notebook 10 Block 10 current Transformative-Weight state does "
        "not reproduce the validated trained Block 9 handover exactly."
    )


# =============================================================================
# Freeze complete post-training architecture
# =============================================================================
#
# Block 9 leaves the 49 Transformative-Weight parameters trainable by design.
# Block 10 is a validation stage, so all parameters are frozen here before any
# post-training recurrent evaluation is performed.
# =============================================================================

for module in NOTEBOOK_10_REPRESENTATION_MODULES.values():

    module.eval()

    for parameter in module.parameters():

        parameter.requires_grad_(
            False
        )

        parameter.grad = None


for module in NOTEBOOK_10_TRANSFORMATIVE_MODULES.values():

    module.eval()

    for parameter in module.parameters():

        parameter.requires_grad_(
            False
        )

        parameter.grad = None


for module in NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES.values():

    module.eval()

    for parameter in module.parameters():

        parameter.requires_grad_(
            False
        )

        parameter.grad = None


# =============================================================================
# Post-freeze immutable state
# =============================================================================

BLOCK_10_REPRESENTATION_STATE_BEFORE = (
    block_10_snapshot_module_collection(
        NOTEBOOK_10_REPRESENTATION_MODULES
    )
)


BLOCK_10_TRANSFORMATIVE_STATE_BEFORE = (
    block_10_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_MODULES
    )
)


BLOCK_10_WEIGHT_STATE_BEFORE = (
    block_10_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES
    )
)


BLOCK_10_WEIGHT_STATE_PRESERVED_BY_FREEZE = (
    block_10_nested_state_exact(
        BLOCK_10_WEIGHT_STATE_AT_ENTRY,
        BLOCK_10_WEIGHT_STATE_BEFORE,
    )
)


if not BLOCK_10_WEIGHT_STATE_PRESERVED_BY_FREEZE:

    raise RuntimeError(
        "Notebook 10 Block 10 freezing changed the trained "
        "Transformative-Weight state unexpectedly."
    )


# =============================================================================
# Parameter accounting and freeze-state validation
# =============================================================================

BLOCK_10_REPRESENTATION_PARAMETER_COUNT = sum(
    parameter.numel()

    for module
    in NOTEBOOK_10_REPRESENTATION_MODULES.values()

    for parameter
    in module.parameters()
)


BLOCK_10_TRANSFORMATIVE_PARAMETER_COUNT = sum(
    parameter.numel()

    for module
    in NOTEBOOK_10_TRANSFORMATIVE_MODULES.values()

    for parameter
    in module.parameters()
)


BLOCK_10_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT = sum(
    parameter.numel()

    for module
    in NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES.values()

    for parameter
    in module.parameters()
)


BLOCK_10_TOTAL_PARAMETER_COUNT = (
    BLOCK_10_REPRESENTATION_PARAMETER_COUNT
    +
    BLOCK_10_TRANSFORMATIVE_PARAMETER_COUNT
    +
    BLOCK_10_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT
)


BLOCK_10_PARAMETER_ACCOUNTING_VALID = all(
    [
        BLOCK_10_REPRESENTATION_PARAMETER_COUNT
        ==
        894_003,

        BLOCK_10_TRANSFORMATIVE_PARAMETER_COUNT
        ==
        8_187,

        BLOCK_10_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT
        ==
        49,

        BLOCK_10_TOTAL_PARAMETER_COUNT
        ==
        902_239,
    ]
)


BLOCK_10_ALL_PARAMETERS_FROZEN_BEFORE = all(
    not parameter.requires_grad

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


BLOCK_10_ALL_MODULES_EVAL_BEFORE = all(
    not module.training

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()
)


BLOCK_10_ALL_GRADIENTS_ABSENT_BEFORE = all(
    parameter.grad is None

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


if not all(
    [
        BLOCK_10_PARAMETER_ACCOUNTING_VALID,
        BLOCK_10_ALL_PARAMETERS_FROZEN_BEFORE,
        BLOCK_10_ALL_MODULES_EVAL_BEFORE,
        BLOCK_10_ALL_GRADIENTS_ABSENT_BEFORE,
    ]
):

    raise RuntimeError(
        "Notebook 10 Block 10 post-training architecture is not in the "
        "required frozen/evaluation/no-gradient state."
    )


# =============================================================================
# Fixed pathway representations
# =============================================================================
#
# The validated runtime contract is article -> pathway -> matrix.
# This is the same structure consumed successfully by Block 9.
# =============================================================================

BLOCK_10_PATHWAY_REPRESENTATIONS = deepcopy(
    NOTEBOOK_10_MULTI_ARTICLE_PATHWAY_REPRESENTATIONS
)


BLOCK_10_PATHWAY_REPRESENTATIONS_VALID = True


for article_id in BLOCK_10_ARTICLE_IDS:

    if article_id not in BLOCK_10_PATHWAY_REPRESENTATIONS:

        BLOCK_10_PATHWAY_REPRESENTATIONS_VALID = False

        continue

    article_indices = (
        BLOCK_10_ARTICLE_RECORD_INDICES[
            article_id
        ]
    )

    article_length = len(
        article_indices
    )

    for pathway_name, pathway_dim in BLOCK_10_PATHWAY_DIMENSIONS.items():

        matrix = (
            BLOCK_10_PATHWAY_REPRESENTATIONS[
                article_id
            ].get(
                pathway_name
            )
        )

        if not torch.is_tensor(
            matrix
        ):

            BLOCK_10_PATHWAY_REPRESENTATIONS_VALID = False

            continue

        if (
            tuple(
                matrix.shape
            )
            !=
            (
                article_length,
                pathway_dim,
            )
        ):

            BLOCK_10_PATHWAY_REPRESENTATIONS_VALID = False

        if not bool(
            torch.isfinite(
                matrix
            ).all()
            .item()
        ):

            BLOCK_10_PATHWAY_REPRESENTATIONS_VALID = False


if not BLOCK_10_PATHWAY_REPRESENTATIONS_VALID:

    raise RuntimeError(
        "Notebook 10 Block 10 inherited article/pathway representations "
        "are invalid."
    )


# =============================================================================
# Fixed transformation targets and objective masks
# =============================================================================

BLOCK_10_TARGETS = {
    pathway_name:
        NOTEBOOK_10_TRANSFORMATION_TARGETS[
            pathway_name
        ]
        .detach()
        .to(
            device=
                BLOCK_10_EXECUTION_DEVICE,
            dtype=
                BLOCK_10_EXECUTION_DTYPE,
        )
        .clone()

    for pathway_name
    in BLOCK_10_PATHWAYS
}


BLOCK_10_OBJECTIVE_MASKS = {
    pathway_name:
        NOTEBOOK_10_TRANSFORMATION_OBJECTIVE_MASKS[
            pathway_name
        ]
        .detach()
        .to(
            device=
                BLOCK_10_EXECUTION_DEVICE,
            dtype=
                torch.bool,
        )
        .clone()

    for pathway_name
    in BLOCK_10_PATHWAYS
}


BLOCK_10_TARGET_MASK_CONTRACT_VALID = True


for pathway_name, pathway_dim in BLOCK_10_PATHWAY_DIMENSIONS.items():

    expected_shape = (
        BLOCK_10_SENTENCE_COUNT,
        pathway_dim,
    )

    target = (
        BLOCK_10_TARGETS[
            pathway_name
        ]
    )

    mask = (
        BLOCK_10_OBJECTIVE_MASKS[
            pathway_name
        ]
    )

    if tuple(
        target.shape
    ) != expected_shape:

        BLOCK_10_TARGET_MASK_CONTRACT_VALID = False

    if tuple(
        mask.shape
    ) != expected_shape:

        BLOCK_10_TARGET_MASK_CONTRACT_VALID = False

    if not bool(
        mask.any()
        .item()
    ):

        BLOCK_10_TARGET_MASK_CONTRACT_VALID = False

    if not bool(
        torch.isfinite(
            target[
                mask
            ]
        ).all()
        .item()
    ):

        BLOCK_10_TARGET_MASK_CONTRACT_VALID = False

    deferred_indices = (
        BLOCK_10_DEFERRED_DIMENSION_INDICES[
            pathway_name
        ]
    )

    if deferred_indices:

        if bool(
            mask[
                :,
                list(
                    deferred_indices
                )
            ].any()
            .item()
        ):

            BLOCK_10_TARGET_MASK_CONTRACT_VALID = False


if not BLOCK_10_TARGET_MASK_CONTRACT_VALID:

    raise RuntimeError(
        "Notebook 10 Block 10 transformation-target/objective-mask "
        "contract is invalid."
    )


# =============================================================================
# Objective-active counts
# =============================================================================

BLOCK_10_OBJECTIVE_ACTIVE_ELEMENTS = {
    pathway_name:
        int(
            BLOCK_10_OBJECTIVE_MASKS[
                pathway_name
            ].sum()
            .item()
        )

    for pathway_name
    in BLOCK_10_PATHWAYS
}


BLOCK_10_OBJECTIVE_COUNTS_MATCH_BLOCK_8 = all(
    BLOCK_10_OBJECTIVE_ACTIVE_ELEMENTS[
        pathway_name
    ]
    ==
    int(
        NOTEBOOK_10_PRETRAIN_OBJECTIVE_ACTIVE_ELEMENTS[
            pathway_name
        ]
    )

    for pathway_name
    in BLOCK_10_PATHWAYS
)


if not BLOCK_10_OBJECTIVE_COUNTS_MATCH_BLOCK_8:

    raise RuntimeError(
        "Notebook 10 Block 10 objective-active counts do not reproduce "
        "the immutable Block 8 contract."
    )


# =============================================================================
# Trained effective Transformative-Weight identity
# =============================================================================

with torch.inference_mode():

    BLOCK_10_EFFECTIVE_WEIGHTS_AT_ENTRY = {
        pathway_name:
            NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES[
                pathway_name
            ]()
            .detach()
            .cpu()
            .clone()

        for pathway_name
        in BLOCK_10_PATHWAYS
    }


NOTEBOOK_10_BLOCK_10_WEIGHT_FORWARD_EXECUTED = True


BLOCK_10_TRAINED_EFFECTIVE_WEIGHT_IDENTITY = all(
    torch.equal(
        BLOCK_10_EFFECTIVE_WEIGHTS_AT_ENTRY[
            pathway_name
        ],
        NOTEBOOK_10_TRAINED_EFFECTIVE_TRANSFORMATIVE_WEIGHTS[
            pathway_name
        ]
        .detach()
        .cpu(),
    )

    for pathway_name
    in BLOCK_10_PATHWAYS
)


if not BLOCK_10_TRAINED_EFFECTIVE_WEIGHT_IDENTITY:

    raise RuntimeError(
        "Notebook 10 Block 10 effective Transformative Weights do not "
        "reproduce the trained Block 9 handover exactly."
    )


# =============================================================================
# Masked MSE helper
# =============================================================================

def block_10_masked_mean_squared_error(
    prediction,
    target,
    mask,
):

    if prediction.shape != target.shape:

        raise RuntimeError(
            "Notebook 10 Block 10 masked-MSE prediction/target "
            "shape mismatch."
        )

    if mask.shape != target.shape:

        raise RuntimeError(
            "Notebook 10 Block 10 masked-MSE mask/target shape mismatch."
        )

    if not bool(
        mask.any()
        .item()
    ):

        raise RuntimeError(
            "Notebook 10 Block 10 masked-MSE received an empty mask."
        )

    squared_error = (
        prediction
        -
        target
    ).pow(
        2
    )

    return squared_error[
        mask
    ].mean()


# =============================================================================
# Complete post-training recurrent evaluation
# =============================================================================

def block_10_evaluate_posttraining(
    evaluation_label,
):

    weighted_predictions = {
        pathway_name:
            torch.zeros(
                (
                    BLOCK_10_SENTENCE_COUNT,
                    pathway_dim,
                ),
                dtype=
                    BLOCK_10_EXECUTION_DTYPE,
                device=
                    BLOCK_10_EXECUTION_DEVICE,
            )

        for pathway_name, pathway_dim
        in BLOCK_10_PATHWAY_DIMENSIONS.items()
    }


    article_trajectories = {
        article_id:
            {}

        for article_id
        in BLOCK_10_ARTICLE_IDS
    }


    effective_weights = {}


    with torch.inference_mode():

        for pathway_name, pathway_dim in BLOCK_10_PATHWAY_DIMENSIONS.items():

            effective_weight = (
                NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES[
                    pathway_name
                ]()
                .detach()
            )

            if (
                tuple(
                    effective_weight.shape
                )
                !=
                (
                    pathway_dim,
                )
            ):

                raise RuntimeError(
                    f"Notebook 10 Block 10 Transformative-Weight output "
                    f"shape invalid for {pathway_name}."
                )

            if not bool(
                torch.isfinite(
                    effective_weight
                ).all()
                .item()
            ):

                raise RuntimeError(
                    f"Notebook 10 Block 10 Transformative-Weight output "
                    f"is non-finite for {pathway_name}."
                )

            effective_weights[
                pathway_name
            ] = (
                effective_weight.detach()
                .cpu()
                .clone()
            )


        for article_id in BLOCK_10_ARTICLE_IDS:

            article_indices = (
                BLOCK_10_ARTICLE_RECORD_INDICES[
                    article_id
                ]
            )

            if not article_indices:

                raise RuntimeError(
                    f"Notebook 10 Block 10 article {article_id!r} "
                    "contains no canonical sentence indices."
                )


            for pathway_name, pathway_dim in BLOCK_10_PATHWAY_DIMENSIONS.items():

                representation_matrix = (
                    BLOCK_10_PATHWAY_REPRESENTATIONS[
                        article_id
                    ][
                        pathway_name
                    ]
                    .to(
                        device=
                            BLOCK_10_EXECUTION_DEVICE,
                        dtype=
                            BLOCK_10_EXECUTION_DTYPE,
                    )
                )


                effective_weight = (
                    effective_weights[
                        pathway_name
                    ]
                    .to(
                        device=
                            BLOCK_10_EXECUTION_DEVICE,
                        dtype=
                            BLOCK_10_EXECUTION_DTYPE,
                    )
                    .reshape(
                        1,
                        pathway_dim,
                    )
                )


                recurrent_state = torch.zeros(
                    (
                        1,
                        pathway_dim,
                    ),
                    dtype=
                        BLOCK_10_EXECUTION_DTYPE,
                    device=
                        BLOCK_10_EXECUTION_DEVICE,
                )


                initial_state = (
                    recurrent_state.detach()
                    .cpu()
                    .clone()
                )


                cumulative_weighted_candidate = torch.zeros_like(
                    recurrent_state
                )


                steps = []


                for local_position, global_index in enumerate(
                    article_indices
                ):

                    current_representation = (
                        representation_matrix[
                            local_position:
                            local_position
                            +
                            1
                        ]
                    )


                    preceding_state = recurrent_state


                    candidate = (
                        NOTEBOOK_10_TRANSFORMATIVE_MODULES[
                            pathway_name
                        ](
                            current_representation,
                            preceding_state,
                        )
                    )


                    weighted_candidate = (
                        candidate
                        *
                        effective_weight
                    )


                    next_state = (
                        preceding_state
                        +
                        weighted_candidate
                    )


                    cumulative_weighted_candidate = (
                        cumulative_weighted_candidate
                        +
                        weighted_candidate
                    )


                    weighted_predictions[
                        pathway_name
                    ][
                        global_index:
                        global_index
                        +
                        1
                    ] = weighted_candidate


                    deferred_indices = (
                        BLOCK_10_DEFERRED_DIMENSION_INDICES[
                            pathway_name
                        ]
                    )


                    deferred_weight_zero = True
                    deferred_contribution_zero = True
                    deferred_state_zero = True


                    if deferred_indices:

                        deferred_index_tensor = torch.tensor(
                            deferred_indices,
                            dtype=
                                torch.long,
                            device=
                                BLOCK_10_EXECUTION_DEVICE,
                        )


                        deferred_weight_zero = bool(
                            torch.equal(
                                effective_weight.index_select(
                                    1,
                                    deferred_index_tensor,
                                ),
                                torch.zeros(
                                    (
                                        1,
                                        len(
                                            deferred_indices
                                        ),
                                    ),
                                    dtype=
                                        BLOCK_10_EXECUTION_DTYPE,
                                    device=
                                        BLOCK_10_EXECUTION_DEVICE,
                                ),
                            )
                        )


                        deferred_contribution_zero = bool(
                            torch.equal(
                                weighted_candidate.index_select(
                                    1,
                                    deferred_index_tensor,
                                ),
                                torch.zeros(
                                    (
                                        1,
                                        len(
                                            deferred_indices
                                        ),
                                    ),
                                    dtype=
                                        BLOCK_10_EXECUTION_DTYPE,
                                    device=
                                        BLOCK_10_EXECUTION_DEVICE,
                                ),
                            )
                        )


                        deferred_state_zero = bool(
                            torch.equal(
                                next_state.index_select(
                                    1,
                                    deferred_index_tensor,
                                ),
                                torch.zeros(
                                    (
                                        1,
                                        len(
                                            deferred_indices
                                        ),
                                    ),
                                    dtype=
                                        BLOCK_10_EXECUTION_DTYPE,
                                    device=
                                        BLOCK_10_EXECUTION_DEVICE,
                                ),
                            )
                        )


                    step = {
                        "global_index":
                            int(
                                global_index
                            ),

                        "current_representation":
                            current_representation.detach()
                            .cpu()
                            .clone(),

                        "preceding_state":
                            preceding_state.detach()
                            .cpu()
                            .clone(),

                        "candidate":
                            candidate.detach()
                            .cpu()
                            .clone(),

                        "effective_weight":
                            effective_weight.detach()
                            .cpu()
                            .clone(),

                        "weighted_candidate":
                            weighted_candidate.detach()
                            .cpu()
                            .clone(),

                        "recurrent_state":
                            next_state.detach()
                            .cpu()
                            .clone(),

                        "candidate_finite":
                            bool(
                                torch.isfinite(
                                    candidate
                                ).all()
                                .item()
                            ),

                        "weighted_finite":
                            bool(
                                torch.isfinite(
                                    weighted_candidate
                                ).all()
                                .item()
                            ),

                        "state_finite":
                            bool(
                                torch.isfinite(
                                    next_state
                                ).all()
                                .item()
                            ),

                        "additive_identity":
                            bool(
                                torch.equal(
                                    next_state,
                                    preceding_state
                                    +
                                    weighted_candidate,
                                )
                            ),

                        "cumulative_identity":
                            bool(
                                torch.equal(
                                    next_state,
                                    initial_state.to(
                                        device=
                                            BLOCK_10_EXECUTION_DEVICE,
                                        dtype=
                                            BLOCK_10_EXECUTION_DTYPE,
                                    )
                                    +
                                    cumulative_weighted_candidate,
                                )
                            ),

                        "deferred_weight_zero":
                            deferred_weight_zero,

                        "deferred_contribution_zero":
                            deferred_contribution_zero,

                        "deferred_state_zero":
                            deferred_state_zero,
                    }


                    steps.append(
                        step
                    )


                    recurrent_state = next_state


                article_trajectories[
                    article_id
                ][
                    pathway_name
                ] = {
                    "initial_state":
                        initial_state,

                    "steps":
                        steps,

                    "final_state":
                        recurrent_state.detach()
                        .cpu()
                        .clone(),

                    "cumulative_weighted_candidate":
                        cumulative_weighted_candidate.detach()
                        .cpu()
                        .clone(),
                }


    pathway_losses = {}


    for pathway_name in BLOCK_10_PATHWAYS:

        prediction = (
            weighted_predictions[
                pathway_name
            ]
            .detach()
            .cpu()
        )


        target = (
            BLOCK_10_TARGETS[
                pathway_name
            ]
            .detach()
            .cpu()
            .to(
                dtype=
                    prediction.dtype,
            )
        )


        mask = (
            BLOCK_10_OBJECTIVE_MASKS[
                pathway_name
            ]
            .detach()
            .cpu()
        )


        pathway_losses[
            pathway_name
        ] = float(
            block_10_masked_mean_squared_error(
                prediction=
                    prediction,
                target=
                    target,
                mask=
                    mask,
            )
            .item()
        )


    total_objective = float(
        sum(
            pathway_losses.values()
        )
        /
        len(
            pathway_losses
        )
    )


    return {
        "evaluation_label":
            evaluation_label,

        "weighted_predictions":
            {
                pathway_name:
                    prediction.detach()
                    .cpu()
                    .clone()

                for pathway_name, prediction
                in weighted_predictions.items()
            },

        "effective_weights":
            effective_weights,

        "article_trajectories":
            article_trajectories,

        "pathway_losses":
            pathway_losses,

        "total_objective":
            total_objective,
    }


# =============================================================================
# Execute post-training validation twice
# =============================================================================

BLOCK_10_EVALUATION_A = (
    block_10_evaluate_posttraining(
        "posttraining_a"
    )
)


NOTEBOOK_10_BLOCK_10_TRANSFORMATIVE_FORWARD_EXECUTED = True
NOTEBOOK_10_BLOCK_10_RECURRENT_STATE_CREATED = True
NOTEBOOK_10_BLOCK_10_RECURRENT_UPDATE_EXECUTED = True
NOTEBOOK_10_BLOCK_10_LOSS_CALCULATED = True


BLOCK_10_EVALUATION_B = (
    block_10_evaluate_posttraining(
        "posttraining_b"
    )
)


NOTEBOOK_10_BLOCK_10_REPEAT_EVALUATION_EXECUTED = True


# =============================================================================
# Deterministic repeated-evaluation validation
# =============================================================================

def block_10_evaluations_exact(
    evaluation_a,
    evaluation_b,
):

    if (
        evaluation_a[
            "pathway_losses"
        ]
        !=
        evaluation_b[
            "pathway_losses"
        ]
    ):

        return False


    if (
        evaluation_a[
            "total_objective"
        ]
        !=
        evaluation_b[
            "total_objective"
        ]
    ):

        return False


    for pathway_name in BLOCK_10_PATHWAYS:

        if not torch.equal(
            evaluation_a[
                "weighted_predictions"
            ][
                pathway_name
            ],
            evaluation_b[
                "weighted_predictions"
            ][
                pathway_name
            ],
        ):

            return False


        if not torch.equal(
            evaluation_a[
                "effective_weights"
            ][
                pathway_name
            ],
            evaluation_b[
                "effective_weights"
            ][
                pathway_name
            ],
        ):

            return False


    for article_id in BLOCK_10_ARTICLE_IDS:

        for pathway_name in BLOCK_10_PATHWAYS:

            trajectory_a = (
                evaluation_a[
                    "article_trajectories"
                ][
                    article_id
                ][
                    pathway_name
                ]
            )


            trajectory_b = (
                evaluation_b[
                    "article_trajectories"
                ][
                    article_id
                ][
                    pathway_name
                ]
            )


            if not torch.equal(
                trajectory_a[
                    "initial_state"
                ],
                trajectory_b[
                    "initial_state"
                ],
            ):

                return False


            if not torch.equal(
                trajectory_a[
                    "final_state"
                ],
                trajectory_b[
                    "final_state"
                ],
            ):

                return False


            if (
                len(
                    trajectory_a[
                        "steps"
                    ]
                )
                !=
                len(
                    trajectory_b[
                        "steps"
                    ]
                )
            ):

                return False


            for step_a, step_b in zip(
                trajectory_a[
                    "steps"
                ],
                trajectory_b[
                    "steps"
                ],
            ):

                if (
                    step_a[
                        "global_index"
                    ]
                    !=
                    step_b[
                        "global_index"
                    ]
                ):

                    return False


                for tensor_name in (
                    "current_representation",
                    "preceding_state",
                    "candidate",
                    "effective_weight",
                    "weighted_candidate",
                    "recurrent_state",
                ):

                    if not torch.equal(
                        step_a[
                            tensor_name
                        ],
                        step_b[
                            tensor_name
                        ],
                    ):

                        return False


    return True


BLOCK_10_COMPLETE_EXECUTION_DETERMINISTIC = (
    block_10_evaluations_exact(
        BLOCK_10_EVALUATION_A,
        BLOCK_10_EVALUATION_B,
    )
)


# =============================================================================
# Recurrent causal validation
# =============================================================================

BLOCK_10_FRESH_ZERO_INITIAL_STATES = True

BLOCK_10_STATE_CHAIN_VALID = True

BLOCK_10_ADDITIVE_UPDATE_VALID = True

BLOCK_10_CUMULATIVE_IDENTITY_VALID = True

BLOCK_10_DEFERRED_WEIGHT_ZERO = True

BLOCK_10_DEFERRED_CONTRIBUTION_ZERO = True

BLOCK_10_DEFERRED_STATE_ZERO = True

BLOCK_10_RECURRENT_VALUES_FINITE = True


for article_id in BLOCK_10_ARTICLE_IDS:

    for pathway_name in BLOCK_10_PATHWAYS:

        trajectory = (
            BLOCK_10_EVALUATION_A[
                "article_trajectories"
            ][
                article_id
            ][
                pathway_name
            ]
        )


        initial_state = (
            trajectory[
                "initial_state"
            ]
        )


        if not torch.equal(
            initial_state,
            torch.zeros_like(
                initial_state
            ),
        ):

            BLOCK_10_FRESH_ZERO_INITIAL_STATES = False


        expected_preceding_state = (
            initial_state
        )


        for step in trajectory[
            "steps"
        ]:

            if not torch.equal(
                step[
                    "preceding_state"
                ],
                expected_preceding_state,
            ):

                BLOCK_10_STATE_CHAIN_VALID = False


            if not step[
                "additive_identity"
            ]:

                BLOCK_10_ADDITIVE_UPDATE_VALID = False


            if not step[
                "cumulative_identity"
            ]:

                BLOCK_10_CUMULATIVE_IDENTITY_VALID = False


            if not step[
                "deferred_weight_zero"
            ]:

                BLOCK_10_DEFERRED_WEIGHT_ZERO = False


            if not step[
                "deferred_contribution_zero"
            ]:

                BLOCK_10_DEFERRED_CONTRIBUTION_ZERO = False


            if not step[
                "deferred_state_zero"
            ]:

                BLOCK_10_DEFERRED_STATE_ZERO = False


            if not all(
                [
                    step[
                        "candidate_finite"
                    ],
                    step[
                        "weighted_finite"
                    ],
                    step[
                        "state_finite"
                    ],
                ]
            ):

                BLOCK_10_RECURRENT_VALUES_FINITE = False


            expected_preceding_state = (
                step[
                    "recurrent_state"
                ]
            )


BLOCK_10_CAUSAL_CONTRACT_VALID = all(
    [
        BLOCK_10_FRESH_ZERO_INITIAL_STATES,
        BLOCK_10_STATE_CHAIN_VALID,
        BLOCK_10_ADDITIVE_UPDATE_VALID,
        BLOCK_10_CUMULATIVE_IDENTITY_VALID,
        BLOCK_10_DEFERRED_WEIGHT_ZERO,
        BLOCK_10_DEFERRED_CONTRIBUTION_ZERO,
        BLOCK_10_DEFERRED_STATE_ZERO,
        BLOCK_10_RECURRENT_VALUES_FINITE,
        NOTEBOOK_10_BLOCK_10_FUTURE_CONTEXT_USED is False,
        NOTEBOOK_10_BLOCK_10_SUPERVISION_USED_AS_MODEL_INPUT is False,
        NOTEBOOK_10_BLOCK_10_CROSS_ARTICLE_PROPAGATION is False,
        NOTEBOOK_10_BLOCK_10_CROSS_PATHWAY_RECURRENCE is False,
    ]
)


# =============================================================================
# Objective comparison against Block 8 and Block 9
# =============================================================================

BLOCK_10_BASELINE_PATHWAY_LOSSES = {
    pathway_name:
        float(
            NOTEBOOK_10_PRETRAIN_PATHWAY_LOSSES[
                pathway_name
            ]
        )

    for pathway_name
    in BLOCK_10_PATHWAYS
}


BLOCK_10_BASELINE_TOTAL_OBJECTIVE = float(
    NOTEBOOK_10_PRETRAIN_TOTAL_OBJECTIVE
)


BLOCK_10_BLOCK_9_PATHWAY_LOSSES = {
    pathway_name:
        float(
            NOTEBOOK_10_POSTTRAIN_PATHWAY_LOSSES[
                pathway_name
            ]
        )

    for pathway_name
    in BLOCK_10_PATHWAYS
}


BLOCK_10_BLOCK_9_TOTAL_OBJECTIVE = float(
    NOTEBOOK_10_POSTTRAIN_TOTAL_OBJECTIVE
)


BLOCK_10_POSTTRAIN_PATHWAY_LOSSES = deepcopy(
    BLOCK_10_EVALUATION_A[
        "pathway_losses"
    ]
)


BLOCK_10_POSTTRAIN_TOTAL_OBJECTIVE = float(
    BLOCK_10_EVALUATION_A[
        "total_objective"
    ]
)


BLOCK_10_OBJECTIVE_DELTA = (
    BLOCK_10_POSTTRAIN_TOTAL_OBJECTIVE
    -
    BLOCK_10_BASELINE_TOTAL_OBJECTIVE
)


BLOCK_10_OBJECTIVE_IMPROVED = (
    BLOCK_10_OBJECTIVE_DELTA
    <
    0.0
)


BLOCK_10_PATHWAY_COMPARISON = {
    pathway_name:
        {
            "baseline":
                BLOCK_10_BASELINE_PATHWAY_LOSSES[
                    pathway_name
                ],

            "block_9":
                BLOCK_10_BLOCK_9_PATHWAY_LOSSES[
                    pathway_name
                ],

            "block_10":
                BLOCK_10_POSTTRAIN_PATHWAY_LOSSES[
                    pathway_name
                ],

            "delta_from_baseline":
                (
                    BLOCK_10_POSTTRAIN_PATHWAY_LOSSES[
                        pathway_name
                    ]
                    -
                    BLOCK_10_BASELINE_PATHWAY_LOSSES[
                        pathway_name
                    ]
                ),

            "improved":
                (
                    BLOCK_10_POSTTRAIN_PATHWAY_LOSSES[
                        pathway_name
                    ]
                    <
                    BLOCK_10_BASELINE_PATHWAY_LOSSES[
                        pathway_name
                    ]
                ),

            "block_9_reproduced":
                math.isclose(
                    BLOCK_10_POSTTRAIN_PATHWAY_LOSSES[
                        pathway_name
                    ],
                    BLOCK_10_BLOCK_9_PATHWAY_LOSSES[
                        pathway_name
                    ],
                    rel_tol=
                        0.0,
                    abs_tol=
                        1e-7,
                ),
        }

    for pathway_name
    in BLOCK_10_PATHWAYS
}


BLOCK_10_BLOCK_9_TOTAL_REPRODUCED = math.isclose(
    BLOCK_10_POSTTRAIN_TOTAL_OBJECTIVE,
    BLOCK_10_BLOCK_9_TOTAL_OBJECTIVE,
    rel_tol=
        0.0,
    abs_tol=
        1e-7,
)


BLOCK_10_ALL_PATHWAYS_IMPROVED = all(
    BLOCK_10_PATHWAY_COMPARISON[
        pathway_name
    ][
        "improved"
    ]

    for pathway_name
    in BLOCK_10_PATHWAYS
)


BLOCK_10_ALL_BLOCK_9_PATHWAYS_REPRODUCED = all(
    BLOCK_10_PATHWAY_COMPARISON[
        pathway_name
    ][
        "block_9_reproduced"
    ]

    for pathway_name
    in BLOCK_10_PATHWAYS
)


BLOCK_10_OBJECTIVE_VALUES_FINITE = all(
    [
        math.isfinite(
            BLOCK_10_POSTTRAIN_TOTAL_OBJECTIVE
        ),

        all(
            math.isfinite(
                pathway_loss
            )

            for pathway_loss
            in BLOCK_10_POSTTRAIN_PATHWAY_LOSSES.values()
        ),
    ]
)


BLOCK_10_POSTTRAIN_OBJECTIVE_VALID = all(
    [
        BLOCK_10_OBJECTIVE_VALUES_FINITE,
        BLOCK_10_OBJECTIVE_IMPROVED,
        BLOCK_10_ALL_PATHWAYS_IMPROVED,
        BLOCK_10_BLOCK_9_TOTAL_REPRODUCED,
        BLOCK_10_ALL_BLOCK_9_PATHWAYS_REPRODUCED,
    ]
)


# =============================================================================
# Transformative-Weight numerical validation
# =============================================================================

BLOCK_10_WEIGHT_NUMERICAL_AUDIT = {}


for pathway_name, pathway_dim in BLOCK_10_PATHWAY_DIMENSIONS.items():

    effective_weight = (
        BLOCK_10_EVALUATION_A[
            "effective_weights"
        ][
            pathway_name
        ]
    )


    finite = bool(
        torch.isfinite(
            effective_weight
        ).all()
        .item()
    )


    bounded = bool(
        (
            (
                effective_weight
                >=
                0.0
            )
            &
            (
                effective_weight
                <=
                1.0
            )
        ).all()
        .item()
    )


    deferred_indices = (
        BLOCK_10_DEFERRED_DIMENSION_INDICES[
            pathway_name
        ]
    )


    if deferred_indices:

        deferred_zero = bool(
            torch.equal(
                effective_weight[
                    list(
                        deferred_indices
                    )
                ],
                torch.zeros(
                    len(
                        deferred_indices
                    ),
                    dtype=
                        effective_weight.dtype,
                ),
            )
        )

    else:

        deferred_zero = True


    BLOCK_10_WEIGHT_NUMERICAL_AUDIT[
        pathway_name
    ] = {
        "dimension":
            pathway_dim,

        "active_indices":
            BLOCK_10_ACTIVE_DIMENSION_INDICES[
                pathway_name
            ],

        "deferred_indices":
            deferred_indices,

        "finite":
            finite,

        "bounded":
            bounded,

        "deferred_zero":
            deferred_zero,

        "effective_weight":
            effective_weight.clone(),
    }


BLOCK_10_WEIGHT_NUMERICAL_CONTRACT_VALID = all(
    all(
        [
            audit[
                "finite"
            ],
            audit[
                "bounded"
            ],
            audit[
                "deferred_zero"
            ],
        ]
    )

    for audit
    in BLOCK_10_WEIGHT_NUMERICAL_AUDIT.values()
)


# =============================================================================
# Post-evaluation architecture invariance
# =============================================================================

BLOCK_10_REPRESENTATION_STATE_AFTER = (
    block_10_snapshot_module_collection(
        NOTEBOOK_10_REPRESENTATION_MODULES
    )
)


BLOCK_10_TRANSFORMATIVE_STATE_AFTER = (
    block_10_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_MODULES
    )
)


BLOCK_10_WEIGHT_STATE_AFTER = (
    block_10_snapshot_module_collection(
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES
    )
)


BLOCK_10_REPRESENTATION_UNCHANGED = (
    block_10_nested_state_exact(
        BLOCK_10_REPRESENTATION_STATE_BEFORE,
        BLOCK_10_REPRESENTATION_STATE_AFTER,
    )
)


BLOCK_10_TRANSFORMATIVE_UNCHANGED = (
    block_10_nested_state_exact(
        BLOCK_10_TRANSFORMATIVE_STATE_BEFORE,
        BLOCK_10_TRANSFORMATIVE_STATE_AFTER,
    )
)


BLOCK_10_WEIGHT_UNCHANGED = (
    block_10_nested_state_exact(
        BLOCK_10_WEIGHT_STATE_BEFORE,
        BLOCK_10_WEIGHT_STATE_AFTER,
    )
)


with torch.inference_mode():

    BLOCK_10_EFFECTIVE_WEIGHT_IDENTITY_AFTER = all(
        torch.equal(
            BLOCK_10_EFFECTIVE_WEIGHTS_AT_ENTRY[
                pathway_name
            ],
            NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES[
                pathway_name
            ]()
            .detach()
            .cpu(),
        )

        for pathway_name
        in BLOCK_10_PATHWAYS
    )


BLOCK_10_ALL_PARAMETERS_FROZEN_AFTER = all(
    not parameter.requires_grad

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


BLOCK_10_ALL_MODULES_EVAL_AFTER = all(
    not module.training

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()
)


BLOCK_10_ALL_GRADIENTS_ABSENT_AFTER = all(
    parameter.grad is None

    for module_collection
    in (
        NOTEBOOK_10_REPRESENTATION_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_MODULES,
        NOTEBOOK_10_TRANSFORMATIVE_WEIGHT_MODULES,
    )

    for module
    in module_collection.values()

    for parameter
    in module.parameters()
)


BLOCK_10_ARCHITECTURE_INVARIANT = all(
    [
        BLOCK_10_REPRESENTATION_UNCHANGED,
        BLOCK_10_TRANSFORMATIVE_UNCHANGED,
        BLOCK_10_WEIGHT_UNCHANGED,
        BLOCK_10_EFFECTIVE_WEIGHT_IDENTITY_AFTER,
        BLOCK_10_ALL_PARAMETERS_FROZEN_AFTER,
        BLOCK_10_ALL_MODULES_EVAL_AFTER,
        BLOCK_10_ALL_GRADIENTS_ABSENT_AFTER,
        BLOCK_10_PARAMETER_ACCOUNTING_VALID,
    ]
)


# =============================================================================
# Generalisation-boundary audit
# =============================================================================
#
# Training and evaluation use the same 3-article / 54-sentence pilot corpus.
# The observed objective change is therefore in-sample pilot optimisation.
# =============================================================================

BLOCK_10_SAME_CORPUS_AS_TRAINING = True

BLOCK_10_HELD_OUT_TEST_CORPUS_USED = False

BLOCK_10_OUT_OF_SAMPLE_EVALUATION = False

BLOCK_10_CROSS_TOPIC_GENERALISATION_TESTED = False

BLOCK_10_CROSS_SOURCE_GENERALISATION_TESTED = False

BLOCK_10_TEMPORAL_GENERALISATION_TESTED = False

BLOCK_10_PRODUCTION_READINESS_CLAIMED = False

BLOCK_10_GENERALISATION_CLAIM_ALLOWED = False

BLOCK_10_EXPANDED_CORPUS_REQUIRED_FOR_GENERALISATION = True

BLOCK_10_RESULT_CLASSIFICATION = (
    "controlled_in_sample_pilot_optimisation"
)


BLOCK_10_GENERALISATION_BOUNDARY_VALID = all(
    [
        BLOCK_10_SAME_CORPUS_AS_TRAINING,
        BLOCK_10_HELD_OUT_TEST_CORPUS_USED is False,
        BLOCK_10_OUT_OF_SAMPLE_EVALUATION is False,
        BLOCK_10_CROSS_TOPIC_GENERALISATION_TESTED is False,
        BLOCK_10_CROSS_SOURCE_GENERALISATION_TESTED is False,
        BLOCK_10_TEMPORAL_GENERALISATION_TESTED is False,
        BLOCK_10_PRODUCTION_READINESS_CLAIMED is False,
        BLOCK_10_GENERALISATION_CLAIM_ALLOWED is False,
        BLOCK_10_EXPANDED_CORPUS_REQUIRED_FOR_GENERALISATION,
    ]
)


# =============================================================================
# Execution-boundary validation
# =============================================================================

BLOCK_10_EXECUTION_BOUNDARY_VALID = all(
    [
        NOTEBOOK_10_BLOCK_10_REPRESENTATION_FORWARD_EXECUTED
        is False,

        NOTEBOOK_10_BLOCK_10_TRANSFORMATIVE_FORWARD_EXECUTED
        is True,

        NOTEBOOK_10_BLOCK_10_WEIGHT_FORWARD_EXECUTED
        is True,

        NOTEBOOK_10_BLOCK_10_RECURRENT_STATE_CREATED
        is True,

        NOTEBOOK_10_BLOCK_10_RECURRENT_UPDATE_EXECUTED
        is True,

        NOTEBOOK_10_BLOCK_10_LOSS_CALCULATED
        is True,

        NOTEBOOK_10_BLOCK_10_REPEAT_EVALUATION_EXECUTED
        is True,

        NOTEBOOK_10_BLOCK_10_OPTIMIZER_CREATED
        is False,

        NOTEBOOK_10_BLOCK_10_ZERO_GRAD_EXECUTED
        is False,

        NOTEBOOK_10_BLOCK_10_BACKWARD_PASS_EXECUTED
        is False,

        NOTEBOOK_10_BLOCK_10_GRADIENT_CLIPPING_EXECUTED
        is False,

        NOTEBOOK_10_BLOCK_10_OPTIMIZER_STEP_EXECUTED
        is False,

        NOTEBOOK_10_BLOCK_10_PARAMETER_UPDATE_EXECUTED
        is False,

        NOTEBOOK_10_BLOCK_10_TRAINING_EXECUTED
        is False,

        NOTEBOOK_10_BLOCK_10_FUTURE_CONTEXT_USED
        is False,

        NOTEBOOK_10_BLOCK_10_SUPERVISION_USED_AS_MODEL_INPUT
        is False,
    ]
)


# =============================================================================
# Final block validation
# =============================================================================

NOTEBOOK_10_BLOCK_10_ERRORS = []


BLOCK_10_VALIDATION_CHECKS = {
    "prerequisites_invalid":
        BLOCK_10_PREREQUISITES_VALID,

    "pathway_dimensions_invalid":
        BLOCK_10_PATHWAY_DIMENSIONS_VALID,

    "dimension_partitions_invalid":
        BLOCK_10_DIMENSION_PARTITIONS_VALID,

    "canonical_corpus_invalid":
        BLOCK_10_CANONICAL_CORPUS_VALID,

    "device_dtype_invalid":
        BLOCK_10_DEVICE_DTYPE_VALID,

    "trained_weight_state_mismatch":
        BLOCK_10_TRAINED_WEIGHT_STATE_EXACT,

    "weight_state_changed_during_freeze":
        BLOCK_10_WEIGHT_STATE_PRESERVED_BY_FREEZE,

    "parameter_accounting_invalid":
        BLOCK_10_PARAMETER_ACCOUNTING_VALID,

    "pathway_representations_invalid":
        BLOCK_10_PATHWAY_REPRESENTATIONS_VALID,

    "target_mask_contract_invalid":
        BLOCK_10_TARGET_MASK_CONTRACT_VALID,

    "objective_counts_changed":
        BLOCK_10_OBJECTIVE_COUNTS_MATCH_BLOCK_8,

    "trained_effective_weight_mismatch":
        BLOCK_10_TRAINED_EFFECTIVE_WEIGHT_IDENTITY,

    "posttrain_objective_invalid":
        BLOCK_10_POSTTRAIN_OBJECTIVE_VALID,

    "posttrain_execution_not_deterministic":
        BLOCK_10_COMPLETE_EXECUTION_DETERMINISTIC,

    "causal_contract_invalid":
        BLOCK_10_CAUSAL_CONTRACT_VALID,

    "weight_numerical_contract_invalid":
        BLOCK_10_WEIGHT_NUMERICAL_CONTRACT_VALID,

    "architecture_not_invariant":
        BLOCK_10_ARCHITECTURE_INVARIANT,

    "generalisation_boundary_invalid":
        BLOCK_10_GENERALISATION_BOUNDARY_VALID,

    "execution_boundary_invalid":
        BLOCK_10_EXECUTION_BOUNDARY_VALID,
}


for error_name, condition in BLOCK_10_VALIDATION_CHECKS.items():

    if not condition:

        NOTEBOOK_10_BLOCK_10_ERRORS.append(
            error_name
        )


NOTEBOOK_10_POSTTRAIN_RECURRENT_VALIDATION_VALID = (
    len(
        NOTEBOOK_10_BLOCK_10_ERRORS
    )
    ==
    0
)


NOTEBOOK_10_BLOCK_10_VALID = (
    NOTEBOOK_10_POSTTRAIN_RECURRENT_VALIDATION_VALID
)


if not NOTEBOOK_10_BLOCK_10_VALID:

    raise RuntimeError(
        "Notebook 10 Block 10 post-training recurrent validation failed. "
        f"Errors: {NOTEBOOK_10_BLOCK_10_ERRORS}"
    )


NOTEBOOK_10_BLOCK_10_COMPLETE = True


# =============================================================================
# Stable Block 10 handover
# =============================================================================

NOTEBOOK_10_VALIDATED_POSTTRAIN_PATHWAY_LOSSES = deepcopy(
    BLOCK_10_POSTTRAIN_PATHWAY_LOSSES
)


NOTEBOOK_10_VALIDATED_POSTTRAIN_TOTAL_OBJECTIVE = float(
    BLOCK_10_POSTTRAIN_TOTAL_OBJECTIVE
)


NOTEBOOK_10_VALIDATED_POSTTRAIN_OBJECTIVE_DELTA = float(
    BLOCK_10_OBJECTIVE_DELTA
)


NOTEBOOK_10_VALIDATED_POSTTRAIN_RECURRENT_TRAJECTORIES = deepcopy(
    BLOCK_10_EVALUATION_A[
        "article_trajectories"
    ]
)


NOTEBOOK_10_VALIDATED_POSTTRAIN_EFFECTIVE_WEIGHTS = {
    pathway_name:
        effective_weight.clone()

    for pathway_name, effective_weight
    in BLOCK_10_EVALUATION_A[
        "effective_weights"
    ].items()
}


NOTEBOOK_10_POSTTRAIN_DETERMINISTIC = (
    BLOCK_10_COMPLETE_EXECUTION_DETERMINISTIC
)


NOTEBOOK_10_POSTTRAIN_ARCHITECTURE_INVARIANT = (
    BLOCK_10_ARCHITECTURE_INVARIANT
)


NOTEBOOK_10_GENERALISATION_BOUNDARY_VALID = (
    BLOCK_10_GENERALISATION_BOUNDARY_VALID
)


NOTEBOOK_10_RESULT_CLASSIFICATION = (
    BLOCK_10_RESULT_CLASSIFICATION
)


NOTEBOOK_10_FINAL_PILOT_AUDIT_READY = (
    NOTEBOOK_10_BLOCK_10_VALID
)


# =============================================================================
# Block summary
# =============================================================================

NOTEBOOK_10_BLOCK_10_SUMMARY = {
    "block":
        NOTEBOOK_10_BLOCK_10,

    "block_name":
        NOTEBOOK_10_BLOCK_10_NAME,

    "version":
        NOTEBOOK_10_BLOCK_10_VERSION,

    "executed_at_utc":
        BLOCK_10_EXECUTED_AT_UTC,

    "article_count":
        BLOCK_10_ARTICLE_COUNT,

    "sentence_count":
        BLOCK_10_SENTENCE_COUNT,

    "baseline_pathway_losses":
        deepcopy(
            BLOCK_10_BASELINE_PATHWAY_LOSSES
        ),

    "baseline_total_objective":
        BLOCK_10_BASELINE_TOTAL_OBJECTIVE,

    "block_9_pathway_losses":
        deepcopy(
            BLOCK_10_BLOCK_9_PATHWAY_LOSSES
        ),

    "block_9_total_objective":
        BLOCK_10_BLOCK_9_TOTAL_OBJECTIVE,

    "validated_posttrain_pathway_losses":
        deepcopy(
            BLOCK_10_POSTTRAIN_PATHWAY_LOSSES
        ),

    "validated_posttrain_total_objective":
        BLOCK_10_POSTTRAIN_TOTAL_OBJECTIVE,

    "objective_delta":
        BLOCK_10_OBJECTIVE_DELTA,

    "objective_improved":
        BLOCK_10_OBJECTIVE_IMPROVED,

    "block_9_reproduced":
        BLOCK_10_BLOCK_9_TOTAL_REPRODUCED,

    "deterministic":
        BLOCK_10_COMPLETE_EXECUTION_DETERMINISTIC,

    "causal_contract_valid":
        BLOCK_10_CAUSAL_CONTRACT_VALID,

    "architecture_invariant":
        BLOCK_10_ARCHITECTURE_INVARIANT,

    "generalisation_boundary_valid":
        BLOCK_10_GENERALISATION_BOUNDARY_VALID,

    "result_classification":
        BLOCK_10_RESULT_CLASSIFICATION,

    "final_pilot_audit_ready":
        NOTEBOOK_10_FINAL_PILOT_AUDIT_READY,

    "block_valid":
        NOTEBOOK_10_BLOCK_10_VALID,

    "block_complete":
        NOTEBOOK_10_BLOCK_10_COMPLETE,
}


# =============================================================================
# Final report
# =============================================================================

print("=" * 72)

print(
    "Media AI — Notebook 10, Block 10: "
    "Post-Training Recurrent Validation, Objective Comparison "
    "and Generalisation-Boundary Audit"
)

print("=" * 72)

print(
    f"Block version                : "
    f"{NOTEBOOK_10_BLOCK_10_VERSION}"
)

print(
    f"Execution device             : "
    f"{BLOCK_10_EXECUTION_DEVICE}"
)

print(
    f"Execution dtype              : "
    f"{BLOCK_10_EXECUTION_DTYPE}"
)

print("-" * 72)

print(
    "Post-training corpus evaluation"
)

print(
    f"Articles evaluated           : "
    f"{BLOCK_10_ARTICLE_COUNT}"
)

print(
    f"Sentences evaluated          : "
    f"{BLOCK_10_SENTENCE_COUNT}"
)

print(
    f"Fresh zero initial states    : "
    f"{BLOCK_10_FRESH_ZERO_INITIAL_STATES}"
)

print(
    f"State chain valid            : "
    f"{BLOCK_10_STATE_CHAIN_VALID}"
)

print(
    f"Additive update valid        : "
    f"{BLOCK_10_ADDITIVE_UPDATE_VALID}"
)

print(
    f"Cumulative identity valid    : "
    f"{BLOCK_10_CUMULATIVE_IDENTITY_VALID}"
)

print(
    f"Deferred weight zero         : "
    f"{BLOCK_10_DEFERRED_WEIGHT_ZERO}"
)

print(
    f"Deferred contribution zero   : "
    f"{BLOCK_10_DEFERRED_CONTRIBUTION_ZERO}"
)

print(
    f"Deferred state zero          : "
    f"{BLOCK_10_DEFERRED_STATE_ZERO}"
)

print(
    f"Future context used          : "
    f"{NOTEBOOK_10_BLOCK_10_FUTURE_CONTEXT_USED}"
)

print(
    f"Supervision used as input    : "
    f"{NOTEBOOK_10_BLOCK_10_SUPERVISION_USED_AS_MODEL_INPUT}"
)

print("-" * 72)

print(
    "Objective comparison"
)


for pathway_name in BLOCK_10_PATHWAYS:

    comparison = (
        BLOCK_10_PATHWAY_COMPARISON[
            pathway_name
        ]
    )


    print(
        f"{pathway_name:<14} baseline masked MSE : "
        f"{comparison['baseline']:.8f}"
    )

    print(
        f"{pathway_name:<14} Block 9 masked MSE  : "
        f"{comparison['block_9']:.8f}"
    )

    print(
        f"{pathway_name:<14} Block 10 masked MSE : "
        f"{comparison['block_10']:.8f}"
    )

    print(
        f"{pathway_name:<14} delta from baseline : "
        f"{comparison['delta_from_baseline']:.8f}"
    )

    print(
        f"{pathway_name:<14} improved            : "
        f"{comparison['improved']}"
    )

    print(
        f"{pathway_name:<14} Block 9 reproduced  : "
        f"{comparison['block_9_reproduced']}"
    )


print(
    f"Immutable Block 8 baseline   : "
    f"{BLOCK_10_BASELINE_TOTAL_OBJECTIVE:.8f}"
)

print(
    f"Block 9 reported objective   : "
    f"{BLOCK_10_BLOCK_9_TOTAL_OBJECTIVE:.8f}"
)

print(
    f"Block 10 reproduced objective: "
    f"{BLOCK_10_POSTTRAIN_TOTAL_OBJECTIVE:.8f}"
)

print(
    f"Objective delta              : "
    f"{BLOCK_10_OBJECTIVE_DELTA:.8f}"
)

print(
    f"Objective improved           : "
    f"{BLOCK_10_OBJECTIVE_IMPROVED}"
)

print(
    f"Block 9 objective reproduced : "
    f"{BLOCK_10_BLOCK_9_TOTAL_REPRODUCED}"
)

print("-" * 72)

print(
    "Deterministic post-training validation"
)

print(
    f"Repeated evaluation executed : "
    f"{NOTEBOOK_10_BLOCK_10_REPEAT_EVALUATION_EXECUTED}"
)

print(
    f"Complete execution deterministic: "
    f"{BLOCK_10_COMPLETE_EXECUTION_DETERMINISTIC}"
)

print(
    f"Trained Weight state exact   : "
    f"{BLOCK_10_TRAINED_WEIGHT_STATE_EXACT}"
)

print(
    f"Trained effective Weight exact: "
    f"{BLOCK_10_TRAINED_EFFECTIVE_WEIGHT_IDENTITY}"
)

print("-" * 72)

print(
    "Transformative-Weight validation"
)


for pathway_name in BLOCK_10_PATHWAYS:

    audit = (
        BLOCK_10_WEIGHT_NUMERICAL_AUDIT[
            pathway_name
        ]
    )


    print(
        f"{pathway_name:<14} dimension           : "
        f"{audit['dimension']}"
    )

    print(
        f"{pathway_name:<14} active dimensions   : "
        f"{list(audit['active_indices'])}"
    )

    print(
        f"{pathway_name:<14} deferred dimensions : "
        f"{list(audit['deferred_indices'])}"
    )

    print(
        f"{pathway_name:<14} finite              : "
        f"{audit['finite']}"
    )

    print(
        f"{pathway_name:<14} bounded             : "
        f"{audit['bounded']}"
    )

    print(
        f"{pathway_name:<14} deferred zero       : "
        f"{audit['deferred_zero']}"
    )


print("-" * 72)

print(
    "Architecture invariance"
)

print(
    f"Representation parameters    : "
    f"{BLOCK_10_REPRESENTATION_PARAMETER_COUNT:,}"
)

print(
    f"Transformative parameters    : "
    f"{BLOCK_10_TRANSFORMATIVE_PARAMETER_COUNT:,}"
)

print(
    f"Transformative Weight params : "
    f"{BLOCK_10_TRANSFORMATIVE_WEIGHT_PARAMETER_COUNT:,}"
)

print(
    f"Total model parameters       : "
    f"{BLOCK_10_TOTAL_PARAMETER_COUNT:,}"
)

print(
    f"Representation unchanged     : "
    f"{BLOCK_10_REPRESENTATION_UNCHANGED}"
)

print(
    f"Transformative unchanged     : "
    f"{BLOCK_10_TRANSFORMATIVE_UNCHANGED}"
)

print(
    f"Transform Weight unchanged   : "
    f"{BLOCK_10_WEIGHT_UNCHANGED}"
)

print(
    f"Effective Weight identity    : "
    f"{BLOCK_10_EFFECTIVE_WEIGHT_IDENTITY_AFTER}"
)

print(
    f"All parameters frozen        : "
    f"{BLOCK_10_ALL_PARAMETERS_FROZEN_AFTER}"
)

print(
    f"All modules eval             : "
    f"{BLOCK_10_ALL_MODULES_EVAL_AFTER}"
)

print(
    f"All gradients absent         : "
    f"{BLOCK_10_ALL_GRADIENTS_ABSENT_AFTER}"
)

print("-" * 72)

print(
    "Generalisation-boundary audit"
)

print(
    f"Evaluation corpus            : "
    f"same_as_training"
)

print(
    f"Held-out test corpus used    : "
    f"{BLOCK_10_HELD_OUT_TEST_CORPUS_USED}"
)

print(
    f"Out-of-sample evaluation     : "
    f"{BLOCK_10_OUT_OF_SAMPLE_EVALUATION}"
)

print(
    f"Cross-topic tested           : "
    f"{BLOCK_10_CROSS_TOPIC_GENERALISATION_TESTED}"
)

print(
    f"Cross-source tested          : "
    f"{BLOCK_10_CROSS_SOURCE_GENERALISATION_TESTED}"
)

print(
    f"Temporal generalisation tested: "
    f"{BLOCK_10_TEMPORAL_GENERALISATION_TESTED}"
)

print(
    f"Generalisation claim allowed : "
    f"{BLOCK_10_GENERALISATION_CLAIM_ALLOWED}"
)

print(
    f"Result classification        : "
    f"{BLOCK_10_RESULT_CLASSIFICATION}"
)

print(
    f"Expanded corpus required     : "
    f"{BLOCK_10_EXPANDED_CORPUS_REQUIRED_FOR_GENERALISATION}"
)

print("-" * 72)

print(
    "Execution boundary"
)

print(
    f"Representation forward       : "
    f"{NOTEBOOK_10_BLOCK_10_REPRESENTATION_FORWARD_EXECUTED}"
)

print(
    f"Transform forward            : "
    f"{NOTEBOOK_10_BLOCK_10_TRANSFORMATIVE_FORWARD_EXECUTED}"
)

print(
    f"Weight forward               : "
    f"{NOTEBOOK_10_BLOCK_10_WEIGHT_FORWARD_EXECUTED}"
)

print(
    f"Recurrent state created      : "
    f"{NOTEBOOK_10_BLOCK_10_RECURRENT_STATE_CREATED}"
)

print(
    f"Recurrent update executed    : "
    f"{NOTEBOOK_10_BLOCK_10_RECURRENT_UPDATE_EXECUTED}"
)

print(
    f"Loss calculated              : "
    f"{NOTEBOOK_10_BLOCK_10_LOSS_CALCULATED}"
)

print(
    f"Optimizer created            : "
    f"{NOTEBOOK_10_BLOCK_10_OPTIMIZER_CREATED}"
)

print(
    f"Backward pass executed       : "
    f"{NOTEBOOK_10_BLOCK_10_BACKWARD_PASS_EXECUTED}"
)

print(
    f"Gradient clipping executed   : "
    f"{NOTEBOOK_10_BLOCK_10_GRADIENT_CLIPPING_EXECUTED}"
)

print(
    f"Parameter update executed    : "
    f"{NOTEBOOK_10_BLOCK_10_PARAMETER_UPDATE_EXECUTED}"
)

print(
    f"Training executed            : "
    f"{NOTEBOOK_10_BLOCK_10_TRAINING_EXECUTED}"
)

print("-" * 72)

print(
    f"Post-training objective valid: "
    f"{BLOCK_10_POSTTRAIN_OBJECTIVE_VALID}"
)

print(
    f"Causal contract valid        : "
    f"{BLOCK_10_CAUSAL_CONTRACT_VALID}"
)

print(
    f"Architecture invariant       : "
    f"{BLOCK_10_ARCHITECTURE_INVARIANT}"
)

print(
    f"Generalisation boundary valid: "
    f"{BLOCK_10_GENERALISATION_BOUNDARY_VALID}"
)

print(
    f"Final pilot audit ready      : "
    f"{NOTEBOOK_10_FINAL_PILOT_AUDIT_READY}"
)

print(
    f"Block valid                  : "
    f"{NOTEBOOK_10_BLOCK_10_VALID}"
)

print(
    f"Block complete               : "
    f"{NOTEBOOK_10_BLOCK_10_COMPLETE}"
)

print("=" * 72)

print(
    f"The trained Media AI recurrent architecture was evaluated across "
    f"{BLOCK_10_ARTICLE_COUNT} article(s) and "
    f"{BLOCK_10_SENTENCE_COUNT} canonical sentence(s) without "
    "additional optimisation."
)

print(
    "The post-training factual, psychological and social objectives "
    "were reproduced from the fixed Block 9 Transformative-Weight state."
)

print(
    f"The equal-pathway objective changed from the immutable Block 8 "
    f"baseline of {BLOCK_10_BASELINE_TOTAL_OBJECTIVE:.8f} to "
    f"{BLOCK_10_POSTTRAIN_TOTAL_OBJECTIVE:.8f}, corresponding to "
    f"a delta of {BLOCK_10_OBJECTIVE_DELTA:.8f}."
)

print(
    "Repeated complete recurrent evaluation reproduced the same weighted "
    "transformations, recurrent trajectories and objective values exactly."
)

print(
    "Article-boundary zero resets, causal state continuity and "
    "deferred-dimension inactivity remained valid."
)

print(
    "The representation, transformative and trained Transformative-Weight "
    "parameter states remained exactly unchanged during post-training "
    "validation."
)

print(
    "No optimiser, backward pass, gradient clipping, parameter update or "
    "additional training operation was executed."
)

print(
    "Because optimisation and evaluation use the same three-article pilot "
    "corpus, the observed objective improvement is classified strictly as "
    "controlled in-sample pilot optimisation and not as evidence of "
    "out-of-sample generalisation."
)

print(
    "Notebook 10 may now proceed to the final pilot evaluation, "
    "architectural completion audit and persistent project handover."
)

print("=" * 72)
```
